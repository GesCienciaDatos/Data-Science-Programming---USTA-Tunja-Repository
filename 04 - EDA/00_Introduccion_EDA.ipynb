## ¿Qué es el Análisis Exploratorio de Datos (EDA)? 📊

El Análisis Exploratorio de Datos (EDA, por sus siglas en inglés) es una fase crítica y fundamental en el proceso de análisis de datos. Su objetivo principal es **comprender las características intrínsecas de los datos** que tenemos entre manos antes de aplicar cualquier modelo complejo.

Implica resumir visual y estadísticamente las características principales de un conjunto de datos, a menudo utilizando representaciones gráficas, para descubrir:
- Patrones ocultos
- Tendencias a lo largo del tiempo o grupos
- Anomalías (valores atípicos o *outliers*)
- Información valiosa inicial (*insights*)

El EDA ayuda a los analistas y científicos de datos a formular hipótesis correctas, identificar relaciones entre variables y guiar la dirección de todo el análisis posterior.
