## Objetivos de aprendizaje 🔎

En este quinto módulo abordaremos una de las tareas más críticas (y a veces frustrantes) para cualquier Científico de Datos: la **Preparación y Limpieza de Datos**.

El material está organizado de forma progresiva en los siguientes cuadernos interactivos:

1. **[Introducción a la Preparación de Datos](00_Introduccion_Data_Preparation.ipynb)** *(Este cuaderno)*: ¿Qué es la Preparación de Datos?
2. **[Valores Faltantes (Missing Values)](01_Valores_Faltantes_Data_Preparation.ipynb)**: Cómo lidiar con datos ausentes mediante eliminación e imputación.
3. **[Escalado de Características (Feature Scaling)](02_Escalado_Caracteristicas_Data_Preparation.ipynb)**: Normalización, estandarización y sensibilidad de los algoritmos.
4. **[Fechas y Datos Inconsistentes](03_Fechas_y_Datos_Inconsistentes_Data_Preparation.ipynb)**: Parsing de fechas y corrección de texto con Fuzzy Matching.

> 🧗 **Nota:** Los temas con mayor dificultad teórica y técnica estarán marcados con un ícono de escalador.

## ¿Qué es la Preparación de Datos? 🧹

La preparación de datos es una parte clave de la ciencia de datos, pero puede llegar a ser profundamente frustrante. ¿Por qué algunos de tus campos de texto aparecen como basura? ¿Qué deberías hacer con esos valores en blanco? ¿Por qué tus fechas no tienen el formato correcto? ¿Cómo puedes limpiar rápidamente un conjunto de datos que fue ingresado inconsistentemente por humanos?

A lo largo de los cuadernos de este módulo, **¡aprenderás a solucionar todos estos problemas!**

### ¿Por qué es tan importante?
De acuerdo con la [Encuesta del Estado de la Ciencia de Datos 2022 de Anaconda](https://www.anaconda.com/resources/whitepapers/state-of-data-science-report-2022), los profesionales de datos gastan consistentemente **la mayor parte de su tiempo** preparando, limpiando y visualizando datos. Un modelo de Inteligencia Artificial solo será tan bueno como los datos con los que se alimenta (*Basura entra, Basura sale*).
