## ¿Qué es la Preparación de Datos y por qué es tan crítica? 🧹

En proyectos reales, los datos casi nunca se presentan en formatos limpios y listos para modelar. Diversos estudios de la industria concuerdan en que los científicos y analistas de datos dedican **entre el 70% y el 80% de su tiempo** a limpiar, transformar y validar los datos antes de poder entrenar un solo modelo.

El principio fundamental es **«Garbage In, Garbage Out» (GIGO)**: si alimentamos un algoritmo sofisticado con datos corruptos, incompletos, en escalas desproporcionadas o con inconsistencias textuales, las predicciones resultantes serán erróneas, sesgadas e inutilizables para la toma de decisiones.

---
## Los Tres Pilares de este Módulo 🏛️

### 1. Manejo Riguroso de Valores Faltantes (*Missing Values*)
- **Diagnóstico de la causa**: ¿El dato falta porque no fue registrado o porque no existe conceptualmente?
- **Estrategias de supresión**: Eliminación de filas completas o descarte de columnas con excesivo porcentaje de nulos.
- **Estrategias de imputación**: Relleno con estadísticos univariados (media, mediana, moda) o algoritmos multivariados basados en vecinos cercanos (KNN).

---
### 2. Escalado de Características (*Feature Scaling*)
- **Sensibilidad de los algoritmos**: Modelos basados en distancias euclidianas (KNN, K-Means), modelos basados en gradientes (Regresión Lineal/Logística, Redes Neuronales) y métodos de regularización (Ridge/Lasso) son altamente vulnerables a diferencias de escala.
- **Técnicas**: Escalado Min-Max (rango de 0 a 1), Estandarización Z-Score (media = 0, varianza = 1) y Escalado Robusto mediante medianas e intercuartiles (IQR) para amortiguar el efecto de valores extremos.

---
### 3. Parseo de Fechas e Inconsistencias de Texto
- **Fechas como tipo temporal nativo**: Conversión de cadenas (`object`) a `datetime64[ns]` para realizar operaciones aritméticas de tiempo, cálculo de duraciones y extracción de estacionalidad.
- **Limpieza de entradas irregulares**: Unificación de mayúsculas/minúsculas, eliminación de espacios en blanco espurios y corrección de variaciones ortográficas mediante coincidencia difusa (*Fuzzy Matching*) basada en la distancia de Levenshtein.

### Demostración Rápida: Diagnóstico de Datos Faltantes

Carguemos un dataset de demostración con datos faltantes e inspeccionemos su impacto porcentual:

```python
# Dataset de prueba con valores nulos integrados (Titanic)
df_titanic = sns.load_dataset('titanic')

# Diagnóstico rápido de valores faltantes por columna
nulos_por_columna = df_titanic.isnull().sum()
porcentaje_nulos = (nulos_por_columna / len(df_titanic)) * 100

resumen_nulos = pd.DataFrame({
    'Valores_Faltantes': nulos_por_columna,
    'Porcentaje (%)': porcentaje_nulos
})

# Mostramos las columnas con al menos un valor nulo
resumen_nulos[resumen_nulos['Valores_Faltantes'] > 0].sort_values(by='Porcentaje (%)', ascending=False)
```
