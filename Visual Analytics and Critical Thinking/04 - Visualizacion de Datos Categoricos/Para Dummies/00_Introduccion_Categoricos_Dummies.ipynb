## ¿Datos categóricos? Piensa en cajones de un armario 🗄️

Imagina que estás ordenando tu clóset. No mides la ropa en centímetros — la **separas en cajones**:
- Cajón "Camisetas"
- Cajón "Pantalones"
- Cajón "Medias"

Eso es exactamente un dato categórico: **una etiqueta que dice a qué grupo pertenece algo**, no un número.

Ahora, no todos los cajones son iguales. Hay tres tipos, y confundirlos es el error #1 al hacer gráficos:

1. **Cajones sin jerarquía** (nominal): "Camisetas", "Pantalones", "Medias" — ninguno es "mejor" que otro.
2. **Cajones con jerarquía** (ordinal): "Ropa de diario", "Ropa buena", "Ropa de gala" — sí hay un orden.
3. **Cajones creados a partir de una medida** (intervalo): "Ropa de bebé (0-2 años)", "Ropa de niño (3-10 años)" — vienen de una edad real, agrupada en rangos.

---
## Los 3 tipos, explicados como si tuvieras 5 años 🧸

| Tipo | Analogía | Ejemplo real de negocio | ¿Puedo reordenar las categorías libremente? |
|---|---|---|---|
| **Nominal** | Sabores de helado (fresa, chocolate, vainilla) — ninguno "vale más" | Ciudad, género, categoría de producto | ✅ Sí, el orden es cosmético |
| **Ordinal** | Medallas olímpicas (oro, plata, bronce) — hay un orden claro | Nivel de satisfacción, calificación (⭐) | ❌ No, romperías la jerarquía |
| **Intervalo (por rangos)** | Tallas de ropa de bebé por edad (0-6m, 6-12m, 1-2 años) | Rango de edad, rango de ingresos | ❌ No, y además la distancia entre cajones significa algo real |

**El truco para no confundirte:** pregúntate *"si mezclo el orden de las categorías en mi gráfico, ¿pierdo información?"* Si la respuesta es sí → es ordinal o de intervalo. Si es no → es nominal.

````python
# ============================================================
# Veámoslo con código: mismo dato, tratado bien y tratado mal
# ============================================================

import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams.update({'axes.spines.top': False, 'axes.spines.right': False})

# Satisfacción de clientes (ORDINAL: Bajo < Medio < Alto)
niveles = ['Bajo', 'Medio', 'Alto']
conteo = [8, 15, 27]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))

# Bien hecho: respeta el orden Bajo -> Medio -> Alto
axes[0].bar(niveles, conteo, color=['#ef4444', '#f59e0b', '#10b981'], edgecolor='white')
axes[0].set_title('✅ Bien: se lee de un vistazo que la satisfacción SUBE', fontweight='bold', fontsize=10)

# Mal hecho: alguien lo ordenó "alfabéticamente" (Alto, Bajo, Medio)
niveles_mal = ['Alto', 'Bajo', 'Medio']
conteo_mal = [27, 8, 15]
axes[1].bar(niveles_mal, conteo_mal, color='#94a3b8', edgecolor='white')
axes[1].set_title('❌ Mal: parece que la satisfacción baja y luego sube al azar', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()

print("👉 Mismo dato, dos gráficos. El de la derecha CONFUNDE porque ignoró que")
print("   'Bajo, Medio, Alto' tiene un orden natural que hay que respetar siempre.")
````

## ¿Y los rangos de edad? El cajón que "sí mide distancia" 📏

Cuando conviertes una edad (23 años, 45 años...) en un rango ("18-25", "26-35"...), sigues teniendo una categoría — pero una especial: sabes que "26-35" está **más cerca** de "18-25" que de "51+". Esa distancia real es lo que la hace "de intervalo" y no puramente ordinal.

```python
# En pandas, esto se hace con pd.cut():
rangos = pd.cut(df['edad'], bins=[0, 25, 35, 50, 100],
                 labels=['18-25', '26-35', '36-50', '51+'])
```

📊 **En Power BI:** puedes crear estos rangos con una columna calculada usando `SWITCH` o agrupando (`Group`) un campo numérico.

📈 **En Tableau:** clic derecho sobre el campo numérico → "Create" → "Bins".

## Resumen de bolsillo 📱

Guarda esta tabla para la próxima vez que abras un dataset nuevo:

| Me pregunto... | Si la respuesta es "sí"... | Entonces es... |
|---|---|---|
| ¿Son solo etiquetas sin ningún orden? | Sí | **Nominal** (ciudad, género, categoría) |
| ¿Hay un orden claro pero no sé "cuánto" separa a cada nivel? | Sí | **Ordinal** (satisfacción, calificación) |
| ¿Viene de agrupar un número real en rangos? | Sí | **Intervalo por rangos** (edad, ingresos) |

> 💡 **Tip final:** antes de hacer cualquier gráfico, identifica el tipo de dato categórico. Te ahorrarás el 90% de los errores de visualización — y en el próximo cuaderno (01) veremos exactamente qué gráfico usar para cada uno.
