## El menú de gráficos: como pedir en un restaurante 🍽️

Ya sabes distinguir tus "cajones" de datos (nominal, ordinal, con rangos). Ahora toca elegir **qué gráfico pedir** según dos preguntas simples, como si fueras a un restaurante:

1. **¿Cuántas categorías tengo?** (¿el menú tiene 4 platillos o 40?)
2. **¿Necesito comparar, ver composición, o ver jerarquía?** (¿quiero un plato fuerte, un combo, o un menú por secciones?)

---
## La Guía Rápida: ¿Qué gráfico uso? 🎯

| Situación | Gráfico recomendado | Por qué |
|---|---|---|
| Tengo 4-6 categorías, quiero comparar sus totales | **Barras verticales** | Simple, el cerebro compara alturas fácilmente |
| Tengo categorías con nombres largos | **Barras horizontales** | El texto cabe mejor |
| Quiero comparar 2023 vs 2024 por categoría | **Barras agrupadas** | Dos barras juntas para comparar |
| Quiero ver la composición interna | **Barras apiladas** | Cada barra = 100%, dividida en partes |
| Solo me importa el % de cada parte, no el total | **Barras 100% apiladas** | Todas las barras miden lo mismo (100%), solo cambia la composición |
| Tengo 2-3 categorías, diferencia obvia | **Donut o waffle** | Reemplazo elegante de la torta |
| Tengo 15+ categorías | **Lollipop (paleta)** | Menos ruido visual que barras anchas |
| Tengo categoría → subcategoría | **Treemap** | El tamaño del rectángulo = valor |

### ¿Y cuándo SÍ se puede usar torta/donut?
Solo cuando tienes **2 o 3 categorías máximo** y las diferencias son grandes y obvias (70% vs 30%, por ejemplo). Si tienes más categorías o diferencias pequeñas → **siempre barras**.

````python
# ============================================================
# Las 4 variantes de barras más útiles — con código simple y comentado
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

categorias = ['Electrónica', 'Ropa', 'Hogar', 'Alimentos', 'Deportes']
ventas_2023 = [45200, 32100, 28900, 38700, 21300]
ventas_2024 = [52300, 35600, 31200, 41800, 25100]
colores = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Las 4 variantes de barras más útiles — Mismos datos, diferentes respuestas', fontsize=12, fontweight='bold')

bars1 = axes[0,0].bar(categorias, ventas_2024, color=colores, edgecolor='white')
for bar, val in zip(bars1, ventas_2024):
    axes[0,0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500, f'${val/1000:.0f}k', ha='center', fontsize=9, fontweight='bold')
axes[0,0].set_title('① Barras Verticales: ¿Cuánto vende cada categoría?', fontweight='bold')
axes[0,0].tick_params(axis='x', rotation=20)

idx = np.argsort(ventas_2024)
cats_ord = [categorias[i] for i in idx]
vals_ord = [ventas_2024[i] for i in idx]
cols_ord = [colores[i] for i in idx]
bars2 = axes[0,1].barh(cats_ord, vals_ord, color=cols_ord, edgecolor='white')
for bar, val in zip(bars2, vals_ord):
    axes[0,1].text(bar.get_width() + 300, bar.get_y() + bar.get_height()/2, f'${val/1000:.0f}k', va='center', fontsize=9)
axes[0,1].set_title('② Barras Horizontales: Mejor para nombres largos', fontweight='bold')

x = np.arange(len(categorias))
w = 0.35
axes[1,0].bar(x - w/2, ventas_2023, w, label='2023', color='#94a3b8', edgecolor='white')
axes[1,0].bar(x + w/2, ventas_2024, w, label='2024', color='#1e3a8a', edgecolor='white')
axes[1,0].set_title('③ Barras Agrupadas: ¿Cómo cambió 2023 → 2024?', fontweight='bold')
axes[1,0].set_xticks(x)
axes[1,0].set_xticklabels(categorias, rotation=20, fontsize=8)
axes[1,0].legend()

axes[1,1].hlines(categorias, 0, ventas_2024, color='#e2e8f0', linewidth=3)
axes[1,1].scatter(ventas_2024, categorias, s=150, c=colores, zorder=5)
for cat, val in zip(categorias, ventas_2024):
    axes[1,1].text(val + 500, cat, f'${val/1000:.0f}k', va='center', fontsize=9)
axes[1,1].set_title('④ Lollipop: Más limpio que barras con muchas categorías', fontweight='bold')

for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

print("📊 En Power BI: estos 4 gráficos son todos nativos, solo arrastra los campos.")
print("📈 En Tableau: Show Me → selecciona el tipo de gráfico según lo que ves aquí.")
````

## El Waffle Chart: la torta convertida en cuadrícula 🧇

Imagina que en vez de cortar una pizza en porciones (torta), reparte 100 cuadraditos de chocolate entre tus amigos según cuánto le toca a cada uno. Eso es un **waffle chart**: una cuadrícula de 10×10 = 100 celdas, donde cada celda pintada es 1%.

**¿Por qué es mejor que la torta?** Porque el ojo humano compara **cuadrados en fila** mucho mejor que **ángulos inclinados**. Contar "20 cuadraditos azules" es más fácil que estimar "un ángulo de 72 grados".

````python
# ============================================================
# Waffle chart casero con matplotlib
# ============================================================

labels = categorias
valores = ventas_2024
porcentajes = (np.array(valores) / sum(valores) * 100).round().astype(int)

celdas = []
for pct, color in zip(porcentajes, colores):
    celdas += [color] * pct
while len(celdas) < 100:
    celdas.append('#e2e8f0')
celdas = celdas[:100]
grid = np.array(celdas).reshape(10, 10)

fig, ax = plt.subplots(figsize=(5.5, 5.5))
for i in range(10):
    for j in range(10):
        ax.add_patch(plt.Rectangle((j, 9 - i), 0.9, 0.9, color=grid[i, j]))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xticks([])
ax.set_yticks([])
ax.set_title('100 cuadraditos = 100% de las ventas', fontweight='bold')
handles = [plt.Rectangle((0,0),1,1, color=c) for c in colores]
ax.legend(handles, [f'{l} ({p}%)' for l, p in zip(labels, porcentajes)],
          loc='upper center', bbox_to_anchor=(0.5, -0.02), ncol=2, fontsize=8, frameon=False)
plt.tight_layout()
plt.show()
````

## El Treemap: cuando tienes jerarquías 🌳

Un **treemap** es como un mapa de un departamento donde cada habitación tiene un tamaño proporcional a su valor. Úsalo cuando tienes **categorías con subcategorías** (por ejemplo: Electrónica → Laptops → Gaming) y quieres ver qué parte del total representa cada elemento.

```
TOTAL VENTAS ($228,300)
┌────────────────────────────────────────────┐
│   ELECTRÓNICA ($52,300)  │  ALIMENTOS     │
│   ┌──────┬──────┐        │  ($41,800)     │
│   │Laptops│ TV  │        ├────────────────┤
│   └──────┴──────┘        │    ROPA        │
│                          │  ($35,600)     │
└──────────────────────────┴────────────────┘
```

📊 **En Power BI:** Visual nativo llamado "Mapa de árbol".

📈 **En Tableau:** Show Me → seleccionar "Treemap" (requiere una dimensión y una medida).

## Resumen de bolsillo 📱

| Tengo... | Quiero mostrar... | Uso... |
|---|---|---|
| 5 categorías con ventas | Cuál vende más | **Barras verticales ordenadas** |
| 5 categorías, nombres largos | Cuál vende más | **Barras horizontales** |
| 5 categorías, 2 años | Cómo cambió | **Barras agrupadas** |
| 5 categorías, su % del total | Composición | **Barras 100% apiladas** |
| 2-3 categorías, diferencia obvia | Proporción | **Donut o waffle** |
| 15+ categorías | Comparar | **Lollipop chart** |
| Categoría + subcategoría | Jerarquía | **Treemap** |

> 💡 **Tip final:** en Power BI y Tableau, si dudas, elige siempre las **barras**. Nunca te arrepentirás. La torta, siempre pregúntate dos veces si es necesaria.
