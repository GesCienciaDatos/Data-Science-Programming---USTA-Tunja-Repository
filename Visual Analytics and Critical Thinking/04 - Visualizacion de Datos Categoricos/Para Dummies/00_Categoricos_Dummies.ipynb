## ¿Qué son datos categóricos? En serio, ¿qué son? 🏷️

Imagina que estás llenando una encuesta. Te preguntan:
- ¿Cuál es tu color favorito? (rojo/azul/verde) → **categórico**
- ¿Cuántos años tienes? → **numérico** (lo vemos en el Módulo 05)

Los datos categóricos son **etiquetas o nombres**, no números. Ejemplos:
- Región: Norte, Sur, Este, Oeste
- Categoría de producto: Electrónica, Ropa, Hogar
- Estado del pedido: Pendiente, Enviado, Entregado

**La pregunta clave con datos categóricos:**
> *¿Cuánto tiene/vende/representa CADA categoría?*

---
## La Guía Rápida: ¿Qué gráfico uso? 🎯

| Situación | Gráfico recomendado | Por qué |
|---|---|---|
| Tengo 4 categorías, quiero comparar sus totales | **Barras verticales** | Simple, el cerebro compara alturas fácilmente |
| Tengo categorías con nombres largos | **Barras horizontales** | El texto cabe mejor |
| Quiero comparar 2023 vs 2024 por categoría | **Barras agrupadas** | Dos barras juntas para comparar |
| Quiero ver la composición interna | **Barras apiladas** | Cada barra = 100%, dividida en partes |
| Tengo muchas categorías y quiero ser elegante | **Lollipop (paleta)** | Menos ruido visual que barras |
| Quiero mostrar jerarquía (categoría → subcategoría) | **Treemap** | El tamaño del rectángulo = valor |

### ¿Y cuándo SÍ se puede usar torta/donut?
Solo cuando tienes **2 o 3 categorías máximo** y las diferencias son grandes y obvias. Por ejemplo:
- 70% vs 30% (Se ve claramente)
- Hombre vs Mujer (2 opciones)

Si tienes más categorías o diferencias pequeñas → **siempre barras**.

````python
# ============================================================
# Las 4 variantes más útiles — con código simple y comentado
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

# Nuestros datos de ejemplo
categorias = ['Electrónica', 'Ropa', 'Hogar', 'Alimentos', 'Deportes']
ventas_2023 = [45200, 32100, 28900, 38700, 21300]
ventas_2024 = [52300, 35600, 31200, 41800, 25100]
colores = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Las 4 variantes de barras más útiles — Mismos datos, diferentes respuestas',
             fontsize=12, fontweight='bold')

# 1. Barras verticales simples (¿cuánto vende cada una?)
bars1 = axes[0,0].bar(categorias, ventas_2024, color=colores, edgecolor='white')
for bar, val in zip(bars1, ventas_2024):
    axes[0,0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500,
                   f'${val/1000:.0f}k', ha='center', fontsize=9, fontweight='bold')
axes[0,0].set_title('① Barras Verticales: ¿Cuánto vende cada categoría?', fontweight='bold')
axes[0,0].tick_params(axis='x', rotation=20)
axes[0,0].set_ylabel('Ventas ($)')
axes[0,0].spines['top'].set_visible(False)
axes[0,0].spines['right'].set_visible(False)

# 2. Barras horizontales ordenadas (para nombres largos)
idx = np.argsort(ventas_2024)
cats_ord = [categorias[i] for i in idx]
vals_ord = [ventas_2024[i] for i in idx]
cols_ord = [colores[i] for i in idx]
bars2 = axes[0,1].barh(cats_ord, vals_ord, color=cols_ord, edgecolor='white')
for bar, val in zip(bars2, vals_ord):
    axes[0,1].text(bar.get_width() + 300, bar.get_y() + bar.get_height()/2,
                   f'${val/1000:.0f}k', va='center', fontsize=9)
axes[0,1].set_title('② Barras Horizontales: Mejor para nombres largos', fontweight='bold')
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)

# 3. Barras agrupadas (2023 vs 2024)
x = np.arange(len(categorias))
w = 0.35
axes[1,0].bar(x - w/2, ventas_2023, w, label='2023', color='#94a3b8', edgecolor='white')
axes[1,0].bar(x + w/2, ventas_2024, w, label='2024', color='#1e3a8a', edgecolor='white')
axes[1,0].set_title('③ Barras Agrupadas: ¿Cómo cambió 2023 → 2024?', fontweight='bold')
axes[1,0].set_xticks(x)
axes[1,0].set_xticklabels(categorias, rotation=20, fontsize=8)
axes[1,0].legend()
axes[1,0].spines['top'].set_visible(False)
axes[1,0].spines['right'].set_visible(False)

# 4. Lollipop (muchas categorías — más elegante)
axes[1,1].hlines(categorias, 0, ventas_2024, color='#e2e8f0', linewidth=3)
scatter = axes[1,1].scatter(ventas_2024, categorias, s=150, c=colores, zorder=5)
for cat, val, col in zip(categorias, ventas_2024, colores):
    axes[1,1].text(val + 500, cat, f'${val/1000:.0f}k', va='center', fontsize=9)
axes[1,1].set_title('④ Lollipop: Más limpio que barras con muchas categorías', fontweight='bold')
axes[1,1].spines['top'].set_visible(False)
axes[1,1].spines['right'].set_visible(False)

for ax in axes.flat:
    ax.set_axisbelow(True)
    ax.yaxis.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📊 En Power BI: estos 4 gráficos son todos nativos.")
print("   Elige el visual → arrastra los campos → selecciona el formato en el panel lateral.")
print("\n📈 En Tableau: Show Me → selecciona el tipo de gráfico según lo que ves aquí.")
````

## El Treemap: cuando tienes jerarquías 🌳

Un **treemap** es como un mapa de un departamento donde cada habitación tiene un tamaño proporcional a su valor. Úsalo cuando:
- Tienes **categorías con subcategorías** (por ejemplo: Electrónica → Laptops → Gaming)
- Quieres ver **qué parte del total** representa cada elemento
- Tienes **muchas categorías** y las barras se vuelven pequeñas

```
TOTAL VENTAS ($228,300)
┌────────────────────────────────────────────┐
│   ELECTRÓNICA ($52,300)  │  ALIMENTOS     │
│                          │  ($41,800)     │
│   ┌──────┬──────┐        ├────────────────┤
│   │Laptops│ TV  │        │    ROPA        │
│   │      │     │        │  ($35,600)     │
│   └──────┴──────┘        ├────────────────┤
│                          │ HOGAR ($31,200)│
│                          ├────────────────┤
│                          │ DEP.($25,100)  │
└──────────────────────────┴────────────────┘
```

📊 **En Power BI:** Visual nativo llamado "Mapa de árbol"

📈 **En Tableau:** Show Me → seleccionar "Treemap" (requiere una dimensión y una medida)

## Resumen de bolsillo 📱

Guarda esta tabla en tu teléfono para la próxima vez que alguien te pida "hacer un gráfico":

| Tengo... | Quiero mostrar... | Uso... |
|---|---|---|
| 5 categorías con ventas | Cuál vende más | **Barras verticales ordenadas** |
| 5 categorías, nombres largos | Cuál vende más | **Barras horizontales** |
| 5 categorías, 2 años | Cómo cambió | **Barras agrupadas** |
| 5 categorías, su % del total | Composición | **Barras 100% apiladas** |
| 15+ categorías | Comparar | **Lollipop chart** |
| Categoría + subcategoría | Jerarquía | **Treemap** |
| 2-3 categorías, diferencia obvia | Proporción | **Donut chart** |

> 💡 **Tip final:** En Power BI y Tableau, si dudas, elige siempre las **barras**. Nunca te arrepentirás. La torta, siempre pregúntate dos veces si es necesaria.
