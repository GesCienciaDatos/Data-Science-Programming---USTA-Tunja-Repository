```python
# ============================================================
# Setup: importaciones y dataset de trabajo
# Dataset: ventas por categoría de producto y región
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

plt.rcParams.update({
    'font.family': 'sans-serif',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'grid.alpha': 0.3
})

np.random.seed(42)
categorias = ['Electrónica', 'Ropa', 'Hogar', 'Alimentos', 'Deportes', 'Libros']
regiones = ['Norte', 'Sur', 'Este', 'Oeste']

ventas_cat = pd.DataFrame({
    'categoria': categorias,
    'ventas_2023': [45200, 32100, 28900, 38700, 21300, 12800],
    'ventas_2024': [52300, 35600, 31200, 41800, 25100, 14200]
})

datos_region = []
for cat in categorias:
    for reg in regiones:
        valor = np.random.randint(5000, 20000)
        datos_region.append({'categoria': cat, 'region': reg, 'ventas': valor})
df_region = pd.DataFrame(datos_region)

print("📦 Dataset cargado correctamente")
print(ventas_cat.to_string(index=False))
```

## 1. Cómo organizar el catálogo de técnicas 🗂️

No todas las técnicas de visualización categórica sirven para todo. La elección correcta depende de dos factores:

| Factor | Pregunta | Impacto |
|---|---|---|
| **Escala** | ¿La variable es nominal u ordinal? | Determina si el orden del gráfico es libre o fijo |
| **Cardinalidad** | ¿Cuántas categorías tiene? | Determina si caben en barras, si conviene lollipop, o si se necesita jerarquía (treemap) |

En las siguientes secciones recorremos las técnicas de menor a mayor complejidad visual, indicando en cada una para qué escala y cardinalidad es ideal.

---
## 2. Pocas categorías (≤ 8): barras simples y horizontales 📊

Con pocas categorías, sean **nominales** (ciudad, categoría de producto) u **ordinales** (satisfacción, calificación), la barra simple es casi siempre la mejor opción.

| Variante | Cuándo usarla |
|---|---|
| **Barra vertical** | Etiquetas cortas (≤ 10 caracteres) |
| **Barra horizontal** | Etiquetas largas — el texto cabe mejor |

```python
# ============================================================
# Barras simples: vertical vs horizontal
# ============================================================

colores_cat = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6', '#06b6d4']

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# Vertical (ordenada, categoría nominal)
df_sorted = ventas_cat.sort_values('ventas_2024', ascending=False)
bars = axes[0].bar(df_sorted['categoria'], df_sorted['ventas_2024'], color=colores_cat, edgecolor='white')
for bar, val in zip(bars, df_sorted['ventas_2024']):
    axes[0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500,
                 f'${val/1000:.0f}k', ha='center', fontsize=8, fontweight='bold')
axes[0].set_title('① Barras Verticales (nominal, ordenada por frecuencia)', fontweight='bold')
axes[0].tick_params(axis='x', rotation=20)

# Horizontal
df_sorted2 = ventas_cat.sort_values('ventas_2024')
h_bars = axes[1].barh(df_sorted2['categoria'], df_sorted2['ventas_2024'], color=colores_cat[::-1], edgecolor='white')
for bar, val in zip(h_bars, df_sorted2['ventas_2024']):
    axes[1].text(bar.get_width() + 300, bar.get_y() + bar.get_height()/2,
                 f'${val/1000:.0f}k', va='center', fontsize=8)
axes[1].set_title('② Barras Horizontales (etiquetas largas)', fontweight='bold')

plt.tight_layout()
plt.show()

print("📊 En Power BI: Visual 'Gráfico de barras' o 'Gráfico de barras horizontal'.")
print("📈 En Tableau: Show Me → Bar chart (barra a la derecha para horizontal).")
```

## 3. Comparaciones y composición: agrupadas, apiladas y 100% apiladas 📶

Cuando se necesita comparar **dos grupos** (por ejemplo, dos años) o ver la **composición interna** de un total, la barra simple no alcanza.

| Variante | Pregunta que responde |
|---|---|
| **Barra agrupada** | ¿Cómo se comparan dos periodos/grupos por categoría? |
| **Barra apilada** | ¿Cuál es la composición del total, manteniendo la magnitud? |
| **Barra 100% apilada** | ¿Qué proporción representa cada parte, sin importar la magnitud total? (ideal para comparar composición entre categorías ordinales)

```python
# ============================================================
# Barras agrupadas, apiladas y 100% apiladas
# ============================================================

fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))

# Agrupadas
x = np.arange(len(categorias))
w = 0.35
axes[0].bar(x - w/2, ventas_cat['ventas_2023'], w, label='2023', color='#94a3b8', edgecolor='white')
axes[0].bar(x + w/2, ventas_cat['ventas_2024'], w, label='2024', color='#1e3a8a', edgecolor='white')
axes[0].set_title('③ Agrupadas (2023 vs 2024)', fontweight='bold')
axes[0].set_xticks(x)
axes[0].set_xticklabels(categorias, rotation=25, fontsize=8)
axes[0].legend(fontsize=8)

# Apiladas
ventas_region_pivot = df_region.pivot_table(values='ventas', index='categoria', columns='region')
bottom = np.zeros(len(ventas_region_pivot))
colores_reg = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444']
for col, color in zip(ventas_region_pivot.columns, colores_reg):
    axes[1].bar(ventas_region_pivot.index, ventas_region_pivot[col], bottom=bottom,
                label=col, color=color, edgecolor='white', linewidth=0.5)
    bottom += ventas_region_pivot[col].values
axes[1].set_title('④ Apiladas (por región)', fontweight='bold')
axes[1].tick_params(axis='x', rotation=25)
axes[1].legend(fontsize=7)

# 100% apiladas
totales = ventas_region_pivot.sum(axis=1)
ventas_pct = ventas_region_pivot.div(totales, axis=0) * 100
bottom_pct = np.zeros(len(ventas_pct))
for col, color in zip(ventas_pct.columns, colores_reg):
    axes[2].bar(ventas_pct.index, ventas_pct[col], bottom=bottom_pct,
                label=col, color=color, edgecolor='white', linewidth=0.5)
    bottom_pct += ventas_pct[col].values
axes[2].set_title('⑤ 100% Apiladas (composición pura)', fontweight='bold')
axes[2].tick_params(axis='x', rotation=25)
axes[2].legend(fontsize=7)

plt.tight_layout()
plt.show()

print("📊 En Power BI: 'Gráfico de barras apiladas' y activar '100% apiladas' en el mismo visual.")
print("📈 En Tableau: Marks → Bar, arrastrar la dimensión a Color para apilar automáticamente.")
```

## 4. Alternativas modernas a la torta: donut y waffle chart 🍩🧇

El gráfico de torta tiene un uso legítimo muy limitado (máximo 2-3 categorías con diferencias claras). Estas son dos alternativas más legibles:

- **Donut chart**: igual que la torta, pero con espacio central (útil para mostrar un total o KPI) y etiquetas externas que facilitan la lectura.
- **Waffle chart**: una cuadrícula de 10×10 celdas donde cada celda representa 1% del total — el ojo humano compara **áreas rectangulares** mucho mejor que **ángulos**.

```python
# ============================================================
# Donut chart (Plotly) y Waffle chart (Matplotlib, cuadrícula 10x10)
# ============================================================

labels = ventas_cat['categoria'].tolist()
values = ventas_cat['ventas_2024'].tolist()
colors = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6', '#06b6d4']

# --- Donut ---
fig_donut = go.Figure(data=[go.Pie(
    labels=labels, values=values, marker_colors=colors,
    textinfo='label+percent', textposition='outside', hole=0.55
)])
fig_donut.update_layout(
    title_text='Donut Chart — Ventas 2024 por Categoría',
    template='plotly_white', showlegend=False, height=420
)
fig_donut.show()

# --- Waffle chart casero: cuadrícula 10x10 = 100 celdas ---
porcentajes = (ventas_cat['ventas_2024'] / ventas_cat['ventas_2024'].sum() * 100).round().astype(int)
celdas = []
for cat, pct, color in zip(labels, porcentajes, colors):
    celdas += [color] * pct
while len(celdas) < 100:
    celdas.append('#e2e8f0')
celdas = celdas[:100]

grid = np.array(celdas).reshape(10, 10)
fig, ax = plt.subplots(figsize=(6, 6))
for i in range(10):
    for j in range(10):
        ax.add_patch(plt.Rectangle((j, 9 - i), 0.9, 0.9, color=grid[i, j]))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xticks([])
ax.set_yticks([])
ax.set_title('Waffle Chart — 1 celda = 1% de las ventas 2024', fontweight='bold')
handles = [plt.Rectangle((0,0),1,1, color=c) for c in colors]
ax.legend(handles, [f'{l} ({p}%)' for l, p in zip(labels, porcentajes)],
          loc='upper center', bbox_to_anchor=(0.5, -0.02), ncol=2, fontsize=8, frameon=False)
plt.tight_layout()
plt.show()

print("📊 En Power BI: Donut = visual nativo 'Gráfico de anillos'. Waffle: AppSource 'Waffle Chart by MAQ Software'.")
print("📈 En Tableau: Donut = Pie + capa blanca central. Waffle: técnica de grid con Index() calculado.")
```

## 5. Muchas categorías (> 8): lollipop chart 🍭

Cuando hay muchas categorías, las barras anchas generan ruido visual. El **lollipop chart** (línea delgada + punto) transmite la misma información con menos "tinta" en pantalla.

```python
# ============================================================
# Lollipop chart
# ============================================================

df_lollipop = ventas_cat.sort_values('ventas_2024')

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hlines(df_lollipop['categoria'], 0, df_lollipop['ventas_2024'], color='#cbd5e1', linewidth=2)
ax.plot(df_lollipop['ventas_2024'], df_lollipop['categoria'], 'o', markersize=12, color='#1e3a8a', zorder=5)
for _, row in df_lollipop.iterrows():
    ax.text(row['ventas_2024'] + 500, row['categoria'], f"${row['ventas_2024']/1000:.0f}k", va='center', fontsize=9)
ax.set_title('Lollipop Chart — Ventas 2024 (elegante con muchas categorías)', fontweight='bold')
ax.set_xlabel('Ventas 2024 ($)')
plt.tight_layout()
plt.show()

print("📊 En Power BI: AppSource → buscar 'Dot Plot by MAQ Software'.")
print("📈 En Tableau: combinar Bar (línea delgada) + Circle en un eje dual.")
```

## 6. Jerarquías categóricas: treemap y sunburst 🌳

Cuando una variable categórica tiene **subcategorías** (categoría → subcategoría), un gráfico de barras plano pierde esa estructura. Treemap y sunburst la preservan visualmente.

```python
# ============================================================
# Treemap y Sunburst con Plotly Express
# ============================================================

datos_jerarquia = {
    'id': ['Electrónica','Laptops','Smartphones','Tablets',
           'Ropa','Hombre','Mujer','Niños',
           'Hogar','Cocina','Sala','Jardín'],
    'parent': ['','Electrónica','Electrónica','Electrónica',
                '','Ropa','Ropa','Ropa',
                '','Hogar','Hogar','Hogar'],
    'value': [0, 28000, 18000, 6300,
               0, 14000, 15000, 6600,
               0, 12000, 10000, 9200]
}
df_hier = pd.DataFrame(datos_jerarquia)

fig_treemap = px.treemap(
    df_hier, ids='id', names='id', parents='parent', values='value',
    color='value', color_continuous_scale='Blues',
    title='Treemap: Ventas por Categoría y Subcategoría', template='plotly_white'
)
fig_treemap.update_layout(height=420)
fig_treemap.show()

fig_sunburst = px.sunburst(
    df_hier, ids='id', names='id', parents='parent', values='value',
    color='value', color_continuous_scale='Teal',
    title='Sunburst: Misma jerarquía en forma radial', template='plotly_white'
)
fig_sunburst.update_layout(height=420)
fig_sunburst.show()

print("📊 En Power BI: Visual nativo 'Mapa de árbol'. Sunburst: AppSource 'Sunburst Chart'.")
print("📈 En Tableau: Show Me → Treemap (dimensión + subdimensión a Detail/Color).")
```

---
## 7. Guía de selección resumida 🧭

| Escala | Cardinalidad | Mejor gráfico | Python | Power BI | Tableau |
|---|---|---|---|---|---|
| Nominal | ≤ 8 categorías | Barras verticales/horizontales | `plt.bar()` / `barh()` | Gráfico de barras | Bar Chart |
| Nominal u Ordinal | 2 grupos a comparar | Barras agrupadas | `px.bar(barmode='group')` | Barras agrupadas | Side-by-side bars |
| Ordinal | Composición + total | Barras apiladas | `px.bar(barmode='stack')` | Barras apiladas | Stacked bars |
| Ordinal | Solo composición | Barras 100% apiladas | `barmode='stack', barnorm='percent'` | 100% barras apiladas | 100% stacked |
| Nominal | 2-3 categorías | Donut | `hole=0.5` en `go.Pie` | Gráfico de anillos | Pie/Donut |
| Nominal | 2-3 categorías, alternativa | Waffle | grid 10×10 con matplotlib | AppSource: Waffle Chart | Grid + Index() |
| Nominal u Ordinal | > 8 categorías | Lollipop | `hlines` + `scatter` | AppSource: Dot Plot | Dual axis |
| Nominal jerárquica | Con subcategorías | Treemap / Sunburst | `px.treemap()` / `px.sunburst()` | Mapa de árbol | Treemap |

---
## Resumen y Puntos Clave 🎯

- Las **barras** (en sus variantes simple, agrupada, apilada y 100% apilada) cubren la gran mayoría de los casos de visualización categórica.
- **Donut** y **waffle** son alternativas legítimas a la torta, pero solo con muy pocas categorías.
- El **lollipop chart** reduce el ruido visual cuando hay muchas categorías.
- **Treemap** y **sunburst** son las herramientas correctas cuando existe una jerarquía categoría → subcategoría.
- La elección final siempre combina dos factores: la **escala** de la variable (nominal/ordinal) y su **cardinalidad** (número de categorías).
