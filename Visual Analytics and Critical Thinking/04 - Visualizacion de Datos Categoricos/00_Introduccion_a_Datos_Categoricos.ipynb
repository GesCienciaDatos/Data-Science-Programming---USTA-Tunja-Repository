```python
# ============================================================
# Setup: importaciones y datasets de ejemplo
# Usaremos un pequeño dataset de clientes para ilustrar las 3 escalas
# ============================================================

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

plt.rcParams.update({
    'font.family': 'sans-serif',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'grid.alpha': 0.3
})

# Dataset de ejemplo: encuesta de satisfacción de clientes
np.random.seed(42)
clientes = pd.DataFrame({
    'cliente_id': range(1, 13),
    'ciudad': ['Tunja', 'Bogotá', 'Tunja', 'Duitama', 'Bogotá', 'Sogamoso',
               'Tunja', 'Bogotá', 'Duitama', 'Tunja', 'Sogamoso', 'Bogotá'],
    'nivel_satisfaccion': ['Alto', 'Medio', 'Bajo', 'Alto', 'Alto', 'Medio',
                            'Bajo', 'Alto', 'Medio', 'Alto', 'Bajo', 'Medio'],
    'edad': [23, 45, 31, 52, 29, 38, 61, 27, 44, 19, 55, 33]
})

print("📋 Dataset de ejemplo — Encuesta de satisfacción")
print(clientes.to_string(index=False))
```

## 1. ¿Qué es una variable categórica? 🏷️

Una **variable categórica** (o cualitativa) representa **etiquetas, nombres o grupos**, no cantidades. Responde a la pregunta *"¿a qué grupo pertenece?"*, no a *"¿cuánto?"*.

| | Variable categórica | Variable numérica |
|---|---|---|
| Ejemplo | Ciudad, género, nivel de satisfacción | Edad, ingresos, ventas |
| Operación válida | Contar, agrupar, comparar frecuencias | Sumar, promediar, calcular varianza |
| Gráfico típico | Barras, donut, treemap | Histograma, boxplot, scatter |

Antes de elegir un gráfico, la primera pregunta que debe hacerse un analista es: **¿esta variable es categórica, y si lo es, de qué tipo?** La respuesta determina qué visualizaciones son válidas y cuáles inducen a error.

---
## 2. Escala Nominal: categorías sin orden natural 🔤

En la escala **nominal**, las categorías son solo **etiquetas distintas**. No existe un orden lógico entre ellas: "Bogotá" no es "mayor" ni "menor" que "Tunja".

**Ejemplos típicos:**
- Ciudad de residencia
- Género
- Categoría de producto (Electrónica, Ropa, Hogar)
- Método de pago (Efectivo, Tarjeta, Transferencia)

**Regla de oro:** con datos nominales, el **orden de las barras en el gráfico es libre** — normalmente se ordenan de mayor a menor frecuencia para facilitar la lectura (nunca alfabéticamente, salvo que el alfabeto sea relevante).

```python
# ============================================================
# Ejemplo: variable nominal — Ciudad (sin orden natural)
# Ordenamos por frecuencia, NO alfabéticamente
# ============================================================

conteo_ciudad = clientes['ciudad'].value_counts().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 4.5))
colores_nom = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444']
bars = ax.bar(conteo_ciudad.index, conteo_ciudad.values, color=colores_nom, edgecolor='white')
for bar, val in zip(bars, conteo_ciudad.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05, str(val),
            ha='center', fontweight='bold')
ax.set_title('Variable Nominal: Ciudad — ordenada por frecuencia, no alfabéticamente', fontweight='bold')
ax.set_ylabel('Número de clientes')
plt.tight_layout()
plt.show()

print("✅ Nominal: reordenar las categorías NO cambia el significado de los datos.")
print("   Bogotá, Tunja, Duitama y Sogamoso podrían aparecer en cualquier orden sin perder información.")
```

---
## 3. Escala Ordinal: categorías con orden natural 📶

En la escala **ordinal**, las categorías tienen un **orden lógico o jerárquico**, pero las distancias entre ellas no están definidas numéricamente (no sabemos "cuánto más" es "Alto" que "Medio").

**Ejemplos típicos:**
- Nivel de satisfacción (Bajo, Medio, Alto)
- Calificación de un producto (1 a 5 estrellas)
- Nivel educativo (Primaria, Secundaria, Pregrado, Posgrado)
- Estrato socioeconómico (1 a 6)

**Regla de oro:** con datos ordinales, el **orden de las categorías en el gráfico NUNCA es libre** — debe respetar siempre su jerarquía natural. Reordenar por frecuencia (como hacemos con nominales) sería un error grave porque destruye la información de orden.

```python
# ============================================================
# Ejemplo: variable ordinal — Nivel de satisfacción
# Usamos pd.Categorical para FORZAR el orden correcto
# ============================================================

orden_satisfaccion = ['Bajo', 'Medio', 'Alto']
clientes['nivel_satisfaccion'] = pd.Categorical(
    clientes['nivel_satisfaccion'], categories=orden_satisfaccion, ordered=True
)
conteo_satisfaccion = clientes['nivel_satisfaccion'].value_counts().reindex(orden_satisfaccion)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# Correcto: respeta el orden natural
colores_ord = ['#ef4444', '#f59e0b', '#10b981']
axes[0].bar(conteo_satisfaccion.index, conteo_satisfaccion.values, color=colores_ord, edgecolor='white')
axes[0].set_title('✅ Correcto: orden natural Bajo → Medio → Alto', fontweight='bold')
axes[0].set_ylabel('Número de clientes')

# Incorrecto: ordenado por frecuencia, como si fuera nominal
conteo_mal_ordenado = clientes['nivel_satisfaccion'].value_counts()
axes[1].bar(conteo_mal_ordenado.index.astype(str), conteo_mal_ordenado.values,
            color='#94a3b8', edgecolor='white')
axes[1].set_title('❌ Incorrecto: ordenado por frecuencia (pierde jerarquía)', fontweight='bold')
axes[1].set_ylabel('Número de clientes')

plt.tight_layout()
plt.show()

print("⚠️ Al ordenar por frecuencia, el lector ya no puede saber de un vistazo")
print("   si la satisfacción sube o baja de izquierda a derecha.")
```

---
## 4. Escala de Intervalo (aplicada a categorías): agrupar lo continuo 📏

Estrictamente, la escala de **intervalo** es propia de variables numéricas (donde sí existen distancias iguales entre valores, como la temperatura en °C). En visualización categórica, la forma más común de encontrarla es cuando **convertimos una variable numérica continua en categorías por rangos** (*binning*), por ejemplo edades agrupadas en rangos etarios.

**Ejemplos típicos:**
- Rango de edad (18-25, 26-35, 36-45, 46+)
- Rango de ingresos (Bajo, Medio-bajo, Medio-alto, Alto)
- Rango de precio de producto ($0-50k, $50k-100k, etc.)

**Diferencia clave con ordinal puro:** aquí las distancias entre categorías **sí tienen un significado cuantitativo** (el rango "26-35" está a una distancia numéricamente medible del rango "36-45"), aunque al graficarlas se traten visualmente como categorías ordenadas.

```python
# ============================================================
# Ejemplo: convertir "edad" (numérica) en rangos etarios (categórica de intervalo)
# ============================================================

bins = [0, 25, 35, 50, 100]
etiquetas = ['18-25', '26-35', '36-50', '51+']
clientes['rango_edad'] = pd.cut(clientes['edad'], bins=bins, labels=etiquetas, right=True)

conteo_edad = clientes['rango_edad'].value_counts().reindex(etiquetas)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(conteo_edad.index.astype(str), conteo_edad.values,
       color='#1e3a8a', edgecolor='white')
ax.set_title('Escala de Intervalo: rangos de edad (orden Y distancia son significativos)', fontweight='bold')
ax.set_ylabel('Número de clientes')
ax.set_xlabel('Rango de edad (años)')
plt.tight_layout()
plt.show()

print(clientes[['cliente_id', 'edad', 'rango_edad']].to_string(index=False))
print("\n💡 A diferencia de la satisfacción (ordinal pura), aquí sabemos EXACTAMENTE")
print("   cuánto separa a '26-35' de '36-50': 10 años de distancia real.")
```

## 5. ¿Por qué importa la escala para elegir el gráfico? 🎯

La escala de medición **no es un detalle académico** — es la decisión que evita gráficos engañosos o simplemente incorrectos.

| Escala | Orden válido en el gráfico | Gráficos recomendados | Error común |
|---|---|---|---|
| **Nominal** | Libre (usualmente por frecuencia) | Barras, donut, treemap | Ordenar alfabéticamente sin razón |
| **Ordinal** | Fijo, respeta la jerarquía | Barras ordenadas, barras apiladas 100% | Ordenar por frecuencia, perdiendo la jerarquía |
| **Intervalo (bins)** | Fijo, respeta la magnitud | Barras ordenadas, histogramas de categorías | Tratar los rangos como si fueran nominales |

## Resumen y Puntos Clave 🎯

- Una variable es **categórica** cuando representa etiquetas o grupos, no cantidades.
- **Nominal**: categorías sin orden (ciudad, género). El orden en el gráfico es libre.
- **Ordinal**: categorías con orden jerárquico pero sin distancias numéricas definidas (satisfacción, calificación). El orden en el gráfico es obligatorio.
- **Intervalo (aplicada a bins)**: categorías creadas al agrupar una variable numérica continua (rangos de edad, de ingreso). El orden y la magnitud de la distancia importan.
- Identificar correctamente la escala **antes** de graficar evita errores de interpretación y visualizaciones engañosas.
