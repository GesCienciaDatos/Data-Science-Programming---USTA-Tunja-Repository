```python
# ============================================================
# Setup: importaciones y dataset de trabajo
# Dataset: ventas por categoría de producto y región
# ============================================================

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

# Configuración global de estilo
plt.rcParams.update({
    'font.family': 'sans-serif',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'grid.alpha': 0.3
})

# Dataset principal
np.random.seed(42)
categorias = ['Electrónica', 'Ropa', 'Hogar', 'Alimentos', 'Deportes', 'Libros']
regiones = ['Norte', 'Sur', 'Este', 'Oeste']

# Ventas por categoría (totales)
ventas_cat = pd.DataFrame({
    'categoria': categorias,
    'ventas_2023': [45200, 32100, 28900, 38700, 21300, 12800],
    'ventas_2024': [52300, 35600, 31200, 41800, 25100, 14200]
})

# Ventas por categoría y región
datos_region = []
for cat in categorias:
    for reg in regiones:
        valor = np.random.randint(5000, 20000)
        datos_region.append({'categoria': cat, 'region': reg, 'ventas': valor})
df_region = pd.DataFrame(datos_region)

print("📦 Dataset cargado correctamente")
print(f"   Ventas por categoría: {ventas_cat.shape}")
print(f"   Ventas por región: {df_region.shape}")
print("\nVista previa:")
print(ventas_cat.to_string(index=False))
```

## 1. Gráfico de Barras: La Base de Todo 📊

El gráfico de barras es el **caballo de batalla** de la visualización categórica. Existe en múltiples variantes, cada una para un propósito específico.

| Variante | Pregunta que responde | Cuándo usarlo |
|---|---|---|
| **Barra simple** | ¿Cuánto tiene cada categoría? | Comparar totales absolutos |
| **Barra horizontal** | Igual, pero con etiquetas largas | Cuando los nombres son largos |
| **Barra agrupada** | ¿Cómo se comparan dos grupos? | Comparar dos métricas paralelas |
| **Barra apilada** | ¿Cuál es la composición del total? | Ver partes y total simultáneamente |
| **Barra 100% apilada** | ¿Qué proporción representa cada parte? | Comparar composición entre grupos |
| **Lollipop** | Igual que barra simple, más elegante | Cuando hay muchas categorías |

```python
# ============================================================
# Galería de Variantes de Gráficos de Barras
# ============================================================

colores_cat = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6', '#06b6d4']

fig, axes = plt.subplots(2, 3, figsize=(18, 11))
fig.suptitle('Galería de Variantes de Gráficos de Barras — Dataset de Ventas',
             fontsize=14, fontweight='bold')

# 1. Barra simple (ordenada descendente)
df_sorted = ventas_cat.sort_values('ventas_2024', ascending=False)
bars = axes[0,0].bar(df_sorted['categoria'], df_sorted['ventas_2024'],
                     color=colores_cat, edgecolor='white')
for bar, val in zip(bars, df_sorted['ventas_2024']):
    axes[0,0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500,
                   f'${val/1000:.0f}k', ha='center', fontsize=8, fontweight='bold')
axes[0,0].set_title('① Barras Simples (ordenadas)', fontweight='bold')
axes[0,0].set_ylabel('Ventas 2024 ($)')
axes[0,0].tick_params(axis='x', rotation=20)

# 2. Barra horizontal
df_sorted2 = ventas_cat.sort_values('ventas_2024')
h_bars = axes[0,1].barh(df_sorted2['categoria'], df_sorted2['ventas_2024'],
                        color=colores_cat[::-1], edgecolor='white')
for bar, val in zip(h_bars, df_sorted2['ventas_2024']):
    axes[0,1].text(bar.get_width() + 300, bar.get_y() + bar.get_height()/2,
                   f'${val/1000:.0f}k', va='center', fontsize=8)
axes[0,1].set_title('② Barras Horizontales', fontweight='bold')
axes[0,1].set_xlabel('Ventas 2024 ($)')

# 3. Barras agrupadas
x = np.arange(len(categorias))
w = 0.35
axes[0,2].bar(x - w/2, ventas_cat['ventas_2023'], w, label='2023',
              color='#94a3b8', edgecolor='white')
axes[0,2].bar(x + w/2, ventas_cat['ventas_2024'], w, label='2024',
              color='#1e3a8a', edgecolor='white')
axes[0,2].set_title('③ Barras Agrupadas (2023 vs 2024)', fontweight='bold')
axes[0,2].set_xticks(x)
axes[0,2].set_xticklabels(categorias, rotation=20, fontsize=8)
axes[0,2].legend(fontsize=9)

# 4. Barras apiladas
ventas_region_pivot = df_region.pivot_table(values='ventas', index='categoria', columns='region')
bottom = np.zeros(len(ventas_region_pivot))
colores_reg = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444']
for i, (col, color) in enumerate(zip(ventas_region_pivot.columns, colores_reg)):
    axes[1,0].bar(ventas_region_pivot.index, ventas_region_pivot[col],
                  bottom=bottom, label=col, color=color, edgecolor='white', linewidth=0.5)
    bottom += ventas_region_pivot[col].values
axes[1,0].set_title('④ Barras Apiladas (por región)', fontweight='bold')
axes[1,0].tick_params(axis='x', rotation=25)
axes[1,0].legend(fontsize=8, loc='upper right')
axes[1,0].set_ylabel('Ventas Totales ($)')

# 5. Barras 100% apiladas
totales = ventas_region_pivot.sum(axis=1)
ventas_pct = ventas_region_pivot.div(totales, axis=0) * 100
bottom_pct = np.zeros(len(ventas_pct))
for i, (col, color) in enumerate(zip(ventas_pct.columns, colores_reg)):
    axes[1,1].bar(ventas_pct.index, ventas_pct[col], bottom=bottom_pct,
                  label=col, color=color, edgecolor='white', linewidth=0.5)
    bottom_pct += ventas_pct[col].values
axes[1,1].set_title('⑤ Barras 100% Apiladas (composición)', fontweight='bold')
axes[1,1].tick_params(axis='x', rotation=25)
axes[1,1].set_ylabel('Proporción (%)')
axes[1,1].legend(fontsize=8)

# 6. Lollipop chart
df_lollipop = ventas_cat.sort_values('ventas_2024')
axes[1,2].hlines(df_lollipop['categoria'], 0, df_lollipop['ventas_2024'],
                 color='#cbd5e1', linewidth=2)
axes[1,2].plot(df_lollipop['ventas_2024'], df_lollipop['categoria'],
               'o', markersize=12, color='#1e3a8a', zorder=5)
for _, row in df_lollipop.iterrows():
    axes[1,2].text(row['ventas_2024'] + 500, row['categoria'],
                   f"${row['ventas_2024']/1000:.0f}k", va='center', fontsize=8)
axes[1,2].set_title('⑥ Lollipop Chart (elegante y limpio)', fontweight='bold')
axes[1,2].set_xlabel('Ventas 2024 ($)')

for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.yaxis.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📊 En Power BI: Todos estos gráficos están disponibles como visuales nativos.")
print("   Lollipop: buscar en AppSource 'Dot Plot by MAQ Software'")
print("\n📈 En Tableau: Bar chart + seleccionar 'Stacked' o 'Side by side' en Marks.")
```

## 2. Más allá de la Torta: Alternativas Modernas 🍩

El gráfico de torta tiene un uso legítimo muy limitado (máximo 2-3 categorías con diferencias claras). Estas son las alternativas modernas:

```python
# ============================================================
# Donut Chart, Waffle Chart y variantes modernas con Plotly
# ============================================================

import plotly.graph_objects as go
from plotly.subplots import make_subplots

labels = ventas_cat['categoria'].tolist()
values = ventas_cat['ventas_2024'].tolist()
colors = ['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6', '#06b6d4']

fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'pie'}, {'type': 'pie'}]],
    subplot_titles=['Torta (❌ difícil comparar)', 'Donut con etiquetas directas (✅)']
)

# Torta clásica
fig.add_trace(go.Pie(
    labels=labels, values=values,
    marker_colors=colors,
    textinfo='percent', hole=0
), row=1, col=1)

# Donut con etiquetas externas
fig.add_trace(go.Pie(
    labels=labels, values=values,
    marker_colors=colors,
    textinfo='label+percent',
    textposition='outside',
    hole=0.5
), row=1, col=2)

fig.update_layout(
    title_text='Torta vs Donut — Mismos Datos, Diferente Legibilidad',
    title_font_size=14,
    template='plotly_white',
    showlegend=False,
    height=450
)
fig.show()

print("📊 En Power BI: Visual 'Gráfico de anillos' (Donut Chart)")
print("   Activar etiquetas: Formato → Etiquetas de datos → Posición: Fuera")
print("\n📈 En Tableau: Marks → Pie → ajustar Angle y Color. Para donut: capa de círculo blanco.")
```

## 3. Treemaps y Sunburst: Jerarquías Visuales 🌳

```python
# ============================================================
# Treemap y Sunburst con Plotly Express
# Excelentes para datos jerárquicos (categoría > subcategoría)
# ============================================================

# Dataset con subcategorías
datos_jerarquia = {
    'id': ['Electrónica','Laptops','Smartphones','Tablets',
           'Ropa','Hombre','Mujer','Niños',
           'Hogar','Cocina','Sala','Jardín'],
    'parent': ['','Electrónica','Electrónica','Electrónica',
                '','Ropa','Ropa','Ropa',
                '','Hogar','Hogar','Hogar'],
    'value': [0, 28000, 18000, 6300,
               0, 14000, 15000, 6600,
               0, 12000, 10000, 9200]
}
df_hier = pd.DataFrame(datos_jerarquia)

# Treemap
fig_treemap = px.treemap(
    df_hier,
    ids='id',
    names='id',
    parents='parent',
    values='value',
    color='value',
    color_continuous_scale='Blues',
    title='Treemap: Ventas por Categoría y Subcategoría — El tamaño = valor',
    template='plotly_white'
)
fig_treemap.update_layout(height=450)
fig_treemap.show()

# Sunburst
fig_sunburst = px.sunburst(
    df_hier,
    ids='id',
    names='id',
    parents='parent',
    values='value',
    color='value',
    color_continuous_scale='Teal',
    title='Sunburst: Misma jerarquía en forma radial',
    template='plotly_white'
)
fig_sunburst.update_layout(height=450)
fig_sunburst.show()

print("📊 En Power BI: Visual nativo 'Mapa de árbol' (Treemap)")
print("   Campos: Grupo = Categoría, Subgrupo = Subcategoría, Valores = Ventas")
print("   Sunburst: buscar en AppSource 'Sunburst Chart'")
print("\n📈 En Tableau: Show Me → Treemap. Arrastrar categoría y subcategoría a Color/Label.")
```

## 4. Resumen: Cuándo Usar Cada Gráfico Categórico 🎯

| Situación | Mejor gráfico | Python | Power BI | Tableau |
|---|---|---|---|---|
| Comparar 6 o menos categorías | Barra vertical | `plt.bar()` | Gráfico barras | Bar Chart |
| Etiquetas largas (>10 chars) | Barra horizontal | `plt.barh()` | Barras horizontales | Bar Chart (horizontal) |
| Comparar 2 años/grupos | Barras agrupadas | `px.bar(barmode='group')` | Barras agrupadas | Side-by-side bars |
| Ver composición + total | Barras apiladas | `px.bar(barmode='stack')` | Barras apiladas | Stacked bars |
| Solo comparar composición | Barras 100% | `px.bar(barmode='stack', barnorm='percent')` | 100% barras apiladas | 100% stacked |
| Muchas categorías, elegante | Lollipop | `hlines + scatter` | AppSource: Dot Plot | Custom via dual axis |
| Proporciones (2-3 cat) | Donut | `px.pie(hole=0.5)` | Gráfico de anillos | Pie/Donut |
| Jerarquía de categorías | Treemap | `px.treemap()` | Mapa de árbol | Treemap |
| Jerarquía radial | Sunburst | `px.sunburst()` | AppSource: Sunburst | AppSource |
