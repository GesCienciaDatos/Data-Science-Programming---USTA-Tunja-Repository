## ¿Qué vamos a aprender aquí? 🕵️

Este cuaderno es la versión **"para no ingenieros"** del primer cuaderno del Módulo 03. Si el cuaderno estándar te resultó denso, aquí vamos paso a paso con analogías de la vida diaria.

Imagina que eres **detective**. Un gráfico o un titular es como un sospechoso: parece inocente, pero puede estar mintiendo. Tu trabajo como detective de datos tiene tres etapas:

1. **Desconfiar** un poco al principio (escepticismo) 🤔
2. **Separar** la historia en pedacitos comprobables (análisis) 🔬
3. **Calificar** qué tan buena es la evidencia (evaluación) 📊

Al terminar podrás explicar con tus propias palabras por qué "no creerle todo a los datos" es en realidad una habilidad técnica, no solo desconfianza.

## 1. Escepticismo: no le creas al primer titular 🤔

Imagina que ves este titular: *"¡9 de cada 10 personas prefieren nuestro producto!"*

Suena convincente... pero antes de creerlo, un buen detective se pregunta:
- ¿A **cuántas** personas le preguntaron? ¿A 10? ¿A 10,000?
- ¿**Quién** hizo la pregunta? ¿La empresa que vende el producto, o alguien independiente?
- ¿Le preguntaron a **cualquiera**, o solo a quienes ya eran fans de la marca?

**El truco:** si solo le preguntaron a 10 personas y 9 dijeron que sí, eso puede pasar por pura casualidad — ¡no significa nada estadísticamente fuerte!

```python
# Vamos a comprobarlo: lanzamos una moneda "justa" (50/50) 12 veces,
# muchas muchas veces, y vemos qué tan seguido sale un resultado "impresionante" por PURA SUERTE

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# Simulamos 10,000 "encuestas" de 12 personas cada una, donde en realidad
# no hay ninguna preferencia real (50% probabilidad para cada opción)
resultados = rng.binomial(n=12, p=0.5, size=10_000) / 12 * 100

plt.figure(figsize=(8, 5))
plt.hist(resultados, bins=13, color="#f59e0b", alpha=0.8, edgecolor="white")
plt.axvline(87.5, color="#dc2626", linewidth=2.5, linestyle="--", label='"87% prefiere el producto"')
plt.title("¿Qué tan seguido sale un 87%+ SIN que exista ninguna preferencia real?")
plt.xlabel("% de \"éxito\" en la encuesta simulada")
plt.ylabel("Cantidad de encuestas simuladas")
plt.legend()
plt.tight_layout()
plt.show()

pct = (resultados >= (11/12*100)).mean() * 100
print(f"🎲 Incluso SIN ninguna preferencia real, un {pct:.1f}% de las encuestas de 12 personas")
print("   dan un resultado de 87% o más, solo por azar.")
print("\n💡 Moraleja: desconfía de titulares basados en pocas personas.")
```

## 2. Análisis: separa la historia en pedacitos 🔬

"Analizar" no significa hacer cálculos complicadísimos. Significa **desarmar** una afirmación como si fuera un Lego, y revisar pieza por pieza:

| Pregunta de detective | Ejemplo |
|---|---|
| ¿Qué significa exactamente esa palabra? | "Cliente satisfecho" — ¿satisfecho cómo? ¿quién lo decidió? |
| ¿Cómo lo midieron? | ¿Fue una encuesta? ¿Un sensor? ¿Alguien "a ojo"? |
| ¿Es una tendencia real o solo ruido? | Un día de ventas altas no es lo mismo que una tendencia de meses |

**Analogía:** si un amigo te dice "¡el equipo de fútbol está en su mejor momento, ganó el último partido!", un buen detective de datos preguntaría: ¿ganó también los 5 partidos anteriores, o fue solo un buen día?

```python
# Separemos "un día bueno" de "una tendencia real" con un ejemplo de ventas

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
dias = np.arange(1, 31)
ventas = 100 + 0.5 * dias + rng.normal(0, 8, size=len(dias))  # crecimiento lento + ruido diario

serie = pd.Series(ventas, index=dias)
promedio_movil = serie.rolling(window=5, center=True).mean()

plt.figure(figsize=(9, 5))
plt.plot(dias, ventas, color="#fbbf24", marker="o", markersize=4, linewidth=1, label="Ventas de cada día (con ruido)")
plt.plot(dias, promedio_movil, color="#b45309", linewidth=3, label="Tendencia real (promedio de 5 días)")
plt.title("¿El 'pico' de un solo día es una señal real... o solo ruido?")
plt.xlabel("Día"); plt.ylabel("Ventas")
plt.legend()
plt.tight_layout()
plt.show()

print("💡 Un solo día espectacular no prueba nada por sí solo.")
print("   La línea café oscuro (promedio de varios días) muestra la tendencia real:")
print("   un crecimiento lento y sostenido — no un milagro de un solo día.")
```

## 3. Evaluación: ponle una calificación a la evidencia 📊

Ya desconfiaste (escepticismo) y ya desarmaste la afirmación en piezas (análisis). El último paso es **calificar qué tan buena es la evidencia**, como si fuera un examen con nota de 0 a 5 en cada criterio:

| Criterio | Pregunta simple |
|---|---|
| Tamaño de muestra | ¿Le preguntaron a poca o mucha gente? |
| Independencia | ¿Quién lo midió tiene interés en el resultado? |
| Transparencia | ¿Puedo ver cómo lo midieron? |
| Repetibilidad | ¿Otros llegaron a la misma conclusión? |
| Lógica | ¿Tiene sentido que A cause B? |

Si una afirmación saca **nota baja** en la mayoría de estos criterios, no la deseches por completo — pero tampoco tomes decisiones importantes basándote solo en ella.

```python
# Comparemos dos afirmaciones con una "tarjeta de calificación" simple

import pandas as pd

afirmaciones = pd.DataFrame([
    {
        "Afirmación": "'Suplemento mágico' (12 personas, estudio de la propia empresa)",
        "Muestra": 1, "Independencia": 0, "Transparencia": 1, "Repetibilidad": 0, "Lógica": 1,
    },
    {
        "Afirmación": "Estudio médico serio (15,000 personas, 40 estudios independientes)",
        "Muestra": 5, "Independencia": 5, "Transparencia": 5, "Repetibilidad": 5, "Lógica": 4,
    },
]).set_index("Afirmación")

afirmaciones["Nota Total (/25)"] = afirmaciones.sum(axis=1)
display(afirmaciones)

print("💡 La segunda afirmación tiene una nota mucho más alta — no porque 'suene mejor',")
print("   sino porque cumple objetivamente más criterios de calidad de evidencia.")
```

## Resumen Express 🎯

| Concepto | Una sola frase |
|---|---|
| **Escepticismo** | No le creas a un titular hasta saber cuántos datos lo respaldan |
| **Análisis** | Desarma la afirmación en piezas: definición, medición, tendencia real vs. ruido |
| **Evaluación** | Ponle nota a la evidencia con criterios claros, no con intuición |
| **Muestra pequeña** | Resultados "impresionantes" con poca gente pueden ser pura casualidad |
