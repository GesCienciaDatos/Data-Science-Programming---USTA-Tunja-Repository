## 1. El Truco más viejo: Cortar el eje Y ✂️

Este es el truco de magia favorito de los políticos, las empresas y los medios de comunicación.

**El truco:** En vez de que el eje Y empiece en 0, lo empiezan en 97 (por ejemplo). Así una diferencia de 0.5% parece una caída apocalíptica.

**La defensa:** Siempre mira dónde empieza el eje Y.

```python
# El truco del eje Y: mismo dato, dos historias completamente diferentes

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
aprobacion = [63.2, 63.8, 64.1, 63.5, 64.3, 63.9]  # Solo varía en 1.1 puntos

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
fig.suptitle('🔍 El mismo dato — ¡dos gráficos totalmente diferentes!',
             fontsize=13, fontweight='bold')

# Versión MANIPULADA (eje Y cortado)
ax1.plot(meses, aprobacion, 'o-', color='#ef4444', linewidth=3, markersize=8)
ax1.fill_between(range(len(meses)), aprobacion, alpha=0.2, color='#ef4444')
ax1.set_ylim(62.5, 65)  # ← EL TRUCO: eje Y empieza en 62.5, no en 0
ax1.set_title('❌ Versión PERIODISTA: "El índice se desplomó"', fontweight='bold', color='#dc2626')
ax1.set_xticks(range(len(meses)))
ax1.set_xticklabels(meses)
ax1.set_ylabel('Índice de Aprobación (%)')
ax1.annotate('Eje Y empieza\nen 62.5 ← TRAMPA', xy=(0, 63.2), xytext=(1, 63.0),
             fontsize=8, color='red', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='red'))
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

# Versión HONESTA (eje Y desde 0)
ax2.plot(meses, aprobacion, 'o-', color='#3b82f6', linewidth=3, markersize=8)
ax2.fill_between(range(len(meses)), aprobacion, alpha=0.2, color='#3b82f6')
ax2.set_ylim(0, 100)  # ← Escala completa (0 a 100%)
ax2.set_title('✅ Versión HONESTA: aprobación estable ~64%', fontweight='bold', color='#1e3a8a')
ax2.set_xticks(range(len(meses)))
ax2.set_xticklabels(meses)
ax2.set_ylabel('Índice de Aprobación (%)')
ax2.axhline(50, color='gray', linestyle='--', linewidth=1, label='Línea del 50%')
ax2.annotate(f'Variación real:\n{max(aprobacion)-min(aprobacion):.1f} puntos', 
             xy=(2, 64), xytext=(3, 40),
             fontsize=9, color='#1e3a8a', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='#1e3a8a'))
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

print(f"\n🎯 El dato: el índice varió solo {max(aprobacion)-min(aprobacion):.1f} puntos porcentuales.")
print("   En el gráfico recortado parece una montaña rusa.")
print("   En el gráfico honesto es una línea casi plana.")
print("\n📊 En Power BI: Ajustar en Formato → Eje Y → Iniciar desde 0 (activar esta opción)")
print("📈 En Tableau: Eje derecho clic → Editar eje → Rango fijo desde 0")
```

## 2. Correlación NO es Causalidad 🦈🍦

Este es el error más famoso en estadística. Se da cuando dos cosas suben y bajan juntas, pero **una no causa la otra**.

**Ejemplo famoso real:**
> Cuando sube el consumo de helados, también suben los ahogamientos en piscinas. 
> ¿Deberíamos prohibir los helados para salvar vidas?

¡No! Ambas cosas suben en **verano** (la causa real es el calor).

| Correlación falsa famosa | Variable oculta |
|---|---|
| Helados 🍦 y ahogamientos 🏊 | Verano / calor |
| Zapatos grandes → mejores notas | Edad (los adultos tienen pies más grandes y más educación) |
| Más hospitales → más muertes | Donde hay más enfermos se construyen más hospitales |
| Más policías → más crimen | Las ciudades con más crimen contratan más policías |

### La pregunta clave siempre es:
**¿Hay una tercera variable que explique ambas cosas al mismo tiempo?**

## 3. Las 5 Preguntas del Detective de Datos 🔍

Antes de creer cualquier gráfico o estadística, hazte estas preguntas:

### ① ¿De dónde vienen los datos?
- ¿Quién los recopiló? ¿Por qué?
- ¿Tienen algún interés en que el resultado sea X?
- ¿Es una empresa mostrando sus propias ventas? (Sesgo de presentación)

### ② ¿Cuántos datos hay?
- Un estudio con 10 personas no dice lo mismo que uno con 10,000.
- "9 de 10 dentistas recomiendan..." — ¿cuántos dentistas preguntaron?

### ③ ¿Cómo midieron?
- "Clientes satisfechos" — ¿midieron solo a los que respondieron? (Sesgo de supervivencia)
- ¿Miden lo que realmente importa?

### ④ ¿Qué NO muestran?
- ¿Está el eje empezando en cero?
- ¿Muestran todo el período o solo los años convenientes?
- ¿Hay datos omitidos que cambiarían la conclusión?

### ⑤ ¿La conclusión sigue de los datos?
- "Las ventas subieron un 200%" — ¿de cuánto a cuánto?
- "Nuestro producto es el mejor" — ¿comparado con qué?
- ¿La causa realmente explica el efecto?

```python
# Ejercicio práctico: aplicar el checklist del detective

print("🕵️ EJERCICIO: Aplica el checklist de detective a este titular:")
print()
print('   "¡Nuestro nuevo suplemento XYZ aumentó la energía en un 87%!"')
print('   (Basado en estudio propio con 12 participantes voluntarios, 2 semanas)"')
print()
print("="*60)

checklist = [
    ("① ¿De dónde vienen los datos?",
     "La PROPIA empresa — sesgo de interés. No hay independencia."),
    ("② ¿Cuántos datos hay?",
     "Solo 12 personas — muestra extremadamente pequeña. No representativa."),
    ("③ ¿Cómo midieron?",
     "'Voluntarios': los que participan ya están motivados (sesgo de selección)."),
    ("④ ¿Qué NO muestran?",
     "¿Hubo grupo control (sin suplemento)? ¿Solo 2 semanas? ¿Efecto placebo?"),
    ("⑤ ¿La conclusión sigue de los datos?",
     "87% puede ser de 'me siento bien' a 'me siento muy bien' en encuesta subjetiva."),
]

for pregunta, respuesta in checklist:
    print(f"\n{pregunta}")
    print(f"   🔍 Hallazgo: {respuesta}")

print("\n" + "="*60)
print("VEREDICTO: Evidencia insuficiente. No compres el suplemento basándote en este estudio.")
```

## 4. Sesgos comunes que debes conocer 🧠

| Sesgo | Qué es | Ejemplo cotidiano |
|---|---|---|
| **Sesgo de confirmación** | Buscar solo datos que confirmen lo que ya creemos | Solo leer noticias que nos dan la razón |
| **Sesgo de supervivencia** | Solo analizar los que "sobrevivieron" | Estudiar solo las startups exitosas sin ver las que fracasaron |
| **Cherry-picking** | Elegir solo los datos convenientes | Mostrar solo los meses buenos de ventas |
| **Sesgo de anclaje** | Dar demasiada importancia al primer número que vemos | Un precio "rebajado" de $1000 a $800 parece bueno, pero puede que nunca valió $1000 |
| **Base rate neglect** | Ignorar la tasa base para calcular probabilidades | "1 en 3 personas vive en Asia" — eso es la mayoría, no una rareza |

### En Power BI y Tableau:

Estos sesgos también se cometen al construir dashboards:
- **Cherry-picking de fechas:** el período del reporte empieza justo después de la caída.
- **Sesgo de presentación:** la métrica del título (KPI) es la que mejor salió ese mes.
- **Sesgo de supervivencia en datos:** si el CRM solo registra ventas exitosas, no ves las perdidas.
