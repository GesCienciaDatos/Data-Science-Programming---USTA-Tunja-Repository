## ¿Qué vamos a aprender aquí? 🎯

Este es el último cuaderno "para no ingenieros" del Módulo 03. Aquí la mala noticia es que un dataset puede estar **perfectamente limpio** (sin huecos, sin repetidos, todo bien escrito) y aun así **mentirte** por completo. El problema no está en el dato — está en **cómo se recolectó** o **cómo se dibujó**.

Vamos a ver tres trampas clásicas:
1. El **sesgo de selección**: preguntarle solo a la gente "equivocada".
2. El **sesgo de supervivencia**: analizar solo a los que "sobrevivieron".
3. El **gráfico tramposo**: el mismo dato, dibujado para engañar.

---
## 1. Sesgo de selección: preguntarle solo a quien ya piensa como tú 🎯

**Analogía:** imagina que quieres saber si a la gente le gusta el fútbol, pero solo le preguntas a las personas que salen de un estadio después de un partido. ¡Por supuesto que la mayoría te va a decir que sí! No preguntaste a nadie que se quedó en casa.

**Ejemplo real de 1936:** una revista quiso predecir quién ganaría la elección presidencial de EE. UU. Le preguntaron a 10 millones de personas... pero sacadas de listas de teléfono y de dueños de carro. En plena crisis económica, eso significaba preguntarle solo a la gente con más dinero. La revista se equivocó feo.

```python
# Comparemos: la opinión de TODOS vs. la opinión de solo quienes "gritan más fuerte"

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(3)

# Opinión real de TODOS los clientes (promedio 6.5 sobre 10)
todos = np.clip(rng.normal(6.5, 1.8, size=3000), 0, 10)

# Pero solo responden la encuesta quienes tienen opiniones MUY extremas (muy contentos o muy enojados)
prob_responde = 0.15 + 0.6 * (np.abs(todos - 6.5) / 5)
responden = rng.random(len(todos)) < prob_responde
muestra = todos[responden]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
axes[0].hist(todos, bins=18, color="#fbbf24", alpha=0.85)
axes[0].set_title(f"✅ TODOS los clientes\nPromedio real: {todos.mean():.1f}")

axes[1].hist(muestra, bins=18, color="#dc2626", alpha=0.85)
axes[1].set_title(f"❌ Solo quienes respondieron\nPromedio de la encuesta: {muestra.mean():.1f}")

plt.tight_layout()
plt.show()

print("💡 ¡La encuesta 'miente' sin que nadie haya mentido! Solo respondieron los extremos.")
```

## 2. Sesgo de supervivencia: solo miras a los que sobrevivieron ✈️

**La historia real:** en la Segunda Guerra Mundial, examinaron los aviones que **regresaban** de sus misiones y vieron dónde tenían más balazos — en las alas y el cuerpo — y querían reforzar justo ahí.

Un matemático llamado **Abraham Wald** dijo: "¡Esperen! Esos son los aviones que SÍ volvieron. Los balazos ahí no eran mortales — por eso volvieron. Hay que reforzar donde estos aviones NO tienen agujeros... como el motor... ¡porque los que fueron alcanzados ahí nunca volvieron para que los contáramos!"

**Ejemplos cotidianos:**
- "Todas las startups exitosas dejaron la universidad" — ¿y las miles que la dejaron y fracasaron? Nadie las cuenta.
- "Mis clientes activos están felices" — ¿y los que se fueron enojados? Ya no están para responder tu encuesta.

## 3. El gráfico tramposo: mismo dato, dos historias 📉

El truco más viejo del mundo: **cortar el eje** de un gráfico. Si el eje no empieza en cero, una diferencia chiquitita se ve como un terremoto.

```python
# El truco del eje cortado: mismo dato, dos gráficos MUY diferentes

import matplotlib.pyplot as plt

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
aprobacion = [52.1, 51.8, 52.5, 51.9, 52.8, 53.2]  # ¡solo varía 1.4 puntos!

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5))
fig.suptitle("🔍 El mismo dato — ¡dos gráficos totalmente diferentes!", fontsize=12, fontweight="bold")

# Versión TRAMPOSA (eje cortado)
ax1.plot(meses, aprobacion, "o-", color="#ef4444", linewidth=3, markersize=7)
ax1.set_ylim(51.5, 53.5)
ax1.set_title("❌ Eje cortado:\n¡\"SE DISPARÓ!\"", color="#dc2626")

# Versión HONESTA (eje completo)
ax2.plot(meses, aprobacion, "o-", color="#1e3a8a", linewidth=3, markersize=7)
ax2.set_ylim(0, 100)
ax2.set_title("✅ Eje completo:\n\"está estable, ~52%\"", color="#1e3a8a")

plt.tight_layout()
plt.show()

print("💡 Truco de detective: SIEMPRE mira en qué número empieza el eje vertical antes de asustarte (o emocionarte).")
```

## Resumen Express 🎯

| Concepto | Una sola frase |
|---|---|
| **Sesgo de selección** | Si solo le preguntas a un grupo particular, tu respuesta ya viene "contaminada" |
| **Sesgo de supervivencia** | Analizar solo a los que "sobrevivieron" te da una foto incompleta y engañosa |
| **Eje cortado** | Si el eje no empieza en cero, cualquier cambio chiquito se ve gigante |
| **Detective de datos** | Ante cualquier gráfico, pregúntate: ¿a quién le preguntaron? ¿qué le pasó a los que faltan? ¿en qué número empieza el eje? |
