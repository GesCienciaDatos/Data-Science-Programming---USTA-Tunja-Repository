## 1.5.3 Técnicas para Identificar Sesgos y Errores en los Datos ⚖️

Un **sesgo** en datos no es un error aleatorio: es una **desviación sistemática** que empuja los resultados consistentemente en una dirección. A diferencia de un dato mal digitado (que se corrige con la auditoría de calidad del cuaderno anterior), un sesgo puede estar presente en un dataset perfectamente "limpio" — porque el problema está en **cómo se recolectaron o se presentaron** los datos, no en su formato.

En este cuaderno cubrimos tres frentes: el **sesgo de selección** (quién queda dentro y fuera de la muestra), el **sesgo de supervivencia** (analizar solo lo que "sobrevivió") y la **manipulación visual** (cómo un gráfico técnicamente correcto puede distorsionar la percepción).

## 2. Sesgo de Selección (Selection Bias) 🎯

El **sesgo de selección** ocurre cuando la muestra analizada no es representativa de la población que se quiere estudiar, porque el proceso de selección favorece sistemáticamente a un subgrupo.

### Caso histórico — la encuesta del *Literary Digest* (1936):
La revista *Literary Digest* envió 10 millones de encuestas por correo para predecir el resultado de la elección presidencial de EE. UU., y predijo la victoria de Alf Landon sobre Franklin D. Roosevelt. Roosevelt ganó por un margen histórico. ¿Qué falló? La lista de envío se construyó a partir de **directorios telefónicos y registros de automóviles** — en plena Gran Depresión, eso sobre-representaba a la población más adinerada, que votaba distinto al resto del electorado.

### Ejemplos comunes de sesgo de selección en analítica de negocio:
- Encuestas de satisfacción que solo responden los clientes muy contentos o muy enojados (los "neutros" no responden).
- Analizar solo usuarios que completaron el registro, ignorando a quienes abandonaron el proceso.
- Estudios de clima laboral con baja tasa de respuesta, donde solo participan los empleados más comprometidos.

```python
# ============================================================
# Simulación de Sesgo de Selección
# Población real vs. muestra sesgada (solo responden los extremos)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(3)

# Población real: satisfacción de TODOS los clientes (escala 0-10), distribución normal centrada en 6.5
poblacion_real = np.clip(rng.normal(6.5, 1.8, size=5000), 0, 10)

# Sesgo de selección: solo responden encuestas quienes tienen opiniones extremas
# (probabilidad de responder aumenta mientras más lejos está la opinión del promedio)
prob_responder = 0.15 + 0.6 * (np.abs(poblacion_real - 6.5) / 5)
responde = rng.random(len(poblacion_real)) < prob_responder
muestra_sesgada = poblacion_real[responde]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(poblacion_real, bins=20, color="#1e3a8a", alpha=0.75, edgecolor="white")
axes[0].axvline(poblacion_real.mean(), color="#16a34a", linewidth=2.5, linestyle="--",
                 label=f"Promedio REAL: {poblacion_real.mean():.2f}")
axes[0].set_title("✅ Población completa (todos los clientes)")
axes[0].set_xlabel("Satisfacción (0-10)"); axes[0].legend()
axes[0].spines["top"].set_visible(False); axes[0].spines["right"].set_visible(False)

axes[1].hist(muestra_sesgada, bins=20, color="#dc2626", alpha=0.75, edgecolor="white")
axes[1].axvline(muestra_sesgada.mean(), color="#dc2626", linewidth=2.5, linestyle="--",
                 label=f"Promedio MUESTRA: {muestra_sesgada.mean():.2f}")
axes[1].set_title(f"❌ Solo quienes respondieron la encuesta\n({len(muestra_sesgada)} de {len(poblacion_real)}, {responde.mean()*100:.0f}%)")
axes[1].set_xlabel("Satisfacción (0-10)"); axes[1].legend()
axes[1].spines["top"].set_visible(False); axes[1].spines["right"].set_visible(False)

plt.tight_layout()
plt.show()

print(f"Promedio en la población real:        {poblacion_real.mean():.2f}")
print(f"Promedio en la muestra sesgada:        {muestra_sesgada.mean():.2f}")
print(f"Desviación estándar poblacional real:  {poblacion_real.std():.2f}")
print(f"Desviación estándar de la muestra:     {muestra_sesgada.std():.2f}  (¡mucho mayor: opiniones polarizadas!)")
print("\n💡 La muestra sesgada no solo desplaza el promedio — también infla la dispersión,")
print("   porque sobre-representa las opiniones extremas y sub-representa a los clientes neutros.")
```

## 3. Sesgo de Supervivencia (Survivorship Bias) ✈️

El **sesgo de supervivencia** ocurre cuando se analizan únicamente los casos que "sobrevivieron" a un proceso de selección, ignorando los que fueron eliminados — y esos casos ausentes son precisamente los más informativos.

### El caso fundacional — Abraham Wald y la Marina de EE. UU. (1943):

Durante la Segunda Guerra Mundial, la Marina de EE. UU. quería reforzar el blindaje de sus aviones de combate, pero blindar todo el avión lo haría demasiado pesado para volar. Los ingenieros examinaron los aviones que **regresaban** de sus misiones y observaron dónde tenían más impactos de bala — principalmente en el fuselaje y las alas — y propusieron reforzar esas zonas.

El estadístico **Abraham Wald** señaló el error: esos eran los aviones que **sí sobrevivieron** a pesar de los impactos. Los impactos en esas zonas eran, precisamente, los que un avión podía resistir. La recomendación correcta era reforzar las zonas **sin impactos en los aviones que volvían** — como el motor — porque los aviones alcanzados ahí probablemente **nunca regresaron** para ser contados.

```python
# ============================================================
# Sesgo de Supervivencia — El error de Abraham Wald (WWII), recreado
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(5)

fig, axes = plt.subplots(1, 2, figsize=(13, 6.5))
fig.suptitle("Sesgo de Supervivencia: el error de Abraham Wald (WWII)\n"
             "Los aviones que REGRESARON vs. el análisis correcto",
             fontsize=12, fontweight="bold")

def dibujar_avion(ax, titulo, marcar_motor=False, marcar_fuselaje=True, color_titulo="#dc2626"):
    ax.set_xlim(0, 10); ax.set_ylim(0, 10); ax.set_aspect("equal")
    ax.set_xticks([]); ax.set_yticks([])
    ax.set_title(titulo, fontsize=10.5, fontweight="bold", color=color_titulo)

    fuselaje = plt.Polygon([[3, 4], [7, 4], [8, 5], [7, 6], [3, 6], [2, 5]],
                            closed=True, facecolor="#e2e8f0", edgecolor="#334155", linewidth=2)
    ax.add_patch(fuselaje)
    ala_sup = plt.Polygon([[4, 6], [6, 6], [6.5, 8], [3.5, 8]],
                           closed=True, facecolor="#cbd5e1", edgecolor="#334155", linewidth=2)
    ala_inf = plt.Polygon([[4, 4], [6, 4], [6.5, 2], [3.5, 2]],
                           closed=True, facecolor="#cbd5e1", edgecolor="#334155", linewidth=2)
    ax.add_patch(ala_sup); ax.add_patch(ala_inf)
    motor = plt.Circle((5, 7), 0.4, facecolor="#94a3b8", edgecolor="#334155", linewidth=1.5)
    ax.add_patch(motor)

    if marcar_fuselaje:
        for _ in range(15):
            x_d = rng.uniform(3.5, 6.5); y_d = rng.uniform(4.2, 5.8)
            ax.plot(x_d, y_d, "rx", markersize=8, markeredgewidth=2)

    if marcar_motor:
        ax.plot(5, 7, "X", markersize=22, color="#16a34a", markeredgewidth=3)

dibujar_avion(axes[0], "❌ Análisis INCORRECTO\n('reforzar donde hay impactos visibles')",
              marcar_motor=False, marcar_fuselaje=True, color_titulo="#dc2626")
axes[0].text(5, 1.0, "Estos aviones SÍ volvieron:\nlos impactos ahí eran sobrevivibles",
             ha="center", fontsize=8.5, color="#dc2626")

dibujar_avion(axes[1], "✅ Análisis de Wald\n('reforzar donde NO hay impactos... porque no volvieron')",
              marcar_motor=True, marcar_fuselaje=False, color_titulo="#16a34a")
axes[1].text(5, 1.0, "Los aviones con daño en el motor\nnunca llegaron a ser contados",
             ha="center", fontsize=8.5, color="#16a34a")

plt.tight_layout()
plt.show()

print("💡 Lección: los datos disponibles (aviones que regresaron) NO son una muestra")
print("   representativa — están sesgados hacia los casos que 'sobrevivieron'.")
print("   Pregunta clave de auditoría: '¿qué le pasó a los casos que NO están en mis datos?'")
print("\n   Ejemplos de negocio: analizar solo clientes activos (¿y los que se fueron?),")
print("   solo empresas que siguen operando (¿y las que quebraron?), solo proyectos exitosos")
print("   (¿y los cancelados antes de completarse?).")
```

## 4. Manipulación Visual: Ejes Truncados y Escalas Engañosas 📉

Incluso con datos perfectamente correctos y sin ningún sesgo de muestreo, un gráfico puede **distorsionar la percepción** mediante decisiones de diseño técnicamente válidas pero engañosas.

| Técnica | Descripción | Efecto en la percepción |
|---|---|---|
| **Eje Y truncado** | El eje Y no empieza en cero | Exagera diferencias pequeñas, las hace ver dramáticas |
| **Escala doble (doble eje Y)** | Dos variables con escalas independientes superpuestas | Sugiere una correlación que puede ser artificial |
| **Cherry-picking de periodo** | Mostrar solo el rango de fechas favorable | Oculta tendencias adversas fuera del rango mostrado |
| **Área vs. longitud** | Usar el área de una figura para representar una cantidad (el ojo no percibe área linealmente) | Exagera las diferencias entre categorías |
| **Perspectiva 3D innecesaria** | Gráficos de torta o barras en 3D | Distorsiona el tamaño percibido de las porciones frontales |
| **Línea base móvil** | Cambiar el punto de referencia de una comparación | Manipula si un cambio se percibe como crecimiento o caída |

La técnica más común — y más fácil de reproducir — es el **truncamiento del eje Y**. A continuación generamos el **mismo dataset** con dos versiones del gráfico: una diseñada para exagerar (engañosa) y otra diseñada para informar con precisión (honesta).

```python
# ============================================================
# El mismo gráfico, dos historias: eje Y truncado vs. eje Y honesto
# ============================================================

import matplotlib.pyplot as plt

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
aprobacion = [52.1, 51.8, 52.5, 51.9, 52.8, 53.2]  # variación real: apenas 1.4 puntos

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
fig.suptitle("El Mismo Dato — Dos Historias Completamente Diferentes", fontsize=13, fontweight="bold")

# --- Versión ENGAÑOSA: eje Y truncado ---
axes[0].plot(meses, aprobacion, "o-", color="#dc2626", linewidth=3,
             markersize=8, markerfacecolor="white", markeredgewidth=2)
axes[0].fill_between(range(len(meses)), aprobacion, min(aprobacion) - 0.1, alpha=0.25, color="#dc2626")
axes[0].set_ylim(51.5, 53.5)  # ¡EJE TRUNCADO!
axes[0].set_xticks(range(len(meses))); axes[0].set_xticklabels(meses)
axes[0].set_ylabel("Aprobación (%)")
axes[0].set_title("❌ ENGAÑOSO: eje Y truncado (51.5–53.5%)\n\"¡La aprobación se DISPARÓ!\"",
                   fontsize=10.5, color="#dc2626")
axes[0].spines["top"].set_visible(False); axes[0].spines["right"].set_visible(False)
axes[0].yaxis.grid(True, alpha=0.3)

# --- Versión HONESTA: eje Y completo desde 0 ---
axes[1].plot(meses, aprobacion, "o-", color="#1e3a8a", linewidth=3,
             markersize=8, markerfacecolor="white", markeredgewidth=2)
axes[1].fill_between(range(len(meses)), aprobacion, 0, alpha=0.1, color="#1e3a8a")
axes[1].set_ylim(0, 100)  # Escala completa y honesta
axes[1].set_xticks(range(len(meses))); axes[1].set_xticklabels(meses)
axes[1].set_ylabel("Aprobación (%)")
axes[1].set_title("✅ HONESTO: eje Y desde 0 (0–100%)\n\"La aprobación es estable, ~52%\"",
                   fontsize=10.5, color="#1e3a8a")
axes[1].spines["top"].set_visible(False); axes[1].spines["right"].set_visible(False)
axes[1].yaxis.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

variacion_real = max(aprobacion) - min(aprobacion)
print(f"📊 Variación real en los datos: {variacion_real:.1f} puntos porcentuales sobre una escala de 100.")
print(f"   Conclusión objetiva: la variación es estadísticamente insignificante.")
print("\n📊 En Power BI: el eje Y por defecto NO siempre empieza en 0.")
print("   Corrección: Panel Formato → Eje Y → 'Inicio' → escribir 0.")
print("📈 En Tableau: clic derecho en el eje → 'Editar eje' → 'Rango fijo' → Min: 0.")
```

## 5. Otras Técnicas de Detección de Errores en los Datos 🕵️

Además del sesgo de selección, el sesgo de supervivencia y la manipulación visual, hay otras señales de alerta que un analista entrenado debe vigilar:

- **Paradoja de Simpson:** una tendencia que aparece en varios grupos de datos separados desaparece o se revierte cuando los grupos se combinan. Siempre vale la pena desagregar un promedio general por subgrupos antes de confiar en él.
- **Negligencia de la tasa base (*base rate neglect*):** ignorar qué tan común es realmente un evento. Un test médico con 99% de precisión puede producir más falsos positivos que verdaderos positivos si la enfermedad es muy poco frecuente en la población.
- **Confusión entre correlación y causalidad:** dos variables que suben juntas (por ejemplo, consumo de helados y ahogamientos) pueden compartir una tercera causa oculta (el calor del verano) sin que una cause la otra.
- **Verificación cruzada con fuente independiente:** cuando sea posible, comparar la cifra reportada contra una fuente externa (por ejemplo, cifras oficiales del DANE) antes de usarla como base de una decisión.

## 6. Checklist de Auditoría de Sesgos y Errores ✅

Aplica este checklist ante cualquier dataset, dashboard o informe visual — en Power BI, Tableau o cualquier herramienta:

### A. Sesgo de Selección
- [ ] ¿Cómo se seleccionó la muestra? ¿Todos los casos tuvieron la misma probabilidad de ser incluidos?
- [ ] ¿La tasa de respuesta o de captura es alta, o solo un subgrupo particular respondió?

### B. Sesgo de Supervivencia
- [ ] ¿El análisis incluye los casos que "no sobrevivieron" (clientes perdidos, proyectos cancelados, empresas cerradas)?
- [ ] ¿Se está sacando una conclusión solo a partir de los casos exitosos o visibles?

### C. Integridad Visual
- [ ] ¿El eje Y empieza en cero (o hay una justificación explícita para truncarlo)?
- [ ] ¿Las áreas y tamaños son proporcionales a los valores reales?
- [ ] ¿Las escalas son consistentes entre gráficos que se están comparando?

### D. Sesgos de Presentación
- [ ] ¿El título del gráfico es descriptivo o ya sugiere una conclusión predeterminada?
- [ ] ¿Se muestran todos los datos relevantes o solo el periodo/segmento favorable?

> 📌 **Regla de oro de Alberto Cairo:** *"Un gráfico honesto es aquel que no esconde ni distorsiona ningún dato que sea relevante para el propósito declarado de la visualización."*
