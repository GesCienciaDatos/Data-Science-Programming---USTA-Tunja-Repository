## 1.5.1 Los Tres Principios del Pensamiento Crítico Aplicados a Datos 🧠

El **pensamiento crítico** es el proceso disciplinado de evaluar activamente información en lugar de aceptarla pasivamente. Cuando el objeto de análisis son datos y visualizaciones, este proceso se apoya en **tres principios rectores**:

| Principio | Pregunta que responde | Rol en el análisis de datos |
|---|---|---|
| **1. Escepticismo** | ¿Debería creer esto? | Filtro inicial: no aceptar una afirmación solo porque viene acompañada de un gráfico o una cifra |
| **2. Análisis** | ¿De qué partes se compone esto? | Descomponer la afirmación, el dato o la visualización en sus elementos verificables |
| **3. Evaluación** | ¿Qué tan buena es la evidencia? | Ponderar la calidad de cada componente con criterios explícitos, no con intuición |

> 📌 **Kahneman (2002, Nobel de Economía):** El cerebro humano opera con dos sistemas: **Sistema 1** (rápido, intuitivo, propenso a sesgos y a aceptar patrones sin cuestionarlos) y **Sistema 2** (lento, deliberado, analítico). Un gráfico bien diseñado —o mal intencionado— habla directamente al Sistema 1. El pensamiento crítico con datos es el hábito de **forzar el Sistema 2** antes de sacar conclusiones.

Estos tres principios no son exclusivos de la ciencia de datos: son la base de la epistemología científica. Lo que hace especial a este módulo es aplicarlos de forma **operacional**, con checklists y código, al contexto específico de tablas, dashboards y gráficos.

## 2. Principio I — Escepticismo: Cuestionar Antes de Creer 🤔

El **escepticismo metodológico** no es desconfianza generalizada ni cinismo — es la disposición a **suspender la aceptación** de una afirmación basada en datos hasta verificar su origen, su método y su contexto.

### Preguntas de escepticismo aplicadas a datos:

- **¿Quién lo dice?** ¿La fuente tiene un interés directo en el resultado (empresa vendiendo su propio producto, estudio financiado por una de las partes)?
- **¿Con qué tamaño de muestra?** Un titular como *"9 de cada 10 personas prefieren X"* puede basarse en 10 encuestados o en 10,000 — el titular suena igual en ambos casos.
- **¿Comparado con qué?** Una cifra aislada ("+35% de crecimiento") no dice nada sin una base de comparación.
- **¿Qué se excluyó?** ¿Se descartaron observaciones, periodos o grupos? ¿Por qué?

El escepticismo es especialmente importante frente a afirmaciones basadas en **muestras pequeñas**, porque la variabilidad estadística natural puede producir resultados espectaculares por puro azar.

```python
# ============================================================
# Escepticismo en acción: ¿por qué desconfiar de una muestra pequeña?
# Simulamos 10,000 "estudios" con una moneda justa (p=0.5) y n=12
# para ver qué tan seguido un resultado "impresionante" ocurre por azar
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

n_estudios = 10_000
n_participantes = 12
p_real = 0.5  # verdad: el suplemento NO tiene ningún efecto (50/50, como una moneda)

# Cada "estudio" simula n_participantes personas, cada una responde sí/no al azar
resultados = rng.binomial(n=n_participantes, p=p_real, size=n_estudios) / n_participantes

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(resultados * 100, bins=np.arange(0, 101, 100 / n_participantes),
        color="#1e3a8a", alpha=0.75, edgecolor="white")
ax.axvline(87.5, color="#dc2626", linewidth=2.5, linestyle="--",
           label='"87% de mejora" (11 de 12)')
ax.set_xlabel("% de \"éxito\" observado en el estudio simulado")
ax.set_ylabel("Número de estudios simulados")
ax.set_title('Con n=12 y CERO efecto real, ¿qué tan seguido "sale" un 87%+ por puro azar?')
ax.legend()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()
plt.show()

pct_87_o_mas = (resultados >= 11 / 12).mean() * 100
print(f"Con una moneda justa (sin ningún efecto real) y n=12 participantes:")
print(f"   → En el {pct_87_o_mas:.1f}% de los estudios simulados, 11 o más de 12 'aciertan' por puro azar.")
print("\n💡 Lección de escepticismo: un resultado llamativo con muestra pequeña")
print("   puede no reflejar ningún efecto real. El tamaño de muestra importa tanto")
print("   como el resultado mismo — quizás más.")
```

## 3. Principio II — Análisis: Descomponer Antes de Concluir 🔬

El **análisis crítico** consiste en separar una afirmación o una visualización en sus **componentes verificables**, en lugar de reaccionar a la impresión general que produce. Frente a un gráfico o una cifra, el analista descompone:

| Componente | Pregunta de análisis |
|---|---|
| **Definición de términos** | ¿Qué significa exactamente "cliente satisfecho" o "usuario activo"? |
| **Método de medición** | ¿Cómo se recolectó el dato? ¿Encuesta, sensor, autoreporte? |
| **Tendencia vs. ruido** | ¿El patrón que veo es una señal real o variación aleatoria? |
| **Unidad de comparación** | ¿Se compara lo mismo en ambos periodos/grupos (misma métrica, misma población)? |

Una habilidad analítica clave es distinguir una **tendencia real** del **ruido estadístico** en series de datos — algo que el ojo humano hace muy mal sin ayuda.

```python
# ============================================================
# Análisis en acción: separar tendencia de ruido
# ¿El "pico" que se ve en el gráfico es una señal real o solo ruido?
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)

dias = np.arange(1, 61)
tendencia_real = 100 + 0.3 * dias                      # crecimiento lento y real
ruido = rng.normal(0, 8, size=len(dias))                # variación aleatoria del día a día
ventas_diarias = tendencia_real + ruido

serie = pd.Series(ventas_diarias, index=dias)
media_movil_7d = serie.rolling(window=7, center=True).mean()

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(dias, ventas_diarias, color="#94a3b8", linewidth=1.2, marker="o", markersize=3)
pico_dia = int(np.argmax(ventas_diarias)) + 1
axes[0].annotate("¡Pico histórico!\n¿tendencia real?", xy=(pico_dia, ventas_diarias.max()),
                  xytext=(pico_dia - 18, ventas_diarias.max() + 5),
                  arrowprops=dict(arrowstyle="->", color="#dc2626"), color="#dc2626", fontsize=9)
axes[0].set_title("❌ Sin análisis: reaccionar al dato crudo del día")
axes[0].set_xlabel("Día"); axes[0].set_ylabel("Ventas")
axes[0].spines["top"].set_visible(False); axes[0].spines["right"].set_visible(False)

axes[1].plot(dias, ventas_diarias, color="#cbd5e1", linewidth=1, alpha=0.6, label="Dato diario (ruido incluido)")
axes[1].plot(dias, media_movil_7d, color="#1e3a8a", linewidth=2.5, label="Media móvil 7 días (tendencia)")
axes[1].set_title("✅ Con análisis: separar tendencia real de ruido")
axes[1].set_xlabel("Día"); axes[1].set_ylabel("Ventas")
axes[1].legend(fontsize=9)
axes[1].spines["top"].set_visible(False); axes[1].spines["right"].set_visible(False)

plt.tight_layout()
plt.show()

print("💡 El 'pico histórico' del día", pico_dia, "es en su mayoría ruido aleatorio.")
print("   La media móvil revela la tendencia real: un crecimiento lento y sostenido,")
print("   no un evento excepcional en ese día puntual.")
```

## 4. Principio III — Evaluación: Ponderar la Evidencia con Criterios 📊

**Evaluar** es asignar un juicio de calidad a la evidencia usando **criterios explícitos y repetibles**, en vez de una impresión subjetiva. Adaptando el espíritu del marco de causalidad de Bradford Hill (1965) al contexto de afirmaciones basadas en datos, podemos evaluar cualquier afirmación con cinco criterios:

| Criterio | Pregunta | Puntaje alto si... |
|---|---|---|
| **Tamaño de muestra** | ¿Cuántas observaciones respaldan la afirmación? | La muestra es grande y representativa |
| **Independencia de la fuente** | ¿Quién midió y quién se beneficia del resultado? | La fuente no tiene conflicto de interés |
| **Transparencia metodológica** | ¿Se explica cómo se midió y se pueden auditar los datos? | El método es público y reproducible |
| **Replicabilidad** | ¿Otros estudios independientes llegan a lo mismo? | El resultado se ha replicado |
| **Plausibilidad** | ¿Existe un mecanismo razonable que explique el resultado? | Hay una explicación causal coherente |

Convertir estos criterios en un **puntaje explícito** evita que la evaluación dependa del estado de ánimo o de qué tan "bonito" se ve el gráfico que acompaña la afirmación.

```python
# ============================================================
# Evaluación en acción: puntaje explícito de calidad de evidencia
# ============================================================

import pandas as pd

def evaluar_evidencia(nombre, muestra, independencia, transparencia, replicabilidad, plausibilidad):
    """Cada criterio se califica de 0 (débil) a 5 (fuerte)."""
    criterios = {
        "Tamaño de muestra": muestra,
        "Independencia de la fuente": independencia,
        "Transparencia metodológica": transparencia,
        "Replicabilidad": replicabilidad,
        "Plausibilidad": plausibilidad,
    }
    puntaje_total = sum(criterios.values())
    return {"Afirmación": nombre, **criterios, "Puntaje Total (/25)": puntaje_total}

evaluaciones = [
    evaluar_evidencia(
        "Suplemento XYZ 'aumenta energía 87%' (estudio propio, n=12, sin control)",
        muestra=0, independencia=0, transparencia=1, replicabilidad=0, plausibilidad=1,
    ),
    evaluar_evidencia(
        "Meta-análisis de 40 ensayos clínicos controlados (n=15,000) sobre un fármaco",
        muestra=5, independencia=4, transparencia=5, replicabilidad=5, plausibilidad=4,
    ),
    evaluar_evidencia(
        "Encuesta interna de satisfacción de una empresa sobre su propio servicio",
        muestra=2, independencia=1, transparencia=2, replicabilidad=1, plausibilidad=2,
    ),
]

df_eval = pd.DataFrame(evaluaciones).set_index("Afirmación")
display(df_eval)

print("\n💡 El puntaje explícito convierte una decisión subjetiva ('me suena creíble')")
print("   en un juicio trazable y defendible ('obtuvo 4/25 en independencia y transparencia').")
```

## 5. Los Tres Principios en Acción: Caso Integrador 🧩

Apliquemos el flujo completo — **escepticismo → análisis → evaluación** — a un caso típico: un dashboard interno reporta que *"la satisfacción del cliente subió 40% este trimestre"*.

| Paso | Pregunta guía | Hallazgo en este caso |
|---|---|---|
| **1. Escepticismo** | ¿Quién mide "satisfacción" y con qué interés? | El equipo que reporta la métrica es el mismo que diseñó la campaña que se está evaluando |
| **2. Análisis** | ¿Qué componentes conforman el 40%? | La "satisfacción" se mide solo entre quienes respondieron una encuesta post-compra (no toda la base de clientes) |
| **3. Evaluación** | ¿Qué tan fuerte es la evidencia? | Tamaño de muestra bajo (8% de tasa de respuesta), sin grupo de comparación, fuente no independiente → evidencia débil |

**Conclusión del caso:** el 40% no es necesariamente falso, pero **no es evidencia suficiente** para tomar decisiones de negocio sin una auditoría adicional — por ejemplo, verificar si la tasa de respuesta introduce **sesgo de selección** (tema del cuaderno 1.5.3 de este módulo).

## 6. Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Escepticismo** | Suspender la aceptación de una afirmación hasta verificar fuente, muestra y comparación |
| **Análisis** | Descomponer la afirmación en sus partes verificables — definiciones, método, tendencia vs. ruido |
| **Evaluación** | Asignar un juicio de calidad con criterios explícitos y repetibles, no con intuición |
| **Sistema 1 vs. Sistema 2** | Los gráficos hablan al Sistema 1 (rápido); el pensamiento crítico exige activar el Sistema 2 (lento) |
| **Muestra pequeña** | Resultados llamativos con pocos datos pueden explicarse enteramente por azar |
