## ¿Qué es Power Query y para qué sirve? ⚡

Power Query es como tener un **asistente de cocina** antes de cocinar:
- La cocina = Power BI (donde haces los gráficos)
- Los ingredientes = tus datos en Excel, CSV, SQL...
- Power Query = el asistente que **lava, pela y corta** los ingredientes antes de cocinar

Sin Power Query, los datos llegan sucios a Power BI y los gráficos quedan mal. Con Power Query, los datos llegan perfectamente preparados.

### ¿Cuándo necesitas Power Query?

| Situación | ¿Necesitas Power Query? |
|---|---|
| Mis datos en Excel ya están limpios y ordenados | ❌ No (carga directa) |
| Tengo datos en múltiples archivos que quiero unir | ✅ Sí (Append) |
| Los headers están en la fila 3, no en la fila 1 | ✅ Sí |
| Las fechas están en texto "20/01/2023" | ✅ Sí |
| Tengo columnas que deberían ser filas (o viceversa) | ✅ Sí (Unpivot) |
| Quiero combinar la tabla de ventas con la tabla de clientes | ✅ Sí (Merge/Join) |

## Power Query vs Pandas: son gemelos 👯

Hacen lo mismo pero con diferente lenguaje:

| Power Query (M) | Pandas (Python) | ¿Qué hace? |
|---|---|---|
| `Tabla → Eliminar filas en blanco` | `df.dropna()` | Quitar filas vacías |
| `Transformar → Reemplazar valores` | `df.fillna(0)` | Poner 0 donde hay vacío |
| `Transformar → Cambiar tipo` | `df['col'].astype(int)` | Convertir texto a número |
| `Transformar → Recortar espacios` | `df['col'].str.strip()` | Quitar espacios al inicio/fin |
| `Combinar → Combinar consultas` | `pd.merge(df1, df2, on='id')` | Unir dos tablas |
| `Combinar → Anexar consultas` | `pd.concat([df1, df2])` | Apilar dos tablas |
| `Transformar → Anular dinamización` | `df.melt(id_vars=['Mes'])` | Convertir columnas en filas |
| `Agregar columna → Columna condicional` | `np.where(cond, 'A', 'B')` | Si [col] > 100 entonces 'Alto' |

## Los 5 problemas más comunes y cómo resolverlos ⚙️

### Problema 1: Las fechas están como texto

**Power Query:**
```
1. Seleccionar la columna de fechas
2. Transformar → Tipo de datos → Fecha
3. Si el formato es dd/mm/yyyy: Configuración regional → Español Colombia
```

**Pandas equivalente:**
```python
df['fecha'] = pd.to_datetime(df['fecha'], format='%d/%m/%Y')
```

---

### Problema 2: La primera columna tiene el nombre de la empresa (no encabezados)

**Power Query:**
```
1. Inicio → Quitar filas → Quitar filas superiores → 2 (o las que haya antes)
2. Inicio → Usar primera fila como encabezado
```

**Pandas equivalente:**
```python
df = pd.read_csv('archivo.csv', skiprows=2, header=0)
```

---

### Problema 3: Tengo 12 archivos de ventas (uno por mes) y quiero uno solo

**Power Query:**
```
1. Inicio → Obtener datos → Carpeta
2. Seleccionar la carpeta donde están los 12 archivos
3. Combinar → Combinar y cargar
¡Power Query los une automáticamente!
```

**Pandas equivalente:**
```python
import glob
archivos = glob.glob('ventas/*.csv')
df_total = pd.concat([pd.read_csv(f) for f in archivos])
```

---

### Problema 4: Las regiones están en columnas (Norte, Sur, Este, Oeste) pero Power BI las necesita en filas

**Power Query:**
```
1. Seleccionar SOLO las columnas Norte, Sur, Este, Oeste
2. Transformar → Anular dinamización de columnas
3. Renombrar: 'Atributo' → 'Region', 'Valor' → 'Ventas'
```

**Pandas equivalente:**
```python
df_largo = df.melt(id_vars=['Mes'], 
                    var_name='Region', 
                    value_name='Ventas')
```

---

### Problema 5: Quiero unir la tabla de ventas con la tabla de productos

**Power Query:**
```
1. En la consulta de Ventas → Inicio → Combinar consultas → Combinar
2. Seleccionar la tabla Productos
3. Seleccionar las columnas clave: id_producto en ambas
4. Tipo: Externa izquierda (todas las ventas + info del producto)
5. Expandir la columna nueva resultante
```

**Pandas equivalente:**
```python
df_completo = pd.merge(df_ventas, df_productos, 
                        on='id_producto', 
                        how='left')
```

## El lenguaje M en 2 minutos ⏱️

No necesitas escribir M manualmente — el GUI de Power Query lo escribe por ti. Pero es bueno saber leerlo:

```m
let
    // Aquí va todo lo que hace Power Query, paso a paso:
    
    Paso1_Cargar = Csv.Document(File.Contents("datos.csv"), ...),
    
    Paso2_Encabezados = Table.PromoteHeaders(Paso1_Cargar),
    
    Paso3_Tipos = Table.TransformColumnTypes(Paso2_Encabezados, 
                                              {{"Fecha", type date},
                                               {"Ventas", type number}}),
    
    Paso4_Filtrar = Table.SelectRows(Paso3_Tipos, 
                                      each [Ventas] > 100),
    
    Paso5_NuevaColumna = Table.AddColumn(Paso4_Filtrar, 
                                          "Categoria",
                                          each if [Ventas] > 1000 then "Alta" else "Normal")
in
    Paso5_NuevaColumna   // ← El resultado final
```

**Traducido al español normal:**
1. Cargar el archivo CSV
2. La primera fila son los nombres de las columnas
3. Convertir Fecha a tipo fecha y Ventas a tipo número
4. Solo quedarse con las ventas mayores a 100
5. Agregar una columna nueva que diga "Alta" si > 1000, si no "Normal"

> 💡 **Pro tip:** Para ver el código M de cualquier consulta en Power Query: `Vista → Editor avanzado`. Así puedes ver exactamente qué está haciendo Power Query detrás de cámaras.

## Resumen en una página ⚡

| ¿Qué necesito hacer? | En Power Query | En Pandas |
|---|---|---|
| Quitar filas vacías | Quitar filas en blanco | `df.dropna()` |
| Poner cero donde hay vacío | Reemplazar valores → null → 0 | `df.fillna(0)` |
| Cambiar texto a número | Transformar → Tipo: Número decimal | `df['col'].astype(float)` |
| Cambiar texto a fecha | Transformar → Tipo: Fecha | `pd.to_datetime(df['col'])` |
| Unir dos tablas | Combinar → Combinar consultas | `pd.merge(df1, df2, on='clave')` |
| Apilar tablas | Combinar → Anexar consultas | `pd.concat([df1, df2])` |
| Columnas → Filas | Anular dinamización | `df.melt(id_vars=...)` |
| Filas → Columnas | Dinamizar columna | `df.pivot_table(...)` |
| Quitar duplicados | Quitar filas → Quitar duplicados | `df.drop_duplicates()` |
| Agregar columna calculada | Agregar columna → Personalizada | `df['nueva'] = ...` |

> ✅ **Conclusión:** Power Query y Pandas son la misma herramienta en dos idiomas diferentes. Si aprendes uno, el otro se aprende mucho más rápido.
