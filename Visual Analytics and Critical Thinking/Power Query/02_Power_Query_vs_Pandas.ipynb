## Propósito de este Cuaderno 🎯

Este cuaderno es una **guía de referencia rápida** — el diccionario bilingüe entre **Power Query (M)** y **Pandas (Python)**. Puedes usarlo para:

- Traducir código M a Pandas cuando necesites automatizar en Python.
- Traducir código Pandas a M cuando implementes en Power BI.
- Entender qué hace un pipeline de Power Query dado su código M.
- Comparar el rendimiento y capacidades de ambas herramientas.

Cada sección incluye:
1. El código **M (Power Query)** equivalente
2. El código **Pandas** ejecutable
3. La operación equivalente en **SQL** (para referencia adicional)

````python
# Setup: Dataset de trabajo
import pandas as pd
import numpy as np

np.random.seed(42)
n = 500

df = pd.DataFrame({
    'id': range(1, n+1),
    'nombre': [f'Producto_{i%50:03d}' for i in range(n)],
    'categoria': np.random.choice(['A', 'B', 'C', 'D'], n),
    'subcategoria': np.random.choice(['X1', 'X2', 'Y1', 'Y2', 'Z1'], n),
    'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], n),
    'fecha': pd.date_range('2023-01-01', periods=n, freq='12h'),
    'ventas': np.random.exponential(1000, n).round(2),
    'costo': np.random.exponential(600, n).round(2),
    'cantidad': np.random.randint(1, 50, n),
    'activo': np.random.choice([True, False], n, p=[0.85, 0.15])
})
df.loc[np.random.choice(n, 30), 'ventas'] = np.nan
df.loc[np.random.choice(n, 15), 'categoria'] = np.nan
df['ganancia'] = (df['ventas'] - df['costo']).round(2)

print(f"Dataset: {df.shape}")
print(f"Nulos: ventas={df['ventas'].isna().sum()}, categoria={df['categoria'].isna().sum()}")
````

## 📋 TABLA MAESTRA: Power Query M vs Pandas vs SQL

````python
# ============================================================
# SECCIÓN 1: CARGA Y CONEXIÓN
# ============================================================

print("🔌 SECCIÓN 1: CARGA Y CONEXIÓN")
print("="*70)

comparativa = [
    ("Cargar CSV",
     'Csv.Document(File.Contents("data.csv"), [Delimiter=","])',
     'pd.read_csv("data.csv")',
     'SELECT * FROM OPENROWSET(...)'),
    ("Cargar Excel",
     'Excel.Workbook(File.Contents("data.xlsx"))',
     'pd.read_excel("data.xlsx", sheet_name="Hoja1")',
     'N/A (importar primero)'),
    ("Conectar SQL Server",
     'Sql.Database("servidor", "basedatos")',
     'pd.read_sql(query, engine)',
     'USE basedatos; SELECT ...'),
    ("Leer desde API",
     'Json.Document(Web.Contents("https://api.ejemplo.com"))',
     'pd.json_normalize(requests.get(url).json())',
     'N/A (función externa)'),
]

for op, m, pandas, sql in comparativa:
    print(f"\n  [{op}]")
    print(f"    M:      {m}")
    print(f"    Pandas: {pandas}")
    print(f"    SQL:    {sql}")
````

````python
# ============================================================
# SECCIÓN 2: INSPECCIÓN Y EXPLORACIÓN
# ============================================================

print("🔍 SECCIÓN 2: INSPECCIÓN")
print("="*70)

comparativa_2 = [
    ("Ver primeras filas",
     'Table.FirstN(Tabla, 5)',
     'df.head(5)',
     'SELECT TOP 5 * FROM tabla'),
    ("Contar filas",
     'Table.RowCount(Tabla)',
     'len(df)  # o df.shape[0]',
     'SELECT COUNT(*) FROM tabla'),
    ("Tipos de columnas",
     'En GUI: Vista → Inspector de columnas',
     'df.dtypes',
     'INFORMATION_SCHEMA.COLUMNS'),
    ("Estadísticas descriptivas",
     'En GUI: Vista → Distribución de columnas',
     'df.describe()',
     'N/A (manual con AVG, MIN, MAX...)'),
    ("Contar nulos",
     'Table.SelectRows(t, each [col] = null) → RowCount',
     'df.isna().sum()',
     'SELECT COUNT(*)-COUNT(col) FROM tabla'),
]

for op, m, pandas, sql in comparativa_2:
    print(f"\n  [{op}]")
    print(f"    M:      {m}")
    print(f"    Pandas: {pandas}")
    print(f"    SQL:    {sql}")

# Ejemplo ejecutable
print("\n📊 Ejemplo ejecutado:")
print(f"   len(df) = {len(df)}")
print(f"   df.isna().sum():")
print(df.isna().sum()[df.isna().sum() > 0].to_string())
````

````python
# ============================================================
# SECCIÓN 3: FILTRADO Y SELECCIÓN
# ============================================================

print("🎯 SECCIÓN 3: FILTRADO Y SELECCIÓN")
print("="*70)

ops_filtrado = [
    ("Filtrar por valor",
     'Table.SelectRows(t, each [categoria] = "A")',
     'df[df["categoria"] == "A"]',
     'WHERE categoria = \'A\''),
    ("Filtrar por lista (IN)",
     'Table.SelectRows(t, each List.Contains({"A", "B"}, [categoria]))',
     'df[df["categoria"].isin(["A", "B"])]',
     'WHERE categoria IN (\'A\', \'B\')'),
    ("Filtrar mayor que",
     'Table.SelectRows(t, each [ventas] > 1000)',
     'df[df["ventas"] > 1000]',
     'WHERE ventas > 1000'),
    ("Filtrar AND",
     'Table.SelectRows(t, each [ventas] > 1000 and [categoria] = "A")',
     'df[(df["ventas"] > 1000) & (df["categoria"] == "A")]',
     'WHERE ventas > 1000 AND categoria = \'A\''),
    ("Filtrar OR",
     'Table.SelectRows(t, each [ventas] > 1000 or [categoria] = "A")',
     'df[(df["ventas"] > 1000) | (df["categoria"] == "A")]',
     'WHERE ventas > 1000 OR categoria = \'A\''),
    ("Eliminar nulos",
     'Table.SelectRows(t, each [ventas] <> null)',
     'df.dropna(subset=["ventas"])',
     'WHERE ventas IS NOT NULL'),
    ("Seleccionar columnas",
     'Table.SelectColumns(t, {"id", "nombre", "ventas"})',
     'df[["id", "nombre", "ventas"]]',
     'SELECT id, nombre, ventas FROM tabla'),
    ("Eliminar columnas",
     'Table.RemoveColumns(t, {"costo", "activo"})',
     'df.drop(columns=["costo", "activo"])',
     'SELECT * EXCEPT(costo, activo) — BigQuery'),
]

for op, m, pandas, sql in ops_filtrado:
    print(f"\n  [{op}]")
    print(f"    M:      {m}")
    print(f"    Pandas: {pandas}")
    print(f"    SQL:    {sql}")

# Demostración
print("\n📊 Demo — Filtrar AND (ventas > 1000 Y categoria = 'A'):")
result = df[(df['ventas'] > 1000) & (df['categoria'] == 'A')]
print(f"   Filas resultado: {len(result)} de {len(df)}")
````

````python
# ============================================================
# SECCIÓN 4: AGRUPACIÓN Y AGREGACIÓN
# ============================================================

print("📊 SECCIÓN 4: AGRUPACIÓN Y AGREGACIÓN")
print("="*70)

ops_grupo = [
    ("Group By básico",
     'Table.Group(t, {"categoria"}, {{"Total", each List.Sum([ventas])}})',
     'df.groupby("categoria")["ventas"].sum()',
     'SELECT categoria, SUM(ventas) FROM t GROUP BY categoria'),
    ("Group By múltiple",
     'Table.Group(t, {"cat", "region"}, {{"Total", each List.Sum([ventas])}})',
     'df.groupby(["categoria", "region"])["ventas"].sum()',
     'GROUP BY categoria, region'),
    ("Múltiples agregaciones",
     'Table.Group(t, {"cat"}, {{"Suma", each List.Sum([v])}, {"Prom", each List.Average([v])}})',
     'df.groupby("categoria").agg(Suma=("ventas","sum"), Prom=("ventas","mean"))',
     'SELECT cat, SUM(v), AVG(v) GROUP BY cat'),
    ("Ranking / RANK()",
     'No nativo — requiere índice manual con List.Sort',
     'df["rank"] = df["ventas"].rank(ascending=False)',
     'RANK() OVER (ORDER BY ventas DESC)'),
    ("Running total (acumulado)",
     'List.Accumulate o columna índice acumulado',
     'df["acum"] = df["ventas"].cumsum()',
     'SUM(ventas) OVER (ORDER BY fecha)'),
]

for op, m, pandas, sql in ops_grupo:
    print(f"\n  [{op}]")
    print(f"    M:      {m}")
    print(f"    Pandas: {pandas}")
    print(f"    SQL:    {sql}")

# Demostración
print("\n📊 Demo — Múltiples agregaciones:")
resumen = df.groupby('categoria').agg(
    Total_Ventas=('ventas', 'sum'),
    Promedio=('ventas', 'mean'),
    Maximo=('ventas', 'max'),
    Conteo=('ventas', 'count')
).round(2)
print(resumen.to_string())
````

````python
# ============================================================
# SECCIÓN 5: FECHAS Y TIEMPO
# ============================================================

print("📅 SECCIÓN 5: FECHAS Y TIEMPO")
print("="*70)

ops_fecha = [
    ("Extraer año",
     'Date.Year([Fecha])',
     'df["fecha"].dt.year',
     'YEAR(fecha)'),
    ("Extraer mes",
     'Date.Month([Fecha])',
     'df["fecha"].dt.month',
     'MONTH(fecha)'),
    ("Extraer trimestre",
     'Date.QuarterOfYear([Fecha])',
     'df["fecha"].dt.quarter',
     'QUARTER(fecha)  -- MySQL/BigQuery'),
    ("Nombre del mes",
     'Date.MonthName([Fecha])',
     'df["fecha"].dt.month_name()',
     'MONTHNAME(fecha)'),
    ("Diferencia en días",
     'Duration.Days([Fecha2] - [Fecha1])',
     '(df["fecha2"] - df["fecha1"]).dt.days',
     'DATEDIFF(day, fecha1, fecha2)'),
    ("Inicio del mes",
     'Date.StartOfMonth([Fecha])',
     'df["fecha"] - pd.offsets.MonthBegin(1)',
     'DATE_TRUNC("month", fecha)'),
    ("Filtrar último año",
     'Table.SelectRows(t, each Date.Year([Fecha]) = 2024)',
     'df[df["fecha"].dt.year == 2024]',
     'WHERE YEAR(fecha) = 2024'),
]

for op, m, pandas, sql in ops_fecha:
    print(f"\n  [{op}]")
    print(f"    M:      {m}")
    print(f"    Pandas: {pandas}")
    print(f"    SQL:    {sql}")

# Demostración
print("\n📊 Demo — Extracción de componentes de fecha:")
df_fechas = df[['fecha']].copy()
df_fechas['año'] = df_fechas['fecha'].dt.year
df_fechas['mes'] = df_fechas['fecha'].dt.month
df_fechas['trimestre'] = df_fechas['fecha'].dt.quarter
df_fechas['nombre_mes'] = df_fechas['fecha'].dt.month_name()
df_fechas['dia_semana'] = df_fechas['fecha'].dt.day_name()
print(df_fechas.head(8).to_string(index=False))
````

## Tabla de Referencia Final 📋

### Cuándo usar cada herramienta:

| Criterio | Usa Power Query (M) | Usa Pandas (Python) |
|---|---|---|
| **Entorno de trabajo** | Power BI Desktop/Excel | Jupyter, VS Code, scripts |
| **Audiencia del resultado** | Equipo de negocio (dashboard PBI) | Técnicos, científicos, automatización |
| **Escala de datos** | Hasta ~500k filas eficientemente | Hasta límite de RAM (millones) |
| **Integración ML** | No nativa | Total (sklearn, pytorch, etc.) |
| **Control de versiones** | Limitado | Git completo |
| **Costo** | Incluido en Power BI | Open source |
| **Velocidad de desarrollo** | Rápida (GUI) | Media-Alta (código) |

### Patrones comunes de uso combinado:

```
Opción 1: Python → Power BI
  Python (limpieza avanzada + ML) → exportar CSV → Power Query (carga) → DAX (métricas) → Visuals

Opción 2: Power Query → Python
  Power Query (ETL en PBI) → exportar datos preparados → Python (análisis avanzado, ML)

Opción 3: Python embebido en Power BI
  Power BI → Visual de Python → código matplotlib/seaborn se ejecuta dentro del informe
```

> 🎓 **Conclusión:** Power Query y Pandas son herramientas complementarias, no competidoras. Un científico de datos competente en el ecosistema Microsoft sabe usar ambas en el contexto correcto.
