---## Tres herramientas, tres preguntas distintas 🧰- **Gráfico de línea:** "¿Cómo cambió esto con el tiempo?"- **Histograma:** "¿Dónde se concentran mis datos?"- **Scatter plot:** "¿Estas dos cosas están relacionadas?"Vamos una por una, con ejemplos de la vida diaria.

---## 1. El gráfico de línea: la foto de tu progreso 📈Imagina que pesas una fruta todos los días y anotas el peso mientras se seca. Si conectas los puntos con una línea, ves de inmediato la **tendencia**: ¿está bajando de peso rápido o lento?⚠️ Ojo: solo conectas con línea cosas que tienen un ORDEN (como los días). Nunca conectes con línea cosas sin orden, como "mi color favorito" y "el de mi hermano" — ¡eso no tiene sentido!

````python
# ============================================================# Line graph cotidiano: temperatura de una taza de café enfriándose# ============================================================import numpy as npimport matplotlib.pyplot as pltminutos = np.arange(0, 30, 2)temperatura = 85 * np.exp(-0.05 * minutos) + 20  # se enfría hacia temp. ambientefig, ax = plt.subplots(figsize=(8, 4))ax.plot(minutos, temperatura, marker='o', color='#dc2626', linewidth=2.5)ax.set_title('Tu café enfriándose minuto a minuto ☕➡️🧊', fontweight='bold')ax.set_xlabel('Minutos desde que serviste el café')ax.set_ylabel('Temperatura (°C)')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()
````

---## 2. El histograma: ver dónde viven los números 🏠Piénsalo como un parking organizado por pisos de precio:```Salarios de 100 personas:Rango $1-2M:    ████████████████████ 35 personasRango $2-3M:    ████████████████████████████ 45 personasRango $3-5M:    ████████████ 15 personasRango $5-10M:   ████ 4 personasRango >$10M:    █ 1 persona   ← OUTLIER/ATÍPICO```| Forma | ¿Cómo se ve? | Ejemplo real ||---|---|---|| **Normal (campana)** | 🔔 Simétrica | Alturas humanas || **Sesgada a la derecha** | 📐 Cola larga a la derecha | Sueldos, precios de casas || **Bimodal** | 🌋🌋 Dos picos | Alturas de hombres y mujeres juntos |

````python
# ============================================================# Histograma cotidiano: salarios simulados# ============================================================import numpy as npimport matplotlib.pyplot as pltnp.random.seed(42)salarios = np.clip(np.random.lognormal(mean=7.5, sigma=0.6, size=1000), 800, 20000)fig, ax = plt.subplots(figsize=(8, 4))ax.hist(salarios, bins=30, color='#3b82f6', edgecolor='white', alpha=0.85)ax.axvline(np.mean(salarios), color='#dc2626', linestyle='--', linewidth=2, label=f'Media: ${np.mean(salarios):.0f}k')ax.axvline(np.median(salarios), color='#16a34a', linestyle='--', linewidth=2, label=f'Mediana: ${np.median(salarios):.0f}k')ax.set_title('¿Cómo se distribuyen 1,000 salarios simulados?', fontweight='bold')ax.set_xlabel('Salario (miles de pesos)')ax.legend()ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()print("La media la 'jalan' hacia arriba los salarios altos; la mediana es más justa aquí.")
````

---## 3. El scatter plot: encontrar relaciones entre dos números 🔗Un scatter plot pone un punto por cada observación. Si los puntos forman una línea imaginaria → **están relacionados**.- **↗ Positiva:** cuando una sube, la otra también (horas de estudio y nota del examen)- **↘ Negativa:** cuando una sube, la otra baja (velocidad del auto y tiempo del viaje)- **➡ Sin relación:** los puntos no forman ningún patrón (número de letras del nombre y el sueldo)📊 **En Power BI:** Visual nativo "Gráfico de dispersión"📈 **En Tableau:** Arrastra una medida a Columns y otra a Rows → scatter automático

---## Cheat Sheet: ¿Qué gráfico numérico usar? 🎯| Pregunta que me hacen | Gráfico a usar ||---|---|| "¿Cómo cambiaron las ventas mes a mes?" | **Gráfico de línea** || "¿Cómo se distribuyen los precios?" | **Histograma** || "¿Hay relación entre el precio y las ventas?" | **Scatter plot** || "¿Cuál es el precio típico? ¿Hay raros?" | **Box plot** |> 💡 **Tip:** Empieza siempre con el histograma. Ver la distribución te dice TODO sobre tus datos: si están sesgados, si hay dos grupos, si hay outliers.

## Resumen y Puntos Clave 🎯- Línea = tendencia en el tiempo. Histograma = dónde se concentran los valores. Scatter = si dos cosas están relacionadas.- Nunca conectes con una línea cosas que no tienen un orden natural.- El histograma es tu punto de partida para entender cualquier variable numérica nueva.
