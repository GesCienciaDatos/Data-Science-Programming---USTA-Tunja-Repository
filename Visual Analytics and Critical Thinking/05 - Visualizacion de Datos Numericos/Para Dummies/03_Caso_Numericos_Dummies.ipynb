---## Ponte el sombrero de detective 🕵️‍♀️Vas a resolver un caso con datos inventados (pero realistas) de una tienda en línea. Tienes dos preguntas del jefe:1. "¿Cómo van las ventas mes a mes?"2. "¿La página está siendo lenta? ¿Por qué?"Vamos a usar las herramientas que ya conoces (línea, histograma, scatter, box plot) para responder — como un detective que junta pistas.

---## 1. Pista #1: Las ventas mes a mes 📈Como las ventas tienen un orden (enero, febrero, marzo...) usamos un **gráfico de línea**.

```python
# ============================================================# Pista 1: ventas mensuales (con "picos" de fin de año)# ============================================================import numpy as npimport pandas as pdimport matplotlib.pyplot as pltnp.random.seed(2024)meses = pd.date_range('2022-01-01', periods=36, freq='MS')tendencia = np.linspace(80, 140, 36)estacionalidad = 15 * np.sin(2 * np.pi * np.arange(36) / 12 - np.pi / 2) + 15ventas = tendencia + estacionalidad + np.random.normal(0, 5, 36)fig, ax = plt.subplots(figsize=(9, 4))ax.plot(meses, ventas, marker='o', color='#1e3a8a', linewidth=2)ax.set_title('Ventas de la tienda mes a mes', fontweight='bold')ax.set_ylabel('Ventas (millones $)')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()print("🔍 Pista encontrada: las ventas suben con el tiempo y tienen picos repetidos cada año (¿Navidad?).")
```

---## 2. Pista #2: ¿La página está lenta? 🐢El jefe dice que el checkout debe responder en menos de 300 ms. Con miles de visitas, un solo número (el promedio) no basta — necesitamos ver **toda la distribución** con un histograma.

```python
# ============================================================# Pista 2: distribución de tiempos de respuesta# ============================================================np.random.seed(2024)normales = np.random.gamma(2.0, 90, 9700)lentos = np.random.gamma(2.0, 350, 300)tiempos = np.concatenate([normales, lentos])fig, ax = plt.subplots(figsize=(9, 4))ax.hist(tiempos, bins=60, color='#0ea5e9', edgecolor='white', alpha=0.85)ax.axvline(300, color='#dc2626', linestyle='--', linewidth=2, label='Límite prometido: 300 ms')ax.set_title('¿Qué tan rápido responde el checkout?', fontweight='bold')ax.set_xlabel('Tiempo de respuesta (ms)')ax.legend()ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()pct = (tiempos > 300).mean() * 100print(f"🔍 Pista encontrada: el {pct:.1f}% de las visitas tardan MÁS de lo prometido.")print("   Hay una 'colita' de casos lentos — no es un problema constante, sino de picos.")
```

---## 3. Pista #3: ¿Por qué se pone lenta? 🔗Sospecha del detective: "¿será que se pone lenta cuando hay mucha gente en la página al mismo tiempo?" Para comprobarlo, comparamos **tráfico diario** contra **lentitud promedio** con un scatter plot.

```python
# ============================================================# Pista 3: scatter plot tráfico vs lentitud# ============================================================np.random.seed(2024)trafico = np.random.uniform(500, 5000, 90)latencia = 60 + 0.045 * trafico + np.random.normal(0, 15, 90)fig, ax = plt.subplots(figsize=(8, 4.5))ax.scatter(trafico, latencia, color='#7c3aed', alpha=0.6, s=40)r = np.corrcoef(trafico, latencia)[0, 1]ax.set_title(f'¿Más visitas = más lentitud? (r = {r:.2f})', fontweight='bold')ax.set_xlabel('Visitas por día')ax.set_ylabel('Lentitud promedio (ms)')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()print("🔍 Caso resuelto: sí hay relación. Entre más gente entra al mismo tiempo, más lenta se pone la página.")print("   Recomendación para el jefe: reforzar los servidores en los días de más tráfico.")
```

---## Resumen y Puntos Clave 🎯- Resolvimos el caso combinando 3 herramientas: **línea** (tendencia de ventas), **histograma** (qué tan lenta es la página) y **scatter** (por qué se pone lenta).- Como un buen detective, ninguna pista por sí sola cuenta toda la historia — hay que combinarlas.- Este es el mismo flujo que usarás en proyectos reales: mirar la tendencia, mirar la distribución, buscar relaciones, y comunicar la conclusión con el gráfico correcto.
