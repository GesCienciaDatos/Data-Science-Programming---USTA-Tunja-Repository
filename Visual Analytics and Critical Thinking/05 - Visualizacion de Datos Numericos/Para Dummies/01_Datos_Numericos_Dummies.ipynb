---## El agua de la llave vs los huevos del cartón 🥚💧Dos formas de pensar los números:- **Agua saliendo de la llave (CONTINUA):** puedes tener 1 litro, 1.5 litros, 1.53782 litros... siempre puedes partirla en pedacitos más pequeños. El tiempo, el peso, el dinero, la temperatura se comportan así.- **Huevos en un cartón (DISCRETA):** tienes 0, 1, 2, 3 huevos... nunca "2.5 huevos" (bueno, a menos que rompas uno, ¡y ya no sería un huevo entero!). El número de hijos, de compras, de errores en un sistema se comportan así.

---## 1. ¿Cómo saber cuál es cuál? 🕵️Pregúntate: **"¿Podría este valor tener decimales que tengan sentido?"**| Pregúntate esto... | Si la respuesta es SÍ → CONTINUA | Si la respuesta es NO → DISCRETA ||---|---|---|| "¿Cuánto pesa el paquete?" | 2.347 kg ✅ | — || "¿Cuántos paquetes envié hoy?" | — | 7 paquetes (nunca 7.3) ✅ || "¿Cuánto tardó en cargar la página?" | 1.284 segundos ✅ | — || "¿Cuántas veces falló la página hoy?" | — | 3 veces (nunca 3.5) ✅ |

```python
# ============================================================# Viendo la diferencia con un ejemplo cotidiano: café ☕# ============================================================import matplotlib.pyplot as pltimport numpy as npnp.random.seed(5)fig, axes = plt.subplots(1, 2, figsize=(11, 4))# Continua: gramos de café usados por taza (se puede medir con decimales)gramos_cafe = np.random.normal(18, 1.5, 400)axes[0].hist(gramos_cafe, bins=30, color='#3b82f6', edgecolor='white')axes[0].set_title('CONTINUA: gramos de café por taza\n(18.347g es un valor perfectamente válido)', fontsize=9, fontweight='bold')axes[0].set_xlabel('Gramos')# Discreta: número de tazas de café tomadas al día (se cuenta)tazas_dia = np.random.poisson(2.5, 400)valores, conteos = np.unique(tazas_dia, return_counts=True)axes[1].bar(valores, conteos, color='#f59e0b', edgecolor='white')axes[1].set_title('DISCRETA: tazas de café al día\n(nunca vas a tomar "2.5 tazas")', fontsize=9, fontweight='bold')axes[1].set_xlabel('Número de tazas')axes[1].set_xticks(valores)for ax in axes:    ax.spines['top'].set_visible(False)    ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()
```

---## 2. ¿Por qué me debería importar esta diferencia? 🤷Porque si la ignoras, tus gráficos **mienten sin que te des cuenta**:- Si agrupas "número de tazas de café" (0, 1, 2, 3, 4...) en bins raros de un histograma genérico, puedes terminar con una barra que mezcle "1 y 2 tazas" — ¡y eso ya no significa nada claro!- Si intentas dibujar una barra separada para CADA valor posible de "gramos de café" (que tiene infinitos decimales posibles), tu gráfico tendría miles de barras microscópicas e ilegibles.La solución: para datos discretos, un bin = un valor entero. Para datos continuos, agrupas en rangos (bins) razonables.

---## Resumen y Puntos Clave 🎯- Continuo = como el agua: siempre se puede partir en pedacitos más pequeños (decimales con sentido).- Discreto = como los huevos: vienen en unidades completas, contables, sin "medios valores".- Truco rápido: si la variable viene de **contar** algo, es discreta. Si viene de **medir** algo con un instrumento, es continua.- Elegir mal la técnica de gráfico según este tipo de dato hace que tu visualización sea confusa o hasta engañosa.
