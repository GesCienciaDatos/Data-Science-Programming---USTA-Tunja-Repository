## ¿Qué son datos numéricos? (y por qué son diferentes a los categóricos) 🔢

Si los datos categóricos son **etiquetas** (Norte, Sur, Electrónica...), los datos numéricos son **medidas** — cosas que puedes sumar, restar, promediar:

- Precio: $45.99, $120.00...
- Temperatura: 18.5°C, 22.3°C...
- Edad: 25, 34, 52 años...
- Tiempo: 3.5 horas, 1.2 minutos...

**La pregunta clave con datos numéricos:**
> *¿Cómo se DISTRIBUYEN los valores? ¿Dónde está la mayoría? ¿Hay valores raros?*

Para responder estas preguntas tenemos herramientas especiales que no aplican a los categóricos.

---
## 1. El Histograma: ver dónde viven los números 🏠

Un histograma responde: **¿Cuántos valores caen en cada rango?**

Piénsalo como un parking organizado por pisos de precio:
```
Salarios de 100 personas:

Rango $1-2M:    ████████████████████ 35 personas
Rango $2-3M:    ████████████████████████████ 45 personas
Rango $3-5M:    ████████████ 15 personas
Rango $5-10M:   ████ 4 personas
Rango >$10M:    █ 1 persona   ← OUTLIER/ATÍPICO
```

### Formas de distribuciones y qué significan:

| Forma | ¿Cómo se ve? | Qué significa | Ejemplo real |
|---|---|---|---|
| **Normal (campana)** | 🔔 Simétrica | La mayoría está en el centro | Alturas humanas |
| **Sesgada a la derecha** | 📐 Cola larga a la derecha | Pocos valores muy altos | Sueldos, precios de casas |
| **Sesgada a la izquierda** | 📐 Cola larga a la izquierda | Pocos valores muy bajos | Notas de un examen fácil |
| **Bimodal** | 🌋🌋 Dos picos | Dos grupos diferentes mezclados | Alturas de hombres y mujeres juntos |

````python
# ============================================================
# El Histograma: código simple para entender la distribución
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

# Simular salarios (distribución típica: sesgada a la derecha)
salarios = np.random.lognormal(mean=7.5, sigma=0.6, size=1000)  # En miles de pesos
salarios = np.clip(salarios, 800, 20000)  # Entre $800k y $20M

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('¿Cómo se distribuyen 1,000 salarios simulados?', fontsize=12, fontweight='bold')

# 1. Histograma básico
axes[0].hist(salarios, bins=30, color='#3b82f6', edgecolor='white', alpha=0.85)
axes[0].axvline(np.mean(salarios), color='#dc2626', linestyle='--', linewidth=2,
                label=f'Media: ${np.mean(salarios):.0f}k')
axes[0].axvline(np.median(salarios), color='#16a34a', linestyle='--', linewidth=2,
                label=f'Mediana: ${np.median(salarios):.0f}k')
axes[0].set_title('① Histograma (30 bins)', fontweight='bold')
axes[0].set_xlabel('Salario (miles de pesos)', fontsize=9)
axes[0].set_ylabel('Número de personas', fontsize=9)
axes[0].legend(fontsize=8)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# 2. Pocos bins vs muchos bins
axes[1].hist(salarios, bins=8, color='#f59e0b', edgecolor='white', alpha=0.85, label='8 bins')
axes[1].set_title('② Pocos bins: pierdes detalle', fontweight='bold')
axes[1].set_xlabel('Salario (miles)', fontsize=9)
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

axes[2].hist(salarios, bins=100, color='#10b981', edgecolor='white', alpha=0.85, label='100 bins')
axes[2].set_title('③ Muchos bins: demasiado ruido', fontweight='bold')
axes[2].set_xlabel('Salario (miles)', fontsize=9)
axes[2].spines['top'].set_visible(False)
axes[2].spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

media = np.mean(salarios)
mediana = np.median(salarios)
print(f"\n💡 Observaciones:")
print(f"   Media:   ${media:.0f}k — La jala hacia arriba los salarios altos")
print(f"   Mediana: ${mediana:.0f}k — El punto donde el 50% gana más y 50% gana menos")
print(f"   → La mediana es más representativa para salarios sesgados")
print("\n📊 En Power BI: Visual 'Histograma' no es nativo — buscar en AppSource.")
print("   Alternativa: crear bins con DAX y hacer barras normales.")
print("\n📈 En Tableau: Show Me → Histogram. Ajustar tamaño del bin en 'Edit Axis'.")
````

## 2. El Box Plot: el gráfico más informativo que nadie entiende a primera vista 📦

El **box plot** (o diagrama de caja) concentra MUCHA información en poco espacio. Vamos por partes:

```
                 ┌───────────────────────┐
    ─────┬───────┤         CAJA          ├──────────┬───────   ← Valores fuera de la línea
         │       │                       │          │
         │      Q1─────────────────────Q3           │
         │                 │                        │
        Min             Mediana                    Max*
                    (Q2 = 50%)
    
    ○ ○     ← Outliers (valores raros)
```

| Parte del box plot | Significado | En español |
|---|---|---|
| **Línea del medio** | Mediana (Q2) | El valor del 50% |
| **Caja** | Del 25% al 75% de los datos (IQR) | El "grueso" de la distribución |
| **Líneas laterales (bigotes)** | Hasta 1.5 × IQR | Los valores "normales" extremos |
| **Puntos sueltos** | Outliers | Valores inusualmente raros |

````python
# Box Plot simplificado — paso a paso

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

np.random.seed(42)

# Salarios por departamento (3 grupos)
salarios_ventas = np.abs(np.random.normal(2500, 600, 100))
salarios_tech = np.abs(np.random.normal(4200, 1200, 100))
salarios_admin = np.abs(np.random.normal(1800, 400, 100))
# Agregar algunos outliers
salarios_tech = np.append(salarios_tech, [12000, 11500, 10800])  # Directivos tech

import pandas as pd
df_sal = pd.DataFrame({
    'salario': np.concatenate([salarios_ventas, salarios_tech, salarios_admin]),
    'depto': ['Ventas']*100 + ['Tecnología']*103 + ['Administración']*100
})

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Box Plot vs Violin — ¿Cuál muestra más?', fontsize=12, fontweight='bold')

# Box plot
sns.boxplot(data=df_sal, x='depto', y='salario', ax=axes[0],
            palette=['#3b82f6', '#10b981', '#f59e0b'])
axes[0].set_title('Box Plot — Compacto y claro', fontweight='bold')
axes[0].set_xlabel('')
axes[0].set_ylabel('Salario mensual (miles $)', fontsize=10)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# Anotar el box plot
axes[0].annotate('Outliers\n(ejecutivos)', xy=(1, 11000), xytext=(0.5, 10000),
                 fontsize=8, color='red', fontweight='bold',
                 arrowprops=dict(arrowstyle='->', color='red'))

# Violin plot
sns.violinplot(data=df_sal, x='depto', y='salario', ax=axes[1],
               palette=['#3b82f6', '#10b981', '#f59e0b'], inner='quartile')
axes[1].set_title('Violin — Muestra la distribución completa', fontweight='bold')
axes[1].set_xlabel('')
axes[1].set_ylabel('Salario mensual (miles $)', fontsize=10)
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

for depto in ['Ventas', 'Tecnología', 'Administración']:
    d = df_sal[df_sal['depto'] == depto]['salario']
    print(f"\n{depto}: Mediana=${d.median():.0f}k | Q1=${d.quantile(0.25):.0f}k | Q3=${d.quantile(0.75):.0f}k | Max=${d.max():.0f}k")

print("\n📊 En Power BI: 'Box and Whisker Chart' de AppSource (gratis).")
print("📈 En Tableau: Show Me → Box-and-Whisker. Arrastra la dimensión a Columns.")
````

## 3. El Scatter Plot: encontrar relaciones entre dos números 🔗

Un scatter plot pone un punto por cada observación. El eje X es un número, el eje Y es otro número. Si los puntos forman una línea → **están relacionados**.

```
Precio ($)  ↑
            │      •
  Alta      │  •      •
            │•   •       •
  Media     │  •    •
            │•        •
  Baja      └─────────────────→ Calificación (1-5 estrellas)
              Baja  Media  Alta
```

### Interpretando la dirección:
- **↗ Positiva:** cuando una sube, la otra también (precio y tamaño de casa)
- **↘ Negativa:** cuando una sube, la otra baja (descuento y ganancias)
- **➡ Sin relación:** los puntos no forman ningún patrón (número de letras del nombre y el sueldo)

📊 **En Power BI:** Visual nativo "Gráfico de dispersión" (Scatter Chart)  
📈 **En Tableau:** Arrastra medida a Columns y otra a Rows → automáticamente scatter

## Cheat Sheet: ¿Qué gráfico numérico usar? 🎯

| Pregunta que me hacen | Gráfico a usar |
|---|---|
| "¿Cómo se distribuyen los precios?" | **Histograma** |
| "¿Cuál es el precio típico? ¿Hay raros?" | **Box plot** |
| "¿Hay relación entre el precio y las ventas?" | **Scatter plot** |
| "¿Cómo cambiaron las ventas mes a mes?" | **Gráfico de línea** (Módulo 06) |
| "¿Están correlacionadas estas 5 variables?" | **Heatmap de correlación** |

> 💡 **Tip:** Empieza siempre con el histograma. Ver la distribución te dice TODO sobre tus datos: si están sesgados, si hay dos grupos, si hay outliers. Con esa información ya sabes qué análisis tiene sentido.
