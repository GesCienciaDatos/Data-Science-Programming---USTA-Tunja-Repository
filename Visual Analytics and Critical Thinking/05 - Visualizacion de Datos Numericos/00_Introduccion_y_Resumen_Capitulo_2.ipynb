---## 3. ¿Por qué los datos numéricos necesitan técnicas distintas? 🔍Un dato categórico (por ejemplo, "región de venta") solo puede compararse por **conteo o frecuencia** dentro de cada categoría. Un dato numérico, en cambio, tiene propiedades matemáticas mucho más ricas:- **Orden natural:** 25 años es más que 20 años; "Norte" no es "más" que "Sur".- **Distancias interpretables:** la diferencia entre 10°C y 20°C es la misma que entre 20°C y 30°C.- **Operaciones aritméticas válidas:** se puede promediar, sumar, calcular varianza — cosas sin sentido en una categoría.- **Distribución:** los valores numéricos se agrupan, se dispersan, tienen forma (simétrica, sesgada, bimodal) — un concepto que no aplica a categorías.Estas propiedades son justamente las que explotan las técnicas del Capítulo 2: histogramas, KDE, box plots, scatter plots y gráficos de línea, que veremos en detalle en los siguientes notebooks de este módulo.

```python
# ============================================================# Comparación rápida: pregunta categórica vs pregunta numérica# ============================================================import matplotlib.pyplot as pltimport numpy as npnp.random.seed(7)fig, ax = plt.subplots(figsize=(9, 4))tiempos_respuesta = np.random.gamma(shape=2.0, scale=40, size=2000)  # msax.hist(tiempos_respuesta, bins=40, color='#0ea5e9', edgecolor='white', alpha=0.85)ax.axvline(np.median(tiempos_respuesta), color='#dc2626', linestyle='--', linewidth=2,           label=f'Mediana: {np.median(tiempos_respuesta):.0f} ms')ax.set_title('Pregunta numérica típica del Capítulo 2:\n"¿Cómo se distribuyen los tiempos de respuesta de un sistema?"',             fontsize=11, fontweight='bold')ax.set_xlabel('Tiempo de respuesta (ms)')ax.set_ylabel('Frecuencia')ax.legend()ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()
```
