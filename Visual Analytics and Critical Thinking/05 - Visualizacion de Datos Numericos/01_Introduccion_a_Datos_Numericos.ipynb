---## 1. ¿Qué es un dato numérico? 🔢Un **dato numérico** (o cuantitativo) es aquel que representa una **cantidad medible**, sobre la cual tienen sentido operaciones aritméticas como suma, resta, promedio o varianza.Esto lo diferencia radicalmente de un **dato categórico**, que representa una **etiqueta o clase** (visto en el Módulo 04) donde esas operaciones no tienen sentido: "Norte + Sur" no es una operación válida, pero "\$45.000 + \$60.000" sí lo es.Dentro de los datos numéricos existen dos subtipos fundamentales que determinan cómo debemos visualizarlos: **continuos** y **discretos**.

```python
# ============================================================# Clasificación rápida de variables con pandas# ============================================================import pandas as pdimport numpy as npnp.random.seed(42)df_demo = pd.DataFrame({    'id_cliente': range(1, 8),    'ciudad': ['Tunja', 'Bogotá', 'Medellín', 'Tunja', 'Cali', 'Bogotá', 'Tunja'],       # categórica    'num_compras': [3, 1, 7, 2, 0, 5, 4],                                                # numérica discreta    'gasto_total': [125430.50, 42100.00, 389250.75, 88000.10, 0.0, 210500.30, 155200.0], # numérica continua})print(df_demo)print("\nTipos de dato inferidos por pandas:")print(df_demo.dtypes)print("\n'ciudad'       -> categórica (etiqueta, no se promedia)")print("'num_compras'  -> numérica DISCRETA (conteo entero, valores separados)")print("'gasto_total'  -> numérica CONTINUA (puede tomar cualquier valor decimal)")
```

---## 2. Datos Numéricos Continuos 🌊Una variable **continua** puede tomar (en teoría) **infinitos valores** dentro de un rango, incluyendo decimales. Entre dos valores cualesquiera siempre existe otro valor posible.| Ejemplo | Rango típico | ¿Por qué es continua? ||---|---|---|| Temperatura (°C) | -40 a 50 | Puede ser 21.347°C || Precio de un producto | $0 a $∞ | Puede ser $45.999,99 || Tiempo de respuesta de un sistema | 0 a ∞ ms | Puede ser 132.456 ms || Estatura de una persona | 40 a 250 cm | Puede ser 172.3 cm || Ingreso mensual | 0 a ∞ | Puede tener cualquier decimal |> 🔑 **Regla práctica:** si la respuesta a "¿podría este valor tener decimales con sentido?" es SÍ, casi siempre es continua.

```python
# ============================================================# Visualizando una variable continua: tiempo de respuesta (ms)# ============================================================import matplotlib.pyplot as pltimport numpy as npnp.random.seed(10)tiempo_respuesta = np.random.gamma(shape=2.2, scale=35, size=3000)  # milisegundos, valores continuosfig, ax = plt.subplots(figsize=(9, 4))ax.hist(tiempo_respuesta, bins=50, color='#1e3a8a', edgecolor='white', alpha=0.85)ax.set_title('Variable continua: Tiempo de respuesta de un sistema (ms)', fontweight='bold')ax.set_xlabel('Tiempo de respuesta (ms) — puede tomar cualquier valor decimal')ax.set_ylabel('Frecuencia')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()print(f"Ejemplos de valores individuales: {np.round(tiempo_respuesta[:5], 3)}")print("Nota que ninguno es un número 'redondo' — así se comportan los datos continuos.")
```

---## 3. Datos Numéricos Discretos 🔲Una variable **discreta** solo puede tomar **valores separados y contables** — típicamente números enteros. No existen valores "intermedios" con sentido entre dos valores discretos consecutivos.| Ejemplo | Valores posibles | ¿Por qué es discreta? ||---|---|---|| Número de hijos | 0, 1, 2, 3... | No existe "1.5 hijos" || Número de compras en un mes | 0, 1, 2, 3... | No existen "2.3 compras" || Número de errores en un log | 0, 1, 2, 3... | Se cuenta, no se mide || Calificación en estrellas (1-5) | 1, 2, 3, 4, 5 | Escala fija de enteros || Número de empleados de una empresa | 0, 1, 2... | Personas completas |> 🔑 **Regla práctica:** si la variable resulta de **contar** cosas (no de medir con un instrumento continuo), casi siempre es discreta.

```python
# ============================================================# Visualizando una variable discreta: número de compras mensuales# ============================================================import matplotlib.pyplot as pltimport numpy as npnp.random.seed(10)num_compras = np.random.poisson(lam=3.2, size=1000)  # conteo entero, valores discretosfig, ax = plt.subplots(figsize=(9, 4))valores, conteos = np.unique(num_compras, return_counts=True)ax.bar(valores, conteos, color='#0ea5e9', edgecolor='white', width=0.7)ax.set_title('Variable discreta: Número de compras por cliente al mes', fontweight='bold')ax.set_xlabel('Número de compras (valores enteros, sin decimales)')ax.set_ylabel('Número de clientes')ax.set_xticks(valores)ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()print(f"Valores únicos observados: {sorted(valores)}")print("Solo existen 'huecos' entre barras — no tiene sentido '2.5 compras'.")
```

---## 4. Por qué esta distinción determina la técnica de visualización correcta ⚠️Confundir continuo con discreto produce gráficos engañosos o, en el mejor de los casos, poco informativos:| Situación | Error común | Técnica correcta ||---|---|---|| Graficar datos discretos con bins arbitrarios de histograma | Un bin puede mezclar los valores 2 y 3, ocultando la verdadera distribución de conteos | Usar un bin por cada valor entero, o un gráfico de barras de frecuencias || Graficar datos continuos como barras separadas por valor exacto | Cientos de "barras" únicas, ilegible, no captura la forma de la distribución | Usar histograma con bins agrupados o KDE || Conectar con línea datos discretos no ordenados temporalmente | Sugiere una tendencia continua inexistente entre categorías de conteo | Usar barras o puntos, no líneas, si no hay orden temporal || Tratar una variable discreta con muchos valores posibles (ej. edad en años) como si necesitara un gráfico de barras por cada edad | Gráfico saturado e ilegible | Agrupar en bins (como si fuera casi-continua) — la frontera continua/discreta se vuelve difusa con rangos amplios |

```python
# ============================================================# Comparación lado a lado: binning correcto vs incorrecto para datos discretos# ============================================================import matplotlib.pyplot as pltimport numpy as npnp.random.seed(3)errores_por_dia = np.random.poisson(lam=4, size=500)  # discreta: número de errores/díafig, axes = plt.subplots(1, 2, figsize=(12, 4.5))fig.suptitle('Mismo dato discreto, dos formas de graficarlo', fontsize=12, fontweight='bold')# INCORRECTO: bins de histograma genérico que agrupan varios enterosaxes[0].hist(errores_por_dia, bins=6, color='#ef4444', edgecolor='white', alpha=0.85)axes[0].set_title('❌ Bins genéricos (agrupan enteros)', fontweight='bold', color='#991b1b')axes[0].set_xlabel('Errores por día')# CORRECTO: un bin por cada valor enterobins_correctos = np.arange(errores_por_dia.min(), errores_por_dia.max() + 2) - 0.5axes[1].hist(errores_por_dia, bins=bins_correctos, color='#16a34a', edgecolor='white', alpha=0.85)axes[1].set_title('✅ Un bin por valor entero', fontweight='bold', color='#14532d')axes[1].set_xlabel('Errores por día')axes[1].set_xticks(range(errores_por_dia.min(), errores_por_dia.max() + 1))for ax in axes:    ax.set_ylabel('Frecuencia (días)')    ax.spines['top'].set_visible(False)    ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()print("El histograma de la izquierda oculta que los valores son enteros exactos.")print("El de la derecha respeta la naturaleza discreta del dato.")
```

---## Resumen y Puntos Clave 🎯- Los **datos numéricos** representan cantidades medibles sobre las que tienen sentido operaciones aritméticas — a diferencia de los datos categóricos.- Los **datos continuos** pueden tomar infinitos valores decimales dentro de un rango (precio, temperatura, tiempo de respuesta).- Los **datos discretos** solo toman valores separados y contables, típicamente enteros (número de compras, número de errores, número de hijos).- Confundir ambos tipos produce gráficos engañosos: usar bins genéricos en datos discretos, o graficar decenas de barras individuales en datos continuos, son errores comunes.- Esta distinción es la base sobre la cual se eligen las técnicas del siguiente notebook: histogramas, gráficos de línea y scatter plots.
