## ¿Qué es "visualización geoespacial"? (sin asustarte) 🗺️

Suena técnico, pero es una idea simple: **poner tus datos sobre un mapa** en vez de sobre ejes X e Y normales.

Piensa en la app del clima 🌦️: no te muestra la temperatura en una tabla — te la muestra **coloreada sobre un mapa**. Eso es visualización geoespacial.

Este cuaderno responde 3 preguntas con analogías cotidianas:

1. **¿Cómo pongo puntos en un mapa?** (como clavar chinches en un mapa de corcho)
2. **¿Cómo coloreo regiones enteras?** (como un mapa político de países)
3. **¿Cómo muestro dónde hay "más gente"?** (como el mapa de tráfico de Google Maps)

## 1. Mapas de Puntos: clavar chinches en un mapa de corcho 📍

Imagina un mapa de corcho en la pared de una oficina de ventas, con una **chinche de color** por cada tienda. Eso es un mapa de puntos.

Si además haces que las chinches más grandes representen tiendas con más ventas, tienes un **mapa de símbolos proporcionales** (bubble map) — la chinche "crece" según el valor.

| Lo que ves en el corcho | Lo que es en datos |
|---|---|
| Posición de la chinche | Latitud y longitud |
| Tamaño de la chinche | Una variable numérica (ventas, población) |
| Color de la chinche | Otra variable (categoría o intensidad) |

````python
# Ejemplo simple: 5 tiendas en un "mapa de corcho" digital

import matplotlib.pyplot as plt

tiendas = {
    'Tienda Centro':  {'lat': 4.60, 'lon': -74.08, 'ventas': 120},
    'Tienda Norte':   {'lat': 4.75, 'lon': -74.03, 'ventas': 80},
    'Tienda Sur':     {'lat': 4.55, 'lon': -74.15, 'ventas': 45},
    'Tienda Oeste':   {'lat': 4.65, 'lon': -74.20, 'ventas': 95},
    'Tienda Este':    {'lat': 4.62, 'lon': -73.98, 'ventas': 60},
}

fig, ax = plt.subplots(figsize=(7, 5))
for nombre, d in tiendas.items():
    ax.scatter(d['lon'], d['lat'], s=d['ventas'] * 4, color='#f59e0b',
               alpha=0.7, edgecolors='#78350f')
    ax.annotate(nombre, (d['lon'], d['lat']), xytext=(5, 5), textcoords='offset points', fontsize=8)

ax.set_title('🗺️ "Mapa de corcho" digital: chinches = tiendas, tamaño = ventas')
ax.set_xlabel('Longitud'); ax.set_ylabel('Latitud')
plt.tight_layout()
plt.show()

print("🎯 Mientras más grande la 'chinche', más ventas tiene esa tienda.")
````

## 2. Mapas de Coropletas: el mapa político del salón de clases 🎨

¿Recuerdas el mapa político de un país, donde cada departamento tiene un color distinto? Un **mapa de coropletas** hace lo mismo pero el color representa un **dato**, no solo una división administrativa.

Por ejemplo: entre más oscuro el azul de un departamento, más alta su tasa de algo (ventas, desempleo, alfabetización).

> ⚠️ **Ojo con la trampa:** si coloreas por el TOTAL en vez de por una TASA (por ejemplo, casos totales en vez de casos por cada 100,000 habitantes), el departamento más grande o más poblado siempre "ganará" aunque no sea realmente el peor. Siempre normaliza antes de colorear.

````python
# Coropletas simplificado: cada "departamento" es una casilla coloreada

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

datos = {'Depto A': 85, 'Depto B': 60, 'Depto C': 45, 'Depto D': 95, 'Depto E': 30}
nombres = list(datos.keys())
valores = np.array(list(datos.values()))
cmap = plt.cm.Blues
norm = plt.Normalize(valores.min(), valores.max())

fig, ax = plt.subplots(figsize=(9, 2.2))
for i, (nombre, valor) in enumerate(zip(nombres, valores)):
    color = cmap(norm(valor))
    ax.add_patch(mpatches.Rectangle((i, 0), 0.9, 0.9, facecolor=color, edgecolor='white', linewidth=2))
    ax.text(i + 0.45, 0.45, f'{nombre}\n{valor}%', ha='center', va='center', fontsize=8,
            color='white' if valor > 60 else '#0f172a', fontweight='bold')

ax.set_xlim(0, len(nombres)); ax.set_ylim(0, 1); ax.axis('off')
ax.set_title('🎨 Mapa político simplificado: más oscuro = valor más alto', fontsize=10)
plt.tight_layout()
plt.show()

print("💡 En un mapa real, esas casillas serían las formas reales de los departamentos.")
````

## 3. Mapas de Calor: el mapa de tráfico de Google Maps 🔥

Cuando abres Google Maps y ves calles en rojo (mucho tráfico), amarillo (medio) o verde (fluido), estás viendo un **mapa de calor geográfico**. No te muestra cada carro individual — te muestra **dónde hay más concentración**.

Lo mismo aplica a datos: si tienes miles de puntos (accidentes, clientes, reportes), un mapa de calor te muestra las "zonas calientes" sin abrumarte con cada punto individual.

````python
# Mapa de calor simplificado: "zonas calientes" de reportes

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(3)
# Dos "zonas calientes" con muchos reportes concentrados
lat = np.concatenate([np.random.normal(4.6, 0.05, 300), np.random.normal(4.75, 0.04, 200)])
lon = np.concatenate([np.random.normal(-74.08, 0.05, 300), np.random.normal(-74.03, 0.04, 200)])

fig, ax = plt.subplots(figsize=(7, 5.5))
hb = ax.hexbin(lon, lat, gridsize=25, cmap='YlOrRd', mincnt=1)
plt.colorbar(hb, ax=ax, label='Cantidad de reportes')
ax.set_title('🔥 "Google Maps del tráfico": rojo = más concentración de reportes')
ax.set_xlabel('Longitud'); ax.set_ylabel('Latitud')
plt.tight_layout()
plt.show()

print("🎯 No ves cada reporte individual — ves DÓNDE se concentran, como el tráfico en tiempo real.")
````

## 4. Mapa Mental: ¿Cuál mapa uso? 🧭

```
¿Tienes ubicaciones exactas de eventos individuales?
    → SÍ, y quieres ver cada una      → Mapa de puntos / burbujas
    → SÍ, pero son miles y se amontonan → Mapa de calor (heatmap)

¿Tienes un valor agregado por región (país, departamento)?
    → SÍ → Mapa de coropletas (¡recuerda normalizar!)

¿Necesitas que alguien haga zoom, clic y explore el mapa?
    → SÍ → Usa una herramienta interactiva: Folium, Plotly, o Power BI/Tableau
    → NO, es solo para un reporte/imagen → matplotlib puro es suficiente
```

> 🎓 **Consejo:** Si solo aprendes UNA cosa de este cuaderno, que sea esta: un mapa de coropletas SIN normalizar (sin dividir por población o área) casi siempre miente. Siempre pregunta "¿esto es un total o una tasa?".
