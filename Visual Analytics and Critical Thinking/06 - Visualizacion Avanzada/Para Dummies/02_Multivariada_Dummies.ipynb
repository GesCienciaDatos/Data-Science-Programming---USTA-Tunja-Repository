## ¿Qué es "visualización multivariada"? (sin asustarte) 🕸️

Un gráfico normal compara 2 cosas: X contra Y. Pero, ¿qué pasa cuando quieres comparar 4 o 5 cosas a la vez? Por ejemplo, describir a un candidato a un empleo por: edad, experiencia, salario esperado, y nivel de estudios — ¡todo al mismo tiempo!

Eso es "multivariado": **muchas variables, un solo vistazo**. Este cuaderno usa una analogía: describir personas con una ficha de "cualidades" tipo videojuego (como las estadísticas de un personaje: fuerza, velocidad, resistencia...).

---
## 1. El Reto: más de 2 cosas a la vez 🎮

Si describes un personaje de videojuego SOLO por su fuerza y velocidad, te falta información: ¿qué tal su resistencia, su magia, su defensa? Necesitas ver **todas las estadísticas juntas** para comparar personajes de verdad.

Con datos pasa lo mismo: comparar clientes solo por "edad vs. ingreso" deja fuera su gasto, antigüedad y satisfacción. Necesitamos formas de ver TODO junto.

````python
# Un dataset sencillo: "ficha de estadísticas" de 6 clientes

import pandas as pd

clientes = pd.DataFrame({
    'cliente':      ['Ana', 'Beto', 'Caro', 'Dani', 'Eli', 'Fer'],
    'edad':         [25, 45, 34, 60, 29, 51],
    'ingreso':      [3.5, 9.0, 5.2, 7.8, 2.9, 8.5],   # millones de pesos
    'gasto':        [1.2, 3.8, 2.1, 2.5, 1.0, 3.2],   # millones de pesos
    'satisfaccion': [7, 9, 6, 8, 5, 9],                # 1-10
})
clientes
````

---
## 2. Pair Plot: la "cuadrícula de comparaciones" 🔲

Un **pair plot** es como poner una tabla de comparación donde cada casilla compara dos estadísticas del personaje. Si tienes 4 estadísticas, obtienes una cuadrícula de 4×4 con todos los "duelos" posibles entre variables — todo de una vez, sin tener que hacer 6 gráficos separados a mano.

````python
import matplotlib.pyplot as plt

try:
    import seaborn as sns
    vars_num = ['edad', 'ingreso', 'gasto', 'satisfaccion']
    g = sns.pairplot(clientes, vars=vars_num, height=1.6,
                      plot_kws={'color': '#f59e0b', 's': 50})
    g.fig.suptitle('🔲 Pair Plot: todas las comparaciones a la vez', y=1.02)
    plt.show()
except ImportError:
    print("ℹ️  seaborn no está instalado. Instalar con: pip install seaborn")

print("🎯 Cada casilla es un 'duelo' entre dos estadísticas — igual que comparar dos")
print("   atributos de un personaje de videojuego al mismo tiempo.")
````

---
## 3. Matriz de Correlación: "¿estas dos estadísticas van de la mano?" 🌡️

En vez de mirar 16 gráficos, a veces solo quieres un **resumen numérico**: ¿el ingreso y el gasto suben juntos? ¿la edad y la satisfacción no tienen relación? Eso es lo que muestra la matriz de correlación — un número entre -1 y 1 por cada par de estadísticas, coloreado como un semáforo.

````python
vars_num = ['edad', 'ingreso', 'gasto', 'satisfaccion']
corr = clientes[vars_num].corr()
print(corr.round(2))

try:
    import seaborn as sns
    fig, ax = plt.subplots(figsize=(5, 4))
    sns.heatmap(corr, annot=True, cmap='RdBu_r', center=0, vmin=-1, vmax=1, ax=ax)
    ax.set_title('🌡️ ¿Qué tan "juntas" se mueven dos estadísticas?')
    plt.tight_layout()
    plt.show()
except ImportError:
    print("ℹ️  seaborn no está instalado.")

print("\n🎯 Cerca de +1 = suben juntas. Cerca de -1 = una sube cuando la otra baja. Cerca de 0 = no hay relación clara.")
````

---
## 4. Coordenadas Paralelas: la "ficha de personaje" en líneas 📶

Piensa en la ficha de un videojuego de rol, donde cada estadística (fuerza, velocidad, magia...) tiene su propia barrita vertical, y el "perfil" del personaje es la línea que conecta todas sus barritas. Eso, exactamente, es un gráfico de **coordenadas paralelas**: un eje vertical por variable, una línea por persona.

````python
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

vars_num = ['edad', 'ingreso', 'gasto', 'satisfaccion']
clientes_norm = clientes.copy()
for v in vars_num:
    vmin, vmax = clientes_norm[v].min(), clientes_norm[v].max()
    clientes_norm[v] = (clientes_norm[v] - vmin) / (vmax - vmin)

fig, ax = plt.subplots(figsize=(8, 4))
parallel_coordinates(clientes_norm[vars_num + ['cliente']], class_column='cliente', ax=ax)
ax.set_title('📶 "Ficha de personaje": una línea = un cliente, un eje = una estadística')
ax.set_ylabel('Valor normalizado (0-1)')
plt.tight_layout()
plt.show()

print("🎯 Si dos líneas siguen casi el mismo camino, esos dos clientes se parecen mucho entre sí.")
````

## 5. Mapa Mental del Módulo Multivariado 🗺️

```
VISUALIZACIÓN MULTIVARIADA (más de 2 variables a la vez)
│
├── PAIR PLOT → cuadrícula de TODOS los duelos 2 a 2
│   sns.pairplot(df)
│
├── MATRIZ DE CORRELACIÓN → un número (-1 a 1) por cada duelo
│   df.corr() + sns.heatmap()
│
├── COORDENADAS PARALELAS → una línea = una persona/fila completa
│   parallel_coordinates(df, class_column=...)
│
└── BUBBLE CHART → hasta 4-5 variables en UN solo gráfico
    posición (X,Y) + tamaño + color
```

> 🎓 **Consejo:** Empieza siempre con un pair plot para explorar, usa la matriz de correlación para confirmar números, y usa coordenadas paralelas o un bubble chart cuando ya sepas qué historia quieres contar.
