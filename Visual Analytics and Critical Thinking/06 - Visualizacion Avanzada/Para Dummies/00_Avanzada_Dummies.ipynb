## 1. Dashboards: ¿qué es y por qué importa? 📊

Un **dashboard** es una pantalla única que muestra varios gráficos al mismo tiempo, como el tablero de un carro: velocidad, combustible, temperatura... de un solo vistazo.

### Los 3 niveles de dashboards (de más fácil a más avanzado):

| Nivel | Herramienta | Cuándo usarlo |
|---|---|---|
| 🟢 **Fácil** | Power BI / Tableau | Para presentar a directivos y no técnicos |
| 🟡 **Medio** | Streamlit (Python) | Para hacer demos rápidas o apps internas |
| 🔴 **Avanzado** | Plotly Dash (Python) | Para apps web de producción, personalizables al 100% |

### ¿Cuándo usar Streamlit vs Power BI?

```
¿Tu audiencia es técnica (puede instalar Python)?
    → SÍ → Streamlit o Dash
    → NO → Power BI (es solo un link en el navegador)

¿Necesitas modelos de Machine Learning integrados?
    → SÍ → Python (Streamlit/Dash)
    → NO → Power BI es suficiente y más rápido de construir

¿Lo usará toda la empresa sin instalar nada?
    → SÍ → Power BI Service (en el navegador)
    → Para tu equipo técnico → Streamlit Cloud (gratis)
```

## 2. Interactividad: de estático a interactivo con Plotly 🎛️

La diferencia entre un gráfico **estático** (PNG, PDF) y uno **interactivo** (HTML, web):

| Característica | Estático | Interactivo |
|---|---|---|
| Hover (pasar el mouse) | ❌ No | ✅ Muestra el valor exacto |
| Zoom | ❌ No | ✅ Acercarse a áreas específicas |
| Filtros | ❌ No | ✅ Hacer clic en la leyenda |
| Exportar | Solo imagen | HTML, imagen, JSON |
| Uso | Informe PDF | Dashboard web, Jupyter, Power BI (visual Python) |

````python
# De matplotlib (estático) a plotly (interactivo) — mismo dato

import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
import numpy as np

np.random.seed(42)
fechas = pd.date_range('2023-01-01', '2024-12-31', freq='W')
ventas = np.abs(50000 + np.cumsum(np.random.randn(len(fechas)) * 2000))
df = pd.DataFrame({'Fecha': fechas, 'Ventas': ventas})

# ❌ Matplotlib: estático
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(df['Fecha'], df['Ventas'], color='#3b82f6', linewidth=1.5)
ax.fill_between(df['Fecha'], df['Ventas'], alpha=0.1, color='#3b82f6')
ax.set_title('❌ Matplotlib (estático): no puedes hacer zoom ni hover', fontweight='bold')
ax.set_ylabel('Ventas ($)')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()

# ✅ Plotly: interactivo
fig_px = px.line(
    df, x='Fecha', y='Ventas',
    title='✅ Plotly (interactivo): haz zoom, hover, y arrastra el tiempo',
    template='plotly_white',
    color_discrete_sequence=['#3b82f6']
)
fig_px.update_traces(fill='tozeroy', fillcolor='rgba(59,130,246,0.1)')
fig_px.update_layout(height=350)
fig_px.show()

print("\n🎯 Prueba en el gráfico interactivo:")
print("   • Pasa el mouse → muestra el valor exacto")
print("   • Haz clic y arrastra → zoom en esa zona")
print("   • Doble clic → vuelve al zoom original")
print("   • Los íconos de arriba a la derecha → exportar, resetear, etc.")
print("\n📊 En Power BI: TODOS los gráficos nativos son interactivos automáticamente.")
print("   No necesitas código — se hace con el GUI.")
print("\n📈 En Tableau: igual, interactividad incluida por defecto.")
````

## 3. Series Temporales: visualizar datos que cambian con el tiempo ⏱️

**Serie temporal** = cualquier conjunto de datos donde el tiempo es el eje X.

Ejemplos del mundo real:
- Ventas diarias/mensuales/anuales
- Temperatura a lo largo del día
- Precio de acciones en bolsa
- Seguidores en Instagram semana a semana

### Los 3 patrones que siempre buscar en una serie temporal:

| Patrón | ¿Qué es? | Ejemplo |
|---|---|---|
| **Tendencia** | La dirección general (↗ sube, ↘ baja) | Las ventas crecen 10% cada año |
| **Estacionalidad** | Ciclos que se repiten regularmente | Ventas siempre más altas en diciembre |
| **Anomalías** | Picos o caídas inesperados | Caída en COVID-19, Black Friday |

📊 **En Power BI:** Usa "Inteligencia de tiempo" (Time Intelligence) en DAX para comparaciones YoY, acumulados, etc.  
📈 **En Tableau:** Arrastra la fecha a Columns y cambia la granularidad (año → mes → semana → día)

## 4. Mapa Mental del Módulo Avanzado 🗺️

```
VISUALIZACIÓN AVANZADA
│
├── DASHBOARDS
│   ├── Power BI → Sin código, para presentaciones
│   ├── Streamlit → Python simple, para demos
│   └── Plotly Dash → Python avanzado, para producción
│
├── MAPAS
│   ├── Folium → Mapas estáticos e interactivos en Python
│   ├── Plotly Maps → Mapas de burbujas y coropleticos
│   ├── Power BI Maps → Visual nativo (Mapa de burbujas, Mapa de Azure)
│   └── Tableau Maps → Automático cuando detecta Lat/Lon o país/ciudad
│
└── SERIES TEMPORALES
    ├── Buscar: Tendencia + Estacionalidad + Anomalías
    ├── Python: px.line() + statsmodels.seasonal_decompose()
    ├── Power BI: Línea + Time Intelligence DAX
    └── Tableau: Fecha en Columns + granularidad ajustable
```
