## ¿Qué es "visualización temporal"? (sin asustarte) ⏱️

Es simplemente graficar **datos que cambian con el tiempo**: tu peso semana a semana, tus calificaciones semestre a semestre, o cuántas personas visitan una tienda cada día.

Pero hay dos "sabores" distintos, y confundirlos es el error más común:

1. **Serie de tiempo (continua)**: un valor por cada intervalo regular — como marcar tu peso cada lunes en la báscula.
2. **Datos a nivel de eventos (irregular)**: un registro cada vez que algo PASA, sin horario fijo — como anotar cada vez que suena el timbre de la puerta.

---
## 1. Series de Tiempo: la báscula del lunes ⚖️

Si te pesas TODOS los lunes a la misma hora, tienes una **serie de tiempo regular**: un dato, siempre al mismo intervalo. Eso es justo lo que hace `pandas.date_range()`: crea ese calendario de "lunes" automáticamente, sin que tengas que escribir cada fecha a mano.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# "Pésate" cada semana durante 6 meses (dato inventado)
np.random.seed(1)
semanas = pd.date_range('2024-01-01', periods=26, freq='W')
peso = 80 - np.linspace(0, 3, 26) + np.random.normal(0, 0.3, 26)  # tendencia a bajar + ruido

serie_peso = pd.Series(peso, index=semanas)

fig, ax = plt.subplots(figsize=(9, 3.5))
ax.plot(serie_peso.index, serie_peso.values, marker='o', color='#f59e0b')
ax.set_title('⚖️ "Serie de tiempo": pesarse cada lunes, mismo intervalo siempre')
ax.set_ylabel('Peso (kg)')
plt.tight_layout()
plt.show()

print("🎯 Un punto por semana, siempre al mismo ritmo — así de simple es una serie de tiempo.")
````

---
## 2. Remuestreo y Media Móvil: de "cada lunes" a "cada mes" 🔁

**Remuestrear** (`resample`) es como agrupar tus pesajes semanales para sacar el promedio del mes — cambias el "ritmo" de tus datos sin perder la esencia.

La **media móvil** es distinto: en vez de agrupar por mes, vas promediando "las últimas 4 semanas" a medida que avanzas — como el promedio de tus últimas 4 calificaciones, que se actualiza cada vez que hay una nueva nota.

````python
# De semanal a mensual (resample) + promedio de las últimas 4 semanas (rolling)

peso_mensual = serie_peso.resample('ME').mean()
peso_media_movil = serie_peso.rolling(4).mean()

fig, ax = plt.subplots(figsize=(9, 3.5))
ax.plot(serie_peso.index, serie_peso.values, color='#cbd5e1', label='Cada semana (original)')
ax.plot(peso_media_movil.index, peso_media_movil.values, color='#dc2626', linewidth=2,
        label='Promedio de últimas 4 semanas')
ax.legend()
ax.set_title('🔁 Media móvil: promedio "que se va moviendo" con el tiempo')
plt.tight_layout()
plt.show()

print("Peso promedio por mes (resample):")
print(peso_mensual.round(2))
````

## 3. Datos a Nivel de Eventos: el timbre de la puerta 🔔

Ahora imagina que en vez de pesarte cada lunes, anotas la hora **cada vez que suena el timbre** de tu casa. Algunos días suena 5 veces, otros ninguna, y nunca a horas fijas. Eso es un **dato a nivel de evento**: irregular, sin ritmo fijo.

| Báscula del lunes (serie de tiempo) | Timbre de la puerta (evento) |
|---|---|
| Un dato SIEMPRE, mismo intervalo | Un dato SOLO cuando algo pasa |
| Se grafica con una línea continua | Se grafica con marcas en una línea de tiempo |
| "¿Cuál es la tendencia?" | "¿Cuándo y con qué frecuencia ocurre?" |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

np.random.seed(5)
# Cada vez que suena el timbre en una semana (horas al azar, sin ritmo fijo)
dia_base = pd.Timestamp('2024-06-03')
n_timbrazos = 18
minutos_azar = np.sort(np.random.randint(0, 7*24*60, n_timbrazos))
timbrazos = [dia_base + pd.Timedelta(minutes=int(m)) for m in minutos_azar]

fig, ax = plt.subplots(figsize=(9, 2.2))
ax.scatter(timbrazos, [0]*len(timbrazos), color='#f59e0b', s=60, zorder=3)
ax.xaxis.set_major_formatter(mdates.DateFormatter('%a %d'))
ax.set_yticks([])
ax.set_title('🔔 "Timbre de la puerta": eventos irregulares en la línea de tiempo')
plt.tight_layout()
plt.show()

# Convertir esos "timbrazos" en una serie de tiempo diaria (conteo por día)
timbrazos_por_dia = pd.Series(1, index=pd.DatetimeIndex(timbrazos)).resample('D').sum()
print("Timbrazos contados por día (¡ahora sí es una serie de tiempo regular!):")
print(timbrazos_por_dia)
````

## 4. Mapa Mental del Módulo Temporal 🗺️

```
VISUALIZACIÓN TEMPORAL
│
├── SERIE DE TIEMPO (como la báscula del lunes)
│   ├── pd.date_range() → crea el calendario automáticamente
│   ├── .resample() → cambia el ritmo (semanal → mensual)
│   └── .rolling() → promedio "que se mueve" (últimas N semanas)
│
└── DATOS A NIVEL DE EVENTO (como el timbre de la puerta)
    ├── Timestamps irregulares, sin ritmo fijo
    ├── Se grafican como línea de tiempo (marcas discretas)
    └── Se pueden CONVERTIR a serie de tiempo contando por día: resample('D').sum()
```

> 🎓 **Consejo:** Si tus datos tienen una hora exacta pero no un "ritmo" fijo (compras, clics, llamadas), probablemente son datos a nivel de evento — no fuerces un gráfico de línea continua sin antes agregarlos.
