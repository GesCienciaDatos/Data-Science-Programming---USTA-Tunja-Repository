## 1. Datos Geoespaciales: ¿Cuándo un dato "vive" en el espacio? 🌍

Un dato es **geoespacial** cuando tiene un componente de ubicación explícito: coordenadas (latitud/longitud), o una referencia a una unidad administrativa (país, departamento, municipio, código postal).

### Tipos de datos geoespaciales

| Tipo | Descripción | Ejemplo |
|---|---|---|
| **Puntual** | Una ubicación exacta (lat/lon) | Sucursal de una tienda, sismo, accidente |
| **Poligonal (área)** | Una región con forma geográfica | Departamento, país, barrio |
| **Lineal** | Una trayectoria o ruta | Río, carretera, ruta de entrega |
| **Raster** | Una grilla continua de valores | Temperatura satelital, elevación |

### Las 3 familias de mapas que veremos

| Familia | Pregunta que responde | Codificación visual |
|---|---|---|
| **Mapa de puntos / símbolos proporcionales** | ¿Dónde ocurren los eventos y qué tan grandes son? | Posición + tamaño/color del símbolo |
| **Mapa de coropletas** | ¿Cómo varía una variable agregada entre regiones? | Color de relleno del polígono |
| **Mapa de calor geográfico** | ¿Dónde se concentra la densidad de eventos? | Intensidad de color por densidad |

> ⚠️ **Cuidado:** Un mapa de coropletas mal usado (sin normalizar por área o población) puede engañar — un departamento grande en superficie no necesariamente tiene el valor más alto per cápita.

## 2. Mapas de Puntos y Símbolos Proporcionales 📍

El mapa de puntos ubica cada observación exactamente donde ocurrió. El **mapa de símbolos proporcionales** (bubble map) añade una segunda variable codificada en el **tamaño** (y opcionalmente el **color**) del símbolo.

Construimos ambos con `matplotlib` puro, sobre coordenadas de ciudades colombianas (sin dependencias externas de mapas).

```python
# ============================================================
# Mapa de puntos y mapa de símbolos proporcionales (bubble map)
# Solo con matplotlib + numpy — sin dependencias geoespaciales externas
# ============================================================

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Datos simulados: PIB per cápita y población por ciudad (Colombia)
ciudades = pd.DataFrame([
    {'ciudad': 'Bogotá',       'lat': 4.711,  'lon': -74.072, 'pib_percapita': 28500, 'poblacion': 7_743_955},
    {'ciudad': 'Medellín',     'lat': 6.244,  'lon': -75.581, 'pib_percapita': 22100, 'poblacion': 2_533_424},
    {'ciudad': 'Cali',         'lat': 3.451,  'lon': -76.532, 'pib_percapita': 18900, 'poblacion': 2_227_642},
    {'ciudad': 'Barranquilla', 'lat': 10.964, 'lon': -74.796, 'pib_percapita': 16800, 'poblacion': 1_232_766},
    {'ciudad': 'Cartagena',    'lat': 10.391, 'lon': -75.479, 'pib_percapita': 15200, 'poblacion': 1_013_389},
    {'ciudad': 'Bucaramanga',  'lat': 7.119,  'lon': -73.122, 'pib_percapita': 14800, 'poblacion': 583_959},
    {'ciudad': 'Manizales',    'lat': 5.067,  'lon': -75.514, 'pib_percapita': 13200, 'poblacion': 441_478},
    {'ciudad': 'Pereira',      'lat': 4.812,  'lon': -75.696, 'pib_percapita': 12900, 'poblacion': 476_000},
    {'ciudad': 'Cúcuta',       'lat': 7.893,  'lon': -72.503, 'pib_percapita': 10100, 'poblacion': 750_000},
    {'ciudad': 'Ibagué',       'lat': 4.438,  'lon': -75.232, 'pib_percapita': 9800,  'poblacion': 557_000},
])

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

# --- Panel 1: mapa de puntos simple ---
ax = axes[0]
ax.scatter(ciudades['lon'], ciudades['lat'], s=60, color='#1e3a8a', zorder=3)
for _, row in ciudades.iterrows():
    ax.annotate(row['ciudad'], (row['lon'], row['lat']), textcoords="offset points",
                xytext=(6, 4), fontsize=8, color='#334155')
ax.set_title('Mapa de Puntos: ubicación de ciudades', fontweight='bold')
ax.set_xlabel('Longitud'); ax.set_ylabel('Latitud')
ax.grid(alpha=0.2)

# --- Panel 2: mapa de símbolos proporcionales (bubble map) ---
ax = axes[1]
size = ciudades['poblacion'] / ciudades['poblacion'].max() * 1500 + 30
sc = ax.scatter(ciudades['lon'], ciudades['lat'], s=size,
                 c=ciudades['pib_percapita'], cmap='Blues',
                 edgecolors='#1e3a8a', linewidths=0.8, alpha=0.85, zorder=3)
for _, row in ciudades.iterrows():
    ax.annotate(row['ciudad'], (row['lon'], row['lat']), textcoords="offset points",
                xytext=(0, -14), fontsize=7.5, ha='center', color='#334155')
cbar = plt.colorbar(sc, ax=ax, shrink=0.8)
cbar.set_label('PIB per cápita (USD)')
ax.set_title('Símbolos Proporcionales: tamaño=población, color=PIB', fontweight='bold')
ax.set_xlabel('Longitud'); ax.set_ylabel('Latitud')
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()

print("📊 En Power BI: Visual 'Mapa de burbujas' — campos Latitud/Longitud, Tamaño, Color.")
print("📈 En Tableau: Show Me → Símbolos proporcionales — arrastra la medida a 'Size'.")
```

## 3. Mapas de Coropletas (Choropleth) 🎨

Un **mapa de coropletas** colorea regiones (polígonos) según el valor de una variable agregada. Es la forma más común de mostrar "cómo varía X entre departamentos/países".

En producción se construyen con **GeoPandas** (que maneja geometrías reales tipo *shapefile*/GeoJSON) o con `plotly.express.choropleth`. Aquí mostramos el patrón con GeoPandas (si está disponible) y una alternativa **100% offline** con `matplotlib` que aproxima cada departamento como una celda de una grilla — útil para entender el concepto sin depender de archivos geográficos externos.

```python
# ============================================================
# Mapa de coropletas
# Intenta usar geopandas (datos reales); si no está disponible,
# cae a una aproximación 100% matplotlib (grilla coloreada)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Datos simulados: tasa de alfabetización digital (%) por departamento
departamentos = {
    'Cundinamarca': 82, 'Antioquia': 78, 'Valle del Cauca': 75,
    'Atlántico': 70, 'Santander': 73, 'Bolívar': 62,
    'Boyacá': 68, 'Caldas': 71, 'Nariño': 58, 'Norte de Santander': 60,
}

try:
    import geopandas as gpd
    print("✅ geopandas disponible: en un proyecto real cargarías un GeoJSON con")
    print("   gpd.read_file('colombia_departamentos.geojson') y unirías (merge) con tus datos.")
    print("   gdf.plot(column='valor', cmap='Blues', legend=True)  # <- una línea genera el mapa")
    GEOPANDAS_OK = True
except ImportError:
    print("ℹ️  geopandas no está instalado (opcional). Usando aproximación con matplotlib puro.")
    GEOPANDAS_OK = False

# --- Aproximación offline: cada departamento = una celda en una grilla ---
nombres = list(departamentos.keys())
valores = np.array(list(departamentos.values()))
ncols = 5
nrows = int(np.ceil(len(nombres) / ncols))

cmap = plt.cm.Blues
norm = plt.Normalize(valores.min(), valores.max())

fig, ax = plt.subplots(figsize=(10, 4.5))
for i, (nombre, valor) in enumerate(zip(nombres, valores)):
    row, col = divmod(i, ncols)
    color = cmap(norm(valor))
    rect = mpatches.Rectangle((col, nrows - row - 1), 0.95, 0.95,
                               facecolor=color, edgecolor='white', linewidth=2)
    ax.add_patch(rect)
    ax.text(col + 0.475, nrows - row - 0.55, nombre, ha='center', va='center',
            fontsize=8, fontweight='bold', color='#0f172a' if valor < 75 else 'white')
    ax.text(col + 0.475, nrows - row - 0.75, f'{valor}%', ha='center', va='center',
            fontsize=8, color='#0f172a' if valor < 75 else 'white')

ax.set_xlim(0, ncols); ax.set_ylim(0, nrows)
ax.set_aspect('equal'); ax.axis('off')
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
plt.colorbar(sm, ax=ax, shrink=0.7, label='Alfabetización digital (%)')
ax.set_title('Mapa de Coropletas (aproximado): Alfabetización Digital por Departamento',
              fontweight='bold', fontsize=12)
plt.tight_layout()
plt.show()

print("\n📊 En Power BI: Visual 'Mapa de formas' (Shape Map) — requiere un archivo TopoJSON.")
print("📈 En Tableau: Show Me → Mapa relleno (Filled Map) — detecta geografías automáticamente.")
```

## 4. Mapas de Calor Geográficos (Heatmaps) 🔥

Un **mapa de calor geográfico** muestra la **densidad** de puntos en el espacio — útil cuando tienes miles de eventos (accidentes, delitos, clientes) y quieres ver dónde se concentran, más que cada punto individual.

Se construye agregando puntos en una grilla 2D (equivalente espacial de un histograma) y coloreando por conteo/densidad.

```python
# ============================================================
# Mapa de calor geográfico (densidad espacial) con matplotlib
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(7)

# Simulamos 2,000 eventos (p.ej. reportes de un servicio) concentrados
# alrededor de tres "focos" urbanos con dispersión distinta
focos = [(4.71, -74.07, 700), (6.24, -75.58, 600), (3.45, -76.53, 400)]
lat_ev, lon_ev = [], []
for lat0, lon0, n in focos:
    lat_ev.extend(np.random.normal(lat0, 0.18, n))
    lon_ev.extend(np.random.normal(lon0, 0.18, n))
lat_ev, lon_ev = np.array(lat_ev), np.array(lon_ev)

fig, ax = plt.subplots(figsize=(8, 6))
hb = ax.hexbin(lon_ev, lat_ev, gridsize=35, cmap='inferno', mincnt=1)
cb = plt.colorbar(hb, ax=ax, shrink=0.8)
cb.set_label('Densidad de eventos')
ax.set_title('Mapa de Calor Geográfico: densidad de eventos reportados', fontweight='bold')
ax.set_xlabel('Longitud'); ax.set_ylabel('Latitud')
plt.tight_layout()
plt.show()

print("💡 hexbin agrupa puntos en celdas hexagonales — evita el 'sobreplotting' de miles de puntos.")
print("\n📊 En Power BI: Actualmente no tiene heatmap geográfico nativo; se aproxima con")
print("   'Mapa de burbujas' + transparencia alta, o un visual personalizado de AppSource.")
print("📈 En Tableau: 'Densidad' está disponible como Mark Type dentro de un mapa (Marks → Density).")
```

## 5. Mapas Interactivos: Folium, Plotly y su lugar en Dashboards 🌐

Los mapas anteriores son estáticos (una imagen). Para explorar interactivamente (zoom, hover, popups con información) se usan librerías como **Folium** (basada en Leaflet.js) o `plotly.express.scatter_mapbox` / `choropleth_mapbox`.

Ambas son **opcionales**: si no están instaladas, el código siguiente lo detecta y explica la alternativa sin fallar.

```python
# ============================================================
# Mapas interactivos — manejo con try/except por si no están instaladas
# ============================================================

import pandas as pd

ciudades_df = ciudades.copy()  # reutiliza el DataFrame de la sección 2

# --- Folium (opcional) ---
try:
    import folium
    mapa = folium.Map(location=[4.5, -74.0], zoom_start=6, tiles='CartoDB positron')
    for _, row in ciudades_df.iterrows():
        folium.CircleMarker(
            location=[row['lat'], row['lon']],
            radius=min(row['poblacion'] / 300_000, 25),
            color='#1e3a8a', fill=True, fill_opacity=0.7,
            popup=f"{row['ciudad']}: ${row['pib_percapita']:,}",
            tooltip=row['ciudad']
        ).add_to(mapa)
    mapa.save('mapa_geoespacial_colombia.html')
    print("✅ folium disponible — mapa interactivo guardado en 'mapa_geoespacial_colombia.html'")
except ImportError:
    print("ℹ️  folium no está instalado (opcional). Instalar con: pip install folium")

# --- Plotly Express (más común, no requiere token de mapbox) ---
try:
    import plotly.express as px
    fig_map = px.scatter_mapbox(
        ciudades_df, lat='lat', lon='lon',
        color='pib_percapita', size='poblacion',
        hover_name='ciudad', size_max=40,
        color_continuous_scale='Blues', mapbox_style='carto-positron',
        title='PIB per cápita por ciudad — mapa interactivo (Plotly)',
        zoom=4.3, center={'lat': 4.5, 'lon': -74.0}
    )
    fig_map.update_layout(height=480, template='plotly_white')
    fig_map.show()
except ImportError:
    print("ℹ️  plotly no está instalado (opcional). Instalar con: pip install plotly")

print("\n🧩 Nota sobre dashboards: estos mapas interactivos suelen ser UN panel dentro de un")
print("   dashboard más amplio (p.ej. construido con Plotly Dash o Streamlit), combinado con")
print("   KPIs, filtros y otros gráficos. Ese ensamblaje se estudia en detalle en el módulo de")
print("   Dashboards; aquí el foco es la construcción del mapa en sí mismo.")
```

## 6. Comparativa: Python vs Power BI vs Tableau para Mapas 🔺

| Capacidad | Python (matplotlib/Folium/Plotly) | Power BI | Tableau |
|---|---|---|---|
| **Mapa de puntos/burbujas** | `scatter` / `scatter_mapbox` | Mapa de burbujas | Symbol map |
| **Coropletas** | GeoPandas + GeoJSON | Shape Map (TopoJSON) | Filled Map (geo-roles automáticos) |
| **Mapa de calor geográfico** | `hexbin` / Folium HeatMap | Visual personalizado | Density (Marks) |
| **Interactividad (zoom, popup)** | Folium / Plotly | Nativa | Nativa |
| **Geocodificación automática** | Requiere librería (`geopy`) | Automática (Bing Maps) | Automática |
| **Costo** | Gratuito | $10-20 USD/usuario/mes | $70+ USD/usuario/mes |

## Resumen y Puntos Clave 🎯

- Un dato es **geoespacial** cuando tiene ubicación explícita: puntual, poligonal, lineal o raster.
- El **mapa de puntos/símbolos proporcionales** codifica ubicación + magnitud (tamaño/color) por observación individual.
- El **mapa de coropletas** colorea regiones agregadas — cuidado con no normalizar variables por área o población.
- El **mapa de calor geográfico** muestra densidad, no valores individuales — ideal ante miles de puntos.
- **GeoPandas/Folium/Plotly** añaden interactividad y geometrías reales; son opcionales frente a una aproximación con `matplotlib` puro.
- Power BI (Shape Map) y Tableau (Filled Map) ofrecen coropletas sin código, pero con menor flexibilidad que Python.
