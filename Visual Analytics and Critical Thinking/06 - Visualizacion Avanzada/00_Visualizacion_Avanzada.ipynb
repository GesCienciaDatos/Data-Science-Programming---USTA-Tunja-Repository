## 1. Dashboards Interactivos con Plotly: Subplots y Widgets 🎛️

Un **dashboard** es una colección de visualizaciones coordinadas que responde múltiples preguntas simultáneamente desde un mismo panel. Los principios de un buen dashboard:

| Principio | Descripción |
|---|---|
| **Una pantalla** | Todo el contenido crítico visible sin scroll |
| **KPIs primero** | Los números más importantes arriba-izquierda |
| **Jerarquía visual** | Lo más importante = más grande y prominente |
| **Paleta consistente** | Mismos colores para las mismas categorías en todos los gráficos |
| **Interactividad con propósito** | Cada filtro/drill-down responde una pregunta específica |
| **Carga rápida** | Máximo 3-5 segundos para cargar — agregar datos solo bajo demanda |

````python
# ============================================================
# Dashboard interactivo con Plotly Subplots
# Simula un dashboard de ventas real
# ============================================================

import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

np.random.seed(42)

# Datos simulados de ventas
fechas = pd.date_range('2023-01-01', '2024-12-31', freq='W')
ventas = 50000 + np.cumsum(np.random.randn(len(fechas)) * 2000) + \
         5000 * np.sin(np.linspace(0, 4*np.pi, len(fechas)))  # Estacionalidad
ventas = np.abs(ventas)

categorias = ['Electrónica', 'Ropa', 'Hogar', 'Alimentos', 'Deportes']
ventas_cat = [45200, 32100, 28900, 38700, 21300]
regiones = ['Norte', 'Sur', 'Este', 'Oeste']
ventas_reg = [35000, 42000, 28000, 38000]

# Dashboard con 4 paneles
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=[
        '📈 Tendencia de Ventas Semanales (2023-2024)',
        '🏆 Top Categorías de Ventas',
        '🗺️ Ventas por Región',
        '📊 Distribución de Tickets de Venta'
    ],
    specs=[[{'type': 'scatter'}, {'type': 'bar'}],
           [{'type': 'pie'}, {'type': 'histogram'}]],
    vertical_spacing=0.12,
    horizontal_spacing=0.1
)

# Panel 1: Serie temporal
fig.add_trace(go.Scatter(
    x=fechas, y=ventas,
    mode='lines', name='Ventas',
    line=dict(color='#3b82f6', width=1.5),
    fill='tozeroy', fillcolor='rgba(59,130,246,0.1)'
), row=1, col=1)

# Línea de media móvil
ventas_ma = pd.Series(ventas).rolling(8).mean()
fig.add_trace(go.Scatter(
    x=fechas, y=ventas_ma,
    mode='lines', name='Media móvil 8 sem',
    line=dict(color='#dc2626', width=2, dash='dot')
), row=1, col=1)

# Panel 2: Barras
fig.add_trace(go.Bar(
    x=ventas_cat, y=categorias,
    orientation='h',
    marker_color=['#3b82f6', '#10b981', '#f59e0b', '#ef4444', '#8b5cf6'],
    name='Ventas por Categoría',
    text=[f'${v/1000:.0f}k' for v in ventas_cat],
    textposition='inside'
), row=1, col=2)

# Panel 3: Donut
fig.add_trace(go.Pie(
    labels=regiones,
    values=ventas_reg,
    hole=0.5,
    name='Regiones',
    marker_colors=['#3b82f6', '#10b981', '#f59e0b', '#ef4444']
), row=2, col=1)

# Panel 4: Histograma
tickets = np.random.lognormal(4.5, 0.8, 500)
fig.add_trace(go.Histogram(
    x=tickets, nbinsx=30,
    name='Tickets',
    marker_color='#10b981',
    opacity=0.8
), row=2, col=2)

fig.update_layout(
    title_text='🏪 Dashboard de Ventas — Empresa Ejemplo (2023-2024)',
    title_font_size=16,
    title_font_family='system-ui',
    height=700,
    template='plotly_white',
    showlegend=False,
    paper_bgcolor='#f8fafc',
    plot_bgcolor='white'
)

fig.show()

print("\n💡 Este dashboard se puede convertir en una app con Dash o Streamlit.")
print("📊 En Power BI: Cada panel = un visual diferente en el lienzo del informe.")
print("   Usar 'Segmentaciones' para filtros interactivos entre visuales.")
print("\n📈 En Tableau: Crear hojas separadas y luego ensamblarlas en un 'Dashboard'.")
print("   Acciones de dashboard = equivalente a filtros cruzados de Power BI.")
````

## 2. Streamlit: Dashboards con Python Puro 🌊

**Streamlit** convierte scripts de Python en aplicaciones web en minutos. Es la herramienta más popular para prototipado rápido de dashboards analíticos.

### Comparativa de frameworks de dashboards en Python:

| Framework | Curva de aprendizaje | Interactividad | Despliegue | Mejor para |
|---|---|---|---|---|
| **Streamlit** | Muy baja | Alta | Streamlit Cloud (gratis) | Prototipado rápido, ML apps |
| **Plotly Dash** | Media | Muy alta | Heroku, AWS, GCP | Dashboards de producción |
| **Panel** | Media | Alta | Bokeh Server | Dashboards científicos |
| **Gradio** | Muy baja | Media | Hugging Face Spaces | Demos de modelos ML |

### Equivalencias con herramientas BI:

| Característica | Streamlit (Python) | Power BI Service | Tableau Server |
|---|---|---|---|
| **Filtros** | `st.selectbox()`, `st.slider()` | Segmentaciones | Acciones de dashboard |
| **KPIs** | `st.metric()` | Tarjeta KPI | KPI Visual |
| **Tablas** | `st.dataframe()` | Tabla/Matriz | Cross-tab |
| **Gráficos** | `st.plotly_chart()` | Visuals nativos | Hojas de trabajo |
| **Alertas** | `st.warning()`, `st.success()` | Alertas de datos | No nativo |

````python
# ============================================================
# Código de Streamlit — para ejecutar FUERA de Jupyter
# Guarda este código como: dashboard_ventas.py
# Luego ejecuta: streamlit run dashboard_ventas.py
# ============================================================

streamlit_code = '''
# dashboard_ventas.py — Ejecutar con: streamlit run dashboard_ventas.py

import streamlit as st
import plotly.express as px
import pandas as pd
import numpy as np

# Configuración de la página
st.set_page_config(
    page_title="Dashboard de Ventas",
    page_icon="📊",
    layout="wide"
)

# Título
st.title("📊 Dashboard de Ventas Interactivo")
st.markdown("*Visual Analytics — Universidad Santo Tomás Tunja*")

# Datos simulados
np.random.seed(42)
fechas = pd.date_range(\'2023-01-01\', \'2024-12-31\', freq=\'W\')
df = pd.DataFrame({
    \'fecha\': fechas,
    \'ventas\': np.abs(50000 + np.cumsum(np.random.randn(len(fechas)) * 2000)),
    \'region\': np.random.choice([\'Norte\', \'Sur\', \'Este\', \'Oeste\'], len(fechas)),
    \'categoria\': np.random.choice([\'Electrónica\', \'Ropa\', \'Hogar\'], len(fechas))
})

# Sidebar: Filtros
with st.sidebar:
    st.header("🔧 Filtros")
    regiones_sel = st.multiselect("Región", df[\'region\'].unique(), default=df[\'region\'].unique())
    rango_fechas = st.date_input("Rango de fechas",
                                  [df[\'fecha\'].min(), df[\'fecha\'].max()])

# Filtrar datos
df_fil = df[df[\'region\'].isin(regiones_sel)]

# KPIs en la parte superior
col1, col2, col3, col4 = st.columns(4)
col1.metric("💰 Ventas Totales", f"${df_fil[\'ventas\'].sum()/1e6:.2f}M", "+12.3%")
col2.metric("📦 Transacciones", f"{len(df_fil):,}", "+8.1%")
col3.metric("🛒 Ticket Promedio", f"${df_fil[\'ventas\'].mean():.0f}", "-2.4%")
col4.metric("🗺️ Regiones Activas", str(df_fil[\'region\'].nunique()), "0%")

# Gráficos
col_left, col_right = st.columns([2, 1])

with col_left:
    fig_line = px.line(df_fil, x=\'fecha\', y=\'ventas\',
                       color=\'region\', title=\'Tendencia de Ventas por Región\',
                       template=\'plotly_white\')
    st.plotly_chart(fig_line, use_container_width=True)

with col_right:
    fig_bar = px.bar(df_fil.groupby(\'categoria\')[\'ventas\'].sum().reset_index(),
                     x=\'ventas\', y=\'categoria\', orientation=\'h\',
                     title=\'Ventas por Categoría\', template=\'plotly_white\')
    st.plotly_chart(fig_bar, use_container_width=True)

# Tabla de datos
st.subheader("📋 Datos Detallados")
st.dataframe(df_fil.head(100), use_container_width=True)
'''

# Guardar el código en un archivo
with open('dashboard_ventas.py', 'w', encoding='utf-8') as f:
    f.write(streamlit_code)

print("✅ Código de Streamlit guardado en 'dashboard_ventas.py'")
print("\n🚀 Para ejecutar:")
print("   1. Instalar Streamlit: pip install streamlit plotly")
print("   2. Ejecutar: streamlit run dashboard_ventas.py")
print("   3. El dashboard se abre automáticamente en http://localhost:8501")
print("\n🌐 Para desplegar gratis: https://streamlit.io/cloud")
````

## 3. Visualización Geoespacial con Folium y Plotly 🗺️

````python
# ============================================================
# Mapa interactivo de Colombia con Folium
# Visualización de datos por departamento
# ============================================================

# Nota: Requiere instalar folium: pip install folium
try:
    import folium
    from folium.plugins import HeatMap
    import json
    
    # Datos simulados: PIB per cápita por departamento (Colombia)
    ciudades_col = {
        'Bogotá': {'lat': 4.711, 'lon': -74.072, 'valor': 28500, 'poblacion': 7743955},
        'Medellín': {'lat': 6.244, 'lon': -75.581, 'valor': 22100, 'poblacion': 2533424},
        'Cali': {'lat': 3.451, 'lon': -76.532, 'valor': 18900, 'poblacion': 2227642},
        'Barranquilla': {'lat': 10.964, 'lon': -74.796, 'valor': 16800, 'poblacion': 1232766},
        'Cartagena': {'lat': 10.391, 'lon': -75.479, 'valor': 15200, 'poblacion': 1013389},
        'Bucaramanga': {'lat': 7.119, 'lon': -73.122, 'valor': 14800, 'poblacion': 583959},
        'Manizales': {'lat': 5.067, 'lon': -75.514, 'valor': 13200, 'poblacion': 441478},
        'Pereira': {'lat': 4.812, 'lon': -75.696, 'valor': 12900, 'poblacion': 476000},
        'Cúcuta': {'lat': 7.893, 'lon': -72.503, 'valor': 10100, 'poblacion': 750000},
        'Ibagué': {'lat': 4.438, 'lon': -75.232, 'valor': 9800, 'poblacion': 557000}
    }
    
    # Crear mapa centrado en Colombia
    mapa = folium.Map(location=[4.5, -74.0], zoom_start=6, 
                      tiles='CartoDB positron')
    
    # Agregar marcadores con información
    for ciudad, datos in ciudades_col.items():
        # Tamaño del círculo proporcional a la población
        radio = datos['poblacion'] / 100000
        color = '#1e3a8a' if datos['valor'] > 18000 else '#3b82f6' if datos['valor'] > 13000 else '#94a3b8'
        
        folium.CircleMarker(
            location=[datos['lat'], datos['lon']],
            radius=min(radio, 25),
            color=color,
            fill=True,
            fill_opacity=0.7,
            popup=folium.Popup(
                f"<b>{ciudad}</b><br>"
                f"PIB per cápita: ${datos['valor']:,}<br>"
                f"Población: {datos['poblacion']:,}",
                max_width=200
            ),
            tooltip=f"{ciudad}: ${datos['valor']:,}"
        ).add_to(mapa)
        
        folium.Marker(
            location=[datos['lat'] + 0.15, datos['lon']],
            icon=folium.DivIcon(
                html=f'<div style="font-size:9px;font-weight:bold;color:#1e3a8a">{ciudad}</div>'
            )
        ).add_to(mapa)
    
    # Heatmap de valores
    heat_data = [[datos['lat'], datos['lon'], datos['valor']/1000] 
                 for datos in ciudades_col.values()]
    HeatMap(heat_data, radius=50, blur=30, min_opacity=0.3).add_to(mapa)
    
    mapa.save('mapa_colombia.html')
    print("✅ Mapa guardado como 'mapa_colombia.html'")
    print("   Abrir en el navegador para ver el mapa interactivo")
    
except ImportError:
    print("⚠️  Folium no está instalado. Para instalar:")
    print("   !pip install folium")

# Mapa con Plotly (no requiere folium)
import plotly.express as px
import pandas as pd

ciudades_df = pd.DataFrame([
    {'ciudad': c, 'lat': d['lat'], 'lon': d['lon'],
     'pib_percapita': d['valor'], 'poblacion': d['poblacion']}
    for c, d in ciudades_col.items()
])

fig_map = px.scatter_mapbox(
    ciudades_df,
    lat='lat', lon='lon',
    color='pib_percapita',
    size='poblacion',
    hover_name='ciudad',
    size_max=40,
    color_continuous_scale='Blues',
    mapbox_style='carto-positron',
    title='PIB per cápita por Ciudad — Colombia (mapa interactivo con Plotly)',
    zoom=5,
    center={'lat': 4.5, 'lon': -74.0}
)
fig_map.update_layout(height=500, template='plotly_white')
fig_map.show()

print("\n📊 En Power BI: Visual 'Mapa de burbujas' o 'Mapa de Azure' (requiere cuenta Azure).")
print("   Campos: Latitud, Longitud, Tamaño (población), Color (PIB)")
print("\n📈 En Tableau: Show Me → Maps. Arrastrar Latitud/Longitud a Rows/Columns.")
````

## 4. Series Temporales: Tendencia, Estacionalidad y Anomalías 📅

````python
# ============================================================
# Descomposición de Series Temporales
# statsmodels: tendencia + estacionalidad + residuos
# ============================================================

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

try:
    from statsmodels.tsa.seasonal import seasonal_decompose
    
    # Serie temporal simulada con estacionalidad
    fechas = pd.date_range('2020-01-01', '2024-12-31', freq='W')
    np.random.seed(42)
    tendencia = np.linspace(50000, 75000, len(fechas))
    estacionalidad = 8000 * np.sin(2 * np.pi * np.arange(len(fechas)) / 52)
    ruido = np.random.randn(len(fechas)) * 2000
    serie = tendencia + estacionalidad + ruido
    
    ts = pd.Series(serie, index=fechas, name='Ventas')
    descomposicion = seasonal_decompose(ts, model='additive', period=52)
    
    # Visualización con Plotly
    fig = make_subplots(rows=4, cols=1, shared_xaxes=True,
                        subplot_titles=['Serie Original', 'Tendencia', 
                                        'Estacionalidad (anual)', 'Residuos (ruido)'],
                        vertical_spacing=0.08)
    
    traces = [
        (ts, '#3b82f6', 1),
        (descomposicion.trend, '#10b981', 2),
        (descomposicion.seasonal, '#f59e0b', 3),
        (descomposicion.resid, '#ef4444', 4)
    ]
    
    for data, color, row in traces:
        fig.add_trace(go.Scatter(
            x=data.index, y=data.values,
            mode='lines', line=dict(color=color, width=1.5),
            showlegend=False
        ), row=row, col=1)
    
    fig.update_layout(
        height=700,
        title='Descomposición de Serie Temporal: Tendencia + Estacionalidad + Ruido',
        template='plotly_white'
    )
    fig.show()
    
except ImportError:
    print("statsmodels no disponible. Instalar con: pip install statsmodels")

print("\n📊 En Power BI: 'Gráfico de líneas' con análisis de inteligencia de tiempo (Time Intelligence).")
print("   DAX para crecimiento YoY: CALCULATE(SUM(Ventas), SAMEPERIODLASTYEAR(Calendario[Fecha]))")
print("\n📈 En Tableau: Usar funciones de fecha (YEAR, QUARTER, MONTH) para granularidad temporal.")
print("   Analytics → Trend Line para añadir tendencia automáticamente.")
````

## 5. Data Storytelling: Narrar con Datos 📖

La **narrativa visual** es el arte de construir una historia coherente y persuasiva usando datos y visualizaciones. Es la competencia que más diferencia a un científico de datos de un analista.

### La estructura narrativa de Cole Nussbaumer Knaflic:

```
1. CONTEXTO    → ¿Quién es la audiencia? ¿Qué decide? ¿Qué sabe?
       ↓
2. TENSIÓN     → ¿Cuál es el problema/oportunidad que muestra el dato?
       ↓
3. PUNTO FOCAL → UN gráfico que muestre la tensión con máxima claridad
       ↓
4. EVIDENCIA   → 2-3 gráficos de soporte que confirman el patrón
       ↓
5. LLAMADA A LA ACCIÓN → ¿Qué debe hacer la audiencia con esta información?
```

### Técnicas de énfasis visual:

| Técnica | Código Python | En Power BI | En Tableau |
|---|---|---|---|
| **Spotlight** (un punto en rojo) | `color=['#cbd5e1']*n; color[i]='#ef4444'` | Formato condicional | Color encoding |
| **Anotaciones** | `ax.annotate(...)` | Cuadro de texto en el lienzo | Annotations |
| **Referencia** | `ax.axhline(meta)` | Línea de referencia en visual | Reference line |
| **Contexto temporal** | `ax.axvspan(inicio, fin, alpha=0.1)` | Elemento de forma en lienzo | Band |

> 📌 **Regla de oro:** Una visualización cuenta una historia — define PRIMERO cuál es la historia y LUEGO elige el gráfico. Nunca al revés.
