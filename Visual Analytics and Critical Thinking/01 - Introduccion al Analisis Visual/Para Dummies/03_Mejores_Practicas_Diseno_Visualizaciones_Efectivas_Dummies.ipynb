## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **"para no ingenieros"** del cuaderno 03: una receta corta y fácil para que tus primeros gráficos no cometan los errores más comunes.

> ⚠️ Esta es la "receta rápida". La teoría completa de diseño (por qué funciona cada regla) está en el Módulo 02 del curso.

---
## 1. Piensa en tu gráfico como en un plato de comida 🍽️

Un buen chef no le pone a un plato **todos** los ingredientes que tiene en la cocina — elige solo los que hacen que el plato sepa bien. Un buen gráfico funciona igual: **menos elementos, pero cada uno con un propósito**.

Si tu gráfico tiene sombras 3D, 7 colores distintos, bordes gruesos y una leyenda gigante... probablemente le sobran "ingredientes".

---
## 2. La receta de 8 pasos para un buen gráfico ✅

| Paso | Qué hacer | Como decirlo simple |
|---|---|---|
| 1 | Elige el tipo de gráfico correcto | ¿quieres comparar, ver tendencia o ver relación? |
| 2 | Quita adornos inútiles | sin sombras 3D, sin bordes gruesos innecesarios |
| 3 | Pon un título que diga el mensaje | no "Ventas 2024", mejor "Las ventas cayeron en marzo" |
| 4 | Empieza el eje en cero | si no, estás "exagerando con lupa" |
| 5 | Usa color solo si significa algo | no pintes de colores random solo por decorar |
| 6 | Ordena de mayor a menor (si aplica) | así se lee de un vistazo, sin buscar |
| 7 | Pon los números directo en la barra si puedes | menos idas y vueltas a la leyenda |
| 8 | Revisa que se entienda sin explicación verbal | si tienes que "traducir" el gráfico hablando, rediseñalo |

```python
# Veamos la receta en accion: mismo dato, antes y despues de aplicar la receta
import matplotlib.pyplot as plt
import numpy as np

productos = ['Producto D', 'Producto A', 'Producto C', 'Producto B', 'Producto E']
ventas = [45, 92, 61, 78, 33]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# ANTES: sin receta
colores_antes = ['#f97316', '#a855f7', '#06b6d4', '#eab308', '#ec4899']
axes[0].bar(productos, ventas, color=colores_antes, edgecolor='black', linewidth=1.5)
axes[0].set_ylim(30, 95)
axes[0].set_title('ANTES (sin receta)', fontsize=10, color='#dc2626', fontweight='bold')
axes[0].tick_params(axis='x', rotation=25)

# DESPUES: con receta
orden = np.argsort(ventas)[::-1]
productos_ord = [productos[i] for i in orden]
ventas_ord = [ventas[i] for i in orden]
colores_despues = ['#2563eb'] * len(ventas_ord)
colores_despues[0] = '#10b981'

bars = axes[1].bar(productos_ord, ventas_ord, color=colores_despues, edgecolor='white')
axes[1].set_ylim(0, 100)
axes[1].set_title('DESPUÉS (con receta)', fontsize=10, color='#059669', fontweight='bold')
axes[1].tick_params(axis='x', rotation=25)
for bar in bars:
    h = bar.get_height()
    axes[1].annotate(f'{h}', xy=(bar.get_x() + bar.get_width()/2, h),
                      xytext=(0, 4), textcoords='offset points', ha='center', fontsize=10, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
print("¿Cuál gráfico entiendes más rápido? Esa es la receta funcionando.")
```
