## 1. No te creas todo lo que ves (ni siquiera tus propios gráficos) 🕵️

Imagina que un amigo te dice: *"¡Como vendo helados y aumentan los ahogamientos en la playa, entonces vender helados causa ahogamientos, deberían prohibirlos!"*

Suena ridículo, ¿verdad? Pero es exactamente el tipo de error que se comete todos los días con gráficos de negocio, noticias y redes sociales — solo que disfrazado de números "serios".

**Pensar críticamente sobre datos** significa hacerte una pregunta simple antes de creer cualquier gráfico: *"¿este gráfico me está mostrando lo que realmente pasa, o solo lo que parece pasar?"*

```python
# El mismo ejemplo del "eje truncado" pero explicado bien simple
import matplotlib.pyplot as plt

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
satisfaccion = [78, 79, 77, 80, 79, 81]  # en realidad casi no cambia

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(meses, satisfaccion, marker='o', color='#ef4444', linewidth=2.5)
axes[0].set_ylim(76, 82)
axes[0].set_title('"¡Mira qué disparada!"\n(truco: el eje casi no muestra rango)', fontsize=10)

axes[1].plot(meses, satisfaccion, marker='o', color='#3b82f6', linewidth=2.5)
axes[1].set_ylim(0, 100)
axes[1].set_title('La verdad: casi sin cambios\n(eje completo 0-100)', fontsize=10)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
print("Mismo dato. Dos graficos. Solo uno te dice la verdad completa.")
```

## 2. Dos trampas comunes (y cómo detectarlas) 🎯

| Trampa | Ejemplo cotidiano | Cómo la detectas |
|---|---|---|
| **El eje "recortado"** | Un anuncio que dice "¡Ventas explotaron!" con un gráfico que empieza en 90 en vez de 0 | Pregúntate: ¿dónde empieza el eje vertical? |
| **"A pasa, B pasa, entonces A causa B"** | "Los días que uso paraguas llueve más" (¡el paraguas no causa la lluvia!) | Pregúntate: ¿hay una tercera razón que explique ambas cosas? |

> 🧠 **Truco mental fácil:** cada vez que un gráfico te sorprenda mucho ("¡wow, qué cambio tan grande!"), es justo el momento de mirarlo con más cuidado, no menos.

## 3. Preguntas rápidas antes de creer un gráfico ❓

La próxima vez que veas un gráfico llamativo (en una noticia, un anuncio, una reunión de trabajo), pregúntate:

1. ¿El eje empieza en cero?
2. ¿Quién hizo este gráfico y qué gana si yo le creo?
3. ¿Se está comparando manzanas con manzanas, o manzanas con naranjas?
4. ¿Esto que "va junto" realmente se causa entre sí, o hay algo más detrás?

> 🔍 En el Módulo 03 aprenderás una lista mucho más completa, con nombres técnicos para cada tipo de "trampa mental" (se llaman *sesgos cognitivos*).
