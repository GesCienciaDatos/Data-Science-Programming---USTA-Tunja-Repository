## 1. Analítica Visual: como tener un asistente con rayos X en los ojos 👁️

Imagina que te dan dos opciones para entender cómo van las ventas de tu empresa este año:

```
Mes    Ventas($)  Mes    Ventas($)  Mes    Ventas($)
Ene    123,450    May    198,320    Sep    267,890
Feb    134,200    Jun    210,450    Oct    289,120
Mar    145,600    Jul    234,670    Nov    312,450
```

**Opción B — El gráfico:** una línea que sube suavemente de enero a diciembre con una aceleración en noviembre-diciembre.

| Hecho | Lo que significa para ti |

## 2. Un poquito de historia: de Napoleón a Tableau 📜

No inventamos los gráficos ayer. Algunos hitos curiosos:

| Año | Qué pasó |
|---|---|
| 1786 | Un escocés (William Playfair) inventó el gráfico de barras que hoy usas en Excel |
| 1854 | Un médico (John Snow) mapeó casos de cólera en Londres y descubrió el pozo de agua contaminado — la primera "analítica visual" que salvó vidas |
| 1869 | Un ingeniero francés dibujó el desastroso avance de Napoleón en Rusia en un solo mapa que aún hoy se estudia |
| 2003-2015 | Nacen Tableau, Power BI y las herramientas modernas de Python que usarás en este curso |

> 💡 **Para recordar:** la gente lleva más de 200 años descubriendo que "un dibujo vale más que mil números". Hoy solo tenemos mejores herramientas para hacerlo.

## 3. El camino de los datos a la decisión 🗺️

Piensa en analítica visual como un viaje con 4 paradas:

```
🗃️ DATOS          👀 VES            🧠 ENTIENDES       ✅ DECIDES
"123,450"    →   una barra alta  →  "enero fue bueno" → "repito la campaña"
números crudos     gráfico           significado          acción
```

### Los trucos visuales que tu cerebro usa sin pedirle permiso

¿Sabías que tu cerebro detecta ciertas cosas **antes de que conscientemente 'mires'**? Se llaman **atributos preatentivos** — una manera fancy de decir "cosas que el ojo capta en automático".

| Truco | Ejemplo cotidiano | Ejemplo en gráficos |
|---|---|---|
| **Color diferente** | Una manzana roja entre naranjas | Una barra roja entre azules = ¡alarma! |
| **Tamaño diferente** | Un elefante en una granja | Una burbuja gigante = número enorme |
| **Posición** | Un edificio más alto | Barra más alta = valor mayor |
| **Movimiento** | Una mosca volando | Animaciones en el dashboard |

> 🧠 **Para recordar:** Un buen visualizador usa estos trucos **con propósito**. Poner todo en rojo = no sirve de nada. Poner **solo** lo importante en rojo = poder total.

````python
# ¡Probemos el truco del color preatentivo!
# Intenta encontrar el punto especial antes de contar mas de 1 segundo.

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(7)
n = 60
x = np.random.rand(n)
y = np.random.rand(n)

colores = ['#cbd5e1'] * n  # todos grises claros
colores[30] = '#ef4444'    # ¡uno rojo!

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x, y, c=colores, s=80, edgecolors='white', linewidths=0.5)
ax.set_title('🔍 ¿Cuánto tardas en encontrar el punto rojo?\n(Tu cerebro lo hizo ANTES de que termines de leer esta pregunta)',
             fontsize=12, fontweight='bold')
ax.set_xticks([])
ax.set_yticks([])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)

plt.tight_layout()
plt.show()
print("👆 Esto es un atributo preatentivo: detección instantánea sin esfuerzo consciente.")
print("   En Power BI o Tableau: es el mismo principio cuando usas 'formato condicional'.")
````

## 4. Power BI, Tableau y Python: ¿cuál uso yo? 🤔

Es como elegir entre una **calculadora**, una **hoja de cálculo** y un **computador completo**:

| Herramienta | Es como... | Úsala cuando... |
|---|---|---|
| **Power BI** | Una hoja de cálculo súper poderosa con gráficos automáticos | Quieres un dashboard bonito rápido, tu empresa ya usa Microsoft, o eres analista de negocio |
| **Tableau** | Un iPad muy intuitivo para explorar datos | Quieres arrastrar y soltar para descubrir patrones, presentaciones ejecutivas |
| **Python** | La herramienta de un constructor — más trabajo, resultados infinitos | Quieres automatizar, conectar con Machine Learning, o crear algo completamente personalizado |

### La buena noticia:
En este curso aprenderás **Python**, que es el más completo. Pero también aprenderás qué hace Power BI y Tableau para que puedas hablar el idioma de tus colegas analistas.

> 💡 **Analogía final:** Python es como aprender a cocinar desde cero (puedes hacer cualquier plato). Power BI y Tableau son como pedir comida a domicilio de un buen restaurante (rápido y bueno, pero limitado al menú).

## 5. Mi primer gráfico en Python 🎉

Sin más explicaciones — ejecuta el código y mira qué pasa. Luego te explico línea por línea.

````python
# Tu primer grafico de datos reales en Python
# No te preocupes por memorizar el codigo ahora -- solo observa el resultado

import matplotlib.pyplot as plt

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
         'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ventas = [123, 134, 146, 179, 198, 210, 235, 251, 268, 289, 312, 390]

fig, ax = plt.subplots(figsize=(11, 5))

ax.plot(meses, ventas, color='#3b82f6', linewidth=2.5, marker='o',
        markersize=7, markerfacecolor='white', markeredgewidth=2)
ax.fill_between(range(len(meses)), ventas, alpha=0.1, color='#3b82f6')

ax.plot(11, 390, 'o', color='#ef4444', markersize=12, label='Mejor mes: Dic')
ax.annotate('🏆 Dic: $390k\n(¡Navidad!)', xy=(11, 390),
            xytext=(9.5, 360), fontsize=10, color='#ef4444',
            arrowprops=dict(arrowstyle='->', color='#ef4444'))

ax.set_title('Ventas Mensuales 2024 — Mi Primera Visualización en Python 🎉',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Mes', fontsize=11)
ax.set_ylabel('Ventas (miles de $)', fontsize=11)
ax.set_xticks(range(len(meses)))
ax.set_xticklabels(meses)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

print("✅ ¡Felicidades! Acabas de crear tu primera visualización con Python.")
print("   En Power BI harías lo mismo: insertar 'Gráfico de líneas' y arrastrar los campos.")
print("   En Tableau: arrastrar 'Mes' a Columns y 'Ventas' a Rows → Show Me → Line.")
````
