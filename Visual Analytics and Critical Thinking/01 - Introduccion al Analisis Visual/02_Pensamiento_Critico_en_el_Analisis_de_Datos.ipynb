## 1. ¿Por qué el pensamiento crítico es indispensable al analizar datos? 🧠

> ⚠️ **Aviso de alcance:** este cuaderno es una **introducción**. La sección 1.3.2 del syllabus se profundiza por completo en el **Módulo 03 — Pensamiento Crítico en Datos**, con sesgos cognitivos, auditoría sistemática de visualizaciones y casos forenses. Aquí solo plantamos la pregunta: **¿por qué importa?**

Los datos no mienten, pero **las personas que los procesan, seleccionan y grafican sí pueden hacerlo** — a veces de forma deliberada, y muchas más veces sin darse cuenta. El pensamiento crítico aplicado al análisis de datos es la disciplina de **cuestionar sistemáticamente** cada paso del proceso:

| Etapa del análisis | Pregunta crítica que deberías hacerte |
|---|---|
| **Recolección** | ¿La muestra representa realmente a la población de interés? |
| **Limpieza** | ¿Qué datos eliminé y por qué? ¿Eso sesga el resultado? |
| **Análisis** | ¿Estoy confundiendo correlación con causalidad? |
| **Visualización** | ¿La escala, el eje o el color exageran o minimizan el patrón real? |
| **Comunicación** | ¿Estoy mostrando el dato que responde la pregunta, o el que confirma lo que quiero probar? |

> 📌 **Idea clave:** un análisis técnicamente correcto puede llevar a una conclusión equivocada si nadie aplicó escépticismo informado en el camino. Las herramientas (Python, Power BI, Tableau) no piensan por ti — **solo ejecutan lo que les pides**.

## 2. Ejemplo 1: el eje que exagera el cambio 📈

Uno de los engaños visuales más comunes (y más fáciles de cometer sin mala intención) es truncar el eje Y. El código a continuación usa los **mismos datos** para construir dos gráficos con impresiones opuestas.

```python
# ============================================================
# Ejemplo de mal uso de graficos: eje Y truncado
# Mismos datos, dos lecturas distintas
# ============================================================

import matplotlib.pyplot as plt

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
satisfaccion_cliente = [78, 79, 77, 80, 79, 81]  # variacion real: apenas 4 puntos sobre 100

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Version enganosa: eje truncado entre 76 y 82
axes[0].plot(meses, satisfaccion_cliente, marker='o', color='#dc2626', linewidth=2.5)
axes[0].set_ylim(76, 82)
axes[0].set_title('Version enganosa: eje truncado\n"¡La satisfacción se disparó!"', fontsize=10, fontweight='bold')
axes[0].set_ylabel('Satisfacción (0-100)')
axes[0].yaxis.grid(True, alpha=0.3)

# Version honesta: eje completo 0-100
axes[1].plot(meses, satisfaccion_cliente, marker='o', color='#2563eb', linewidth=2.5)
axes[1].set_ylim(0, 100)
axes[1].set_title('Version honesta: eje completo (0-100)\n"Cambio real: prácticamente estable"', fontsize=10, fontweight='bold')
axes[1].set_ylabel('Satisfacción (0-100)')
axes[1].yaxis.grid(True, alpha=0.3)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

print("Cambio real de enero a junio:", satisfaccion_cliente[-1] - satisfaccion_cliente[0], "puntos sobre 100")
print("El grafico de la izquierda no es 'falso' -- solo elige una escala que exagera la percepcion visual del cambio.")
```

## 3. Ejemplo 2: correlación disfrazada de causalidad 🔗

El segundo error crítico más frecuente no está en el gráfico en sí, sino en la **interpretación** que se hace de él. Dos variables pueden moverse juntas sin que una cause la otra.

```python
# ============================================================
# Ejemplo de mal uso de graficos: correlacion vs causalidad
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(3)
anios = np.arange(2015, 2025)

# Dos series que suben juntas por una tercera variable oculta (ej. crecimiento poblacional/economico)
ventas_helados = np.array([100, 108, 115, 126, 138, 120, 145, 160, 172, 185])
ahogamientos = ventas_helados * 0.08 + np.random.normal(0, 0.4, len(anios))

fig, ax1 = plt.subplots(figsize=(9, 5))

color1 = '#f59e0b'
ax1.set_xlabel('Año')
ax1.set_ylabel('Ventas de helados (miles $)', color=color1)
ax1.plot(anios, ventas_helados, color=color1, marker='o', linewidth=2.5)
ax1.tick_params(axis='y', labelcolor=color1)

ax2 = ax1.twinx()
color2 = '#dc2626'
ax2.set_ylabel('Casos de ahogamiento', color=color2)
ax2.plot(anios, ahogamientos, color=color2, marker='s', linewidth=2.5)
ax2.tick_params(axis='y', labelcolor=color2)

plt.title('¿Los helados causan ahogamientos?\n(Ambas suben en verano — la variable oculta es la temporada)',
          fontsize=11, fontweight='bold')
fig.tight_layout()
plt.show()

correlacion = np.corrcoef(ventas_helados, ahogamientos)[0, 1]
print(f"Correlación estadística: {correlacion:.2f} (fuerte)")
print("¿Causalidad real? No. La variable oculta ('confounder') es la temporada/clima cálido:")
print("más calor -> más gente compra helado Y más gente nada -> más riesgo de ahogamiento.")
```

## 4. Preguntas que un pensador crítico siempre se hace ante un gráfico ❓

Antes de aceptar (o publicar) cualquier visualización, hazte estas preguntas rápidas:

1. **¿El eje empieza en cero?** Si no, ¿hay una razón legítima (ej. rango muy estrecho de interés científico) o solo exagera el cambio?
2. **¿Qué periodo o muestra se eligió, y por qué ese y no otro?** Cambiar el rango temporal puede invertir una conclusión.
3. **¿Se está comparando algo comparable?** (misma unidad, misma población, mismo periodo)
4. **¿La correlación visual implica causalidad, o hay una variable oculta?**
5. **¿Quién produjo este gráfico y qué interés tiene en la conclusión que sugiere?**

> 🔍 Esta lista es apenas la punta del iceberg. El **Módulo 03** retoma cada uno de estos puntos con un marco formal de sesgos cognitivos (anclaje, confirmación, disponibilidad) y una checklist completa de auditoría de visualizaciones.
