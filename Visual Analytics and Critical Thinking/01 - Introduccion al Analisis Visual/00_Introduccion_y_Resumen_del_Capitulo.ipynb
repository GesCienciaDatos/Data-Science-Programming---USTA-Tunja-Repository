## 2. ¿Por qué ahora, y por qué juntos? 🤔

Es tentador pensar que "visualización" y "pensamiento crítico" son temas separados: uno es técnico (cómo dibujar), el otro es filosófico (cómo razonar). En la práctica profesional están **inseparablemente unidos**:

- Cada decisión de diseño (qué tipo de gráfico, qué escala, qué color) **es también** una decisión ética: puede aclarar o puede engañar.
- Un analista sin pensamiento crítico puede construir gráficos técnicamente correctos pero **conceptualmente engañosos** (ejes truncados, correlaciones disfrazadas de causalidad, muestras sesgadas).
- Un analista sin habilidades de visualización puede tener el razonamiento más riguroso del mundo, y **nadie lo escuchará**, porque su comunicación visual no logra transmitirlo.

Por eso el syllabus del curso dedica un capítulo introductorio compartido antes de separar los caminos: **generalidades de la disciplina** (sección 1.3.1), **por qué pensar críticamente** (sección 1.3.2) y **cómo empezar a diseñar bien** (sección 1.3.3).

```python
# ============================================================
# Teaser del capitulo: el mismo dato, dos formas de mostrarlo
# Un vistazo rapido a por que "como" se muestra un dato importa tanto como el dato mismo
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

trimestres = ['T1', 'T2', 'T3', 'T4']
ingresos = [102, 104, 103, 106]  # variacion real: apenas 4%

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Panel 1: eje truncado -> exagera visualmente el cambio (lo que NO se debe hacer)
axes[0].bar(trimestres, ingresos, color='#dc2626', alpha=0.85, edgecolor='white')
axes[0].set_ylim(100, 107)
axes[0].set_title('Version A: eje Y truncado\n(¿"explosivo" crecimiento?)', fontsize=11, fontweight='bold')
axes[0].set_ylabel('Ingresos (millones $)')

# Panel 2: eje desde cero -> representacion honesta de la magnitud real
axes[1].bar(trimestres, ingresos, color='#2563eb', alpha=0.85, edgecolor='white')
axes[1].set_ylim(0, 120)
axes[1].set_title('Version B: eje desde cero\n(crecimiento real: ~4%)', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Ingresos (millones $)')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.yaxis.grid(True, alpha=0.3)
    ax.set_axisbelow(True)

fig.suptitle('Mismos datos, dos historias distintas — este capítulo explica por qué', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

print("Diferencia real entre T1 y T4:", round((ingresos[-1]/ingresos[0] - 1) * 100, 1), "%")
print("Este es el tipo de decision de diseño que este capitulo te enseña a detectar y a evitar.")
```
