## 1. Por qué empezar con buenas prácticas, antes de la teoría formal ✨

> ⚠️ **Aviso de alcance:** este cuaderno entrega un **checklist inicial**, pensado para que empieces a diseñar gráficos correctos desde ya. La teoría formal —gramática de gráficos, codificación visual, principios de Tufte y data-ink ratio— se desarrolla en profundidad en el **Módulo 02 — Fundamentos de Visualización**.

No hace falta dominar la teoría completa del diseño de información para dejar de cometer los errores más comunes. Igual que aprendes las reglas básicas de ortografía antes de estudiar gramática avanzada, aquí aprenderás un **checklist práctico** que puedes aplicar desde tu primer gráfico.

---
## 2. Checklist de Buenas Prácticas Iniciales ✅

| # | Práctica | Por qué importa |
|---|---|---|
| 1 | **Elige el tipo de gráfico correcto** para tu pregunta (¿comparación? ¿tendencia? ¿distribución? ¿relación?) | Un gráfico equivocado oculta el patrón en lugar de revelarlo |
| 2 | **Elimina el "chartjunk"** (bordes, sombras 3D, fondos decorativos innecesarios) | Cada elemento visual que no aporta información distrae |
| 3 | **Usa títulos que comuniquen el mensaje**, no solo el tema (ej. "Las ventas cayeron 12% en Q3", no "Ventas por trimestre") | El título es lo primero (y a veces lo único) que se lee |
| 4 | **Empieza el eje numérico en cero** salvo justificación explícita | Evita exagerar o minimizar cambios (ver cuaderno 02) |
| 5 | **Usa el color con propósito**, no decorativamente | El color debe codificar una categoría o resaltar un dato clave |
| 6 | **Ordena las categorías de forma significativa** (por valor, no alfabéticamente, salvo que el orden importe) | Facilita comparar de un vistazo |
| 7 | **Etiqueta directamente** cuando sea posible, en vez de depender solo de leyendas | Reduce el "ir y venir" visual entre gráfico y leyenda |
| 8 | **Verifica accesibilidad de color** (daltonismo: evita depender solo de rojo/verde) | ~8% de los hombres tienen algún tipo de daltonismo |

> 📌 **Regla de oro:** si tienes que explicar verbalmente lo que el gráfico "en realidad" quiere decir, el gráfico necesita rediseño.

## 3. Antes / Después: aplicando el checklist 🔧

Vamos a tomar un gráfico que viola varias reglas del checklist y lo corregimos paso a paso.

```python
# ============================================================
# Antes / Despues: aplicando el checklist de buenas practicas
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

productos = ['Producto D', 'Producto A', 'Producto C', 'Producto B', 'Producto E']
ventas = [45, 92, 61, 78, 33]

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# ANTES: colores aleatorios sin proposito, orden alfabetico/arbitrario, eje truncado, sin etiquetas de valor
colores_antes = ['#f97316', '#a855f7', '#06b6d4', '#eab308', '#ec4899']
axes[0].bar(productos, ventas, color=colores_antes, edgecolor='black', linewidth=1.5)
axes[0].set_ylim(30, 95)
axes[0].set_title('ANTES\n(colores sin propósito, sin orden, eje truncado)', fontsize=10, fontweight='bold', color='#dc2626')
axes[0].tick_params(axis='x', rotation=25)

# DESPUES: orden por valor, un solo color con el maximo resaltado, eje desde cero, etiquetas directas
orden = np.argsort(ventas)[::-1]
productos_ord = [productos[i] for i in orden]
ventas_ord = [ventas[i] for i in orden]
colores_despues = ['#2563eb'] * len(ventas_ord)
colores_despues[0] = '#10b981'  # resalta el mejor producto con proposito

bars = axes[1].bar(productos_ord, ventas_ord, color=colores_despues, edgecolor='white')
axes[1].set_ylim(0, 100)
axes[1].set_title('DESPUÉS\n(ordenado, color con propósito, eje desde cero)', fontsize=10, fontweight='bold', color='#059669')
axes[1].tick_params(axis='x', rotation=25)
for bar in bars:
    h = bar.get_height()
    axes[1].annotate(f'{h}', xy=(bar.get_x() + bar.get_width()/2, h),
                      xytext=(0, 4), textcoords='offset points', ha='center', fontsize=10, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

print("El grafico 'DESPUES' aplica: orden por valor, color con proposito, eje desde cero y etiquetas directas.")
print("La teoria formal detras de cada una de estas decisiones se profundiza en el Modulo 02.")
```

## 4. Lo que viene después 📐

Este checklist es un punto de partida, no el destino final. En el **Módulo 02 — Fundamentos de Visualización** profundizarás en:

- La **Gramática de Gráficos** (Wilkinson): cómo cualquier gráfico se construye a partir de componentes reutilizables (datos, estética, geometría, escalas).
- La **codificación visual** formal: qué tipo de dato (categórico, ordinal, numérico) se codifica mejor con qué canal visual (posición, color, tamaño).
- Los principios de **Edward Tufte**, incluyendo el *data-ink ratio* (maximizar la tinta que representa datos, minimizar la decorativa).
