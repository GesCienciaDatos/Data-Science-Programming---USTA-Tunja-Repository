## 1. Las 4 familias de visualización, explicadas fácil 🍽️

Piensa en 4 formas distintas de contar la misma historia:

| Familia | Es como... | Cuándo usarla |
|---|---|---|
| **Chart** (gráfico) | Una foto que resume un patrón | "¿Quién vendió más este mes?" |
| **Table** (tabla) | Un recibo detallado | "¿Cuánto pagué exactamente el 15 de marzo?" |
| **Map** (mapa) | El GPS de tu celular | "¿En qué ciudades tengo más clientes?" |
| **Graph** (grafo/red) | El árbol genealógico de tu familia | "¿Quién está conectado con quién?" |

> 🧠 La pregunta que te haces (no el gusto estético) decide qué familia usar.

## 2. ¿Gráfico o tabla? El truco para decidir 🧾

Imagina que estás en el banco:
- Si preguntas **"¿cuánto gasté en marzo?"** → necesitas un **número exacto** → **tabla** (como tu estado de cuenta).
- Si preguntas **"¿en qué mes gasté más?"** → necesitas **comparar de un vistazo** → **gráfico** (como una barra).

| Pregunta | ¿Tabla o gráfico? |
|---|---|
| "¿Cuánto exactamente pagué el 15 de marzo?" | 🧾 Tabla |
| "¿En qué mes gasté más este año?" | 📊 Gráfico |
| "¿Cuál fue el producto #4,382 que vendí?" | 🧾 Tabla |
| "¿Las ventas están subiendo o bajando?" | 📊 Gráfico |

### El combo ganador: tabla con colorcitos
Excel, Power BI y Tableau tienen una opción llamada "tabla con formato condicional": es una tabla normal, pero las celdas se pintan según su valor (más oscuro = más alto). Tienes el número exacto **y** el patrón de color al mismo tiempo.

```python
# Una tabla normal vs una tabla "con colorcitos" (highlight table)

import pandas as pd
import numpy as np

rng = np.random.default_rng(3)
productos = ['Pan', 'Leche', 'Huevos', 'Café', 'Azúcar']
meses = ['Ene', 'Feb', 'Mar']
datos = rng.integers(10, 100, size=(len(productos), len(meses)))

tabla = pd.DataFrame(datos, index=productos, columns=meses)
tabla_bonita = tabla.style.background_gradient(cmap='Blues', axis=None) \
                          .set_caption('Unidades vendidas: mismo numero, ahora con color')
tabla_bonita

print("💡 Sigues viendo el numero exacto, pero el color te dice de un vistazo")
print("   donde estan los valores mas altos, sin tener que leer celda por celda.")
```

## 3. ¿Cuándo usar un mapa? La lógica del GPS 🗺️

Si tu pregunta menciona un **lugar** (país, ciudad, dirección), lo más probable es que un mapa cuente la historia mejor que una barra.

| Tipo de mapa | Es como... |
|---|---|
| **Mapa de puntos** | Alfileres clavados en un mapa de corcho, uno por cada tienda |
| **Mapa coroplético** (de colores por región) | Un mapa del clima, donde cada país se pinta según la temperatura |

> ⚠️ **Cuidado:** un país grande en tamaño (como Rusia o Canadá) "se ve más importante" en un mapa aunque su valor sea bajo. El tamaño del país en el mapa NO es el dato — el color sí lo es. No los confundas.

## 4. ¿Qué es un "graph" (grafo)? El árbol genealógico 🌳

Cuando la gente dice "gráfico" normalmente piensa en barras o líneas. Pero en visualización de datos, un **graph** es otra cosa muy específica: **círculos conectados por líneas**, donde cada círculo es "alguien" o "algo", y cada línea es "una relación".

Ejemplos de la vida diaria:
- El árbol genealógico de tu familia (cada persona = un nodo, cada línea = "es hijo de")
- El mapa del metro (cada estación = un nodo, cada línea = una ruta)
- Quién sigue a quién en Instagram

En estos casos, **la posición en la pantalla no importa** — lo único que importa es **qué está conectado con qué**. Por eso un grafo NO se lee igual que un gráfico de barras.
