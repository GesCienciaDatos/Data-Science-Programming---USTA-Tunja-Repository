## 1. El Panorama de Herramientas: Low-Code vs Code-First 🧭

Toda herramienta de visualización se ubica en algún punto de un espectro entre dos extremos:

- **Low-code / No-code**: interfaces gráficas de "arrastrar y soltar" (*drag-and-drop*). Priorizan la **velocidad** y la **accesibilidad** para usuarios de negocio. Ejemplos: Tableau, Power BI.
- **Code-first**: todo se define mediante código. Priorizan el **control total** y la **personalización**, a costa de una curva de aprendizaje más alta. Ejemplos: D3.js, y el ecosistema Python (matplotlib/seaborn/plotly).

```
Menos control, más velocidad                          Más control, menos velocidad
◄──────────────────────────────────────────────────────────────────────────────►
   Power BI          Tableau          Plotly/Seaborn          Matplotlib          D3.js
  (low-code)      (low-code)         (code, alto nivel)     (code, medio nivel)  (code, bajo nivel)
```

No existe una herramienta "mejor" en abstracto — existe la herramienta correcta según el **usuario**, el **objetivo** y el **tiempo disponible**.

## 2. Tableau: Analítica Visual "Drag-and-Drop" 📊

**Qué es:** plataforma de Business Intelligence (BI) especializada en análisis visual interactivo. Pionera del paradigma *drag-and-drop* (arrastrar campos a los ejes "Rows/Columns/Marks" para construir gráficos sin escribir código).

| Aspecto | Detalle |
|---|---|
| **Tipo de usuario** | Analistas de negocio, consultores, equipos de BI sin perfil técnico profundo |
| **Curva de aprendizaje** | Baja-media: interfaz intuitiva, pero dominar *calculated fields* y *LOD expressions* toma tiempo |
| **Fortaleza principal** | Exploración visual muy rápida y fluida; el mejor "sentir" de arrastrar-y-ver-resultado del mercado |
| **Lenguaje propio** | Tableau Calculation Language (fórmulas tipo Excel + funciones de nivel de detalle) |
| **Modelo de licenciamiento** | Comercial (Tableau Public es gratuito pero los workbooks son públicos) |
| **Ecosistema** | Independiente; se conecta a casi cualquier fuente de datos (SQL, Excel, la nube) |

---
## 3. Power BI: BI Empresarial Integrado al Ecosistema Microsoft 💼

**Qué es:** plataforma de BI de Microsoft, fuertemente integrada con Excel, Azure y Microsoft 365. Combina un motor de reportes visuales con **Power Query** (ETL) y **DAX** (lenguaje de fórmulas analíticas).

| Aspecto | Detalle |
|---|---|
| **Tipo de usuario** | Analistas de negocio dentro de organizaciones que ya usan el ecosistema Microsoft |
| **Curva de aprendizaje** | Baja para reportes básicos; media-alta para modelado de datos con DAX |
| **Fortaleza principal** | Integración nativa con Excel/Azure/Teams y un motor ETL propio (Power Query) incluido gratis |
| **Lenguajes propios** | **M** (Power Query, transformación de datos) y **DAX** (medidas y cálculos analíticos) |
| **Modelo de licenciamiento** | Power BI Desktop gratuito; Power BI Service/Pro es comercial |
| **Ecosistema** | Muy integrado con productos Microsoft; ideal para empresas que ya usan Excel intensivamente |

> 🔗 **Recurso complementario de este curso:** el motor ETL de Power BI (**Power Query** y su lenguaje **M**) tiene un taller propio y completo en este repositorio: **[Ir al taller de Power Query](../Power%20Query/)**. Allí verás Merge, Append, Pivot/Unpivot y una tabla exhaustiva de equivalencias M ↔ Pandas.

## 4. D3.js: Control Total mediante Código 💻

**Qué es:** librería de JavaScript para manipular documentos basados en datos (*Data-Driven Documents*). No es una herramienta de "gráficos predefinidos" — es un conjunto de utilidades de bajo nivel para unir datos a elementos SVG/HTML y controlar cada píxel.

| Aspecto | Detalle |
|---|---|
| **Tipo de usuario** | Desarrolladores front-end y *data visualization engineers* |
| **Curva de aprendizaje** | Alta: requiere JavaScript, SVG y conceptos de *data binding* (`enter/update/exit`) |
| **Fortaleza principal** | Personalización absoluta — cualquier visualización interactiva que has visto en el New York Times probablemente usa D3.js |
| **Lenguaje propio** | JavaScript (librería, no lenguaje) |
| **Modelo de licenciamiento** | Open source (gratuito) |
| **Ecosistema** | Nativo de la web; se integra en cualquier sitio o aplicación sin depender de un vendor de BI |

> 💡 D3.js es a la visualización web lo que matplotlib es a Python: la capa de más bajo nivel, sobre la cual otras librerías (Vega-Lite, Observable Plot, muchos dashboards corporativos) construyen abstracciones más simples.

---
## 5. El Ecosistema Python: matplotlib, seaborn y plotly 🐍

**Qué es:** no una sola herramienta, sino un ecosistema de librerías *code-first* con distintos niveles de abstracción, usado tanto por analistas como por científicos de datos e ingenieros.

| Librería | Nivel de abstracción | Cuándo usarla |
|---|---|---|
| **matplotlib** | Bajo (control total, sintaxis verbosa) | Gráficos estáticos para publicaciones científicas, control fino de cada elemento |
| **seaborn** | Medio (sobre matplotlib, sintaxis declarativa) | Gráficos estadísticos rápidos con buen estilo por defecto |
| **plotly / plotly express** | Alto (declarativo, interactivo por defecto) | Dashboards interactivos, exploración rápida, gráficos web embebibles |

**Tipo de usuario:** científicos de datos, analistas técnicos, ingenieros de ML — cualquiera que ya trabaje en un notebook de Python y necesite visualizar como parte de un flujo de análisis reproducible (versionable en Git, automatizable, integrable a un pipeline).

## 6. Tabla Comparativa: Tableau vs Power BI vs D3.js vs Python 📋

| Criterio | Tableau | Power BI | D3.js | Python (mpl/seaborn/plotly) |
|---|---|---|---|---|
| **Paradigma** | Low-code | Low-code | Code-first | Code-first |
| **Usuario típico** | Analista de negocio | Analista de negocio (Microsoft) | Desarrollador web | Científico de datos / analista técnico |
| **Curva de aprendizaje** | Media | Media | Alta | Media (varía por librería) |
| **Interactividad out-of-the-box** | Alta | Alta | Total (requiere construirla) | Alta (plotly) / Baja (matplotlib estático) |
| **Personalización visual** | Media | Media | Total (pixel a pixel) | Alta (código = control total) |
| **ETL integrado** | Limitado | Power Query (M) — muy fuerte | No (requiere backend externo) | Pandas (fuera del gráfico) |
| **Reproducibilidad / versionado en Git** | Baja (archivos binarios) | Baja (archivos binarios) | Alta (es código) | Alta (es código) |
| **Costo** | Comercial (Public gratis) | Desktop gratis, Service comercial | Gratis (open source) | Gratis (open source) |
| **Mejor caso de uso** | Exploración visual ágil de negocio | BI empresarial + ecosistema Microsoft | Visualización web a medida, alto impacto | Análisis reproducible dentro de un pipeline de datos |

````python
# ============================================================
# Low-code vs code-first: el MISMO grafico, dos filosofias
#
# En Tableau/Power BI este grafico se construye arrastrando:
#   "Region" -> eje X, "Ventas" -> eje Y, click en 'Bar Chart'
# En Python (code-first) se construye escribiendo cada paso.
# ============================================================

import matplotlib.pyplot as plt
import pandas as pd

datos = pd.DataFrame({
    'Region': ['Norte', 'Sur', 'Este', 'Oeste', 'Centro'],
    'Ventas': [230, 190, 280, 170, 210],
})
datos = datos.sort_values('Ventas', ascending=True)

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.barh(datos['Region'], datos['Ventas'], color='#1e3a8a', edgecolor='white')
ax.set_xlabel('Ventas (miles de $)', fontsize=10, color='#64748b')
ax.set_title('Ventas por Región — construido 100% con código (Python)',
             fontsize=12, fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
for bar, val in zip(bars, datos['Ventas']):
    ax.text(bar.get_width() + 3, bar.get_y() + bar.get_height()/2,
             f'{val}', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

print("⏱️  Comparación de esfuerzo para ESTE grafico puntual:")
print("   Power BI / Tableau: ~15 segundos (arrastrar 2 campos + elegir 'Bar Chart')")
print("   Python (matplotlib): ~15 líneas de código, pero 100% reproducible y versionable en Git")
print()
print("💡 La ventaja de low-code es la VELOCIDAD para un análisis puntual.")
print("   La ventaja de code-first es la REPRODUCIBILIDAD cuando el mismo")
print("   análisis se debe repetir cada semana con datos nuevos (pipeline).")
````
