## 1. Simplicidad: El Data-Ink Ratio de Tufte 📏

Edward Tufte, en *The Visual Display of Quantitative Information* (1983), estableció principios fundamentales para visualizaciones honestas y eficientes:

### El Data-Ink Ratio

$$\text{Data-Ink Ratio} = \frac{\text{Tinta que representa datos}}{\text{Tinta total del gráfico}}$$

**Objetivo: maximizar este ratio** — cada elemento visual debe justificar su existencia con información.

### Chartjunk: lo que hay que eliminar

| Elemento | ¿Es chartjunk? | Alternativa |
|---|---|---|
| Fondo gris oscuro | ✅ Sí | Fondo blanco o sin fondo |
| Cuadrícula gruesa | ✅ Sí | Líneas finas con alpha bajo |
| Sombras 3D en barras | ✅ Sí | Barras 2D planas |
| Bordes en todos los ejes | ✅ Sí | Solo eje X e Y necesarios |
| Leyenda redundante | ✅ Sí | Etiqueta directa en el gráfico |
| Decimales innecesarios | ✅ Sí | Redondear a cifras significativas |
| Colores decorativos | ✅ Sí | Color solo con significado analítico |

```python
# ============================================================
# Antes vs Después: Aplicar principios de Tufte
# El mismo dataset — resultado completamente diferente
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

categorias = ['Q1', 'Q2', 'Q3', 'Q4']
ventas_2023 = [145, 167, 189, 234]
ventas_2024 = [158, 172, 201, 267]

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# ---- ANTES: grafico cargado de chartjunk ----
ax_before = axes[0]
x = np.arange(len(categorias))
w = 0.35
ax_before.set_facecolor('#2d2d2d')  # Fondo oscuro innecesario
ax_before.bar(x - w/2, ventas_2023, w, color='#ff6b35', label='2023',
              edgecolor='black', linewidth=2)  # Borde grueso
ax_before.bar(x + w/2, ventas_2024, w, color='#00d4aa', label='2024',
              edgecolor='black', linewidth=2)
ax_before.set_xticks(x)
ax_before.set_xticklabels(categorias, color='white', fontsize=14, fontweight='bold')
ax_before.tick_params(colors='white')
ax_before.yaxis.grid(True, color='white', linewidth=1.5)
ax_before.set_title('❌ ANTES: Chartjunk excesivo', fontsize=12, fontweight='bold', color='red')
ax_before.legend(facecolor='gray', edgecolor='white', labelcolor='white')
for spine in ax_before.spines.values():
    spine.set_edgecolor('white')
    spine.set_linewidth(2)

# ---- DESPUES: principios de Tufte ----
ax_after = axes[1]
ax_after.bar(x - w/2, ventas_2023, w, color='#94a3b8', label='2023', edgecolor='white')
ax_after.bar(x + w/2, ventas_2024, w, color='#1e3a8a', label='2024', edgecolor='white')

# Etiquetas directas (eliminar leyenda)
ax_after.text(x[-1] - w/2, ventas_2023[-1] + 5, '2023', ha='center', color='#64748b', fontsize=9)
ax_after.text(x[-1] + w/2, ventas_2024[-1] + 5, '2024', ha='center', color='#1e3a8a',
              fontsize=9, fontweight='bold')

ax_after.set_xticks(x)
ax_after.set_xticklabels(categorias, fontsize=11)
ax_after.spines['top'].set_visible(False)
ax_after.spines['right'].set_visible(False)
ax_after.spines['left'].set_visible(False)
ax_after.yaxis.grid(True, alpha=0.25, linewidth=0.8)
ax_after.set_axisbelow(True)
ax_after.tick_params(left=False)
ax_after.set_ylabel('Ventas (miles de $)', fontsize=10, color='#64748b')
ax_after.set_title('✅ DESPUÉS: Tufte — foco en los datos', fontsize=12, fontweight='bold')

plt.suptitle('Comparativa de Ventas por Trimestre 2023 vs 2024', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n📊 En Power BI: Desactiva 'Sombra', 'Borde', 'Fondo' en el panel de formato.")
print("   Usa la cuadricula solo en el eje Y con transparencia ~30%.")
print("\n📈 En Tableau: Menu Format > Lines > desactiva Grid lines o ponlas suaves.")
print("   Usa 'Label' directamente en los marcadores en lugar de leyendas externas.")
```

## 2. Claridad: Percepción Visual y los Principios de la Gestalt 🧩

La **psicología de la Gestalt** (Wertheimer, Köhler y Koffka, escuela alemana de principios del siglo XX) describe cómo el cerebro humano agrupa elementos visuales de forma automática, **antes** de razonar conscientemente sobre ellos. Estos principios explican por qué algunos gráficos "se leen solos" y otros generan confusión, sin que el espectador sepa explicar por qué.

| Principio | Descripción | Aplicación en visualización de datos |
|---|---|---|
| **Proximidad** | Elementos cercanos entre sí se perciben como un grupo | Espaciar categorías distintas; agrupar barras relacionadas dentro de un mismo clúster |
| **Similitud** | Elementos con el mismo color, forma o tamaño se perciben como relacionados | Usar siempre el mismo color para la misma serie en todos los gráficos de un dashboard |
| **Cierre** | El cerebro completa formas incompletas por sí solo | Se puede omitir parte de un borde o eje sin perder legibilidad (menos tinta, misma información) |
| **Continuidad** | El ojo sigue líneas y patrones continuos con naturalidad | Ordenar categorías de forma lógica (alfabético, por valor, cronológico) para crear un "flujo" de lectura |
| **Conexión** | Elementos unidos por una línea se perciben más relacionados que los agrupados solo por color | Usar líneas conectoras en *slope charts* para comparar el mismo elemento en dos momentos |
| **Figura-fondo** | El cerebro separa activamente un "objeto" de su "fondo" | Garantizar contraste suficiente entre los datos (figura) y el fondo del gráfico |

> 🧠 **Idea clave:** la claridad no depende solo de "poner menos cosas" (eso es simplicidad) — depende de **organizar** las cosas que sí están, aprovechando cómo el cerebro humano ya está predispuesto a agruparlas.

```python
# ============================================================
# Demostracion: Proximidad y Similitud en accion
# Los mismos 30 puntos, organizados de dos formas distintas
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
fig.suptitle('Gestalt en la práctica: mismos datos, distinta legibilidad',
             fontsize=13, fontweight='bold')

# ---- Sin agrupar: dispersos y de un solo color ----
x_random = rng.uniform(0, 10, 30)
y_random = rng.uniform(0, 10, 30)
axes[0].scatter(x_random, y_random, s=90, color='#94a3b8', edgecolor='white')
axes[0].set_title('❌ Sin proximidad ni similitud\n(¿cuántos grupos hay aquí?)', fontsize=11)
axes[0].set_xticks([]); axes[0].set_yticks([])
for sp in axes[0].spines.values():
    sp.set_visible(False)

# ---- Agrupados por proximidad y coloreados por similitud ----
centros = [(2, 2), (7, 3), (4.5, 8)]
colores = ['#1e3a8a', '#10b981', '#f59e0b']
nombres = ['Grupo A', 'Grupo B', 'Grupo C']
for (cx, cy), color, nombre in zip(centros, colores, nombres):
    xs = cx + rng.normal(0, 0.6, 10)
    ys = cy + rng.normal(0, 0.6, 10)
    axes[1].scatter(xs, ys, s=90, color=color, edgecolor='white', label=nombre)
axes[1].set_title('✅ Con proximidad + similitud\n(3 grupos evidentes al instante)', fontsize=11)
axes[1].legend(frameon=False, loc='lower right')
axes[1].set_xticks([]); axes[1].set_yticks([])
for sp in axes[1].spines.values():
    sp.set_visible(False)

plt.tight_layout()
plt.show()

print("\n💡 El cerebro no necesita etiquetas ni cálculos para 'ver' los 3 grupos")
print("   del panel derecho: la proximidad espacial y el color ya hicieron el trabajo.")
```

## 3. Consistencia: Paletas de Color y Estilo Visual 🎨

El color es el canal visual más poderoso — y el más mal usado. Existen tres tipos de paletas para tres tipos de datos, y usarlas de forma **consistente** en todo un reporte o dashboard es lo que permite a la audiencia aprender el "lenguaje visual" una sola vez:

| Tipo de paleta | Para qué datos | Ejemplos |
|---|---|---|
| **Cualitativa** | Variables categóricas (sin orden) | `Set2`, `Paired`, `tab10` |
| **Secuencial** | Variables numéricas (bajo → alto) | `Blues`, `Greens`, `Viridis` |
| **Divergente** | Variables con punto medio neutro (negativo/positivo) | `RdBu`, `PiYG`, `coolwarm` |

### La regla de oro de la consistencia
Si "Norte" es azul en el primer gráfico de un reporte, **debe seguir siendo azul en todos los demás gráficos** del mismo reporte. Cambiar de paleta entre visualizaciones rompe la continuidad cognitiva que el lector ya construyó.

> ⚠️ **Regla de accesibilidad:** Usa paletas **daltonismo-seguras** (colorblind-friendly). El rojo-verde es inaccesible para ~8% de los hombres. Alternativa: azul-naranja (`RdBu`, `Set2`).

```python
# ============================================================
# Demostracion de Paletas de Color con Seaborn y Matplotlib
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

fig, axes = plt.subplots(3, 1, figsize=(12, 8))
fig.suptitle('Las Tres Familias de Paletas de Color en Visualización',
             fontsize=13, fontweight='bold')

# 1. Cualitativa
paleta_cual = sns.color_palette('Set2', 6)
for i, c in enumerate(paleta_cual):
    axes[0].bar(i, 1, color=c, width=0.8, edgecolor='white')
axes[0].set_title('1. Paleta Cualitativa (Set2) → Para categorías sin orden',
                  fontsize=10, loc='left')
axes[0].set_xticks(range(6))
axes[0].set_xticklabels(['Cat A', 'Cat B', 'Cat C', 'Cat D', 'Cat E', 'Cat F'])
axes[0].set_yticks([])
for sp in ['top', 'right', 'left']: axes[0].spines[sp].set_visible(False)

# 2. Secuencial
cmap_seq = plt.cm.Blues
vals_seq = np.linspace(0.3, 1.0, 8)
for i, v in enumerate(vals_seq):
    axes[1].bar(i, v * 100, color=cmap_seq(v), width=0.8, edgecolor='white')
axes[1].set_title('2. Paleta Secuencial (Blues) → Para valores numéricos bajo→alto',
                  fontsize=10, loc='left')
axes[1].set_xticks(range(8))
axes[1].set_xticklabels([f'{int(v*100)}' for v in vals_seq])
axes[1].set_ylabel('Valor', fontsize=9)
for sp in ['top', 'right']: axes[1].spines[sp].set_visible(False)

# 3. Divergente
cmap_div = plt.cm.RdBu
vals_div = np.linspace(0, 1, 9)
labels_div = ['-40', '-30', '-20', '-10', '0', '+10', '+20', '+30', '+40']
for i, (v, lbl) in enumerate(zip(vals_div, labels_div)):
    axes[2].bar(i, abs(float(lbl)) + 5, color=cmap_div(v), width=0.8, edgecolor='white')
axes[2].set_title('3. Paleta Divergente (RdBu) → Para valores con punto neutro (0)',
                  fontsize=10, loc='left')
axes[2].set_xticks(range(9))
axes[2].set_xticklabels(labels_div)
axes[2].set_xlabel('Variación respecto al promedio (%)', fontsize=9)
for sp in ['top', 'right']: axes[2].spines[sp].set_visible(False)

plt.tight_layout()
plt.show()

print("\n📊 En Power BI: Panel Formato → 'Colores de datos' → 'Formato condicional'")
print("   Permite aplicar escalas de color secuenciales o divergentes automaticamente.")
print("\n📈 En Tableau: Arrastrar variable numerica a 'Color' en la tarjeta Marks.")
print("   Clic en Color → 'Editar colores' → seleccionar paleta, y reutilizarla en todo el workbook.")
```

## Resumen y Puntos Clave 🎯

| Principio | Regla práctica |
|---|---|
| **Simplicidad** (Data-Ink Ratio) | Elimina todo elemento visual que no represente datos |
| **Claridad** (Gestalt) | Organiza mediante proximidad, similitud y figura-fondo para que el cerebro agrupe sin esfuerzo |
| **Consistencia** (Paletas de color) | La misma categoría debe usar el mismo color en todos los gráficos de un mismo reporte |
| **Regla general** | Un gráfico bien diseñado no necesita explicación oral — se explica solo |
