## ¿De qué va este notebook? 🎯

La analítica visual no se usa igual en una empresa, en un laboratorio y en una alcaldía. Aquí vemos **3 ejemplos con analogías simples** para entender cómo cambia el mismo "truco" (hacer gráficos que ayudan a decidir) según quién los use.

> 💡 **La idea central:** los mismos ingredientes (datos + gráfico + historia) se cocinan distinto según quién se va a comer el plato.

## Caso A: Empresa — "¿Dónde está la plata?" 🏪

Imagina que tienes una tienda con 3 secciones: tecnología, muebles y papelería. Vendes mucho de todo, pero **vender mucho no es lo mismo que ganar mucho**.

- Si regalas descuentos enormes, vendes más unidades pero **pierdes plata en cada una**.
- El gráfico de este caso es como el tablero de un carro: de un vistazo, el gerente ve qué sección "va bien" y cuál "va en rojo".

**La lección:** en el mundo empresarial, el gráfico tiene que responder RÁPIDO a la pregunta "¿dónde meto o saco plata?".

```python
# Demo simplificada: el efecto de "regalar" descuentos
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
descuento = np.random.choice([0, 10, 20, 30, 40, 50], 200)
ganancia = 25 - 0.6 * descuento + np.random.normal(0, 5, 200)

colores = ['#10b981' if g > 0 else '#ef4444' for g in ganancia]
plt.figure(figsize=(7, 4.5))
plt.scatter(descuento, ganancia, c=colores, alpha=0.6)
plt.axhline(0, color='black', linestyle='--')
plt.xlabel('Descuento regalado (%)')
plt.ylabel('Ganancia (%)')
plt.title('Entre mas descuento regalas, mas rojo (perdida) ves')
plt.show()

print("Como en el mercado: si regalas demasiado, terminas pagando tu por vender.")
```

## Caso B: Laboratorio — "¿De verdad funcionó o fue casualidad?" 🔬

Imagina que un científico prueba una vitamina nueva en 60 personas y compara con otras 60 que NO la tomaron (grupo control).

- Si solo mira el promedio, podría equivocarse: quizás la diferencia es pura casualidad.
- Por eso en ciencia **siempre se dibuja una "nube" alrededor de la línea** (el intervalo de confianza) — es como decir "estoy bastante seguro de que el valor real está en este rango, no solo en este punto exacto".

**La lección:** en ciencia no basta con "se ve diferente" — hay que demostrar que la diferencia es real y no ruido.

```python
# Demo simplificada: promedio con "nube" de incertidumbre
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(2)
semanas = np.arange(0, 9)
control = 14 - 0.15 * semanas
tratamiento = 14 - 0.75 * semanas
error = 0.8

plt.figure(figsize=(7, 4.5))
plt.plot(semanas, control, color='#94a3b8', marker='o', label='Sin vitamina (control)')
plt.fill_between(semanas, control - error, control + error, color='#94a3b8', alpha=0.2)
plt.plot(semanas, tratamiento, color='#1e3a8a', marker='o', label='Con vitamina (tratamiento)')
plt.fill_between(semanas, tratamiento - error, tratamiento + error, color='#1e3a8a', alpha=0.2)
plt.xlabel('Semana'); plt.ylabel('Nivel de inflamacion')
plt.title('Cuando las "nubes" ya no se tocan, la diferencia es creible')
plt.legend()
plt.show()

print("Si las nubes de colores dejan de tocarse, es mas probable que el efecto sea real.")
```

## Caso C: Alcaldía — "¿A quién le llega primero la ayuda?" 🏛️

Imagina que un alcalde tiene plata para mejorar el transporte público, pero no le alcanza para toda la ciudad. Tiene que elegir barrios.

- Dibuja un "mapa" sencillo: entre más rojo, peor acceso a transporte tiene ese barrio; entre más grande el círculo, más gente vive ahí.
- Así, cualquier ciudadano —sin saber de estadística— entiende por qué se eligieron esos barrios primero.

**La lección:** en el sector público, el gráfico tiene que poderse explicar en una reunión de barrio, no solo en una sala de expertos.

```python
# Demo simplificada: mapa de prioridad de inversion
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(3)
n = 8
x = np.random.uniform(0, 10, n)
y = np.random.uniform(0, 10, n)
acceso = np.random.uniform(10, 90, n)
poblacion = np.random.randint(10000, 80000, n)

plt.figure(figsize=(6.5, 5))
sc = plt.scatter(x, y, s=poblacion/300, c=acceso, cmap='RdYlGn', edgecolor='black')
plt.colorbar(sc, label='Acceso a transporte (%)')
plt.title('Rojo = necesita ayuda primero | Circulo grande = mucha gente')
plt.xticks([]); plt.yticks([])
plt.show()

print("Con un vistazo, cualquiera entiende que hay que empezar por los circulos rojos y grandes.")
```
