## 2. El Checklist Maestro de Diseño de Visualizaciones ✅

Úsalo antes de publicar **cualquier** visualización, sin importar la herramienta.

### 🎯 A. Propósito y audiencia (Módulo 01)
- [ ] ¿Tengo clara la pregunta que este gráfico responde?
- [ ] ¿Sé quién lo va a leer y qué decisión debe permitirle tomar?
- [ ] ¿Elegí el nivel de detalle adecuado para esa audiencia (ejecutivo vs. técnico)?

### 📐 B. Diseño y codificación visual (Módulo 02)
- [ ] ¿Elegí el tipo de gráfico según la tarea (comparar, componer, distribuir, relacionar, evolución)?
- [ ] ¿Maximicé el *data-ink ratio* (eliminé bordes, sombras y grillas innecesarias)?
- [ ] ¿Eliminé todo *chartjunk* (3D innecesario, íconos decorativos, efectos)?
- [ ] ¿El eje Y de mis gráficos de barras empieza en cero (a menos que esté justificado y advertido)?

### 🧠 C. Pensamiento crítico (Módulo 03)
- [ ] ¿Verifiqué la fuente y la fecha de los datos?
- [ ] ¿Evité el *cherry-picking* de período o de categorías favorables?
- [ ] ¿Distinguí claramente correlación de causalidad en el texto que acompaña el gráfico?
- [ ] ¿Revisé si algún sesgo cognitivo (anclaje, confirmación) influyó en cómo elegí presentar los datos?

### 🔤 D. Datos categóricos (Módulo 04)
- [ ] Si tengo pocas categorías (≤7), ¿usé barras ordenadas en vez de un pie chart?
- [ ] Si tengo jerarquías, ¿consideré un treemap en vez de múltiples pie charts?

### 🔢 E. Datos numéricos (Módulo 05)
- [ ] ¿Usé histograma/KDE para distribución, boxplot para comparar grupos, scatter para relación entre dos variables?
- [ ] ¿Mostré la dispersión/incertidumbre y no solo el promedio?

### 🚀 F. Técnicas avanzadas y narrativa (Módulo 06)
- [ ] Si el dato es geoespacial, ¿un mapa agrega valor real (y no solo estética)?
- [ ] ¿La interactividad (si existe) ayuda a explorar, sin sustituir una conclusión clara?
- [ ] ¿Mi visualización cuenta una historia con inicio (contexto), nudo (hallazgo) y desenlace (recomendación)?

### 🌐 G. Adaptación al contexto (Módulo 07)
- [ ] ¿Elegí la herramienta correcta (Python/Power BI/Tableau) según audiencia, urgencia y escala?
- [ ] ¿El nivel de rigor estadístico corresponde al sector (BI, investigación o política pública)?

```python
# ============================================================
# Makeover: aplicando el checklist a una visualizacion deficiente
# ============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(10)
trimestres = ['T1', 'T2', 'T3', 'T4']
ventas = [820, 845, 810, 910]

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
fig.suptitle('Makeover: del grafico "chartjunk" al grafico que respeta el checklist',
             fontsize=13, fontweight='bold')

# ANTES: eje que no empieza en cero, colores random, grilla pesada, sin foco
colores_antes = np.random.rand(4, 3)
axes[0].bar(trimestres, ventas, color=colores_antes, edgecolor='black', linewidth=2)
axes[0].set_ylim(780, 930)  # eje manipulado: exagera la diferencia
axes[0].grid(True, linewidth=1.5, color='gray')
axes[0].set_title('❌ ANTES: eje manipulado, colores sin significado,\nexagera una variacion pequeña')
for spine in axes[0].spines.values():
    spine.set_linewidth(2)

# DESPUES: eje en cero, un solo color con foco en el ultimo trimestre, sin grilla pesada
colores_despues = ['#94a3b8', '#94a3b8', '#94a3b8', '#1e3a8a']
axes[1].bar(trimestres, ventas, color=colores_despues)
axes[1].set_ylim(0, 1000)
for spine in ['top', 'right']:
    axes[1].spines[spine].set_visible(False)
axes[1].set_title('✅ DESPUES: eje en cero, color con foco,\ncontexto proporcional real')
for i, v in enumerate(ventas):
    axes[1].text(i, v + 15, str(v), ha='center', fontsize=9)

plt.tight_layout()
plt.show()

print("Checklist aplicado en la version 'DESPUES':")
print("  [x] Eje Y empieza en cero -> la variacion real (11%) ya no se ve como un salto dramatico")
print("  [x] Color con proposito -> resalta solo el trimestre relevante (T4)")
print("  [x] Sin chartjunk -> se eliminaron bordes gruesos, grilla pesada y colores aleatorios")
print("  [x] Etiquetas de datos -> el numero exacto queda visible, no hay que adivinar")
```

```python
# ============================================================
# Cuantificando el makeover: puntaje de checklist antes vs. despues
# ============================================================
import matplotlib.pyplot as plt
import numpy as np

criterios = ['Proposito\nclaro', 'Eje en\ncero', 'Color con\nsentido', 'Sin\nchartjunk',
             'Etiquetas\nde datos', 'Facil de\nleer sin ayuda']
puntaje_antes = [1, 0, 0, 0, 0, 0]
puntaje_despues = [1, 1, 1, 1, 1, 1]

x = np.arange(len(criterios))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.bar(x - width/2, puntaje_antes, width, label='Version ANTES', color='#ef4444')
ax.bar(x + width/2, puntaje_despues, width, label='Version DESPUES', color='#10b981')
ax.set_xticks(x)
ax.set_xticklabels(criterios, fontsize=9)
ax.set_yticks([0, 1])
ax.set_yticklabels(['No cumple', 'Cumple'])
ax.set_title('Checklist aplicado al makeover: de 1/6 a 6/6 criterios', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()

print(f"Puntaje ANTES: {sum(puntaje_antes)}/6 criterios cumplidos")
print(f"Puntaje DESPUES: {sum(puntaje_despues)}/6 criterios cumplidos")
print("-> Aplicar el checklist de forma sistematica es lo que separa un grafico amateur de uno profesional.")
```
