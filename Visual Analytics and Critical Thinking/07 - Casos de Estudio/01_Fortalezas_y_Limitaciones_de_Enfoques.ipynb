## 1. Dos filosofías de visualización 🧩

| | **Código (Python)** | **BI de bajo código (Tableau / Power BI)** |
|---|---|---|
| **Interfaz** | Scripts, notebooks | Arrastrar y soltar (*drag-and-drop*) |
| **Audiencia natural** | Data scientists, ingenieros | Analistas de negocio, gerencia |
| **Reproducibilidad** | Alta (código versionable en Git) | Media (archivos binarios, versionado limitado) |
| **Curva de aprendizaje** | Alta al inicio | Baja al inicio, media para lógica avanzada (DAX/Calculated Fields) |
| **Personalización visual** | Ilimitada (cualquier librería) | Limitada a lo que la herramienta permite |
| **Automatización / pipelines** | Nativa (cron, Airflow, CI/CD) | Requiere conectores/gateways adicionales |
| **Interactividad out-of-the-box** | Requiere Plotly/Dash/Streamlit | Nativa e inmediata |
| **Colaboración con negocio** | Requiere traducir hallazgos | Nativa, autoservicio para el usuario de negocio |
| **Costo** | Gratuito (open source) | Licencias por usuario/mes |
| **Escalabilidad de datos** | Alta con el motor correcto (Spark, DuckDB) | Alta pero depende del backend conectado |

No existe un "ganador" universal: la elección correcta depende del **contexto de uso**, no de cuál herramienta es "mejor" en abstracto.

````python
# ============================================================
# Comparacion cuantitativa ilustrativa: Codigo vs BI de bajo codigo
# (puntajes 0-10, ilustrativos, con fines pedagogicos)
# ============================================================
import numpy as np
import matplotlib.pyplot as plt

dimensiones = ['Reproducibilidad', 'Personalizacion', 'Velocidad de\ndesarrollo inicial',
               'Interactividad\nnativa', 'Colaboracion\ncon negocio', 'Automatizacion',
               'Escalabilidad', 'Costo\n(10=mas barato)']

puntaje_python = [9, 10, 4, 5, 4, 9, 8, 10]
puntaje_bi     = [5, 5, 9, 9, 9, 6, 7, 5]

angulos = np.linspace(0, 2 * np.pi, len(dimensiones), endpoint=False).tolist()
puntaje_python += puntaje_python[:1]
puntaje_bi += puntaje_bi[:1]
angulos += angulos[:1]

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
ax.plot(angulos, puntaje_python, color='#1e3a8a', linewidth=2, label='Python (codigo)')
ax.fill(angulos, puntaje_python, color='#1e3a8a', alpha=0.15)
ax.plot(angulos, puntaje_bi, color='#f59e0b', linewidth=2, label='BI bajo codigo (Tableau/Power BI)')
ax.fill(angulos, puntaje_bi, color='#f59e0b', alpha=0.15)

ax.set_xticks(angulos[:-1])
ax.set_xticklabels(dimensiones, fontsize=9)
ax.set_yticks([2, 4, 6, 8, 10])
ax.set_title('Comparacion ilustrativa de enfoques por dimension', fontsize=13, fontweight='bold', pad=20)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.tight_layout()
plt.show()

print("Nota: puntajes ilustrativos para fines didacticos, no una medicion de mercado.")
print("Python domina en reproducibilidad, personalizacion y automatizacion.")
print("BI bajo codigo domina en velocidad inicial, interactividad nativa y colaboracion con negocio.")
````

## 2. Fortalezas y limitaciones según el contexto 🧭

### a) Equipo técnico vs. equipo de negocio

| Contexto | Enfoque recomendado | Por qué |
|---|---|---|
| Equipo de data science/ingeniería, con Git y CI/CD | **Python** | Se integra al flujo de trabajo existente, versiona el análisis junto al código del modelo |
| Equipo comercial/marketing sin conocimientos de programación | **BI de bajo código** | Autoservicio: ellos mismos arman y filtran sus reportes sin depender de TI |

### b) Tiempo real vs. reporte estático

| Contexto | Enfoque recomendado | Por qué |
|---|---|---|
| Monitoreo operativo en vivo (ventas del día, sensores IoT) | **BI de bajo código** (con conexión directa/DirectQuery) | Conectores nativos a bases de datos en streaming, refresco automático |
| Reporte académico/paper reproducible | **Python** | El notebook documenta exactamente cómo se generó cada cifra y gráfico |

### c) Escala de datos

| Contexto | Enfoque recomendado | Por qué |
|---|---|---|
| Terabytes de datos, requiere procesamiento distribuido | **Python** (+ Spark/Dask/DuckDB) | Control total del pipeline y del motor de cómputo |
| Miles a millones de filas ya modeladas en un data warehouse | **BI de bajo código** | El motor tabular de Power BI/Tableau ya está optimizado para ese volumen sin programar nada |

> ⚠️ **Limitación importante a reconocer:** un analista que solo sabe BI de bajo código queda limitado por lo que la herramienta permite (p. ej. una visualización estadística poco común). Un analista que solo sabe Python puede tardar más en entregar un dashboard interactivo simple que un usuario de negocio necesita hoy mismo. El perfil más valioso domina **ambos**.

````python
# ============================================================
# Mismo insight, dos tiempos de entrega: ilustracion del trade-off
# ============================================================
import matplotlib.pyplot as plt
import numpy as np

tareas = ['Explorar datos\nnuevos', 'Grafico estandar\n(barras/lineas)',
          'Dashboard\ninteractivo simple', 'Modelo estadistico\navanzado',
          'Automatizar\nreporte recurrente', 'Compartir con\nusuario de negocio']

tiempo_python_horas = [1.5, 0.5, 4, 1, 0.5, 2]
tiempo_bi_horas =     [3, 0.3, 0.5, 6, 0.2, 0.1]

x = np.arange(len(tareas))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 5))
ax.bar(x - width/2, tiempo_python_horas, width, label='Python', color='#1e3a8a')
ax.bar(x + width/2, tiempo_bi_horas, width, label='BI bajo codigo', color='#f59e0b')
ax.set_xticks(x)
ax.set_xticklabels(tareas, fontsize=9)
ax.set_ylabel('Horas estimadas para completar la tarea')
ax.set_title('Tiempo estimado por tarea segun el enfoque (ilustrativo)', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()

print("Ningun enfoque gana en todas las tareas: por eso el enfoque 'triangulo'")
print("(Python + Power BI + Tableau) del curso busca que domines ambos mundos.")
````

## 3. Marco de decisión rápido 🗺️

```
¿Necesitas reproducibilidad total y control del pipeline?
        │
   Sí ──┴── No
   │          │
 PYTHON    ¿La audiencia final es de negocio y necesita autoservicio?
              │
        Sí ──┴── No
        │          │
   BI BAJO       ¿Es un análisis exploratorio puntual o investigación?
   CÓDIGO           │
                Sí ──┴── No
                │          │
             PYTHON    BI BAJO CÓDIGO
                        (reporte operativo)
```

Este marco es una guía inicial, no una receta rígida: en la práctica, los mejores equipos de analítica **combinan ambos** — Python para el modelado y la preparación de datos, BI de bajo código para la distribución final a la organización.

## Resumen y Puntos Clave 🎯

- Ni Python ni las herramientas BI de bajo código son "superiores" de forma absoluta — cada uno resuelve mejor **problemas distintos**.
- **Python** gana en reproducibilidad, personalización, automatización y escalabilidad técnica.
- **BI de bajo código** gana en velocidad de entrega inicial, interactividad nativa y autoservicio para usuarios de negocio.
- La decisión correcta depende del **contexto**: tipo de equipo, urgencia de tiempo real, escala de datos y a quién va dirigido el resultado.
- Los equipos más maduros en analítica usan el **enfoque triángulo** (Python ↔ Power BI ↔ Tableau) combinando fortalezas de cada uno según la etapa del proyecto.
