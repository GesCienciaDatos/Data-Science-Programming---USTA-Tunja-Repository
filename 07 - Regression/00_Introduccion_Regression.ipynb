## 1. ¿Qué es el Aprendizaje Supervisado y qué es la Regresión? 🧠

En el **Aprendizaje Supervisado (*Supervised Learning*)**, los algoritmos aprenden a partir de datos históricos etiquetados donde cada observación consiste en un conjunto de características de entrada $\mathbf{x} = [x_1, x_2, \dots, x_p]^T$ y una respuesta correcta conocida $y$:

$$\mathcal{D} = \{(\mathbf{x}_1, y_1), (\mathbf{x}_2, y_2), \dots, (\mathbf{x}_n, y_n)\}$$

El objetivo es aprender una función de mapeo $\hat{y} = f(\mathbf{x})$ que generalice con alta precisión sobre nuevos datos no observados.

```
                           ┌──────────────────────────────┐
                           │   Aprendizaje Supervisado    │
                           └──────────────┬───────────────┘
                                          │
                  ┌───────────────────────┴───────────────────────┐
                  ▼                                               ▼
      ┌───────────────────────┐                       ┌───────────────────────┐
      │  Regresión (Continuo) │                       │ Clasificación (Categ.)│
      └───────────────────────┘                       └───────────────────────┘
        • Precio de inmuebles ($)                       • Detección de Fraude (0 / 1)
        • Temperatura ambiental (°C)                    • Diagnóstico Médico (A / B)
        • Demanda de energía (MWh)                      • Cancelación de Clientes (Churn)
```

> 📌 **Definición de Regresión:**  
> La **Regresión** es la tarea específica de predecir un valor numérico cuantitativo continuo ($y \in \mathbb{R}$). Implica ajustar un modelo matemático que describe cómo cambia el valor esperado de la variable objetivo en función de las variables predictoras:
> $$\hat{y} = \mathbb{E}[Y \mid \mathbf{X} = \mathbf{x}]$$

## 2. ¿Por qué utilizar Modelos de Regresión Lineal? 💡

A pesar del surgimiento de algoritmos complejos como Redes Neuronales Profundas o Ensambles de Árboles (XGBoost, LightGBM), los modelos de **Regresión Lineal** siguen siendo la piedra angular de la Ciencia de Datos aplicada por cuatro razones fundamentales:

1. **Interpretabilidad Directa y Transparencia:** Cada coeficiente $\beta_j$ tiene un significado físico o de negocio exacto (cuánto cambia $y$ por cada unidad de cambio en $x_j$, manteniendo constantes las demás variables).
2. **Eficiencia Computacional y Escalabilidad:** El ajuste por Mínimos Cuadrados Ordinarios (OLS) o Gradiente Descendente se ejecuta en fracciones de segundo incluso sobre millones de registros.
3. **Inferencia Estadística Rigurosa:** Permite realizar pruebas de hipótesis ($p$-valores, intervalos de confianza del 95% y significancia $t$/$F$) para validar si una relación es genuina o producto del azar.
4. **El Modelo Base Indispensable (*Baseline Model*):** En todo proyecto profesional de Machine Learning, el primer modelo a entrenar debe ser una regresión lineal simple/múltiple para establecer el estándar mínimo de desempeño que modelos más complejos deben superar.
