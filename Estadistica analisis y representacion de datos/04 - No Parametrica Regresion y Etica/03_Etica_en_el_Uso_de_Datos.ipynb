## 1. Marco Ético y Privacidad: de lo Personal a lo Anonimizado 🛡️

**Principios.** El *Informe Belmont* (1979) resume tres: **respeto por las personas** (consentimiento informado, autonomía), **beneficencia** (maximizar beneficios, minimizar daños) y **justicia** (quién carga con los riesgos y quién recibe los beneficios). Las guías de práctica estadística de las asociaciones profesionales añaden **integridad** (no manipular resultados), **transparencia** y **reconocimiento de la incertidumbre**. En Colombia la **Ley 1581 de 2012** (*Habeas Data*) regula el tratamiento de datos personales, con principios como **finalidad**, **libertad** (consentimiento), **veracidad**, **transparencia**, **acceso restringido** y **seguridad**, y reglas más estrictas para los **datos sensibles** (por ejemplo, salud, origen étnico, orientación política o datos biométricos); la autoridad de control es la Superintendencia de Industria y Comercio *(a nivel conceptual; verifica el texto vigente y consulta a un profesional del derecho para casos reales)*.

| Término | Significado | ¿Sigue siendo dato personal? |
|---|---|---|
| **Dato personal** | Información que identifica o hace identificable a una persona | Sí |
| **Dato sensible** | Afecta la intimidad o puede dar lugar a discriminación (salud, etnia, ideología...) | Sí, con protección reforzada |
| **Seudonimización** | Se reemplaza el identificador por un código; **existe una forma de revertirlo** (tabla, clave) | **Sí** |
| **Anonimización** | No es razonablemente posible reidentificar, ni siquiera combinando con otras fuentes | No (si es efectiva) |

**El error clásico.** Quitar nombres y cédulas **no** anonimiza. (a) Un *hash* sin secreto de un identificador de espacio pequeño se invierte probando todas las posibilidades (**ataque de diccionario**). (b) Combinaciones de **cuasi-identificadores** (edad, sexo, municipio, ocupación) que individualmente no identifican a nadie pueden **singularizar** a una persona, y cruzadas con otra fuente pública (**ataque de vinculación**) revelan quién es y su dato sensible. El **$k$-anonimato** (Sweeney, 2002) exige que cada combinación de cuasi-identificadores aparezca en **al menos $k$ registros**, de modo que cada persona sea indistinguible de otras $k-1$; se logra **generalizando** (edad exacta $\to$ rango) y **suprimiendo** registros raros. Pero tiene una falla: si todos los de un grupo comparten el mismo valor sensible (**ataque de homogeneidad**), basta saber el grupo. La **$l$-diversidad** pide al menos $l$ valores sensibles distintos por grupo. Probamos todo con un registro de salud **100 % sintético**.

````python
# ============================================================
# 1. Hash vs HMAC, cuasi-identificadores, ataque de vinculacion y k-anonimato
# ============================================================
import hashlib, hmac, time

def huella(documento, sal=""):
    return hashlib.sha256((sal + str(documento)).encode()).hexdigest()

# --- (a) Un hash de un identificador de espacio pequeno se invierte probando todos los candidatos
rg = np.random.default_rng(1)
docs = rg.choice(np.arange(100_000, 1_000_000), 5, replace=False)       # numeros de 6 digitos SINTETICOS (no son documentos reales)
publicados = [huella(d) for d in docs]
t0 = time.perf_counter()
diccionario = {huella(d): d for d in range(100_000, 1_000_000)}         # ataque de diccionario: 900 000 candidatos
recuperados = [diccionario[h] for h in publicados]
seg = time.perf_counter() - t0
assert recuperados == list(docs)
clave = rg.bytes(16)                                                    # secreto que solo conoce la organizacion
con_clave = [hmac.new(clave, str(d).encode(), "sha256").hexdigest() for d in docs]
assert not any(h in diccionario for h in con_clave)                     # sin la clave, el diccionario no sirve
print(f"Hash SHA-256 sin secreto: se recuperaron {len(recuperados)} de {len(publicados)} identificadores en {seg:.1f} s (900 000 candidatos).")
print(f"HMAC con clave secreta: {sum(h in diccionario for h in con_clave)} coincidencias con el diccionario. (Sigue siendo SEUDONIMIZACION: quien tenga la clave puede revertirlo.)")

# --- (b) Registro de salud SINTETICO (5 000 personas). Cuasi-identificadores: municipio, sexo, edad, ocupacion. Sensible: diagnostico.
def genera_registros(n=5000, seed=7):
    rg = np.random.default_rng(seed)
    w = np.array([20, 15, 12, 10, 9, 8, 7, 5, 4, 4, 3, 3], float)
    municipio = rg.choice([f"M{i:02d}" for i in range(1, 13)], n, p=w / w.sum())
    edad = np.clip(rg.normal(48, 16, n).round(), 18, 85).astype(int)
    sexo = rg.choice(["F", "M"], n)
    ocupacion = rg.choice(["agricultor", "ganadero", "comerciante", "docente", "transportador", "otro"], n, p=[0.40, 0.15, 0.15, 0.10, 0.10, 0.10])
    u = rg.random(n); p_h, p_d, p_a = 0.02 + 0.004 * (edad - 18), 0.01 + 0.002 * (edad - 18), 0.05
    diag = np.select([u < p_h, u < p_h + p_d, u < p_h + p_d + p_a], ["hipertension", "diabetes", "asma"], "ninguno")
    return pd.DataFrame({"id": [f"P{i:04d}" for i in range(n)], "municipio": municipio, "sexo": sexo, "edad": edad, "ocupacion": ocupacion, "diagnostico": diag})

registros = genera_registros()
QI = ["municipio", "sexo", "edad", "ocupacion"]

def k_anonimato(df, qi):
    return int(df.groupby(qi, observed=True).size().min())

def frac_unicos(df, qi):
    tam = df.groupby(qi, observed=True)[qi[0]].transform("size")
    return float((tam == 1).mean())

filas = [[", ".join(q), k_anonimato(registros, q), frac_unicos(registros, q)] for q in (["municipio"], ["municipio", "sexo"], ["municipio", "sexo", "edad"], QI)]
print(pd.DataFrame(filas, columns=["cuasi-identificadores", "k (minimo grupo)", "% de personas UNICAS"]).set_index("cuasi-identificadores").assign(**{"% de personas UNICAS": lambda d: (100 * d["% de personas UNICAS"]).round(1)}).to_string())

# --- (c) Ataque de vinculacion: un 'padron publico' con nombres (70 % de las personas) cruzado con la tabla de salud 'anonima'
rg = np.random.default_rng(3)
padron = registros.drop(columns="diagnostico").sample(frac=0.70, random_state=3)
cruce = registros[["id"] + QI].merge(padron.rename(columns={"id": "id_padron"}), on=QI, how="left")
candidatos = cruce.groupby("id")["id_padron"].nunique()                                   # cuantas personas del padron comparten sus cuasi-identificadores
unico = cruce[cruce["id"].map(candidatos) == 1]
reidentificadas = (unico["id"] == unico["id_padron"]).sum()
print(f"\nAtaque de vinculacion: {reidentificadas} de {len(registros)} personas ({reidentificadas / len(registros):.1%}) quedan identificadas con certeza, con su diagnostico, aunque la tabla de salud NO tenia nombres.")
assert reidentificadas / len(registros) > 0.2

# --- (d) Generalizacion y supresion para lograr k-anonimato
def anonimiza(df, k=5, ancho_edad=10, region=True, grupo_ocup=True):
    g = df.copy()
    g["edad"] = (g["edad"] // ancho_edad * ancho_edad).astype(str) + "-" + (g["edad"] // ancho_edad * ancho_edad + ancho_edad - 1).astype(str)
    if region:
        g["municipio"] = g["municipio"].map(lambda m: "R" + str((int(m[1:]) - 1) // 4 + 1))
    if grupo_ocup:
        g["ocupacion"] = g["ocupacion"].map(lambda o: "agropecuario" if o in ("agricultor", "ganadero") else "servicios")
    tam = g.groupby(QI)["id"].transform("size")
    publicado = g[tam >= k]
    return dict(k_final=k_anonimato(publicado, QI), suprimido=1 - len(publicado) / len(df), clases=publicado.groupby(QI).ngroups, publicado=publicado)

configs = {"sin generalizar": dict(ancho_edad=1, region=False, grupo_ocup=False), "edad 5 + region": dict(ancho_edad=5, region=True, grupo_ocup=False),
           "edad 10 + region + ocup.": dict(ancho_edad=10, region=True, grupo_ocup=True), "edad 20 + region + ocup.": dict(ancho_edad=20, region=True, grupo_ocup=True)}
res = {nom: anonimiza(registros, k=5, **cfg) for nom, cfg in configs.items()}
tab = pd.DataFrame({nom: {"k final": r["k_final"], "% suprimido": round(100 * r["suprimido"], 1), "clases": r["clases"]} for nom, r in res.items()}).T
print("\nGeneralizacion + supresion para k = 5:"); print(tab.to_string())
assert all(r["k_final"] >= 5 for r in res.values())

# --- (e) Ataque de homogeneidad: k-anonimato no basta; l-diversidad
pub = res["edad 10 + region + ocup."]["publicado"]
div = pub.groupby(QI)["diagnostico"].nunique()
print(f"\nCon k >= 5, hay {int((div == 1).sum())} de {len(div)} grupos donde TODOS comparten el mismo diagnostico (l = 1): saber el grupo basta para conocer el dato sensible.")
print("Diversidad minima l:", int(div.min()), "| mediana de valores distintos por grupo:", int(div.median()))
# utilidad: la prevalencia de diabetes por banda de edad casi no cambia
prev_orig = registros.assign(banda=(registros["edad"] // 10 * 10).astype(str)).groupby("banda")["diagnostico"].apply(lambda s: (s == "diabetes").mean())
prev_anon = pub.groupby("edad")["diagnostico"].apply(lambda s: (s == "diabetes").mean())
prev_anon.index = prev_anon.index.str.split("-").str[0]
print(f"Diferencia absoluta maxima en la prevalencia de diabetes por decada (original vs anonimizado): {(prev_orig - prev_anon).abs().max():.3f}")
````

**Lectura.** (1) **Hash:** con un identificador de 6 dígitos (900 000 candidatos) el ataque de diccionario recuperó **5 de 5** en **1.8 s**; con **HMAC** y clave secreta, 0 coincidencias. Seguridad no es anonimato: quien tenga la clave revierte todo, por eso **sigue siendo dato personal**. (2) **Cuasi-identificadores:** por separado no identifican (municipio: $k=137$; municipio y sexo: $k=64$, 0 % de personas únicas), pero con la **edad exacta** aparecen personas únicas (6.2 %, $k=1$) y con la **ocupación** el **41.5 %** de las personas es única. (3) **Vinculación:** con un padrón público que cubre el 70 % de las personas, **1 802 de 5 000 (36.0 %)** quedan identificadas **con certeza**, con su diagnóstico, aunque la tabla de salud no tenía nombres. (4) **$k$-anonimato con $k=5$:** sin generalizar habría que suprimir **90.8 %** de los registros (inviable); agrupar la edad de 5 en 5 años y los municipios en 3 regiones suprime **9.0 %** y deja 307 clases; con edad por décadas y ocupación agrupada se suprime **0.3 %** pero quedan 92 clases, y con rangos de 20 años, 56. Más generalización $\Rightarrow$ más privacidad y menos detalle. (5) **Límites:** con $k\ge5$ hay **1 grupo de 92** en el que todos comparten el diagnóstico ($l=1$), así que saber el grupo revela el dato sensible; la mediana es de 4 diagnósticos distintos por grupo. La utilidad se conserva: la prevalencia de diabetes por década cambia a lo sumo **0.009**. Además, la Práctica 1 muestra que, con tolerancia de 10 % de supresión, la mejor opción conserva la edad en rangos de 5 años y las regiones, sin agrupar ocupaciones.

## 2. Privacidad Diferencial: el Mecanismo de Laplace 🔊

**Idea.** En vez de proteger la tabla, se protege la **respuesta**: se añade **ruido calibrado** de modo que el resultado casi no dependa de que **una persona cualquiera** esté o no en los datos. Un mecanismo aleatorio $M$ es **$\varepsilon$-diferencialmente privado** si para todo par de bases **vecinas** $D,D'$ (que difieren en una persona) y todo conjunto de salidas $S$,

$$P\big(M(D)\in S\big)\le e^{\varepsilon}\,P\big(M(D')\in S\big).$$

$\varepsilon$ pequeño $\Rightarrow$ más privacidad (y más ruido). **Mecanismo de Laplace:** para una consulta numérica $f$ con **sensibilidad** $\Delta f=\max_{D\sim D'}\lvert f(D)-f(D')\rvert$, se publica

$$M(D)=f(D)+\operatorname{Lap}\!\left(0,\ \frac{\Delta f}{\varepsilon}\right),\qquad \text{densidad }\ \frac{1}{2b}\,e^{-\lvert x\rvert/b},\quad b=\frac{\Delta f}{\varepsilon},\quad E\lvert\operatorname{Lap}(b)\rvert=b .$$

Un **conteo** tiene $\Delta f=1$; una **media** de $n$ valores acotados en $[a,b_{\max}]$ (hay que **recortar** los datos) tiene $\Delta f=(b_{\max}-a)/n$: **a mayor $n$, menos ruido relativo**. **Propiedades:** *postprocesamiento* (cualquier cálculo sobre la salida privada sigue siendo $\varepsilon$-privado), *composición secuencial* (consultas sucesivas **suman** sus $\varepsilon$: el «presupuesto de privacidad») y *composición paralela* (consultas sobre grupos disjuntos de personas cuestan el máximo, no la suma). **Respuesta aleatorizada** (Warner, 1965; privacidad **local**: el ruido se añade en cada persona, antes de llegar al analista): cada persona dice la verdad con probabilidad $p=e^{\varepsilon}/(1+e^{\varepsilon})$ y miente con $1-p$, lo que cumple $\varepsilon$-DP, y la prevalencia se recupera con el estimador **insesgado** $\hat\pi=(q-(1-p))/(2p-1)$, donde $q$ es la fracción observada de «sí».

````python
# ============================================================
# 2. Mecanismo de Laplace desde cero, compromiso privacidad-utilidad, composicion y respuesta aleatorizada
# ============================================================
def laplace(valor, sens, eps, rg):
    """Mecanismo de Laplace: valor + Lap(0, sens/eps)."""
    return valor + rg.laplace(0.0, sens / eps, size=np.shape(valor))

# Consulta: cuantas personas con diabetes hay en el municipio M03 (conteo, sensibilidad 1)
c = int(((registros["municipio"] == "M03") & (registros["diagnostico"] == "diabetes")).sum())
print(f"Conteo real (solo lo conoce la organizacion): {c}")

# (a) La garantia: cociente de densidades entre bases vecinas (cuenta c y c + 1) acotado por e^eps
eps = 0.5
x = c + np.linspace(-40, 40, 4001)
cociente = stats.laplace.pdf(x, loc=c, scale=1 / eps) / stats.laplace.pdf(x, loc=c + 1, scale=1 / eps)
assert cociente.max() <= np.exp(eps) + 1e-9 and cociente.min() >= np.exp(-eps) - 1e-9
rg = np.random.default_rng(11)
sal_D, sal_Dp = laplace(np.full(400_000, c), 1, eps, rg), laplace(np.full(400_000, c + 1), 1, eps, rg)
razon = (sal_D <= c).mean() / (sal_Dp <= c).mean()                       # evento {salida <= c}: alli se alcanza el limite e^eps
print(f"eps = {eps}: cociente maximo de densidades = {cociente.max():.4f} (e^eps = {np.exp(eps):.4f}); razon empirica P(S|D)/P(S|D') = {razon:.3f}")
assert abs(razon - np.exp(eps)) / np.exp(eps) < 0.03

# (b) Compromiso privacidad-utilidad: error absoluto medio = sensibilidad / eps
epsilons = np.array([0.1, 0.25, 0.5, 1, 2, 5])
mae = np.array([np.abs(laplace(np.full(100_000, c), 1, e, rg) - c).mean() for e in epsilons])
assert np.allclose(mae, 1 / epsilons, rtol=0.03)
print(pd.DataFrame({"eps": epsilons, "error medio simulado": mae, "teorico 1/eps": 1 / epsilons}).round(3).to_string(index=False))

# (c) El ruido pesa mas en grupos pequenos; y la media privada mejora con n
c_peq = int(((registros["municipio"] == "M12") & (registros["diagnostico"] == "diabetes")).sum())
print(f"\nError relativo esperado con eps = 0.5 (ruido medio = 2): grupo pequeno (conteo {c_peq}) = {2 / max(c_peq, 1):.0%} | grupo grande (conteo {int((registros['diagnostico'] == 'diabetes').sum())}) = {2 / (registros['diagnostico'] == 'diabetes').sum():.1%}")
def media_privada(v, a, b, eps, rg):
    v = np.clip(v, a, b)
    return laplace(v.mean(), (b - a) / len(v), eps, rg)
edad = registros["edad"].to_numpy(float)
for n_ in (50, 500, 5000):
    errs = [abs(media_privada(edad[:n_], 18, 85, 1.0, rg) - np.clip(edad[:n_], 18, 85).mean()) for _ in range(5000)]
    print(f"Media de edad con eps = 1, n = {n_:4d}: sensibilidad {(85 - 18) / n_:.4f} | error medio = {np.mean(errs):.3f} anos")

# (d) Composicion: repetir la misma consulta con ruido nuevo y promediar DESTRUYE la privacidad (el presupuesto se suma)
eps_q = 0.1
filas = []
for m in (1, 10, 100, 1000):
    prom = laplace(np.full((2000, m), c), 1, eps_q, rg).mean(axis=1)
    filas.append((m, m * eps_q, np.abs(prom - c).mean()))
comp = pd.DataFrame(filas, columns=["consultas iguales", "eps total gastado", "error medio del promedio"]).set_index("consultas iguales")
print("\nComposicion (cada consulta con eps = 0.1):"); print(comp.round(3).to_string())
assert comp["error medio del promedio"].iloc[-1] < 0.2 * comp["error medio del promedio"].iloc[0]

# (e) Respuesta aleatorizada (privacidad local)
eps_rr = np.log(3); p_ver = np.exp(eps_rr) / (1 + np.exp(eps_rr))                 # p = 0.75
assert abs(p_ver / (1 - p_ver) - np.exp(eps_rr)) < 1e-12                           # razon de probabilidades de respuesta = e^eps
pi_real = (registros["diagnostico"] == "diabetes").mean(); n_ = len(registros)
est = np.empty(2000); rg2 = np.random.default_rng(12)
for i in range(2000):
    veraz = rg2.random(n_) < p_ver
    resp = np.where(veraz, registros["diagnostico"].eq("diabetes").to_numpy(), ~registros["diagnostico"].eq("diabetes").to_numpy())
    est[i] = (resp.mean() - (1 - p_ver)) / (2 * p_ver - 1)
sd_teo = np.sqrt((pi_real * p_ver + (1 - pi_real) * (1 - p_ver)) * (1 - (pi_real * p_ver + (1 - pi_real) * (1 - p_ver))) / n_) / (2 * p_ver - 1)
print(f"\nRespuesta aleatorizada (p = {p_ver:.2f}, eps = ln 3 = {eps_rr:.2f}): prevalencia real = {pi_real:.4f} | estimador: media = {est.mean():.4f}, desv. est. = {est.std():.4f} (teorica {sd_teo:.4f})")
assert abs(est.mean() - pi_real) < 0.002 and abs(est.std() - sd_teo) / sd_teo < 0.06

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].plot(x, stats.laplace.pdf(x, c, 1 / eps), color=PALETA[0], label="base D (conteo c)"); ax[0].plot(x, stats.laplace.pdf(x, c + 1, 1 / eps), color=PALETA[1], label="base D' (conteo c+1)")
ax[0].set(title=f"Salidas casi indistinguibles (eps = {eps})", xlim=(c - 15, c + 15), xlabel="valor publicado"); ax[0].legend()
ax[1].loglog(epsilons, mae, "o-", color=PALETA[2], label="simulado"); ax[1].loglog(epsilons, 1 / epsilons, "k--", label="1/eps"); ax[1].set(title="Mas privacidad = mas error", xlabel="eps", ylabel="error absoluto medio"); ax[1].legend()
ax[2].loglog(comp.index, comp["error medio del promedio"], "o-", color=PALETA[3], label="error del promedio"); ax[2].set(title="Repetir la consulta: el ruido se promedia, el presupuesto no", xlabel="consultas iguales", ylabel="error medio"); ax[2].legend()
plt.tight_layout(); plt.show()
````

## 3. *P-hacking* y Comparaciones Múltiples 🎣

**El valor $p$ correcto.** Es la probabilidad, **si $H_0$ y todos los supuestos son ciertos**, de obtener un resultado al menos tan extremo como el observado. **No** es la probabilidad de que $H_0$ sea cierta ni de que el resultado «se replique». Esa garantía vale para **una** prueba planeada **de antemano**. Cuando se hacen muchas, o se decide qué probar **después de mirar los datos**, la tasa de falsos positivos se dispara:

| Práctica | Qué ocurre |
|---|---|
| **Comparaciones múltiples** | Con $m$ pruebas independientes y $H_0$ cierta en todas, $P(\text{al menos un } p<0.05)=1-0.95^{m}$ ($m=20\Rightarrow64\,\%$) |
| **Subgrupos, variables y transformaciones a la carta** («jardín de los senderos que se bifurcan») | Se reporta el análisis que «salió» sin contar los que no |
| **Detención opcional** | Mirar los datos cada cierto tiempo y parar al encontrar $p<0.05$ infla el error tipo I muy por encima del 5 % |

**Qué controlar.** *FWER* (probabilidad de **algún** falso positivo) y *FDR* (**proporción esperada** de falsos entre los descubrimientos). Con $p_{(1)}\le\dots\le p_{(m)}$:

| Procedimiento | Regla | Controla |
|---|---|---|
| **Bonferroni** | rechazar si $p_i\le\alpha/m$ | FWER |
| **Holm** (paso a paso) | rechazar $p_{(j)}$ mientras $p_{(j)}\le\alpha/(m-j+1)$ | FWER, más potente que Bonferroni |
| **Benjamini-Hochberg** | rechazar los $p_{(j)}$ con $j\le k$, donde $k=\max\{j:\ p_{(j)}\le \tfrac{j}{m}\,q\}$ | FDR $\le q$ (con independencia o dependencia positiva) |

**Por qué «significativo» no es «cierto».** Si una fracción $\pi$ de las hipótesis son realmente ciertas, la prueba tiene potencia $1-\beta$ y se usa nivel $\alpha$, la fracción de resultados significativos que son **falsos** es $\dfrac{\alpha(1-\pi)}{\alpha(1-\pi)+(1-\beta)\pi}$ (el argumento de Ioannidis, 2005). Además, si solo se reportan los estudios significativos, los efectos publicados se **exageran** (la «maldición del ganador»): con baja potencia, para pasar el umbral el efecto **estimado** tiene que ser, por azar, mucho mayor que el verdadero.

````python
# ============================================================
# 3a. Comparaciones multiples, correcciones desde cero y detencion opcional
# ============================================================
from statsmodels.stats.multitest import multipletests

def prueba_t(X, Y):
    """p-valores bilaterales y d de Cohen de dos muestras independientes (filas = experimentos)."""
    n1, n2 = X.shape[1], Y.shape[1]
    m1, m2, v1, v2 = X.mean(1), Y.mean(1), X.var(1, ddof=1), Y.var(1, ddof=1)
    sp = np.sqrt(((n1 - 1) * v1 + (n2 - 1) * v2) / (n1 + n2 - 2))
    t = (m1 - m2) / (sp * np.sqrt(1 / n1 + 1 / n2))
    return 2 * stats.t.sf(np.abs(t), n1 + n2 - 2), (m1 - m2) / sp

def bonferroni(p): return np.minimum(1.0, np.asarray(p) * len(p))
def holm(p):
    p = np.asarray(p, float); m = len(p); o = np.argsort(p); aj = np.empty(m); corrido = 0.0
    for j, i in enumerate(o):
        corrido = max(corrido, (m - j) * p[i]); aj[i] = min(1.0, corrido)
    return aj
def benjamini_hochberg(p):
    """p ajustados (q-valores) de Benjamini-Hochberg."""
    p = np.asarray(p, float); m = len(p); o = np.argsort(p)
    escalado = p[o] * m / np.arange(1, m + 1)
    aj = np.minimum.accumulate(escalado[::-1])[::-1]
    out = np.empty(m); out[o] = np.minimum(1.0, aj)
    return out

# verificacion contra statsmodels con 50 p-valores al azar (mezcla de nulos y senales)
rg = np.random.default_rng(21)
p_prueba = np.r_[rg.random(40), rg.beta(0.3, 8, 10)]
for nombre, f, met in [("bonferroni", bonferroni, "bonferroni"), ("holm", holm, "holm"), ("BH", benjamini_hochberg, "fdr_bh")]:
    assert np.allclose(f(p_prueba), multipletests(p_prueba, method=met)[1]), nombre

# (a) 20 desenlaces, TODOS sin efecto real: cuantos experimentos 'encuentran algo'
R_, m_, n_g = 5000, 20, 30
p20, _ = prueba_t(rg.normal(size=(R_ * m_, n_g)), rg.normal(size=(R_ * m_, n_g)))
p20 = p20.reshape(R_, m_)
alguno = (p20.min(1) < 0.05).mean(); alguno_bonf = (p20.min(1) < 0.05 / m_).mean()
print(f"(a) 20 desenlaces sin efecto: P(al menos un p < 0.05) simulada = {alguno:.3f} | teorica 1 - 0.95^20 = {1 - 0.95 ** 20:.3f} | con Bonferroni = {alguno_bonf:.3f}")
assert abs(alguno - (1 - 0.95 ** 20)) < 0.02 and alguno_bonf < 0.07

# (b) 1 000 pruebas: 900 sin efecto y 100 con d = 1.0 (n = 30 por grupo); 300 repeticiones
R_, m_, n_g, n_ver = 300, 1000, 30, 100
Xs = rg.normal(size=(R_ * m_, n_g)); Ys = rg.normal(size=(R_ * m_, n_g)); verdadera = np.zeros(m_, bool); verdadera[:n_ver] = True
Ys = Ys.reshape(R_, m_, n_g); Ys[:, verdadera, :] += 1.0; Ys = Ys.reshape(R_ * m_, n_g)
P = prueba_t(Xs, Ys)[0].reshape(R_, m_)
orden = np.sort(P, axis=1); alfa = 0.05
rechazos = {"sin corregir": P < alfa, "Bonferroni": P < alfa / m_}
ok = orden <= alfa / (m_ - np.arange(m_)); falla = ~ok; k_h = np.where(falla.any(1), falla.argmax(1), m_)
umbral_h = np.where(k_h > 0, orden[np.arange(R_), np.maximum(k_h - 1, 0)], -1); rechazos["Holm"] = P <= umbral_h[:, None]
cond = orden <= (np.arange(1, m_ + 1) * alfa / m_); umbral_bh = np.where(cond, orden, -1).max(1); rechazos["Benjamini-Hochberg"] = P <= umbral_bh[:, None]
assert (rechazos["Holm"][0] == (holm(P[0]) < alfa)).all() and (rechazos["Benjamini-Hochberg"][0] == (benjamini_hochberg(P[0]) < alfa)).all()   # version vectorizada = escalar
filas = []
for nombre, rch in rechazos.items():
    falsos = (rch & ~verdadera).sum(1); total = rch.sum(1)
    filas.append([nombre, (falsos > 0).mean(), np.mean(np.where(total > 0, falsos / np.maximum(total, 1), 0)), (rch & verdadera).sum(1).mean() / n_ver])
tab = pd.DataFrame(filas, columns=["procedimiento", "FWER (algun falso)", "FDR (falsos / descubrimientos)", "potencia"]).set_index("procedimiento")
print("\n(b) 1 000 pruebas (900 nulas, 100 con efecto d = 1.0, n = 30 por grupo), 300 repeticiones:"); print(tab.round(3).to_string())
assert tab.loc["sin corregir", "FWER (algun falso)"] > 0.99 and tab.loc["Bonferroni", "FWER (algun falso)"] < 0.08 and tab.loc["Holm", "FWER (algun falso)"] < 0.08
assert tab.loc["Benjamini-Hochberg", "FDR (falsos / descubrimientos)"] < 0.06 and tab.loc["Benjamini-Hochberg", "potencia"] > tab.loc["Bonferroni", "potencia"]

# (c) Detencion opcional: dos grupos SIN efecto, se mira cada 5 observaciones por grupo desde n = 10 hasta 100
R_ = 4000; mx = 100
A, B = rg.normal(size=(R_, mx)), rg.normal(size=(R_, mx))
sA, sB, qA, qB = A.cumsum(1), B.cumsum(1), (A ** 2).cumsum(1), (B ** 2).cumsum(1)
rechaza_alguna, rechaza_fija = np.zeros(R_, bool), None
for nn in range(10, mx + 1, 5):
    mA, mB = sA[:, nn - 1] / nn, sB[:, nn - 1] / nn
    vA, vB = (qA[:, nn - 1] - nn * mA ** 2) / (nn - 1), (qB[:, nn - 1] - nn * mB ** 2) / (nn - 1)
    pp = 2 * stats.t.sf(np.abs((mA - mB) / np.sqrt((vA + vB) / nn)), 2 * nn - 2)
    rechaza_alguna |= pp < 0.05
    if nn == mx: rechaza_fija = pp < 0.05
print(f"\n(c) Mirar los datos {len(range(10, mx + 1, 5))} veces y parar en el primer p < 0.05: error tipo I = {rechaza_alguna.mean():.3f} | con n fijo = 100: {rechaza_fija.mean():.3f}")
assert rechaza_alguna.mean() > 0.15 and abs(rechaza_fija.mean() - 0.05) < 0.015

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].hist(p20.min(1), bins=40, color=PALETA[0]); ax[0].axvline(0.05, color="black", ls="--"); ax[0].set(title="Mejor p de 20 pruebas sin efecto", xlabel="p minimo")
tab[["FWER (algun falso)", "FDR (falsos / descubrimientos)", "potencia"]].plot.bar(ax=ax[1], color=PALETA[:3], rot=20); ax[1].set(title="Procedimientos (1000 pruebas)", xlabel="")
ax[2].bar(["n fijo = 100", "mirar 19 veces"], [rechaza_fija.mean(), rechaza_alguna.mean()], color=[PALETA[2], PALETA[1]]); ax[2].axhline(0.05, color="black", ls="--"); ax[2].set(title="Error tipo I real", ylabel="tasa de falsos positivos")
plt.tight_layout(); plt.show()
````

````python
# ============================================================
# 3b. Por que 'significativo' no es 'cierto': FDR de la literatura, maldicion del ganador y un ejemplo para practicar
# ============================================================
rg = np.random.default_rng(22)
# (a) 20 000 estudios: 10 % de hipotesis verdaderas (d = 0.5), n = 30 por grupo, alfa = 0.05
S, n_g, pi, d_ver, alfa = 20_000, 30, 0.10, 0.5, 0.05
es_verdad = rg.random(S) < pi
X = rg.normal(size=(S, n_g)); Y = rg.normal(size=(S, n_g)) + d_ver * es_verdad[:, None]
p, d_obs = prueba_t(Y, X)
sig = p < alfa
potencia = sig[es_verdad].mean(); fdr_sim = (~es_verdad[sig]).mean()
fdr_teo = alfa * (1 - pi) / (alfa * (1 - pi) + potencia * pi)
print(f"(a) Potencia simulada = {potencia:.2f} | de los resultados 'significativos', son FALSOS: simulado = {fdr_sim:.1%} | formula = {fdr_teo:.1%}")
assert abs(fdr_sim - fdr_teo) < 0.02

# (b) Maldicion del ganador: efecto verdadero d = 0.3, n = 15 por grupo (potencia baja)
S, n_g, d_ver = 40_000, 15, 0.3
X = rg.normal(size=(S, n_g)); Y = rg.normal(size=(S, n_g)) + d_ver
p2, d2 = prueba_t(Y, X)
sig2 = (p2 < 0.05) & (d2 > 0)
print(f"(b) Efecto verdadero d = {d_ver} | potencia = {sig2.mean():.2f} | d observado promedio, TODOS los estudios = {d2.mean():.2f} | solo los significativos = {d2[sig2].mean():.2f} (exagera x{d2[sig2].mean() / d_ver:.1f})")
assert abs(d2.mean() - d_ver) < 0.02 and d2[sig2].mean() > 2 * d_ver

# (c) 12 valores p de un estudio ilustrativo (inventados, ordenados): para la Practica 3
pvals_demo = np.array([0.001, 0.004, 0.006, 0.012, 0.018, 0.022, 0.031, 0.040, 0.200, 0.350, 0.600, 0.900])
print("\n(c) 12 p-valores de ejemplo:", pvals_demo)
print(pd.DataFrame({"p": pvals_demo, "Bonferroni": bonferroni(pvals_demo), "Holm": holm(pvals_demo), "BH": benjamini_hochberg(pvals_demo)}).round(3).T.to_string())
print("Rechazos con q = 0.05 -> sin corregir:", int((pvals_demo < 0.05).sum()), "| Bonferroni:", int((bonferroni(pvals_demo) < 0.05).sum()), "| Holm:", int((holm(pvals_demo) < 0.05).sum()), "| BH:", int((benjamini_hochberg(pvals_demo) < 0.05).sum()))
````

## 4. Sesgo y Equidad: Auditar un Modelo 🧭

Un modelo aprendido de **decisiones históricas** hereda sus sesgos aunque nunca vea la variable protegida (género, etnia, zona...): basta con que otras variables funcionen como **proxies**. Simulamos un caso **inventado**: una entidad de **crédito agrícola** decide si aprueba préstamos. Hay dos grupos (A y B) con la **misma capacidad de pago** a igual ingreso e historial, pero con ingresos medios algo distintos por razones estructurales, y una variable de **zona** muy asociada al grupo (proxy). Las decisiones históricas **penalizaron al grupo B** sin base en su capacidad de pago. Como los datos son sintéticos conocemos el **desenlace real** (si pagaría), algo que en la práctica **no se observa** para quienes fueron rechazados (el problema de las **etiquetas selectivas**).

**Métricas por grupo** ($\hat Y$ = decisión del modelo, $Y$ = paga de verdad, $G$ = grupo):

| Criterio | Definición | Pregunta |
|---|---|---|
| **Paridad demográfica** | $P(\hat Y=1\mid G=g)$ (tasa de selección) | ¿se aprueba la misma proporción? (regla de las 4/5: razón mínima 0.8, solo una **alerta**) |
| **Igualdad de oportunidades** | $P(\hat Y=1\mid Y=1,G=g)$ (TPR) | de los que **sí pagarían**, ¿a cuántos se aprueba? |
| **Tasas de error** | FPR $=P(\hat Y=1\mid Y=0,G=g)$ | ¿a quién se aprueba por error? |
| **Paridad predictiva** | $P(Y=1\mid\hat Y=1,G=g)$ (VPP) | de los aprobados, ¿qué fracción paga? |

**Imposibilidad.** Si las **tasas base** $P(Y=1\mid G)$ difieren, un clasificador imperfecto **no puede** igualar a la vez el VPP, el FPR y el FNR entre grupos (Chouldechova, 2017; Kleinberg *et al.*, 2016): hay que **elegir** qué criterio importa según el daño que se quiere evitar, y decirlo. Ningún ajuste técnico sustituye la discusión sobre **justicia** y sobre qué decisiones se delegan en un modelo.

````python
# ============================================================
# 4. Auditoria de equidad con datos SINTETICOS: sesgo en las etiquetas, proxies y umbrales por grupo
# ============================================================
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix

def sigmoide(z): return 1 / (1 + np.exp(-z))

def genera_credito(n=20_000, seed=5):
    rg = np.random.default_rng(seed)
    g = (rg.random(n) < 0.40).astype(int)                                    # 1 = grupo B (40 %)
    zona = (rg.random(n) < np.where(g == 1, 0.80, 0.20)).astype(int)         # zona 'sur': proxy del grupo
    ing = rg.normal(2.5 - 0.25 * g, 0.5, n)                                  # log del ingreso: B algo menor (estructural)
    hist = np.clip(rg.normal(6, 1.5, n), 0, 10)                              # puntualidad de pagos (mismo reparto en A y B)
    z = (ing - 2.4) / 0.5 + 0.6 * (hist - 6) / 1.5
    paga = (rg.random(n) < sigmoide(0.3 + 1.0 * z)).astype(int)              # desenlace real: SOLO depende de ingreso e historial
    aprob_hist = (rg.random(n) < sigmoide(0.3 + 1.0 * z - 1.0 * g)).astype(int)   # decision historica: penaliza al grupo B
    return pd.DataFrame(dict(grupo=g, zona=zona, ing=ing, hist=hist, paga=paga, aprob_hist=aprob_hist))

cred = genera_credito()
perm = np.random.default_rng(6).permutation(len(cred)); tr, te = cred.iloc[perm[:10_000]], cred.iloc[perm[10_000:]]
feats = ["ing", "hist", "zona"]                                                # el modelo NO ve 'grupo' (equidad por desconocimiento)
print("Tasa base de pago real: grupo A = %.3f | grupo B = %.3f | aprobacion historica: A = %.3f | B = %.3f" % tuple(
    [cred.loc[cred.grupo == g, c].mean() for c in ("paga", "aprob_hist") for g in (0, 1)]))

def metricas(y, yhat, g):
    """Metricas por grupo desde la matriz de confusion (escrita a mano; se verifica con sklearn)."""
    out = {}
    for gg, nom in ((0, "A"), (1, "B")):
        m = g == gg; tp = ((yhat == 1) & (y == 1) & m).sum(); fp = ((yhat == 1) & (y == 0) & m).sum()
        fn = ((yhat == 0) & (y == 1) & m).sum(); tn = ((yhat == 0) & (y == 0) & m).sum()
        out[nom] = dict(seleccion=(tp + fp) / m.sum(), TPR=tp / (tp + fn), FPR=fp / (fp + tn), VPP=tp / (tp + fp), tn=tn, fp=fp, fn=fn, tp=tp)
    return out

def ajusta(y_entrena):
    mod = LogisticRegression(max_iter=1000).fit(tr[feats], y_entrena)
    return mod, mod.predict_proba(te[feats])[:, 1]

mod1, s1 = ajusta(tr["aprob_hist"])            # Modelo 1: imita las decisiones historicas
mod2, s2 = ajusta(tr["paga"])                  # Modelo 2: entrenado con el desenlace real
y_te, g_te = te["paga"].to_numpy(), te["grupo"].to_numpy()
M1, M2 = metricas(y_te, (s1 >= 0.5).astype(int), g_te), metricas(y_te, (s2 >= 0.5).astype(int), g_te)
cm = confusion_matrix(y_te[g_te == 1], (s1 >= 0.5).astype(int)[g_te == 1])        # verificacion contra sklearn (grupo B, modelo 1)
assert (cm.ravel() == [M1["B"]["tn"], M1["B"]["fp"], M1["B"]["fn"], M1["B"]["tp"]]).all()
print("\nCoeficientes del Modelo 1 (imita decisiones historicas):", {f: round(float(c), 2) for f, c in zip(feats, mod1.coef_[0])}, "| del Modelo 2 (desenlace real):", {f: round(float(c), 2) for f, c in zip(feats, mod2.coef_[0])})

# Modelo 1 con umbrales por grupo que igualan el TPR (postprocesamiento, igualdad de oportunidades)
def umbral_para_tpr(s, y, objetivo):
    cand = np.sort(np.unique(s)); tpr = np.array([((s >= t) & (y == 1)).sum() / (y == 1).sum() for t in cand[::25]])
    return cand[::25][np.argmin(np.abs(tpr - objetivo))]
objetivo = M1["A"]["TPR"]
umbral_B = umbral_para_tpr(s1[g_te == 1], y_te[g_te == 1], objetivo)
yhat3 = np.where(g_te == 1, s1 >= umbral_B, s1 >= 0.5).astype(int); M3 = metricas(y_te, yhat3, g_te)

filas = []
for nombre, M, yh in (("1. Imita decisiones historicas", M1, (s1 >= 0.5).astype(int)), ("2. Entrenado con el desenlace real", M2, (s2 >= 0.5).astype(int)), ("3. Modelo 1 + umbral por grupo", M3, yhat3)):
    filas.append({"modelo": nombre, "sel. A": M["A"]["seleccion"], "sel. B": M["B"]["seleccion"], "razon B/A": M["B"]["seleccion"] / M["A"]["seleccion"],
                  "TPR A": M["A"]["TPR"], "TPR B": M["B"]["TPR"], "FPR A": M["A"]["FPR"], "FPR B": M["B"]["FPR"], "VPP A": M["A"]["VPP"], "VPP B": M["B"]["VPP"], "exactitud": (yh == y_te).mean()})
tab = pd.DataFrame(filas).set_index("modelo")
print("\nAuditoria por grupo (conjunto de prueba, desenlace REAL como referencia):"); print(tab.round(3).T.to_string())

brecha = lambda M: M["A"]["TPR"] - M["B"]["TPR"]
assert brecha(M1) > 0.08 and brecha(M2) < 0.5 * brecha(M1)                   # el sesgo historico se hereda; entrenar con el desenlace real lo reduce mucho
assert abs(brecha(M3)) < 0.02 and M3["B"]["VPP"] < M3["A"]["VPP"] + 1e-9     # igualar TPR deja de igualar el VPP (imposibilidad)
assert tab.loc["1. Imita decisiones historicas", "exactitud"] <= tab.loc["2. Entrenado con el desenlace real", "exactitud"] + 0.005

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
nombres = ["Modelo 1", "Modelo 2", "Modelo 3"]
for a, met, tit in zip(ax, ("sel.", "TPR", "VPP"), ("Tasa de seleccion", "TPR: aprobados entre quienes SI pagarian", "VPP: pagan entre los aprobados")):
    a.bar(np.arange(3) - 0.2, tab[f"{met} A"], 0.4, color=PALETA[0], label="grupo A"); a.bar(np.arange(3) + 0.2, tab[f"{met} B"], 0.4, color=PALETA[1], label="grupo B"); a.set(title=tit, xticks=range(3), xticklabels=nombres); a.legend()
plt.tight_layout(); plt.show()
````

**Lectura.** (1) **Los datos:** la tasa base de pago real es **0.593** en A y **0.504** en B (diferencia estructural de ingresos), pero la aprobación histórica fue **0.595** en A y solo **0.312** en B: se penalizó a B por encima de lo que su capacidad de pago justificaba. (2) **Modelo 1** (imita decisiones históricas, sin ver el grupo): hereda el sesgo mediante el **proxy** `zona` (coeficiente $-0.60$, frente a $0.03$ cuando se entrena con el desenlace real). Aprueba al **59.6 %** de A y al **30.4 %** de B (razón 0.51, muy por debajo de 0.8); de quienes **sí pagarían**, aprueba al **75.1 %** de A y al **46.4 %** de B (TPR); y su FPR es 0.374 frente a 0.147. Obsérvese que el **VPP es casi igual** (0.743 y 0.757): mirando solo la «paridad predictiva» el modelo parecería justo. **Las métricas pueden contradecirse.** (3) **Modelo 2** (entrenado con el desenlace real): la brecha del TPR baja de **0.287 a 0.118** y la razón de selección sube a **0.76**; lo que queda se debe a que los ingresos de B son menores (diferencia de tasas base), no a un sesgo de las etiquetas, y **tampoco desaparece** por mejorar el modelo. (4) **Modelo 3** (umbral por grupo): con un umbral más bajo para B se iguala el TPR (0.751 en ambos), la selección es 0.596 y 0.560 y los FPR son 0.374 y 0.372, pero el VPP pasa a **0.743 y 0.665**: al igualar un criterio **otro se rompe** (imposibilidad). La exactitud respecto al desenlace real es 0.683, 0.700 y 0.695: aquí corregir **no** costó exactitud porque los errores del Modelo 1 eran justamente el sesgo; en la práctica, el desenlace de los rechazados **no se observa** (etiquetas selectivas) y esa comprobación no es posible. (5) Además, usar umbrales distintos por grupo implica **utilizar el atributo protegido** en la decisión, lo que puede ser legalmente problemático según el contexto: la técnica no resuelve la decisión ética ni la jurídica.

## 5. Interpretación Responsable y Desafíos Contemporáneos 🧩

Ninguna prueba reemplaza el criterio. Lista de verificación para **reportar** un análisis estadístico:

| Pregunta | Riesgo si se ignora | Buena práctica |
|---|---|---|
| ¿Qué se **planeó** antes de ver los datos? | *p-hacking*, hipótesis a la carta | **Preregistrar** hipótesis y análisis; declarar lo exploratorio como exploratorio |
| ¿Cuántas cosas se **probaron**? | Falsos positivos inflados | Reportar **todas** las pruebas; corregir (Holm, BH) |
| ¿Solo el **valor $p$**? | Confundir significancia con importancia | Reportar **tamaño del efecto e intervalo de confianza**; nunca «la probabilidad de que $H_0$ sea cierta» |
| ¿Hay **supuestos** y se revisaron? | Conclusiones inválidas | Diagnóstico (cuadernos 00-02), análisis de sensibilidad, alternativas robustas |
| ¿**Correlación o causalidad**? | Decisiones erróneas por confusores o regresión a la media | Diseño (aleatorización, grupo control), lenguaje cuidadoso («se asocia con») |
| ¿Quién está **ausente** de los datos? | Sesgo de selección y de supervivencia | Describir cómo se obtuvo la muestra y a quién **no** representa |
| ¿De dónde vienen los datos y para qué se **recogieron**? | Uso fuera de la finalidad, falta de consentimiento | Documentar procedencia (*datasheets*), finalidad, base legal y retención |
| ¿Se puede **reproducir**? | Resultados irrepetibles | Semillas, versiones, código y datos (o datos sintéticos) compartidos; principios **FAIR** |
| ¿Quién puede salir **perjudicado** por el modelo? | Discriminación automatizada | Auditar por grupos (sección 4), documentar con *model cards*, supervisión humana, vías de apelación |

**Desafíos contemporáneos.** (1) **IA generativa y datos de entrenamiento:** ¿había consentimiento?, ¿se pueden reconstruir datos personales a partir del modelo? (2) **Reidentificación a escala:** con muchas fuentes, los cuasi-identificadores se multiplican y el anonimato se vuelve frágil (caso de la desanonimización de datos de calificaciones de películas, Narayanan y Shmatikov, 2008 *(de memoria, verificar)*). (3) **Datos sintéticos:** ayudan, pero **no son automáticamente privados** si copian demasiado de los reales. (4) **Decisiones automatizadas** en crédito, salud, empleo o justicia: transparencia y derecho a explicación. (5) **Soberanía de datos de los pueblos indígenas** (principios **CARE**: beneficio colectivo, autoridad para controlar, responsabilidad, ética) y datos comunitarios agrícolas. (6) **Costo ambiental y de acceso:** quién puede pagar el cómputo y quién queda fuera. **Regla práctica:** si no puedes explicar a las personas afectadas qué haces con sus datos y por qué, o no puedes mostrar la incertidumbre de tu conclusión, **todavía no estás listo para publicarla**.

| Concepto | Mensaje Central |
|---|---|
| **Marco** | Belmont (respeto, beneficencia, justicia); Ley 1581 de 2012 (finalidad, consentimiento, seguridad; datos sensibles con protección reforzada); seudonimizar **no** es anonimizar |
| **Hash y HMAC** | Un *hash* de un identificador de 900 000 posibilidades se invierte en segundos (ataque de diccionario); con **clave secreta** (HMAC) no, pero sigue siendo seudonimización |
| **k-anonimato** | Los cuasi-identificadores **singularizan** y permiten la **vinculación**; generalizar y suprimir da $k\ge5$ con pérdida de detalle; **l-diversidad** frente al ataque de homogeneidad |
| **Privacidad diferencial** | Laplace con escala $\Delta f/\varepsilon$; el cociente de densidades se acota por $e^{\varepsilon}$ (verificado); error medio $=\Delta f/\varepsilon$; **el presupuesto se suma** (repetir la consulta destruye la privacidad); más $n$, menos ruido relativo; respuesta aleatorizada con corrección insesgada |
| **Comparaciones múltiples** | 20 pruebas sin efecto: 64 % de «hallazgos»; Bonferroni/Holm controlan el FWER, **BH** el FDR con mayor potencia; mirar los datos 19 veces infla el error tipo I a ≈ 4-5 veces el nominal |
| **Significativo $\neq$ cierto** | Con hipótesis poco probables y potencia baja gran parte de lo «significativo» es falso, y el efecto publicado **se exagera** (maldición del ganador) |
| **Equidad** | Un modelo hereda el sesgo de sus etiquetas y lo reproduce mediante **proxies**; entrenar con el desenlace real lo reduce; igualar un criterio rompe otro (**imposibilidad**); no hay arreglo técnico sin decisión ética |
| **Interpretación responsable** | Preregistrar, reportar todo lo probado, tamaño del efecto e IC, causalidad con cuidado, documentar y poder reproducir |
