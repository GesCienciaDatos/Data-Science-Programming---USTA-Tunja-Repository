Este cuaderno es la versión **«para no matemáticos»** del [cuaderno 00 del Módulo 04 (Estadística no paramétrica)](../00_Estadistica_No_Parametrica_Kruskal_Wallis_y_Mann_Whitney.ipynb). Imagina que quieres saber **qué finca da papas más grandes**. Lo normal es comparar promedios, pero ¿qué pasa si en una finca apareció **una papa gigante** (o alguien anotó mal un peso)? El promedio se deja engañar; **el orden de llegada**, no.

| Lo que decimos | Lo que significa en la vida diaria |
|---|---|
| **Rango** | El puesto en que llega cada papa si las ordenas de la más chica a la más grande |
| **Carrera de pares** | Enfrentar cada papa de una finca con cada papa de la otra y ver quién gana más veces |
| **Barajar** | Mezclar al azar las papas entre las fincas para ver qué tan raro es el resultado que obtuviste |
| **«Qué tan raro»** | De cada 100 barajadas, cuántas producen una diferencia tan grande como la real (si es muy pocas, la diferencia **no parece obra del azar**) |

Al terminar podrás explicar:
1. Por qué **una sola papa gigante** engaña al promedio pero no al orden.
2. Cómo se compara con una **carrera de pares** y cómo se mide **qué tan raro** es el resultado **barajando**.
3. Cómo se comparan **tres o más** fincas a la vez y por qué, si hay diferencia, **luego** hay que mirar quién es el distinto.

## 1. La Papa Gigante Engaña al Promedio 🥔

Pesamos **12 papas** de la finca A y **12 de la finca B** (datos **inventados**, en gramos). En la B, una de las papas pesa **1 500 g** (¡una gigante, o un error al anotar!). Las demás papas de la B son un poco **más chicas** que las de la A. ¿Qué finca tiene papas más grandes?

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)                      # semilla fija: siempre salen los mismos numeros
plt.rcParams.update({"axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})

finca_a = np.round(rng.normal(200, 25, 12))              # gramos: papas "normales" de la finca A
finca_b = np.round(rng.normal(185, 25, 12))              # la finca B tiene papas un poco mas chicas...
finca_b[3] = 1500                                        # ...pero UNA papa gigante (o un error al anotar)

print(f"Promedio  A: {finca_a.mean():6.0f} g | B: {finca_b.mean():6.0f} g   <- el promedio dice que la B es mucho mayor")
print(f"Mediana   A: {np.median(finca_a):6.0f} g | B: {np.median(finca_b):6.0f} g   <- la papa 'del medio' dice lo contrario")

# El puesto (rango) de cada papa si ordenamos las 24 de la mas chica (1) a la mas grande (24)
todas = pd.Series(np.r_[finca_a, finca_b])
puestos = todas.rank().to_numpy()
print(f"Puesto de la papa gigante: {int(puestos[12 + 3])} de 24 (solo una posicion mas que la siguiente, aunque pese 7 veces mas)")
print(f"Puesto promedio  A: {puestos[:12].mean():.1f} | B: {puestos[12:].mean():.1f}   (el promedio de los puestos de A es mayor)")

plt.figure(figsize=(9, 3.6))
plt.scatter(np.arange(12), finca_a, color="#2563eb", label="finca A", s=45); plt.scatter(np.arange(12), finca_b, color="#dc2626", label="finca B", s=45)
plt.xlabel("numero de papa"); plt.ylabel("peso (g)"); plt.title("Una sola papa gigante estira todo el eje"); plt.legend(); plt.tight_layout(); plt.show()

assert finca_b.mean() > finca_a.mean() + 50                  # el promedio se deja enganar
assert np.median(finca_a) > np.median(finca_b)               # la mediana no
assert puestos[12 + 3] == 24 and puestos[:12].mean() > puestos[12:].mean()
````

---
**Lección:** el **promedio** dice que la finca B pesa mucho más (**299 g contra 196 g**), pero solo porque una papa pesa 1 500 g. La papa «del medio» (la **mediana**) cuenta la historia contraria: **202 g en la A y 190 g en la B**. Y en la lista de puestos, la papa gigante ocupa el puesto **24 de 24**: es «solo una posición más» que la siguiente, aunque pese siete veces más. El **puesto promedio** de A (13.2) es algo mayor que el de B (11.8). Ordenar le quita a una papa extrema el poder de arrastrar todo.

## 2. La Carrera de Pares: ¿Quién Gana Más Veces? 🏁

Enfrentamos **cada** papa de la finca A con **cada** papa de la B (12 × 12 = 144 duelos) y contamos en cuántos gana la A. Si las dos fincas fueran iguales, la A ganaría **la mitad** (72). Luego **barajamos**: mezclamos las 24 papas al azar, las repartimos en dos montones de 12 y repetimos el conteo **10 000 veces**. Así vemos **qué tan raro** es el número real si todo fuera pura suerte.

````python
duelos = finca_a[:, None] > finca_b[None, :]            # tabla de 12 x 12: True si gana la papa de A
gana_a = int(duelos.sum())
print(f"La finca A gana {gana_a} de {duelos.size} duelos ({gana_a / duelos.size:.0%}); si fueran iguales esperariamos 72 (50 %).")

# Barajar: mezclar las 24 papas y repartirlas al azar en dos montones de 12
baraja = np.random.default_rng(1)
ganadas_azar = np.empty(10_000)
for i in range(10_000):
    mezcla = baraja.permutation(np.r_[finca_a, finca_b])
    ganadas_azar[i] = (mezcla[:12][:, None] > mezcla[12:][None, :]).sum()
raro = np.mean(np.abs(ganadas_azar - 72) >= abs(gana_a - 72))
print(f"Barajando, una diferencia TAN grande (a favor de cualquiera de las dos) aparece en {raro:.1%} de las veces.")

plt.figure(figsize=(9, 3.6))
plt.hist(ganadas_azar, bins=30, color="#94a3b8"); plt.axvline(gana_a, color="#dc2626", lw=2, label=f"resultado real: {gana_a}")
plt.xlabel("duelos ganados por A (cuando todo es suerte)"); plt.ylabel("barajadas"); plt.title("¿Es raro lo que vimos?"); plt.legend(); plt.tight_layout(); plt.show()

# Otro ejemplo para la practica: minutos de espera de dos rutas de bus (inventados)
ruta_1 = np.array([12, 15, 11, 14, 13, 16, 12, 15])
ruta_2 = np.array([18, 14, 20, 17, 19, 13, 21, 18])

assert gana_a == int((finca_a[:, None] > finca_b[None, :]).sum()) and gana_a > 72
assert 0 <= raro <= 1
````

## 3. Tres Fincas a la Vez: ¿Alguna Es Distinta? 🏆

Ahora contamos **cuántos insectos** caen en una trampa en tres fincas (datos **inventados**: Norte, Centro y Sur). Ordenamos **todas** las trampas juntas, le damos a cada una su puesto y miramos el **puesto promedio** de cada finca. Si fueran iguales, los tres puestos promedio serían parecidos; medimos **qué tan separados** están y barajamos las trampas entre fincas para ver qué tan raro es.

````python
g = np.random.default_rng(7)
fincas = {"Norte": g.poisson(4, 14), "Centro": g.poisson(8, 15), "Sur": g.poisson(9, 13)}      # insectos por trampa (inventados)
etiqueta = np.concatenate([[nombre] * len(v) for nombre, v in fincas.items()])
valores = np.concatenate(list(fincas.values()))
puestos = pd.Series(valores).rank().to_numpy()                                                  # empates: puesto promedio
promedio_puestos = pd.Series(puestos).groupby(etiqueta).mean()
print(pd.DataFrame({"trampas": [len(v) for v in fincas.values()], "insectos (promedio)": [v.mean() for v in fincas.values()], "puesto promedio": promedio_puestos[list(fincas)].values}, index=list(fincas)).round(1).to_string())

def separacion(puestos_, etiqueta_):
    return pd.Series(puestos_).groupby(etiqueta_).mean().max() - pd.Series(puestos_).groupby(etiqueta_).mean().min()

real = separacion(puestos, etiqueta)
azar = np.array([separacion(puestos, g.permutation(etiqueta)) for _ in range(5_000)])
raro3 = (azar >= real).mean()
print(f"\nSeparacion real entre el puesto promedio mas alto y el mas bajo: {real:.1f} | barajando, tan grande o mayor en {raro3:.2%} de las veces.")

fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
ax[0].boxplot(list(fincas.values())); ax[0].set_xticklabels(list(fincas)); ax[0].set_title("Insectos por trampa"); ax[0].set_ylabel("conteo")
ax[1].hist(azar, bins=30, color="#94a3b8"); ax[1].axvline(real, color="#dc2626", lw=2, label="real"); ax[1].set_title("Separacion cuando todo es suerte"); ax[1].legend()
plt.tight_layout(); plt.show()

assert raro3 < 0.05 and promedio_puestos["Norte"] < promedio_puestos["Centro"]
````

---
**Lección:** los **puestos promedio** son 9.7 (Norte), 25.5 (Centro) y 29.7 (Sur): el Norte está muy lejos de los otros dos, mientras Centro y Sur son parecidos. La separación real es de **20 puestos** y **ninguna** de las 5 000 barajadas llegó a tanto: la pregunta «¿alguna es distinta?» tiene respuesta **sí**. Pero esto **no** dice cuál: el siguiente paso sería comparar las fincas **de a dos** (y tener cuidado de que, con muchas comparaciones, alguna salga «rara» solo por suerte). Y el hecho de que Centro y Sur parezcan parecidas no prueba que sean iguales.

## Resumen en una frase 🎯

> **Cuando hay datos raros, una papa gigante o conteos con muchos empates, comparar por el orden de llegada es más justo que comparar promedios; se juntan todos los datos, se ordenan y se barajan para ver qué tan raro es el resultado, y si salen tres o más grupos y uno es distinto, después hay que mirar cuál.**

### Lo que aprendiste hoy:
- ✅ **El promedio se deja engañar** por un valor extremo; el **orden** (y la mediana) no.
- ✅ **Carrera de pares:** contar quién gana más duelos entre dos grupos y comparar con la mitad (lo que daría el azar).
- ✅ **Barajar** para medir **qué tan raro** es lo que viste, sin fórmulas.
- ✅ Con **tres o más grupos** primero se pregunta «¿alguno es distinto?» y, solo si sí, «¿cuál?».
