## 1. El Modelo Lineal Múltiple en Forma Matricial 🧱

**Modelo.** Con $n$ observaciones y $p$ coeficientes (incluido el intercepto), $\mathbf y=\mathbf X\boldsymbol\beta+\boldsymbol\varepsilon$, con $\boldsymbol\varepsilon\sim N(\mathbf 0,\sigma^2\mathbf I)$. Los **MCO** minimizan $\lVert\mathbf y-\mathbf X\boldsymbol\beta\rVert^2$ y resuelven las **ecuaciones normales** $\mathbf X^\top\mathbf X\,\hat{\boldsymbol\beta}=\mathbf X^\top\mathbf y$:

$$\hat{\boldsymbol\beta}=(\mathbf X^\top\mathbf X)^{-1}\mathbf X^\top\mathbf y,\qquad \operatorname{Var}(\hat{\boldsymbol\beta})=\sigma^2(\mathbf X^\top\mathbf X)^{-1},\qquad s^2=\frac{SCE}{n-p},\qquad \mathbf H=\mathbf X(\mathbf X^\top\mathbf X)^{-1}\mathbf X^\top .$$

$\mathbf H$ es la **matriz sombrero** ($\hat{\mathbf y}=\mathbf H\mathbf y$): simétrica, idempotente y con $\operatorname{tr}(\mathbf H)=p$; su diagonal $h_{ii}$ es el **apalancamiento** de la observación $i$. **Bondad de ajuste e inferencia global:**

$$R^2=\frac{SCR}{SCT},\qquad R^2_{\text{aj}}=1-(1-R^2)\frac{n-1}{n-p},\qquad F=\frac{SCR/(p-1)}{SCE/(n-p)}\sim F_{p-1,\,n-p}\ \text{bajo }H_0:\beta_1=\dots=\beta_{p-1}=0.$$

**Interpretación.** $\beta_j$ es el cambio esperado en $y$ por **una unidad más de $x_j$ manteniendo constantes las demás variables del modelo** (*ceteris paribus*): cambia al agregar o quitar variables y **no** es causal sin un diseño que lo respalde. **Variables categóricas:** una variable de $k$ niveles se codifica con $k-1$ **indicadoras**; el nivel omitido es la **referencia**. Incluir las $k$ indicadoras *y* el intercepto produce columnas **linealmente dependientes** (la **trampa de la variable ficticia**) y $\mathbf X^\top\mathbf X$ deja de ser invertible.

**Numérica.** Resolver `solve(X.T @ X, X.T @ y)` **eleva al cuadrado** el número de condición; `lstsq` (SVD) o la descomposición **QR** son más estables. Aquí las comparamos.

````python
# ============================================================
# 1a. Modelo lineal multiple desde cero: ecuaciones normales, QR, lstsq, matriz sombrero
# ============================================================
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy import linalg

def genera_fincas(n=200, seed=2024):
    """Fincas de papa SINTETICAS. Modelo verdadero: rend = 10 + 0.06 fert + 0.025 lluvia + 2.5 riego - 0.004 (alt - 2800) + efecto_variedad + N(0, 2.2^2).
    'temp' depende casi solo de la altitud (colinealidad) y NO esta en el modelo verdadero; riego es mas frecuente con mas fertilizante."""
    rg = np.random.default_rng(seed)
    fert = rg.uniform(20, 140, n)                                    # kg de N por ha
    lluvia = rg.normal(120, 30, n).clip(40, None)                   # mm por mes
    alt = rg.uniform(2500, 3200, n)                                  # msnm
    temp = 28 - 0.0065 * alt + rg.normal(0, 0.15, n)                 # temperatura media (C)
    riego = (rg.random(n) < 1 / (1 + np.exp(-(fert - 80) / 25))).astype(int)
    variedad = rg.choice(["Pastusa", "Criolla", "Capiro"], n, p=[0.45, 0.30, 0.25])
    efecto = pd.Series(variedad).map({"Pastusa": 0.0, "Criolla": 1.0, "Capiro": 2.0}).to_numpy()
    rend = 10 + 0.06 * fert + 0.025 * lluvia + 2.5 * riego - 0.004 * (alt - 2800) + efecto + rg.normal(0, 2.2, n)
    return pd.DataFrame(dict(rend=rend, fert=fert, lluvia=lluvia, riego=riego, alt=alt, temp=temp, variedad=variedad))

fincas = genera_fincas()
dummies = pd.get_dummies(fincas["variedad"], dtype=float)[["Criolla", "Capiro"]]               # referencia: Pastusa
nombres_X = ["intercepto", "fert", "lluvia", "riego", "alt", "Criolla", "Capiro"]
X = np.c_[np.ones(len(fincas)), fincas[["fert", "lluvia", "riego", "alt"]].to_numpy(), dummies.to_numpy()]
y = fincas["rend"].to_numpy()
n, p = X.shape

def mco(X, y):
    """MCO completo desde cero (ecuaciones normales)."""
    n, p = X.shape
    XtX_inv = np.linalg.inv(X.T @ X)
    beta = XtX_inv @ X.T @ y
    H = X @ XtX_inv @ X.T
    e = y - X @ beta
    SCE, SCT = (e ** 2).sum(), ((y - y.mean()) ** 2).sum()
    s2 = SCE / (n - p); se = np.sqrt(s2 * np.diag(XtX_inv))
    t = beta / se; pv = 2 * stats.t.sf(np.abs(t), n - p)
    R2 = 1 - SCE / SCT; R2aj = 1 - (1 - R2) * (n - 1) / (n - p)
    F = ((SCT - SCE) / (p - 1)) / s2
    return dict(beta=beta, se=se, t=t, p=pv, e=e, s2=s2, SCE=SCE, SCT=SCT, R2=R2, R2aj=R2aj, F=F, pF=stats.f.sf(F, p - 1, n - p), h=np.diag(H), H=H, XtX_inv=XtX_inv)

aj = mco(X, y)
# tres formas de resolver lo mismo
beta_ne = np.linalg.solve(X.T @ X, X.T @ y)
Q, R = np.linalg.qr(X); beta_qr = linalg.solve_triangular(R, Q.T @ y)
beta_ls = np.linalg.lstsq(X, y, rcond=None)[0]
assert np.allclose(aj["beta"], beta_ne) and np.allclose(beta_ne, beta_qr) and np.allclose(beta_qr, beta_ls)

# contraste con statsmodels: con formula (categorica, referencia = Pastusa) y con la matriz X
sm_form = smf.ols("rend ~ fert + lluvia + riego + alt + C(variedad, Treatment('Pastusa'))", data=fincas).fit()
assert np.allclose(np.sort(sm_form.params.to_numpy()), np.sort(aj["beta"]))                      # mismos coeficientes (orden distinto)
sm_fit = sm.OLS(y, X).fit()
assert np.allclose(sm_fit.params, aj["beta"]) and np.allclose(sm_fit.bse, aj["se"]) and np.allclose(sm_fit.pvalues, aj["p"])
assert abs(sm_fit.rsquared - aj["R2"]) < 1e-12 and abs(sm_fit.rsquared_adj - aj["R2aj"]) < 1e-12
assert abs(sm_fit.fvalue - aj["F"]) < 1e-8 and abs(sm_fit.f_pvalue - aj["pF"]) < 1e-12

# propiedades de la matriz sombrero
H = aj["H"]
assert np.allclose(H, H.T) and np.allclose(H @ H, H) and abs(np.trace(H) - p) < 1e-9 and np.allclose(H @ y, X @ aj["beta"])
assert (aj["h"] > 0).all() and (aj["h"] < 1).all()

verdad = [21.2, 0.06, 0.025, 2.5, -0.004, 1.0, 2.0]                     # intercepto verdadero = 10 + 0.004 * 2800
ci = sm_fit.conf_int()
tabla = pd.DataFrame({"estimador": aj["beta"], "EE": aj["se"], "t": aj["t"], "p": aj["p"], "IC inf": ci[:, 0], "IC sup": ci[:, 1], "verdadero": verdad}, index=nombres_X)
print(tabla.round(4).to_string())
print(f"\ns = {np.sqrt(aj['s2']):.3f} (verdadero 2.2) | R2 = {aj['R2']:.3f} | R2 ajustado = {aj['R2aj']:.3f} | F({p - 1}, {n - p}) = {aj['F']:.1f}, p = {aj['pF']:.1e}")
print(f"traza de H = {np.trace(H):.3f} = p | apalancamiento medio = p/n = {p / n:.4f}, maximo = {aj['h'].max():.4f}")
print(f"Numero de condicion: cond(X) = {np.linalg.cond(X):.2e} | cond(X'X) = {np.linalg.cond(X.T @ X):.2e} = cond(X)^2")
print(f"IC 95 % contienen el valor verdadero: {int(((tabla['IC inf'] <= tabla['verdadero']) & (tabla['verdadero'] <= tabla['IC sup'])).sum())} de {p}")
````

````python
# ============================================================
# 1b. F parcial (modelos anidados), sesgo por variable omitida y trampa de la variable ficticia
# ============================================================
def sce_modelo(cols):
    Xm = np.c_[np.ones(n), fincas[cols].to_numpy()]
    b = np.linalg.lstsq(Xm, y, rcond=None)[0]
    return ((y - Xm @ b) ** 2).sum(), Xm.shape[1], b

SCEs = [sce_modelo(c) for c in (["fert"], ["fert", "lluvia", "riego"], ["fert", "lluvia", "riego", "alt"])]
SCEs.append((aj["SCE"], p, aj["beta"]))                                    # + variedad = modelo completo
nombres_pasos = ["fert", "+ lluvia, riego", "+ altitud", "+ variedad (2 indicadoras)"]
filas = []
for k in range(1, 4):
    (sce_r, p_r, _), (sce_c, p_c, _) = SCEs[k - 1], SCEs[k]
    q = p_c - p_r; Fp = ((sce_r - sce_c) / q) / (sce_c / (n - p_c)); filas.append([nombres_pasos[k], q, sce_r, sce_c, Fp, stats.f.sf(Fp, q, n - p_c)])
tab_F = pd.DataFrame(filas, columns=["se agrega", "q", "SCE reducido", "SCE completo", "F parcial", "p"]).set_index("se agrega")
print(tab_F.round(4).to_string())

# verificacion contra statsmodels.anova_lm sobre los modelos anidados
m0 = smf.ols("rend ~ fert", fincas).fit(); m1 = smf.ols("rend ~ fert + lluvia + riego", fincas).fit()
m2 = smf.ols("rend ~ fert + lluvia + riego + alt", fincas).fit(); m3 = smf.ols("rend ~ fert + lluvia + riego + alt + C(variedad)", fincas).fit()
for k, (ma, mb) in enumerate([(m0, m1), (m1, m2), (m2, m3)]):                 # comparacion de cada par anidado (con varias a la vez, anova_lm usa el error del modelo mayor)
    at = sm.stats.anova_lm(ma, mb)
    assert np.isclose(at["F"].iloc[1], tab_F["F parcial"].iloc[k]) and np.isclose(at["Pr(>F)"].iloc[1], tab_F["p"].iloc[k])

# sesgo por variable omitida: el coeficiente de fert cambia al incluir 'riego' (correlacionada con fert)
b_simple, b_con_riego, b_completo = SCEs[0][2][1], SCEs[1][2][1], aj["beta"][1]
print(f"\nCorrelacion fert-riego = {np.corrcoef(fincas['fert'], fincas['riego'])[0, 1]:.2f}")
print(f"Coeficiente de fert: solo fert = {b_simple:.4f} | + lluvia y riego = {b_con_riego:.4f} | modelo completo = {b_completo:.4f} | verdadero = 0.0600")
assert b_simple > b_completo + 0.01 and abs(b_completo - 0.06) < 0.015                             # omitir 'riego' infla el efecto del fertilizante

# trampa de la variable ficticia: intercepto + las 3 indicadoras
X_trampa = np.c_[np.ones(n), pd.get_dummies(fincas["variedad"], dtype=float).to_numpy()]
print(f"\nIntercepto + 3 indicadoras: rango de X = {np.linalg.matrix_rank(X_trampa)} de {X_trampa.shape[1]} columnas | cond(X'X) = {np.linalg.cond(X_trampa.T @ X_trampa):.1e}")
assert np.linalg.matrix_rank(X_trampa) == X_trampa.shape[1] - 1
````

## 2. Multicolinealidad: Cuando los Predictores se Estorban 🧲

**Qué es.** Si un predictor es (casi) una combinación lineal de los otros, el modelo **no puede separar sus efectos**: predice bien, pero los coeficientes individuales son **inestables** y sus errores estándar enormes. El **factor de inflación de la varianza** lo cuantifica: se ajusta $x_j$ contra los demás predictores, con coeficiente de determinación $R_j^2$, y

$$\operatorname{VIF}_j=\frac{1}{1-R_j^2},\qquad \operatorname{Var}(\hat\beta_j)=\frac{\sigma^2}{(1-R_j^2)\,S_{jj}},\qquad \operatorname{VIF}_j=\big[\mathbf R^{-1}\big]_{jj},$$

donde $\mathbf R$ es la **matriz de correlaciones** de los predictores. Regla práctica (convención, no ley): $\operatorname{VIF}>5$ merece revisión y $>10$ indica un problema serio. **Qué hacer:** quitar o combinar predictores redundantes, usar conocimiento del dominio, o regularización (Ridge, en el curso de Machine Learning). **Qué no hacer:** quitar la variable «porque salió no significativa» sin entender por qué. Si solo importa **predecir** y la colinealidad se mantiene en los datos nuevos, el ajuste sigue siendo bueno.

````python
# ============================================================
# 2. VIF desde cero, inflacion de errores estandar e inestabilidad (simulado y con 'diabetes')
# ============================================================
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.datasets import load_diabetes

def vif(M):
    """VIF de cada columna de M (sin constante) = diagonal de la inversa de la matriz de correlaciones."""
    return np.diag(np.linalg.inv(np.corrcoef(M, rowvar=False)))

pred_A = ["fert", "lluvia", "riego", "alt"]; pred_B = pred_A + ["temp"]
vif_A, vif_B = vif(fincas[pred_A].to_numpy()), vif(fincas[pred_B].to_numpy())
M_B = sm.add_constant(fincas[pred_B].to_numpy())
assert np.allclose(vif_B, [variance_inflation_factor(M_B, j + 1) for j in range(len(pred_B))])        # igual que statsmodels
print(pd.DataFrame({"VIF sin temp": pd.Series(vif_A, pred_A), "VIF con temp": pd.Series(vif_B, pred_B)}).round(2).to_string())
print(f"Correlacion altitud-temperatura = {np.corrcoef(fincas['alt'], fincas['temp'])[0, 1]:.4f}")

fA = smf.ols("rend ~ fert + lluvia + riego + alt", fincas).fit(); fB = smf.ols("rend ~ fert + lluvia + riego + alt + temp", fincas).fit()
print(f"\nR2: sin temp = {fA.rsquared:.4f} | con temp = {fB.rsquared:.4f}  (la prediccion casi no cambia)")
print(f"Coef. de alt: sin temp = {fA.params['alt']:+.5f} (EE {fA.bse['alt']:.5f}) | con temp = {fB.params['alt']:+.5f} (EE {fB.bse['alt']:.5f}) | verdadero = -0.00400")
print(f"Coef. de temp (verdadero 0): {fB.params['temp']:+.3f} (EE {fB.bse['temp']:.3f}, p = {fB.pvalues['temp']:.3f})")
assert fB.bse["alt"] > 8 * fA.bse["alt"] and vif_B[pred_B.index("alt")] > 50 and vif_A.max() < 3

# inestabilidad: 500 remuestreos bootstrap del coeficiente de alt
rg = np.random.default_rng(31)
bs = {"sin temp": [], "con temp": []}
for _ in range(500):
    idx = rg.integers(0, len(fincas), len(fincas)); d = fincas.iloc[idx]
    bs["sin temp"].append(np.linalg.lstsq(np.c_[np.ones(len(d)), d[pred_A]], d["rend"], rcond=None)[0][4])
    bs["con temp"].append(np.linalg.lstsq(np.c_[np.ones(len(d)), d[pred_B]], d["rend"], rcond=None)[0][4])
print(f"Bootstrap del coef. de alt: desv. est. sin temp = {np.std(bs['sin temp']):.5f} | con temp = {np.std(bs['con temp']):.5f}")

# datos reales incluidos en scikit-learn (sin red): diabetes
dia = load_diabetes(as_frame=True).data
v_dia = pd.Series(vif(dia.to_numpy()), dia.columns)
print("\nVIF en 'diabetes' (10 variables clinicas estandarizadas):"); print(v_dia.round(1).to_string())
assert v_dia.max() > 10 and v_dia.idxmax() in ("s1", "s2")

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].scatter(fincas["alt"], fincas["temp"], s=10, color=PALETA[0]); ax[0].set(title="Altitud y temperatura: casi la misma variable", xlabel="altitud (msnm)", ylabel="temperatura (C)")
ax[1].hist(bs["sin temp"], bins=30, alpha=0.7, color=PALETA[2], label="sin temp"); ax[1].hist(bs["con temp"], bins=30, alpha=0.7, color=PALETA[1], label="con temp")
ax[1].axvline(-0.004, color="black", ls="--"); ax[1].set(title="Coeficiente de altitud (bootstrap)", xlabel="coeficiente"); ax[1].legend()
ax[2].barh(v_dia.index, v_dia.values, color=PALETA[3]); ax[2].axvline(10, color="black", ls="--"); ax[2].set(title="VIF en 'diabetes'", xlabel="VIF")
plt.tight_layout(); plt.show()
````

## 3. Residuos, Apalancamiento e Influencia 🕵️

No todas las observaciones pesan igual. Con $e_i$ el residuo, $h_{ii}$ el apalancamiento y $s$ la estimación de $\sigma$:

| Cantidad | Fórmula | Qué detecta |
|---|---|---|
| **Apalancamiento** | $h_{ii}$ (diagonal de $\mathbf H$); alerta si $h_{ii}>2p/n$ | observaciones **lejanas en el espacio de los predictores** |
| **Residuo estandarizado** | $r_i=\dfrac{e_i}{s\sqrt{1-h_{ii}}}$ | residuos **grandes**, con varianza corregida |
| **Residuo studentizado externo** | $t_i=r_i\sqrt{\dfrac{n-p-1}{n-p-r_i^2}}\sim t_{n-p-1}$ | atípicos en $y$ (estimando $\sigma$ **sin** la observación $i$) |
| **Distancia de Cook** | $D_i=\dfrac{r_i^2}{p}\,\dfrac{h_{ii}}{1-h_{ii}}$ | cuánto **cambia el ajuste completo** al quitar la observación $i$ (regla: $D_i>4/n$ merece revisión; $>1$ es seria) |

Un punto **apalancado** no es necesariamente malo; es **influyente** cuando además su residuo es grande. Para ver esto, contaminamos 3 filas de las fincas: un **error de digitación en $x$** (420 kg/ha en lugar de 42), una **cosecha perdida** ($y$ muy bajo, $x$ normales) y una finca **legítimamente extrema** (140 kg/ha, 3 200 msnm) cuyo $y$ sí sigue el modelo.

````python
# ============================================================
# 3. Apalancamiento, residuos studentizados y distancia de Cook, desde cero
# ============================================================
from statsmodels.stats.outliers_influence import OLSInfluence

fincas_c = fincas.copy()
fincas_c.loc[0, "fert"] = 420.0                                           # error de digitacion (era ~42): apalancamiento alto, y no se corrigio
fincas_c.loc[1, "rend"] = fincas_c.loc[1, "rend"] - 18.0                  # cosecha perdida por helada: residuo grande
fincas_c.loc[2, ["fert", "alt"]] = [140.0, 3200.0]                        # finca extrema pero coherente con el modelo: su rend sigue la recta ajustada
fincas_c.loc[2, "rend"] = aj["beta"][:5] @ np.array([1, 140.0, fincas_c.loc[2, "lluvia"], fincas_c.loc[2, "riego"], 3200.0]) + aj["beta"][5:] @ dummies.iloc[2].to_numpy()
dum_c = pd.get_dummies(fincas_c["variedad"], dtype=float)[["Criolla", "Capiro"]]
Xc = np.c_[np.ones(n), fincas_c[["fert", "lluvia", "riego", "alt"]].to_numpy(), dum_c.to_numpy()]; yc = fincas_c["rend"].to_numpy()

ac = mco(Xc, yc); h, e, s2c = ac["h"], ac["e"], ac["s2"]
r_est = e / np.sqrt(s2c * (1 - h))                                         # estandarizados (internos)
t_est = r_est * np.sqrt((n - p - 1) / (n - p - r_est ** 2))                # studentizados externos
D = r_est ** 2 / p * h / (1 - h)                                           # distancia de Cook

infl = OLSInfluence(sm.OLS(yc, Xc).fit())                                  # verificacion contra statsmodels
assert np.allclose(h, infl.hat_matrix_diag) and np.allclose(r_est, infl.resid_studentized_internal)
assert np.allclose(t_est, infl.resid_studentized_external) and np.allclose(D, infl.cooks_distance[0])

umbral_h, umbral_D = 2 * p / n, 4 / n
tabla = pd.DataFrame({"h": h, "residuo student.": t_est, "Cook D": D})
print(f"Umbrales: h > {umbral_h:.3f} | |t| > 3 | D > {umbral_D:.3f}")
print(tabla.loc[[0, 1, 2]].round(3).to_string())
sospechosas = tabla.index[(tabla["h"] > umbral_h) | (tabla["residuo student."].abs() > 3) | (tabla["Cook D"] > umbral_D)]
print(f"Observaciones senaladas por al menos un criterio: {len(sospechosas)} de {n}; las 3 con mayor D: {list(tabla['Cook D'].nlargest(3).index)}")

# cuanto cambia el modelo al quitar las observaciones 0, 1 y 2
b_c = ac["beta"]; b_sin = mco(np.delete(Xc, [0, 1, 2], axis=0), np.delete(yc, [0, 1, 2]))["beta"]
print(pd.DataFrame({"contaminado": b_c, "sin las 3 filas": b_sin, "verdadero": verdad}, index=nombres_X).round(4).to_string())
assert set(tabla["Cook D"].nlargest(2).index) == {0, 1} and D[0] > 1 and D[2] < D[1]                  # el error de x y la cosecha perdida lideran; la finca extrema coherente casi no influye

fig, ax = plt.subplots(2, 2, figsize=(12, 8))
aj_c = Xc @ b_c
ax[0, 0].scatter(aj_c, r_est, s=12, color=PALETA[0]); ax[0, 0].axhline(0, color="black"); ax[0, 0].set(title="Residuos estandarizados vs ajustados", xlabel="ajustados", ylabel="r_i")
(osm, osr), _ = stats.probplot(t_est, dist=stats.t, sparams=(n - p - 1,))
ax[0, 1].scatter(osm, osr, s=12, color=PALETA[0]); ax[0, 1].plot(osm, osm, color="black"); ax[0, 1].set(title="Q-Q de los residuos studentizados (t)", xlabel="cuantiles teoricos", ylabel="observados")
ax[1, 0].scatter(aj_c, np.sqrt(np.abs(r_est)), s=12, color=PALETA[0]); ax[1, 0].set(title="Escala-ubicacion (varianza constante?)", xlabel="ajustados", ylabel="raiz |r_i|")
ax[1, 1].scatter(h, t_est, s=20 + 400 * D, color=PALETA[0], alpha=0.5)
for i in (0, 1, 2): ax[1, 1].annotate(str(i), (h[i], t_est[i]), fontsize=11, color=PALETA[1])
ax[1, 1].axvline(umbral_h, color="grey", ls="--"); ax[1, 1].axhline(3, color="grey", ls="--"); ax[1, 1].axhline(-3, color="grey", ls="--")
ax[1, 1].set(title="Apalancamiento vs residuo (tamano = Cook D)", xlabel="h_ii", ylabel="t_i")
plt.tight_layout(); plt.show()
````

## 4. Contrastes de Supuestos y Errores Estándar Robustos 🧪

Los gráficos de residuos se complementan con **contrastes formales** (que nunca reemplazan el juicio: con muchas observaciones rechazan desviaciones triviales, con pocas no ven nada):

| Supuesto | Contraste | Estadístico |
|---|---|---|
| Varianza constante | **Breusch-Pagan** (versión de Koenker) | regresar $e_i^2$ sobre $\mathbf X$; $LM=nR^2_{\text{aux}}\sim\chi^2_{p-1}$ |
| Independencia (series ordenadas en el tiempo) | **Durbin-Watson** | $DW=\dfrac{\sum_{t=2}^n(e_t-e_{t-1})^2}{\sum_{t=1}^ne_t^2}\approx2(1-\hat\rho)$; ≈ 2 si no hay autocorrelación, $<2$ si es positiva |
| Normalidad de los errores | **Shapiro-Wilk** sobre los residuos | menos relevante con $n$ grande (el TLC protege la inferencia sobre $\boldsymbol\beta$) |

**Si hay heterocedasticidad**, los coeficientes de MCO siguen siendo insesgados pero los **errores estándar clásicos son incorrectos**. La corrección más usada es el **estimador sándwich HC3** (MacKinnon y White):

$$\widehat{\operatorname{Var}}_{\text{HC3}}(\hat{\boldsymbol\beta})=(\mathbf X^\top\mathbf X)^{-1}\,\mathbf X^\top\operatorname{diag}\!\left(\frac{e_i^2}{(1-h_{ii})^2}\right)\mathbf X\,(\mathbf X^\top\mathbf X)^{-1}.$$

No cambia los coeficientes, solo los errores estándar (y por tanto $t$, valores $p$ e IC). Lo evaluamos por simulación: si la corrección funciona, la cobertura de sus IC debe acercarse al 95 %.

````python
# ============================================================
# 4. Breusch-Pagan, Durbin-Watson, Shapiro y errores estandar HC3, desde cero
# ============================================================
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson

def breusch_pagan(X, e):
    """LM = n R^2 de la regresion auxiliar e^2 ~ X (Koenker); p-valor chi-cuadrado con p-1 gl."""
    z = e ** 2; b = np.linalg.lstsq(X, z, rcond=None)[0]
    R2a = 1 - ((z - X @ b) ** 2).sum() / ((z - z.mean()) ** 2).sum()
    lm = len(e) * R2a
    return lm, stats.chi2.sf(lm, X.shape[1] - 1)

def dw(e):
    return (np.diff(e) ** 2).sum() / (e ** 2).sum()

def cov_hc3(X, e, h):
    A = np.linalg.inv(X.T @ X)
    return A @ (X.T * (e ** 2 / (1 - h) ** 2)) @ X @ A

# (a) datos de fincas con varianza constante vs con varianza creciente (sd proporcional a fert)
rg = np.random.default_rng(41)
mu = X @ np.array(verdad)
y_homo = mu + rg.normal(0, 2.2, n)
y_hetero = mu + rg.normal(0, 1, n) * (0.045 * fincas["fert"].to_numpy())
filas = []
for nombre, yy in {"homocedastico": y_homo, "heterocedastico (sd crece con fert)": y_hetero}.items():
    a = mco(X, yy); lm, plm = breusch_pagan(X, a["e"])
    ref = het_breuschpagan(a["e"], X)
    assert abs(lm - ref[0]) < 1e-8 and abs(plm - ref[1]) < 1e-10
    filas.append([nombre, lm, plm, stats.shapiro(a["e"]).pvalue, dw(a["e"])])
print(pd.DataFrame(filas, columns=["datos", "LM Breusch-Pagan", "p BP", "p Shapiro", "Durbin-Watson"]).set_index("datos").round(4).to_string())

# (b) Durbin-Watson en una serie temporal: 80 semanas de una parcela con errores AR(1)
rg = np.random.default_rng(42)
lluvia_t = rg.normal(30, 8, 80); u = np.zeros(80)
for t_ in range(1, 80): u[t_] = 0.7 * u[t_ - 1] + rg.normal(0, 1.0)
y_t = 5 + 0.05 * lluvia_t + u
Xt = np.c_[np.ones(80), lluvia_t]; at_ = mco(Xt, y_t)
rho_hat = (at_["e"][1:] @ at_["e"][:-1]) / (at_["e"] @ at_["e"])
assert abs(dw(at_["e"]) - durbin_watson(at_["e"])) < 1e-12
print(f"\nSerie semanal: Durbin-Watson = {dw(at_['e']):.3f} (2(1 - rho_hat) = {2 * (1 - rho_hat):.3f}, rho_hat = {rho_hat:.2f}) frente a {dw(mco(X, y_homo)['e']):.2f} en las fincas sin orden temporal")

# (c) HC3 desde cero vs statsmodels, con y_hetero
a = mco(X, y_hetero)
V3 = cov_hc3(X, a["e"], a["h"]); se_hc3 = np.sqrt(np.diag(V3))
assert np.allclose(se_hc3, sm.OLS(y_hetero, X).fit(cov_type="HC3").bse)
print("\nError estandar de fert con y_hetero: clasico =", round(a["se"][1], 5), "| HC3 =", round(se_hc3[1], 5))

# (d) simulacion: 3000 repeticiones con X fija, mismo modelo heterocedastico
R_ = 3000; rg = np.random.default_rng(43)
beta_v = np.array(verdad); A_ = np.linalg.inv(X.T @ X) @ X.T; hh = np.diag(X @ A_)
Y = (X @ beta_v)[:, None] + rg.normal(size=(n, R_)) * (0.045 * fincas["fert"].to_numpy())[:, None]
B = A_ @ Y; E = Y - X @ B
se_cl = np.sqrt((E ** 2).sum(0) / (n - p) * np.diag(np.linalg.inv(X.T @ X))[1])          # error estandar clasico de fert
se_h3 = np.sqrt((A_[1] ** 2) @ (E ** 2 / (1 - hh)[:, None] ** 2))                           # HC3 para fert
q = stats.t.ppf(0.975, n - p)
cob_cl = (np.abs(B[1] - 0.06) <= q * se_cl).mean(); cob_h3 = (np.abs(B[1] - 0.06) <= q * se_h3).mean()
print(f"\nDesv. est. real de b_fert = {B[1].std():.5f} | EE clasico medio = {se_cl.mean():.5f} | EE HC3 medio = {se_h3.mean():.5f}")
print(f"Cobertura del IC 95 % de fert: clasico = {cob_cl:.1%} | HC3 = {cob_h3:.1%}")
assert abs(cob_h3 - 0.95) <= abs(cob_cl - 0.95) + 1e-9 and abs(cob_h3 - 0.95) < 0.02
assert filas[0][2] > 0.05 and filas[1][2] < 0.001 and dw(at_["e"]) < 1.0 < dw(mco(X, y_homo)["e"])
````

---
**Lectura.** (1) **Breusch-Pagan** (coincide con `statsmodels`): con varianza constante $LM=2.88$ ($p=0.82$); con desviación estándar proporcional al fertilizante, $LM=37.1$ ($p<10^{-4}$). Shapiro-Wilk, en cambio, apenas lo nota ($p=0.229$ frente a $0.048$): **cada contraste ve un tipo de falla**. (2) **Durbin-Watson:** en las fincas (sin orden temporal) vale **2.02**; en la serie semanal con errores AR(1) vale **0.523**, que es $2(1-\hat\rho)=0.530$ con $\hat\rho=0.73$: la independencia se rompe y el valor $p$ clásico no sirve (cuaderno 01: cobertura 59 %). (3) **HC3 desde cero** coincide con `statsmodels(cov_type="HC3")`: el error estándar de `fert` pasa de 0.00787 a 0.00851 con los datos heterocedásticos, **sin cambiar el coeficiente**. (4) **Simulación (3 000 repeticiones):** la desviación estándar real de $\hat\beta_{\text{fert}}$ es **0.00985**; el error estándar clásico promedia **0.00838** (subestima ≈ 15 %) y HC3 **0.00992**. Los IC «del 95 %» cubren **90.0 %** con el error clásico y **94.6 %** con HC3. Con $n=200$ la corrección funciona; con muestras pequeñas HC3 también puede quedar corto, y no corrige la autocorrelación (ahí se requieren modelos de series de tiempo, Módulo 07).

## 5. Elegir entre Modelos: Más Variables No Es Mejor 🧮

$R^2$ **nunca disminuye** al agregar un predictor (aunque sea ruido puro). Criterios que **penalizan la complejidad**, con la verosimilitud gaussiana $\ell=-\frac n2\left[\ln(2\pi)+\ln\frac{SCE}{n}+1\right]$:

$$R^2_{\text{aj}},\qquad \text{AIC}=-2\ell+2p,\qquad \text{BIC}=-2\ell+p\ln n,\qquad \text{PRESS}=\sum_{i=1}^{n}\left(\frac{e_i}{1-h_{ii}}\right)^2,\quad R^2_{\text{pred}}=1-\frac{\text{PRESS}}{SCT}.$$

(Convención de `statsmodels`: $p$ cuenta los coeficientes de regresión.) **AIC** y **BIC** favorecen modelos con menor valor; BIC penaliza más con $n$ grande. **PRESS** es exactamente el error de la **validación cruzada «deja uno fuera»** (LOOCV) y se obtiene **sin reajustar** $n$ veces gracias a la matriz sombrero. Es la forma rápida de ver si el modelo generaliza; la validación cruzada $k$-fold y el sobreajuste en general se estudian en el Módulo 06 y en [Machine Learning, Módulo 07](../../Machine%20Learning/07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/).

````python
# ============================================================
# 5. R2, R2 ajustado, AIC, BIC y PRESS al agregar variables de ruido
# ============================================================
def criterios(Xm, yy):
    a = mco(Xm, yy); nn, pp = Xm.shape
    ll = -nn / 2 * (np.log(2 * np.pi) + np.log(a["SCE"] / nn) + 1)
    press = ((a["e"] / (1 - a["h"])) ** 2).sum()
    return dict(R2=a["R2"], R2_aj=a["R2aj"], AIC=-2 * ll + 2 * pp, BIC=-2 * ll + pp * np.log(nn), R2_pred=1 - press / a["SCT"], PRESS=press)

# verificacion de AIC/BIC contra statsmodels y de PRESS contra LOOCV explicito
c0 = criterios(X, y); s0 = sm.OLS(y, X).fit()
assert abs(c0["AIC"] - s0.aic) < 1e-8 and abs(c0["BIC"] - s0.bic) < 1e-8
loo = np.array([y[i] - X[i] @ np.linalg.lstsq(np.delete(X, i, 0), np.delete(y, i), rcond=None)[0] for i in range(n)])
assert abs((loo ** 2).sum() - c0["PRESS"]) < 1e-6

rg = np.random.default_rng(51)
ruido = rg.normal(size=(n, 40))                                           # 40 columnas de ruido puro, independientes de y
filas = []
for k in (0, 2, 5, 10, 20, 40):
    filas.append({"variables de ruido": k, **criterios(np.c_[X, ruido[:, :k]], y)})
tab = pd.DataFrame(filas).set_index("variables de ruido")
print(tab.round(4).to_string())
assert (np.diff(tab["R2"]) > 0).all()                                                      # R2 SIEMPRE sube
assert tab.loc[40, "AIC"] > tab.loc[0, "AIC"] and tab.loc[40, "BIC"] > tab.loc[0, "BIC"] and tab.loc[40, "R2_pred"] < tab.loc[0, "R2_pred"]

fig, ax = plt.subplots(1, 3, figsize=(14, 3.8))
ax[0].plot(tab.index, tab["R2"], "o-", color=PALETA[1], label="R2"); ax[0].plot(tab.index, tab["R2_aj"], "s-", color=PALETA[0], label="R2 ajustado"); ax[0].plot(tab.index, tab["R2_pred"], "^-", color=PALETA[2], label="R2 predictivo (PRESS)")
ax[0].set(title="Ajuste vs generalizacion", xlabel="columnas de ruido"); ax[0].legend()
ax[1].plot(tab.index, tab["AIC"], "o-", color=PALETA[3], label="AIC"); ax[1].plot(tab.index, tab["BIC"], "s-", color=PALETA[4], label="BIC"); ax[1].set(title="AIC y BIC (menos es mejor)", xlabel="columnas de ruido"); ax[1].legend()
ax[2].plot(tab.index, tab["PRESS"], "o-", color=PALETA[5]); ax[2].set(title="PRESS = error LOOCV", xlabel="columnas de ruido")
plt.tight_layout(); plt.show()
````

---
**Lectura.** Agregar columnas de **ruido puro** (generadas sin relación alguna con el rendimiento) sube $R^2$ de **0.618 a 0.719** con 40 columnas, una «mejora» completamente falsa. El $R^2$ ajustado oscila (0.606, 0.609, 0.609, 0.616, 0.613, 0.634): **penaliza poco** con $n=200$ y 40 columnas, y hasta puede subir por azar, así que **no basta** como único criterio. En cambio **AIC** pasa de 911.2 a 930.1, **BIC** de 934.3 a 1 085.1 (castiga más con cada variable) y el **$R^2$ predictivo** (PRESS, equivalente a la validación cruzada «deja uno fuera») **baja de 0.590 a 0.524**, mientras PRESS sube de 1 116 a 1 296. Nuestro AIC y BIC coinciden con `statsmodels`, y el PRESS por la matriz sombrero coincide con 200 reajustes explícitos. **Moraleja:** el ajuste en los datos de entrenamiento premia el sobreajuste; lo que cuenta es el desempeño con datos que el modelo **no vio** (Módulo 06).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Forma matricial** | $\hat{\boldsymbol\beta}=(\mathbf X^\top\mathbf X)^{-1}\mathbf X^\top\mathbf y$; ecuaciones normales, QR y `lstsq` coinciden, pero `lstsq`/QR son más estables ($\operatorname{cond}(\mathbf X^\top\mathbf X)=\operatorname{cond}(\mathbf X)^2$); $\mathbf H$ es simétrica, idempotente, con traza $p$ |
| **Interpretación** | Cada $\beta_j$ es el efecto *ceteris paribus*; **cambia** si omites una variable correlacionada (sesgo por variable omitida) y **no es causal** por sí solo |
| **Categóricas** | $k-1$ indicadoras con nivel de referencia; con intercepto y las $k$ indicadoras, $\mathbf X$ pierde rango (trampa de la variable ficticia) |
| **F parcial** | Compara modelos anidados con $F=\frac{(SCE_R-SCE_C)/q}{SCE_C/(n-p_C)}$ (idéntico a `anova_lm`) |
| **Multicolinealidad** | $\operatorname{VIF}_j=1/(1-R_j^2)=[\mathbf R^{-1}]_{jj}$; con altitud y temperatura casi iguales los EE se multiplican por más de 8 y el coeficiente se vuelve inestable, **sin** dañar la predicción |
| **Influencia** | $h_{ii}$, $t_i$ studentizado y **Cook** $D_i$ distinguen apalancar, ser atípico e influir; investiga y documenta antes de borrar |
| **Supuestos** | Breusch-Pagan, Durbin-Watson y Shapiro apoyan (no sustituyen) los gráficos; con heterocedasticidad, **HC3** corrige los EE y la cobertura de los IC |
| **Selección** | $R^2$ siempre sube; $R^2_{\text{aj}}$, **AIC, BIC y PRESS (LOOCV)** castigan variables de ruido; la generalización se mide con datos que el modelo no vio |
