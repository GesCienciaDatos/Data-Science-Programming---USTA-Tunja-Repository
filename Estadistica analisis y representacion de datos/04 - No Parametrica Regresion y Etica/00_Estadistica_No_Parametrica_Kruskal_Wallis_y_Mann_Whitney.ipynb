## 1. Rangos: Ordenar en Lugar de Promediar 🏁

**Idea.** Las pruebas paramétricas usan los **valores** (medias, varianzas). Las pruebas de **rangos** reemplazan cada observación por su **posición** en el orden de la muestra combinada: el menor valor recibe el rango 1, el siguiente el 2, y así hasta $N$. Si varios valores **empatan**, cada uno recibe el **promedio** de las posiciones que ocuparían (tres valores iguales en las posiciones 2, 3 y 4 reciben rango 3).

| Situación | Prueba paramétrica | Equivalente de rangos | Qué contrasta |
|---|---|---|---|
| Dos grupos independientes | $t$ de Student / Welch | **Mann-Whitney** (o Wilcoxon de suma de rangos) | $P(X>Y)=\tfrac12$ (dominancia estocástica) |
| Dos mediciones pareadas | $t$ pareada | **Wilcoxon de rangos con signo** | simetría de las diferencias alrededor de 0 |
| Tres o más grupos independientes | ANOVA de un factor | **Kruskal-Wallis** | que todos los grupos provienen de la misma distribución |
| Asociación entre dos variables | correlación de Pearson | **Spearman** / **Kendall** (cuaderno 01) | asociación **monótona** |

**Por qué sirven.** (1) No suponen normalidad: bajo la hipótesis nula de que todos los datos provienen de la **misma distribución continua**, los rangos son una permutación al azar de $1,\dots,N$, de modo que la distribución del estadístico es **exacta** y **no depende de la población**. (2) Son **resistentes** a valores extremos: un dato 10 veces mayor sigue teniendo «solo» el rango $N$. (3) Sirven con datos **ordinales** (escala de satisfacción, nivel de infestación bajo/medio/alto). **El precio:** se descarta la información sobre las **distancias** entre valores, de modo que, cuando los supuestos de la $t$ sí se cumplen, la prueba de rangos pierde algo de potencia (veremos cuánto).

````python
# ============================================================
# 1. Rangos con empates, desde cero
# ============================================================
def rangos(x):
    """Rangos 1..n; los empates reciben el promedio de las posiciones que ocuparian."""
    x = np.asarray(x, dtype=float)
    orden = np.argsort(x, kind="mergesort")
    xs = x[orden]
    bloque = np.cumsum(np.r_[True, xs[1:] != xs[:-1]]) - 1             # indice de cada grupo de valores iguales
    posicion = np.arange(1, len(x) + 1)
    promedio = np.bincount(bloque, weights=posicion) / np.bincount(bloque)
    r = np.empty(len(x))
    r[orden] = promedio[bloque]
    return r

# (a) verificacion contra SciPy, con empates
x_demo = np.array([3.1, 5.0, 5.0, 2.2, 9.9, 5.0, 2.2])
assert np.allclose(rangos(x_demo), stats.rankdata(x_demo))
print("valores:", x_demo)
print("rangos :", rangos(x_demo), "  (los tres 5.0 ocupan las posiciones 4, 5 y 6 -> rango 5)")

# (b) un dato extremo cambia mucho la media y nada los rangos
rg = np.random.default_rng(11)
datos = np.round(rg.normal(50, 8, size=12), 1)                          # 12 pesos (kg) SINTETICOS
con_atipico = datos.copy()
con_atipico[np.argmax(datos)] = 500.0                                   # error de digitacion: 50.0 -> 500.0
print(f"\nMedia    sin/con atipico: {datos.mean():6.1f} / {con_atipico.mean():6.1f}")
print(f"Mediana  sin/con atipico: {np.median(datos):6.1f} / {np.median(con_atipico):6.1f}")
print(f"Desv. est. sin/con atipico: {datos.std(ddof=1):5.1f} / {con_atipico.std(ddof=1):5.1f}")
assert np.array_equal(rangos(datos), rangos(con_atipico))               # los rangos NO cambian
print("Rangos identicos con y sin el dato extremo:", np.array_equal(rangos(datos), rangos(con_atipico)))

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(datos, "o-", color=PALETA[0], label="original"); ax[0].plot(con_atipico, "s--", color=PALETA[1], label="con un atipico")
ax[0].set(title="Valores: el atipico domina la escala", xlabel="observacion", ylabel="peso (kg)"); ax[0].legend()
ax[1].plot(rangos(datos), "o-", color=PALETA[0], label="original"); ax[1].plot(rangos(con_atipico), "s--", color=PALETA[1], label="con un atipico")
ax[1].set(title="Rangos: exactamente los mismos", xlabel="observacion", ylabel="rango"); ax[1].legend()
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Nuestro `rangos` coincide con `scipy.stats.rankdata`, incluidos los empates (los tres valores 5.0 reciben el rango 5). (2) Al cambiar **un solo** peso de 5x kg a 500 (un error de digitación), la **media** pasa de **51.5 a 87.9** y la **desviación estándar** de **7.8 a 130.0**, mientras la **mediana** sigue en **50.0** y los **rangos son exactamente los mismos**. Cualquier prueba basada solo en los rangos es **inmune** a ese error; una basada en medias y varianzas, no. (3) **Ojo:** esa resistencia tiene un costo, porque ignorar las distancias entre valores desperdicia información cuando los datos sí son «bien portados» (sección 3).

## 2. Mann-Whitney: Dos Grupos Independientes 🥊

**Estadístico.** Con muestras de tamaños $n_1$ y $n_2$ ($N=n_1+n_2$), se ordenan los $N$ datos juntos y se suman los rangos de la primera muestra, $R_1$. El estadístico es

$$U_1 = R_1 - \frac{n_1(n_1+1)}{2},\qquad U_1+U_2=n_1 n_2 ,$$

y **cuenta, entre los $n_1 n_2$ pares $(x_i,y_j)$, cuántos tienen $x_i>y_j$** (los empates cuentan $\tfrac12$). Bajo $H_0$ (misma distribución), $\mu_U=\dfrac{n_1 n_2}{2}$ y, con **corrección por empates** ($t_g$ = tamaño de cada grupo de valores iguales),

$$\sigma_U^2=\frac{n_1 n_2}{12}\left[(N+1)-\frac{\sum_g (t_g^3-t_g)}{N(N-1)}\right],\qquad z=\frac{\lvert U-\mu_U\rvert-\tfrac12}{\sigma_U}\ \ (\text{con corrección de continuidad}).$$

Con muestras pequeñas y **sin empates** se usa la distribución **exacta**: el número de ordenamientos con $U=u$ obedece la recurrencia $c(m,n,u)=c(m-1,n,u-n)+c(m,n-1,u)$ (el mayor valor viene de la muestra 1 y supera a los $n$ de la muestra 2, o viene de la muestra 2 y no suma nada).

**Qué contrasta (y qué no).** $H_0$: $P(X>Y)+\tfrac12P(X=Y)=\tfrac12$. Solo equivale a «igualdad de medianas» si las dos distribuciones tienen la **misma forma** (solo cambia la ubicación). El **tamaño del efecto** natural es la **probabilidad de superioridad** $\theta=U_1/(n_1n_2)$ —coincide con el **área bajo la curva ROC** (AUC) del Módulo de métricas de Machine Learning— y la **correlación biserial por rangos** $r_{rb}=2\theta-1\in[-1,1]$. Para un efecto en las **unidades** de los datos se usa el estimador de **Hodges-Lehmann**, la mediana de las $n_1n_2$ diferencias $x_i-y_j$.

````python
# ============================================================
# 2. Mann-Whitney desde cero: asintotico, exacto, permutacion, efectos
# ============================================================
from functools import lru_cache
from math import comb
from sklearn.metrics import roc_auc_score

def mann_whitney(x, y, continuidad=True):
    """U de Mann-Whitney con varianza corregida por empates y valor p bilateral asintotico."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    n1, n2 = len(x), len(y); N = n1 + n2
    r = rangos(np.r_[x, y])
    U1 = r[:n1].sum() - n1 * (n1 + 1) / 2
    U2 = n1 * n2 - U1
    _, t = np.unique(np.r_[x, y], return_counts=True)
    var = n1 * n2 / 12 * ((N + 1) - (t ** 3 - t).sum() / (N * (N - 1)))
    z = (abs(U1 - n1 * n2 / 2) - (0.5 if continuidad else 0.0)) / np.sqrt(var)
    return {"U1": U1, "U2": U2, "z": z, "p": min(1.0, 2 * stats.norm.sf(max(z, 0)))}

@lru_cache(maxsize=None)
def cuenta_u(m, n, u):
    """Cuantos ordenamientos de m+n valores distintos tienen estadistico U = u."""
    if u < 0: return 0
    if m == 0 or n == 0: return 1 if u == 0 else 0
    return cuenta_u(m - 1, n, u - n) + cuenta_u(m, n - 1, u)

def p_exacto_u(U1, n1, n2):
    """Valor p bilateral exacto (sin empates): 2 * P(U <= min(U1, U2))."""
    u_min = int(min(U1, n1 * n2 - U1))
    cola = sum(cuenta_u(n1, n2, k) for k in range(u_min + 1))
    return min(1.0, 2 * cola / comb(n1 + n2, n1))

# --- Datos SINTETICOS: kg por parcela en dos fincas (lognormales: asimetricas, como suelen ser los rendimientos)
rg = np.random.default_rng(101)
finca_a = np.round(rg.lognormal(3.0, 0.6, 18), 1)
finca_b = np.round(rg.lognormal(3.4, 0.6, 22), 1)

mw = mann_whitney(finca_a, finca_b)
ref = stats.mannwhitneyu(finca_a, finca_b, method="asymptotic", use_continuity=True)
t_welch = stats.ttest_ind(finca_a, finca_b, equal_var=False)
assert abs(mw["U1"] - ref.statistic) < 1e-9 and abs(mw["p"] - ref.pvalue) < 1e-12      # coincide con SciPy
print(f"Finca A: mediana {np.median(finca_a):.1f} kg (n={len(finca_a)}) | Finca B: mediana {np.median(finca_b):.1f} kg (n={len(finca_b)})")
print(f"Mann-Whitney: U1 = {mw['U1']:.1f}, z = {mw['z']:.3f}, p = {mw['p']:.4f}   (SciPy: U1 = {ref.statistic:.1f}, p = {ref.pvalue:.4f})")
print(f"t de Welch  : t = {t_welch.statistic:.3f}, p = {t_welch.pvalue:.4f}")

# --- Valor p exacto con muestras pequenas y sin empates (datos continuos sin redondear)
x_peq, y_peq = rg.lognormal(3.0, 0.6, 8), rg.lognormal(3.5, 0.6, 9)
U_peq = mann_whitney(x_peq, y_peq)["U1"]
p_ex = p_exacto_u(U_peq, 8, 9)
p_ex_sp = stats.mannwhitneyu(x_peq, y_peq, method="exact").pvalue
assert abs(p_ex - p_ex_sp) < 1e-12
print(f"\nMuestras pequenas (8 y 9): U1 = {U_peq:.0f} | p exacto = {p_ex:.4f} (SciPy exacto: {p_ex_sp:.4f}) | p asintotico = {mann_whitney(x_peq, y_peq)['p']:.4f}")

# --- Prueba de permutacion: reasignar al azar las etiquetas de grupo (20 000 veces)
n1, n2 = len(finca_a), len(finca_b); N = n1 + n2
r_comb = rangos(np.r_[finca_a, finca_b])
perm = np.random.default_rng(7).random((20_000, N)).argsort(axis=1)[:, :n1]          # n1 posiciones al azar por permutacion
U_perm = r_comb[perm].sum(axis=1) - n1 * (n1 + 1) / 2
p_perm = (np.abs(U_perm - n1 * n2 / 2) >= abs(mw["U1"] - n1 * n2 / 2) - 1e-9).mean()
print(f"p por permutacion = {p_perm:.4f} (asintotico: {mw['p']:.4f})")
assert abs(p_perm - mw["p"]) < 0.02

# --- Tamanos del efecto
theta = mw["U1"] / (n1 * n2)                                    # P(A > B) + 0.5 P(A = B)
r_rb = 2 * theta - 1
auc = roc_auc_score(np.r_[np.zeros(n1), np.ones(n2)], -np.r_[finca_a, finca_b])         # AUC de "ser de A" usando -valor como puntaje
assert abs(theta - auc) < 1e-12
dif = (finca_a[:, None] - finca_b[None, :]).ravel()
hl = np.median(dif)                                              # estimador de Hodges-Lehmann (A - B)
bs = np.random.default_rng(8)
hl_boot = np.array([np.median((bs.choice(finca_a, n1)[:, None] - bs.choice(finca_b, n2)[None, :])) for _ in range(2000)])
ic = np.percentile(hl_boot, [2.5, 97.5])
print(f"\nP(A > B) = {theta:.3f} (AUC = {auc:.3f}) | r biserial por rangos = {r_rb:.3f}")
print(f"Hodges-Lehmann (A - B) = {hl:.1f} kg | IC 95 % bootstrap = [{ic[0]:.1f}, {ic[1]:.1f}] kg")
````

---
**Lectura.** (1) Nuestro $U$ y su valor $p$ asintótico coinciden con `scipy.stats.mannwhitneyu` (aprobado por `assert`): $U_1=128$, $z=1.890$, **$p=0.0588$**. La $t$ de Welch sobre los mismos datos da $p=0.0470$: **las dos pruebas quedan a ambos lados de 0.05** con la misma muestra. Eso **no** es una contradicción que haya que «resolver» escogiendo la prueba que da significancia (sería *p-hacking*, cuaderno 03): ambas dicen que hay evidencia **moderada** de que la finca B rinde más, y el 0.05 es una convención, no un precipicio. La decisión sobre qué prueba usar se toma **antes** de ver los resultados, según los supuestos (datos asimétricos: Mann-Whitney). (2) Con muestras pequeñas (8 y 9) el $p$ **exacto** (0.0927) y el asintótico (0.0922) casi coinciden, y el exacto de SciPy es idéntico al nuestro. (3) La **permutación** (20 000 reasignaciones) da $p=0.0600$, de acuerdo con el asintótico. (4) **Tamaño del efecto:** $\theta=0.323$ (idéntico al AUC de scikit-learn): al tomar una parcela al azar de cada finca, la de A supera a la de B en solo **32 %** de los casos; $r_{rb}=-0.354$ (efecto mediano). **Hodges-Lehmann:** la finca A rinde **9.7 kg menos** por parcela, con IC 95 % *bootstrap* de **[−20.6, 0.2] kg**, un intervalo que apenas roza el cero: consistente con la evidencia moderada. Reportar solo «$p=0.059$, no significativo» **ocultaría** que el efecto estimado es grande.

##### 🛠️ Práctica 1: U de Mann-Whitney y tamaño del efecto en dos lotes

Con los arreglos `finca_a` y `finca_b` de la sección 2 y la función `rangos` definida arriba:

## 3. ¿Cuándo Gana Mann-Whitney y Cuándo No? Una Simulación Honesta 🎲

Que una prueba «no necesite normalidad» **no** significa que sea siempre mejor. Simulamos **4 000 experimentos** por escenario y medimos (a) el **error tipo I** (rechazos cuando $H_0$ es cierta; debería ser ≈ 5 %) y (b) la **potencia** (rechazos cuando sí hay un desplazamiento), para la $t$ de **Student** (varianzas iguales), la $t$ de **Welch** y **Mann-Whitney**. Escenarios: normal, lognormal (asimétrica), $t$ con 2 grados de libertad (colas muy pesadas) y dos casos de **varianzas distintas con tamaños distintos**, donde $H_0$ se cumple en el sentido de «mismo centro, simétricas».

````python
# ============================================================
# 3. Error tipo I y potencia: t de Student, t de Welch y Mann-Whitney
# ============================================================
def simula(gen_x, gen_y, n1, n2, reps=4000, alpha=0.05, seed=0):
    """Tasa de rechazo (p < alpha) de tres pruebas bilaterales; datos continuos (sin empates)."""
    rg = np.random.default_rng(seed)
    X, Y = gen_x(rg, (reps, n1)), gen_y(rg, (reps, n2))
    m1, m2, v1, v2 = X.mean(1), Y.mean(1), X.var(1, ddof=1), Y.var(1, ddof=1)
    sp2 = ((n1 - 1) * v1 + (n2 - 1) * v2) / (n1 + n2 - 2)
    p_s = 2 * stats.t.sf(np.abs((m1 - m2) / np.sqrt(sp2 * (1 / n1 + 1 / n2))), n1 + n2 - 2)
    e1, e2 = v1 / n1, v2 / n2
    gl = (e1 + e2) ** 2 / (e1 ** 2 / (n1 - 1) + e2 ** 2 / (n2 - 1))
    p_w = 2 * stats.t.sf(np.abs((m1 - m2) / np.sqrt(e1 + e2)), gl)
    R = np.concatenate([X, Y], axis=1).argsort(1).argsort(1) + 1
    U1 = R[:, :n1].sum(1) - n1 * (n1 + 1) / 2
    z = np.maximum(np.abs(U1 - n1 * n2 / 2) - 0.5, 0) / np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)
    p_u = 2 * stats.norm.sf(z)
    return np.array([(p_s < alpha).mean(), (p_w < alpha).mean(), (p_u < alpha).mean()])

normal = lambda s=1.0, c=0.0: (lambda rg, size: c + s * rg.normal(size=size))
lognorm = lambda c=0.0: (lambda rg, size: c + rg.lognormal(0, 1, size=size))
t2 = lambda c=0.0: (lambda rg, size: c + rg.standard_t(2, size=size))

esc = [  # (nombre, generador X, generador Y bajo H0, generador Y bajo H1, n1, n2)
    ("Normal (n=20 y 20, desplazamiento 0.8)", normal(), normal(), normal(c=0.8), 20, 20),
    ("Lognormal asimetrica (20 y 20, desplaz. 1.0)", lognorm(), lognorm(), lognorm(c=1.0), 20, 20),
    ("t con 2 gl, colas pesadas (20 y 20, desplaz. 1.5)", t2(), t2(), t2(c=1.5), 20, 20),
]
filas = []
for nombre, gx, gy0, gy1, n1_, n2_ in esc:
    filas.append([nombre, *simula(gx, gy0, n1_, n2_, seed=1), *simula(gx, gy1, n1_, n2_, seed=2)])
# Varianzas distintas y tamanos distintos: H0 cierta en el sentido de 'mismo centro, simetricas'
filas.append(["Normal, sigma 1 (n=10) vs sigma 3 (n=40)", *simula(normal(1), normal(3), 10, 40, seed=3), np.nan, np.nan, np.nan])
filas.append(["Normal, sigma 3 (n=10) vs sigma 1 (n=40)", *simula(normal(3), normal(1), 10, 40, seed=4), np.nan, np.nan, np.nan])
cols = pd.MultiIndex.from_product([["Error tipo I (H0 cierta)", "Potencia (hay desplazamiento)"], ["t Student", "t Welch", "Mann-Whitney"]])
tabla = pd.DataFrame([f[1:] for f in filas], index=[f[0] for f in filas], columns=cols)
print(tabla.round(3).to_string())

fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
anchos = np.arange(len(filas)); w = 0.27
for k, nombre_p in enumerate(["t Student", "t Welch", "Mann-Whitney"]):
    ax[0].bar(anchos + (k - 1) * w, tabla[("Error tipo I (H0 cierta)", nombre_p)], w, label=nombre_p, color=PALETA[k])
    ax[1].bar(anchos[:3] + (k - 1) * w, tabla[("Potencia (hay desplazamiento)", nombre_p)].iloc[:3], w, label=nombre_p, color=PALETA[k])
ax[0].axhline(0.05, color="black", ls="--"); ax[0].set(title="Error tipo I (objetivo: 0.05)", xticks=anchos, xticklabels=["normal", "lognormal", "t(2)", "s1 n10\ns3 n40", "s3 n10\ns1 n40"], ylabel="tasa de rechazo")
ax[1].set(title="Potencia (mas alto es mejor)", xticks=anchos[:3], xticklabels=["normal", "lognormal", "t(2)"]); ax[0].legend()
plt.tight_layout(); plt.show()

tipo1 = tabla["Error tipo I (H0 cierta)"]; pot = tabla["Potencia (hay desplazamiento)"].iloc[:3]
assert (tipo1.iloc[:3] < 0.07).all().all() and (tipo1.iloc[:3] > 0.02).all().all()    # con mismas varianzas las tres estan calibradas
assert pot.loc[pot.index[1], "Mann-Whitney"] > pot.loc[pot.index[1], "t Student"] + 0.10   # lognormal: gana Mann-Whitney
assert pot.loc[pot.index[2], "Mann-Whitney"] > pot.loc[pot.index[2], "t Student"] + 0.10   # colas pesadas: gana Mann-Whitney
assert pot.loc[pot.index[0], "t Student"] >= pot.loc[pot.index[0], "Mann-Whitney"] - 0.01  # normal: la t es (casi) igual o mejor
````

---
**Lectura.** (Cada tasa viene de 4 000 experimentos; el error de simulación es de ±0.7 puntos.) (1) **Datos normales:** las tres pruebas están calibradas (5.0 %, 5.0 % y 4.8 %) y la potencia de Mann-Whitney (**0.660**) es solo ≈ 2.5 puntos menor que la de la $t$ (**0.685**): perder tan poco es el precio de no suponer normalidad (la eficiencia asintótica relativa teórica es $3/\pi\approx0.955$). (2) **Lognormal y colas pesadas:** Mann-Whitney **gana por mucho** (potencia **0.844** frente a **0.450** con la lognormal, y **0.872** frente a **0.604** con la $t(2)$), porque la media y la varianza son estimadores pésimos en esas poblaciones. (3) **Varianzas distintas con tamaños distintos** (mismo centro, distribuciones simétricas): la $t$ de Student se rompe **en ambas direcciones** (**0.1 %** de rechazos cuando el grupo pequeño tiene la menor varianza, **25.6 %** cuando tiene la mayor); **Welch se mantiene en ≈ 5 %**. Y **Mann-Whitney tampoco está a salvo**: da 1.0 % y **13.6 %**, porque su varianza bajo $H_0$ supone que **las distribuciones son idénticas**. Si hay heterocedasticidad, se necesita una prueba pensada para ello (Welch para medias; Brunner-Munzel, `scipy.stats.brunnermunzel`, para $P(X>Y)$ *(citado de memoria, verificar)*). **«No paramétrica» no significa «sin supuestos»**: cambia el supuesto de normalidad por el de forma/dispersión comparable.

## 4. Muestras Pareadas: Wilcoxon de Rangos con Signo 🔗

**Situación.** La misma unidad se mide **dos veces** (antes y después de un cambio) y se analizan las diferencias $d_i$. La $t$ pareada exige que las $d_i$ sean aproximadamente normales. El **Wilcoxon de rangos con signo** solo exige que las $d_i$ sean **simétricas** alrededor de su mediana: se descartan las diferencias nulas, se ordenan las $\lvert d_i\rvert$ y se suman los rangos de las diferencias **positivas**,

$$W^{+}=\sum_{d_i>0} R(\lvert d_i\rvert),\qquad \mu_W=\frac{n(n+1)}{4},\qquad \sigma_W^2=\frac{n(n+1)(2n+1)}{24}.$$

Bajo $H_0$ cada signo es $+$ o $-$ con probabilidad $\tfrac12$, **independiente** de los rangos: con $n$ pares hay $2^n$ combinaciones de signos igualmente probables, y la distribución **exacta** se obtiene **enumerándolas** ($n\le 15$ es trivial). La **prueba del signo** (solo cuenta cuántas $d_i$ son positivas, $\text{Binomial}(n,\tfrac12)$) es aún más robusta pero **menos potente**, porque ignora la magnitud.

````python
# ============================================================
# 4. Wilcoxon de rangos con signo: exacto por enumeracion de los 2^n signos
# ============================================================
def wilcoxon_exacto(d):
    """(W+, p bilateral exacto) por enumeracion de todas las combinaciones de signos."""
    d = np.asarray(d, float); d = d[d != 0]; n = len(d)
    r = rangos(np.abs(d)); w_obs = r[d > 0].sum()
    signos = (np.arange(2 ** n)[:, None] >> np.arange(n)) & 1             # matriz 2^n x n de ceros y unos
    W = signos @ r
    centro = n * (n + 1) / 4
    return w_obs, float(np.mean(np.abs(W - centro) >= abs(w_obs - centro) - 1e-9))

def wilcoxon_asintotico(d, continuidad=True):
    d = np.asarray(d, float); d = d[d != 0]; n = len(d)
    r = rangos(np.abs(d)); w = r[d > 0].sum()
    z = (abs(w - n * (n + 1) / 4) - (0.5 if continuidad else 0)) / np.sqrt(n * (n + 1) * (2 * n + 1) / 24)
    return 2 * stats.norm.sf(z)

# --- Datos SINTETICOS: kg por parcela en las MISMAS 12 parcelas, antes y despues de cambiar el riego
rg = np.random.default_rng(202)
antes = rg.normal(30, 4, 12)
despues = antes + rg.normal(1.0, 1.5, 12)
despues[-1] += 40.0                                                       # error de digitacion: se anoto 40 kg de mas en la ultima parcela
d = despues - antes
w, p_ex = wilcoxon_exacto(d)
ref = stats.wilcoxon(d, method="exact")
assert abs(w - ref.statistic) < 1e-9 or abs(w - (12 * 13 / 2 - ref.statistic)) < 1e-9     # SciPy reporta el menor de W+ y W-
assert abs(p_ex - ref.pvalue) < 1e-12
assert abs(wilcoxon_asintotico(d) - stats.wilcoxon(d, method="approx", correction=True).pvalue) < 1e-9
t_par = stats.ttest_rel(despues, antes)
p_signo = stats.binomtest(int((d > 0).sum()), len(d), 0.5).pvalue
print(f"Diferencias (kg): {np.round(d, 1)}")
print(f"Diferencia media = {d.mean():.2f} kg | mediana = {np.median(d):.2f} kg | positivas: {(d > 0).sum()} de {len(d)}")
print(f"Wilcoxon: W+ = {w:.1f}, p exacto = {p_ex:.4f} (asintotico {wilcoxon_asintotico(d):.4f}, SciPy {ref.pvalue:.4f})")
print(f"t pareada: t = {t_par.statistic:.3f}, p = {t_par.pvalue:.4f} | prueba del signo: p = {p_signo:.4f}")

# Efecto del error de digitacion: quitar la ultima parcela
d_sin = d[:-1]
print(f"\nSin la parcela con error: media = {d_sin.mean():.2f} | t pareada p = {stats.ttest_1samp(d_sin, 0).pvalue:.4f} | Wilcoxon p exacto = {wilcoxon_exacto(d_sin)[1]:.4f}")
````

---
**Lectura.** (1) Las 12 parcelas mejoran casi todas entre 0 y 3 kg, pero a la última se le **anotaron 40 kg de más** por error: la diferencia media salta a **4.85 kg** frente a una mediana de **1.77 kg**. (2) La **$t$ pareada** queda **destruida** por el dato erróneo: $t=1.52$, **$p=0.156$**, porque la desviación estándar de las diferencias explota. **Wilcoxon** da $W^{+}=76$ (de un máximo de 78) y **$p$ exacto $=0.0015$**; el atípico solo aporta el rango 12, sin importar su tamaño. La **prueba del signo** (11 positivas de 12) da $p=0.0063$: también robusta pero con menos potencia, porque ignora las magnitudes. (3) Nuestro valor exacto coincide con `scipy.stats.wilcoxon(method="exact")`; la aproximación normal con corrección de continuidad ($p=0.0042$) es **conservadora** con $n=12$, por eso con muestras pequeñas conviene el valor exacto. (4) Quitando la parcela con error, la $t$ pareada recupera su potencia ($p=0.0004$) y Wilcoxon sigue concluyendo lo mismo ($p=0.0029$; con $n=11$ el mínimo posible es $2/2^{11}\approx0.001$). **Moraleja:** la mejor solución para un error de digitación es **corregirlo en los datos** (cuaderno de limpieza, Módulo 05); la prueba de rangos protege cuando el valor extremo es **real** y no se puede descartar.

## 5. Kruskal-Wallis y Comparaciones por Pares (Dunn + Holm) 🏆

**Estadístico.** Con $k$ grupos de tamaños $n_i$ se ordenan los $N$ datos juntos y $R_i$ es la suma de rangos del grupo $i$:

$$H=\frac{12}{N(N+1)}\sum_{i=1}^{k}\frac{R_i^{2}}{n_i}-3(N+1),\qquad H_{\text{corr}}=\frac{H}{C},\quad C=1-\frac{\sum_g (t_g^3-t_g)}{N^3-N}.$$

Mide qué tan lejos están los **rangos promedio** de los grupos del rango promedio global $(N+1)/2$. Bajo $H_0$ y con grupos no demasiado pequeños, $H_{\text{corr}}\approx\chi^2_{k-1}$. Con **conteos** (muchos empates) la corrección $C$ **importa**. **Tamaño del efecto:** $\varepsilon^2=H/(N-1)\in[0,1]$ (la fracción de la «variabilidad de los rangos» asociada al grupo). Hace falta además que los grupos tengan **la misma forma** (y dispersión parecida) para leer el resultado como «diferencia de ubicación»; si no, solo contrasta «alguna distribución difiere».

**Post hoc.** Si $H_0$ se rechaza, la prueba de **Dunn** compara cada par con los rangos promedio $\bar R_i=R_i/n_i$:

$$z_{ij}=\frac{\bar R_i-\bar R_j}{\sqrt{\left[\dfrac{N(N+1)}{12}-\dfrac{\sum_g (t_g^3-t_g)}{12(N-1)}\right]\left(\dfrac1{n_i}+\dfrac1{n_j}\right)}}$$

y los $\binom{k}{2}$ valores $p$ se ajustan por **comparaciones múltiples** con **Holm** (ordena los $p_{(1)}\le\dots\le p_{(m)}$ y usa $\tilde p_{(j)}=\max_{l\le j}\,\min\{1,(m-l+1)\,p_{(l)}\}$; es uniformemente más potente que Bonferroni con el mismo control del error por familia). Con $k=2$, $z^2=H$ exactamente: Kruskal-Wallis **generaliza** a Mann-Whitney.

````python
# ============================================================
# 5. Kruskal-Wallis y Dunn con ajuste de Holm, desde cero
# ============================================================
import itertools

def kruskal_wallis(grupos):
    """H corregida por empates, valor p chi-cuadrado, epsilon^2 y rangos promedio por grupo."""
    n = np.array([len(g) for g in grupos]); N, k = n.sum(), len(grupos)
    r = rangos(np.concatenate(grupos))
    R = np.array([s.sum() for s in np.split(r, np.cumsum(n)[:-1])])
    H = 12 / (N * (N + 1)) * np.sum(R ** 2 / n) - 3 * (N + 1)
    _, t = np.unique(np.concatenate(grupos), return_counts=True)
    H /= 1 - (t ** 3 - t).sum() / (N ** 3 - N)
    return H, stats.chi2.sf(H, k - 1), H / (N - 1), R / n

def holm(p):
    """p ajustados por Holm (control del error por familia)."""
    p = np.asarray(p, float); m = len(p); orden = np.argsort(p)
    ajust = np.empty(m); corrido = 0.0
    for j, i in enumerate(orden):
        corrido = max(corrido, (m - j) * p[i]); ajust[i] = min(1.0, corrido)
    return ajust

def dunn(grupos, nombres):
    n = np.array([len(g) for g in grupos]); N = n.sum()
    todos = np.concatenate(grupos); _, t = np.unique(todos, return_counts=True)
    r = rangos(todos); Rm = np.array([s.mean() for s in np.split(r, np.cumsum(n)[:-1])])
    s2 = N * (N + 1) / 12 - (t ** 3 - t).sum() / (12 * (N - 1))
    filas = []
    for i, j in itertools.combinations(range(len(grupos)), 2):
        z = (Rm[i] - Rm[j]) / np.sqrt(s2 * (1 / n[i] + 1 / n[j]))
        filas.append((f"{nombres[i]} vs {nombres[j]}", Rm[i] - Rm[j], z, 2 * stats.norm.sf(abs(z))))
    df = pd.DataFrame(filas, columns=["par", "dif. rangos prom.", "z", "p crudo"]).set_index("par")
    df["p Holm"] = holm(df["p crudo"]); df["p Bonferroni"] = np.minimum(1, df["p crudo"] * len(df))
    return df

# --- Datos SINTETICOS: insectos por trampa en tres fincas (conteos sobredispersos -> muchos empates)
rg = np.random.default_rng(303)
nombres = ["Norte", "Centro", "Sur"]
media_g, tam = [4.0, 5.5, 9.0], [15, 16, 14]
grupos = [rg.negative_binomial(3, 3 / (3 + m), size=n_) for m, n_ in zip(media_g, tam)]
H, p_H, eps2, rbar = kruskal_wallis(grupos)
ref = stats.kruskal(*grupos)
assert abs(H - ref.statistic) < 1e-9 and abs(p_H - ref.pvalue) < 1e-12
f_oneway = stats.f_oneway(*grupos)

# permutacion de H (distribucion exacta condicional a los datos), 10 000 reasignaciones
todos = np.concatenate(grupos); N = len(todos); n_g = np.array(tam)
r_all = rangos(todos); inicio = np.r_[0, np.cumsum(n_g)[:-1]]
perm_r = r_all[np.random.default_rng(9).random((10_000, N)).argsort(axis=1)]
R_perm = np.add.reduceat(perm_r, inicio, axis=1)
_, t_emp = np.unique(todos, return_counts=True)
H_perm = (12 / (N * (N + 1)) * (R_perm ** 2 / n_g).sum(1) - 3 * (N + 1)) / (1 - (t_emp ** 3 - t_emp).sum() / (N ** 3 - N))
p_perm = (H_perm >= H - 1e-9).mean()

print(pd.DataFrame({"n": tam, "media": [g.mean() for g in grupos], "mediana": [np.median(g) for g in grupos], "rango promedio": rbar}, index=nombres).round(2))
print(f"\nKruskal-Wallis: H = {H:.3f}, gl = 2, p = {p_H:.2e} (SciPy: {ref.statistic:.3f}, {ref.pvalue:.2e}) | p por permutacion = {p_perm:.4f} (con 10 000 reasignaciones, el minimo posible es 1e-4)")
print(f"epsilon^2 = {eps2:.3f}  | ANOVA clasico sobre los mismos conteos: F = {f_oneway.statistic:.3f}, p = {f_oneway.pvalue:.5f}")
print(f"Empates: {len(todos) - len(t_emp)} observaciones repetidas; sin corregir, H seria {H * (1 - (t_emp ** 3 - t_emp).sum() / (N ** 3 - N)):.3f}")
tabla_dunn = dunn(grupos, nombres)
print("\nPost hoc de Dunn:"); print(tabla_dunn.round(4).to_string())

# Kruskal-Wallis con k = 2 es Mann-Whitney: H = z^2 (sin correccion de continuidad)
g2 = [grupos[0], grupos[2]]
assert abs(kruskal_wallis(g2)[0] - mann_whitney(*g2, continuidad=False)["z"] ** 2) < 1e-9
assert abs(dunn(g2, ["N", "S"])["z"].iloc[0] ** 2 - kruskal_wallis(g2)[0]) < 1e-9
assert np.allclose(holm(tabla_dunn["p crudo"]), __import__("statsmodels.stats.multitest", fromlist=["x"]).multipletests(tabla_dunn["p crudo"], method="holm")[1])

fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
ax[0].boxplot(grupos, showmeans=True); ax[0].set_xticklabels(nombres)
for i, g in enumerate(grupos): ax[0].scatter(np.full(len(g), i + 1) + rg.uniform(-0.12, 0.12, len(g)), g, s=14, color=PALETA[i], alpha=0.7)
ax[0].set(title="Insectos por trampa (sintetico)", ylabel="conteo")
ax[1].bar(nombres, rbar, color=PALETA[:3]); ax[1].axhline((N + 1) / 2, color="black", ls="--", label="rango promedio esperado bajo H0")
ax[1].set(title="Rango promedio por finca", ylabel="rango promedio"); ax[1].legend()
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Conteos de insectos por trampa en tres fincas sintéticas: rangos promedio de **10.5** (Norte), **27.7** (Centro) y **31.0** (Sur), frente a un esperado de 23 bajo $H_0$. Nuestro $H=21.304$ y su valor $p$ ($2.4\times10^{-5}$) coinciden con `scipy.stats.kruskal`; la **permutación** (10 000 reasignaciones) da $p<10^{-4}$, de acuerdo. **$\varepsilon^2=0.484$:** efecto grande. El ANOVA clásico sobre los mismos conteos también rechaza ($F=9.69$, $p=0.00035$), pero sus supuestos son dudosos con conteos sobredispersos. (2) Hay **30 observaciones repetidas**; sin la corrección por empates $H$ sería **20.966** en vez de **21.304**: aquí el cambio es pequeño, pero con escalas de pocos niveles (por ejemplo, de 1 a 5) puede ser grande. (3) **Dunn + Holm:** Norte difiere de Centro ($p_{\text{Holm}}=0.0005$) y de Sur ($p_{\text{Holm}}=0.0001$); **Centro y Sur no se distinguen** ($p=0.487$). Eso es **ausencia de evidencia, no evidencia de igualdad**: con 16 y 14 trampas, la potencia para una diferencia pequeña es baja. Ajustar con Holm o Bonferroni evita que tres comparaciones dejen la probabilidad de algún falso positivo por encima del 5 %. (4) Con $k=2$ comprobamos que $H=z^2$ del Mann-Whitney sin corrección de continuidad, y que nuestro Holm coincide con `statsmodels`.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Rangos** | Reemplazar valores por posiciones (empates = promedio) vuelve a la prueba **resistente a extremos** y **libre de distribución** bajo $H_0$; descarta la información de las distancias |
| **Mann-Whitney** | $U_1=R_1-n_1(n_1+1)/2$ cuenta pares $x>y$; p asintótico con **corrección por empates**, p **exacto** por recurrencia y p por **permutación** coinciden; contrasta $P(X>Y)=\tfrac12$, **no** «medianas iguales» salvo igualdad de forma |
| **Tamaño del efecto** | $\theta=P(X>Y)+\tfrac12P(X=Y)$ (= AUC), $r_{rb}=2\theta-1$ y **Hodges-Lehmann** con IC *bootstrap* en las unidades de los datos; el valor $p$ solo no basta |
| **Comparación por simulación** | Con datos normales la $t$ gana **poco**; con asimetría o colas pesadas Mann-Whitney gana **mucho**; con varianzas y tamaños distintos **ninguna** de las tres se salva siempre (Welch es la más fiable en medias) |
| **Wilcoxon con signo** | Mide la suma de rangos de las diferencias positivas; exacto por enumeración de $2^n$ signos; protege de una parcela atípica, pero **supone simetría** de las diferencias |
| **Kruskal-Wallis** | $H$ generaliza a Mann-Whitney ($z^2=H$ con $k=2$); la **corrección por empates** es relevante con conteos; $\varepsilon^2=H/(N-1)$ mide el efecto |
| **Dunn + Holm** | Un $H$ significativo no dice **cuáles** grupos difieren: hay que comparar por pares y **ajustar** los valores $p$ por multiplicidad (Holm ≥ Bonferroni en potencia) |
