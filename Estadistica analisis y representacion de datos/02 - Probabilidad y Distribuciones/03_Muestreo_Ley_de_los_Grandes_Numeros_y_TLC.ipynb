## 1. La Distribución Muestral de la Media y el Error Estándar 📏

Sea $X_1,\dots,X_n$ una **muestra aleatoria simple** (independientes e idénticamente distribuidas, i. i. d.) de una población con media $\mu$ y varianza $\sigma^2$. La **media muestral** $\bar X=\frac1n\sum_i X_i$ es una variable aleatoria con

$$E[\bar X]=\mu \quad(\text{insesgada}), \qquad \text{Var}(\bar X)=\frac{\sigma^2}{n}, \qquad \text{EE}(\bar X)=\frac{\sigma}{\sqrt n}.$$

La **desviación estándar** $\sigma$ describe la dispersión de los **datos**; el **error estándar** $\text{EE}$ describe la dispersión del **estimador** y se reduce con $\sqrt n$: para dividir el error a la mitad hay que **cuadruplicar** la muestra. Además:

- **Sin reemplazo** de una población finita de tamaño $N$: $\text{Var}(\bar X)=\dfrac{\sigma^2}{n}\cdot\dfrac{N-n}{N-1}$. El segundo factor es el **factor de población finita** (FPC); es casi 1 cuando $n\ll N$ y se anula cuando $n=N$ (un censo no tiene error de muestreo).
- **Varianza muestral:** $S^2=\frac{1}{n-1}\sum_i (X_i-\bar X)^2$ cumple $E[S^2]=\sigma^2$. Con divisor $n$ el estimador sería **sesgado a la baja** por el factor $(n-1)/n$ (**corrección de Bessel**), porque se mide la dispersión respecto de $\bar X$, que ya está «ajustado» a la muestra.

**Población sintética:** el peso (kg) de $N=5\,000$ bultos de papa de una bodega ficticia, con una distribución **asimétrica a la derecha** (lognormal), para que las conclusiones no dependan de una población ya normal. Calculamos $\mu$ y $\sigma$ de **toda** la población (son parámetros) y comprobamos por simulación con 20 000 muestras repetidas por cada tamaño.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib scipy

import math
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import stats

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")
````

````python
# ============================================================
# 1. Distribucion muestral de la media: E[xbar] = mu, EE = sigma/sqrt(n), FPC sin reemplazo y correccion de Bessel
# ============================================================
rng = np.random.default_rng(SEED)
N = 5000
poblacion = rng.lognormal(np.log(45), 0.35, N)                                  # kg por bulto (SINTETICO, asimetrico a la derecha)
mu, sigma = poblacion.mean(), poblacion.std()                                   # parametros de la poblacion (divisor N)
print(f"Poblacion: N = {N}, mu = {mu:.3f} kg, sigma = {sigma:.3f} kg, asimetria = {stats.skew(poblacion):.2f}")


def medias_muestrales(n, reemplazo=True, R=20_000, rng=rng):
    """R medias muestrales de tamano n; con reemplazo (iid) o sin reemplazo (de la poblacion finita)."""
    if reemplazo:
        return poblacion[rng.integers(0, N, size=(R, n))].mean(axis=1)
    return np.array([poblacion[rng.choice(N, n, replace=False)].mean() for _ in range(R)])


tam = np.array([5, 10, 20, 50, 100, 200, 500])
ee_sim = np.array([medias_muestrales(n).std() for n in tam])
ee_teo = sigma / np.sqrt(tam)
print(pd.DataFrame({"n": tam, "EE simulado": ee_sim, "sigma/sqrt(n)": ee_teo, "error relativo": ee_sim / ee_teo - 1}).round(4).to_string(index=False))
pendiente = np.polyfit(np.log(tam), np.log(ee_sim), 1)[0]
print(f"Pendiente de log(EE) frente a log(n): {pendiente:.3f}  (teoria: -0.5)")
assert np.abs(ee_sim / ee_teo - 1).max() < 0.03 and abs(pendiente + 0.5) < 0.02          # EE = sigma/sqrt(n)
m30 = medias_muestrales(30)
assert abs(m30.mean() - mu) < 4 * sigma / math.sqrt(30 * len(m30))                     # insesgada: E[xbar] = mu

# Sin reemplazo: factor de poblacion finita (n = 1000 de N = 5000)
n_fp = 1000
sin_rem = medias_muestrales(n_fp, reemplazo=False, R=3000)
fpc = math.sqrt((N - n_fp) / (N - 1))
print(f"\nn = {n_fp}: EE con reemplazo = {sigma / math.sqrt(n_fp):.4f} | EE sin reemplazo teorico = {sigma / math.sqrt(n_fp) * fpc:.4f} | simulado = {sin_rem.std():.4f} (FPC = {fpc:.3f})")
assert abs(sin_rem.std() / (sigma / math.sqrt(n_fp) * fpc) - 1) < 0.04 and sin_rem.std() < 0.95 * sigma / math.sqrt(n_fp)

# Correccion de Bessel: n = 5, con reemplazo, E[S^2] = sigma^2 si se divide por n-1
muestras = poblacion[rng.integers(0, N, size=(40_000, 5))]
v_bessel, v_sesgada = muestras.var(axis=1, ddof=1).mean(), muestras.var(axis=1, ddof=0).mean()
print(f"\nVarianza poblacional = {sigma**2:.2f} | E[S^2] con n-1 = {v_bessel:.2f} | con n = {v_sesgada:.2f} (= {v_sesgada / sigma**2:.3f} sigma^2; teoria (n-1)/n = 0.8)")
assert abs(v_bessel / sigma**2 - 1) < 0.03 and abs(v_sesgada / sigma**2 - 0.8) < 0.03

fig, ejes = plt.subplots(1, 3, figsize=(15, 4.2))
ejes[0].hist(poblacion, bins=50, color="#94a3b8", density=True); ejes[0].axvline(mu, color="black", ls="--")
ejes[0].set(title="Población (5 000 bultos): asimétrica", xlabel="kg por bulto")
for n, c in zip((5, 30, 200), PALETA):
    ejes[1].hist(medias_muestrales(n), bins=50, density=True, alpha=0.5, color=c, label=f"n = {n}")
ejes[1].axvline(mu, color="black", ls="--"); ejes[1].set(title="Medias muestrales: centradas en μ y cada vez más concentradas", xlabel="media muestral (kg)"); ejes[1].legend()
ejes[2].loglog(tam, ee_sim, "o", color=PALETA[1], label="EE simulado"); ejes[2].loglog(tam, ee_teo, color="black", label="σ/√n (pendiente −1/2)")
ejes[2].set(title="Error estándar frente a n", xlabel="n", ylabel="EE"); ejes[2].legend()
plt.tight_layout(); plt.show()
````

---
**Lectura.** La población de bultos es asimétrica (asimetría 1.15; $\mu=47.51$ kg, $\sigma=17.17$ kg), pero las medias muestrales se comportan como dice la teoría: el error estándar simulado difiere del $\sigma/\sqrt n$ en menos del 1 % para todos los tamaños y la pendiente log-log es $-0.497$ (teoría $-0.5$). Sin reemplazo, con $n=1000$ de $N=5000$, el error teórico es 0.4857 y el simulado 0.4860 (FPC = 0.895), frente a 0.5430 con reemplazo: **muestrear una quinta parte de la población ya reduce el error un 10 %**. Con $n=5$, la varianza con divisor $n$ da 235.3, es decir **0.798** de la varianza poblacional (294.8; teoría 0.8), mientras que con $n-1$ da 294.1: la corrección de Bessel elimina el sesgo. El histograma de la derecha muestra el mensaje central: con $n=5$ la distribución de $\bar X$ aún hereda algo de la asimetría; con $n=200$ es estrecha y casi simétrica.

## 2. La Ley de los Grandes Números 🔁

**Ley débil.** Si $X_i$ son i. i. d. con media finita $\mu$, entonces para todo $\varepsilon>0$, $\;P\big(\lvert\bar X_n-\mu\rvert>\varepsilon\big)\to 0$ cuando $n\to\infty$. Con varianza finita la prueba es **una línea** con Chebyshev (cuaderno 2):

$$P\big(\lvert\bar X_n-\mu\rvert\ge\varepsilon\big)\ \le\ \frac{\sigma^2}{n\,\varepsilon^2}\ \xrightarrow[n\to\infty]{}\ 0.$$

**Ley fuerte (Kolmogorov).** Bajo $E\lvert X\rvert<\infty$, $\bar X_n\to\mu$ **casi seguramente**: *casi toda* trayectoria $(\bar X_1,\bar X_2,\dots)$ converge a $\mu$. Es lo que justifica la interpretación frecuentista de la probabilidad (cuaderno 0) y los métodos Monte Carlo.

**Dos límites que hay que conocer:**

1. **Sin esperanza no hay ley.** Para la **Cauchy** (cociente de dos normales estándar) no existe $E[X]$; de hecho $\bar X_n$ tiene **la misma distribución Cauchy** para todo $n$: promediar no ayuda. Las colas pesadas (ingresos extremos, pérdidas financieras) son el equivalente práctico.
2. **La ley NO compensa deudas (falacia del apostador).** La **proporción** de caras converge a $1/2$, pero la **diferencia absoluta** $\lvert S_n-n/2\rvert$ crece como $\sqrt n$. Tras una racha de caras, la moneda no «debe» sacar sellos.

````python
# ============================================================
# 2. LGN: medias acumuladas, cota de Chebyshev, el contraejemplo de Cauchy y la falacia del apostador
# ============================================================
rng = np.random.default_rng(SEED)
n_max = 100_000
idx = np.arange(1, n_max + 1)
exp5 = np.cumsum(rng.exponential(1.0, size=(5, n_max)), axis=1) / idx             # media 1
ber = np.cumsum(rng.random(n_max) < 0.3) / idx                                      # proporcion de exitos, p = 0.3
cau5 = np.cumsum(rng.standard_cauchy(size=(5, n_max)), axis=1) / idx                # no tiene esperanza

fig, ejes = plt.subplots(1, 3, figsize=(16, 4.2))
for i in range(5):
    ejes[0].semilogx(idx, exp5[i], lw=1, color=PALETA[i % 5])
    ejes[2].semilogx(idx, cau5[i], lw=1, color=PALETA[i % 5])
ejes[0].axhline(1, color="black", ls="--"); ejes[0].set(title="Exponencial(1): converge a μ = 1", xlabel="n", ylabel="media acumulada")
ejes[1].semilogx(idx, ber, color=PALETA[0], lw=1.2); ejes[1].axhline(0.3, color="black", ls="--"); ejes[1].set(title="Bernoulli(0.3): la proporción converge a p", xlabel="n")
ejes[2].axhline(0, color="black", ls="--"); ejes[2].set(title="Cauchy: la media acumulada NO converge", xlabel="n", ylim=(-30, 30))
plt.tight_layout(); plt.show()
assert np.abs(exp5[:, -1] - 1).max() < 0.03 and abs(ber[-1] - 0.3) < 0.01

# Cota de Chebyshev frente a la probabilidad exacta (Exp(1): xbar_n ~ Gamma(n, escala 1/n))
eps = 0.1
filas = []
for n in (100, 400, 1600):
    exacta = stats.gamma(n, scale=1 / n).sf(1 + eps) + stats.gamma(n, scale=1 / n).cdf(1 - eps)
    filas.append({"n": n, "P(|xbar - 1| >= 0.1) exacta": exacta, "cota sigma^2/(n eps^2)": min(1.0, 1 / (n * eps**2))})
cheb = pd.DataFrame(filas); print(cheb.round(5).to_string(index=False))
assert (cheb.iloc[:, 1] <= cheb.iloc[:, 2] + 1e-12).all() and cheb.iloc[:, 1].is_monotonic_decreasing and cheb.iloc[-1, 1] < 1e-3

# Cauchy: el rango intercuartilico de xbar_n NO se reduce al aumentar n (el de la exponencial, si)
iqr = lambda a: np.subtract(*np.percentile(a, [75, 25]))
c100 = rng.standard_cauchy((3000, 100)).mean(axis=1); c1000 = rng.standard_cauchy((3000, 1000)).mean(axis=1)
e100 = rng.exponential(1, (3000, 100)).mean(axis=1); e1000 = rng.exponential(1, (3000, 1000)).mean(axis=1)
print(f"\nRIC de xbar: Cauchy n=100: {iqr(c100):.2f}, n=1000: {iqr(c1000):.2f} (IQR de la Cauchy estandar = 2) | Exponencial n=100: {iqr(e100):.3f}, n=1000: {iqr(e1000):.3f}")
assert 0.8 < iqr(c1000) / iqr(c100) < 1.25 and abs(iqr(c1000) - 2) < 0.25 and 0.25 < iqr(e1000) / iqr(e100) < 0.40

# Falacia del apostador: la proporcion converge, pero la diferencia absoluta (S_n - n/2) crece como sqrt(n)
caras = np.cumsum(rng.integers(0, 2, size=(1000, 10_000), dtype=np.int8), axis=1, dtype=np.int32)
for n in (100, 10_000):
    dif = caras[:, n - 1] - n / 2
    print(f"n = {n:>6}: sd de la proporcion = {np.std(caras[:, n - 1] / n):.4f} | sd de la diferencia S_n - n/2 = {dif.std():.1f} (teoria sqrt(n)/2 = {math.sqrt(n) / 2:.1f})")
assert np.std(caras[:, -1] / 10_000) < np.std(caras[:, 99] / 100) / 5                                  # la proporcion se concentra...
assert (caras[:, -1] - 5000).std() > 5 * (caras[:, 99] - 50).std()                                      # ...pero la diferencia absoluta se ensancha
````

---
**Lectura.** Las cinco trayectorias exponenciales terminan a menos de 0.03 de $\mu=1$, y la proporción de la Bernoulli en 0.3 con error menor a 0.01. **Chebyshev demuestra la ley pero es muy holgado:** para $n=400$ la cota es 0.25 y la probabilidad exacta de desviarse más de 0.1 es 0.045; para $n=1600$, 0.0625 frente a 0.00007 (con $n=100$ la cota vale 1 y no dice nada, aunque la probabilidad real es 0.32). **Cauchy:** el rango intercuartílico de $\bar X_n$ es 1.99 con $n=100$ y 1.95 con $n=1000$ (el de la Cauchy estándar es 2): promediar 10 veces más datos **no mejora nada**, mientras que en la exponencial pasa de 0.136 a 0.043 (se divide por ≈ $\sqrt{10}$). **Apostador:** al pasar de $n=100$ a $n=10\,000$, la desviación estándar de la **proporción** baja de 0.0492 a 0.0048 (÷10), pero la de la **diferencia** $S_n-n/2$ **sube** de 4.9 a 47.6 (×10): la ley no compensa desviaciones, las diluye.

## 3. El Teorema del Límite Central 🔔

**Enunciado.** Si $X_i$ son i. i. d. con media $\mu$ y varianza $0<\sigma^2<\infty$, entonces

$$Z_n=\frac{\bar X_n-\mu}{\sigma/\sqrt n}\ \xrightarrow{\ d\ }\ N(0,1) \qquad (n\to\infty),$$

es decir, $\sup_x\lvert P(Z_n\le x)-\Phi(x)\rvert\to 0$. **No importa la forma de la población** (discreta, asimétrica, bimodal): lo único que se exige es varianza finita. Por eso la normal aparece en todas partes y por eso casi toda la inferencia clásica se apoya en ella.

**¿Qué tan rápido?** Depende de la forma de la población. El teorema de **Berry-Esseen** acota la distancia $\sup_x\lvert F_{Z_n}(x)-\Phi(x)\rvert\le C\,\dfrac{\rho}{\sigma^3\sqrt n}$ con $\rho=E\lvert X-\mu\rvert^3$ y $C\le 0.4748$ *(constante citada de memoria, verificar)*. Una consecuencia práctica: la **asimetría** de $\bar X_n$ es la de la población dividida por $\sqrt n$, $\gamma_1(\bar X_n)=\gamma_1(X)/\sqrt n$, así que las poblaciones muy asimétricas necesitan muchas más observaciones. La regla «$n\ge 30$» es una **heurística**, no un teorema.

Medimos la distancia a la normal con el **estadístico de Kolmogorov-Smirnov** $D_n=\sup_x\lvert \hat F(x)-\Phi(x)\rvert$ sobre 20 000 medias estandarizadas, para cuatro poblaciones: **uniforme** (simétrica), **exponencial** (asimétrica, $\gamma_1=2$), **Bernoulli(0.1)** (discreta y muy asimétrica, $\gamma_1\approx 2.67$) y una **mezcla bimodal** (simétrica pero con dos picos).

````python
# ============================================================
# 3. TLC: distancia a la normal (KS) segun la poblacion y n, asimetria de xbar y cota de Berry-Esseen
# ============================================================
rng = np.random.default_rng(SEED)
R = 20_000
fuentes = {   # nombre: (generador de matriz (R, n), mu, sigma, asimetria)
    "Uniforme(0,1)": (lambda n: rng.random((R, n)), 0.5, math.sqrt(1 / 12), 0.0),
    "Exponencial(1)": (lambda n: rng.exponential(1.0, (R, n)), 1.0, 1.0, 2.0),
    "Bimodal 50/50": (lambda n: rng.normal(rng.choice([-2.0, 2.0], size=(R, n)), 0.5), 0.0, math.sqrt(4.25), 0.0),
}
p_b = 0.1
ns = [1, 2, 5, 10, 30, 100, 200]


def medias_std(nombre, n):
    """R medias de tamano n, estandarizadas con mu y sigma de la poblacion."""
    if nombre == "Bernoulli(0.1)":
        xbar = rng.binomial(n, p_b, R) / n; m, s = p_b, math.sqrt(p_b * (1 - p_b))
    else:
        g, m, s, _ = fuentes[nombre]; xbar = g(n).mean(axis=1)
    return (xbar - m) / (s / math.sqrt(n))


nombres = list(fuentes) + ["Bernoulli(0.1)"]
ks = pd.DataFrame({nom: [stats.kstest(medias_std(nom, n), stats.norm.cdf).statistic for n in ns] for nom in nombres}, index=ns)
ks.index.name = "n"
print(ks.round(4))

# Cota de Berry-Esseen para la exponencial: rho = E|X-1|^3 = 12/e - 2
rho_exp = 12 / math.e - 2
be = 0.4748 * rho_exp / np.sqrt(np.array(ns))
print(f"\nExponencial: rho/sigma^3 = {rho_exp:.4f}; cota de Berry-Esseen en n=30: {be[4]:.3f} vs KS observado {ks.loc[30, 'Exponencial(1)']:.4f}")
assert (ks["Exponencial(1)"].to_numpy()[1:] <= be[1:] + 0.01).all()                              # la cota se cumple (con holgura)

# Asimetria de la media = asimetria de la poblacion / sqrt(n)
for n in (5, 30, 200):
    sk = stats.skew(medias_std("Exponencial(1)", n))
    print(f"Exponencial n = {n:>3}: asimetria de xbar = {sk:.3f} (teoria 2/sqrt(n) = {2 / math.sqrt(n):.3f})")
    assert abs(sk - 2 / math.sqrt(n)) < 0.08

assert ks.loc[5, "Exponencial(1)"] > 3 * ks.loc[5, "Uniforme(0,1)"]                              # la asimetria frena la convergencia
assert ks.loc[200, "Exponencial(1)"] < ks.loc[5, "Exponencial(1)"] / 3 and ks.loc[200, "Bimodal 50/50"] < 0.02
assert ks.loc[30, "Bernoulli(0.1)"] > ks.loc[30, "Exponencial(1)"]                              # la mas asimetrica y discreta tarda mas
assert ks.loc[200, "Bernoulli(0.1)"] < ks.loc[10, "Bernoulli(0.1)"]

fig, ejes = plt.subplots(2, 4, figsize=(16, 6.5)); zz = np.linspace(-4, 4, 300)
for fila, nom in enumerate(("Exponencial(1)", "Bimodal 50/50")):
    for col, n in enumerate((1, 5, 30, 200)):
        a = ejes[fila, col]; a.hist(medias_std(nom, n), bins=60, density=True, color=PALETA[fila], alpha=0.6)
        a.plot(zz, stats.norm.pdf(zz), "k", lw=1.5); a.set_xlim(-4, 4); a.set_title(f"{nom}, n = {n}", fontsize=10)
plt.tight_layout(); plt.show()

fig, a = plt.subplots(figsize=(9, 4.5))
for nom, c in zip(nombres, PALETA):
    a.loglog(ns, ks[nom], "o-", color=c, label=nom)
a.loglog(ns, 0.5 / np.sqrt(ns), "k--", label="referencia ∝ 1/√n"); a.axhline(0.005, color="gray", ls=":", label="piso de ruido de la simulación")
a.set(xlabel="n", ylabel="distancia KS a N(0,1)", title="El TLC: la distancia a la normal baja ~ 1/√n, más lento si hay asimetría"); a.legend()
plt.tight_layout(); plt.show()
````

---
**Lectura.** Con $n=5$ la media de uniformes ya es casi indistinguible de la normal (KS 0.008, del orden del ruido de la simulación, ≈ 0.005), la mezcla bimodal necesita $n\approx 10$ (0.011) y la **exponencial**, asimétrica, sigue en 0.023 con $n=30$ y en 0.013 con $n=200$: **el «$n\ge 30$» no basta para poblaciones asimétricas**. La asimetría de $\bar X$ medida (0.887, 0.363 y 0.125 para $n=5,30,200$) sigue a $2/\sqrt n$ (0.894, 0.365 y 0.141). La cota de Berry-Esseen se cumple (0.209 frente a 0.023 en $n=30$) pero es muy conservadora. La **Bernoulli(0.1)** parece la peor (KS 0.056 con $n=200$): en parte es real (asimetría 2.67) y en parte un **artefacto de la discretización**, porque comparar una acumulada en escalera con una continua deja una diferencia de al menos la mitad del mayor salto (≈ 0.05); con la **corrección de continuidad** (cuaderno 3) la aproximación es mucho mejor de lo que sugiere este número. En los histogramas, la exponencial pasa de una cola derecha marcada ($n=1$ y $n=5$) a una campana casi simétrica ($n=200$); la mezcla bimodal pierde sus dos picos con pocos sumandos.

## 4. Aplicaciones: Cobertura Real, Tamaño de Muestra y Monte Carlo 🎯

**(a) Cobertura de un intervalo.** Con el TLC, un intervalo de confianza de nivel nominal 95 % para la media es $\bar x\pm 1.96\,s/\sqrt n$ (o con el cuantil $t_{n-1}$, que además corrige por haber estimado $\sigma$). Pero «95 %» es una promesa que depende del TLC; la **cobertura real** (la fracción de intervalos que contienen a $\mu$ al repetir el muestreo) se mide por simulación. Probamos con una población **muy asimétrica** (lognormal con $\sigma=1$, cuya media es $e^{1/2}$).

**(b) Tamaño de muestra para una proporción.** El error estándar de una proporción muestral es $\sqrt{p(1-p)/n}$. Para un **margen de error** $E$ con confianza 95 %:

$$n\ \ge\ \frac{z_{0.975}^2\,p(1-p)}{E^2}, \qquad \text{prudente: } p=0.5 \text{ (el máximo de } p(1-p)=0.25).$$

**(c) Monte Carlo.** Estimar $\pi$ lanzando puntos al azar en el cuadrado unitario y contando la fracción que cae dentro del cuarto de círculo: $\hat\pi=4\hat p$, con $p=\pi/4$. Su error típico es $4\sqrt{p(1-p)/n}\approx 1.64/\sqrt n$: **cada dígito decimal adicional cuesta 100 veces más puntos**.

````python
# ============================================================
# 4a. Cobertura real del intervalo xbar +- c * s/sqrt(n) en una poblacion muy asimetrica (lognormal sigma = 1)
# ============================================================
rng = np.random.default_rng(SEED)
mu_ln = math.exp(0.5)                                                           # E[lognormal(0, 1)] = e^{1/2}
filas = []
for n in (5, 15, 30, 100, 400):
    x = rng.lognormal(0, 1, size=(5000, n))
    xbar, s = x.mean(axis=1), x.std(axis=1, ddof=1)
    cob = {}
    for nombre, c in (("z (1.96)", 1.96), ("t (n-1 g.l.)", stats.t.ppf(0.975, n - 1))):
        cob[nombre] = float((np.abs(xbar - mu_ln) <= c * s / math.sqrt(n)).mean())
    filas.append({"n": n, **cob})
cobertura = pd.DataFrame(filas).set_index("n")
print("Cobertura nominal 95 %; cobertura real (5000 muestras por n):"); print(cobertura.round(3))
assert (cobertura["t (n-1 g.l.)"] >= cobertura["z (1.96)"]).all()               # el intervalo t contiene al z: cubre al menos tanto
assert cobertura.loc[5, "z (1.96)"] < 0.85 and cobertura.loc[400, "z (1.96)"] > 0.92
assert cobertura["z (1.96)"].is_monotonic_increasing or cobertura.loc[400, "z (1.96)"] > cobertura.loc[5, "z (1.96)"]

fig, a = plt.subplots(figsize=(8.5, 4.2))
a.semilogx(cobertura.index, cobertura["z (1.96)"], "o-", color=PALETA[1], label="z = 1.96"); a.semilogx(cobertura.index, cobertura["t (n-1 g.l.)"], "s-", color=PALETA[0], label="t de Student")
a.axhline(0.95, color="black", ls="--", label="nivel nominal"); a.set(xlabel="n", ylabel="cobertura real", title="La promesa del 95 % depende del TLC: con asimetría y n pequeña, no se cumple"); a.legend()
plt.tight_layout(); plt.show()
````

````python
# ============================================================
# 4b. Tamano de muestra para una proporcion y error de Monte Carlo para estimar pi
# ============================================================
z = stats.norm.ppf(0.975)
n_req = lambda E, p=0.5: math.ceil(z**2 * p * (1 - p) / E**2)
tabla_n = pd.DataFrame({"margen de error E": [0.10, 0.05, 0.03, 0.01]}); tabla_n["n necesario (p = 0.5)"] = tabla_n["margen de error E"].map(n_req)
print(tabla_n.to_string(index=False))
assert n_req(0.05) == 385 and n_req(0.03) == 1068 and n_req(0.01) == 9604
assert abs(n_req(0.025) / n_req(0.05) - 4) < 0.02                                # mitad del margen = cuadruple de la muestra

rng = np.random.default_rng(SEED)
p_real = 0.40                                                                  # proporcion verdadera (la desconocida en la practica)
p_hat = rng.binomial(385, p_real, 50_000) / 385
cob_p = float((np.abs(p_hat - p_real) <= 0.05).mean())
print(f"\nCon n = 385 y p real = {p_real}: la proporcion muestral cae dentro de ±0.05 de p en el {cob_p:.1%} de las muestras (nominal >= 95 % por el diseno prudente)")
assert 0.94 < cob_p < 0.98

# Monte Carlo para pi: error tipico ~ 1.64/sqrt(n)
filas = []
for n, reps in ((100, 400), (1000, 400), (10_000, 200), (100_000, 60)):
    est = np.array([4 * (rng.random(n) ** 2 + rng.random(n) ** 2 <= 1).mean() for _ in range(reps)])
    filas.append({"n": n, "RMSE simulado": float(np.sqrt(np.mean((est - math.pi) ** 2))), "teoria 4 sqrt(p(1-p)/n)": 4 * math.sqrt(math.pi / 4 * (1 - math.pi / 4) / n)})
mc = pd.DataFrame(filas).set_index("n"); print("\n", mc.round(5))
pend = np.polyfit(np.log(mc.index), np.log(mc["RMSE simulado"]), 1)[0]
print(f"Pendiente log-log del error: {pend:.2f} (teoria -0.5): 100 veces mas puntos -> solo 10 veces menos error")
assert (mc["RMSE simulado"] / mc["teoria 4 sqrt(p(1-p)/n)"]).between(0.7, 1.4).all() and abs(pend + 0.5) < 0.1
````

---
**Lectura.** **Cobertura:** el intervalo «95 %» con $z$ cubre solo 74.4 % con $n=5$, 87.8 % con $n=30$ y 91.0 % con $n=100$; llega a 94.0 % recién con $n=400$. El intervalo $t$ ayuda al principio (81.9 % con $n=5$) pero **no resuelve** la asimetría (88.8 % con $n=30$): corrige por haber estimado $\sigma$, no por la forma de la población. Es una advertencia seria para la inferencia con datos reales asimétricos (ingresos, tiempos): los intervalos «de libro» pueden prometer más de lo que cumplen. **Tamaños de muestra:** 97, 385, 1 068 y 9 604 para márgenes de 0.10, 0.05, 0.03 y 0.01 (con $p=0.5$); con $p$ real de 0.4 y $n=385$, el 95.7 % de las encuestas cae dentro de ±0.05. **Monte Carlo:** el error de $\hat\pi$ baja de 0.170 con 100 puntos a 0.0053 con 100 000 (pendiente $-0.50$): para obtener un dígito más hay que pagar 100 veces más simulaciones, la misma ley $1/\sqrt n$ que rige el muestreo.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Distribución muestral** | $\bar X$ es una variable aleatoria: $E[\bar X]=\mu$ y $\text{EE}=\sigma/\sqrt n$ (pendiente $-1/2$ en escala log-log); **cuadruplicar $n$ divide el error a la mitad** |
| **Población finita** | Sin reemplazo, el EE se multiplica por $\sqrt{(N-n)/(N-1)}$ (FPC); con $n=N$ no hay error de muestreo |
| **Bessel** | Dividir por $n$ subestima la varianza en el factor $(n-1)/n$ (0.8 con $n=5$); dividir por $n-1$ la hace insesgada |
| **LGN** | $\bar X_n\to\mu$ (débil por Chebyshev, fuerte por Kolmogorov); **falla sin esperanza** (Cauchy: $\bar X_n$ sigue siendo Cauchy) y **no compensa** desviaciones pasadas: la proporción converge pero $\lvert S_n-n/2\rvert$ crece como $\sqrt n$ |
| **TLC** | $\sqrt n(\bar X-\mu)/\sigma\to N(0,1)$ para casi cualquier población con varianza finita; la **velocidad** depende de la asimetría (la de $\bar X$ es $\gamma_1/\sqrt n$) y de la discretización; $n\ge 30$ es una **heurística** |
| **Cobertura real** | Con una población lognormal ($\sigma=1$) el intervalo «95 %» con $z$ cubre solo 74 % con $n=5$ y 88 % con $n=30$; el intervalo $t$ ayuda pero no lo arregla |
| **Tamaño de muestra** | $n\ge z^2p(1-p)/E^2$: con $p=0.5$, 385 para $E=0.05$ y 9 604 para $E=0.01$; Monte Carlo sigue la misma ley $1/\sqrt n$ |
