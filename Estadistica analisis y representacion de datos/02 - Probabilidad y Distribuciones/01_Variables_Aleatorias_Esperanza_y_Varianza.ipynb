## 1. Variable Aleatoria: Masa, Densidad y Distribución Acumulada 🔢

Una **variable aleatoria** (v. a.) es una **función** $X: \Omega \to \mathbb{R}$ que asigna un número a cada resultado del experimento. Ejemplo: al lanzar dos dados, $X(d_1, d_2) = d_1 + d_2$. La probabilidad de un valor se **hereda** de $\Omega$: $P(X = x) = P(\{\omega : X(\omega) = x\})$.

| | **Discreta** | **Continua** |
|---|---|---|
| Valores | conjunto finito o numerable | un intervalo de $\mathbb{R}$ |
| Descripción | **masa** $p(x) = P(X = x)$, con $\sum_x p(x) = 1$ | **densidad** $f(x) \ge 0$, con $\int f = 1$ |
| Probabilidad de un valor | $p(x) > 0$ | $P(X = x) = 0$ (!); solo hay probabilidad en **intervalos**: $P(a \le X \le b) = \int_a^b f(x)\,dx$ |
| Acumulada $F(x) = P(X \le x)$ | $\sum_{t \le x} p(t)$ (escalera) | $\int_{-\infty}^{x} f(t)\,dt$, y $f = F'$ |

**La acumulada** $F$ existe siempre y cumple: es **no decreciente**, **continua por la derecha**, $\lim_{x\to-\infty}F = 0$ y $\lim_{x\to+\infty}F = 1$. Con ella, $P(a < X \le b) = F(b) - F(a)$, y su inversa $F^{-1}(p)$ es la **función cuantil**.

> ⚠️ **Una densidad no es una probabilidad:** $f(x)$ puede ser **mayor que 1** (una densidad uniforme en $[0, 0.5]$ vale 2); lo que debe valer 1 es el **área** bajo la curva.

Implementamos una clase `VADiscreta` desde su función de masa, con **fracciones exactas**, y la contrastamos con simulación y con `scipy.stats.rv_discrete`.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib scipy

import itertools
import math
import warnings
from fractions import Fraction

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import integrate, stats

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")
````

````python
# ============================================================
# 1. Clase VADiscreta (desde cero) y la suma de dos dados
# ============================================================
class VADiscreta:
    """Variable aleatoria discreta definida por su funcion de masa {valor: probabilidad}, con fracciones exactas."""

    def __init__(self, pmf):
        self.pmf = {k: Fraction(v) for k, v in pmf.items()}
        assert all(p >= 0 for p in self.pmf.values()) and sum(self.pmf.values()) == 1, "una masa debe ser >= 0 y sumar 1"

    def cdf(self, x):
        return sum(p for k, p in self.pmf.items() if k <= x)

    def esperanza(self, g=lambda x: x):
        """E[g(X)] = sum g(x) p(x)  (regla del estadistico inconsciente)."""
        return sum(g(k) * p for k, p in self.pmf.items())

    def varianza(self):
        mu = self.esperanza()
        return self.esperanza(lambda x: (x - mu) ** 2)

    def muestrear(self, n, rng):
        valores = list(self.pmf)
        return rng.choice(valores, size=n, p=[float(self.pmf[v]) for v in valores])


# X = suma de dos dados: una funcion sobre el espacio muestral Omega
conteo = {}
for d1, d2 in itertools.product(range(1, 7), repeat=2):
    conteo[d1 + d2] = conteo.get(d1 + d2, 0) + Fraction(1, 36)
X = VADiscreta(conteo)

mu, var = X.esperanza(), X.varianza()
print(f"E[X] = {mu} | Var(X) = {var} = {float(var):.4f} | sigma = {math.sqrt(var):.4f}")

rng = np.random.default_rng(SEED)
muestra = X.muestrear(200_000, rng)
print(f"Muestra (n=200000): media = {muestra.mean():.4f} | varianza (ddof=0) = {muestra.var():.4f}")

# Contraste con scipy.stats.rv_discrete
xs = np.array(sorted(X.pmf)); ps = np.array([float(X.pmf[k]) for k in xs])
rv = stats.rv_discrete(values=(xs, ps))
assert mu == 7 and var == Fraction(35, 6)
assert abs(rv.mean() - float(mu)) < 1e-12 and abs(rv.var() - float(var)) < 1e-12
assert all(abs(float(X.cdf(k)) - rv.cdf(k)) < 1e-12 for k in xs)             # la acumulada coincide en cada punto
assert abs(muestra.mean() - 7) < 0.03 and abs(muestra.var() - 35 / 6) < 0.1
assert X.cdf(1) == 0 and X.cdf(12) == 1 and X.cdf(7) - X.cdf(6) == X.pmf[7]   # F(7)-F(6) = p(7)

fig, ejes = plt.subplots(1, 2, figsize=(11, 4.2))
ejes[0].bar(xs, ps, color=PALETA[0], alpha=0.8, label="masa exacta p(x)")
ejes[0].plot(xs, [(muestra == k).mean() for k in xs], "o", color=PALETA[1], label="frecuencia simulada")
ejes[0].set(title="Función de masa de X = suma de dos dados", xlabel="x", ylabel="P(X = x)"); ejes[0].legend()
ejes[1].step(xs, np.cumsum(ps), where="post", color=PALETA[0], lw=2)
ejes[1].set(title="Acumulada F(x): escalera no decreciente", xlabel="x", ylabel="P(X ≤ x)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** La suma de dos dados tiene $E[X]=7$ y $\text{Var}(X)=35/6\approx 5.833$ ($\sigma\approx 2.415$), **exactas** con fracciones; la simulación (media 7.000, varianza 5.836) y `scipy.stats.rv_discrete` coinciden. La acumulada es una **escalera** que sube de 0 a 1 y $F(7)-F(6)=p(7)=1/6$: la masa es el *salto* de la acumulada. El valor más probable (la moda, 7) coincide aquí con la esperanza porque la distribución es simétrica, pero **en general la esperanza no es un valor que la variable pueda tomar** (el promedio de un dado es 3.5).

## 2. Esperanza y Varianza: Propiedades y Valor Esperado de una Decisión 💰

La **esperanza** (media o valor esperado) es el **centro de masa** de la distribución, y la **varianza** mide la dispersión alrededor de ella:

$$\mu = E[X] = \sum_x x\,p(x) \;\Big(\text{o } \int x f(x)\,dx\Big), \qquad \sigma^2 = \text{Var}(X) = E\big[(X-\mu)^2\big] = E[X^2] - \mu^2, \qquad \sigma = \sqrt{\sigma^2}.$$

Para una función $g$, la **regla del estadístico inconsciente** dice $E[g(X)] = \sum_x g(x)\,p(x)$: no hace falta hallar primero la distribución de $g(X)$. Propiedades (todas verificadas abajo):

| Propiedad | Fórmula | Condición |
|---|---|---|
| **Linealidad** | $E[aX + b] = a\,E[X] + b$; $E[X+Y] = E[X] + E[Y]$ | **siempre** (aunque $X$ y $Y$ sean dependientes) |
| **Escala y traslación** | $\text{Var}(aX + b) = a^2\,\text{Var}(X)$ | trasladar no cambia la dispersión |
| **Suma de varianzas** | $\text{Var}(X+Y) = \text{Var}(X) + \text{Var}(Y) + 2\,\text{Cov}(X,Y)$ | se simplifica solo si $\text{Cov}=0$ (p. ej. independientes) |
| **Jensen** | $g$ convexa $\Rightarrow E[g(X)] \ge g(E[X])$ | en general $E[g(X)] \ne g(E[X])$ |

**La esperanza como promedio a largo plazo.** Si repites el experimento $n$ veces, el promedio de los resultados tiende a $E[X]$ (ley de los grandes números, cuaderno 4). Por eso es la cifra correcta para decidir entre apuestas **repetidas**, pero **no basta**: dos decisiones con la misma esperanza pueden tener riesgos muy distintos, y ahí entra la varianza.

**Ejemplo (rifa hipotética).** Una rifa de una ternera: 1 000 boletas de \$5 000 y un premio de \$3 000 000. Si compras una boleta, tu ganancia neta es $+2\,995\,000$ con probabilidad $1/1000$ y $-5\,000$ con probabilidad $999/1000$. Las cifras son **inventadas**.

````python
# ============================================================
# 2. Propiedades de E y Var (verificadas con fracciones exactas) y la rifa de la ternera
# ============================================================
# Linealidad y escala: Y = aX + b
a, b = 3, -2
Y = VADiscreta({a * k + b: p for k, p in X.pmf.items()})
assert Y.esperanza() == a * mu + b and Y.varianza() == a**2 * var
assert X.esperanza(lambda x: x**2) - mu**2 == var                           # Var = E[X^2] - mu^2
print(f"Y = 3X - 2 -> E[Y] = {Y.esperanza()} (= 3*7 - 2) | Var(Y) = {Y.varianza()} (= 9 * 35/6)")

# E[g(X)] != g(E[X]) (Jensen): g(x) = x^2 es convexa
e_x2, g_ex = X.esperanza(lambda x: x**2), mu**2
print(f"E[X^2] = {e_x2} = {float(e_x2):.3f} > (E[X])^2 = {g_ex}  -> la diferencia es exactamente Var(X) = {e_x2 - g_ex}")
assert e_x2 > g_ex and e_x2 - g_ex == var

# Linealidad de E sin independencia: X1 = primer dado, X2 = 7 - X1 (totalmente dependiente de X1)
D = VADiscreta({k: Fraction(1, 6) for k in range(1, 7)})
assert D.esperanza() + VADiscreta({7 - k: Fraction(1, 6) for k in range(1, 7)}).esperanza() == 7   # E[X1 + X2] = E[X1] + E[X2] aunque X2 dependa de X1
assert VADiscreta({7: Fraction(1)}).varianza() == 0 != 2 * D.varianza()      # pero Var(X1 + X2) = 0, no Var X1 + Var X2

# Rifa: valor esperado negativo, varianza enorme
rifa = VADiscreta({2_995_000: Fraction(1, 1000), -5_000: Fraction(999, 1000)})
e_rifa, sd_rifa = rifa.esperanza(), math.sqrt(rifa.varianza())
print(f"\nRifa: E[ganancia] = {float(e_rifa):,.0f} pesos por boleta | sigma = {sd_rifa:,.0f} pesos")
rng = np.random.default_rng(SEED)
juegos = rifa.muestrear(300_000, rng)
prom = np.cumsum(juegos) / np.arange(1, len(juegos) + 1)
print(f"Promedio tras 10 boletas: {prom[9]:,.0f} | tras 300000: {prom[-1]:,.0f} (esperanza teorica {float(e_rifa):,.0f})")

fig, ax = plt.subplots(figsize=(9, 4.2))
ax.plot(np.arange(1, len(prom) + 1), prom, color=PALETA[0], lw=1.2, label="promedio acumulado de la ganancia")
ax.axhline(float(e_rifa), color="black", ls="--", label="E[ganancia] = −2 000"); ax.set_xscale("log")
ax.set_xlabel("boletas compradas (escala log)"); ax.set_ylabel("ganancia promedio por boleta (pesos)")
ax.set_title("La esperanza como promedio a largo plazo: la rifa pierde en promedio", fontweight="bold"); ax.legend()
plt.tight_layout(); plt.show()
assert e_rifa == -2000 and abs(prom[-1] - float(e_rifa)) < 4 * sd_rifa / math.sqrt(len(juegos))   # dentro de ~4 errores estandar
````

---
**Lectura.** Para $Y=3X-2$ se cumple $E[Y]=19=3\cdot 7-2$ y $\text{Var}(Y)=105/2=9\cdot 35/6$: **la constante no afecta la dispersión y el factor entra al cuadrado**. También $E[X^2]=329/6\approx 54.83 > 49=(E[X])^2$, y la diferencia es exactamente $\text{Var}(X)$ (Jensen). El ejemplo de $X_2=7-X_1$ muestra que $E[X_1+X_2]=E[X_1]+E[X_2]$ **aun con dependencia total**, pero $\text{Var}(X_1+X_2)=0\ne \text{Var}(X_1)+\text{Var}(X_2)$: **las varianzas no se suman sin ausencia de covarianza**. La rifa tiene esperanza **−2 000 pesos** por boleta con $\sigma\approx 94\,800$: tras 10 boletas el promedio fue −5 000 (nadie ganó) y tras 300 000 fue −2 110, a menos de un error estándar (≈ 173 pesos) del valor teórico. Una esperanza negativa es una **pérdida segura a largo plazo**, aunque cada boleta ofrezca la ilusión de ganar mucho.

## 3. Variables Continuas: Integración Numérica, Momentos y Transformada Inversa 📈

Para una v. a. continua con densidad $f$:

$$E[X] = \int_{-\infty}^{\infty} x\,f(x)\,dx, \qquad \text{Var}(X) = \int (x-\mu)^2 f(x)\,dx, \qquad P(a \le X \le b) = F(b) - F(a) = \int_a^b f(x)\,dx.$$

Los **momentos estandarizados** describen la forma: la **asimetría** $\gamma_1 = E[(X-\mu)^3]/\sigma^3$ y la **curtosis** $\gamma_2 = E[(X-\mu)^4]/\sigma^4$ (la *curtosis en exceso* es $\gamma_2 - 3$, que vale 0 para la normal). Los calculamos por **integración numérica** (regla del trapecio escrita desde cero) y contrastamos con `scipy.integrate.quad`.

**Transformada inversa (muestreo).** Si $U \sim \text{Uniforme}(0,1)$ y $F$ es una acumulada invertible, entonces $X = F^{-1}(U)$ tiene acumulada $F$, porque $P(F^{-1}(U) \le x) = P(U \le F(x)) = F(x)$. Es el método base de **toda** simulación de variables aleatorias. Para la exponencial con tasa $\lambda$: $F(x) = 1 - e^{-\lambda x}$ y $X = -\ln(1-U)/\lambda$.

Usamos dos modelos **hipotéticos**: (a) la **proporción de papa de primera** en un lote, con densidad $f(x) = 6x(1-x)$ en $[0,1]$ (una Beta(2,2)); (b) el **tiempo entre llegadas de buses** a un paradero, exponencial con tasa $\lambda = 0.5$ por minuto (media de 2 minutos).

````python
# ============================================================
# 3. Densidad continua: integracion numerica (trapecio desde cero), momentos y transformada inversa
# ============================================================
def trapecio(y, x):
    """Regla del trapecio: area bajo la curva (x, y)."""
    return float(np.sum((y[1:] + y[:-1]) * np.diff(x)) / 2)


f_prop = lambda x: 6 * x * (1 - x)                                           # densidad de la proporcion de papa de primera (Beta(2,2))
xs = np.linspace(0, 1, 100_001)
fx = f_prop(xs)
area = trapecio(fx, xs)
mu_p = trapecio(xs * fx, xs)
var_p = trapecio((xs - mu_p) ** 2 * fx, xs)
xc = np.linspace(0.3, 0.7, 40_001)
p_central = trapecio(f_prop(xc), xc)                                          # P(0.3 <= X <= 0.7)
F = lambda x: 3 * x**2 - 2 * x**3                                            # acumulada exacta
print(f"Area = {area:.8f} | E = {mu_p:.6f} | Var = {var_p:.6f} | P(0.3 <= X <= 0.7) = {p_central:.6f} (exacta F(.7)-F(.3) = {F(.7) - F(.3):.6f})")

q_area = integrate.quad(f_prop, 0, 1)[0]
q_mu = integrate.quad(lambda x: x * f_prop(x), 0, 1)[0]
assert abs(area - 1) < 1e-8 and abs(q_area - area) < 1e-8 and abs(mu_p - q_mu) < 1e-8
assert abs(mu_p - 0.5) < 1e-8 and abs(var_p - 0.05) < 1e-8 and abs(p_central - (F(.7) - F(.3))) < 1e-8
dist = stats.beta(2, 2)
assert abs(dist.mean() - mu_p) < 1e-8 and abs(dist.var() - var_p) < 1e-8 and abs(dist.cdf(.7) - dist.cdf(.3) - p_central) < 1e-8

# Momentos estandarizados de la exponencial (lambda=0.5) por integracion numerica
lam = 0.5
fe = lambda x: lam * np.exp(-lam * x)
xe = np.linspace(0, 120, 400_001)                                            # la cola mas alla de 120 pesa e^-60: despreciable
mu_e = trapecio(xe * fe(xe), xe); sd_e = math.sqrt(trapecio((xe - mu_e) ** 2 * fe(xe), xe))
asim = trapecio((xe - mu_e) ** 3 * fe(xe), xe) / sd_e**3
curt = trapecio((xe - mu_e) ** 4 * fe(xe), xe) / sd_e**4
print(f"Exponencial(0.5): E = {mu_e:.4f} (1/lambda) | sigma = {sd_e:.4f} | asimetria = {asim:.3f} | curtosis en exceso = {curt - 3:.3f}")
assert abs(mu_e - 2) < 1e-4 and abs(sd_e - 2) < 1e-4 and abs(asim - 2) < 1e-3 and abs(curt - 3 - 6) < 5e-3

# Transformada inversa: X = -ln(1-U)/lambda
rng = np.random.default_rng(SEED)
u = rng.random(100_000)
x_inv = -np.log(1 - u) / lam
ks = stats.kstest(x_inv, stats.expon(scale=1 / lam).cdf)
print(f"Transformada inversa: media = {x_inv.mean():.3f} | var = {x_inv.var():.3f} | estadistico KS = {ks.statistic:.4f} (p = {ks.pvalue:.2f})")
assert abs(x_inv.mean() - 2) < 0.03 and abs(x_inv.var() - 4) < 0.15 and ks.statistic < 0.006

fig, ejes = plt.subplots(1, 3, figsize=(14, 4))
ejes[0].plot(xs, fx, color=PALETA[0], lw=2); m = (xs >= 0.3) & (xs <= 0.7)
ejes[0].fill_between(xs[m], fx[m], color=PALETA[0], alpha=0.3, label=f"P(0.3 ≤ X ≤ 0.7) = {p_central:.3f}")
ejes[0].set(title="Densidad f(x) = 6x(1−x)", xlabel="proporción de papa de primera"); ejes[0].legend()
ejes[1].plot(xs, F(xs), color=PALETA[1], lw=2); ejes[1].set(title="Acumulada F(x) = 3x² − 2x³", xlabel="x")
ejes[2].hist(x_inv, bins=60, density=True, color=PALETA[2], alpha=0.6, label="muestra por transformada inversa")
ejes[2].plot(np.linspace(0, 12, 300), fe(np.linspace(0, 12, 300)), color="black", lw=2, label="densidad exacta")
ejes[2].set(title="Exponencial(0.5): minutos entre buses", xlabel="minutos"); ejes[2].legend()
plt.tight_layout(); plt.show()
````

---
**Lectura.** La regla del trapecio integra la densidad $6x(1-x)$ con área $1.00000000$, $E=0.5$, $\text{Var}=0.05$ y $P(0.3\le X\le 0.7)=0.568000$, idénticos a la fórmula exacta, a `integrate.quad` y a `scipy.stats.beta(2,2)` (tolerancia $10^{-8}$). Para la exponencial, la asimetría vale **2** y la curtosis en exceso **6**, **sin importar $\lambda$** (son propiedades de la forma, no de la escala): una cola derecha larga y pesada frente a la normal (0 y 0). La transformada inversa produce una muestra con media 2.003 y varianza 4.002 (teóricas 2 y 4); el estadístico de Kolmogorov-Smirnov es 0.0030 ($p=0.33$), es decir, **no hay evidencia contra** la hipótesis de que la muestra proviene de la exponencial, lo que era de esperar porque se generó así.

## 4. Distribución Conjunta, Covarianza y Correlación 🔗

Con **dos** variables, la información está en la **distribución conjunta** $p(x,y) = P(X=x, Y=y)$. De ella salen las **marginales** (sumando sobre la otra variable) y la medida de asociación lineal:

$$\text{Cov}(X,Y) = E[(X-\mu_X)(Y-\mu_Y)] = E[XY] - \mu_X\mu_Y, \qquad \rho_{XY} = \frac{\text{Cov}(X,Y)}{\sigma_X\,\sigma_Y} \in [-1, 1].$$

La cota $|\rho| \le 1$ es la **desigualdad de Cauchy-Schwarz**. Tres hechos que se confunden con frecuencia:

1. **Independientes $\Rightarrow$ $\text{Cov}=0$**, pero **no al revés**: $\rho$ solo mide asociación **lineal**. Si $X$ es simétrica alrededor de 0 y $Y = X^2$, entonces $Y$ es función exacta de $X$ y sin embargo $\text{Cov}(X,Y)=0$.
2. La covarianza **depende de las unidades**; el coeficiente $\rho$ no.
3. $\text{Var}(X+Y) = \text{Var}(X) + \text{Var}(Y) + 2\,\text{Cov}(X,Y)$: sumar variables **positivamente correlacionadas** aumenta la varianza total (el riesgo de una cartera, el error de una suma de mediciones).

**Ejemplo (hipotético).** Número semanal de **quejas** ($X$) y de **devoluciones** ($Y$) de un punto de venta, ambos en $\{0,1,2\}$, con la siguiente distribución conjunta inventada. Verificamos también con `numpy.cov` sobre una muestra simulada.

````python
# ============================================================
# 4. Distribucion conjunta discreta: marginales, covarianza, correlacion y el contraejemplo Y = X^2
# ============================================================
conj = pd.DataFrame([[20, 10, 2], [10, 25, 8], [2, 8, 15]], index=[0, 1, 2], columns=[0, 1, 2]).map(lambda v: Fraction(v, 100))
conj.index.name, conj.columns.name = "X (quejas)", "Y (devoluciones)"
assert conj.to_numpy().sum() == 1
px, py = conj.sum(axis=1), conj.sum(axis=0)                                  # marginales
vx, vy = np.array(conj.index), np.array(conj.columns)
E = lambda g: sum(g(i, j) * conj.loc[i, j] for i in vx for j in vy)           # esperanza sobre la conjunta
mx, my = E(lambda i, j: i), E(lambda i, j: j)
cov = E(lambda i, j: i * j) - mx * my
vX, vY = E(lambda i, j: i**2) - mx**2, E(lambda i, j: j**2) - my**2
rho = float(cov) / math.sqrt(float(vX) * float(vY))
print(conj.astype(float).round(2)); print(f"\nE[X] = {mx} | E[Y] = {my} | Cov = {cov} = {float(cov):.4f} | rho = {rho:.4f}")
var_suma = E(lambda i, j: (i + j) ** 2) - (mx + my) ** 2
assert var_suma == vX + vY + 2 * cov                                          # Var(X+Y) = Var X + Var Y + 2 Cov (exacto)
assert -1 <= rho <= 1 and cov > 0
indep = all(conj.loc[i, j] == px[i] * py[j] for i in vx for j in vy)
print("¿Son independientes? (p(x,y) = p(x)p(y) en todas las celdas):", indep)
assert not indep

# Verificacion por muestra: np.cov (ddof=1) sobre 300 000 pares
rng = np.random.default_rng(SEED)
celdas = [(i, j) for i in vx for j in vy]
idx = rng.choice(len(celdas), size=300_000, p=[float(conj.loc[c]) for c in celdas])
xy = np.array(celdas)[idx]
cov_m = np.cov(xy[:, 0], xy[:, 1])[0, 1]; rho_m = np.corrcoef(xy[:, 0], xy[:, 1])[0, 1]
print(f"Muestra: cov = {cov_m:.4f} | rho = {rho_m:.4f}")
assert abs(cov_m - float(cov)) < 0.01 and abs(rho_m - rho) < 0.01

# Contraejemplo: X uniforme en {-2,...,2}, Y = X^2 -> Cov = 0 pero Y es funcion de X
Xs = VADiscreta({k: Fraction(1, 5) for k in range(-2, 3)})
cov_sq = Xs.esperanza(lambda x: x * x**2) - Xs.esperanza() * Xs.esperanza(lambda x: x**2)
p_y4 = Xs.esperanza(lambda x: 1 if x**2 == 4 else 0)                         # P(Y = 4) = P(X = +-2) = 2/5
print(f"Cov(X, X^2) = {cov_sq} (correlacion 0) pero P(Y=4) = {p_y4} mientras que P(Y=4 | X=2) = 1")
assert cov_sq == 0 and p_y4 == Fraction(2, 5) != 1                            # dependientes pese a Cov = 0

fig, ejes = plt.subplots(1, 2, figsize=(11, 4.2))
ejes[0].imshow(conj.astype(float).to_numpy(), cmap="Blues", origin="lower")
for i in range(3):
    for j in range(3):
        ejes[0].text(j, i, f"{float(conj.iloc[i, j]):.2f}", ha="center", va="center")
ejes[0].set(title="Conjunta p(x, y): masa concentrada en la diagonal", xlabel="Y (devoluciones)", ylabel="X (quejas)", xticks=range(3), yticks=range(3)); ejes[0].grid(False)
xx = np.linspace(-2, 2, 100)
ejes[1].plot(xx, xx**2, color="gray", lw=1); ejes[1].scatter(range(-2, 3), [k**2 for k in range(-2, 3)], color=PALETA[1], s=70, zorder=3)
ejes[1].set(title="Y = X²: dependencia perfecta con correlación 0", xlabel="X", ylabel="Y")
plt.tight_layout(); plt.show()
````

---
**Lectura.** En la tabla de quejas y devoluciones, $E[X]=E[Y]=0.93$, $\text{Cov}=0.3051$ y $\rho\approx 0.540$: asociación positiva moderada (la masa se concentra en la diagonal). **No** son independientes (p. ej. $p(0,0)=0.20\ne 0.32\cdot 0.32$), la identidad $\text{Var}(X+Y)=\text{Var}X+\text{Var}Y+2\text{Cov}$ se cumple **exacta**, y `np.cov` sobre 300 000 pares simulados da 0.3051 y $\rho=0.540$. El contraejemplo $Y=X^2$ tiene $\text{Cov}=0$ aunque $Y$ **es** función de $X$ ($P(Y=4)=2/5$ pero $P(Y=4\mid X=2)=1$). **Moraleja:** $\rho=0$ descarta una relación *lineal*, no una relación; siempre se mira el gráfico de dispersión.

## 5. Desigualdades de Markov y Chebyshev, y Esperanza y Varianza Totales 🧮

Cuando no conocemos la distribución completa pero sí su media y su varianza, dos desigualdades dan **cotas garantizadas**:

$$\textbf{Markov } (X \ge 0,\ a>0): \quad P(X \ge a) \le \frac{E[X]}{a}, \qquad\qquad \textbf{Chebyshev: } P\big(\lvert X - \mu\rvert \ge k\sigma\big) \le \frac{1}{k^2}.$$

Valen para **cualquier** distribución (con esperanza, o varianza, finita), por eso son **conservadoras**: casi siempre la probabilidad real es mucho menor. Su valor es teórico: son la base de la **ley de los grandes números** y justifican por qué la desviación estándar es una unidad de medida natural para la dispersión.

Para una v. a. $X$ y un factor condicionante $G$ (el grupo, la finca, el segmento), las leyes de **esperanza y varianza totales** descomponen lo que se observa en el conjunto:

$$E[X] = E\big[E[X \mid G]\big], \qquad \text{Var}(X) = \underbrace{E\big[\text{Var}(X \mid G)\big]}_{\text{variabilidad dentro de los grupos}} + \underbrace{\text{Var}\big(E[X \mid G]\big)}_{\text{variabilidad entre los grupos}}.$$

Esta es la idea que está detrás del **ANOVA** (Módulo 03): separar la variación *dentro* de la variación *entre* grupos. **Ejemplo (sintético):** el rendimiento de parcelas de papa (kg por parcela) en dos fincas: la finca A aporta el 60 % de las parcelas con media 20 y desviación 3; la B el 40 % con media 28 y desviación 5.

````python
# ============================================================
# 5. Markov y Chebyshev vs la probabilidad real; esperanza y varianza totales con dos fincas
# ============================================================
# Exponencial(lambda=0.5): mu = sigma = 2
dist_e = stats.expon(scale=2.0)
filas = []
for k in (1.5, 2, 3, 4):
    exacta = dist_e.sf(2 + k * 2) + dist_e.cdf(2 - k * 2)                      # P(|X-mu| >= k sigma)
    norm_ = 2 * stats.norm.sf(k)                                              # lo mismo para una normal
    filas.append({"k": k, "cota de Chebyshev 1/k^2": 1 / k**2, "exponencial": exacta, "normal": norm_})
cheb = pd.DataFrame(filas).set_index("k")
print(cheb.round(4)); print()
assert (cheb["exponencial"] <= cheb["cota de Chebyshev 1/k^2"]).all() and (cheb["normal"] <= cheb["cota de Chebyshev 1/k^2"]).all()
assert cheb.loc[4, "cota de Chebyshev 1/k^2"] / cheb.loc[4, "normal"] > 100   # para la normal la cota es muy holgada

# Markov: P(X >= 6) <= E[X]/6 para la exponencial de media 2
markov = 2 / 6; real = dist_e.sf(6)
print(f"Markov: P(X >= 6) <= {markov:.3f}; real = {real:.4f}")
assert real <= markov

# La cota de Chebyshev NO se puede mejorar en general: existe una distribucion que la alcanza (k = 2)
k = 2
alcanza = VADiscreta({-k: Fraction(1, 2 * k**2), 0: 1 - Fraction(1, k**2), k: Fraction(1, 2 * k**2)})
assert alcanza.esperanza() == 0 and alcanza.varianza() == 1                    # mu = 0, sigma = 1
p_cola = sum(p for v, p in alcanza.pmf.items() if abs(v) >= k * 1)             # P(|X - mu| >= k sigma)
print(f"Distribucion extremal: P(|X| >= {k} sigma) = {p_cola} = 1/k^2 = {Fraction(1, k**2)}")
assert p_cola == Fraction(1, 4)

# Esperanza y varianza totales: dos fincas
peso = np.array([0.6, 0.4]); medias = np.array([20.0, 28.0]); sds = np.array([3.0, 5.0])
E_total = (peso * medias).sum()
var_dentro = (peso * sds**2).sum()
var_entre = (peso * (medias - E_total) ** 2).sum()
var_total = var_dentro + var_entre
rng = np.random.default_rng(SEED)
M = 1_000_000
g = rng.choice(2, size=M, p=peso)
rend = rng.normal(medias[g], sds[g])
print(f"E[X] = {E_total:.2f} (sim {rend.mean():.3f}) | Var dentro = {var_dentro:.2f} + entre = {var_entre:.2f} = {var_total:.2f} (sim {rend.var():.2f})")
assert abs(E_total - 23.2) < 1e-12 and abs(var_total - 30.76) < 1e-12
assert abs(rend.mean() - E_total) < 0.02 and abs(rend.var() - var_total) < 0.1
print(f"Fraccion de la varianza explicada por la finca: {var_entre / var_total:.1%}")
````

---
**Lectura.** **Chebyshev y Markov son correctas pero holgadas.** Con $k=2$ la cota es 0.25 y las probabilidades reales son 0.0498 (exponencial) y 0.0455 (normal); con $k=4$ la cota es 0.0625 frente a $6\times 10^{-5}$ en la normal (≈ 1 000 veces mayor). La de Markov para $P(X\ge 6)$ da $\le 0.333$ y la exponencial real es 0.0498. Aun así, la distribución de tres puntos construida **alcanza** la cota $1/k^2=1/4$: no existe una cota universal mejor. Con las dos fincas, $E[X]=23.2$ y la varianza total es $30.76 = 15.40$ (dentro) $+\ 15.36$ (entre); la simulación (23.203 y 30.74) lo confirma. La finca explica el **49.9 %** de la variabilidad: la mitad de la dispersión que ves en las parcelas viene de *a qué finca pertenecen*, no de lo que pasa dentro de cada una.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Variable aleatoria** | Una **función** $X:\Omega\to\mathbb{R}$; discreta (masa $p$, $\sum p = 1$) o continua (densidad $f$, $\int f = 1$, $P(X=x)=0$). La acumulada $F$ existe siempre y su inversa es la **función cuantil** |
| **Esperanza** | Centro de masa y **promedio a largo plazo**; es **lineal siempre** ($E[X+Y]=E[X]+E[Y]$ aun con dependencia) y $E[g(X)]\neq g(E[X])$ (Jensen: $E[X^2]-\mu^2=\text{Var}(X)$) |
| **Varianza** | $E[X^2]-\mu^2$; $\text{Var}(aX+b)=a^2\text{Var}(X)$; $\text{Var}(X+Y)=\text{Var}X+\text{Var}Y+2\,\text{Cov}$. La rifa tiene $E=-2000$ pesos y $\sigma\approx 94\,800$: la esperanza sola no describe el riesgo |
| **Continuas** | Probabilidad = área; la integración numérica (trapecio) coincide con `quad` y con `scipy.stats` hasta $10^{-8}$; **transformada inversa** $X=F^{-1}(U)$ simula cualquier distribución con $F$ invertible |
| **Conjuntas** | Marginales por suma; $\rho\in[-1,1]$ mide asociación **lineal**; **independiente $\Rightarrow$ $\rho=0$, no al revés** (con $Y=X^2$, $\text{Cov}=0$ pero dependencia total) |
| **Desigualdades** | Markov y Chebyshev valen para **cualquier** distribución, por eso son **holgadas** (para la normal, la cota con $k=4$ es más de 100 veces mayor que la probabilidad real), pero Chebyshev **no se puede mejorar** en general |
| **Totales** | $E[X]=E[E[X\mid G]]$ y $\text{Var}(X)=E[\text{Var}(X\mid G)]+\text{Var}(E[X\mid G])$: con dos fincas, el **49.9 %** de la varianza total (30.76) es variación **entre** fincas, la idea base del ANOVA |
