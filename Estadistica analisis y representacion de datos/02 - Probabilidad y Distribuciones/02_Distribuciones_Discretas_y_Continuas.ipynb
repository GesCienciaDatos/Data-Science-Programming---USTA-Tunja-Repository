## 1. La Galería de Distribuciones y Cómo Elegir ✨

Cada familia responde a un **mecanismo**. La pregunta correcta no es «¿qué forma tiene mi histograma?» sino «**¿cómo se generó el dato?**»:

| Familia (parámetros) | Soporte | Mecanismo / cuándo usarla | $E[X]$ | $\text{Var}(X)$ |
|---|---|---|---|---|
| **Bernoulli**($p$) | $\{0,1\}$ | un ensayo éxito/fracaso | $p$ | $p(1-p)$ |
| **Binomial**($n,p$) | $0,\dots,n$ | nº de éxitos en $n$ ensayos **independientes** con la misma $p$ | $np$ | $np(1-p)$ |
| **Geométrica**($p$) | $1,2,\dots$ | nº de ensayos hasta el **primer** éxito | $1/p$ | $(1-p)/p^2$ |
| **Poisson**($\lambda$) | $0,1,2,\dots$ | nº de eventos «raros» e independientes por unidad de tiempo o espacio | $\lambda$ | $\lambda$ |
| **Hipergeométrica**($N,K,n$) | $\max(0,n-N+K),\dots,\min(n,K)$ | éxitos en $n$ extracciones **sin reemplazo** de una población finita con $K$ éxitos | $nK/N$ | $n\frac{K}{N}\big(1-\frac{K}{N}\big)\frac{N-n}{N-1}$ |
| **Uniforme**($a,b$) | $(a,b)$ | ignorancia total dentro de un intervalo; base de la simulación | $(a+b)/2$ | $(b-a)^2/12$ |
| **Normal**($\mu,\sigma^2$) | $\mathbb{R}$ | suma de muchos efectos pequeños (TLC); errores de medición | $\mu$ | $\sigma^2$ |
| **Exponencial**($\lambda$) | $x \ge 0$ | tiempo de espera entre eventos de Poisson; **sin memoria** | $1/\lambda$ | $1/\lambda^2$ |
| **Gamma**($\alpha,\lambda$) | $x > 0$ | suma de $\alpha$ exponenciales (si $\alpha$ es entero); tiempos de servicio | $\alpha/\lambda$ | $\alpha/\lambda^2$ |
| **Beta**($a,b$) | $(0,1)$ | proporciones y probabilidades desconocidas | $\frac{a}{a+b}$ | $\frac{ab}{(a+b)^2(a+b+1)}$ |
| **Lognormal**($\mu,\sigma^2$) | $x > 0$ | producto de muchos factores positivos; ingresos, tamaños | $e^{\mu+\sigma^2/2}$ | $(e^{\sigma^2}-1)e^{2\mu+\sigma^2}$ |
| **Chi-cuadrado**($\nu$) | $x > 0$ | suma de $\nu$ normales estándar al cuadrado; base de varias pruebas | $\nu$ | $2\nu$ |
| **t de Student**($\nu$) | $\mathbb{R}$ | $Z/\sqrt{\chi^2_\nu/\nu}$: media muestral cuando se estima $\sigma$ | $0$ ($\nu>1$) | $\frac{\nu}{\nu-2}$ ($\nu>2$) |

> 💡 **API de `scipy.stats`:** toda distribución tiene los mismos métodos: `pmf` o `pdf`, `cdf`, `sf` ($=1-F$), `ppf` (cuantil, la inversa de `cdf`), `rvs` (muestras), `mean`, `var`, `fit` (máxima verosimilitud). Nota: `scipy` parametriza la exponencial y la gamma con `scale` $=1/\lambda$.

Empezamos por las **discretas**: escribimos cada masa desde cero (usando `math.comb` y, para Poisson, logaritmos con `math.lgamma` para evitar desbordamientos) y las comparamos con `scipy.stats`. Los parámetros de los ejemplos son **hipotéticos**: 20 semillas con 80 % de germinación (binomial); 15 % de probabilidad de cerrar una venta por llamada (geométrica); 4 buses por hora en un paradero (Poisson); un lote de 50 piezas con 6 defectuosas del que se inspeccionan 10 (hipergeométrica).

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib scipy

import math
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import stats

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")
````

````python
# ============================================================
# 1. Discretas desde cero: binomial, geometrica, Poisson e hipergeometrica (contra scipy.stats y por momentos)
# ============================================================
def binom_pmf(k, n, p):
    return math.comb(n, k) * p**k * (1 - p) ** (n - k)


def geom_pmf(k, p):                                                            # k = 1, 2, ... ensayos hasta el primer exito
    return (1 - p) ** (k - 1) * p


def pois_pmf(k, lam):                                                          # en escala log para evitar desbordamientos con k grande
    return math.exp(-lam + k * math.log(lam) - math.lgamma(k + 1))


def hyper_pmf(k, N, K, n):
    return math.comb(K, k) * math.comb(N - K, n - k) / math.comb(N, n)


n_b, p_b = 20, 0.80
p_g = 0.15
lam = 4.0
N_h, K_h, n_h = 50, 6, 10

assert all(abs(binom_pmf(k, n_b, p_b) - stats.binom.pmf(k, n_b, p_b)) < 1e-14 for k in range(n_b + 1))
assert all(abs(geom_pmf(k, p_g) - stats.geom.pmf(k, p_g)) < 1e-14 for k in range(1, 40))
assert all(abs(pois_pmf(k, lam) - stats.poisson.pmf(k, lam)) < 1e-14 for k in range(60))
assert all(abs(hyper_pmf(k, N_h, K_h, n_h) - stats.hypergeom.pmf(k, N_h, K_h, n_h)) < 1e-14 for k in range(0, min(K_h, n_h) + 1))

# E y Var calculadas sumando la masa vs formulas cerradas (soporte truncado donde es infinito)
def momentos(pmf, soporte):
    m = sum(k * pmf(k) for k in soporte); v = sum((k - m) ** 2 * pmf(k) for k in soporte)
    return m, v


f_h = n_h * K_h / N_h
filas = {
    "Binomial(20, 0.8)": (momentos(lambda k: binom_pmf(k, n_b, p_b), range(n_b + 1)), (n_b * p_b, n_b * p_b * (1 - p_b))),
    "Geometrica(0.15)": (momentos(lambda k: geom_pmf(k, p_g), range(1, 400)), (1 / p_g, (1 - p_g) / p_g**2)),
    "Poisson(4)": (momentos(lambda k: pois_pmf(k, lam), range(100)), (lam, lam)),
    "Hipergeom(50, 6, 10)": (momentos(lambda k: hyper_pmf(k, N_h, K_h, n_h), range(0, 7)), (f_h, n_h * (K_h / N_h) * (1 - K_h / N_h) * (N_h - n_h) / (N_h - 1))),
}
tabla = pd.DataFrame({k: {"E (suma)": v[0][0], "E (formula)": v[1][0], "Var (suma)": v[0][1], "Var (formula)": v[1][1]} for k, v in filas.items()}).T
print(tabla.round(4))
assert all(abs(v[0][0] - v[1][0]) < 1e-9 and abs(v[0][1] - v[1][1]) < 1e-9 for v in filas.values())

fig, ejes = plt.subplots(1, 4, figsize=(15, 3.6))
ks = np.arange(0, 21); ejes[0].bar(ks, [binom_pmf(k, n_b, p_b) for k in ks], color=PALETA[0]); ejes[0].set_title("Binomial(20, 0.8)\nsemillas que germinan")
ks = np.arange(1, 31); ejes[1].bar(ks, [geom_pmf(k, p_g) for k in ks], color=PALETA[1]); ejes[1].set_title("Geométrica(0.15)\nllamadas hasta la 1.ª venta")
ks = np.arange(0, 15); ejes[2].bar(ks, [pois_pmf(k, lam) for k in ks], color=PALETA[2]); ejes[2].set_title("Poisson(4)\nbuses por hora")
ks = np.arange(0, 7); ejes[3].bar(ks, [hyper_pmf(k, N_h, K_h, n_h) for k in ks], color=PALETA[3]); ejes[3].set_title("Hipergeométrica(50, 6, 10)\ndefectuosas en la muestra")
for a in ejes: a.set_xlabel("k")
ejes[0].set_ylabel("P(X = k)")
plt.tight_layout(); plt.show()
print("P(exactamente 16 de 20 germinan) =", round(binom_pmf(16, n_b, p_b), 4), "| P(al menos 18) =", round(1 - stats.binom.cdf(17, n_b, p_b), 4))
print("P(ningun defectuoso en la muestra de 10) =", round(hyper_pmf(0, N_h, K_h, n_h), 4), "vs con reemplazo (binomial) =", round(binom_pmf(0, n_h, K_h / N_h), 4))
````

**Lectura.** Las cuatro masas escritas desde cero coinciden con `scipy.stats` hasta $10^{-14}$, y sumar la masa reproduce las fórmulas: binomial $E=16$, $\text{Var}=3.2$; Poisson $E=\text{Var}=4$; hipergeométrica $E=1.2$, $\text{Var}=0.862$. La **geométrica** espera en promedio $1/p=6.7$ llamadas hasta la primera venta, pero con una desviación estándar de ≈ 6.1: la **variabilidad es enorme** (a veces la venta llega en la primera llamada y a veces en la vigésima). Sin reemplazo, la probabilidad de no hallar defectuosos en la muestra de 10 es **0.2415**, menor que la 0.2785 de la binomial (con reemplazo): al no repetir piezas, la muestra informa más, y la varianza hipergeométrica es la binomial multiplicada por $\frac{N-n}{N-1}=0.816$.

## 2. Aproximaciones Entre Familias: Poisson, Normal y Binomial 🔀

Las familias no son islas. Tres aproximaciones clásicas (y **cuándo fallan**):

| Aproximación | Condición típica | Idea |
|---|---|---|
| **Binomial $\to$ Poisson** | $n$ grande, $p$ pequeña, $\lambda = np$ moderada | eventos raros e independientes: $\text{Bin}(n,p)\approx\text{Poisson}(np)$; el error en variación total es a lo sumo $np^2$ (cota de Le Cam) |
| **Binomial $\to$ Normal** | $np\ge 10$ y $n(1-p)\ge 10$ (regla práctica) | $\text{Bin}(n,p)\approx N\big(np,\,np(1-p)\big)$; para una acumulada discreta se usa la **corrección de continuidad**: $P(X\le k)\approx\Phi\Big(\dfrac{k+0.5-np}{\sqrt{np(1-p)}}\Big)$ |
| **Hipergeométrica $\to$ Binomial** | población $N$ mucho mayor que la muestra $n$ | extraer sin reemplazo de una población enorme es casi como extraer con reemplazo; la diferencia es el **factor de población finita** $\frac{N-n}{N-1}$ |

Medimos la calidad de cada aproximación con la **distancia de variación total** entre dos masas $p$ y $q$: $\text{TV}(p,q) = \tfrac{1}{2}\sum_k \lvert p(k)-q(k)\rvert \in [0,1]$ (0 = idénticas; es la mayor diferencia posible entre las probabilidades que ambas asignan a un mismo evento).

````python
# ============================================================
# 2. Aproximaciones: Poisson a binomial, normal a binomial (con y sin correccion de continuidad), binomial a hipergeometrica
# ============================================================
def tv(p, q):
    """Distancia de variacion total entre dos vectores de masa alineados (completando con ceros)."""
    m = max(len(p), len(q)); p = np.pad(p, (0, m - len(p))); q = np.pad(q, (0, m - len(q)))
    return 0.5 * float(np.abs(p - q).sum())


# (a) Binomial -> Poisson: n grande, p pequena
casos = [(20, 0.20), (100, 0.04), (200, 0.02), (1000, 0.004)]
filas = []
for n, p in casos:
    k = np.arange(0, n + 1)
    filas.append({"n": n, "p": p, "np": n * p, "TV(Bin, Poisson)": tv(stats.binom.pmf(k, n, p), stats.poisson.pmf(np.arange(0, 80), n * p)), "cota n p^2": n * p**2})
aprox_poi = pd.DataFrame(filas)
print(aprox_poi.round(5).to_string(index=False))
assert (aprox_poi["TV(Bin, Poisson)"] <= aprox_poi["cota n p^2"] + 1e-12).all()      # la cota de Le Cam se cumple
assert aprox_poi["TV(Bin, Poisson)"].is_monotonic_decreasing and aprox_poi["TV(Bin, Poisson)"].iloc[-1] < 0.004

# (b) Binomial -> Normal con y sin correccion de continuidad
def cdf_norm_aprox(k, n, p, cc=True):
    mu, sd = n * p, math.sqrt(n * p * (1 - p))
    return stats.norm.cdf((k + (0.5 if cc else 0.0) - mu) / sd)


filas = []
for n, p, k in [(100, 0.30, 35), (400, 0.30, 130), (20, 0.05, 0), (20, 0.50, 12)]:
    ex = stats.binom.cdf(k, n, p)
    filas.append({"n": n, "p": p, "k": k, "np": n * p, "exacta": ex, "normal sin cc": cdf_norm_aprox(k, n, p, False), "normal con cc": cdf_norm_aprox(k, n, p, True)})
aprox_n = pd.DataFrame(filas)
aprox_n["error sin cc"] = (aprox_n["normal sin cc"] - aprox_n["exacta"]).abs(); aprox_n["error con cc"] = (aprox_n["normal con cc"] - aprox_n["exacta"]).abs()
print("\n" + aprox_n.round(4).to_string(index=False))
assert (aprox_n.loc[[0, 1, 3], "error con cc"] < aprox_n.loc[[0, 1, 3], "error sin cc"]).all()    # la correccion mejora cuando np >= 10
assert aprox_n.loc[2, "error con cc"] > 0.04                                                       # np = 1: la normal NO sirve (honestidad)

# (c) Hipergeometrica -> Binomial segun la fraccion de muestreo n/N
filas = []
for N, K, n in [(50, 10, 20), (200, 40, 20), (1000, 200, 20), (10_000, 2000, 20)]:
    k = np.arange(0, n + 1)
    filas.append({"N": N, "n/N": n / N, "TV(Hiper, Binomial)": tv(stats.hypergeom.pmf(k, N, K, n), stats.binom.pmf(k, n, K / N)), "var hiper / var binom": ((N - n) / (N - 1))})
aprox_h = pd.DataFrame(filas)
print("\n" + aprox_h.round(4).to_string(index=False))
assert aprox_h["TV(Hiper, Binomial)"].is_monotonic_decreasing and aprox_h["TV(Hiper, Binomial)"].iloc[-1] < 0.01

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
k = np.arange(0, 16)
ejes[0].bar(k - 0.2, stats.binom.pmf(k, 200, 0.02), width=0.4, color=PALETA[0], label="Binomial(200, 0.02)")
ejes[0].bar(k + 0.2, stats.poisson.pmf(k, 4), width=0.4, color=PALETA[1], label="Poisson(4)"); ejes[0].legend(); ejes[0].set_title("Poisson aproxima a la binomial (eventos raros)")
k = np.arange(10, 55); ejes[1].bar(k, stats.binom.pmf(k, 100, 0.3), color=PALETA[0], alpha=0.5, label="Binomial(100, 0.3)")
xx = np.linspace(10, 55, 300); ejes[1].plot(xx, stats.norm.pdf(xx, 30, math.sqrt(21)), color=PALETA[1], lw=2, label="Normal(30, 21)"); ejes[1].legend(); ejes[1].set_title("Normal aproxima a la binomial (np = 30)")
k = np.arange(0, 8); ejes[2].bar(k, stats.binom.pmf(k, 20, 0.05), color=PALETA[0], alpha=0.5, label="Binomial(20, 0.05)")
xx = np.linspace(-2, 8, 300); ejes[2].plot(xx, stats.norm.pdf(xx, 1, math.sqrt(0.95)), color=PALETA[1], lw=2, label="Normal(1, 0.95)"); ejes[2].legend(); ejes[2].set_title("...pero falla con np = 1: asimétrica y acotada en 0")
plt.tight_layout(); plt.show()
````

**Lectura.** **Poisson a binomial:** la distancia de variación total baja de 0.056 ($n=20$, $p=0.2$) a 0.001 ($n=1000$, $p=0.004$) manteniendo $np=4$, siempre por debajo de la cota $np^2$ (que es muy holgada pero cierta). **Normal a binomial:** la corrección de continuidad reduce el error de la acumulada de 0.0215 a 0.0010 (con $np=30$) y de 0.054 a 0.0002 (con $n=20$, $p=0.5$); pero con $np=1$ ($n=20$, $p=0.05$) la normal **sigue fallando** aun corregida: $P(X\le 0)$ exacta 0.3585 frente a 0.3040 (error 0.055), porque la binomial es asimétrica y está acotada en 0. ## 3. Distribuciones Continuas: Normal, Exponencial, Gamma, $\chi^2$ y $t$ 📐

**La normal.** $X\sim N(\mu,\sigma^2)$ tiene densidad $f(x)=\dfrac{1}{\sigma\sqrt{2\pi}}\exp\Big(-\dfrac{(x-\mu)^2}{2\sigma^2}\Big)$. Su acumulada **no tiene forma cerrada elemental**, pero se expresa con la función de error: $\Phi(z)=\tfrac12\big(1+\text{erf}(z/\sqrt2)\big)$. La **estandarización** $Z=(X-\mu)/\sigma\sim N(0,1)$ reduce todo a una sola tabla; la **regla empírica** dice que $P(\lvert Z\rvert<1)\approx 0.683$, $P(\lvert Z\rvert<2)\approx 0.954$, $P(\lvert Z\rvert<3)\approx 0.997$. Los cuantiles $\Phi^{-1}(p)$ (como 1.645, 1.960, 2.576) son la base de los intervalos de confianza del Módulo 03.

**La exponencial y Poisson.** Si los eventos ocurren según un **proceso de Poisson** de tasa $\lambda$, el **número** de eventos en un tiempo $t$ es $\text{Poisson}(\lambda t)$ y el **tiempo entre eventos** es $\text{Exponencial}(\lambda)$. Es la única continua **sin memoria**: $P(T>s+t\mid T>s)=P(T>t)$; que el bus lleve 3 minutos sin pasar no lo hace «más cercano».

**Sumas y razones que fabrican familias** (todas verificadas por simulación abajo):

- $\text{Gamma}(\alpha,\lambda)$ = suma de $\alpha$ exponenciales independientes (para $\alpha$ entero).
- $\chi^2_\nu = Z_1^2+\cdots+Z_\nu^2$ (suma de $\nu$ normales estándar al cuadrado). Es una gamma con $\alpha=\nu/2$, $\lambda=1/2$.
- $t_\nu = Z\big/\sqrt{\chi^2_\nu/\nu}$: la normal estándar con la incertidumbre de haber **estimado** $\sigma$; tiene **colas más pesadas** que la normal y tiende a ella cuando $\nu\to\infty$.

````python
# ============================================================
# 3a. Normal desde cero (erf + biseccion), regla empirica, exponencial sin memoria y familias derivadas por simulacion
# ============================================================
def norm_pdf(x, mu=0.0, sigma=1.0):
    z = (np.asarray(x, dtype=float) - mu) / sigma
    return np.exp(-0.5 * z**2) / (sigma * math.sqrt(2 * math.pi))


def norm_cdf(x, mu=0.0, sigma=1.0):
    return 0.5 * (1 + np.vectorize(math.erf)((np.asarray(x, dtype=float) - mu) / (sigma * math.sqrt(2))))


def norm_ppf(p, mu=0.0, sigma=1.0):
    """Cuantil por biseccion sobre la acumulada (escalar)."""
    lo, hi = -12.0, 12.0
    for _ in range(200):
        mid = (lo + hi) / 2
        lo, hi = (mid, hi) if float(norm_cdf(mid)) < p else (lo, mid)
    return mu + sigma * (lo + hi) / 2


xs = np.linspace(-4, 4, 41)
assert np.abs(norm_pdf(xs, 1.5, 2.0) - stats.norm.pdf(xs, 1.5, 2.0)).max() < 1e-14
assert np.abs(norm_cdf(xs, 1.5, 2.0) - stats.norm.cdf(xs, 1.5, 2.0)).max() < 1e-14
assert all(abs(norm_ppf(p) - stats.norm.ppf(p)) < 1e-9 for p in (0.001, 0.05, 0.5, 0.975, 0.999))
print("Cuantiles: z(0.95) =", round(norm_ppf(0.95), 4), "| z(0.975) =", round(norm_ppf(0.975), 4), "| z(0.995) =", round(norm_ppf(0.995), 4))

# Regla empirica con 'pesos de papas' sinteticos N(180 g, 25 g)
rng = np.random.default_rng(SEED)
pesos = rng.normal(180, 25, 200_000)
regla = pd.DataFrame({"teorica": [float(norm_cdf(k) - norm_cdf(-k)) for k in (1, 2, 3)],
                      "en la muestra": [float((np.abs(pesos - 180) < k * 25).mean()) for k in (1, 2, 3)]}, index=["±1 sigma", "±2 sigma", "±3 sigma"])
print("\n", regla.round(4))
assert np.allclose(regla["teorica"], [0.6827, 0.9545, 0.9973], atol=1e-4) and np.abs(regla["teorica"] - regla["en la muestra"]).max() < 0.003
z230 = (230 - 180) / 25
print(f"Una papa de 230 g tiene z = {z230:.1f} y percentil {float(norm_cdf(z230)):.1%}")

# Exponencial: falta de memoria y relacion con Poisson (tasa 0.5 eventos/min)
lam_e = 0.5
t = rng.exponential(1 / lam_e, 1_000_000)
s_, t_ = 3.0, 2.0
cond = (t[t > s_] > s_ + t_).mean(); incond = (t > t_).mean()
print(f"\nSin memoria: P(T > 5 | T > 3) = {cond:.4f} | P(T > 2) = {incond:.4f} | exacta e^-1 = {math.exp(-1):.4f}")
assert abs(cond - incond) < 0.005 and abs(cond - math.exp(-lam_e * t_)) < 0.005

gaps = rng.exponential(1 / lam_e, size=(100_000, 40))                          # llegadas = suma acumulada de esperas exponenciales
llegadas = np.cumsum(gaps, axis=1)
conteo = (llegadas <= 10).sum(axis=1)                                          # eventos en una ventana de 10 minutos
frec = np.bincount(conteo, minlength=20)[:20] / len(conteo)
d_tv = tv(frec, stats.poisson.pmf(np.arange(20), lam_e * 10))
print(f"Conteos en 10 min con esperas exponenciales vs Poisson({lam_e * 10:.0f}): TV = {d_tv:.4f}, media = {conteo.mean():.3f}, var = {conteo.var():.3f}")
assert llegadas[:, -1].min() > 10 and d_tv < 0.01 and abs(conteo.mean() - 5) < 0.05 and abs(conteo.var() - 5) < 0.1

# Familias derivadas: gamma, chi-cuadrado y t (estadistico de Kolmogorov-Smirnov contra la distribucion teorica)
M = 200_000
g3 = rng.exponential(2.0, size=(M, 3)).sum(axis=1)                             # suma de 3 exponenciales (media 2) ~ Gamma(3, scale=2)
chi4 = (rng.standard_normal((M, 4)) ** 2).sum(axis=1)                          # suma de 4 normales^2 ~ chi2(4)
t5 = rng.standard_normal(M) / np.sqrt((rng.standard_normal((M, 5)) ** 2).sum(axis=1) / 5)    # Z / sqrt(chi2_5 / 5) ~ t(5)
ks_gamma = stats.kstest(g3, stats.gamma(3, scale=2).cdf).statistic
ks_chi = stats.kstest(chi4, stats.chi2(4).cdf).statistic
ks_t = stats.kstest(t5, stats.t(5).cdf).statistic
print(f"KS: suma de 3 exponenciales vs Gamma(3): {ks_gamma:.4f} | suma de 4 Z^2 vs chi2(4): {ks_chi:.4f} | Z/sqrt(chi2/5) vs t(5): {ks_t:.4f}")
assert max(ks_gamma, ks_chi, ks_t) < 0.005                                      # ~ 1.95/sqrt(M) = 0.0044 es el nivel 0.001
print(f"Colas: P(|T|>2) con t(5) = {2 * stats.t.sf(2, 5):.4f} frente a la normal = {2 * stats.norm.sf(2):.4f}  (la t tiene colas mas pesadas)")
assert 2 * stats.t.sf(2, 5) > 2 * 2 * stats.norm.sf(2) > 0.09 and abs(stats.chi2(4).mean() - 4) < 1e-12
````

## 4. Ajuste y Diagnóstico: ¿Qué Familia Describe Mis Datos? 🔍

Ajustar una distribución es **estimar sus parámetros** y **verificar que el modelo es razonable**. Dos métodos de estimación:

- **Método de los momentos:** igualar los momentos teóricos a los muestrales (para Poisson, $\hat\lambda=\bar x$; para la gamma, $\hat\alpha=\bar x^2/s^2$ y $\hat\lambda=\bar x/s^2$).
- **Máxima verosimilitud (MLE):** el valor del parámetro que **maximiza** la probabilidad de lo observado, $\hat\theta=\arg\max_\theta \sum_i \log f(x_i;\theta)$ (`dist.fit(datos)` en `scipy.stats`).

**Diagnóstico** (nunca basta un solo número):

1. **Gráfico Q-Q:** cuantiles de los datos contra cuantiles del modelo; si el modelo es bueno, los puntos caen sobre la diagonal.
2. **Índice de dispersión** para conteos: $D=s^2/\bar x$. En Poisson $E=\text{Var}$, así que $D\approx 1$; $D>1$ es **sobredispersión** (frecuente en conteos reales: heterogeneidad entre unidades), y la **binomial negativa** la modela.
3. **Criterio de Akaike:** $\text{AIC}=2k-2\ln\hat L$ (con $k$ parámetros); **menor es mejor** y penaliza la complejidad. Sirve para **comparar** modelos, no para declarar que uno sea «verdadero».
4. **Prueba de Kolmogorov-Smirnov / chi-cuadrado de ajuste:** ojo, cuando los parámetros se **estiman con los mismos datos**, los valores $p$ de KS son **demasiado optimistas** (hace falta la corrección de Lilliefors o *bootstrap*). Por eso aquí usamos principalmente AIC y gráficos.

**Dos casos sintéticos.** (A) **Conteos de plagas por parcela** (400 parcelas) generados con una **binomial negativa** (heterogeneidad entre parcelas): ¿resiste el modelo Poisson «natural»? (B) **Tiempos de atención** (500 clientes, en minutos) generados con una **gamma**: ¿distinguimos entre exponencial, gamma y lognormal?

````python
# ============================================================
# 4. Ajuste y diagnostico: sobredispersion en conteos (Poisson vs binomial negativa) y tiempos (exponencial, gamma, lognormal)
# ============================================================
rng = np.random.default_rng(SEED)

# (A) Conteos de plagas por parcela (sinteticos: binomial negativa r=3, p=0.5 -> media 3, varianza 6)
conteos_plaga = stats.nbinom.rvs(3, 0.5, size=400, random_state=rng)
xbar, s2 = conteos_plaga.mean(), conteos_plaga.var(ddof=1)
D = s2 / xbar
chi_disp = (len(conteos_plaga) - 1) * D                                          # bajo Poisson ~ chi2(n-1)
p_disp = stats.chi2.sf(chi_disp, len(conteos_plaga) - 1)
print(f"Plagas: media = {xbar:.3f}, varianza = {s2:.3f}, indice de dispersion D = {D:.2f} | prueba de dispersion: chi2 = {chi_disp:.0f}, p = {p_disp:.1e}")

r_hat = xbar**2 / (s2 - xbar); p_hat = xbar / s2                                 # binomial negativa por momentos
ll_pois = stats.poisson.logpmf(conteos_plaga, xbar).sum()
ll_nb = stats.nbinom.logpmf(conteos_plaga, r_hat, p_hat).sum()
aic_pois, aic_nb = 2 * 1 - 2 * ll_pois, 2 * 2 - 2 * ll_nb
print(f"Poisson(lambda={xbar:.2f}): logL = {ll_pois:.1f}, AIC = {aic_pois:.1f} | NegBin(r={r_hat:.2f}, p={p_hat:.2f}): logL = {ll_nb:.1f}, AIC = {aic_nb:.1f}")

obs = np.bincount(conteos_plaga, minlength=12)[:12]
ks_ = np.arange(12)
fig, ejes = plt.subplots(1, 3, figsize=(16, 4.2))
ejes[0].bar(ks_ - 0.27, obs / len(conteos_plaga), width=0.27, color="#94a3b8", label="observado")
ejes[0].bar(ks_, stats.poisson.pmf(ks_, xbar), width=0.27, color=PALETA[1], label="Poisson ajustada")
ejes[0].bar(ks_ + 0.27, stats.nbinom.pmf(ks_, r_hat, p_hat), width=0.27, color=PALETA[2], label="Binomial negativa ajustada")
ejes[0].set(title="Conteos de plagas: la Poisson subestima el 0 y las colas", xlabel="plagas por parcela", ylabel="proporción"); ejes[0].legend()

# (B) Tiempos de atencion (sinteticos: gamma de forma 2.5 y escala 2 -> media 5 min)
tiempos = stats.gamma.rvs(2.5, scale=2.0, size=500, random_state=rng)
cand = {}
cand["Exponencial"] = (stats.expon, stats.expon.fit(tiempos, floc=0), 1)
cand["Gamma"] = (stats.gamma, stats.gamma.fit(tiempos, floc=0), 2)
cand["Lognormal"] = (stats.lognorm, stats.lognorm.fit(tiempos, floc=0), 2)
filas = {}
for nombre, (dist, par, kpar) in cand.items():
    ll = dist.logpdf(tiempos, *par).sum()
    filas[nombre] = {"logL": ll, "AIC": 2 * kpar - 2 * ll, "KS (optimista)": stats.kstest(tiempos, dist(*par).cdf).statistic}
res = pd.DataFrame(filas).T.sort_values("AIC")
res["dAIC"] = res["AIC"] - res["AIC"].min()
print("\nTiempos de atencion (media muestral %.2f min):" % tiempos.mean()); print(res.round(2))
a_hat, _, esc_hat = cand["Gamma"][1]
print(f"Gamma por MLE: forma = {a_hat:.2f} (verdadera 2.5), escala = {esc_hat:.2f} (verdadera 2.0)")

q = (np.arange(1, len(tiempos) + 1) - 0.5) / len(tiempos)
for nombre, c in (("Exponencial", PALETA[1]), ("Gamma", PALETA[2])):
    dist, par, _ = cand[nombre]
    ejes[1 if nombre == "Exponencial" else 2].scatter(dist(*par).ppf(q), np.sort(tiempos), s=8, color=c)
    ejes[1 if nombre == "Exponencial" else 2].plot([0, 25], [0, 25], "k--", lw=1)
    ejes[1 if nombre == "Exponencial" else 2].set(title=f"Gráfico Q-Q: {nombre}", xlabel="cuantiles del modelo", ylabel="cuantiles de los datos")
plt.tight_layout(); plt.show()

assert D > 1.5 and p_disp < 1e-6 and aic_nb < aic_pois - 20                       # sobredispersion clara: la Poisson se rechaza y la binomial negativa gana
assert res.loc["Gamma", "AIC"] < res.loc["Exponencial", "AIC"] - 20               # la exponencial es claramente peor que la gamma
assert abs(a_hat - 2.5) < 0.6 and abs(esc_hat - 2.0) < 0.6
print("\nOrden por AIC:", list(res.index), "| distancia entre gamma y lognormal:", round(abs(res.loc["Gamma", "AIC"] - res.loc["Lognormal", "AIC"]), 1), "puntos de AIC")
````

**Lectura.** **Conteos:** la media de las plagas es 2.925 pero la varianza es 6.466: el índice de dispersión es **$D=2.21$** y la prueba de dispersión da $\chi^2=882$ con 399 grados de libertad ($p\approx 10^{-38}$), así que el modelo **Poisson es incompatible** con los datos (subestima las parcelas sin plagas y las de muchas plagas, como se ve en el primer gráfico). La binomial negativa ajustada por momentos ($\hat r=2.42$, $\hat p=0.45$; los valores verdaderos eran 3 y 0.5: con 400 parcelas la estimación es ruidosa) tiene un AIC de 1 740.5 frente a 1 892.9 de la Poisson, una diferencia enorme. **Tiempos:** la exponencial queda 187.7 puntos de AIC por encima de la mejor; la gamma ($\hat\alpha=2.52$, escala 2.05, cercanos a los verdaderos 2.5 y 2.0) gana a la lognormal por 39 puntos, y el gráfico Q-Q de la exponencial se curva lejos de la diagonal. Dos advertencias: (1) los KS (0.02, 0.05 y 0.19) usan parámetros estimados con los mismos datos y son **optimistas**; (2) el AIC compara solo a los candidatos considerados, y **sabemos** que la gamma gana porque los datos se generaron con ella: con datos reales y pocos casos, gamma y lognormal suelen ser difíciles de separar.

assert abs(p_10_o_mas - stats.poisson.sf(9, 6)) < 1e-12 and 0.08 < p_10_o_mas < 0.09
assert abs(p_cero_15min - p_espera_15min) < 1e-12
print("Son el MISMO evento: 'no hay llamadas en 15 minutos' equivale a 'la proxima llamada tarda mas de 15 minutos'. Poisson cuenta eventos; la exponencial mide esperas.")

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Elegir una familia** | Se elige por el **mecanismo** (ensayos $\to$ binomial/geométrica; conteos de eventos raros $\to$ Poisson; sin reemplazo $\to$ hipergeométrica; suma de efectos $\to$ normal; esperas $\to$ exponencial/gamma; proporciones $\to$ beta), no por la forma del histograma |
| **Desde cero** | Las masas y densidades escritas con `math.comb`, `math.lgamma` y `math.erf` coinciden con `scipy.stats` hasta $10^{-14}$, y las sumas de la masa reproducen $E$ y $\text{Var}$ de las fórmulas |
| **Aproximaciones** | Poisson para la binomial con $p$ pequeña (error acotado por $np^2$); normal con **corrección de continuidad** cuando $np\ge 10$ (**falla** con $np=1$); la binomial aproxima a la hipergeométrica cuando $n/N$ es pequeña |
| **Normal** | Estandarización, cuantiles ($1.645$, $1.960$, $2.576$) y regla empírica 68-95-99.7 verificada en una muestra |
| **Exponencial y Poisson** | Mismo proceso, dos caras: el **conteo** es Poisson y la **espera** es exponencial (**sin memoria**); $P(\text{0 eventos en }t)=P(T>t)$ |
| **Familias derivadas** | Suma de exponenciales $\to$ gamma; suma de $Z^2\to\chi^2$; $Z/\sqrt{\chi^2/\nu}\to t$, de **colas más pesadas** que la normal |
| **Ajuste honesto** | Conteos con **sobredispersión** ($D\approx 2$) **rechazan** la Poisson y favorecen la binomial negativa (AIC); para tiempos, la exponencial pierde por 188 puntos de AIC y la gamma gana a la lognormal por 39 (los datos *eran* gamma; con pocos datos reales la separación sería menos clara); los $p$ de KS con parámetros estimados son optimistas |
