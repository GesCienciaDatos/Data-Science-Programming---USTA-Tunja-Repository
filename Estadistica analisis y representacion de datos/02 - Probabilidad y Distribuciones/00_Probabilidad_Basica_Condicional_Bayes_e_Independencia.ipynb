## 1. Espacio Muestral, Eventos y los Axiomas de Kolmogorov 🎲

Un **experimento aleatorio** es un proceso cuyo resultado no se puede predecir con certeza, aunque sí se conocen los resultados posibles. Tres objetos lo describen:

- **Espacio muestral** $\Omega$: el conjunto de **todos** los resultados posibles (lanzar dos dados: 36 pares $(d_1, d_2)$).
- **Evento** $A \subseteq \Omega$: un subconjunto de resultados («la suma es 7»). Ocurre si el resultado observado $\omega \in A$.
- **Probabilidad** $P$: una función que asigna a cada evento un número, y que cumple los **tres axiomas de Kolmogorov (1933)**:

$$\text{(1)}\ P(A) \ge 0, \qquad \text{(2)}\ P(\Omega) = 1, \qquad \text{(3)}\ A_i \cap A_j = \emptyset \ (i \ne j) \ \Rightarrow\ P\Big(\bigcup_{i} A_i\Big) = \sum_i P(A_i).$$

De ellos se **deducen** (no se suponen) las reglas que usamos todo el tiempo:

| Regla | Fórmula | Por qué |
|---|---|---|
| **Complemento** | $P(A^c) = 1 - P(A)$ | $A$ y $A^c$ son excluyentes y su unión es $\Omega$ |
| **Evento imposible** | $P(\emptyset) = 0$ | caso particular del complemento con $A = \Omega$ |
| **Monotonía** | $A \subseteq B \Rightarrow P(A) \le P(B)$ | $B = A \cup (B \setminus A)$, y $P(B \setminus A) \ge 0$ |
| **Inclusión-exclusión** | $P(A \cup B) = P(A) + P(B) - P(A \cap B)$ | sumar $P(A)+P(B)$ cuenta dos veces la intersección |

**Tres formas de interpretar $P$** (los axiomas son los mismos en las tres):

| Interpretación | Idea | Cuándo sirve | Límite |
|---|---|---|---|
| **Clásica** | $P(A) = \lvert A \rvert / \lvert \Omega \rvert$ si los resultados son **equiprobables** | dados, cartas, sorteos | exige simetría; no sirve para «lloverá mañana» |
| **Frecuentista** | $P(A) = \lim_{n \to \infty} n_A / n$ (frecuencia relativa) | experimentos repetibles | no aplica a eventos únicos |
| **Subjetiva (bayesiana)** | grado de creencia, actualizado con evidencia | eventos únicos, decisiones | depende de la creencia inicial (*prior*) |

En la celda siguiente enumeramos $\Omega$ para dos dados con `itertools.product` y calculamos las probabilidades con **fracciones exactas** (`Fraction`, sin errores de redondeo); después **simulamos** para ver la interpretación frecuentista.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib scipy

import itertools
import math
import warnings
from fractions import Fraction

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import special

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")
````

````python
# ============================================================
# 1. Espacio muestral de dos dados, axiomas verificados y convergencia de la frecuencia relativa
# ============================================================
Omega = list(itertools.product(range(1, 7), repeat=2))                     # 36 resultados equiprobables (dado 1, dado 2)


def prob(evento, espacio=Omega):
    """Probabilidad clasica |A|/|Omega| como fraccion exacta."""
    return Fraction(sum(1 for w in espacio if evento(w)), len(espacio))


A = lambda w: w[0] + w[1] == 7                                              # la suma es 7
B = lambda w: 6 in w                                                        # sale al menos un 6
C = lambda w: w[0] == w[1]                                                  # dobles

eventos = {"A: suma = 7": A, "B: al menos un 6": B, "C: dobles": C,
           "A y B": lambda w: A(w) and B(w), "A o B": lambda w: A(w) or B(w), "A o C": lambda w: A(w) or C(w), "no A": lambda w: not A(w)}
tabla = pd.DataFrame({"P(evento)": {k: str(prob(f)) for k, f in eventos.items()}, "decimal": {k: round(float(prob(f)), 4) for k, f in eventos.items()}})
print(tabla, "\n")

# Axiomas y consecuencias, con aritmetica exacta (== sobre fracciones, sin tolerancias)
assert len(Omega) == 36 and prob(lambda w: True) == 1                       # axioma 2
assert all(prob(f) >= 0 for f in eventos.values())                          # axioma 1
assert prob(lambda w: not A(w)) == 1 - prob(A)                              # complemento
assert prob(lambda w: A(w) or C(w)) == prob(A) + prob(C)                    # axioma 3: A y C son excluyentes (7 es impar, los dobles suman par)
assert prob(lambda w: A(w) or B(w)) == prob(A) + prob(B) - prob(lambda w: A(w) and B(w))   # inclusion-exclusion
assert prob(lambda w: A(w) and B(w)) <= prob(B)                             # monotonia

# Interpretacion frecuentista: la frecuencia relativa de 'suma = 7' se estabiliza en 1/6
rng = np.random.default_rng(SEED)
N = 20_000
dados = rng.integers(1, 7, size=(5, N, 2))                                  # 5 trayectorias independientes de N lanzamientos de dos dados
frec = np.cumsum(dados.sum(axis=2) == 7, axis=1) / np.arange(1, N + 1)
fig, ax = plt.subplots(figsize=(9, 4.5))
for i in range(5):
    ax.plot(np.arange(1, N + 1), frec[i], lw=1.1, color=PALETA[i], alpha=0.85)
ax.axhline(1 / 6, color="black", ls="--", lw=1.3, label="P(A) = 1/6 (clásica)")
ax.set_xscale("log"); ax.set_ylim(0, 0.4)
ax.set_xlabel("número de lanzamientos n (escala log)"); ax.set_ylabel("frecuencia relativa de 'suma = 7'")
ax.set_title("Interpretación frecuentista: la frecuencia relativa converge a la probabilidad", fontweight="bold"); ax.legend()
plt.tight_layout(); plt.show()

print("Frecuencia relativa final de las 5 trayectorias:", np.round(frec[:, -1], 4), "| 1/6 =", round(1 / 6, 4))
print("Dispersion en n=10:", round(frec[:, 9].std(), 3), "| en n=20000:", round(frec[:, -1].std(), 4))
assert np.abs(frec[:, -1] - 1 / 6).max() < 0.01                             # error tipico ~ sqrt(p(1-p)/n) = 0.0026
assert frec[:, -1].std() < frec[:, 9].std()                                 # la variabilidad entre trayectorias se reduce con n
````

---
**Lectura.** La enumeración confirma las reglas **con igualdades exactas**: $P(A)=1/6$, $P(B)=11/36$, $P(A\cap B)=1/18$ y $P(A\cup B)=5/12 = 1/6 + 11/36 - 1/18$. Los eventos $A$ (suma 7) y $C$ (dobles) son **excluyentes** (7 es impar y los dobles suman par), por eso $P(A\cup C)=1/3$ es la suma simple. En el gráfico, con 10 lanzamientos las cinco trayectorias están muy dispersas (desviación estándar ≈ 0.117) y con 20 000 quedan entre 0.166 y 0.170 (desviación ≈ 0.002): **cerca** de $1/6$ pero **no iguales**. La interpretación frecuentista promete convergencia, no igualdad con $n$ finito; el error típico decrece como $\sqrt{p(1-p)/n}$ (≈ 0.0026 con $n=20\,000$), una idea que formalizamos en el cuaderno 4 con la ley de los grandes números.

## 2. Conteo: Permutaciones, Combinaciones y el Problema de los Cumpleaños 🎂

Con resultados **equiprobables**, calcular probabilidades es **contar**. Dos herramientas bastan:

- **Regla del producto:** si un proceso tiene $k$ etapas con $n_1, \dots, n_k$ opciones cada una, hay $n_1 n_2 \cdots n_k$ resultados.
- **Permutaciones** (el orden importa): ordenar $k$ de $n$ objetos distintos $\;P(n,k) = \dfrac{n!}{(n-k)!}$.
- **Combinaciones** (el orden no importa): elegir $k$ de $n$ $\;\binom{n}{k} = \dfrac{n!}{k!\,(n-k)!}$.

**Problema de los cumpleaños.** En un grupo de $n$ personas, suponiendo **365 fechas equiprobables e independientes** (una simplificación: los nacimientos reales no son uniformes, lo que *aumenta* las coincidencias), la probabilidad de que **al menos dos** cumplan el mismo día se calcula por el complemento:

$$P(\text{coincidencia}) = 1 - \frac{365 \cdot 364 \cdots (365 - n + 1)}{365^{\,n}} = 1 - \prod_{i=0}^{n-1} \frac{365 - i}{365}.$$

El resultado sorprende porque la intuición compara *mi* cumpleaños con los demás ($n$ comparaciones), pero el evento incluye **todas las parejas** posibles: $\binom{n}{2}$, que crece como $n^2$.

````python
# ============================================================
# 2. Conteo y problema de los cumpleaños: exacto vs simulacion
# ============================================================
# Conteo basico, contrastado con scipy.special
assert math.comb(45, 6) == int(special.comb(45, 6, exact=True)) == 8_145_060        # elegir 6 numeros de 45 (loteria ficticia)
assert math.perm(10, 3) == 720 == 10 * 9 * 8
assert math.comb(10, 3) * math.factorial(3) == math.perm(10, 3)                      # P(n,k) = C(n,k) k!
print("C(45,6) =", f"{math.comb(45, 6):,}", "-> probabilidad de acertar una combinacion al azar =", f"{1 / math.comb(45, 6):.2e}")


def p_cumpleanos(n, dias=365):
    """P(al menos dos coinciden) = 1 - prod (dias-i)/dias  (exacto, producto acumulado)."""
    if n > dias:
        return 1.0
    return 1.0 - float(np.prod((dias - np.arange(n)) / dias))


def p_cumpleanos_sim(n, M=20_000, dias=365, semilla=SEED):
    rng = np.random.default_rng(semilla)
    cumples = np.sort(rng.integers(0, dias, size=(M, n)), axis=1)
    return float((np.diff(cumples, axis=1) == 0).any(axis=1).mean())           # hay coincidencia si dos vecinos ordenados son iguales


ns = np.arange(1, 81)
exacta = np.array([p_cumpleanos(n) for n in ns])
puntos = {n: p_cumpleanos_sim(n) for n in (10, 23, 30, 50)}
print(pd.DataFrame({"n": list(puntos), "exacta": [round(p_cumpleanos(n), 4) for n in puntos], "simulacion (M=20000)": [round(v, 4) for v in puntos.values()]}).to_string(index=False))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(ns, exacta, color=PALETA[0], lw=2, label="exacta")
ax.scatter(list(puntos), list(puntos.values()), color=PALETA[1], zorder=3, s=45, label="simulación")
ax.axhline(0.5, color="gray", ls=":"); ax.axvline(23, color="gray", ls=":")
ax.set_xlabel("tamaño del grupo n"); ax.set_ylabel("P(al menos dos cumplen el mismo día)")
ax.set_title("Problema de los cumpleaños: con 23 personas ya es más probable que no", fontweight="bold"); ax.legend()
plt.tight_layout(); plt.show()

assert abs(p_cumpleanos(23) - 0.5073) < 5e-4 and p_cumpleanos(22) < 0.5 < p_cumpleanos(23)
assert all(abs(p_cumpleanos(n) - v) < 0.015 for n, v in puntos.items())
assert p_cumpleanos(366) == 1.0 and np.all(np.diff(exacta) > 0)                      # principio del palomar y monotonia
print("Parejas posibles con 23 personas:", math.comb(23, 2))
````

---
**Lectura.** El conteo da $\binom{45}{6}=8\,145\,060$ combinaciones (la probabilidad de acertar una al azar es ≈ $1.2\times 10^{-7}$). En los cumpleaños, la probabilidad exacta con **23 personas es 0.5073** (con 22 aún es menor que 0.5) y con 50 llega a 0.970; la simulación (0.509 y 0.969) coincide con el cálculo exacto dentro de su error de Monte Carlo. La sorpresa se explica contando **parejas**: 23 personas forman $\binom{23}{2}=253$ parejas, cada una con probabilidad $1/365$ de coincidir. Con fechas de nacimiento **no uniformes** la probabilidad real es *mayor*, así que el resultado es una cota prudente.

## 3. Probabilidad Condicional, Probabilidad Total y Teorema de Bayes 🔄

**Condicionar** es **recalcular la probabilidad cuando ya sabemos que ocurrió un evento**: el espacio muestral se *reduce* a $B$. Si $P(B) > 0$,

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}.$$

De la definición salen tres resultados centrales. Sea $B_1, \dots, B_k$ una **partición** de $\Omega$ (excluyentes y exhaustivos, todos con $P(B_i) > 0$):

$$\underbrace{P(A \cap B) = P(B)\,P(A \mid B)}_{\text{regla del producto}}, \qquad \underbrace{P(A) = \sum_{i=1}^{k} P(A \mid B_i)\,P(B_i)}_{\text{probabilidad total}}, \qquad \underbrace{P(B_j \mid A) = \frac{P(A \mid B_j)\,P(B_j)}{\sum_{i} P(A \mid B_i)\,P(B_i)}}_{\text{teorema de Bayes}}.$$

El teorema de Bayes **invierte el condicionamiento**: pasa de $P(\text{evidencia} \mid \text{hipótesis})$ a $P(\text{hipótesis} \mid \text{evidencia})$. Es la herramienta para **actualizar creencias** al ver datos: la probabilidad *a priori* $P(B_j)$ se convierte en la *a posteriori* $P(B_j \mid A)$.

> ⚠️ **Error clásico (la falacia del fiscal):** $P(A \mid B) \ne P(B \mid A)$ en general. «Si es culpable, la evidencia coincide con probabilidad altísima» **no** significa «si la evidencia coincide, es culpable con probabilidad altísima»: falta el *a priori*.

**Ejemplo 1: planta de empaque (cifras hipotéticas).** Tres líneas de producción empacan papa. Cada paquete proviene de una sola línea y puede salir defectuoso. **Datos inventados:**

| Línea | Participación $P(L_i)$ | Tasa de defectos $P(D \mid L_i)$ |
|---|---|---|
| A | 50 % | 2 % |
| B | 30 % | 3 % |
| C | 20 % | 6 % |

Si un paquete sale defectuoso, ¿de qué línea es más probable que venga? Aplicamos probabilidad total y Bayes, y **verificamos por simulación** (sorteando 1 000 000 de paquetes).

````python
# ============================================================
# 3a. Probabilidad total y Bayes en la planta de empaque (cifras hipoteticas) + verificacion por simulacion
# ============================================================
lineas = pd.DataFrame({"P(L)": [0.50, 0.30, 0.20], "P(D|L)": [0.02, 0.03, 0.06]}, index=["A", "B", "C"])
assert abs(lineas["P(L)"].sum() - 1) < 1e-12                                # es una particion
lineas["P(D y L)"] = lineas["P(L)"] * lineas["P(D|L)"]                       # regla del producto
p_def = lineas["P(D y L)"].sum()                                            # probabilidad total
lineas["P(L|D)"] = lineas["P(D y L)"] / p_def                               # Bayes
print(lineas.round(4))
print(f"\nP(defecto) = {p_def:.4f}  |  la linea C aporta {lineas.loc['C', 'P(L)']:.0%} de la produccion pero {lineas.loc['C', 'P(L|D)']:.1%} de los defectuosos")

rng = np.random.default_rng(SEED)
M = 1_000_000
linea = rng.choice(3, size=M, p=lineas["P(L)"].to_numpy())
defecto = rng.random(M) < lineas["P(D|L)"].to_numpy()[linea]
sim_def = defecto.mean()
sim_post = np.array([(linea[defecto] == k).mean() for k in range(3)])
print("Simulacion: P(D) =", round(sim_def, 4), "| P(L|D) =", np.round(sim_post, 4))

assert abs(p_def - 0.031) < 1e-12 and abs(lineas["P(L|D)"].sum() - 1) < 1e-12        # las posteriores suman 1
assert abs(sim_def - p_def) < 0.002 and np.abs(sim_post - lineas["P(L|D)"].to_numpy()).max() < 0.01
assert lineas["P(L|D)"].idxmax() == "C" and lineas.loc["C", "P(L|D)"] > lineas.loc["C", "P(L)"]   # C pasa de 20 % a mas

# Condicional por enumeracion: P(A|B) != P(B|A) con los dados
P_A_dado_B = prob(lambda w: A(w) and B(w)) / prob(B)                        # P(suma 7 | al menos un 6)
P_B_dado_A = prob(lambda w: A(w) and B(w)) / prob(A)                        # P(al menos un 6 | suma 7)
print(f"\nP(suma=7 | al menos un 6) = {P_A_dado_B}  |  P(al menos un 6 | suma=7) = {P_B_dado_A}")
assert P_A_dado_B == Fraction(2, 11) and P_B_dado_A == Fraction(1, 3) and P_A_dado_B != P_B_dado_A
````

---
**Lectura.** La línea **C** produce solo el 20 % de los paquetes pero aporta el **38.7 % de los defectuosos**: es la línea más probable dado un defecto, aunque **A** produzca más (su probabilidad pasó de 50 % *a priori* a 32.3 % *a posteriori*). La probabilidad total de defecto es $P(D)=0.031$; las posteriores (0.323, 0.290, 0.387) suman 1, como debe ser, y la simulación (0.318, 0.292, 0.391) las reproduce dentro del error de Monte Carlo. Con los dados, $P(\text{suma}=7\mid \text{al menos un }6)=2/11\approx 0.18$ pero $P(\text{al menos un }6\mid\text{suma}=7)=1/3$: **condicionar no es simétrico**.

**Ejemplo 2: la prueba de detección (cifras hipotéticas).** Una enfermedad (o una plaga en un cultivo) tiene **prevalencia** $\pi = P(E)$. Una prueba tiene **sensibilidad** $P(+ \mid E)$ y **especificidad** $P(- \mid E^c)$. Lo que le importa a quien recibe un positivo es el **valor predictivo positivo (VPP)**:

$$\text{VPP} = P(E \mid +) = \frac{\text{sens}\cdot \pi}{\text{sens}\cdot \pi + (1-\text{espec})\,(1-\pi)}.$$

Hay tres formas equivalentes de verlo: (1) la fórmula; (2) **frecuencias naturales** (imaginar 10 000 personas y contar cuántas caen en cada celda, mucho más intuitivo); y (3) en **odds**: $\text{odds}_{\text{posterior}} = \text{LR}_+ \times \text{odds}_{\text{previo}}$, con razón de verosimilitud $\text{LR}_+ = \dfrac{\text{sens}}{1-\text{espec}}$ y $\text{odds} = p/(1-p)$.

Usaremos **prevalencia 1 %, sensibilidad 90 %, especificidad 95 %** (hipotéticas), y veremos además qué pasa al **encadenar dos pruebas**. *Supuesto crítico:* las dos pruebas son **condicionalmente independientes dado el estado real** (los errores de una no predicen los de la otra); repetir exactamente la misma prueba con la misma causa de error **no** cumple esto.

````python
# ============================================================
# 3b. Teorema de Bayes en una prueba de deteccion: VPP, frecuencias naturales, odds y dos pruebas en serie
# ============================================================
def bayes_ppv(prev, sens, espec):
    """Valor predictivo positivo P(E|+) por el teorema de Bayes."""
    return sens * prev / (sens * prev + (1 - espec) * (1 - prev))


def bayes_vpn(prev, sens, espec):
    """Valor predictivo negativo P(no E|-)."""
    return espec * (1 - prev) / (espec * (1 - prev) + (1 - sens) * prev)


prev, sens, espec = 0.01, 0.90, 0.95
ppv = bayes_ppv(prev, sens, espec)

# (2) Frecuencias naturales: 10 000 personas
n = 10_000
enfermos = round(n * prev); sanos = n - enfermos
vp = round(enfermos * sens); fn = enfermos - vp; fp = round(sanos * (1 - espec)); vn = sanos - fp
print(pd.DataFrame({"prueba +": [vp, fp, vp + fp], "prueba -": [fn, vn, fn + vn], "total": [enfermos, sanos, n]}, index=["enfermos", "sanos", "total"]))
print(f"\nVPP = {ppv:.4f}  (frecuencias naturales: {vp}/{vp + fp} = {vp / (vp + fp):.4f}) | VPN = {bayes_vpn(prev, sens, espec):.5f}")

# (3) Odds: odds_posterior = LR+ x odds_previo
odds_prev = prev / (1 - prev); lr_pos = sens / (1 - espec); odds_post = lr_pos * odds_prev
assert abs(odds_post / (1 + odds_post) - ppv) < 1e-12 and abs(vp / (vp + fp) - ppv) < 1e-12

# Dos pruebas en serie (condicionalmente independientes): el posterior de la primera es el previo de la segunda
ppv2 = bayes_ppv(ppv, sens, espec)
print(f"Tras un segundo positivo (otra prueba independiente): P(E|++) = {ppv2:.4f}")
assert ppv2 > ppv > prev and abs(ppv2 - lr_pos**2 * odds_prev / (1 + lr_pos**2 * odds_prev)) < 1e-12

# Verificacion por simulacion
rng = np.random.default_rng(SEED)
M = 2_000_000
e = rng.random(M) < prev
pos1 = np.where(e, rng.random(M) < sens, rng.random(M) < 1 - espec)
pos2 = np.where(e, rng.random(M) < sens, rng.random(M) < 1 - espec)
print(f"Simulacion: P(E|+) = {e[pos1].mean():.4f} | P(E|++) = {e[pos1 & pos2].mean():.4f}")
assert abs(e[pos1].mean() - ppv) < 0.01 and abs(e[pos1 & pos2].mean() - ppv2) < 0.03

# El VPP depende de la prevalencia: misma prueba, distintas poblaciones
prevs = np.linspace(0.001, 0.5, 300)
fig, ax = plt.subplots(figsize=(9, 4.5))
for (s, sp), col in zip([(0.90, 0.95), (0.90, 0.99), (0.99, 0.99)], PALETA):
    ax.plot(prevs, bayes_ppv(prevs, s, sp), color=col, lw=2, label=f"sens {s:.0%} / espec {sp:.0%}")
ax.axvline(prev, color="gray", ls=":"); ax.scatter([prev], [ppv], color=PALETA[0], zorder=3)
ax.set_xlabel("prevalencia π"); ax.set_ylabel("VPP = P(E | +)")
ax.set_title("Valor predictivo positivo según la prevalencia: la rareza del evento pesa", fontweight="bold"); ax.legend()
plt.tight_layout(); plt.show()
assert bayes_ppv(0.5, 0.9, 0.95) > 0.9 > ppv and abs(ppv - 0.1538) < 5e-4
assert abs(bayes_ppv(prev, 0.90, 0.99) - 0.4762) < 5e-4 and abs(bayes_ppv(prev, 0.99, 0.99) - 0.5) < 1e-12   # con prevalencia baja manda la especificidad
````

---
**Lectura.** Con 10 000 personas hay 100 enfermas (90 positivos verdaderos) y 9 900 sanas (**495 falsos positivos**): de 585 positivos solo 90 están enfermos, o sea un **VPP de 15.4 %** pese a una sensibilidad de 90 % y una especificidad de 95 %. El **VPN es 99.89 %**: un negativo casi descarta la enfermedad. Un **segundo positivo** (bajo el supuesto de independencia condicional) eleva la probabilidad a 76.6 %; la simulación (0.153 y 0.767) coincide con Bayes. El gráfico muestra que, con prevalencia baja, **la especificidad manda**: pasar de 95 % a 99 % de especificidad lleva el VPP de 15 % a ≈ 48 %, mientras que subir la sensibilidad de 90 % a 99 % lo mueve apenas de 48 % a 50 %. Por eso las pruebas de tamizaje de eventos raros se **confirman** con una segunda prueba.

## 4. Independencia 🔗

Dos eventos son **independientes** si saber que uno ocurrió **no cambia** la probabilidad del otro:

$$A \perp B \iff P(A \cap B) = P(A)\,P(B) \iff P(A \mid B) = P(A) \quad (\text{si } P(B) > 0).$$

Tres matices que el libro y la práctica exigen distinguir:

1. **Independiente $\ne$ excluyente.** Si $A$ y $B$ son excluyentes con $P(A), P(B) > 0$, entonces $P(A \cap B) = 0 \ne P(A)P(B)$: son **muy dependientes** (saber que $A$ ocurrió garantiza que $B$ no).
2. **Por parejas $\ne$ mutua.** Para $n$ eventos, **independencia mutua** exige que la regla del producto valga para **toda subfamilia**; que valga solo por parejas **no basta**.
3. **Condicional.** $A \perp B \mid C$ significa $P(A \cap B \mid C) = P(A \mid C)\,P(B \mid C)$. Dos variables pueden ser dependientes en general pero independientes dado un tercer factor (o al revés).

**Por qué importa en ciencia de datos:** la independencia es el supuesto que **permite multiplicar probabilidades** (Naive Bayes, verosimilitud de una muestra aleatoria, fiabilidad de sistemas) y que fundamenta casi toda la inferencia. Cuando es falsa, los cálculos pueden fallar **por órdenes de magnitud**.

Verificamos con enumeración exacta (dados, monedas) y con un **sistema de componentes** cuya fiabilidad se calcula por independencia, para luego añadir una **causa común** (un apagón que tumba todo a la vez).

````python
# ============================================================
# 4. Independencia: dados, por parejas vs mutua, y fiabilidad de sistemas con causa comun
# ============================================================
def independientes(f, g, espacio=Omega):
    return prob(lambda w: f(w) and g(w), espacio) == prob(f, espacio) * prob(g, espacio)


par_d1 = lambda w: w[0] % 2 == 0                                             # el primer dado es par
suma7 = A                                                                    # suma = 7
suma8 = lambda w: w[0] + w[1] == 8
print("¿'primer dado par' y 'suma = 7' son independientes?", independientes(par_d1, suma7))
print("¿'primer dado par' y 'suma = 8' son independientes?", independientes(par_d1, suma8))
print("¿'suma = 7' y 'dobles' (excluyentes) son independientes?", independientes(suma7, C), "| P(A y C) =", prob(lambda w: suma7(w) and C(w)))
assert independientes(par_d1, suma7) and not independientes(par_d1, suma8)
assert not independientes(suma7, C) and prob(C) > 0 and prob(suma7) > 0       # excluyentes con P>0 => dependientes

# Por parejas vs mutua: dos monedas X, Y y Z = X xor Y
omega2 = [(x, y, x ^ y) for x, y in itertools.product((0, 1), repeat=2)]       # 4 resultados equiprobables
ev = {k: (lambda w, k=k: w[k] == 1) for k in range(3)}                       # X=1, Y=1, Z=1
pares = all(prob(lambda w, i=i, j=j: ev[i](w) and ev[j](w), omega2) == prob(ev[i], omega2) * prob(ev[j], omega2) for i, j in itertools.combinations(range(3), 2))
triple = prob(lambda w: all(ev[k](w) for k in range(3)), omega2)
print(f"\nIndependientes por parejas: {pares} | P(X=1, Y=1, Z=1) = {triple} frente a P(X=1)P(Y=1)P(Z=1) = {Fraction(1, 8)}")
assert pares and triple == 0 and triple != Fraction(1, 8)                    # NO son mutuamente independientes

# Fiabilidad de un bloque de 3 componentes en paralelo: independencia vs causa comun
p_ok = 0.95                                                                  # fiabilidad de cada componente (hipotetica)
R_serie, R_paralelo = p_ok**3, 1 - (1 - p_ok)**3
q = 0.02                                                                     # probabilidad de un apagon que inutiliza a TODOS a la vez (causa comun)
R_paralelo_cc = (1 - q) * R_paralelo
rng = np.random.default_rng(SEED)
M = 1_000_000
apagon = rng.random(M) < q
falla = rng.random((M, 3)) < (1 - p_ok)
sim_serie = (~falla.any(axis=1)).mean()
sim_par_ind = (~falla.all(axis=1)).mean()
sim_par_cc = ((~falla.all(axis=1)) & ~apagon).mean()
print(pd.DataFrame({"formula": [R_serie, R_paralelo, R_paralelo_cc], "simulacion": [sim_serie, sim_par_ind, sim_par_cc]},
                   index=["serie (3 en cadena)", "paralelo, independientes", "paralelo con causa comun"]).round(5))
print(f"Probabilidad de fallar: independientes {1 - R_paralelo:.6f} vs con causa comun {1 - R_paralelo_cc:.4f} ({(1 - R_paralelo_cc) / (1 - R_paralelo):.0f} veces mayor)")
assert abs(R_paralelo - 0.999875) < 1e-12 and abs(sim_serie - R_serie) < 0.002 and abs(sim_par_cc - R_paralelo_cc) < 0.002
assert (1 - R_paralelo_cc) / (1 - R_paralelo) > 100                          # la redundancia 'independiente' era una ilusion
````

---
**Lectura.** «Primer dado par» y «suma = 7» **son independientes** (la suma 7 tiene probabilidad $1/6$ sea cual sea el primer dado), pero con «suma = 8» **no**: $P(\text{par}\cap\text{suma }8)=3/36 \ne 1/2\cdot 5/36$. Los excluyentes $A$ y $C$ cumplen $P(A\cap C)=0\neq 1/36$: son **dependientes**. Con dos monedas $X$, $Y$ y $Z = X \oplus Y$, las tres variables son independientes **por parejas**, pero $P(X=Y=Z=1)=0\neq 1/8$: conocer dos determina la tercera, así que **no** son mutuamente independientes. En el sistema de tres componentes en paralelo, la fiabilidad por independencia es 0.99988 (falla con probabilidad 0.000125); con un apagón que afecta a todos con probabilidad 2 %, la fiabilidad real baja a 0.9799 y la probabilidad de falla es **161 veces mayor**. Las simulaciones (0.99988 y 0.97994) coinciden con las fórmulas. **La redundancia tiene un techo: la causa común.**

## 5. Monty Hall: el Mecanismo que Genera la Información Importa 🚪🚪🚪

Hay tres puertas: tras una hay un auto y tras las otras dos, cabras. Eliges la puerta 1. El presentador, que **sabe** dónde está el auto, abre una de las otras dos puertas **siempre con una cabra** y te ofrece **cambiar**. ¿Conviene? La intuición dice «50-50»; el cálculo exacto dice otra cosa.

Lo decisivo es **cómo** el presentador escoge la puerta. Modelamos dos mecanismos:

- **Presentador informado:** nunca abre la puerta con el auto (si hay dos cabras, elige una al azar).
- **Presentador ignorante:** abre al azar una de las otras dos; si por casualidad revela el auto, ese caso **se descarta** (condicionamos a «salió una cabra»).

Es el mismo dato observado (una cabra tras la puerta abierta), pero con **mecanismos distintos** el condicionamiento da **resultados distintos**. Esto es exactamente lo que ocurre con datos reales: *¿cómo llegó este dato a mi base?* (muestreo, filtros, supervivencia) cambia lo que se puede concluir. Calculamos por **enumeración exacta** y verificamos por **simulación**.

````python
# ============================================================
# 5. Monty Hall: enumeracion exacta con Fraction y simulacion, para dos mecanismos del presentador
# ============================================================
def monty_exacto(host_sabe=True):
    """P(ganar | cambiar) con el jugador en la puerta 0; el presentador abre la puerta 1 o 2 (se condiciona a que sea cabra)."""
    gana, total = Fraction(0), Fraction(0)
    for auto in range(3):
        for abre in (1, 2):
            if host_sabe:
                posibles = [d for d in (1, 2) if d != auto]
                p_abre = Fraction(1, len(posibles)) if abre in posibles else Fraction(0)
            else:
                p_abre = Fraction(1, 2)
            p = Fraction(1, 3) * p_abre
            if abre == auto:                                                  # el ignorante revelo el auto: caso descartado
                continue
            total += p
            if auto == 3 - abre:                                              # la otra puerta cerrada (1 <-> 2) tiene el auto
                gana += p
    return gana / total


def monty_sim(host_sabe=True, M=400_000, semilla=SEED):
    rng = np.random.default_rng(semilla)
    auto = rng.integers(0, 3, M)
    if host_sabe:
        abre = np.where(auto == 0, rng.integers(1, 3, M), 3 - auto)           # si el auto esta en 0 elige 1 o 2; si no, abre la otra cabra
    else:
        abre = rng.integers(1, 3, M)
    valido = abre != auto                                                     # condicionamos a 'salio una cabra'
    gana_cambiar = (auto == 3 - abre)[valido].mean()
    gana_quedarse = (auto == 0)[valido].mean()
    return gana_cambiar, gana_quedarse, valido.mean()


res = {}
for nombre, sabe in (("informado", True), ("ignorante", False)):
    exacto = monty_exacto(sabe)
    sim_c, sim_q, frac_valida = monty_sim(sabe)
    res[nombre] = (exacto, sim_c, sim_q, frac_valida)
print(pd.DataFrame({k: {"P(ganar | cambiar) exacta": str(v[0]), "simulacion cambiar": round(v[1], 4), "simulacion quedarse": round(v[2], 4), "casos validos": round(v[3], 3)} for k, v in res.items()}))

assert res["informado"][0] == Fraction(2, 3) and res["ignorante"][0] == Fraction(1, 2)
assert abs(res["informado"][1] - 2 / 3) < 0.005 and abs(res["informado"][2] - 1 / 3) < 0.005
assert abs(res["ignorante"][1] - 0.5) < 0.005 and abs(res["ignorante"][3] - 2 / 3) < 0.005
````

---
**Lectura.** Con el presentador **informado**, cambiar gana **2/3** y quedarse 1/3: la puerta inicial tenía 1/3 y el presentador **nunca iba a abrirla**, de modo que su gesto no le aporta información; los 2/3 restantes se concentran en la otra puerta. Con el **ignorante**, cambiar y quedarse empatan en **1/2**, y solo 2/3 de las partidas son «válidas» (en 1/3 revela el auto y se descartan). Ambas situaciones muestran lo mismo que se observa (una cabra), pero el **mecanismo** que produjo el dato cambia la conclusión. En ciencia de datos esto es el **sesgo de selección y de supervivencia**: los datos que llegan a tu base pasaron por un filtro, y condicionar sin modelar ese filtro conduce a errores. Las simulaciones (0.6665 y 0.4999) confirman los cálculos exactos.

| Concepto | Mensaje Central |
|---|---|
| **Axiomas** | $P(A)\ge 0$, $P(\Omega)=1$ y aditividad para excluyentes; de ahí salen el complemento, la monotonía y la **inclusión-exclusión** (verificadas con `Fraction`, sin redondeo) |
| **Interpretaciones** | Clásica (conteo de equiprobables), frecuentista (límite de la frecuencia relativa: la variabilidad entre trayectorias baja con $n$) y subjetiva; los axiomas son los mismos |
| **Conteo** | $P(n,k)=n!/(n-k)!$ y $\binom{n}{k}$; en los **cumpleaños** bastan 23 personas para superar el 50 % porque hay $\binom{23}{2}=253$ parejas |
| **Condicional y total** | $P(A\mid B)=P(A\cap B)/P(B)$; regla del producto y probabilidad total sobre una partición; $P(A\mid B)\neq P(B\mid A)$ |
| **Bayes** | Posterior $\propto$ verosimilitud $\times$ *a priori*; en *odds*, $\text{LR}\times\text{odds}$; con prevalencia 1 %, sens 90 % y espec 95 % el **VPP es solo ≈ 15 %** y un segundo positivo independiente lo sube a ≈ 77 % |
| **Independencia** | $P(A\cap B)=P(A)P(B)$; **no** es lo mismo que excluyente; **por parejas $\neq$ mutua** (contraejemplo X, Y, X xor Y); una **causa común** multiplica la probabilidad de falla de un sistema «redundante» por más de 100 |
| **Mecanismo de datos** | Monty Hall: cambiar gana 2/3 con presentador informado, pero **1/2** con el ignorante: el cálculo correcto depende de **cómo** llegó el dato |
