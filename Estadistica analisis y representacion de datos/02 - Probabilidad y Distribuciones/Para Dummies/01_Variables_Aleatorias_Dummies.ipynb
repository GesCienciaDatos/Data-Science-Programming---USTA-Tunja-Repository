## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **«para no matemáticos»** del [cuaderno 01 del Módulo 02 (Variables aleatorias, esperanza y varianza)](../01_Variables_Aleatorias_Esperanza_y_Varianza.ipynb). Una **variable aleatoria** es, simplemente, **un número que depende del azar**: lo que sale al lanzar un dado, lo que ganas en una rifa, los litros de leche que da una vaca mañana.

| Lo que decimos | Lo que significa en la vida diaria |
|---|---|
| **Promedio a la larga** (valor esperado) | Lo que ganarías **en promedio** si repites el juego muchísimas veces |
| **Variación** | Qué tan lejos **suelen quedar** los resultados del promedio |
| **Relación entre dos cosas** | Si una sube cuando la otra sube (horas de sol y cosecha) |

Al terminar podrás explicar:
1. Por qué un juego puede **perder dinero en promedio** aunque alguien se saque el premio.
2. Por qué **dos fincas con el mismo promedio** pueden ser muy distintas.
3. Por qué un número que mide la relación entre dos cosas puede decir «**cero**» y aun así haber una relación.

## 1. La Rifa de la Vereda: el Promedio a la Larga 🎟️

En la vereda rifan una ternera: se venden **100 boletas** de **\$5 000** y el premio vale **\$250 000** (cifras **inventadas**). Si compras una boleta, hay **1 posibilidad en 100** de ganar (te quedan \$245 000, porque ya pagaste la boleta) y **99 en 100** de perder los \$5 000. ¿Cuánto ganas o pierdes **en promedio**? Lo calculamos con la tabla y lo comprobamos **comprando muchas boletas** en el computador.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)                      # semilla fija: siempre salen los mismos numeros
plt.rcParams.update({"axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})

# La tabla de la rifa: cada resultado con su probabilidad
resultados = np.array([245_000, -5_000])           # gano el premio (descontando la boleta) / pierdo la boleta
probabilidades = np.array([1 / 100, 99 / 100])
promedio = (resultados * probabilidades).sum()
print(f"Promedio por boleta = 245 000 x 0.01 + (-5 000) x 0.99 = {promedio:,.0f} pesos")

# Comprar 100 000 boletas de verdad (simuladas): ¿hacia donde va el promedio?
boletas = rng.choice(resultados, size=100_000, p=probabilidades)
acumulado = np.cumsum(boletas) / np.arange(1, 100_001)
print(f"Promedio tras 10 boletas: {acumulado[9]:,.0f} | tras 1 000: {acumulado[999]:,.0f} | tras 100 000: {acumulado[-1]:,.0f}")

plt.figure(figsize=(9, 3.6))
plt.semilogx(np.arange(1, 100_001), acumulado, color="#2563eb")
plt.axhline(promedio, color="black", ls="--", label="promedio de la tabla")
plt.xlabel("boletas compradas (escala logaritmica)"); plt.ylabel("ganancia promedio por boleta"); plt.title("A la larga, la rifa le cuesta plata al comprador"); plt.legend()
plt.tight_layout(); plt.show()

# Un dado: el promedio (3.5) es un numero que el dado NUNCA muestra
promedio_dado = np.mean([1, 2, 3, 4, 5, 6])
assert abs(promedio - (-2500)) < 1e-9
assert abs(acumulado[-1] - promedio) < 600           # el promedio de muchas boletas queda cerca del de la tabla
assert promedio_dado == 3.5
````

---
**Lección:** la tabla dice que, **en promedio**, cada boleta te cuesta **\$2 500**; después de 10 boletas casi seguro **no has ganado nada** y tu promedio es −5 000; con 100 000 boletas el promedio es −2 382, ya cerca del de la tabla. Ojo: el promedio **no tiene que ser un resultado posible** (un dado nunca sale 3.5). Sirve para **decidir si un juego conviene a la larga**, no para saber qué pasará en un solo intento.

## 2. Mismo Promedio, Distinta Variación: Dos Fincas Lecheras 🐄

La **finca A** y la **finca B** dan, **en promedio, 20 litros por vaca al día**. Pero la A es muy **pareja** y la B es muy **irregular** (datos **inventados**). Si cada vaca debe dar al menos 15 litros para que el negocio sea rentable, ¿importa que tengan el mismo promedio?

````python
litros_a = rng.normal(20, 1, 1000)            # finca A: casi todas las vacas cerca de 20
litros_b = rng.normal(20, 6, 1000)            # finca B: unas dan 8 y otras 32

resumen = pd.DataFrame({"promedio": [litros_a.mean(), litros_b.mean()],
                        "desviacion (que tan lejos suelen quedar)": [litros_a.std(), litros_b.std()],
                        "% de vacas con menos de 15 litros": [(litros_a < 15).mean() * 100, (litros_b < 15).mean() * 100]}, index=["Finca A", "Finca B"])
print(resumen.round(1))

fig, ejes = plt.subplots(1, 2, figsize=(10, 3.4), sharex=True, sharey=True)
for a, datos, nombre, c in zip(ejes, (litros_a, litros_b), ("Finca A (pareja)", "Finca B (irregular)"), ("#16a34a", "#dc2626")):
    a.hist(datos, bins=30, color=c, alpha=0.7); a.axvline(15, color="black", ls=":", label="minimo rentable (15 L)"); a.set_title(nombre); a.set_xlabel("litros por vaca")
ejes[0].legend(); plt.tight_layout(); plt.show()

assert abs(litros_a.mean() - 20) < 0.5 and abs(litros_b.mean() - 20) < 0.8        # promedios casi iguales
assert abs(litros_a.std() - 1) < 0.2 and abs(litros_b.std() - 6) < 0.5
assert (litros_a < 15).mean() < 0.001 and 0.15 < (litros_b < 15).mean() < 0.25
````

---
**Lección:** las dos fincas tienen el mismo promedio (≈ 20 L), pero en la B **22 de cada 100 vacas** (cerca de 1 de cada 5) dan menos de 15 litros y en la A **ninguna**. El promedio solo cuenta una parte de la historia: hay que mirar también **cuánto varían** los datos. Por eso en estadística siempre se reporta el **centro** *y* la **dispersión**.

## 3. Cuando Dos Cosas Suben Juntas (y Cuando el Número No Ve la Relación) ☀️

Un número entre **−1 y 1** resume si dos cosas **suben juntas** (cerca de 1), **una sube cuando la otra baja** (cerca de −1) o **no se acompañan** (cerca de 0). Probamos con dos casos **inventados**: las **horas de sol** y la **cosecha** de una parcela (suben juntas), y la **cantidad de abono** y la cosecha, donde **poco** y **demasiado** abono dañan la planta (una relación curva).

````python
sol = rng.uniform(3, 9, 200)                                          # horas de sol al dia
cosecha = 10 * sol + rng.normal(0, 8, 200)                            # mas sol, mas cosecha (con ruido)
abono = rng.uniform(-2, 2, 200)                                       # 0 = abono ideal; negativo = poco, positivo = demasiado
dano = abono ** 2 + rng.normal(0, 0.2, 200)                           # el 'dano' crece si te alejas del punto ideal

r_sol = np.corrcoef(sol, cosecha)[0, 1]
r_abono = np.corrcoef(abono, dano)[0, 1]
print(f"Numero para sol y cosecha: {r_sol:.2f} (suben juntas) | numero para abono y dano: {r_abono:.2f} (¡cerca de cero!)")

fig, ejes = plt.subplots(1, 2, figsize=(10, 3.6))
ejes[0].scatter(sol, cosecha, s=12, color="#16a34a"); ejes[0].set(title=f"Sol y cosecha: número = {r_sol:.2f}", xlabel="horas de sol", ylabel="cosecha")
ejes[1].scatter(abono, dano, s=12, color="#dc2626"); ejes[1].set(title=f"Abono y daño: número = {r_abono:.2f}", xlabel="abono (0 = ideal)", ylabel="daño a la planta")
plt.tight_layout(); plt.show()

assert r_sol > 0.7 and abs(r_abono) < 0.15
````

## Resumen en una frase 🎯

> **Una variable aleatoria es un número que depende del azar: su promedio a la larga dice si un juego conviene, su variación dice cuánto riesgo hay, y el número que mide la relación entre dos cosas solo ve las relaciones en línea recta.**

### Lo que aprendiste hoy:
- ✅ **Promedio a la larga:** la rifa le cuesta \$2 500 por boleta al comprador, aunque alguien gane; con 100 000 boletas el promedio se acerca al de la tabla.
- ✅ **Variación:** las fincas A y B promedian 20 litros, pero en la B cerca de 1 de cada 5 vacas da menos de 15.
- ✅ **Relación:** sol y cosecha suben juntos (número alto); abono y daño tienen relación curva pero el número da casi cero.
