| Lo que decimos | Lo que significa en la vida diaria |
|---|---|
| **Probabilidad** | Cuántas veces de cada 100 pasa algo (si llueve 30 de cada 100 tardes, la probabilidad es 30 %) |
| **Condicional** | La probabilidad **cuando ya sabes algo más** (llueve más si el cielo está negro) |
| **Falsa alarma** | La alarma suena, pero no había nada |
| **Independientes** | Dos cosas que **no se afectan** entre sí (lo que sale en un dado no cambia el otro) |

## 1. La Probabilidad es una Fracción: Contar y Luego Simular 🎲

Lanzas **dos dados**. ¿Qué tan probable es que la suma sea 7? Primero **cuenta** todos los resultados posibles (36) y cuántos suman 7. Después **simula**: lanza los dados miles de veces con el computador y mira qué fracción de las veces salió 7. Si el conteo es correcto, la simulación se acerca a él.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)                      # semilla fija: siempre salen los mismos numeros
plt.rcParams.update({"axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})

# 1. Contar: todas las parejas posibles de dos dados
casos = [(a, b) for a in range(1, 7) for b in range(1, 7)]
suman_7 = [c for c in casos if sum(c) == 7]
prob_contada = len(suman_7) / len(casos)
print(f"Casos posibles: {len(casos)} | casos que suman 7: {len(suman_7)} -> probabilidad = {len(suman_7)}/{len(casos)} = {prob_contada:.3f}")

# 2. Simular: lanzar los dos dados 20 000 veces y mirar la fraccion acumulada de 'suma 7'
lanzamientos = rng.integers(1, 7, size=(20_000, 2)).sum(axis=1) == 7
fraccion = np.cumsum(lanzamientos) / np.arange(1, 20_001)
print(f"Despues de 10 lanzamientos: {fraccion[9]:.2f} | de 100: {fraccion[99]:.3f} | de 20 000: {fraccion[-1]:.4f}")

plt.figure(figsize=(9, 3.6))
plt.semilogx(np.arange(1, 20_001), fraccion, color="#2563eb")
plt.axhline(prob_contada, color="black", ls="--", label="lo que dice el conteo (1/6)")
plt.xlabel("lanzamientos (escala logaritmica)"); plt.ylabel("fraccion de veces que salio 7"); plt.title("Con pocos lanzamientos hay saltos; con muchos, se acerca al conteo")
plt.legend(); plt.tight_layout(); plt.show()

assert len(casos) == 36 and len(suman_7) == 6
assert abs(fraccion[-1] - prob_contada) < 0.02          # con muchos lanzamientos, la simulacion se parece al conteo
````

---
**Lección:** con **10 lanzamientos** la fracción de sietes salta mucho (en la corrida, 0.10 tras 10 lanzamientos y 0.170 tras 100); con **20 000** queda en 0.166, muy cerca de **1/6 ≈ 0.167**. Eso es la probabilidad en la práctica: *no promete lo que pasa en un lanzamiento, sino lo que pasa a la larga*. Un solo día de lluvia no «confirma» ni «desmiente» que la probabilidad sea 30 %.

## 2. El Detector de Plaga que se Equivoca (Casi Siempre que Suena) 🚨

Un técnico del ICA ficticio usa un **detector de plaga** en matas de papa. Los números son **inventados**: de cada 1 000 matas, **2 de cada 100** están enfermas; el detector **acierta en el 90 %** de las enfermas y da **falsa alarma en el 10 %** de las sanas. Cuando suena la alarma, ¿qué tan probable es que la mata esté enferma?

La clave es **contar con personas reales** (matas), no con porcentajes sueltos: se arma una tabla de 1 000 matas.

````python
def contar_alarmas(total, pct_enfermas, pct_detecta, pct_falsa):
    """Cuenta cuantas matas hay en cada caso (numeros redondeados a matas enteras)."""
    enfermas = round(total * pct_enfermas / 100)
    sanas = total - enfermas
    return {"enfermas": enfermas, "sanas": sanas,
            "alarmas_ciertas": round(enfermas * pct_detecta / 100),      # enfermas que el detector marca
            "alarmas_falsas": round(sanas * pct_falsa / 100)}              # sanas que el detector marca por error


res = contar_alarmas(1000, 2, 90, 10)
suenan = res["alarmas_ciertas"] + res["alarmas_falsas"]
print(pd.DataFrame({"matas": [res["enfermas"], res["sanas"]], "suena la alarma": [res["alarmas_ciertas"], res["alarmas_falsas"]]}, index=["enfermas", "sanas"]))
print(f"\nSuena la alarma en {suenan} matas, pero solo {res['alarmas_ciertas']} estan enfermas -> {res['alarmas_ciertas'] / suenan:.0%} de las alarmas son ciertas.")

# Misma maquina, pero en un campo donde la plaga es MUY comun (20 de cada 100)
res2 = contar_alarmas(1000, 20, 90, 10)
suenan2 = res2["alarmas_ciertas"] + res2["alarmas_falsas"]
print(f"Si 20 de cada 100 estan enfermas: suenan {suenan2} alarmas y {res2['alarmas_ciertas']} son ciertas -> {res2['alarmas_ciertas'] / suenan2:.0%}")

fig, ejes = plt.subplots(1, 2, figsize=(10, 3.6), sharey=True)
for a, r, t in zip(ejes, (res, res2), ("Plaga poco común (2 %)", "Plaga común (20 %)")):
    a.bar(["alarmas ciertas", "falsas alarmas"], [r["alarmas_ciertas"], r["alarmas_falsas"]], color=["#16a34a", "#dc2626"]); a.set_title(t)
ejes[0].set_ylabel("matas en las que suena la alarma"); plt.tight_layout(); plt.show()

assert suenan == 116 and res["alarmas_ciertas"] == 18
assert res["alarmas_ciertas"] / suenan < 0.16 and res2["alarmas_ciertas"] / suenan2 > 0.65
````

---
**Lección:** de **116 alarmas**, solo **18 son ciertas** (≈ 16 %): el detector acierta el 90 % de las veces cuando hay plaga, pero **como hay pocas matas enfermas**, las falsas alarmas de las muchas matas sanas **son más** que los aciertos. Cuando la plaga es común (20 %), la misma máquina acierta en ≈ 69 % de sus alarmas. **Moraleja:** antes de creerle a una alarma, pregunta **qué tan común es lo que busca**. Por eso un resultado positivo de una prueba médica casi siempre se **confirma** con una segunda prueba.

## 3. Dos Cosas que Parecen Independientes... Hasta que Llueve 🚌

Dos buses de rutas distintas llegan tarde cada uno el **20 % de los días**. Si no tienen nada que ver entre sí (son **independientes**), la probabilidad de que **los dos** lleguen tarde es 20 % de 20 % = **4 %**. Pero en Tunja, cuando **llueve** los dos se atrasan por el mismo motivo. Simulamos 100 000 días con (a) buses realmente independientes y (b) buses que **comparten la lluvia**.

````python
dias = 100_000
# (a) Independientes: cada bus tarde con 20 % de probabilidad, sin relacion con el otro
tarde_a, tarde_b = rng.random(dias) < 0.20, rng.random(dias) < 0.20
ambos_indep = (tarde_a & tarde_b).mean()

# (b) Comparten la lluvia: llueve el 30 % de los dias; con lluvia cada bus se atrasa el 50 %, sin lluvia el 8 %
llueve = rng.random(dias) < 0.30
p_tarde = np.where(llueve, 0.50, 0.08)
tarde_c, tarde_d = rng.random(dias) < p_tarde, rng.random(dias) < p_tarde
ambos_lluvia = (tarde_c & tarde_d).mean()
cada_uno = tarde_c.mean()

print(f"(a) Independientes: cada bus tarde {tarde_a.mean():.1%} de los dias | los dos tarde {ambos_indep:.1%} (esperado 4 %)")
print(f"(b) Con lluvia compartida: cada bus tarde {cada_uno:.1%} | los dos tarde {ambos_lluvia:.1%} (si fueran independientes serian {cada_uno ** 2:.1%})")

plt.figure(figsize=(7, 3.4))
plt.bar(["independientes", "comparten la lluvia"], [ambos_indep * 100, ambos_lluvia * 100], color=["#2563eb", "#dc2626"])
plt.ylabel("% de dias con LOS DOS buses tarde"); plt.title("Una causa en comun hace que fallen juntos"); plt.tight_layout(); plt.show()

assert abs(ambos_indep - 0.04) < 0.006
assert ambos_lluvia > 1.5 * cada_uno ** 2                  # fallan juntos bastante mas de lo que daria la cuenta 'independiente'
````

## Resumen en una frase 🎯

> **La probabilidad cuenta cuántas veces de cada cien pasa algo (y la simulación se acerca al conteo cuando repites mucho); una alarma casi perfecta se equivoca a menudo si lo que busca es raro; y dos cosas solo son independientes si no comparten ninguna causa.**

### Lo que aprendiste hoy:
- ✅ **Contar y simular:** 6 de 36 parejas suman 7 (≈ 0.167); con 20 000 lanzamientos la simulación queda a menos de 0.02 del conteo.
- ✅ **Contar con personas:** de 116 alarmas, solo 18 eran ciertas; con la plaga más común, la misma máquina acierta mucho más.
- ✅ **Causa en común:** si dos cosas comparten la lluvia, fallan juntas más veces de lo que dice la cuenta «independiente».
