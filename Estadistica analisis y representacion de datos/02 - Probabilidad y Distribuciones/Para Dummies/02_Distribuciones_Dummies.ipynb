Muchos azares de la vida real tienen **la misma forma**, como los **moldes** de una panadería: cambia el tamaño, pero la figura se repite. A esas figuras las llamamos **distribuciones**.

| El molde | Dónde lo ves | Qué cuenta o mide |
|---|---|---|
| **Éxitos de varios intentos** (binomial) | semillas que germinan | cuántas de 10 semillas brotan |
| **Eventos sueltos por hora** (Poisson) | buses que pasan, llamadas | cuántos pasan en una hora |
| **La espera** (exponencial) | tiempo hasta el próximo bus | cuántos minutos faltan |
| **La campana** (normal) | peso de las papas, estatura | casi todos cerca del promedio |

## 1. Sembrar 10 Semillas: Cuántas Brotan 🌱

Un agricultor siembra **10 semillas** de papa y cada una brota con un **80 % de probabilidad**, sin afectar a las demás (datos **inventados**). ¿Cuántas brotarán? Es distinto cada vez, así que **repetimos la siembra 10 000 veces** en el computador y miramos qué tan seguido sale cada cantidad.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)                      # semilla fija: siempre salen los mismos numeros
plt.rcParams.update({"axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})

siembras = rng.binomial(n=10, p=0.80, size=10_000)                 # 10 000 siembras de 10 semillas; cada una cuenta cuantas brotaron
frecuencia = np.bincount(siembras, minlength=11) / len(siembras) * 100

print(pd.DataFrame({"semillas que brotan": range(11), "% de las siembras": frecuencia.round(1)}).query("`% de las siembras` > 0").to_string(index=False))
print(f"\nLo mas comun: {frecuencia.argmax()} semillas | 9 o mas: {(siembras >= 9).mean():.0%} de las veces | 5 o menos: {(siembras <= 5).mean():.1%}")

plt.figure(figsize=(8, 3.6))
plt.bar(range(11), frecuencia, color=["#16a34a" if k == frecuencia.argmax() else "#94a3b8" for k in range(11)])
plt.xlabel("semillas que brotaron (de 10)"); plt.ylabel("% de las siembras"); plt.title("Sembrar 10 semillas con 80 % de brote: lo normal es 8"); plt.tight_layout(); plt.show()

assert frecuencia.argmax() == 8
assert abs((siembras >= 9).mean() - 0.376) < 0.02 and (siembras <= 5).mean() < 0.05
````

---
**Lección:** lo más común es que broten **8** (cerca de 3 de cada 10 siembras), pero **9 o más** pasa casi 4 de cada 10 veces y que broten **5 o menos** es muy raro (3 de cada 100). Esa figura —un montículo con su pico cerca del 80 %— es la forma típica de «**cuántos éxitos en varios intentos**». Si cambias el 80 % o el número de semillas, cambia el tamaño y la posición del montículo, pero la figura sigue siendo la misma familia.

## 2. Buses por Hora y el Bus que «No se Acerca» 🚌

Por un paradero pasan **en promedio 4 buses por hora**, de forma desordenada (cualquier momento, sin horario), y eso equivale a **un bus cada 15 minutos en promedio** (datos **inventados**). Simulamos 10 000 horas para ver **cuántos buses** pasan en cada una, y 100 000 esperas para ver **cuánto se tarda el próximo**.

````python
buses_por_hora = rng.poisson(lam=4, size=10_000)                     # cuantos buses pasan en cada una de 10 000 horas
esperas = rng.exponential(scale=15, size=100_000)                    # minutos hasta el proximo bus

print(f"Promedio de buses por hora: {buses_por_hora.mean():.2f} | horas SIN ningun bus: {(buses_por_hora == 0).mean():.1%}")
print(f"Espera promedio: {esperas.mean():.1f} min | esperas de mas de 30 minutos: {(esperas > 30).mean():.1%}")

# ¿El bus 'se acerca' mientras esperas? Miramos solo a quienes YA esperaron 10 minutos
ya_esperaron = esperas[esperas > 10] - 10                            # minutos que les faltaban despues de esos 10
print(f"Si ya esperaste 10 minutos, aun te faltan en promedio {ya_esperaron.mean():.1f} minutos (igual que al llegar!)")

fig, ejes = plt.subplots(1, 2, figsize=(10, 3.6))
ejes[0].bar(range(13), np.bincount(buses_por_hora, minlength=13)[:13] / len(buses_por_hora) * 100, color="#2563eb"); ejes[0].set(title="Buses que pasan en una hora", xlabel="buses", ylabel="% de las horas")
ejes[1].hist(esperas, bins=60, range=(0, 90), color="#f59e0b"); ejes[1].set(title="Minutos hasta el próximo bus", xlabel="minutos", ylabel="veces")
plt.tight_layout(); plt.show()

assert abs(buses_por_hora.mean() - 4) < 0.1 and abs((buses_por_hora == 0).mean() - 0.0183) < 0.006
assert abs(esperas.mean() - 15) < 0.5 and abs(ya_esperaron.mean() - 15) < 0.5
````

---
**Lección:** la cantidad de buses por hora se concentra alrededor de 4 (y solo en 1.9 % de las horas no pasa ningún bus); la espera, en cambio, tiene **muchas esperas cortas y unas pocas muy largas**. Y lo más curioso: quien **ya esperó 10 minutos** todavía espera, en promedio, **lo mismo que al llegar** (≈ 15). Cuando los buses pasan al azar, el bus **«no tiene memoria»**: haber esperado no lo acerca. (Con buses de **horario fijo** esto no ocurre, ahí sí vale la pena seguir esperando.)

## 3. La Campana: el Peso de las Papas 🥔

El peso de las papas de una cosecha suele tener forma de **campana**: la mayoría cerca del promedio y cada vez menos hacia los extremos. Con papas **inventadas** de **180 g de promedio** y una variación típica de **25 g**, contamos cuántas caen a menos de 1, 2 y 3 «pasos» de 25 g del promedio.

````python
pesos = rng.normal(180, 25, 100_000)                                  # peso de 100 000 papas (gramos)
prom, paso = 180, 25
dentro = {k: (np.abs(pesos - prom) < k * paso).mean() for k in (1, 2, 3)}
tabla = pd.DataFrame({"rango de pesos (g)": [f"{prom - k * paso} a {prom + k * paso}" for k in (1, 2, 3)],
                      "% de las papas dentro": [dentro[k] * 100 for k in (1, 2, 3)]}, index=["1 paso", "2 pasos", "3 pasos"])
print(tabla.round(1))
print(f"\nPapas de mas de 255 g (3 pasos arriba): {(pesos > 255).mean():.2%} | de menos de 105 g: {(pesos < 105).mean():.2%}")

plt.figure(figsize=(9, 3.6))
plt.hist(pesos, bins=70, color="#94a3b8")
for k, c in zip((1, 2, 3), ("#16a34a", "#f59e0b", "#dc2626")):
    plt.axvspan(prom - k * paso, prom + k * paso, color=c, alpha=0.12, label=f"{k} paso(s): {dentro[k]:.0%}")
plt.xlabel("peso (g)"); plt.ylabel("papas"); plt.title("La campana: casi todas las papas están cerca del promedio"); plt.legend(); plt.tight_layout(); plt.show()

assert abs(dentro[1] - 0.68) < 0.01 and abs(dentro[2] - 0.95) < 0.01 and dentro[3] > 0.99
````

## Resumen en una frase 🎯

> **Muchos azares tienen formas conocidas: contar éxitos de varios intentos da un montículo (semillas), contar eventos sueltos por hora da otro (buses), la espera tiene muchas cortas y pocas largas (y no tiene memoria), y la suma de muchos efectos pequeños da la campana.**

### Lo que aprendiste hoy:
- ✅ **Semillas:** de 10 con 80 % de brote, lo común es 8 y casi nunca 5 o menos.
- ✅ **Buses:** 4 por hora en promedio; el bus «no tiene memoria»: esperar 10 minutos no lo acerca.
- ✅ **Campana:** 68 %, 95 % y 99.7 % de los datos a 1, 2 y 3 pasos del promedio.
