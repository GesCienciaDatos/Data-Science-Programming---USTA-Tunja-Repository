Si la olla está bien revuelta, una cucharada sirve; pero **cuántas** cucharadas pruebes cambia **cuánta confianza** tienes.

| Lo que decimos | Lo que significa en la vida diaria |
|---|---|

## 1. Probar una Cucharada: Muestras Pequeñas y Grandes 🥄

Una bodega tiene **5 000 bultos de papa** (peso **inventado**; unos pocos bultos son muy pesados). Como tenemos toda la bodega en el computador, conocemos el **peso promedio verdadero**. Vamos a hacer como si no lo conociéramos: sacamos **8 muestras pequeñas** (5 bultos) y **8 muestras grandes** (100 bultos) y comparamos qué tan cerca queda cada promedio del verdadero.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)                      # semilla fija: siempre salen los mismos numeros
plt.rcParams.update({"axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})

bodega = rng.lognormal(mean=np.log(45), sigma=0.35, size=5000)       # peso (kg) de 5 000 bultos: la mayoria cerca de 45, algunos muy pesados
verdadero = bodega.mean()
print(f"Peso promedio verdadero de la bodega: {verdadero:.1f} kg")

peque = np.array([rng.choice(bodega, size=5).mean() for _ in range(8)])       # 8 cucharadas de 5 bultos
grande = np.array([rng.choice(bodega, size=100).mean() for _ in range(8)])    # 8 cucharadas de 100 bultos
print(pd.DataFrame({"muestras de 5 bultos": peque.round(1), "muestras de 100 bultos": grande.round(1)}, index=[f"muestra {i + 1}" for i in range(8)]))
print(f"\nLas de 5 bultos van de {peque.min():.1f} a {peque.max():.1f} kg (abanico de {peque.max() - peque.min():.1f}) | las de 100, de {grande.min():.1f} a {grande.max():.1f} kg (abanico de {grande.max() - grande.min():.1f})")

plt.figure(figsize=(8, 3.6))
plt.axhline(verdadero, color="black", ls="--", label="verdadero (toda la bodega)")
plt.plot(np.arange(8), peque, "o", color="#dc2626", label="muestras de 5"); plt.plot(np.arange(8), grande, "s", color="#16a34a", label="muestras de 100")
plt.xlabel("numero de muestra"); plt.ylabel("peso promedio (kg)"); plt.title("Las muestras grandes caen mas cerca de la verdad"); plt.legend(); plt.tight_layout(); plt.show()

assert (grande.max() - grande.min()) < (peque.max() - peque.min())           # el abanico de las muestras grandes es mas angosto
assert np.abs(grande - verdadero).max() < np.abs(peque - verdadero).max()    # y su peor error es menor
````

---
**Lección:** con muestras de **5 bultos** los promedios se riegan en un abanico ancho (de 37.4 a 54.4 kg, con el verdadero en 47.5); con **100 bultos** quedan todos cerca (de 44.5 a 48.8 kg). Ninguna cucharada acierta *exactamente*, pero **la cucharada grande se equivoca menos**. Si repites la misma encuesta con otra muestra, no esperes el mismo número: espera un número **parecido**, y más parecido mientras más grande sea la muestra.

## 2. Los Promedios Forman una Campana 🔔

Los pesos de la bodega **no** forman una campana: son **torcidos** (hay bultos muy pesados pero ninguno que pese casi cero). Aun así, si sacas **muchas muestras** y miras **sus promedios**, estos sí forman una campana, cada vez más angosta. Lo comprobamos con 5 000 muestras de 1, 5, 30 y 200 bultos.

````python
tamanos = (1, 5, 30, 200)
promedios = {n: bodega[rng.integers(0, len(bodega), size=(5000, n))].mean(axis=1) for n in tamanos}     # 5 000 cucharadas de cada tamano
arriba = {n: (promedios[n] > verdadero).mean() for n in tamanos}
print(pd.DataFrame({"% de promedios por encima del verdadero": [arriba[n] * 100 for n in tamanos],
                    "abanico tipico (desviacion)": [promedios[n].std() for n in tamanos]}, index=[f"muestras de {n}" for n in tamanos]).round(1))

fig, ejes = plt.subplots(1, 4, figsize=(15, 3.4), sharex=True)
for a, n, c in zip(ejes, tamanos, ("#dc2626", "#f59e0b", "#2563eb", "#16a34a")):
    a.hist(promedios[n], bins=40, color=c, alpha=0.75); a.axvline(verdadero, color="black", ls="--"); a.set_title(f"promedios de {n} bulto(s)"); a.set_xlabel("kg")
ejes[0].set_ylabel("muestras"); plt.tight_layout(); plt.show()

assert arriba[1] < 0.45 and abs(arriba[200] - 0.5) < 0.04               # los datos individuales son torcidos; los promedios de 200 quedan parejos a ambos lados
assert promedios[200].std() < promedios[30].std() < promedios[5].std() < promedios[1].std()
````

---
**Lección:** con **1 bulto** (los datos tal cual) la distribución es torcida: solo el 44 % de los bultos pesa más que el promedio, porque unos pocos muy pesados lo jalan hacia arriba. Con **200 bultos**, los promedios quedan **repartidos casi por igual** a ambos lados (48.7 % por encima) y forman una **campana angosta**. Esa es la magia: **el promedio de muchas cosas torcidas se vuelve una campana**. Por eso casi todos los métodos para estimar y comparar promedios se apoyan en ella.

## 3. Más Muestra, Menos Error: ¿Cuánto Menos? 📉

Si cuadruplicas la muestra, ¿el error baja a la cuarta parte? No: **baja a la mitad**. Medimos el «error típico» (qué tan lejos suele quedar el promedio de la muestra del verdadero) para muestras de 25, 100, 400 y 1 600 bultos, y de paso vemos qué pasa al tirar una moneda pocas o muchas veces.

````python
tamanos = (25, 100, 400, 1600)
error_tipico = {n: bodega[rng.integers(0, len(bodega), size=(3000, n))].mean(axis=1).std() for n in tamanos}
tabla = pd.DataFrame({"error tipico (kg)": list(error_tipico.values())}, index=[f"muestra de {n}" for n in tamanos])
tabla["cuantas veces mas chico que el anterior"] = [np.nan] + [error_tipico[a] / error_tipico[b] for a, b in zip(tamanos[:-1], tamanos[1:])]
print(tabla.round(2))

plt.figure(figsize=(7, 3.6))
plt.bar([str(n) for n in tamanos], list(error_tipico.values()), color="#2563eb")
plt.xlabel("tamano de la muestra (bultos)"); plt.ylabel("error tipico (kg)"); plt.title("Cuadruplicar la muestra reduce el error a la mitad"); plt.tight_layout(); plt.show()

# Una moneda: pocas tiradas vs muchas
for veces in (10, 1000, 100_000):
    caras = (rng.random(veces) < 0.5).mean()
    print(f"{veces:>7} tiradas: {caras:.1%} de caras")

razones = [error_tipico[a] / error_tipico[b] for a, b in zip(tamanos[:-1], tamanos[1:])]
assert all(1.85 < r < 2.15 for r in razones)
assert abs((rng.random(100_000) < 0.5).mean() - 0.5) < 0.01
````

## Resumen en una frase 🎯

> **Probar una cucharada sirve si la olla está revuelta, pero cada cucharada da un resultado distinto: las muestras grandes se equivocan menos, los promedios de muchas muestras forman una campana aunque los datos sean torcidos, y para reducir el error a la mitad hay que cuadruplicar la muestra.**

### Lo que aprendiste hoy:
- ✅ **Muestras pequeñas vs grandes:** las de 5 bultos se riegan en un abanico ancho; las de 100 quedan cerca del peso verdadero.
- ✅ **Campana de promedios:** aunque los pesos son torcidos, los promedios de 200 bultos quedan repartidos a ambos lados.
- ✅ **Regla del cuádruple:** 4 veces más muestra = la mitad del error; y con una moneda, más tiradas = más cerca del 50 %.
