## 1. Métricas de Error y Bondad de Ajuste 📐

Con $n$ observaciones, valores ajustados $\hat y_i$ y media $\bar y$, la descomposición de la variabilidad de un modelo lineal **con intercepto** es

$$\underbrace{\sum_i(y_i-\bar y)^2}_{\text{SST}}=\underbrace{\sum_i(\hat y_i-\bar y)^2}_{\text{SSR}}+\underbrace{\sum_i(y_i-\hat y_i)^2}_{\text{SSE}},\qquad R^2=1-\frac{\text{SSE}}{\text{SST}}=\frac{\text{SSR}}{\text{SST}}.$$

| Métrica | Fórmula | Se lee como |
|---|---|---|
| $R^2$ | $1-\text{SSE}/\text{SST}$ | Fracción de la variabilidad de $y$ que explica el modelo (frente a predecir siempre $\bar y$) |
| $R^2$ ajustado | $1-(1-R^2)\dfrac{n-1}{n-p-1}$ | $R^2$ penalizado por el número $p$ de predictores |
| MSE | $\text{SSE}/n$ | Error cuadrático medio (unidades de $y$ al cuadrado) |
| RMSE | $\sqrt{\text{MSE}}$ | Error «típico» en **unidades de $y$**; penaliza más los errores grandes |
| MAE | $\frac1n\sum\lvert y_i-\hat y_i\rvert$ | Error absoluto medio; más robusto a atípicos |

**Tres advertencias.** (1) Con intercepto y **datos de entrenamiento**, $R^2\in[0,1]$ y **no disminuye** al añadir predictores, aunque sean ruido puro; el $R^2$ ajustado corrige ese optimismo **en promedio**, pero no refleja el costo de estimar coeficientes inútiles sobre las predicciones nuevas. (2) Calculado **fuera de muestra** con la media del conjunto de prueba, $R^2=1-\text{SSE}_{\text{prueba}}/\text{SST}_{\text{prueba}}$ **puede ser negativo**: significa que el modelo predice **peor que la media**. (3) En estadística, la estimación de $\sigma^2$ usa $\text{SSE}/(n-p-1)$ (`mse_resid` de `statsmodels`), mientras el MSE descriptivo usa $\text{SSE}/n$.

Verificamos las métricas con `sklearn` y `statsmodels` y medimos el efecto de añadir **variables de ruido** (120 observaciones de entrenamiento, 2 000 de prueba, promedio de 30 simulaciones): los predictores verdaderos son 3; el resto son ruido.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import (KFold, RepeatedKFold, LeaveOneOut, GroupKFold, TimeSeriesSplit, StratifiedKFold,
                                     cross_val_score, cross_val_predict, train_test_split)
from sklearn.pipeline import make_pipeline
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.ensemble import RandomForestRegressor
import statsmodels.api as sm

plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})


def generar_rendimiento(n=200, p_ruido=0, semilla=1):
    '''SINTETICO. y = 10 + 2 x1 - 1.5 x2 + 1 x3 + N(0, 3^2); las demas columnas son ruido puro.'''
    rng = np.random.default_rng(semilla)
    X = rng.normal(size=(n, 3 + p_ruido))
    y = 10 + 2.0 * X[:, 0] - 1.5 * X[:, 1] + 1.0 * X[:, 2] + rng.normal(0, 3, n)
    return X, y


def metricas(y, yhat, p=None):
    '''SSE, SST, R2, (R2 ajustado si se da p = num. de predictores), MSE, RMSE y MAE.'''
    n = len(y)
    sse, sst = np.sum((y - yhat) ** 2), np.sum((y - y.mean()) ** 2)
    out = {"SSE": sse, "SST": sst, "R2": 1 - sse / sst, "MSE": sse / n, "RMSE": np.sqrt(sse / n), "MAE": np.mean(np.abs(y - yhat))}
    if p is not None:
        out["R2_aj"] = 1 - (1 - out["R2"]) * (n - 1) / (n - p - 1)
    return out


X_ev, y_ev = generar_rendimiento(n=200, p_ruido=0)
mod = LinearRegression().fit(X_ev, y_ev)
m = metricas(y_ev, mod.predict(X_ev), p=3)
ols = sm.OLS(y_ev, sm.add_constant(X_ev)).fit()
print({k: round(v, 4) for k, v in m.items()})
assert np.isclose(m["R2"], r2_score(y_ev, mod.predict(X_ev))) and np.isclose(m["R2"], ols.rsquared) and np.isclose(m["R2_aj"], ols.rsquared_adj)
assert np.isclose(m["MSE"], mean_squared_error(y_ev, mod.predict(X_ev))) and np.isclose(m["MAE"], mean_absolute_error(y_ev, mod.predict(X_ev)))
assert np.isclose(m["SSE"] / (200 - 3 - 1), ols.mse_resid) and np.isclose(m["SST"], ols.centered_tss)       # sigma^2 usa n-p-1
assert np.isclose(m["SST"], np.sum((mod.predict(X_ev) - y_ev.mean()) ** 2) + m["SSE"])                       # SST = SSR + SSE

# --- Efecto de anadir variables de ruido (promedio de 30 simulaciones) ---
def curva_ruido(ps=(0, 5, 10, 20, 40, 60, 80), reps=30):
    filas = []
    for p_r in ps:
        r = {"entr": [], "aj": [], "prueba": []}
        for s in range(reps):
            Xa, ya = generar_rendimiento(120, p_r, semilla=100 + s)
            Xb, yb = generar_rendimiento(2000, p_r, semilla=5000 + s)
            mod = LinearRegression().fit(Xa, ya)
            mt = metricas(ya, mod.predict(Xa), p=3 + p_r)
            r["entr"].append(mt["R2"]); r["aj"].append(mt["R2_aj"]); r["prueba"].append(r2_score(yb, mod.predict(Xb)))
        filas.append({"ruido": p_r, "R2 entrenamiento": np.mean(r["entr"]), "R2 ajustado": np.mean(r["aj"]), "R2 prueba": np.mean(r["prueba"])})
    return pd.DataFrame(filas).set_index("ruido")


curva = curva_ruido()
print("\n" + curva.round(3).to_string())
fig, ax = plt.subplots(figsize=(7, 3.8))
for c, col in zip(curva.columns, ["#94a3b8", "#16a34a", "#dc2626"]):
    ax.plot(curva.index, curva[c], marker="o", color=col, label=c)
ax.axhline(0, color="k", lw=0.8); ax.set_xlabel("variables de ruido añadidas"); ax.set_ylabel("$R^2$ (promedio)"); ax.legend(fontsize=8)
ax.set_title("Más ruido: el $R^2$ de entrenamiento sube; el de prueba cae"); plt.tight_layout(); plt.show()

assert curva["R2 entrenamiento"].iloc[-1] > curva["R2 entrenamiento"].iloc[0] + 0.3 and curva["R2 prueba"].iloc[-1] < 0 < curva["R2 prueba"].iloc[0]
assert (curva["R2 ajustado"] - curva["R2 ajustado"].iloc[0]).abs().max() < 0.06      # el ajustado se queda plano: no "ve" el costo de estimar coeficientes inutiles
````

## 2. Criterios de Información: AIC y BIC 🧭

Penalizar la complejidad sin separar datos: con **log-verosimilitud** maximizada $\hat\ell$ y $k$ parámetros,

$$\text{AIC}=-2\hat\ell+2k,\qquad \text{BIC}=-2\hat\ell+k\ln n.$$

Menor es mejor. Para un modelo lineal gaussiano, $\hat\sigma^2_{\text{MV}}=\text{SSE}/n$ y $\hat\ell=-\tfrac n2\bigl[\ln(2\pi\hat\sigma^2_{\text{MV}})+1\bigr]$. (Como `statsmodels`, contamos $k$ = número de columnas de $\mathbf{X}$; contar también $\sigma^2$ suma una constante a todos los modelos y no altera las comparaciones.) El **AIC** apunta al mejor **predictor** (es asintóticamente equivalente a la validación cruzada *leave-one-out*, Stone 1977); el **BIC** penaliza más ($\ln n>2$ si $n\ge8$) y apunta a recuperar el **modelo verdadero** si está entre los candidatos. Solo se comparan modelos ajustados **a los mismos datos y a la misma $y$**.

**Experimento:** los datos salen de un polinomio **cúbico** verdadero ($n=60$, ruido $\sigma=1.5$); ajustamos grados 1 a 8 y vemos qué grado elige cada criterio, repitiendo **300 veces**. Incluimos el $R^2$ ajustado y la validación cruzada de 5 pliegues.

````python
def ll_gauss(y, yhat):
    n = len(y)
    return -n / 2 * (np.log(2 * np.pi * np.mean((y - yhat) ** 2)) + 1)


def aic_bic(y, yhat, k):
    ll = ll_gauss(y, yhat)
    return -2 * ll + 2 * k, -2 * ll + k * np.log(len(y))


aic, bic = aic_bic(y_ev, mod.predict(X_ev), k=4)
assert np.isclose(ll_gauss(y_ev, mod.predict(X_ev)), ols.llf) and np.isclose(aic, ols.aic) and np.isclose(bic, ols.bic)
print(f"AIC = {aic:.2f}, BIC = {bic:.2f}  (identicos a statsmodels)")


def vander(x, grado):
    return np.vander(x / 2, grado + 1, increasing=True)               # x en [-2, 2] -> [-1, 1]: bien condicionado


def cv5_mse(X, y, rng):
    idx = rng.permutation(len(y))
    err = 0.0
    for test in np.array_split(idx, 5):
        tr = np.setdiff1d(idx, test)
        b = np.linalg.lstsq(X[tr], y[tr], rcond=None)[0]
        err += np.sum((y[test] - X[test] @ b) ** 2)
    return err / len(y)


def seleccionar_grado(reps=300, n=60, grado_max=8, semilla=0):
    rng = np.random.default_rng(semilla)
    elegidos = {"AIC": [], "BIC": [], "R2 ajustado": [], "CV 5-pliegues": []}
    for _ in range(reps):
        x = rng.uniform(-2, 2, n)
        y = 1 + x - 0.8 * x ** 2 + 0.5 * x ** 3 + rng.normal(0, 1.5, n)
        a, b, r, c = [], [], [], []
        for g in range(1, grado_max + 1):
            X = vander(x, g)
            yh = X @ np.linalg.lstsq(X, y, rcond=None)[0]
            ai, bi = aic_bic(y, yh, g + 1)
            r2 = 1 - np.sum((y - yh) ** 2) / np.sum((y - y.mean()) ** 2)
            a.append(ai); b.append(bi); r.append(-(1 - (1 - r2) * (n - 1) / (n - g - 1))); c.append(cv5_mse(X, y, rng))
        for k, v in zip(elegidos, (a, b, r, c)):
            elegidos[k].append(1 + int(np.argmin(v)))
    return pd.DataFrame(elegidos)


sel = seleccionar_grado()
resumen = pd.DataFrame({"elige grado 3 (verdadero)": (sel == 3).mean(), "elige grado < 3": (sel < 3).mean(), "elige grado > 3": (sel > 3).mean(),
                        "grado medio elegido": sel.mean()})
print("\nGrado elegido (verdadero = 3), 300 replicas:\n" + resumen.round(3).to_string())
assert resumen.loc["BIC", "elige grado 3 (verdadero)"] > resumen.loc["AIC", "elige grado 3 (verdadero)"]
assert resumen.loc["BIC", "elige grado > 3"] < resumen.loc["AIC", "elige grado > 3"]
assert resumen.loc["R2 ajustado", "elige grado > 3"] > resumen.loc["BIC", "elige grado > 3"]
````

**Lectura.** (1) **Métricas.** Las funciones propias coinciden con `sklearn` y `statsmodels` (incluida la identidad $\text{SST}=\text{SSR}+\text{SSE}$ y el hecho de que $\hat\sigma^2$ usa $n-p-1$). (2) **Ruido.** El $R^2$ de entrenamiento sube de 0.441 a 0.824 al añadir 80 variables que **no** significan nada, mientras el $R^2$ **de prueba** cae de 0.432 a −0.903 (**negativo**: mucho peor que predecir la media). El $R^2$ ajustado se queda en ≈ 0.42–0.44: corrige el optimismo **en promedio**, pero no refleja el costo de estimar 80 coeficientes inútiles sobre las predicciones nuevas; sigue siendo una medida **dentro de muestra**. El $R^2$ ajustado sobreajusta casi siempre que se equivoca (66.7 % elige grado > 3) y la validación cruzada de 5 pliegues queda en medio (37.3 % acierta; 55 % sobreajusta). Ningún criterio es infalible con $n=60$ y ruido considerable, y que el BIC gane aquí **no** significa que sea mejor para *predecir* (el AIC apunta a otro objetivo).

## 3. Hold-out, *k*-fold, *Leave-One-Out* y Validación Repetida 🔁

El error de entrenamiento es optimista. La alternativa es **estimar el error de generalización** sobre datos que el modelo no vio:

- ***Hold-out*:** una partición entrenamiento/prueba. Sencilla, pero el resultado **depende de la partición** (alta varianza) y desperdicia datos.
- ***k-fold*:** se parten los datos en $k$ pliegues; cada uno hace de prueba una vez; se promedian los $k$ errores. Con $k$ pequeño el modelo se entrena con menos datos y el error estimado es **pesimista** (sesgo); con $k$ grande, el sesgo baja pero los conjuntos de entrenamiento se parecen mucho y la estimación puede ser más **variable** y costosa.
- ***Leave-one-out* (LOO):** $k=n$. En regresión lineal tiene **atajo exacto** sin reajustar: el residuo LOO es $e_i^{(-i)}=\dfrac{e_i}{1-h_{ii}}$, con $h_{ii}$ el elemento diagonal de la matriz sombrero $\mathbf{H}=\mathbf{X}(\mathbf{X}^{\top}\mathbf{X})^{-1}\mathbf{X}^{\top}$, y $\text{PRESS}=\sum_i\bigl(e_i/(1-h_{ii})\bigr)^2$.
- **Validación repetida:** repetir $k$-fold con distintas particiones y promediar reduce la varianza **debida a la partición**. No elimina la incertidumbre debida a los datos (no hay estimador insesgado de la varianza de la validación cruzada: Bengio y Grandvalet, 2004).

Implementamos $k$-fold y LOO desde cero y los comparamos con `sklearn`; luego medimos por simulación el **sesgo y la dispersión** de cada esquema ($n=40$, 8 predictores de los cuales 3 son reales), frente al **error real** (calculado en 5 000 puntos nuevos), y la **variabilidad del hold-out**.

````python
def pliegues(n, k, semilla=0):
    '''k-fold desde cero: permutacion aleatoria y k trozos; produce (indices_entrenamiento, indices_prueba).'''
    trozos = np.array_split(np.random.default_rng(semilla).permutation(n), k)
    for i in range(k):
        yield np.concatenate(trozos[:i] + trozos[i + 1:]), trozos[i]


def cv_mse(X, y, lista_pliegues):
    '''MSE medio de los pliegues (regresion lineal con intercepto).'''
    Xc = np.column_stack([np.ones(len(y)), X])
    errores = []
    for tr, te in lista_pliegues:
        b = np.linalg.lstsq(Xc[tr], y[tr], rcond=None)[0]
        errores.append(np.mean((y[te] - Xc[te] @ b) ** 2))
    return np.mean(errores)


lista = list(pliegues(200, 5, semilla=3))
todos_te = np.concatenate([te for _, te in lista])
assert sorted(todos_te) == list(range(200))                                                    # cada fila es prueba exactamente una vez
assert all(len(set(tr) & set(te)) == 0 for tr, te in lista)
mse_propio = cv_mse(X_ev, y_ev, lista)
mse_sk = -cross_val_score(LinearRegression(), X_ev, y_ev, cv=lista, scoring="neg_mean_squared_error").mean()
assert np.isclose(mse_propio, mse_sk)                                                          # igual a scikit-learn con los mismos pliegues

# LOO: atajo con la matriz sombrero vs reajustar n veces
Xc = np.column_stack([np.ones(200), X_ev])
H_diag = np.einsum("ij,jk,ik->i", Xc, np.linalg.inv(Xc.T @ Xc), Xc)
e = y_ev - Xc @ np.linalg.lstsq(Xc, y_ev, rcond=None)[0]
loo_atajo = np.mean((e / (1 - H_diag)) ** 2)
loo_bruto = cv_mse(X_ev, y_ev, [(np.delete(np.arange(200), i), np.array([i])) for i in range(200)])
assert np.isclose(loo_atajo, loo_bruto)
print(f"RMSE: entrenamiento = {np.sqrt(np.mean(e ** 2)):.3f} | 5-fold = {np.sqrt(mse_propio):.3f} | LOO (atajo = bucle) = {np.sqrt(loo_atajo):.3f}")

# --- Sesgo y dispersion de cada esquema (n = 40, p = 8; 3 predictores reales) ---
def experimento_k(reps=200, n=40, p=8, semilla=0):
    rng = np.random.default_rng(semilla)
    beta = np.r_[2.0, -1.5, 1.0, np.zeros(p - 3)]
    est = {"2-fold": [], "5-fold": [], "10-fold": [], "LOO": []}
    real = []
    for _ in range(reps):
        X = rng.normal(size=(n, p)); y = 10 + X @ beta + rng.normal(0, 3, n)
        Xn = rng.normal(size=(5000, p)); yn = 10 + Xn @ beta + rng.normal(0, 3, 5000)
        Xc = np.column_stack([np.ones(n), X])
        b = np.linalg.lstsq(Xc, y, rcond=None)[0]
        real.append(np.mean((yn - np.column_stack([np.ones(5000), Xn]) @ b) ** 2))                # error real del modelo ajustado con los n datos
        for k in (2, 5, 10):
            est[f"{k}-fold"].append(cv_mse(X, y, list(pliegues(n, k, semilla=int(rng.integers(1e9))))))
        h = np.einsum("ij,jk,ik->i", Xc, np.linalg.inv(Xc.T @ Xc), Xc)
        est["LOO"].append(np.mean(((y - Xc @ b) / (1 - h)) ** 2))
    real = np.array(real)
    return pd.DataFrame({k: {"MSE estimado medio": np.mean(v), "sesgo vs error real": np.mean(v) - real.mean(), "desv. est. del estimador": np.std(v)}
                         for k, v in est.items()}).T, real.mean()


tabla_k, mse_real = experimento_k()
print(f"\nError real medio (MSE en 5000 puntos nuevos) = {mse_real:.2f}\n" + tabla_k.round(2).to_string())
assert tabla_k.loc["2-fold", "sesgo vs error real"] > tabla_k.loc["10-fold", "sesgo vs error real"] > -1.0
assert tabla_k.loc["2-fold", "sesgo vs error real"] > 0.8

# --- Variabilidad del hold-out vs validacion repetida (datos de la seccion 1) ---
rmse_ho = np.array([np.sqrt(mean_squared_error(yb, LinearRegression().fit(Xa, ya).predict(Xb)))
                    for Xa, Xb, ya, yb in (train_test_split(X_ev, y_ev, test_size=0.3, random_state=s) for s in range(300))])
rmse_rep = np.array([np.sqrt(-cross_val_score(LinearRegression(), X_ev, y_ev, cv=KFold(10, shuffle=True, random_state=s), scoring="neg_mean_squared_error").mean())
                     for s in range(300)])
print(f"\nRMSE hold-out 70/30 (300 particiones): media {rmse_ho.mean():.2f}, rango [{rmse_ho.min():.2f}, {rmse_ho.max():.2f}], d.e. {rmse_ho.std():.3f}")
print(f"RMSE 10-fold (300 particiones):         media {rmse_rep.mean():.2f}, rango [{rmse_rep.min():.2f}, {rmse_rep.max():.2f}], d.e. {rmse_rep.std():.3f}")
assert rmse_ho.std() > 2 * rmse_rep.std()

# --- Incertidumbre del RMSE de prueba: bootstrap de los errores al cuadrado ---
Xa, Xb, ya, yb = train_test_split(X_ev, y_ev, test_size=0.3, random_state=0)
err2 = (yb - LinearRegression().fit(Xa, ya).predict(Xb)) ** 2
boot = np.sqrt(np.random.default_rng(1).choice(err2, (4000, len(err2))).mean(axis=1))
ic = np.percentile(boot, [2.5, 97.5])
print(f"\nRMSE de prueba (60 observaciones) = {np.sqrt(err2.mean()):.2f}, IC bootstrap 95% = [{ic[0]:.2f}, {ic[1]:.2f}] (el sigma verdadero es 3)")
assert ic[0] < np.sqrt(err2.mean()) < ic[1]
````

---
**Lectura (sección 3).** (1) Nuestros pliegues y el atajo de la matriz sombrero reproducen a `sklearn` y al bucle explícito de LOO. (2) **Sesgo:** con $n=40$, entrenar con la mitad de los datos (2-fold) da un MSE estimado de 17.4 frente a un error real de 11.6 (sesgo **pesimista** de +5.9); 5-fold (+1.1) y 10-fold (+0.5) se acercan, y el LOO (+0.3) es casi insesgado. La desviación del estimador también baja con $k$ (7.0, 3.5, 3.1, 3.0). (3) **Dispersión:** un solo *hold-out* 70/30 produce RMSE entre 2.59 y 3.94 según la partición (d.e. 0.25), mientras que repetir el 10-fold da una estimación estable (d.e. 0.018). Pero estabilizar la partición no vuelve «verdadero» al número: la incertidumbre por tener **pocos datos** sigue ahí, y por eso reportamos el intervalo *bootstrap* del RMSE de prueba (60 observaciones: 3.04, IC 95 % [2.55, 3.50]).

## 4. Validación Cruzada que Respeta la Estructura: Grupos y Tiempo 🧱

La validación $k$-fold supone que las observaciones son **intercambiables** (independientes). Dos situaciones comunes lo rompen:

1. **Mediciones repetidas (grupos).** Varias medidas de la **misma finca**. Si un pliegue aleatorio deja filas de la finca A en entrenamiento y prueba, un modelo flexible **reconoce la finca** (por cualquier covariable constante en ella, como la altitud) y «predice» su nivel sin haber aprendido nada generalizable. Si el modelo se usará en **fincas nuevas**, hay que validar con ***GroupKFold***: cada finca está **entera** en entrenamiento o en prueba.
2. **Series temporales.** Un pliegue aleatorio entrena con el **futuro** para predecir el pasado (interpola entre vecinos temporales). Si el modelo predecirá el **futuro**, la validación debe ser ***forward chaining*** (`TimeSeriesSplit`): entrenar con el pasado y probar en el bloque siguiente.

La regla general: **el esquema de validación debe imitar cómo se usará el modelo**. Simulamos (a) 40 fincas con 10 mediciones cada una, donde el nivel de cada finca (efecto aleatorio, sd = 3) **no** depende de la altitud; y (b) 20 años de datos mensuales con tendencia, estacionalidad y errores autocorrelacionados. Comparamos un bosque aleatorio (flexible) con una regresión lineal (rígida).

````python
# ---------- (a) Mediciones repetidas por finca ----------
rng = np.random.default_rng(2)
n_f, m_f = 40, 10
finca = np.repeat(np.arange(n_f), m_f)
altitud = rng.uniform(2400, 3000, n_f)[finca]                      # constante dentro de la finca
efecto = rng.normal(0, 3.0, n_f)[finca]                             # nivel propio de la finca (NO depende de la altitud)
humedad = rng.normal(size=n_f * m_f)
y_f = 15 + 1.0 * humedad + efecto + rng.normal(0, 1.0, n_f * m_f)
X_f = np.column_stack([altitud, humedad])

modelos = {"Bosque aleatorio": RandomForestRegressor(100, random_state=0, n_jobs=1), "Regresión lineal": LinearRegression()}
filas = {}
for nombre, mdl in modelos.items():
    r2_az = cross_val_score(mdl, X_f, y_f, cv=KFold(5, shuffle=True, random_state=0), scoring="r2").mean()
    r2_gr = cross_val_score(mdl, X_f, y_f, cv=GroupKFold(5), groups=finca, scoring="r2").mean()
    filas[nombre] = {"KFold aleatorio": r2_az, "GroupKFold": r2_gr}
tab_g = pd.DataFrame(filas).T
print("R2 de validacion cruzada, mediciones repetidas por finca:\n" + tab_g.round(3).to_string())
for tr, te in GroupKFold(5).split(X_f, y_f, finca):
    assert set(finca[tr]).isdisjoint(finca[te])                      # ninguna finca aparece en entrenamiento y prueba a la vez

# ---------- (b) Serie mensual con tendencia, estacionalidad y errores autocorrelacionados ----------
rng = np.random.default_rng(4)
T = 240
t = np.arange(T)
ruido = np.zeros(T)
for i in range(1, T):
    ruido[i] = 0.8 * ruido[i - 1] + rng.normal(0, 0.8)               # AR(1)
y_t = 20 + 0.05 * t + 3 * np.sin(2 * np.pi * t / 12) + ruido
X_bosque = np.column_stack([t, t % 12])                             # indice de tiempo y mes
X_lineal = np.column_stack([t, np.sin(2 * np.pi * t / 12), np.cos(2 * np.pi * t / 12)])


def pliegues_temporales(n, k):
    '''Forward chaining: entrena con el pasado, prueba en el bloque siguiente (igual a TimeSeriesSplit).'''
    tam = n // (k + 1)
    for i in range(k):
        ini = n - k * tam + i * tam
        yield np.arange(ini), np.arange(ini, ini + tam)


for (a1, b1), (a2, b2) in zip(pliegues_temporales(T, 5), TimeSeriesSplit(5).split(t)):
    assert np.array_equal(a1, a2) and np.array_equal(b1, b2)         # igual a scikit-learn
assert all(tr.max() < te.min() for tr, te in pliegues_temporales(T, 5))   # el entrenamiento siempre es anterior a la prueba

filas = {}
for nombre, mdl, X in [("Bosque aleatorio (t, mes)", RandomForestRegressor(100, random_state=0), X_bosque),
                       ("Regresión: tendencia + armónico", LinearRegression(), X_lineal)]:
    filas[nombre] = {"KFold aleatorio": cross_val_score(mdl, X, y_t, cv=KFold(5, shuffle=True, random_state=0), scoring="r2").mean(),
                     "Forward chaining": cross_val_score(mdl, X, y_t, cv=list(pliegues_temporales(T, 5)), scoring="r2").mean()}
tab_t = pd.DataFrame(filas).T
print("\nR2 de validacion cruzada, serie mensual:\n" + tab_t.round(3).to_string())

# ---------- Estratificada: clasificacion con 10 % de positivos ----------
y_cl = np.r_[np.ones(20), np.zeros(180)]
prop = [y_cl[te].mean() for _, te in StratifiedKFold(5, shuffle=True, random_state=0).split(np.zeros(200), y_cl)]
prop_ale = [y_cl[te].mean() for _, te in KFold(5, shuffle=True, random_state=0).split(np.zeros(200))]
print(f"\nProporcion de positivos por pliegue: estratificado = {np.round(prop, 2)} | aleatorio = {np.round(prop_ale, 2)}")

assert tab_g.loc["Bosque aleatorio", "KFold aleatorio"] > tab_g.loc["Bosque aleatorio", "GroupKFold"] + 0.4
assert tab_g.loc["Bosque aleatorio", "GroupKFold"] < 0.25
assert tab_t.iloc[0]["KFold aleatorio"] > tab_t.iloc[0]["Forward chaining"] + 0.2
assert max(prop) - min(prop) < 0.05
````

## 5. Sobreajuste Honesto: la Fuga al Seleccionar Variables 🕳️

El error de validación más común no está en el modelo sino en el **procedimiento**: usar **toda** la información (incluida $y$) para decidir algo —qué variables conservar, qué transformar, qué hiperparámetro usar— **antes** de dividir en pliegues. Los pliegues de prueba ya «participaron» en la decisión y el resultado es optimista.

**Caso extremo (Hastie et al., cap. 7).** $n=60$ observaciones, $p=500$ predictores y una respuesta que es **ruido puro**, independiente de todo. Sabemos que ningún modelo puede predecir: el $R^2$ de prueba esperado es $\le0$. Si primero **elegimos las 10 variables más correlacionadas con $y$ usando todos los datos** y luego hacemos validación cruzada, obtendremos un $R^2$ «brillante». La salida honesta es meter la **selección dentro del *pipeline***, de modo que se repita **dentro de cada pliegue de entrenamiento**.

````python
def datos_ruido(n=60, p=500, semilla=0):
    r = np.random.default_rng(semilla)
    return r.normal(size=(n, p)), r.normal(size=n)                  # y es RUIDO PURO: independiente de X


def r2_selection(X, y, semilla=0):
    cv = KFold(5, shuffle=True, random_state=semilla)
    X_fuga = SelectKBest(f_regression, k=10).fit_transform(X, y)                 # MAL: la seleccion ve TODOS los datos (incluida la prueba)
    mal = cross_val_score(LinearRegression(), X_fuga, y, cv=cv, scoring="r2").mean()
    pipe = make_pipeline(SelectKBest(f_regression, k=10), LinearRegression())      # BIEN: la seleccion se repite dentro de cada pliegue
    bien = cross_val_score(pipe, X, y, cv=cv, scoring="r2").mean()
    return mal, bien


X_rn, y_rn = datos_ruido()
mal0, bien0 = r2_selection(X_rn, y_rn)
res = np.array([r2_selection(*datos_ruido(semilla=s), semilla=s) for s in range(30)])
print(f"Una muestra:  R2 con fuga = {mal0:.3f} | R2 honesto (pipeline) = {bien0:.3f}")
print(f"Promedio de 30 muestras de ruido puro: R2 con fuga = {res[:, 0].mean():.3f} | R2 honesto = {res[:, 1].mean():.3f}")
assert res[:, 0].mean() > 0.15 > 0 > res[:, 1].mean()
````

---
**Lectura (secciones 4 y 5).** (1) **Grupos.** El bosque aleatorio obtiene un $R^2$ de **0.669** con `KFold` aleatorio y **−0.854** con `GroupKFold`: con pliegues aleatorios **reconoce cada finca por su altitud** y recita su nivel; con fincas nuevas no hay nada que recordar, y como el nivel de cada finca no depende de la altitud, el modelo flexible aprende un patrón falso y queda **muy por debajo de la media**. La regresión lineal, que no puede memorizar, da casi lo mismo con ambos esquemas (−0.000 frente a −0.063): **la fuga infla sobre todo a los modelos flexibles**. (2) **Tiempo.** Con pliegues aleatorios el bosque interpola entre meses vecinos ($R^2=0.944$); con *forward chaining* tiene que **extrapolar la tendencia**, que un árbol no sabe hacer (0.273). La regresión con tendencia y armónico también baja (0.920 a 0.638, por la autocorrelación y porque los bloques de prueba son cortos), pero **mucho menos**. (3) **Fuga por selección.** Con respuesta de ruido puro, seleccionar variables con todos los datos arroja un $R^2$ medio de **0.228**, mientras que el *pipeline* honesto da **−0.608** (negativo, como debe ser: con 500 predictores y 60 datos, cualquier ajuste empeora a la media). **Moraleja:** todo lo que mire $y$ —selección, imputación por objetivo, escalado, codificación por objetivo, elección de hiperparámetros— debe vivir **dentro** de la validación. La estratificación, por su parte, mantiene la proporción de la clase rara en cada pliegue (0.10 en los cinco, frente a 0.02–0.15 con pliegues aleatorios).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Métricas** | $R^2=1-\text{SSE}/\text{SST}$ (igual a `sklearn` y `statsmodels`), RMSE en las unidades de $y$, MAE más robusto. $\hat\sigma^2$ usa $n-p-1$, no $n$ |
| **$R^2$ de entrenamiento** | **Nunca baja** al añadir variables, aunque sean ruido (de 0.44 a 0.82 con 80 columnas inútiles); fuera de muestra **puede ser negativo** |
| **AIC / BIC** | $-2\hat\ell+2k$ y $-2\hat\ell+k\ln n$ (idénticos a `statsmodels`). BIC más parsimonioso; AIC apunta a predecir. Ningún criterio es infalible con $n$ pequeño |
| **Hold-out** | Sencillo pero de **alta varianza** según la partición |
| ***k*-fold, LOO, repetida** | $k$ pequeño $\Rightarrow$ error **pesimista**; LOO casi insesgado y con atajo $e_i/(1-h_{ii})$; repetir reduce la varianza de la partición, no la incertidumbre por pocos datos |
| **Por grupos y temporal** | Mediciones repetidas $\Rightarrow$ `GroupKFold`; series $\Rightarrow$ *forward chaining*. Los esquemas aleatorios **inflan** el desempeño sobre todo de los modelos flexibles |
| **Sobreajuste honesto** | Seleccionar variables con todos los datos antes de validar produce un $R^2$ positivo sobre **ruido puro**; la solución es el ***pipeline*** |
| **Reportar** | Una métrica sin su incertidumbre (intervalo *bootstrap*) es solo una anécdota |
