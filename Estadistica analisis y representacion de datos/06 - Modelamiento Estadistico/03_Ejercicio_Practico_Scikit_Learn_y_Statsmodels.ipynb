## 1. El Flujo de Trabajo y las Fórmulas de *patsy* 🧭

**Flujo de modelamiento:** (1) pregunta y tipo de respuesta → (2) datos y codificación → (3) **ajuste** → (4) **diagnóstico** de supuestos → (5) **validación** fuera de muestra → (6) **reporte** con incertidumbre. Las dos bibliotecas se reparten el trabajo:

| | `statsmodels` | `scikit-learn` |
|---|---|---|
| Objetivo | **Inferir**: ¿cuánto y con qué incertidumbre? | **Predecir**: ¿qué tan bien en datos nuevos? |
| Interfaz | Fórmulas (`y ~ x1 + C(g)`) o matrices (añadir constante a mano) | Matrices/*DataFrames* y *pipelines* |
| Salida | `summary()`: coeficientes, EE, $t$, $p$, IC, $R^2$, AIC/BIC, pruebas | `predict`, `score`, validación cruzada, búsqueda de hiperparámetros |
| Intercepto | Explícito (constante) | Automático (`fit_intercept=True`) |
| Regularización | Limitada | `Ridge`, `Lasso`, `ElasticNet` y más |

**Fórmulas.** La notación de *patsy* describe la matriz de diseño: `~` separa respuesta y predictores; `+` añade términos; `C(x)` fuerza categórica (con contraste de tratamiento: se omite la primera categoría, o la indicada con `C(x, Treatment('ref'))`); `a:b` es la interacción; `a*b` equivale a `a + b + a:b`; `I(x**2)` calcula tal cual; `np.log(x)` aplica funciones de NumPy; `- 1` elimina el intercepto. Trabajamos con el **conjunto de diabetes** de scikit-learn (442 pacientes): diez variables basales (`age`, `sex`, `bmi`, `bp` y seis mediciones de suero `s1`–`s6`) y la respuesta `target`, «una medida cuantitativa de la progresión de la enfermedad un año después del inicio» (según la documentación de scikit-learn). Es un conjunto **real**, y como tal no conocemos el mecanismo verdadero: la pregunta es descriptiva y predictiva, **no causal**. La variable `sex` aparece con los códigos 1 y 2 de la fuente; la tratamos como categoría nominal.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import patsy
from scipy import stats
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, RidgeCV, GammaRegressor
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import RepeatedKFold, cross_validate, train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler, SplineTransformer
from sklearn.impute import SimpleImputer
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import jarque_bera, durbin_watson
from statsmodels.stats.outliers_influence import variance_inflation_factor, OLSInfluence

plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})

diab = load_diabetes(as_frame=True, scaled=False).frame
diab["sex"] = diab["sex"].astype(int).astype(str)                 # nominal: codigos '1' y '2' de la fuente
print(diab.shape, "| faltantes:", int(diab.isna().sum().sum()))
print(diab.describe().T[["mean", "std", "min", "max"]].round(2).to_string())

formula = "target ~ age + C(sex) + bmi + bp + s1 + s2 + s3 + s4 + s5 + s6"
y_dm, X_dm = patsy.dmatrices(formula, diab, return_type="dataframe")
print("\ncolumnas de la matriz de diseno:", list(X_dm.columns))

# La matriz de diseno de patsy == la que armariamos a mano
X_mano = np.column_stack([np.ones(len(diab)), (diab["sex"] == "2").astype(float), diab["age"], diab[["bmi", "bp", "s1", "s2", "s3", "s4", "s5", "s6"]]])
assert np.allclose(X_dm.to_numpy(), X_mano)

m_full = smf.ols(formula, diab).fit()
print(m_full.summary())

# --- Reconstruir el summary() con algebra matricial ---
X, y = X_dm.to_numpy(), y_dm.to_numpy().ravel()
n, k = X.shape
beta = np.linalg.solve(X.T @ X, X.T @ y)                          # ecuaciones normales
e = y - X @ beta
s2 = e @ e / (n - k)
ee = np.sqrt(s2 * np.diag(np.linalg.inv(X.T @ X)))
t = beta / ee
p_val = 2 * stats.t.sf(np.abs(t), n - k)
sst, sse = np.sum((y - y.mean()) ** 2), e @ e
F = ((sst - sse) / (k - 1)) / (sse / (n - k))
assert np.allclose(beta, m_full.params) and np.allclose(ee, m_full.bse) and np.allclose(t, m_full.tvalues) and np.allclose(p_val, m_full.pvalues)
assert np.isclose(1 - sse / sst, m_full.rsquared) and np.isclose(F, m_full.fvalue) and np.isclose(s2, m_full.mse_resid)
ic = np.c_[beta - stats.t.ppf(0.975, n - k) * ee, beta + stats.t.ppf(0.975, n - k) * ee]
assert np.allclose(ic, m_full.conf_int().to_numpy())
print("\nCoeficientes, EE, t, p, IC95%, R2 y F reconstruidos con (X'X)^-1 X'y: IGUALES al summary().")
````

## 2. Leer el Resumen y Diagnosticar 🩺

**Cómo leer el `summary()`.** (i) *Cabecera:* $R^2$ y $R^2$ ajustado, el estadístico $F$ (¿hay *algún* predictor útil?; $H_0$: todos los $\beta_j=0$), log-verosimilitud, AIC y BIC. (ii) *Tabla de coeficientes:* estimación, error estándar, $t=\hat\beta/\operatorname{EE}$, valor $p$ de $H_0:\beta_j=0$ **dados los demás términos** e IC 95 %. (iii) *Pie:* *Omnibus* y *Jarque–Bera* (normalidad de los residuos), *Durbin–Watson* (autocorrelación; ≈ 2 indica ninguna) y el **número de condición** (valores grandes sugieren multicolinealidad). **Cuidado:** un $p$ grande en un coeficiente **no** significa que la variable no sirva; con predictores correlacionados los efectos individuales se confunden.

**Diagnóstico de supuestos.**
- ***VIF*** (factor de inflación de la varianza): $\text{VIF}_j=1/(1-R_j^2)$, donde $R_j^2$ es el de regresar $x_j$ sobre los demás predictores. Valores $>5$–$10$ indican multicolinealidad fuerte: el EE de $\hat\beta_j$ se infla por $\sqrt{\text{VIF}_j}$.
- ***Breusch–Pagan***: $H_0$ = varianza constante. ***Jarque–Bera***: $H_0$ = normalidad (asimetría 0, curtosis 3). ***Durbin–Watson***: independencia serial.
- **Apalancamiento** $h_{ii}$ (elementos diagonales de $\mathbf{H}=\mathbf{X}(\mathbf{X}^{\top}\mathbf{X})^{-1}\mathbf{X}^{\top}$) y **distancia de Cook** $D_i=\dfrac{e_i^2}{k\,s^2}\cdot\dfrac{h_{ii}}{(1-h_{ii})^2}$: cuánto cambian los coeficientes al quitar la observación $i$.
- **Errores estándar robustos (HC3)** cuando la varianza no es constante.

Después **simplificamos**: el modelo reducido (`sex`, `bmi`, `bp`, `s3`, `s5`) está **anidado** en el completo, así que se compara con la prueba $F$ y con AIC/BIC.

````python
# --- Diagnostico del modelo completo ---
vif = pd.Series([variance_inflation_factor(X, j) for j in range(1, k)], index=X_dm.columns[1:]).round(1)
bp_lm, bp_p = het_breuschpagan(m_full.resid, m_full.model.exog)[:2]
jb, jb_p = jarque_bera(m_full.resid)[:2]
dw = durbin_watson(m_full.resid)
print("VIF:", vif.to_dict())
print(f"Breusch-Pagan: LM = {bp_lm:.2f}, p = {bp_p:.4f} | Jarque-Bera: JB = {jb:.2f}, p = {jb_p:.3f} | Durbin-Watson = {dw:.2f}")

infl = OLSInfluence(m_full)
h, cook = infl.hat_matrix_diag, infl.cooks_distance[0]
print(f"Apalancamiento maximo = {h.max():.3f} (promedio k/n = {k / n:.3f}) | Cook maxima = {cook.max():.3f} | obs. con h > 3k/n: {(h > 3 * k / n).sum()}")

fig, ax = plt.subplots(2, 2, figsize=(10, 7))
ax[0, 0].scatter(m_full.fittedvalues, m_full.resid, s=10, alpha=0.6, color="#1e3a8a"); ax[0, 0].axhline(0, color="k", lw=1)
ax[0, 0].set(title="Residuos vs. ajustados", xlabel="ajustado", ylabel="residuo")
sm.qqplot(infl.resid_studentized_internal, line="45", ax=ax[0, 1], markersize=3); ax[0, 1].set_title("QQ de los residuos estudentizados")
ax[1, 0].scatter(m_full.fittedvalues, np.sqrt(np.abs(infl.resid_studentized_internal)), s=10, alpha=0.6, color="#1e3a8a")
ax[1, 0].set(title="Escala-ubicación (varianza)", xlabel="ajustado", ylabel="√|residuo estudentizado|")
ax[1, 1].scatter(h, infl.resid_studentized_internal, s=400 * cook / cook.max(), alpha=0.4, color="#dc2626")
ax[1, 1].set(title="Apalancamiento vs. residuo (tamaño = Cook)", xlabel="apalancamiento $h_{ii}$", ylabel="residuo estudentizado")
plt.tight_layout(); plt.show()

# --- Errores estandar clasicos vs. robustos (HC3) ---
m_hc3 = smf.ols(formula, diab).fit(cov_type="HC3")
cmp_ee = pd.DataFrame({"EE clasico": m_full.bse, "EE HC3": m_hc3.bse, "razon": m_hc3.bse / m_full.bse}).round(3)
print("\n" + cmp_ee.to_string())

# --- Modelo reducido (anidado) y comparacion ---
m_red = smf.ols("target ~ C(sex) + bmi + bp + s3 + s5", diab).fit()
f_nested = anova_lm(m_red, m_full)
print(f"\nCompleto: R2 = {m_full.rsquared:.3f}, R2aj = {m_full.rsquared_adj:.3f}, AIC = {m_full.aic:.1f}, BIC = {m_full.bic:.1f}")
print(f"Reducido: R2 = {m_red.rsquared:.3f}, R2aj = {m_red.rsquared_adj:.3f}, AIC = {m_red.aic:.1f}, BIC = {m_red.bic:.1f}")
print(f"Prueba F anidada (5 terminos eliminados): F = {f_nested['F'].iloc[1]:.2f}, p = {f_nested['Pr(>F)'].iloc[1]:.3f}")

assert vif[["s1", "s2"]].min() > 10 and vif.drop(["s1", "s2", "s3", "s4", "s5"]).max() < 2.5          # s1, s2 (colesterol) casi colineales
assert np.isclose(h.sum(), k) and (cook > 0).all()                                                  # traza de H = numero de parametros
assert m_red.bic < m_full.bic and f_nested["df_diff"].iloc[1] == 5
````

## 3. El Puente: el Mismo Modelo en las Dos Bibliotecas 🌉

`LinearRegression` y `OLS` resuelven **el mismo problema de mínimos cuadrados**; lo que cambia es la interfaz y lo que devuelven. Lo mismo ocurre con `GammaRegressor` (enlace log, penalización $\alpha=0$) y `GLM(..., family=Gamma(Log()))`. La respuesta `target` es positiva y asimétrica, así que el GLM Gamma es un candidato natural (sección 1 del cuaderno 01). Verificamos las equivalencias y luego **comparamos los modelos fuera de muestra** con validación cruzada 5×3 repetida: ¿mejora algo el modelo más elaborado?

Ojo con una diferencia práctica: el $R^2$ de `statsmodels` (0.518) es **dentro de muestra**; el de la validación cruzada es la estimación honesta del desempeño.

````python
Xd, yd = diab.drop(columns="target"), diab["target"]
Xdum = pd.get_dummies(Xd, columns=["sex"], drop_first=True, dtype=float)

# 1) LinearRegression == OLS
lr = LinearRegression().fit(Xdum, yd)
sm_ols = sm.OLS(yd, sm.add_constant(Xdum)).fit()
assert np.allclose(np.r_[lr.intercept_, lr.coef_], sm_ols.params.to_numpy())
assert np.allclose(lr.predict(Xdum), m_full.fittedvalues)                                    # y tambien == la version con formula

# 2) GammaRegressor(alpha=0) == GLM Gamma con enlace log
glm_g = smf.glm(formula, diab, family=sm.families.Gamma(sm.families.links.Log())).fit()
gr = GammaRegressor(alpha=0, solver="newton-cholesky", tol=1e-10, max_iter=1000).fit(Xdum, yd)
assert np.allclose(gr.predict(Xdum), glm_g.predict(diab), rtol=1e-6)
assert np.allclose(np.r_[gr.intercept_, gr.coef_], glm_g.params.to_numpy()[[0] + list(range(2, 11)) + [1]], atol=1e-5)
print("LinearRegression == OLS y GammaRegressor(alpha=0) == GLM Gamma(log): IGUALES (coeficientes y predicciones).")

# 3) Comparacion fuera de muestra: 5 pliegues x 3 repeticiones
cv = RepeatedKFold(n_splits=5, n_repeats=3, random_state=0)
num_all = [c for c in Xd.columns if c != "sex"]
prep_full = ColumnTransformer([("num", StandardScaler(), num_all), ("sex", OneHotEncoder(drop="first"), ["sex"])])
prep_red = ColumnTransformer([("num", StandardScaler(), ["bmi", "bp", "s3", "s5"]), ("sex", OneHotEncoder(drop="first"), ["sex"])])
candidatos = {"Solo la media": DummyRegressor(),
              "OLS completo": make_pipeline(prep_full, LinearRegression()),
              "OLS reducido": make_pipeline(prep_red, LinearRegression()),
              "Ridge (alfa por CV)": make_pipeline(prep_full, RidgeCV(alphas=np.logspace(-2, 3, 13))),
              "GLM Gamma (log)": make_pipeline(prep_full, GammaRegressor(alpha=0, solver="newton-cholesky"))}
filas = {}
for nombre, mdl in candidatos.items():
    r = cross_validate(mdl, Xd, yd, cv=cv, scoring=("neg_root_mean_squared_error", "r2"))
    filas[nombre] = {"RMSE (CV)": -r["test_neg_root_mean_squared_error"].mean(), "R2 (CV)": r["test_r2"].mean(), "d.e. RMSE entre pliegues": (-r["test_neg_root_mean_squared_error"]).std()}
tab_cv = pd.DataFrame(filas).T
print("\nR2 dentro de muestra del OLS completo (statsmodels) = %.3f\n" % m_full.rsquared + tab_cv.round(3).to_string())

assert tab_cv.loc["OLS completo", "RMSE (CV)"] < 0.75 * tab_cv.loc["Solo la media", "RMSE (CV)"]
assert tab_cv.loc["OLS completo", "R2 (CV)"] < m_full.rsquared                                  # el R2 honesto es menor que el de entrenamiento
spread = tab_cv.drop("Solo la media")["RMSE (CV)"]
assert spread.max() - spread.min() < 0.05 * spread.min()                                        # ninguno de los 4 modelos mejora de forma material
````

## 4. *Pipeline* con `ColumnTransformer` sobre Datos Mixtos 🧩

Los datos reales mezclan **numéricas con faltantes**, **categorías nominales**, **ordinales** y relaciones **no lineales**. Un `ColumnTransformer` aplica a cada grupo de columnas su propio tratamiento, y un `Pipeline` lo encadena con el modelo, de modo que **todo lo que aprende de los datos (medianas de imputación, medias y desviaciones, categorías) se ajusta solo con el entrenamiento de cada pliegue**: sin fuga (cuaderno 02, sección 5). Simulamos 500 fincas **sintéticas** con: `altitud`, `fert` (8 % faltante) y `precip` (5 % faltante) numéricas; `variedad` (3 % faltante) y `zona` nominales; `manejo` ordinal (Bajo < Medio < Alto). La respuesta `rend` es positiva y asimétrica, y **depende de la altitud de forma cuadrática** (óptimo cerca de 2 800 m): un modelo lineal en la altitud no puede captarlo, pero con *splines* sí.

Compararemos OLS, Ridge y GLM Gamma, con y sin *splines* en la altitud, con validación cruzada 5×3; elegiremos con la validación y **evaluaremos una sola vez** en un conjunto de prueba reservado. Para **inferir**, ajustaremos después un GLM Gamma con `statsmodels` y leeremos los efectos como **razones de medias** con su intervalo.

````python
def generar_fincas_mixtas(n=500, semilla=8):
    '''SINTETICO. log(media) = 2.3 - 0.14 z_alt^2 + 0.15 log1p(fert/100) + 0.0003 (precip - 900) + efectos de variedad, manejo y zona. Gamma(forma=25).'''
    rng = np.random.default_rng(semilla)
    zonas = [f"Zona {c}" for c in "ABCDEF"]
    zona = rng.choice(zonas, n)
    variedad = rng.choice(["Capiro", "Criolla", "Pastusa", "Sabanera"], n, p=[0.30, 0.20, 0.35, 0.15])
    manejo = rng.choice(["Bajo", "Medio", "Alto"], n, p=[0.35, 0.40, 0.25])
    altitud, fert, precip = rng.normal(2800, 150, n), rng.gamma(4, 50, n), rng.normal(900, 120, n)
    ef_z = pd.Series(zona).map(dict(zip(zonas, [0, 0.05, -0.05, 0.10, -0.10, 0.02]))).to_numpy()
    ef_v = pd.Series(variedad).map({"Capiro": 0, "Criolla": -0.20, "Pastusa": 0.25, "Sabanera": -0.05}).to_numpy()
    ef_m = pd.Series(manejo).map({"Bajo": 0, "Medio": 0.12, "Alto": 0.20}).to_numpy()
    log_mu = 2.3 - 0.14 * ((altitud - 2800) / 150) ** 2 + 0.15 * np.log1p(fert / 100) + 0.0003 * (precip - 900) + ef_v + ef_m + ef_z
    d = pd.DataFrame({"altitud": altitud, "fert": fert, "precip": precip, "variedad": variedad, "zona": zona, "manejo": manejo,
                      "rend": rng.gamma(25, np.exp(log_mu) / 25)})
    d.loc[rng.choice(n, int(0.08 * n), replace=False), "fert"] = np.nan
    d.loc[rng.choice(n, int(0.05 * n), replace=False), "precip"] = np.nan
    d.loc[rng.choice(n, int(0.03 * n), replace=False), "variedad"] = np.nan
    return d


fincas = generar_fincas_mixtas()
print("faltantes:", fincas.isna().sum()[lambda s: s > 0].to_dict(), "| rend: media = %.2f, d.e. = %.2f, min = %.2f" % (fincas.rend.mean(), fincas.rend.std(), fincas.rend.min()))
Xf, yf = fincas.drop(columns="rend"), fincas["rend"]
Xf_ent, Xf_pru, yf_ent, yf_pru = train_test_split(Xf, yf, test_size=0.2, random_state=0)

num, nom, ordi = ["altitud", "fert", "precip"], ["variedad", "zona"], ["manejo"]


def preprocesador(splines=False):
    pasos = [("num", make_pipeline(SimpleImputer(strategy="median"), StandardScaler()), num),
             ("nom", make_pipeline(SimpleImputer(strategy="most_frequent"), OneHotEncoder(drop="first", handle_unknown="ignore")), nom),
             ("ord", OrdinalEncoder(categories=[["Bajo", "Medio", "Alto"]]), ordi)]
    if splines:
        pasos.append(("alt", make_pipeline(SimpleImputer(strategy="median"), SplineTransformer(n_knots=5, degree=3, include_bias=False)), ["altitud"]))
    return ColumnTransformer(pasos)


modelos = {"Solo la media": DummyRegressor(),
           "OLS": make_pipeline(preprocesador(), LinearRegression()),
           "Ridge": make_pipeline(preprocesador(), RidgeCV(alphas=np.logspace(-3, 3, 13))),
           "GLM Gamma": make_pipeline(preprocesador(), GammaRegressor(alpha=0, solver="newton-cholesky")),
           "OLS + splines(altitud)": make_pipeline(preprocesador(True), LinearRegression()),
           "Ridge + splines(altitud)": make_pipeline(preprocesador(True), RidgeCV(alphas=np.logspace(-3, 3, 13))),
           "GLM Gamma + splines(altitud)": make_pipeline(preprocesador(True), GammaRegressor(alpha=0, solver="newton-cholesky"))}
cv = RepeatedKFold(n_splits=5, n_repeats=3, random_state=0)
filas = {}
for nombre, mdl in modelos.items():
    r = cross_validate(mdl, Xf_ent, yf_ent, cv=cv, scoring=("neg_root_mean_squared_error", "r2"))
    filas[nombre] = {"RMSE (CV)": -r["test_neg_root_mean_squared_error"].mean(), "R2 (CV)": r["test_r2"].mean()}
tab_f = pd.DataFrame(filas).T
print("\nValidacion cruzada 5x3 sobre el entrenamiento (400 fincas):\n" + tab_f.round(3).to_string())

# Se elige con la validacion y se evalua UNA sola vez en la prueba
mejor_nombre = tab_f["RMSE (CV)"].idxmin()
mejor = modelos[mejor_nombre].fit(Xf_ent, yf_ent)
err2 = (yf_pru - mejor.predict(Xf_pru)) ** 2
rmse_pru = np.sqrt(err2.mean())
ic = np.percentile(np.sqrt(np.random.default_rng(0).choice(err2.to_numpy(), (4000, len(err2))).mean(axis=1)), [2.5, 97.5])
r2_pru = 1 - err2.sum() / ((yf_pru - yf_pru.mean()) ** 2).sum()
print(f"\nMejor por CV: {mejor_nombre}. Prueba (100 fincas): RMSE = {rmse_pru:.3f} (IC bootstrap 95% [{ic[0]:.3f}, {ic[1]:.3f}]), R2 = {r2_pru:.3f}")

# Equivalencia con statsmodels (casos completos): pipeline lineal sin escalado == formula
comp = fincas.dropna().copy()
comp["manejo_cod"] = comp["manejo"].map({"Bajo": 0, "Medio": 1, "Alto": 2})
f_sm = smf.ols("rend ~ altitud + fert + precip + C(variedad) + C(zona) + manejo_cod", comp).fit()
pre_simple = ColumnTransformer([("num", "passthrough", num), ("nom", OneHotEncoder(drop="first"), nom),
                                ("ord", OrdinalEncoder(categories=[["Bajo", "Medio", "Alto"]]), ordi)])
pip_simple = make_pipeline(pre_simple, LinearRegression()).fit(comp.drop(columns=["rend", "manejo_cod"]), comp["rend"])
assert np.allclose(pip_simple.predict(comp.drop(columns=["rend", "manejo_cod"])), f_sm.fittedvalues)
assert np.isclose(pip_simple[-1].coef_[1], f_sm.params["fert"])
print("Pipeline lineal (ColumnTransformer + LinearRegression) == statsmodels con formula: IGUALES.")

# Inferencia: GLM Gamma con statsmodels (casos completos), efectos como razones de medias
comp["z_alt"] = (comp["altitud"] - 2800) / 150
glm_f = smf.glm("rend ~ z_alt + I(z_alt**2) + np.log1p(fert / 100) + precip + C(variedad) + C(zona) + C(manejo, Treatment('Bajo'))",
                comp, family=sm.families.Gamma(sm.families.links.Log())).fit()
razon = pd.DataFrame({"razon de medias": np.exp(glm_f.params), "IC95% inf": np.exp(glm_f.conf_int()[0]), "IC95% sup": np.exp(glm_f.conf_int()[1])})
print("\nGLM Gamma (statsmodels, %d fincas completas): razones de medias\n" % len(comp) + razon.drop("Intercept").round(3).to_string())

assert tab_f.loc["OLS + splines(altitud)", "RMSE (CV)"] < 0.9 * tab_f.loc["OLS", "RMSE (CV)"]               # el spline capta la no linealidad
assert tab_f["RMSE (CV)"].drop("Solo la media").min() < 0.7 * tab_f.loc["Solo la media", "RMSE (CV)"]
assert ic[0] < rmse_pru < ic[1]
assert razon.loc["I(z_alt ** 2)", "IC95% sup"] < 1                                                          # curvatura negativa: optimo intermedio
````

**Lectura.** (1) **Resumen.** El modelo completo tiene $R^2=0.518$ ($R^2$ ajustado 0.507) y el $F$ rechaza con claridad «ningún predictor sirve» ($p\approx4\times10^{-62}$). Pero solo `sex`, `bmi`, `bp` y `s5` tienen $p<0.001$; `s1` queda en $p=0.058$ y el resto no se distingue de 0 **dado lo demás**. Eso **no** quiere decir que `s1` o `s2` no se relacionen con la respuesta: el VIF de `s1` (59.2) y `s2` (39.2) delata que miden casi lo mismo, y con predictores tan correlacionados los efectos individuales no se pueden separar (el número de condición, $7.2\times10^3$, avisa lo mismo). (2) **Diagnóstico.** Breusch–Pagan rechaza varianza constante ($p=0.0035$), pero los errores estándar HC3 difieren de los clásicos como máximo en ≈ 4 %: la heterocedasticidad es **estadísticamente detectable y prácticamente inocua** aquí. No hay evidencia contra la normalidad (Jarque–Bera, $p=0.50$) ni autocorrelación (Durbin–Watson = 2.03), y la mayor distancia de Cook es 0.024 (muy lejos de 1). (3) **Simplificación.** Eliminar cinco términos (`age`, `s1`, `s2`, `s4`, `s6`) no empeora el ajuste de forma detectable ($F$ anidada, $p=0.151$); el BIC prefiere el reducido (4 816.8 frente a 4 839.0) y el AIC los deja casi iguales (4 792.3 frente a 4 794.0). (4) **Fuera de muestra.** El $R^2$ honesto es ≈ 0.49 (no 0.518) y el RMSE baja de 77.0 (predecir la media) a 54.6. Los cuatro modelos —OLS completo, reducido, Ridge y GLM Gamma— son **prácticamente indistinguibles** (RMSE 54.6–54.8, con desviación entre pliegues ≈ 2.5–2.9, mayor que las diferencias): ante un empate, el criterio es la **parsimonia** y la interpretabilidad. (5) **Fincas sintéticas.** Los *splines* en la altitud reducen el RMSE de validación de 3.57 a 2.85 (≈ −20 %), mucho más que cambiar de OLS a Ridge o GLM; el mejor es el GLM Gamma con *splines* (2.77). En la prueba reservada (100 fincas) el RMSE es 2.23 (IC 95 % [1.88, 2.60]) y $R^2=0.625$, por encima del 0.571 de la validación, una diferencia compatible con el ruido de una prueba tan pequeña. El GLM de `statsmodels` recupera, con su intervalo, los efectos con los que se simularon los datos (Pastusa ≈ 1.32 veces la media de Capiro, IC [1.26, 1.39], frente a $e^{0.25}=1.28$; manejo Alto frente a Bajo 1.22, IC [1.15, 1.29], frente a $e^{0.20}=1.22$) y una curvatura negativa en la altitud (0.862 por unidad $z^2$).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Dos bibliotecas, dos propósitos** | `statsmodels` **infiere** (EE, IC, pruebas, diagnósticos); `scikit-learn` **predice** (*pipelines*, validación cruzada). Producen el **mismo modelo**: `LinearRegression` = `OLS` y `GammaRegressor(alpha=0)` = `GLM Gamma(log)` |
| **Fórmulas de *patsy*** | `y ~ a + C(g) + np.log(x) + a:b` construye la matriz de diseño; se verificó igual a la armada a mano y las ecuaciones normales reproducen el `summary()` completo |
| **Leer el `summary()`** | $R^2$, $F$ (¿algún predictor?), tabla $t$/$p$/IC (cada efecto **dado los demás**) y pruebas de supuestos. Un $p$ grande **no** dice «no sirve» |
| **Diagnóstico** | VIF (colinealidad: `s1` 59, `s2` 39), Breusch–Pagan (detecta heterocedasticidad pero HC3 difiere ≤ 4 %), Jarque–Bera, Durbin–Watson, apalancamiento y Cook (máx. 0.024) |
| **Comparar modelos** | Prueba $F$ anidada ($p=0.151$), AIC/BIC y, sobre todo, **validación cruzada repetida**: el $R^2$ honesto (0.49) < el de entrenamiento (0.518); cuatro modelos empatados $\Rightarrow$ elegir el más simple |
| **`ColumnTransformer` + `Pipeline`** | Imputación, escalado, *one-hot* (referencia), ordinal y *splines* por grupo de columnas, **ajustados solo con el entrenamiento** de cada pliegue (sin fuga). Los *splines* de la altitud mejoraron ≈ 20 % el RMSE; cambiar de modelo, casi nada |
| **Disciplina de evaluación** | Elegir con la validación, evaluar **una vez** en la prueba reservada y reportar el intervalo (RMSE 2.23, IC [1.88, 2.60]) |
| **Producción** | Categorías desconocidas ($\to$ referencia) y datos faltantes se manejan dentro del *pipeline*; registrar cuántos casos llegan así |
