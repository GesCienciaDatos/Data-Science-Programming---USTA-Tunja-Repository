## 1. Del Modelo Lineal al GLM y el Algoritmo IRLS 🔧

Un **GLM** tiene tres componentes:

1. **Componente aleatorio.** $y_i$ tiene una distribución de la **familia exponencial** con media $\mu_i=E[y_i]$ y varianza $\operatorname{Var}(y_i)=\phi\,V(\mu_i)$. La **función de varianza** $V(\cdot)$ define la familia; $\phi$ es el parámetro de **dispersión** (fijo en 1 para Poisson y Bernoulli).
2. **Componente sistemático.** El **predictor lineal** $\eta_i=\mathbf{x}_i^{\top}\boldsymbol\beta$ (más un *offset* conocido, si lo hay).
3. **Función de enlace.** Una función monótona $g$ con $g(\mu_i)=\eta_i$, es decir $\mu_i=g^{-1}(\eta_i)$.

El modelo lineal es el caso particular normal con enlace identidad. Las combinaciones más usadas:

| Respuesta | Familia | $V(\mu)$ | Enlace canónico | Interpretación de $e^{\beta_j}$ |
|---|---|---|---|---|
| Continua simétrica | Normal | $1$ | identidad | (aditiva: $\beta_j$ es una diferencia de medias) |
| Conteo | **Poisson** | $\mu$ | $\log\mu$ | **razón de tasas** (IRR) |
| Sí/No (0/1) | **Binomial** (Bernoulli) | $\mu(1-\mu)$ | $\operatorname{logit}\mu=\log\frac{\mu}{1-\mu}$ | ***odds ratio*** |
| Positiva y asimétrica | **Gamma** | $\mu^2$ | $1/\mu$ (en la práctica se usa $\log\mu$) | **razón de medias** |
| Conteo sobredisperso | **Binomial negativa** (NB2) | $\mu+\alpha\mu^2$ | $\log\mu$ | razón de tasas |

**Por qué no basta MCO.** Con conteos o proporciones la varianza **depende de la media** (heterocedasticidad inherente) y las predicciones lineales pueden salirse del rango válido (negativas, o fuera de $[0,1]$). El GLM modela la **media** con el enlace adecuado y la **varianza** con $V(\mu)$.

**Ajuste: mínimos cuadrados ponderados iterativos (IRLS).** La verosimilitud se maximiza con el método de *scoring* de Fisher. En cada iteración se calculan la **respuesta de trabajo** $z_i$ y los **pesos de trabajo** $w_i$:

$$z_i=\eta_i+(y_i-\mu_i)\,\frac{d\eta_i}{d\mu_i},\qquad w_i=\frac{1}{V(\mu_i)\,\bigl(d\eta_i/d\mu_i\bigr)^2},\qquad \boldsymbol\beta^{(t+1)}=(\mathbf{X}^{\top}\mathbf{W}\mathbf{X})^{-1}\mathbf{X}^{\top}\mathbf{W}\mathbf{z},$$

con $\mathbf{W}=\operatorname{diag}(w_i)$, y se repite hasta que la **desviación** deje de cambiar. La **desviación** $D=2\sum_i[\ell(y_i;y_i)-\ell(y_i;\hat\mu_i)]$ generaliza la suma de cuadrados del error. La matriz de covarianzas aproximada es $\widehat{\operatorname{Cov}}(\hat{\boldsymbol\beta})=\hat\phi\,(\mathbf{X}^{\top}\mathbf{W}\mathbf{X})^{-1}$, con $\hat\phi=\chi^2_P/(n-p)$ (estadístico de Pearson) cuando $\phi$ no está fijo.

Implementamos una clase mínima `Familia` (enlace, varianza y desviación) y la función `irls`, y la probamos con **conteos de plagas** de 400 parcelas **sintéticas**, con un ***offset*** $\log(\text{área})$: si cada parcela tiene un área distinta, el conteo esperado es proporcional al área, $\mu_i=\text{área}_i\cdot e^{\mathbf{x}_i^{\top}\boldsymbol\beta}$, y el modelo estima **tasas por hectárea**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats, optimize
from scipy.special import expit, gammaln
import statsmodels.api as sm

plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})


def _ylogy(y, mu):
    '''y * log(y / mu), con el convenio 0 * log(0) = 0.'''
    return np.where(y > 0, y * np.log(np.where(y > 0, y, 1.0) / mu), 0.0)


class Familia:
    '''Una familia GLM = enlace + funcion de varianza + desviacion.'''
    def __init__(self, nombre, enlace, enlace_inv, dmu_deta, varianza, desviacion, inicio, escala_fija=True):
        self.nombre, self.enlace, self.inv, self.dmu_deta = nombre, enlace, enlace_inv, dmu_deta
        self.varianza, self.desviacion, self.inicio, self.escala_fija = varianza, desviacion, inicio, escala_fija


def poisson():
    return Familia("Poisson", np.log, np.exp, np.exp, lambda mu: mu,
                   lambda y, mu: 2 * np.sum(_ylogy(y, mu) - (y - mu)), lambda y: (y + y.mean()) / 2)


def binomial():
    return Familia("Binomial", lambda mu: np.log(mu / (1 - mu)), expit, lambda eta: expit(eta) * (1 - expit(eta)),
                   lambda mu: mu * (1 - mu), lambda y, mu: 2 * np.sum(_ylogy(y, mu) + _ylogy(1 - y, 1 - mu)),
                   lambda y: (y + 0.5) / 2)


def gamma_log():
    return Familia("Gamma", np.log, np.exp, np.exp, lambda mu: mu ** 2,
                   lambda y, mu: 2 * np.sum(-np.log(y / mu) + (y - mu) / mu), lambda y: (y + y.mean()) / 2, escala_fija=False)


def irls(X, y, familia, offset=None, tol=1e-10, max_iter=100):
    '''Ajusta un GLM por IRLS (scoring de Fisher). Devuelve un diccionario con el ajuste.'''
    X, y = np.asarray(X, float), np.asarray(y, float)
    off = np.zeros(len(y)) if offset is None else np.asarray(offset, float)
    mu = familia.inicio(y)
    eta = familia.enlace(mu)
    dev_ant = np.inf
    for it in range(1, max_iter + 1):
        d = familia.dmu_deta(eta)
        z = eta - off + (y - mu) / d                       # respuesta de trabajo (sin el offset)
        w = d ** 2 / familia.varianza(mu)                  # pesos de trabajo
        beta = np.linalg.solve((X.T * w) @ X, (X.T * w) @ z)
        eta = X @ beta + off
        mu = familia.inv(eta)
        dev = familia.desviacion(y, mu)
        if abs(dev - dev_ant) < tol * (abs(dev) + 0.1):
            break
        dev_ant = dev
    d = familia.dmu_deta(eta)
    w = d ** 2 / familia.varianza(mu)
    n, p = X.shape
    pearson = np.sum((y - mu) ** 2 / familia.varianza(mu))
    phi = 1.0 if familia.escala_fija else pearson / (n - p)
    cov = phi * np.linalg.inv((X.T * w) @ X)
    return dict(beta=beta, mu=mu, eta=eta, desviacion=dev, iteraciones=it, w=w, phi=phi, pearson=pearson,
                gl_resid=n - p, cov=cov, ee=np.sqrt(np.diag(cov)), X=X, y=y, familia=familia)


def generar_plagas(n=400, semilla=7):
    '''SINTETICO. Conteo de plagas ~ Poisson(area * exp(0.3 + 0.5*humedad - 0.3*altitud - 0.7*insecticida)).'''
    rng = np.random.default_rng(semilla)
    d = pd.DataFrame({"humedad": rng.normal(0, 1, n), "altitud": rng.normal(0, 1, n),
                      "insecticida": rng.binomial(1, 0.5, n), "area_ha": rng.uniform(0.5, 3.0, n)})
    mu = d["area_ha"] * np.exp(0.3 + 0.5 * d["humedad"] - 0.3 * d["altitud"] - 0.7 * d["insecticida"])
    d["plagas"] = rng.poisson(mu)
    return d


plagas = generar_plagas()
X_p = sm.add_constant(plagas[["humedad", "altitud", "insecticida"]]).to_numpy()
y_p, off_p = plagas["plagas"].to_numpy(float), np.log(plagas["area_ha"].to_numpy())

res_p = irls(X_p, y_p, poisson(), offset=off_p)
ref_p = sm.GLM(y_p, X_p, family=sm.families.Poisson(), offset=off_p).fit()

nombres = ["intercepto", "humedad", "altitud", "insecticida"]
tab = pd.DataFrame({"beta": res_p["beta"], "EE": res_p["ee"], "z": res_p["beta"] / res_p["ee"],
                    "IRR=exp(beta)": np.exp(res_p["beta"]),
                    "IC95% inf": np.exp(res_p["beta"] - 1.96 * res_p["ee"]), "IC95% sup": np.exp(res_p["beta"] + 1.96 * res_p["ee"])}, index=nombres)
print(tab.round(3).to_string())
print(f"\nIteraciones IRLS: {res_p['iteraciones']} | desviacion = {res_p['desviacion']:.3f} | chi2 de Pearson / gl = {res_p['pearson'] / res_p['gl_resid']:.3f}")

assert np.allclose(res_p["beta"], ref_p.params, atol=1e-7) and np.allclose(res_p["ee"], ref_p.bse, atol=1e-7)
assert np.isclose(res_p["desviacion"], ref_p.deviance, rtol=1e-9) and np.isclose(res_p["pearson"], ref_p.pearson_chi2, rtol=1e-8)
````

## 2. Logística y Gamma: Interpretar y Comparar con MCO 📊

**Regresión logística (Binomial, enlace logit).** Con $y_i\in\{0,1\}$ y $p_i=P(y_i=1)$, el modelo es $\log\frac{p_i}{1-p_i}=\mathbf{x}_i^{\top}\boldsymbol\beta$. El cociente $\frac{p}{1-p}$ son los ***odds***; $e^{\beta_j}$ es el ***odds ratio*** al subir una unidad $x_j$, con las demás fijas. **No** es una razón de probabilidades (solo lo aproxima cuando el evento es raro). Para contrastar el modelo completo con el nulo (solo intercepto) se usa la prueba de razón de verosimilitudes: $D_0-D\sim\chi^2_{q}$ con $q$ el número de predictores. Aquí simulamos el **ataque de un hongo** (sí/no) según lluvia y temperatura estandarizadas.

**Regresión Gamma (enlace log).** Para respuestas positivas y asimétricas con coeficiente de variación aproximadamente constante ($\operatorname{Var}=\phi\mu^2$). Dos alternativas habituales —y problemáticas— son (a) MCO sobre $y$, que ignora que la varianza crece con la media, y (b) MCO sobre $\log y$ con la inversa $e^{\hat y}$ para volver a la escala original: por la **desigualdad de Jensen**, $E[e^{\log y}]\ne e^{E[\log y]}$, así que se **subestima la media** de $y$ (sesgo de retransformación). El GLM Gamma con enlace log modela **directamente $\log E[y]$**: sin ese sesgo y con la varianza correcta. Simulamos el **ingreso** (en millones de pesos, **inventado**) de 400 fincas.

````python
# ---------- Logistica: ataque de un hongo (sintetico) ----------
rng = np.random.default_rng(21)
n_h = 500
lluvia, temp = rng.normal(size=n_h), rng.normal(size=n_h)
hongo = rng.binomial(1, expit(-0.5 + 1.0 * lluvia - 0.8 * temp)).astype(float)
X_h = np.column_stack([np.ones(n_h), lluvia, temp])

res_h = irls(X_h, hongo, binomial())
ref_h = sm.GLM(hongo, X_h, family=sm.families.Binomial()).fit()
assert np.allclose(res_h["beta"], ref_h.params, atol=1e-7) and np.allclose(res_h["ee"], ref_h.bse, atol=1e-7)
assert np.allclose(res_h["beta"], sm.Logit(hongo, X_h).fit(disp=0).params, atol=1e-6)    # Logit = GLM binomial con enlace logit

OR = np.exp(res_h["beta"])
lo, hi = np.exp(res_h["beta"] - 1.96 * res_h["ee"]), np.exp(res_h["beta"] + 1.96 * res_h["ee"])
dev0 = irls(X_h[:, :1], hongo, binomial())["desviacion"]
lrt = dev0 - res_h["desviacion"]
print("Logistica (hongo ~ lluvia + temperatura)")
for nm, o, a, b in zip(["intercepto", "lluvia", "temperatura"], OR, lo, hi):
    print(f"   {nm:12s} OR = {o:5.2f}   IC95% [{a:4.2f}, {b:4.2f}]")
print(f"   Prueba de razon de verosimilitudes vs modelo nulo: chi2 = {lrt:.1f} (2 gl), p = {stats.chi2.sf(lrt, 2):.2e}")
assert np.isclose(lrt, ref_h.null_deviance - ref_h.deviance, rtol=1e-8) and lo[1] > 1 > hi[2]

# ---------- Gamma: ingreso por finca (sintetico, millones de pesos inventados) ----------
rng = np.random.default_rng(5)
n_g = 400
area_z, coop = rng.normal(size=n_g), rng.binomial(1, 0.5, n_g)
mu_g = np.exp(1.0 + 0.4 * area_z + 0.3 * coop)
ingreso = rng.gamma(shape=4.0, scale=mu_g / 4.0)                    # media mu_g, CV = 1/sqrt(4) = 0.5
X_g = np.column_stack([np.ones(n_g), area_z, coop])

res_g = irls(X_g, ingreso, gamma_log())
ref_g = sm.GLM(ingreso, X_g, family=sm.families.Gamma(sm.families.links.Log())).fit()
assert np.allclose(res_g["beta"], ref_g.params, atol=1e-7) and np.allclose(res_g["ee"], ref_g.bse, atol=1e-6)
assert np.isclose(res_g["phi"], ref_g.scale, rtol=1e-6)

b_mco = np.linalg.lstsq(X_g, ingreso, rcond=None)[0]                 # MCO sobre y
b_log = np.linalg.lstsq(X_g, np.log(ingreso), rcond=None)[0]         # MCO sobre log(y)
res_log = np.log(ingreso) - X_g @ b_log
pred = {"MCO sobre y": X_g @ b_mco, "MCO sobre log y, e^yhat": np.exp(X_g @ b_log),
        "... con correccion de Duan": np.exp(X_g @ b_log) * np.mean(np.exp(res_log)), "GLM Gamma (log)": res_g["mu"]}
print(f"\nGamma: dispersion phi = {res_g['phi']:.3f} (verdadera 1/forma = 0.25) | razon de medias de 'coop' = {np.exp(res_g['beta'][2]):.3f} (verdadera e^0.3 = {np.exp(0.3):.3f})")
print(f"{'metodo':28s} {'media(pred)/media(y)':>20s} {'RMSE':>7s}")
for k, v in pred.items():
    print(f"{k:28s} {v.mean() / ingreso.mean():20.3f} {np.sqrt(np.mean((ingreso - v) ** 2)):7.3f}")
print(f"MCO sobre y: {(pred['MCO sobre y'] <= 0).sum()} predicciones <= 0")

fig, ax = plt.subplots(1, 2, figsize=(11, 3.6))
ax[0].scatter(pred["MCO sobre y"], np.abs(ingreso - pred["MCO sobre y"]), s=10, alpha=0.5, color="#dc2626")
ax[0].set_title("MCO sobre $y$: |residuo| crece con el ajuste"); ax[0].set_xlabel("valor ajustado"); ax[0].set_ylabel("|residuo|")
pear = (ingreso - res_g["mu"]) / res_g["mu"]
ax[1].scatter(res_g["mu"], pear, s=10, alpha=0.5, color="#1e3a8a")
ax[1].axhline(0, color="k", lw=1); ax[1].set_title("Gamma: residuo de Pearson sin embudo"); ax[1].set_xlabel("valor ajustado")
plt.tight_layout(); plt.show()

embudo_mco = np.corrcoef(pred["MCO sobre y"], np.abs(ingreso - pred["MCO sobre y"]))[0, 1]
embudo_gamma = np.corrcoef(res_g["mu"], np.abs(pear))[0, 1]
print(f"correlacion (ajuste, |residuo|): MCO = {embudo_mco:.2f} | Gamma = {embudo_gamma:.2f}")
assert embudo_mco > 0.3 and abs(embudo_gamma) < 0.15
assert pred["MCO sobre log y, e^yhat"].mean() < 0.93 * ingreso.mean()                      # sesgo de retransformacion
assert abs(pred["... con correccion de Duan"].mean() / ingreso.mean() - 1) < 0.05
assert abs(pred["GLM Gamma (log)"].mean() / ingreso.mean() - 1) < 0.02
````

**Lectura.** (1) **Plagas (Poisson):** el IRLS converge en pocas iteraciones y reproduce `statsmodels` en coeficientes, errores estándar, desviación y $\chi^2$ de Pearson. Cada desviación estándar más de humedad multiplica la tasa de plagas por $\approx 1.6$ y el insecticida la reduce a $\approx 0.5$ veces; el intervalo del insecticida no contiene a 1. El *offset* hace que el modelo hable de **plagas por hectárea**. (2) **Logística:** `Logit` y el GLM binomial coinciden; la lluvia aumenta los *odds* y la temperatura los reduce, y la razón de verosimilitudes rechaza claramente el modelo nulo. Recuerda: un *odds ratio* de 2.7 **no** significa «2.7 veces más probable». (3) **Gamma:** el embudo de MCO se ve en el gráfico (el residuo crece con el valor ajustado) mientras el residuo de Pearson del GLM no lo muestra. La retransformación ingenua $e^{\hat y}$ **subestima la media** (≈ 0.89 de la media observada; la teoría predice el factor $e^{\psi(4)-\log4}\approx0.88$ para forma 4); la corrección de Duan o el GLM Gamma la eliminan. Los **coeficientes** de MCO sobre $\log y$ siguen siendo razonables (el sesgo está en los **niveles**), pero la varianza y los intervalos de nivel solo los da bien el GLM.

## 3. Sobredispersión y Binomial Negativa 📈

En el modelo de **Poisson** la varianza es **igual a la media**: $\operatorname{Var}(y)=\mu$. Los conteos reales suelen tener **más** variabilidad (parcelas con focos de plaga, heterogeneidad no medida): $\operatorname{Var}(y)>\mu$, la ***sobredispersión***. Los **coeficientes** de Poisson siguen siendo consistentes (si la media está bien especificada), pero los **errores estándar son demasiado pequeños**: los intervalos son demasiado angostos y las pruebas, demasiado optimistas.

**Cómo detectarla.** (i) El cociente de Pearson $\hat\phi=\chi^2_P/(n-p)$ debería ser $\approx1$. (ii) La **prueba de Cameron–Trivedi**: se regresa $\bigl[(y_i-\hat\mu_i)^2-y_i\bigr]/\hat\mu_i$ sobre $\hat\mu_i$ **sin intercepto**; el coeficiente estima $\alpha$ en $\operatorname{Var}(y)=\mu+\alpha\mu^2$ y un $t$ grande indica $\alpha>0$.

**Tres remedios.**
1. ***Casi-Poisson*:** mismos coeficientes, errores estándar multiplicados por $\sqrt{\hat\phi}$.
2. **Errores estándar robustos (sándwich):** $\widehat{\operatorname{Cov}}(\hat{\boldsymbol\beta})=\mathbf{A}^{-1}\mathbf{B}\mathbf{A}^{-1}$, con $\mathbf{A}=\mathbf{X}^{\top}\mathbf{W}\mathbf{X}$ y $\mathbf{B}=\sum_i s_i^2\,\mathbf{x}_i\mathbf{x}_i^{\top}$, donde $s_i=\frac{d\mu_i/d\eta_i}{V(\mu_i)}(y_i-\mu_i)$ es la contribución de la observación $i$ al *score*. No supone la forma de la varianza.
3. **Binomial negativa (NB2):** modela la sobredispersión con $\operatorname{Var}(y)=\mu+\alpha\mu^2$ (una mezcla Poisson–Gamma). Dado $\alpha$, los coeficientes se obtienen con el mismo IRLS (con $V(\mu)=\mu+\alpha\mu^2$); $\alpha$ se estima por **máxima verosimilitud de perfil**: se maximiza sobre $\alpha$ la verosimilitud evaluada en el $\hat{\boldsymbol\beta}(\alpha)$ del IRLS. La log-verosimilitud es $\ell=\sum_i\bigl[\log\Gamma(y_i+\tfrac1\alpha)-\log\Gamma(\tfrac1\alpha)-\log y_i!+\tfrac1\alpha\log\frac{1/\alpha}{1/\alpha+\mu_i}+y_i\log\frac{\mu_i}{1/\alpha+\mu_i}\bigr]$. Poisson es el caso $\alpha\to0$.

Para comparar Poisson y NB se usa el **AIC** $=-2\ell+2k$ (se desarrolla en el cuaderno 02). Simulamos 400 parcelas con $\alpha=0.8$ y evaluamos **por simulación** qué tan cierto es el «95 %» de cada intervalo cuando se ignora la sobredispersión.

````python
def binomial_negativa(a):
    '''NB2 con alfa conocido: V(mu) = mu + a mu^2, enlace log.'''
    dev = lambda y, mu: 2 * np.sum(_ylogy(y, mu) - (y + 1 / a) * np.log((1 + a * y) / (1 + a * mu)))
    return Familia("NB2", np.log, np.exp, np.exp, lambda mu: mu + a * mu ** 2, dev, lambda y: (y + y.mean()) / 2)


def ll_poisson(y, mu):
    return np.sum(y * np.log(mu) - mu - gammaln(y + 1))


def ll_nb(y, mu, a):
    r = 1 / a
    return np.sum(gammaln(y + r) - gammaln(r) - gammaln(y + 1) + r * np.log(r / (r + mu)) + y * np.log(mu / (r + mu)))


def ajustar_nb(X, y):
    '''Maxima verosimilitud de perfil: para cada alfa, IRLS; se optimiza log(alfa) en 1-D.'''
    perfil = lambda la: -ll_nb(y, irls(X, y, binomial_negativa(np.exp(la)))["mu"], np.exp(la))
    la = optimize.minimize_scalar(perfil, bounds=(-6, 3), method="bounded", options={"xatol": 1e-9}).x
    ajuste = irls(X, y, binomial_negativa(np.exp(la)))
    ajuste["alfa"] = np.exp(la)
    ajuste["ll"] = ll_nb(y, ajuste["mu"], np.exp(la))
    return ajuste


def sandwich(res):
    '''Covarianza robusta HC0 de un GLM: A^-1 B A^-1.'''
    f, X = res["familia"], res["X"]
    s = f.dmu_deta(res["eta"]) * (res["y"] - res["mu"]) / f.varianza(res["mu"])
    pan = np.linalg.inv((X.T * res["w"]) @ X)
    return pan @ ((X.T * s ** 2) @ X) @ pan


def prueba_ct(y, mu):
    '''Cameron-Trivedi: coeficiente (alfa) y t de la regresion auxiliar sin intercepto.'''
    f = sm.OLS(((y - mu) ** 2 - y) / mu, mu).fit()
    return f.params[0], f.tvalues[0]


rng = np.random.default_rng(11)
n_s, alfa_verd = 400, 0.8
hum, alt = rng.normal(size=n_s), rng.normal(size=n_s)
X_s = np.column_stack([np.ones(n_s), hum, alt])
mu_s = np.exp(1.0 + 0.5 * hum - 0.3 * alt)
y_s = rng.negative_binomial(1 / alfa_verd, 1 / (1 + alfa_verd * mu_s)).astype(float)       # NB2 con alfa = 0.8

res_ps = irls(X_s, y_s, poisson())
phi_hat = res_ps["pearson"] / res_ps["gl_resid"]
a_ct, t_ct = prueba_ct(y_s, res_ps["mu"])
print(f"Poisson: chi2 de Pearson / gl = {phi_hat:.2f} (esperado 1) | Cameron-Trivedi: alfa = {a_ct:.2f}, t = {t_ct:.1f}")

# Sandwich == statsmodels HC0
ref_hc = sm.GLM(y_s, X_s, family=sm.families.Poisson()).fit(cov_type="HC0")
ee_sand = np.sqrt(np.diag(sandwich(res_ps)))
assert np.allclose(ee_sand, ref_hc.bse, rtol=1e-6)

# NB desde cero vs statsmodels (MLE conjunta de beta y alfa)
res_nb = ajustar_nb(X_s, y_s)
ref_nb = sm.NegativeBinomial(y_s, X_s).fit(disp=0)
assert np.isclose(res_nb["alfa"], ref_nb.params[-1], rtol=1e-3) and np.isclose(res_nb["ll"], ref_nb.llf, rtol=1e-8)
assert np.allclose(res_nb["beta"], ref_nb.params[:-1], atol=1e-4)
glm_nb = sm.GLM(y_s, X_s, family=sm.families.NegativeBinomial(alpha=res_nb["alfa"])).fit()
assert np.allclose(res_nb["beta"], glm_nb.params, atol=1e-6) and np.allclose(res_nb["ee"], glm_nb.bse, atol=1e-6)

aic_p, aic_nb = -2 * ll_poisson(y_s, res_ps["mu"]) + 2 * 3, -2 * res_nb["ll"] + 2 * 4
assert np.isclose(aic_nb, ref_nb.aic, rtol=1e-8)
tab = pd.DataFrame({"beta Poisson": res_ps["beta"], "EE Poisson": res_ps["ee"], "EE casi-Poisson": res_ps["ee"] * np.sqrt(phi_hat),
                    "EE sandwich": ee_sand, "beta NB": res_nb["beta"], "EE NB": res_nb["ee"]}, index=["intercepto", "humedad", "altitud"])
print("\n" + tab.round(3).to_string())
print(f"\nalfa NB = {res_nb['alfa']:.3f} (verdadero {alfa_verd}) | AIC Poisson = {aic_p:.1f} | AIC NB = {aic_nb:.1f}")
assert phi_hat > 2 and t_ct > 3 and aic_nb < aic_p - 50

# Cobertura del IC 95% de beta_humedad (verdadero 0.5) con n = 100, 1000 replicas
def cobertura_poisson(reps=1000, n=100, semilla=0):
    r = np.random.default_rng(semilla)
    aciertos = np.zeros(3)
    for _ in range(reps):
        h, a = r.normal(size=(2, n))
        X = np.column_stack([np.ones(n), h, a])
        m = np.exp(1.0 + 0.5 * h - 0.3 * a)
        y = r.negative_binomial(1 / alfa_verd, 1 / (1 + alfa_verd * m)).astype(float)
        ajuste = irls(X, y, poisson())
        ee = [ajuste["ee"][1], ajuste["ee"][1] * np.sqrt(ajuste["pearson"] / ajuste["gl_resid"]), np.sqrt(sandwich(ajuste)[1, 1])]
        aciertos += [abs(ajuste["beta"][1] - 0.5) < 1.96 * e for e in ee]
    return {k: float(v) for k, v in zip(["Poisson ingenuo", "casi-Poisson", "sandwich HC0"], aciertos / reps)}


cob = cobertura_poisson()
print("\nCobertura empirica del IC 95% (datos NB, n = 100, 1000 replicas):", {k: round(v, 3) for k, v in cob.items()})
assert cob["Poisson ingenuo"] < 0.80 and cob["casi-Poisson"] > 0.86 and cob["sandwich HC0"] > 0.86
````

## 4. Métodos Robustos y No Paramétricos: Cuantiles, Huber y Theil–Sen 🛡️

MCO minimiza $\sum e_i^2$: un solo valor extremo puede **arrastrar** la recta, y su eficiencia depende de errores normales y de varianza constante. Tres alternativas que no suponen una distribución concreta del error:

1. **Regresión por cuantiles** (Koenker–Bassett). Para $\tau\in(0,1)$ minimiza la **pérdida «check»** $\sum_i\rho_\tau(e_i)$, con $\rho_\tau(u)=u\,(\tau-\mathbf{1}\{u<0\})$. Con $\tau=0.5$ es la **regresión de la mediana** ($\sum|e_i|$). Estima el **cuantil condicional** $Q_\tau(y\mid x)$: no solo el centro, sino también los extremos (útil con varianza no constante). Es un **problema de programación lineal**: con $e_i=u_i-v_i$ ($u_i,v_i\ge0$) se minimiza $\tau\sum u_i+(1-\tau)\sum v_i$ sujeto a $\mathbf{X}\boldsymbol\beta+\mathbf{u}-\mathbf{v}=\mathbf{y}$.
2. **M-estimador de Huber.** Minimiza $\sum\rho(e_i/s)$ con $\rho$ cuadrática para residuos pequeños y **lineal** para los grandes. Se resuelve por IRLS con pesos $w_i=\min\{1,\,k/|e_i/s|\}$, $k=1.345$ (≈ 95 % de eficiencia bajo normalidad) y escala $s=\operatorname{mediana}|e_i|/0.6745$.
3. **Theil–Sen** (regresión simple). La pendiente es la **mediana de todas las pendientes entre pares** de puntos, $\hat\beta_1=\operatorname{mediana}\{(y_j-y_i)/(x_j-x_i)\}$; su **punto de ruptura** es ≈ 29 %. La ordenada (convención de SciPy) es $\operatorname{mediana}(y)-\hat\beta_1\operatorname{mediana}(x)$.

**Costo honesto.** Estos métodos pierden algo de eficiencia cuando los errores son normales y limpios. La simulación de abajo mide ambos escenarios: **sin atípicos** y con **10 % de contaminación**. Datos **sintéticos**: rendimiento frente a fertilizante, con varianza creciente y 8 atípicos gruesos.

````python
def cuantil_lp(X, y, tau):
    '''Regresion por cuantiles via programacion lineal (solver HiGHS).'''
    n, p = X.shape
    c = np.r_[np.zeros(p), tau * np.ones(n), (1 - tau) * np.ones(n)]
    A = np.c_[X, np.eye(n), -np.eye(n)]
    r = optimize.linprog(c, A_eq=A, b_eq=y, bounds=[(None, None)] * p + [(0, None)] * (2 * n), method="highs")
    return r.x[:p]


def huber_irls(X, y, k=1.345, tol=1e-12, max_iter=500):
    '''M-estimador de Huber: pesos min(1, k/|u|), escala = mediana(|e|)/0.6745 (como statsmodels).'''
    b = np.linalg.lstsq(X, y, rcond=None)[0]
    for _ in range(max_iter):
        e = y - X @ b
        s = np.median(np.abs(e)) / 0.6744897501960817
        u = np.abs(e / s)
        w = np.where(u <= k, 1.0, k / np.maximum(u, 1e-12))
        b_nuevo = np.linalg.solve((X.T * w) @ X, (X.T * w) @ y)
        if np.abs(b_nuevo - b).max() < tol:
            return b_nuevo
        b = b_nuevo
    return b


def theil_sen(x, y):
    '''Pendiente = mediana de las pendientes entre pares; ordenada = mediana(y) - pendiente * mediana(x).'''
    i, j = np.triu_indices(len(x), 1)
    pend = np.median((y[j] - y[i]) / (x[j] - x[i]))
    return np.median(y) - pend * np.median(x), pend


rng = np.random.default_rng(11)
n_c = 150
x_c = rng.uniform(0, 10, n_c)
y_c = 2 + 1.5 * x_c + rng.normal(0, 1 + 0.4 * x_c)                      # verdadero: ordenada 2, pendiente 1.5
atip = rng.choice(n_c, 8, replace=False)
y_c[atip] += rng.choice([-1, 1], 8) * rng.uniform(25, 40, 8)           # 8 atipicos gruesos
X_c = np.column_stack([np.ones(n_c), x_c])

b_mco = np.linalg.lstsq(X_c, y_c, rcond=None)[0]
b_hub, b_med = huber_irls(X_c, y_c), cuantil_lp(X_c, y_c, 0.5)
b_ts = np.array(theil_sen(x_c, y_c))

# Verificacion contra statsmodels y SciPy
for tau in (0.1, 0.5, 0.9):
    assert np.allclose(cuantil_lp(X_c, y_c, tau), sm.QuantReg(y_c, X_c).fit(q=tau, max_iter=5000, p_tol=1e-10).params, atol=1e-4)
rlm = sm.RLM(y_c, X_c, M=sm.robust.norms.HuberT()).fit(conv="coefs", tol=1e-12, maxiter=500)
assert np.allclose(b_hub, rlm.params, atol=1e-7)
ts = stats.theilslopes(y_c, x_c)
assert np.isclose(b_ts[1], ts.slope) and np.isclose(b_ts[0], ts.intercept)

print("Pendiente (verdadera 1.5):", {k: round(float(v[1]), 3) for k, v in {"MCO": b_mco, "Huber": b_hub, "mediana": b_med, "Theil-Sen": b_ts}.items()})

fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.scatter(x_c, y_c, s=12, alpha=0.5, color="#64748b")
xx = np.linspace(0, 10, 50)
for b, nm, col in [(b_mco, "MCO", "#dc2626"), (b_hub, "Huber", "#16a34a"), (b_med, "mediana (τ=0.5)", "#1e3a8a")]:
    ax.plot(xx, b[0] + b[1] * xx, color=col, lw=2, label=nm)
for tau in (0.1, 0.9):
    b = cuantil_lp(X_c, y_c, tau)
    ax.plot(xx, b[0] + b[1] * xx, color="#1e3a8a", lw=1.2, ls="--", label=f"cuantil τ={tau}")
ax.set_xlabel("fertilizante (unidades inventadas)"); ax.set_ylabel("rendimiento"); ax.set_title("MCO arrastrado por atípicos; los cuantiles capturan el abanico")
ax.legend(fontsize=8); plt.tight_layout(); plt.show()


def simular(contaminado, reps=200, n=60, semilla=0):
    r = np.random.default_rng(semilla)
    est = {"MCO": [], "Huber": [], "Mediana": [], "Theil-Sen": []}
    for _ in range(reps):
        x = r.uniform(0, 10, n)
        e = r.normal(0, 2, n)
        if contaminado:
            m = r.random(n) < 0.10
            e[m] += r.choice([-1, 1], m.sum()) * r.uniform(15, 30, m.sum())
        y = 2 + 1.5 * x + e
        X = np.column_stack([np.ones(n), x])
        est["MCO"].append(np.linalg.lstsq(X, y, rcond=None)[0][1]); est["Huber"].append(huber_irls(X, y)[1])
        est["Mediana"].append(cuantil_lp(X, y, 0.5)[1]); est["Theil-Sen"].append(theil_sen(x, y)[1])
    return pd.Series({k: np.sqrt(np.mean((np.array(v) - 1.5) ** 2)) for k, v in est.items()})


rmse = pd.DataFrame({"sin atipicos": simular(False), "10% contaminado": simular(True)})
print("\nRMSE de la pendiente (200 replicas, n = 60; menor es mejor):\n" + rmse.round(3).to_string())
assert rmse.loc["MCO", "10% contaminado"] > 2.5 * rmse.loc["Huber", "10% contaminado"]
assert rmse.loc["Huber", "sin atipicos"] < 1.15 * rmse.loc["MCO", "sin atipicos"]
assert rmse.loc["Mediana", "sin atipicos"] > rmse.loc["MCO", "sin atipicos"]
````

## 5. *Bootstrap* para Regresión: Inferencia sin Suponer Varianza Constante 🥾

El IC clásico de un coeficiente, $\hat\beta_1\pm t_{n-2,\,0.975}\,\widehat{\operatorname{EE}}$, descansa en que los errores tengan **varianza constante**. Si la varianza crece con $x$, el $\widehat{\operatorname{EE}}$ clásico es **incorrecto** (en la mayoría de los casos, demasiado pequeño). El ***bootstrap* de pares** remuestrea las **filas** $(x_i,y_i)$ con reemplazo, recalcula $\hat\beta_1^{\ast}$ muchas veces y toma los **percentiles** $2.5\,\%$ y $97.5\,\%$: no usa la fórmula del error estándar ni supone homocedasticidad. (El *bootstrap* de **residuos** conserva los $x$ y es válido bajo homocedasticidad.) Una alternativa analítica es el error estándar **HC3** (sándwich con corrección por apalancamiento).

Probamos las tres con **1 000 muestras sintéticas** de $n=40$, $y=1+2x+\varepsilon$, con $x$ asimétrica y $\operatorname{sd}(\varepsilon)\propto x$, y medimos la **cobertura real** del «95 %». Después aplicamos el *bootstrap* de pares al GLM de Poisson **sobredisperso** de la sección 3 (500 remuestras con nuestro IRLS).

````python
def cobertura_pendiente(reps=1000, n=40, B=400, semilla=1):
    r = np.random.default_rng(semilla)
    cob = {"t clasico": 0, "HC3": 0, "bootstrap de pares": 0}
    for _ in range(reps):
        x = r.lognormal(0, 0.6, n)
        y = 1 + 2 * x + r.normal(0, 1, n) * x                              # sd(error) proporcional a x
        X = np.column_stack([np.ones(n), x])
        b = np.linalg.lstsq(X, y, rcond=None)[0]
        e = y - X @ b
        inv = np.linalg.inv(X.T @ X)
        q = stats.t.ppf(0.975, n - 2)
        ee_clasico = np.sqrt(e @ e / (n - 2) * inv[1, 1])
        h = np.einsum("ij,jk,ik->i", X, inv, X)
        ee_hc3 = np.sqrt((inv @ ((X.T * (e / (1 - h)) ** 2) @ X) @ inv)[1, 1])
        idx = r.integers(0, n, (B, n))                                      # B remuestras de pares, vectorizado
        xb, yb = x[idx], y[idx]
        xc, yc = xb - xb.mean(1, keepdims=True), yb - yb.mean(1, keepdims=True)
        pend = (xc * yc).sum(1) / (xc ** 2).sum(1)
        lo, hi = np.percentile(pend, [2.5, 97.5])
        cob["t clasico"] += abs(b[1] - 2) < q * ee_clasico
        cob["HC3"] += abs(b[1] - 2) < q * ee_hc3
        cob["bootstrap de pares"] += lo < 2 < hi
    return {k: v / reps for k, v in cob.items()}


# HC3 propio == statsmodels (una muestra)
r0 = np.random.default_rng(3)
x0 = r0.lognormal(0, 0.6, 40); y0 = 1 + 2 * x0 + r0.normal(0, 1, 40) * x0
X0 = np.column_stack([np.ones(40), x0]); ajuste0 = sm.OLS(y0, X0).fit(cov_type="HC3")
e0 = y0 - X0 @ ajuste0.params; inv0 = np.linalg.inv(X0.T @ X0); h0 = np.einsum("ij,jk,ik->i", X0, inv0, X0)
assert np.allclose(np.sqrt(np.diag(inv0 @ ((X0.T * (e0 / (1 - h0)) ** 2) @ X0) @ inv0)), ajuste0.bse)

cob = cobertura_pendiente()
print("Cobertura empirica del IC 95% de la pendiente (n = 40, 1000 muestras, varianza ~ x^2):")
for k, v in cob.items():
    print(f"   {k:20s} {v:.3f}")
assert cob["t clasico"] < 0.80 and cob["bootstrap de pares"] > 0.88 and cob["HC3"] > 0.88

# Bootstrap de pares sobre el GLM Poisson sobredisperso de la seccion 3
rb = np.random.default_rng(2024)
beta_b = np.array([irls(X_s[i], y_s[i], poisson())["beta"] for i in (rb.integers(0, n_s, n_s) for _ in range(500))])
ee_boot = beta_b.std(axis=0, ddof=1)
comp = pd.DataFrame({"Poisson ingenuo": res_ps["ee"], "sandwich": ee_sand, "bootstrap de pares": ee_boot, "NB": res_nb["ee"]},
                    index=["intercepto", "humedad", "altitud"])
print("\nErrores estandar de los coeficientes (datos sobredispersos):\n" + comp.round(3).to_string())
assert (ee_boot > 1.3 * res_ps["ee"]).all()
assert np.allclose(ee_boot, ee_sand, rtol=0.25)
````

**Lectura.** (1) **Sobredispersión.** Los datos NB dan $\hat\phi=3.83$ (esperado 1) y la prueba de Cameron–Trivedi estima $\hat\alpha=0.68$ con $t=6.0$. Los coeficientes de Poisson y NB son parecidos (p. ej. humedad 0.52 frente a 0.53), pero el EE de Poisson (0.027) es **la mitad** del robusto (0.054): con $n=100$ la cobertura real del «95 %» es **63.2 %** para el Poisson ingenuo, **89.8 %** para el casi-Poisson y **88.9 %** para el sándwich (que, con $n$ moderado, tampoco llega a 95 %). La NB ($\hat\alpha=0.834$ frente a 0.8 verdadero) gana por AIC por un margen enorme (1760.9 frente a 2257.7). (2) **Robustez.** Con 8 atípicos, MCO se desvía (pendiente 1.333 frente a 1.5), mientras Huber (1.455), la mediana (1.415) y Theil–Sen (1.470) quedan cerca. **Costo honesto:** sin atípicos, Huber (RMSE 0.093) y Theil–Sen (0.092) pierden casi nada frente a MCO (0.091), pero la **mediana** sí pierde eficiencia (0.114); con 10 % de contaminación, MCO (0.334) es ≈ 3 veces peor que Huber (0.109). Los cuantiles 0.1 y 0.9 no son «robustez»: describen el **abanico** de la distribución condicional. (3) ***Bootstrap*.** Con varianza creciente con $x$, el IC $t$ clásico cubre solo **69.0 %**; el *bootstrap* de pares (92.2 %) y HC3 (91.8 %) mejoran mucho, pero quedan por debajo de 95 % con $n=40$: ninguna es magia. En el GLM sobredisperso, el EE *bootstrap* (0.055) coincide con el sándwich (0.054) y duplica al Poisson ingenuo (0.027).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **GLM** | Componente aleatorio (familia exponencial, $V(\mu)$) + predictor lineal + enlace $g(\mu)=\eta$. Poisson/log para conteos, Binomial/logit para 0-1, Gamma/log para positivas asimétricas |
| **IRLS** | $\boldsymbol\beta\leftarrow(\mathbf{X}^{\top}\mathbf{W}\mathbf{X})^{-1}\mathbf{X}^{\top}\mathbf{W}\mathbf{z}$ hasta que la desviación se estabiliza. Implementado desde cero e **igual a `statsmodels`** (Poisson con *offset*, Binomial, Gamma) |
| **Interpretación** | $e^{\beta}$ = razón de tasas (Poisson), ***odds ratio*** (logística; **no** una razón de probabilidades) o razón de medias (Gamma) |
| **Gamma vs. MCO** | MCO sobre $y$ tiene varianza no constante; $e^{\hat y}$ desde $\log y$ **subestima la media** (≈ 0.89 aquí): usar GLM Gamma o la corrección de Duan |
| **Sobredispersión** | Con Var > media, los coeficientes de Poisson valen pero los EE **subestiman**; el IC «95 %» cubre mucho menos. Remedios: casi-Poisson, sándwich, **binomial negativa** (MLE de perfil igual a `statsmodels`); se compara por AIC |
| **Robustos y no paramétricos** | Cuantiles (programación lineal), Huber (IRLS) y Theil–Sen resisten atípicos; **cuestan** algo de eficiencia con datos limpios (la mediana más que Huber y Theil–Sen) |
| ***Bootstrap* de pares** | Inferencia sin suponer homocedasticidad; mejora la cobertura del IC $t$ pero con $n$ pequeño no llega al 95 % exacto |
