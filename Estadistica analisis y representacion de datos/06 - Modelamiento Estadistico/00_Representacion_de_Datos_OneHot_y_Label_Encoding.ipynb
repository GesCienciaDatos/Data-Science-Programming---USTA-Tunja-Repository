## 1. De Categorías a Números: la Matriz de Diseño 🧮

Un modelo lineal escribe la respuesta como $\mathbf{y}=\mathbf{X}\boldsymbol{\beta}+\boldsymbol{\varepsilon}$, donde $\mathbf{X}\in\mathbb{R}^{n\times p}$ es la **matriz de diseño**: una fila por observación y una columna por término del modelo (el intercepto es una columna de unos). El estimador de mínimos cuadrados, $\hat{\boldsymbol\beta}=(\mathbf{X}^{\top}\mathbf{X})^{-1}\mathbf{X}^{\top}\mathbf{y}$, solo existe si $\mathbf{X}^{\top}\mathbf{X}$ es **invertible**, es decir, si las columnas de $\mathbf{X}$ son **linealmente independientes**. Todo lo que sigue gira alrededor de esta frase: **cómo escribir una categoría como columnas de $\mathbf{X}$** sin romper la invertibilidad ni inventar supuestos.

La elección depende de la **escala de medición** de la variable (módulo 00, Stevens):

| Escala | Ejemplo en una finca | Qué relación tienen los valores | Codificación habitual |
|---|---|---|---|
| **Nominal** | variedad (Capiro, Criolla, Pastusa, Sabanera) | Solo *igual/distinto*: **no hay orden** | ***One-hot*** (o contrastes) |
| **Ordinal** | nivel de manejo (Bajo < Medio < Alto) | Hay **orden**, pero **no** distancias conocidas | Códigos ordenados (supone pasos iguales) o *one-hot* |
| **Nominal de alta cardinalidad** | vereda (decenas o cientos de niveles) | Igual/distinto, **muchos** niveles | *Target encoding* validado, frecuencia, agrupar raras |
| **Cuantitativa** | fertilizante (kg/ha) | Distancias con sentido | Se usa tal cual |

Generamos una base **sintética** de 400 fincas donde **conocemos el mecanismo**: la variedad tiene efectos que **no** siguen el orden alfabético, el manejo tiene efectos crecientes pero **no equidistantes** y el fertilizante suma un poco.

````python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from itertools import permutations
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder, TargetEncoder
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_score, train_test_split
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm

plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})


def generar_fincas(n=400, semilla=42):
    '''Fincas SINTETICAS (hipoteticas). Mecanismo verdadero (t/ha):
    rend = 18 + efecto_variedad + efecto_manejo + 0.01*fert + ruido(0, 2.5).'''
    rng = np.random.default_rng(semilla)
    variedad = rng.choice(["Capiro", "Criolla", "Pastusa", "Sabanera"], n, p=[0.30, 0.20, 0.35, 0.15])
    manejo = rng.choice(["Bajo", "Medio", "Alto"], n, p=[0.35, 0.40, 0.25])
    e_var = pd.Series(variedad).map({"Capiro": 0.0, "Criolla": -3.0, "Pastusa": 4.0, "Sabanera": -1.0}).to_numpy()
    e_man = pd.Series(manejo).map({"Bajo": 0.0, "Medio": 2.0, "Alto": 3.0}).to_numpy()
    fert = rng.gamma(4, 50, n)
    rend = 18 + e_var + e_man + 0.01 * fert + rng.normal(0, 2.5, n)
    return pd.DataFrame({"variedad": variedad, "manejo": manejo, "fert_kg_ha": fert, "rend_t_ha": rend})


fincas = generar_fincas()
print(fincas.head(6).round(2).to_string())
print()
print(fincas.groupby("variedad")["rend_t_ha"].agg(["count", "mean", "std"]).round(2).to_string())
print(fincas.groupby("manejo")["rend_t_ha"].mean().reindex(["Bajo", "Medio", "Alto"]).round(2).to_string())

assert fincas.shape == (400, 4) and fincas.isna().sum().sum() == 0
assert set(fincas["variedad"]) == {"Capiro", "Criolla", "Pastusa", "Sabanera"}
````

## 2. Codificación por Etiquetas: Label y Ordinal ⚠️

La codificación más sencilla asigna un entero a cada categoría: $\text{Capiro}\to0$, $\text{Criolla}\to1$, $\text{Pastusa}\to2$, $\text{Sabanera}\to3$. Es lo que hacen `LabelEncoder` (pensado para la **respuesta** $y$ en clasificación) y `OrdinalEncoder` (para **predictores**). Por defecto el orden es **alfabético**, es decir, **arbitrario**.

**El riesgo.** Si metemos ese código como **una sola columna** en un modelo lineal, el modelo es $\hat y=\beta_0+\beta_1\cdot\text{código}$, que impone tres supuestos: (i) las categorías tienen un **orden**; (ii) pasar de la 0 a la 1 «vale lo mismo» que de la 2 a la 3 (**pasos iguales**); (iii) el efecto es **monótono**. Para una variable **nominal** los tres son falsos y el resultado depende de un orden que nadie eligió con criterio. Lo demostramos de tres maneras: (a) implementando el codificador y verificándolo, (b) midiendo el $R^2$ de validación cruzada, y (c) **probando las $4!=24$ asignaciones posibles de códigos**: si la codificación fuera inocua, el $R^2$ no cambiaría.

**Cuando el orden sí es real** (ordinal), el código de una columna es una **simplificación razonable**: usa **1 parámetro** en lugar de $k-1$. Se puede **contrastar** con una prueba $F$ de modelos anidados: $F=\dfrac{(\text{SCE}_0-\text{SCE}_1)/(q)}{\text{SCE}_1/(n-p_1)}$, donde el modelo 0 (código ordinal) está contenido en el modelo 1 (una ficticia por nivel), $q$ es la diferencia de parámetros y $\text{SCE}$ la suma de cuadrados del error.

````python
def codificar_etiquetas(col, categorias=None):
    '''Asigna 0..k-1 a cada categoria. Sin 'categorias' el orden es alfabetico (arbitrario).'''
    cats = np.array(sorted(pd.unique(col))) if categorias is None else np.asarray(categorias)
    return pd.Series(col).map({c: i for i, c in enumerate(cats)}).to_numpy(), cats


cod, cats = codificar_etiquetas(fincas["variedad"])
assert (cod == LabelEncoder().fit_transform(fincas["variedad"])).all()
assert (cod.reshape(-1, 1) == OrdinalEncoder().fit_transform(fincas[["variedad"]])).all()
print("categorias en orden alfabetico:", {str(c): i for i, c in enumerate(cats)})

y = fincas["rend_t_ha"].to_numpy()


def r2_cv(X, y, semilla=0):
    '''R2 medio de validacion cruzada de 5 pliegues (regresion lineal).'''
    cv = KFold(5, shuffle=True, random_state=semilla)
    return cross_val_score(LinearRegression(), X, y, cv=cv, scoring="r2").mean()


X_onehot = OneHotEncoder(drop="first", sparse_output=False).fit_transform(fincas[["variedad"]])
r2_codigo, r2_onehot = r2_cv(cod.reshape(-1, 1), y), r2_cv(X_onehot, y)

# Las 24 asignaciones posibles de codigos 0..3 a las 4 variedades
r2_perm = np.array([r2_cv(fincas["variedad"].map(dict(zip(cats, p))).to_numpy().reshape(-1, 1), y)
                    for p in permutations(range(4))])
print(f"R2 (CV) con UNA columna de codigos: alfabetico = {r2_codigo:.3f} | min/max sobre 24 ordenes = {r2_perm.min():.3f} / {r2_perm.max():.3f}")
print(f"R2 (CV) con one-hot (3 columnas):   {r2_onehot:.3f}  (identico para cualquier orden de las categorias)")

fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
medias = fincas.groupby("variedad")["rend_t_ha"].mean().reindex(cats)
ax[0].scatter(range(4), medias.values, s=90, color="#1e3a8a", zorder=3, label="media real por variedad")
b1, b0 = np.polyfit(cod, y, 1)
ax[0].plot(range(4), b0 + b1 * np.arange(4), color="#dc2626", lw=2, label="recta del label encoding")
ax[0].set_xticks(range(4)); ax[0].set_xticklabels([f"{c}\n(código {i})" for i, c in enumerate(cats)], fontsize=8)
ax[0].set_ylabel("rendimiento medio (t/ha)"); ax[0].set_title("La recta impone un orden y pasos iguales"); ax[0].legend(fontsize=8)
ax[1].hist(r2_perm, bins=10, color="#94a3b8", edgecolor="white", label="24 órdenes posibles (una columna)")
ax[1].axvline(r2_onehot, color="#16a34a", lw=2.5, label="one-hot (3 columnas)")
ax[1].set_xlabel("$R^2$ de validación cruzada"); ax[1].set_title("El orden arbitrario cambia el modelo"); ax[1].legend(fontsize=8)
plt.tight_layout(); plt.show()

assert r2_perm.max() - r2_perm.min() > 0.10 and r2_onehot > r2_perm.max()

# --- Codificacion ORDINAL: aqui el orden SI es real (Bajo < Medio < Alto) ---
orden_manejo = ["Bajo", "Medio", "Alto"]
fincas["manejo_cod"] = OrdinalEncoder(categories=[orden_manejo]).fit_transform(fincas[["manejo"]]).ravel()
assert (fincas["manejo_cod"] == codificar_etiquetas(fincas["manejo"], orden_manejo)[0]).all()
alfabetico = {c: i for i, c in enumerate(sorted(fincas["manejo"].unique()))}
print("\nOrden alfabetico (INCORRECTO para una ordinal):", alfabetico, "-> 'Alto' queda en 0")
assert alfabetico["Alto"] == 0 and alfabetico["Medio"] == 2

m_ord = smf.ols("rend_t_ha ~ manejo_cod + fert_kg_ha", fincas).fit()                       # 1 parametro para el manejo
m_dum = smf.ols("rend_t_ha ~ C(manejo, Treatment('Bajo')) + fert_kg_ha", fincas).fit()      # 2 parametros (una por nivel)
prueba_F = anova_lm(m_ord, m_dum)
b_med = m_dum.params["C(manejo, Treatment('Bajo'))[T.Medio]"]
b_alto = m_dum.params["C(manejo, Treatment('Bajo'))[T.Alto]"]
paso1, paso2 = b_med, b_alto - b_med
print(f"R2 ordinal = {m_ord.rsquared:.3f} | R2 ficticias = {m_dum.rsquared:.3f}")
print(f"Paso Bajo->Medio = {paso1:.2f} t/ha; paso Medio->Alto = {paso2:.2f} t/ha (el codigo ordinal los supone iguales a {m_ord.params['manejo_cod']:.2f})")
print(f"Prueba F anidada: F = {prueba_F['F'].iloc[1]:.2f}, p = {prueba_F['Pr(>F)'].iloc[1]:.3f}")
assert m_dum.rsquared >= m_ord.rsquared and prueba_F["df_diff"].iloc[1] == 1
````

**Lectura.** (1) Con **una sola columna de códigos**, el $R^2$ de validación cruzada depende del orden que le pongamos a las variedades: con el orden alfabético es apenas **0.033** y entre las 24 asignaciones posibles va de **−0.040 a 0.359**, y **ninguna** alcanza al *one-hot* (**0.395**). El modelo con códigos no está «un poco peor»: está resolviendo **otra pregunta** (una recta sobre un eje inventado). (2) En el manejo, donde el orden **sí** es real, el modelo ordinal es una simplificación aceptable *si* los pasos son parecidos; la prueba $F$ anidada es la forma de **comprobarlo** en lugar de suponerlo. Aquí los pasos estimados son 1.71 (Bajo→Medio) y 0.72 t/ha (Medio→Alto), frente a los 2 y 1 verdaderos, y la prueba da $F=1.70$, $p=0.19$: no hay evidencia para abandonar el código ordinal. Pero un $p$ grande **no prueba** que los pasos sean iguales (no lo son: 2 y 1); solo indica que **esta muestra** no puede distinguirlos. Si el orden es dudoso o los pasos claramente distintos, las ficticias son la opción segura, a costa de más parámetros.

## 3. One-Hot y la Trampa de la Variable Ficticia 🪤

El ***one-hot encoding*** crea **una columna indicadora por categoría**: $D_j=1$ si la observación pertenece a la categoría $j$ y $0$ en otro caso. No supone orden ni distancias. Pero hay una trampa: las $k$ indicadoras **suman 1 en todas las filas**, es decir, $D_1+\cdots+D_k=\mathbf{1}$, **idéntica a la columna del intercepto**. Con intercepto y las $k$ ficticias, $\mathbf{X}$ tiene $k+1$ columnas pero **rango $k$**: $\mathbf{X}^{\top}\mathbf{X}$ es **singular**, $\boldsymbol\beta$ no es único (hay infinitas soluciones que producen las **mismas predicciones**).

**Soluciones equivalentes:** (a) **descartar una categoría** (la **referencia**) y conservar el intercepto, o (b) quitar el intercepto y dejar las $k$ ficticias. En (a), con $\hat y=\beta_0+\sum_{j\ne r}\beta_jD_j$:
- $\beta_0$ es la media de la categoría de **referencia** $r$;
- $\beta_j$ es la **diferencia de medias** entre la categoría $j$ y la referencia (ajustada por las demás variables, si las hay).

Cambiar la referencia **cambia los coeficientes**, pero **no las predicciones** ni el $R^2$. Es el origen de `drop='first'` en `OneHotEncoder`, `drop_first=True` en `get_dummies` y del contraste de tratamiento `C(x, Treatment(...))` de *patsy*.

````python
def one_hot(col, categorias=None, descartar=None):
    '''Matriz de indicadoras (n x k) con broadcasting. 'descartar' quita la categoria de referencia.'''
    cats = np.array(sorted(pd.unique(col))) if categorias is None else np.asarray(categorias)
    M = (np.asarray(col)[:, None] == cats[None, :]).astype(float)
    if descartar is not None:
        j = list(cats).index(descartar)
        M, cats = np.delete(M, j, axis=1), np.delete(cats, j)
    return M, cats


M, cats = one_hot(fincas["variedad"])
assert np.array_equal(M, pd.get_dummies(fincas["variedad"]).to_numpy(float))
assert np.array_equal(M, OneHotEncoder(sparse_output=False).fit_transform(fincas[["variedad"]]))
assert (M.sum(axis=1) == 1).all()                            # cada fila tiene exactamente un 1

# --- La trampa: intercepto + las k ficticias ---
X_trampa = np.column_stack([np.ones(len(M)), M])             # 1 + 4 = 5 columnas
rango = np.linalg.matrix_rank(X_trampa)
print(f"columnas = {X_trampa.shape[1]} | rango = {rango} | numero de condicion = {np.linalg.cond(X_trampa):.2e}")
print(f"menor autovalor de X'X = {np.linalg.eigvalsh(X_trampa.T @ X_trampa).min():.2e}  (cero salvo error numerico: X'X es singular)")
assert rango == 4 < X_trampa.shape[1]

# statsmodels no falla: usa la pseudo-inversa y devuelve UNA de las infinitas soluciones
with warnings.catch_warnings(record=True) as avisos:
    warnings.simplefilter("always")
    res_trampa = sm.OLS(y, X_trampa).fit()
print("aviso de statsmodels:", avisos[0].category.__name__)
print("coeficientes de la solucion de norma minima (sin interpretacion unica):", res_trampa.params.round(2))

# --- Solucion: descartar la referencia ---
def ajustar_ols(X, y):
    Xc = np.column_stack([np.ones(len(X)), X])
    beta = np.linalg.lstsq(Xc, y, rcond=None)[0]
    return beta, Xc @ beta


medias = fincas.groupby("variedad")["rend_t_ha"].mean()
resultados = {}
for ref in ["Capiro", "Pastusa"]:
    Mr, cr = one_hot(fincas["variedad"], descartar=ref)
    beta, ajust = ajustar_ols(Mr, y)
    resultados[ref] = (beta, ajust, cr)
    print(f"\nReferencia = {ref}: intercepto = {beta[0]:.3f} (media de {ref} = {medias[ref]:.3f})")
    for c, b in zip(cr, beta[1:]):
        print(f"   {c:9s} beta = {b:+.3f}   (media {c} - media {ref} = {medias[c] - medias[ref]:+.3f})")
        assert abs(b - (medias[c] - medias[ref])) < 1e-9
    assert abs(beta[0] - medias[ref]) < 1e-9

# Mismas predicciones, aunque los coeficientes cambien
assert np.allclose(resultados["Capiro"][1], resultados["Pastusa"][1])
assert np.allclose(resultados["Capiro"][1], res_trampa.fittedvalues)
print("\nPredicciones identicas con cualquier referencia y con la solucion 'trampa' -> solo cambia la interpretacion de los coeficientes.")

# statsmodels/patsy: el contraste de tratamiento hace lo mismo, y '- 1' elimina el intercepto (coeficientes = medias por grupo)
f_tr = smf.ols("rend_t_ha ~ C(variedad, Treatment('Pastusa'))", fincas).fit()
assert np.allclose(f_tr.params.to_numpy(), resultados["Pastusa"][0], atol=1e-9)
f_sin = smf.ols("rend_t_ha ~ C(variedad) - 1", fincas).fit()
assert np.allclose(f_sin.params.to_numpy(), medias.to_numpy())
print("patsy 'C(variedad) - 1' devuelve las medias por variedad:", f_sin.params.round(2).to_numpy())
````

## 4. Target Encoding y la Fuga de Información 💧

Cuando una variable nominal tiene **muchos niveles** (60 veredas, 500 municipios), el *one-hot* crea decenas de columnas con pocos datos cada una. El ***target encoding*** reemplaza cada categoría por **un solo número**: la media de la respuesta en esa categoría. Con suavizado hacia la media global $\bar y$:

$$\tilde{x}_c=\frac{n_c\,\bar{y}_c+m\,\bar{y}}{n_c+m},$$

donde $n_c$ es el número de observaciones de la categoría $c$, $\bar y_c$ su media y $m\ge0$ el **peso del prior** (cuántas observaciones «virtuales» de la media global aportamos). Con $m=0$ no hay suavizado; con $m$ grande todas las categorías se acercan a $\bar y$. Es un **estimador de contracción** (*shrinkage*): las categorías con pocos datos, cuya media es muy ruidosa, se acercan a la media global.

**El peligro: la fuga (*leakage*).** Si $\tilde x_c$ se calcula con **todas** las filas de entrenamiento y luego se usa como predictor **para esas mismas filas**, la respuesta $y_i$ ya está **dentro** de su propio predictor: en una categoría con $n_c=1$ el predictor *es* $y_i$. El modelo parece excelente en entrenamiento, pero en datos nuevos no existe esa pista. La solución estándar es el ***out-of-fold*** (codificar cada fila con un valor calculado **sin** esa fila, mediante validación cruzada interna) y/o suavizar. `scikit-learn.preprocessing.TargetEncoder` aplica este esquema en `fit_transform`.

**Experimento.** Simulamos 600 observaciones con 60 veredas y comparamos, en **40 particiones aleatorias** entrenamiento/prueba, cuatro formas de usar la vereda, en dos mundos: **(A)** la vereda **no tiene efecto** real y **(B)** la vereda **sí** lo tiene. Son las dos caras del mismo problema: la codificación debe ayudar cuando hay señal y **no inventarla** cuando no la hay.

````python
def generar_veredas(n=600, n_veredas=60, sd_efecto=0.0, semilla=3):
    '''SINTETICO. y = 20 + 1.5 x + efecto_vereda + ruido(0, 2). sd_efecto=0 -> la vereda NO importa.'''
    rng = np.random.default_rng(semilla)
    v = rng.integers(0, n_veredas, n)
    efectos = rng.normal(0, sd_efecto, n_veredas)
    x = rng.normal(size=n)
    return pd.DataFrame({"vereda": [f"V{i:02d}" for i in v], "x": x, "y": 20 + 1.5 * x + efectos[v] + rng.normal(0, 2, n)})


def te_ajustar(cat, y, m=0.0):
    '''Media suavizada por categoria: (n_c * media_c + m * media_global) / (n_c + m).'''
    g = y.mean()
    s = y.groupby(cat).agg(["sum", "count"])
    return ((s["sum"] + m * g) / (s["count"] + m)).to_dict(), g


def te_aplicar(cat, mapa, g):
    '''Categorias no vistas reciben la media global.'''
    return cat.map(mapa).fillna(g).to_numpy()


def te_fuera_de_pliegue(cat, y, m=0.0, k=5, semilla=0):
    '''Codifica cada fila con un mapa calculado SIN esa fila (k pliegues): evita la fuga.'''
    salida = np.empty(len(cat))
    for a, b in KFold(k, shuffle=True, random_state=semilla).split(cat):
        mapa, g = te_ajustar(cat.iloc[a], y.iloc[a], m)
        salida[b] = te_aplicar(cat.iloc[b], mapa, g)
    return salida


def comparar_codificaciones(datos, n_particiones=40):
    res = {}
    for s in range(n_particiones):
        tr, te = train_test_split(datos, test_size=0.4, random_state=s)
        def anotar(nombre, A, B):
            mod = LinearRegression().fit(A, tr["y"])
            res.setdefault(nombre, []).append((mod.score(A, tr["y"]), mod.score(B, te["y"])))
        anotar("ignorar la vereda", tr[["x"]], te[["x"]])
        mapa, g = te_ajustar(tr["vereda"], tr["y"], 0)
        anotar("TE ingenuo (m=0)", np.c_[tr["x"], te_aplicar(tr["vereda"], mapa, g)], np.c_[te["x"], te_aplicar(te["vereda"], mapa, g)])
        mapa, g = te_ajustar(tr["vereda"], tr["y"], 5)
        anotar("TE ingenuo suavizado (m=5)", np.c_[tr["x"], te_aplicar(tr["vereda"], mapa, g)], np.c_[te["x"], te_aplicar(te["vereda"], mapa, g)])
        anotar("TE fuera de pliegue (m=5)", np.c_[tr["x"], te_fuera_de_pliegue(tr["vereda"], tr["y"], 5)], np.c_[te["x"], te_aplicar(te["vereda"], mapa, g)])
    return pd.DataFrame({k: np.mean(v, axis=0) for k, v in res.items()}, index=["R2 entrenamiento", "R2 prueba"]).T


veredas_A, veredas_B = generar_veredas(sd_efecto=0.0), generar_veredas(sd_efecto=1.5)
tab_A, tab_B = comparar_codificaciones(veredas_A), comparar_codificaciones(veredas_B)
print("Mundo A: la vereda NO tiene efecto real\n", tab_A.round(3).to_string())
print("\nMundo B: la vereda SI tiene efecto real (sd = 1.5 t/ha)\n", tab_B.round(3).to_string())

# TargetEncoder de scikit-learn: con 'fit().transform()' reproduce nuestra formula (suavizado m=5)...
tr_B, te_B = train_test_split(veredas_B, test_size=0.4, random_state=0)
mapa, g = te_ajustar(tr_B["vereda"], tr_B["y"], 5.0)
tes = TargetEncoder(smooth=5.0, target_type="continuous").fit(tr_B[["vereda"]], tr_B["y"])
assert np.allclose(tes.transform(tr_B[["vereda"]]).ravel(), te_aplicar(tr_B["vereda"], mapa, g))
# ...pero 'fit_transform' usa validacion cruzada interna (cross fitting): NO es lo mismo, y es lo correcto para entrenar
ft = TargetEncoder(smooth=5.0, target_type="continuous", cv=5).fit_transform(tr_B[["vereda"]], tr_B["y"]).ravel()
assert not np.allclose(ft, te_aplicar(tr_B["vereda"], mapa, g))
print("\nscikit-learn TargetEncoder: transform() = nuestra formula; fit_transform() usa cross-fitting (distinto a proposito).")

fig, ax = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for a, tab, t in zip(ax, [tab_A, tab_B], ["A: la vereda NO importa", "B: la vereda SÍ importa"]):
    pos = np.arange(len(tab)); w = 0.38
    a.bar(pos - w / 2, tab["R2 entrenamiento"], w, color="#94a3b8", label="entrenamiento")
    a.bar(pos + w / 2, tab["R2 prueba"], w, color="#1e3a8a", label="prueba")
    a.set_xticks(pos); a.set_xticklabels([s.replace(" (", "\n(") for s in tab.index], fontsize=7); a.set_title(t)
ax[0].set_ylabel("$R^2$ (media de 40 particiones)"); ax[0].legend(fontsize=8)
plt.tight_layout(); plt.show()

# Fuga: en el mundo A el TE ingenuo mejora el entrenamiento y EMPEORA la prueba
assert tab_A.loc["TE ingenuo (m=0)", "R2 entrenamiento"] > tab_A.loc["ignorar la vereda", "R2 entrenamiento"]
assert tab_A.loc["TE ingenuo (m=0)", "R2 prueba"] < tab_A.loc["ignorar la vereda", "R2 prueba"] - 0.05
# Fuera de pliegue no inventa senal en A y la aprovecha en B
assert abs(tab_A.loc["TE fuera de pliegue (m=5)", "R2 prueba"] - tab_A.loc["ignorar la vereda", "R2 prueba"]) < 0.02
assert tab_B.loc["TE fuera de pliegue (m=5)", "R2 prueba"] > tab_B.loc["ignorar la vereda", "R2 prueba"] + 0.10
````

---
**Lectura.** (1) **Mundo A (sin efecto real).** El *target encoding* ingenuo **sube** el $R^2$ de entrenamiento (0.359 → 0.437) y **baja** el de prueba (0.362 → 0.261) respecto a ignorar la vereda: el modelo aprendió ruido que le «sopló» la propia respuesta. El esquema *out-of-fold* queda **a la par** de ignorar la variable (0.359 frente a 0.362): no inventa señal. (2) **Mundo B (efecto real).** Codificar bien la vereda mejora la prueba de forma clara (0.238 → 0.502 con *out-of-fold* y $m=5$); además su $R^2$ de entrenamiento (0.492) es **honesto** —parecido al de prueba—, mientras que el ingenuo muestra 0.639 en entrenamiento y 0.472 en prueba (la señal de alarma de la fuga: una brecha de ≈ 0.17). (3) **Matiz honesto:** el suavizado ($m=5$) por sí solo **reduce** el daño del enfoque ingenuo (prueba 0.261 → 0.284 en A; 0.472 → 0.495 en B, casi igualando al *out-of-fold*), pero **no elimina** la fuga ni sustituye al esquema *out-of-fold*: en A sigue perdiendo frente a ignorar la vereda. (4) El valor de $m$ es un **hiperparámetro**: se elige por validación cruzada, no a ojo.

## 5. En Producción: Categorías Nuevas, Raras y Columnas Desalineadas 🏭

Un codificador se **ajusta una vez** (con los datos de entrenamiento) y se **aplica muchas veces**. Tres fallos clásicos al pasar a datos nuevos:

1. **Columnas desalineadas.** Llamar `pd.get_dummies` por separado sobre entrenamiento y prueba genera columnas **distintas** si falta (o sobra) una categoría. El modelo recibe una matriz con otra forma o con columnas en otro orden.
2. **Categoría nunca vista.** Aparece una vereda o variedad nueva. `OneHotEncoder()` lanza un error; con `handle_unknown="ignore"` la fila queda en **ceros** (equivale a «ninguna categoría conocida»). En un modelo con referencia eso se **confunde con la propia referencia**, por lo que conviene registrar cuántas filas caen allí.
3. **Categorías raras.** Con 3 observaciones, el coeficiente de una ficticia es ruido. `min_frequency` agrupa las categorías poco frecuentes en una categoría común «infrecuente».

````python
entr, prueba = fincas.iloc[:300].copy(), fincas.iloc[300:].copy()
prueba.iloc[0, prueba.columns.get_loc("variedad")] = "Tuquerreña"      # categoria nueva (inventada)
prueba = prueba[prueba["variedad"] != "Sabanera"]                     # y en prueba falta una categoria de entrenamiento

# 1) get_dummies por separado: columnas distintas
A, B = pd.get_dummies(entr["variedad"]), pd.get_dummies(prueba["variedad"])
print("columnas entrenamiento:", list(A.columns), "\ncolumnas prueba:       ", list(B.columns))
assert list(A.columns) != list(B.columns)

# 2) OneHotEncoder ajustado en entrenamiento: misma forma siempre; la categoria nueva queda en ceros
enc = OneHotEncoder(handle_unknown="ignore", sparse_output=False).fit(entr[["variedad"]])
Z = enc.transform(prueba[["variedad"]])
print("\nforma con OneHotEncoder:", Z.shape, "| fila de 'Tuquerreña':", Z[0], "| filas sin ninguna categoria conocida:", int((Z.sum(axis=1) == 0).sum()))
assert Z.shape[1] == A.shape[1] and Z[0].sum() == 0 and (Z.sum(axis=1) == 0).sum() == 1

# 3) sin handle_unknown: error explicito (mejor fallar que aprender en silencio)
try:
    OneHotEncoder().fit(entr[["variedad"]]).transform(prueba[["variedad"]])
except ValueError as e:
    print("\nError esperado:", str(e).splitlines()[0][:80])

# 4) categorias infrecuentes agrupadas
enc_raras = OneHotEncoder(min_frequency=0.16, handle_unknown="infrequent_if_exist", sparse_output=False).fit(entr[["variedad"]])
print("\ncategorias infrecuentes:", enc_raras.infrequent_categories_[0], "| columnas:", list(enc_raras.get_feature_names_out()))
assert "Sabanera" in enc_raras.infrequent_categories_[0]
assert enc_raras.transform(prueba[["variedad"]]).shape[1] == len(enc_raras.get_feature_names_out())
````

**Guía de decisión.**

| Situación | Codificación recomendada | Cuidado |
|---|---|---|
| Nominal, pocos niveles, modelo lineal/GLM | ***One-hot*** con referencia (`drop="first"`) | Elegir una referencia **interpretable** (la más frecuente o la «de control») |
| Ordinal con pasos aproximadamente iguales | Código ordinal con **orden explícito** | Contrastar con ficticias; nunca orden alfabético |
| Ordinal con pasos desiguales o dudosos | *One-hot* | Más parámetros |
| Nominal, **muchos** niveles | *Target encoding* **out-of-fold** + suavizado; o agrupar raras | Fuga si se calcula con toda la muestra |
| Respuesta categórica (clasificación) | `LabelEncoder` solo sobre $y$ | No usarlo en predictores |
| Modelos de árboles | Códigos ordinales suelen bastar | Los lineales y los GLM **no** |
| Producción | Codificador **ajustado en entrenamiento** dentro de un `Pipeline` | Categorías nuevas y columnas desalineadas |

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Matriz de diseño** | Un modelo lineal necesita columnas numéricas **linealmente independientes**: $\mathbf{X}^{\top}\mathbf{X}$ debe ser invertible |
| **Label encoding** | Una sola columna de códigos impone **orden y pasos iguales**. Con una nominal, el $R^2$ **depende del orden arbitrario** (24 órdenes = 24 modelos distintos, $R^2$ de −0.04 a 0.36) y queda por debajo del *one-hot* (0.395) |
| **Ordinal** | Si el orden es real, el código es una simplificación de 1 parámetro; se **contrasta** con una prueba $F$ anidada y se declara el orden **explícitamente** |
| **One-hot** | Una indicadora por categoría, sin orden. Verificado igual a `get_dummies` y `OneHotEncoder` |
| **Trampa de la variable ficticia** | Intercepto + $k$ ficticias $\Rightarrow$ rango $k$ (singular). Se descarta una **referencia**; los coeficientes son **diferencias respecto a ella** y las **predicciones no cambian** con la referencia |
| **Target encoding** | $\tilde x_c=(n_c\bar y_c+m\bar y)/(n_c+m)$. Ingenuo = **fuga**: sube el entrenamiento y baja la prueba cuando no hay efecto. Usar *out-of-fold* + suavizado, y elegir $m$ por validación |
| **Producción** | Ajustar el codificador en entrenamiento, manejar categorías nuevas e infrecuentes y evitar `get_dummies` separado |
