```python
# ============================================================
# Preparación (no modificar): imports y datos sintéticos
# ============================================================
import unicodedata
import warnings

import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
from scipy import stats
from sklearn.cluster import DBSCAN, KMeans
from sklearn.covariance import EmpiricalCovariance, MinCovDet
from sklearn.datasets import make_moons
from sklearn.inspection import permutation_importance
from sklearn.metrics import adjusted_rand_score, silhouette_score
from sklearn.model_selection import GroupKFold, GroupShuffleSplit, KFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import acf, adfuller

warnings.filterwarnings("ignore")      # los ajustes de statsmodels avisan de cosas que aquí no importan
SEMILLA = 2027


# --- Reto 1: ventas mensuales (toneladas) de insumos de un almacén agropecuario, 10 años ---
def generar_ventas(semilla=SEMILLA, n=120):
    """Nivel con deriva estocástica (camino aleatorio con tendencia) + estacionalidad fija de periodo 12 + ruido."""
    rng = np.random.default_rng(semilla)
    nivel = 200 + np.cumsum(0.8 + rng.normal(0, 2.5, n))
    mes = np.arange(n) % 12
    estacion = 30 * np.sin(2 * np.pi * (mes - 2) / 12) + 10 * np.cos(4 * np.pi * mes / 12)
    y = nivel + estacion + rng.normal(0, 6, n)
    return pd.Series(y, index=pd.date_range("2015-01-01", periods=n, freq="MS"), name="ventas_t")


# --- Reto 2: 240 estaciones meteorológicas (4 climas) y dos lunas para DBSCAN ---
def generar_estaciones(semilla=SEMILLA, por_clima=60):
    """Temperatura (°C) y humedad (%) separan 4 climas; la precipitación (mm/año) los mezcla y está en una escala enorme."""
    rng = np.random.default_rng(semilla + 1)
    centros = [(10, 55), (10, 85), (20, 55), (20, 85)]
    filas, real = [], []
    for k, (t, h) in enumerate(centros):
        temp, hum = rng.normal(t, 0.9, por_clima), rng.normal(h, 3.5, por_clima)
        lluvia = 1500 + 45 * (temp - 15) + 15 * (hum - 70) + rng.normal(0, 60, por_clima)
        filas.append(np.column_stack([temp, hum, lluvia]))
        real += [k] * por_clima
    orden = rng.permutation(len(real))
    X = np.vstack(filas)[orden]
    return pd.DataFrame(X, columns=["temperatura", "humedad", "precipitacion_mm"]), np.array(real)[orden]


estaciones, clima_real = generar_estaciones()
lunas, lunas_real = make_moons(n_samples=300, noise=0.06, random_state=SEMILLA)


# --- Reto 3: fincas y lotes de un distrito de riego (varios lotes por finca) ---
def generar_fincas_lotes(semilla=SEMILLA, n_f=150):
    rng = np.random.default_rng(semilla + 2)
    ids = [f"F-{i + 1:03d}" for i in range(n_f)]
    fincas = pd.DataFrame({"id_finca": ids, "vereda": rng.choice(["Alta", "Media", "Baja", "Centro", "Norte"], n_f),
                           "altitud": np.round(rng.uniform(1800, 3200, n_f)), "area_ha": np.round(rng.lognormal(1.0, 0.6, n_f), 2),
                           "riego": rng.integers(0, 2, n_f)})
    u = rng.normal(0, 1.5, n_f)                                           # efecto de finca NO observado
    n_lotes = rng.integers(2, 7, n_f)
    idx = np.repeat(np.arange(n_f), n_lotes)
    lotes = pd.DataFrame({"id_lote": [f"L-{i + 1:04d}" for i in range(len(idx))], "id_finca": np.array(ids)[idx],
                          "fert": np.round(rng.uniform(0, 4, len(idx)), 2), "ruido": np.round(rng.normal(size=len(idx)), 3)})
    lotes["rend"] = np.round(8 + 0.9 * fincas["riego"].to_numpy()[idx] + 0.5 * lotes["fert"] + u[idx] + rng.normal(0, 0.8, len(idx)), 3)
    fincas = pd.concat([fincas, fincas.iloc[[3, 40, 77]]]).reset_index(drop=True)   # 3 filas de finca repetidas (defecto)
    return fincas, lotes


fincas, lotes = generar_fincas_lotes()
GRILLA_ALFA = [0.0, 0.001, 0.002, 0.005, 0.01, 0.02, 0.04]


# --- Reto 4: un censo agropecuario SINTÉTICO (¡no son datos del DANE!) de un municipio ficticio, con defectos ---
MUNICIPIOS = ["Alto Verde", "Río Claro", "Villa Esperanza", "San Rafael", "La Pradera"]


def _sin_tilde(t):
    return "".join(c for c in unicodedata.normalize("NFKD", t) if not unicodedata.combining(c))


def generar_censo(semilla=SEMILLA, n=700):
    """Devuelve el censo sucio. `atipico_sim` es la verdad de la simulación (solo para evaluar; en la vida real no existe)."""
    rng = np.random.default_rng(semilla + 3)
    mun = rng.choice(MUNICIPIOS, n)
    mun = [m.upper() if k == 1 else ("  " + m.lower().replace(" ", "   ") + " " if k == 2 else (_sin_tilde(m) if k == 3 else m))
           for m, k in zip(mun, rng.integers(0, 4, n))]
    lx = rng.multivariate_normal([1.0, 2.0], [[0.64, 0.56], [0.56, 0.64]], n)      # log(área sembrada), log(producción): r = 0.875
    atip = np.zeros(n, bool)
    atip[rng.choice(n, int(0.12 * n), replace=False)] = True                        # 12 % con la relación invertida
    lx[atip] = np.column_stack([rng.normal(1.7, 0.3, atip.sum()), rng.normal(0.8, 0.3, atip.sum())])
    sembrada, produccion = np.round(np.exp(lx[:, 0]), 2), np.round(np.exp(lx[:, 1]), 2)
    total = np.round(sembrada * (1 + rng.exponential(0.6, n)) + 0.05, 2)
    cen = pd.DataFrame({"id_upa": [f"U-{i + 1:04d}" for i in range(n)], "municipio": mun, "area_total_ha": total,
                        "area_sembrada_ha": sembrada, "produccion_t": produccion, "atipico_sim": atip})
    cen.loc[rng.choice(n, 14, replace=False), "area_total_ha"] = -1.0               # centinela de «sin dato»
    k = rng.choice(np.flatnonzero(~atip & (cen["area_total_ha"] > 0)), 18, replace=False)
    cen.loc[k, "area_total_ha"] = cen.loc[k, "area_sembrada_ha"] * 0.5               # sembrada > total: imposible
    cen = pd.concat([cen, cen.iloc[rng.choice(n, 12, replace=False)]])               # 12 filas repetidas
    return cen.iloc[rng.permutation(len(cen))].reset_index(drop=True)


censo = generar_censo()
print(f"Ventas: {len(generar_ventas())} meses | Estaciones: {len(estaciones)} | Lotes: {len(lotes)} en {lotes['id_finca'].nunique()} fincas ({len(fincas)} filas de finca) | Censo: {len(censo)} filas")
```
