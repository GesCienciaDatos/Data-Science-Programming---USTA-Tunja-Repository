```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np
import pandas as pd

# 1 000 estudiantes: ¿compran el almuerzo en la cafetería? y qué nota (0 a 10) le dan al almuerzo escolar
_rng = np.random.default_rng(7)
compra = _rng.random(1000) < 0.30
nota = np.clip(np.where(compra, _rng.normal(8, 1, 1000), _rng.normal(5, 1.5, 1000)), 0, 10)

# Los resultados de dos profesores de matemáticas (alumnos y cuántos aprobaron en cada curso)
cursos = pd.DataFrame({"profe": ["Ana", "Ana", "Beto", "Beto"],
                       "curso": ["facil", "dificil", "facil", "dificil"],
                       "alumnos": [20, 80, 80, 20],
                       "aprueban": [19, 56, 72, 12]})
print("Estudiantes:", len(nota), "| compran en la cafetería:", int(compra.sum()))
```

## Reto 2: Dos profes y una tabla que engaña 🔄

La tabla `cursos` dice cuántos alumnos tuvo cada profe en un curso **fácil** y en uno **difícil**, y cuántos aprobaron (cifras inventadas). Ana tuvo casi todos sus alumnos en el curso difícil y Beto casi todos en el fácil.

**Pasos**
1. En `cursos`, crea la columna `tasa` = `aprueban / alumnos` (la fracción que aprobó en cada fila).
2. `total`: suma `alumnos` y `aprueban` por profe (`groupby("profe")`) y calcula `tasa_global` = aprueban totales / alumnos totales de cada uno.
3. `tasa_justa`: compara con **justicia**: usa los **mismos pesos** para los dos profes, la mitad fácil y la mitad difícil (`0.5` y `0.5`): `0.5 * tasa en el fácil + 0.5 * tasa en el difícil`.

**Puntaje sugerido:** 35 puntos.

```python
# Casillas que debes llenar
total = tasa_global = tasa_justa = None

# 💡 1) La fracción que aprobó en cada fila
cursos["tasa"] = cursos["aprueban"] / cursos[...]            # pista: dividir entre los alumnos

# 💡 2) El total de cada profe (sumando sus dos cursos)
total = cursos.groupby("profe")[["alumnos", "aprueban"]].sum()
tasa_global = total["aprueban"] / total[...]

# 💡 3) Comparar con los mismos pesos: la mitad de la nota viene del curso fácil y la mitad del difícil
tabla = cursos.pivot(index="profe", columns="curso", values="tasa")
tasa_justa = 0.5 * tabla["facil"] + ... * tabla["dificil"]
```
