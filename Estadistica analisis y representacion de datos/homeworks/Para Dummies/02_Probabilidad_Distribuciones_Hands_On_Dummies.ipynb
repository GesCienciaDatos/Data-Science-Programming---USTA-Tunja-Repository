## 🎈 ¿Qué vamos a hacer en este taller?

En la **kermés del colegio** hay dos cosas que pasan por tus manos: el grupo de WhatsApp de padres, que tiene un **filtro de mensajes basura** (*spam*), y un **juego de dados** con premios. Vas a usar la probabilidad para responder dos preguntas con cuentas y con simulaciones:

```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import itertools

import numpy as np

# El filtro de mensajes basura del grupo de padres (cifras inventadas)
N_MENSAJES = 1000     # mensajes del mes
P_SPAM = 0.10         # 10 % de los mensajes son basura
SENS = 0.90           # el filtro marca el 90 % de la basura
FPR = 0.05            # y marca por error el 5 % de los mensajes buenos

# El juego de dados de la kermés: se lanzan 2 dados y se cobra $2 000 por jugar
COBRO = 2000
print("Mensajes:", N_MENSAJES, "| el filtro acierta con la basura el", f"{SENS:.0%}", "de las veces")
```

## Reto 1: Suena el filtro... ¿era basura? 📱

Imagina los **1 000 mensajes del mes**. 100 son basura (10 %) y 900 son buenos. El filtro marca el 90 % de la basura, pero **también** marca el 5 % de los buenos. Cuando un mensaje aparece marcado, ¿qué fracción de los marcados es basura de verdad? Primero con cuentas de personas «enteras», luego **simulando** 200 000 mensajes.

**Pasos**
1. `basura_marcada`: cuántos de los 1 000 mensajes son basura **y** están marcados. `buenos_marcados`: cuántos son buenos **y** están marcados.
2. `p_basura_si_marca` $=$ basura marcada $\big/$ (basura marcada $+$ buenos marcados).
3. Simulación: `es_spam` (verdadero con probabilidad 0.10), `marca` (si es basura, se marca con probabilidad 0.90; si no, con 0.05) y `p_sim`: la fracción de basura **entre los marcados**.

**Puntaje sugerido:** 40 puntos.

```python
# Casillas que debes llenar (empiezan vacías; no las borres)
basura_marcada = buenos_marcados = p_basura_si_marca = p_sim = None

# 💡 1) Con cuentas: ¿cuántos mensajes de cada tipo quedan marcados?
basura_marcada = N_MENSAJES * P_SPAM * SENS                 # basura (100) × 90 %
buenos_marcados = N_MENSAJES * (1 - P_SPAM) * ...           # buenos (900) × 5 %

# 💡 2) De los marcados, ¿qué fracción es basura de verdad?
p_basura_si_marca = basura_marcada / (basura_marcada + ...)

# 💡 3) Lo mismo, pero simulando 200 000 mensajes
rng = np.random.default_rng(5)
n = 200_000
es_spam = rng.random(n) < P_SPAM
marca = np.where(es_spam, rng.random(n) < SENS, rng.random(n) < ...)   # basura: 0.90; buenos: 0.05
p_sim = es_spam[marca].mean()                                          # fracción de basura entre los marcados
```

```python
# Autoverificación del Reto 1 (no modificar)
assert basura_marcada is not None and abs(basura_marcada - 90) < 1e-9, "Pista 🧸: basura_marcada = 1000 × 0.10 × 0.90 = 90 mensajes."
assert buenos_marcados is not None and abs(buenos_marcados - 45) < 1e-9, "Pista 🧸: buenos_marcados = 1000 × 0.90 × 0.05 = 45 mensajes buenos que el filtro marca por error (usa FPR)."
assert p_basura_si_marca is not None and abs(p_basura_si_marca - 90 / 135) < 1e-9, "Pista 🧸: de los 135 marcados (90 + 45), la basura es 90/135 ≈ 0.667. ¡No es 0.90!"
assert p_sim is not None and abs(p_sim - 2 / 3) < 0.02, "Pista 🧸: p_sim = es_spam[marca].mean(): entre los mensajes marcados, la fracción que es basura. Debe quedar cerca de 0.67."
assert p_basura_si_marca < SENS, "Aunque el filtro acierta el 90 % de la basura, la fracción de basura ENTRE LOS MARCADOS es menor que 0.90."
print(f"✅ ¡Reto 1 superado! De cada 100 mensajes marcados, ≈ {p_basura_si_marca * 100:.0f} son basura (simulación: {p_sim * 100:.0f}); los otros ≈ {100 - p_basura_si_marca * 100:.0f} eran buenos.")
```

## Reto 2: ¿Conviene jugar a los dados? 🎲

En la kermés pagas **$2 000** por jugar: se lanzan **2 dados**. Si la suma es **7** te dan **$5 000**; si la suma es **12** te dan **$20 000**; con cualquier otra suma, nada. El resultado de cada lanzamiento depende del azar, pero **a la larga** tiene un promedio: el **valor esperado**.

**Pasos**
1. `sumas`: la lista de las 36 sumas posibles de dos dados (ya está armada con `itertools.product`). `p_7` y `p_12`: la probabilidad de cada suma (casos favorables entre 36).
2. `ganancia_esperada` $=5000\,p_7+20000\,p_{12}-2000$ (lo que se espera ganar **por jugada**, ya descontado el cobro).
3. `ganancia_media_sim`: juega **100 000** veces en el computador y calcula la ganancia media por jugada.

**Puntaje sugerido:** 35 puntos.

```python
# Casillas que debes llenar
p_7 = p_12 = ganancia_esperada = ganancia_media_sim = None

# 💡 1) Las 36 formas de lanzar dos dados y sus sumas
sumas = [a + b for a, b in itertools.product(range(1, 7), repeat=2)]
p_7 = sumas.count(7) / 36
p_12 = sumas.count(...) / 36                                  # pista: ¿qué suma da $20 000?

# 💡 2) Valor esperado de la ganancia neta por jugada
ganancia_esperada = 5000 * p_7 + 20000 * ... - COBRO

# 💡 3) Jugar 100 000 veces en el computador
rng = np.random.default_rng(11)
lanzamientos = rng.integers(1, 7, size=(100_000, 2)).sum(axis=1)
premio = np.where(lanzamientos == 7, 5000, 0) + np.where(lanzamientos == 12, 20000, 0)
ganancia_media_sim = (premio - COBRO).mean()
```

```python
# Autoverificación del Reto 2 (no modificar)
assert len(sumas) == 36, "Pista 🧸: dos dados dan 6 × 6 = 36 combinaciones; sumas ya viene armada."
assert p_7 is not None and abs(p_7 - 6 / 36) < 1e-12, "Pista 🧸: la suma 7 sale de 6 de las 36 formas (1+6, 2+5, 3+4, 4+3, 5+2, 6+1): p_7 = 6/36."
assert p_12 is not None and abs(p_12 - 1 / 36) < 1e-12, "Pista 🧸: la suma 12 solo sale como 6+6: p_12 = 1/36. Cuenta sumas.count(12)."
assert ganancia_esperada is not None and abs(ganancia_esperada - (5000 * 6 / 36 + 20000 / 36 - 2000)) < 1e-9, "Pista 🧸: ganancia_esperada = 5000 × p_7 + 20000 × p_12 - COBRO."
assert ganancia_esperada < 0, "El valor esperado debe salir NEGATIVO: a la larga, el que juega pierde (≈ -611 pesos por jugada)."
assert ganancia_media_sim is not None and abs(ganancia_media_sim - ganancia_esperada) < 60, "Pista 🧸: la ganancia media de las 100 000 jugadas debe quedar cerca del valor esperado (ley de los grandes números)."
print(f"✅ ¡Reto 2 superado! Valor esperado por jugada: {ganancia_esperada:,.0f} pesos | jugando 100 000 veces: {ganancia_media_sim:,.0f} pesos | el juego favorece a la kermés.")
```
