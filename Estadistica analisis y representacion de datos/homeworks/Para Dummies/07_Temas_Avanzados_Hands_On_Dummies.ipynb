```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.covariance import EmpiricalCovariance
from sklearn.metrics import adjusted_rand_score
from sklearn.tree import DecisionTreeClassifier

SEMILLA = 11
rng = np.random.default_rng(SEMILLA)

# Reto 1: ruanas vendidas en 36 meses (3 años): más ventas en los meses fríos, y cada mes un poquito más que el anterior
estacion = np.array([12, 9, 4, 0, -4, -9, -12, -8, -2, 5, 10, 14])
ventas = np.round(100 + 0.8 * np.arange(36) + np.tile(estacion, 3) + rng.normal(0, 2, 36))

# Reto 2: 40 canastas de mora de dos tipos (tipo_mora: 0 o 1). El dulzor las separa; el peso de la canasta (g) es puro azar
tipo_mora = np.repeat([0, 1], 20)
dulzor = np.where(tipo_mora == 0, rng.normal(4, 0.5, 40), rng.normal(8, 0.5, 40))
moras = np.column_stack([dulzor, rng.uniform(500, 3500, 40)])
# ... y 600 lotes de mora con lluvia (mm) y altitud (m); enferma = 1 si llovió mucho (con 20 % de errores de anotación)
lluvia, altitud = rng.uniform(50, 150, 600), rng.uniform(2000, 3000, 600)
enferma = ((lluvia > 100) ^ (rng.random(600) < 0.2)).astype(int)
lotes = np.column_stack([lluvia, altitud])
lotes_ent, lotes_prb, enf_ent, enf_prb = lotes[:400], lotes[400:], enferma[:400], enferma[400:]

# Reto 3: la libreta de las veredas (cabezas de ganado; -1 significa «no se contó») y 40 terneros (altura en cm, peso en kg)
libreta = pd.DataFrame({"vereda": ["El Rosal", "EL ROSAL", " el rosal ", "La Esperanza", "la esperanza", "LA ESPERANZA ",
                                   "San Isidro", "san isidro", " SAN ISIDRO", "El Rosal", "La Esperanza", "San Isidro"],
                        "cabezas": [12, 8, 15, 20, -1, 9, 14, 7, -1, 11, 18, 6]})
alt = rng.normal(100, 8, 40)
terneros = np.column_stack([alt, 120 + 2.5 * (alt - 100) + rng.normal(0, 4, 40)])
terneros = np.vstack([terneros, [[108, 100], [92, 140]]])      # los dos últimos: ¡los sospechosos de Doña Rosalía!

print("Meses:", len(ventas), "| canastas:", len(moras), "| lotes:", len(lotes), "| filas de la libreta:", len(libreta), "| terneros:", len(terneros))
```

## Reto 1: «Vendo igual que el año pasado» 📈

Mira `ventas`: 36 meses de ruanas. Hay dos cosas a la vez: **el frío** (cada año se repite el mismo patrón) y **una subida lenta** (cada mes se vende un poquito más). Para saber si un pronóstico sirve, hay que hacerle un **examen sorpresa**: guardas los últimos 12 meses sin mirarlos, pronosticas y comparas. Nunca se baraja, porque el futuro no se puede usar para adivinar el pasado.

**Pasos**
1. `entrenamiento`: los primeros 24 meses (lo ya vivido); `prueba`: los últimos 12 (el examen).
2. `pron_ingenuo`: «igual que el año pasado» = los últimos 12 meses de `entrenamiento`. `error_ingenuo`: el promedio de las diferencias absolutas entre `prueba` y `pron_ingenuo`.
3. `pron_prom`: el **promedio de los dos años** vividos, mes a mes (año 1 + año 2, entre 2), y su `error_prom`.
4. `mejor`: `"ingenuo"` o `"promedio"`, según cuál se equivoca menos.

**Puntaje sugerido:** 25 puntos.

```python
# Casillas que debes llenar (empiezan vacías; no las borres)
entrenamiento = prueba = pron_ingenuo = error_ingenuo = pron_prom = error_prom = mejor = None

# 💡 1) Los primeros 24 meses para aprender y los últimos 12 para el examen
entrenamiento = ventas[:...]
prueba = ventas[...:]

# 💡 2) «Igual que el año pasado»: repite los últimos 12 meses vividos
pron_ingenuo = entrenamiento[-...:]
error_ingenuo = np.mean(np.abs(prueba - ...))

# 💡 3) Promedio de los dos años vividos (mes a mes)
pron_prom = (entrenamiento[:12] + entrenamiento[12:]) / ...
error_prom = np.mean(np.abs(prueba - pron_prom))

# 💡 4) ¿Cuál se equivoca menos?
mejor = "ingenuo" if error_ingenuo < ... else "promedio"
```

```python
# Autoverificación del Reto 1 (no modificar)
assert entrenamiento is not None and len(entrenamiento) == 24 and len(prueba) == 12, "Pista 🧸: ventas[:24] son los 24 primeros meses y ventas[24:] los 12 últimos."
assert np.array_equal(np.r_[entrenamiento, prueba], ventas), "Pista 🧸: no barajes nada; entrenamiento y prueba juntos deben dar exactamente la serie original."
assert pron_ingenuo is not None and np.array_equal(pron_ingenuo, entrenamiento[-12:]), "Pista 🧸: «igual que el año pasado» son los últimos 12 meses de entrenamiento: entrenamiento[-12:]."
assert error_ingenuo is not None and abs(error_ingenuo - np.mean(np.abs(prueba - pron_ingenuo))) < 1e-9, "Pista 🧸: error_ingenuo = np.mean(np.abs(prueba - pron_ingenuo))."
assert pron_prom is not None and len(pron_prom) == 12 and np.allclose(pron_prom, (ventas[:12] + ventas[12:24]) / 2), "Pista 🧸: suma el año 1 y el año 2 y divide entre 2."
assert abs(error_prom - np.mean(np.abs(prueba - pron_prom))) < 1e-9, "Pista 🧸: error_prom = np.mean(np.abs(prueba - pron_prom))."
assert error_ingenuo < error_prom, "Revisa 🧸: como las ventas suben cada mes, el año pasado se parece más al año que viene que el promedio de dos años; el error del «ingenuo» debería ser menor."
assert mejor == "ingenuo", "Pista 🧸: mejor es el pronóstico con menor error."
print(f"✅ ¡Reto 1 superado! Error de «igual que el año pasado»: {error_ingenuo:.1f} ruanas por mes; error del promedio de dos años: {error_prom:.1f}. Gana: {mejor}.")
```

## Reto 2: Moras en montoncitos y un árbol que memoriza 🫐🌳

**Parte A: agrupar sin que nadie diga cuáles son.** Las 40 canastas de `moras` (dos columnas: **dulzor** y **peso de la canasta en gramos**) son de dos tipos (`tipo_mora`, que el algoritmo NO ve). El dulzor las separa, pero el peso está en cientos de gramos: si no pones ambas columnas en la **misma escala**, el algoritmo solo mira el peso (que es puro azar). El `ARI` mide qué tanto se parecen los grupos hallados a los reales: **1 = idénticos, 0 = como tirar monedas**.

**Parte B: un árbol de preguntas.** Con `lotes_ent` / `enf_ent` (400 lotes para aprender) y `lotes_prb` / `enf_prb` (200 lotes para el examen) vas a comparar un árbol **chiquito** (2 pisos) con uno que crece **sin límite**.

**Pasos**
1. `moras_esc`: `moras` con cada columna a promedio 0 y desviación 1.
2. `ari_crudo`: el ARI de `KMeans` (2 grupos) sobre `moras` tal cual; `ari_esc`: el mismo, sobre `moras_esc`.
3. `arbol_chico` (`max_depth=2`) y `arbol_grande` (sin límite), ambos con `random_state=0` y ajustados con `lotes_ent`.
4. `tr_chico`, `te_chico`, `tr_grande`, `te_grande`: su exactitud (`.score`) en los lotes de **aprender** (`tr`) y en los del **examen** (`te`).

**Puntaje sugerido:** 30 puntos.

```python
# Casillas que debes llenar
moras_esc = ari_crudo = ari_esc = None
arbol_chico = arbol_grande = tr_chico = te_chico = tr_grande = te_grande = None

# 💡 1) Misma escala: resta el promedio y divide entre la desviación de CADA columna
moras_esc = (moras - moras.mean(axis=0)) / moras.std(axis=...)

# 💡 2) K-means con 2 grupos, sin y con la misma escala
etq_crudo = KMeans(n_clusters=2, n_init=10, random_state=0).fit_predict(moras)
etq_esc = KMeans(n_clusters=2, n_init=10, random_state=0).fit_predict(...)
ari_crudo = adjusted_rand_score(tipo_mora, etq_crudo)
ari_esc = adjusted_rand_score(tipo_mora, etq_esc)

# 💡 3) Un árbol de 2 pisos y otro sin límite, ambos aprenden con lotes_ent
arbol_chico = DecisionTreeClassifier(max_depth=..., random_state=0).fit(lotes_ent, enf_ent)
arbol_grande = DecisionTreeClassifier(random_state=0).fit(..., enf_ent)

# 💡 4) Exactitud al aprender (tr) y en el examen (te)
tr_chico, te_chico = arbol_chico.score(lotes_ent, enf_ent), arbol_chico.score(lotes_prb, enf_prb)
tr_grande, te_grande = arbol_grande.score(lotes_ent, enf_ent), arbol_grande.score(..., enf_prb)
```

```python
# Autoverificación del Reto 2 (no modificar)
assert moras_esc is not None and np.allclose(moras_esc.mean(axis=0), 0, atol=1e-9) and np.allclose(moras_esc.std(axis=0), 1, atol=1e-9), "Pista 🧸: (moras - moras.mean(axis=0)) / moras.std(axis=0): así cada columna queda con promedio 0 y desviación 1."
assert ari_esc is not None and abs(ari_esc - adjusted_rand_score(tipo_mora, KMeans(2, n_init=10, random_state=0).fit_predict(moras_esc))) < 1e-9, "Pista 🧸: ari_esc usa los grupos de KMeans sobre moras_esc (las columnas en la misma escala)."
assert ari_esc > 0.8, "Con la misma escala el dulzor manda y los dos tipos de mora se separan casi perfecto (ARI > 0.8)."
assert ari_crudo is not None and ari_crudo < 0.5, "Sin la misma escala el peso (cientos de gramos) manda sobre el dulzor, y el ARI queda cerca de 0."
assert arbol_chico is not None and arbol_chico.get_depth() <= 2, "Pista 🧸: DecisionTreeClassifier(max_depth=2, random_state=0)."
assert arbol_grande is not None and arbol_grande.get_depth() > 4, "Pista 🧸: el árbol grande no lleva max_depth: crece hasta aprender cada lote."
assert abs(tr_chico - arbol_chico.score(lotes_ent, enf_ent)) < 1e-9 and abs(te_grande - arbol_grande.score(lotes_prb, enf_prb)) < 1e-9, "Pista 🧸: tr_* usa lotes_ent y enf_ent; te_* usa lotes_prb y enf_prb."
assert tr_grande == 1.0 and tr_grande > tr_chico, "El árbol grande memoriza: al repasar los lotes que ya vio acierta el 100 %."
assert te_chico > te_grande, "En el examen (lotes nuevos) el árbol chiquito debe ganarle al que memorizó, porque el 20 % de las anotaciones tiene errores y el grande los aprendió."
print(f"✅ ¡Reto 2 superado! ARI con la misma escala {ari_esc:.2f} y sin ella {ari_crudo:.2f}. Árbol grande: {tr_grande:.2f} al aprender pero {te_grande:.2f} en el examen; el chiquito: {tr_chico:.2f} y {te_chico:.2f}.")
```

## Reto 3: El cuaderno de Doña Rosalía y los terneros sospechosos 📒🐄

**Parte A: la libreta.** En `libreta` las veredas están escritas de **nueve formas** (mayúsculas, minúsculas, espacios de más) pero en realidad son **tres**. Además, `cabezas = -1` quiere decir «no se contó»: **no** es un dato. Las filas raras no se botan a la caneca: se separan en una **cuarentena** y se lleva la cuenta (lo que entra = lo limpio + la cuarentena).

**Parte B: los terneros.** `terneros` tiene dos columnas (altura en cm, peso en kg) y los **dos últimos** terneros son los sospechosos de Doña Rosalía. Por separado, su altura y su peso parecen normales (a menos de 2 desviaciones del promedio); pero un ternero alto debería pesar más, y estos **no cuadran juntos**. La herramienta que mira las dos medidas a la vez es la **distancia de Mahalanobis** (`d2`): si pasa de **9.21**, el ternero es raro con una confianza del 99 % (para dos medidas).

**Pasos**
1. `vereda_limpia`: la columna `vereda` sin espacios en los bordes y en minúsculas; `n_antes` y `n_despues`: cuántas veredas distintas había antes y después.
2. `cuarentena`: las filas con `cabezas` negativas; `limpias`: las demás; `entran`: cuántas filas tiene `libreta`.
3. `z`: cada columna de `terneros` en desviaciones (promedio 0, desviación 1); `z_max_sosp`: el mayor valor absoluto de `z` entre los **dos últimos** terneros.
4. `d2`: la distancia de Mahalanobis de cada ternero; `d2_sosp`: la de los dos últimos; `n_marcados`: cuántos terneros pasan de 9.21.

**Puntaje sugerido:** 30 puntos.

```python
# Casillas que debes llenar
vereda_limpia = n_antes = n_despues = cuarentena = limpias = entran = None
z = z_max_sosp = d2 = d2_sosp = n_marcados = None

# 💡 1) Nombres: quita espacios de los bordes y pasa a minúsculas
vereda_limpia = libreta["vereda"].str.strip().str.lower()
n_antes = libreta["..."].nunique()
n_despues = vereda_limpia.nunique()

# 💡 2) Cuarentena para el -1 (no se contó); lo demás es limpio
cuarentena = libreta[libreta["cabezas"] < ...]
limpias = libreta[libreta["cabezas"] >= 0]
entran = len(libreta)

# 💡 3) Cada medida por separado, en desviaciones
z = (terneros - terneros.mean(axis=0)) / terneros.std(axis=0)
z_max_sosp = np.abs(z[-...:]).max()

# 💡 4) Las dos medidas juntas (distancia de Mahalanobis)
d2 = EmpiricalCovariance().fit(terneros).mahalanobis(...)
d2_sosp = d2[-2:]
n_marcados = int(np.sum(d2 > ...))
```

```python
# Autoverificación del Reto 3 (no modificar)
assert vereda_limpia is not None and n_antes == 9, "Pista 🧸: libreta['vereda'].nunique() cuenta las nueve formas de escribir (son 9)."
assert n_despues == 3, "Pista 🧸: tras .str.strip().str.lower() deben quedar solo 3 veredas distintas."
assert cuarentena is not None and len(cuarentena) == 2 and (cuarentena["cabezas"] == -1).all(), "Pista 🧸: son 2 filas con cabezas = -1; filtra con libreta[libreta['cabezas'] < 0]."
assert limpias is not None and len(limpias) == 10 and (limpias["cabezas"] >= 0).all(), "Pista 🧸: las limpias son las filas con cabezas mayor o igual que 0 (deben ser 10)."
assert entran == len(limpias) + len(cuarentena) == 12, "Pista 🧸: entran = 12 filas y debe ser igual a limpias + cuarentena (no se perdió ninguna)."
assert z is not None and z.shape == terneros.shape and np.allclose(z.mean(axis=0), 0, atol=1e-9), "Pista 🧸: z = (terneros - terneros.mean(axis=0)) / terneros.std(axis=0)."
assert abs(z_max_sosp - np.abs(z[-2:]).max()) < 1e-9 and z_max_sosp < 2, "Pista 🧸: z_max_sosp = np.abs(z[-2:]).max(). Por separado, ninguna de sus medidas pasa de 2 desviaciones: ¡parecen normales!"
assert d2 is not None and len(d2) == len(terneros) and np.allclose(d2, EmpiricalCovariance().fit(terneros).mahalanobis(terneros)), "Pista 🧸: EmpiricalCovariance().fit(terneros).mahalanobis(terneros) da una distancia por ternero."
assert np.all(np.asarray(d2_sosp) > 9.21), "Revisa 🧸: d2_sosp son las dos últimas distancias, d2[-2:]; juntas, las dos medidas de estos terneros SÍ son raras (pasan de 9.21)."
assert n_marcados == int(np.sum(d2 > 9.21)) and 2 <= n_marcados <= 6, "Pista 🧸: cuenta cuántos d2 pasan de 9.21 con np.sum(d2 > 9.21); deben estar los 2 sospechosos y pocos más."
print(f"✅ ¡Reto 3 superado! {n_antes} formas de escribir → {n_despues} veredas | entran {entran} = {len(limpias)} limpias + {len(cuarentena)} en cuarentena | los sospechosos: |z| máximo {z_max_sosp:.2f} por separado, pero distancia {d2_sosp.min():.0f} o más juntos | {n_marcados} terneros marcados.")
```
