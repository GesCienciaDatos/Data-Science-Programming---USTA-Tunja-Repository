## 🎈 ¿Qué vamos a hacer en este taller?

Doña Rosa vende **panes de 100 gramos** (eso dice la etiqueta). Un cliente sospecha que pesan menos, y además asegura que el pan de Don Pedro, el vecino, es más liviano que el de ella. No puedes pesar todos los panes del mundo, pero sí unos cuantos. Vas a:

1. 🎣 **Lanzar la red:** con 25 panes pesados, construir un **intervalo** que atrape el peso promedio de *todos* los panes de Doña Rosa, «volviendo a sacar de la bolsa» los mismos panes muchas veces (*bootstrap*).
2. ⚖️ **Dictar el juicio:** con 15 panes de cada panadería, decidir si la diferencia es **real o suerte**, revolviendo las etiquetas.
3. ✍️ **Contarlo con tus palabras.**

```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np

_rng = np.random.default_rng(21)
panes = _rng.normal(96, 5, 25).round(1)       # 25 panes de Doña Rosa pesados al azar (gramos); la etiqueta dice 100 g
rosa = _rng.normal(100, 4, 15).round(1)       # 15 panes de Doña Rosa (gramos)
pedro = _rng.normal(96.5, 4, 15).round(1)       # 15 panes de Don Pedro (gramos)
print("Panes pesados:", len(panes), "| promedio de los 25:", round(panes.mean(), 1), "g")
```

## Reto 1: La red de pesca con los mismos 25 panes 🎣

El promedio de tus 25 panes es un solo número, pero otra tanda de 25 panes habría dado otro. Para ver cuánto «baila», **vuelve a sacar de la bolsa**: arma una muestra nueva de 25 panes **sacando con reemplazo** de tus 25 (un pan puede salir repetido), calcula su promedio y repite **5 000 veces**. Los promedios que quedan en el medio, entre el 2.5 % y el 97.5 %, forman tu **intervalo de 95 %**.

**Pasos**
1. `promedio`: el promedio de `panes`.
2. `medias_boot`: lista de 5 000 promedios de muestras de 25 panes sacadas **con reemplazo** de `panes`.
3. `ic_inf` e `ic_sup`: los percentiles 2.5 y 97.5 de `medias_boot` (`np.percentile`).
4. `incluye_100`: `True` si 100 g queda **dentro** del intervalo.

**Puntaje sugerido:** 40 puntos.

```python
# Casillas que debes llenar (empiezan vacías; no las borres)
promedio = medias_boot = ic_inf = ic_sup = incluye_100 = None

rng = np.random.default_rng(1)

# 💡 1) El promedio de los 25 panes
promedio = panes.mean()

# 💡 2) Volver a sacar de la bolsa 5 000 veces (con reemplazo)
medias_boot = [rng.choice(panes, ..., replace=True).mean() for _ in range(5000)]   # pista: cada muestra tiene 25 panes

# 💡 3) El intervalo de 95 %: lo que queda entre el 2.5 % y el 97.5 %
ic_inf = np.percentile(medias_boot, ...)
ic_sup = np.percentile(medias_boot, 97.5)

# 💡 4) ¿Cabe el 100 g de la etiqueta dentro del intervalo?
incluye_100 = bool(ic_inf <= ... <= ic_sup)
```

```python
# Autoverificación del Reto 1 (no modificar)
assert promedio is not None and abs(promedio - panes.mean()) < 1e-12, "Pista 🧸: promedio = panes.mean()."
assert medias_boot is not None and len(medias_boot) == 5000, "Pista 🧸: son 5 000 promedios, uno por cada repetición del ciclo."
assert abs(np.mean(medias_boot) - panes.mean()) < 0.3, "Pista 🧸: cada muestra debe tener 25 panes sacados con reemplazo: rng.choice(panes, 25, replace=True). Sus promedios deben girar alrededor del promedio original."
assert 0.5 < np.std(medias_boot) < 2.5, "Pista 🧸: los promedios deben bailar poco (≈ 1 g). Si bailan mucho o nada, revisa el tamaño de cada muestra (25)."
assert ic_inf is not None and ic_sup is not None and ic_inf < promedio < ic_sup, "Pista 🧸: ic_inf = np.percentile(medias_boot, 2.5) e ic_sup = np.percentile(medias_boot, 97.5); el promedio debe quedar adentro."
assert abs(ic_inf - np.percentile(medias_boot, 2.5)) < 1e-9 and abs(ic_sup - np.percentile(medias_boot, 97.5)) < 1e-9, "Los extremos son los percentiles 2.5 y 97.5 de medias_boot."
assert incluye_100 == bool(ic_inf <= 100 <= ic_sup), "Pista 🧸: incluye_100 es True solo si ic_inf <= 100 <= ic_sup."
print(f"✅ ¡Reto 1 superado! Con 95 % de confianza, el peso promedio de los panes de Doña Rosa está entre {ic_inf:.1f} y {ic_sup:.1f} g | ¿incluye los 100 g de la etiqueta? {incluye_100}")
```

## Reto 2: ¿El pan de Rosa pesa más que el de Pedro, o fue suerte? ⚖️

Pesaste 15 panes de cada panadería (`rosa` y `pedro`). El promedio de Rosa es mayor, pero ¿fue suerte? Si las dos panaderías fueran **iguales**, las etiquetas «Rosa» y «Pedro» no importarían: puedes **revolverlas** 5 000 veces y ver qué tan seguido sale, solo por revolver, una diferencia tan grande como la que viste.

**Pasos**
1. `dif_obs`: promedio de `rosa` menos promedio de `pedro`.
2. Junta los 30 panes en `todos`. En cada una de las 5 000 repeticiones revuelve (`rng.permutation(todos)`), toma los primeros 15 como «Rosa» y los otros 15 como «Pedro», y guarda la diferencia de promedios en `difs`.
3. `p_perm`: la fracción de repeticiones en que la diferencia revuelta, **en valor absoluto**, es igual o mayor que `abs(dif_obs)`.
4. `decision`: `"diferencia real"` si `p_perm < 0.05`; si no, `"puede ser suerte"`.

**Puntaje sugerido:** 35 puntos.
