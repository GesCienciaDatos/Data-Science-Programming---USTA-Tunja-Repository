```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA

SEMILLA = 11
rng = np.random.default_rng(SEMILLA)

# Reto 1: ventas diarias de la tienda (miles de pesos). ¡Una está mal anotada!
ventas = [120, 135, 128, 150, 142, 138, 5000, 131, 147, 126, 133, 140, 129, 145, 136]
# y 100 personas que lanzan 10 monedas (cuántas caras sacó cada una)
caras = rng.binomial(10, 0.5, 100)

# Reto 2: cuaderno de 4 semanas. Los domingos (cuando más se vende) NO se anotó nada
dias = ["lun", "mar", "mié", "jue", "vie", "sáb", "dom"] * 4
tipico = {"lun": 120, "mar": 125, "mié": 130, "jue": 135, "vie": 160, "sáb": 220, "dom": 330}
venta_real = np.array([tipico[d] for d in dias]) + rng.integers(-10, 11, 28)
cuaderno = pd.DataFrame({"dia": dias, "venta_real": venta_real,
                         "venta_cuaderno": np.where(np.array(dias) == "dom", np.nan, venta_real)})

# Reto 3: 12 papas con su altura (cm) y su peso (decenas de gramos)
altura = np.round(rng.normal(8, 1.5, 12), 1)
peso = np.round(1.1 * altura + rng.normal(0, 0.4, 12), 1)
papas = np.column_stack([altura, peso])

print("Ventas anotadas:", len(ventas), "| personas con monedas:", len(caras), "| días del cuaderno:", len(cuaderno), "| papas:", len(papas))
```

## Reto 2: Las hojas con huecos 🕳️

En el `cuaderno` de Don Hernando no se anotó **ningún domingo** (los huecos se ven como `NaN`). Justo los domingos son los días que **más** se vende. Como conocemos la columna `venta_real` (solo aquí, ¡en la vida real no la tendrías!), podemos medir el daño. Luego probarás la tentación de «rellenar con el promedio».

**Pasos**
1. `n_huecos`: cuántos días no tienen venta anotada (`venta_cuaderno`).
2. `prom_real` (de `venta_real`) y `prom_visible` (de `venta_cuaderno`; `pandas` ignora los `NaN`). Luego `sesgo = prom_visible - prom_real`.
3. `relleno`: la columna `venta_cuaderno` con los huecos rellenados con `prom_visible`; `prom_relleno` y las desviaciones `sd_visible` (de `venta_cuaderno`) y `sd_relleno` (de `relleno`).

## Reto 3: La sombra de la papa 🥔

Las 12 `papas` tienen dos medidas, **altura** y **peso**, que casi dicen lo mismo: las papas altas pesan más. Cuando dos columnas se parecen tanto, se pueden resumir en **una sola**, como la **sombra** de las papas proyectada en una pared. Esa es la idea del PCA: guardar la sombra que **más información conserva**.

**Pasos**
1. `pca`: un `PCA(n_components=1)` ajustado con `papas`. `porcentaje`: el porcentaje de la información (varianza) que conserva la sombra (`explained_variance_ratio_[0]` por 100).
2. `sombra`: los 12 números de la sombra (`pca.transform(papas)`); `regreso`: lo que se recupera de ellos (`pca.inverse_transform(sombra)`); `error_medio`: el promedio de las diferencias absolutas entre `papas` y `regreso`.
3. `error_dos`: lo mismo pero con **dos** sombras (`PCA(n_components=2)`): ¿se pierde algo?

**Puntaje sugerido:** 25 puntos.

```python
# Casillas que debes llenar
pca = porcentaje = sombra = regreso = error_medio = error_dos = None

# 💡 1) Una sola sombra
pca = PCA(n_components=...).fit(papas)
porcentaje = pca.explained_variance_ratio_[0] * 100

# 💡 2) La sombra y lo que se recupera de ella
sombra = pca.transform(...)
regreso = pca.inverse_transform(sombra)
error_medio = float(np.mean(np.abs(papas - regreso)))

# 💡 3) Con dos sombras no se pierde nada
pca2 = PCA(n_components=...).fit(papas)
error_dos = float(np.mean(np.abs(papas - pca2.inverse_transform(pca2.transform(papas)))))
```
