## Reto 1: ¿Cuánto se espera «normalmente»? 🕒

Si sumas los 15 tiempos y divides, sale un número; si ordenas y tomas el del medio, sale otro. Con un dato raro (95 minutos) **no coinciden**. Un dato es **raro** si pasa de `q3 + 1.5 * iqr` (o es menor que `q1 - 1.5 * iqr`).

**Pasos**
1. `promedio` y `mediana` de `espera`.
2. `promedio_sin_puntas`: ordena los tiempos y promedia **sin** el menor y **sin** el mayor.
