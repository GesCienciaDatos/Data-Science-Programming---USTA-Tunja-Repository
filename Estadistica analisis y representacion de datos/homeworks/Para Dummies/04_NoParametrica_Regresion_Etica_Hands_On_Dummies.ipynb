## Reto 2: La línea de la panela y la correlación tramposa 📏

Don Joaquín anotó cuántos kilos de **panela** salieron de cada cantidad de **arrobas de caña** (`cana`, `panela`). Quiere una **línea** para predecir.

**Pasos**
1. `pendiente` y `corte`: la línea que menos se equivoca. La pendiente es la suma de (x − promedio de x) × (y − promedio de y), dividida entre la suma de (x − promedio de x) al cuadrado; el corte es el promedio de y menos la pendiente por el promedio de x.
2. `pred_15`: los kilos esperados con **15** arrobas (¡dentro del rango observado!) y `fuera_de_rango`: `True` si pedirle a la línea una predicción para **40** arrobas sería salirse de los datos (extrapolar).
3. **Correlación tramposa:** `x_u` va de −5 a 5 y `y_u = x_u ** 2` (una relación perfecta, pero en forma de U). `r_u` es su correlación con `np.corrcoef`.

**Puntaje sugerido:** 30 puntos.

```python
# Casillas que debes llenar
pendiente = corte = pred_15 = fuera_de_rango = r_u = None

# 💡 1) La línea que menos se equivoca
xm, ym = np.mean(cana), np.mean(...)
pendiente = np.sum((cana - xm) * (panela - ...)) / np.sum((cana - xm) ** 2)
corte = ym - pendiente * ...

# 💡 2) Predecir dentro (15) y ver si 40 se sale del rango de los datos
pred_15 = corte + pendiente * ...
fuera_de_rango = 40 > cana.max() or 40 < cana.min()     # ¿40 está fuera de [mínimo, máximo]?

# 💡 3) Una relación perfecta que la correlación no ve
x_u = np.arange(-5, 6)
y_u = x_u ** 2
r_u = np.corrcoef(x_u, ...)[0, 1]
```

```python
# Autoverificación del Reto 2 (no modificar)
_b1, _b0 = np.polyfit(cana, panela, 1)
assert pendiente is not None and abs(pendiente - _b1) < 1e-9, "Pista 🧸: pendiente = suma((cana - xm) * (panela - ym)) / suma((cana - xm) ** 2). En el hueco va ym, el promedio de la panela."
assert corte is not None and abs(corte - _b0) < 1e-9, "Pista 🧸: corte = ym - pendiente * xm (el promedio de la caña va en el último hueco)."
assert 1.0 < pendiente < 1.3, "Revisa: por cada arroba de caña deben salir poco más de un kilo de panela, no más ni menos."
assert pred_15 is not None and abs(pred_15 - (_b0 + _b1 * 15)) < 1e-9, "Pista 🧸: pred_15 = corte + pendiente * 15."
assert fuera_de_rango is not None and bool(fuera_de_rango) is True, "Pista 🧸: los datos van de 2 a 20 arrobas; 40 está fuera, así que predecir ahí es EXTRAPOLAR y puede fallar."
assert r_u is not None and abs(r_u) < 1e-9, "Pista 🧸: r_u = np.corrcoef(x_u, y_u)[0, 1]. Si lo hiciste bien da 0: la correlación solo ve líneas rectas."
print(f"✅ ¡Reto 2 superado! Cada arroba de caña da ≈ {pendiente:.2f} kg de panela; con 15 arrobas esperas {pred_15:.1f} kg. Predecir con 40 sería extrapolar. Y la U perfecta tiene correlación {r_u:.2f}: ¡la correlación no ve curvas!")
```

## Reto 3: La libreta de la tienda y los veinte colores de camisa 🔐

Don Joaquín quiere compartir su **libreta** (`libreta`) sin los nombres. Pero hay personas **únicas** con su vereda, grupo de edad y oficio: quien conozca a esa persona la reconoce. Además, un amigo probó **20 colores de camisa** buscando cuál «trae buena suerte» (`p_colores`, un número entre 0 y 1 por color: mientras más pequeño, más «sorprendente»).

**Pasos**
1. `tamano_grupo`: cuántas personas comparten **exactamente** la misma vereda, edad y oficio (`groupby(...).size()`), y `n_unicas`: cuántos grupos tienen **una sola** persona.
2. `prob_falso`: la probabilidad de que **al menos uno** de 20 colores salga «significativo» (p < 0.05) por puro azar, aunque ninguno sirva: `1 - 0.95 ** 20`.
3. `n_crudo`: cuántos colores tienen `p < 0.05`; `umbral`: el umbral de **Bonferroni** (0.05 dividido entre 20) y `n_bonf`: cuántos colores quedan con `p <= umbral`.

**Puntaje sugerido:** 20 puntos.

```python
# Casillas que debes llenar
tamano_grupo = n_unicas = prob_falso = n_crudo = umbral = n_bonf = None

# 💡 1) ¿Cuántas personas comparten vereda, edad y oficio?
tamano_grupo = libreta.groupby(["vereda", "edad", ...]).size()
n_unicas = int((tamano_grupo == ...).sum())       # grupos con UNA sola persona

# 💡 2) Veinte intentos, cada uno con 5 % de falsa alarma
prob_falso = 1 - 0.95 ** ...

# 💡 3) Hallazgos «crudos» frente a Bonferroni
n_crudo = int((p_colores < ...).sum())
umbral = 0.05 / ...
n_bonf = int((p_colores <= umbral).sum())
```

```python
# Autoverificación del Reto 3 (no modificar)
_g = libreta.groupby(["vereda", "edad", "oficio"]).size()
assert tamano_grupo is not None and len(tamano_grupo) == len(_g) and tamano_grupo.sum() == 12, "Pista 🧸: agrupa por las TRES columnas: groupby(['vereda', 'edad', 'oficio']).size()."
assert n_unicas == int((_g == 1).sum()) and n_unicas >= 3, "Pista 🧸: cuenta los grupos de tamaño 1 con (tamano_grupo == 1).sum(). Esas personas se pueden reconocer."
assert prob_falso is not None and abs(prob_falso - (1 - 0.95 ** 20)) < 1e-12, "Pista 🧸: prob_falso = 1 - 0.95 ** 20 (se multiplica 20 veces el 95 % de «no me equivoco»)."
assert n_crudo == 4, "Pista 🧸: con p < 0.05 hay 4 colores «ganadores»: 0.0008, 0.04, 0.012 y 0.047."
assert umbral is not None and abs(umbral - 0.0025) < 1e-12, "Pista 🧸: umbral = 0.05 / 20 = 0.0025."
assert n_bonf == 1, "Pista 🧸: solo el color con p = 0.0008 cumple p <= 0.0025."
print(f"✅ ¡Reto 3 superado! {n_unicas} grupos tienen una sola persona (reconocible aunque no haya nombres). Con 20 colores la probabilidad de algún «hallazgo» falso es {prob_falso:.0%}: sin corregir salen {n_crudo}, con Bonferroni solo {n_bonf}.")
```
