## 1. La Distribución Chi-Cuadrado y la Prueba de Bondad de Ajuste 🎲

**Distribución.** Si $Z_1,\dots,Z_k$ son normales estándar independientes, $Q=\sum_{i=1}^{k} Z_i^2\sim\chi^2_k$, con $\mathbb{E}[Q]=k$ y $\mathrm{Var}[Q]=2k$. Es **asimétrica a la derecha** y se acerca a la normal al crecer $k$. De ahí salen la prueba de bondad de ajuste, la de independencia y el intervalo para la varianza del primer cuaderno.

**Bondad de ajuste.** Con $n$ observaciones clasificadas en $k$ categorías mutuamente excluyentes y exhaustivas, $H_0:\;p_i=p_{i0}$ para todo $i$. Con las **frecuencias esperadas** $E_i=n\,p_{i0}$ y las **observadas** $O_i$, el estadístico de **Pearson** es

$$\chi^2=\sum_{i=1}^{k}\frac{(O_i-E_i)^2}{E_i}\;\overset{\text{aprox.}}{\sim}\;\chi^2_{k-1-m}\quad(\text{bajo }H_0),$$

donde $m$ es el número de parámetros **estimados** de los datos (0 si las $p_{i0}$ están fijadas de antemano). **Condiciones:** observaciones **independientes** y esperadas **no demasiado pequeñas** (la regla práctica es $E_i\ge5$; es una guía, lo comprobaremos). La prueba es **omnibus**: dice **que** hay desviación, no **dónde**; para eso se miran los **residuos de Pearson** $(O_i-E_i)/\sqrt{E_i}$. Una medida de tamaño del efecto es la $w$ de Cohen, $w=\sqrt{\chi^2/n}$.

La celda comprueba la definición por simulación, implementa la prueba, la contrasta con `scipy.stats.chisquare`, **mide su calibración y su potencia** con dados sintéticos y muestra cuándo la aproximación se resiente.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas scipy statsmodels matplotlib

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy import stats
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from statsmodels.stats.oneway import anova_oneway

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {__import__('scipy').__version__} | statsmodels {__import__('statsmodels').__version__}")
````

````python
# ============================================================
# 1. La distribucion chi-cuadrado y la prueba de bondad de ajuste
# ============================================================
# (a) chi2_k = suma de k normales estandar al cuadrado (comprobacion por simulacion)
k = 3
Q = (rng.standard_normal((100000, k)) ** 2).sum(axis=1)
ks = stats.kstest(Q, "chi2", args=(k,)).statistic
print(f"chi2({k}): media = {Q.mean():.3f} (teoria {k}) | varianza = {Q.var():.3f} (teoria {2 * k}) | KS D = {ks:.4f}")
assert abs(Q.mean() - k) < 0.05 and abs(Q.var() - 2 * k) < 0.2 and ks < 0.01


# (b) Bondad de ajuste: estadistico de Pearson desde cero
def chi2_bondad(obs, probs):
    obs = np.asarray(obs, float)
    esp = obs.sum() * np.asarray(probs, float)
    est = ((obs - esp) ** 2 / esp).sum()
    gl = len(obs) - 1
    return est, gl, stats.chi2.sf(est, gl), esp


# Dos dados SINTETICOS lanzados 300 veces: uno justo y otro cargado (la cara 6 sale con prob. 0.25 en vez de 0.167)
justo = np.bincount(rng.integers(1, 7, 300), minlength=7)[1:]
cargado = np.bincount(rng.choice(np.arange(1, 7), 300, p=[0.15, 0.15, 0.15, 0.15, 0.15, 0.25]), minlength=7)[1:]
p0 = np.full(6, 1 / 6)
for nombre, obs in (("justo", justo), ("cargado", cargado)):
    est, gl, p, esp = chi2_bondad(obs, p0)
    ref = stats.chisquare(obs, esp)
    assert np.allclose((est, p), (ref.statistic, ref.pvalue))
    print(f"Dado {nombre:8s} frecuencias = {obs.tolist()} | chi2 = {est:.2f} (gl = {gl}) | p = {p:.4f}")

# (c) Calibracion: tasa de rechazo con 5 000 dados justos (error tipo I) y con 5 000 cargados (potencia), n = 300
def stat_matriz(C, probs):
    esp = C.sum(axis=1, keepdims=True) * probs
    return ((C - esp) ** 2 / esp).sum(axis=1)


Cj = rng.multinomial(300, p0, 5000)
Cc = rng.multinomial(300, [0.15, 0.15, 0.15, 0.15, 0.15, 0.25], 5000)
crit = stats.chi2.ppf(0.95, 5)
tI, pot = np.mean(stat_matriz(Cj, p0) > crit), np.mean(stat_matriz(Cc, p0) > crit)
print(f"n = 300: rechazos con dado justo = {tI:.3f} (nominal 0.05) | rechazos con dado cargado (potencia) = {pot:.3f}")
assert abs(tI - 0.05) < 0.015 and 0.2 < pot < 0.9

# (d) Frecuencias esperadas pequenas: la regla "E >= 5" es una guia conservadora, pero con E << 1 en varias categorias el nivel real se infla
C20 = rng.multinomial(20, p0, 20000)                                          # equiprobable, n = 20 -> E = 3.3 en todas
p_sesgada = np.array([0.01, 0.02, 0.03, 0.04, 0.10, 0.80])                    # n = 30 -> E = 0.3, 0.6, 0.9, 1.2, 3.0, 24
C30 = rng.multinomial(30, p_sesgada, 20000)
tI20 = np.mean(stat_matriz(C20, p0) > crit)
tI30 = np.mean(stat_matriz(C30, p_sesgada) > crit)
print(f"Rechazos con H0 cierta usando la chi2 asintotica: equiprobable n = 20 (E = 3.3): {tI20:.3f} | muy desigual n = 30 (E de 0.3 a 24): {tI30:.3f} (nominal 0.05)")
assert abs(tI20 - 0.05) < 0.02 and tI30 > 0.07

xs = np.linspace(0.01, 25, 500)
est_c = chi2_bondad(cargado, p0)[0]
fig, ax = plt.subplots(figsize=(9, 4.2))
for df_, col in zip((1, 3, 5, 10), PALETA):
    ax.plot(xs, stats.chi2.pdf(xs, df_), color=col, label=f"gl = {df_}")
ax.fill_between(xs, stats.chi2.pdf(xs, 5), where=xs >= est_c, color=PALETA[2], alpha=0.3)
ax.axvline(est_c, color="black", ls="--"); ax.annotate(f"dado cargado: chi2 = {est_c:.1f}", (est_c, 0.15), xytext=(6, 0), textcoords="offset points")
ax.set(ylim=(0, 0.5), xlabel="chi2", ylabel="densidad", title="Distribucion chi-cuadrado (el area sombreada es el valor p para gl = 5)"); ax.legend()
plt.tight_layout(); plt.show()
````

**Lectura.** (1) La suma de 3 normales al cuadrado tiene media **3.004** y varianza **6.069** (teoría: 3 y 6), con $D=0.0016$ de Kolmogorov–Smirnov: la definición se cumple. (2) **Dado justo** ($n=300$): $\chi^2=4.64$ con 5 gl, **$p=0.4614$** → no se rechaza. **Dado cargado:** $\chi^2=15.00$, **$p=0.0104$** → se rechaza; el culpable se ve en las frecuencias: la cara 6 salió **73 veces frente a 50 esperadas** (residuo de Pearson $\approx 3.25$). (3) **Calibración:** con dados justos la prueba rechazó **5.0 %** de las veces (nominal 5 %) y con el dado cargado detectó la trampa en **82.2 %** (potencia): con $n=300$ un sesgo de 0.25 frente a 0.167 es detectable **casi siempre**, pero **no siempre**. (4) **Frecuencias esperadas pequeñas:** con un dado equiprobable y $n=20$ ($E=3.3$) la tasa real fue **4.7 %**: la regla $E\ge5$ es **conservadora**. Pero con probabilidades muy desiguales (esperadas de 0.3 a 24) y $n=30$ el nivel real subió a **8.2 %** en vez de 5 %: ahí conviene **agrupar categorías** o usar una prueba exacta o por simulación (Monte Carlo).

## 2. Independencia en Tablas de Contingencia 📊

Una **tabla de contingencia** $r\times c$ cuenta los casos por pareja de categorías de dos variables. $H_0$: las variables son **independientes** (el uso de riego **no** depende del cultivo); la misma cuenta sirve para **homogeneidad** (comparar la distribución de una variable en varias poblaciones; cambia el diseño de muestreo, no el cálculo). Bajo independencia, las esperadas son

$$E_{ij}=\frac{(\text{total fila } i)(\text{total columna } j)}{N},\qquad \chi^2=\sum_{i,j}\frac{(O_{ij}-E_{ij})^2}{E_{ij}}\sim\chi^2_{(r-1)(c-1)}\ (\text{aprox.}).$$

**Más allá del valor p:** el **V de Cramér**, $V=\sqrt{\chi^2/\big(N\min(r-1,\,c-1)\big)}\in[0,1]$, mide la **fuerza** de la asociación (con $N$ grande, $p$ diminuto con $V$ casi nulo es posible); los **residuos estandarizados ajustados**, $\dfrac{O_{ij}-E_{ij}}{\sqrt{E_{ij}\,(1-\text{fila}_i/N)(1-\text{col}_j/N)}}$, que son aproximadamente $N(0,1)$, indican **qué celdas** se apartan (|valor| > 2 es notable).

**Tablas pequeñas.** Si alguna esperada es < 5, la aproximación chi-cuadrado es dudosa. En $2\times2$ se usa la **corrección de continuidad de Yates**, $\sum(|O-E|-0.5)^2/E$ (conservadora), o mejor la **prueba exacta de Fisher**, basada en la distribución hipergeométrica. **Asociación no es causalidad** (ni descarta confusores: ver la paradoja de Simpson del Módulo 00).

Usamos una **encuesta sintética** de 400 fincas (cultivo × riego tecnificado, con una asociación que fijamos nosotros) y la verificamos con `chi2_contingency`, con `association` de SciPy y con una **prueba de permutación**.

````python
# ============================================================
# 2. Prueba de independencia con tablas de contingencia
# ============================================================
# Encuesta SINTETICA de 400 fincas: tipo de cultivo y uso de riego tecnificado (la asociacion la fijamos nosotros)
cultivos, n_c, p_riego = ["papa", "cebolla", "maiz"], [150, 130, 120], [0.55, 0.35, 0.30]
cultivo = np.repeat(cultivos, n_c)
riego = np.concatenate([np.where(rng.random(n) < p, "si", "no") for n, p in zip(n_c, p_riego)])
encuesta = pd.DataFrame({"cultivo": cultivo, "riego": riego})
tabla = pd.crosstab(encuesta.cultivo, encuesta.riego).loc[cultivos]
print(tabla, "\n")


def chi2_independencia(tab):
    O = np.asarray(tab, float)
    N = O.sum()
    fila, col = O.sum(axis=1, keepdims=True), O.sum(axis=0, keepdims=True)
    E = fila @ col / N                                                       # esperadas bajo independencia
    est = ((O - E) ** 2 / E).sum()
    gl = (O.shape[0] - 1) * (O.shape[1] - 1)
    V = np.sqrt(est / (N * min(O.shape[0] - 1, O.shape[1] - 1)))             # V de Cramer
    res = (O - E) / np.sqrt(E * (1 - fila / N) * (1 - col / N))              # residuos estandarizados ajustados
    return est, gl, stats.chi2.sf(est, gl), E, V, res


est, gl, p, E, V, res = chi2_independencia(tabla)
ref = stats.chi2_contingency(tabla, correction=False)
assert np.allclose((est, p, gl), (ref.statistic, ref.pvalue, ref.dof)) and np.allclose(E, ref.expected_freq)
assert np.allclose(stats.contingency.association(tabla.values, method="cramer"), V)
print(f"chi2 = {est:.3f} | gl = {gl} | p = {p:.5f} | V de Cramer = {V:.3f} | esperada minima = {E.min():.1f}")
print("Residuos estandarizados ajustados:\n", pd.DataFrame(res, index=tabla.index, columns=tabla.columns).round(2).to_string())
print("Proporcion con riego por cultivo:", (tabla["si"] / tabla.sum(axis=1)).round(3).to_dict())
assert E.min() >= 5 and p < 0.05

# Verificacion por permutacion: barajar la columna 'riego' rompe cualquier asociacion
codigos_c = pd.factorize(encuesta.cultivo, sort=False)[0]
codigos_r = (encuesta.riego == "si").astype(int).to_numpy()
rp = np.random.default_rng(3)
stats_perm = np.empty(5000)
for b in range(5000):
    O = np.bincount(codigos_c * 2 + rp.permutation(codigos_r), minlength=6).reshape(3, 2).astype(float)
    Eb = O.sum(axis=1, keepdims=True) @ O.sum(axis=0, keepdims=True) / O.sum()
    stats_perm[b] = ((O - Eb) ** 2 / Eb).sum()
p_perm = (1 + np.sum(stats_perm >= est)) / 5001
print(f"p por permutacion = {p_perm:.5f} | p chi2 = {p:.5f} | cuantil 95 % de la nula simulada = {np.quantile(stats_perm, 0.95):.2f} (chi2 teorico {stats.chi2.ppf(0.95, 2):.2f})")
assert abs(np.quantile(stats_perm, 0.95) - stats.chi2.ppf(0.95, 2)) < 0.3

fig, axs = plt.subplots(1, 2, figsize=(11, 4))
im = axs[0].imshow(res, cmap="RdBu_r", vmin=-4, vmax=4)
axs[0].set(xticks=[0, 1], xticklabels=tabla.columns, yticks=range(3), yticklabels=tabla.index, title="Residuos estandarizados ajustados")
for i in range(3):
    for j in range(2):
        axs[0].text(j, i, f"{res[i, j]:.1f}", ha="center", va="center", fontweight="bold")
plt.colorbar(im, ax=axs[0], fraction=0.046); axs[0].grid(False)
axs[1].hist(stats_perm, bins=50, density=True, color=PALETA[4], alpha=0.6, label="chi2 bajo permutaciones")
xs = np.linspace(0.01, 14, 300); axs[1].plot(xs, stats.chi2.pdf(xs, 2), color="black", label="chi2 (gl = 2)")
axs[1].axvline(est, color=PALETA[1], ls="--", label=f"observado = {est:.1f}"); axs[1].set(xlabel="chi2", title="La nula simulada coincide con la teorica"); axs[1].legend()
plt.tight_layout(); plt.show()

# Una tabla 2 x 2 pequena (para la Practica 1): las frecuencias esperadas son < 5
tabla_peq = np.array([[8, 2], [3, 7]])
E_peq = tabla_peq.sum(axis=1, keepdims=True) @ tabla_peq.sum(axis=0, keepdims=True) / tabla_peq.sum()
print("\nTabla pequena:\n", tabla_peq, "| esperadas:\n", E_peq.round(2))
assert E_peq.min() < 5
````

**Lectura.** (1) Hay riego tecnificado en **48.7 %** de las fincas de papa, **33.1 %** de cebolla y **24.2 %** de maíz. La prueba da $\chi^2=18.16$ con 2 gl y **$p=0.00011$**: los datos son **muy poco compatibles** con independencia. Todas las esperadas son ≥ 43.5, así que la aproximación es fiable; coincide **exactamente** con `chi2_contingency` y con `association`. (2) **Fuerza:** $V=0.213$, una asociación **modesta**: el cultivo explica una parte pequeña de quién usa riego, aunque sea «muy significativa». ## 3. ANOVA de un Factor 🧮

**Modelo.** Para $k$ grupos con $n_i$ observaciones, $y_{ij}=\mu+\alpha_i+\varepsilon_{ij}$, con errores $\varepsilon_{ij}$ **independientes** y $N(0,\sigma^2)$. $H_0:\ \mu_1=\dots=\mu_k$ frente a «al menos una media difiere». La idea es **comparar la variación entre grupos con la variación dentro de los grupos**:

$$\underbrace{\sum_{i,j}(y_{ij}-\bar{y})^2}_{\mathrm{SCT}}=\underbrace{\sum_i n_i(\bar{y}_i-\bar{y})^2}_{\mathrm{SCE}\ (\text{entre})}+\underbrace{\sum_{i,j}(y_{ij}-\bar{y}_i)^2}_{\mathrm{SCD}\ (\text{dentro})},\qquad F=\frac{\mathrm{SCE}/(k-1)}{\mathrm{SCD}/(N-k)}=\frac{\mathrm{CM}_E}{\mathrm{CM}_D}\sim F_{k-1,\,N-k}\ (\text{bajo }H_0).$$

Si $H_0$ es cierta, ambos cuadrados medios estiman $\sigma^2$ y $F\approx1$; si las medias difieren, el numerador **crece**. **Tamaño del efecto:** $\eta^2=\mathrm{SCE}/\mathrm{SCT}$ (proporción de varianza explicada; sesgada hacia arriba) y $\omega^2=\dfrac{\mathrm{SCE}-(k-1)\,\mathrm{CM}_D}{\mathrm{SCT}+\mathrm{CM}_D}$ (menos sesgado).

**Supuestos:** (i) **independencia** (diseño); (ii) **normalidad de los residuos** (el ANOVA es bastante robusto con grupos balanceados y $n$ moderado); (iii) **varianzas iguales** (**homocedasticidad**): Levene (robusta) y Bartlett (sensible a la no normalidad). El ANOVA es **omnibus**: un $F$ significativo dice que **alguna** media difiere, **no cuál** (sección siguiente). Con $k=2$ el ANOVA **es** la prueba $t$ de Student agrupada ($F=t^2$). La extensión a **dos factores** (con interacción) sigue la misma lógica de descomposición de la variación (`anova_lm(ols("y ~ C(A) * C(B)", df).fit())`), pero no se desarrolla aquí.

Datos **sintéticos**: rendimiento de 4 variedades de papa, 12 parcelas cada una, con $\sigma=4$ común y medias reales 22, 24, 23 y 27 t/ha.

````python
# ============================================================
# 3. ANOVA de un factor desde cero
# ============================================================
# Rendimiento (t/ha) de 4 variedades de papa, 12 parcelas cada una (SINTETICO; sigma comun = 4)
medias_reales = {"A": 22.0, "B": 24.0, "C": 23.0, "D": 27.0}
gen3 = np.random.default_rng(12)                                           # semilla propia (con otras semillas F va de ~1.2 a ~9.4: potencia moderada)
rend = {v: gen3.normal(m, 4.0, 12) for v, m in medias_reales.items()}
datos = pd.DataFrame({"variedad": np.repeat(list(rend), 12), "rend": np.concatenate(list(rend.values()))})
grupos = list(rend.values())
k, N = len(grupos), sum(len(g) for g in grupos)

gm = datos.rend.mean()
ss_b = sum(len(g) * (g.mean() - gm) ** 2 for g in grupos)                  # entre grupos
ss_w = sum(((g - g.mean()) ** 2).sum() for g in grupos)                    # dentro de grupos
ss_t = ((datos.rend - gm) ** 2).sum()
ms_b, ms_w = ss_b / (k - 1), ss_w / (N - k)
F = ms_b / ms_w
p = stats.f.sf(F, k - 1, N - k)
eta2 = ss_b / ss_t                                                         # proporcion de varianza explicada
omega2 = (ss_b - (k - 1) * ms_w) / (ss_t + ms_w)                           # menos sesgado
print(pd.DataFrame({"gl": [k - 1, N - k, N - 1], "SC": [ss_b, ss_w, ss_t], "CM": [ms_b, ms_w, np.nan]}, index=["entre", "dentro", "total"]).round(3).to_string())
print(f"\nF = {F:.3f} (gl = {k - 1}, {N - k}) | p = {p:.5f} | eta^2 = {eta2:.3f} | omega^2 = {omega2:.3f}")
print("Medias muestrales:", {v: round(g.mean(), 2) for v, g in rend.items()})

# Verificaciones: descomposicion SCT = SCE + SCD, SciPy y statsmodels
assert np.isclose(ss_t, ss_b + ss_w)
assert np.allclose((F, p), stats.f_oneway(*grupos))
modelo = ols("rend ~ C(variedad)", datos).fit()
tab_sm = sm.stats.anova_lm(modelo, typ=1)
assert np.allclose(tab_sm.loc["C(variedad)", ["sum_sq", "F", "PR(>F)"]].astype(float), (ss_b, F, p))
assert np.isclose(tab_sm.loc["Residual", "sum_sq"], ss_w)
# Con k = 2 grupos, F = t^2 (la t de Student agrupada es un caso particular)
t2 = stats.ttest_ind(grupos[0], grupos[1])
f2 = stats.f_oneway(grupos[0], grupos[1])
assert np.allclose((t2.statistic**2, t2.pvalue), (f2.statistic, f2.pvalue))
assert 0 < omega2 < eta2 < 1

# Supuestos sobre los residuos
res_m = modelo.resid
print(f"Shapiro-Wilk de los residuos: p = {stats.shapiro(res_m).pvalue:.3f} | Levene: p = {stats.levene(*grupos).pvalue:.3f} | Bartlett: p = {stats.bartlett(*grupos).pvalue:.3f}")

fig, axs = plt.subplots(1, 3, figsize=(14, 4.2))
axs[0].boxplot(grupos, widths=0.5, showmeans=True); axs[0].set_xticks(range(1, 5), list(rend)); axs[0].set(ylabel="rendimiento (t/ha)", title=f"F = {F:.2f}, p = {p:.4f}")
stats.probplot(res_m, plot=axs[1]); axs[1].set_title("Q-Q de los residuos"); axs[1].get_lines()[0].set(color=PALETA[0], ms=4)
axs[2].scatter(modelo.fittedvalues, res_m, color=PALETA[4], alpha=0.7); axs[2].axhline(0, color="black", lw=1)
axs[2].set(xlabel="valores ajustados (media del grupo)", ylabel="residuo", title="Residuos vs. ajustados")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **Tabla ANOVA:** SCE = 244.4 (3 gl), SCD = 555.2 (44 gl), SCT = 799.5 (47 gl); $\mathrm{CM}_E=81.45$, $\mathrm{CM}_D=12.62$, **$F=6.455$ y $p=0.00102$**: se rechaza la igualdad de las cuatro medias. La descomposición SCT = SCE + SCD se cumple, y $F$, $p$ y las sumas de cuadrados coinciden con `f_oneway` y con `anova_lm`. (2) **Efecto:** $\eta^2=0.306$ y $\omega^2=0.254$: las variedades explican ≈ 25–31 % de la variación del rendimiento (el resto es variabilidad entre parcelas). (3) Las medias muestrales son 22.78, 24.40, 21.29 y 27.37 (reales: 22, 24, 23, 27): **el orden** de la variedad D se recupera, el de B y C **no**. (4) **Supuestos:** residuos compatibles con la normalidad (Shapiro–Wilk $p=0.641$; el Q–Q casi sigue la recta) y con varianzas iguales (Levene 0.652, Bartlett 0.759); el gráfico de residuos contra ajustados no muestra patrón evidente. (5) **Nota de honestidad:** esta semilla se eligió entre las 12 primeras por dar un caso «típico»; con otras semillas el $F$ va de ≈ 1.2 a ≈ 9.4 con **exactamente los mismos parámetros**: con 12 parcelas por grupo la **potencia es moderada** y a veces **no se detectan diferencias que existen**. (6) Un $F$ significativo **no dice cuáles** variedades difieren.

## 4. Comparaciones Múltiples: Tukey HSD 🔍

Con $k$ grupos hay $\binom{k}{2}$ pares. Si cada par se prueba al 5 %, la probabilidad de **algún falso positivo** (la tasa de error **por familia**, FWER) crece: con $k=4$ son 6 pruebas. Las soluciones:

- **Bonferroni:** rechazar el par $i$ si $\binom{k}{2}\,p_i<\alpha$ (simple, conservador). **Holm** (cuaderno anterior) lo mejora.
- **Tukey HSD** (*honestly significant difference*): usa la distribución del **rango estudentizado**; para $n_i$ distintos (Tukey–Kramer), el par $(i,j)$ es significativo si $|\bar{y}_i-\bar{y}_j|\big/\sqrt{\tfrac{\mathrm{CM}_D}{2}\left(\tfrac1{n_i}+\tfrac1{n_j}\right)}>q_{1-\alpha;\,k,\,N-k}$. Produce **intervalos de confianza simultáneos**.
- Si las comparaciones se **planearon antes** de ver los datos (**contrastes planificados**), se necesita menos corrección; con varianzas desiguales, **Games–Howell**.

La celda **(a)** simula 5 000 experimentos con **todas las medias iguales** para ver la FWER de cada estrategia y **(b)** implementa Tukey–Kramer desde cero con las 4 variedades, contrastándolo con `scipy.stats.tukey_hsd` y `pairwise_tukeyhsd` de statsmodels.

````python
# ============================================================
# 4. Comparaciones multiples: por que no basta con muchas pruebas t, y Tukey HSD
# ============================================================
# (a) Con k = 4 grupos hay 6 comparaciones por pares. Todas las medias IGUALES (H0 cierta), n = 12, 5 000 experimentos
reps, kk, n_ = 5000, 4, 12
X = rng.normal(0, 1, (reps, kk, n_))
m, msw = X.mean(axis=2), X.var(axis=2, ddof=1).mean(axis=1)
gl_w = kk * (n_ - 1)
pares = [(i, j) for i in range(kk) for j in range(i + 1, kk)]
d = np.stack([np.abs(m[:, i] - m[:, j]) for i, j in pares], axis=1)           # (reps, 6)
p_t = 2 * stats.t.sf(d / np.sqrt(msw * 2 / n_)[:, None], gl_w)                 # t por pares con el CM dentro
q_crit = stats.studentized_range.ppf(0.95, kk, gl_w)                            # Tukey: rechaza algun par <=> el rango estudentizado supera q_crit
fwer = {"F global": np.mean(stats.f_oneway(*[X[:, g] for g in range(kk)], axis=1).pvalue < 0.05),
        "6 pruebas t sin corregir": np.mean((p_t < 0.05).any(axis=1)),
        "6 pruebas t con Bonferroni": np.mean((p_t * 6 < 0.05).any(axis=1)),
        "Tukey HSD": np.mean(d.max(axis=1) / np.sqrt(msw / n_) > q_crit)}
print("Prob. de algun falso positivo con H0 cierta (nominal 0.05):\n", pd.Series(fwer).round(3).to_string())
assert 0.15 < fwer["6 pruebas t sin corregir"] < 0.27
assert all(fwer[k_] < 0.065 for k_ in ("F global", "6 pruebas t con Bonferroni", "Tukey HSD"))


# (b) Tukey-Kramer desde cero con las 4 variedades de la seccion 3
def tukey(grupos, nombres, conf=0.95):
    k, N = len(grupos), sum(len(g) for g in grupos)
    msw = sum(((g - g.mean()) ** 2).sum() for g in grupos) / (N - k)
    qc = stats.studentized_range.ppf(conf, k, N - k)
    filas = []
    for i in range(k):
        for j in range(i + 1, k):
            dif = grupos[i].mean() - grupos[j].mean()
            ee = np.sqrt(msw / 2 * (1 / len(grupos[i]) + 1 / len(grupos[j])))
            filas.append({"par": f"{nombres[i]}-{nombres[j]}", "dif": dif, "p_tukey": stats.studentized_range.sf(abs(dif) / ee, k, N - k),
                          "inf": dif - qc * ee, "sup": dif + qc * ee})
    return pd.DataFrame(filas).set_index("par")


tk = tukey(grupos, list(rend))
ref = stats.tukey_hsd(*grupos)
sm_tk = pairwise_tukeyhsd(datos.rend, datos.variedad)
iu = np.triu_indices(4, 1)
assert np.allclose(tk.p_tukey.values, ref.pvalue[iu], atol=1e-4)
assert np.allclose(tk.p_tukey.values, sm_tk.pvalues, atol=5e-3)
tk["p_bonferroni"] = np.minimum(2 * stats.t.sf(np.abs(tk["dif"]) / np.sqrt(ms_w * (2 / 12)), N - k) * 6, 1)
print("\nTukey HSD (95 %) y Bonferroni con el CM dentro de la seccion 3:\n", tk.round(4).to_string())
assert ((tk.inf < tk["dif"]) & (tk["dif"] < tk.sup)).all() and (tk.p_bonferroni >= tk.p_tukey - 1e-12).all()

fig, ax = plt.subplots(figsize=(8, 4))
ys = np.arange(len(tk))[::-1]
for y, (par, r) in zip(ys, tk.iterrows()):
    col = PALETA[1] if r.inf > 0 or r.sup < 0 else PALETA[0]
    ax.plot([r.inf, r.sup], [y, y], color=col, lw=3); ax.plot(r["dif"], y, "o", color="black")
ax.axvline(0, color="gray", ls="--"); ax.set_yticks(ys, tk.index)
ax.set(xlabel="diferencia de medias (t/ha)", title="IC simultaneos de Tukey (rojo = excluye el 0)")
plt.tight_layout(); plt.show()
````

**Lectura.** (1) **FWER con $H_0$ cierta:** el $F$ global rechaza **5.2 %**; las **6 pruebas $t$ sin corregir** producen **al menos un falso positivo en 20.0 %** de los experimentos (cuatro veces el nivel nominal; sería 26.5 % si las pruebas fueran independientes, pero comparten grupos); **Bonferroni** (4.1 %) y **Tukey** (5.1 %) respetan el 5 %. (2) **Tukey en las variedades:** D–C es la mayor diferencia (**6.08 t/ha**, IC simultáneo $[2.21,\,9.95]$, $p=0.0007$) y A–D (**4.59**, IC $[0.72,\,8.47]$, $p=0.0143$) también excluye al 0; **B–D** ($p=0.186$, IC $[-6.84,\,0.90]$) y el resto **no**. Los valores propios coinciden con SciPy (a $10^{-4}$) y con statsmodels. (3) **Bonferroni** es algo más conservador (A–D: 0.0168 frente a 0.0143 de Tukey). (4) **Cuidado con la lectura:** «B no difiere de D» **no** significa que sean iguales, y que D difiera de A y C pero B no se distinga de ninguna es un patrón **normal** con pocos datos (la «no diferencia» no es transitiva). La conclusión honesta es: «**D rinde más que A y C**; sobre B los datos no alcanzan».

## 5. Cuando Fallan los Supuestos: ANOVA de Welch 🛠️

El $F$ clásico es sensible a la **heterocedasticidad**, sobre todo con **tamaños desiguales**: si el grupo **pequeño** tiene la varianza **grande**, el denominador subestima la variabilidad de las medias y el $F$ **rechaza demasiado**; si el grupo **grande** tiene la varianza grande, **rechaza muy poco** (pierde potencia). El **ANOVA de Welch** pondera cada grupo por $w_i=n_i/s_i^2$ y ajusta los grados de libertad:

$$F_W=\frac{\sum_i w_i(\bar{y}_i-\tilde{y})^2/(k-1)}{1+\frac{2(k-2)}{k^2-1}\,\Lambda},\quad \tilde{y}=\frac{\sum_i w_i\bar{y}_i}{\sum_i w_i},\quad \Lambda=\sum_i\frac{(1-w_i/W)^2}{n_i-1},\quad \nu_2=\frac{k^2-1}{3\Lambda},\ \ W=\sum_i w_i .$$

La alternativa **no paramétrica**, **Kruskal–Wallis** (Módulo 04), compara **rangos**; ojo: **no** resuelve la heterocedasticidad (sigue suponiendo distribuciones de la misma forma). Simulamos tres escenarios con **medias iguales** y medimos cuántas veces rechaza cada prueba.

````python
# ============================================================
# 5. Cuando fallan los supuestos: ANOVA de Welch y Kruskal-Wallis
# ============================================================
def anova_welch(grupos):
    '''ANOVA de Welch (no supone varianzas iguales). Cada grupo es un arreglo (reps, n_i) o (n_i,).'''
    g = [np.asarray(x, float) for x in grupos]
    k = len(g)
    n = np.array([x.shape[-1] for x in g])
    m = np.stack([x.mean(axis=-1) for x in g], axis=-1)
    w = n / np.stack([x.var(axis=-1, ddof=1) for x in g], axis=-1)
    W = w.sum(axis=-1, keepdims=True)
    mt = (w * m).sum(axis=-1, keepdims=True) / W
    A = (w * (m - mt) ** 2).sum(axis=-1) / (k - 1)
    lam = (((1 - w / W) ** 2) / (n - 1)).sum(axis=-1)
    Fw = A / (1 + 2 * (k - 2) / (k**2 - 1) * lam)
    gl2 = (k**2 - 1) / (3 * lam)
    return Fw, stats.f.sf(Fw, k - 1, gl2), gl2


# Verificacion contra statsmodels en un conjunto con varianzas distintas
g_dem = [rng.normal(0, s_, n) for s_, n in ((1, 10), (2, 15), (5, 20))]
Fw, pw, glw = anova_welch(g_dem)
ref = anova_oneway(g_dem, use_var="unequal")
assert np.allclose((Fw, pw, glw), (ref.statistic, ref.pvalue, ref.df_denom))

# Simulacion de error tipo I: medias IGUALES, varianzas y tamanos distintos (3 000 experimentos por escenario)
def tasas(sds, ns, reps=3000):
    G = [rng.normal(0, s_, (reps, n)) for s_, n in zip(sds, ns)]
    return {"F clasica": np.mean(stats.f_oneway(*G, axis=1).pvalue < 0.05),
            "ANOVA de Welch": np.mean(anova_welch(G)[1] < 0.05),
            "Kruskal-Wallis": np.mean(np.array([stats.kruskal(*[x[r] for x in G]).pvalue for r in range(reps)]) < 0.05)}


escenarios = {"Varianzas iguales (sd 3,3,3; n 12,12,12)": tasas((3, 3, 3), (12, 12, 12)),
              "Grupo PEQUENO con varianza GRANDE (sd 1,2,5; n 40,20,10)": tasas((1, 2, 5), (40, 20, 10)),
              "Grupo GRANDE con varianza GRANDE (sd 5,2,1; n 40,20,10)": tasas((5, 2, 1), (40, 20, 10))}
res = pd.DataFrame(escenarios).T
print("Tasa de rechazo con H0 cierta (nominal 0.05):\n", res.round(3).to_string())
assert (res.loc[res.index[0]] - 0.05).abs().max() < 0.02                       # con varianzas iguales los tres controlan el error
assert res.loc[res.index[1], "F clasica"] > 0.10                                # el F clasico se INFLA
assert abs(res.loc[res.index[1], "ANOVA de Welch"] - 0.05) < 0.025             # Welch lo controla
assert res.loc[res.index[2], "F clasica"] < 0.04                                # y en el otro caso es CONSERVADOR (pierde potencia)

fig, ax = plt.subplots(figsize=(10, 4.2))
res.plot.bar(ax=ax, color=PALETA[:3], width=0.75, rot=0)
ax.axhline(0.05, color="black", ls="--", lw=1); ax.set(ylabel="tasa de rechazo con H0 cierta", title="Error tipo I del ANOVA cuando las varianzas difieren")
ax.set_xticklabels(["Varianzas\niguales", "Grupo pequeno,\nvarianza grande", "Grupo grande,\nvarianza grande"]); ax.legend(loc="upper left", fontsize=8)
plt.tight_layout(); plt.show()
````

**Lectura.** (1) Nuestro ANOVA de Welch coincide con `statsmodels.anova_oneway(use_var="unequal")` (estadístico, valor p y grados de libertad). (2) **Con varianzas iguales** las tres pruebas controlan el error (F: 5.1 %, Welch: 5.0 %, Kruskal–Wallis: 4.8 %). (3) **Grupo pequeño con varianza grande** (sd 1, 2, 5; n 40, 20, 10): el $F$ clásico rechaza **28.6 %** de las veces **aunque las medias sean iguales** (casi seis veces el nivel nominal), Kruskal–Wallis **14.8 %** y **Welch 5.3 %**. (4) **Grupo grande con varianza grande** (sd 5, 2, 1): el $F$ clásico rechaza solo **0.7 %** (es **ciego**: perdería casi toda su potencia) y Welch mantiene **4.4 %**. Moraleja: con **tamaños y varianzas desiguales** el $F$ clásico puede **errar en cualquier dirección**; el costo de usar Welch **siempre** es mínimo cuando las varianzas son iguales, y conviene **no decidir según una pre-prueba de Levene**.

```python
eta2_desde_F = F * (k - 1) / (F * (k - 1) + (N - k))
print(f"eta^2 = {eta2:.3f} | eta^2 desde F = {eta2_desde_F:.3f} | omega^2 = {omega2:.3f}")

assert np.isclose(eta2, eta2_desde_F) and 0 < omega2 < eta2
print("omega^2 descuenta lo que se 'explicaria' solo por azar al ajustar k medias; eta^2 sobrestima en muestras pequenas (aqui ~25 % de varianza explicada).")
```
</details>

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Chi-cuadrado** | $\chi^2_k$ = suma de $k$ normales² (media $k$, varianza $2k$). Bondad de ajuste: $\sum(O-E)^2/E$ con $k-1-m$ gl; con dados justos rechazó 5.0 %, con el cargado 82.2 % ($n=300$) |
| **Esperadas pequeñas** | $E\ge5$ es una guía conservadora (equiprobable $n=20$: 4.7 %); con probabilidades muy desiguales ($E$ de 0.3 a 24) el nivel real fue **8.2 %**: agrupar categorías o usar Fisher/Monte Carlo |
| **Independencia** | $E_{ij}=\text{fila}\times\text{columna}/N$, gl $=(r-1)(c-1)$. **V de Cramér** (0.213: modesta aunque $p=0.0001$) y **residuos ajustados** dicen cuánto y dónde; coincide con la permutación |
| **2×2 pequeñas** | Sin corrección $p\approx0.025$, Yates y Fisher $p\approx0.07$: **la decisión cambia**; preferir Fisher con esperadas < 5 |
| **ANOVA** | SCT = SCE + SCD; $F=\mathrm{CM}_E/\mathrm{CM}_D$ (aquí 6.455, $p=0.001$); con $k=2$, $F=t^2$. **Reportar $\eta^2$/$\omega^2$** (0.31/0.25), no solo el valor p |
| **Supuestos** | Independencia (diseño), residuos aproximadamente normales, varianzas iguales (Levene, Bartlett). El $F$ es **omnibus**: no dice cuáles grupos difieren |
| **Comparaciones múltiples** | 6 pruebas $t$ sin corregir: **20 %** de falsos positivos con $H_0$ cierta; **Tukey** (5.1 %) o **Bonferroni** (4.1 %) lo controlan y Tukey da IC simultáneos |
| **Welch** | Con varianzas y tamaños desiguales el $F$ clásico erró en ambas direcciones (**28.6 %** y **0.7 %** de rechazos con $H_0$ cierta); **Welch** se mantuvo en 5.3 % y 4.4 %. Kruskal–Wallis **no** arregla la heterocedasticidad |
