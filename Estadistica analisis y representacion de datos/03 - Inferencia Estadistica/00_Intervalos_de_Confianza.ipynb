## 1. De la Muestra a la Población: Estimar con Incertidumbre 🎯

**Vocabulario.** Un **parámetro** $\theta$ es un número fijo (y desconocido) de la población, por ejemplo la media $\mu$ o la proporción $p$. Un **estimador** $\hat\theta$ es una **función de la muestra** (por eso es una variable aleatoria: cambia de muestra en muestra) y una **estimación** es el valor que toma en **nuestra** muestra. La variabilidad del estimador entre muestras se describe con su **distribución muestral**; su desviación estándar es el **error estándar** (EE). Para la media de una muestra aleatoria simple:

$$\mathrm{EE}(\bar{X}) = \frac{\sigma}{\sqrt{n}}, \qquad \widehat{\mathrm{EE}}(\bar{x}) = \frac{s}{\sqrt{n}}.$$

Una **estimación puntual** ($\bar{x}=22.07$) no dice **cuánto confiar** en ella. Un **intervalo de confianza** de nivel $1-\alpha$ es un par de estadísticos $[L(X), U(X)]$ construido de modo que, **antes de observar los datos**,

$$\mathbb{P}\big(L(X) \le \theta \le U(X)\big) = 1-\alpha .$$

La receta general (cuando el estimador es aproximadamente normal) es **estimación ± valor crítico × error estándar**. Lo que cambia entre casos es el **error estándar** y el **valor crítico**:

| Parámetro | Estimador | Error estándar | Intervalo | Sección |
|---|---|---|---|---|
| Media $\mu$ (σ conocida) | $\bar{x}$ | $\sigma/\sqrt{n}$ | $\bar{x} \pm z_{1-\alpha/2}\,\sigma/\sqrt{n}$ | 2 |
| Media $\mu$ (σ desconocida) | $\bar{x}$ | $s/\sqrt{n}$ | $\bar{x} \pm t_{n-1,\,1-\alpha/2}\,s/\sqrt{n}$ | 2 |
| Proporción $p$ | $\hat{p}$ | $\sqrt{\hat{p}(1-\hat{p})/n}$ | Wald, Wilson, Clopper–Pearson… | 4 |
| Diferencia $\mu_1-\mu_2$ | $\bar{x}_1-\bar{x}_2$ | $\sqrt{s_1^2/n_1+s_2^2/n_2}$ | Welch ($t$ con gl de Satterthwaite) | 5 |
| Varianza $\sigma^2$ | $s^2$ | (no simétrico) | basado en $\chi^2_{n-1}$ | 5 |
| Cualquier $\theta$ | $\hat\theta$ | estimado por remuestreo | **bootstrap** (percentil, BCa) | 6 |

**La interpretación correcta (y las incorrectas).** «Confianza del 95 %» es una propiedad del **procedimiento**: si repitiéramos el muestreo muchas veces y construyéramos el intervalo cada vez, **≈ 95 % de los intervalos contendría al parámetro**. Una vez calculado, un intervalo concreto **contiene o no contiene** a $\theta$ (que es fijo): no tiene sentido decir «hay 95 % de probabilidad de que $\mu$ esté en $[20.9, 23.2]$» en el marco frecuentista.

| Afirmación | ¿Correcta? |
|---|---|
| «Si repitiéramos el estudio, ≈ 95 % de los intervalos construidos así contendría a $\mu$» | ✅ |
| «Hay un 95 % de probabilidad de que $\mu$ esté en *este* intervalo» | ❌ (en el enfoque frecuentista $\mu$ es fija; eso lo dice un intervalo de *credibilidad* bayesiano) |
| «El 95 % de las **observaciones** cae dentro del intervalo» | ❌ (eso es un intervalo de **predicción** o de **tolerancia**, mucho más ancho) |
| «Un IC que no incluye a $\mu_0$ equivale a rechazar $H_0:\mu=\mu_0$ al 5 %» | ✅ (dualidad, sección 5) |

El **bootstrap** (sección 6) es una alternativa que no exige una fórmula para el error estándar. Empecemos por la configuración.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas scipy statsmodels matplotlib

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.stats.proportion import proportion_confint
from statsmodels.stats.weightstats import CompareMeans, DescrStatsW

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {__import__('scipy').__version__} | statsmodels {__import__('statsmodels').__version__}")
````

## 2. IC para la Media: $z$ (σ conocida) y $t$ de Student (σ desconocida) 📏

Si $X_1,\dots,X_n$ son independientes y $N(\mu,\sigma^2)$, entonces $\bar{X}\sim N(\mu,\sigma^2/n)$ y

$$Z=\frac{\bar{X}-\mu}{\sigma/\sqrt{n}} \sim N(0,1) \;\Longrightarrow\; \bar{x} \pm z_{1-\alpha/2}\,\frac{\sigma}{\sqrt{n}} .$$

En la práctica $\sigma$ **no se conoce** y se estima con $s$. Al sustituir, el estadístico deja de ser normal: tiene **colas más pesadas** porque $s$ también fluctúa. Gosset («Student», 1908) demostró que, bajo normalidad,

$$T=\frac{\bar{X}-\mu}{S/\sqrt{n}} \sim t_{n-1} \;\Longrightarrow\; \bar{x} \pm t_{n-1,\,1-\alpha/2}\,\frac{s}{\sqrt{n}} .$$

A medida que $n$ crece, $t_{n-1}\to N(0,1)$. **Supuestos:** (i) muestra **aleatoria e independiente**; (ii) población **aproximadamente normal** o $n$ **suficientemente grande** (TLC); con asimetría fuerte y $n$ pequeño, ver el *bootstrap* (sección 6).

El **ancho** $2\,t\,s/\sqrt{n}$ depende de tres palancas: el **nivel de confianza** (más confianza, más ancho), la **variabilidad** $s$ y el **tamaño de muestra** $n$ (el ancho decrece como $1/\sqrt{n}$). Despejando $n$ para un **margen de error** $E$ se obtiene $n \ge (z\,\sigma/E)^2$. Un **intervalo de predicción** para *una observación nueva* añade la variabilidad individual: $\bar{x} \pm t_{n-1,\,1-\alpha/2}\,s\sqrt{1+1/n}$.

La celda implementa todo **desde cero** y lo contrasta con SciPy y statsmodels.

````python
# ============================================================
# 2. IC para la media: z (sigma conocida) y t (sigma desconocida)
# ============================================================
# Datos SINTETICOS: rendimiento (t/ha) de 30 parcelas de papa; el "parametro real" lo conocemos porque lo fijamos nosotros
MU_REAL, SIGMA_REAL = 22.0, 4.0
parcelas = rng.normal(MU_REAL, SIGMA_REAL, 30)


def t_crit(conf, gl):
    return stats.t.ppf(1 - (1 - conf) / 2, gl)


def ic_media_z(x, sigma, conf=0.95):
    x = np.asarray(x, float)
    me = stats.norm.ppf(1 - (1 - conf) / 2) * sigma / np.sqrt(len(x))
    return x.mean() - me, x.mean() + me


def ic_media_t(x, conf=0.95):
    x = np.asarray(x, float)
    me = t_crit(conf, len(x) - 1) * x.std(ddof=1) / np.sqrt(len(x))
    return x.mean() - me, x.mean() + me


n = len(parcelas)
xbar, s = parcelas.mean(), parcelas.std(ddof=1)
ee = s / np.sqrt(n)                                                      # error estandar estimado
z_lo, z_hi = ic_media_z(parcelas, SIGMA_REAL)
t_lo, t_hi = ic_media_t(parcelas)
print(f"n = {n} | media = {xbar:.2f} | s = {s:.2f} | error estandar = {ee:.3f}")
print(f"IC 95% z (sigma = {SIGMA_REAL} conocida): [{z_lo:.2f}, {z_hi:.2f}]  ancho = {z_hi - z_lo:.2f}")
print(f"IC 95% t (sigma desconocida)    : [{t_lo:.2f}, {t_hi:.2f}]  ancho = {t_hi - t_lo:.2f}")

# Contraste con SciPy y statsmodels
assert np.allclose((t_lo, t_hi), stats.t.interval(0.95, n - 1, loc=xbar, scale=stats.sem(parcelas)))
assert np.allclose((t_lo, t_hi), DescrStatsW(parcelas).tconfint_mean(alpha=0.05))
assert np.allclose((z_lo, z_hi), stats.norm.interval(0.95, loc=xbar, scale=SIGMA_REAL / np.sqrt(n)))
assert t_lo < xbar < t_hi and np.isclose((t_lo + t_hi) / 2, xbar)

# Efecto del nivel de confianza: mas confianza -> intervalo mas ancho
tabla = pd.DataFrame([{"confianza": c, "inferior": ic_media_t(parcelas, c)[0], "superior": ic_media_t(parcelas, c)[1],
                       "ancho": np.subtract(*ic_media_t(parcelas, c)[::-1])} for c in (0.80, 0.90, 0.95, 0.99)]).round(3)
print("\n", tabla.to_string(index=False))
assert tabla["ancho"].is_monotonic_increasing

# Intervalo de PREDICCION para UNA observacion nueva (mas ancho que el IC de la media) y su cobertura por simulacion
def ic_prediccion(x, conf=0.95):
    x = np.asarray(x, float)
    me = t_crit(conf, len(x) - 1) * x.std(ddof=1) * np.sqrt(1 + 1 / len(x))
    return x.mean() - me, x.mean() + me


reps, n_ = 5000, 20
M = rng.normal(MU_REAL, SIGMA_REAL, (reps, n_))
nuevo = rng.normal(MU_REAL, SIGMA_REAL, reps)                                # una observacion futura por muestra
xb, sd = M.mean(axis=1), M.std(axis=1, ddof=1)
tc = t_crit(0.95, n_ - 1)
cob_ic = np.mean(np.abs(nuevo - xb) <= tc * sd / np.sqrt(n_))
cob_pi = np.mean(np.abs(nuevo - xb) <= tc * sd * np.sqrt(1 + 1 / n_))
pi_lo, pi_hi = ic_prediccion(parcelas)
print(f"\nParcelas: IC 95% de la media = [{t_lo:.2f}, {t_hi:.2f}] (ancho {t_hi - t_lo:.2f}) | intervalo de prediccion 95% = [{pi_lo:.2f}, {pi_hi:.2f}] (ancho {pi_hi - pi_lo:.2f})")
print(f"Fraccion de observaciones NUEVAS dentro del IC de la media: {cob_ic:.3f} | dentro del intervalo de prediccion: {cob_pi:.3f}")
assert cob_ic < 0.40 and abs(cob_pi - 0.95) < 0.015 and (pi_hi - pi_lo) > 3 * (t_hi - t_lo)


# Tamano de muestra para estimar la media con margen de error E (sigma planificada) y cuanto cuesta la precision
def n_para_media(sigma, E, conf=0.95):
    z = stats.norm.ppf(1 - (1 - conf) / 2)
    return int(np.ceil((z * sigma / E) ** 2))


ancho = lambda n, sigma=SIGMA_REAL: 2 * 1.96 * sigma / np.sqrt(n)
print(f"\nPara estimar el rendimiento medio con margen E = 1 t/ha (sigma = {SIGMA_REAL}): n = {n_para_media(SIGMA_REAL, 1.0)}; con E = 0.5: n = {n_para_media(SIGMA_REAL, 0.5)}")
assert n_para_media(SIGMA_REAL, 1.0) == 62 and 1.96 * SIGMA_REAL / np.sqrt(62) <= 1.0
assert np.isclose(ancho(4 * 50), ancho(50) / 2)                             # cuadruplicar n reduce el ancho a la mitad

ns = np.arange(5, 500)
fig, ax = plt.subplots(figsize=(9, 4.2))
ax.plot(ns, ancho(ns), color=PALETA[0])
for m in (25, 100, 400):
    ax.plot(m, ancho(m), "o", color=PALETA[1]); ax.annotate(f"n={m}: {ancho(m):.2f}", (m, ancho(m)), textcoords="offset points", xytext=(8, 8))
ax.set(xlabel="tamano de muestra n", ylabel="ancho del IC 95 % (t/ha)", title="Ancho del IC vs. n (sigma = 4): rendimientos decrecientes (1/raiz n)")
plt.tight_layout(); plt.show()
````

**Lectura.** (1) Las funciones propias coinciden con `scipy.stats.t.interval`, `DescrStatsW.tconfint_mean` y `norm.interval` (los `assert` pasan). (2) Con $n=30$ y $\bar{x}=22.07$, el IC 95 % con $t$ es $[20.91,\,23.23]$. **Atención:** aquí el IC con $z$ (usando el $\sigma=4$ real) salió *más ancho* (2.86) que el de $t$ (2.32) solo porque **esta** muestra tuvo $s=3.11<4$; **en promedio** el intervalo $t$ es el más ancho, porque paga el precio de **no conocer** $\sigma$ (la simulación de la sección siguiente lo mostrará). (3) Subir la confianza de 80 % a 99 % **multiplica el ancho por ≈ 2.1** (1.49 → 3.13): **no hay almuerzo gratis**, más certeza cuesta precisión. (4) El **intervalo de predicción** (ancho 12.9) es ≈ 5.6 veces el IC de la media (2.3): la media se conoce con precisión, **una parcela individual no**; solo **35.4 %** de las observaciones nuevas cae dentro del IC de la media, frente a **95.0 %** del de predicción. (5) Para un margen de ±1 t/ha con $\sigma=4$ hacen falta **62 parcelas**; para ±0.5, **246** (el doble de precisión cuesta el **cuádruple** de datos).

## 3. ¿Qué Significa «95 %»? Simulación de Cobertura 🎲

Como en esta guía **conocemos** $\mu$, podemos repetir el muestreo miles de veces y contar **cuántos intervalos atrapan** al parámetro: la **cobertura empírica**. Es la forma más honesta de ver qué promete un método. Comparamos tres procedimientos con $n\in\{3,5,10,30,100\}$: (a) el IC $t$ (correcto), (b) el IC $z$ con $s$ en lugar de $\sigma$ (el **error común**) y (c) el IC $z$ con $\sigma$ conocida. Después dibujamos 100 intervalos $t$ con $n=15$.

````python
# ============================================================
# 3. Que significa "95 %": simulacion de cobertura
# ============================================================
def cobertura_media(n, reps=5000, rng=rng):
    M = rng.normal(MU_REAL, SIGMA_REAL, (reps, n))
    xb, sd = M.mean(axis=1), M.std(axis=1, ddof=1)
    ee = sd / np.sqrt(n)
    c_t = np.mean(np.abs(xb - MU_REAL) <= t_crit(0.95, n - 1) * ee)             # correcto: cuantil t
    c_zs = np.mean(np.abs(xb - MU_REAL) <= 1.96 * ee)                           # error comun: z con s
    c_zsig = np.mean(np.abs(xb - MU_REAL) <= 1.96 * SIGMA_REAL / np.sqrt(n))    # z con sigma conocida
    return c_t, c_zs, c_zsig


cob = pd.DataFrame([(m, *cobertura_media(m)) for m in (3, 5, 10, 30, 100)],
                   columns=["n", "t (correcto)", "z con s (error comun)", "z con sigma conocida"]).set_index("n")
print("Cobertura empirica de intervalos nominales del 95 % (5 000 muestras por fila):\n", cob.round(3).to_string())
assert (cob["t (correcto)"] - 0.95).abs().max() < 0.015                          # t cubre ~95 % para todo n
assert (cob["z con sigma conocida"] - 0.95).abs().max() < 0.015
assert cob.loc[5, "z con s (error comun)"] < 0.91 and cob.loc[3, "z con s (error comun)"] < 0.86   # z con s se queda corto con n pequeno
assert cob.loc[100, "z con s (error comun)"] > 0.93

# 100 intervalos t con n = 15: cuantos atrapan al parametro?
M = rng.normal(MU_REAL, SIGMA_REAL, (100, 15))
xb, ee = M.mean(axis=1), M.std(axis=1, ddof=1) / np.sqrt(15)
lo, hi = xb - t_crit(0.95, 14) * ee, xb + t_crit(0.95, 14) * ee
acierta = (lo <= MU_REAL) & (MU_REAL <= hi)
fig, ax = plt.subplots(figsize=(9, 6))
for i in range(100):
    ax.plot([lo[i], hi[i]], [i, i], color=PALETA[0] if acierta[i] else PALETA[1], lw=1.6)
    ax.plot(xb[i], i, "o", color=PALETA[0] if acierta[i] else PALETA[1], ms=2.5)
ax.axvline(MU_REAL, color="black", ls="--", label=f"parametro real mu = {MU_REAL}")
ax.set(xlabel="rendimiento (t/ha)", ylabel="muestra n.º", title=f"100 intervalos t del 95 %: {acierta.sum()} atrapan a mu, {100 - acierta.sum()} no (rojo)")
ax.legend(); plt.tight_layout(); plt.show()
print(f"De los 100 intervalos, {acierta.sum()} atrapan a mu y {100 - acierta.sum()} no.")
assert 88 <= acierta.sum() <= 100
````

---
**Lectura.** El IC $t$ cubre ≈ 95 % **para cualquier $n$** (0.948–0.956, dentro del error de simulación), y el $z$ con $\sigma$ conocida también. El **error común** (poner $s$ dentro de la fórmula $z$) **se queda corto** con muestras pequeñas: con $n=3$ cubre solo **81 %** y con $n=5$, **87 %** (teóricamente $\mathbb{P}(|T_4|<1.96)\approx 0.88$); recién con $n\ge 30$ la diferencia es pequeña (94.5 %) y con $n=100$ desaparece (94.8 %). En el gráfico, **96 de 100** intervalos atrapan a $\mu$ y **4 no** (en rojo): es exactamente lo que promete «95 %», **ni más ni menos**; en otra tanda podrían ser 93 o 98. Moraleja: el nivel de confianza **no es una promesa sobre un intervalo concreto**, sino sobre la **tasa de aciertos a largo plazo**.

## 4. IC para una Proporción: Wald, Wilson, Agresti–Coull y Clopper–Pearson ⚖️

Para $X\sim\mathrm{Binomial}(n,p)$ con $\hat{p}=x/n$, el intervalo más enseñado es el de **Wald**, $\hat{p}\pm z\sqrt{\hat{p}(1-\hat{p})/n}$, que usa la aproximación normal **evaluando el error estándar en $\hat{p}$**. Falla cuando $p$ está cerca de 0 o 1 o $n$ es pequeño (el intervalo puede salirse de $[0,1]$ o colapsar a $[0,0]$ si $x=0$). Alternativas:

- **Wilson** (1927): invierte la prueba $z$ con el EE evaluado en $p$ (no en $\hat{p}$): $\dfrac{\hat{p}+\frac{z^2}{2n}}{1+\frac{z^2}{n}}\;\pm\;\dfrac{z}{1+\frac{z^2}{n}}\sqrt{\dfrac{\hat{p}(1-\hat{p})}{n}+\dfrac{z^2}{4n^2}}$.
- **Agresti–Coull** (1998): Wald aplicado a $\tilde{p}=(x+z^2/2)/(n+z^2)$ con $\tilde{n}=n+z^2$ («suma ≈ 2 éxitos y 2 fracasos»).
- **Clopper–Pearson** (1934, «exacto»): invierte la prueba binomial con cuantiles de la distribución $\mathrm{Beta}$; **garantiza cobertura ≥ 1-α**, a costa de ser conservador.

Como $X$ es discreta, la cobertura de cada método se puede calcular **exactamente** (sin simulación): para cada $p$, se suma $\mathbb{P}(X=x)$ sobre los $x$ cuyo intervalo contiene a $p$. Lo hacemos para $n=30$ y una rejilla de 991 valores de $p$. La celda también incluye el tamaño de muestra para una proporción, $n \ge z^2p(1-p)/E^2$ (máximo en $p=0.5$) y la **corrección por población finita** $n=n_0/(1+(n_0-1)/N)$.

````python
# ============================================================
# 4. IC para una proporcion: Wald, Wilson, Agresti-Coull y Clopper-Pearson
# ============================================================
def ic_wald(x, n, conf=0.95):
    p, z = x / n, stats.norm.ppf(1 - (1 - conf) / 2)
    me = z * np.sqrt(p * (1 - p) / n)
    return np.maximum(p - me, 0.0), np.minimum(p + me, 1.0)                 # se recorta a [0, 1] (en bruto puede salirse)


def ic_wilson(x, n, conf=0.95):
    p, z = x / n, stats.norm.ppf(1 - (1 - conf) / 2)
    centro = (p + z**2 / (2 * n)) / (1 + z**2 / n)
    me = z / (1 + z**2 / n) * np.sqrt(p * (1 - p) / n + z**2 / (4 * n**2))
    return centro - me, centro + me


def ic_agresti_coull(x, n, conf=0.95):
    z = stats.norm.ppf(1 - (1 - conf) / 2)
    nt = n + z**2
    pt = (x + z**2 / 2) / nt
    me = z * np.sqrt(pt * (1 - pt) / nt)
    return max(pt - me, 0.0), min(pt + me, 1.0)                              # se recorta a [0, 1]


def ic_clopper_pearson(x, n, conf=0.95):
    a = 1 - conf
    x = np.asarray(x, float)
    lo = np.where(x == 0, 0.0, stats.beta.ppf(a / 2, np.maximum(x, 1e-12), n - x + 1))
    hi = np.where(x == n, 1.0, stats.beta.ppf(1 - a / 2, x + 1, np.maximum(n - x, 1e-12)))
    return lo, hi


# Caso tipico: 7 parcelas con plaga de n = 40, y caso extremo: 0 de 30
for x, n_ in ((7, 40), (0, 30)):
    print(f"\nx = {x}, n = {n_}  (p_hat = {x / n_:.3f})")
    for nombre, f, met in (("Wald", ic_wald, "normal"), ("Wilson", ic_wilson, "wilson"),
                           ("Agresti-Coull", ic_agresti_coull, "agresti_coull"), ("Clopper-Pearson", ic_clopper_pearson, "beta")):
        lo, hi = f(x, n_)
        ref = proportion_confint(x, n_, alpha=0.05, method=met)
        print(f"  {nombre:16s} [{float(lo):.4f}, {float(hi):.4f}]")
        assert np.allclose((lo, hi), ref, atol=1e-8), (nombre, (lo, hi), ref)           # igual que statsmodels
assert ic_wald(0, 30) == (0.0, 0.0)                                           # Wald "dice" que con 0 exitos la proporcion es exactamente 0: absurdo
assert ic_wilson(0, 30)[1] > 0.1                                              # Wilson si deja ver que p podria ser ~11 %

# Cobertura EXACTA (sin simulacion) para n = 30: suma de la binomial sobre los x cuyo intervalo contiene a p
n_b = 30
xs = np.arange(n_b + 1)
ps = np.linspace(0.005, 0.995, 991)
pmf = stats.binom.pmf(xs[None, :], n_b, ps[:, None])                         # (991, 31)
cobertura = {}
for nombre, f in (("Wald", ic_wald), ("Wilson", ic_wilson), ("Clopper-Pearson", ic_clopper_pearson)):
    lo, hi = f(xs, n_b)
    dentro = (np.asarray(lo)[None, :] <= ps[:, None]) & (ps[:, None] <= np.asarray(hi)[None, :])
    cobertura[nombre] = (pmf * dentro).sum(axis=1)
resumen = pd.DataFrame({k: {"minima": v.min(), "media": v.mean(), "p < 0.95 (% de la rejilla)": 100 * np.mean(v < 0.95)} for k, v in cobertura.items()}).T.round(3)
print("\nCobertura exacta, n = 30 (nominal 0.95):\n", resumen.to_string())
assert cobertura["Wald"].min() < 0.40 and cobertura["Wald"].mean() < 0.94
assert cobertura["Clopper-Pearson"].min() >= 0.95 - 1e-9                      # garantia exacta (conservador)
assert 0.93 < cobertura["Wilson"].mean() < 0.97

fig, ax = plt.subplots(figsize=(10, 5))
for (nombre, v), col in zip(cobertura.items(), (PALETA[1], PALETA[0], PALETA[2])):
    ax.plot(ps, v, label=nombre, color=col, lw=1.2)
ax.axhline(0.95, color="black", ls="--", lw=1)
ax.set(xlabel="proporcion verdadera p", ylabel="cobertura exacta", ylim=(0.2, 1.02), title="Cobertura de tres IC para una proporcion (n = 30, nominal 95 %)")
ax.legend(loc="lower center", ncol=3); plt.tight_layout(); plt.show()


# Tamano de muestra para una proporcion (peor caso p = 0.5) y correccion por poblacion finita
def n_para_proporcion(E, p=0.5, conf=0.95):
    z = stats.norm.ppf(1 - (1 - conf) / 2)
    return int(np.ceil(z**2 * p * (1 - p) / E**2))


def corregir_poblacion_finita(n0, N):
    return int(np.ceil(n0 / (1 + (n0 - 1) / N)))


tab_n = pd.DataFrame({"margen E": [0.10, 0.05, 0.03, 0.02, 0.01]})
tab_n["n (p = 0.5)"] = [n_para_proporcion(E) for E in tab_n["margen E"]]
print("\n", tab_n.to_string(index=False))
assert n_para_proporcion(0.03) == 1068 and n_para_proporcion(0.05) == 385
assert n_para_proporcion(0.03, p=0.1) < n_para_proporcion(0.03)               # p = 0.5 es el peor caso
print(f"Con N = 600 parcelas y E = 0.05: n0 = 385 -> con poblacion finita n = {corregir_poblacion_finita(385, 600)}")
assert corregir_poblacion_finita(385, 600) == 235
````

---
**Lectura.** (1) Las cuatro implementaciones **coinciden con `proportion_confint`** (a $10^{-8}$). (2) Con $0$ éxitos de $30$, Wald da $[0,0]$: «seguro que $p=0$», una conclusión **absurda** con 30 observaciones. Wilson y Clopper–Pearson dicen que $p$ podría ser **hasta ≈ 11–12 %** (la «regla del tres» da $3/30=10\%$). (3) La **cobertura exacta** con $n=30$: Wald tiene **mínimo 0.14**, media **0.882** y queda por debajo de 0.95 en **97 %** de la rejilla; Wilson tiene media **0.953** pero **oscila** (mínimo 0.86); Clopper–Pearson **nunca baja de 0.954** (media 0.973): es **conservador** (intervalos algo más anchos de lo necesario). (4) **Recomendación:** no usar Wald salvo con $n$ grande y $p$ lejos de los extremos; usar **Wilson** (o Agresti–Coull) por defecto y **Clopper–Pearson** cuando se necesite **garantía**. (5) Para un margen de ±3 puntos al 95 % hacen falta **1 068** encuestados *independientemente del tamaño de la población*, salvo que la muestra sea una fracción grande de ella: con $N=600$ parcelas y $E=0.05$ bastan **235** en vez de 385.

## 5. Diferencia de Medias (Welch), Varianza y su Relación con las Pruebas 🔁

**Dos grupos independientes.** Para $\mu_1-\mu_2$ **no** se asume que las varianzas sean iguales (Welch, 1947):

$$(\bar{x}_1-\bar{x}_2)\pm t_{\nu,\,1-\alpha/2}\sqrt{\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}},\qquad \nu=\frac{\left(\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}\right)^2}{\frac{(s_1^2/n_1)^2}{n_1-1}+\frac{(s_2^2/n_2)^2}{n_2-1}}\quad(\text{Welch–Satterthwaite}).$$

**Varianza.** Bajo normalidad, $(n-1)S^2/\sigma^2\sim\chi^2_{n-1}$, de donde $\left[\dfrac{(n-1)s^2}{\chi^2_{n-1,\,1-\alpha/2}},\;\dfrac{(n-1)s^2}{\chi^2_{n-1,\,\alpha/2}}\right]$. A diferencia del IC de la media, este es **muy sensible a la no normalidad** (depende de la curtosis).

**Dualidad IC ↔ prueba.** La prueba de Welch de $H_0:\mu_1=\mu_2$ al nivel $\alpha$ **rechaza si y solo si** el IC de $1-\alpha$ de la diferencia **excluye al 0**. El IC es **más informativo**: además de «¿hay diferencia?» dice **cuán grande** podría ser. Datos: dos fertilizantes con varianzas y tamaños distintos (**sintéticos**).

````python
# ============================================================
# 5. IC para la diferencia de medias (Welch) y para la varianza
# ============================================================
# Dos fertilizantes (SINTETICO): varianzas distintas y tamanos distintos
rend_a = rng.normal(21.0, 3.5, 18)
rend_b = rng.normal(24.0, 6.0, 14)


def ic_dif_welch(x, y, conf=0.95):
    x, y = np.asarray(x, float), np.asarray(y, float)
    vx, vy = x.var(ddof=1) / len(x), y.var(ddof=1) / len(y)
    gl = (vx + vy) ** 2 / (vx**2 / (len(x) - 1) + vy**2 / (len(y) - 1))      # Welch-Satterthwaite
    d, me = x.mean() - y.mean(), t_crit(conf, gl) * np.sqrt(vx + vy)
    return d - me, d + me, gl


lo, hi, gl = ic_dif_welch(rend_a, rend_b)
cm = CompareMeans(DescrStatsW(rend_a), DescrStatsW(rend_b))
ref = cm.tconfint_diff(alpha=0.05, usevar="unequal")
print(f"Media A = {rend_a.mean():.2f}, media B = {rend_b.mean():.2f}, diferencia A - B = {rend_a.mean() - rend_b.mean():.2f}")
print(f"IC 95% de mu_A - mu_B (Welch): [{lo:.2f}, {hi:.2f}] con gl = {gl:.1f}")
assert np.allclose((lo, hi), ref) and np.isclose(gl, cm.dof_satt())

# Dualidad IC <-> prueba: la prueba t de Welch rechaza H0: mu_A = mu_B al 5 % si y solo si el IC 95 % de la diferencia excluye al 0
def dualidad(a, b):
    lo, hi, _ = ic_dif_welch(a, b)
    p = stats.ttest_ind(a, b, equal_var=False).pvalue
    return (p < 0.05) == (not lo <= 0 <= hi), p


print(f"p de Welch = {stats.ttest_ind(rend_a, rend_b, equal_var=False).pvalue:.3f}; el IC contiene al 0: {lo <= 0 <= hi}")
rd = np.random.default_rng(3)
res = [dualidad(rd.normal(0, 1, 15), rd.normal(d, 1.5, 12)) for d in np.linspace(0, 2.5, 200)]
assert all(ok for ok, _ in res) and any(p < 0.05 for _, p in res) and any(p >= 0.05 for _, p in res)


def ic_varianza(x, conf=0.95):
    x = np.asarray(x, float)
    n, s2, a = len(x), x.var(ddof=1), 1 - conf
    return (n - 1) * s2 / stats.chi2.ppf(1 - a / 2, n - 1), (n - 1) * s2 / stats.chi2.ppf(a / 2, n - 1)


print("IC 95% de la desviacion estandar de A:", np.sqrt(ic_varianza(rend_a)).round(2), "| s =", rend_a.std(ddof=1).round(2))

# Sensibilidad a la normalidad: el IC chi-cuadrado de sigma^2 NO es robusto (a diferencia del IC t de la media)
def cobertura_var(gen, var_real, n=30, reps=4000):
    M = gen((reps, n))
    s2 = M.var(axis=1, ddof=1)
    lo = (n - 1) * s2 / stats.chi2.ppf(0.975, n - 1)
    hi = (n - 1) * s2 / stats.chi2.ppf(0.025, n - 1)
    return np.mean((lo <= var_real) & (var_real <= hi))


cv_norm = cobertura_var(lambda sh: rng.normal(0, 2.0, sh), 4.0)
cv_exp = cobertura_var(lambda sh: rng.exponential(2.0, sh), 4.0)
print(f"Cobertura del IC de la varianza (n = 30): normal = {cv_norm:.3f} | exponencial (asimetrica) = {cv_exp:.3f}")
assert abs(cv_norm - 0.95) < 0.015 and cv_exp < 0.85
````

Pero **tampoco prueban que no la haya**: el intervalo es compatible con diferencias de hasta ≈ 6 t/ha a favor de B, lo que **sería agronómicamente relevante**. Con muestras pequeñas, «no significativo» suele significar «**poca información**», no «sin efecto». (2) La **dualidad** se verificó con 200 pares de muestras: en **todas**, «p < 0.05» coincidió con «el IC excluye 0». (3) El IC de la **desviación estándar** de A es $[2.80,\,5.59]$ para $s=3.73$: la varianza se estima con **mucha** menos precisión que la media. (4) Con $n=30$, el IC chi-cuadrado de $\sigma^2$ cubre **95.1 %** con datos normales pero solo **71.6 %** con datos exponenciales: **no es robusto**; ante asimetría, usar *bootstrap*.

## 6. Bootstrap: Intervalos sin Fórmula 🥾

Idea (Efron, 1979): la **muestra** es nuestra mejor imagen de la población; **remuestreamos con reemplazo** de ella $B$ veces, recalculamos el estadístico $\hat\theta^{\ast}_b$ y usamos la **distribución de los $\hat\theta^{\ast}_b$** como aproximación a la distribución muestral de $\hat\theta$.

- **Error estándar bootstrap:** la desviación estándar de los $\hat\theta^{\ast}_b$.
- **IC percentil:** los cuantiles $\alpha/2$ y $1-\alpha/2$ de los $\hat\theta^{\ast}_b$. Simple, pero puede tener sesgo.
- **IC BCa** (*bias-corrected and accelerated*): ajusta los cuantiles con una **corrección de sesgo** $z_0=\Phi^{-1}\big(\#\{\hat\theta^{\ast}_b<\hat\theta\}/B\big)$ y una **aceleración** $a$ (calculada con el *jackknife* «deja uno fuera»): se usan los percentiles $\Phi\!\left(z_0+\dfrac{z_0+z_\alpha}{1-a\,(z_0+z_\alpha)}\right)$.

**Cuándo ayuda:** estadísticos sin error estándar sencillo (mediana, coeficiente de variación, razones), distribuciones desconocidas. **Cuándo NO es magia:** con $n$ muy pequeño o colas pesadas el remuestreo no puede inventar información que la muestra no tiene; además **no corrige un muestreo sesgado**. La celda implementa percentil y BCa vectorizados, los compara con `scipy.stats.bootstrap` sobre una muestra **asimétrica** de ingresos (sintética) y mide su **cobertura** frente al IC $t$ con $n=25$ y asimetría fuerte.

````python
# ============================================================
# 7. Bootstrap: percentil y BCa desde cero
# ============================================================
def dist_bootstrap(x, estad, B=10000, rng=rng):
    x = np.asarray(x)
    idx = rng.integers(0, len(x), (B, len(x)))                                # B remuestras con reemplazo
    return estad(x[idx], axis=-1)


def ic_percentil(x, estad, B=10000, conf=0.95, rng=rng):
    boot = dist_bootstrap(x, estad, B, rng)
    return tuple(np.quantile(boot, [(1 - conf) / 2, 1 - (1 - conf) / 2]))


def ic_bca(x, estad, B=10000, conf=0.95, rng=rng):
    x = np.asarray(x, float)
    n, th = len(x), estad(x, axis=-1)
    boot = dist_bootstrap(x, estad, B, rng)
    z0 = stats.norm.ppf(np.mean(boot < th) + 0.5 * np.mean(boot == th))      # correccion de sesgo
    jack = np.array([estad(np.delete(x, i), axis=-1) for i in range(n)])      # remuestras "deja uno fuera"
    d = jack.mean() - jack
    a = (d**3).sum() / (6 * ((d**2).sum()) ** 1.5)                           # aceleracion
    zq = stats.norm.ppf([(1 - conf) / 2, 1 - (1 - conf) / 2])
    alfas = stats.norm.cdf(z0 + (z0 + zq) / (1 - a * (z0 + zq)))
    return tuple(np.quantile(boot, alfas))


# Muestra ASIMETRICA (SINTETICA): ingreso mensual de 40 hogares (millones de pesos), lognormal
ingreso = rng.lognormal(mean=0.8, sigma=0.7, size=40)
print(f"media = {ingreso.mean():.3f} | mediana = {np.median(ingreso):.3f} | asimetria = {stats.skew(ingreso):.2f}")

ic_t_ = ic_media_t(ingreso)
ic_p_ = ic_percentil(ingreso, np.mean, rng=np.random.default_rng(1))
ic_b_ = ic_bca(ingreso, np.mean, rng=np.random.default_rng(1))
ref = stats.bootstrap((ingreso,), np.mean, confidence_level=0.95, n_resamples=9999, method="BCa", random_state=np.random.default_rng(5))
ref_p = stats.bootstrap((ingreso,), np.mean, confidence_level=0.95, n_resamples=9999, method="percentile", random_state=np.random.default_rng(5))
print(f"IC 95% de la media: t = [{ic_t_[0]:.3f}, {ic_t_[1]:.3f}] | percentil = [{ic_p_[0]:.3f}, {ic_p_[1]:.3f}] | BCa = [{ic_b_[0]:.3f}, {ic_b_[1]:.3f}]")
print(f"                    SciPy percentil = [{ref_p.confidence_interval.low:.3f}, {ref_p.confidence_interval.high:.3f}] | SciPy BCa = [{ref.confidence_interval.low:.3f}, {ref.confidence_interval.high:.3f}]")
anch = ic_b_[1] - ic_b_[0]
assert abs(ic_p_[0] - ref_p.confidence_interval.low) < 0.05 * anch and abs(ic_p_[1] - ref_p.confidence_interval.high) < 0.05 * anch
assert abs(ic_b_[0] - ref.confidence_interval.low) < 0.05 * anch and abs(ic_b_[1] - ref.confidence_interval.high) < 0.05 * anch
assert ic_b_[0] < ingreso.mean() < ic_b_[1]

# La MEDIANA no tiene formula sencilla para su error estandar: el bootstrap si la da
med_ic = ic_bca(ingreso, np.median, rng=np.random.default_rng(2))
boot_med = dist_bootstrap(ingreso, np.median, rng=np.random.default_rng(2))
print(f"Mediana = {np.median(ingreso):.3f} | error estandar bootstrap = {boot_med.std(ddof=1):.3f} | IC BCa 95% = [{med_ic[0]:.3f}, {med_ic[1]:.3f}]")
assert med_ic[0] < np.median(ingreso) < med_ic[1]

boot_media = dist_bootstrap(ingreso, np.mean, rng=np.random.default_rng(3))
fig, axs = plt.subplots(1, 2, figsize=(11, 4.2))
for ax, boot, nom, ic in ((axs[0], boot_media, "media", ic_b_), (axs[1], boot_med, "mediana", med_ic)):
    ax.hist(boot, bins=50, color=PALETA[0], alpha=0.75, density=True)
    ax.axvline(ic[0], color=PALETA[1], ls="--"); ax.axvline(ic[1], color=PALETA[1], ls="--", label="IC BCa 95 %")
    ax.set(title=f"Distribucion bootstrap de la {nom}", xlabel="valor del estadistico"); ax.legend()
plt.tight_layout(); plt.show()

# Cobertura: t vs. bootstrap con datos asimetricos (lognormal(0, 0.9), n = 25, 1 000 muestras)
SIG_LN, reps_b, n_b = 0.9, 1000, 25
mu_ln = np.exp(SIG_LN**2 / 2)                                              # media verdadera de la lognormal
rb = np.random.default_rng(7)
aciertos = {"t": 0, "percentil": 0, "BCa": 0}
for _ in range(reps_b):
    x = rb.lognormal(0.0, SIG_LN, n_b)
    lo, hi = ic_media_t(x);                              aciertos["t"] += lo <= mu_ln <= hi
    lo, hi = ic_percentil(x, np.mean, B=1000, rng=rb);   aciertos["percentil"] += lo <= mu_ln <= hi
    lo, hi = ic_bca(x, np.mean, B=1000, rng=rb);         aciertos["BCa"] += lo <= mu_ln <= hi
cob_b = pd.Series({k: v / reps_b for k, v in aciertos.items()})
print(f"\nCobertura (nominal 0.95) con n = {n_b} y asimetria fuerte, {reps_b} muestras:\n", cob_b.round(3).to_string())
assert cob_b.max() < 0.95 and cob_b.min() > 0.80                          # ninguno llega al 95 %, pero todos son razonables
assert cob_b["BCa"] >= cob_b["percentil"] - 0.02
````

**Lectura.** (1) Con 40 hogares de ingreso muy asimétrico (asimetría 2.69; media 2.72, mediana 2.35), el IC $t$ es $[2.04,\,3.40]$, el percentil $[2.13,\,3.42]$ y el BCa $[2.23,\,3.63]$: el BCa **se desplaza a la derecha**, reflejando la cola larga que el IC $t$ (simétrico) ignora. Nuestro BCa coincide con el de `scipy.stats.bootstrap` ($[2.229,\,3.628]$) salvo ruido de Monte Carlo. (2) La **mediana**, que no tiene fórmula sencilla, queda con error estándar bootstrap 0.32 y IC BCa $[1.68,\,2.89]$. (3) **Resultado honesto sobre cobertura:** con $n=25$ y asimetría fuerte, **ningún** método llega al 95 %: $t$ cubre **89.6 %**, percentil **88.5 %** y BCa **88.5 %**. El *bootstrap* **no mejoró** al IC $t$ en este escenario; con muestras pequeñas tiende a producir intervalos **demasiado estrechos**. Su ventaja aparece con $n$ moderado/grande o con estadísticos para los que $t$ no existe. Más datos o transformar (p. ej., logaritmo) suelen servir más que un método sofisticado.

Una **limitación general:** todos estos intervalos suponen una **muestra aleatoria** de la población de interés. Si el muestreo está sesgado (p. ej., solo parcelas a la orilla del camino), el intervalo será **preciso pero equivocado**.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Estimar con incertidumbre** | Un **estimador** es una variable aleatoria; su variabilidad es el **error estándar**. La receta general es **estimación ± valor crítico × EE** |
| **Interpretación del IC** | «95 %» describe la **tasa de aciertos a largo plazo** del procedimiento (verificada por simulación: 96 de 100 en el gráfico), **no** la probabilidad de que $\mu$ esté en *este* intervalo |
| **Media: $z$ vs. $t$** | Con σ desconocida se usa $t_{n-1}$. Usar $z$ con $s$ **subcubre** con $n$ pequeño (81 % con $n=3$; 87 % con $n=5$); el IC $t$ cubre ≈ 95 % para todo $n$ bajo normalidad |
| **Ancho y tamaño de muestra** | El ancho crece con la confianza y la variabilidad y cae como $1/\sqrt{n}$; $n\ge(z\sigma/E)^2$. **Cuadruplicar** $n$ **reduce a la mitad** el ancho |
| **IC vs. predicción** | El IC de la media (ancho 2.3) **no** contiene a las observaciones individuales (solo 35 % de las nuevas); el intervalo de predicción (ancho 12.9) sí (95 %) |
| **Proporciones** | **Wald** subcubre (cobertura media 0.882, mínima 0.14 con $n=30$) y colapsa con 0 éxitos; **Wilson/Agresti–Coull** son la opción por defecto; **Clopper–Pearson** garantiza ≥ 95 % |
| **Diferencia de medias** | **Welch** (sin suponer varianzas iguales). El IC es la otra cara de la prueba: rechaza al 5 % ⇔ excluye al 0, y además muestra **qué tan grande** podría ser el efecto |
| **Varianza** | El IC $\chi^2$ **no es robusto**: cubre 95 % con normales pero 72 % con exponenciales |
| **Bootstrap** | Percentil y BCa implementados desde cero y contrastados con SciPy. Sirve para la **mediana**, el CV y otros estadísticos sin fórmula, pero **no hace magia**: con $n=25$ y asimetría fuerte ningún método llegó al 95 % (t 89.6 %, percentil 88.5 %, BCa 88.5 %) |
