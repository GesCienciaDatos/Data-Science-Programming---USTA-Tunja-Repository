Una prueba de hipótesis funciona como un **juicio**: se parte de que **«no pasa nada»** (como la presunción de inocencia) y solo se «condena» cuando los datos serían **muy difíciles de explicar por pura casualidad**.

| En el juicio | En estadística |
|---|---|
| Presunción de inocencia | La hipótesis de que **«no pasa nada»** (la moneda es justa, los abonos rinden igual) |
| Las pruebas | Los **datos** |
| «¿Qué tan raro sería esto si el acusado fuera inocente?» | El **valor p** |
| Condenar | **Rechazar** que «no pasa nada» |
| Condenar a un inocente | **Falsa alarma** (como una alarma de humo que suena sin fuego) |
| Dejar libre a un culpable | **Alarma que no suena** aunque haya fuego |
| «No hubo pruebas suficientes» | **No rechazar**: *no* es lo mismo que «demostrado inocente» |

## 1. ¿Es Sospechosa Esta Moneda? 🪙

Alguien te dice que su moneda es justa. La tiras **100 veces** y salen **62 caras**. ¿Es sospechoso? Para saberlo preguntamos: *«si la moneda fuera justa, ¿qué tan seguido pasaría algo tan raro como esto?»*. No hace falta ninguna fórmula: **simulamos 10 000 veces** «100 tiros de una moneda justa» y contamos. Esa fracción es el **valor p**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# Una moneda que dice ser justa. La tiramos 100 veces y salen 62 caras. ¿Es sospechoso?
# Idea: simulamos 10 000 veces "100 tiros de una moneda JUSTA" y vemos que tan seguido pasa algo tan raro.
caras = rng.binomial(100, 0.5, 10000)
tan_raro = np.mean((caras >= 62) | (caras <= 38))                              # 62 o mas caras, o 38 o menos (igual de raro hacia el otro lado)
print(f"Con una moneda justa, un resultado tan extremo como 62 caras ocurre {tan_raro:.1%} de las veces (ese es el 'valor p')")
assert 0.01 < tan_raro < 0.04

con_55 = np.mean((caras >= 55) | (caras <= 45))
print(f"En cambio, 55 caras (o 45) ocurre {con_55:.1%} de las veces: eso es muy normal, no sorprende.")
assert 0.25 < con_55 < 0.45

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(caras, bins=range(30, 71), color="#f59e0b", alpha=0.8)
ax.axvline(62, color="#dc2626", ls="--"); ax.axvline(38, color="#dc2626", ls="--")
ax.set(xlabel="caras en 100 tiros de una moneda justa", ylabel="veces (de 10 000)", title="Lo que pasa 'por pura suerte' con una moneda justa")
plt.show()
````

---
**Lección:** con una moneda **justa**, un resultado tan extremo como 62 caras (o 38 o menos) pasa **≈ 2.5 %** de las veces: es **poco común**, así que hay **motivos para sospechar**. En cambio 55 caras pasa **37 %** de las veces: es **normal**, no sorprende. Ese «2.5 %» es el **valor p**, y se lee **así**: «*si la moneda fuera justa*, solo 2.5 de cada 100 tandas darían algo tan raro». **No** es «la probabilidad de que la moneda sea justa». Y un valor p de 2.5 % **tampoco prueba** que la moneda esté cargada: puede ser que justo te tocara esa tanda rara (le pasa a 1 de cada 40 monedas justas).

## 2. Dos Abonos: ¿Diferencia Real o Suerte? 🌱

Probamos dos abonos en **12 parcelas cada uno** (rendimiento en toneladas por hectárea, **inventado**). El abono B rindió más en promedio, pero **¿y si fue suerte?** Si los abonos fueran **iguales**, las etiquetas «A» y «B» no importarían: podemos **revolverlas al azar** 10 000 veces y ver **qué tan seguido** aparece, solo por revolver, una diferencia tan grande como la que vimos. Al final, una advertencia importante sobre **«significativo» vs. «importante»**.

````python
# Dos abonos en 12 parcelas cada uno (rendimiento en toneladas por hectarea, INVENTADO)
abono_a = rng.normal(20, 4, 12).round(1)
abono_b = rng.normal(23, 4, 12).round(1)
dif_obs = abono_b.mean() - abono_a.mean()
print(f"Promedio A = {abono_a.mean():.1f} | promedio B = {abono_b.mean():.1f} | diferencia observada = {dif_obs:.1f} t/ha")

# Si los abonos fueran IGUALES, las etiquetas 'A' y 'B' no importarian: las revolvemos 10 000 veces
todos = np.concatenate([abono_a, abono_b])
diferencias = np.empty(10000)
for i in range(10000):
    r = rng.permutation(todos)
    diferencias[i] = r[12:].mean() - r[:12].mean()
p = np.mean(np.abs(diferencias) >= abs(dif_obs))
print(f"Revolviendo las etiquetas, una diferencia tan grande (en cualquier direccion) pasa solo {p:.1%} de las veces")
assert 0 < p < 0.2 and dif_obs > 0

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(diferencias, bins=40, color="#7c3aed", alpha=0.7)
ax.axvline(dif_obs, color="#dc2626", ls="--", label="lo que observamos"); ax.axvline(-dif_obs, color="#dc2626", ls="--")
ax.set(xlabel="diferencia entre B y A cuando las etiquetas se revuelven", title="¿Suerte o diferencia real?"); ax.legend()
plt.show()

# 'Muy significativo' NO es 'muy importante': con muchisimas parcelas, hasta 0.3 t/ha de diferencia sale 'rara'
gr_a, gr_b = rng.normal(20.0, 4, 3000), rng.normal(20.3, 4, 3000)
dif_gr = gr_b.mean() - gr_a.mean()
tod = np.concatenate([gr_a, gr_b])
dif_perm = np.array([(lambda r: r[3000:].mean() - r[:3000].mean())(rng.permutation(tod)) for _ in range(2000)])
p_gr = np.mean(np.abs(dif_perm) >= abs(dif_gr))
print(f"\nCon 3000 parcelas por abono: diferencia = {dif_gr:.2f} t/ha (apenas {dif_gr / 20:.1%} del rendimiento) | valor p = {p_gr:.3f}")
assert abs(dif_gr) < 0.6 and p_gr < 0.05
````

---
**Lección:** B rindió **3.5 t/ha** más que A, pero al revolver las etiquetas una diferencia **así de grande (en cualquier dirección) aparece 6.8 %** de las veces: **no alcanza** el 5 % que se usa por costumbre, así que **no podemos descartar que sea suerte**; ojo: eso **no demuestra** que los abonos sean iguales; con solo 12 parcelas por abono, **faltan datos**. Y fíjate en el otro extremo: con **3 000 parcelas por abono**, una diferencia de apenas **0.29 t/ha (1.4 % del rendimiento)** da un valor p de **0.005**, o sea «muy significativo»: **la significancia solo dice que no parece suerte, no que valga la pena**. Siempre hay que mirar **cuánto** cambia el rendimiento y **cuánto cuesta** el abono. (Con 12 parcelas no se «ve» ni un efecto grande; con 3 000, se «ve» hasta uno ínfimo.)

## Resumen en una frase 🎯

> **Un valor p dice qué tan raro sería lo que viste si «no pasara nada»: si es muy chico, desconfías de la casualidad; si es grande, simplemente no tienes pruebas suficientes (no que no haya diferencia), y aun cuando sea chico hay que preguntarse si la diferencia importa.**

### Lo que aprendiste hoy:
- ✅ **Valor p** = con qué frecuencia ocurre algo **tan raro como lo observado** si «no pasa nada» (aquí lo medimos simulando).
- ✅ **Revolver las etiquetas** responde «¿diferencia real o suerte?» sin fórmulas.
- ✅ **No rechazar ≠ demostrar que no hay diferencia**: pueden faltar datos.
- ✅ **Muy significativo ≠ muy importante**: con muchísimos datos, lo diminuto sale «significativo».
- ✅ Habrá **falsas alarmas** (≈ 1 de cada 20 con el criterio del 5 %) y **alarmas que no suenan**.
