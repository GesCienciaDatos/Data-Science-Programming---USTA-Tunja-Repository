## 1. ¿El Cultivo Tiene que Ver con el Riego? 🥔🧅

Una encuesta **inventada** a **200 fincas**: 100 de papa y 100 de cebolla; se pregunta si usan **riego tecnificado**. Si el cultivo **no** tuviera que ver con el riego, **las dos filas se verían iguales**. Calculamos cuántas fincas **esperaríamos** en cada casilla en ese caso y medimos la **sorpresa total** (qué tanto se alejan los conteos reales). Después **revolvemos** al azar quién tiene riego, 5 000 veces, para ver cuánta sorpresa produce **la pura suerte**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# Encuesta INVENTADA a 200 fincas: ¿usan riego tecnificado? (100 de papa y 100 de cebolla)
tabla = pd.DataFrame({"con riego": [48, 30], "sin riego": [52, 70]}, index=["papa", "cebolla"])
print(tabla)

# Si el cultivo NO tuviera que ver con el riego, las dos filas se verian iguales: ¿cuantas fincas ESPERARIAMOS en cada casilla?
total_fila = tabla.sum(axis=1).values.reshape(-1, 1)
total_col = tabla.sum(axis=0).values.reshape(1, -1)
esperados = total_fila * total_col / tabla.values.sum()
print("\nEsperado si no hubiera relacion:\n", pd.DataFrame(esperados, index=tabla.index, columns=tabla.columns))

# 'Sorpresa total': que tanto se aleja lo observado de lo esperado (cada casilla pesa segun cuanto se esperaba)
sorpresa = ((tabla.values - esperados) ** 2 / esperados).sum()
print(f"\nSorpresa total = {sorpresa:.2f}")

# ¿Cuanta sorpresa produce la pura suerte? Revolvemos al azar quien tiene riego, 5 000 veces
cultivo = np.repeat([0, 1], 100)
riego = np.array([1] * 48 + [0] * 52 + [1] * 30 + [0] * 70)
sorpresas = np.empty(5000)
for i in range(5000):
    r = rng.permutation(riego)
    obs = np.array([[r[:100].sum(), 100 - r[:100].sum()], [r[100:].sum(), 100 - r[100:].sum()]])
    esp = obs.sum(axis=1, keepdims=True) * obs.sum(axis=0, keepdims=True) / obs.sum()
    sorpresas[i] = ((obs - esp) ** 2 / esp).sum()
p = np.mean(sorpresas >= sorpresa)
print(f"Con pura suerte, una sorpresa de {sorpresa:.1f} o mas ocurre {p:.1%} de las veces")
assert abs(sorpresa - 6.9) < 0.3 and p < 0.05

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(sorpresas, bins=40, color="#0891b2", alpha=0.7); ax.axvline(sorpresa, color="#dc2626", ls="--", label="sorpresa observada")
ax.set(xlabel="sorpresa total cuando se revuelve el riego", title="¿Tiene que ver el cultivo con el riego?"); ax.legend()
plt.show()
````

---
**Lección:** si el cultivo no importara, esperaríamos **39 fincas con riego y 61 sin riego en cada cultivo**; lo observado fue 48 y 30 con riego. La **sorpresa total** es **6.81**, y revolviendo el riego al azar una sorpresa así de grande ocurre solo **≈ 1.2 %** de las veces: **los datos son poco compatibles con «no tiene que ver»**. Ojo con lo que **no** dice: **no** prueba que el cultivo **cause** el riego (pueden influir el tamaño de la finca, el agua disponible o el crédito), y **no** dice **cuánta** relación hay, solo que no parece casualidad.

## 2. ¿Rinden Distinto Cuatro Variedades de Papa? 📊

Cuatro variedades, **10 parcelas cada una** (**inventado**). Ya vimos los promedios; ¿qué tan en serio tomarlos? Medimos un **contraste**: cuánto **se separan los promedios** de las variedades **frente a** cuánto **varían las parcelas** de una misma variedad. Si las variedades fueran iguales, ese contraste sería chico; para ver **cuán grande puede ser por suerte** revolvemos las 40 parcelas entre las variedades 5 000 veces. Al final, una trampa: **comparar todas las parejas**.

````python
# Cuatro variedades de papa, 10 parcelas cada una (INVENTADO)
variedades = ["A", "B", "C", "D"]
rend = np.array([rng.normal(m, 4, 10) for m in (22, 24, 23, 28)])             # fila = variedad
prom = rend.mean(axis=1)
print("Promedios por variedad:", {v: round(float(x), 1) for v, x in zip(variedades, prom)})


def contraste(r):
    '''Cuanto se separan los promedios de los grupos (entre) frente a cuanto varian las parcelas de una misma variedad (dentro).'''
    entre = r.mean(axis=1).var(ddof=1)
    dentro = r.var(axis=1, ddof=1).mean() / r.shape[1]
    return entre / dentro


c_obs = contraste(rend)
# ¿Cuanto contraste produce la pura suerte? Revolvemos las 40 parcelas entre las 4 variedades, 5 000 veces
planas = rend.ravel()
c_azar = np.array([contraste(rng.permutation(planas).reshape(4, 10)) for _ in range(5000)])
p = np.mean(c_azar >= c_obs)
print(f"Contraste observado = {c_obs:.1f} | con pura suerte, un contraste tan alto ocurre {p:.2%} de las veces")
assert c_obs > 1 and p < 0.05

# Pero OJO: si comparamos TODAS las parejas (6 comparaciones) en variedades que son IGUALES, a veces alguna 'sale distinta' por suerte
n_exp = 3000
grupos = rng.normal(0, 1, (n_exp, 4, 10))                                      # 4 variedades IDENTICAS
m = grupos.mean(axis=2)
se = np.sqrt(grupos.var(axis=2, ddof=1).mean(axis=1) * 2 / 10)                 # margen tipico de una diferencia
alguna = np.zeros(n_exp, dtype=bool)
for i in range(4):
    for j in range(i + 1, 4):
        alguna |= np.abs(m[:, i] - m[:, j]) > 2.02 * se
print(f"Con 4 variedades IGUALES y 6 comparaciones, en {alguna.mean():.1%} de los experimentos alguna pareja 'salio distinta' solo por suerte")
assert 0.12 < alguna.mean() < 0.30

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(variedades, prom, color=["#2563eb", "#16a34a", "#f59e0b", "#7c3aed"])
ax.set(ylabel="rendimiento promedio (t/ha)", title="Promedios por variedad"); plt.show()
````

---
**Lección:** los promedios son A **20.0**, B **22.1**, C **24.3** y D **28.6** t/ha. El contraste observado es **12.1** y, de **5 000** revueltas al azar, **ninguna** llegó a un contraste así: no parece casualidad que las variedades rindan distinto. Pero **no sabemos aún cuáles** difieren de cuáles: este resultado solo dice que **al menos una** es distinta. Y mira la **trampa**: con **4 variedades idénticas** y **6 parejas** que comparar, en **20.5 %** de los experimentos **alguna pareja «salió distinta» solo por suerte** (y eso que cada pareja se juzgó con un criterio «del 5 %»). Por eso, cuando se comparan **muchas parejas**, se usan métodos que **corrigen** ese riesgo (en el cuaderno estándar: Tukey y Bonferroni).

## Resumen en una frase 🎯

> **Para conteos en categorías se compara lo observado con lo esperado si «nada tuviera que ver»; para varios grupos se compara cuánto se separan los promedios frente a cuánto varía cada grupo; en ambos casos «revolver» los datos muestra qué produce la pura suerte, y al mirar muchas parejas siempre habrá alguna «distinta» por azar.**

### Lo que aprendiste hoy:
- ✅ **Esperado vs. observado:** en la encuesta, «sorpresa total» 6.81 (≈ 1.2 % por pura suerte).
- ✅ **Varios grupos:** un contraste de 12.1 que ninguna de las 5 000 revueltas igualó; pero **no dice cuál** variedad es distinta.
- ✅ **Comparar muchas parejas** → ≈ 20 % de **falsas alarmas** aunque todo sea igual.
- ✅ **Asociación no es causalidad**, y «significativo» no dice **cuánto** ni **por qué**.
