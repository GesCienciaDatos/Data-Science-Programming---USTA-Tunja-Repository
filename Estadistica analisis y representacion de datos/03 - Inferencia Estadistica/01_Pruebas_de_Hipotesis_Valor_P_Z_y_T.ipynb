## 1. La Lógica de una Prueba de Hipótesis ⚖️

**Protocolo en seis pasos:**

| Paso | Qué se hace | Ejemplo (llenado de bultos) |
|---|---|---|
| 1 | Plantear $H_0$ («no pasa nada») y $H_1$ (la alternativa), **antes de ver los datos** | $H_0:\mu=50$ kg frente a $H_1:\mu\neq 50$ (o $\mu<50$) |
| 2 | Fijar el **nivel de significancia** $\alpha$ (tolerancia al error tipo I) | $\alpha=0.05$ |
| 3 | Elegir un **estadístico** cuya distribución se conozca **si $H_0$ fuera cierta** (distribución nula) | $Z=\dfrac{\bar{x}-\mu_0}{\sigma/\sqrt{n}}\sim N(0,1)$ |
| 4 | Calcular el estadístico con los datos | $z=-1.96$ |
| 5 | Calcular el **valor p** o la **región crítica** | $p=2\,\mathbb{P}(Z\ge\lvert z\rvert)$ |
| 6 | **Decidir**: rechazar $H_0$ si $p\le\alpha$ (equivale a estar en la región crítica) | y **reportar tamaño del efecto e IC** |

**Definición correcta del valor p.** Es la probabilidad, **calculada suponiendo que $H_0$ es cierta** (y que se cumplen los supuestos del modelo), de obtener un estadístico **al menos tan extremo** como el observado:

$$p=\mathbb{P}_{H_0}\big(\,|T|\ge|t_{\text{obs}}|\,\big)\quad(\text{dos colas}).$$

**Errores de decisión.**

| | $H_0$ cierta | $H_0$ falsa |
|---|---|---|
| **Rechazar $H_0$** | **Error tipo I** (falso positivo), probabilidad $\alpha$ | Decisión correcta, probabilidad $1-\beta$ = **potencia** |
| **No rechazar $H_0$** | Decisión correcta, probabilidad $1-\alpha$ | **Error tipo II** (falso negativo), probabilidad $\beta$ |

**Lo que el valor p NO es** (Declaración de la ASA, 2016):

| Afirmación | ¿Correcta? |
|---|---|
| «$p=0.03$ significa que hay 3 % de probabilidad de que $H_0$ sea cierta» | ❌ $H_0$ es cierta o no; $p$ se calcula **asumiendo** que lo es |
| «$p=0.03$ significa que hay 97 % de probabilidad de que el efecto sea real» | ❌ |
| «$p>0.05$ demuestra que no hay efecto» | ❌ «No rechazar» ≠ «aceptar $H_0$»: puede faltar potencia |
| «$p<0.05$ significa que el efecto es importante» | ❌ la significancia depende de $n$; hay que ver el **tamaño del efecto** |
| «Si $H_0$ fuera cierta, solo el 3 % de las muestras daría un resultado tan extremo» | ✅ |

**Una sola cola o dos.** $H_1:\mu\neq\mu_0$ (**bilateral**) es lo habitual; $H_1:\mu<\mu_0$ (**unilateral**) solo se justifica si **antes de ver los datos** un efecto en la otra dirección sería irrelevante o imposible. **Elegir la cola después de ver los datos duplica el error tipo I real.**

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas scipy statsmodels matplotlib

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.stats.multitest import multipletests
from statsmodels.stats.power import TTestIndPower
from statsmodels.stats.weightstats import ztest

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {__import__('scipy').__version__} | statsmodels {__import__('statsmodels').__version__}")
````

## 2. Prueba Z para una Media (σ Conocida) 📐

Si $X_1,\dots,X_n\sim N(\mu,\sigma^2)$ con $\sigma$ **conocida** (por ejemplo, la precisión certificada de una báscula), bajo $H_0:\mu=\mu_0$ el estadístico

$$Z=\frac{\bar{X}-\mu_0}{\sigma/\sqrt{n}}\sim N(0,1),$$

y los valores p son $2\,\mathbb{P}(Z\ge\lvert z\rvert)$ (bilateral), $\mathbb{P}(Z\le z)$ ($H_1:\mu<\mu_0$) o $\mathbb{P}(Z\ge z)$ ($H_1:\mu>\mu_0$). La **región crítica** al nivel $\alpha=0.05$ bilateral es $|z|\ge 1.96$. Usamos 36 bultos de papa **sintéticos** con etiqueta de 50 kg y $\sigma=1.5$ kg conocida. `statsmodels.ztest` estima la desviación con la muestra, así que para **contrastar la mecánica** llamamos a nuestra función con $\sigma=s$.

````python
# ============================================================
# 2. Prueba Z para una media (sigma conocida)
# ============================================================
# Datos SINTETICOS: peso (kg) de 36 bultos de papa; la etiqueta dice 50 kg y la bascula tiene sigma conocida = 1.5 kg
MU0, SIGMA = 50.0, 1.5
bultos = rng.normal(49.4, SIGMA, 36)                                         # (el llenado real esta ~0.6 kg por debajo)


def prueba_z(x, mu0, sigma, alternativa="dos_colas"):
    x = np.asarray(x, float)
    z = (x.mean() - mu0) / (sigma / np.sqrt(len(x)))
    if alternativa == "dos_colas":
        p = 2 * stats.norm.sf(abs(z))
    elif alternativa == "menor":
        p = stats.norm.cdf(z)
    else:
        p = stats.norm.sf(z)
    return z, p


z, p2 = prueba_z(bultos, MU0, SIGMA)
_, p_menor = prueba_z(bultos, MU0, SIGMA, "menor")
print(f"n = {len(bultos)} | media = {bultos.mean():.3f} kg | z = {z:.3f} | p (dos colas) = {p2:.4f} | p (H1: mu < 50) = {p_menor:.4f}")
zc = stats.norm.ppf(0.975)
print(f"Valor critico (alpha = 0.05, dos colas): +-{zc:.3f} -> {'se rechaza' if abs(z) > zc else 'no se rechaza'} H0")

# Contraste con statsmodels (que estima sigma con s): usamos sigma = s para comparar la mecanica
s = bultos.std(ddof=1)
for alt_sm, alt in (("two-sided", "dos_colas"), ("smaller", "menor"), ("larger", "mayor")):
    assert np.allclose(prueba_z(bultos, MU0, s, alt), ztest(bultos, value=MU0, alternative=alt_sm))
assert np.isclose(p2, 2 * p_menor)                                           # simetria: dos colas = 2 * una cola (si la cola es la del signo observado)
# Dualidad con el IC: se rechaza H0 (alpha = 0.05) <=> mu0 queda fuera del IC 95 %
ic = bultos.mean() + np.array([-1, 1]) * zc * SIGMA / np.sqrt(len(bultos))
assert (p2 < 0.05) == (not ic[0] <= MU0 <= ic[1])
print(f"IC 95% de mu (sigma conocida) = [{ic[0]:.3f}, {ic[1]:.3f}] -> contiene a {MU0}: {ic[0] <= MU0 <= ic[1]}")

# Region critica y valor p dibujados
xs = np.linspace(-4, 4, 600)
fig, ax = plt.subplots(figsize=(9, 4.2))
ax.plot(xs, stats.norm.pdf(xs), color="black", lw=1.5)
ax.fill_between(xs, stats.norm.pdf(xs), where=np.abs(xs) >= zc, color=PALETA[1], alpha=0.25, label="region critica (alpha = 0.05)")
ax.fill_between(xs, stats.norm.pdf(xs), where=np.abs(xs) >= abs(z), color=PALETA[0], alpha=0.5, label=f"area = valor p = {p2:.3f}")
ax.axvline(z, color=PALETA[0], ls="--"); ax.annotate(f"z observado = {z:.2f}", (z, 0.30), xytext=(8, 0), textcoords="offset points")
ax.set(xlabel="z", ylabel="densidad", title="Distribucion nula N(0, 1): region critica y valor p"); ax.legend(loc="upper right")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) El peso medio es 49.51 kg ($n=36$) y $z=-1.957$, **a 0.003 del valor crítico** $-1.960$: el valor p bilateral es **0.0504**, de modo que **no se rechaza** $H_0$ al 5 %; el IC 95 % $[49.02,\,50.00]$ contiene a 50 por un margen mínimo. (2) Con $H_1:\mu<50$ el valor p es **0.0252** y **sí se rechazaría**: pero eso solo es legítimo si la hipótesis unilateral se planteó **antes** de ver los datos. (3) **No hay un abismo** entre $p=0.0504$ y $p=0.0496$: la dicotomía «significativo/no significativo» es una **convención**; lo informativo es el **tamaño** del déficit (≈ 0.5 kg por bulto) con su IC. (4) La mecánica coincide con `statsmodels.ztest` en las tres alternativas, y la **dualidad** con el IC se cumple (se rechaza $\Leftrightarrow$ $\mu_0$ queda fuera del intervalo).

## 3. Pruebas t: Una Muestra, Welch, Student y Pareada 🎯

Cuando $\sigma$ es **desconocida** se estima con $s$ y el estadístico sigue una $t$ de Student (cuaderno anterior). Las cuatro variantes:

| Prueba | Estadístico | Grados de libertad | Cuándo |
|---|---|---|---|
| **Una muestra** | $t=\dfrac{\bar{x}-\mu_0}{s/\sqrt{n}}$ | $n-1$ | Comparar una media con un valor de referencia |
| **Welch** (dos muestras) | $t=\dfrac{\bar{x}_1-\bar{x}_2}{\sqrt{s_1^2/n_1+s_2^2/n_2}}$ | Welch–Satterthwaite (no entero) | Dos grupos **independientes**, varianzas posiblemente distintas (**recomendada por defecto**) |
| **Student agrupada** | $t=\dfrac{\bar{x}_1-\bar{x}_2}{s_p\sqrt{1/n_1+1/n_2}}$, con $s_p^2=\dfrac{(n_1-1)s_1^2+(n_2-1)s_2^2}{n_1+n_2-2}$ | $n_1+n_2-2$ | Dos grupos independientes **con varianzas iguales** |
| **Pareada** | $t$ de una muestra sobre las **diferencias** $d_i=x_i-y_i$ | $n-1$ | **Mismas** unidades medidas dos veces (antes/después) |

**Supuestos:** (i) **independencia** (garantizada por el diseño, no verificable con una prueba); (ii) **normalidad** de los datos (o de las diferencias en la pareada) o $n$ grande por el TLC; (iii) para Student, **varianzas iguales**. Se explora con Shapiro–Wilk y Levene, pero ojo: con $n$ pequeño estas pruebas tienen **poca potencia** (que no rechacen no *demuestra* el supuesto) y **no se recomienda elegir la prueba según el resultado de una pre-prueba**: Welch pierde muy poco cuando las varianzas son iguales y gana mucho cuando no lo son.

Datos **sintéticos**: rendimiento (t/ha) de dos variedades de papa en parcelas distintas, y de las **mismas 15 parcelas** antes y después de un programa de riego. Todo se verifica contra `scipy.stats` (las tres alternativas).

````python
# ============================================================
# 3. Pruebas t: una muestra, Welch, Student (agrupada) y pareada
# ============================================================
def t_una_muestra(x, mu0, alt="dos_colas"):
    x = np.asarray(x, float)
    t, gl = (x.mean() - mu0) / (x.std(ddof=1) / np.sqrt(len(x))), len(x) - 1
    return t, gl, _p_t(t, gl, alt)


def _p_t(t, gl, alt):
    return {"dos_colas": 2 * stats.t.sf(abs(t), gl), "menor": stats.t.cdf(t, gl), "mayor": stats.t.sf(t, gl)}[alt]


def t_welch(x, y, alt="dos_colas"):
    x, y = np.asarray(x, float), np.asarray(y, float)
    vx, vy = x.var(ddof=1) / len(x), y.var(ddof=1) / len(y)
    t = (x.mean() - y.mean()) / np.sqrt(vx + vy)
    gl = (vx + vy) ** 2 / (vx**2 / (len(x) - 1) + vy**2 / (len(y) - 1))
    return t, gl, _p_t(t, gl, alt)


def t_agrupada(x, y, alt="dos_colas"):                                        # Student: supone varianzas iguales
    x, y = np.asarray(x, float), np.asarray(y, float)
    gl = len(x) + len(y) - 2
    sp2 = ((len(x) - 1) * x.var(ddof=1) + (len(y) - 1) * y.var(ddof=1)) / gl
    t = (x.mean() - y.mean()) / np.sqrt(sp2 * (1 / len(x) + 1 / len(y)))
    return t, gl, _p_t(t, gl, alt)


def t_pareada(a, b, alt="dos_colas"):                                         # = t de una muestra sobre las diferencias
    return t_una_muestra(np.asarray(a, float) - np.asarray(b, float), 0.0, alt)


# Datos SINTETICOS: rendimiento (t/ha) de dos variedades de papa en parcelas distintas
var_a = rng.normal(24.0, 4.0, 20)
var_b = rng.normal(21.5, 5.0, 24)
# Y rendimiento de las MISMAS 15 parcelas antes y despues de un programa de riego (datos pareados)
antes = rng.normal(20.0, 4.0, 15)
despues = antes + rng.normal(1.5, 1.2, 15)

# Contraste con SciPy
assert np.allclose(t_una_muestra(bultos, 50.0), (*stats.ttest_1samp(bultos, 50.0)[:1], len(bultos) - 1, stats.ttest_1samp(bultos, 50.0).pvalue))
for alt_sp, alt in (("two-sided", "dos_colas"), ("less", "menor"), ("greater", "mayor")):
    r = stats.ttest_ind(var_a, var_b, equal_var=False, alternative=alt_sp)
    assert np.allclose(t_welch(var_a, var_b, alt)[::2], (r.statistic, r.pvalue))
    r = stats.ttest_ind(var_a, var_b, equal_var=True, alternative=alt_sp)
    assert np.allclose(t_agrupada(var_a, var_b, alt)[::2], (r.statistic, r.pvalue))
    r = stats.ttest_rel(despues, antes, alternative=alt_sp)
    assert np.allclose(t_pareada(despues, antes, alt)[::2], (r.statistic, r.pvalue))
assert np.isclose(t_welch(var_a, var_b)[1], stats.ttest_ind(var_a, var_b, equal_var=False).df)

tw, glw, pw = t_welch(var_a, var_b)
ta, gla, pa = t_agrupada(var_a, var_b)
tp, glp, pp = t_pareada(despues, antes)
tp_mal = t_welch(despues, antes)                                              # error: ignorar el emparejamiento
print(f"Welch    : t = {tw:6.3f}  gl = {glw:5.1f}  p = {pw:.4f}   (medias {var_a.mean():.2f} vs {var_b.mean():.2f})")
print(f"Agrupada : t = {ta:6.3f}  gl = {gla:5.1f}  p = {pa:.4f}")
print(f"Pareada  : t = {tp:6.3f}  gl = {glp:5.1f}  p = {pp:.2e}  (diferencia media = {(despues - antes).mean():.2f})")
print(f"Welch aplicada (mal) a los datos pareados: p = {tp_mal[2]:.4f}  <- ignorar el emparejamiento pierde potencia")
assert pp < tp_mal[2]

# Supuestos: normalidad (Shapiro-Wilk) e igualdad de varianzas (Levene, centrada en la mediana)
print(f"\nShapiro-Wilk: A p = {stats.shapiro(var_a).pvalue:.3f} | B p = {stats.shapiro(var_b).pvalue:.3f} | diferencias pareadas p = {stats.shapiro(despues - antes).pvalue:.3f}")
print(f"Levene (varianzas iguales): p = {stats.levene(var_a, var_b).pvalue:.3f} | s_A = {var_a.std(ddof=1):.2f}, s_B = {var_b.std(ddof=1):.2f}")

fig, axs = plt.subplots(1, 2, figsize=(11, 4.4))
axs[0].boxplot([var_a, var_b], widths=0.5, showmeans=True)
axs[0].set_xticks([1, 2], ["Variedad A", "Variedad B"])
for k, v in enumerate((var_a, var_b), start=1):
    axs[0].plot(k + rng.uniform(-0.12, 0.12, len(v)), v, "o", color=PALETA[k - 1], alpha=0.6, ms=4)
axs[0].set(ylabel="rendimiento (t/ha)", title=f"Dos grupos independientes (Welch p = {pw:.3f})")
for a_, d_ in zip(antes, despues):
    axs[1].plot([0, 1], [a_, d_], "-o", color=PALETA[2], alpha=0.5, ms=3)
axs[1].set(xticks=[0, 1], xticklabels=["Antes", "Despues"], ylabel="rendimiento (t/ha)", title=f"Datos pareados (t pareada p = {pp:.1e})")
plt.tight_layout(); plt.show()
````

**Lectura.** (1) Las cuatro funciones propias coinciden con `ttest_1samp`, `ttest_ind` (con y sin `equal_var`) y `ttest_rel` en las **tres alternativas**. (2) Variedades A y B: medias 23.99 y 21.39; Welch $t=2.60$ con $\nu=39.6$ gl y **$p=0.0130$**; Student agrupada $p=0.0163$. Se parecen porque los tamaños son similares (20 y 24); con **tamaños muy desiguales y varianzas distintas** pueden divergir bastante. (3) **Supuestos:** Shapiro–Wilk no rechaza la normalidad (A: 0.238; B: 0.769; diferencias pareadas: 0.443) y Levene sí detecta varianzas distintas ($p=0.035$; $s_A=2.57$, $s_B=4.00$): otra razón para preferir **Welch**. (4) **El emparejamiento importa:** la mejora media de las 15 parcelas es de 1.00 t/ha con **$p=0.0030$** en la prueba pareada, pero si se **ignora** el emparejamiento (Welch sobre los mismos datos) el valor p sube a **0.357**: la variabilidad **entre parcelas** esconde la mejora **dentro de cada una**. Usar la prueba que respeta el diseño es una cuestión de **validez y de potencia**.

## 4. ¿Qué Es (y Qué No Es) el Valor p? Simulación y Permutación 🎲

Para entender el valor p, simulamos lo que describe su definición. **(a)** Repetimos 10 000 veces un experimento en el que **$H_0$ es cierta** (dos grupos de 15 con la misma media) y registramos el valor p de Welch: si la prueba es válida, **el valor p es uniforme en $[0,1]$** y rechazar con $\alpha=0.05$ ocurre el 5 % de las veces (**error tipo I controlado**). Con un efecto real de 0.5 desviaciones, los valores p **se apilan cerca de 0**, pero **no todos**: la fracción que cae bajo 0.05 es la **potencia**.

**(b) Prueba de permutación.** Si $H_0$ dice que «las etiquetas A/B no importan», podemos **barajarlas** muchas veces, recalcular la diferencia de medias y ver **qué tan extrema** es la observada dentro de esa distribución nula. No asume normalidad: es la definición del valor p hecha con una computadora (y la misma idea del ***target shuffling*** del libro de Nisbet *et al.*). Usamos la corrección $(1+\#\{\text{extremos}\})/(B+1)$, que cuenta a la muestra observada como una permutación más.

````python
# ============================================================
# 4. Que es (y que no es) el valor p: simulacion y prueba de permutacion
# ============================================================
# (a) Distribucion de p bajo H0 (uniforme) y bajo H1 (apilada cerca de 0): 10 000 pruebas de Welch con n = 15 por grupo
R, n_g = 10000, 15
p_h0 = stats.ttest_ind(rng.normal(0, 1, (R, n_g)), rng.normal(0, 1, (R, n_g)), axis=1, equal_var=False).pvalue
p_h1 = stats.ttest_ind(rng.normal(0.5, 1, (R, n_g)), rng.normal(0, 1, (R, n_g)), axis=1, equal_var=False).pvalue
print(f"Bajo H0: P(p < 0.05) = {np.mean(p_h0 < 0.05):.4f} | P(p < 0.5) = {np.mean(p_h0 < 0.5):.4f} | KS contra U(0,1): D = {stats.kstest(p_h0, 'uniform').statistic:.4f}")
print(f"Bajo H1 (efecto 0.5 sd): P(p < 0.05) = {np.mean(p_h1 < 0.05):.4f}  <- esto es la POTENCIA con n = {n_g}")
assert abs(np.mean(p_h0 < 0.05) - 0.05) < 0.01 and stats.kstest(p_h0, "uniform").statistic < 0.02
assert 0.18 < np.mean(p_h1 < 0.05) < 0.34

# (b) Prueba de permutacion desde cero (diferencia de medias entre las variedades A y B)
def p_permutacion(x, y, estad=lambda a, b: a.mean(axis=-1) - b.mean(axis=-1), B=20000, rng=rng):
    x, y = np.asarray(x, float), np.asarray(y, float)
    todos = np.concatenate([x, y])
    obs = estad(x, y)
    idx = np.argsort(rng.random((B, len(todos))), axis=1)                      # B reasignaciones aleatorias de etiquetas
    nula = estad(todos[idx[:, :len(x)]], todos[idx[:, len(x):]])
    p = (1 + np.sum(np.abs(nula) >= abs(obs))) / (B + 1)                       # +1: el dato observado es una de las permutaciones
    return obs, p, nula


obs, p_perm, nula = p_permutacion(var_a, var_b, rng=np.random.default_rng(10))
ref = stats.permutation_test((var_a, var_b), lambda a, b, axis: a.mean(axis=axis) - b.mean(axis=axis), permutation_type="independent",
                             n_resamples=20000, alternative="two-sided", vectorized=True, random_state=np.random.default_rng(11))
p_w = stats.ttest_ind(var_a, var_b, equal_var=False).pvalue
print(f"\nDiferencia observada = {obs:.3f} | p (permutacion, propia) = {p_perm:.4f} | p (SciPy permutation_test) = {ref.pvalue:.4f} | p (Welch) = {p_w:.4f}")
assert abs(p_perm - ref.pvalue) < 0.01 and abs(p_perm - p_w) < 0.03

fig, axs = plt.subplots(1, 3, figsize=(13, 3.8))
axs[0].hist(p_h0, bins=20, range=(0, 1), color=PALETA[0], alpha=0.8); axs[0].set(title="p bajo H0: uniforme", xlabel="valor p", ylabel="frecuencia")
axs[1].hist(p_h1, bins=20, range=(0, 1), color=PALETA[1], alpha=0.8); axs[1].set(title="p bajo H1: apilado cerca de 0", xlabel="valor p")
axs[2].hist(nula, bins=50, color=PALETA[4], alpha=0.7)
axs[2].axvline(obs, color="black", ls="--"); axs[2].axvline(-obs, color="black", ls="--")
axs[2].set(title=f"Permutaciones (p = {p_perm:.3f})", xlabel="diferencia de medias bajo H0")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Bajo $H_0$, **4.68 %** de las pruebas rechazó con $\alpha=0.05$ y **48.8 %** tuvo $p<0.5$: la distribución es prácticamente uniforme (estadístico de Kolmogorov–Smirnov $D=0.015$): **un valor p de 0.03 no es «raro»: ocurre el 3 % de las veces aunque no haya efecto**. (2) Con un efecto real de 0.5 desviaciones y 15 por grupo, solo **25.1 %** de las pruebas rechaza: **tres de cada cuatro estudios no «encontrarían» un efecto que existe**; eso es una **potencia baja**, y es el tema de la sección siguiente. (3) La **prueba de permutación** da $p=0.0158$ para la diferencia observada de 2.60 t/ha (SciPy: 0.0174; Welch: 0.0130): mismo orden de magnitud sin suponer normalidad. (4) **Lectura correcta:** «si las variedades fueran intercambiables, solo ≈ 1.6 % de las reasignaciones de etiquetas daría una diferencia tan grande», **no** «hay 98.4 % de probabilidad de que B sea peor».

## 5. Errores Tipo I y II, Potencia y Tamaño del Efecto 🔋

**Potencia** $=1-\beta=\mathbb{P}(\text{rechazar }H_0\mid H_1\text{ cierta})$. Para la prueba $t$ de dos muestras con $n$ **por grupo**, efecto estandarizado $d=(\mu_1-\mu_2)/\sigma$ y varianzas iguales, bajo $H_1$ el estadístico sigue una **$t$ no central** con parámetro de no centralidad $\lambda=d\sqrt{n/2}$ y $2n-2$ grados de libertad; la potencia exacta es $\mathbb{P}(T>t_c)+\mathbb{P}(T<-t_c)$ con $T\sim t_{2n-2}(\lambda)$. La potencia **crece** con el tamaño del efecto, con $n$ y con $\alpha$, y **decrece** con la variabilidad; **$\alpha$ y $\beta$ se compensan**: bajar uno sube el otro, salvo que aumentemos $n$.

**Tamaño del efecto.** El valor p mezcla efecto y tamaño de muestra; el **tamaño del efecto** los separa. Para dos medias, la **$d$ de Cohen** y la **$g$ de Hedges** (corrige el sesgo con $n$ pequeño):

$$d=\frac{\bar{x}_1-\bar{x}_2}{s_p},\qquad g=d\left(1-\frac{3}{4(n_1+n_2)-9}\right).$$

Las etiquetas de Cohen (0.2 «pequeño», 0.5 «mediano», 0.8 «grande») son **convenciones**, no verdades: lo «grande» depende del campo y del costo. La celda calcula la potencia con la fórmula exacta, la contrasta con **statsmodels** y con una **simulación de Monte Carlo**, resuelve el $n$ necesario y dibuja el compromiso $\alpha$–$\beta$ y las curvas de potencia.

````python
# ============================================================
# 5. Errores tipo I y II, potencia y tamano del efecto
# ============================================================
def cohen_d(x, y):
    x, y = np.asarray(x, float), np.asarray(y, float)
    sp = np.sqrt(((len(x) - 1) * x.var(ddof=1) + (len(y) - 1) * y.var(ddof=1)) / (len(x) + len(y) - 2))
    return (x.mean() - y.mean()) / sp


def hedges_g(x, y):
    return cohen_d(x, y) * (1 - 3 / (4 * (len(x) + len(y)) - 9))               # correccion de sesgo para n pequenos


def ic_d(x, y, conf=0.95):                                                      # aproximacion normal (Hedges y Olkin)
    n1, n2, d = len(x), len(y), cohen_d(x, y)
    ee = np.sqrt((n1 + n2) / (n1 * n2) + d**2 / (2 * (n1 + n2)))
    z = stats.norm.ppf(1 - (1 - conf) / 2)
    return d - z * ee, d + z * ee


def potencia_t(d, n, alpha=0.05):                                               # prueba t de dos muestras, n POR GRUPO, dos colas
    gl, ncp = 2 * n - 2, d * np.sqrt(n / 2)
    tc = stats.t.ppf(1 - alpha / 2, gl)
    return stats.nct.sf(tc, gl, ncp) + stats.nct.cdf(-tc, gl, ncp)


d_obs, g_obs = cohen_d(var_a, var_b), hedges_g(var_a, var_b)
print(f"Variedades A y B: d de Cohen = {d_obs:.3f} | g de Hedges = {g_obs:.3f} | IC 95% de d ~ [{ic_d(var_a, var_b)[0]:.2f}, {ic_d(var_a, var_b)[1]:.2f}]")
assert abs(g_obs) < abs(d_obs)

# Potencia exacta (t no central) vs. statsmodels y vs. simulacion de Monte Carlo
for d, n in ((0.5, 30), (0.8, 15), (0.2, 100)):
    assert np.isclose(potencia_t(d, n), TTestIndPower().power(effect_size=d, nobs1=n, alpha=0.05, ratio=1.0), atol=1e-6)
rs = np.random.default_rng(12)
sim = np.mean(stats.ttest_ind(rs.normal(0.5, 1, (5000, 30)), rs.normal(0, 1, (5000, 30)), axis=1).pvalue < 0.05)
print(f"Potencia (d = 0.5, n = 30 por grupo): formula = {potencia_t(0.5, 30):.4f} | statsmodels = {TTestIndPower().power(0.5, 30, 0.05):.4f} | simulacion = {sim:.4f}")
assert abs(sim - potencia_t(0.5, 30)) < 0.02

def n_minimo(d, potencia=0.8, alpha=0.05):
    n = 2
    while potencia_t(d, n, alpha) < potencia:
        n += 1
    return n


tabla = pd.DataFrame({"d": [0.2, 0.5, 0.8]})
tabla["n por grupo (80 % potencia)"] = [n_minimo(d) for d in tabla.d]
tabla["n (90 % potencia)"] = [n_minimo(d, 0.9) for d in tabla.d]
print("\n", tabla.to_string(index=False))
assert n_minimo(0.5) == int(np.ceil(TTestIndPower().solve_power(effect_size=0.5, power=0.8, alpha=0.05))) == 64

# Compromiso alfa-beta (dibujo) y curvas de potencia
d_alt = 2.0                                                                      # efecto de 2 errores estandar (prueba z unilateral, para el dibujo)
xs = np.linspace(-4, 6, 800)
zc = stats.norm.ppf(0.95)                                                        # corte unilateral al 5 %
fig, axs = plt.subplots(1, 2, figsize=(12, 4.2))
axs[0].plot(xs, stats.norm.pdf(xs), color="black", label="H0 cierta"); axs[0].plot(xs, stats.norm.pdf(xs, d_alt), color=PALETA[0], label="H1 cierta (efecto = 2 EE)")
axs[0].fill_between(xs, stats.norm.pdf(xs), where=xs >= zc, color=PALETA[1], alpha=0.35, label=f"alfa (error I) = {stats.norm.sf(zc):.2f}")
axs[0].fill_between(xs, stats.norm.pdf(xs, d_alt), where=xs <= zc, color=PALETA[3], alpha=0.35, label=f"beta (error II) = {stats.norm.cdf(zc - d_alt):.2f}")
axs[0].axvline(zc, color="gray", ls="--"); axs[0].set(title="Errores tipo I y II (prueba z unilateral)", xlabel="estadistico"); axs[0].legend(fontsize=8)
ns = np.arange(5, 201)
for d, col in zip((0.2, 0.5, 0.8), PALETA):
    axs[1].plot(ns, [potencia_t(d, n) for n in ns], color=col, label=f"d = {d}")
axs[1].axhline(0.8, color="gray", ls="--"); axs[1].set(xlabel="n por grupo", ylabel="potencia", title="Curvas de potencia (t de dos muestras, alfa = 0.05)"); axs[1].legend()
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Entre las variedades A y B, $d=0.758$ y $g=0.744$: un efecto «mediano a grande» por convención, pero el IC aproximado de $d$ es **$[0.14,\,1.37]$**: con 20 y 24 parcelas el tamaño del efecto **no está bien determinado** (puede ser casi trivial o enorme). (2) Para $d=0.5$ y $n=30$ por grupo, la potencia es **0.4779** con la fórmula exacta, **idéntica a statsmodels** y **0.4898** por simulación: con ese diseño **fallamos en detectar un efecto mediano más de la mitad de las veces**. (3) Para llegar a **80 %** de potencia hacen falta **64 por grupo** si $d=0.5$, **26** si $d=0.8$ y **394** si $d=0.2$ (**15 veces** más datos para un efecto pequeño que para uno grande); para 90 % se necesitan 86, 34 y 527. (4) El dibujo muestra el **compromiso**: moviendo el umbral para reducir el error tipo I crece el II.

## 6. Significancia No Es Relevancia; y las Pruebas Múltiples 🧪

**(a) Significancia vs. relevancia.** Con $n$ enorme **cualquier** diferencia, por trivial que sea, produce $p$ diminuto; con $n$ pequeño, hasta efectos importantes pueden dar $p>0.05$, y los efectos que **sí** salen significativos tienden a estar **exagerados** (error tipo M, «maldición del ganador»). Comparamos dos «estudios» sintéticos y repetimos el de pocos datos 4 000 veces.

**(b) Pruebas múltiples.** Con $m$ pruebas independientes y **todas** las $H_0$ ciertas, $\mathbb{P}(\text{al menos un falso positivo})=1-(1-\alpha)^m$ (para $m=20$: ≈ 64 %). Correcciones: **Bonferroni** ($p_i\cdot m$) y **Holm** (secuencial, uniformemente más potente) controlan la tasa de error **por familia** (FWER); **Benjamini–Hochberg** controla la **tasa de falsos descubrimientos** (FDR), menos estricta y más potente. Las implementamos desde cero y las comparamos con `multipletests`.

````python
# ============================================================
# 6. Significancia no es relevancia; y el problema de las pruebas multiples
# ============================================================
# (a) Mismo "veredicto" p < 0.05 con efectos muy distintos; y p > 0.05 con efecto grande y pocos datos
def resumen(x, y, nombre):
    t, gl, p = t_welch(x, y)
    dif = x.mean() - y.mean()
    ee = np.sqrt(x.var(ddof=1) / len(x) + y.var(ddof=1) / len(y))
    return {"estudio": nombre, "n por grupo": len(x), "diferencia": dif, "IC 95 % dif.": f"[{dif - stats.t.ppf(.975, gl) * ee:.2f}, {dif + stats.t.ppf(.975, gl) * ee:.2f}]",
            "p": p, "d de Cohen": cohen_d(x, y)}


gen = np.random.default_rng(5)
est_grande = resumen(gen.normal(22.3, 6, 20000), gen.normal(22.0, 6, 20000), "A: muestra enorme, efecto real 0.3 t/ha")
est_chico = resumen(gen.normal(24, 6, 8), gen.normal(20, 6, 8), "B: pocas parcelas, efecto real 4 t/ha")
tab = pd.DataFrame([est_grande, est_chico])
tab["p"] = tab["p"].map("{:.2e}".format)
print(tab.round(3).to_string(index=False))
assert float(est_grande["p"]) < 0.05 and abs(est_grande["d de Cohen"]) < 0.1                  # "significativo" pero practicamente irrelevante
print(f"Potencia del diseno B (n = 8 por grupo) si el efecto real es d = 0.67 (4 t/ha con sd = 6): {potencia_t(4 / 6, 8):.2f}")
assert potencia_t(4 / 6, 8) < 0.3

# Repetimos el diseno B 4 000 veces: poca potencia => casi nunca "sale", y cuando sale EXAGERA el efecto (error tipo M, "maldicion del ganador")
XB, YB = gen.normal(24, 6, (4000, 8)), gen.normal(20, 6, (4000, 8))
pB = stats.ttest_ind(XB, YB, axis=1, equal_var=False).pvalue
difB = XB.mean(axis=1) - YB.mean(axis=1)
print(f"Diseno B repetido: significativo en {np.mean(pB < 0.05):.1%} de las veces | diferencia media estimada cuando SI es significativa = {difB[pB < 0.05].mean():.2f} (real = 4.00)")
assert 0.15 < np.mean(pB < 0.05) < 0.32 and difB[pB < 0.05].mean() > 1.5 * 4

# (b) Pruebas multiples: 20 pruebas con TODAS las H0 ciertas
m, reps = 20, 2000
P = stats.ttest_ind(rng.normal(0, 1, (reps, m, 15)), rng.normal(0, 1, (reps, m, 15)), axis=2).pvalue     # (reps, m)
print(f"\nP(al menos un p < 0.05) con {m} pruebas nulas: teoria = {1 - 0.95**m:.3f} | simulacion = {np.mean((P < 0.05).any(axis=1)):.3f}")
assert abs(np.mean((P < 0.05).any(axis=1)) - (1 - 0.95**m)) < 0.04


def ajustar(p, metodo):
    p = np.asarray(p, float); m = len(p); orden = np.argsort(p); aj = np.empty(m)
    if metodo == "bonferroni":
        aj = np.minimum(p * m, 1)
    elif metodo == "holm":
        aj[orden] = np.minimum(np.maximum.accumulate((m - np.arange(m)) * p[orden]), 1)
    elif metodo == "bh":                                                                   # Benjamini-Hochberg (controla la FDR)
        aj[orden] = np.minimum(np.minimum.accumulate((m / np.arange(1, m + 1) * p[orden])[::-1])[::-1], 1)
    return aj


# 20 hipotesis: 5 con efecto real (d = 1) y 15 nulas
efectos = np.r_[np.ones(5), np.zeros(15)]
p_mix = np.array([stats.ttest_ind(rng.normal(d, 1, 20), rng.normal(0, 1, 20)).pvalue for d in efectos])
for met, sm in (("bonferroni", "bonferroni"), ("holm", "holm"), ("bh", "fdr_bh")):
    assert np.allclose(ajustar(p_mix, met), multipletests(p_mix, method=sm)[1])
res = pd.DataFrame({"sin corregir": p_mix < 0.05, "Bonferroni": ajustar(p_mix, "bonferroni") < 0.05, "Holm": ajustar(p_mix, "holm") < 0.05, "BH (FDR 5 %)": ajustar(p_mix, "bh") < 0.05})
tabla_dec = pd.DataFrame({"descubrimientos": res.sum(), "verdaderos (de 5)": res[efectos == 1].sum(), "falsos (de 15)": res[efectos == 0].sum()})
print("\n", tabla_dec.to_string())
assert (res["Bonferroni"] <= res["Holm"]).all() and (res["Holm"] <= res["BH (FDR 5 %)"]).all()   # Holm domina a Bonferroni y BH es mas liberal
assert res["Bonferroni"].sum() <= res["sin corregir"].sum()
````

**Lectura.** (1) **Estudio A** ($n=20\,000$ por grupo): una diferencia de **0.37 t/ha** (IC $[0.25,\,0.49]$) da $p=8.8\times10^{-10}$ pero $d=0.061$: **altamente significativa y prácticamente irrelevante** (≈ 1.7 % del rendimiento medio); si mejorarla cuesta algo, quizá no valga la pena. (2) **Estudio B** ($n=8$): el efecto real es de 4 t/ha, pero la **potencia es solo 0.24**; en 4 000 repeticiones solo **22.4 %** resultó significativo y, **cuando lo fue, la diferencia estimada promedió 7.69 t/ha, casi el doble de la real**. El caso mostrado (diferencia 7.66, $p=0.014$, IC $[1.83,\,13.49]$) es justo uno de esos «éxitos» **exagerados**: un IC tan ancho debería **frenar el entusiasmo**. (3) **Multiplicidad:** con 20 pruebas nulas, $\mathbb{P}(\text{≥ 1 falso positivo})$ = **0.642** (teoría) y **0.640** (simulación). Con 20 hipótesis de las que 5 son reales: sin corregir hay **6 descubrimientos** (5 verdaderos y **1 falso**); **Bonferroni y Holm** dan 3 (todos verdaderos, pero se **pierden 2 reales**) y **BH** da 4 (todos verdaderos). Es **una** simulación: con otra semilla los números cambian, pero el patrón (corregir **evita falsos positivos a cambio de potencia**) es general. ## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Lógica** | $H_0$/$H_1$ antes de ver los datos → estadístico con distribución nula conocida → valor p → decisión. **Nunca «aceptar $H_0$»**: se rechaza o no se rechaza |
| **Valor p** | $\mathbb{P}_{H_0}(\text{resultado al menos tan extremo})$. **No** es $\mathbb{P}(H_0)$. Bajo $H_0$ es **uniforme** (4.7 % rechazó con $\alpha=0.05$); con $H_1$ y $n=15$ solo 25 % de las pruebas rechaza |
| **Z y t** | $Z$ con σ conocida; $t$ con σ estimada. **Welch** por defecto para dos grupos; **pareada** cuando hay emparejamiento (en el ejemplo, $p=0.003$ frente a 0.357 si se ignora el emparejamiento) |
| **Supuestos** | Independencia (diseño), normalidad aproximada, varianzas (Welch no las exige). No elegir la prueba según una pre-prueba de Levene o Shapiro |
| **Una o dos colas** | Bilateral por defecto; unilateral solo si se decide **antes** de ver los datos (con $p=0.0504$ bilateral frente a 0.0252 unilateral la decisión **cambia**) |
| **Permutación** | Barajar etiquetas reproduce la definición del valor p sin suponer normalidad ($p=0.016$ frente a Welch 0.013) |
| **Potencia** | Exacta con la $t$ no central (= statsmodels); $d=0.5$ con 30 por grupo: **48 %**. Para 80 %: 64 por grupo si $d=0.5$; **394** si $d=0.2$ |
| **Tamaño del efecto** | $d$ de Cohen / $g$ de Hedges **con su IC**: significancia ≠ relevancia ($n=20\,000$: $p\approx10^{-9}$, $d=0.06$). Con poca potencia, los «éxitos» **exageran** el efecto (7.7 frente a 4 reales) |
| **Pruebas múltiples** | $\mathbb{P}(\ge1\text{ falso positivo})=1-0.95^{20}\approx0.64$. **Bonferroni/Holm** (FWER) y **BH** (FDR): evitan falsos positivos a cambio de potencia |
