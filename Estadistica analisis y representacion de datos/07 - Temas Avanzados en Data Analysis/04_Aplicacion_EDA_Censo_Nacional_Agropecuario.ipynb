## 1. Un Censo Agropecuario con Estructura Real y Datos Sintéticos 🌾

**Censo frente a muestra.** En una **muestra** se estima un parámetro con incertidumbre de muestreo; en un **censo** se intenta observar **toda** la población, así que los estadísticos descriptivos **son** (casi) los parámetros: no hay error de muestreo. Eso **no** vuelve perfectos los datos: persisten los **errores ajenos al muestreo** (cobertura, no respuesta, medición, digitación). Por eso, los intervalos de confianza de un censo solo tienen sentido como medida de **sensibilidad** (¿cuánto cambiaría el resultado si los datos fueran una realización de un proceso?) o al comparar periodos, y hay que decirlo. En la tabla usamos IC *bootstrap* con esa lectura.

**Estructura.** La unidad de observación es la **unidad de producción agropecuaria (UPA)** *(término usado por el DANE, citado de memoria, verificar)*; la jerarquía es departamento $\to$ municipio $\to$ UPA, y suele haber **varias tablas** (UPA, cultivos, lotes, personas) unidas por llaves. Nuestro archivo sintético tiene **una fila por UPA**:

| Grupo | Variables (sintéticas) | Tipo | Defectos que inyectamos |
|---|---|---|---|
| Ubicación | `departamento`, `municipio` | categórica | mayúsculas y escritura distinta |
| Tamaño y uso | `area_total_ha`, `area_sembrada_ha`, `area_pastos_ha`, `area_bosque_ha` | numérica (ha) | sembrada $>$ total (error $\times10$), pastos $>$ total, faltantes **MAR** |
| Producción | `cultivo_principal`, `riego`, `maquinaria`, `ventas_mm` | categórica / numérica | ventas con 6 % de faltantes y cola muy pesada |
| Tenencia y crédito | `tenencia`, `credito`, `monto_credito_mm`, `asistencia_tecnica` | categórica / numérica | tildes y espacios; monto **vacío por diseño** si no hay crédito |
| Productor | `edad_productor`, `sexo_productor` | numérica / categórica | centinelas 999 y 0; 1 % sin dato |
| Control | `id_upa` | entero | 40 filas repetidas |

El mecanismo que generó los datos **es inventado**: el acceso al crédito crece con el tamaño de la finca, con tener título y con recibir asistencia técnica; los faltantes de área sembrada son más frecuentes en fincas pequeñas sin asistencia. Conservamos la **verdad** (el área sembrada limpia) aparte, solo para evaluar nuestras decisiones. El código siguiente genera el archivo «sucio» y muestra su aspecto inicial.

````python
import unicodedata
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

pd.set_option("display.width", 150); pd.set_option("display.max_columns", 30)
rng = np.random.default_rng(1964)

# ============ Generador de un "censo agropecuario" SINTÉTICO (NO son datos del DANE) ============
deptos = ["Boyacá", "Cundinamarca", "Antioquia", "Nariño", "Santander", "Tolima", "Huila", "Meta"]
p_depto = [0.17, 0.15, 0.16, 0.14, 0.11, 0.09, 0.09, 0.09]
sigma_area = dict(zip(deptos, [1.1, 1.2, 1.3, 0.9, 1.2, 1.3, 1.2, 1.7]))     # concentración de la tierra INVENTADA por departamento
mu_area = dict(zip(deptos, [0.9, 1.0, 1.2, 0.5, 1.1, 1.2, 1.1, 2.3]))
cultivos = ["papa", "café", "maíz", "plátano", "caña panelera", "frijol", "pastos/ganadería", "hortalizas"]
w_cult = {"Boyacá": [.30, .02, .12, .02, .03, .08, .35, .08], "Cundinamarca": [.18, .10, .10, .06, .08, .06, .30, .12],
          "Antioquia": [.05, .30, .08, .12, .05, .05, .30, .05], "Nariño": [.30, .12, .12, .05, .04, .10, .15, .12],
          "Santander": [.05, .22, .12, .06, .15, .10, .25, .05], "Tolima": [.03, .25, .12, .10, .05, .10, .30, .05],
          "Huila": [.02, .35, .10, .10, .04, .12, .22, .05], "Meta": [.01, .03, .13, .12, .02, .04, .60, .05]}
N = 6000
depto = rng.choice(deptos, N, p=p_depto)
municipio = np.array([f"{d[:3].upper()}-M{rng.integers(1, 6)}" for d in depto])
area_total = np.maximum(rng.lognormal([mu_area[d] for d in depto], [sigma_area[d] for d in depto]), 0.05).round(2)
cultivo = np.array([rng.choice(cultivos, p=w_cult[d]) for d in depto])
# Uso del suelo (ha): sembrada + pastos + bosque + otros <= total
frac = rng.dirichlet([2, 1.5, 1, 0.7], N)
frac[cultivo == "pastos/ganadería"] = rng.dirichlet([0.6, 4, 1, 0.7], (cultivo == "pastos/ganadería").sum())
a_sem, a_pas, a_bos = (area_total * frac[:, k] for k in range(3))
# Tenencia (las fincas grandes tienden más a tener título: mecanismo INVENTADO)
p_titulo = 1 / (1 + np.exp(-(-0.4 + 0.55 * np.log(area_total))))
tenencia = np.where(rng.random(N) < p_titulo, "propia con título",
                    rng.choice(["propia sin título", "arriendo", "aparcería", "comodato", "ocupación de hecho"], N, p=[.38, .27, .13, .12, .10]))
asistencia = rng.random(N) < 0.28
riego = rng.random(N) < 1 / (1 + np.exp(-(-1.6 + 0.35 * np.log(area_total) + 0.5 * np.isin(cultivo, ["hortalizas", "caña panelera"]))))
maquinaria = rng.random(N) < 1 / (1 + np.exp(-(-2.2 + 0.7 * np.log(area_total))))
credito = rng.random(N) < 1 / (1 + np.exp(-(-2.4 + 0.5 * np.log(area_total) + 0.9 * (tenencia == "propia con título") + 0.7 * asistencia)))
monto = np.where(credito, rng.lognormal(2.6 + 0.3 * np.log(area_total), 0.7), np.nan).round(1)
edad = np.clip(rng.normal(53, 13, N), 18, 90).round(0)
mujer = rng.random(N) < 0.30
ventas = rng.lognormal(0.9 + 0.5 * np.log(a_sem + 0.1) + 0.4 * maquinaria + 0.3 * riego + 0.2 * asistencia, 0.9).round(2)   # millones de pesos, cola pesada

limpio = pd.DataFrame({"id_upa": np.arange(1, N + 1), "departamento": depto, "municipio": municipio, "area_total_ha": area_total,
                       "area_sembrada_ha": a_sem.round(2), "area_pastos_ha": a_pas.round(2), "area_bosque_ha": a_bos.round(2),
                       "cultivo_principal": cultivo, "tenencia": tenencia, "riego": np.where(riego, "Sí", "No"),
                       "maquinaria": np.where(maquinaria, "Sí", "No"), "credito": np.where(credito, "Sí", "No"), "monto_credito_mm": monto,
                       "asistencia_tecnica": np.where(asistencia, "Sí", "No"), "edad_productor": edad, "sexo_productor": np.where(mujer, "Mujer", "Hombre"),
                       "ventas_mm": ventas})
verdad_sembrada = limpio.set_index("id_upa")["area_sembrada_ha"].copy()              # guardamos la VERDAD para poder evaluar la limpieza

# ============ Ensuciar el archivo con defectos plausibles ============
cna = limpio.copy()
i = rng.choice(N, 120, replace=False); cna.loc[i, "departamento"] = cna.loc[i, "departamento"].str.upper()                 # MAYÚSCULAS
i = rng.choice(N, 90, replace=False); cna.loc[i, "tenencia"] = cna.loc[i, "tenencia"].str.upper() + " "                    # mayúsculas y espacios
i = rng.choice(N, 70, replace=False); cna.loc[i, "tenencia"] = cna.loc[i, "tenencia"].str.replace("í", "i").str.replace("é", "e")   # tildes perdidas
p_falta = 0.04 + 0.16 * (~asistencia) * (area_total < 2)                              # faltantes MAR: más en fincas pequeñas sin asistencia
cna.loc[rng.random(N) < p_falta, "area_sembrada_ha"] = np.nan
cna.loc[rng.random(N) < 0.06, "ventas_mm"] = np.nan
cna.loc[rng.random(N) < 0.01, "sexo_productor"] = np.nan
i = rng.choice(N, 45, replace=False); cna.loc[i, "edad_productor"] = 999                                                   # centinela de "no sabe"
i = rng.choice(N, 15, replace=False); cna.loc[i, "edad_productor"] = 0
i = rng.choice(N, 90, replace=False); cna.loc[i, "area_sembrada_ha"] = cna.loc[i, "area_total_ha"] * 10                    # error de unidades/digitación (x10)
i = rng.choice(N, 60, replace=False); cna.loc[i, "area_pastos_ha"] = cna.loc[i, "area_total_ha"] * 1.5                     # pastos mayores que la finca
cna = pd.concat([cna, cna.sample(40, random_state=3)], ignore_index=True)                                                  # 40 filas repetidas
print(f"Archivo sintético: {cna.shape[0]} filas x {cna.shape[1]} columnas (6 000 UPA + 40 repetidas)")
print(cna.head(4).to_string())
print(cna.dtypes.value_counts().to_dict())
````

**Lectura.** El archivo sintético tiene **6 040 filas y 17 columnas** (6 000 UPA más 40 repetidas); 9 columnas son texto (`object`), 7 decimales y 1 entera. A simple vista **no se ven los defectos**: están en las columnas que no se miran, en los extremos y en la escritura de las categorías. Un EDA de verdad **empieza por perfilar**, no por graficar; es lo que hacemos a continuación.

## 2. Perfilar, Normalizar y Validar con Reglas 🔍

La limpieza honesta sigue cuatro principios: (1) **perfilar antes de tocar**; (2) **normalizar** sin perder información (minúsculas, espacios, tildes); (3) escribir las **reglas de consistencia como código** (cada regla produce una máscara booleana y un conteo) y **marcar** en vez de borrar; (4) llevar una **bitácora** de cada decisión.

**Faltantes: estructural, MCAR, MAR.** Si `monto_credito_mm` está vacío cuando `credito = no`, el dato **no se perdió**: no existe (falta **estructural**); imputarlo sería un error conceptual. Si el área sembrada falta **más** en fincas pequeñas sin asistencia, la falta es **MAR** (depende de variables observadas) y los **casos completos** estarán sesgados. Como conocemos la verdad, comparamos la media del área sembrada en cuatro escenarios: la **verdad**, el archivo **sin limpiar**, los **casos completos** y la **imputación** (razón mediana área sembrada/área total por cultivo, multiplicada por el área total). La prueba de independencia $\chi^2$ cuantifica la asociación entre «finca pequeña sin asistencia» y «dato faltante».

````python
# ---------- 1) Perfilado: qué hay en cada columna ----------
def perfil(df):
    return pd.DataFrame({"tipo": df.dtypes.astype(str), "faltante_%": (df.isna().mean() * 100).round(1),
                         "únicos": df.nunique(), "ejemplo": df.iloc[0]})
print(perfil(cna).iloc[[1, 4, 8, 12, 14, 15]].to_string())

# ---------- 2) Normalizar texto y quitar duplicados exactos ----------
def norm(s):
    s = unicodedata.normalize("NFKD", str(s).strip().lower())
    return "".join(c for c in s if not unicodedata.combining(c))
antes = {c: cna[c].nunique() for c in ["departamento", "tenencia"]}
bitacora = {"filas originales": len(cna), "duplicados exactos": int(cna.duplicated().sum())}
df = cna.drop_duplicates().copy()
for c in ["departamento", "tenencia", "cultivo_principal", "sexo_productor", "riego", "maquinaria", "credito", "asistencia_tecnica"]:
    df[c] = df[c].map(lambda v: v if pd.isna(v) else norm(v))
despues = {c: df[c].nunique() for c in ["departamento", "tenencia"]}
print("categorías distintas antes -> después:", {c: (antes[c], despues[c]) for c in antes})
assert df["id_upa"].is_unique and despues == {"departamento": 8, "tenencia": 6}

# ---------- 3) Reglas de consistencia: se MARCAN, no se borran en silencio ----------
reglas = {"sembrada > total": df["area_sembrada_ha"] > df["area_total_ha"] * 1.001,
          "usos suman más que el total": (df[["area_sembrada_ha", "area_pastos_ha", "area_bosque_ha"]].sum(axis=1, min_count=1) > df["area_total_ha"] * 1.001),
          "edad centinela o imposible": ~df["edad_productor"].between(15, 100),
          "monto de crédito sin crédito": df["monto_credito_mm"].notna() & (df["credito"] == "no"),
          "crédito sin monto": df["monto_credito_mm"].isna() & (df["credito"] == "si")}
viol = pd.DataFrame({"n": {k: int(v.sum()) for k, v in reglas.items()}})
viol["%"] = (viol["n"] / len(df) * 100).round(2)
print(viol)
bitacora["filas únicas"] = len(df)
df["flag_area"] = reglas["sembrada > total"] | reglas["usos suman más que el total"]
df.loc[reglas["edad centinela o imposible"], "edad_productor"] = np.nan          # centinela -> NaN (y queda en la bitácora)
df["sembrada_limpia"] = df["area_sembrada_ha"].where(~reglas["sembrada > total"])

# ---------- 4) Faltantes: estructural frente a MAR ----------
falta = df["area_sembrada_ha"].isna()
df["pequena"] = np.where(df["area_total_ha"] < 2, "< 2 ha", ">= 2 ha")
tab = pd.crosstab([df["pequena"], df["asistencia_tecnica"]], falta, normalize="index")[True].mul(100).round(1)
print("% sin área sembrada según tamaño y asistencia:\n", tab.to_string())
chi2, p, _, _ = stats.chi2_contingency(pd.crosstab(df["pequena"].eq("< 2 ha") & df["asistencia_tecnica"].eq("no"), falta))
print(f"chi² de independencia (finca pequeña sin asistencia vs dato faltante): chi² = {chi2:.1f}, p = {p:.2e}")
estr = df.loc[df["credito"] == "no", "monto_credito_mm"].isna().all()
print("monto_credito_mm vacío en TODAS las UPA sin crédito (falta ESTRUCTURAL, no es un dato perdido):", estr)

# ---------- 5) Impacto en la media del área sembrada (conocemos la verdad) ----------
verd = verdad_sembrada.loc[df["id_upa"]].to_numpy()
valida = df["sembrada_limpia"].notna()
cc = df.loc[valida, "sembrada_limpia"].mean()
ratio = (df["sembrada_limpia"] / df["area_total_ha"]).groupby(df["cultivo_principal"]).transform("median")
imput = df["sembrada_limpia"].fillna(ratio * df["area_total_ha"])
print(f"media del área sembrada (ha): verdad = {verd.mean():.3f} | sin limpiar = {df['area_sembrada_ha'].mean():.3f} | "
      f"casos completos = {cc:.3f} | imputada = {imput.mean():.3f}")
assert abs(imput.mean() - verd.mean()) < abs(cc - verd.mean()) < abs(df["area_sembrada_ha"].mean() - verd.mean())
df["area_sembrada_imp"] = imput
````

---
**Lectura.** (1) **Perfil:** `departamento` tiene **16** valores distintos (8 reales más sus variantes en mayúsculas) y `tenencia` **15** (6 reales); `monto_credito_mm` tiene **72.6 %** de vacíos y `area_sembrada_ha` **8.8 %**. Tras normalizar quedan **8** y **6** categorías y **6 000** UPA únicas (`id_upa` ya es llave). (2) **Reglas:** **90 UPA (1.50 %)** declaran más área sembrada que área total (el error $\times10$ que inyectamos); **160 (2.67 %)** tienen usos que suman más que el total (incluye las 90 anteriores y las de pastos); **60 (1.00 %)** tienen edad 999 o 0 (centinelas). ## 3. Concentración de la Tierra: Lorenz y Gini 📉

En un censo agropecuario la pregunta de política suele ser **la distribución de la tierra**. La **curva de Lorenz** grafica la fracción acumulada del área $L(p)$ frente a la fracción acumulada de UPA $p$ (de la más pequeña a la más grande); con igualdad perfecta es la diagonal. El **coeficiente de Gini** es dos veces el área entre la diagonal y la curva. Con los datos ordenados $x_{(1)}\le\cdots\le x_{(n)}$:
$$G=\frac{2\sum_{i=1}^{n} i\,x_{(i)}}{n\sum_{i=1}^{n}x_{(i)}}-\frac{n+1}{n}=\frac{\sum_{i}\sum_{j}\lvert x_i-x_j\rvert}{2n^2\bar x}.$$
Va de 0 (todas las UPA iguales) a casi 1 (una sola UPA tiene todo). La segunda forma lo interpreta como la **diferencia media relativa** entre dos UPA tomadas al azar. Implementamos la primera, la contrastamos con la segunda (por *broadcasting*, en una submuestra) y calculamos el Gini por departamento con IC *bootstrap* percentil. Para una distribución **lognormal** con parámetro de escala $\sigma$, el Gini teórico es $2\Phi(\sigma/\sqrt2)-1$, con $\Phi$ la función de distribución normal estándar; como inventamos $\sigma$ distinto por departamento, la comparación es posible. **Ojo:** el Gini no distingue *dónde* ocurre la desigualdad (dos curvas distintas pueden dar el mismo Gini) y el Gini **nacional** mezcla la desigualdad **dentro** de cada departamento con las diferencias **entre** ellos.

````python
def gini(x):
    x = np.sort(np.asarray(x, float)); n = len(x); i = np.arange(1, n + 1)
    return 2 * (i * x).sum() / (n * x.sum()) - (n + 1) / n          # fórmula con datos ordenados

# Verificación: coincide con la definición por diferencias medias G = sum|xi-xj| / (2 n² media)
sub = df["area_total_ha"].sample(1500, random_state=0).to_numpy()
G_dm = np.abs(sub[:, None] - sub[None, :]).sum() / (2 * len(sub) ** 2 * sub.mean())
assert np.isclose(gini(sub), G_dm)
assert np.isclose(gini(np.ones(50)), 0) and gini(np.r_[np.zeros(999), 1.0]) > 0.99          # igualdad perfecta y concentración extrema

at = df["area_total_ha"].to_numpy()
orden = np.sort(at)
print(f"UPA: {len(at)} | área total mediana = {np.median(at):.2f} ha, media = {at.mean():.2f} ha | Gini nacional (sintético) = {gini(at):.3f}")
print(f"el 10 % de las UPA más grandes concentra el {orden[-len(at) // 10:].sum() / at.sum():.1%} del área; el 50 % más pequeño, el {orden[:len(at) // 2].sum() / at.sum():.1%}")

# Gini por departamento con IC bootstrap percentil (B = 500 remuestras por departamento)
rng_b = np.random.default_rng(11)
filas = []
for d, g in df.groupby("departamento"):
    x = g["area_total_ha"].to_numpy()
    bs = [gini(rng_b.choice(x, len(x))) for _ in range(500)]
    filas.append((d, len(x), gini(x), *np.percentile(bs, [2.5, 97.5])))
gd = pd.DataFrame(filas, columns=["departamento", "n", "gini", "ic_inf", "ic_sup"]).set_index("departamento").sort_values("gini")
print(gd.round(3))

fig, ax = plt.subplots(1, 3, figsize=(17, 4.2))
ax[2].hist(np.log10(at), bins=40, color="#93c5fd", edgecolor="white"); ax[2].set_xlabel("log10(área total en ha)"); ax[2].set_ylabel("UPA"); ax[2].set_title("Área total en escala logarítmica")
for d in ["narino", "boyaca", "meta"]:
    x = np.sort(df.loc[df["departamento"] == d, "area_total_ha"].to_numpy())
    ax[0].plot(np.r_[0, np.arange(1, len(x) + 1) / len(x)], np.r_[0, np.cumsum(x) / x.sum()], label=f"{d} (G = {gini(x):.2f})")
ax[0].plot([0, 1], [0, 1], "--", color="grey", label="igualdad")
ax[0].set_xlabel("fracción acumulada de UPA (de la más pequeña a la más grande)"); ax[0].set_ylabel("fracción acumulada del área"); ax[0].set_title("Curvas de Lorenz"); ax[0].legend()
ax[1].errorbar(gd["gini"], np.arange(len(gd)), xerr=[gd["gini"] - gd["ic_inf"], gd["ic_sup"] - gd["gini"]], fmt="o", color="#1e3a8a", capsize=3)
ax[1].set_yticks(np.arange(len(gd)), gd.index); ax[1].set_xlabel("Gini del área total"); ax[1].set_title("Gini por departamento (IC bootstrap 95 %)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) La fórmula con datos ordenados **coincide** con la de diferencias medias (`assert`), vale 0 con igualdad perfecta y supera 0.99 con concentración extrema. (2) En el archivo sintético el **área total mediana es 2.79 ha y la media 8.43 ha** (una media 3 veces mayor que la mediana: cola derecha). **El 10 % de las UPA más grandes concentra el 61.9 % del área y el 50 % más pequeño solo el 7.7 %**; el Gini «nacional» es **0.716**. (3) **Por departamento:** de **0.452** (Nariño, el más igualitario: $\sigma=0.9$) a **0.738** (Meta, el más concentrado: $\sigma=1.7$), con IC estrechos para $n\approx$ 500-1 000 por departamento; el IC de Santander (0.570-0.702) es el más ancho aunque tiene más UPA que Meta (656 frente a 543): probablemente unas pocas UPA muy grandes mueven su Gini, que es muy sensible a la cola derecha. Con $\sigma=0.9$ y $1.7$ el Gini teórico es 0.475 y 0.771: las estimaciones quedan algo por debajo (el error estándar muestral es ≈ 0.012 y 0.029, así que están a ≈ 2 y 1 errores estándar), y el *bootstrap* percentil tiende a subestimar en colas pesadas, por lo que **conviene no leer los IC como garantía exacta**. (4) El histograma en escala logarítmica es casi simétrico: la variable es aproximadamente **lognormal**, por lo que **la media es un mal resumen** y los métodos para atípicos deberán trabajar en escala log (cuaderno 05). Recuerda además que, como es un censo, estos Gini **describen** la población (sintética) y los IC miden la sensibilidad de ese resumen.

## 4. Asociaciones Categóricas con Intervalos, y el Efecto de un Confusor 🔗

**Independencia.** Para una tabla $r\times c$ de frecuencias observadas $O_{ij}$, la frecuencia esperada bajo independencia es $E_{ij}=\dfrac{(\text{total fila}_i)(\text{total col}_j)}{N}$ y $\chi^2=\sum_{ij}\dfrac{(O_{ij}-E_{ij})^2}{E_{ij}}$ con $(r-1)(c-1)$ grados de libertad. Un valor $p$ diminuto solo dice que **no** es independencia; el **tamaño de la asociación** lo da la **V de Cramér**: $V=\sqrt{\chi^2/\big(N\min(r-1,c-1)\big)}\in[0,1]$. En un conjunto grande, cualquier asociación mínima es «significativa»: por eso se reporta $V$.

**Proporciones con intervalo de Wilson.** Para $k$ éxitos en $n$ ensayos, $\hat p=k/n$ y el IC de Wilson es $\dfrac{\hat p+\frac{z^2}{2n}\pm z\sqrt{\frac{\hat p(1-\hat p)}{n}+\frac{z^2}{4n^2}}}{1+\frac{z^2}{n}}$ con $z=1.96$ para el 95 %; funciona mucho mejor que el intervalo de Wald cuando $n$ es pequeño o $\hat p$ está cerca de 0 o 1. Lo implementamos y lo contrastamos con `statsmodels`.

**Cruda frente a ajustada.** ¿Tener título aumenta el acceso al crédito? Como las fincas grandes tienen más título **y** más crédito, el tamaño es un **confusor**: la razón de momios (OR) **cruda** exagera. La regresión logística **ajustada** por `log(área)` y asistencia técnica separa los efectos. Recuerda que, aun ajustada, es una **asociación**, no necesariamente un efecto causal.

````python
from statsmodels.stats.proportion import proportion_confint
import statsmodels.formula.api as smf
from scipy.stats.contingency import association

df["titulo"] = (df["tenencia"] == "propia con titulo").astype(int)
df["cred"] = (df["credito"] == "si").astype(int)
df["asist"] = (df["asistencia_tecnica"] == "si").astype(int)
df["log_area"] = np.log(df["area_total_ha"])

# ---------- 1) Tabla de contingencia, chi² y V de Cramér desde cero ----------
ct = pd.crosstab(df["tenencia"], df["cred"])
obs = ct.to_numpy().astype(float)
esp = obs.sum(1, keepdims=True) * obs.sum(0, keepdims=True) / obs.sum()
chi2_np = ((obs - esp) ** 2 / esp).sum()
chi2_sp, p_sp, dof, _ = stats.chi2_contingency(ct, correction=False)
V = np.sqrt(chi2_np / (obs.sum() * (min(obs.shape) - 1)))
assert np.isclose(chi2_np, chi2_sp) and np.isclose(V, association(ct, method="cramer", correction=False))
print(f"tenencia x crédito: chi² = {chi2_np:.1f} ({dof} g.l.), p = {p_sp:.2e}, V de Cramér = {V:.3f} (esperado mínimo por celda: {esp.min():.1f})")

# ---------- 2) Proporción con crédito por tenencia con IC de Wilson (desde cero y contra statsmodels) ----------
def wilson(k, n, z=1.959964):
    p = k / n; c = (p + z ** 2 / (2 * n)) / (1 + z ** 2 / n); h = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2)) / (1 + z ** 2 / n)
    return c - h, c + h
g = df.groupby("tenencia")["cred"].agg(k="sum", n="size")
lo, hi = wilson(g["k"].to_numpy(), g["n"].to_numpy())
lo_sm, hi_sm = proportion_confint(g["k"], g["n"], method="wilson")
assert np.allclose(lo, lo_sm) and np.allclose(hi, hi_sm)
g["p"], g["lo"], g["hi"] = g["k"] / g["n"], lo, hi
g = g.sort_values("p")
print(g.round(3).to_string())

# ---------- 3) Asociación cruda frente a ajustada: el tamaño de la finca confunde ----------
cruda = smf.logit("cred ~ titulo", data=df).fit(disp=0)
ajust = smf.logit("cred ~ titulo + log_area + asist", data=df).fit(disp=0)
OR = lambda m, k: (np.exp(m.params[k]), *np.exp(m.conf_int().loc[k]))
print("OR del título -> crédito:  cruda = %.2f [%.2f, %.2f] | ajustada por tamaño y asistencia = %.2f [%.2f, %.2f]" % (*OR(cruda, "titulo"), *OR(ajust, "titulo")))
print("OR de la asistencia técnica (ajustada) = %.2f [%.2f, %.2f]" % OR(ajust, "asist"))
assert OR(cruda, "titulo")[0] > OR(ajust, "titulo")[0]

plt.figure(figsize=(7.5, 3.8))
plt.errorbar(g["p"] * 100, np.arange(len(g)), xerr=[(g["p"] - g["lo"]) * 100, (g["hi"] - g["p"]) * 100], fmt="o", color="#1e3a8a", capsize=3)
plt.yticks(np.arange(len(g)), g.index); plt.xlabel("% de UPA con crédito (IC de Wilson 95 %)"); plt.title("Acceso a crédito según tenencia (datos sintéticos)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **Tenencia $\times$ crédito:** $\chi^2=419.1$ con 5 g.l., $p\approx2\times10^{-88}$ y $V=0.264$ (asociación **moderada**; el mínimo esperado por celda es 69.8, sin problemas para la aproximación). El cálculo desde cero y el de `scipy` coinciden (`assert`). (2) **Proporciones con Wilson:** las UPA con **título** acceden al crédito en **38.0 %** (IC [36.3, 39.6]); el resto de las formas de tenencia, entre **10.2 %** (ocupación de hecho, [7.0, 14.5]) y **18.3 %** (comodato, [14.4, 22.9]); los intervalos de las categorías pequeñas (256 a 349 UPA) son claramente más anchos y **se solapan** entre sí: no hay evidencia de diferencia entre arriendo, sin título, aparcería y comodato, pero sí entre todas ellas y «con título». (3) **El confusor importa:** el OR crudo del título es **3.60** (IC [3.17, 4.10]) y el **ajustado por tamaño y asistencia, 2.70** ([2.35, 3.09]); el valor verdadero inventado es $e^{0.9}\approx2.46$, que queda dentro del IC ajustado pero **fuera** del crudo. La asistencia técnica ajustada da un OR de **2.03** ([1.78, 2.32]; verdadero $e^{0.7}\approx2.01$). Si hubiéramos reportado solo la tabla cruzada, habríamos reportado una razón de momios **≈ 33 % mayor** que la ajustada. (4) Con **6 000** UPA, intervalos y $V$ son más informativos que los valores $p$.

## 5. Del Dato Sintético al Censo Real: Guía de Adaptación 🛠️

Cuando tengas el archivo real, el flujo de este cuaderno **no cambia**; lo que cambia es la **capa de entrada**. Lista de pasos:

1. **Ubicar la fuente.** Entra al portal del **DANE** <https://www.dane.gov.co> *(verificar)* o a **Datos Abiertos Colombia** <https://www.datos.gov.co> *(verificar)* y busca el **Censo Nacional Agropecuario**; descarga **junto con los datos** la **ficha metodológica** y el **diccionario de variables**. No usamos enlaces a archivos concretos porque las rutas cambian. Anota **fecha de descarga, versión y licencia**.
2. **Leer los metadatos antes que los datos:** unidad de observación, periodo de referencia, universo, **tablas y llaves** (puede haber una tabla por módulo), códigos de **no sabe / no aplica** y unidades (¿hectáreas, metros cuadrados, fanegadas?).
3. **Leer sin dañar:** `pd.read_csv(ruta, sep=";", encoding="latin-1", dtype=str)` es un punto de partida típico, pero el **separador, la codificación y el decimal dependen del archivo** *(verificar en la documentación)*. Leer todo como texto (`dtype=str`) evita perder ceros a la izquierda en **códigos de departamento y municipio**; después se convierte cada columna con `pd.to_numeric(..., errors="coerce")`.
4. **Mapear nombres** del archivo real a los de nuestro esquema con un diccionario y **validar** que no falte ninguna columna (la función `adaptar_censo` de abajo; los nombres `P_*` son **inventados** para ilustrar).
5. **Unir tablas** con `merge(..., validate="m:1")` (cuaderno 03) y repetir el **perfilado**, las **reglas** de consistencia y la **bitácora**.
6. **Interpretar como censo:** sin factores de expansión *(los censos no suelen requerirlos; si la fuente fuera una encuesta, sí; verificar en la ficha)*, con los IC como medidas de sensibilidad y con atención a **errores ajenos al muestreo**.
7. **Ética y privacidad:** los microdatos publicados suelen estar **anonimizados**; **no** intentes reidentificar productores (cuaderno 03 del Módulo 04) y respeta la licencia. **Nunca** compares con cifras sintéticas.

El código siguiente **simula** el archivo descargado (separador `;`, coma decimal, todo texto y otros nombres) y lo adapta; el validador falla con un mensaje útil si falta una columna.

````python
import io

# --- Ejemplo de ADAPTACIÓN: un archivo "crudo" con otros nombres, separador ';', coma decimal y todo como texto ---
mapa_inventado = {"P_DEPTO": "departamento", "P_AREA_TOT": "area_total_ha", "P_AREA_SEMB": "area_sembrada_ha", "P_CULT": "cultivo_principal",
                  "P_TENENCIA": "tenencia", "P_CREDITO": "credito", "P_ID": "id_upa"}              # nombres INVENTADOS para ilustrar el mapeo
inverso = {v: k for k, v in mapa_inventado.items()}
crudo_txt = cna[list(inverso)].rename(columns=inverso).to_csv(sep=";", decimal=",", index=False)      # lo que sería el archivo descargado
crudo = pd.read_csv(io.StringIO(crudo_txt), sep=";", dtype=str)                                      # dtype=str: no perder ceros ni formatos
print(crudo.head(3).to_string(), "\n")

ESQUEMA = {"id_upa": "int", "departamento": "cat", "area_total_ha": "num", "area_sembrada_ha": "num",
           "cultivo_principal": "cat", "tenencia": "cat", "credito": "cat"}

def adaptar_censo(crudo, mapa, esquema):
    d = crudo.rename(columns=mapa)
    faltan = set(esquema) - set(d.columns)
    assert not faltan, f"columnas que no están en el archivo: {sorted(faltan)}"
    d = d[list(esquema)].copy()
    for c, t in esquema.items():
        if t == "num":
            d[c] = pd.to_numeric(d[c].str.replace(",", ".", regex=False), errors="coerce")           # coma decimal -> punto; basura -> NaN
        elif t == "int":
            d[c] = pd.to_numeric(d[c], errors="raise").astype(int)
        else:
            d[c] = d[c].map(lambda v: v if pd.isna(v) else norm(v))
    return d

adaptado = adaptar_censo(crudo, mapa_inventado, ESQUEMA)
assert np.allclose(adaptado["area_total_ha"], cna["area_total_ha"]) and adaptado["area_sembrada_ha"].isna().sum() == cna["area_sembrada_ha"].isna().sum()
print(adaptado.dtypes.to_dict())
try:
    adaptar_censo(crudo.drop(columns="P_CREDITO"), mapa_inventado, ESQUEMA)
except AssertionError as e:
    print("El validador protege:", e)
````
