## 1. Componentes de una Serie y Descomposición Clásica 🧩

Casi toda serie económica o agropecuaria mezcla **cuatro ingredientes**:

| Componente | Qué es | Ejemplo (leche acopiada) |
|---|---|---|
| **Tendencia** $T_t$ | Dirección de largo plazo | Crece porque entran más productores |
| **Estacionalidad** $S_t$ | Patrón que se repite con **periodo fijo** $m$ (12 meses) | Más leche cuando hay pasto abundante |
| **Ciclo** | Oscilaciones **sin periodo fijo** (varios años) | Ciclos de precios o de clima |
| **Ruido / residuo** $R_t$ | Lo que queda: fluctuación aleatoria | Variaciones del día a día |

El **modelo aditivo** supone $y_t = T_t + S_t + R_t$ (la amplitud estacional no cambia con el nivel); el **multiplicativo**, $y_t = T_t \times S_t \times R_t$ (la amplitud crece con el nivel; se vuelve aditivo tomando logaritmos).

**Descomposición clásica (aditiva), paso a paso.** (1) Estima la tendencia con una **media móvil centrada** de orden $m$; con $m$ par se usa la «$2\times m$»: pesos $\tfrac{1}{2m},\tfrac{1}{m},\dots,\tfrac{1}{m},\tfrac{1}{2m}$ (así queda centrada en un mes). (2) Resta: $y_t - \hat{T}_t$. (3) El **factor estacional** de cada mes es el promedio de esos valores para ese mes, **centrado** para que sume 0. (4) El residuo es $\hat{R}_t = y_t - \hat{T}_t - \hat{S}_t$. Se pierden $m/2$ valores de tendencia en cada extremo.

Generamos una serie **sintética** de 144 meses con tendencia lineal ($500 + 2t$), un patrón estacional de periodo 12 y ruido **AR(1)** con $\phi=0.6$ (ruido con memoria, que será importante después). Como **conocemos** los componentes verdaderos, podemos medir cuánto se aproxima la descomposición.

````python
# ============================================================
# Configuracion inicial, datos sinteticos y descomposicion clasica
# ============================================================
# !pip install -q numpy pandas scipy statsmodels scikit-learn matplotlib

import os
import warnings

for v in ("OMP_NUM_THREADS", "OPENBLAS_NUM_THREADS", "MKL_NUM_THREADS"):   # un hilo: las series son pequenas y asi se evita el exceso de hilos
    os.environ.setdefault(v, "1")

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({"figure.figsize": (10, 4.5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | statsmodels {__import__('statsmodels').__version__}")

# Serie SINTETICA: leche acopiada por mes (miles de litros) en un centro de acopio ficticio, 12 anios (144 meses)
n, m = 144, 12
t = np.arange(n)
idx = pd.date_range("2012-01-01", periods=n, freq="MS")
tendencia_real = 500 + 2.0 * t
estacional_real = 60 * np.sin(2 * np.pi * (t % m - 2) / m) + 15 * np.cos(4 * np.pi * (t % m) / m)
ruido = np.zeros(n)
eps = rng.normal(0, 12, n)
for i in range(1, n):                                    # ruido AR(1): phi = 0.6
    ruido[i] = 0.6 * ruido[i - 1] + eps[i]
leche = pd.Series(tendencia_real + estacional_real + ruido, index=idx, name="leche_miles_L")
print(leche.head(3).round(1).to_string(), "\n...\n", leche.tail(2).round(1).to_string())

# --- Descomposicion clasica ADITIVA desde cero: y = tendencia + estacional + residuo ---
def descomposicion_aditiva(y, m):
    y = np.asarray(y, float)
    w = np.r_[0.5, np.ones(m - 1), 0.5] / m               # media movil centrada 2 x m (m par)
    tend = np.full(len(y), np.nan)
    tend[m // 2: len(y) - m // 2] = np.convolve(y, w, mode="valid")
    sin_tend = y - tend
    fig = np.array([np.nanmean(sin_tend[k::m]) for k in range(m)])
    fig -= fig.mean()                                     # los factores estacionales suman 0
    est = np.tile(fig, len(y) // m + 1)[: len(y)]
    return tend, est, y - tend - est

tend, est, res = descomposicion_aditiva(leche.values, m)
ref = seasonal_decompose(leche, model="additive", period=m)
assert np.allclose(tend, ref.trend.values, equal_nan=True)
assert np.allclose(est, ref.seasonal.values) and np.allclose(res, ref.resid.values, equal_nan=True)
assert abs(est[:m].sum()) < 1e-9
ok = ~np.isnan(tend)
print(f"\nDescomposicion propia == statsmodels (tendencia, estacional y residuo).")
print(f"Error de la tendencia estimada vs la real (RMSE): {np.sqrt(np.mean((tend[ok] - tendencia_real[ok]) ** 2)):.2f} mil L")
print(f"Error del factor estacional vs el real (RMSE)  : {np.sqrt(np.mean((est - (estacional_real - estacional_real[:m].mean())) ** 2)):.2f} mil L")
print(f"Residuo: media {np.nanmean(res):.2f}, desv. est. {np.nanstd(res, ddof=1):.2f}  (ruido AR(1) generador: sd marginal = 12/sqrt(1-0.6^2) = {12 / np.sqrt(1 - 0.36):.1f})")

fig, ax = plt.subplots(4, 1, figsize=(10, 8.5), sharex=True)
for a, (nom, v, c) in zip(ax, [("Serie", leche.values, PALETA[0]), ("Tendencia (MM 2x12)", tend, PALETA[1]), ("Estacional", est, PALETA[2]), ("Residuo", res, PALETA[3])]):
    a.plot(idx, v, color=c, lw=1.4); a.set_ylabel(nom)
ax[0].set_title("Descomposicion aditiva clasica de la leche acopiada (datos sinteticos)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) La descomposición propia **coincide** con `seasonal_decompose` de `statsmodels` en tendencia, estacionalidad y residuo (los `assert` pasan). (2) Recupera bien lo que pusimos: la tendencia estimada se aparta de la real con un RMSE de **5.64 mil L** y el factor estacional con **3.28 mil L**, pequeños frente a una amplitud estacional de ±60 mil L. (3) **Cuidado con el residuo:** su desviación estándar es **9.78**, *menor* que la del ruido que generamos (**15.0**). No es magia: la media móvil de 12 meses **absorbe parte del ruido persistente** (el AR(1) tiene memoria y se confunde con tendencia), de modo que el residuo subestima el ruido real. Lección: **un componente estimado no es el componente verdadero**. (4) La descomposición **describe**, no pronostica: no dice cómo se prolongan la tendencia y la estacionalidad más allá de los datos. Para eso hacen falta modelos, y para los modelos hay que hablar de **estacionariedad**.

## 2. Estacionariedad y Autocorrelación: el Lenguaje de los Modelos 🔍

Una serie es **(débilmente) estacionaria** si su **media** y su **varianza** son constantes y la covarianza entre $y_t$ y $y_{t-k}$ depende **solo del rezago** $k$, no del instante $t$. Una tendencia o una estacionalidad **violan** la estacionariedad. Los modelos clásicos (AR, MA, ARMA) se definen para series estacionarias; el ARIMA los aplica a series no estacionarias **diferenciándolas** ($\Delta y_t = y_t - y_{t-1}$; con estacionalidad, $y_t - y_{t-m}$).

**Dos pruebas con hipótesis nulas opuestas** (por eso se usan juntas):

- **ADF** (*Augmented Dickey–Fuller*): ajusta $\Delta y_t = c + \gamma\, y_{t-1} + \sum_{i}\delta_i\,\Delta y_{t-i} + \varepsilon_t$ y contrasta $H_0:\gamma=0$ (**raíz unitaria**, no estacionaria) con el estadístico $t$ de $\hat\gamma$. Ese $t$ **no** sigue una $t$ de Student, sino la distribución de Dickey–Fuller (valores críticos de MacKinnon).
- **KPSS**: $H_0$: la serie es **estacionaria** (alrededor de una constante o de una tendencia). Con $e_t=y_t-\bar{y}$ y sumas parciales $S_t=\sum_{i\le t}e_i$, el estadístico es $\eta=\dfrac{\sum_t S_t^2}{N^2\,\hat\sigma^2_\ell}$, donde $\hat\sigma^2_\ell$ es la **varianza de largo plazo** (ponderación de Bartlett con $\ell$ rezagos).

| Resultado | ADF (rechaza raíz unitaria) | KPSS (rechaza estacionariedad) | Conclusión |
|---|---|---|---|
| 1 | Sí | No | **Estacionaria** |
| 2 | No | Sí | **Raíz unitaria**: diferenciar |
| 3 | No | No | **Inconcluso** (pocos datos / poca potencia) |
| 4 | Sí | Sí | Señales contradictorias (p. ej. cambio estructural) |

**ACF y PACF.** La **autocorrelación** de rezago $k$ es $\hat\rho_k=\dfrac{\sum_{t=k+1}^{N}(y_t-\bar y)(y_{t-k}-\bar y)}{\sum_{t=1}^{N}(y_t-\bar y)^2}$. La **autocorrelación parcial** $\phi_{kk}$ es la correlación con el rezago $k$ **después de descontar** los rezagos intermedios; se calcula con la recursión de **Durbin–Levinson**: $\phi_{kk}=\dfrac{\rho_k-\sum_{j=1}^{k-1}\phi_{k-1,j}\,\rho_{k-j}}{1-\sum_{j=1}^{k-1}\phi_{k-1,j}\,\rho_j}$. Bajo ruido blanco, el 95 % de las autocorrelaciones cae en $\pm 1.96/\sqrt{N}$. La **prueba de Ljung–Box** resume $h$ autocorrelaciones: $Q=N(N+2)\sum_{k=1}^{h}\dfrac{\hat\rho_k^2}{N-k}\sim\chi^2_{h-q}$ ($q$ = parámetros ARMA ajustados), con $H_0$: ruido blanco.

| Proceso | ACF | PACF |
|---|---|---|
| AR($p$) | Decae (exponencial u oscilante) | **Se corta** después del rezago $p$ |
| MA($q$) | **Se corta** después del rezago $q$ | Decae |
| ARMA($p,q$) | Decae | Decae |

Implementamos ADF (sin rezagos), KPSS, ACF, PACF y Ljung–Box **desde cero**, los contrastamos con `statsmodels` y aplicamos las pruebas a cuatro series sintéticas con «personalidades» distintas.

````python
# ============================================================
# 2. Estacionariedad (ADF, KPSS) y autocorrelacion (ACF, PACF, Ljung-Box)
# ============================================================
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.tsa.stattools import acf, adfuller, kpss, pacf

# --- ADF desde cero (sin rezagos, con constante): dy_t = c + g * y_{t-1} + e_t ; estadistico = t de g ---
def adf_stat(y):
    y = np.asarray(y, float)
    dy, x = np.diff(y), np.c_[np.ones(len(y) - 1), y[:-1]]
    b, *_ = np.linalg.lstsq(x, dy, rcond=None)
    e = dy - x @ b
    s2 = e @ e / (len(dy) - 2)
    return b[1] / np.sqrt(s2 * np.linalg.inv(x.T @ x)[1, 1])

# --- KPSS (nivel) desde cero: suma acumulada de residuos / varianza de largo plazo (Bartlett) ---
def kpss_stat(y, l):
    y = np.asarray(y, float); e = y - y.mean(); N = len(y)
    s2 = e @ e / N + 2 / N * sum((1 - k / (l + 1)) * (e[k:] @ e[:-k]) for k in range(1, l + 1))
    return np.sum(np.cumsum(e) ** 2) / (N ** 2 * s2)

# Cuatro series SINTETICAS de 200 puntos con "personalidades" distintas
N = 200
e = rng.normal(size=N)
ar1 = np.zeros(N); rw = np.zeros(N)
for i in range(1, N):
    ar1[i] = 0.6 * ar1[i - 1] + e[i]                      # estacionaria
    rw[i] = rw[i - 1] + e[i]                              # caminata aleatoria: raiz unitaria
ts = 0.05 * np.arange(N) + ar1                            # estacionaria alrededor de una tendencia determinista
series = {"AR(1) estacionaria": ar1, "Caminata aleatoria": rw, "Tendencia determinista + AR(1)": ts, "Leche (nivel)": leche.values}

for nombre, v in series.items():                          # nuestras funciones == statsmodels
    assert np.isclose(adf_stat(v), adfuller(v, maxlag=0, regression="c", autolag=None)[0])
    assert np.isclose(kpss_stat(v, 4), kpss(v, regression="c", nlags=4)[0])

filas = []
for nombre, v in series.items():
    a = adfuller(v, regression="c", autolag="AIC")        # H0: raiz unitaria
    k = kpss(v, regression="c", nlags="auto")             # H0: estacionaria (alrededor de una constante)
    kt = kpss(v, regression="ct", nlags="auto")           # H0: estacionaria alrededor de una tendencia
    filas.append({"serie": nombre, "ADF_p": a[1], "KPSS_p(c)": k[1], "KPSS_p(ct)": kt[1]})
tabla = pd.DataFrame(filas).set_index("serie").round(3)
print("p-valores (ADF: H0 = raiz unitaria | KPSS: H0 = estacionaria; statsmodels trunca KPSS a [0.01, 0.10])\n", tabla.to_string())

# La receta: diferenciar. La serie de leche pide diferencia estacional (12) y/o ordinaria (1)
d12 = leche.diff(12).dropna()
d1_12 = leche.diff(12).diff(1).dropna()
for nombre, v in {"Leche, diferencia 12": d12.values, "Leche, diferencia 12 y 1": d1_12.values, "Caminata, diferencia 1": np.diff(rw)}.items():
    print(f"{nombre:26s} | ADF p = {adfuller(v, regression='c', autolag='AIC')[1]:.4f} | KPSS p = {kpss(v, regression='c', nlags='auto')[1]:.3f}")
assert adfuller(rw)[1] > 0.05 and adfuller(np.diff(rw))[1] < 0.01 and adfuller(ar1)[1] < 0.01

# --- ACF y PACF desde cero ---
def acf_propia(y, nlags):
    y = np.asarray(y, float) - np.mean(y)
    return np.array([y[k:] @ y[: len(y) - k] for k in range(nlags + 1)]) / (y @ y)

def pacf_propia(y, nlags):                                # Durbin-Levinson sobre la ACF
    r = acf_propia(y, nlags)
    phi, out = np.zeros((nlags + 1, nlags + 1)), [1.0]
    for k in range(1, nlags + 1):
        num = r[k] - (phi[k - 1, 1:k] @ r[1:k][::-1] if k > 1 else 0.0)
        den = 1 - (phi[k - 1, 1:k] @ r[1:k] if k > 1 else 0.0)
        phi[k, k] = num / den
        for j in range(1, k):
            phi[k, j] = phi[k - 1, j] - phi[k, k] * phi[k - 1, k - j]
        out.append(phi[k, k])
    return np.array(out)

def ljung_box(y, h):
    r, N_ = acf_propia(y, h), len(y)
    return N_ * (N_ + 2) * np.sum(r[1:] ** 2 / (N_ - np.arange(1, h + 1)))

# Dos procesos de identificacion clasica: AR(2) (ACF decae, PACF corta en 2) y MA(1) (ACF corta en 1, PACF decae)
M = 600
z = rng.normal(size=M + 2)
ar2 = np.zeros(M + 2); ma1 = np.zeros(M + 2)
for i in range(2, M + 2):
    ar2[i] = 0.5 * ar2[i - 1] - 0.3 * ar2[i - 2] + z[i]
    ma1[i] = z[i] + 0.8 * z[i - 1]
ar2, ma1 = ar2[2:], ma1[2:]
for v in (ar2, ma1, d1_12.values):
    assert np.allclose(acf_propia(v, 20), acf(v, nlags=20, fft=False))
    assert np.allclose(pacf_propia(v, 20), pacf(v, nlags=20, method="ywm"))
    assert np.isclose(ljung_box(v, 10), acorr_ljungbox(v, lags=[10])["lb_stat"].iloc[0])
print("\nACF, PACF (Durbin-Levinson) y Ljung-Box propios == statsmodels (assert OK)")
print("AR(2): PACF rezagos 1-4 =", pacf_propia(ar2, 4)[1:].round(2), "| MA(1): ACF rezagos 1-4 =", acf_propia(ma1, 4)[1:].round(2))
print(f"Banda de ruido blanco +-1.96/sqrt(n) = +-{1.96 / np.sqrt(M):.3f} (n = {M})")
print(f"Leche con dif. 12 y 1 (n = {len(d1_12)}): ACF rezago 12 = {acf_propia(d1_12.values, 12)[12]:.2f}, PACF rezago 12 = {pacf_propia(d1_12.values, 12)[12]:.2f}, banda +-{1.96 / np.sqrt(len(d1_12)):.2f}")

fig, ax = plt.subplots(2, 3, figsize=(12, 6), sharey=True)
for j, (nom, v) in enumerate([("AR(2)", ar2), ("MA(1)", ma1), ("Leche dif. 12 y 1", d1_12.values)]):
    for i, (f, lab) in enumerate([(acf_propia, "ACF"), (pacf_propia, "PACF")]):
        r_ = f(v, 24); b = 1.96 / np.sqrt(len(v))
        ax[i, j].vlines(range(1, 25), 0, r_[1:], color=PALETA[i], lw=2)
        ax[i, j].axhspan(-b, b, color="grey", alpha=0.2); ax[i, j].axhline(0, color="k", lw=0.8)
        ax[i, j].set_title(f"{lab} - {nom}"); ax[i, j].set_xlabel("rezago")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Las funciones propias **coinciden** con `adfuller`, `kpss`, `acf`, `pacf` y `acorr_ljungbox` (los `assert` pasan). (2) **La tabla pide matices, no recetas.** La **AR(1)** es estacionaria para las dos pruebas (ADF $p<0.001$; KPSS no rechaza). La **tendencia determinista + AR(1)** y la **leche** (por construcción también tendencia determinista + estacionalidad + AR(1)) tienen ADF $p=0.756$ y $0.929$ y KPSS **con constante** $p=0.01$: parecen no estacionarias; pero el KPSS **con tendencia** ya **no rechaza** ($p=0.10$): son **estacionarias alrededor de una tendencia** (*trend-stationary*), no raíces unitarias. La distinción importa: se **quita la tendencia** (o se diferencia) según el caso. (3) **Una sola realización puede engañar:** la **caminata aleatoria** (raíz unitaria *verdadera*) obtuvo ADF $p=0.181$ (no rechaza, bien), pero su KPSS **con constante** *tampoco* rechazó ($p=0.10$): caso **inconcluso** (fila 3 de la tabla); solo el KPSS con tendencia ($p=0.01$) la delata. Los valores $0.10$ y $0.01$ son los **límites de truncamiento** de `statsmodels` (el $p$ real es $\ge 0.10$ o $\le 0.01$). (4) Diferenciar funciona: la caminata diferenciada y la leche con diferencia 12 (y con 12 y 1) pasan ADF ($p=0.0085$ y $<0.0001$) y KPSS. (5) **ACF/PACF reconocen el proceso:** en el AR(2) la PACF vale $0.43$ y $-0.35$ en los rezagos 1 y 2 y luego queda dentro de la banda ($\pm 0.080$); en el MA(1) la ACF vale $0.51$ en el rezago 1 (teórica: $0.8/1.64=0.49$) y se anula después. ## 3. Suavizado Exponencial: de Pronóstico Plano a Holt–Winters 📉

El **suavizado exponencial simple (SES)** pronostica con un **promedio ponderado** del pasado en el que los pesos **decaen geométricamente**: $\ell_t=\alpha\,y_t+(1-\alpha)\,\ell_{t-1}$ con $0<\alpha<1$, y el pronóstico a cualquier horizonte es el último nivel: $\hat y_{t+h|t}=\ell_t$. Un $\alpha$ alto da peso a lo reciente; uno bajo, a la historia. Se estima minimizando la suma de cuadrados de los **errores a un paso**, $\sum_t (y_t-\ell_{t-1})^2$.

**Holt–Winters aditivo** agrega una ecuación para la **tendencia** $b_t$ y otra para la **estacionalidad** $s_t$ (con parámetros $\beta$ y $\gamma$): $\hat y_{t+h|t}=\ell_t+h\,b_t+s_{t+h-m(k+1)}$, donde $k=\lfloor (h-1)/m\rfloor$. Particionamos **en el tiempo**: los primeros 120 meses para estimar y los últimos 24 para evaluar. Implementamos el SES desde cero (contrastado con `SimpleExpSmoothing`) y ajustamos Holt–Winters con `statsmodels`.

````python
# ============================================================
# 3. Suavizado exponencial: SES desde cero y Holt-Winters
# ============================================================
from scipy.optimize import minimize_scalar
from statsmodels.tsa.holtwinters import ExponentialSmoothing, SimpleExpSmoothing

# Particion TEMPORAL (nunca aleatoria): 10 anios de entrenamiento y 2 de prueba
n_tr = 120
train, test = leche.iloc[:n_tr], leche.iloc[n_tr:]

def ses(y, alpha, nivel0=None):
    """Nivel l_t = alpha*y_t + (1-alpha)*l_{t-1}. Pronostico a un paso para y_t = l_{t-1}. Devuelve (ajustados, ultimo nivel)."""
    y = np.asarray(y, float)
    l = y[0] if nivel0 is None else nivel0
    ajust = np.empty(len(y))
    for i, v in enumerate(y):
        ajust[i] = l
        l = alpha * v + (1 - alpha) * l
    return ajust, l

def sse_ses(alpha, y):
    return np.sum((np.asarray(y, float) - ses(y, alpha)[0]) ** 2)

# alpha optimo por minimos cuadrados del error a un paso, contrastado con statsmodels (mismo nivel inicial fijo)
opt = minimize_scalar(sse_ses, bounds=(0.001, 0.999), args=(train.values,), method="bounded", options={"xatol": 1e-8})
alpha_ses = opt.x
sm_ses = SimpleExpSmoothing(train, initialization_method="known", initial_level=train.iloc[0]).fit(optimized=True)
a_prop, nivel_final = ses(train.values, alpha_ses)
assert abs(sm_ses.params["smoothing_level"] - alpha_ses) < 5e-3
assert np.allclose(ses(train.values, 0.3)[0], SimpleExpSmoothing(train, initialization_method="known", initial_level=train.iloc[0]).fit(smoothing_level=0.3, optimized=False).fittedvalues.values)
print(f"SES: alpha optimo propio = {alpha_ses:.3f} | statsmodels = {sm_ses.params['smoothing_level']:.3f}")
print(f"El pronostico SES es PLANO: {nivel_final:.1f} mil L para todos los horizontes (ignora tendencia y estacionalidad)")

# Holt-Winters aditivo (nivel + tendencia + estacionalidad de periodo 12)
hw = ExponentialSmoothing(train, trend="add", seasonal="add", seasonal_periods=12, initialization_method="estimated").fit()
f_ses = np.full(len(test), nivel_final)
f_hw = hw.forecast(len(test)).values
mae = lambda a, f: np.mean(np.abs(np.asarray(a) - np.asarray(f)))
print(f"MAE en prueba (24 meses): SES = {mae(test, f_ses):.1f} | Holt-Winters = {mae(test, f_hw):.1f} mil L")
print("Parametros HW: alpha = %.3f, beta = %.3f, gamma = %.3f" % (hw.params["smoothing_level"], hw.params["smoothing_trend"], hw.params["smoothing_seasonal"]))
assert mae(test, f_hw) < mae(test, f_ses)

# Efecto de alpha: mas alto = recuerda menos; mas bajo = mas suave
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
for a_, c in zip((0.1, 0.5, 0.9), PALETA[1:4]):
    ax[0].plot(train.index[-36:], ses(train.values, a_)[0][-36:], color=c, label=f"alpha = {a_}")
ax[0].plot(train.index[-36:], train.values[-36:], "k.", ms=4, label="serie"); ax[0].legend(); ax[0].set_title("SES: el efecto de alpha")
ax[1].plot(leche.index[-60:], leche.values[-60:], color="k", lw=1.2, label="serie")
ax[1].plot(test.index, f_ses, color=PALETA[1], label="SES (plano)"); ax[1].plot(test.index, f_hw, color=PALETA[0], label="Holt-Winters")
ax[1].axvline(test.index[0], color="grey", ls="--"); ax[1].legend(); ax[1].set_title("Pronostico a 24 meses")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) El SES propio **reproduce exactamente** los valores ajustados de `SimpleExpSmoothing` para un $\alpha$ dado, y el $\alpha$ óptimo coincide (**0.999** propio, **1.000** statsmodels). (2) Que el óptimo sea ≈ 1 **no es un fallo**: con una tendencia fuerte la mejor «memoria» de un solo nivel es la última observación, es decir, el SES degenera en el **pronóstico ingenuo** (de hecho sus errores de prueba coinciden: MAE **72.7** para ambos en la tabla de la sección siguiente). El pronóstico es **plano** (693.1 mil L) mientras la serie sigue subiendo y oscilando. (3) **Holt–Winters baja el error de 72.7 a 17.3 mil L** porque modela justo los dos componentes que el SES ignora. Estimó $\beta=0$ y $\gamma=0$ (tendencia y patrón estacional **fijos**), coherente con cómo generamos los datos; en una serie real esos parámetros rara vez serían cero. (4) **Moraleja:** elegir el método según los componentes presentes (nivel → SES; nivel + tendencia → Holt; + estacionalidad → Holt–Winters) y no por moda.

## 4. Modelos ARIMA y SARIMA 🧮

Un **AR($p$)** regresa la serie sobre sus propios $p$ rezagos: $y_t=c+\phi_1y_{t-1}+\dots+\phi_py_{t-p}+\varepsilon_t$. Un **MA($q$)** la expresa como combinación de $q$ choques pasados: $y_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\dots+\theta_q\varepsilon_{t-q}$. El **ARIMA($p,d,q$)** aplica ARMA($p,q$) a la serie diferenciada $d$ veces; el **SARIMA($p,d,q$)($P,D,Q$)$_m$** añade la parte estacional de periodo $m$. En operadores de rezago $B$: $\phi(B)\,\Phi(B^m)\,(1-B)^d(1-B^m)^D\,y_t=c+\theta(B)\,\Theta(B^m)\,\varepsilon_t$.

La estimación habitual es por **máxima verosimilitud**; la de un AR($p$) puro puede aproximarse por **mínimos cuadrados** (regresión sobre los rezagos), que hacemos desde cero y comparamos. Para elegir el orden se minimiza el **AIC** $=-2\ln\hat L+2k$ o el **BIC** $=-2\ln\hat L+k\ln N$ (penaliza más la complejidad); y se **diagnostican los residuos**: si el modelo captó la dependencia, los residuos deben parecer **ruido blanco** (Ljung–Box con $\chi^2_{h-q}$). Con la serie de leche diferenciada una vez estacional y otra ordinaria, probamos $p\in\{0,1,2\}$, $q\in\{0,1\}$ con la parte estacional $(0,1,1)_{12}$ fija.

````python
# ============================================================
# 4. ARIMA: AR(p) por minimos cuadrados desde cero vs statsmodels; seleccion por AIC; diagnostico
# ============================================================
from statsmodels.tsa.arima.model import ARIMA

# AR(2) por minimos cuadrados condicionales (regresion de y_t sobre y_{t-1}, y_{t-2}) sobre el AR(2) simulado antes
def ar_mco(y, p):
    y = np.asarray(y, float) - np.mean(y)
    X = np.column_stack([y[p - j - 1: len(y) - j - 1] for j in range(p)])
    return np.linalg.lstsq(X, y[p:], rcond=None)[0]

phi_mco = ar_mco(ar2, 2)
phi_sm = ARIMA(ar2, order=(2, 0, 0), trend="c").fit().params[1:3]
print(f"AR(2) real = [0.50, -0.30] | MCO propio = {phi_mco.round(3)} | ARIMA (max. verosimilitud) = {phi_sm.round(3)}")
assert np.allclose(phi_mco, phi_sm, atol=0.02) and np.allclose(phi_mco, [0.5, -0.3], atol=0.12)

# SARIMA(p,1,q)(0,1,1)_12 sobre la serie de entrenamiento: rejilla pequena y criterio AIC
filas = []
for p in range(3):
    for q in range(2):
        r = ARIMA(train, order=(p, 1, q), seasonal_order=(0, 1, 1, 12)).fit()
        filas.append({"p": p, "q": q, "AIC": r.aic, "BIC": r.bic})
rej = pd.DataFrame(filas).sort_values("AIC").reset_index(drop=True)
print("\nRejilla SARIMA(p,1,q)(0,1,1)_12 ordenada por AIC (menor = mejor):\n", rej.round(1).to_string(index=False))
p_b, q_b = int(rej.loc[0, "p"]), int(rej.loc[0, "q"])
sarima = ARIMA(train, order=(p_b, 1, q_b), seasonal_order=(0, 1, 1, 12)).fit()

# Diagnostico: los residuos deben parecerse a ruido blanco (Ljung-Box con gl = rezagos - parametros ARMA)
resid = sarima.resid.iloc[13:]                                   # se descartan los primeros 13 (inicializacion)
lb = acorr_ljungbox(resid, lags=[12, 24], model_df=p_b + q_b + 1)
print(f"\nModelo elegido por AIC: SARIMA({p_b},1,{q_b})(0,1,1)_12 | residuos: media {resid.mean():.2f}, sd {resid.std(ddof=1):.2f}")
print("Ljung-Box de los residuos (H0: ruido blanco):\n", lb.round(3).to_string())
fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
ax[0].plot(resid.index, resid.values, color=PALETA[0], lw=1); ax[0].axhline(0, color="k", lw=0.8); ax[0].set_title("Residuos del SARIMA")
r_ = acf_propia(resid.values, 24)
ax[1].vlines(range(1, 25), 0, r_[1:], color=PALETA[1], lw=2); ax[1].axhspan(-1.96 / np.sqrt(len(resid)), 1.96 / np.sqrt(len(resid)), color="grey", alpha=0.2)
ax[1].set_title("ACF de los residuos"); ax[1].set_xlabel("rezago")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Sobre el AR(2) simulado, los mínimos cuadrados propios dan $\hat\phi=(0.577,\,-0.351)$ y la máxima verosimilitud de `statsmodels` $(0.576,\,-0.350)$: **prácticamente idénticos**, y ambos cerca de los valores reales $(0.5,\,-0.3)$; la diferencia con los reales (≈ 0.08) es **error muestral** de una sola realización de 600 puntos, no un sesgo del método. (2) El AIC elige **SARIMA(1,1,1)(0,1,1)$_{12}$** (AIC **826.5** frente a 828.3 del (2,1) y 841.7 del (0,0)); el BIC elige el mismo (837.2). La ganancia de pasar de (0,0) a (1,1) es de **15 puntos de AIC**: la parte no estacional **sí** hacía falta. Nota: la ganancia de agregar más términos es pequeña y un AIC más bajo por 1–2 puntos no es evidencia fuerte. (3) **Diagnóstico:** Ljung–Box no rechaza ruido blanco en los residuos ($p=0.263$ a 12 rezagos y $0.634$ a 24), y la ACF de los residuos queda dentro de la banda: el modelo captó la dependencia lineal. Pero **«residuos blancos» no significa «buen pronóstico»**: solo que no se dejó estructura lineal evidente; la prueba real es la que sigue, **fuera de muestra**.

## 5. Validar sin Hacer Trampa: Orden Temporal y *Baselines* ⚖️

Tres reglas para evaluar pronósticos con honestidad:

1. **Respeta el tiempo.** El conjunto de prueba debe estar **después** del de entrenamiento. Mezclar al azar (validación cruzada ordinaria) deja que el modelo «vea el futuro» (fuga de información) y, como las observaciones vecinas son parecidas, **hace creer que el modelo es mejor de lo que es**. Las variantes correctas son la **partición temporal** y el **origen móvil** (*rolling origin*: se reajusta el modelo con datos hasta cada origen y se evalúan los siguientes $H$ meses).
2. **Compara contra *baselines*.** Un modelo sofisticado solo importa si **supera a los pronósticos ingenuos**: el **ingenuo** ($\hat y_{n+h}=y_n$), el **ingenuo estacional** ($\hat y_{n+h}=y_{n+h-m}$), la **deriva** (recta entre el primer y el último dato) y el **estacional con deriva**.
3. **Usa métricas sin escala.** El **MASE** divide el MAE por el MAE del ingenuo estacional **dentro de la muestra de entrenamiento**: $\text{MASE}=\dfrac{\frac1h\sum_{j}|y_{n+j}-\hat y_{n+j}|}{\frac1{n-m}\sum_{t=m+1}^{n}|y_t-y_{t-m}|}$. Un MASE $<1$ indica mejor que ese ingenuo de referencia; permite comparar series de distinta escala.

Evaluamos siete métodos en los 24 meses de prueba, ponemos a prueba el **intervalo del 95 %** del SARIMA y demostramos la **fuga** con un bosque aleatorio sobre variables de rezago (misma evaluación sobre las **mismas** observaciones con `KFold` aleatorio y con `TimeSeriesSplit`).

````python
# ============================================================
# 5. Validacion temporal, baselines y comparacion honesta
# ============================================================
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, TimeSeriesSplit

def f_ingenuo(y, h):
    return np.full(h, y[-1])

def f_estacional(y, h, m=12):
    return np.array([y[len(y) - m + (i % m)] for i in range(h)])

def f_deriva(y, h):
    return y[-1] + (y[-1] - y[0]) / (len(y) - 1) * np.arange(1, h + 1)

def f_estacional_deriva(y, h, m=12):                    # ingenuo estacional + cambio medio anual acumulado
    anual = np.mean(y[m:] - y[:-m])
    return f_estacional(y, h, m) + anual * (np.arange(h) // m + 1)

def mase(real, pron, y_train, m=12):
    escala = np.mean(np.abs(y_train[m:] - y_train[:-m]))  # MAE del ingenuo estacional DENTRO de la muestra
    return np.mean(np.abs(np.asarray(real) - np.asarray(pron))) / escala

h = len(test)
y_tr = train.values
sar_f = sarima.get_forecast(h)
pron = {
    "Ingenuo": f_ingenuo(y_tr, h), "Ingenuo estacional": f_estacional(y_tr, h), "Deriva": f_deriva(y_tr, h),
    "Estacional + deriva": f_estacional_deriva(y_tr, h), "SES": f_ses, "Holt-Winters": f_hw, "SARIMA": sar_f.predicted_mean.values,
}
tab = pd.DataFrame({k: {"MAE": mae(test, v), "RMSE": np.sqrt(np.mean((test.values - v) ** 2)), "MASE": mase(test.values, v, y_tr)} for k, v in pron.items()}).T
print("Evaluacion en los 24 meses de PRUEBA (el modelo nunca los vio):\n", tab.sort_values("MASE").round(2).to_string())
print("\nMASE < 1 => mejor que el ingenuo estacional medido en la muestra de entrenamiento; > 1 => peor.")
assert np.isclose(mase(test.values, pron["Ingenuo estacional"], y_tr), tab.loc["Ingenuo estacional", "MASE"])

# --- Pronostico del SARIMA con intervalo del 95 % ---
ic = sar_f.conf_int(alpha=0.05)
dentro = np.mean((test.values >= ic.iloc[:, 0].values) & (test.values <= ic.iloc[:, 1].values))
print(f"\nSARIMA: ancho medio del IC 95 % = {np.mean(ic.iloc[:, 1] - ic.iloc[:, 0]):.1f} mil L | el IC contiene {dentro:.0%} de los 24 valores reales")

# --- Fuga de informacion: validacion cruzada ALEATORIA vs TEMPORAL con variables de rezago ---
df = pd.DataFrame({"y": leche.values, "lag1": leche.shift(1).values, "lag12": leche.shift(12).values,
                   "sen": np.sin(2 * np.pi * t / 12), "cos": np.cos(2 * np.pi * t / 12)}).dropna()
X, yy = df[["lag1", "lag12", "sen", "cos"]].values, df["y"].values
def pred_cv(cv):
    p = np.full(len(yy), np.nan)
    for tr, te in cv.split(X):
        p[te] = RandomForestRegressor(n_estimators=40, min_samples_leaf=2, random_state=SEED).fit(X[tr], yy[tr]).predict(X[te])
    return p
p_kfold, p_temporal = pred_cv(KFold(5, shuffle=True, random_state=SEED)), pred_cv(TimeSeriesSplit(5))
ev = ~np.isnan(p_temporal)                              # mismas observaciones evaluadas por ambos esquemas
mae_kfold, mae_temporal = mae(yy[ev], p_kfold[ev]), mae(yy[ev], p_temporal[ev])
print(f"\nBosque aleatorio con rezagos, MAE sobre las MISMAS {ev.sum()} observaciones | KFold aleatorio = {mae_kfold:.1f} | TimeSeriesSplit = {mae_temporal:.1f} mil L")
assert mae_temporal > 1.3 * mae_kfold

fig, ax = plt.subplots(figsize=(10, 4.3))
ax.plot(leche.index[-60:], leche.values[-60:], color="k", lw=1.2, label="serie")
for k, c in [("Ingenuo estacional", PALETA[3]), ("Holt-Winters", PALETA[2]), ("SARIMA", PALETA[0])]:
    ax.plot(test.index, pron[k], color=c, label=k)
ax.fill_between(test.index, ic.iloc[:, 0], ic.iloc[:, 1], color=PALETA[0], alpha=0.15, label="IC 95 % SARIMA")
ax.axvline(test.index[0], color="grey", ls="--"); ax.legend(ncol=3, fontsize=8); ax.set_title("Prueba temporal de 24 meses: modelos frente a baselines")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **Orden de la tabla (MASE):** SARIMA **0.63**, Holt–Winters **0.70**, estacional + deriva **0.93**, ingenuo estacional **1.79**, deriva **2.33**, SES e ingenuo **2.95**. Ambos modelos le ganan al *baseline* **más fuerte** (estacional + deriva) y los dos están por debajo de 1. El ingenuo estacional, en cambio, tiene MASE $>1$ en prueba porque **ignora la tendencia**: pronostica 24 meses con el nivel de hace un año. (2) **Ojo con la honestidad:** estos datos se generaron con tendencia y estacionalidad regulares, que son justo lo que SARIMA y Holt–Winters modelan; que ganen **no** prueba que ganarían en cualquier serie real. Con series ruidosas o con cambios de régimen, el ingenuo estacional suele ser difícil de superar. (3) **El intervalo del 95 % cubrió solo 79 %** de los 24 valores reales (19 de 24; con solo 24 puntos la cobertura es muy variable, pero la señal de subcobertura se confirma abajo). (4) **La fuga existe:** el bosque con `KFold` aleatorio obtuvo un MAE de **14.8** y con `TimeSeriesSplit` **21.5** (sobre las mismas 110 observaciones): la validación aleatoria **subestima el error ≈ 31 %** porque entrena con datos posteriores. Parte de la diferencia es que cada pliegue temporal entrena con **menos** datos (y un bosque no extrapola la tendencia); esa es justamente la situación real al pronosticar.

## 6. Origen Móvil, Comparación Honesta y Cobertura de los Intervalos 🎯

Una sola partición (24 meses) es **una** muestra del desempeño; el error depende de qué meses cayeron en la prueba. Con el **origen móvil** repetimos el experimento en 11 orígenes (cada 6 meses, de t = 72 a t = 132), reajustando cada modelo con los datos disponibles y pronosticando 12 meses. Esto da una distribución del error y permite calcular la **cobertura empírica** del intervalo del 95 %: *¿qué fracción de los valores reales cae dentro?* Un intervalo bien calibrado cubre ≈ 95 %; por debajo, **subestima la incertidumbre**. Las causas típicas: la **incertidumbre de los parámetros** (el intervalo asume que los parámetros estimados son los verdaderos), el supuesto de **normalidad** y la posibilidad de que el modelo esté **mal especificado**.

````python
# ============================================================
# 6. Origen movil (rolling origin): comparacion honesta y cobertura de los intervalos
# ============================================================
import time
H = 12
origenes = list(range(72, n - H + 1, 6))                 # el modelo se re-ajusta con datos hasta cada origen
reg = []
t0 = time.time()
for o in origenes:
    y_o, real = leche.values[:o], leche.values[o:o + H]
    serie_o = leche.iloc[:o]
    sar = ARIMA(serie_o, order=(1, 1, 1), seasonal_order=(0, 1, 1, 12)).fit()
    fc = sar.get_forecast(H)
    hw_o = ExponentialSmoothing(serie_o, trend="add", seasonal="add", seasonal_periods=12, initialization_method="estimated").fit()
    cand = {"Ingenuo estacional": f_estacional(y_o, H), "Estacional + deriva": f_estacional_deriva(y_o, H),
            "Holt-Winters": hw_o.forecast(H).values, "SARIMA": fc.predicted_mean.values}
    ci = fc.conf_int(alpha=0.05).values
    for nombre, f in cand.items():
        reg.append({"origen": o, "modelo": nombre, "MAE": mae(real, f), "MASE": mase(real, f, y_o)})
    reg.append({"origen": o, "modelo": "_cobertura", "MAE": np.mean((real >= ci[:, 0]) & (real <= ci[:, 1])), "MASE": np.mean(ci[:, 1] - ci[:, 0])})
print(f"{len(origenes)} origenes x {H} meses de horizonte | tiempo: {time.time() - t0:.1f} s")
R = pd.DataFrame(reg)
res_mod = R[R.modelo != "_cobertura"]
resumen = res_mod.groupby("modelo")[["MAE", "MASE"]].mean().sort_values("MASE")
print("\nPromedio sobre los origenes moviles:\n", resumen.round(2).to_string())

piv = res_mod.pivot(index="origen", columns="modelo", values="MAE")
dif = piv["Estacional + deriva"] - piv["SARIMA"]            # >0 => SARIMA mejor
se = dif.std(ddof=1) / np.sqrt(len(dif))
print(f"\nSARIMA mejora al 'estacional + deriva' en {(dif > 0).sum()} de {len(dif)} origenes | diferencia media de MAE = {dif.mean():.2f} (EE aprox. {se:.2f}; los origenes se traslapan, es solo orientativo)")
cobertura = R[R.modelo == "_cobertura"]["MAE"].mean()
print(f"Cobertura empirica del IC 95 % del SARIMA a {H} meses: {cobertura:.1%} (nominal 95 %) | ancho medio {R[R.modelo == '_cobertura']['MASE'].mean():.1f} mil L")
assert 0.5 < cobertura <= 1.0 and resumen.index[0] in ("SARIMA", "Holt-Winters")

fig, ax = plt.subplots(1, 2, figsize=(12, 4.3))
for k, c in zip(["Ingenuo estacional", "Estacional + deriva", "Holt-Winters", "SARIMA"], PALETA[3:] + PALETA[:1]):
    ax[0].plot(piv.index, piv[k], marker="o", ms=4, color=c, label=k)
ax[0].set_xlabel("origen del pronostico (mes)"); ax[0].set_ylabel("MAE a 12 meses"); ax[0].legend(fontsize=8); ax[0].set_title("Error por origen movil")
ancho_h = (ic.iloc[:, 1] - ic.iloc[:, 0]).values            # intervalo del SARIMA ajustado con 120 meses (seccion anterior)
ax[1].plot(range(1, 25), ancho_h, marker="o", ms=4, color=PALETA[0]); ax[1].set_xlabel("horizonte (meses)"); ax[1].set_ylabel("ancho del IC 95 % (mil L)")
ax[1].set_title("La incertidumbre crece con el horizonte")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Con 11 orígenes y 12 meses de horizonte, el **SARIMA** tiene el menor error (MAE **10.89**, MASE **0.44**), seguido de **estacional + deriva** (13.96; 0.56), **Holt–Winters** (17.78; 0.71) y el **ingenuo estacional** (25.41; 1.03). Con una sola partición Holt–Winters parecía mejor que el *baseline*; con origen móvil **queda por detrás del estacional + deriva**: otra razón para no decidir con una prueba. (2) El SARIMA mejora al *baseline* fuerte en **9 de 11** orígenes, con diferencia media de MAE de **3.07** (EE ≈ 1.09). Las ventanas se traslapan, así que el error estándar es solo **orientativo**: la ventaja existe pero no es enorme (≈ 22 % del MAE), y un baseline de una línea ya hace un trabajo muy bueno. (3) **Cobertura:** el intervalo del 95 % cubrió **90.9 %** a 12 meses (y 79 % en la partición de 24 meses): **subestima** la incertidumbre, como era de esperar. El ancho medio es ≈ 51 mil L, y **crece con el horizonte** (figura de la derecha). Si la decisión depende del intervalo, conviene **ampliarlo** (p. ej. por simulación con remuestreo de residuos o recalibrándolo con la cobertura observada). (4) **Receta honesta de pronóstico:** descomponer y mirar los datos → comprobar estacionariedad y diferenciar → probar *baselines* y modelos → evaluar con **origen móvil** y MASE → **reportar intervalos con su cobertura observada**.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Componentes** | Tendencia + estacionalidad + ruido (aditivo o multiplicativo). La descomposición clásica por media móvil coincide con `seasonal_decompose`, **describe pero no pronostica**, y subestima el ruido persistente (residuo sd 9.8 frente a 15.0 real) |
| **Estacionariedad** | ADF ($H_0$: raíz unitaria) y KPSS ($H_0$: estacionaria) se leen **juntas**; una sola realización puede dar señales inconclusas (caminata aleatoria: ADF $p=0.18$ y KPSS $p=0.10$). La leche era estacionaria **alrededor de una tendencia** (KPSS con constante $p=0.01$, con tendencia $p=0.10$) |
| **ACF / PACF** | Identifican el proceso: AR($p$) → PACF se corta en $p$; MA($q$) → ACF se corta en $q$. Ljung–Box resume la autocorrelación residual. Implementadas desde cero e iguales a `statsmodels` |
| **Suavizado exponencial** | SES con tendencia degenera en el ingenuo ($\alpha\approx1$, MAE 72.7); Holt–Winters modela tendencia y estacionalidad (MAE 17.3) |
| **ARIMA / SARIMA** | MCO propio ≈ máxima verosimilitud en un AR($p$); AIC/BIC eligen el orden (SARIMA(1,1,1)(0,1,1)$_{12}$) y los residuos se diagnostican con Ljung–Box |
| **Validación temporal** | Nunca al azar: la validación aleatoria dio un MAE 31 % menor (14.8 frente a 21.5) por **fuga de información**. Usar partición temporal y **origen móvil** |
| **Comparación honesta** | Todo modelo debe vencer a los *baselines*; el MASE lo cuantifica. SARIMA ganó (MASE 0.44) al mejor baseline (0.56) en 9 de 11 orígenes, pero con una ventaja moderada y con datos que cumplen justo sus supuestos |
| **Intervalos** | El IC 95 % cubrió 90.9 % (12 meses) y 79 % (24 meses): subestima la incertidumbre. **Hay que medir la cobertura**, no asumirla |
