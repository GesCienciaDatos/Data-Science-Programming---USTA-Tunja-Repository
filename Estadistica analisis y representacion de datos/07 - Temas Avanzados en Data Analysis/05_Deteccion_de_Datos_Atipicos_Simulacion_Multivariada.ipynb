## 1. Simular Datos Multivariados con Contaminación Controlada 🎲

**Modelo de contaminación (Huber).** Cada observación viene, con probabilidad $1-\varepsilon$, de la distribución «buena» $F=N_p(\boldsymbol\mu,\Sigma)$ y, con probabilidad $\varepsilon$, de una distribución contaminante $H$ que **no conocemos en la práctica pero sí en la simulación**: $X\sim(1-\varepsilon)F+\varepsilon H$. Como en el libro, generamos $F$ con la normal multivariada: en NumPy, `rng.multivariate_normal(mu, Sigma, size=n)` (el método del `Generator` moderno) y la función clásica `np.random.multivariate_normal` (semilla global) producen la **misma distribución**; verificamos que la covarianza muestral recupera $\Sigma$. Tomamos $\Sigma_{ij}=\rho^{|i-j|}$ con $\rho=0.8$ (correlación AR(1)) y tres tipos de contaminante $H$:

| Escenario | $H$ | Qué lo hace difícil |
|---|---|---|
| **marginal** | una coordenada se desplaza $\pm6$ desviaciones estándar | Fácil: es extremo en **una** variable |
| **correlación** | misma media y varianzas, pero **correlaciones con el signo invertido** | **Cada marginal es normal**: ninguna regla univariada puede verlo |
| **cluster** | un **grupo compacto** centrado en $(3.5,\dots,3.5)$ con covarianza $0.15\,\Sigma$ | Muchos atípicos juntos **arrastran** la media y la covarianza clásicas (enmascaramiento) |

**Verificación teórica.** Si $X\sim N_p(\boldsymbol\mu,\Sigma)$, la distancia de Mahalanobis al cuadrado $d^2=(X-\boldsymbol\mu)^\top\Sigma^{-1}(X-\boldsymbol\mu)$ sigue una $\chi^2_p$ (suma de $p$ normales estándar independientes al cuadrado). Lo comprobamos con una prueba de Kolmogorov–Smirnov en datos limpios; este hecho es el que justifica **el corte con un cuantil de la $\chi^2$** más adelante.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

pd.set_option("display.width", 150); pd.set_option("display.max_columns", 20)

# ---------- Simulación multivariada: np.random.multivariate_normal ----------
# (a) La función clásica (semilla global) y (b) el método del Generador moderno: misma distribución
mu = np.array([0.0, 0.0]); Sigma = np.array([[1.0, 0.8], [0.8, 1.0]])
np.random.seed(0)
X_legacy = np.random.multivariate_normal(mu, Sigma, size=20000)                  # API clásica
rng = np.random.default_rng(42)
X_gen = rng.multivariate_normal(mu, Sigma, size=20000)                           # API moderna (la que usaremos)
for nombre, X0 in [("np.random.multivariate_normal", X_legacy), ("Generator.multivariate_normal", X_gen)]:
    assert np.allclose(np.cov(X0.T), Sigma, atol=0.03) and np.allclose(X0.mean(0), mu, atol=0.03)
print("covarianza muestral (Generator):\n", np.cov(X_gen.T).round(3))

# ---------- Generador de datos con contaminación CONTROLADA ----------
def ar1(p, rho): return rho ** np.abs(np.subtract.outer(np.arange(p), np.arange(p)))

def simular(n, p, eps, tipo, rng, rho=0.8):
    """n observaciones, fracción eps de atípicos de un `tipo`; devuelve X (n x p) y y (1 = atípico)."""
    S = ar1(p, rho); m = int(round(eps * n))
    X = rng.multivariate_normal(np.zeros(p), S, size=n); y = np.zeros(n, int)
    idx = rng.choice(n, m, replace=False); y[idx] = 1
    if tipo == "marginal":            # una coordenada se dispara (+6 desviaciones estándar)
        j = rng.integers(0, p, m); X[idx, j] += 6 * np.sqrt(S[j, j]) * rng.choice([-1, 1], m)
    elif tipo == "correlacion":       # mismas marginales, pero la correlación apunta al revés
        D = np.diag((-1.0) ** np.arange(p))
        X[idx] = rng.multivariate_normal(np.zeros(p), D @ S @ D, size=m)
    elif tipo == "cluster":           # un grupo compacto desplazado
        X[idx] = rng.multivariate_normal(np.full(p, 3.5), 0.15 * S, size=m)
    return X, y

# Verificación: sin contaminación, la distancia de Mahalanobis² al cuadrado sigue una chi² con p grados de libertad
p = 2
Xc, _ = simular(5000, p, 0.0, "marginal", rng)
d2 = np.einsum("ij,jk,ik->i", Xc, np.linalg.inv(ar1(p, 0.8)), Xc)
ks = stats.kstest(d2, stats.chi2(df=p).cdf)
print(f"Mahalanobis² de datos limpios vs chi²({p}): KS p = {ks.pvalue:.3f}; cuantil 97.5 %: {np.quantile(d2, 0.975):.2f} (teórico {stats.chi2.ppf(0.975, p):.2f})")
assert ks.pvalue > 0.01

# ---------- Los tres escenarios en el plano ----------
esc = {k: simular(600, 2, 0.06, k, np.random.default_rng(1)) for k in ["marginal", "correlacion", "cluster"]}
fig, ax = plt.subplots(1, 3, figsize=(15, 4.3))
for a, (k, (X, y)) in zip(ax, esc.items()):
    a.scatter(*X[y == 0].T, s=9, color="#93c5fd", label="datos buenos"); a.scatter(*X[y == 1].T, s=26, color="#dc2626", label="contaminación")
    a.set_title(f"Escenario «{k}»"); a.set_xlabel("$x_1$"); a.set_ylabel("$x_2$"); a.legend(loc="upper left", fontsize=8)
plt.tight_layout(); plt.show()
print({k: (int(y.sum()), len(y)) for k, (X, y) in esc.items()})
````

---
**Lectura.** (1) Con 20 000 puntos, la covarianza muestral del `Generator` es $\begin{pmatrix}0.995&0.788\\0.788&0.993\end{pmatrix}$, casi la verdadera $\begin{pmatrix}1&0.8\\0.8&1\end{pmatrix}$ (los `assert` pasan con la API clásica **y** con la moderna). (2) **La teoría se cumple:** con datos limpios, $d^2$ sigue una $\chi^2_2$ (KS $p=0.452$) y su cuantil 97.5 % es 7.26 frente al teórico 7.38. (3) Los tres escenarios tienen **36 atípicos de 600** (6 %) y son visualmente distintos: en **«marginal»** los puntos rojos quedan lejos de la nube a lo largo de un eje; en **«cluster»** forman un grupo compacto aparte; y en **«correlación»** los puntos rojos **se mezclan con los azules**: están dentro del mismo rango en $x_1$ y en $x_2$, pero fuera de la elipse inclinada de los datos buenos. Esa dificultad es **intrínseca**: veremos que ni el detector óptimo recupera todos.

## 2. Reglas Univariadas: Z, IQR y MAD (y el Enmascaramiento) 📏

Tres reglas clásicas para una variable $x$:

| Regla | Marca $x_i$ si... | Resumen que usa | Falsas alarmas con datos normales |
|---|---|---|---|
| **Z** | $\lvert x_i-\bar x\rvert/s>3$ | media y desviación (**sensibles**) | $2\,[1-\Phi(3)]\approx0.27\,\%$ |
| **IQR** (Tukey) | $x_i<Q_1-1.5\,\mathrm{IQR}$ o $x_i>Q_3+1.5\,\mathrm{IQR}$ | cuartiles (rompe con ≈ 25 % de contaminación) | $\approx0.70\,\%$ |
| **MAD** (Iglewicz–Hoaglin) | $\lvert M_i\rvert>3.5$, con $M_i=0.6745\,(x_i-\tilde x)/\mathrm{MAD}$ | mediana y $\mathrm{MAD}=\mathrm{mediana}\,\lvert x_i-\tilde x\rvert$ (**robustos**, rompen al 50 %) | $\approx0.05\,\%$ |

La constante $0.6745=\Phi^{-1}(0.75)$ hace que $\mathrm{MAD}/0.6745=1.4826\,\mathrm{MAD}$ estime $\sigma$ cuando los datos son normales. **Enmascaramiento:** los propios atípicos inflan $\bar x$ y $s$, de modo que ninguno parece extremo; las reglas basadas en mediana y cuartiles resisten. Primero verificamos las implementaciones contra `scipy.stats.median_abs_deviation` y medimos las falsas alarmas en un millón de datos limpios; luego creamos un caso con 12 % de contaminación y, por último, **aplicamos las reglas coordenada por coordenada** a los tres escenarios bidimensionales (una observación se marca si **alguna** coordenada se marca).

````python
from sklearn.metrics import precision_score, recall_score, roc_auc_score, average_precision_score

# ---------- Tres reglas univariadas desde cero ----------
def z_flag(x, k=3.0):
    return np.abs((x - x.mean()) / x.std(ddof=1)) > k
def iqr_flag(x, k=1.5):
    q1, q3 = np.percentile(x, [25, 75]); i = q3 - q1
    return (x < q1 - k * i) | (x > q3 + k * i)
def mad_flag(x, k=3.5):
    med = np.median(x); mad = np.median(np.abs(x - med))
    return np.abs(0.6745 * (x - med) / mad) > k            # z modificado de Iglewicz y Hoaglin

# La constante 0.6745 = 1/1.4826: la MAD escalada es un estimador consistente de sigma bajo normalidad
x = np.random.default_rng(0).normal(size=1000)
assert np.allclose(np.abs(0.6745 * (x - np.median(x)) / np.median(np.abs(x - np.median(x)))),
                   np.abs((x - np.median(x)) / stats.median_abs_deviation(x, scale="normal")), rtol=1e-3)
assert np.isclose(1 / 0.6745, stats.norm.ppf(0.75) ** -1, rtol=1e-3)

# ---------- Falsas alarmas con datos limpios normales (n = 1 000 000) ----------
big = np.random.default_rng(1).normal(size=1_000_000)
fa = {nombre: f(big).mean() for nombre, f in [("Z > 3", z_flag), ("IQR 1.5", iqr_flag), ("MAD 3.5", mad_flag)]}
print({k: f"{v:.4%}" for k, v in fa.items()}, "| teórico Z>3:", f"{2 * stats.norm.sf(3):.4%}", "| IQR 1.5:", f"{2 * stats.norm.sf(2.698):.4%}")

# ---------- Enmascaramiento: los atípicos inflan la media y la desviación y se esconden ----------
r = np.random.default_rng(3)
buenos = r.normal(0, 1, 176); malos = r.normal(8, 0.5, 24)               # 12 % de contaminación
xx = np.r_[buenos, malos]; yy = np.r_[np.zeros(176), np.ones(24)]
filas = []
for nombre, f in [("Z > 3", z_flag), ("IQR 1.5", iqr_flag), ("MAD 3.5", mad_flag)]:
    fl = f(xx)
    filas.append({"método": nombre, "marcados": int(fl.sum()), "recall": recall_score(yy, fl), "precisión": precision_score(yy, fl, zero_division=0)})
print(f"media = {xx.mean():.2f}, desviación = {xx.std(ddof=1):.2f} (la limpia es 1) | mediana = {np.median(xx):.2f}, MAD escalada = {stats.median_abs_deviation(xx, scale='normal'):.2f}")
print(pd.DataFrame(filas).set_index("método").round(3))

# ---------- En el plano: lo que ve cada coordenada por separado (escenarios de la sección anterior) ----------
filas = []
for k, (X, y) in esc.items():
    for nombre, f in [("Z>3", z_flag), ("IQR", iqr_flag), ("MAD", mad_flag)]:
        fl = np.any([f(X[:, j]) for j in range(X.shape[1])], axis=0)
        filas.append({"escenario": k, "regla": nombre, "recall": recall_score(y, fl), "falsas alarmas": int((fl & (y == 0)).sum())})
print(pd.DataFrame(filas).pivot(index="regla", columns="escenario", values="recall").round(2))
````

---
**Lectura.** (1) **Falsas alarmas:** con un millón de datos limpios, Z marca **0.280 %** (teórico 0.270 %), IQR **0.715 %** (0.698 %) y MAD **0.050 %**: las reglas son **muy distintas** en severidad, y «3.5» no es «3». (2) **Enmascaramiento:** con 24 valores en 8 (12 %) y 176 buenos $N(0,1)$, la media se corre a 0.99 y la desviación **sube de 1 a 2.82**; el $z$ de los atípicos cae a ≈ 2.5 y **la regla Z no marca ninguno (recall 0)**. IQR marca 25 (recall 1.0, precisión 0.96) y MAD marca 24 (recall y precisión 1.0), porque mediana (0.20) y MAD escalada (1.14) casi no se mueven. (3) **En el plano, solo «marginal» se ve coordenada por coordenada** (recall 1.00 con IQR y MAD, 0.94 con Z). En **«correlación»** el recall es **0.03 (IQR) y 0.00 (MAD y Z)**: las marginales son normales y **ninguna regla univariada puede detectar** que la correlación apunta al revés. En **«cluster»** (centrado en $(3.5,3.5)$) el IQR halla 0.94 pero MAD solo 0.17 y Z 0.08: el grupo está solo ≈ 3.5 desviaciones por encima del centro y la regla estricta lo deja pasar. **Lección:** mirar variable por variable **no basta**; el atípico multivariado vive en las **relaciones**.

## 3. Mahalanobis Clásica y Robusta (MCD) 🎯

La distancia de Mahalanobis mide la lejanía **teniendo en cuenta la forma de la nube**: $d_i^2=(\mathbf x_i-\hat{\boldsymbol\mu})^\top\hat\Sigma^{-1}(\mathbf x_i-\hat{\boldsymbol\mu})$. Con parámetros verdaderos y datos normales, $d^2\sim\chi^2_p$, así que se marca la observación si $d_i^2>\chi^2_{p,\,1-\alpha}$ (por ejemplo, $\alpha=0.025$). El problema es **estimar $\hat{\boldsymbol\mu},\hat\Sigma$ con datos contaminados**: la media y la covarianza muestral tienen **punto de ruptura** 0 (un solo atípico puede arrastrarlas) y los atípicos agrupados **inflan** la elipse y se esconden (**enmascaramiento**).

El **estimador de covarianza de determinante mínimo (MCD)** elige el subconjunto de $h$ observaciones ($n/2\le h\le n$) cuya matriz de covarianza tiene el **determinante más pequeño** y estima $\hat{\boldsymbol\mu},\hat\Sigma$ con él; con $h\approx n/2$ resiste hasta ≈ 50 % de contaminación. Se calcula con el algoritmo *FAST-MCD* (pasos C) y `MinCovDet` de *scikit-learn* añade un factor de consistencia y un reponderado. Implementamos $d^2$ con `einsum`, la contrastamos con SciPy y *scikit-learn* y la comparamos en los tres escenarios con tres versiones: **clásica**, **robusta (MCD)** y **oráculo** (con los parámetros **verdaderos**, el mejor caso posible). Esa tercera fila es clave para no culpar al método de lo que es **solapamiento intrínseco** entre buenos y atípicos.

````python
from scipy.spatial.distance import mahalanobis as sp_mahal
from sklearn.covariance import MinCovDet, EmpiricalCovariance

def mahal2(X, mu, S):
    """Distancia de Mahalanobis al cuadrado de cada fila: (x - mu)' S^{-1} (x - mu)."""
    D = X - mu
    return np.einsum("ij,jk,ik->i", D, np.linalg.inv(S), D)

# Verificación contra SciPy y scikit-learn
X, y = esc["cluster"]
mu_c, S_c = X.mean(0), np.cov(X.T)
d2_c = mahal2(X, mu_c, S_c)
assert np.isclose(np.sqrt(d2_c[0]), sp_mahal(X[0], mu_c, np.linalg.inv(S_c)))
assert np.allclose(mahal2(X, mu_c, np.cov(X.T, bias=True)), EmpiricalCovariance().fit(X).mahalanobis(X), rtol=1e-6)    # sklearn: distancia AL CUADRADO y covarianza con divisor n
assert np.allclose(d2_c * len(X) / (len(X) - 1), mahal2(X, mu_c, np.cov(X.T, bias=True)) )       # divisor n-1 frente a n: factor (n-1)/n
print("Mahalanobis propia == SciPy == scikit-learn (EmpiricalCovariance)")

# ---------- Clásica frente a robusta (MCD): corte con chi² ----------
p = X.shape[1]; corte = stats.chi2.ppf(0.975, p)
def evalua(X, y, nombre):
    mu_c, S_c = X.mean(0), np.cov(X.T)
    mcd = MinCovDet(random_state=0).fit(X)
    out = []
    for met, d2 in [("clásica", mahal2(X, mu_c, S_c)), ("robusta (MCD)", mcd.mahalanobis(X)),
                    ("oráculo (parámetros verdaderos)", mahal2(X, np.zeros(p), ar1(p, 0.8)))]:
        fl = d2 > corte
        out.append({"escenario": nombre, "método": met, "marcados": int(fl.sum()), "recall": recall_score(y, fl),
                    "precisión": precision_score(y, fl, zero_division=0), "falsas alarmas": int((fl & (y == 0)).sum())})
    return out, mcd, (mu_c, S_c)
filas = []
for k, (Xk, yk) in esc.items():
    o, _, _ = evalua(Xk, yk, k); filas += o
print(f"corte chi²(p={p}, 0.975) = {corte:.2f}")
print(pd.DataFrame(filas).set_index(["escenario", "método"]).round(3))

# ---------- Por qué falla la clásica con el grupo desplazado: la covarianza "se infla" ----------
_, mcd, (mu_c, S_c) = evalua(*esc["cluster"], "cluster")
S_true = ar1(2, 0.8)
print(f"determinante de la covarianza: verdadera = {np.linalg.det(S_true):.3f} | clásica = {np.linalg.det(S_c):.3f} | MCD = {np.linalg.det(mcd.covariance_):.3f}")
print(f"centro: verdadero (0, 0) | clásico = {mu_c.round(2)} | MCD = {mcd.location_.round(2)}")

def elipse(ax, mu, S, c, **kw):
    t = np.linspace(0, 2 * np.pi, 200); L = np.linalg.cholesky(S)
    pts = mu[:, None] + np.sqrt(c) * L @ np.vstack([np.cos(t), np.sin(t)]); ax.plot(*pts, **kw)
fig, ax = plt.subplots(figsize=(6.2, 5))
ax.scatter(*X[y == 0].T, s=9, color="#93c5fd"); ax.scatter(*X[y == 1].T, s=26, color="#dc2626", label="contaminación (grupo desplazado)")
elipse(ax, mu_c, S_c, corte, color="#d97706", label="elipse clásica 97.5 %"); elipse(ax, mcd.location_, mcd.covariance_, corte, color="#16a34a", label="elipse robusta (MCD) 97.5 %")
ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$"); ax.legend(fontsize=8, loc="upper left"); ax.set_title("El grupo desplazado arrastra la elipse clásica"); plt.show()
````

---
**Lectura.** (1) **Verificaciones:** la $d^2$ propia coincide con `scipy.spatial.distance.mahalanobis` y con `EmpiricalCovariance().mahalanobis` (que devuelve la distancia **al cuadrado** y usa la covarianza con divisor $n$, no $n-1$: factor $n/(n-1)$). (2) **Marginal:** las tres versiones hallan los 36 (recall 1.00); la clásica tiene precisión **0.947** (2 falsas alarmas) y la MCD **0.643** (20 falsas alarmas): **el MCD tiene más falsas alarmas porque su elipse es más estrecha** (con $\alpha=0.025$ esperaríamos ≈ 14 de 564 y el oráculo también da 19), así que su precisión baja aunque su recall no. (3) **Cluster:** aquí está la lección: la clásica solo halla **25 %** (recall 0.250), mientras que la MCD halla el **100 %** (igual que el oráculo): el grupo desplazado **arrastró la media a (0.27, 0.25)** y **infló la covarianza** (determinante 0.544 frente a 0.360 verdadero), de modo que sus 36 puntos «caben» en la elipse clásica; el MCD (centro (0.05, 0.05), determinante 0.321) **no se deja arrastrar**. (4) **Correlación en $p=2$:** las tres versiones, **incluido el oráculo**, quedan en recall **0.333**: con $\rho=0.8$ y $p=2$ los contaminantes se solapan mucho con la región de los buenos y **ningún** corte con 2.5 % de falsas alarmas puede hacer más. (Con $p=5$, en la sección siguiente, el solapamiento baja y la brecha se abre.) *No es un fallo del método: es un límite de los datos.*

## 4. Isolation Forest, LOF y la Evaluación Honesta 📊

Los dos detectores de aprendizaje automático parten de ideas opuestas:

- **Isolation Forest** aísla cada punto con particiones aleatorias de una variable: los atípicos quedan solos en **pocos cortes**. Con $h(x)$ la profundidad media de aislamiento y $c(n)=2H_{n-1}-2(n-1)/n$ (con $H_k$ el número armónico), el puntaje es $s(x)=2^{-\mathbb E[h(x)]/c(n)}$ (cercano a 1 = atípico). Como corta **una variable a la vez**, detecta extremos marginales y grupos aislados, pero **no la correlación invertida**.
- **LOF** (*Local Outlier Factor*) compara la **densidad local** de un punto con la de sus $k$ vecinos: $\mathrm{LOF}_k(A)=\dfrac{1}{k}\sum_{B\in N_k(A)}\dfrac{\mathrm{lrd}(B)}{\mathrm{lrd}(A)}$. Valores $\gg1$ indican un punto en una zona **menos densa** que sus vecinos. Si el grupo contaminado tiene más de $k$ puntos, cada uno es vecino de otros atípicos y **parece normal**: $k$ debe superar el tamaño de los grupos de atípicos.

**Cómo se evalúa.** Cada método da un **puntaje** de rareza. El **AUC** (área bajo la ROC) es la probabilidad de que un atípico reciba un puntaje mayor que uno bueno (0.5 = azar) y **no depende del umbral**; la **precisión promedio (AP)** resume la curva precisión–recall y es más exigente con clases raras; y **precisión@$m$ / recall@$m$** marcan los $m$ puntajes más altos, con $m$ el número de contaminantes **conocido en la simulación** (en la práctica $m$ no se conoce). La *exactitud* (*accuracy*) se **descarta**: con 5 % de atípicos, no marcar nada da 95 %. Para cada método usamos puntajes comparables: Z y MAD toman el **máximo** de $|z|$ por coordenada, Mahalanobis la $d^2$ y los otros dos sus puntajes nativos. Primero **un experimento** por escenario ($n=1000$, $p=5$) y luego **10 repeticiones** (con semillas distintas) para ver la variabilidad.

````python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor

def puntajes(X, seed=0):
    """Un puntaje de 'rareza' por observación (más alto = más atípico) para cada método."""
    Z = (X - np.median(X, 0)) / stats.median_abs_deviation(X, axis=0, scale="normal")
    mcd = MinCovDet(random_state=seed).fit(X)
    return {"Z máx.": np.abs((X - X.mean(0)) / X.std(0, ddof=1)).max(1),
            "MAD máx.": np.abs(Z).max(1),
            "Mahalanobis clásica": mahal2(X, X.mean(0), np.cov(X.T)),
            "Mahalanobis MCD": mcd.mahalanobis(X),
            "Isolation Forest": -IsolationForest(n_estimators=100, random_state=seed).fit(X).score_samples(X),
            "LOF (k=30)": -LocalOutlierFactor(n_neighbors=30).fit(X).negative_outlier_factor_}

def evalua_puntajes(X, y, seed=0):
    m = int(y.sum()); filas = {}
    for nombre, s in puntajes(X, seed).items():
        top = np.zeros(len(y), bool); top[np.argsort(s)[-m:]] = True                  # marcamos los m más raros (m = contaminación conocida)
        filas[nombre] = {"AUC": roc_auc_score(y, s), "AP": average_precision_score(y, s),
                         "precisión@m": precision_score(y, top), "recall@m": recall_score(y, top)}
    return pd.DataFrame(filas).T

# Un solo experimento: n = 1 000, p = 5, 5 % de contaminación
res_un = {}
for k in ["marginal", "correlacion", "cluster"]:
    X5, y5 = simular(1000, 5, 0.05, k, np.random.default_rng(7))
    res_un[k] = evalua_puntajes(X5, y5)
    print(f"\n--- Escenario «{k}» (n = 1000, p = 5, {int(y5.sum())} atípicos) ---"); print(res_un[k].round(3))
# Verificación: puntajes inocuos (sin información) darían AUC ≈ 0.5
alea = np.random.default_rng(0).random(len(y5))
assert abs(roc_auc_score(y5, alea) - 0.5) < 0.1
assert res_un["cluster"].loc["Mahalanobis MCD", "AUC"] > res_un["cluster"].loc["Mahalanobis clásica", "AUC"] - 1e-9

# LOF y el tamaño del grupo: con k menor que el grupo contaminado (50), los vecinos de cada atípico son del mismo grupo y parece "normal"
X5, y5 = simular(1000, 5, 0.05, "cluster", np.random.default_rng(7))
for k in (30, 60, 100):
    print(f"LOF con k = {k} en el escenario «cluster» (50 atípicos): AUC = {roc_auc_score(y5, -LocalOutlierFactor(n_neighbors=k).fit(X5).negative_outlier_factor_):.3f}")
assert roc_auc_score(y5, -LocalOutlierFactor(n_neighbors=100).fit(X5).negative_outlier_factor_) > 0.9
````

````python
import time
R, n_rep = 10, 400
t0 = time.perf_counter()
todo = []
for k in ["marginal", "correlacion", "cluster"]:
    for r in range(R):
        X5, y5 = simular(n_rep, 5, 0.05, k, np.random.default_rng(1000 + r))
        d = evalua_puntajes(X5, y5, seed=r)["AUC"]
        todo += [{"escenario": k, "método": m, "AUC": v} for m, v in d.items()]
rep = pd.DataFrame(todo)
tab = rep.pivot_table(index="método", columns="escenario", values="AUC", aggfunc=["mean", "std"]).round(3)
media = rep.pivot_table(index="método", columns="escenario", values="AUC").loc[["Z máx.", "MAD máx.", "Mahalanobis clásica", "Mahalanobis MCD", "Isolation Forest", "LOF (k=30)"]]
print(f"{R} repeticiones por escenario (n = {n_rep}, p = 5, 5 % de contaminación); tiempo {time.perf_counter() - t0:.0f} s")
print(tab.round(3).to_string())

fig, ax = plt.subplots(figsize=(10, 4.2))
ancho = 0.25
for j, (k, col) in enumerate(zip(["marginal", "correlacion", "cluster"], ["#93c5fd", "#1e3a8a", "#dc2626"])):
    sd = rep[rep["escenario"] == k].groupby("método")["AUC"].std().loc[media.index]
    ax.bar(np.arange(len(media)) + (j - 1) * ancho, media[k], ancho, yerr=sd, color=col, label=k, capsize=2)
ax.set_xticks(np.arange(len(media)), media.index, rotation=15); ax.axhline(0.5, color="grey", ls="--"); ax.set_ylabel("AUC medio (± desviación estándar)")
ax.set_ylim(0.4, 1.02); ax.legend(title="escenario", ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.3)); ax.set_title("AUC por método y escenario"); plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **Marginal** (fácil): todos los métodos tienen AUC ≈ 1.00, salvo *Isolation Forest* (0.979 en el experimento; precisión@$m$ de 0.60): con 5 coordenadas, un solo eje extremo se «diluye» en el promedio de profundidades de aislamiento. (2) **Correlación invertida** (el difícil): **Z y MAD dan AUC 0.49 (azar)**, como predijo la sección 2; **Mahalanobis MCD** es el mejor (AUC 0.879; precisión@$m$ = 0.70), seguida de LOF (0.872) y Mahalanobis clásica (0.836), y *Isolation Forest* solo llega a 0.690. Con $p=5$ el problema ya **se puede** resolver, pero no del todo: ningún método pasa de ≈ 70 % de precisión@$m$. (3) **Cluster:** la Mahalanobis **clásica cae a AUC 0.913 y precisión@$m$ 0.30** (enmascaramiento), la **MCD sube a 0.992 (0.80)** e *Isolation Forest* a 0.991. **LOF (k = 30) da AUC 0.343, peor que el azar:** el grupo tiene 50 puntos, más que $k=30$, y cada uno es vecino de otros atípicos; con $k=60$ el AUC es 0.661 y con **$k=100$, 0.994**: *el parámetro $k$ debe superar el tamaño del grupo contaminado*. (4) **Con 10 repeticiones ($n=400$, 20 atípicos por réplica)** el panorama se confirma con su variabilidad: en «correlación», MCD 0.828 ± 0.042, LOF 0.795, clásica 0.781, IF 0.648 y Z/MAD ≈ 0.51; en «cluster», MCD 0.988, clásica 0.921 ± 0.017, y LOF **0.987** (con 20 atípicos, menos que $k=30$, funciona: el mismo método, **otra conclusión** por el tamaño del grupo). Las barras de error son desviaciones estándar entre réplicas y las diferencias de 0.01-0.03 **no** son concluyentes con 10 réplicas. **Ningún detector gana siempre:** depende del *tipo* de atípico, y eso no se sabe de antemano; por eso se prueba con simulaciones que imiten los atípicos plausibles del problema.

## 5. El Equivalente en R y Cómo Decidir qué Hacer con un Atípico 🧭

**Equivalente en R** *(bloque de texto, **no se ejecuta aquí**; los nombres de funciones se citan de memoria, verificar en la documentación de `MASS`, `stats` y `robustbase`)*. El libro de la asignatura simula con `MASS::mvrnorm` y calcula distancias con `mahalanobis`:

```r
library(MASS)
set.seed(42)
Sigma <- matrix(c(1, 0.8, 0.8, 1), nrow = 2)
X <- mvrnorm(n = 1000, mu = c(0, 0), Sigma = Sigma)     # equivale a rng.multivariate_normal(mu, Sigma, 1000)

# Mahalanobis clasica (OJO: mahalanobis() devuelve la distancia AL CUADRADO)
d2 <- mahalanobis(X, center = colMeans(X), cov = cov(X))
corte <- qchisq(0.975, df = 2)                           # equivale a stats.chi2.ppf(0.975, 2)
atipicos <- which(d2 > corte)

# Mahalanobis robusta con MCD
rob <- MASS::cov.rob(X, method = "mcd")                  # alternativa: robustbase::covMcd(X)
d2_rob <- mahalanobis(X, center = rob$center, cov = rob$cov)
atipicos_rob <- which(d2_rob > corte)
```

| Tarea | Python (este cuaderno) | R |
|---|---|---|
| Simular normal multivariada | `rng.multivariate_normal(mu, Sigma, n)` | `MASS::mvrnorm(n, mu, Sigma)` |
| Distancia de Mahalanobis al cuadrado | `mahal2(X, mu, S)` / `EmpiricalCovariance().mahalanobis` | `mahalanobis(x, center, cov)` |
| Corte con $\chi^2$ | `stats.chi2.ppf(q, p)` | `qchisq(q, df = p)` |
| Covarianza robusta (MCD) | `MinCovDet().fit(X)` | `MASS::cov.rob(X, method = "mcd")` |
| MAD escalada | `stats.median_abs_deviation(x, scale="normal")` | `mad(x)` (ya incluye 1.4826) |

Las **secuencias aleatorias de R y de NumPy no coinciden** aunque se use la «misma» semilla (generadores y algoritmos de factorización distintos): se comparan **distribuciones**, no números individuales.

**Detectar no es decidir.** Marcar un dato **no** autoriza a borrarlo. Una regla práctica:

1. **Investiga la causa:** error de digitación o de unidades (cuaderno 04: el $\times10$ del área sembrada), medición defectuosa, **o un caso real y raro** (una finca realmente enorme). Los primeros se corrigen; el último **es información**.
2. **Documenta** (bitácora) qué se marcó, con qué regla y qué se hizo; reporta el resultado **con y sin** los datos marcados (análisis de sensibilidad).
3. **Prefiere métodos robustos** (mediana, MAD, regresión robusta, MCD) a la eliminación; usa **transformaciones** (el logaritmo de `ventas_mm`, práctica 2 del cuaderno 04) cuando la asimetría es la causa.
4. **Conoce el costo de las falsas alarmas:** con 1 000 000 de datos normales, Z>3 marca 2 800 que no tienen nada de raro.
5. **Evalúa tu detector** con simulaciones que imiten el tipo de contaminación que esperas, no con una sola métrica.

|---|---|
| **Simulación** | $X\sim(1-\varepsilon)N_p(\mu,\Sigma)+\varepsilon H$ con `multivariate_normal`; con datos limpios $d^2\sim\chi^2_p$ (KS $p=0.452$). La verdad conocida permite **evaluar** detectores |
| **Reglas univariadas** | Z (0.28 % de falsas alarmas), IQR (0.72 %) y MAD (0.05 %) son muy distintas en severidad. **Enmascaramiento:** con 12 % de contaminación la desviación pasa de 1 a 2.82 y Z halla 0 de 24; MAD halla los 24 |
| **Límite univariado** | Con la correlación invertida, IQR/MAD/Z tienen recall 0.03/0.00/0.00: el atípico multivariado vive en las **relaciones** |

| **Evaluación** | AUC, AP y precisión/recall@$m$; nunca *accuracy*. Repetir la simulación: ningún método gana siempre y las diferencias pequeñas no son concluyentes |
