## 1. La Libreta: Pegar Listas sin Contar Doble 📒

Tenemos **dos listas inventadas**: una con **150 fincas** (cada una con su vereda) y otra con las **cosechas** (cada finca cosechó una o dos veces, con el abono usado y las toneladas). Para ver las toneladas por vereda hay que **pegar** las listas por el **número de finca**. Pero ojo con la trampa: si una finca está **anotada dos veces** en la lista de fincas, **todas sus cosechas se cuentan doble**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(5)

# Lista 1: 150 fincas INVENTADAS (una fila por finca)
fincas = pd.DataFrame({"id": np.arange(1, 151),
                       "vereda": rng.choice(["La Esperanza", "El Retiro", "Alto Bonito"], 150)})
# Lista 2: cosechas (cada finca cosechó 1 o 2 veces)
filas = [(i, 1 + (rng.random() < 0.5)) for i in fincas["id"]]
cosechas = pd.DataFrame([(i, s) for i, k in filas for s in range(1, k + 1)], columns=["id", "semestre"])
cosechas["abono_kg"] = rng.gamma(6, 35, len(cosechas)).round(0)
cosechas["toneladas"] = (8 + 0.012 * cosechas["abono_kg"] + rng.normal(0, 1.5, len(cosechas))).round(2)       # INVENTADO
print("Fincas:", len(fincas), "| cosechas:", len(cosechas))

# Pegar las dos listas por el número de finca
todo = cosechas.merge(fincas, on="id")
print("Filas después de pegar:", len(todo), "(deben ser las mismas que las cosechas)")

# Trampa: si por error una finca está repetida en la lista 1, sus cosechas se cuentan DOBLE
fincas_mal = pd.concat([fincas, fincas.iloc[:10]])           # 10 fincas anotadas dos veces por error
todo_mal = cosechas.merge(fincas_mal, on="id")
print("Con 10 fincas repetidas salen", len(todo_mal), "filas: sobran", len(todo_mal) - len(todo), "(cosechas contadas dos veces)")

# Promedio de toneladas por vereda
print(todo.groupby("vereda")["toneladas"].agg(["count", "mean"]).round(2))
````

---
**Lección:** al pegar bien, **las 229 cosechas siguen siendo 229** (ni una más, ni una menos): esa es la **primera revisión** que se hace siempre. Con **10 fincas anotadas dos veces por error**, salieron **242 filas**: **13 cosechas contadas doble**, sin que el computador avisara de nada. Por eso se cuentan las filas antes y después de pegar. Y fíjate en el resumen: las tres veredas rinden casi lo mismo (**≈ 10.5 toneladas**) porque en estos datos inventados la vereda **no importa**; lo que sí importa es el abono, y eso lo veremos ahora.

## 2. El Contador Explica, el Adivino Predice 🧮🔮

Separamos las fincas en dos grupos: **100 para estudiar** y **50 para el examen** (el adivino **nunca las vio**; y una finca jamás está en los dos grupos, así que no hay trampa). El **contador** estudia cuánto aporta el abono y dice **cuánto, con un margen de error**. El **adivino** usa esa misma idea para predecir las fincas del examen y lo comparamos con la regla perezosa de **«adivinar siempre el promedio»**.

````python
import statsmodels.formula.api as smf
from sklearn.linear_model import LinearRegression
from sklearn.dummy import DummyRegressor

# Las fincas para estudiar (100 primeras) y las del examen (50 últimas): una finca nunca está en las dos
estudio = todo[todo["id"] <= 100]
examen = todo[todo["id"] > 100]

# El "contador" (statsmodels): explica cuánto aporta el abono y con qué margen de error
modelo = smf.ols("toneladas ~ abono_kg", data=estudio).fit()
b, (lo, hi) = modelo.params["abono_kg"], modelo.conf_int().loc["abono_kg"]
print(f"Por cada 100 kg más de abono: {100 * b:+.2f} toneladas (margen: de {100 * lo:+.2f} a {100 * hi:+.2f})")

# El "adivino" (scikit-learn): predice fincas que NO vio, y lo comparamos con adivinar siempre el promedio
X_e, y_e = estudio[["abono_kg"]], estudio["toneladas"]
X_x, y_x = examen[["abono_kg"]], examen["toneladas"]
adivino = LinearRegression().fit(X_e, y_e)
promedio = DummyRegressor().fit(X_e, y_e)
err = lambda m: np.abs(y_x - m.predict(X_x)).mean()
print(f"Error promedio en el examen: adivino = {err(adivino):.2f} t | 'siempre el promedio' = {err(promedio):.2f} t")
# Los dos usan la misma recta: el contador y el adivino coinciden
assert np.isclose(adivino.coef_[0], b)

plt.figure(figsize=(7, 4.2))
plt.scatter(X_x, y_x, s=14, color="#1e3a8a", label="fincas del examen")
xs = np.linspace(X_x["abono_kg"].min(), X_x["abono_kg"].max(), 50)
plt.plot(xs, adivino.predict(pd.DataFrame({"abono_kg": xs})), color="#dc2626", label="recta del adivino")
plt.axhline(y_e.mean(), color="grey", ls="--", label="siempre el promedio")
plt.xlabel("abono (kg)"); plt.ylabel("toneladas"); plt.legend(); plt.title("Predecir fincas que no vio"); plt.show()
````

---
**Lección:** el contador dice que **cada 100 kg más de abono** dan **+1.39 toneladas**, con un margen de **+1.08 a +1.70**. (El valor que pusimos al inventar los datos era **+1.2**: cae dentro del margen.) El margen es importante: **un número sin su margen de error engaña**. En el examen, el adivino se equivoca en promedio **1.26 toneladas** y «adivinar siempre el promedio» **1.50**: el adivino **mejora**, pero **no hace milagros** (el azar de cada cosecha no se puede predecir). Y el contador y el adivino **usan la misma recta** (el `assert` lo comprueba): la diferencia es **para qué la usan**, uno para **explicar** y otro para **predecir**.
