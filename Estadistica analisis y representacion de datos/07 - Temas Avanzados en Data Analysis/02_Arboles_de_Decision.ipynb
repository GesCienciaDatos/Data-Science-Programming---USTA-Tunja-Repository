## 1. Impureza, Ganancia y el Algoritmo CART 🌳

Un nodo $t$ contiene $N_t$ observaciones. Su **impureza** $I(t)$ mide cuán mezcladas están las clases:

| Criterio | Fórmula ($p_k$ = proporción de la clase $k$ en el nodo) | Nodo puro | Máximo (2 clases) |
|---|---|---|---|
| **Gini** | $G=1-\sum_k p_k^2$ | 0 | 0.5 |
| **Entropía** | $H=-\sum_k p_k\log_2 p_k$ | 0 | 1 bit |
| **Varianza** (regresión, MSE) | $\frac{1}{N_t}\sum_{i\in t}(y_i-\bar y_t)^2$ | 0 | — |

**Algoritmo CART** (*recursive binary splitting*): en cada nodo se prueban **todos** los cortes $x_j\le s$ (con $s$ en los puntos medios entre valores consecutivos) y se elige el de mayor **ganancia**
$$\Delta=I(t)-\frac{N_L}{N_t}I(t_L)-\frac{N_R}{N_t}I(t_R),$$
y se repite en cada hijo hasta una regla de parada (profundidad máxima, mínimo de observaciones por hoja o nodo puro). Es un algoritmo **voraz**: elige el mejor corte *local*, sin garantía de que el árbol completo sea óptimo. La predicción de una hoja es la **proporción de clases** (o la **media** en regresión). Un efecto práctico: cada corte involucra **una sola variable y un umbral**, así que el árbol es invariante a transformaciones monótonas (no requiere escalar).

Generamos 500 parcelas **sintéticas**: la clase `enferma` depende de la humedad y del pH (con **interacción**: el pH solo importa si la humedad supera 50) y de la temperatura, más ruido. Implementamos la clase `Arbol` (clasificación y regresión) y la contrastamos con scikit-learn **predicción por predicción** y también en la importancia de variables.

````python
# ============================================================
# Configuracion inicial y arbol CART desde cero (clasificacion y regresion)
# ============================================================
# !pip install -q numpy pandas scipy scikit-learn matplotlib

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, export_text

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scikit-learn {__import__('sklearn').__version__}")

# Datos SINTETICOS: 500 parcelas; la clase "enferma" depende de la humedad y el pH (con interaccion) y de la temperatura, mas ruido
n = 500
parcelas = pd.DataFrame({"humedad": rng.uniform(20, 90, n), "ph": rng.uniform(4.5, 7.5, n),
                         "temperatura": rng.uniform(8, 24, n), "altitud": rng.uniform(2400, 3200, n)})
z = 0.06 * (parcelas.humedad - 55) + 1.2 * (parcelas.ph - 6.0) * (parcelas.humedad > 50) + 0.15 * (parcelas.temperatura - 16) + rng.normal(0, 0.6, n)
parcelas["enferma"] = (z > 0.4).astype(int)
Xp, yp = parcelas[["humedad", "ph", "temperatura", "altitud"]].values, parcelas["enferma"].values
print("Proporcion de parcelas enfermas:", yp.mean().round(3))

# --- Impureza de un nodo ---
def gini(y):
    p = np.bincount(y, minlength=2) / len(y)
    return 1 - np.sum(p ** 2)

def entropia(y):
    p = np.bincount(y, minlength=2) / len(y)
    p = p[p > 0]
    return -np.sum(p * np.log2(p))

# Ejemplo a mano: nodo con 8 enfermas y 2 sanas
nodo = np.array([1] * 8 + [0] * 2)
print(f"Nodo 8/2: Gini = {gini(nodo):.3f} (= 1 - 0.8^2 - 0.2^2) | Entropia = {entropia(nodo):.3f} bits | Gini de un nodo puro = {gini(np.ones(5, int)):.1f}")
assert np.isclose(gini(nodo), 0.32) and np.isclose(gini(np.array([0, 1])), 0.5) and np.isclose(entropia(np.array([0, 1])), 1.0)

# --- CART desde cero ---
class Arbol:
    """Arbol binario CART. tarea='clf' (Gini/entropia, 2 clases 0/1) o 'reg' (varianza = MSE). Cortes por X[:, j] <= umbral (punto medio)."""
    def __init__(self, tarea="clf", criterio="gini", max_depth=None, min_samples_leaf=1):
        self.tarea, self.criterio, self.max_depth, self.msl = tarea, criterio, max_depth, min_samples_leaf

    def _imp(self, y):
        if self.tarea == "reg":
            return y.var()
        return gini(y) if self.criterio == "gini" else entropia(y)

    def _mejor_corte(self, X, y):
        n_, mejor = len(y), (0.0, None, None)                          # (disminucion ponderada, variable, umbral)
        imp_padre = self._imp(y)
        for j in range(X.shape[1]):
            orden = np.argsort(X[:, j], kind="stable"); xs, ys = X[orden, j], y[orden]
            for i in range(self.msl, n_ - self.msl + 1):               # corte entre las posiciones i-1 e i
                if xs[i - 1] == xs[i]:
                    continue
                hijos = (i * self._imp(ys[:i]) + (n_ - i) * self._imp(ys[i:])) / n_
                if imp_padre - hijos > mejor[0] + 1e-12:
                    mejor = (imp_padre - hijos, j, (xs[i - 1] + xs[i]) / 2)
        return mejor

    def fit(self, X, y):
        X, y = np.asarray(X, float), np.asarray(y)
        self.n_total, self.d = len(y), X.shape[1]
        self.nodos = []                                                # cada nodo: dict con j, umbral, izq, der, n, imp, valor
        self._crece(X, y, 0)
        self.importancias_ = self._importancias()
        return self

    def _crece(self, X, y, prof):
        idx = len(self.nodos)
        valor = y.mean() if self.tarea == "reg" else np.bincount(y, minlength=2) / len(y)
        nodo = dict(j=-1, umbral=None, izq=-1, der=-1, n=len(y), imp=self._imp(y), valor=valor, ganancia=0.0)
        self.nodos.append(nodo)
        if (self.max_depth is not None and prof >= self.max_depth) or len(y) < 2 * self.msl or nodo["imp"] < 1e-12:
            return idx
        gan, j, u = self._mejor_corte(X, y)
        if j is None:
            return idx
        mask = X[:, j] <= u
        nodo.update(j=j, umbral=u, ganancia=gan * len(y) / self.n_total)   # disminucion de impureza ponderada por N_t / N
        nodo["izq"] = self._crece(X[mask], y[mask], prof + 1)
        nodo["der"] = self._crece(X[~mask], y[~mask], prof + 1)
        return idx

    def _importancias(self):
        imp = np.zeros(self.d)
        for nd in self.nodos:
            if nd["j"] >= 0:
                imp[nd["j"]] += nd["ganancia"]
        return imp / imp.sum() if imp.sum() > 0 else imp

    def _hoja(self, x):
        i = 0
        while self.nodos[i]["j"] >= 0:
            i = self.nodos[i]["izq"] if x[self.nodos[i]["j"]] <= self.nodos[i]["umbral"] else self.nodos[i]["der"]
        return i

    def predict_proba(self, X):
        return np.array([self.nodos[self._hoja(x)]["valor"] for x in np.asarray(X, float)])

    def predict(self, X):
        P = self.predict_proba(X)
        return P if self.tarea == "reg" else P.argmax(1)

    @property
    def n_hojas(self):
        return sum(nd["j"] < 0 for nd in self.nodos)

# Verificacion contra scikit-learn (variables continuas: sin empates entre cortes)
Xtr, ytr, Xte, yte = Xp[:350], yp[:350], Xp[350:], yp[350:]
for crit in ("gini", "entropy"):
    mio = Arbol("clf", crit, max_depth=4, min_samples_leaf=5).fit(Xtr, ytr)
    sk = DecisionTreeClassifier(criterion=crit, max_depth=4, min_samples_leaf=5, random_state=SEED).fit(Xtr, ytr)
    assert np.array_equal(mio.predict(Xp), sk.predict(Xp)) and mio.n_hojas == sk.get_n_leaves()
    assert np.allclose(mio.importancias_, sk.feature_importances_, atol=1e-6)
    print(f"[{crit:7s}] arbol propio == DecisionTreeClassifier | hojas: {mio.n_hojas} | exactitud train/test = {(mio.predict(Xtr) == ytr).mean():.3f} / {(mio.predict(Xte) == yte).mean():.3f}")

arbol = Arbol("clf", "gini", max_depth=3, min_samples_leaf=5).fit(Xtr, ytr)
print("\nReglas del arbol propio (profundidad 3):")
nombres = ["humedad", "ph", "temperatura", "altitud"]
def reglas(a, i=0, sang=""):
    nd = a.nodos[i]
    if nd["j"] < 0:
        print(f"{sang}-> P(enferma) = {nd['valor'][1]:.2f}  (n = {nd['n']}, gini = {nd['imp']:.2f})")
        return
    print(f"{sang}si {nombres[nd['j']]} <= {nd['umbral']:.2f}:"); reglas(a, nd["izq"], sang + "    ")
    print(f"{sang}si no ({nombres[nd['j']]} > {nd['umbral']:.2f}):"); reglas(a, nd["der"], sang + "    ")
reglas(arbol)
````

## 2. Árboles de Regresión y el Compromiso Sesgo–Varianza 📈

En regresión la impureza es la **varianza** del nodo y la ganancia de un corte es la **reducción de la suma de cuadrados** dentro de los nodos (dividida por $N$). La predicción es la **media de la hoja**, así que la curva ajustada es una **escalera**. Con datos sintéticos de rendimiento frente a humedad del suelo verificamos nuestro árbol contra `DecisionTreeRegressor`.

**Sesgo y varianza.** Para un punto $x_0$, el error cuadrático esperado de un predictor $\hat f$ entrenado con muestras aleatorias se descompone en $\mathbb{E}\big[(y_0-\hat f(x_0))^2\big]=\underbrace{\big(\mathbb{E}\hat f(x_0)-f(x_0)\big)^2}_{\text{sesgo}^2}+\underbrace{\operatorname{Var}\big(\hat f(x_0)\big)}_{\text{varianza}}+\underbrace{\sigma^2}_{\text{ruido}}$. Un árbol **poco profundo** tiene sesgo alto (escalera muy gruesa) y un árbol **muy profundo** tiene varianza alta (se adapta al ruido de cada muestra). Como conocemos la función verdadera, estimamos ambos términos **simulando 300 conjuntos de entrenamiento** nuevos para cada profundidad.

````python
# ============================================================
# 2. Arbol de REGRESION (MSE) desde cero y el compromiso sesgo-varianza
# ============================================================
# Datos SINTETICOS: rendimiento (t/ha) segun la humedad del suelo (relacion no lineal con tope) + ruido
def f_real(x):
    return 12 + 6 * np.sin((x - 20) / 15) - 0.002 * (x - 55) ** 2

xr = np.sort(rng.uniform(20, 90, 200))
yr = f_real(xr) + rng.normal(0, 1.2, 200)

mio_r = Arbol("reg", max_depth=3, min_samples_leaf=5).fit(xr[:, None], yr)
sk_r = DecisionTreeRegressor(max_depth=3, min_samples_leaf=5, random_state=SEED).fit(xr[:, None], yr)
malla = np.linspace(20, 90, 400)[:, None]
assert np.allclose(mio_r.predict(malla), sk_r.predict(malla)) and mio_r.n_hojas == sk_r.get_n_leaves()
assert np.allclose(mio_r.importancias_, sk_r.feature_importances_)
print(f"Arbol de regresion propio == DecisionTreeRegressor (profundidad 3, {mio_r.n_hojas} hojas). Prediccion = media de la hoja (escalera).")

# Impureza de regresion = varianza; la ganancia de un corte = reduccion de la suma de cuadrados dentro de los nodos
y_ = yr
corte = mio_r.nodos[0]["umbral"]
sse = lambda v: np.sum((v - v.mean()) ** 2)
red_sse = sse(y_) - sse(y_[xr <= corte]) - sse(y_[xr > corte])
assert np.isclose(red_sse / len(y_), mio_r.nodos[0]["ganancia"])
print(f"Raiz: corte en humedad <= {corte:.1f} | reduccion de SSE = {red_sse:.1f} (= N x disminucion de varianza)")

# Sesgo y varianza por simulacion: 300 conjuntos de entrenamiento nuevos (n = 100) para cada profundidad
xt = np.linspace(22, 88, 120)
verdad = f_real(xt)
filas = []
for prof in (1, 2, 3, 4, 6, 8, 12):
    P = np.empty((300, len(xt)))
    for b in range(300):
        xb = rng.uniform(20, 90, 100)
        P[b] = DecisionTreeRegressor(max_depth=prof, random_state=0).fit(xb[:, None], f_real(xb) + rng.normal(0, 1.2, 100)).predict(xt[:, None])
    sesgo2, var = np.mean((P.mean(0) - verdad) ** 2), np.mean(P.var(0))
    filas.append({"profundidad": prof, "sesgo2": sesgo2, "varianza": var, "ECM_sin_ruido": sesgo2 + var})
bv = pd.DataFrame(filas).set_index("profundidad")
print("\nSesgo^2 + varianza (el ruido irreducible 1.44 se suma aparte):\n", bv.round(3).to_string())
assert bv["sesgo2"].iloc[-1] < 0.01 * bv["sesgo2"].iloc[0] and bv["varianza"].iloc[-1] > 1.5 * bv["varianza"].loc[4] and bv["ECM_sin_ruido"].idxmin() in (3, 4)

fig, ax = plt.subplots(1, 3, figsize=(15, 4.3))
ax[0].scatter(xr, yr, s=10, color="grey")
for prof, c in zip((1, 3, 12), PALETA[:3]):
    ax[0].plot(malla, DecisionTreeRegressor(max_depth=prof, random_state=0).fit(xr[:, None], yr).predict(malla), color=c, label=f"profundidad {prof}")
ax[0].plot(malla, f_real(malla[:, 0]), "k--", lw=1, label="funcion real"); ax[0].legend(fontsize=8); ax[0].set_xlabel("humedad"); ax[0].set_title("Escaleras de distinta profundidad")
ax[1].plot(bv.index, bv["sesgo2"], "o-", color=PALETA[0], label="sesgo^2"); ax[1].plot(bv.index, bv["varianza"], "o-", color=PALETA[1], label="varianza")
ax[1].plot(bv.index, bv["ECM_sin_ruido"], "o-", color="k", label="suma"); ax[1].legend(); ax[1].set_xlabel("profundidad"); ax[1].set_title("Sesgo vs varianza")
ax[2].step(malla[:, 0], mio_r.predict(malla), color=PALETA[2], where="post"); ax[2].scatter(xr, yr, s=8, color="grey"); ax[2].set_title("Arbol propio (profundidad 3)"); ax[2].set_xlabel("humedad")
plt.tight_layout(); plt.show()
````

**Lectura.** (1) El árbol de regresión propio **coincide** con `DecisionTreeRegressor` (8 hojas con profundidad 3 y `min_samples_leaf=5`) en toda la malla y en la importancia. La ganancia de la raíz (corte en humedad ≤ 69.2) es exactamente la reducción de la suma de cuadrados dividida por $N$ (3 133.4 / 200). (2) **Sesgo–varianza:** el sesgo² cae de **3.72** (profundidad 1) a **0.095** (profundidad 4) y a ≈ 0.006 desde la 6; la varianza no es monótona: **1.205** con un solo corte (la *posición* del único corte cambia mucho entre muestras), baja a **0.744** con profundidad 4 y vuelve a subir hasta **1.433** con profundidad 12. El error total sin ruido es mínimo con profundidad **4** (0.839), mientras que con profundidad 12 (1.439) es casi el doble: el árbol profundo tiene sesgo casi cero **pero** paga con varianza. ## 3. Sobreajuste y Poda por Complejidad de Costo ✂️

Un árbol sin límite crece hasta que cada hoja es pura y **memoriza** el entrenamiento. Hay dos formas de controlarlo: **pre-poda** (limitar `max_depth`, `min_samples_leaf`) y **post-poda**: crecer un árbol grande y recortarlo. La post-poda de CART (**cost-complexity pruning**) penaliza el número de hojas $|\tilde T|$ con un parámetro $\alpha\ge 0$:
$$R_\alpha(T)=R(T)+\alpha\,|\tilde T|,\qquad R(T)=\sum_{t\in\tilde T}\frac{N_t}{N}\,I(t).$$
Para cada nodo interno $t$ se calcula su **alfa efectivo** $\alpha_{\mathrm{ef}}(t)=\dfrac{R(t)-R(T_t)}{|\tilde T_t|-1}$ (cuánto aumenta el costo por hoja ahorrada al colapsar la rama $T_t$ en una hoja) y se poda el nodo con **menor** alfa efectivo, el **eslabón más débil**. Repetido hasta la raíz, produce una **secuencia anidada de subárboles** $T_0\supset T_1\supset\dots$ con $\alpha$ creciente. Se elige $\alpha$ por validación cruzada; la **regla del 1 error estándar** prefiere el árbol **más simple** cuya exactitud CV esté a menos de 1 EE de la mejor. Implementamos el camino de poda desde cero y lo contrastamos con `cost_complexity_pruning_path`.

````python
# ============================================================
# 3. Sobreajuste y poda por complejidad de costo (cost-complexity pruning) desde cero
# ============================================================
from sklearn.model_selection import StratifiedKFold, cross_val_score

# Un arbol sin limite memoriza: curva de profundidad con validacion cruzada
skf = StratifiedKFold(5, shuffle=True, random_state=SEED)
filas = []
for prof in (1, 2, 3, 4, 5, 6, 8, 10, None):
    t_ = DecisionTreeClassifier(max_depth=prof, random_state=SEED)
    filas.append({"max_depth": "sin limite" if prof is None else prof, "train": t_.fit(Xp, yp).score(Xp, yp), "cv": cross_val_score(t_, Xp, yp, cv=skf).mean(),
                  "hojas": t_.get_n_leaves()})
prof_df = pd.DataFrame(filas).set_index("max_depth")
print(prof_df.round(3).to_string())

# --- Poda: eliminar el "eslabon mas debil". Costo de un subarbol T: R_a(T) = R(T) + a * |hojas(T)|, R(T) = suma de N_t/N * impureza de las hojas ---
def camino_poda(arbol):
    """Devuelve (alphas, impurezas totales, n.o de hojas) de la secuencia de subarboles anidados por poda del eslabon mas debil."""
    N = arbol.nodos
    n0 = N[0]["n"]
    R = lambda i: N[i]["n"] / n0 * N[i]["imp"]
    activo = [nd["j"] >= 0 for nd in N]                       # nodos que siguen siendo internos
    def hojas(i):
        return [i] if not activo[i] else hojas(N[i]["izq"]) + hojas(N[i]["der"])
    def internos(i):                                          # nodos internos alcanzables desde la raiz
        return [] if not activo[i] else [i] + internos(N[i]["izq"]) + internos(N[i]["der"])
    alphas, imps, nh = [0.0], [sum(R(i) for i in hojas(0))], [len(hojas(0))]
    while activo[0]:
        cand = {i: (R(i) - sum(R(h) for h in hojas(i))) / (len(hojas(i)) - 1) for i in internos(0)}
        a = min(cand.values())
        for i, v in cand.items():                             # se podan TODOS los nodos con alfa efectivo minimo (empates incluidos)
            if np.isclose(v, a, atol=1e-12):
                activo[i] = False
        alphas.append(max(a, alphas[-1])); imps.append(sum(R(i) for i in hojas(0))); nh.append(len(hojas(0)))
    return np.array(alphas), np.array(imps), np.array(nh)

grande = Arbol("clf", "gini", min_samples_leaf=1).fit(Xtr, ytr)                    # arbol completo (sobreajustado)
al, im, nh = camino_poda(grande)
sk_g = DecisionTreeClassifier(random_state=SEED).fit(Xtr, ytr)
ruta = sk_g.cost_complexity_pruning_path(Xtr, ytr)
assert np.allclose(np.unique(np.round(al, 10)), np.unique(np.round(ruta.ccp_alphas, 10)), atol=1e-9)
assert np.allclose(np.unique(np.round(im, 10)), np.unique(np.round(ruta.impurities, 10)), atol=1e-9)
print(f"\nArbol completo: {grande.n_hojas} hojas. Camino de poda propio == cost_complexity_pruning_path: {len(np.unique(np.round(al, 10)))} valores distintos de alfa")

# Elegir alfa por validacion cruzada (en el entrenamiento) con la regla del 1 error estandar
alphas = np.unique(np.round(ruta.ccp_alphas, 8))
alphas = alphas[alphas < alphas.max()]
cv_scores = np.array([cross_val_score(DecisionTreeClassifier(ccp_alpha=a, random_state=SEED), Xtr, ytr, cv=skf) for a in alphas])
cv_mean, cv_se = cv_scores.mean(1), cv_scores.std(1, ddof=1) / np.sqrt(cv_scores.shape[1])
i_best = cv_mean.argmax()
alpha_best = alphas[i_best]
alpha_1se = alphas[np.where(cv_mean >= cv_mean[i_best] - cv_se[i_best])[0].max()]
for nombre, a in (("alfa = 0 (sin poda)", 0.0), ("alfa de maxima exactitud CV", alpha_best), ("alfa regla 1-SE", alpha_1se)):
    m_ = DecisionTreeClassifier(ccp_alpha=a, random_state=SEED).fit(Xtr, ytr)
    print(f"{nombre:30s} alfa = {a:.5f} | hojas = {m_.get_n_leaves():3d} | exactitud CV = {cv_mean[np.argmin(np.abs(alphas - a))]:.3f} | test = {m_.score(Xte, yte):.3f}")

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
ax[0].plot(range(len(prof_df)), prof_df["train"], "o-", color=PALETA[1], label="entrenamiento"); ax[0].plot(range(len(prof_df)), prof_df["cv"], "o-", color=PALETA[0], label="validacion cruzada")
ax[0].set_xticks(range(len(prof_df))); ax[0].set_xticklabels(prof_df.index.astype(str), fontsize=8); ax[0].legend(); ax[0].set_title("Profundidad: el sobreajuste"); ax[0].set_xlabel("max_depth")
ax[1].step(al, nh, where="post", color=PALETA[2]); ax[1].set_xlabel("alfa"); ax[1].set_ylabel("hojas"); ax[1].set_title("Camino de poda: hojas vs alfa")
ax[2].errorbar(alphas, cv_mean, yerr=cv_se, fmt="o-", color=PALETA[0], ms=3); ax[2].axvline(alpha_best, color="k", ls="--", label="maximo"); ax[2].axvline(alpha_1se, color=PALETA[1], ls="--", label="1-SE")
ax[2].set_xlabel("alfa"); ax[2].set_ylabel("exactitud CV"); ax[2].legend(); ax[2].set_title("Eleccion de alfa")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **El sobreajuste es evidente:** la exactitud en entrenamiento sube a 1.000 con profundidad 10, mientras la de validación cruzada se estanca en ≈ 0.90 (profundidad 6–8: 0.900 y 0.902) y no mejora más. (2) Nuestro camino de poda **coincide** con `cost_complexity_pruning_path` (21 valores distintos de $\alpha$ y las mismas impurezas totales), y también el número de hojas de cada subárbol. (3) **Honestidad sobre el efecto de la poda:** el árbol completo (31 hojas) tiene exactitud CV 0.860 y 0.873 en prueba; el podado por CV ($\alpha=0.0132$, **5 hojas**) tiene 0.877 en CV y **0.867 en prueba**. Es decir: la poda mejoró la exactitud en CV (+0.017), pero en prueba **no mejora** (−0.006, con 150 observaciones): la ventaja es **parsimonia e interpretabilidad** (5 hojas frente a 31), no una ganancia clara de exactitud. La regla del 1-SE eligió también 5 hojas ($\alpha=0.0185$). (4) Que la exactitud CV de las profundidades 6–8 (≈ 0.90) supere a la del árbol podado a 5 hojas (0.877) se debe a que ahí la CV usa los 500 datos y aquí 350, y a que **no** son comparables directamente; lo relevante es la forma de las curvas. (5) Recomendación: **podar con CV** y quedarse con el árbol pequeño si la exactitud es equivalente.

## 4. ¿Qué Variables Importan? MDI frente a Permutación 🔎

Hay dos formas habituales de medir la importancia de una variable:

- **MDI** (*Mean Decrease in Impurity*, la `feature_importances_` de scikit-learn): $\mathrm{Imp}(j)=\sum_{t:\,\text{se corta por }j}\frac{N_t}{N}\,\Delta_t$, normalizada para que sume 1. Es **gratuita** (sale del entrenamiento), pero se calcula **sobre los datos de entrenamiento**, así que premia a las variables que el árbol **usó para memorizar**, incluidas las que no tienen relación con la clase. Además está **sesgada hacia variables con muchos valores distintos**, que ofrecen más cortes candidatos (Strobl et al., 2007).
- **Importancia por permutación**: se mide cuánto **baja la exactitud en el conjunto de prueba** al barajar al azar una columna (rompiendo su relación con $y$). Es agnóstica al modelo, usa datos no vistos y no premia el ruido; su límite es que, con variables **muy correlacionadas**, repartir o diluir la importancia entre ellas.

Experimento: agregamos a las parcelas dos variables **sin relación alguna con la clase**: `ruido` (500 valores continuos distintos) y `sorteo` (entera de 0 a 4). Además `altitud` ya era irrelevante por construcción. Comparamos las dos medidas con un bosque de 150 árboles y la MDI de un árbol completo.

````python
# ============================================================
# 4. Importancia de variables: MDI (impureza) vs permutacion
# ============================================================
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance

# Agregamos dos variables SIN relacion con la clase: "ruido" (continua, 500 valores distintos) y "sorteo" (entera 0-4)
rg2 = np.random.default_rng(7)
Xr = np.column_stack([Xp, rg2.uniform(0, 1, n), rg2.integers(0, 5, n)])
nom = ["humedad", "ph", "temperatura", "altitud", "ruido", "sorteo"]
Xr_tr, Xr_te = Xr[:350], Xr[350:]

# MDI = suma de las disminuciones de impureza ponderadas por N_t/N (verificada contra sklearn en la seccion 1; aqui hay empates
# entre variables enteras y scikit-learn los rompe al azar, por eso no se exige igualdad exacta).
arb = Arbol("clf", "gini", min_samples_leaf=1).fit(Xr_tr, ytr)                # arbol completo (sin poda)
assert np.isclose(arb.importancias_.sum(), 1.0)

def imp_permutacion(modelo, X, y, rep=20, seed=0):
    """Caida media de la exactitud al barajar UNA columna del conjunto de PRUEBA (rompe su relacion con y)."""
    r = np.random.default_rng(seed); base = (modelo.predict(X) == y).mean(); out = np.zeros((rep, X.shape[1]))
    for j in range(X.shape[1]):
        for b in range(rep):
            Xp_ = X.copy(); Xp_[:, j] = r.permutation(Xp_[:, j])
            out[b, j] = base - (modelo.predict(Xp_) == y).mean()
    return out.mean(0), out.std(0, ddof=1)

rf = RandomForestClassifier(150, random_state=SEED, n_jobs=1).fit(Xr_tr, ytr)
pm, ps = imp_permutacion(rf, Xr_te, yte)
sk_perm = permutation_importance(rf, Xr_te, yte, n_repeats=20, random_state=SEED, n_jobs=1)
assert np.allclose(pm, sk_perm.importances_mean, atol=0.04)
tabi = pd.DataFrame({"MDI arbol": arb.importancias_, "MDI bosque": rf.feature_importances_, "permutacion bosque (test)": pm, "+-": ps}, index=nom)
print("Importancia de variables (humedad, ph y temperatura son reales; 'ruido' y 'sorteo' no):\n", tabi.round(3).to_string())
print(f"\nExactitud del bosque: train = {rf.score(Xr_tr, ytr):.3f} | test = {rf.score(Xr_te, yte):.3f}")
assert rf.feature_importances_[[3, 4]].sum() > 0.10 and np.abs(pm[[3, 4, 5]]).max() < 0.02

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
w = 0.38
ax[0].barh(np.arange(6) - w / 2, rf.feature_importances_, height=w, color=PALETA[0], label="MDI (impureza)")
ax[0].barh(np.arange(6) + w / 2, pm, xerr=ps, height=w, color=PALETA[1], label="permutacion (test)")
ax[0].set_yticks(range(6)); ax[0].set_yticklabels(nom); ax[0].invert_yaxis(); ax[0].legend(); ax[0].set_title("Bosque: dos medidas de importancia")
ax[1].barh(range(6), arb.importancias_, color=PALETA[2]); ax[1].set_yticks(range(6)); ax[1].set_yticklabels(nom); ax[1].invert_yaxis(); ax[1].set_title("MDI de UN arbol completo")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) La MDI de un árbol completo y la del bosque ordenan bien a las tres variables **reales**: humedad (0.46 y 0.44), pH (0.36 y 0.26) y temperatura (0.16 y 0.14). (2) **Pero la MDI le asigna importancia al ruido:** en el bosque, `altitud` (0.066), `ruido` (0.077) y `sorteo` (0.025) suman **0.17** (17 % de la «importancia» total) para variables que no aportan nada, y el bosque tiene exactitud 1.000 en entrenamiento y **0.907 en prueba**: memorizó. (3) La **permutación en prueba** corrige el problema: para las tres variables irrelevantes la caída de exactitud es ≈ 0 (−0.009, −0.007, −0.001, con incertidumbres de ≈ 0.006–0.010 que incluyen el 0), mientras humedad (0.271), pH (0.171) y temperatura (0.076) quedan claramente por encima. Nuestra implementación coincide con `permutation_importance` dentro de la tolerancia del azar. (4) **Reglas de uso:** para *decidir* qué variables importan, usar permutación (idealmente sobre datos no vistos); la MDI sirve como diagnóstico rápido, no como prueba. (5) Una importancia **no** dice nada del *sentido* del efecto ni de causalidad: solo cuánto usa el modelo la variable.

## 5. Inestabilidad del Árbol y Comparación con un Bosque Aleatorio 🌲🌲🌲

Un árbol es un estimador de **alta varianza**: la elección voraz de cada corte hace que un cambio pequeño en los datos (una remuestra) pueda alterar el corte de la raíz y, con él, **toda la estructura** de abajo. Es la razón de ser de los **bosques aleatorios** (Breiman, 2001): promediar muchos árboles entrenados con remuestras *bootstrap* y subconjuntos aleatorios de variables reduce la varianza, a costa de la **interpretabilidad** (un bosque no se lee como reglas).

Con el conjunto *breast_cancer* de scikit-learn (569 tumores, 30 variables, benigno/maligno) hacemos tres cosas: (a) **medir la inestabilidad** con 100 árboles de profundidad 3 sobre remuestras bootstrap; (b) **comparar** árbol sin poda, árbol podado y bosque con **validación cruzada 5×5 repetida** (mismos pliegues), reportando diferencias con un **intervalo corregido** de Nadeau y Bengio (los pliegues comparten datos y no son independientes, así que la varianza ingenua es optimista); y (c) ver cómo varía la **predicción de un solo caso** entre árboles bootstrap profundos.

````python
# ============================================================
# 5. Inestabilidad del arbol y comparacion con un bosque aleatorio (breast_cancer de scikit-learn)
# ============================================================
from scipy import stats
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import RepeatedStratifiedKFold

BC = load_breast_cancer()
Xc, yc = BC.data, BC.target
print(f"breast_cancer: {Xc.shape[0]} casos, {Xc.shape[1]} variables, clases {np.bincount(yc)} (0 = maligno, 1 = benigno)")

# (a) Inestabilidad: 100 arboles de profundidad 3 sobre remuestras bootstrap; que variable sale en la raiz y en que umbral
raices, umbrales, estruct = [], [], []
rb = np.random.default_rng(SEED)
for b in range(100):
    ib = rb.choice(len(yc), len(yc))
    t_ = DecisionTreeClassifier(max_depth=3, random_state=0).fit(Xc[ib], yc[ib]).tree_
    raices.append(BC.feature_names[t_.feature[0]]); umbrales.append(t_.threshold[0])
    estruct.append((t_.feature[0], t_.feature[t_.children_left[0]], t_.feature[t_.children_right[0]]))
cuenta = pd.Series(raices).value_counts()
print("\nVariable en la RAIZ de 100 arboles bootstrap:\n", cuenta.to_string())
print(f"Umbral de la raiz (cuando es '{cuenta.index[0]}'): sd = {np.std([u for r, u in zip(raices, umbrales) if r == cuenta.index[0]]):.2f} | estructuras (raiz, hijo izq., hijo der.) distintas: {len(set(estruct))} de 100")

# (b) Comparacion honesta: validacion cruzada 5 x 5 repetida; mismas particiones para todos los modelos
rskf = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=SEED)
modelos = {"Arbol sin poda": DecisionTreeClassifier(random_state=SEED), "Arbol podado (alfa = 0.01)": DecisionTreeClassifier(ccp_alpha=0.01, random_state=SEED),
           "Bosque (100 arboles)": RandomForestClassifier(100, random_state=SEED, n_jobs=1)}
acc = {k: [] for k in modelos}
for tr, te in rskf.split(Xc, yc):
    for k, m_ in modelos.items():
        acc[k].append(m_.fit(Xc[tr], yc[tr]).score(Xc[te], yc[te]))
acc = pd.DataFrame(acc)
print("\nExactitud en 25 pliegues (media +- desv. est.):\n", pd.DataFrame({"media": acc.mean(), "sd": acc.std(ddof=1)}).round(4).to_string())

def dif_corregida(a, b, n_te_over_n_tr=0.25):                 # prueba t re-muestreada corregida de Nadeau y Bengio (2003)
    d = np.asarray(a) - np.asarray(b); k = len(d)
    se = np.sqrt(d.var(ddof=1) * (1 / k + n_te_over_n_tr))
    tcrit = stats.t.ppf(0.975, k - 1)
    return d.mean(), (d.mean() - tcrit * se, d.mean() + tcrit * se), 2 * stats.t.sf(abs(d.mean() / se), k - 1)
for a, b in (("Bosque (100 arboles)", "Arbol sin poda"), ("Bosque (100 arboles)", "Arbol podado (alfa = 0.01)"), ("Arbol podado (alfa = 0.01)", "Arbol sin poda")):
    dm, ic, pv = dif_corregida(acc[a], acc[b])
    print(f"{a} - {b}: dif = {dm:+.4f} | IC 95 % corregido = [{ic[0]:+.4f}, {ic[1]:+.4f}] | p = {pv:.4f}")
assert acc["Bosque (100 arboles)"].mean() > acc["Arbol sin poda"].mean()

# (c) Varianza de la prediccion de UN caso: arboles bootstrap profundos vs bosque
caso = Xc[[0]]
pr = np.array([DecisionTreeClassifier(random_state=0).fit(Xc[ib], yc[ib]).predict_proba(caso)[0, 1] for ib in (rb.choice(len(yc), len(yc)) for _ in range(100))])
print(f"\nProbabilidad de 'benigno' para el caso 0 (real: maligno) en 100 arboles bootstrap profundos: media {pr.mean():.2f}, sd {pr.std():.2f}")

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
cuenta.head(6).plot.barh(ax=ax[0], color=PALETA[0]); ax[0].invert_yaxis(); ax[0].set_title("Variable en la raiz (100 bootstrap)"); ax[0].set_xlabel("veces")
ax[1].boxplot([acc[k] for k in modelos]); ax[1].set_xticklabels(["sin poda", "podado", "bosque"]); ax[1].set_title("Exactitud en 25 pliegues"); ax[1].set_ylabel("exactitud")
ax[2].hist(pr, bins=np.linspace(0, 1, 11), color=PALETA[1]); ax[2].set_title("Prediccion de un caso segun la remuestra"); ax[2].set_xlabel("P(benigno)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **Inestabilidad:** la variable de la raíz cambia con la remuestra: `worst concave points` sale en 35 de los 100 árboles, `worst perimeter` en 32, y `mean concave points`, `worst radius` y `worst area` en el resto. Cuando la raíz es `worst concave points`, su umbral casi no se mueve (sd 0.01), y las variables que compiten están muy **correlacionadas** (miden tamaño y forma del tumor), así que cambiar de raíz no implica mucha diferencia en rendimiento; aun así, entre los 100 árboles hay **58 estructuras distintas** en los tres primeros nodos (raíz y sus dos hijos). Un árbol *singular* es una foto, no «la» verdad. (2) **Comparación en 25 pliegues:** árbol sin poda 0.9307, podado 0.9262, bosque **0.9568**. El bosque supera al árbol sin poda en **+2.6 puntos** (IC 95 % corregido [−0.000, +0.052], $p=0.050$) y al podado en +3.1 puntos (IC [−0.001, +0.062], $p=0.057$): la ventaja es **consistente en el signo pero estadísticamente marginal** con este tamaño de muestra y la corrección conservadora. Podar no cambió la exactitud (−0.5 puntos, IC [−0.028, +0.018], $p=0.68$). No se debe vender el bosque como «significativamente mejor»: se dice que **probablemente** es mejor por unos 2–3 puntos. (3) **Predicción de un caso:** para el caso 0 (maligno), la probabilidad de «benigno» según 100 árboles profundos bootstrap tiene media 0.10 y desviación **0.30**: la mayoría acierta, pero alrededor de uno de cada diez árboles da el veredicto contrario con total seguridad (probabilidad 1). Es la varianza del árbol individual que el bosque promedia. (4) **Qué se gana y qué se pierde:** el bosque mejora la exactitud y estabiliza la predicción, pero se pierde la regla legible; en contextos regulados (salud, crédito) un árbol pequeño y podado, o un bosque acompañado de importancia por permutación, pueden ser un mejor compromiso. Para un tratamiento extenso de bosques, véase [Data Mining, Módulo 04](../../Data%20Mining/04%20-%20Arboles%20de%20Decision%20y%20Bosques%20Aleatorios/04_Bosques_Aleatorios_Random_Forests.ipynb).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Impureza y ganancia** | Gini $1-\sum p_k^2$, entropía $-\sum p_k\log_2p_k$, varianza (regresión). El corte elegido maximiza la reducción ponderada de impureza. CART propio == scikit-learn (clasificación y regresión) |
| **Reglas e interacciones** | El árbol se lee como reglas y capta interacciones (humedad × pH), pero con regiones **rectangulares**; Gini y entropía dieron resultados equivalentes (0.867 y 0.873 en prueba, dentro del azar) |
| **Sesgo–varianza** | Profundidad baja = sesgo alto (3.72); profundidad alta = varianza alta (1.43). El mínimo fue la profundidad 4 (error 0.84 frente a 1.44 con profundidad 12) |
| **Poda** | Cost-complexity: se colapsa el eslabón más débil; camino propio == scikit-learn (21 alfas). Podar redujo 31 hojas a 5 con exactitud equivalente en prueba (0.867 frente a 0.873): gana **parsimonia**, no exactitud |
| **Importancia** | La MDI premia el ruido (0.17 de importancia total para 3 variables irrelevantes); la **permutación** en prueba lo corrige (≈ 0). Importancia no es causalidad |
| **Inestabilidad** | La raíz cambió de variable según la remuestra (35/32/12/12/9) y hubo 58 estructuras distintas en 100 árboles; un árbol individual es una foto |
| **Árbol frente a bosque** | El bosque ganó en 2.6–3.1 puntos de exactitud (0.957 frente a 0.931 y 0.926), pero el IC corregido casi incluye 0 ($p\approx0.05$): evidencia **marginal**. Se pierde la legibilidad |
