## 1. pandas: de Tablas Crudas a una Tabla Analítica 🐼

Una cooperativa **ficticia** entrega dos tablas: **fincas** (una fila por finca: municipio, altitud, área, riego) y **cosechas** (varias filas por finca: semestre, variedad, fertilizante, lluvia, rendimiento en t/ha y si hubo enfermedad). Antes de modelar hay que **limpiar, unir y resumir**, y cada paso tiene su trampa:

| Operación | Pregunta que responde | Trampa típica |
|---|---|---|
| `drop_duplicates` | ¿Hay registros repetidos? | Borrar filas que **no** son duplicados reales (mismo valor, distinto evento) |
| `merge(..., validate="m:1")` | ¿Cada cosecha encuentra **una** finca? | Un `merge` con claves repetidas **multiplica filas en silencio**; `validate` y `indicator` lo detectan |
| `groupby(...).agg(...)` | ¿Cuál es el resumen **por grupo**? | Olvidar el tamaño $n$ de cada grupo |
| `groupby(...).transform(...)` | ¿Cómo se compara cada fila con **su** grupo? | Usar `agg` y luego un `merge` manual (más código, más errores) |
| `pivot_table` | ¿Cómo cambia una medida entre **dos** factores? | Celdas vacías que parecen ceros |
| `groupby(...).shift(1)` | ¿Qué pasó en la cosecha **anterior** de la misma finca? | Mezclar fincas al ordenar mal, o **usar el futuro** como predictor |

Generamos los datos con el mecanismo siguiente (todo **inventado**, semilla fija): $\text{rendimiento}=14+0.012(\text{fert}-210)+\text{efecto de variedad}+2\cdot\text{riego}-0.004(\text{altitud}-2750)+u_{\text{finca}}+\varepsilon$, con $u_{\text{finca}}\sim N(0,1.2^2)$ un efecto **no observado** de cada finca y $\varepsilon\sim N(0,2^2)$. Después ensuciamos la tabla con 3 filas duplicadas y 12 fertilizantes faltantes. Cada resumen de *pandas* se contrasta con un cálculo equivalente en *NumPy* (`assert`).

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.width", 140); pd.set_option("display.max_columns", 20)
rng = np.random.default_rng(2026)
municipios = ["Tunja", "Duitama", "Sogamoso", "Ventaquemada", "Siachoque", "Toca", "Samacá", "Tibasosa", "Soracá", "Motavita"]
semestres = ["2022A", "2022B", "2023A", "2023B", "2024A", "2024B"]

# --- Tabla 1: 400 fincas SINTÉTICAS (una fila por finca) ---
n_f = 400
fincas = pd.DataFrame({
    "id_finca": np.arange(1, n_f + 1),
    "municipio": rng.choice(municipios, n_f),
    "altitud": rng.normal(2750, 150, n_f).round(0),
    "area_ha": rng.lognormal(0.8, 0.6, n_f).round(2),
    "riego": rng.random(n_f) < 0.35,
})
u_finca = rng.normal(0, 1.2, n_f)            # efecto NO observado de cada finca (suelo, manejo, ...)

# --- Tabla 2: cosechas (varias filas por finca: 1 a 5 semestres) ---
filas = [(i + 1, semestres[s]) for i in range(n_f) for s in sorted(rng.choice(6, int(np.clip(1 + rng.poisson(1.6), 1, 5)), replace=False))]
cosechas = pd.DataFrame(filas, columns=["id_finca", "semestre"])
m = len(cosechas)
cosechas["variedad"] = rng.choice(["Capiro", "Criolla", "Pastusa"], m, p=[0.4, 0.2, 0.4])
cosechas["fert_kg_ha"] = rng.gamma(6, 35, m).round(0)
cosechas["lluvia_mm"] = rng.normal(620, 110, m).round(0)

# Mecanismo verdadero (lo conocemos porque lo inventamos): rendimiento en t/ha y enfermedad
tmp = cosechas.merge(fincas, on="id_finca")
efecto_var = tmp["variedad"].map({"Capiro": 0.0, "Criolla": -2.5, "Pastusa": 1.5}).to_numpy()
cosechas["rend_t_ha"] = (14 + 0.012 * (tmp["fert_kg_ha"] - 210) + efecto_var + 2.0 * tmp["riego"] - 0.004 * (tmp["altitud"] - 2750)
                         + u_finca[tmp["id_finca"] - 1] + rng.normal(0, 2.0, m)).round(2)
logit = -1.4 + 0.012 * (tmp["lluvia_mm"] - 620) + 0.5 * (tmp["variedad"] == "Criolla") - 0.5 * tmp["riego"]
cosechas["enferma"] = (rng.random(m) < 1 / (1 + np.exp(-logit))).astype(int)

# Suciedad plausible: 3 filas duplicadas y 12 fertilizantes faltantes
cosechas = pd.concat([cosechas, cosechas.sample(3, random_state=1)], ignore_index=True)
cosechas.loc[rng.choice(len(cosechas), 12, replace=False), "fert_kg_ha"] = np.nan
print(f"fincas: {fincas.shape} | cosechas: {cosechas.shape} | duplicados: {cosechas.duplicated().sum()} | fert faltante: {cosechas['fert_kg_ha'].isna().sum()}")

# --- 1) Limpiar y unir: validate evita el producto cartesiano silencioso ---
cos = cosechas.drop_duplicates()
datos = cos.merge(fincas, on="id_finca", how="left", validate="m:1", indicator=True)
assert len(datos) == len(cos) and (datos["_merge"] == "both").all()          # ninguna cosecha sin finca, ninguna fila duplicada por el join
datos = datos.drop(columns="_merge")
datos["variedad"] = datos["variedad"].astype("category")
datos["fert_kg_ha"] = datos["fert_kg_ha"].fillna(datos.groupby("variedad", observed=True)["fert_kg_ha"].transform("median"))   # imputación simple por variedad
assert datos["fert_kg_ha"].notna().all()

# --- 2) Agrupar: resumen por variedad con nombres explícitos ---
resumen = datos.groupby("variedad", observed=True).agg(n=("rend_t_ha", "size"), media=("rend_t_ha", "mean"), sd=("rend_t_ha", "std"),
                                                       prop_enferma=("enferma", "mean")).round(3)
print(resumen)

# Verificación cruzada con NumPy: la media por grupo con bincount
cod = datos["variedad"].cat.codes.to_numpy()
media_np = np.bincount(cod, weights=datos["rend_t_ha"].to_numpy()) / np.bincount(cod)
assert np.allclose(media_np, resumen["media"].to_numpy(), atol=1e-3)

# --- 3) transform: desviación respecto al promedio del municipio; pivot_table: municipio x semestre ---
datos["rend_rel_mun"] = datos["rend_t_ha"] - datos.groupby("municipio")["rend_t_ha"].transform("mean")
assert abs(datos.groupby("municipio")["rend_rel_mun"].mean()).max() < 1e-9
tabla = datos.pivot_table(index="municipio", columns="semestre", values="rend_t_ha", aggfunc="mean").round(1)
print(tabla.iloc[:4])
print("celdas vacías del pivote:", int(tabla.isna().sum().sum()), "de", tabla.size)

# --- 4) Panel: rezago por finca (el rendimiento de la cosecha anterior) ---
datos = datos.sort_values(["id_finca", "semestre"])
datos["rend_previo"] = datos.groupby("id_finca")["rend_t_ha"].shift(1)
print("filas con cosecha previa:", int(datos["rend_previo"].notna().sum()), "de", len(datos),
      "| fincas con 1 sola cosecha:", int((datos.groupby("id_finca").size() == 1).sum()))
r_prev = datos[["rend_t_ha", "rend_previo"]].dropna().corr().iloc[0, 1]
print(f"correlación entre cosechas consecutivas de la misma finca: {r_prev:.3f}  (efecto de finca no observado)")
````

---
**Lectura.** (1) La tabla de cosechas llegó con **1 024 filas, 3 duplicadas y 12 fertilizantes faltantes**; tras limpiar quedan **1 021**. El `merge` con `validate="m:1"` y `indicator` confirma que **todas** las cosechas tienen finca y que la unión no multiplicó filas (los `assert` pasan). (2) **Resumen por variedad:** Pastusa rinde en promedio **15.89 t/ha**, Capiro **14.59** y Criolla **12.23**; la Criolla es además la más enferma (**34.3 %** frente a 20.7 % y 19.7 %). La media por grupo calculada con `np.bincount` **coincide** con la de `groupby` (verificación independiente). (3) El pivote municipio $\times$ semestre no tiene celdas vacías (0 de 60) porque hay muchas cosechas por celda; con menos datos, las celdas vacías serían `NaN` y habría que decidir qué hacer con ellas. (4) **Panel:** solo **621 de 1 021** filas tienen cosecha previa y **84 fincas** aparecen una sola vez; la correlación entre cosechas **consecutivas de la misma finca** es **0.22** pese a que las cosechas son «independientes» dadas las covariables: es la huella del efecto de finca $u_{\text{finca}}$ que **no observamos**. Esa dependencia entre filas es la pieza que más importará en las secciones siguientes.

## 2. NumPy: Vectorizar, Remuestrear y *Broadcasting* ⚡

*NumPy* es el motor de casi todo lo demás. Dos ideas bastan para este cuaderno.

**Vectorización con una matriz de índices.** Un *bootstrap* de $B$ réplicas sobre $n$ datos se escribe con una sola matriz de enteros `idx` de forma $(B, n)$: `y[idx]` produce las $B$ muestras de una vez y las medias por fila dan las $B$ estadísticas. Comparamos el bucle `for` con la versión vectorizada **usando los mismos índices** (deben dar **exactamente** lo mismo) y construimos el intervalo percentil para la diferencia de medias $\hat\Delta=\bar y_{\text{riego}}-\bar y_{\text{secano}}$.

**Independencia de las filas.** Remuestrear **cosechas** supone que las filas son independientes, pero varias cosechas de una misma finca comparten $u_{\text{finca}}$. El **bootstrap por conglomerados** remuestrea **fincas completas**. Como el número de cosechas con y sin riego de cada finca cambia en cada réplica, la diferencia de medias es un **cociente**: $\hat\Delta^\ast=\dfrac{\sum_{f\in\text{remuestra}}S_{1f}}{\sum_{f\in\text{remuestra}}N_{1f}}-\dfrac{\sum_{f\in\text{remuestra}}S_{0f}}{\sum_{f\in\text{remuestra}}N_{0f}}$, con $S_{gf}$ la suma de rendimientos y $N_{gf}$ el número de cosechas de la finca $f$ en el grupo $g$ (se calculan una sola vez con `np.bincount`). El efecto de diseño aproximado es $\mathrm{deff}\approx 1+(\bar m-1)\rho$, con $\bar m$ las cosechas por finca y $\rho$ la correlación intraclase.

**Broadcasting.** Al restar arreglos de formas $(n,1,3)$ y $(1,n,3)$ se obtiene un tensor $(n,n,3)$ sin bucles: así se calculan **todas** las distancias euclidianas entre cosechas (se contrasta con `scipy.spatial.distance.cdist`). El costo es **memoria** $O(n^2)$.

````python
import time
from scipy import stats
from scipy.spatial.distance import cdist

y = datos["rend_t_ha"].to_numpy()
g = datos["riego"].to_numpy()
B = 4000
rng = np.random.default_rng(7)

# --- 1) Bootstrap ingenuo: bucle frente a matriz de índices (mismos índices => mismo resultado) ---
idx = rng.integers(0, len(y), (B, len(y)))
t0 = time.perf_counter()
dif_bucle = np.array([y[i][g[i]].mean() - y[i][~g[i]].mean() for i in idx])
t_bucle = time.perf_counter() - t0
t0 = time.perf_counter()
yy, gg = y[idx], g[idx]                                                           # (B, n)
dif_vec = (yy * gg).sum(1) / gg.sum(1) - (yy * ~gg).sum(1) / (~gg).sum(1)
t_vec = time.perf_counter() - t0
assert np.allclose(dif_bucle, dif_vec)
print(f"bucle {t_bucle:.2f} s | vectorizado {t_vec:.2f} s | resultados idénticos")

# --- 2) Bootstrap por conglomerados: se remuestrean FINCAS, no cosechas (las cosechas de una finca no son independientes) ---
f = datos["id_finca"].to_numpy() - 1
nf = f.max() + 1
S1, N1 = np.bincount(f, weights=y * g, minlength=nf), np.bincount(f, weights=g, minlength=nf)
S0, N0 = np.bincount(f, weights=y * ~g, minlength=nf), np.bincount(f, weights=~g, minlength=nf)
idf = rng.integers(0, nf, (B, nf))
dif_cl = S1[idf].sum(1) / N1[idf].sum(1) - S0[idf].sum(1) / N0[idf].sum(1)

obs = y[g].mean() - y[~g].mean()
w = stats.ttest_ind(y[g], y[~g], equal_var=False)
ic_w = w.confidence_interval(0.95)
ic_ing = np.percentile(dif_vec, [2.5, 97.5])
ic_cl = np.percentile(dif_cl, [2.5, 97.5])
print(f"diferencia observada riego - secano: {obs:.3f} t/ha (efecto verdadero inventado del riego: 2.0 t/ha)")
print(f"IC Welch        : [{ic_w.low:.3f}, {ic_w.high:.3f}] ancho {ic_w.high - ic_w.low:.3f}")
print(f"IC bootstrap    : [{ic_ing[0]:.3f}, {ic_ing[1]:.3f}] ancho {ic_ing[1] - ic_ing[0]:.3f}")
print(f"IC conglomerados: [{ic_cl[0]:.3f}, {ic_cl[1]:.3f}] ancho {ic_cl[1] - ic_cl[0]:.3f}")
assert abs((ic_ing[1] - ic_ing[0]) - (ic_w.high - ic_w.low)) < 0.06 and (ic_cl[1] - ic_cl[0]) > (ic_ing[1] - ic_ing[0])

# --- 3) Broadcasting: distancias entre cosechas en 3 variables estandarizadas ---
Z = datos[["fert_kg_ha", "lluvia_mm", "altitud"]].to_numpy()
Z = (Z - Z.mean(0)) / Z.std(0, ddof=1)
D = np.sqrt(((Z[:, None, :] - Z[None, :, :]) ** 2).sum(-1))                      # (n, n, 3) -> (n, n)
assert np.allclose(D, cdist(Z, Z))
np.fill_diagonal(D, np.inf)
print(f"matriz {D.shape}, memoria del tensor intermedio ≈ {Z.shape[0] ** 2 * 3 * 8 / 1e6:.0f} MB; distancia media al vecino más cercano: {D.min(1).mean():.3f}")
````

---
**Lectura.** (1) **Vectorizar da el mismo resultado, más rápido:** con 4 000 réplicas el bucle tardó 0.2-0.3 s y la versión por matriz ≈ 0.1 s en esta máquina (los tiempos varían con la carga; lo que **no** varía es la igualdad de resultados, que verificamos con `np.allclose`). La ventaja crece con $n$ y $B$, pero el costo es memoria: la matriz `y[idx]` guarda $B\times n$ números. (2) La diferencia observada riego frente a secano es **1.755 t/ha** (el efecto verdadero inventado es 2.0; la estimación cae a **0.25 t/ha** del valor verdadero y su intervalo lo cubre). El IC *bootstrap* ingenuo **[1.369, 2.145]** casi coincide con el de Welch **[1.369, 2.141]**: ambos suponen filas independientes. (3) **El de conglomerados es más ancho:** **[1.294, 2.212]**, un **18 % más ancho** (0.918 frente a 0.776), porque el riego es una característica **de la finca** y las cosechas de una misma finca no aportan información independiente sobre él. Los intervalos ingenuos son **demasiado optimistas**: reportan una precisión que los datos no tienen. (4) La matriz de distancias ($1\,021\times1\,021$) se obtiene por *broadcasting* y coincide con `cdist`; el tensor intermedio ocupa ≈ 25 MB: con 100 000 filas serían ≈ 240 GB, y habría que usar `cdist` o *scikit-learn*.

assert np.allclose(pond[nombres_m].to_numpy(), chk)
simple = datos.groupby("municipio")["rend_t_ha"].mean()
dif = (pond - simple).abs()

## 3. statsmodels: Explicar con Incertidumbre 📐

*statsmodels* responde **«¿cuánto, y con qué margen de error?»** Tres modelos del curso, ahora sobre la tabla analítica:

**ANOVA de una vía.** $F=\dfrac{SC_{\text{entre}}/(k-1)}{SC_{\text{dentro}}/(N-k)}$ compara la variación **entre** variedades con la variación **dentro** de ellas; el tamaño del efecto es $\eta^2=SC_{\text{entre}}/SC_{\text{total}}$ (proporción de la varianza explicada por el factor). Si $F$ es significativo, **Tukey HSD** compara los pares controlando la tasa de error de la familia. Contrastamos $F$ y $p$ con `scipy.stats.f_oneway`.

**Regresión múltiple con errores robustos por conglomerado.** Con varias cosechas por finca, los errores $u_{ij}$ de una misma finca están correlacionados y los errores estándar clásicos son optimistas para las variables **constantes dentro de la finca** (riego, altitud). El estimador «sándwich» por conglomerados es $\widehat V=(X^\top X)^{-1}\Big(\sum_{g}X_g^\top\hat u_g\hat u_g^\top X_g\Big)(X^\top X)^{-1}$, donde $g$ indexa las fincas; los **coeficientes no cambian**, solo su incertidumbre. Como conocemos el valor verdadero, podemos ver si los intervalos lo cubren.

**Regresión logística.** Modela $\log\dfrac{p}{1-p}=\beta_0+\beta_1x_1+\cdots$; $e^{\beta_j}$ es la **razón de momios** (*odds ratio*) por unidad de $x_j$. Un modelo logístico de *scikit-learn* **sin penalización** debe coincidir con el de *statsmodels* (por defecto *scikit-learn* aplica una penalización $L_2$ con $C=1$; usamos $C$ enorme para eliminarla).

````python
import warnings
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from sklearn.linear_model import LogisticRegression

datos["fert100"] = datos["fert_kg_ha"] / 100          # unidades de 100 kg/ha: coeficientes legibles
datos["alt100"] = (datos["altitud"] - 2750) / 100     # centrada en 2750 m, en cientos de metros
datos["lluvia100"] = (datos["lluvia_mm"] - 620) / 100
datos["criolla"] = (datos["variedad"] == "Criolla").astype(int)
datos["riego_i"] = datos["riego"].astype(int)

# --- 1) ANOVA de una vía a mano (statsmodels) frente a SciPy, tamaño del efecto y Tukey ---
m0 = smf.ols("rend_t_ha ~ C(variedad)", data=datos).fit()
an = anova_lm(m0)
F_sm = an.loc["C(variedad)", "F"]
grupos = [d["rend_t_ha"].to_numpy() for _, d in datos.groupby("variedad", observed=True)]
F_sp, p_sp = stats.f_oneway(*grupos)
assert np.isclose(F_sm, F_sp) and np.isclose(an.loc["C(variedad)", "PR(>F)"], p_sp)
eta2 = an.loc["C(variedad)", "sum_sq"] / an["sum_sq"].sum()
print(f"ANOVA variedad: F = {F_sm:.1f}, p = {p_sp:.2e}, eta² = {eta2:.3f}")
tk = pairwise_tukeyhsd(datos["rend_t_ha"], datos["variedad"].astype(str))
print(tk.summary())

# --- 2) Regresión múltiple: errores estándar clásicos frente a robustos por conglomerado (finca) ---
fm = "rend_t_ha ~ fert100 + C(variedad) + riego_i + alt100"
ols = smf.ols(fm, data=datos).fit()
olc = smf.ols(fm, data=datos).fit(cov_type="cluster", cov_kwds={"groups": datos["id_finca"]})
verdad = {"Intercept": np.nan, "fert100": 1.2, "C(variedad)[T.Criolla]": -2.5, "C(variedad)[T.Pastusa]": 1.5, "riego_i": 2.0, "alt100": -0.4}
tab = pd.DataFrame({"coef": ols.params, "ee_clasico": ols.bse, "ee_finca": olc.bse, "cociente": olc.bse / ols.bse,
                    "IC_inf": olc.conf_int()[0], "IC_sup": olc.conf_int()[1], "verdad": pd.Series(verdad)}).round(3)
print(tab)
print(f"R² = {ols.rsquared:.3f}; R² ajustado = {ols.rsquared_adj:.3f}")
assert np.allclose(ols.params, olc.params)                         # mismos coeficientes: solo cambia la incertidumbre
assert olc.bse["riego_i"] > ols.bse["riego_i"]                      # riego es una variable de finca: el error clásico es optimista

# --- 3) Regresión logística: razones de momios (OR) con IC y el mismo ajuste en scikit-learn ---
lg = smf.logit("enferma ~ lluvia100 + criolla + riego_i", data=datos).fit(disp=0)
orr = np.exp(pd.concat([lg.params, lg.conf_int()], axis=1)).set_axis(["OR", "IC_inf", "IC_sup"], axis=1).round(3)
orr["p"] = lg.pvalues.round(4)
print(orr)
Xl = datos[["lluvia100", "criolla", "riego_i"]].to_numpy()
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    sk = LogisticRegression(C=1e10, max_iter=5000, tol=1e-10).fit(Xl, datos["enferma"])          # C enorme = sin penalización
assert np.allclose(sk.coef_.ravel(), lg.params.iloc[1:].to_numpy(), atol=2e-3) and np.isclose(sk.intercept_[0], lg.params.iloc[0], atol=2e-3)
print("coeficientes de scikit-learn (sin penalización) == statsmodels")
````

---
**Lectura.** (1) **ANOVA:** $F=121.0$, $p\approx7\times10^{-48}$ y $\eta^2=0.192$: la variedad explica ≈ **19 %** de la varianza del rendimiento (efecto grande, pero **81 %** queda sin explicar). Tukey rechaza los tres pares: Pastusa $-$ Capiro $=+1.31$, Criolla $-$ Capiro $=-2.36$ y Criolla $-$ Pastusa $=+3.67$ t/ha. Un valor $p$ diminuto dice que la diferencia **no es azar**, no que sea grande: eso lo dicen los intervalos y $\eta^2$. (2) **Regresión:** todos los coeficientes quedan cerca de su valor verdadero y **los cinco IC por finca lo cubren** (fert $1.19$ frente a 1.2; riego $2.03$ frente a 2.0; altitud $-0.33$ frente a $-0.4$ por 100 m; Pastusa $1.55$ frente a 1.5; Criolla $-2.32$ frente a $-2.5$). $R^2=0.421$ (ajustado 0.418): con ruido $\sigma=2$ y un efecto de finca no observado, ese es el techo. Los coeficientes de las dos regresiones son idénticos, pero **el error estándar del riego pasa de 0.158 a 0.200 (×1.27) y el de la altitud de 0.048 a 0.059 (×1.23)**, mientras que los de variables que **varían dentro de la finca** casi no cambian (fertilizante 0.085 frente a 0.082). Es el mismo patrón que vimos con el *bootstrap*. (3) **Logística:** por cada 100 mm más de lluvia los momios de enfermedad se multiplican por **3.46** (IC [2.85, 4.20]; el valor verdadero inventado es $e^{1.2}\approx3.32$), la variedad Criolla los multiplica por **2.25** ([1.52, 3.34]; verdadero $e^{0.5}\approx1.65$) y el riego por **0.65** ([0.45, 0.95]; verdadero $e^{-0.5}\approx0.61$). Los coeficientes de *scikit-learn* sin penalización coinciden con los de *statsmodels* (`assert`). *Cuidado:* una razón de momios **no** es una razón de probabilidades; solo se parecen cuando el evento es raro (aquí la prevalencia es 23 %).

## 4. scikit-learn: Predecir sin Hacerse Trampa 🤖

*scikit-learn* responde **«¿qué tan bien predice con datos que no vio?»** Tres reglas que ya conocemos del curso, ahora aplicadas juntas:

1. **Todo en el *pipeline*:** el escalado y la codificación se **ajustan solo con el entrenamiento** de cada pliegue (`make_pipeline(ColumnTransformer(...), modelo)`); si se ajustan antes de validar, hay **fuga**.
2. **Un modelo de referencia (*baseline*):** `DummyRegressor` predice la media; un modelo que no lo supere por un margen claro no sirve. El $R^2$ fuera de muestra del *baseline* es ≈ 0 (puede ser levemente negativo).
3. **Validar por grupos:** si varias filas pertenecen a la misma finca, `KFold` mezcla cosechas de una misma finca entre entrenamiento y prueba, y un modelo flexible puede **reconocer a la finca** por su altitud y su área (son una «huella digital»). `GroupKFold` mantiene cada finca en un solo lado.

Para la clasificación de la **enfermedad** usamos probabilidades **fuera de muestra** (`cross_val_predict` con `GroupKFold`) y evaluamos con el **AUC** (probabilidad de que una cosecha enferma reciba mayor puntaje que una sana), la **precisión** $=\dfrac{VP}{VP+FP}$ y la **sensibilidad** $=\dfrac{VP}{VP+FN}$ para dos umbrales: mover el umbral **intercambia** una por otra.

````python
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyRegressor, DummyClassifier
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import roc_auc_score, precision_score, recall_score, roc_curve
from sklearn.model_selection import GroupKFold, KFold, cross_val_predict, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

num = ["fert_kg_ha", "lluvia_mm", "altitud", "area_ha"]
cat = ["variedad", "municipio"]
prep = lambda: ColumnTransformer([("num", StandardScaler(), num), ("cat", OneHotEncoder(handle_unknown="ignore"), cat + ["riego"])])
Xs = datos[num + cat + ["riego"]].assign(riego=datos["riego"].astype(str))
ys, grp = datos["rend_t_ha"], datos["id_finca"]

modelos = {"Media (baseline)": DummyRegressor(),
           "Regresión lineal": LinearRegression(),
           "Bosque aleatorio": RandomForestRegressor(n_estimators=150, min_samples_leaf=3, random_state=0, n_jobs=1)}
cv_ing = KFold(5, shuffle=True, random_state=0)      # mezcla cosechas de la misma finca entre entrenamiento y prueba
cv_fin = GroupKFold(5)                               # una finca nunca queda en ambos lados
filas = {}
for nombre, est in modelos.items():
    pipe = make_pipeline(prep(), est)
    filas[nombre] = {"R² KFold": cross_val_score(pipe, Xs, ys, cv=cv_ing, scoring="r2").mean(),
                     "R² GroupKFold": cross_val_score(pipe, Xs, ys, cv=cv_fin, groups=grp, scoring="r2").mean()}
res = pd.DataFrame(filas).T.round(3)
print(res)
assert res.loc["Bosque aleatorio", "R² KFold"] > res.loc["Bosque aleatorio", "R² GroupKFold"]          # la fuga entre cosechas de la misma finca infla el bosque
assert res.loc["Regresión lineal", "R² GroupKFold"] > res.loc["Media (baseline)", "R² GroupKFold"]

# --- Clasificación de la enfermedad: probabilidades fuera de muestra y umbral ---
yc = datos["enferma"]
clf = make_pipeline(prep(), LogisticRegression(max_iter=2000))
proba = cross_val_predict(clf, Xs, yc, cv=cv_fin, groups=grp, method="predict_proba")[:, 1]
auc = roc_auc_score(yc, proba)
base = cross_val_predict(make_pipeline(prep(), DummyClassifier(strategy="prior")), Xs, yc, cv=cv_fin, groups=grp, method="predict_proba")[:, 1]
print(f"AUC fuera de muestra: {auc:.3f} (baseline {roc_auc_score(yc, base):.3f}); prevalencia de enfermedad: {yc.mean():.3f}")
for umbral in (0.5, 0.3):
    pred = proba >= umbral
    print(f"umbral {umbral}: precisión = {precision_score(yc, pred):.3f}, sensibilidad (recall) = {recall_score(yc, pred):.3f}, alertas = {pred.sum()}")
````

**Lectura.** (1) **Regresión:** el *baseline* tiene $R^2\approx-0.01$ (como esperamos: predice la media); la **regresión lineal** llega a **0.39** con `GroupKFold` y el **bosque aleatorio** a **0.34**. Con el mecanismo **lineal** que inventamos, el modelo lineal gana: *el modelo más complejo no es el mejor cuando la verdad es simple*. (2) **La fuga entre cosechas de una misma finca existe pero es pequeña aquí:** el bosque pasa de 0.336 (por grupos) a 0.349 con `KFold`, y el lineal de 0.390 a 0.407. Con más ruido de finca y modelos con más capacidad de memorizar (árboles profundos, vecinos cercanos), la brecha sería mayor; **validar por grupos no cuesta nada** y protege. (3) **Clasificación:** AUC fuera de muestra **0.80** (el *baseline* da 0.47, es decir, azar). Con umbral 0.5 el modelo emite 124 alertas con precisión **0.58** y detecta solo el **31 %** de las cosechas enfermas; bajando el umbral a 0.3 emite 282 alertas, detecta **59 %** pero la precisión cae a **0.49**. No existe «el» umbral correcto: depende del **costo** de no detectar una enfermedad frente al de una alarma falsa. (4) La gráfica reúne las tres conclusiones: los coeficientes de *statsmodels* cubren el valor verdadero, el modelo lineal supera al bosque en $R^2$ y la curva ROC queda claramente sobre la diagonal.
