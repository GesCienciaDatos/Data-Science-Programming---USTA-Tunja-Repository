## 1. k-means: la Idea, el Algoritmo y sus Óptimos Locales 🎯

**Objetivo.** Dado $X=\{x_1,\dots,x_n\}\subset\mathbb{R}^d$ y un número de grupos $k$, k-means busca la partición $C_1,\dots,C_k$ que minimiza la **inercia** (suma de cuadrados dentro de los grupos):
$$J=\sum_{j=1}^{k}\sum_{x_i\in C_j}\lVert x_i-\mu_j\rVert^2,\qquad \mu_j=\frac{1}{|C_j|}\sum_{x_i\in C_j}x_i.$$
Encontrar el mínimo global es **NP-difícil**; el **algoritmo de Lloyd** alterna dos pasos: **(1) asignar** cada punto al centroide más cercano y **(2) recalcular** cada centroide como la media de su grupo. Cada paso **no puede aumentar** $J$, y como hay un número finito de particiones, el algoritmo **converge**, pero **solo a un óptimo local** que depende de los centroides iniciales. De ahí que se use ***k-means++*** (cada centro nuevo se sortea con probabilidad proporcional a $D(x)^2$, la distancia al centro más cercano ya elegido) y se **repita varias veces** quedándose con la menor inercia.

**Supuestos implícitos** (de la inercia): grupos **convexos y aproximadamente esféricos**, de **tamaño y varianza parecidos**, y variables en **escala comparable** (usa distancia euclidiana). Los iremos poniendo a prueba. Implementamos Lloyd + k-means++ desde cero y comprobamos que, con los **mismos centros iniciales**, produce **exactamente** las etiquetas, los centroides y la inercia de `KMeans`.

````python
# ============================================================
# Configuracion inicial y k-means desde cero (Lloyd + k-means++)
# ============================================================
# !pip install -q numpy pandas scipy scikit-learn matplotlib

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import adjusted_rand_score

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scikit-learn {__import__('sklearn').__version__}")

# Datos SINTETICOS: 4 grupos de fincas con dispersiones distintas (2 variables estandarizadas); conocemos las etiquetas reales
X, y_real = make_blobs(n_samples=400, centers=[(-4, -3), (0, 4), (5, 0), (-1, -9)], cluster_std=[1.1, 1.6, 0.8, 1.3], random_state=SEED)
print("X:", X.shape, "| tamanos reales:", np.bincount(y_real))

def dist2(A, B):                                          # distancias euclidianas al cuadrado, todas contra todas
    return np.maximum((A ** 2).sum(1)[:, None] - 2 * A @ B.T + (B ** 2).sum(1)[None, :], 0.0)

def kmeanspp(X, k, rng):                                  # k-means++: cada centro nuevo ~ proporcional a D(x)^2
    C = [X[rng.integers(len(X))]]
    for _ in range(k - 1):
        d = dist2(X, np.array(C)).min(1)
        C.append(X[rng.choice(len(X), p=d / d.sum())])
    return np.array(C)

def lloyd(X, C0, max_iter=300):
    C = C0.copy()
    lab = None
    for it in range(max_iter):
        nuevo = dist2(X, C).argmin(1)                     # (1) asignar al centroide mas cercano
        if lab is not None and np.array_equal(nuevo, lab):
            break
        lab = nuevo
        for j in range(len(C)):                           # (2) recalcular cada centroide como el promedio de su grupo
            if np.any(lab == j):
                C[j] = X[lab == j].mean(0)
    return lab, C, ((X - C[lab]) ** 2).sum(), it + 1

def kmeans(X, k, n_init=10, seed=0):
    r = np.random.default_rng(seed)
    mejor = None
    for _ in range(n_init):
        lab, C, inercia, it = lloyd(X, kmeanspp(X, k, r))
        if mejor is None or inercia < mejor[2]:
            mejor = (lab, C, inercia, it)
    return mejor

# Verificacion exacta: mismos centros iniciales -> mismas etiquetas e inercia que scikit-learn
C0 = kmeanspp(X, 4, np.random.default_rng(1))
lab, C, inercia, iters = lloyd(X, C0)
sk = KMeans(n_clusters=4, init=C0, n_init=1, algorithm="lloyd", tol=0, max_iter=300).fit(X)
assert np.array_equal(lab, sk.labels_) and np.allclose(C, sk.cluster_centers_) and np.isclose(inercia, sk.inertia_)
print(f"Lloyd propio == KMeans(init=C0): inercia {inercia:.2f}, {iters} iteraciones (sklearn: {sk.n_iter_})")

lab_b, C_b, in_b, _ = kmeans(X, 4)
print(f"Mejor de 10 inicializaciones k-means++: inercia {in_b:.2f} | ARI vs grupos reales = {adjusted_rand_score(y_real, lab_b):.3f}")

# La inercia (suma de cuadrados dentro de los grupos) baja en cada iteracion; mostramos un mal inicio vs k-means++
def inercias_por_iter(X, C0, n_it=8):
    C, out = C0.copy(), []
    for _ in range(n_it):
        lab = dist2(X, C).argmin(1); out.append(((X - C[lab]) ** 2).sum())
        C = np.array([X[lab == j].mean(0) if np.any(lab == j) else C[j] for j in range(len(C))])
    return out
malo = X[np.argsort(X[:, 0])[:4]]                         # 4 puntos extremos de la izquierda como centros iniciales
tr_malo, tr_pp = inercias_por_iter(X, malo), inercias_por_iter(X, C0)
assert all(np.diff(tr_malo) <= 1e-9) and all(np.diff(tr_pp) <= 1e-9)

fig, ax = plt.subplots(1, 3, figsize=(15, 4.3))
for j in range(4):
    ax[0].scatter(*X[lab_b == j].T, s=14, color=PALETA[j])
ax[0].scatter(*C_b.T, marker="X", s=140, color="k", label="centroides"); ax[0].legend(); ax[0].set_title("k-means (k = 4)")
ax[1].plot(range(1, 9), tr_malo, "o-", color=PALETA[1], label="inicio malo"); ax[1].plot(range(1, 9), tr_pp, "o-", color=PALETA[0], label="k-means++")
ax[1].set_xlabel("iteracion"); ax[1].set_ylabel("inercia"); ax[1].legend(); ax[1].set_title("La inercia solo puede bajar")
finales = [kmeans(X, 4, n_init=1, seed=s)[2] for s in range(60)]
ax[2].hist(finales, bins=20, color=PALETA[2]); ax[2].axvline(in_b, color="k", ls="--", label="mejor de 10"); ax[2].legend()
ax[2].set_title("Optimos locales: 60 inicios, 1 corrida c/u"); ax[2].set_xlabel("inercia final")
plt.tight_layout(); plt.show()
print(f"Inercia final en 60 corridas de 1 inicio: min {min(finales):.1f}, max {max(finales):.1f} | corridas peores que el optimo: {np.mean(np.array(finales) > in_b + 1e-6):.0%}")
````

---
**Lectura.** (1) Con los mismos centros iniciales, el Lloyd propio **coincide** con `KMeans` en etiquetas, centroides e inercia, e incluso en el número de iteraciones (7). (2) **Pero ese inicio en particular era malo:** convergió a una inercia de **3 013.6**, casi el triple de la mejor solución encontrada (**1 120.3**). Es el óptimo local en acción: dos grupos reales quedaron fusionados y uno partido. (3) Con **60 corridas de un solo inicio**, **7 %** terminó en un óptimo peor que el mejor; por eso se usan varias inicializaciones (`n_init`). (4) La mejor solución recupera los grupos con ARI **0.942** (no 1, porque unos pocos puntos de grupos vecinos quedan del lado equivocado). (5) La figura central muestra la propiedad clave: la inercia **solo puede bajar** en cada iteración, sea cual sea el inicio.

## 2. ¿Cuántos Grupos? Codo, Silueta y Calinski–Harabasz 📐

La inercia **siempre baja** al aumentar $k$ (en el límite, $k=n$ da inercia 0), así que **no sirve sola** para elegir $k$. Los criterios habituales:

| Criterio | Definición | Regla |
|---|---|---|
| **Codo** | Inercia en función de $k$ | Buscar donde la caída se aplana (subjetivo) |
| **Silueta** | $s_i=\dfrac{b_i-a_i}{\max(a_i,b_i)}\in[-1,1]$; $a_i$ = distancia media al propio grupo, $b_i$ = distancia media al grupo vecino más cercano | **Maximizar** la silueta media (cerca de 1 = bien separado; < 0 = mal asignado) |
| **Calinski–Harabasz** | $\mathrm{CH}=\dfrac{SS_B/(k-1)}{SS_W/(n-k)}$ (varianza entre grupos / dentro) | **Maximizar** |
| **BIC** (GMM) | $-2\ln\hat L+p\ln n$ | **Minimizar** (sección 5) |
| **Estabilidad** | ARI entre agrupaciones de remuestras bootstrap | Preferir $k$ cuyos grupos se **repiten** (sección 6) |

Implementamos la silueta desde cero (la contrastamos con `silhouette_samples`) y comparamos $k=2,\dots,8$.

````python
# ============================================================
# 2. Elegir k: codo (inercia), silueta (desde cero) y Calinski-Harabasz
# ============================================================
from sklearn.metrics import calinski_harabasz_score, silhouette_samples, silhouette_score

def silueta(X, lab):
    """s_i = (b_i - a_i) / max(a_i, b_i); a = distancia media a su grupo, b = distancia media al grupo vecino mas cercano."""
    D = np.sqrt(dist2(X, X))
    s = np.zeros(len(X))
    for i in range(len(X)):
        propio = lab == lab[i]
        if propio.sum() == 1:
            continue                                      # s_i = 0 para grupos de un solo punto
        a = D[i, propio].sum() / (propio.sum() - 1)
        b = min(D[i, lab == c].mean() for c in np.unique(lab) if c != lab[i])
        s[i] = (b - a) / max(a, b)
    return s

s_prop = silueta(X, lab_b)
assert np.allclose(s_prop, silhouette_samples(X, lab_b)) and np.isclose(s_prop.mean(), silhouette_score(X, lab_b))
print(f"Silueta propia == silhouette_samples (k = 4): media = {s_prop.mean():.3f}")

filas = []
for k in range(2, 9):
    lb, _, ine, _ = kmeans(X, k, n_init=10, seed=k)
    filas.append({"k": k, "inercia": ine, "silueta": silhouette_score(X, lb), "Calinski_Harabasz": calinski_harabasz_score(X, lb)})
sel = pd.DataFrame(filas).set_index("k")
print(sel.round(3).to_string())
print(f"k elegido por silueta = {sel['silueta'].idxmax()} | por Calinski-Harabasz = {sel['Calinski_Harabasz'].idxmax()}")
assert sel["silueta"].idxmax() == 4 and np.all(np.diff(sel["inercia"]) < 0)          # la inercia SIEMPRE baja con k: por eso no sirve sola

# Segundo conjunto SINTETICO (3 tipos de cafe: variables estandarizadas), para la practica
X3, y3 = make_blobs(n_samples=300, centers=3, cluster_std=1.0, center_box=(-8, 8), random_state=7)

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].plot(sel.index, sel["inercia"], "o-", color=PALETA[0]); ax[0].set_title("Codo: inercia (siempre baja con k)")
ax[1].plot(sel.index, sel["silueta"], "o-", color=PALETA[1]); ax[1].axvline(4, color="grey", ls="--"); ax[1].set_title("Silueta media (maximo)")
ax[2].plot(sel.index, sel["Calinski_Harabasz"], "o-", color=PALETA[2]); ax[2].set_title("Calinski-Harabasz (maximo)")
for a in ax: a.set_xlabel("k")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) La silueta propia **coincide** con `silhouette_samples` punto por punto (media **0.676** con $k=4$). (2) **El codo es ambiguo:** la inercia baja de 5 385 ($k=2$) a 3 143 ($k=3$) y a 1 120 ($k=4$), y después cae poco (978, 850, ...): hay un codo en $k=4$, pero hay que *mirarlo*; y la inercia es estrictamente decreciente (el `assert` lo comprueba). (3) La **silueta** es máxima en $k=4$ (**0.676**; con $k=3$, 0.595 y con $k=5$, 0.601) y **Calinski–Harabasz** también (**1 556.9** frente a 706.8 y 1 348.0): los tres criterios apuntan a 4, que es el valor verdadero. (4) **No siempre coincidirán:** en datos con grupos solapados o de tamaños muy distintos, la silueta tiende a favorecer valores bajos de $k$. (5) Estos criterios dicen qué $k$ da **grupos compactos y separados**, no qué $k$ es **útil para el negocio**: una segmentación en 6 tipos de finca puede ser más accionable que una «óptima» en 3.

## 3. Clustering Jerárquico Aglomerativo 🌳

En lugar de fijar $k$ de antemano, el método **aglomerativo** empieza con cada punto como su propio grupo y, en cada paso, **une los dos grupos más cercanos**, hasta tener uno solo. El resultado es un **dendrograma** (árbol) que se **corta** a la altura deseada. Todo depende de cómo se mida la distancia entre **grupos** (el *enlace*):

| Enlace | Distancia entre grupos $A$ y $B$ | Tendencia |
|---|---|---|
| **Simple** | $\min_{a\in A,b\in B}d(a,b)$ | **Efecto cadena**: une grupos por un puente de puntos; detecta valores atípicos |
| **Completo** | $\max_{a\in A,b\in B}d(a,b)$ | Grupos compactos de diámetro parecido |
| **Promedio** | $\frac{1}{|A||B|}\sum_{a,b}d(a,b)$ | Intermedio |
| **Ward** | Aumento de la suma de cuadrados al unir: $\sqrt{\tfrac{2|A||B|}{|A|+|B|}}\,\lVert\bar a-\bar b\rVert$ (altura de SciPy) | Grupos esféricos y parecidos, como k-means |

La **correlación cofenética** mide cuánto respeta el dendrograma las distancias originales (cerca de 1 = fiel). Implementamos el algoritmo ingenuo ($O(n^3)$, solo para 60 puntos) y comprobamos que las **alturas de unión** y el **corte en 4 grupos** coinciden con `scipy.cluster.hierarchy.linkage` para los cuatro enlaces; luego comparamos los enlaces con los 400 puntos.

````python
# ============================================================
# 3. Clustering jerarquico aglomerativo desde cero (4 enlaces) vs SciPy
# ============================================================
from scipy.cluster.hierarchy import cophenet, dendrogram, fcluster, linkage
from scipy.spatial.distance import pdist

def aglomerativo(X, metodo, k_corte=4):
    """Une en cada paso el par de grupos mas cercano. Devuelve las alturas de union y las etiquetas al llegar a k_corte grupos."""
    X = np.asarray(X, float); D = np.sqrt(dist2(X, X))
    grupos = {i: [i] for i in range(len(X))}
    alturas, etiquetas = [], None
    while len(grupos) > 1:
        mejor = (np.inf, None, None)
        claves = list(grupos)
        for ia, a in enumerate(claves):
            for b in claves[ia + 1:]:
                A, B = grupos[a], grupos[b]
                if metodo == "single":
                    d = D[np.ix_(A, B)].min()
                elif metodo == "complete":
                    d = D[np.ix_(A, B)].max()
                elif metodo == "average":
                    d = D[np.ix_(A, B)].mean()
                else:                                     # Ward: aumento de la suma de cuadrados al unir (altura de SciPy)
                    ca, cb = X[A].mean(0), X[B].mean(0)
                    d = np.sqrt(2 * len(A) * len(B) / (len(A) + len(B))) * np.linalg.norm(ca - cb)
                if d < mejor[0]:
                    mejor = (d, a, b)
        d, a, b = mejor
        grupos[a] = grupos[a] + grupos.pop(b)
        alturas.append(d)
        if len(grupos) == k_corte:
            etiquetas = np.empty(len(X), int)
            for j, g in enumerate(grupos.values()):
                etiquetas[g] = j
    return np.array(alturas), etiquetas

# Muestra de 60 puntos (el algoritmo ingenuo es O(n^3)): alturas de union identicas a scipy.cluster.hierarchy.linkage
sub = np.random.default_rng(3).choice(len(X), 60, replace=False)
Xs = X[sub]
for met in ("single", "complete", "average", "ward"):
    h_prop, lab_prop = aglomerativo(Xs, met)
    Z = linkage(Xs, method=met)
    assert np.allclose(h_prop, Z[:, 2]), met
    assert adjusted_rand_score(lab_prop, fcluster(Z, 4, criterion="maxclust")) == 1.0, met
print("Alturas de union y cortes en 4 grupos: propios == scipy (single, complete, average, ward)")

# Con los 400 puntos: calidad del corte en 4 grupos y correlacion cofenetica por tipo de enlace
D400 = pdist(X)
filas = []
for met in ("single", "complete", "average", "ward"):
    Z = linkage(X, method=met)
    lab_c = fcluster(Z, 4, criterion="maxclust")
    filas.append({"enlace": met, "ARI_vs_real": adjusted_rand_score(y_real, lab_c), "tamanos": sorted(np.bincount(lab_c)[1:].tolist()), "cofenetica": cophenet(Z, D400)[0]})
tabj = pd.DataFrame(filas).set_index("enlace")
print(tabj.round(3).to_string())

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5), gridspec_kw={"width_ratios": [1.5, 1]})
Zw = linkage(X, "ward")
dendrogram(Zw, truncate_mode="lastp", p=16, ax=ax[0], color_threshold=Zw[-3, 2] + 1e-9, above_threshold_color="grey")
ax[0].axhline(Zw[-3, 2], color="k", ls="--", lw=0.8); ax[0].set_title("Dendrograma de Ward (16 hojas); corte = 4 grupos"); ax[0].set_ylabel("altura de union")
lab_s = fcluster(linkage(X, "single"), 4, criterion="maxclust")
for j in np.unique(lab_s):
    ax[1].scatter(*X[lab_s == j].T, s=10, color=PALETA[j % 10])
ax[1].set_title("Enlace simple: efecto cadena (un grupo enorme)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Las alturas de unión y los cortes propios son **idénticos** a SciPy para los cuatro enlaces (los `assert` pasan). (2) **El enlace importa mucho:** con el corte en 4 grupos, **Ward** recupera los grupos casi a la perfección (ARI **0.987**, tamaños 98/100/100/102), **promedio** 0.948 y **completo** 0.936, mientras que el **simple** obtiene ARI **0.496** con tamaños (1, 1, 198, 200): aísla dos valores extremos como grupos de un punto y **fusiona** los grupos reales de dos en dos (efecto cadena). (3) **La correlación cofenética no mide acierto:** la más alta es la del enlace promedio (**0.876**), pero no la del que mejor agrupa (Ward, 0.866); mide fidelidad a las distancias, no calidad de los grupos. (4) Ward es el mejor aquí porque los grupos son casi esféricos y de tamaño similar, **justo lo que asume**; con otras formas dejaría de serlo (sección 6). (5) Ventaja del jerárquico: **no hace falta fijar $k$** de antemano y el dendrograma muestra la estructura en varios niveles; su costo, $O(n^2)$ de memoria, lo limita a decenas de miles de puntos.

## 4. DBSCAN: Grupos por Densidad 🌌

**DBSCAN** define los grupos como **regiones densas** separadas por regiones ralas. Dos parámetros: el radio $\varepsilon$ y el mínimo de puntos `min_pts`. Un punto es **núcleo** si su vecindad $\{x':\lVert x-x'\rVert\le\varepsilon\}$ (incluido él) tiene al menos `min_pts` puntos; es **frontera** si no es núcleo pero está en la vecindad de uno; y es **ruido** en otro caso. Un grupo es el conjunto de núcleos **alcanzables por densidad** entre sí más sus fronteras. Ventajas: **no exige $k$**, encuentra **formas arbitrarias** y **marca el ruido**. Límites: un único $\varepsilon$ para todo el conjunto (falla con **densidades distintas**) y alta sensibilidad a $\varepsilon$. Para elegirlo se usa el **gráfico de $k$-distancias**: se ordenan las distancias de cada punto a su `min_pts`-ésimo vecino y se busca el codo.

Aplicamos DBSCAN desde cero (idéntico a scikit-learn) a **dos lunas entrelazadas** con 20 puntos de ruido uniforme, donde k-means falla por construcción.

````python
# ============================================================
# 4. DBSCAN desde cero vs scikit-learn; eps con el grafico de k-distancias
# ============================================================
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

def dbscan(X, eps, min_pts):
    """Etiquetas 0..K-1 y -1 para ruido. Nucleo: punto con >= min_pts vecinos a distancia <= eps (contandose a si mismo)."""
    n = len(X)
    vecinos = [np.flatnonzero(r <= eps ** 2) for r in dist2(X, X)]
    nucleo = np.array([len(v) >= min_pts for v in vecinos])
    lab = np.full(n, -1)
    c = 0
    for i in range(n):
        if lab[i] != -1 or not nucleo[i]:
            continue
        lab[i] = c; pila = [i]
        while pila:                                       # expandir el grupo por alcanzabilidad de densidad
            j = pila.pop()
            if not nucleo[j]:
                continue                                  # un punto frontera se une pero no expande
            for v in vecinos[j]:
                if lab[v] == -1:
                    lab[v] = c; pila.append(v)
        c += 1
    return lab, nucleo

# Datos SINTETICOS: dos lunas entrelazadas + 20 puntos de ruido uniforme
Xl, yl = make_moons(n_samples=300, noise=0.07, random_state=SEED)
Xl = np.vstack([Xl, np.random.default_rng(SEED).uniform([-1.5, -1], [2.5, 1.5], (20, 2))])
yl = np.r_[yl, -np.ones(20, int)]

eps, min_pts = 0.2, 5
lab_d, nucleo = dbscan(Xl, eps, min_pts)
sk_d = DBSCAN(eps=eps, min_samples=min_pts).fit(Xl)
assert np.array_equal(lab_d, sk_d.labels_) and np.array_equal(np.flatnonzero(nucleo), sk_d.core_sample_indices_)
print(f"DBSCAN propio == sklearn | grupos: {lab_d.max() + 1} | ruido: {np.sum(lab_d == -1)} | nucleos: {nucleo.sum()} | fronteras: {np.sum((lab_d >= 0) & ~nucleo)}")

# k-means (k = 2) sobre las mismas lunas: falla porque supone grupos convexos y redondos
lab_km = KMeans(2, n_init=10, random_state=SEED).fit_predict(Xl)
real = yl >= 0
print(f"ARI sobre los puntos no-ruido: DBSCAN = {adjusted_rand_score(yl[real], lab_d[real]):.3f} | k-means = {adjusted_rand_score(yl[real], lab_km[real]):.3f}")
assert adjusted_rand_score(yl[real], lab_d[real]) > 0.9 > adjusted_rand_score(yl[real], lab_km[real])

# Sensibilidad a eps (min_pts = 5): numero de grupos y de puntos de ruido
filas = [{"eps": e, "grupos": (l := dbscan(Xl, e, min_pts)[0]).max() + 1, "ruido": int(np.sum(l == -1))} for e in (0.05, 0.1, 0.15, 0.2, 0.3, 0.5, 0.8)]
sens = pd.DataFrame(filas).set_index("eps")
print("\nSensibilidad a eps:\n", sens.to_string())

# Grafico de k-distancias (k = min_pts): el "codo" sugiere eps
kd = np.sort(np.sqrt(np.sort(dist2(Xl, Xl), axis=1)[:, min_pts - 1]))
fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
ax[0].plot(kd, color=PALETA[0]); ax[0].axhline(eps, color="k", ls="--", label=f"eps = {eps}"); ax[0].legend()
ax[0].set_title(f"Distancia al vecino {min_pts} (ordenada)"); ax[0].set_xlabel("puntos ordenados")
for j in range(-1, lab_d.max() + 1):
    ax[1].scatter(*Xl[lab_d == j].T, s=12, color="grey" if j == -1 else PALETA[j], marker="x" if j == -1 else "o")
ax[1].set_title("DBSCAN (gris = ruido)")
for j in range(2):
    ax[2].scatter(*Xl[lab_km == j].T, s=12, color=PALETA[j + 3])
ax[2].set_title("k-means (k = 2)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) DBSCAN propio **coincide** con scikit-learn en etiquetas y puntos núcleo (2 grupos, 9 puntos de ruido, 306 núcleos y 5 fronteras). (2) Sobre los puntos no-ruido, DBSCAN recupera las lunas con ARI **1.000**, y k-means con $k=2$ solo **0.247**: corta cada luna por la mitad porque su criterio (la distancia al centroide) solo produce fronteras **rectas** entre grupos convexos. (3) **Sensibilidad a $\varepsilon$:** con 0.05 casi todo es ruido (296 de 320 puntos); con 0.1 hay 9 grupos y 61 de ruido; entre **0.15 y 0.3** el resultado es estable (2 grupos); con 0.5 las dos lunas se **fusionan** en uno y con 0.8 desaparece el ruido. El gráfico de $k$-distancias sugiere el rango correcto, pero **hay que verificarlo mirando el resultado**. (4) De los 20 puntos de ruido uniforme, DBSCAN marcó 9: los demás cayeron **dentro** de las lunas y es razonable que se absorban (la etiqueta «ruido» de partida no es una verdad absoluta para DBSCAN). (5) Limitación no vista aquí: con grupos de **densidades muy distintas** no existe un $\varepsilon$ único bueno (se verá en la sección 6).

## 5. Mezclas Gaussianas (GMM) y el Algoritmo EM 🔔

K-means asigna cada punto a **un** grupo. Un **modelo de mezclas gaussianas** es probabilístico: supone que cada punto sale de una de $k$ normales con **pesos** $\pi_j$, **medias** $\mu_j$ y **covarianzas** $\Sigma_j$:
$$p(x)=\sum_{j=1}^{k}\pi_j\,\mathcal{N}(x\mid\mu_j,\Sigma_j),\qquad \sum_j\pi_j=1.$$
Se estima por máxima verosimilitud con el algoritmo **EM**, que alterna: **E** (calcular las **responsabilidades** $r_{ij}=P(\text{grupo }j\mid x_i)=\dfrac{\pi_j\mathcal{N}(x_i\mid\mu_j,\Sigma_j)}{\sum_l\pi_l\mathcal{N}(x_i\mid\mu_l,\Sigma_l)}$) y **M** (reestimar $\pi_j=N_j/n$, $\mu_j=\frac1{N_j}\sum_i r_{ij}x_i$ y $\Sigma_j=\frac1{N_j}\sum_i r_{ij}(x_i-\mu_j)(x_i-\mu_j)^{\top}$, con $N_j=\sum_i r_{ij}$). Cada iteración **no disminuye** la log-verosimilitud. Ventajas sobre k-means: grupos **elípticos con orientaciones y tamaños distintos**, **asignación blanda** (incertidumbre) y un criterio **estadístico** para elegir $k$: el **BIC** $=-2\ln\hat L+p\ln n$, con $p=(k-1)+kd+kd(d+1)/2$ parámetros, que **penaliza** la complejidad. (K-means es el caso límite de una GMM con covarianzas esféricas iguales y asignaciones duras.)

Implementamos EM con covarianza completa, inicializada con k-means, y lo contrastamos con `GaussianMixture` (mismos valores iniciales, tolerancia muy estricta).

````python
# ============================================================
# 5. Mezclas gaussianas (GMM) por EM desde cero; k con BIC
# ============================================================
from scipy.special import logsumexp
from sklearn.mixture import GaussianMixture

def logpdf_gauss(X, mu, S):                               # log N(x | mu, S) con Cholesky
    L = np.linalg.cholesky(S)
    z = np.linalg.solve(L, (X - mu).T)
    return -0.5 * (z ** 2).sum(0) - np.log(np.diag(L)).sum() - 0.5 * X.shape[1] * np.log(2 * np.pi)

def gmm_em(X, pesos, medias, covs, tol=1e-10, max_iter=1000, reg=1e-6):
    n, d = X.shape; k = len(pesos); ll_ant = -np.inf
    for it in range(max_iter):
        lp = np.column_stack([np.log(pesos[j]) + logpdf_gauss(X, medias[j], covs[j]) for j in range(k)])
        ll = logsumexp(lp, axis=1).sum()                  # log-verosimilitud
        R = np.exp(lp - logsumexp(lp, axis=1, keepdims=True))   # E: responsabilidades P(grupo j | x_i)
        if abs(ll - ll_ant) < tol * n:
            break
        ll_ant = ll
        Nk = R.sum(0)                                     # M: pesos, medias y covarianzas ponderados
        pesos = Nk / n
        medias = (R.T @ X) / Nk[:, None]
        covs = np.array([((R[:, j, None] * (X - medias[j])).T @ (X - medias[j])) / Nk[j] + reg * np.eye(d) for j in range(k)])
    return pesos, medias, covs, ll, R, it + 1

def bic_gmm(ll, n, d, k):                                 # parametros: k-1 pesos + k*d medias + k*d(d+1)/2 covarianzas
    return -2 * ll + (k - 1 + k * d + k * d * (d + 1) / 2) * np.log(n)

def init_kmeans(X, k, seed=0):
    lab = kmeans(X, k, n_init=5, seed=seed)[0]
    return (np.bincount(lab, minlength=k) / len(X), np.array([X[lab == j].mean(0) for j in range(k)]),
            np.array([np.cov(X[lab == j].T) + 1e-6 * np.eye(X.shape[1]) for j in range(k)]))

n_, d_ = X.shape
w0, m0, S0 = init_kmeans(X, 4)
w, mu, S, ll, R, iters = gmm_em(X, w0, m0, S0)
sk_g = GaussianMixture(4, covariance_type="full", weights_init=w0, means_init=m0, precisions_init=np.linalg.inv(S0),
                       tol=1e-10, max_iter=1000, reg_covar=1e-6).fit(X)
assert np.isclose(ll, sk_g.score(X) * n_, rtol=1e-6) and np.allclose(np.sort(w), np.sort(sk_g.weights_), atol=1e-4)
assert np.isclose(bic_gmm(ll, n_, d_, 4), sk_g.bic(X), rtol=1e-6)
print(f"EM propio == GaussianMixture: log-verosimilitud {ll:.2f} ({iters} iteraciones) | BIC {bic_gmm(ll, n_, d_, 4):.2f}")

# BIC para k = 1..8 (menor = mejor): mide ajuste penalizando parametros
filas = []
for k in range(1, 9):
    w_, m_, S_, ll_, R_, _ = gmm_em(X, *init_kmeans(X, k, seed=k))
    filas.append({"k": k, "log_verosimilitud": ll_, "BIC": bic_gmm(ll_, n_, d_, k)})
bics = pd.DataFrame(filas).set_index("k")
print(bics.round(1).to_string(), f"\nk elegido por BIC = {bics['BIC'].idxmin()}")
assert bics["BIC"].idxmin() == 4

# Asignacion BLANDA: a diferencia de k-means, dice cuanta incertidumbre hay
dura = R.argmax(1)
incierto = R.max(1) < 0.9
print(f"ARI del GMM vs grupos reales = {adjusted_rand_score(y_real, dura):.3f} | puntos con responsabilidad maxima < 0.9: {incierto.sum()} de {n_}")

fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
ax[0].plot(bics.index, bics["BIC"], "o-", color=PALETA[0]); ax[0].axvline(4, color="grey", ls="--"); ax[0].set_xlabel("k"); ax[0].set_title("BIC del GMM (menor = mejor)")
ax[1].scatter(*X.T, c=[PALETA[j] for j in dura], s=np.where(incierto, 45, 12), edgecolor=np.where(incierto, "k", "none"))
for j in range(4):
    vals, vecs = np.linalg.eigh(S[j]); ang = np.degrees(np.arctan2(vecs[1, 1], vecs[0, 1]))
    ax[1].add_patch(plt.matplotlib.patches.Ellipse(mu[j], 4 * np.sqrt(vals[1]), 4 * np.sqrt(vals[0]), angle=ang, fill=False, color="k"))
ax[1].set_title("GMM: elipses a 2 desv. est. y puntos inciertos (borde negro)")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) El EM propio alcanza la **misma log-verosimilitud** que `GaussianMixture` ($-1\,782.01$, en 9 iteraciones desde el inicio de k-means) y el mismo BIC (**3 701.82**). (2) **El BIC elige $k=4$**: baja de 4 522 ($k=1$) a 3 995, 3 882 y 3 702, y luego **sube** (3 718, 3 741, ...) porque la mejora de la verosimilitud ya no compensa los parámetros extra (cada grupo añadido cuesta 6 parámetros en 2 dimensiones). Es la ventaja de un modelo probabilístico: un criterio de selección con fundamento. (3) El GMM recupera los grupos con ARI **0.987** (k-means: 0.942): como los grupos tienen dispersiones distintas (0.8 a 1.6) y se tocan en los bordes, las elipses ajustan mejor que las esferas. (4) La asignación **blanda** señala dónde está la duda: solo **1 de 400** puntos tiene responsabilidad máxima < 0.9; con grupos más solapados serían muchos más, y eso es **información** que k-means (asignación dura) esconde. (5) Cuidado: el BIC elige entre modelos **gaussianos**; si los grupos reales no son normales (por ejemplo, las lunas), un GMM los descompone en varias componentes y el BIC puede preferir un $k$ grande que no corresponde a «grupos» interpretables.

## 6. Limitaciones y Validez: Forma, Escalado, Ruido y Estabilidad ⚠️

Cuatro advertencias que todo análisis de agrupamientos debe atender:

1. **Cada algoritmo asume una forma de grupo.** No hay un «mejor» método universal; hay que comprobar el supuesto.
2. **El escalado decide el resultado.** Las distancias suman diferencias al cuadrado de **todas** las variables: una con valores grandes domina. Casi siempre se **estandariza** ($z$-scores) antes de agrupar.
3. **Siempre hay grupos.** Un algoritmo devuelve $k$ grupos aun si los datos son ruido. Antes de agrupar conviene medir la **tendencia al agrupamiento**: el **estadístico de Hopkins** $H=\dfrac{\sum u_i}{\sum u_i+\sum w_i}$ compara la distancia al vecino más cercano de puntos **uniformes** ($u_i$) con la de puntos **reales** ($w_i$): ≈ 0.5 si los datos son aleatorios, cerca de 1 si hay estructura.
4. **Validar.** La validación **interna** (silueta, BIC) no usa etiquetas; la **externa** (índice de Rand ajustado, **ARI**: 1 = coincidencia perfecta, ≈ 0 = azar) las necesita y solo se tiene en ejercicios. La **estabilidad** (¿se repiten los grupos al remuestrear?) es una validación sin etiquetas.

Probamos cuatro conjuntos sintéticos × cuatro algoritmos (con DBSCAN y $\varepsilon$ elegido mirando cada conjunto), el escalado con el conjunto *wine* (13 variables químicas, 3 cultivares, incluido en scikit-learn), el ruido uniforme con Hopkins y la estabilidad bootstrap.

````python
# ============================================================
# 6. Limitaciones y validez: forma de los grupos, escalado, ruido y estabilidad
# ============================================================
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler

# (a) Cuatro conjuntos SINTETICOS x cuatro algoritmos: ARI frente a los grupos verdaderos (1 = perfecto, ~0 = azar)
rg = np.random.default_rng(11)
Xb, yb = make_blobs(300, centers=[(-5, 0), (0, 5), (5, 0)], cluster_std=1.0, random_state=11)
band = rg.normal(size=(300, 2)) * [3.0, 0.4]                       # tres bandas alargadas y paralelas, giradas 35 grados
ya = np.arange(300) % 3
band[:, 1] += (ya - 1) * 2.0
th = np.radians(35)
Xa = band @ np.array([[np.cos(th), -np.sin(th)], [np.sin(th), np.cos(th)]]).T
Xv, yv = make_blobs(300, centers=[(-6, 0), (0, 0), (6, 0)], cluster_std=[0.6, 3.0, 0.6], random_state=11)
Xm, ym = make_moons(300, noise=0.06, random_state=11)
# (datos, etiquetas reales, k, eps de DBSCAN elegido MIRANDO el grafico de k-distancias de cada conjunto)
conjuntos = {"Blobs redondos": (Xb, yb, 3, 0.4), "Bandas alargadas": (Xa, ya, 3, 0.25), "Varianzas distintas": (Xv, yv, 3, 0.2), "Dos lunas": (Xm, ym, 2, 0.3)}

filas = []
for nombre, (Xd, yd, k, e_) in conjuntos.items():
    Z = StandardScaler().fit_transform(Xd)
    l_km = KMeans(k, n_init=10, random_state=SEED).fit_predict(Z)
    l_wd = fcluster(linkage(Z, "ward"), k, criterion="maxclust")
    l_db = dbscan(Z, e_, 5)[0]
    l_gm = GaussianMixture(k, covariance_type="full", n_init=5, random_state=SEED).fit_predict(Z)
    filas.append({"conjunto": nombre, "k-means": adjusted_rand_score(yd, l_km), "Ward": adjusted_rand_score(yd, l_wd),
                  "DBSCAN": adjusted_rand_score(yd, l_db), "GMM": adjusted_rand_score(yd, l_gm), "ruido_DBSCAN": int(np.sum(l_db == -1))})
res_f = pd.DataFrame(filas).set_index("conjunto")
ruido_db = res_f.pop("ruido_DBSCAN")
print("ARI por conjunto y algoritmo (datos estandarizados):\n", res_f.round(2).to_string())
print("Puntos que DBSCAN dejo como ruido (de 300):", ruido_db.to_dict())
assert res_f.loc["Dos lunas", "DBSCAN"] > 0.9 > res_f.loc["Dos lunas", "k-means"]
assert res_f.loc["Bandas alargadas", "GMM"] > 0.9 > res_f.loc["Bandas alargadas", "k-means"]

# (b) El ESCALADO decide: vino (13 variables quimicas, 3 cultivares) con y sin estandarizar
W = load_wine()
Xw, yw = W.data, W.target
Zw = StandardScaler().fit_transform(Xw)
ari_raw = adjusted_rand_score(yw, KMeans(3, n_init=10, random_state=SEED).fit_predict(Xw))
ari_std = adjusted_rand_score(yw, KMeans(3, n_init=10, random_state=SEED).fit_predict(Zw))
dom = pd.Series(Xw.std(0), index=W.feature_names).sort_values(ascending=False)
print(f"\nVino: variable de mayor desviacion = '{dom.index[0]}' (sd {dom.iloc[0]:.0f}) frente a '{dom.index[-1]}' (sd {dom.iloc[-1]:.2f})")
print(f"ARI k-means sin escalar = {ari_raw:.3f} | estandarizado = {ari_std:.3f}")
assert ari_std > ari_raw + 0.4

# (c) Siempre hay grupos: k-means sobre RUIDO UNIFORME y el estadistico de Hopkins (tendencia al agrupamiento)
def hopkins(Z, m=60, seed=0):
    r = np.random.default_rng(seed); n_, d = Z.shape
    U = r.uniform(Z.min(0), Z.max(0), (m, d)); idx_ = r.choice(n_, m, replace=False)
    u = np.sqrt(dist2(U, Z).min(1))                                # distancia de puntos uniformes al dato mas cercano
    D = np.sqrt(dist2(Z[idx_], Z)); D[np.arange(m), idx_] = np.inf
    w = D.min(1)                                                   # distancia de puntos reales a su vecino real
    return u.sum() / (u.sum() + w.sum())
ruido = np.random.default_rng(5).uniform(0, 1, (300, 2))
lr = KMeans(3, n_init=10, random_state=SEED).fit_predict(ruido)
H_ruido, H_blobs = np.mean([hopkins(ruido, seed=s) for s in range(20)]), np.mean([hopkins(StandardScaler().fit_transform(Xb), seed=s) for s in range(20)])
print(f"\nRuido uniforme con k-means (k = 3): silueta = {silhouette_score(ruido, lr):.3f}  (no hay grupos reales, y aun asi 'se ve razonable')")
print(f"Hopkins (0.5 = aleatorio, cerca de 1 = agrupado): ruido = {H_ruido:.2f} | blobs = {H_blobs:.2f}")
assert H_ruido < 0.65 < H_blobs

# (d) Estabilidad: ARI medio entre agrupaciones de remuestras bootstrap (si el k es bueno, los grupos se repiten)
def estabilidad(Z, k, B=25, seed=0):
    r = np.random.default_rng(seed); aris = []
    for _ in range(B):
        i1, i2 = r.choice(len(Z), len(Z)), r.choice(len(Z), len(Z))
        m1, m2 = KMeans(k, n_init=3, random_state=0).fit(Z[i1]), KMeans(k, n_init=3, random_state=0).fit(Z[i2])
        aris.append(adjusted_rand_score(m1.predict(Z), m2.predict(Z)))          # se comparan en TODOS los puntos
    return np.mean(aris)
est = pd.Series({k: estabilidad(X, k) for k in range(2, 8)})
print("\nEstabilidad (ARI medio entre remuestras) por k:\n", est.round(3).to_string())

fig, ax = plt.subplots(1, 3, figsize=(15, 4.3))
im = ax[0].imshow(res_f.values, vmin=0, vmax=1, cmap="Blues"); ax[0].set_xticks(range(4)); ax[0].set_xticklabels(res_f.columns, fontsize=8)
ax[0].set_yticks(range(4)); ax[0].set_yticklabels(res_f.index, fontsize=8); ax[0].grid(False); ax[0].set_title("ARI: quien funciona donde")
for i in range(4):
    for j in range(4):
        ax[0].text(j, i, f"{res_f.values[i, j]:.2f}", ha="center", va="center", color="k" if res_f.values[i, j] < 0.6 else "w")
ax[1].scatter(*ruido.T, c=[PALETA[j] for j in lr], s=12); ax[1].set_title("k-means (k = 3) sobre ruido uniforme")
ax[2].plot(est.index, est.values, "o-", color=PALETA[2]); ax[2].set_xlabel("k"); ax[2].set_title("Estabilidad bootstrap por k")
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) **Cada método tiene su terreno** (ARI): en *blobs* redondos los cuatro aciertan (1.00; DBSCAN lo logra con $\varepsilon$ ajustado). En **bandas alargadas**, k-means (**0.22**) y Ward (0.14) fallan porque cortan perpendicularmente a la dirección larga; solo el **GMM** (0.96, covarianzas completas) las recupera; DBSCAN llega a 0.57 y marca 53 puntos como ruido. Con **varianzas distintas** el grupo central difuso confunde a k-means (0.51) y Ward (0.56); el GMM (0.91) y DBSCAN (0.92) lo resuelven, pero DBSCAN deja **96 de 300 puntos como ruido**, es decir, su ARI es alto a costa de renunciar a un tercio de los datos. En **dos lunas** solo DBSCAN (1.00). (2) **El escalado cambia todo:** en *wine*, `proline` tiene desviación **314** frente a **0.12** de `nonflavanoid_phenols`, y k-means sin escalar domina por la prolina: ARI **0.371**; **estandarizando**, **0.897**. (3) **Siempre hay grupos:** k-means sobre ruido uniforme da silueta **0.403**, un valor que en otros contextos pasaría por «razonable». Lo delata Hopkins: **0.49** (aleatorio) frente a **0.82** en los *blobs*. Una silueta de 0.4 **no prueba** que haya grupos. (4) **Estabilidad:** $k=4$ es muy estable (ARI medio **0.996**), $k=2$ también (1.000, trivialmente: casi cualquier estructura se parte en dos de forma estable), mientras que $k=3$ (0.821), $k=5$ (0.856) y $k=7$ (0.738) son menos estables; la estabilidad **no basta sola** (favorece $k$ pequeños) pero **confirma** la elección de 4. (5) **Conclusión práctica:** estandariza, mira la forma, mide la tendencia al agrupamiento, compara más de un algoritmo y un criterio, y **valida con el conocimiento del dominio**.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **k-means** | Minimiza la inercia con Lloyd; converge pero a un **óptimo local** (un inicio malo dio inercia 3 013.6 frente a 1 120.3; 7 % de 60 inicios fallaron). Usar *k-means++* y varias inicializaciones. Propio == `KMeans` |
| **Elegir $k$** | La inercia siempre baja: no sirve sola. Silueta (0.676), Calinski–Harabasz (1 556.9) y BIC (3 701.8) coincidieron en $k=4$; la estabilidad lo confirmó. Ninguno es infalible ni sustituye al criterio del dominio |
| **Jerárquico** | El enlace decide: Ward ARI 0.987, simple 0.496 (efecto cadena). La correlación cofenética mide fidelidad a las distancias, no acierto. Alturas propias == SciPy |
| **DBSCAN** | Grupos por densidad, formas arbitrarias y ruido explícito; sensible a $\varepsilon$ (estable entre 0.15 y 0.3 en las lunas) y con un solo $\varepsilon$ no maneja densidades distintas. Propio == scikit-learn |
| **GMM** | EM con covarianzas completas, asignación blanda y **BIC**. Recuperó las bandas alargadas (ARI 0.96) donde k-means falló (0.22). Propio == `GaussianMixture` |
| **Escalado** | Sin estandarizar, la variable de mayor escala domina (wine: ARI 0.371 frente a 0.897) |
| **Validez** | Siempre se obtienen grupos: silueta 0.40 sobre ruido uniforme. Hopkins (0.49 frente a 0.82) y la estabilidad bootstrap ayudan; la validación externa (ARI) solo existe en ejercicios |
