## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **«para no ingenieros»** del cuaderno 00 del Módulo 01 (**cuaderno estándar**). Imagina que llegas a la **plaza de mercado** a comprar papa y cada puesto te dice un precio distinto. Tu pregunta es sencilla: **¿cuánto vale, en general, una libra de papa?** Para responder, los datos se resumen **con un solo número**, y hay tres formas famosas de elegirlo:

| Medida | En la plaza de mercado | ¿Cómo se obtiene? |
|---|---|---|
| **Promedio** (o *media*) | «Si todos los puestos cobraran lo mismo y la plaza recaudara igual» | Se **suman** todos los precios y se **reparte** entre los puestos |
| **Mediana** | «El puesto que queda **en el medio** de la fila, ordenada de barato a caro» | Se **ordenan** los precios y se toma el del centro |
| **Moda** | «El precio que **más se repite**» | Se cuenta cuál aparece más veces |

Al terminar podrás explicar:
1. Por qué **un solo precio absurdo** mueve el promedio pero casi no mueve la mediana.
2. Qué significa que «el **90 %** de las veces el bus llega en menos de tantos minutos» (los **percentiles**).

> 🔗 Si luego quieres ver todo **desde cero y con rigor** (qué minimiza cada medida, cuánto aguanta cada una frente a datos malos): **cuaderno estándar**. Después de esto: **Dos fincas, el mismo promedio**.

## 1. Nueve Puestos y un Precio Absurdo 🥔

Visitamos **9 puestos** de la plaza y anotamos el precio de una libra de papa (en pesos; los datos son **inventados**). Calculamos el promedio, la mediana y la moda. Después **llega un puesto turístico** que cobra 9 000 pesos por libra (¡un precio exagerado!) y volvemos a calcular. ¿Cuál de los tres números se deja engañar?

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Precio de una libra de papa en 9 puestos (pesos, datos INVENTADOS)
precios = np.array([1800, 1900, 1900, 2000, 2000, 2000, 2100, 2200, 2300])


def tres_medidas(x):
    """Devuelve (promedio, mediana, moda) de una lista de precios."""
    return x.mean(), np.median(x), pd.Series(x).mode().iloc[0]


prom, med, moda = tres_medidas(precios)
print(f"Los 9 puestos  -> promedio: {prom:.0f} | mediana: {med:.0f} | moda: {moda:.0f}")
assert round(prom) == 2022 and med == 2000 and moda == 2000

# Llega un puesto turistico con un precio exagerado
con_turistico = np.append(precios, 9000)
prom2, med2, moda2 = tres_medidas(con_turistico)
print(f"Con el turistico -> promedio: {prom2:.0f} | mediana: {med2:.0f} | moda: {moda2:.0f}")
assert prom2 > 2700 and med2 == 2000 and moda2 == 2000                                   # el promedio sube MUCHO; la mediana y la moda no se mueven

fig, ejes = plt.subplots(1, 2, figsize=(11, 3.4), sharex=True)
for ax, datos, titulo in [(ejes[0], precios, "Solo los 9 puestos"), (ejes[1], con_turistico, "Con el puesto turístico")]:
    p_, m_, mo_ = tres_medidas(datos)
    ax.scatter(datos, np.zeros(len(datos)), s=90, color="#2563eb", alpha=0.6, zorder=3)
    ax.axvline(p_, color="#dc2626", lw=2.5, label=f"promedio {p_:.0f}"); ax.axvline(m_, color="#16a34a", lw=2.5, ls="--", label=f"mediana {m_:.0f}")
    ax.set_yticks([]); ax.set_xlabel("precio de la libra (pesos)"); ax.set_title(titulo, fontweight="bold", fontsize=10); ax.legend(fontsize=8, loc="upper center")
plt.tight_layout(); plt.show()
````

---
**Lección:** con los 9 puestos normales, los tres números cuentan casi lo mismo (promedio ≈ **2 022**, mediana y moda **2 000**). Pero con **un solo** puesto exagerado el **promedio salta a 2 720** (¡subió casi 700 pesos por culpa de un puesto!) mientras que **la mediana y la moda siguen en 2 000**. El promedio «se deja arrastrar» por los valores extremos porque **suma todo**; la mediana solo mira **quién queda en el medio**. Por eso, cuando hay datos muy raros (o una cola larga, como los ingresos), la mediana suele describir mejor a **la persona típica**.

## 2. La Fila de los Buses: ¿Qué es un Percentil? 🚌

Anotamos cuántos minutos esperaste el bus en **20 días** (datos **inventados**). Ordenados de menor a mayor forman una fila. El **percentil 90** es el tiempo que **deja atrás al 90 % de los días**: «el 90 % de las veces el bus llegó en ese tiempo o menos». La **mediana** es el percentil 50 (el puesto del medio). Así se hacen afirmaciones como «**casi siempre** esperas menos de X minutos», más útiles que solo el promedio.

````python
# Minutos de espera del bus en 20 dias (INVENTADOS)
espera = np.array([4, 5, 5, 6, 6, 7, 7, 8, 8, 9, 9, 10, 10, 11, 12, 13, 15, 18, 25, 40])

mediana_espera = np.percentile(espera, 50)
p90 = np.percentile(espera, 90)
promedio_espera = espera.mean()
print(f"Promedio: {promedio_espera:.1f} min | mediana (percentil 50): {mediana_espera:.1f} min | percentil 90: {p90:.1f} min")
assert mediana_espera == 9.0 and promedio_espera > mediana_espera                         # unos pocos dias terribles (25 y 40 min) jalan el promedio hacia arriba
assert (espera <= p90).mean() >= 0.9                                                      # al menos el 90 % de los dias esperaste p90 minutos o menos

fig, ax = plt.subplots(figsize=(9, 3.6))
ax.bar(range(1, 21), np.sort(espera), color=["#dc2626" if v > p90 else "#2563eb" for v in np.sort(espera)])
ax.axhline(mediana_espera, color="#16a34a", ls="--", lw=2, label=f"mediana ({mediana_espera:.1f} min)")
ax.axhline(p90, color="#f59e0b", ls="--", lw=2, label=f"percentil 90 ({p90:.1f} min)")
ax.set_xlabel("los 20 días, ordenados del más corto al más largo"); ax.set_ylabel("minutos de espera"); ax.legend()
ax.set_title("La fila de los 20 días: los rojos son el 10 % de peores esperas", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** la **mediana** es 9 minutos (la mitad de los días esperaste menos), pero el **promedio** es 11.4: **dos días terribles** (25 y 40 minutos) lo jalan hacia arriba. El **percentil 90** es 18.7 minutos: «en 9 de cada 10 días esperaste **menos de 19 minutos**». Para planear tu salida **con seguridad** te sirve más el percentil 90 que el promedio. (Las barras rojas son los pocos días que superan ese tiempo.)

## Resumen en una frase 🎯

> **El promedio reparte parejo pero se deja arrastrar por los valores extremos; la mediana y los percentiles dicen dónde cae un dato en la fila y no se asustan de los precios absurdos.**

### Lo que aprendiste hoy:
- ✅ **Promedio, mediana y moda** son tres formas de resumir; con datos «normales» se parecen, con datos raros **se separan**.
- ✅ Un **solo puesto exagerado** subió el promedio de 2 022 a 2 720, pero la **mediana siguió en 2 000**.
- ✅ Un **percentil** es un puesto en la fila: «el 90 % de los días esperaste menos de 19 minutos»; sirve para planear con seguridad.
- ✅ En la vereda con una finca grande, **solo 1 de 7 familias** ganaba más que el promedio: ahí la mediana es más honesta.
