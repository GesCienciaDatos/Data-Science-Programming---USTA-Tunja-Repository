## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **«para no ingenieros»** del cuaderno 02 del Módulo 01 (**cuaderno estándar**). Un dato sin gráfico es como una receta sin foto: se entiende, pero cuesta. Sin embargo, **no todo gráfico sirve para todo**: es como elegir la herramienta de la finca. Con un **machete** no destapas una botella; con un **destapador** no cortas caña. Hay que elegir el gráfico según **la pregunta**.

Al terminar podrás explicar:
1. **Qué gráfico usar** para ver cómo se reparten los datos, comparar grupos, ver si dos cosas se relacionan o seguir el tiempo.
2. Por qué **el mismo dato** puede verse distinto según **cuántas cajitas** tenga el histograma.
3. Cómo un **eje que no empieza en cero** hace parecer enorme una diferencia pequeña (¡la trampa más vieja de los gráficos!).

| Mi pregunta es... | Uso un gráfico de... | Se parece a... |
|---|---|---|
| ¿Cómo se **reparten** los precios? | **Histograma** | Apilar costales en cajones por rango de precio |
| ¿Cuál plaza es **más cara**? | **Cajas** (una por grupo) | Resumir cada plaza en una cajita: medio, mitad central y extremos |
| ¿**Se relacionan** lluvia y cosecha? | **Dispersión** (puntos) | Un mapa de la finca con un punto por parcela |
| ¿Cómo **cambia** el precio con los días? | **Línea** | El electrocardiograma del mercado |

> 🔗 Si luego quieres ver todo **desde cero y con medidas** (reglas para el ancho de las cajitas, cómo se calcula una caja, factor de mentira de un eje, colores accesibles): **cuaderno estándar**. Antes: **Dos fincas, el mismo promedio**.

## 1. Cuatro Preguntas, Cuatro Gráficos 🖼️

Con datos **inventados** de una plaza de mercado (30 días de precios, tres plazas y 12 parcelas) dibujamos **un gráfico para cada pregunta**. Cada uno se acompaña de una comprobación de que dibuja **todos** los datos.

````python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
# 30 dias de precio de la libra de papa (pesos, INVENTADOS), con una pequena subida
precio = np.round(2000 + 8 * np.arange(30) + rng.normal(0, 120, 30))
# 15 precios en cada una de tres plazas (INVENTADOS)
plazas = {"Plaza A": np.round(rng.normal(1900, 90, 15)), "Plaza B": np.round(rng.normal(2100, 90, 15)), "Plaza C": np.round(rng.normal(2300, 150, 15))}
# 12 parcelas (INVENTADAS): lluvia (mm) y rendimiento (t/ha)
lluvia = np.array([60, 75, 80, 90, 95, 100, 110, 115, 120, 130, 140, 150], dtype=float)
rendimiento = np.array([15, 17, 16, 19, 20, 21, 22, 21, 24, 25, 24, 27], dtype=float)

fig, ejes = plt.subplots(2, 2, figsize=(11, 7))
conteos, _, _ = ejes[0, 0].hist(precio, bins=6, color="#2563eb", edgecolor="white")
ejes[0, 0].set_title("Histograma: ¿cómo se reparten los precios?", fontweight="bold", fontsize=10); ejes[0, 0].set_xlabel("precio (pesos)"); ejes[0, 0].set_ylabel("días")
ejes[0, 1].boxplot(list(plazas.values())); ejes[0, 1].set_xticklabels(list(plazas)); ejes[0, 1].set_title("Cajas: ¿cuál plaza es más cara?", fontweight="bold", fontsize=10); ejes[0, 1].set_ylabel("precio (pesos)")
ejes[1, 0].scatter(lluvia, rendimiento, s=60, color="#16a34a"); ejes[1, 0].set_title("Dispersión: ¿se relacionan lluvia y cosecha?", fontweight="bold", fontsize=10); ejes[1, 0].set_xlabel("lluvia (mm)"); ejes[1, 0].set_ylabel("t/ha")
ejes[1, 1].plot(np.arange(1, 31), precio, color="#dc2626", lw=2, marker="o", ms=3); ejes[1, 1].set_title("Línea: ¿cómo cambia el precio día a día?", fontweight="bold", fontsize=10); ejes[1, 1].set_xlabel("día"); ejes[1, 1].set_ylabel("precio (pesos)")
plt.tight_layout(); plt.show()

assert conteos.sum() == 30                                                             # el histograma cuenta los 30 dias, ni uno mas ni uno menos
assert len(plazas["Plaza A"]) == len(plazas["Plaza B"]) == len(plazas["Plaza C"]) == 15 and np.median(plazas["Plaza A"]) < np.median(plazas["Plaza C"])
print("Mediana por plaza:", {k: float(np.median(v)) for k, v in plazas.items()})
````

---
**Lección:** cada gráfico responde **una** pregunta: el histograma muestra **cómo se reparten** los 30 precios (sumando las barras da 30); las cajas dicen enseguida que la **Plaza A es la más barata y la C la más cara** (y la C, además, **varía más**); la dispersión muestra que **a más lluvia, más cosecha**; y la línea muestra que el precio **sube con los días**. Si hubieras usado una línea para comparar plazas o un histograma para ver el tiempo, la gráfica **no respondería** tu pregunta.

## 2. Las Cajitas del Histograma: el Mismo Dato, Tres Caras 📊

Un histograma mete los datos en **cajitas** (rangos de precio) y cuenta cuántos caen en cada una. ¿**Cuántas** cajitas poner? Con **muy pocas**, no se ve la forma (todo queda en una o dos barras). Con **demasiadas**, casi cada barra tiene un solo dato y **aparece ruido** que parece forma pero no lo es. Probamos con **3**, **8** y **30** cajitas sobre los **mismos** 30 precios.

````python
fig, ejes = plt.subplots(1, 3, figsize=(12, 3.4), sharey=False)
resultado = {}
for ax, k in zip(ejes, (3, 8, 30)):
    conteos, bordes, _ = ax.hist(precio, bins=k, color="#7c3aed", edgecolor="white")
    resultado[k] = conteos
    ax.set_title(f"{k} cajitas", fontweight="bold", fontsize=10); ax.set_xlabel("precio (pesos)")
ejes[0].set_ylabel("días")
plt.tight_layout(); plt.show()

for k, conteos in resultado.items():
    assert len(conteos) == k and conteos.sum() == 30                                    # siempre los mismos 30 dias, solo cambia como se agrupan
print({k: int(c.max()) for k, c in resultado.items()}, "<- el mayor número de días en una sola cajita")
pocas = int((resultado[30] <= 1).sum())
print(f"Con 30 cajitas, {pocas} de las 30 tienen 0 o 1 día")
assert resultado[30].max() <= 4 and resultado[3].max() >= 10 and pocas >= 18               # con 30 cajitas casi todas tienen 0 o 1 dato; con 3, una se llena
````

---
**Lección:** son **los mismos 30 datos**, pero con **3 cajitas** una sola junta **13 días** (se pierde la forma) y con **30 cajitas** ninguna pasa de **4 días** y **23 de las 30** tienen 0 o 1 día (casi cada barra es un solo dato: puro ruido). No hay un número mágico, pero sí **reglas** para escoger (el cuaderno estándar las implementa) y un buen hábito: **probar varias cantidades** y confiar solo en lo que se ve **en todas**.

## 3. La Trampa del Eje: ¡Una Diferencia Pequeña que Parece Enorme! ✂️

Dos cooperativas produjeron **520** y **550 toneladas** (datos **inventados**): una diferencia de **solo el 6 %**. En un gráfico de barras, la **altura** de cada barra debe ser **proporcional** al dato. Pero si el eje vertical **no empieza en cero** (por ejemplo, empieza en 500), la barra solo muestra **lo que pasa de 500**: ¡20 contra 50! Calculamos **cuántas veces más alta parece** la barra de la cooperativa Y frente a la X con cada eje.

````python
x_ton, y_ton = 520, 550                                                                # toneladas INVENTADAS

def cuantas_veces_mas_alta(v0, v1, base_del_eje):
    """Altura VISIBLE de la barra v1 dividida entre la de v0, cuando el eje empieza en `base_del_eje`."""
    return (v1 - base_del_eje) / (v0 - base_del_eje)

parece_truncado = cuantas_veces_mas_alta(x_ton, y_ton, 500)
parece_cero = cuantas_veces_mas_alta(x_ton, y_ton, 0)
print(f"Con el eje desde 500, la barra de Y parece {parece_truncado:.2f} veces la de X | con el eje desde 0, {parece_cero:.2f} veces (la verdad: {y_ton / x_ton:.2f})")
assert parece_truncado == 2.5 and round(parece_cero, 2) == round(y_ton / x_ton, 2)

fig, ejes = plt.subplots(1, 2, figsize=(10, 3.8))
for ax, base, titulo in [(ejes[0], 500, "Eje desde 500: ¡Y parece 2.5 veces X!"), (ejes[1], 0, "Eje desde 0: la diferencia real")]:
    ax.bar(["Cooperativa X", "Cooperativa Y"], [x_ton, y_ton], color=["#2563eb", "#dc2626"]); ax.set_ylim(base, 570); ax.set_ylabel("toneladas"); ax.set_title(titulo, fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````
