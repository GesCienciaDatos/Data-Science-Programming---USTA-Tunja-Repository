## 1. Dos Fincas con el Mismo Promedio 🐄

Anotamos los **litros por vaca** durante **10 días** en dos fincas (datos **inventados**). Calculamos tres números: el **promedio**, el **rango** (la distancia entre el mejor y el peor día) y la **desviación típica**, que resume **qué tan lejos suelen quedar los días del promedio** (cuanto más grande, más «desordenados» están los datos).

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Litros por vaca durante 10 dias (INVENTADOS): las dos fincas promedian exactamente 20
finca_a = np.array([20, 21, 19, 20, 21, 19, 20, 20, 21, 19], dtype=float)
finca_b = np.array([10, 30, 5, 35, 20, 25, 15, 30, 12, 18], dtype=float)


def resumen(x):
    return {"promedio": x.mean(), "rango (mejor - peor)": x.max() - x.min(), "desviación típica": x.std()}


tabla = pd.DataFrame({"Finca A": resumen(finca_a), "Finca B": resumen(finca_b)})
print(tabla.round(2).to_string())
assert finca_a.mean() == finca_b.mean() == 20
assert finca_b.std() > 10 * finca_a.std()                                              # misma media, pero la B varía MUCHISIMO mas

fig, ax = plt.subplots(figsize=(9, 3.2))
ax.scatter(finca_a, np.full(10, 1), s=90, color="#2563eb", alpha=0.6, label="Finca A"); ax.scatter(finca_b, np.full(10, 0), s=90, color="#dc2626", alpha=0.6, label="Finca B")
ax.axvline(20, color="black", ls="--", lw=1.5); ax.text(20.3, 1.45, "promedio = 20", fontsize=9)
ax.set_yticks([0, 1]); ax.set_yticklabels(["Finca B", "Finca A"]); ax.set_ylim(-0.5, 1.8); ax.set_xlabel("litros por vaca"); ax.set_title("Mismo promedio, historias muy distintas", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** las dos fincas promedian **20 litros**, pero la **Finca A** tiene un rango de **2 litros** y una desviación típica de **0.77**, mientras que la **Finca B** tiene un rango de **30 litros** y una desviación de **9.3** (¡**12 veces más**!). Si fueras a firmar un contrato de **entregar al menos 15 litros por vaca cada día**, la Finca A casi siempre cumple y la B **falla muchos días** aunque «en promedio» sea igual. Siempre que veas un promedio, pregunta: **¿y qué tanto varían los datos?**

## 2. El Dato Mal Digitado 🖊️

En la **Finca A**, el cuarto día alguien digita **90** en lugar de **19** (¡un error de digitación!). Comparamos qué le pasa al **promedio**, al **rango**, a la **desviación típica** y a una medida más «tranquila»: el **rango del medio**, que es la distancia entre el puesto 25 y el puesto 75 de la fila ordenada (es decir, **qué tan ancho es el 50 % central**). También miramos la **mediana**.

````python
rango_medio = lambda x: np.percentile(x, 75) - np.percentile(x, 25)

finca_a_error = finca_a.copy()
finca_a_error[3] = 90                                                                  # el cuarto dia: 90 en vez de 19 (error de digitacion)

filas = {}
for nombre, x in [("Finca A (correcta)", finca_a), ("Finca A (con el error)", finca_a_error)]:
    filas[nombre] = {"promedio": x.mean(), "mediana": np.median(x), "rango": x.max() - x.min(), "desviación típica": x.std(), "rango del medio": rango_medio(x)}
tabla = pd.DataFrame(filas)
print(tabla.round(2).to_string())

assert finca_a_error.std() > 20 * finca_a.std()                                         # la desviacion tipica se dispara
assert np.median(finca_a_error) == np.median(finca_a) == 20 and rango_medio(finca_a_error) < 2 * rango_medio(finca_a)   # mediana y rango del medio casi ni se enteran
````

---
**Lección:** un **solo** dato equivocado hizo que el promedio pasara de 20 a **27**, el rango de 2 a **71** y la desviación típica de 0.77 a **21.0** (¡27 veces mayor!). En cambio, la **mediana** siguió en **20** y el **rango del medio** apenas pasó de 1.5 a 1.75. Moraleja: las medidas que usan **todos** los datos (promedio, rango, desviación) son **muy sensibles** a los errores; las que miran **el centro de la fila** (mediana, rango del medio) **resisten**. Y antes de borrar un dato raro, **investiga si es un error o un caso real**.

## 3. ¿Cuando una Sube, la Otra También? La Correlación 🌧️

Medimos **12 parcelas** de papa (datos **inventados**): los milímetros de **lluvia** y las toneladas por hectárea que se cosecharon. La **correlación** es un número entre **−1 y 1** que dice qué tan bien se alinean dos cosas en una **línea recta**:

- cerca de **1**: cuando una sube, la otra **también sube** (juntas, casi en línea);
- cerca de **0**: **no hay línea** (una no ayuda a adivinar la otra);
- cerca de **−1**: cuando una sube, la otra **baja**.

Además, probamos otra pareja: las **ventas de paraguas** y de **chocolate caliente** a lo largo de 12 meses.

````python
# 12 parcelas (INVENTADAS): lluvia (mm) y rendimiento (toneladas por hectarea)
lluvia = np.array([60, 75, 80, 90, 95, 100, 110, 115, 120, 130, 140, 150], dtype=float)
rendimiento = np.array([15, 17, 16, 19, 20, 21, 22, 21, 24, 25, 24, 27], dtype=float)
r_papa = np.corrcoef(lluvia, rendimiento)[0, 1]
print(f"Correlación entre lluvia y rendimiento: {r_papa:.2f}")
assert r_papa > 0.95

# 12 meses (INVENTADOS): ventas de paraguas y de chocolate caliente; ambas dependen de cuanto llueve
lluvia_mes = np.array([30, 45, 60, 90, 120, 140, 150, 130, 100, 70, 50, 35], dtype=float)
paraguas = np.array([105, 142, 165, 200, 294, 327, 332, 309, 245, 184, 140, 118], dtype=float)
chocolate = np.array([66, 96, 114, 172, 210, 236, 246, 222, 180, 132, 121, 95], dtype=float)
r_pc = np.corrcoef(paraguas, chocolate)[0, 1]
print(f"Correlación entre ventas de paraguas y de chocolate: {r_pc:.2f}")
assert r_pc > 0.95

fig, ejes = plt.subplots(1, 2, figsize=(11, 3.8))
ejes[0].scatter(lluvia, rendimiento, s=70, color="#2563eb"); ejes[0].set_xlabel("lluvia (mm)"); ejes[0].set_ylabel("toneladas por hectárea"); ejes[0].set_title(f"Lluvia y cosecha: correlación {r_papa:.2f}", fontweight="bold", fontsize=10)
ejes[1].scatter(paraguas, chocolate, s=70, color="#dc2626"); ejes[1].set_xlabel("ventas de paraguas"); ejes[1].set_ylabel("ventas de chocolate caliente"); ejes[1].set_title(f"Paraguas y chocolate: correlación {r_pc:.2f}", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## Resumen en una frase 🎯

> **El promedio no basta: hay que saber cuánto varían los datos; un dato mal digitado dispara las medidas que usan todos los datos; y la correlación dice qué tan bien se alinean dos cosas, pero no por qué.**

### Lo que aprendiste hoy:
- ✅ **Mismo promedio, distinta variación:** las fincas A y B promedian 20 litros, pero la B varía **12 veces más** (desviación 9.3 frente a 0.77).
- ✅ **Un error de digitación** (90 en vez de 19) llevó la desviación de 0.77 a 21.0, pero la **mediana** siguió en 20 y el **rango del medio** casi no cambió.
- ✅ **Correlación** (de −1 a 1): lluvia y cosecha, 0.97; **pero** paraguas y chocolate (0.98) suben juntos sin causarse: la causa es la lluvia.
- ✅ **Una sola parcela rara** bajó la correlación de 0.97 a 0.14: **mira siempre el gráfico**.
