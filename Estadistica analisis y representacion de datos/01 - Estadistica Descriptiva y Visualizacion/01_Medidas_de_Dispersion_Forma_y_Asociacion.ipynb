## 1. Dispersión: Cuánto se Alejan los Datos de su Centro 📏

Una medida central sin una de dispersión es **media historia**. Para una muestra $x_1,\dots,x_n$:

| Medida | Definición | Unidad | Punto de ruptura | Observaciones |
|---|---|---|---|---|
| **Rango** | $R=x_{(n)}-x_{(1)}$ | la de $x$ | $0$ | Depende de **dos** datos y **crece con $n$** |
| **RIC / IQR** | $Q_3-Q_1$ | la de $x$ | $25\,\%$ | La amplitud del **50 % central**; robusta |
| **Varianza** | $s^2=\dfrac{1}{n-1}\displaystyle\sum_{i=1}^{n}(x_i-\bar{x})^2$ | **cuadrada** | $0$ | La más usada en teoría (aditiva); muy sensible a atípicos |
| **Desviación estándar** | $s=\sqrt{s^2}$ | la de $x$ | $0$ | «Distancia típica» a la media |
| **MAD** | $\text{mediana}\big(\lvert x_i-\tilde{x}\rvert\big)$, con $\tilde{x}$ la mediana | la de $x$ | $50\,\%$ | Multiplicada por $1.4826=1/\Phi^{-1}(0.75)$ **estima $\sigma$ si los datos son normales** |
| **CV** | $\mathrm{CV}=s/\lvert\bar{x}\rvert$ | **sin unidad** | $0$ | Solo en **escala de razón** y con media $\ne 0$: compara variabilidad entre magnitudes distintas |

**¿Por qué $n-1$?** Las desviaciones $x_i-\bar{x}$ suman cero, así que solo $n-1$ son «libres» (grados de libertad); además $\bar{x}$ está **más cerca de los datos** que $\mu$, por lo que el divisor $n$ **subestima** $\sigma^2$: $\mathbb{E}\big[\tfrac1n\sum(x_i-\bar{x})^2\big]=\tfrac{n-1}{n}\sigma^2$. Con $n-1$ la **varianza** queda insesgada, pero **$s$ sigue siendo sesgada hacia abajo** (desigualdad de Jensen). **Cuidado con el software:** `np.var` y `np.std` usan `ddof=0` por defecto; `pandas` usa `ddof=1`.

**Cuánto de los datos cae cerca de la media.** La **desigualdad de Chebyshev** garantiza, para **cualquier** distribución, que al menos $1-1/k^2$ de los datos está a menos de $k$ desviaciones de la media (≥ 75 % a $k=2$, ≥ 88.9 % a $k=3$). Solo para datos **normales** vale la **regla empírica** 68.27 – 95.45 – 99.73 %.

La primera celda configura el entorno, genera dos zonas con **la misma media y distinta variabilidad** (más los ingresos asimétricos del cuaderno anterior) e implementa las medidas **desde cero**.

````python
# ============================================================
# Configuracion inicial, datos SINTETICOS y medidas de dispersion DESDE CERO
# ============================================================
# !pip install -q numpy pandas scipy matplotlib scikit-learn

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import stats

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")

# --- datos FICTICIOS: dos zonas con la MISMA media de rendimiento (t/ha) y distinta variabilidad; ingresos asimetricos
rng = np.random.default_rng(SEED)
zona_A = np.round(rng.normal(22.0, 2.0, size=150), 1)
zona_B = np.round(rng.normal(22.0, 5.0, size=150), 1)
ingresos = np.round(rng.lognormal(mean=np.log(1.8), sigma=0.55, size=400), 2)                  # millones de pesos al mes


def rango(x):
    return float(np.max(x) - np.min(x))


def cuantil(x, p):
    """Cuantil muestral tipo 7 (interpolacion lineal), el mismo de NumPy y pandas."""
    x = np.sort(np.asarray(x, dtype=float))
    h = (x.size - 1) * p
    j = int(np.floor(h))
    return float(x[j] + (h - j) * (x[min(j + 1, x.size - 1)] - x[j]))


def iqr(x):
    return cuantil(x, 0.75) - cuantil(x, 0.25)


def varianza(x, ddof=1):
    x = np.asarray(x, dtype=float)
    return float(((x - x.sum() / x.size) ** 2).sum() / (x.size - ddof))


def desv(x, ddof=1):
    return varianza(x, ddof) ** 0.5


def mad(x, escala=1.0):
    """Mediana de las desviaciones absolutas respecto a la mediana, por `escala` (1.4826 la hace consistente con sigma en datos normales)."""
    x = np.asarray(x, dtype=float)
    return float(np.median(np.abs(x - np.median(x))) * escala)


ESCALA_NORMAL = 1 / stats.norm.ppf(0.75)                                                        # = 1.4826...
cv = lambda x: desv(x) / abs(np.mean(x))                                                        # coeficiente de variacion (solo escala de razon, media != 0)

# --- contraste con NumPy / SciPy / pandas
for x in (zona_A, zona_B, ingresos):
    assert np.isclose(varianza(x), np.var(x, ddof=1)) and np.isclose(desv(x), np.std(x, ddof=1)) and np.isclose(varianza(x, 0), np.var(x))
    assert np.isclose(iqr(x), stats.iqr(x)) and np.isclose(rango(x), np.ptp(x))
    assert np.isclose(mad(x, ESCALA_NORMAL), stats.median_abs_deviation(x, scale="normal"))
assert not np.isclose(np.std(zona_A), pd.Series(zona_A).std())                                  # OJO: NumPy usa ddof=0 y pandas ddof=1 por defecto
assert np.isclose(np.std(zona_A) * np.sqrt(150 / 149), pd.Series(zona_A).std())

tabla = pd.DataFrame({"zona A": [np.mean(zona_A), rango(zona_A), desv(zona_A), iqr(zona_A), mad(zona_A, ESCALA_NORMAL), cv(zona_A)],
                      "zona B": [np.mean(zona_B), rango(zona_B), desv(zona_B), iqr(zona_B), mad(zona_B, ESCALA_NORMAL), cv(zona_B)]},
                     index=["media", "rango", "desviación estándar", "RIC (IQR)", "MAD escalada", "CV"])
print(tabla.round(3).to_string())
assert abs(np.mean(zona_A) - np.mean(zona_B)) < 1.0 and desv(zona_B) > 2 * desv(zona_A)         # misma media, mas del doble de dispersion

# --- el CV exige escala de razon: cambiar el cero (Celsius -> Kelvin) cambia el CV, cambiar la unidad (m -> cm) no
temp_c = np.array([8.0, 10.0, 12.0, 14.0, 16.0])
assert not np.isclose(cv(temp_c), cv(temp_c + 273.15)) and np.isclose(cv(ingresos), cv(ingresos * 100))
print(f"\nCV de temperaturas en °C: {cv(temp_c):.3f} frente a Kelvin: {cv(temp_c + 273.15):.4f} (cero arbitrario); el CV de ingresos no cambia al pasar de millones a decenas de miles.")
````

````python
# ============================================================
# Tres experimentos: (a) por que n-1, (b) calculo numerico de la varianza, (c) Chebyshev y regla empirica
# ============================================================
# (a) Sesgo de la varianza con divisor n frente a n-1 (muestras de tamano 5 de una N(0, sigma^2 = 4))
rng_a = np.random.default_rng(SEED + 1)
muestras = rng_a.normal(0.0, 2.0, size=(20000, 5))
v_n, v_n1 = muestras.var(axis=1, ddof=0), muestras.var(axis=1, ddof=1)
print(f"(a) sigma^2 = 4.00 | E[s^2 con n] ≈ {v_n.mean():.3f} (teoría 3.2) | E[s^2 con n-1] ≈ {v_n1.mean():.3f} | E[s] ≈ {np.sqrt(v_n1).mean():.3f} (< sigma = 2)")
assert abs(v_n.mean() - 3.2) < 0.1 and abs(v_n1.mean() - 4.0) < 0.1 and np.sqrt(v_n1).mean() < 2.0   # n-1 corrige la VARIANZA pero s sigue sesgada hacia abajo


# (b) Formula ingenua E[x^2] - E[x]^2 frente a Welford (una pasada, estable)
def var_ingenua(x):
    return float((x ** 2).mean() - x.mean() ** 2)


def var_welford(x):
    n, m, M2 = 0, 0.0, 0.0
    for v in x:
        n += 1
        d = v - m
        m += d / n
        M2 += d * (v - m)
    return M2 / (n - 1)


datos_b = 1e9 + np.random.default_rng(SEED + 2).normal(0, 1, size=10000)                         # varianza verdadera ≈ 1, pero con un desplazamiento enorme
ref = np.var(datos_b, ddof=1)
print(f"(b) np.var = {ref:.6f} | Welford = {var_welford(datos_b):.6f} | fórmula ingenua (ddof=0) = {var_ingenua(datos_b):.1f}")
assert abs(var_welford(datos_b) - ref) / ref < 1e-6 and abs(var_ingenua(datos_b) - 1.0) > 1.0      # la ingenua pierde TODAS las cifras

# (c) Chebyshev (cualquier distribucion) frente a la regla empirica (solo normales)
normal_grande = np.random.default_rng(SEED + 3).normal(size=200_000)
filas = []
for k in (1.5, 2, 3):
    fila = {"k": k, "cota de Chebyshev ≥": 1 - 1 / k ** 2}
    for nombre, x in {"normal": normal_grande, "ingresos (lognormal)": ingresos}.items():
        fila[nombre] = np.mean(np.abs(x - x.mean()) <= k * np.std(x))
        assert fila[nombre] >= fila["cota de Chebyshev ≥"]                                          # la cota se cumple SIEMPRE (con sigma poblacional ddof=0)
    filas.append(fila)
cheb = pd.DataFrame(filas).set_index("k")
print("\n(c) fracción dentro de k desviaciones de la media:\n" + cheb.round(4).to_string())
assert abs(cheb.loc[2, "normal"] - 0.9545) < 0.005 and abs(cheb.loc[3, "normal"] - 0.9973) < 0.002
````

## 2. Forma: Asimetría y Curtosis 🔔

Con los **momentos centrales** $m_r=\dfrac1n\sum_i (x_i-\bar{x})^r$, la **asimetría** y la **curtosis en exceso** (respecto a la normal) son

$$g_1=\frac{m_3}{m_2^{3/2}},\qquad g_2=\frac{m_4}{m_2^{2}}-3 .$$

Las versiones con **corrección de sesgo** para muestras pequeñas son $G_1=g_1\dfrac{\sqrt{n(n-1)}}{n-2}$ y $G_2=\dfrac{n-1}{(n-2)(n-3)}\big[(n+1)g_2+6\big]$ (las de `bias=False` en SciPy; las de Excel). Cómo leerlas:

- $g_1>0$: **cola larga a la derecha**; $g_1<0$: a la izquierda. $g_1=0$ **no garantiza simetría** (basta que el tercer momento se anule). Alternativas basadas en cuantiles: **Bowley** $\dfrac{Q_3+Q_1-2Q_2}{Q_3-Q_1}$ y el **segundo coeficiente de Pearson** $3(\bar{x}-\tilde{x})/s$.
- $g_2$ mide el **peso de las colas** (valores extremos), **no** lo «picudo» del centro (Westfall, 2014): Normal $0$, Uniforme $-1.2$ (sin colas), Laplace $3$ (colas pesadas). Al depender de la **cuarta potencia**, **unos pocos extremos dominan** y la curtosis muestral es **muy inestable** con colas pesadas.
- **Transformar para simetrizar:** el logaritmo convierte una lognormal en normal; la familia de **Box-Cox** $x^{(\lambda)}=\dfrac{x^{\lambda}-1}{\lambda}$ (con $x^{(0)}=\ln x$) estima $\lambda$ por máxima verosimilitud.

La celda implementa los momentos, contrasta con SciPy (con y sin corrección), compara cinco distribuciones con su valor teórico, muestra la **inestabilidad** de la curtosis y aplica transformaciones a los ingresos.

````python
# ============================================================
# 2. Forma: momentos, asimetria y curtosis desde cero; transformaciones
# ============================================================
def momento_central(x, r):
    x = np.asarray(x, dtype=float)
    return float(((x - x.mean()) ** r).mean())


def asimetria(x, corregida=False):
    n = len(x)
    g1 = momento_central(x, 3) / momento_central(x, 2) ** 1.5
    return g1 * np.sqrt(n * (n - 1)) / (n - 2) if corregida else g1


def curtosis_exceso(x, corregida=False):
    n = len(x)
    g2 = momento_central(x, 4) / momento_central(x, 2) ** 2 - 3
    return (n - 1) / ((n - 2) * (n - 3)) * ((n + 1) * g2 + 6) if corregida else g2


for x in (zona_A, ingresos):
    assert np.isclose(asimetria(x), stats.skew(x)) and np.isclose(asimetria(x, True), stats.skew(x, bias=False))
    assert np.isclose(curtosis_exceso(x), stats.kurtosis(x)) and np.isclose(curtosis_exceso(x, True), stats.kurtosis(x, bias=False))

# --- cinco distribuciones con n grande: la asimetria y la curtosis muestrales frente a sus valores teoricos
rng_f = np.random.default_rng(SEED + 4)
M = 200_000
dist = {"Normal": (rng_f.standard_normal(M), 0.0, 0.0), "Uniforme": (rng_f.uniform(-1, 1, M), 0.0, -1.2), "Laplace": (rng_f.laplace(size=M), 0.0, 3.0),
        "Exponencial": (rng_f.exponential(size=M), 2.0, 6.0), "Lognormal (σ=0.55)": (rng_f.lognormal(0, 0.55, M), 1.99, 7.80)}
filas = [{"distribución": k, "asimetría muestral": asimetria(v[0]), "asimetría teórica": v[1], "curtosis exc. muestral": curtosis_exceso(v[0]), "curtosis exc. teórica": v[2]} for k, v in dist.items()]
tab_forma = pd.DataFrame(filas).set_index("distribución")
print(tab_forma.round(2).to_string())
for k in ("Normal", "Uniforme", "Laplace"):
    assert abs(asimetria(dist[k][0]) - dist[k][1]) < 0.05 and abs(curtosis_exceso(dist[k][0]) - dist[k][2]) < 0.15
assert abs(asimetria(dist["Exponencial"][0]) - 2.0) < 0.15

# --- la curtosis muestral de una cola pesada es MUY inestable (con n = 1000)
curt = np.array([curtosis_exceso(np.random.default_rng(s).lognormal(0, 0.55, 1000)) for s in range(300)])
print(f"\nCurtosis en exceso de la lognormal con n = 1000 (300 repeticiones): mediana {np.median(curt):.2f}, RIC {iqr(curt):.2f}, rango [{curt.min():.1f}, {curt.max():.1f}]; teórica 7.80")
assert iqr(curt) > 1.0 and curt.min() < 7.80 < curt.max()

# --- coeficientes alternativos y transformaciones sobre los ingresos
q1, q2, q3 = cuantil(ingresos, 0.25), cuantil(ingresos, 0.5), cuantil(ingresos, 0.75)
pearson2 = 3 * (ingresos.mean() - q2) / desv(ingresos)
bowley = (q3 + q1 - 2 * q2) / (q3 - q1)
print(f"\ningresos: g1 = {asimetria(ingresos):.3f} | Pearson-2 = {pearson2:.3f} | Bowley = {bowley:.3f}")
lam = stats.boxcox(ingresos)[1]
trans = {"original": ingresos, "raíz cuadrada": np.sqrt(ingresos), "logaritmo": np.log(ingresos), "Box-Cox (λ óptimo)": stats.boxcox(ingresos)[0]}
print({k: round(float(asimetria(v)), 3) for k, v in trans.items()}, f"| λ de Box-Cox = {lam:.3f}")
assert asimetria(ingresos) > 0.8 > abs(asimetria(np.log(ingresos))) + 0.5 and abs(lam) < 0.45   # el log casi simetriza una lognormal (lambda ≈ 0)

fig, ejes = plt.subplots(2, 3, figsize=(13, 6.4))
for ax, (k, v), col in zip(ejes.ravel(), list(dist.items()) + [("log(Lognormal)", (np.log(dist["Lognormal (σ=0.55)"][0]), 0, 0))], PALETA):
    sub = v[0][:20000]
    ax.hist(sub, bins=60, density=True, color=col, alpha=0.8, edgecolor="white")
    ax.set_title(f"{k}\nasimetría {asimetria(v[0]):.2f} | curtosis exc. {curtosis_exceso(v[0]):.2f}", fontsize=9.5, fontweight="bold"); ax.set_yticks([])
plt.tight_layout(); plt.show()
````

---
**Lectura.** Con $n=200\,000$ la asimetría y la curtosis muestrales reproducen las teóricas (Normal 0 y 0; Uniforme 0 y $-1.20$; Laplace 0 y 2.89 frente a 3; Exponencial 1.99 y 5.92 frente a 2 y 6). Pero con colas pesadas y $n=1000$ la **curtosis de una lognormal** (teórica 7.80) tuvo **mediana 5.68, RIC 3.76 y rango de 2.0 a 34.0** en 300 muestras: **un solo número de curtosis de una muestra no es fiable**; conviene acompañarlo de gráficos o de intervalos (módulo 03, *bootstrap*). Los ingresos del cuaderno anterior tienen $g_1=1.49$ (Pearson-2: 0.66; Bowley: 0.13, menor porque los cuartiles ignoran la cola); la **raíz cuadrada** lo reduce a 0.67, el **logaritmo** a $-0.09$ y **Box-Cox** halla $\hat{\lambda}=0.06$, es decir, casi exactamente el logaritmo, como corresponde a una lognormal.

## 3. Valores Atípicos: Detectar sin Eliminar a Ciegas 🚨

Un **atípico** es una observación **alejada del patrón del resto**; puede ser un **error** (sensor dañado), una **población distinta** (otro tipo de hogar) o simplemente **cola legítima** de una distribución asimétrica. Tres reglas clásicas:

| Regla | Marca $x_i$ si... | Resistente a la contaminación |
|---|---|---|
| **$z$ clásico** | $\lvert x_i-\bar{x}\rvert / s > 3$ | **No**: $\bar{x}$ y $s$ **los infla el propio atípico** |
| **Tukey** (1977) | $x_i<Q_1-1.5\,\text{RIC}$ o $x_i>Q_3+1.5\,\text{RIC}$ | Sí, hasta ≈ 25 % |
| **$z$ robusto (modificado)** | $\lvert x_i-\tilde{x}\rvert / (1.4826\cdot\text{MAD}) > 3.5$ | Sí, hasta ≈ 50 % |

Si hay **varios** atípicos, el $z$ clásico sufre **enmascaramiento** (*masking*): inflan la desviación y ninguno parece extremo. Y la regla de Tukey asume una forma aproximadamente **simétrica**: sobre datos asimétricos marca como «atípicos» valores perfectamente legítimos. La decisión (corregir, eliminar, conservar, modelar aparte) **no es estadística sino de dominio**: una marca es **una pregunta, no una orden de borrado**. Simulamos 200 lecturas de un sensor de temperatura con **10 % defectuosas** (verdad conocida) para medir *recall* (¿cuántas defectuosas se atrapan?) y precisión (¿cuántas marcas son defectuosas?).

````python
# ============================================================
# 3. Valores atipicos: Tukey, z clasico y z robusto (enmascaramiento)
# ============================================================
rng_o = np.random.default_rng(SEED + 5)
lecturas = np.concatenate([rng_o.normal(20, 2, 180), rng_o.normal(34, 1.5, 20)])                # 180 lecturas buenas + 20 lecturas falladas (10 %)
etiqueta = np.concatenate([np.zeros(180, bool), np.ones(20, bool)])
perm = rng_o.permutation(lecturas.size)
lecturas, etiqueta = lecturas[perm], etiqueta[perm]                                             # etiqueta = True si la lectura es realmente defectuosa


def marca_tukey(x, k=1.5):
    q1, q3 = cuantil(x, 0.25), cuantil(x, 0.75)
    return (x < q1 - k * (q3 - q1)) | (x > q3 + k * (q3 - q1))


def marca_z(x, umbral=3.0):
    return np.abs(x - x.mean()) / desv(x) > umbral


def marca_z_robusto(x, umbral=3.5):
    """z modificado: usa la mediana y la MAD escalada en vez de la media y la desviacion (Iglewicz y Hoaglin)."""
    return np.abs(x - np.median(x)) / mad(x, ESCALA_NORMAL) > umbral


def evaluar(marcas, verdad):
    tp = int((marcas & verdad).sum()); fp = int((marcas & ~verdad).sum())
    return {"marcadas": int(marcas.sum()), "aciertos": tp, "falsas alarmas": fp, "recall": float(tp / verdad.sum()), "precisión": float(tp / max(1, marcas.sum()))}


reglas = {"z clásico (|z|>3)": marca_z(lecturas), "Tukey (1.5·RIC)": marca_tukey(lecturas), "z robusto (>3.5)": marca_z_robusto(lecturas)}
res = pd.DataFrame({k: evaluar(m, etiqueta) for k, m in {**reglas, "z robusto (>2.5)": marca_z_robusto(lecturas, 2.5)}.items()}).T
res[["marcadas", "aciertos", "falsas alarmas"]] = res[["marcadas", "aciertos", "falsas alarmas"]].astype(int)
print(res.round(3).to_string())
print(f"\nmedia = {lecturas.mean():.2f}, desviación = {desv(lecturas):.2f} (inflada por los defectuosos) | mediana = {np.median(lecturas):.2f}, MAD escalada = {mad(lecturas, ESCALA_NORMAL):.2f}")
assert res.loc["z robusto (>2.5)", "recall"] >= 0.95 and res.loc["z robusto (>2.5)", "falsas alarmas"] >= res.loc["z robusto (>3.5)", "falsas alarmas"]   # umbral bajo: mas falsas alarmas
assert res.loc["z clásico (|z|>3)", "recall"] < 0.5 and res.loc["z robusto (>3.5)", "recall"] >= 0.95 and res.loc["Tukey (1.5·RIC)", "recall"] >= 0.95   # enmascaramiento

# --- "atipico" NO es "error": la regla de Tukey sobre datos normales y sobre datos asimetricos
normales = np.random.default_rng(SEED + 6).normal(size=100_000)
lognorm = np.random.default_rng(SEED + 6).lognormal(0, 0.55, size=100_000)
f_norm, f_logn = marca_tukey(normales).mean(), marca_tukey(lognorm).mean()
print(f"Fracción marcada por Tukey: normal {f_norm:.2%} | lognormal {f_logn:.2%} | log(lognormal) {marca_tukey(np.log(lognorm)).mean():.2%}")
assert f_logn > 5 * f_norm and abs(marca_tukey(np.log(lognorm)).mean() - f_norm) < 0.003

fig, ejes = plt.subplots(1, 3, figsize=(13, 4.2), sharey=True)
for ax, (nombre, m) in zip(ejes, reglas.items()):
    ax.scatter(np.where(~m)[0], lecturas[~m], s=14, color="#94a3b8", label="no marcadas")
    ax.scatter(np.where(m)[0], lecturas[m], s=26, color=PALETA[1], label="marcadas")
    ax.scatter(np.where(etiqueta & ~m)[0], lecturas[etiqueta & ~m], s=60, facecolors="none", edgecolors=PALETA[3], lw=1.6, label="defectuosa NO detectada")
    ax.set_title(nombre, fontweight="bold", fontsize=10); ax.set_xlabel("índice de la lectura")
ejes[0].set_ylabel("lectura (°C)"); ejes[0].legend(fontsize=7.5, loc="center left", bbox_to_anchor=(0.0, 0.56))
plt.tight_layout(); plt.show()
````

---
**Lectura.** El $z$ clásico detecta solo **4 de 20** defectuosas (recall 0.20): los 20 valores defectuosos inflan la desviación a 4.71 (la de las lecturas buenas es ≈ 2) y el umbral $\bar{x}+3s$ queda en ≈ 35.6, **por encima de casi todos los defectuosos** (enmascaramiento). Tukey y el $z$ robusto, basados en la mediana (20.39) y la MAD escalada (2.19), detectan las 20 **sin falsas alarmas** en este experimento. Pero **atípico no es error**: sobre datos **normales** Tukey marca 0.69 % de los valores, sobre una **lognormal** 4.17 % (6 veces más) sin que haya un solo error; al aplicar el logaritmo vuelve a 0.69 %. Con asimetría se transforma **antes** de aplicar la regla, o se usa una regla ajustada a la asimetría. Bajar el umbral del $z$ robusto de 3.5 a 2.5 sigue detectando las 20 defectuosas pero marca **una lectura buena de más** (precisión 0.952): es el compromiso habitual, un umbral bajo favorece el *recall* y uno alto la precisión, y se elige según el **costo** de cada error.

## 4. Asociación entre Dos Variables Numéricas 🔗

Para pares $(x_i,y_i)$, la **covarianza** muestral es $s_{xy}=\dfrac{1}{n-1}\sum_i (x_i-\bar{x})(y_i-\bar{y})$ (su signo indica el sentido, pero **depende de las unidades**). Estandarizándola se obtiene la **correlación de Pearson**

$$r=\frac{s_{xy}}{s_x\,s_y}\in[-1,1],$$

que mide la asociación **lineal**: $r^2$ es la fracción de la varianza de $y$ explicada por la recta de mínimos cuadrados. Es **invariante a cambios lineales** $ax+b$ (cambia de signo si $ac<0$) y muy **sensible a atípicos**. Dos alternativas **basadas en rangos**:

- **Spearman** $\rho_s$: la correlación de Pearson aplicada a los **rangos** (con rango promedio en empates). Capta cualquier relación **monótona** y resiste atípicos.
- **Kendall** $\tau_b=\dfrac{n_c-n_d}{\sqrt{(n_0-n_1)(n_0-n_2)}}$, con $n_c$ y $n_d$ los pares **concordantes** y **discordantes**, $n_0=\binom{n}{2}$ y $n_1,n_2$ los pares empatados en $x$ y en $y$. Se interpreta como diferencia de probabilidades de concordancia y suele ser menor en magnitud que $\rho_s$.

**Un número no sustituye al gráfico:** el **cuarteto de Anscombe** (1973) son cuatro conjuntos con **las mismas** medias, varianzas, correlación y recta de regresión, y formas totalmente distintas. Y recuerda: **correlación no implica causalidad** (confusores, causalidad inversa, azar) —tema del Módulo 00. La celda implementa covarianza, Pearson, rangos con empates, Spearman y Kendall **desde cero**, las contrasta con NumPy y SciPy y reproduce Anscombe.

````python
# ============================================================
# 4. Asociacion: covarianza, Pearson, Spearman y Kendall desde cero; el cuarteto de Anscombe
# ============================================================
def covarianza(x, y, ddof=1):
    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    return float(((x - x.mean()) * (y - y.mean())).sum() / (x.size - ddof))


def pearson(x, y):
    return covarianza(x, y) / (desv(x) * desv(y))


def rangos(x):
    """Rangos 1..n; los empates reciben el rango PROMEDIO."""
    x = np.asarray(x)
    r = np.empty(x.size)
    r[np.argsort(x, kind="mergesort")] = np.arange(1, x.size + 1)
    _, inv, cuentas = np.unique(x, return_inverse=True, return_counts=True)
    return (np.bincount(inv, weights=r) / cuentas)[inv]


def spearman(x, y):
    return pearson(rangos(x), rangos(y))                                                       # Spearman = Pearson sobre los rangos


def kendall(x, y):
    """tau-b: (concordantes - discordantes) / sqrt((n0-n1)(n0-n2)); O(n^2) para ser transparente."""
    sx, sy = np.sign(np.subtract.outer(x, x)), np.sign(np.subtract.outer(y, y))
    return float((sx * sy).sum() / np.sqrt((sx ** 2).sum() * (sy ** 2).sum()))


rng_r = np.random.default_rng(SEED + 8)
xr = np.round(rng_r.normal(size=120), 1)                                                         # redondeados a proposito: hay empates
yr = np.round(0.7 * xr + rng_r.normal(0, 0.7, size=120), 1)
assert np.isclose(covarianza(xr, yr), np.cov(xr, yr)[0, 1]) and np.isclose(pearson(xr, yr), stats.pearsonr(xr, yr)[0])
assert np.allclose(rangos(xr), stats.rankdata(xr)) and np.isclose(spearman(xr, yr), stats.spearmanr(xr, yr)[0])
assert np.isclose(kendall(xr, yr), stats.kendalltau(xr, yr)[0])
print(f"Con empates: Pearson {pearson(xr, yr):.4f} | Spearman {spearman(xr, yr):.4f} | Kendall {kendall(xr, yr):.4f} (las tres coinciden con SciPy)")

# --- el cuarteto de Anscombe (1973): cuatro conjuntos con los MISMOS resumenes y formas muy distintas
x123 = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5], dtype=float)
anscombe = {"I": (x123, np.array([8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68])),
            "II": (x123, np.array([9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74])),
            "III": (x123, np.array([7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73])),
            "IV": (np.array([8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8], dtype=float), np.array([6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]))}
filas = []
for nombre, (x, y) in anscombe.items():
    pend, orig = np.polyfit(x, y, 1)
    filas.append({"conjunto": nombre, "media x": x.mean(), "media y": y.mean(), "var x": varianza(x), "var y": varianza(y), "Pearson r": pearson(x, y), "pendiente": pend, "intercepto": orig, "Spearman": spearman(x, y)})
    assert abs(x.mean() - 9) < 1e-9 and abs(y.mean() - 7.50) < 0.01 and abs(pearson(x, y) - 0.816) < 0.001 and abs(pend - 0.5) < 0.01
print(pd.DataFrame(filas).set_index("conjunto").round(3).to_string())

fig, ejes = plt.subplots(2, 2, figsize=(9.5, 6.8), sharex=True, sharey=True)
for ax, (nombre, (x, y)), col in zip(ejes.ravel(), anscombe.items(), PALETA):
    pend, orig = np.polyfit(x, y, 1)
    ax.scatter(x, y, color=col, s=45, edgecolor="white", zorder=3); xs = np.array([3, 20]); ax.plot(xs, orig + pend * xs, color="black", lw=1.4)
    ax.set_title(f"Conjunto {nombre}: r = {pearson(x, y):.3f}", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** Con **empates** (datos redondeados) las tres funciones desde cero coinciden con SciPy: Pearson 0.716, Spearman 0.678 y Kendall 0.504 (Kendall suele ser el menor). En Anscombe los cuatro conjuntos tienen $\bar{x}=9$, $\bar{y}\approx 7.50$, $s_x^2=11$, $s_y^2\approx 4.12$, $r=0.816$ y la recta $y\approx 3+0.5x$, pero el **Spearman** ya los separa: 0.818, 0.691, 0.991 y 0.500. Mirando el gráfico: el I es una relación lineal con ruido; el II es una **curva**; el III es lineal con **un atípico**; el IV es un **único punto** que fabrica la recta (todos los demás tienen $x=8$). **Resumen sin gráfico = conclusión peligrosa.**

## 4b. Qué Capta Cada Coeficiente 🔍

Los coeficientes responden **preguntas distintas**: Pearson, «¿qué tan lineal?»; Spearman y Kendall, «¿qué tan monótona?»; y **ninguno** detecta una dependencia **no monótona** (una parábola). Para eso existen medidas de dependencia general como la **información mutua** $I(X;Y)\ge 0$, que vale 0 **si y solo si** hay independencia (aquí estimada con `mutual_info_regression`, en *nats*). Probamos cinco situaciones y dos **invariancias**: Pearson respeta cambios lineales, Spearman y Kendall **cualquier transformación monótona**.

````python
# ============================================================
# 4b. Que capta cada coeficiente: lineal, monotona no lineal, no monotona, un solo atipico; invariancias
# ============================================================
from sklearn.feature_selection import mutual_info_regression

rng_q = np.random.default_rng(SEED + 9)
n_q = 200
x_lin = rng_q.uniform(0, 5, n_q); y_lin = 2 * x_lin + rng_q.normal(0, 1, n_q)
x_exp = rng_q.uniform(0, 5, n_q); y_exp = np.exp(x_exp)                                         # monotona creciente, NO lineal
x_cua = rng_q.uniform(-3, 3, n_q); y_cua = x_cua ** 2 + rng_q.normal(0, 0.3, n_q)               # dependencia perfecta pero NO monotona
x_ind = rng_q.normal(size=30); y_ind = rng_q.normal(size=30)                                      # independientes...
x_out, y_out = np.append(x_ind, 10.0), np.append(y_ind, 10.0)                                     # ...mas UN atipico en (10, 10)
casos = {"lineal con ruido": (x_lin, y_lin), "exponencial (monótona)": (x_exp, y_exp), "parábola (no monótona)": (x_cua, y_cua),
         "independientes (n=30)": (x_ind, y_ind), "independientes + 1 atípico": (x_out, y_out)}
filas = []
for nombre, (x, y) in casos.items():
    mi = mutual_info_regression(np.asarray(x).reshape(-1, 1), y, random_state=0)[0]
    filas.append({"caso": nombre, "Pearson": pearson(x, y), "Spearman": spearman(x, y), "Kendall": kendall(x, y), "información mutua": mi})
tab_c = pd.DataFrame(filas).set_index("caso")
print(tab_c.round(3).to_string())
assert tab_c.loc["exponencial (monótona)", "Spearman"] > 0.999 and tab_c.loc["exponencial (monótona)", "Pearson"] < 0.9
assert max(abs(tab_c.loc["parábola (no monótona)", ["Pearson", "Spearman", "Kendall"]])) < 0.15 and tab_c.loc["parábola (no monótona)", "información mutua"] > 0.5
assert tab_c.loc["independientes + 1 atípico", "Pearson"] > 0.5 > abs(tab_c.loc["independientes (n=30)", "Pearson"]) and tab_c.loc["independientes + 1 atípico", "Spearman"] < 0.5   # 1 punto fabrica una correlacion

# --- invariancias
a_, b_, c_, d_ = 3.0, -5.0, -0.5, 100.0
assert np.isclose(pearson(a_ * x_lin + b_, c_ * y_lin + d_), -pearson(x_lin, y_lin))              # afin: solo cambia el signo si a*c < 0
assert np.isclose(spearman(np.exp(x_lin), y_lin), spearman(x_lin, y_lin)) and not np.isclose(pearson(np.exp(x_lin), y_lin), pearson(x_lin, y_lin), atol=0.05)
print("\nPearson es invariante a cambios lineales de escala (a·x+b); Spearman y Kendall lo son a CUALQUIER transformación monótona.")
````

---
**Lectura.** (1) **Lineal:** los tres coeficientes son altos (0.94, 0.94 y 0.79). (2) **Exponencial (monótona, no lineal):** Spearman y Kendall valen **1.00** y Pearson solo 0.87: la dependencia es perfecta pero no lineal. (3) **Parábola:** la relación es casi determinista ($y=x^2$) pero los tres coeficientes están en $[-0.12,-0.06]$, **como si no hubiera relación**; la información mutua (1.77 nats) sí la detecta. (4) **Un solo punto:** 30 pares independientes tienen $r=0.08$; añadir **un** atípico en $(10,10)$ lleva Pearson a **0.80** (Spearman 0.19 y Kendall 0.14): **un dato fabricó una correlación fuerte**. Las invariancias se verificaron con `assert`. La lección: **elegir el coeficiente según la pregunta** y **siempre mirar el diagrama de dispersión**.

## 5. Matriz de Covarianza y de Correlación; Asociación entre Categóricas 🧮

Para $p$ variables, la **matriz de covarianza** $S=\dfrac{1}{n-1}X_c^{\top}X_c$ (con $X_c$ la matriz centrada) es **simétrica y semidefinida positiva**; la **matriz de correlación** es $R=D^{-1/2}\,S\,D^{-1/2}$ con $D=\operatorname{diag}(s_1^2,\dots,s_p^2)$, y coincide con la covarianza de las variables **estandarizadas**. Se visualiza con un mapa de calor con **colormap divergente centrado en 0**. Usamos los **178 vinos** del conjunto *wine* de scikit-learn (13 variables químicas reales de tres cultivares).

Para **dos variables categóricas** se parte de la **tabla de contingencia** (observados $O_{ij}$) y de los valores esperados bajo independencia $E_{ij}=\dfrac{(\text{total fila}_i)(\text{total columna}_j)}{n}$; el estadístico $\chi^2=\sum_{ij}\dfrac{(O_{ij}-E_{ij})^2}{E_{ij}}$ se normaliza en la **V de Cramér**

$$V=\sqrt{\frac{\chi^2}{n\,\big(\min(r,c)-1\big)}}\in[0,1],$$

que aquí se usa **solo como medida descriptiva** (la prueba de hipótesis del $\chi^2$ se ve en el Módulo 03). Para comparar con una variable numérica se **discretiza en cuartiles**.

````python
# ============================================================
# 5. Matriz de covarianza y correlacion (datos reales de scikit-learn: vinos) y V de Cramer
# ============================================================
from sklearn.datasets import load_wine

vino = load_wine(as_frame=True)
X, clase = vino.data, vino.target                                                                # 178 vinos, 13 variables quimicas; 3 cultivares
Xc = X - X.mean()
S = Xc.T @ Xc / (len(X) - 1)                                                                     # matriz de covarianza desde cero
D = np.sqrt(np.diag(S))
R = S / np.outer(D, D)                                                                           # R = D^-1/2 S D^-1/2
assert np.allclose(S, X.cov()) and np.allclose(R, X.corr()) and np.linalg.eigvalsh(S.to_numpy()).min() > -1e-9   # S es semidefinida positiva
assert np.allclose(((Xc / D).T @ (Xc / D) / (len(X) - 1)), R)                                    # la covarianza de variables estandarizadas ES la correlacion
pares = R.where(np.triu(np.ones(R.shape, bool), 1)).stack().sort_values(key=np.abs, ascending=False)
print("Pares más correlacionados (Pearson):\n" + pares.head(4).round(3).to_string())
assert pares.abs().iloc[0] > 0.8

fig, ejes = plt.subplots(1, 2, figsize=(13, 5.4), gridspec_kw={"width_ratios": [1.15, 1]})
im = ejes[0].imshow(R, cmap="coolwarm", vmin=-1, vmax=1)
ejes[0].set_xticks(range(13)); ejes[0].set_yticks(range(13)); ejes[0].set_xticklabels(X.columns, rotation=90, fontsize=8); ejes[0].set_yticklabels(X.columns, fontsize=8); ejes[0].grid(False)
plt.colorbar(im, ax=ejes[0], fraction=0.046, label="correlación de Pearson"); ejes[0].set_title("Matriz de correlación (divergente, centrada en 0)", fontweight="bold", fontsize=10)
for k, col in zip(range(3), PALETA):
    ejes[1].scatter(X.loc[clase == k, "flavanoids"], X.loc[clase == k, "total_phenols"], s=22, color=col, alpha=0.8, label=f"cultivar {k}")
ejes[1].set_xlabel("flavanoids"); ejes[1].set_ylabel("total_phenols"); ejes[1].legend()
ejes[1].set_title(f"El par más correlacionado: r = {pearson(X['flavanoids'], X['total_phenols']):.3f}", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()


# --- asociacion entre categoricas: V de Cramer = sqrt(chi2 / (n (min(filas, columnas) - 1)))
def cramer_v(tab):
    obs = np.asarray(tab, dtype=float)
    esp = np.outer(obs.sum(axis=1), obs.sum(axis=0)) / obs.sum()
    chi2 = ((obs - esp) ** 2 / esp).sum()
    return float(np.sqrt(chi2 / (obs.sum() * (min(obs.shape) - 1))))


cuartil = lambda col: pd.qcut(X[col], 4, labels=["Q1", "Q2", "Q3", "Q4"])
tabla_alc = pd.crosstab(clase, cuartil("alcohol"))
print("\nCultivar frente a cuartil de alcohol:\n" + tabla_alc.to_string())
v_alc, v_fla = cramer_v(tabla_alc), cramer_v(pd.crosstab(clase, cuartil("flavanoids")))
assert np.isclose(v_alc, stats.contingency.association(tabla_alc, method="cramer", correction=False))
print(f"V de Cramér: alcohol {v_alc:.3f} | flavanoids {v_fla:.3f}")
rng_p = np.random.default_rng(SEED)
v_nulo = [cramer_v(pd.crosstab(rng_p.permutation(clase.to_numpy()), cuartil("alcohol"))) for _ in range(500)]
print(f"Con la etiqueta BARAJADA (sin asociación real) V promedio = {np.mean(v_nulo):.3f}: en tablas pequeñas V no baja de 0 por azar")
assert v_fla > v_alc > np.percentile(v_nulo, 95) and np.mean(v_nulo) > 0.05
````

---
**Lectura.** Los vinos más correlacionados son `total_phenols` y `flavanoids` ($r=0.865$), seguidos de `flavanoids` con `od280/od315_of_diluted_wines` (0.787): un grupo de variables **redundantes** (aparece como bloque rojo en el mapa), algo que importará al modelar (multicolinealidad, Módulos 04 y 06). La V de Cramér entre cultivar y cuartiles es **0.577 para el alcohol** y **0.743 para los flavanoides**: estos separan mejor los cultivares. **Cautela honesta:** con la etiqueta **barajada** (sin asociación real) la V promedio no es 0 sino **0.125**; en tablas pequeñas la V tiene un **sesgo positivo** y no se interpreta sola (véase la prueba de $\chi^2$ en el Módulo 03).
