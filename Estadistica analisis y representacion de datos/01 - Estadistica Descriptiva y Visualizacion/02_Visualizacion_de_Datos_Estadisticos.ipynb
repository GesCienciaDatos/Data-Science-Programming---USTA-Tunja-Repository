## 1. Principios: un Gráfico es una Afirmación Estadística 🧭

**Dos propósitos distintos.** *Explorar* (para uno mismo: muchos gráficos rápidos para descubrir estructura, atípicos, errores) y *comunicar* (para otros: pocos gráficos, cuidados, que digan **una** cosa verdadera). Los dos se rigen por tres ideas:

- **Jerarquía perceptual (Cleveland y McGill, 1984).** Los humanos leemos con más exactitud, de mejor a peor: **posición sobre una escala común** > posición en escalas no alineadas > **longitud**, dirección y ángulo > **área** > volumen y curvatura > **saturación y tono del color**. Por eso los puntos y las barras (posición y longitud) superan al gráfico de pastel (ángulo y área) y al 3D.
- **Tinta de datos (Tufte).** Maximiza la fracción de tinta que **representa datos** y elimina la decoración que no informa (*chartjunk*): rejillas pesadas, sombras, fondos.
- **Integridad gráfica.** El efecto mostrado debe ser **proporcional** al efecto en los datos: el **factor de mentira** $\text{FM}=\dfrac{\text{efecto mostrado en el gráfico}}{\text{efecto en los datos}}$ debe valer **1**. Además: accesibilidad (contraste, daltonismo) y honestidad sobre la incertidumbre.

**Qué gráfico para qué pregunta:**

| Pregunta / tipo de variable | Gráfico adecuado | Evitar |
|---|---|---|
| Distribución de **una numérica** | Histograma, **KDE**, **ECDF**, diagrama de caja | Pastel; barras con eje truncado |
| Numérica **por grupos** | Cajas, **violines**, **tira de puntos** (*strip*), ECDF por grupo | Barras de promedio con «bigotes» sin mostrar los datos |
| **Una categórica** | Barras (ordenadas, **desde cero**) | Pastel con muchas categorías, 3D |
| **Dos numéricas** | Dispersión; con muchos puntos, **transparencia** o ***hexbin***; tendencia **LOWESS** | Unir con líneas puntos que no son una serie |
| **Tiempo** | Líneas, **media móvil**, múltiplos pequeños | Ejes duales sin justificación |
| **Muchas variables** | Mapa de calor de correlación (colormap **divergente**), matriz de dispersión | Mapas de color arcoíris (*jet*) |
| ¿Es **normal**? | **Q-Q plot** (más informativo que una sola prueba) | Decidir solo con un valor $p$ |

Primero la configuración y los datos: ingresos de 600 hogares **ficticios** en tres zonas, 5 000 parcelas **ficticias** (lluvia y rendimiento) y un precio semanal **ficticio** durante tres años.

````python
# ============================================================
# Configuracion inicial y datos SINTETICOS (semilla fija)
# ============================================================
# !pip install -q numpy pandas scipy matplotlib statsmodels

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import stats

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
OKABE_ITO = ["#000000", "#E69F00", "#56B4E9", "#009E73", "#F0E442", "#0072B2", "#D55E00", "#CC79A7"]      # paleta apta para daltonismo (Okabe e Ito)
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")

rng = np.random.default_rng(SEED)
# --- (1) ingresos de 600 hogares FICTICIOS en tres zonas: lognormales con distinta mediana
N = 600
zona = rng.choice(["Norte", "Centro", "Sur"], size=N, p=[0.30, 0.40, 0.30])
mediana_zona = {"Norte": 1.5, "Centro": 1.9, "Sur": 2.4}
ingreso = np.round(rng.lognormal(np.log([mediana_zona[z] for z in zona]), 0.5), 2)                       # millones de pesos al mes
# --- (2) 5 000 parcelas FICTICIAS: lluvia (mm/mes) y rendimiento (t/ha) con relacion lineal + ruido
M = 5000
lluvia = rng.normal(100, 25, size=M)
rend = 8 + 0.12 * lluvia + rng.normal(0, 2.5, size=M)
# --- (3) precio semanal FICTICIO (3 anos): tendencia + estacionalidad anual + ruido
semana = np.arange(156)
precio = 1000 + 2.0 * semana + 120 * np.sin(2 * np.pi * semana / 52) + rng.normal(0, 60, size=156)       # pesos por kg

desv = lambda x: float(np.std(x, ddof=1))
iqr = lambda x: float(np.subtract(*np.percentile(x, [75, 25])))
print(f"ingreso: n={N}, mediana por zona = { {z: float(np.median(ingreso[zona == z])) for z in mediana_zona} }")
print(f"parcelas: n={M}, r(lluvia, rendimiento) = {np.corrcoef(lluvia, rend)[0, 1]:.3f} | precio semanal: n={semana.size}, rango = [{precio.min():.0f}, {precio.max():.0f}]")
assert ingreso.shape == (N,) and rend.shape == (M,) and precio.shape == (156,) and np.corrcoef(lluvia, rend)[0, 1] > 0.5
````

---
**Lectura.** El ingreso mediano crece de la zona Norte (1.50) a la Sur (2.42): ese contraste lo mostraremos con cajas y violines. Las 5 000 parcelas tienen $r=0.763$ entre lluvia y rendimiento (relación lineal con ruido, por construcción) y el precio semanal oscila entre 865 y 1 446 pesos/kg.

## 2. Distribuciones de una Variable: Histograma, KDE y ECDF 📊

**Histograma.** Divide el rango $R=x_{(n)}-x_{(1)}$ en $k$ clases de ancho $w$ y cuenta datos por clase. **El aspecto depende de $k$**: con muy pocas clases se **oculta** la forma; con demasiadas aparece **ruido**. Reglas para elegirlo (con $k=\lceil R/w\rceil$):

| Regla | Ancho / número de clases | Supuesto |
|---|---|---|
| **Sturges** | $k=\lceil\log_2 n+1\rceil$ | Aprox. normal; **se queda corta** si $n$ es grande |
| **Scott** | $w=3.49\,s\,n^{-1/3}$ | Minimiza el error cuadrático para datos **normales** |
| **Freedman-Diaconis** | $w=2\,\text{RIC}\,n^{-1/3}$ | Usa el **RIC** (robusto a atípicos y colas largas) |

**KDE (estimación de densidad por núcleos).** Suaviza el histograma: $\hat{f}_h(t)=\dfrac{1}{nh}\sum_{i=1}^{n}K\!\Big(\dfrac{t-x_i}{h}\Big)$ con $K$ la campana gaussiana. El **ancho de banda $h$** controla el compromiso **sesgo–varianza**: $h$ pequeño sigue el ruido, $h$ grande aplasta la forma (un KDE gaussiano también «derrama» densidad fuera del soporte, p. ej. por debajo de 0 para ingresos). **ECDF:** $\hat{F}_n(t)=\#\{x_i\le t\}/n$ no necesita elegir ni clases ni $h$, y permite leer **cuantiles** y **comparar grupos** con exactitud (ver el cuaderno de posición).

La celda implementa el **número de clases** de cada regla y la verifica contra `np.histogram_bin_edges`, el **KDE** desde cero (verificado contra `scipy.stats.gaussian_kde`) y la ECDF.

````python
# ============================================================
# 2. Histograma (reglas de ancho de clase), KDE y ECDF desde cero
# ============================================================
def n_clases(x, regla):
    """Numero de clases segun la regla; el ancho es: Scott 3.49·s·n^(-1/3), Freedman-Diaconis 2·RIC·n^(-1/3)."""
    x = np.asarray(x, dtype=float)
    n = x.size
    if regla == "sturges":
        return int(np.ceil(np.log2(n) + 1))
    ancho = 3.4908 * np.std(x) * n ** (-1 / 3) if regla == "scott" else 2 * iqr(x) * n ** (-1 / 3)    # np.std con ddof=0, como NumPy
    return int(np.ceil(np.ptp(x) / ancho))


def kde_gauss(x, malla, h):
    """Estimador de densidad con nucleo gaussiano: promedio de campanas de desviacion h centradas en cada dato."""
    z = (malla[:, None] - np.asarray(x)[None, :]) / h
    return np.exp(-0.5 * z ** 2).sum(axis=1) / (len(x) * h * np.sqrt(2 * np.pi))


def ecdf(x):
    xs = np.sort(np.asarray(x, dtype=float))
    return xs, np.arange(1, xs.size + 1) / xs.size


filas = {}
for nombre, x in {"ingreso (n=600)": ingreso, "rendimiento (n=5000)": rend}.items():
    filas[nombre] = {regla: n_clases(x, regla) for regla in ("sturges", "scott", "fd")}
    for regla in ("sturges", "scott", "fd"):
        assert filas[nombre][regla] == len(np.histogram_bin_edges(x, bins=regla)) - 1, (nombre, regla)      # contraste con NumPy
print(pd.DataFrame(filas).T.rename(columns={"fd": "Freedman-Diaconis"}).to_string())

# --- KDE desde cero frente a SciPy (factor de Silverman de SciPy: (3n/4)^(-1/5) por la desviacion estandar)
malla = np.linspace(ingreso.min() - 0.5, ingreso.max() + 0.5, 400)
h_scipy = (0.75 * N) ** (-0.2) * desv(ingreso)
assert np.allclose(kde_gauss(ingreso, malla, h_scipy), stats.gaussian_kde(ingreso, bw_method="silverman")(malla))
dens = kde_gauss(ingreso, malla, h_scipy)
area = float(((dens[:-1] + dens[1:]) / 2 * np.diff(malla)).sum())                                              # regla del trapecio
h_silverman = 0.9 * min(desv(ingreso), iqr(ingreso) / 1.34) * N ** (-0.2)                                       # regla de Silverman "robusta" (de memoria, verificar)
print(f"\nAncho de banda del KDE: SciPy 'silverman' h = {h_scipy:.3f} | regla robusta 0.9·min(s, RIC/1.34)·n^(-1/5) = {h_silverman:.3f} | área bajo el KDE = {area:.3f}")
assert abs(area - 1) < 0.02 and h_silverman < h_scipy                                                           # con asimetria, la regla robusta suaviza menos

# --- ECDF: la lectura exacta, sin elegir clases ni ancho de banda
xs, Fs = ecdf(ingreso)
assert Fs[-1] == 1.0 and np.all(np.diff(xs) >= 0) and np.isclose(np.interp(np.median(ingreso), xs, Fs), 0.5, atol=0.01)

fig, ejes = plt.subplots(1, 4, figsize=(16, 3.9))
for ax, k, titulo in zip(ejes[:3], (4, n_clases(ingreso, "fd"), 150), ("demasiado pocas clases (4)", "regla Freedman-Diaconis", "demasiadas clases (150)")):
    ax.hist(ingreso, bins=k, color=PALETA[0], edgecolor="white", alpha=0.85); ax.set_title(titulo + (f" ({k})" if k > 4 and k < 150 else ""), fontsize=9.5, fontweight="bold"); ax.set_xlabel("ingreso (millones/mes)")
for h, col, et in [(h_scipy / 4, PALETA[1], "h/4 (subsuaviza)"), (h_scipy, PALETA[2], "h de Silverman"), (4 * h_scipy, PALETA[3], "4h (sobresuaviza)")]:
    ejes[3].plot(malla, kde_gauss(ingreso, malla, h), color=col, lw=2, label=et)
ejes[3].hist(ingreso, bins="fd", density=True, color="#e2e8f0", edgecolor="white"); ejes[3].legend(fontsize=8); ejes[3].set_title("KDE: el ancho de banda decide", fontsize=9.5, fontweight="bold"); ejes[3].set_xlabel("ingreso (millones/mes)")
ejes[0].set_ylabel("frecuencia")
plt.tight_layout(); plt.show()
````

---
**Lectura.** Las tres reglas **no coinciden** y la diferencia crece con $n$: para los 600 ingresos piden 11 (Sturges), 20 (Scott) y 30 (Freedman-Diaconis) clases; para las 5 000 parcelas, 14, 37 y 49. Sturges crece como $\log_2 n$ (muy lento) y las otras como $n^{1/3}$. Los cuatro gráficos lo muestran: con **4 clases** se ve una rampa sin forma, con **150** hay picos de puro ruido (los ingresos están redondeados al centésimo), y con la regla de Freedman-Diaconis aparece la **cola derecha** limpia. En el KDE, $h/4$ dibuja ondulaciones que no existen y $4h$ **aplasta** el pico y lo **corre** hacia valores bajos. El $h$ de SciPy ('silverman', 0.355) es mayor que el de la regla robusta $0.9\min(s,\text{RIC}/1.34)n^{-1/5}$ (0.253) porque los datos son asimétricos; el área bajo el KDE es 0.999, es decir, una densidad válida.

## 3. Comparar Grupos y Revisar la Forma: Caja, Violín y Q-Q 📦

**Diagrama de caja (Tukey).** La caja va de $Q_1$ a $Q_3$ con una línea en la mediana; los **bigotes** llegan al dato más extremo que queda **dentro** de las vallas $[Q_1-1.5\,\text{RIC},\;Q_3+1.5\,\text{RIC}]$; lo que cae fuera se dibuja como puntos. Resume en cinco números **pero no muestra la forma** (una distribución bimodal tiene la misma caja que una unimodal): por eso se **superpone la tira de puntos** o se usa un **violín** (un KDE reflejado). Con datos asimétricos, la regla $1.5\,\text{RIC}$ marca como «atípicos» valores legítimos de la cola (cuaderno anterior).

**Q-Q plot normal.** Ordena los datos y grafica $\big(q_i,\,x_{(i)}\big)$ con $q_i=\Phi^{-1}(m_i)$ los **cuantiles teóricos** de una normal en las posiciones $m_i$ (aproximación de Filliben: $m_i=\dfrac{i-0.3175}{n+0.365}$ y, en los extremos, $1-0.5^{1/n}$ y $0.5^{1/n}$). Si los datos son normales, los puntos caen sobre una **recta** cuya pendiente es la escala $\sigma$ y cuyo intercepto es la ubicación $\mu$. **Cómo leerlo:** curva convexa = **asimetría a la derecha**; forma de S = colas más pesadas o más ligeras que la normal. La **correlación** de los puntos con su recta resume qué tan recto es.

La celda calcula la **anatomía de la caja** a mano (y la compara con la que dibuja Matplotlib), las posiciones de Filliben (comparadas con `scipy.stats.probplot`) y arma los cuatro gráficos.

````python
# ============================================================
# 3. Comparar grupos: diagrama de caja (anatomia desde cero), violin, Q-Q (posiciones de Filliben)
# ============================================================
from matplotlib.cbook import boxplot_stats


def anatomia_caja(x, k=1.5):
    x = np.sort(np.asarray(x, dtype=float))
    q1, med, q3 = np.percentile(x, [25, 50, 75])
    riq = q3 - q1
    bigote_inf, bigote_sup = x[x >= q1 - k * riq].min(), x[x <= q3 + k * riq].max()                        # el dato mas extremo DENTRO de las vallas
    atipicos = x[(x < bigote_inf) | (x > bigote_sup)]
    return {"q1": q1, "mediana": med, "q3": q3, "bigote_inf": bigote_inf, "bigote_sup": bigote_sup, "atipicos": atipicos}


filas = []
for z in ("Norte", "Centro", "Sur"):
    a, ref = anatomia_caja(ingreso[zona == z]), boxplot_stats(ingreso[zona == z])[0]                        # ref: lo que dibuja Matplotlib
    assert np.isclose(a["q1"], ref["q1"]) and np.isclose(a["q3"], ref["q3"]) and np.isclose(a["bigote_inf"], ref["whislo"])
    assert np.isclose(a["bigote_sup"], ref["whishi"]) and a["atipicos"].size == len(ref["fliers"])
    filas.append({"zona": z, "n": int((zona == z).sum()), "mediana": a["mediana"], "RIC": a["q3"] - a["q1"], "marcados por la regla 1.5·RIC": a["atipicos"].size})
tabla = pd.DataFrame(filas).set_index("zona")
print(tabla.round(3).to_string())
assert tabla.loc["Norte", "mediana"] < tabla.loc["Centro", "mediana"] < tabla.loc["Sur", "mediana"]


def cuantiles_teoricos(n):
    """Cuantiles normales en las 'medianas de los estadisticos de orden uniformes' (aproximacion de Filliben, la de scipy.stats.probplot)."""
    m = (np.arange(1, n + 1) - 0.3175) / (n + 0.365)
    m[0], m[-1] = 1 - 0.5 ** (1 / n), 0.5 ** (1 / n)
    return stats.norm.ppf(m)


def r_qq(x):
    return float(np.corrcoef(cuantiles_teoricos(len(x)), np.sort(x))[0, 1])


(osm, osr), (pend, orig, r_sp) = stats.probplot(ingreso, dist="norm")
assert np.allclose(osm, cuantiles_teoricos(N)) and np.isclose(r_qq(ingreso), r_sp)
print(f"\nCorrelación del Q-Q normal: ingreso {r_qq(ingreso):.4f} | log(ingreso) {r_qq(np.log(ingreso)):.4f} | rendimiento {r_qq(rend):.4f}")
assert r_qq(np.log(ingreso)) > r_qq(ingreso) and r_qq(rend) > 0.99

fig, ejes = plt.subplots(2, 2, figsize=(12.5, 8))
grupos = [ingreso[zona == z] for z in ("Norte", "Centro", "Sur")]
bp = ejes[0, 0].boxplot(grupos, patch_artist=True, medianprops=dict(color="black", lw=2))
ejes[0, 0].set_xticklabels(["Norte", "Centro", "Sur"])
for caja, col in zip(bp["boxes"], PALETA):
    caja.set(facecolor=col, alpha=0.55)
for i, g in enumerate(grupos, start=1):
    ejes[0, 0].scatter(i + np.random.default_rng(i).uniform(-0.18, 0.18, g.size), g, s=5, color="#334155", alpha=0.35)       # tira de puntos (jitter): se ve cada dato
ejes[0, 0].set_title("Caja + puntos: resumen y datos a la vez", fontweight="bold", fontsize=10); ejes[0, 0].set_ylabel("ingreso (millones/mes)")
vp = ejes[0, 1].violinplot(grupos, showmedians=True, showextrema=False)
for cuerpo, col in zip(vp["bodies"], PALETA):
    cuerpo.set(facecolor=col, alpha=0.6)
ejes[0, 1].set_xticks([1, 2, 3]); ejes[0, 1].set_xticklabels(["Norte", "Centro", "Sur"]); ejes[0, 1].set_title("Violín: la forma de cada distribución", fontweight="bold", fontsize=10)
for ax, x, tit in ((ejes[1, 0], ingreso, "Q-Q normal del ingreso: curva = asimetría"), (ejes[1, 1], np.log(ingreso), "Q-Q normal del log(ingreso): casi una recta")):
    teo = cuantiles_teoricos(len(x)); xs_ = np.sort(x); b, a = np.polyfit(teo, xs_, 1)
    ax.scatter(teo, xs_, s=8, color=PALETA[0]); ax.plot(teo, a + b * teo, color=PALETA[1], lw=1.5)
    ax.set_xlabel("cuantiles teóricos (normal)"); ax.set_ylabel("cuantiles de la muestra"); ax.set_title(f"{tit}\nr = {r_qq(x):.4f}", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

Las medianas suben de Norte (1.50) a Sur (2.42) y el RIC también (0.89, 1.21 y 1.62): **la dispersión crece con el centro**, típico de datos multiplicativos (lognormales). ## 4. Relaciones entre Variables y Series de Tiempo 🔗

- **Sobreposición de puntos (*overplotting*).** Con miles de puntos opacos, la nube se vuelve una **mancha** y la densidad desaparece. Soluciones: **transparencia** (`alpha` pequeño, así las zonas densas se oscurecen), **hexbin** (conteos por celdas hexagonales con colormap secuencial) o gráficos de contorno.
- **Tendencia suave (LOWESS).** *Locally weighted scatterplot smoothing*: en cada $x$ ajusta una **regresión lineal ponderada** con los datos cercanos (una fracción `frac` del total, con pesos que decaen con la distancia) y toma su valor. No supone una forma funcional: dibuja lo que hay, y se compara con la recta de mínimos cuadrados.
- **Series de tiempo.** Línea (el orden **sí** importa); la **media móvil** de $m$ periodos, $\bar{x}_t=\frac1m\sum_{j=0}^{m-1}x_{t-j}$, reduce el ruido y deja ver la tendencia (cuanto mayor $m$, más suave y **más retrasada** la curva; ventana de 52 semanas elimina la estacionalidad anual).

La celda implementa la media móvil con **sumas acumuladas** (verificada contra `pandas.rolling`), ajusta LOWESS y dibuja los cuatro paneles. Se comprueba además que **los conteos del *hexbin* suman las 5 000 parcelas**.

````python
# ============================================================
# 4. Relaciones y series: sobreposicion de puntos, tendencia (LOWESS), hexbin, media movil
# ============================================================
from statsmodels.nonparametric.smoothers_lowess import lowess


def media_movil(x, ventana):
    """Promedio de las ultimas `ventana` observaciones (los primeros ventana-1 valores quedan NaN), con sumas acumuladas."""
    c = np.cumsum(np.insert(np.asarray(x, dtype=float), 0, 0.0))
    out = np.full(len(x), np.nan)
    out[ventana - 1:] = (c[ventana:] - c[:-ventana]) / ventana
    return out


mm = media_movil(precio, 13)
assert np.allclose(mm, pd.Series(precio).rolling(13).mean().to_numpy(), equal_nan=True)                  # contraste con pandas
suav = lowess(rend, lluvia, frac=0.3, return_sorted=True)
ajuste_100 = float(np.interp(100.0, suav[:, 0], suav[:, 1]))
print(f"LOWESS en lluvia = 100 mm: {ajuste_100:.2f} t/ha (verdad: 8 + 0.12·100 = 20.00) | media móvil de 13 semanas, último valor: {mm[-1]:.1f}")
assert abs(ajuste_100 - 20.0) < 0.5

fig, ejes = plt.subplots(2, 2, figsize=(12.5, 8))
ejes[0, 0].scatter(lluvia, rend, s=12, color=PALETA[0]); ejes[0, 0].set_title("5 000 puntos opacos: la nube es una mancha", fontweight="bold", fontsize=10)
ejes[0, 1].scatter(lluvia, rend, s=12, color=PALETA[0], alpha=0.06, edgecolor="none"); ejes[0, 1].set_title("Con transparencia (alpha = 0.06) aparece la densidad", fontweight="bold", fontsize=10)
hb = ejes[1, 0].hexbin(lluvia, rend, gridsize=30, cmap="viridis", mincnt=1); plt.colorbar(hb, ax=ejes[1, 0], label="parcelas por celda")
ejes[1, 0].plot(suav[:, 0], suav[:, 1], color="white", lw=3); ejes[1, 0].plot(suav[:, 0], suav[:, 1], color=PALETA[1], lw=1.6, label="LOWESS"); ejes[1, 0].legend()
ejes[1, 0].set_title("Hexbin (conteos) + tendencia suave", fontweight="bold", fontsize=10)
for ax in (ejes[0, 0], ejes[0, 1], ejes[1, 0]):
    ax.set_xlabel("lluvia (mm/mes)"); ax.set_ylabel("rendimiento (t/ha)")
assert int(hb.get_array().sum()) == M                                                                     # los conteos de las celdas suman las 5 000 parcelas
ejes[1, 1].plot(semana, precio, color="#94a3b8", lw=1, label="precio semanal"); ejes[1, 1].plot(semana, mm, color=PALETA[1], lw=2.2, label="media móvil (13 semanas)")
ejes[1, 1].set_xlabel("semana"); ejes[1, 1].set_ylabel("pesos por kg"); ejes[1, 1].legend(); ejes[1, 1].set_title("Serie de tiempo: el suavizado revela la tendencia", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** Con 5 000 puntos opacos (panel superior izquierdo) solo se ve una mancha; con `alpha = 0.06` aparece el **núcleo denso** de la nube y el *hexbin* lo cuantifica (los conteos suman 5 000). El LOWESS estima 20.03 t/ha a 100 mm, prácticamente la verdad ($8+0.12\cdot100=20$): la tendencia es **lineal**, y LOWESS lo descubre sin suponerlo. En la serie, la media móvil de 13 semanas (último valor 1 209.8) elimina el ruido semanal y muestra la **tendencia creciente con ciclo anual**.

## 5. Gráficos Honestos: Escala, Color y Accesibilidad ⚖️

- **Eje de las barras.** Las barras codifican **longitud**, así que su base debe ser **0**. Un eje truncado multiplica visualmente la diferencia. El **factor de mentira** lo mide: con $v_0$ y $v_1$ y un eje que empieza en $b$, $\text{FM}=\dfrac{(v_1-v_0)/(v_0-b)}{(v_1-v_0)/v_0}=\dfrac{v_0}{v_0-b}$. (Un eje truncado es admisible en **puntos** o **líneas**, donde se codifica posición, si se avisa.)
- **Contraste.** La **luminancia relativa** WCAG de un color sRGB es $Y=0.2126\,R_{\rm lin}+0.7152\,G_{\rm lin}+0.0722\,B_{\rm lin}$ (con la linealización gamma) y el **contraste** entre dos colores es $\dfrac{Y_{\max}+0.05}{Y_{\min}+0.05}$, de 1 (iguales) a 21 (negro sobre blanco). Para elementos gráficos se recomienda **al menos 3:1** *(WCAG 2.x; de memoria, verificar)*.
- **Daltonismo.** Alrededor del 8 % de los hombres y del 0.5 % de las mujeres tienen alguna deficiencia de visión de color *(cifras de memoria, verificar)*; evita depender solo del contraste rojo-verde y usa paletas como **Okabe-Ito**, además de formas o etiquetas.
- **Mapas de color.** Para datos **ordenados** usa un colormap **secuencial perceptualmente uniforme** (*viridis*): su **luminancia crece de forma monótona** y se ve bien en escala de grises. Para datos con un centro natural (correlaciones), uno **divergente** centrado (*coolwarm*). **Evita *jet*** (arcoíris): su luminancia sube y baja y crea «franjas» falsas.

La celda calcula el **factor de mentira**, el **contraste** de la paleta de Okabe-Ito contra el blanco y verifica la **monotonía de la luminancia** de tres mapas de color.

````python
# ============================================================
# 5. Graficos honestos: factor de mentira, contraste de colores (WCAG) y luminancia de los mapas de color
# ============================================================
def factor_mentira(v0, v1, base_eje):
    """Tufte: (efecto mostrado en el grafico) / (efecto en los datos); con barras, la altura visible es valor - base del eje."""
    efecto_grafico = ((v1 - base_eje) - (v0 - base_eje)) / (v0 - base_eje)
    efecto_datos = (v1 - v0) / v0
    return efecto_grafico / efecto_datos


coop = {"Cooperativa X": 520, "Cooperativa Y": 550}                                                      # toneladas FICTICIAS
lf_trunc, lf_cero = factor_mentira(520, 550, 500), factor_mentira(520, 550, 0)
print(f"Factor de mentira: eje desde 500 -> {lf_trunc:.1f} | eje desde 0 -> {lf_cero:.1f} (ideal = 1; datos: +{(550 - 520) / 520:.1%}, gráfico: +{(50 - 20) / 20:.0%})")
assert np.isclose(lf_cero, 1.0) and np.isclose(lf_trunc, 26.0, atol=0.1)


def luminancia(color):
    """Luminancia relativa WCAG de un color sRGB (0 = negro, 1 = blanco); acepta '#rrggbb' o una tupla RGB en [0, 1]."""
    c = np.array([int(color[i:i + 2], 16) for i in (1, 3, 5)]) / 255 if isinstance(color, str) else np.asarray(color[:3], dtype=float)
    lin = np.where(c <= 0.03928, c / 12.92, ((c + 0.055) / 1.055) ** 2.4)
    return float(lin @ np.array([0.2126, 0.7152, 0.0722]))


def contraste(c1, c2):
    a, b = sorted((luminancia(c1), luminancia(c2)), reverse=True)
    return (a + 0.05) / (b + 0.05)


assert np.isclose(contraste("#000000", "#ffffff"), 21.0) and np.isclose(contraste("#ffffff", "#ffffff"), 1.0)
tabla_c = pd.DataFrame({"color": OKABE_ITO, "contraste con blanco": [contraste(c, "#ffffff") for c in OKABE_ITO]}).set_index("color")
tabla_c["≥ 3:1 (gráficos)"] = tabla_c["contraste con blanco"] >= 3
print(tabla_c.round(2).T.to_string())
fallan = list(tabla_c.index[~tabla_c["≥ 3:1 (gráficos)"]])
print("Colores de Okabe-Ito que NO alcanzan 3:1 sobre fondo blanco:", fallan)
assert "#F0E442" in fallan and tabla_c.loc["#0072B2", "≥ 3:1 (gráficos)"]                              # ser apta para daltonismo no basta: el amarillo casi no se ve sobre blanco

import matplotlib as mpl
t_ = np.linspace(0, 1, 256)
Y = {nombre: np.array([luminancia(mpl.colormaps[nombre](v)) for v in t_]) for nombre in ("viridis", "jet", "coolwarm")}
mono = {nombre: bool(np.all(np.diff(y) > 0)) for nombre, y in Y.items()}                                   # 255 saltos entre colores consecutivos
print("¿Luminancia estrictamente creciente? ", mono)
assert mono["viridis"] and not mono["jet"] and not mono["coolwarm"]                                      # viridis es secuencial; jet engaña (franjas brillantes); coolwarm es divergente

fig, ejes = plt.subplots(2, 2, figsize=(12.5, 8))
for ax, base, tit in ((ejes[0, 0], 500, f"Eje truncado en 500: factor de mentira {lf_trunc:.0f}"), (ejes[0, 1], 0, "Eje desde 0: factor de mentira 1")):
    ax.bar(list(coop), list(coop.values()), color=[PALETA[0], PALETA[1]]); ax.set_ylim(base, 560); ax.set_ylabel("toneladas"); ax.set_title(tit, fontweight="bold", fontsize=10)
ejes[1, 0].barh(range(len(OKABE_ITO)), tabla_c["contraste con blanco"], color=OKABE_ITO, edgecolor="#64748b"); ejes[1, 0].axvline(3, color="black", ls="--", lw=1)
ejes[1, 0].set_yticks(range(len(OKABE_ITO))); ejes[1, 0].set_yticklabels(OKABE_ITO, fontsize=8); ejes[1, 0].set_xlabel("contraste con fondo blanco (línea: 3:1)"); ejes[1, 0].set_title("Paleta apta para daltonismo, pero no todos los colores se leen", fontweight="bold", fontsize=10)
for nombre, y in Y.items():
    ejes[1, 1].plot(t_, y, lw=2, label=nombre)
ejes[1, 1].set_xlabel("posición en el mapa de color"); ejes[1, 1].set_ylabel("luminancia relativa"); ejes[1, 1].legend(); ejes[1, 1].set_title("Luminancia: viridis crece; jet y coolwarm no", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Con 520 y 550 toneladas (+5.8 %), un eje que empieza en 500 muestra barras de 20 y 50: **+150 %**, un factor de mentira de **26**; con el eje desde 0 el factor es **1**. (2) La paleta de Okabe-Ito, pensada para daltonismo, **no garantiza legibilidad**: sobre fondo blanco el amarillo `#F0E442` solo tiene contraste 1.32, y el naranja `#E69F00` y el celeste `#56B4E9` tampoco llegan a 3:1 (2.25 y 2.31); funcionan como **relleno grande** o con borde oscuro, pero no como línea fina o texto. El negro llega a 21 y el azul `#0072B2` a 5.19. (3) La luminancia de **viridis crece sin saltos** (255 de 255 pasos positivos); *jet* sube, baja y vuelve a subir, y *coolwarm* (divergente) sube y baja **simétricamente**, como debe ser en un colormap con un centro.

| **Histograma** | El ancho de clase decide la forma: Sturges (11 clases en n=600; 14 en n=5 000) se queda corta; Scott (20; 37) y **Freedman-Diaconis** (30; 49, usa el RIC) crecen como $n^{1/3}$. Las tres se verificaron con `np.histogram_bin_edges` |
| **KDE y ECDF** | El **ancho de banda** equilibra sesgo y varianza ($h/4$ ondula, $4h$ aplasta); la **ECDF** no tiene parámetros. El KDE desde cero coincide con `gaussian_kde` |
| **Caja, violín y Q-Q** | La caja resume cinco números pero **no la forma**: añadir puntos o violín. Anatomía verificada contra `boxplot_stats`. El Q-Q del ingreso se curva (r = 0.9350) y el del logaritmo es casi recto (0.9990) |
| **Muchos puntos** | Transparencia, ***hexbin*** (los conteos suman $n$) y **LOWESS** (20.03 t/ha frente a 20.00 reales) evitan la mancha; la **media móvil** (verificada contra pandas) revela la tendencia |
| **Integridad** | **Factor de mentira** = 26 con un eje truncado (520 vs 550, eje en 500) frente a 1 desde cero: las barras empiezan en **0** |
| **Color y accesibilidad** | Okabe-Ito sirve para daltonismo pero el amarillo tiene contraste 1.32 sobre blanco (< 3:1); *viridis* tiene luminancia monótona y *jet* no: **evita el arcoíris** |
