## 1. Datos de Trabajo y Escalas de Medición 🧭

Un **estadístico descriptivo** es una función de la muestra que condensa información; no hay uno «correcto» sino **uno adecuado a la pregunta, a la escala y a la forma de los datos**. Trabajaremos con **400 hogares agrícolas ficticios** (todo es **inventado**; los municipios, ingresos y rendimientos solo ambientan el **mecanismo**) y cuatro variables:

| Variable | Contenido | Escala | Medida de tendencia central con sentido |
|---|---|---|---|
| `variedad` | Pastusa, Criolla, Capiro, Diacol | **Nominal** | Solo la **moda** (no hay orden ni distancias) |
| `talla` | pequeña < mediana < grande | **Ordinal** | **Mediana** y cuantiles (hay orden, **no** distancias iguales); la media **no** está definida |
| `rendimiento` | t/ha de papa, aproximadamente simétrico | **Razón** | Media, mediana, recortada: coinciden aproximadamente |
| `ingreso` | millones de pesos al mes, **asimétrico a la derecha** | **Razón** | Mediana y cuantiles como resumen; la media tiene sentido pero **la arrastran** los ingresos altos |

La regla práctica: **cuanto más débil la escala, menos operaciones tienen sentido**. Y para variables de razón **positivas** con efectos multiplicativos (tasas, precios, tiempos) existen además las medias **geométrica** y **armónica**. La celda siguiente prepara el entorno, genera los datos y comprueba la tabla anterior en código.

````python
# ============================================================
# Configuracion inicial y datos SINTETICOS (semilla fija)
# ============================================================
# !pip install -q numpy pandas scipy matplotlib

import warnings
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
from scipy import stats

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__}")

# ------------------------------------------------------------
# 400 hogares agricolas FICTICIOS (todo es inventado: solo describe el MECANISMO, no un caso real)
# ------------------------------------------------------------
rng = np.random.default_rng(SEED)
N = 400
hogares = pd.DataFrame({
    "ingreso": np.round(rng.lognormal(mean=np.log(1.8), sigma=0.55, size=N), 2),            # millones de pesos al mes: asimetrico a la derecha
    "rendimiento": np.round(rng.normal(22.0, 3.5, size=N), 1),                              # t/ha de papa: aprox. simetrico (normal)
    "variedad": rng.choice(["Pastusa", "Criolla", "Capiro", "Diacol"], size=N, p=[0.40, 0.15, 0.30, 0.15]),   # escala nominal
    "talla": rng.choice(["pequeña", "mediana", "grande"], size=N, p=[0.25, 0.50, 0.25]),                       # escala ordinal
})
ingresos = hogares["ingreso"].to_numpy()
rendimiento = hogares["rendimiento"].to_numpy()
print(hogares.head(5).to_string())
print(f"\nn = {N} | ingreso: min {ingresos.min():.2f}, max {ingresos.max():.2f} | rendimiento: min {rendimiento.min():.1f}, max {rendimiento.max():.1f}")

# ------------------------------------------------------------
# Que medida de tendencia central tiene SENTIDO segun la escala
# ------------------------------------------------------------
moda_nominal = Counter(hogares["variedad"]).most_common(1)[0][0]                              # nominal: solo la moda
assert moda_nominal == hogares["variedad"].mode().iloc[0]

orden = ["pequeña", "mediana", "grande"]
codigos = np.sort(hogares["talla"].map({t: i for i, t in enumerate(orden)}).to_numpy())       # ordinal: la mediana es el valor central EN EL ORDEN
mediana_ordinal = orden[int(codigos[N // 2])]                                                 # (con N par y valores ordinales se toma el central superior)
print(f"\nvariedad (nominal) -> moda: {moda_nominal} ({(hogares['variedad'] == moda_nominal).mean():.1%} de los hogares)")
print(f"talla (ordinal)    -> mediana en el orden {orden}: {mediana_ordinal}")
print(f"ingreso (razon)    -> la media SI tiene sentido: media {ingresos.mean():.3f}, mediana {np.median(ingresos):.3f}")
assert mediana_ordinal == "mediana" and ingresos.mean() > np.median(ingresos)
````

---
**Lectura.** Los 400 hogares tienen un ingreso mínimo de 0.44 y un máximo de 8.94 millones/mes: el máximo es **20 veces** el mínimo, señal de una cola larga. La variedad más frecuente es la Pastusa (38.5 %), la talla mediana es la central en el orden, y para el ingreso **media (2.064) > mediana (1.815)**: aún no hemos mirado la forma, pero ya hay indicio de asimetría a la derecha.

## 2. La Media Aritmética y sus Variantes 📐

Para una muestra $x_1,\dots,x_n$, la **media aritmética** es

$$\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i ,$$

y la media poblacional se denota $\mu$. Propiedades que verificaremos con `assert`: (i) las desviaciones suman cero, $\sum_i (x_i-\bar{x})=0$ (la media es el **punto de equilibrio**); (ii) es **lineal**: $\overline{ax+b}=a\bar{x}+b$ (cambia de unidad sin sorpresas); (iii) **minimiza** $\sum_i (x_i-c)^2$ sobre $c$. Su costo es la **sensibilidad**: un solo valor extremo puede moverla tanto como se quiera (*punto de ruptura* $1/n$).

| Variante | Fórmula | Cuándo usarla |
|---|---|---|
| **Ponderada** | $\bar{x}_w=\dfrac{\sum_i w_i x_i}{\sum_i w_i}$ | Observaciones con pesos distintos; combinar **medias de grupos de tamaño distinto** (el «promedio de promedios» sin pesos es un error clásico) |
| **Recortada** al $\alpha$ | media de los datos ordenados tras descartar $\lfloor \alpha n\rfloor$ por cada cola | Compromiso entre media y mediana; robusta hasta una fracción $\alpha$ de errores |
| **Winsorizada** al $\alpha$ | como la recortada, pero los extremos se **reemplazan** por el valor más cercano que se conserva | Mantiene el tamaño $n$ (útil para varianzas) |
| **Geométrica** | $\bar{x}_G=\left(\prod_i x_i\right)^{1/n}=\exp\!\Big(\tfrac1n\sum_i\ln x_i\Big)$, con $x_i>0$ | **Factores multiplicativos**: crecimiento compuesto, razones |
| **Armónica** | $\bar{x}_H=\dfrac{n}{\sum_i 1/x_i}$, con $x_i>0$ | **Tasas** sobre una cantidad fija: velocidades en tramos iguales, precios por unidad |

Entre ellas, para datos positivos no todos iguales: $\bar{x}_H<\bar{x}_G<\bar{x}$ (**desigualdad de las medias**). Implementamos las seis desde cero y las contrastamos con NumPy y SciPy; después, tres aplicaciones: el **promedio de promedios**, la **tasa media de variación de un precio** (¿aritmética o geométrica?) y la **velocidad media** de un viaje de ida y vuelta.

````python
# ============================================================
# 2. La media y sus variantes, DESDE CERO y contrastadas con NumPy / SciPy
# ============================================================
def media(x):
    x = np.asarray(x, dtype=float)
    return x.sum() / x.size


def media_ponderada(x, w):
    x, w = np.asarray(x, dtype=float), np.asarray(w, dtype=float)
    return (w * x).sum() / w.sum()


def media_recortada(x, prop):
    """Descarta la fraccion `prop` de datos de CADA cola (int(prop*n), como SciPy) y promedia el resto."""
    x = np.sort(np.asarray(x, dtype=float))
    k = int(prop * x.size)
    return media(x[k: x.size - k])


def media_winsorizada(x, prop):
    """En vez de descartar, REEMPLAZA los extremos por el valor mas cercano que se conserva."""
    x = np.sort(np.asarray(x, dtype=float))
    k = int(prop * x.size)
    if k:
        x[:k], x[x.size - k:] = x[k], x[x.size - k - 1]
    return media(x)


def media_geometrica(x):
    return float(np.exp(np.log(np.asarray(x, dtype=float)).mean()))


def media_armonica(x):
    x = np.asarray(x, dtype=float)
    return x.size / (1.0 / x).sum()


# --- contraste con las implementaciones de referencia
assert np.isclose(media(ingresos), np.mean(ingresos))
assert np.isclose(media_recortada(ingresos, 0.10), stats.trim_mean(ingresos, 0.10))
assert np.isclose(media_winsorizada(ingresos, 0.10), stats.mstats.winsorize(ingresos, limits=(0.10, 0.10)).mean())
assert np.isclose(media_geometrica(ingresos), stats.gmean(ingresos)) and np.isclose(media_armonica(ingresos), stats.hmean(ingresos))

# --- propiedades de la media
assert abs((ingresos - media(ingresos)).sum()) < 1e-9                                         # las desviaciones respecto a la media suman 0
a, b = 3.0, -2.0
assert np.isclose(media(a * ingresos + b), a * media(ingresos) + b)                           # linealidad: la media respeta cambios de unidad
malla = np.linspace(ingresos.min(), ingresos.max(), 4001)
sse = ((ingresos[:, None] - malla[None, :]) ** 2).sum(axis=0)                                 # suma de cuadrados al "centro" c
assert abs(malla[sse.argmin()] - media(ingresos)) < 0.01                                      # la media MINIMIZA la suma de cuadrados

# --- la familia de medias sobre los ingresos
filas = {"aritmética": media(ingresos), "mediana": np.median(ingresos), "recortada 10 %": media_recortada(ingresos, 0.10),
         "recortada 25 %": media_recortada(ingresos, 0.25), "winsorizada 10 %": media_winsorizada(ingresos, 0.10),
         "geométrica": media_geometrica(ingresos), "armónica": media_armonica(ingresos)}
print(pd.Series(filas, name="ingreso (millones/mes)").round(3).to_string())
assert media_armonica(ingresos) <= media_geometrica(ingresos) <= media(ingresos)              # desigualdad HM <= GM <= AM

# --- media ponderada: el "promedio de promedios" NO es la media global si los grupos tienen tamanos distintos
veredas = {"A": (120, 1.5), "B": (60, 2.2), "C": (20, 3.4)}                                  # (hogares, ingreso medio) FICTICIOS
n_v = np.array([n for n, _ in veredas.values()]); m_v = np.array([m for _, m in veredas.values()])
global_ = media(np.concatenate([np.full(n, m) for n, m in veredas.values()]))
print(f"\nPromedio simple de las 3 medias: {m_v.mean():.3f} | media ponderada por hogares: {media_ponderada(m_v, n_v):.3f} | media de los 200 hogares: {global_:.3f}")
assert np.isclose(media_ponderada(m_v, n_v), global_) and not np.isclose(m_v.mean(), global_)

# --- media geometrica: variacion PROMEDIO de una serie de precios (bulto de papa, pesos, FICTICIO)
precios = np.array([80_000, 96_000, 72_000, 90_000, 117_000, 99_000], dtype=float)
factores = precios[1:] / precios[:-1]                                                         # 1 + tasa de cada periodo
g_geo = media_geometrica(factores) - 1
g_ari = (factores - 1).mean()
print(f"\nTasa media aritmética: {g_ari:.2%} | tasa media geométrica: {g_geo:.2%} (la que reproduce el precio final)")
assert np.isclose(precios[0] * (1 + g_geo) ** 5, precios[-1]) and precios[0] * (1 + g_ari) ** 5 > precios[-1] * 1.02   # la aritmetica SOBRESTIMA

# --- media armonica: velocidad media con dos tramos de IGUAL longitud (60 km/h de ida, 40 km/h de vuelta)
v_media = media_armonica([60, 40])
print(f"Velocidad media ida y vuelta: {v_media:.1f} km/h (no 50): total 2d / (d/60 + d/40)")
assert np.isclose(v_media, 48.0)
````

**Lectura.** (1) La **familia de medias** ordena los ingresos como $1.570<1.795<1.815<1.824<1.888<1.955<2.064$ (armónica, geométrica, mediana, recortada 25 %, recortada 10 %, winsorizada 10 % y aritmética): cuanto más peso se da a los valores altos, mayor el resultado, y el `assert` de $\text{HM}\le\text{GM}\le\text{AM}$ se cumple. (2) **Promedio de promedios:** el promedio simple de las medias de tres veredas (2.367) **no** es la media de los 200 hogares (1.900); solo la media **ponderada** por hogares lo es. (3) **Tasa de variación:** la media aritmética de las variaciones del precio (6.92 %) **sobrestima** el crecimiento, porque ignora que cada tasa se aplica sobre una base distinta; la **geométrica** (4.35 %) es la única que reproduce el precio final. (4) **Velocidad:** ir a 60 km/h y volver a 40 km/h da **48 km/h** de media (armónica), no 50, porque se pasa más tiempo a la velocidad menor.

## 3. Mediana y Moda: qué Minimizan y qué Dicen de la Forma 🎯

- **Mediana.** Valor que deja la mitad de los datos a cada lado: con los datos ordenados $x_{(1)}\le\dots\le x_{(n)}$, es $x_{((n+1)/2)}$ si $n$ es impar y $\tfrac12\big(x_{(n/2)}+x_{(n/2+1)}\big)$ si es par. **Minimiza** la suma de desviaciones **absolutas** $\sum_i |x_i-c|$ (cuando $n$ es par cualquier $c$ entre los dos centrales es un mínimo). Su punto de ruptura es de casi **50 %**.
- **Moda.** Valor (o intervalo) más frecuente. En datos discretos puede **no ser única** (bimodal, multimodal); en datos continuos solo tiene sentido como **máximo de una densidad estimada** (la altura depende de la ventana o del *kernel*). Es la única medida válida en escala nominal.
- **Media frente a mediana: la forma.** Con **cola a la derecha** (ingresos, precios, tiempos de espera) lo habitual es $\text{moda}<\text{mediana}<\text{media}$; con cola a la izquierda, al revés; en distribuciones simétricas unimodales las tres coinciden. Ojo: es una **regla práctica, no un teorema** (existen contraejemplos, véase von Hippel, 2005); la forma se mide con asimetría y curtosis (cuaderno siguiente).

La celda comprueba mediana y moda contra NumPy y SciPy, muestra por **búsqueda en malla** qué minimiza cada una, y dibuja los ingresos con las tres marcas.

````python
# ============================================================
# 3. Mediana y moda desde cero; que minimiza cada una; media vs mediana vs moda en datos asimetricos
# ============================================================
def mediana(x):
    x = np.sort(np.asarray(x, dtype=float))
    n = x.size
    return x[n // 2] if n % 2 else (x[n // 2 - 1] + x[n // 2]) / 2


def modas(x):
    """Todos los valores con la frecuencia maxima (puede haber mas de uno: distribucion multimodal)."""
    valores, cuentas = np.unique(np.asarray(x), return_counts=True)
    return valores[cuentas == cuentas.max()]


def moda_continua(x, puntos=2000):
    """Moda de datos continuos: el maximo de una estimacion de densidad (KDE gaussiana)."""
    malla = np.linspace(np.min(x), np.max(x), puntos)
    return float(malla[stats.gaussian_kde(x)(malla).argmax()])


rng_t = np.random.default_rng(SEED + 7)
for n_ in (7, 8, 101, 250):                                                                   # tamanos impares y pares
    z = rng_t.normal(size=n_)
    assert np.isclose(mediana(z), np.median(z))
assert list(modas([1, 2, 2, 3, 3])) == [2, 3] and modas([1, 2, 2, 3, 3])[0] == stats.mode([1, 2, 2, 3, 3]).mode   # bimodal; SciPy devuelve la menor

# --- que minimiza cada medida: la media minimiza la suma de CUADRADOS; la mediana la suma de VALORES ABSOLUTOS
sad = np.abs(ingresos[:, None] - malla[None, :]).sum(axis=0)
x_ord = np.sort(ingresos)
paso = malla[1] - malla[0]
assert x_ord[N // 2 - 1] - paso <= malla[sad.argmin()] <= x_ord[N // 2] + paso                # el minimo cae entre los dos centrales

med, mea, mod = mediana(ingresos), media(ingresos), moda_continua(ingresos)
print(f"ingresos -> moda ≈ {mod:.3f} < mediana = {med:.3f} < media = {mea:.3f}   (asimetría a la derecha)")
assert mod < med < mea

fig, ejes = plt.subplots(1, 2, figsize=(13, 4.8))
ejes[0].plot(malla, sse / N, color=PALETA[0], lw=2, label="promedio de $(x-c)^2$")
ejes[0].plot(malla, sad / N, color=PALETA[1], lw=2, label="promedio de $|x-c|$")
ejes[0].axvline(mea, color=PALETA[0], ls="--"); ejes[0].axvline(med, color=PALETA[1], ls="--")
ejes[0].set_ylim(0, 3); ejes[0].set_xlabel("centro propuesto $c$ (millones/mes)"); ejes[0].set_ylabel("error promedio")
ejes[0].set_title("La media minimiza el error cuadrático;\nla mediana, el error absoluto", fontweight="bold", fontsize=10); ejes[0].legend()
ejes[1].hist(ingresos, bins=30, color="#cbd5e1", edgecolor="white", density=True)
for valor, nombre, col in [(mod, "moda", PALETA[2]), (med, "mediana", PALETA[1]), (mea, "media", PALETA[0])]:
    ejes[1].axvline(valor, color=col, lw=2.2, label=f"{nombre} = {valor:.2f}")
ejes[1].set_xlabel("ingreso (millones/mes)"); ejes[1].set_ylabel("densidad"); ejes[1].legend()
ejes[1].set_title("Con cola a la derecha: moda < mediana < media", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** (Izquierda) El error cuadrático promedio es mínimo en la **media** (2.064) y el error absoluto promedio en la **mediana** (1.815): son dos respuestas distintas a «¿cuál es el centro?», según cómo se **castigue** el error (los cuadrados castigan mucho los valores lejanos, por eso la media se acerca a la cola). (Derecha) los ingresos cumplen $\text{moda}\approx 1.51<\text{mediana}=1.82<\text{media}=2.06$. Para **resumir un ingreso típico** de estos hogares la mediana es más honesta que la media.

## 4. Robustez y Eficiencia: el Costo de Cada Estimador 🛡️

Dos preguntas distintas, que a veces se confunden:

- **Robustez:** ¿qué tanto se mueve el estimador cuando una parte de los datos es **basura** (errores de digitación, sensores dañados)? El **punto de ruptura** es la mayor fracción de datos que se puede reemplazar por valores arbitrarios sin que el estimador pueda volverse arbitrariamente malo: **0** para la media (basta un dato), $\alpha$ para la media recortada al $\alpha$ y cerca de **0.5** para la mediana.
- **Eficiencia:** si los datos **sí son limpios**, ¿cuál estimador de la ubicación varía **menos** de muestra en muestra? Se compara el **error estándar** (desviación de la distribución muestral del estimador). Para una población **normal**, la varianza asintótica de la mediana es $\frac{\pi}{2}\cdot\frac{\sigma^2}{n}$, es decir $\text{EE(mediana)}/\text{EE(media)}\to\sqrt{\pi/2}\approx 1.253$: la mediana paga un 25 % más de error. Pero con **colas pesadas** la comparación se invierte.

**No hay estimador mejor en todo:** robustez y eficiencia **compiten**. Lo simulamos: (a) a 200 datos limpios $N(50,5^2)$ se les reemplaza una fracción $\varepsilon$ por el valor absurdo 500 y se sigue el desplazamiento de cuatro estimadores; (b) se estima el cociente de errores estándar mediana/media en 5 000 muestras de tamaño 50 de tres poblaciones.

````python
# ============================================================
# 4. Robustez (contaminacion) y eficiencia (que estimador varia menos al muestrear)
# ============================================================
rng_c = np.random.default_rng(SEED)
base = rng_c.normal(50, 5, size=200)                                                          # datos "limpios"


def contaminar(x, frac, valor=500.0):
    y = x.copy()
    y[: int(round(frac * x.size))] = valor                                                     # reemplaza una fraccion por un valor absurdo (error de digitacion)
    return y


epsilons = np.round(np.arange(0.0, 0.5501, 0.01), 2)
estimadores = {"media": media, "mediana": mediana, "recortada 10 %": lambda x: media_recortada(x, 0.10), "recortada 25 %": lambda x: media_recortada(x, 0.25)}
curvas = {nombre: np.array([f(contaminar(base, e)) for e in epsilons]) for nombre, f in estimadores.items()}
desp = {nombre: v - v[0] for nombre, v in curvas.items()}                                     # desplazamiento respecto a los datos limpios
i10, i12, i45, i55 = [int(np.where(epsilons == e)[0][0]) for e in (0.10, 0.12, 0.45, 0.55)]
print(f"{'':>14} | {'ε=10 %':>8} | {'ε=12 %':>8} | {'ε=45 %':>8} | {'ε=55 %':>8}   (desplazamiento respecto a los datos limpios)")
for nombre, d in desp.items():
    print(f"{nombre:>14} | {d[i10]:8.2f} | {d[i12]:8.2f} | {d[i45]:8.2f} | {d[i55]:8.2f}")
assert desp["media"][i10] > 40                                                                # 10 % de errores desplazan la media ~ 0.10 * 450
assert abs(desp["mediana"][i10]) < 2 and abs(desp["recortada 10 %"][i10]) < 2                 # la recortada al 10 % absorbe justo el 10 %...
assert desp["recortada 10 %"][i12] > 5                                                        # ...pero si hay 12 % de errores, "se rompe"
assert desp["mediana"][i45] < 20 and desp["mediana"][i55] > 400                              # la mediana aguanta hasta casi el 50 %, y luego colapsa

# --- eficiencia: error estandar de la media y de la mediana muestral segun la forma de la poblacion
B, n_m = 5000, 50
rng_e = np.random.default_rng(SEED + 1)
poblaciones = {"Normal": lambda size: rng_e.standard_normal(size), "t de Student (3 gl)": lambda size: rng_e.standard_t(3, size), "Laplace": lambda size: rng_e.laplace(size=size)}
razon = {}
for nombre, gen in poblaciones.items():
    muestras = gen((B, n_m))
    razon[nombre] = np.median(muestras, axis=1).std(ddof=1) / muestras.mean(axis=1).std(ddof=1)
    print(f"{nombre:>20}: EE(mediana) / EE(media) = {razon[nombre]:.3f}")
print(f"Valor teórico asintótico para la Normal: sqrt(π/2) = {np.sqrt(np.pi / 2):.3f}")
assert 1.15 < razon["Normal"] < 1.36 and razon["t de Student (3 gl)"] < 0.95 and razon["Laplace"] < 0.85

fig, ejes = plt.subplots(1, 2, figsize=(13, 4.6))
for (nombre, v), col in zip(curvas.items(), PALETA):
    ejes[0].plot(epsilons * 100, v, lw=2, color=col, label=nombre)
ejes[0].set_ylim(40, 300); ejes[0].set_xlabel("% de datos reemplazados por 500"); ejes[0].set_ylabel("valor del estimador")
ejes[0].set_title("Punto de ruptura: media 0 %, recortada 10 %, mediana ≈ 50 %", fontweight="bold", fontsize=10); ejes[0].legend()
ejes[1].bar(list(razon), list(razon.values()), color=[PALETA[0], PALETA[2], PALETA[3]])
ejes[1].axhline(1, color="black", ls="--", lw=1); ejes[1].set_ylabel("EE(mediana) / EE(media)")
ejes[1].set_title("<1: la mediana varía menos que la media (colas pesadas)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** (a) Con **10 %** de datos absurdos la media se desplaza **+45.0**, la mediana **+0.9** y la recortada al 10 % **+1.0**. Con **12 %** la recortada al 10 % ya **se rompe** (+12.2): solo absorbe lo que recorta; en cambio la mediana sigue casi quieta (+1.1) y aún con **45 %** de datos corruptos se desplaza solo +7.6; con **55 %** colapsa (+450). (b) Con datos limpios y normales, la mediana varía **1.23 veces** lo que la media (teoría: 1.253); con colas pesadas (*t* de 3 gl, Laplace) la relación se invierte (**0.80** y **0.78**): **la mediana es más eficiente que la media**. Moraleja honesta: la media es la mejor opción con datos limpios y normales; en cuanto haya errores o colas pesadas, un estimador robusto (mediana, media recortada) cuesta poco y protege mucho.

## 5. Medidas de Posición: Cuantiles, ECDF y Puntaje $z$ 📍

Las medidas de posición responden «**¿dónde cae esta observación?**» o «**¿qué valor deja una fracción $p$ de los datos por debajo?**». Sus herramientas:

- **Función de distribución empírica (ECDF):** $\hat{F}_n(t)=\dfrac{1}{n}\,\#\{i: x_i\le t\}$, una escalera que sube $1/n$ en cada dato. Su valor en $t$ por cien es el **rango percentil** de $t$.
- **Cuantil de orden $p$:** el valor $Q(p)$ tal que una fracción $p$ de los datos queda por debajo (la «lectura inversa» de la ECDF). Casos con nombre propio: **cuartiles** $Q_1,Q_2,Q_3$ ($p=0.25,0.5,0.75$), **deciles** y **percentiles**. El **resumen de cinco números** (mín, $Q_1$, mediana, $Q_3$, máx) es la base del diagrama de caja.
- **Problema:** con $n$ finito $\hat F_n$ es una escalera y el cuantil **no está definido de forma única**. **Hyndman y Fan (1996)** catalogan nueve definiciones; los tipos 4 a 9 interpolan linealmente: con $h=np+m(p)$, $j=\lfloor h\rfloor$ y $g=h-j$,

$$Q(p)=(1-g)\,x_{(j)}+g\,x_{(j+1)},\qquad x_{(0)}:=x_{(1)},\; x_{(n+1)}:=x_{(n)} .$$

| Tipo | $m(p)$ | Quién lo usa (de memoria, verificar) |
|---|---|---|
| 4 | $0$ | interpolación de la ECDF; `interpolated_inverted_cdf` |
| 5 | $0.5$ | Hazen; `hazen` |
| 6 | $p$ | Weibull; *Excel* `QUARTILE.EXC`, SPSS, Minitab; `weibull` |
| **7** | $1-p$ | **R, NumPy y pandas por defecto**; *Excel* `QUARTILE.INC`; `linear` |
| 8 | $(p+1)/3$ | insesgado para la mediana; `median_unbiased` |
| 9 | $p/4+3/8$ | insesgado bajo normalidad; `normal_unbiased` |

- **Puntaje $z$:** $z=(x-\bar{x})/s$, la distancia a la media en desviaciones estándar. Equivale a un rango percentil **solo si la distribución es (aproximadamente) normal**, porque entonces el percentil es $100\,\Phi(z)$.

La celda implementa `cuantil` para los seis tipos (un solo código), la contrasta con `np.quantile` en cada `method`, muestra **cuándo la definición importa** y compara el percentil normal con el empírico.

````python
# ============================================================
# 5. Cuantiles: los 6 tipos de Hyndman y Fan (4-9) desde cero, ECDF, percentiles y puntaje z
# ============================================================
_M = {4: lambda p: 0.0 * p, 5: lambda p: 0.5 + 0.0 * p, 6: lambda p: p, 7: lambda p: 1 - p, 8: lambda p: (p + 1) / 3, 9: lambda p: p / 4 + 3 / 8}
_NP = {4: "interpolated_inverted_cdf", 5: "hazen", 6: "weibull", 7: "linear", 8: "median_unbiased", 9: "normal_unbiased"}


def cuantil(x, p, tipo=7):
    """Cuantil muestral de orden p. Posicion (base 1) h = n*p + m(p); se interpola entre x_(j) y x_(j+1), j = floor(h)."""
    x = np.sort(np.asarray(x, dtype=float))
    n = x.size
    p = np.atleast_1d(np.asarray(p, dtype=float))
    h = n * p + _M[tipo](p)
    j = np.floor(h).astype(int)
    g = h - j
    lo, hi = x[np.clip(j - 1, 0, n - 1)], x[np.clip(j, 0, n - 1)]
    q = lo + g * (hi - lo)
    return q if q.size > 1 else float(q[0])


ps = np.linspace(0.05, 0.95, 19)
for tipo, metodo in _NP.items():                                                              # contraste con NumPy en dos muestras (una con empates)
    for muestra in (rendimiento, np.round(ingresos[:60], 1)):
        assert np.allclose(cuantil(muestra, ps, tipo), np.quantile(muestra, ps, method=metodo)), (tipo, metodo)
assert np.allclose(cuantil(ingresos, ps, 7), pd.Series(ingresos).quantile(ps).to_numpy())     # pandas usa el tipo 7 (el de R por defecto)

# --- en muestras pequenas la definicion IMPORTA
pequena = rendimiento[:11]
tabla = pd.DataFrame({f"tipo {t}": [cuantil(pequena, q, t) for q in (0.25, 0.50, 0.75)] for t in _NP}, index=["Q1", "Q2", "Q3"])
print(f"n = {pequena.size}\n" + tabla.round(2).to_string())
assert tabla.loc["Q1"].max() - tabla.loc["Q1"].min() > 0.3 and np.allclose(tabla.loc["Q2", "tipo 5":], np.median(pequena))      # los tipos 5-9 son simetricos: dan la mediana usual
grande = pd.DataFrame({f"tipo {t}": [cuantil(rendimiento, q, t) for q in (0.25, 0.75)] for t in _NP}, index=["Q1", "Q3"])
print(f"\nn = {N}\n" + grande.round(2).to_string())
assert grande.max(axis=1).sub(grande.min(axis=1)).max() < 0.1                                 # con n = 400 las definiciones casi coinciden

# --- ECDF, rango percentil y resumen de cinco numeros
def ecdf(x):
    xs = np.sort(np.asarray(x, dtype=float))
    return xs, np.arange(1, xs.size + 1) / xs.size


def rango_percentil(x, v):
    """Porcentaje de observaciones <= v: la ECDF evaluada en v, por cien."""
    return 100 * np.searchsorted(np.sort(x), v, side="right") / len(x)


for v in (1.0, 1.8, 2.55, 4.0):
    assert np.isclose(rango_percentil(ingresos, v), stats.percentileofscore(ingresos, v, kind="weak"))
cinco = [float(ingresos.min()), cuantil(ingresos, 0.25), cuantil(ingresos, 0.5), cuantil(ingresos, 0.75), float(ingresos.max())]
assert np.allclose(cinco, pd.Series(ingresos).describe()[["min", "25%", "50%", "75%", "max"]])
print("\nResumen de cinco números del ingreso:", dict(zip(["mín", "Q1", "mediana", "Q3", "máx"], np.round(cinco, 2))))
print("Deciles:", np.round(cuantil(ingresos, np.arange(0.1, 1.0, 0.1)), 2))

# --- puntaje z frente a rango percentil: coinciden solo si la distribucion es (aprox.) normal
def puntaje_z(x, v):
    return (v - media(x)) / np.std(x, ddof=1)


v = 4.0
z, p_norm, p_emp = puntaje_z(ingresos, v), 100 * stats.norm.cdf(puntaje_z(ingresos, v)), rango_percentil(ingresos, v)
z_r, p_norm_r, p_emp_r = puntaje_z(rendimiento, 28.0), 100 * stats.norm.cdf(puntaje_z(rendimiento, 28.0)), rango_percentil(rendimiento, 28.0)
print(f"\ningreso 4.0 -> z = {z:.2f}: percentil si fuera normal = {p_norm:.1f}, percentil empírico = {p_emp:.1f}")
print(f"rendimiento 28.0 -> z = {z_r:.2f}: percentil si fuera normal = {p_norm_r:.1f}, percentil empírico = {p_emp_r:.1f}")
assert abs(p_norm - p_emp) > 1.0 and abs(p_norm_r - p_emp_r) < 2.0                           # con asimetria el z engana; con simetria funciona

fig, ejes = plt.subplots(1, 2, figsize=(13, 4.6))
xs, Fs = ecdf(ingresos)
ejes[0].step(xs, Fs, where="post", color=PALETA[0], lw=2, label="ECDF empírica")
ejes[0].plot(xs, stats.norm.cdf(xs, media(ingresos), np.std(ingresos, ddof=1)), color=PALETA[1], ls="--", lw=2, label="normal con la misma media y desv.")
for q, c in zip((0.25, 0.5, 0.75), PALETA[2:5]):
    ejes[0].plot([0, cuantil(ingresos, q)], [q, q], color=c, ls=":"); ejes[0].plot([cuantil(ingresos, q)] * 2, [0, q], color=c, ls=":")
ejes[0].set_xlabel("ingreso (millones/mes)"); ejes[0].set_ylabel("proporción acumulada"); ejes[0].legend(loc="lower right")
ejes[0].set_title("ECDF: el cuantil es la 'lectura inversa'", fontweight="bold", fontsize=10)
ejes[1].boxplot(ingresos, vert=False, widths=0.5, patch_artist=True, boxprops=dict(facecolor="#bfdbfe"), medianprops=dict(color=PALETA[1], lw=2))
ejes[1].scatter([media(ingresos)], [1], marker="D", color=PALETA[0], zorder=3, label="media"); ejes[1].legend(); ejes[1].set_yticks([])
ejes[1].set_xlabel("ingreso (millones/mes)"); ejes[1].set_title("Cinco números: la caja resume el 50 % central", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** (1) Con **n = 11** el primer cuartil va de **17.32 a 19.05** según el tipo (casi 1.7 t/ha de diferencia); con **n = 400** los seis tipos difieren en menos de 0.1 t/ha. Por eso **se debe reportar la definición** cuando $n$ es pequeño y los resultados deben **reproducirse** entre programas (R y pandas usan el tipo 7; *Excel* tiene dos funciones distintas). (2) El resumen de cinco números del ingreso es $\{0.44,\,1.25,\,1.82,\,2.50,\,8.94\}$: la mediana queda más cerca de $Q_1$ que de $Q_3$, otra marca de la cola derecha. (3) **Puntaje $z$:** un hogar con ingreso 4.0 tiene $z=1.62$; si los ingresos fueran normales estaría en el percentil **94.7**, pero empíricamente está en el **92.0**: el $z$ **engaña** con asimetría. Con el rendimiento (simétrico) los dos percentiles prácticamente coinciden (95.9 y 95.8). La ECDF lo muestra: la normal con la misma media y desviación se separa de la curva real.

## 6. Datos Agrupados: Resumir desde una Tabla de Frecuencias 🧮

A veces **no se tienen los datos crudos** sino la **tabla de frecuencias** (censos publicados, informes): clases $[L_i,U_i)$ de ancho $w$, frecuencia $f_i$ y frecuencia acumulada $F_i$. Se trabaja con la **marca de clase** $m_i$ (punto medio), asumiendo que los datos se reparten **uniformemente** dentro de cada clase, y se obtienen aproximaciones:

$$\bar{x}_g=\frac{\sum_i f_i m_i}{n},\qquad \text{Mediana}\approx L_k+\frac{n/2-F_{k-1}}{f_k}\,w,\qquad \text{Moda}\approx L_m+\frac{d_1}{d_1+d_2}\,w ,$$

donde $k$ es la **clase mediana** (la primera con $F_k\ge n/2$), $m$ la **clase modal** (mayor $f$), $d_1=f_m-f_{m-1}$ y $d_2=f_m-f_{m+1}$. El error depende del **ancho de clase**: agrupar es otra pérdida de información, y se puede **medir** comparando con los datos crudos.

````python
# ============================================================
# 6. Datos agrupados (tabla de frecuencias): media, mediana y moda desde la tabla
# ============================================================
def resumen_agrupado(x, ancho):
    bordes = np.arange(0.0, np.ceil(np.max(x) / ancho) * ancho + ancho, ancho)
    f = np.histogram(x, bins=bordes)[0]                                                       # clases [a, b); la ultima es cerrada
    marca = (bordes[:-1] + bordes[1:]) / 2                                                    # marca de clase = punto medio
    F = f.cumsum()
    n = f.sum()
    k = int(np.searchsorted(F, n / 2))                                                        # clase mediana: primera con F >= n/2
    F_prev = F[k - 1] if k else 0
    mediana_g = bordes[k] + (n / 2 - F_prev) / f[k] * ancho                                   # interpolacion lineal dentro de la clase
    km = int(f.argmax())                                                                      # clase modal
    d1 = f[km] - (f[km - 1] if km else 0)
    d2 = f[km] - (f[km + 1] if km + 1 < f.size else 0)
    moda_g = bordes[km] + d1 / (d1 + d2) * ancho                                              # formula de King/Czuber
    return pd.DataFrame({"clase": [f"[{a:.1f}, {b:.1f})" for a, b in zip(bordes[:-1], bordes[1:])], "f": f, "F": F, "marca": marca}), (f * marca).sum() / n, mediana_g, moda_g


tab, media_g, mediana_g, moda_g = resumen_agrupado(ingresos, 0.5)
print(tab.head(8).to_string(index=False))
print(f"\n(ancho 0.5) media agrupada {media_g:.3f} vs exacta {media(ingresos):.3f} | mediana agrupada {mediana_g:.3f} vs exacta {mediana(ingresos):.3f} | moda agrupada {moda_g:.3f}")
assert abs(media_g - media(ingresos)) < 0.25 and abs(mediana_g - mediana(ingresos)) < 0.25

errores = []
for ancho in (0.1, 0.25, 0.5, 1.0, 2.0):
    _, mg, md_, _ = resumen_agrupado(ingresos, ancho)
    errores.append({"ancho de clase": ancho, "error en la media": abs(mg - media(ingresos)), "error en la mediana": abs(md_ - mediana(ingresos))})
errores = pd.DataFrame(errores).set_index("ancho de clase")
print("\n" + errores.round(4).to_string())
assert (errores.index.to_numpy() >= errores.max(axis=1).to_numpy()).all()                      # el error nunca supera el ancho de la clase
assert errores.iloc[-1].max() > errores.iloc[0].max()                                         # clases muy anchas pierden informacion
````

---
**Lectura.** Con clases de ancho 0.5, la media agrupada (2.069) y la mediana agrupada (1.824) quedan muy cerca de las exactas (2.064 y 1.815); la **moda agrupada** (1.432) es una aproximación gruesa (la del KDE fue 1.507): la moda es la medida más sensible a la elección de las clases. El error **nunca superó el ancho de clase**, y con clases de ancho 2 el error de la media sube a 0.024 y el de la mediana a 0.076, unas veinte veces el que se tiene con ancho 0.1 (0.0038): **agrupar es otra pérdida de información que se puede medir**. Si se tienen los datos crudos, no se deben usar las fórmulas agrupadas.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Escala y medida** | Nominal: **moda**; ordinal: **mediana y cuantiles**; intervalo y razón: **media**, con geométrica (factores) y armónica (tasas). Cuanto más débil la escala, menos operaciones tienen sentido |
| **La media** | Punto de equilibrio y mínimo de la suma de **cuadrados**; lineal; pero con **punto de ruptura 0**: con 10 % de datos absurdos se desplazó +45.0. El «promedio de promedios» exige **pesos**; la tasa media de variación es la **geométrica** (4.35 % frente a 6.92 % de la aritmética) |
| **Mediana y moda** | La mediana minimiza la suma de **valores absolutos** y resiste hasta ≈ 50 % de contaminación (+7.6 con 45 %; colapsa con 55 %). La moda puede no ser única y en continuos depende de la densidad estimada. Con cola derecha: moda (≈ 1.51) < mediana (1.82) < media (2.06) *(regla práctica, no teorema)* |
| **Robustez frente a eficiencia** | Con datos normales y limpios la mediana tiene 1.23 veces el error estándar de la media (teoría 1.253); con colas pesadas es **más** eficiente (≈ 0.8). La recortada al $\alpha$ resiste solo hasta $\alpha$ |
| **Cuantiles** | Hay **nueve definiciones** (Hyndman y Fan); con $n=11$ el $Q_1$ varió de 17.32 a 19.05, con $n=400$ casi nada. NumPy, pandas y R: **tipo 7**. Se **reporta la definición** y se verifica contra una referencia |
| **Posición** | ECDF = escalera acumulada; rango percentil = ECDF por cien; el **$z$ equivale al percentil solo bajo normalidad** (ingreso 4.0: 94.7 si fuera normal frente a 92.0 empírico) |
| **Datos agrupados** | Marca de clase + interpolación dentro de la clase aproximan media, mediana y moda; el error crece con el ancho de clase (0.024 en la media con clases de ancho 2) |
