## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **«para no ingenieros»** del **cuaderno estándar**. Una tendera anota sus ventas diarias en un cuaderno, pero **a algunas páginas les falta información**. La pregunta no es solo «¿cómo relleno el hueco?», sino **«¿por qué faltan esas páginas?»**, porque la razón decide si el promedio de lo que quedó es de fiar.

| ¿Por qué falta la página? | Ejemplo en la tienda | ¿El promedio de lo que quedó sirve? |
|---|---|---|
| **Por azar** | La lluvia mojó páginas sueltas | **Sí**, más o menos (solo hay menos datos) |
| **Por algo que yo sé** | Los domingos no se anota nada | **No**, pero se puede **corregir** usando lo que sé (el día) |
| **Por el propio dato** | No se anotan las ventas **malas** | **No**, y **no hay forma de corregirlo** solo con el cuaderno |

> 🔗 Si luego quieres ver todo **con rigor y desde cero** (mecanismos, imputación múltiple, interpolación): **cuaderno estándar**. Antes: **Conociendo la finca**. Después: **La sombra de la papa**.

## 1. Cuatro Semanas de Ventas y Tres Formas de Perder Páginas 📒

Inventamos las ventas de **28 días** (miles de pesos). Los **domingos** se vende mucho más. Conocemos **todas** las ventas, así que podemos comprobar qué tan bien funciona el promedio de lo que «sobrevive» en cada caso.

````python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
dias = np.arange(28)
es_dom = (dias % 7 == 6)                                              # 4 domingos (días 6, 13, 20 y 27)
ventas = np.round(100 + 100 * es_dom + rng.normal(0, 8, 28))           # miles de pesos (INVENTADAS)
verdad = ventas.mean()

f_lluvia = np.zeros(28, bool); f_lluvia[rng.choice(28, 8, replace=False)] = True      # 1) la lluvia moja 8 páginas AL AZAR
f_dom = es_dom.copy()                                                                 # 2) los domingos NO se anotan
f_malas = ventas < np.percentile(ventas, 30)                                          # 3) NO se anotan las ventas MALAS (el 30 % más bajo)

print(f"Promedio verdadero (con las 28 páginas): {verdad:.1f}")
resultados = {}
for nombre, f in [("lluvia (al azar)", f_lluvia), ("no anota domingos", f_dom), ("no anota ventas malas", f_malas)]:
    resultados[nombre] = ventas[~f].mean()
    print(f"  {nombre:24s} faltan {f.sum():2d} páginas -> promedio de lo anotado = {resultados[nombre]:.1f}  (error {resultados[nombre] - verdad:+.1f})")

assert abs(resultados["lluvia (al azar)"] - verdad) < 12                         # al azar: más o menos bien
assert resultados["no anota domingos"] < verdad - 10                              # sin los domingos buenos: se SUBESTIMA
assert resultados["no anota ventas malas"] > verdad + 8                           # sin las ventas malas: se SOBREESTIMA

fig, ax = plt.subplots(figsize=(9, 3))
ax.bar(dias, ventas, color=["#f59e0b" if d else "#2563eb" for d in es_dom])
for f, c, n in [(f_lluvia, "#6b7280", "lluvia")]:
    ax.scatter(dias[f], ventas[f] + 4, marker="v", color=c, label="página mojada (lluvia)")
ax.axhline(verdad, color="#16a34a", ls="--", label=f"promedio verdadero {verdad:.0f}")
ax.set_xlabel("día (barras naranjas = domingos)"); ax.set_ylabel("ventas (miles)"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
````

---
**Lección:** con la **lluvia** (azar) el promedio de lo anotado queda cerca de la verdad. Cuando **no se anotan los domingos**, el promedio **baja** (faltan justo los días buenos) y cuando **no se anotan las ventas malas**, **sube** (se queda solo lo bonito). El tercer caso es el peligroso: **la propia página explica por qué falta**, y el cuaderno, por sí solo, no puede delatarlo.

## 2. Rellenar con Quien Se Parece 🤝

Supón que **solo falta el primer domingo** (día 6) y quieres adivinar cuánto vendió. Hay dos formas de rellenar:

- **Con el promedio de todos los días anotados** (la forma «perezosa»).
- **Con el promedio de los otros domingos** (rellenar con quien **se parece**).

Comparamos cuál se acerca más a la venta real de ese día.

````python
dia = 6                                                            # el primer domingo
real = ventas[dia]
perezoso = np.delete(ventas, dia).mean()                           # promedio de los 27 días restantes
otros_domingos = ventas[es_dom & (dias != dia)].mean()             # promedio de los otros 3 domingos
print(f"Venta real del domingo {dia}: {real:.0f}")
print(f"  Rellenar con el promedio de todo:        {perezoso:.0f}  (error {abs(perezoso - real):.0f})")
print(f"  Rellenar con el promedio de los domingos: {otros_domingos:.0f}  (error {abs(otros_domingos - real):.0f})")
assert abs(otros_domingos - real) < abs(perezoso - real) / 3
````

---
**Lección:** el promedio de todo (110) se equivoca por **100 mil pesos** porque mezcla días normales con domingos; el de los **otros domingos** (197) se equivoca por solo **13 mil**. **Rellenar bien es buscar a quien se parece al dato que falta** (misma tienda, mismo día de la semana, misma zona). Y cuidado: ningún relleno arregla el caso en que **faltan justo las ventas malas**.

## Resumen en una frase 🎯

> **Antes de rellenar un hueco, pregunta por qué falta: si falta por azar casi no pasa nada, si falta por algo que conoces se puede corregir con quien se parece, y si falta por el propio dato, ningún relleno lo arregla.**

### Lo que aprendiste hoy:
- ✅ Hay **tres razones** de falta: **azar**, **algo que yo sé** y **el propio dato**; la razón decide si el promedio de lo anotado sirve.
- ✅ Sin los domingos el promedio **bajó** y sin las ventas malas **subió**: el sesgo tiene dirección.
- ✅ Rellenar con **quien se parece** (los otros domingos) acierta mucho más que con el promedio de todo.
- ✅ **Rellenar con el promedio no cambia el promedio**: no arregla un sesgo, solo lo disimula.
