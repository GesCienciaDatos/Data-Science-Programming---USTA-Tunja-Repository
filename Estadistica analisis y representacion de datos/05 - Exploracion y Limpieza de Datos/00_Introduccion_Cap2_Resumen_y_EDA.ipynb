## 2. Los Datos: Fincas Paperas **Sintéticas** 🥔

Trabajaremos con **600 fincas ficticias** generadas con semilla fija (`default_rng(42)`): zona, altitud, precipitación, área, fertilizante, variedad, fecha de siembra y **rendimiento** (t/ha). El generador es **multiplicativo** —$\ln(\text{rend})$ depende de $\ln(1+\text{fert}/100)$ y de una **parábola** en la altitud— y por eso conocemos la «verdad»: podemos comprobar si cada técnica la recupera. **Ninguna cifra describe cultivos reales.** Dejamos a propósito 12 valores faltantes en el fertilizante (tema del [cuaderno 01](01_Datos_Faltantes_e_Inconsistentes.ipynb)).

Lo primero siempre es el **perfil de datos**: ¿cuántas filas?, ¿qué tipo tiene cada columna?, ¿cuántos valores únicos?, ¿qué porcentaje falta?, ¿hay filas duplicadas?

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})


def generar_fincas(n=600, semilla=42):
    """Fincas paperas SINTÉTICAS (hipotéticas). El rendimiento es multiplicativo:
    log(rend) = a + b·log(1 + fert/100) − ((altitud − 2800)/400)² + efecto variedad + ruido."""
    rng = np.random.default_rng(semilla)
    zonas = np.array(["Zona A", "Zona B", "Zona C", "Zona D", "Zona E", "Zona F"])
    alt_zona = rng.uniform(2500, 3100, size=zonas.size)                     # altitud media inventada de cada zona
    z = rng.integers(0, zonas.size, size=n)
    altitud = alt_zona[z] + rng.normal(0, 60, n)
    precip = 700 + 0.4 * (altitud - 2800) + rng.normal(0, 90, n)            # mm/año, inventado
    area = rng.lognormal(mean=0.3, sigma=0.8, size=n)                       # asimétrica a la derecha
    fert = rng.gamma(shape=4, scale=60, size=n)                             # kg/ha
    variedad = rng.choice(["Pastusa", "Criolla", "Capiro"], size=n, p=[0.5, 0.2, 0.3])
    efecto = pd.Series(variedad).map({"Pastusa": 0.0, "Criolla": -0.25, "Capiro": 0.10}).to_numpy()
    log_rend = 1.6 + 0.45 * np.log1p(fert / 100) - 1.0 * ((altitud - 2800) / 400) ** 2 + efecto + rng.normal(0, 0.18, n)
    fecha = pd.Timestamp("2022-01-01") + pd.to_timedelta(rng.integers(0, 730, size=n), unit="D")
    d = pd.DataFrame({"zona": zonas[z], "altitud_m": altitud, "precip_mm": precip, "area_ha": area,
                      "fert_kg_ha": fert, "variedad": variedad, "fecha_siembra": fecha, "rend_t_ha": np.exp(log_rend)})
    d.loc[rng.choice(n, size=12, replace=False), "fert_kg_ha"] = np.nan   # 12 faltantes al azar (el tema del cuaderno 01)
    return d


fincas = generar_fincas()


def perfil(d):
    """Perfil de datos: tipo, cardinalidad, % de faltantes y un ejemplo por columna."""
    return pd.DataFrame({"tipo": d.dtypes.astype(str), "n_unicos": d.nunique(),
                         "pct_faltante": (d.isna().mean() * 100).round(1), "ejemplo": d.iloc[0]})


print(fincas.shape, "| filas duplicadas:", int(fincas.duplicated().sum()))
print(perfil(fincas).to_string())
assert fincas.shape == (600, 8) and fincas.fert_kg_ha.isna().sum() == 12 and fincas.duplicated().sum() == 0
assert (fincas.rend_t_ha > 0).all() and fincas.zona.nunique() == 6
````

---
**Lectura del perfil:** 600 filas y 8 columnas; `zona` (6) y `variedad` (3) son **categóricas** de baja cardinalidad; `fecha_siembra` es un `datetime64` (de ella saldrá el mes); `fert_kg_ha` tiene **2 % de faltantes**; no hay duplicados. Un perfil así detecta de inmediato tipos mal leídos (un número guardado como texto), cardinalidades sospechosas (una «categoría» con 600 valores únicos es un identificador) y columnas casi vacías.

## 3. Resumen Estadístico con Rigor 📏 (Sección 2.3.1)

Para cada variable numérica reportamos cuatro familias de medidas (desarrolladas en el [Módulo 01](../01%20-%20Estadistica%20Descriptiva%20y%20Visualizacion/00_Medidas_de_Tendencia_Central_y_Posicion.ipynb)):

| Familia | Medidas | Fórmula / idea | Sensible a atípicos |
|---|---|---|---|
| **Posición** | media $\bar{x}$, mediana | $\bar{x}=\frac{1}{n}\sum x_i$ | media **sí**, mediana **no** |
| **Escala** | desviación típica $s$, rango intercuartílico (IQR), MAD | $s=\sqrt{\frac{1}{n-1}\sum (x_i-\bar{x})^2}$; $\text{MAD}=\text{mediana}\lvert x_i-\tilde{x}\rvert$ | $s$ **sí**; IQR y MAD **no** |
| **Forma** | asimetría $g_1$, curtosis en exceso $g_2$ | $g_1=\frac{m_3}{m_2^{3/2}}$, $g_2=\frac{m_4}{m_2^{2}}-3$ con $m_k=\frac{1}{n}\sum (x_i-\bar{x})^k$ | ambas **sí** |
| **Incertidumbre** | IC del 95 % para la media | $\bar{x}\pm t_{n-1,\,0.975}\,\dfrac{s}{\sqrt{n}}$ | sí |

La varianza muestral divide por $n-1$ (corrección de Bessel), no por $n$; `pandas` lo hace así por defecto y `NumPy` **no** (`ddof=0`). Para que el MAD sea comparable con $s$ en datos normales se multiplica por $1/\Phi^{-1}(0.75)\approx 1.4826$. El IC con la $t$ de Student supone que $\bar{x}$ es aproximadamente normal (TLC; cuaderno 03 del [Módulo 02](../02%20-%20Probabilidad%20y%20Distribuciones/03_Muestreo_Ley_de_los_Grandes_Numeros_y_TLC.ipynb)), algo dudoso con $n$ pequeño y datos muy asimétricos. Lo implementamos **desde cero** y lo verificamos contra `pandas` y `SciPy`.

````python
def resumen(x):
    """Resumen estadístico desde cero (sin pandas.describe): posición, escala, forma e IC t del 95 % para la media."""
    x = np.asarray(x, float)
    x = x[~np.isnan(x)]
    n = x.size
    m = x.sum() / n
    s = np.sqrt(((x - m) ** 2).sum() / (n - 1))                       # varianza MUESTRAL: divide por n-1
    q1, med, q3 = np.percentile(x, [25, 50, 75])
    mad = np.median(np.abs(x - med))
    m2, m3, m4 = [((x - m) ** k).mean() for k in (2, 3, 4)]
    se = s / np.sqrt(n)
    tcrit = stats.t.ppf(0.975, n - 1)
    return {"n": n, "media": m, "mediana": med, "sd": s, "cv": s / m, "iqr": q3 - q1,
            "MAD_normal": mad / stats.norm.ppf(0.75), "asimetria": m3 / m2 ** 1.5, "curtosis_exc": m4 / m2 ** 2 - 3,
            "ic95_inf": m - tcrit * se, "ic95_sup": m + tcrit * se}


cols = ["altitud_m", "precip_mm", "area_ha", "fert_kg_ha", "rend_t_ha"]
tabla = pd.DataFrame({c: resumen(fincas[c]) for c in cols}).T
print(tabla.round(3).to_string())

# Verificación contra pandas y SciPy
for c in cols:
    x = fincas[c].dropna()
    r = resumen(x)
    assert np.isclose(r["sd"], x.std(ddof=1)) and np.isclose(r["mediana"], x.median())
    assert np.isclose(r["asimetria"], stats.skew(x)) and np.isclose(r["curtosis_exc"], stats.kurtosis(x))
    assert np.isclose(r["MAD_normal"], stats.median_abs_deviation(x, scale="normal"))
    lo, hi = stats.t.interval(0.95, len(x) - 1, loc=x.mean(), scale=stats.sem(x))
    assert np.isclose(r["ic95_inf"], lo) and np.isclose(r["ic95_sup"], hi)
print("\nResumen desde cero == pandas/SciPy en las 5 variables.")

# Media vs mediana según la asimetría: cuanto más asimétrica, más se separan
for c in ["altitud_m", "area_ha", "rend_t_ha"]:
    t = tabla.loc[c]
    print(f"{c:10s} asimetría={t.asimetria:5.2f}  (media − mediana)/sd = {(t.media - t.mediana) / t.sd:5.2f}")
assert tabla.loc["area_ha", "asimetria"] > 1.5 > abs(tabla.loc["altitud_m", "asimetria"])
````

---
**Interpretación:**

- **`area_ha`** es el caso extremo: asimetría **3.38** y curtosis en exceso **18.1** (colas muy pesadas), CV de 0.94 y una media (1.65 ha) bastante mayor que la mediana (1.25 ha). Para describir «una finca típica» sirve la **mediana** y el **MAD**, no la media y $s$.
- **`altitud_m`** tiene asimetría **negativa** (−0.61) y curtosis en exceso −0.69 (más plana que una normal) porque **mezcla seis zonas** con altitudes medias distintas: una medida de forma no revela esa estructura de grupos; un histograma por zona sí.
- **`rend_t_ha`** (media 6.73, mediana 6.46) es asimétrica a la derecha (0.76) porque el generador es multiplicativo.
- El **IC95 de la media** es estrecho (± 0.19 t/ha) porque $n=600$; con $n=15$ sería $\sqrt{40}\approx 6$ veces más ancho. **Un IC dice dónde está la media poblacional, no dónde caen las fincas individuales** (para eso sirven los percentiles).

## 4. EDA Visual: Mirar Antes de Modelar 👀 (Sección 2.3.1)

Los números resumen; los **gráficos revelan estructura** que un resumen aplasta. El panel siguiente combina las herramientas básicas de Tukey: **histograma + densidad** (forma), **caja** (comparación entre grupos), **dispersión con tendencia por deciles** (relaciones) y **mapa de correlaciones**. La matriz de correlación de Pearson, $r_{jk}=\frac{\sum (x_{ij}-\bar{x}_j)(x_{ik}-\bar{x}_k)}{(n-1)s_js_k}$, se calcula desde cero y se verifica contra `np.corrcoef`; la de **Spearman** es la de Pearson aplicada a los **rangos** (mide asociación *monótona*).

````python
def corr_pearson(M):
    """Matriz de correlación de Pearson desde cero: covarianza estandarizada."""
    Z = (M - M.mean(0)) / M.std(0, ddof=1)
    return Z.T @ Z / (len(M) - 1)


num = fincas[["altitud_m", "precip_mm", "area_ha", "fert_kg_ha", "rend_t_ha"]].dropna()
M = num.to_numpy()
R = corr_pearson(M)
Rs = corr_pearson(num.rank().to_numpy())                       # Spearman = Pearson de los rangos
assert np.allclose(R, np.corrcoef(M.T)) and np.allclose(Rs, stats.spearmanr(M).statistic)

fig, ax = plt.subplots(2, 3, figsize=(15, 8))
x = fincas.rend_t_ha
ax[0, 0].hist(x, bins=30, density=True, color="#93c5fd", edgecolor="white")
gx = np.linspace(x.min(), x.max(), 200)
ax[0, 0].plot(gx, stats.gaussian_kde(x)(gx), color="#1e3a8a", lw=2)
ax[0, 0].axvline(x.mean(), color="#dc2626", ls="--", label=f"media {x.mean():.2f}")
ax[0, 0].axvline(x.median(), color="#16a34a", ls="--", label=f"mediana {x.median():.2f}")
ax[0, 0].set(title="Rendimiento (t/ha): cola a la derecha", xlabel="t/ha"); ax[0, 0].legend(fontsize=8)

ax[0, 1].hist(np.log(fincas.area_ha), bins=30, color="#fcd34d", edgecolor="white")
ax[0, 1].set(title="log(área): casi simétrica (el área cruda tiene asimetría 3.4)", xlabel="log(ha)")

orden = ["Criolla", "Pastusa", "Capiro"]
ax[0, 2].boxplot([fincas.rend_t_ha[fincas.variedad == v] for v in orden], tick_labels=orden, showmeans=True)
ax[0, 2].set(title="Rendimiento por variedad", ylabel="t/ha")

f = fincas.dropna(subset=["fert_kg_ha"])
ax[1, 0].scatter(f.fert_kg_ha, f.rend_t_ha, s=8, alpha=0.4, color="#2563eb")
cortes = pd.qcut(f.fert_kg_ha, 10)
med = f.groupby(cortes, observed=True).agg(fx=("fert_kg_ha", "median"), ry=("rend_t_ha", "median"))
ax[1, 0].plot(med.fx, med.ry, "o-", color="#dc2626", label="mediana por decil")
ax[1, 0].set(title="Fertilizante vs rendimiento (efecto con saturación suave)", xlabel="kg/ha", ylabel="t/ha"); ax[1, 0].legend(fontsize=8)

ax[1, 1].scatter(fincas.altitud_m, fincas.rend_t_ha, s=8, alpha=0.4, color="#2563eb")
cortes = pd.qcut(fincas.altitud_m, 10)
med = fincas.groupby(cortes, observed=True).agg(ax_=("altitud_m", "median"), ry=("rend_t_ha", "median"))
ax[1, 1].plot(med.ax_, med.ry, "o-", color="#dc2626", label="mediana por decil")
ax[1, 1].set(title="Altitud vs rendimiento: relación en U invertida", xlabel="m", ylabel="t/ha"); ax[1, 1].legend(fontsize=8)

im = ax[1, 2].imshow(R, cmap="RdBu_r", vmin=-1, vmax=1)
nombres = ["altitud", "precip", "área", "fert", "rend"]
ax[1, 2].set_xticks(range(5), nombres, rotation=45); ax[1, 2].set_yticks(range(5), nombres); ax[1, 2].grid(False)
for i in range(5):
    for j in range(5):
        ax[1, 2].text(j, i, f"{R[i, j]:.2f}", ha="center", va="center", fontsize=8, color="white" if abs(R[i, j]) > 0.6 else "black")
ax[1, 2].set_title("Correlación de Pearson")
plt.tight_layout(); plt.show()

# r lineal vs. R² de un ajuste cuadrático: el coeficiente de Pearson no "ve" la U invertida
alt, rend = fincas.altitud_m.to_numpy(), fincas.rend_t_ha.to_numpy()
r2_lin = np.corrcoef(alt, rend)[0, 1] ** 2
res = rend - np.polyval(np.polyfit(alt, rend, 2), alt)
r2_cuad = 1 - res.var() / rend.var()
print(f"altitud–rendimiento: Pearson r = {np.corrcoef(alt, rend)[0, 1]:+.3f} (r² = {r2_lin:.3f}) | R² del ajuste cuadrático = {r2_cuad:.3f}")
print(f"fert–rendimiento:    Pearson = {R[3, 4]:+.3f} | Spearman = {Rs[3, 4]:+.3f}  (casi iguales: aquí la relación es casi lineal)")
assert r2_cuad > 3 * r2_lin and abs(R[3, 4] - Rs[3, 4]) < 0.05
````

### La trampa de explorar demasiado: comparaciones múltiples 🎰

Con $p$ variables hay $m=\binom{p}{2}$ parejas. Si todas las variables fueran **independientes** y probáramos cada correlación con $\alpha=0.05$, esperaríamos $m\alpha$ falsos «hallazgos». La **corrección de Bonferroni** exige $p_i<\alpha/m$ y controla la probabilidad de **al menos un** falso positivo. Lo comprobamos simulando **puro ruido** ($n=100$, $p=20$, $m=190$) 500 veces.

````python
# ¿Cuántas "correlaciones significativas" aparecen en datos de PURO RUIDO al mirar todas las parejas?
def n_significativas(n=100, p=20, alpha=0.05, rng=None):
    Z = rng.normal(size=(n, p))
    R_ = np.corrcoef(Z.T)[np.triu_indices(p, 1)]                  # 190 correlaciones
    t = R_ * np.sqrt((n - 2) / (1 - R_ ** 2))
    pv = 2 * stats.t.sf(np.abs(t), n - 2)
    return (pv < alpha).sum(), (pv < alpha / pv.size).sum(), pv.size


rng = np.random.default_rng(7)
sim = np.array([n_significativas(rng=rng)[:2] for _ in range(500)])
m_ = n_significativas(rng=rng)[2]
print(f"Parejas por revisar: {m_} | 'hallazgos' por azar con α=0.05: {sim[:, 0].mean():.1f} en promedio (esperado {0.05 * m_:.1f})")
print(f"Con Bonferroni (α/{m_}): {sim[:, 1].mean():.3f} en promedio (esperado ≤ 0.05)")
print(f"Proporción de simulaciones con al menos 1 'hallazgo' sin corregir: {(sim[:, 0] > 0).mean():.3f}")
assert 8.5 < sim[:, 0].mean() < 10.5 and sim[:, 1].mean() < 0.08 and (sim[:, 0] > 0).mean() > 0.99
````

---
**Lectura:** en datos sin ninguna relación real aparecen **≈ 9.6 correlaciones «significativas» por tabla** (esperado: $190\times 0.05=9.5$) y **prácticamente siempre hay al menos una**. Con Bonferroni baja a ≈ 0.04. Por eso un EDA con muchas variables:

- se reporta como **generador de hipótesis**, no como evidencia;
- se confirma en **datos nuevos** (reservar una muestra de validación *antes* de explorar) o con procedimientos que controlan el error (Bonferroni, Holm, tasa de falsos descubrimientos; Módulo 03);
- exige reportar **cuántas cosas se miraron**, no solo la que salió bien.

## 5. Ingeniería de Características: Transformar para Cumplir Supuestos 🔧 (Sección 2.3.1)

Muchas técnicas (regresión lineal, PCA, $k$-vecinos, distancias en general) funcionan mejor cuando las variables tienen **escalas comparables** y **distribuciones no muy asimétricas**.

### 5.1 Transformaciones de potencia (Box-Cox)

Para $x>0$, la familia de Box-Cox es $y(\lambda)=\dfrac{x^{\lambda}-1}{\lambda}$ si $\lambda\neq0$ y $y(0)=\ln x$ (así $\lambda=1$ no transforma, $\lambda=\tfrac12$ es una raíz y $\lambda=0$ es el logaritmo). El $\lambda$ se estima por **máxima verosimilitud** suponiendo que $y(\lambda)$ es normal: la log-verosimilitud perfilada es $\ell(\lambda)=-\frac{n}{2}\ln\hat{\sigma}^{2}_{y}(\lambda)+(\lambda-1)\sum_i\ln x_i$, donde el segundo término es el jacobiano de la transformación. Un **IC95 por razón de verosimilitud** es $\{\lambda:\ \ell(\hat\lambda)-\ell(\lambda)\le\tfrac12\chi^2_{1,\,0.95}\}$; si contiene a 0, el logaritmo es una elección defendible. Para datos con ceros o negativos existe la variante de Yeo-Johnson (`PowerTransformer`).

### 5.2 Escalamientos

| Escalador | Fórmula | Cuándo | Debilidad |
|---|---|---|---|
| **z-score** | $z=\dfrac{x-\bar{x}}{s}$ | distancias, regresión regularizada, PCA | $\bar{x}$ y $s$ los arrastran los atípicos |
| **min-max** | $\dfrac{x-x_{\min}}{x_{\max}-x_{\min}}$ | redes neuronales, rangos acotados | **un solo** extremo comprime a todos los demás |
| **robusto** | $\dfrac{x-\tilde{x}}{Q_3-Q_1}$ | datos con atípicos | no deja el rango en $[0,1]$ |

(`StandardScaler` usa $\sigma$ poblacional, divide por $n$; por eso la comparación con scikit-learn se hace con `x.std()`.) Inyectamos **un** valor absurdo (25 000 kg/ha de fertilizante, claramente un error de captura) para medir cuánto del rango ocupa el grueso de los datos (ancho entre los percentiles 5 y 95).

````python
from scipy.optimize import minimize_scalar
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, PowerTransformer


def boxcox(x, lmb):
    return np.log(x) if abs(lmb) < 1e-12 else (x ** lmb - 1) / lmb


def loglik_boxcox(lmb, x):
    """Log-verosimilitud perfilada de Box-Cox: −(n/2)·ln σ̂²(λ) + (λ−1)·Σ ln x."""
    y = boxcox(x, lmb)
    return -0.5 * x.size * np.log(y.var()) + (lmb - 1) * np.log(x).sum()


def lambda_mv(x):
    """λ de máxima verosimilitud: malla gruesa y refinamiento."""
    malla = np.linspace(-3, 3, 121)
    l0 = malla[np.argmax([loglik_boxcox(l, x) for l in malla])]
    return minimize_scalar(lambda l: -loglik_boxcox(l, x), bounds=(l0 - 0.1, l0 + 0.1), method="bounded", options={"xatol": 1e-8}).x


area = fincas.area_ha.to_numpy()
lam = lambda_mv(area)
lam_scipy = stats.boxcox(area)[1]
print(f"λ̂ (desde cero) = {lam:.4f} | λ̂ (scipy.stats.boxcox) = {lam_scipy:.4f}")
assert abs(lam - lam_scipy) < 1e-3
for nombre, v in [("área cruda", area), ("raíz cuadrada", np.sqrt(area)), ("log", np.log(area)), (f"Box-Cox(λ={lam:.2f})", boxcox(area, lam))]:
    print(f"  {nombre:20s} asimetría = {stats.skew(v):+.3f}")
assert abs(stats.skew(boxcox(area, lam))) < 0.15 < stats.skew(area)
# Interpretación honesta: el área se generó lognormal, así que λ̂ ≈ 0 (el log) es lo esperable; el IC de λ̂ lo contiene:
corte = loglik_boxcox(lam, area) - stats.chi2.ppf(0.95, 1) / 2                      # IC95 por razón de verosimilitud
grid = np.linspace(-0.6, 0.6, 241)
ic = grid[[loglik_boxcox(l, area) >= corte for l in grid]]
print(f"IC95 aprox. de λ (razón de verosimilitud): [{ic.min():.2f}, {ic.max():.2f}]  ->  ¿contiene 0 (log)?  {ic.min() <= 0 <= ic.max()}")
assert ic.min() <= 0 <= ic.max()

# --- Escaladores: fórmulas explícitas contra scikit-learn ---
x = fincas.fert_kg_ha.dropna().to_numpy()
z = (x - x.mean()) / x.std()                                               # z-score (σ poblacional, como sklearn)
mm = (x - x.min()) / (x.max() - x.min())
q1, q2, q3 = np.percentile(x, [25, 50, 75])
rb = (x - q2) / (q3 - q1)
assert np.allclose(z, StandardScaler().fit_transform(x[:, None]).ravel())
assert np.allclose(mm, MinMaxScaler().fit_transform(x[:, None]).ravel())
assert np.allclose(rb, RobustScaler().fit_transform(x[:, None]).ravel())

# Un solo valor absurdo (error de captura: 25 000 kg/ha) arruina min-max y z-score, pero casi no toca al robusto
x_sucio = np.append(x, 25000.0)
def amplitud_del_grueso(esc):                                              # ancho del rango P5-P95 del escalado
    p5, p95 = np.percentile(esc, [5, 95])
    return p95 - p5
for nombre, f_ in [("min-max", lambda v: (v - v.min()) / (v.max() - v.min())), ("z-score", lambda v: (v - v.mean()) / v.std()),
                   ("robusto", lambda v: (v - np.median(v)) / np.subtract(*np.percentile(v, [75, 25])))]:
    print(f"  {nombre:8s} ancho de P5–P95: limpio = {amplitud_del_grueso(f_(x)):.3f} | con 1 valor absurdo = {amplitud_del_grueso(f_(x_sucio)):.3f}")
assert amplitud_del_grueso((x_sucio - x_sucio.min()) / np.ptp(x_sucio)) < 0.05
assert abs(amplitud_del_grueso((x_sucio - np.median(x_sucio)) / np.subtract(*np.percentile(x_sucio, [75, 25]))) - amplitud_del_grueso(rb)) < 0.05
````

---
**Resultados:**

- El $\hat\lambda$ desde cero (0.0113) coincide con el de `scipy.stats.boxcox` al cuarto decimal. Como el área se generó **lognormal**, esperábamos $\lambda\approx0$: el IC95 es $[-0.07,\,0.09]$ y **contiene al 0**, o sea que el logaritmo es una elección tan buena como Box-Cox. La raíz cuadrada ($\lambda=\tfrac12$) **no** basta (asimetría 1.33).
- Con **un** valor absurdo, el ancho P5–P95 en min-max pasa de 0.52 a **0.02** (todo se aplasta contra 0) y en z-score de 3.1 a 0.4; el **escalador robusto no cambia** (2.41 → 2.46). Moraleja: *revisa los atípicos antes de escalar* (Módulo 07, cuaderno 05).

### 5.3 Fechas cíclicas y discretización · 5.4 ¿Realmente ayuda? ⚖️

- **Fechas cíclicas.** Codificar el mes como número 1–12 dice que diciembre está a 11 unidades de enero. Mapearlo al círculo $(\sin\theta,\cos\theta)$ con $\theta=2\pi(m-1)/12$ pone diciembre **tan cerca de enero como enero de febrero** (0.518 en ambos casos).
- **Discretización por cuantiles** (`pd.qcut`) crea grupos de igual tamaño (150 fincas por cuartil); sirve para tablas y relaciones muy irregulares, pero **descarta información** dentro de cada intervalo.
- **Evaluación honesta.** Una transformación «se justifica» solo si mejora el desempeño **fuera de muestra**. Comparamos cinco configuraciones con **validación cruzada 5×3** (15 pliegues), con *todo* el preprocesamiento dentro del `Pipeline` (imputación por mediana, `log1p`, polinomio de la altitud estandarizada, *one-hot* de la variedad) para **evitar fugas**. A: modelo lineal con variables crudas; B: + transformaciones (log del fertilizante, parábola de la altitud); C: B + objetivo en log (se predice $\ln(\text{rend})$ y se mide $R^2$ en la **escala original**); D: C + mes cíclico, que en el generador **no tiene efecto**; E: bosque aleatorio sobre variables crudas.

````python
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer, PolynomialFeatures
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RepeatedKFold, cross_validate

# --- Fechas: el mes como círculo (diciembre está al lado de enero) ---
mes = np.arange(1, 13)
ang = 2 * np.pi * (mes - 1) / 12
pts = np.c_[np.sin(ang), np.cos(ang)]
d_dic_ene = np.linalg.norm(pts[11] - pts[0])
d_ene_feb = np.linalg.norm(pts[0] - pts[1])
print(f"Distancia dic–ene: número crudo = {abs(12 - 1)} | círculo = {d_dic_ene:.3f} (igual que ene–feb = {d_ene_feb:.3f})")
assert np.isclose(d_dic_ene, d_ene_feb) and abs(12 - 1) > abs(2 - 1)

# --- Discretización por cuantiles: grupos de igual tamaño ---
cuartiles = pd.qcut(fincas.area_ha, 4, labels=["Q1", "Q2", "Q3", "Q4"])
print("Fincas por cuartil de área:", cuartiles.value_counts().sort_index().to_dict())
assert cuartiles.value_counts().between(149, 151).all()

# --- ¿Sirve la ingeniería de características? Validación cruzada (todo DENTRO del pipeline: sin fuga) ---
d = fincas.copy()
d["mes"] = d.fecha_siembra.dt.month
y = d.rend_t_ha.to_numpy()
X = d[["altitud_m", "precip_mm", "area_ha", "fert_kg_ha", "variedad", "mes"]]
log1p100 = FunctionTransformer(lambda a: np.log1p(a / 100), feature_names_out="one-to-one")
ciclo = FunctionTransformer(lambda a: np.c_[np.sin(2 * np.pi * (a - 1) / 12), np.cos(2 * np.pi * (a - 1) / 12)])
imp_med = SimpleImputer(strategy="median")
cat = ("cat", OneHotEncoder(drop="first"), ["variedad"])


def construir(tipo):
    if tipo == "A crudo":
        ct = ColumnTransformer([("num", imp_med, ["altitud_m", "precip_mm", "area_ha", "fert_kg_ha"]), cat])
        return Pipeline([("ct", ct), ("m", LinearRegression())])
    if tipo == "E bosque (crudo)":
        ct = ColumnTransformer([("num", imp_med, ["altitud_m", "precip_mm", "area_ha", "fert_kg_ha"]), cat])
        return Pipeline([("ct", ct), ("m", RandomForestRegressor(n_estimators=150, min_samples_leaf=3, random_state=0))])
    transf = [("fert", make_pipeline(imp_med, log1p100), ["fert_kg_ha"]),
              ("alt", make_pipeline(StandardScaler(), PolynomialFeatures(2, include_bias=False)), ["altitud_m"]),
              ("otras", "passthrough", ["precip_mm", "area_ha"]), cat]
    if tipo.startswith("D"):
        transf.append(("mes", ciclo, ["mes"]))
    ct = ColumnTransformer(transf)
    base = Pipeline([("ct", ct), ("m", LinearRegression())])
    if tipo.startswith("B"):
        return base
    return TransformedTargetRegressor(regressor=base, func=np.log, inverse_func=np.exp)    # objetivo en log, R² en la escala original


cv = RepeatedKFold(n_splits=5, n_repeats=3, random_state=0)
tipos = ["A crudo", "B + transformaciones", "C + objetivo en log", "D + mes cíclico (sin efecto real)", "E bosque (crudo)"]
res = {t: cross_validate(construir(t), X, y, cv=cv, scoring=("r2", "neg_root_mean_squared_error")) for t in tipos}
r2 = {t: res[t]["test_r2"] for t in tipos}
tab = pd.DataFrame({"R² medio": {t: r2[t].mean() for t in tipos}, "sd entre pliegues": {t: r2[t].std() for t in tipos},
                    "RMSE medio (t/ha)": {t: -res[t]["test_neg_root_mean_squared_error"].mean() for t in tipos}})
print(tab.round(3).to_string())
dif = r2["B + transformaciones"] - r2["A crudo"]
print(f"\nΔR² (B − A) por pliegue: media {dif.mean():+.3f}, positivo en {(dif > 0).sum()}/{dif.size} pliegues")
dif_d = r2["D + mes cíclico (sin efecto real)"] - r2["C + objetivo en log"]
print(f"ΔR² (D − C): media {dif_d.mean():+.4f}, positivo en {(dif_d > 0).sum()}/{dif_d.size} pliegues  (ruido: el mes no aporta)")
assert r2["B + transformaciones"].mean() > r2["A crudo"].mean() + 0.1 and (dif > 0).all()
assert abs(dif_d.mean()) < 0.02
````

---
**Lectura honesta:**

- **B vs A:** pasar de $R^2=0.36$ a $0.67$ (**+0.31 en los 15 pliegues**) porque el modelo lineal con variables crudas **no puede** representar ni la saturación del fertilizante ni la U de la altitud. La mejora es enorme y consistente.
- **C vs B:** predecir el logaritmo sube a $0.70$ y baja el RMSE de 1.38 a 1.31 t/ha: el ruido es multiplicativo, así que el log lo «aplana».
- **D vs C:** el mes cíclico **no mejora** (Δ$R^2=-0.001$, positivo en 8 de 15 pliegues): agregar variables sin señal solo añade ruido. *No todo lo que se puede construir ayuda.*
- **E:** el bosque con datos **crudos** logra $0.62$: captura parte de la no linealidad sin ingeniería, pero un modelo lineal **bien transformado** (C) lo supera en este generador, que tiene estructura log-lineal. Con otras estructuras el orden podría invertirse: la conclusión depende de los datos.
- **Cuidado inferencial:** los 15 pliegues **no son independientes** (comparten observaciones); la desviación entre pliegues es orientativa y subestima la incertidumbre real de la comparación (Nadeau & Bengio, 2003).

```python
print(f"Diferencia máxima entre escalar sin y con fuga: {dif_max:.4f}")
assert dif_max > 0
print("Con un escalador robusto la fuga es mínima, pero el principio es el mismo que en la imputación, la selección de variables y el 'target encoding': cuanto más usa la transformación la etiqueta o muchos parámetros, mayor el sesgo optimista (lo medimos en el cuaderno 02).")
```
