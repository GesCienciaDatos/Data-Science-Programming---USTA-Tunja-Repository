## 1. La Maldición de la Dimensionalidad 🌌

Más variables **no** es siempre mejor. Dos fenómenos independientes lo muestran:

1. **Concentración de distancias.** En el cubo unitario $[0,1]^d$ los puntos «cercanos» y «lejanos» se vuelven casi indistinguibles cuando $d$ crece: el **contraste relativo** $(d_{\max}-d_{\min})/d_{\min}$ tiende a 0. Toda técnica basada en distancias (KNN, clustering, t-SNE) sufre.
2. **Sobreajuste por exceso de variables.** Con $n$ fijo, agregar variables **aunque sean ruido** hace que el $R^2$ de **entrenamiento** suba hacia 1 (con $p\to n$ una regresión interpola los datos), mientras el $R^2$ en **validación cruzada** cae e incluso se vuelve **negativo** (peor que predecir el promedio).

Lo comprobamos con 500 puntos uniformes (distancias al primer punto) y con 60 observaciones en las que **solo 3 variables importan**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import cross_val_score, KFold

plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})

# --- (1) Concentración de distancias: en alta dimensión "cercano" y "lejano" se parecen ---
rng = np.random.default_rng(0)
filas = []
for d in (2, 10, 100, 1000):
    P = rng.random((500, d))                                   # 500 puntos uniformes en el cubo [0,1]^d
    dist = np.linalg.norm(P[1:] - P[0], axis=1)                # distancias desde el primer punto a los demás
    filas.append({"dimensión": d, "dist. mínima": dist.min(), "dist. máxima": dist.max(),
                  "contraste (máx − mín)/mín": (dist.max() - dist.min()) / dist.min()})
conc = pd.DataFrame(filas).set_index("dimensión")
print(conc.round(3).to_string())
assert conc["contraste (máx − mín)/mín"].is_monotonic_decreasing and conc.loc[1000, "contraste (máx − mín)/mín"] < 0.35

# --- (2) Más variables que información: el R² de entrenamiento sube aunque sean puro ruido ---
n = 60
x_info = rng.normal(size=(n, 3))
y = x_info @ np.array([2.0, -1.0, 1.5]) + rng.normal(0, 1.5, n)         # solo 3 variables importan
res = []
for p_ruido in (0, 7, 27, 50):
    Xp = np.c_[x_info, rng.normal(size=(n, p_ruido))]
    r2_tr = LinearRegression().fit(Xp, y).score(Xp, y)
    r2_cv = cross_val_score(LinearRegression(), Xp, y, cv=KFold(5, shuffle=True, random_state=0), scoring="r2").mean()
    res.append({"variables de ruido": p_ruido, "p total": 3 + p_ruido, "R² entrenamiento": r2_tr, "R² validación cruzada": r2_cv})
sob = pd.DataFrame(res).set_index("variables de ruido")
print(sob.round(3).to_string())
assert sob["R² entrenamiento"].is_monotonic_increasing and sob.loc[50, "R² entrenamiento"] > 0.9
assert sob.loc[0, "R² validación cruzada"] > 0.5 > sob.loc[50, "R² validación cruzada"] and sob.loc[50, "R² validación cruzada"] < 0

fig, ax = plt.subplots(1, 2, figsize=(11, 3.6))
ax[0].plot(conc.index, conc.iloc[:, 2], "o-", color="#2563eb"); ax[0].set(xscale="log", title="Concentración de distancias", xlabel="dimensión d", ylabel="(máx − mín) / mín")
ax[1].plot(sob["p total"], sob["R² entrenamiento"], "o-", color="#dc2626", label="entrenamiento")
ax[1].plot(sob["p total"], sob["R² validación cruzada"], "o-", color="#16a34a", label="validación cruzada")
ax[1].axhline(0, color="gray", lw=0.8); ax[1].set(title="Agregar ruido: el ajuste sube, la predicción cae", xlabel="número de variables p (n = 60)", ylabel="R²"); ax[1].legend()
plt.tight_layout(); plt.show()
````

---
**Lectura:**

- El contraste de distancias pasa de **66.1** en $d=2$ a **0.36** en $d=100$ y **0.11** en $d=1000$: en 1 000 dimensiones el punto más lejano está apenas ≈ 11 % más lejos que el más cercano.
- Con $n=60$: al pasar de 3 a 53 variables el $R^2$ de entrenamiento sube de **0.83 a 0.99**, pero el de validación cruzada baja de **0.76 a −0.87**. *Un ajuste excelente en entrenamiento no es evidencia de nada.*

## 2. Selección de Características: Filtro, Envoltura e Incrustada 🔎 (Sección 2.3.3)

| Familia | Idea | Costo | Ve interacciones y redundancia | Ejemplos |
|---|---|---|---|---|
| **Filtro** | puntúa cada variable **por separado**, sin modelo | bajo | **No** | $F$ de la correlación, **información mutua** $I(X;Y)=\sum p(x,y)\log\frac{p(x,y)}{p(x)p(y)}$ |
| **Envoltura** (*wrapper*) | busca el subconjunto que **mejor predice con un modelo** (validación cruzada) | alto | Sí, **para ese modelo** | selección hacia adelante/atrás, RFE |
| **Incrustada** (*embedded*) | la selección ocurre **dentro del entrenamiento** | medio | Sí | **Lasso** ($\min_\beta\frac{1}{2n}\lVert y-X\beta\rVert^2+\alpha\lVert\beta\rVert_1$), importancias de un bosque |

Para un regresor simple, el estadístico $F$ es $F=\dfrac{r^2(n-2)}{1-r^2}$ (lo verificamos contra `f_regression`). La selección hacia adelante añade, en cada paso, la variable que más sube el $R^2$ de validación cruzada (la verificamos contra `SequentialFeatureSelector`).

**Verdad conocida** ($n=300$, $p=30$): $x_0,\dots,x_4$ con efecto **lineal**; $x_5$ con efecto **cuadrático** (su correlación lineal con $y$ es casi 0); $x_6,x_7,x_8$ **copias ruidosas** de $x_0,x_1,x_2$ (redundantes); $x_9,\dots,x_{29}$ **ruido**. Cada método elige 6 variables (el Lasso, las que no anula) y contamos cuántas de $\{x_0,\dots,x_5\}$ recupera, cuántas son redundantes y cuántas son ruido.

````python
from sklearn.feature_selection import f_regression, mutual_info_regression, SequentialFeatureSelector
from sklearn.linear_model import LassoCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance

# Datos SINTÉTICOS con verdad conocida (n = 300, p = 30):
#   x0..x4 -> efecto lineal;  x5 -> efecto CUADRÁTICO (correlación lineal ≈ 0);  x6..x8 -> copias ruidosas de x0..x2 (redundantes);  x9..x29 -> ruido
rng = np.random.default_rng(3)
n, p = 300, 30
Xs = rng.normal(size=(n, p))
Xs[:, 6:9] = Xs[:, 0:3] + rng.normal(0, 0.15, (n, 3))
ys = Xs[:, :5] @ np.array([3.0, 2.0, -2.0, 1.5, 1.0]) + 2.5 * Xs[:, 5] ** 2 + rng.normal(0, 1.5, n)
VERDAD = set(range(6))
nom = [f"x{i}" for i in range(p)]
tipo = {**{i: "lineal" for i in range(5)}, 5: "cuadrática", **{i: "redundante" for i in (6, 7, 8)}, **{i: "ruido" for i in range(9, 30)}}

# (1) FILTRO univariado: F de la correlación (desde cero) e información mutua
r = np.array([np.corrcoef(Xs[:, j], ys)[0, 1] for j in range(p)])
F_cero = r ** 2 * (n - 2) / (1 - r ** 2)
assert np.allclose(F_cero, f_regression(Xs, ys)[0])                            # F de un regresor simple == r²(n−2)/(1−r²)
mi = mutual_info_regression(Xs, ys, random_state=0)
top = lambda s, k=6: set(np.argsort(s)[::-1][:k].tolist())
sel = {"filtro F (|r|)": top(F_cero), "filtro información mutua": top(mi)}

# (2) ENVOLTURA: selección hacia adelante con validación cruzada (desde cero) == SequentialFeatureSelector
cv5 = KFold(5, shuffle=True, random_state=0)
def forward(X, y, k):
    elegido, restante = [], list(range(X.shape[1]))
    for _ in range(k):
        puntaje = {j: cross_val_score(LinearRegression(), X[:, elegido + [j]], y, cv=cv5, scoring="r2").mean() for j in restante}
        mejor = max(puntaje, key=puntaje.get)
        elegido.append(mejor); restante.remove(mejor)
    return elegido
fw = forward(Xs, ys, 6)
sk_fw = set(np.flatnonzero(SequentialFeatureSelector(LinearRegression(), n_features_to_select=6, direction="forward", cv=cv5, scoring="r2").fit(Xs, ys).get_support()))
assert set(fw) == sk_fw
sel["envoltura (hacia adelante, lineal)"] = set(fw)

# (3) INCRUSTADOS: Lasso (coeficientes ≠ 0) e importancia por permutación de un bosque
lasso = LassoCV(cv=5, random_state=0).fit((Xs - Xs.mean(0)) / Xs.std(0), ys)
sel["Lasso (coef. ≠ 0)"] = set(np.flatnonzero(np.abs(lasso.coef_) > 1e-8).tolist())
rf = RandomForestRegressor(n_estimators=200, min_samples_leaf=3, random_state=0, n_jobs=1).fit(Xs, ys)
pi = permutation_importance(rf, Xs, ys, n_repeats=5, random_state=0, n_jobs=1).importances_mean
sel["bosque (permutación, top 6)"] = top(pi)

filas = []
for m, s in sel.items():
    filas.append({"método": m, "n elegidas": len(s), "recupera de {x0..x5}": len(s & VERDAD), "redundantes": sum(tipo[j] == "redundante" for j in s),
                  "ruido": sum(tipo[j] == "ruido" for j in s), "¿incluye x5 (cuadrática)?": 5 in s})
tabla_sel = pd.DataFrame(filas).set_index("método")
print(tabla_sel.to_string())
print("\nCorrelación lineal de x5 con y:", round(r[5], 3), "| información mutua de x5:", round(mi[5], 3), "| de un ruido típico:", round(np.median(mi[9:]), 3))
assert abs(r[5]) < 0.15 and 5 not in sel["filtro F (|r|)"] and 5 in sel["filtro información mutua"]
assert 5 not in sel["envoltura (hacia adelante, lineal)"] and 5 in sel["bosque (permutación, top 6)"]
assert tabla_sel.loc["Lasso (coef. ≠ 0)", "ruido"] <= 8 and tabla_sel.loc["bosque (permutación, top 6)", "recupera de {x0..x5}"] >= 5
````

---
**Lectura (cada método ve algo distinto):**

- **Filtro $F$:** eligió $\{x_0,x_1,x_2\}$ **y sus tres copias** (3 redundantes): puntúa **una por una**, así que la redundancia le es invisible y se «gasta» la mitad del cupo en repetir información; **no ve $x_5$** (correlación lineal 0.097).
- **Filtro de información mutua:** **sí detecta $x_5$** (MI 0.165 contra 0.013 de un ruido típico), pero su estimación con $n=300$ es **ruidosa**: incluyó 2 variables de ruido y solo 2 de las verdaderas.
- **Envoltura lineal:** la más limpia (5 verdaderas, 1 redundante, 0 ruido), pero **no puede ver $x_5$**: *una envoltura solo encuentra lo que su modelo sabe usar*.
- **Lasso:** retuvo 10 variables (5 verdaderas, 1 redundante, **4 de ruido**): el $\alpha$ de validación cruzada favorece la predicción, no la parsimonia; y también es lineal.
- **Bosque (permutación):** 5 verdaderas, incluida **$x_5$**, 1 redundante, 0 ruido: captura efectos no lineales, a mayor costo.
- No hay un método «mejor»: **la elección depende de si esperas relaciones no lineales, de cuánta redundancia hay y de cuánto costo tolera el problema**.

### La fuga de datos al seleccionar 🚰

El error más caro en la práctica: **elegir las variables con todos los datos y luego validar**. La selección «vio» las etiquetas de los pliegues de validación. Lo medimos con **2 000 variables de puro ruido** y 60 etiquetas aleatorias (la exactitud real **tiene que** ser 0.50), 20 repeticiones: (a) seleccionar las 20 mejores con **todos** los datos y luego validar, y (b) seleccionar **dentro** de cada pliegue con un `Pipeline`.

````python
# La fuga de datos más frecuente: elegir variables con TODOS los datos y recién después validar.
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline

rng = np.random.default_rng(1)
fuga, bien = [], []
for _ in range(20):
    Xr = rng.normal(size=(60, 2000))                           # 2000 variables de PURO RUIDO
    yr = rng.integers(0, 2, 60)                                # etiquetas sin relación alguna con X: la exactitud verdadera es 0.5
    cv_ = StratifiedKFold(5, shuffle=True, random_state=0)
    Xk = SelectKBest(f_classif, k=20).fit_transform(Xr, yr)    # selección con TODAS las filas (incluye las de validación)
    fuga.append(cross_val_score(LogisticRegression(max_iter=1000), Xk, yr, cv=cv_).mean())
    pipe = make_pipeline(SelectKBest(f_classif, k=20), LogisticRegression(max_iter=1000))   # selección DENTRO de cada pliegue
    bien.append(cross_val_score(pipe, Xr, yr, cv=cv_).mean())
print(f"Exactitud con selección ANTES de validar (con fuga): {np.mean(fuga):.3f} ± {np.std(fuga, ddof=1):.3f}")
print(f"Exactitud con selección DENTRO del pipeline:         {np.mean(bien):.3f} ± {np.std(bien, ddof=1):.3f}   (verdad: 0.500)")
assert np.mean(fuga) > 0.70 and abs(np.mean(bien) - 0.5) < 0.05
````

## 3. PCA desde Cero con la SVD 🧮 (Sección 2.3.3)

El **análisis de componentes principales** busca las direcciones ortogonales de **máxima varianza**. Con $X_c$ la matriz centrada ($n\times p$) y su descomposición en valores singulares $X_c=U\,S\,V^\top$:

- Las **filas de $V^\top$** son los **componentes** (direcciones; sus entradas son las **cargas**), y los **autovalores** de la covarianza son $\lambda_j=\dfrac{s_j^2}{n-1}$.
- Las **puntuaciones** (coordenadas de cada observación) son $X_cV=US$.
- La **fracción de varianza explicada** del componente $j$ es $\lambda_j/\sum_i\lambda_i$.
- Con $k$ componentes, la reconstrucción $\hat{X}_k=X_cV_kV_k^\top$ tiene error cuadrático total $\lVert X_c-\hat{X}_k\rVert_F^2=(n-1)\sum_{j>k}\lambda_j$: **lo que se pierde es exactamente la varianza descartada**.

**Estandarizar o no:** PCA maximiza **varianza**, y la varianza depende de las **unidades**. Con variables en escalas distintas, la de mayor escala domina. Estandarizar equivale a usar la **matriz de correlación**.

**¿Cuántos componentes?** Tres criterios, que no siempre coinciden: (i) **varianza acumulada** (p. ej. 90 %); (ii) **Kaiser**: con datos estandarizados, conservar los $\lambda_j>1$ (explican más que una variable original); (iii) **análisis paralelo** (Horn, 1965): conservar los $\lambda_j$ mayores que el **percentil 95 de los autovalores de datos permutados** (cada columna se permuta por separado, lo que destruye las correlaciones y deja el «ruido de muestreo»). Usamos los **178 vinos** de `load_wine` (13 variables químicas; datos reales incluidos en `scikit-learn`).

````python
from sklearn.datasets import load_wine, load_diabetes, load_digits
from sklearn.decomposition import PCA

wine = load_wine()
Xw, yw, nomw = wine.data, wine.target, wine.feature_names


def pca_svd(X, escalar=True):
    """PCA desde cero con la SVD de la matriz centrada: X_c = U·S·Vᵀ  ->  autovalores λ_j = s_j²/(n−1), componentes = filas de Vᵀ."""
    Xc = X - X.mean(0)
    if escalar:
        Xc = Xc / X.std(0, ddof=1)
    U, s, Vt = np.linalg.svd(Xc, full_matrices=False)
    return {"Xc": Xc, "comp": Vt, "lam": s ** 2 / (len(X) - 1), "scores": Xc @ Vt.T}


A = pca_svd(Xw)
n_w = len(Xw)
# Verificación 1: autovalores de la matriz de covarianza (correlación, al estandarizar) == s²/(n−1)
lam_eig = np.sort(np.linalg.eigvalsh(np.cov(A["Xc"].T)))[::-1]
assert np.allclose(lam_eig, A["lam"])
# Verificación 2: contra scikit-learn (los signos de los componentes son arbitrarios)
sk = PCA().fit(A["Xc"])
assert np.allclose(sk.explained_variance_, A["lam"]) and np.allclose(sk.explained_variance_ratio_, A["lam"] / A["lam"].sum())
assert np.allclose(np.abs(sk.components_), np.abs(A["comp"])) and np.allclose(np.abs(sk.transform(A["Xc"])), np.abs(A["scores"]))
# Verificación 3: error de reconstrucción con k componentes = (n−1)·(suma de los autovalores descartados)
for k in (1, 3, 5, 8):
    Xk = A["scores"][:, :k] @ A["comp"][:k]
    assert np.isclose(((A["Xc"] - Xk) ** 2).sum(), (n_w - 1) * A["lam"][k:].sum())
print("PCA desde cero == scikit-learn; el error de reconstrucción es (n−1)·Σ λ_descartados.")

# ¿Con o sin estandarizar?
B_ = pca_svd(Xw, escalar=False)
print(f"PC1 explica {B_['lam'][0] / B_['lam'].sum():.1%} sin estandarizar (domina «{nomw[np.argmax(np.abs(B_['comp'][0]))]}») frente a {A['lam'][0] / A['lam'].sum():.1%} estandarizando")
assert B_["lam"][0] / B_["lam"].sum() > 0.99 and nomw[np.argmax(np.abs(B_["comp"][0]))] == "proline" and A["lam"][0] / A["lam"].sum() < 0.45

# ¿Cuántos componentes? varianza acumulada, Kaiser (λ > 1) y ANÁLISIS PARALELO (percentil 95 de autovalores de datos permutados)
rng = np.random.default_rng(0)
lam_perm = np.array([pca_svd(np.column_stack([rng.permutation(c) for c in Xw.T]))["lam"] for _ in range(200)])
umbral = np.percentile(lam_perm, 95, axis=0)
acum = np.cumsum(A["lam"]) / A["lam"].sum()
k90 = int(np.searchsorted(acum, 0.90) + 1)
k_kaiser, k_par = int((A["lam"] > 1).sum()), int((A["lam"] > umbral).sum())
print(f"k por 90 % de varianza: {k90} | Kaiser (λ>1): {k_kaiser} | análisis paralelo: {k_par}")
assert k_kaiser == 3 and k_par == 3 and k90 == 8

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
ax[0].plot(range(1, 14), A["lam"], "o-", color="#2563eb", label="observado")
ax[0].plot(range(1, 14), umbral, "s--", color="#dc2626", label="percentil 95 (datos permutados)")
ax[0].axhline(1, color="gray", lw=0.8); ax[0].set(title="Scree plot y análisis paralelo", xlabel="componente", ylabel="autovalor"); ax[0].legend(fontsize=8)
ax[1].bar(range(1, 14), A["lam"] / A["lam"].sum(), color="#93c5fd"); ax[1].plot(range(1, 14), acum, "o-", color="#1e3a8a")
ax[1].axhline(0.9, color="#dc2626", ls="--"); ax[1].set(title="Varianza explicada y acumulada", xlabel="componente")
for c, col in zip(range(3), ["#2563eb", "#dc2626", "#16a34a"]):
    ax[2].scatter(A["scores"][yw == c, 0], A["scores"][yw == c, 1], s=14, color=col, alpha=0.7, label=wine.target_names[c])
ax[2].set(title=f"Vinos en PC1–PC2 ({acum[1]:.0%} de la varianza)", xlabel="PC1", ylabel="PC2"); ax[2].legend(fontsize=8)
plt.tight_layout(); plt.show()
carga = pd.DataFrame(A["comp"][:2].T, index=nomw, columns=["PC1", "PC2"])
print(carga.reindex(carga.PC1.abs().sort_values(ascending=False).index).head(5).round(3).to_string())
````

- **Verificación:** la PCA desde cero coincide con `sklearn.decomposition.PCA` (autovalores, razones y componentes, salvo el **signo**, que es arbitrario) y el error de reconstrucción cumple la identidad de la varianza descartada.
- **Sin estandarizar**, PC1 explica **99.8 %** de la varianza y es casi solo **`proline`** (la variable de mayor escala, en cientos); estandarizando, **36.2 %**. *Un PCA sin estandarizar solo describe qué variable tiene las unidades más grandes.*

## 4. Los Límites de PCA: Reducir No Siempre Ayuda ⚖️

PCA es **no supervisado**: ordena direcciones por **varianza**, no por **relevancia para $y$**. Dos consecuencias que medimos:

1. **Regresión sobre componentes (PCR)** en `load_diabetes` (442 pacientes, 10 variables): ¿mejora reducir? Con todos los componentes, PCR **es** la regresión original.
2. **La señal en un componente de baja varianza:** dos clases que se separan solo a lo largo de un eje de **poca varianza**, rotado para que no sea evidente. PC1 (gran varianza, sin información) contra PC2 (poca varianza, toda la información).

## 5. t-SNE: Útil para Mirar, Peligroso para Interpretar 🌀

**t-SNE** (van der Maaten & Hinton, 2008) es una proyección **no lineal** para **visualizar** estructura local. Construye afinidades entre puntos en el espacio original y busca un mapa 2-D con afinidades similares:

1. **Afinidades originales.** Para cada punto $i$: $p_{j\mid i}=\dfrac{\exp(-\lVert x_i-x_j\rVert^2/2\sigma_i^2)}{\sum_{k\neq i}\exp(-\lVert x_i-x_k\rVert^2/2\sigma_i^2)}$, con $\sigma_i$ elegido por bisección para que la **perplejidad** $\exp(H_i)$, con $H_i=-\sum_j p_{j\mid i}\ln p_{j\mid i}$, sea la fijada por el usuario (≈ «número efectivo de vecinos»). Se simetriza: $p_{ij}=\dfrac{p_{j\mid i}+p_{i\mid j}}{2n}$.
2. **Afinidades del mapa** (colas pesadas, $t$ con 1 grado de libertad): $q_{ij}=\dfrac{(1+\lVert y_i-y_j\rVert^2)^{-1}}{\sum_{k\neq l}(1+\lVert y_k-y_l\rVert^2)^{-1}}$.
3. **Objetivo:** minimizar $\mathrm{KL}(P\,\Vert\,Q)=\sum_{i\neq j}p_{ij}\ln\dfrac{p_{ij}}{q_{ij}}$ por descenso de gradiente, con $\dfrac{\partial\,\mathrm{KL}}{\partial y_i}=4\sum_j(p_{ij}-q_{ij})(y_i-y_j)(1+\lVert y_i-y_j\rVert^2)^{-1}$.

Medimos la **confiabilidad** (*trustworthiness*, $0\!\to\!1$): qué fracción de los vecinos que aparecen en el mapa **eran** vecinos reales. Con 500 dígitos de `load_digits` (64 variables) comparamos PCA y t-SNE con perplejidades 5, 30 y 100, y luego mostramos **tres advertencias** con datos construidos para que la verdad sea conocida:

| Advertencia | Experimento | Verdad conocida |
|---|---|---|
| **1. Tamaño y densidad no se conservan** | dos nubes: sd 1 y sd 5 | razón de dispersión = 5 |
| **2. El ruido parece estructura** | 400 puntos gaussianos de 20 variables, perplejidad 5 | **no hay grupos** |
| **3. Las distancias entre grupos no significan nada** | 8 grupos **equidistantes** (vértices de un simplex) | todas las distancias iguales (CV = 0) |

````python
import time
from scipy.stats import spearmanr
from scipy.spatial.distance import pdist
from sklearn.manifold import TSNE, trustworthiness
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

dg = load_digits()
idx = np.random.default_rng(0).choice(len(dg.data), 500, replace=False)
Xg, yg = dg.data[idx], dg.target[idx]                                  # 500 dígitos escritos a mano (8×8 = 64 variables), incluidos en scikit-learn
t0 = time.time()
emb, fila = {}, []
pca2 = PCA(2).fit_transform(Xg)
fila.append({"proyección": "PCA (2 componentes)", "confiabilidad (k=10)": trustworthiness(Xg, pca2, n_neighbors=10), "KL": np.nan})
for perp in (5, 30, 100):
    ts = TSNE(2, perplexity=perp, init="pca", random_state=0)
    emb[perp] = ts.fit_transform(Xg)
    fila.append({"proyección": f"t-SNE perplejidad {perp}", "confiabilidad (k=10)": trustworthiness(Xg, emb[perp], n_neighbors=10), "KL": ts.kl_divergence_})
tab_ts = pd.DataFrame(fila).set_index("proyección")
print(tab_ts.round(3).to_string())

fig, ax = plt.subplots(1, 4, figsize=(17, 4))
for a_, (titulo, E) in zip(ax, [("PCA", pca2)] + [(f"t-SNE, perplejidad {p_}", emb[p_]) for p_ in (5, 30, 100)]):
    sc = a_.scatter(E[:, 0], E[:, 1], c=yg, cmap="tab10", s=8); a_.set_title(titulo, fontsize=10); a_.set_xticks([]); a_.set_yticks([]); a_.grid(False)
ax[0].legend(*sc.legend_elements(), fontsize=6, ncol=2, loc="best", title="dígito", title_fontsize=7)
plt.tight_layout(); plt.show()
assert tab_ts.loc["t-SNE perplejidad 30", "confiabilidad (k=10)"] > tab_ts.loc["PCA (2 componentes)", "confiabilidad (k=10)"] + 0.05

# ADVERTENCIA 1: el tamaño y la densidad de los grupos NO se conservan (t-SNE adapta la escala local a la perplejidad)
rng = np.random.default_rng(2)
G = np.vstack([rng.normal(0, 1.0, (200, 10)), rng.normal(0, 5.0, (200, 10)) + 40])      # un grupo compacto (sd 1) y uno disperso (sd 5)
Eg = TSNE(2, perplexity=30, init="pca", max_iter=500, random_state=0).fit_transform(G)
sd_orig = G[:200].std() , G[200:].std()
sd_2d = Eg[:200].std(0).mean(), Eg[200:].std(0).mean()
print(f"ADVERTENCIA 1 — razón de dispersión disperso/compacto: original = {sd_orig[1] / sd_orig[0]:.1f} | en t-SNE = {sd_2d[1] / sd_2d[0]:.1f}")

# ADVERTENCIA 2: con perplejidad baja aparecen «grupos» en puro ruido
N = rng.normal(size=(400, 20))
En = TSNE(2, perplexity=5, init="pca", max_iter=500, random_state=0).fit_transform(N)
sil_o = silhouette_score(N, KMeans(5, n_init=10, random_state=0).fit_predict(N))
sil_t = silhouette_score(En, KMeans(5, n_init=10, random_state=0).fit_predict(En))
print(f"ADVERTENCIA 2 — silueta de K-means (k=5) sobre RUIDO gaussiano: espacio original = {sil_o:.3f} | t-SNE perplejidad 5 = {sil_t:.3f}")

# ADVERTENCIA 3: las distancias ENTRE grupos no son interpretables. 8 grupos equidistantes (vértices de un simplex) -> t-SNE los coloca a distancias muy distintas
rng = np.random.default_rng(6)
S = np.vstack([rng.normal(0, 1.0, (40, 8)) + 15 * np.eye(8)[i] for i in range(8)])
lab = np.repeat(np.arange(8), 40)
cvs = []
for sem in (0, 1):
    Es = TSNE(2, perplexity=30, init="random", max_iter=500, random_state=sem).fit_transform(S)
    dc = pdist(np.array([Es[lab == c].mean(0) for c in range(8)]))
    cvs.append(dc.std() / dc.mean())
    print(f"ADVERTENCIA 3 — semilla {sem}: distancias entre los 28 pares de centroides: mín = {dc.min():.1f}, máx = {dc.max():.1f}, coef. de variación = {cvs[-1]:.2f}  (en el espacio original es 0.00)")
print(f"(tiempo de esta celda: {time.time() - t0:.0f} s)")
assert sd_2d[1] / sd_2d[0] < 0.5 * sd_orig[1] / sd_orig[0]
assert sil_t > 2 * sil_o
assert min(cvs) > 0.2
````

**Lectura:**

- **t-SNE vs PCA:** la confiabilidad sube de **0.83 (PCA)** a **≈ 0.98** para las tres perplejidades: para **vecindades locales** en datos de dígitos t-SNE es mucho mejor. La divergencia KL **no es comparable entre perplejidades** (cambia el objetivo) y el aspecto del gráfico cambia: con 5 se fragmenta, con 100 se acerca.
- **Advertencia 1:** la razón de dispersión pasó de **5.0** a **1.0**: t-SNE **iguala la densidad local**; el tamaño de un grupo en el mapa **no** dice qué tan disperso es.
- **Advertencia 2:** sobre ruido gaussiano, la silueta de K-means es **0.04** en el espacio original (sin estructura) y **0.35** tras t-SNE con perplejidad 5: *aparecen «grupos» donde no los hay*.
- **Advertencia 3:** 8 grupos **equidistantes** quedaron a distancias entre 15 y 46 (**CV ≈ 0.35**, no 0): la **posición relativa de los grupos y el espacio vacío entre ellos no son interpretables**. Solo la **pertenencia local** (qué puntos son vecinos) es confiable.
- **Reglas de uso:** (i) prueba **varias perplejidades** y semillas; (ii) no midas distancias, tamaños ni densidades en el mapa; (iii) valida lo que veas con **otro método** (clustering en el espacio original, PCA); (iv) no uses t-SNE para alimentar un modelo sin cuidado: no define una función para puntos nuevos. (Existen alternativas, como UMAP, que no usamos aquí porque no está en el entorno estándar.)

### t-SNE desde cero (exacto, $O(n^2)$)

Para comprobar que entendemos el algoritmo, lo implementamos con **300 dígitos**: **bisección** de $\sigma_i$ (se verifica que la entropía de cada $P_{\cdot\mid i}$ es $\ln 30$), simetrización, gradiente con **exageración temprana** ($\times12$ las primeras 100 iteraciones), **momento** y **ganancias adaptativas**. No esperamos coincidir punto por punto con `scikit-learn` (usa la aproximación de Barnes-Hut y otra inicialización), pero sí que la **KL baje** y la **confiabilidad** sea parecida.

````python
def afinidades(X, perp):
    """P_{j|i}: para cada punto i se busca por bisección el ancho gaussiano (β_i = 1/2σ_i²) cuya entropía de Shannon es ln(perplejidad)."""
    n = len(X)
    D = ((X[:, None, :] - X[None, :, :]) ** 2).sum(-1)
    P, objetivo = np.zeros((n, n)), np.log(perp)
    for i in range(n):
        d = np.delete(D[i], i)
        lo, hi = 1e-12, 1e12
        for _ in range(100):
            beta = np.sqrt(lo * hi)
            w = np.exp(-(d - d.min()) * beta)
            p_ = w / w.sum()
            H = -(p_ * np.log(p_ + 1e-300)).sum()
            lo, hi = (beta, hi) if H > objetivo else (lo, beta)               # H decrece con β: H grande -> subir β
        P[i, np.arange(n) != i] = p_
    return P


def tsne_cero(X, perp=30, n_iter=600, lr=100.0, semilla=0):
    """t-SNE exacto (O(n²)): afinidades simétricas P, Student-t con 1 g.l. en 2-D, descenso por gradiente con momento, ganancias y exageración temprana."""
    n = len(X)
    Pc = afinidades(X, perp)
    P = (Pc + Pc.T) / (2 * n)
    Y = PCA(2, random_state=semilla).fit_transform(X)
    Y = 1e-2 * Y / Y[:, 0].std()
    upd, gan, kl = np.zeros_like(Y), np.ones_like(Y), []
    for it in range(n_iter):
        ex = 12.0 if it < 100 else 1.0
        D2 = ((Y[:, None, :] - Y[None, :, :]) ** 2).sum(-1)
        num = 1.0 / (1.0 + D2); np.fill_diagonal(num, 0)
        Q = np.maximum(num / num.sum(), 1e-12)
        PQ = (ex * P - Q) * num
        grad = 4 * (PQ.sum(1)[:, None] * Y - PQ @ Y)
        gan = np.where(np.sign(grad) != np.sign(upd), gan + 0.2, gan * 0.8).clip(0.01)
        upd = (0.5 if it < 100 else 0.8) * upd - lr * gan * grad
        Y = Y + upd
        Y -= Y.mean(0)
        if it % 50 == 49:
            kl.append((it + 1, float((P * np.log(np.maximum(P, 1e-12) / Q)).sum())))
    return Y, Pc, kl


sub = np.random.default_rng(1).choice(len(dg.data), 300, replace=False)
Xs_, ys_ = dg.data[sub], dg.target[sub]
t0 = time.time()
Y0, Pc, kl = tsne_cero(Xs_)
H = -(Pc * np.log(Pc + 1e-300)).sum(1)
print("Entropía de cada P_{j|i} == ln(perplejidad):", np.allclose(H, np.log(30), atol=1e-3), "| perplejidad efectiva mínima/máxima:", np.exp(H).min().round(2), np.exp(H).max().round(2))
print("KL (cada 50 iteraciones):", [(i, round(k, 3)) for i, k in kl])
sk_Y = TSNE(2, perplexity=30, init="pca", random_state=0).fit_transform(Xs_)
tr = {"PCA": trustworthiness(Xs_, PCA(2).fit_transform(Xs_), n_neighbors=10), "t-SNE desde cero": trustworthiness(Xs_, Y0, n_neighbors=10),
      "t-SNE scikit-learn": trustworthiness(Xs_, sk_Y, n_neighbors=10)}
print("Confiabilidad (k=10):", {k: round(float(v), 3) for k, v in tr.items()}, f"| {time.time() - t0:.0f} s")
fig, ax = plt.subplots(1, 2, figsize=(11, 4.2))
for a_, E, tt in zip(ax, [Y0, sk_Y], ["t-SNE desde cero (NumPy)", "t-SNE de scikit-learn"]):
    a_.scatter(E[:, 0], E[:, 1], c=ys_, cmap="tab10", s=10); a_.set_title(tt, fontsize=10); a_.set_xticks([]); a_.set_yticks([]); a_.grid(False)
plt.tight_layout(); plt.show()
assert np.allclose(H, np.log(30), atol=1e-3)
assert kl[-1][1] < kl[2][1] and all(np.isfinite(k) for _, k in kl)
assert tr["t-SNE desde cero"] > tr["PCA"] + 0.05 and abs(tr["t-SNE desde cero"] - tr["t-SNE scikit-learn"]) < 0.05
````
