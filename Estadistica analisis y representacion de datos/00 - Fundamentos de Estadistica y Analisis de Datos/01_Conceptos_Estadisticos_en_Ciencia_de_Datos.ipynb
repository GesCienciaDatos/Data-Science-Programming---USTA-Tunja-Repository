## 1. Sesgo de Muestreo: Cuando Más Datos No Ayudan 🎯

Un estimador tiene **dos fuentes de error** que conviene no confundir:

- **Variabilidad (varianza):** cuánto cambia el estimador de una muestra a otra. **Disminuye** al crecer $n$ (como $1/\sqrt{n}$).
- **Sesgo:** cuánto se **desvía sistemáticamente** del parámetro: $\operatorname{Sesgo}(\hat\theta)=E[\hat\theta]-\theta$. **No** disminuye al crecer $n$ si la muestra se obtiene con un mecanismo defectuoso.

Se combinan en el **error cuadrático medio**:

$$\text{ECM}(\hat\theta)=E\big[(\hat\theta-\theta)^2\big]=\operatorname{Var}(\hat\theta)+\operatorname{Sesgo}(\hat\theta)^2 .$$

Por eso **100 000 respuestas sesgadas pueden valer menos que 100 respuestas bien muestreadas**. Los sesgos de muestreo más frecuentes:

| Sesgo | Qué ocurre | Ejemplo |
|---|---|---|
| **Cobertura / selección** | El marco muestral omite a una parte de la población | Encuesta telefónica a quienes tienen línea fija |
| **No respuesta** | Responden más unos grupos que otros | Encuesta en línea: responden sobre todo los jóvenes |
| **Autoselección** | La gente decide si entra a la muestra | Reseñas de producto: opinan los muy contentos y los muy molestos |
| **Supervivencia** | Solo se observa lo que «sobrevivió» | Analizar solo las empresas que siguen operando |
| **Medición** | El instrumento mide mal de forma sistemática | Báscula descalibrada |

**Corrección por ponderación.** Si cada persona $i$ responde con probabilidad $\pi_i$ **conocida** (o estimable a partir de variables observadas), se pondera a cada respondiente con $w_i=1/\pi_i$ (idea de Horvitz–Thompson; en su forma normalizada, el estimador de Hájek):

$$\hat\mu_w=\frac{\sum_{i\in\text{resp}} w_i\,y_i}{\sum_{i\in\text{resp}} w_i}.$$

Esto corrige el sesgo **solo si** la respuesta depende únicamente de variables que **sí observamos** (por ejemplo, el grupo de edad). Si depende del **propio valor $y$** (los inconformes contestan más) no hay ponderación por grupo que lo arregle del todo.

La celda crea una población **ficticia** de 100 000 clientes con tres grupos de edad y simula una encuesta en línea con dos mecanismos: **(A)** la probabilidad de responder depende solo del grupo; **(B)** además depende de la propia satisfacción (los menos satisfechos responden más). Para cada tamaño de invitación, 300 repeticiones.

````python
# ============================================================
# 1. Sesgo de no respuesta: el sesgo no se va con mas datos; la ponderacion corrige solo lo observable
# ============================================================
rng = np.random.default_rng(SEED)
N = 100_000
grupo = rng.choice(3, size=N, p=[0.4, 0.4, 0.2])                       # 0 joven, 1 adulto, 2 mayor
nombres = np.array(["joven", "adulto", "mayor"])
satisf = np.clip(rng.normal(np.array([6.0, 7.0, 8.0])[grupo], 1.5), 0, 10)   # puntaje 0-10 de satisfaccion (FICTICIO)
mu_pob = satisf.mean()
p_grupo = np.array([0.30, 0.10, 0.02])                                  # prob. de responder por grupo (mecanismo A)
p_A = p_grupo[grupo]
p_B = np.clip(p_A * np.exp(-0.35 * (satisf - mu_pob)), 0, 1)           # (B): los menos satisfechos responden mas
print(f"Media poblacional de satisfaccion (parametro) = {mu_pob:.3f} | tasa de respuesta global: A = {p_A.mean():.3f}, B = {p_B.mean():.3f}")


def media_ponderada(y, w):
    y, w = np.asarray(y, float), np.asarray(w, float)
    return (w * y).sum() / w.sum()


assert np.isclose(media_ponderada([1, 2, 3], [1, 1, 2]), np.average([1, 2, 3], weights=[1, 1, 2]))


def encuesta(rng, n_inv, p):
    invitados = rng.choice(N, size=n_inv, replace=False)                # muestreo aleatorio simple de INVITADOS
    resp = invitados[rng.random(n_inv) < p[invitados]]                  # responden con probabilidad p_i
    return resp


R = 300
filas = []
for n_inv in (200, 2_000, 20_000):
    for mec, p in (("A (solo grupo)", p_A), ("B (grupo y satisfaccion)", p_B)):
        r = np.random.default_rng(SEED + n_inv)
        ing, pon = [], []
        for _ in range(R):
            idx = encuesta(r, n_inv, p)
            ing.append(satisf[idx].mean())
            pon.append(media_ponderada(satisf[idx], 1 / p_grupo[grupo[idx]]))           # pesos 1/p_grupo (conocidos)
        for nombre, e in (("ingenuo", np.array(ing)), ("ponderado", np.array(pon))):
            filas.append({"invitados": n_inv, "mecanismo": mec, "estimador": nombre, "sesgo": e.mean() - mu_pob, "desv. est.": e.std(ddof=1), "RMSE": np.sqrt(((e - mu_pob) ** 2).mean())})
res = pd.DataFrame(filas)
print(res.round(3).to_string(index=False))

# Descomposicion ECM = varianza + sesgo^2 (identidad exacta con momentos empiricos, ddof = 0)
r = np.random.default_rng(1)
e = np.array([satisf[encuesta(r, 2_000, p_A)].mean() for _ in range(R)])
assert np.isclose(((e - mu_pob) ** 2).mean(), e.var(ddof=0) + (e.mean() - mu_pob) ** 2)

g = res.set_index(["mecanismo", "estimador", "invitados"])
ing_A = g.loc[("A (solo grupo)", "ingenuo")]
pon_A = g.loc[("A (solo grupo)", "ponderado")]
pon_B = g.loc[("B (grupo y satisfaccion)", "ponderado")]
ing_B = g.loc[("B (grupo y satisfaccion)", "ingenuo")]
assert abs(ing_A.loc[20_000, "sesgo"]) > 0.4                                              # el sesgo ingenuo sigue ahi con 100 veces mas invitados
assert ing_A.loc[20_000, "desv. est."] < ing_A.loc[200, "desv. est."] / 5                 # ...aunque la variabilidad si cae
assert abs(pon_A.loc[20_000, "sesgo"]) < 0.05                                            # la ponderacion corrige el mecanismo A
assert abs(pon_B.loc[20_000, "sesgo"]) > 0.1                                             # pero NO del todo el mecanismo B
assert abs(pon_B.loc[20_000, "sesgo"]) < abs(ing_B.loc[20_000, "sesgo"])                 # aunque mejora

fig, ax = plt.subplots(1, 2, figsize=(12, 4.2), sharey=True)
for a, mec in zip(ax, ("A (solo grupo)", "B (grupo y satisfaccion)")):
    for est, color in (("ingenuo", PALETA[1]), ("ponderado", PALETA[2])):
        d = g.loc[(mec, est)]
        a.plot(d.index, d["RMSE"], "o-", color=color, lw=2, label=f"RMSE {est}")
    a.axhline(abs(ing_A.loc[20_000, "sesgo"]) if mec.startswith("A") else abs(ing_B.loc[20_000, "sesgo"]), color=PALETA[1], ls=":", label="piso: |sesgo| ingenuo")
    a.set_xscale("log"); a.set_title(f"Mecanismo {mec}", fontweight="bold", fontsize=10); a.set_xlabel("personas invitadas"); a.legend(fontsize=8)
ax[0].set_ylabel("error (RMSE, puntos de satisfacción)")
plt.tight_layout(); plt.show()

# Una encuesta concreta (mecanismo A, 20 000 invitados) para la practica
r = np.random.default_rng(11)
idx_obs = encuesta(r, 20_000, p_A)
encuesta_obs = pd.DataFrame({"satisfaccion": satisf[idx_obs], "grupo": nombres[grupo[idx_obs]]})
p_resp = {"joven": 0.30, "adulto": 0.10, "mayor": 0.02}                                      # probabilidades de respuesta por grupo (conocidas)
print(f"\nEncuesta concreta: {len(encuesta_obs)} respondientes de 20 000 invitados | composicion {encuesta_obs['grupo'].value_counts(normalize=True).round(2).to_dict()}")
````

## 2. Variabilidad: Ruido, Error Estándar y Falsos Descubrimientos 🎲

La **variabilidad** es la materia prima de la estadística, y hay tres tipos que se confunden: la **natural** (las unidades difieren entre sí), la de **muestreo** (otra muestra daría otro estadístico) y la de **medición** (el instrumento no es exacto). La de muestreo se cuantifica con el **error estándar**; para la media, $\operatorname{EE}(\bar{x})=\sigma/\sqrt{n}$, de modo que **para reducir el error a la mitad hace falta cuadruplicar la muestra** y los rendimientos son decrecientes.

Ese ruido tiene una consecuencia incómoda: **si se miran suficientes métricas, alguna parecerá «significativa» por pura casualidad**. Para una prueba individual con nivel $\alpha=0.05$, el **valor p** es la probabilidad, **calculada suponiendo que la hipótesis nula $H_0$ es cierta**, de observar un resultado **al menos tan extremo** como el obtenido. Bajo $H_0$ rechazaremos por error el $5\,\%$ de las veces. Pero si se hacen $m$ comparaciones independientes, la probabilidad de **al menos un falso positivo** (*FWER*) es

$$\text{FWER}=1-(1-\alpha)^m\;\xrightarrow{\;m=20\;}\;1-0.95^{20}\approx 0.64 ,$$

y la **corrección de Bonferroni** (usar $\alpha/m$ en cada prueba) la devuelve a $\le\alpha$. El valor p **no** es la probabilidad de que $H_0$ sea cierta (eso requiere un enfoque bayesiano y una probabilidad *a priori*); el Módulo 03 lo desarrolla con detalle.

La celda hace dos experimentos: **(a)** mide cómo cae el error estándar con $n$ y ajusta la pendiente en escala log-log (teoría: $-0.5$); **(b)** simula **2 000 «experimentos A/B» en los que no hay ningún efecto real** (ambos grupos vienen de la misma población), cada uno con 20 métricas.

````python
# ============================================================
# 2. Error estandar vs n, y comparaciones multiples (todo bajo H0: no hay efecto)
# ============================================================
rng = np.random.default_rng(SEED)
tiempos = rng.gamma(shape=4.0, scale=5.0, size=200_000)               # tiempo de entrega (min) de una poblacion FICTICIA: media 20, sd 10
sigma = tiempos.std()
ns = np.array([10, 40, 160, 640, 2_560])
ee = np.array([rng.choice(tiempos, size=(1_000, n)).mean(axis=1).std(ddof=1) for n in ns])
pendiente = np.polyfit(np.log(ns), np.log(ee), 1)[0]
tabla = pd.DataFrame({"n": ns, "EE simulado": ee, "sigma/raiz(n)": sigma / np.sqrt(ns), "cociente EE(n)/EE(4n)": np.r_[ee[:-1] / ee[1:], np.nan]}).set_index("n")
print(tabla.round(3).to_string())
print(f"\nPendiente log-log del EE frente a n = {pendiente:.3f} (teoria: -0.5)")
assert -0.55 < pendiente < -0.45
assert np.allclose(ee, sigma / np.sqrt(ns), rtol=0.08)
assert np.all(np.abs(tabla["cociente EE(n)/EE(4n)"].dropna() - 2) < 0.2)           # cuadruplicar n divide el error entre ~2

# (b) 2 000 "experimentos" con 20 metricas cada uno, SIN efecto real (A y B son iguales)
E, M, n_g = 2_000, 20, 50
r = np.random.default_rng(SEED + 1)
a = r.normal(0, 1, size=(E, M, n_g))
b = r.normal(0, 1, size=(E, M, n_g))
p = stats.ttest_ind(a, b, axis=2, equal_var=False).pvalue            # forma (E, M): una prueba t de Welch por metrica
tasa_1 = (p < 0.05).mean()                                           # falsos positivos por prueba individual
hay_alguna = (p < 0.05).any(axis=1).mean()                           # P(al menos un "descubrimiento" en el experimento)
hay_alguna_bonf = (p < 0.05 / M).any(axis=1).mean()                  # con Bonferroni
print(f"\nBajo H0: falsos positivos por prueba = {tasa_1:.3f} | P(al menos uno de {M}) = {hay_alguna:.3f} (teoria {1 - 0.95 ** M:.3f}) | con Bonferroni = {hay_alguna_bonf:.3f}")
assert 0.04 < tasa_1 < 0.06
assert abs(hay_alguna - (1 - 0.95 ** M)) < 0.04
assert hay_alguna_bonf < 0.08
assert abs(np.histogram(p.ravel(), bins=10, range=(0, 1))[0] / p.size - 0.1).max() < 0.01   # bajo H0 el valor p es ~ uniforme

fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
ax[0].loglog(ns, ee, "o-", color=PALETA[0], lw=2, label="EE simulado")
ax[0].loglog(ns, sigma / np.sqrt(ns), "k--", label="$\\sigma/\\sqrt{n}$ (pendiente $-1/2$)")
ax[0].set_xlabel("n"); ax[0].set_ylabel("error estándar de la media"); ax[0].set_title("Cuadruplicar n divide el error entre 2", fontweight="bold", fontsize=10); ax[0].legend()
ax[1].hist(p.ravel(), bins=20, range=(0, 1), color=PALETA[4], alpha=0.8, density=True)
ax[1].axhline(1, color="black", ls="--"); ax[1].set_xlabel("valor p"); ax[1].set_title("Bajo $H_0$ el valor p es uniforme: el 5 % cae bajo 0.05 por azar", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 3. Correlación Frente a Causalidad 🔗

La **correlación de Pearson** mide la **asociación lineal** entre dos variables:

$$r=\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i-\bar{x})^2\,\sum_{i=1}^{n}(y_i-\bar{y})^2}}\in[-1,1].$$

La de **Spearman** es la de Pearson aplicada a los **rangos**; mide asociación **monótona** y es robusta a atípicos. Dos advertencias: (1) $r\approx 0$ **no** implica independencia (solo ausencia de relación *lineal*); (2) $r\neq 0$ **no** implica causalidad.

**¿Por qué una asociación puede no ser causal?** Hay tres explicaciones además de «$X$ causa $Y$»: **azar** (variabilidad), **causalidad inversa** ($Y$ causa $X$) y un **confusor** $Z$ que causa a ambas: $X\leftarrow Z\rightarrow Y$. Con un confusor, controlar por $Z$ hace desaparecer la asociación espuria. La **correlación parcial** de $X$ e $Y$ controlando $Z$ es

$$r_{xy\cdot z}=\frac{r_{xy}-r_{xz}\,r_{yz}}{\sqrt{(1-r_{xz}^2)(1-r_{yz}^2)}},$$

que coincide con la correlación entre los **residuos** de regresar $X$ sobre $Z$ y $Y$ sobre $Z$. La herramienta más fuerte para establecer causalidad es el **experimento aleatorizado**: al asignar $X$ al azar, se rompe cualquier vínculo con $Z$ y la diferencia en $Y$ se atribuye a $X$. También hay sesgos que se **crean al condicionar**: si se selecciona por una variable que **depende de ambas** (un *colisionador* $X\rightarrow S\leftarrow Y$), aparece una asociación entre $X$ e $Y$ aunque sean independientes.

La celda: **(1)** calcula Pearson y Spearman desde cero; **(2)** muestra una dependencia perfecta con $r\approx 0$; **(3)** simula el confusor *temperatura* que genera una asociación entre ventas de helado y casos de insolación **sin** causalidad; **(4)** repite con un efecto real $\beta=0.5$ y compara el efecto **ingenuo**, **ajustado** y **aleatorizado**; **(5)** crea una correlación negativa solo por **selección**. Todos los datos son sintéticos.

````python
# ============================================================
# 3. Correlacion, confusores, aleatorizacion y colisionadores
# ============================================================
def pearson(x, y):
    x, y = np.asarray(x, float), np.asarray(y, float)
    xc, yc = x - x.mean(), y - y.mean()
    return (xc * yc).sum() / np.sqrt((xc ** 2).sum() * (yc ** 2).sum())


def rangos(x):
    return stats.rankdata(x)                                   # promedia empates, igual que scipy.stats.spearmanr


def spearman(x, y):
    return pearson(rangos(x), rangos(y))


def parcial(rxy, rxz, ryz):
    return (rxy - rxz * ryz) / np.sqrt((1 - rxz ** 2) * (1 - ryz ** 2))


rng = np.random.default_rng(SEED)
n = 5_000

# (1) Contraste con scipy
x1 = rng.normal(size=300); y1 = 0.6 * x1 + rng.normal(size=300)
assert np.isclose(pearson(x1, y1), stats.pearsonr(x1, y1)[0]) and np.isclose(spearman(x1, y1), stats.spearmanr(x1, y1)[0])

# (2) Dependencia perfecta (casi) con r ~ 0
xq = rng.uniform(-3, 3, n); yq = xq ** 2 + rng.normal(0, 0.3, n)
print(f"(2) y = x^2 + ruido: Pearson = {pearson(xq, yq):.3f} | Spearman = {spearman(xq, yq):.3f} | pero corr(y, x^2) = {pearson(yq, xq ** 2):.3f}")
assert abs(pearson(xq, yq)) < 0.1 and abs(spearman(xq, yq)) < 0.1 and pearson(yq, xq ** 2) > 0.95

# (3) Confusor: temperatura -> helado, temperatura -> insolacion; helado NO causa insolacion
temp = rng.normal(14, 4, n)
helado = 5 + 2.0 * temp + rng.normal(0, 4, n)
insol = 1 + 0.8 * temp + rng.normal(0, 3, n)
rxy, rxz, ryz = pearson(helado, insol), pearson(helado, temp), pearson(insol, temp)
r_par = parcial(rxy, rxz, ryz)
res_x = helado - np.polyval(np.polyfit(temp, helado, 1), temp)
res_y = insol - np.polyval(np.polyfit(temp, insol, 1), temp)
print(f"(3) corr(helado, insolacion) = {rxy:.3f} | controlando temperatura (parcial) = {r_par:.3f} | via residuos = {pearson(res_x, res_y):.3f}")
assert rxy > 0.4 and abs(r_par) < 0.05 and np.isclose(r_par, pearson(res_x, res_y), atol=1e-8)
df_conf = pd.DataFrame({"temp": temp, "helado": helado, "insolacion": insol})        # lo reutiliza la practica


# (4) Efecto causal verdadero beta = 0.5 de X sobre Y, con confusor Z (motivacion)
def mco(X, y):
    X1 = np.column_stack([np.ones(len(X)), X])
    beta = np.linalg.solve(X1.T @ X1, X1.T @ y)
    return beta


beta_v = 0.5
Z = rng.normal(0, 1, n)
X_obs = 2.0 * Z + rng.normal(0, 1, n)                                   # observacional: X depende de Z
Y_obs = beta_v * X_obs + 1.5 * Z + rng.normal(0, 1, n)
X_alea = rng.normal(0, np.sqrt(5), n)                                   # aleatorizado: X independiente de Z (misma varianza que X_obs)
Y_alea = beta_v * X_alea + 1.5 * Z + rng.normal(0, 1, n)
b_ing = mco(X_obs[:, None], Y_obs)[1]
b_aju = mco(np.column_stack([X_obs, Z]), Y_obs)[1]
b_ale = mco(X_alea[:, None], Y_alea)[1]
assert np.allclose(mco(np.column_stack([X_obs, Z]), Y_obs), sm.OLS(Y_obs, sm.add_constant(np.column_stack([X_obs, Z]))).fit().params)   # contra statsmodels
print(f"(4) efecto verdadero = {beta_v} | ingenuo (observacional) = {b_ing:.3f} | ajustado por Z = {b_aju:.3f} | aleatorizado = {b_ale:.3f}")
assert b_ing > 1.0 and abs(b_aju - beta_v) < 0.05 and abs(b_ale - beta_v) < 0.05

# (5) Colisionador: talento y esfuerzo independientes; solo "pasan" quienes suman mucho
talento = rng.normal(size=n); esfuerzo = rng.normal(size=n)
pasa = (talento + esfuerzo) > 1.0
print(f"(5) corr(talento, esfuerzo) en todos = {pearson(talento, esfuerzo):.3f} | solo entre quienes pasan = {pearson(talento[pasa], esfuerzo[pasa]):.3f}")
assert abs(pearson(talento, esfuerzo)) < 0.05 and pearson(talento[pasa], esfuerzo[pasa]) < -0.3

fig, ax = plt.subplots(1, 3, figsize=(14, 4))
ax[0].scatter(xq, yq, s=4, alpha=0.4, color=PALETA[0]); ax[0].set_title(f"$y=x^2$: Pearson = {pearson(xq, yq):.3f}", fontweight="bold", fontsize=10)
sc = ax[1].scatter(helado[:1500], insol[:1500], c=temp[:1500], s=6, cmap="coolwarm"); ax[1].set_xlabel("ventas de helado"); ax[1].set_ylabel("casos de insolación")
ax[1].set_title(f"Confusor: r = {rxy:.2f}, |parcial| = {abs(r_par):.2f}", fontweight="bold", fontsize=10); plt.colorbar(sc, ax=ax[1], label="temperatura (°C)")
ax[2].scatter(talento[~pasa][:800], esfuerzo[~pasa][:800], s=4, alpha=0.3, color="#94a3b8", label="no pasa"); ax[2].scatter(talento[pasa][:800], esfuerzo[pasa][:800], s=5, alpha=0.6, color=PALETA[1], label="pasa")
ax[2].set_xlabel("talento"); ax[2].set_ylabel("esfuerzo"); ax[2].set_title("Colisionador: la selección crea correlación", fontweight="bold", fontsize=10); ax[2].legend(fontsize=8)
plt.tight_layout(); plt.show()
````

## 4. La Paradoja de Simpson 🔄

La **paradoja de Simpson** ocurre cuando una tendencia que aparece en **cada subgrupo** se **invierte o desaparece** al **agregar** los datos. No es una paradoja lógica sino aritmética: los promedios agregados son **promedios ponderados** de los de cada subgrupo, y los **pesos** (cuánta gente hay en cada subgrupo en cada tratamiento) pueden estar muy desbalanceados:

$$\underbrace{\frac{\sum_k n_{Ak}\,\hat p_{Ak}}{\sum_k n_{Ak}}}_{\text{tasa global de }A}\quad\text{frente a}\quad\underbrace{\frac{\sum_k n_{Bk}\,\hat p_{Bk}}{\sum_k n_{Bk}}}_{\text{tasa global de }B},$$

con $n_{Ak}\neq n_{Bk}$ en proporción. Para comparar con justicia se **estandariza**: se aplican a ambos tratamientos **los mismos pesos** $w_k$ (por ejemplo, la composición de la población total): $\hat p_A^{\text{est}}=\sum_k w_k\hat p_{Ak}$.

**¿Cuál es la lectura correcta, la agregada o la estratificada?** **Depende de la historia causal**, no de la aritmética. Si la variable del subgrupo es un **confusor** (determina a quién se le aplica cada tratamiento y también el resultado, como el tipo de terreno), hay que **estratificar**. Si el subgrupo es una **consecuencia** del tratamiento (un *mediador*), estratificar puede ocultar el efecto real, y la lectura agregada es la correcta. Ninguna regla estadística reemplaza el conocimiento del contexto *(véase Pearl, 2014)*.

La celda construye una tabla **con números inventados** (dos métodos de riego, dos tipos de terreno, éxito = «la parcela alcanzó la meta de rendimiento»), la agrega, la estandariza, y muestra además la versión **continua**: tres grupos con pendiente positiva dentro de cada uno y pendiente **negativa** global.

````python
# ============================================================
# 4. Paradoja de Simpson: tabla (conteos inventados) y version continua
# ============================================================
tabla_riego = pd.DataFrame({
    "metodo": ["A", "A", "B", "B"],
    "terreno": ["plano", "ladera", "plano", "ladera"],
    "exitos": [36, 120, 204, 16],
    "total": [40, 240, 240, 40],
})
tabla_riego["tasa"] = tabla_riego["exitos"] / tabla_riego["total"]
por_terreno = tabla_riego.pivot(index="terreno", columns="metodo", values="tasa")
global_ = tabla_riego.groupby("metodo")[["exitos", "total"]].sum()
global_["tasa"] = global_["exitos"] / global_["total"]
print("Tasa de exito por terreno:\n", por_terreno.round(3).to_string())
print("\nTasa de exito GLOBAL:\n", global_.round(3).to_string())

pesos = tabla_riego.groupby("terreno")["total"].sum() / tabla_riego["total"].sum()          # composicion de TODA la poblacion
est_A = (por_terreno["A"] * pesos).sum(); est_B = (por_terreno["B"] * pesos).sum()
print(f"\nEstandarizando (mismos pesos): A = {est_A:.3f} | B = {est_B:.3f} | pesos = {pesos.round(2).to_dict()}")
assert (por_terreno["A"] > por_terreno["B"]).all()                    # A gana en CADA terreno...
assert global_.loc["A", "tasa"] < global_.loc["B", "tasa"]            # ...pero pierde en el agregado
assert est_A > est_B                                                  # la comparacion justa devuelve la ventaja a A
comp = tabla_riego.pivot(index="metodo", columns="terreno", values="total")
print("\nComposicion (cuantas parcelas de cada terreno recibio cada metodo):\n", comp.to_string())

# Version continua: tres grupos, pendiente positiva dentro, negativa global
rng = np.random.default_rng(SEED)
mu_x = np.array([2.0, 5.0, 8.0]); a_g = np.array([12.0, 8.0, 4.0])
gx = np.repeat(np.arange(3), 150)
x = rng.normal(mu_x[gx], 1.0)
y = a_g[gx] + 0.8 * (x - mu_x[gx]) + rng.normal(0, 0.8, x.size)
pend_global = np.polyfit(x, y, 1)[0]
pend_grupo = [np.polyfit(x[gx == k], y[gx == k], 1)[0] for k in range(3)]
print(f"\nPendiente global = {pend_global:.2f} | pendientes dentro de cada grupo = {np.round(pend_grupo, 2).tolist()}")
assert pend_global < 0 and all(p > 0 for p in pend_grupo)
assert pearson(x, y) < 0

fig, ax = plt.subplots(1, 2, figsize=(12, 4.3))
xs = np.arange(3); w = 0.35
for i, m in enumerate(["A", "B"]):
    ax[0].bar(xs + i * w - w / 2, [por_terreno.loc["plano", m], por_terreno.loc["ladera", m], global_.loc[m, "tasa"]], w, color=PALETA[i], label=f"método {m}")
ax[0].set_xticks(xs); ax[0].set_xticklabels(["plano", "ladera", "GLOBAL"]); ax[0].set_ylabel("tasa de éxito"); ax[0].set_ylim(0, 1)
ax[0].set_title("A gana en cada terreno y pierde en el global", fontweight="bold", fontsize=10); ax[0].legend()
for k in range(3):
    ax[1].scatter(x[gx == k], y[gx == k], s=8, alpha=0.5, color=PALETA[k], label=f"grupo {k + 1}")
    xx = np.linspace(x[gx == k].min(), x[gx == k].max(), 10); ax[1].plot(xx, np.polyval(np.polyfit(x[gx == k], y[gx == k], 1), xx), color=PALETA[k], lw=2)
xx = np.linspace(x.min(), x.max(), 10); ax[1].plot(xx, np.polyval(np.polyfit(x, y, 1), xx), "k--", lw=2, label="global")
ax[1].set_xlabel("x"); ax[1].set_ylabel("y"); ax[1].set_title("Pendiente positiva en cada grupo, negativa global", fontweight="bold", fontsize=10); ax[1].legend(fontsize=8)
plt.tight_layout(); plt.show()
````

## 5. Estadística Frente a *Machine Learning*: Explicar o Predecir 🤖

Breiman (2001) describió **dos culturas**: la que modela los datos para **inferir** el mecanismo que los generó, y la que usa algoritmos para **predecir** bien, tratando el mecanismo como una caja negra. No son enemigas; responden **preguntas distintas**:

| | **Inferencia (estadística clásica)** | **Predicción (aprendizaje automático)** |
|---|---|---|
| **Pregunta** | ¿Cómo y cuánto afecta $X$ a $Y$? ¿Con qué incertidumbre? | ¿Qué valor tendrá $Y$ para un caso nuevo? |
| **Producto** | Parámetros interpretables, **intervalos de confianza**, valores p | Predicciones y su **error en datos no vistos** |
| **Evaluación** | Supuestos del modelo, tamaños del efecto, diagnóstico de residuos | **Validación** con datos de prueba o validación cruzada |
| **Modelos típicos** | Regresión lineal/logística con *statsmodels* | Bosques aleatorios, *boosting*, redes con *scikit-learn* |
| **Riesgo principal** | Supuestos falsos, confusores, *p-hacking* | Sobreajuste, fuga de datos, deriva |

La **regresión lineal** sirve de puente. Con matriz de diseño $X$ (con columna de unos) el estimador de mínimos cuadrados es $\hat\beta=(X^\top X)^{-1}X^\top y$, con varianza estimada $\widehat{\operatorname{Var}}(\hat\beta)=\hat\sigma^2(X^\top X)^{-1}$ y $\hat\sigma^2=\sum e_i^2/(n-p)$. De ahí salen los errores estándar, los **intervalos de confianza** y los valores p que *statsmodels* informa; *scikit-learn* entrega los **mismos coeficientes** pero ni errores estándar ni valores p, porque su objetivo es predecir. Además **significancia estadística $\neq$ relevancia práctica**: con $n$ enorme, un efecto minúsculo es «significativo»; con $n$ pequeño, uno enorme puede no serlo.

| **Inferencia y predicción** | Elegir entre interpretar un modelo (regresión) o maximizar su precisión (bosque aleatorio) | 04, 06 |

La celda: **(a)** implementa los mínimos cuadrados desde cero y los contrasta con *statsmodels* y *scikit-learn*; **(b)** compara la **predicción** en datos de prueba cuando la verdad es lineal y cuando no lo es; **(c)** muestra un efecto **«significativo» pero irrelevante**.

````python
# ============================================================
# 5. Inferencia (statsmodels) frente a prediccion (scikit-learn), con MCO desde cero
# ============================================================
rng = np.random.default_rng(41)                                         # con otras semillas, x3 sale "significativo" ~5 % de las veces (el nivel alfa), p. ej. con 42
n = 1_000
X = rng.normal(size=(n, 3))
beta_real = np.array([2.0, 1.5, -1.0, 0.0])                            # intercepto, x1, x2, x3 (x3 NO tiene efecto)
y = beta_real[0] + X @ beta_real[1:] + rng.normal(0, 2.0, n)

# (a) MCO desde cero
X1 = np.column_stack([np.ones(n), X])
beta_hat = np.linalg.solve(X1.T @ X1, X1.T @ y)
res = y - X1 @ beta_hat
sigma2_hat = (res ** 2).sum() / (n - X1.shape[1])
ee_beta = np.sqrt(np.diag(sigma2_hat * np.linalg.inv(X1.T @ X1)))
t_val = beta_hat / ee_beta
p_val = 2 * stats.t.sf(np.abs(t_val), df=n - X1.shape[1])
mod = sm.OLS(y, X1).fit()
lin = LinearRegression().fit(X, y)
assert np.allclose(beta_hat, mod.params) and np.allclose(ee_beta, mod.bse) and np.allclose(p_val, mod.pvalues)    # = statsmodels
assert np.allclose(np.r_[lin.intercept_, lin.coef_], beta_hat)                                                    # = scikit-learn (coeficientes)
assert not hasattr(lin, "bse") and not hasattr(lin, "pvalues_")                                                   # sklearn no da EE ni p
ic = mod.conf_int()
tab = pd.DataFrame({"verdadero": beta_real, "estimado": beta_hat, "EE": ee_beta, "IC95 inf": ic[:, 0], "IC95 sup": ic[:, 1], "valor p": p_val}, index=["intercepto", "x1", "x2", "x3"])
print(tab.round(3).to_string())
assert all(ic[i, 0] <= beta_real[i] <= ic[i, 1] for i in range(4))                                                 # con esta semilla los IC contienen al verdadero (en el 95 % de las semillas lo haran)

# (b) Prediccion en datos de prueba: verdad lineal vs verdad no lineal
def comparar(X, y, nombre):
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=SEED)
    r2_lin = r2_score(yte, LinearRegression().fit(Xtr, ytr).predict(Xte))
    r2_rf = r2_score(yte, RandomForestRegressor(n_estimators=150, random_state=SEED, n_jobs=1).fit(Xtr, ytr).predict(Xte))
    return {"escenario": nombre, "R2 prueba (lineal)": r2_lin, "R2 prueba (bosque)": r2_rf}


y_nl = 3 * np.sin(X[:, 0]) + X[:, 1] ** 2 + rng.normal(0, 0.5, n)       # verdad NO lineal
comp = pd.DataFrame([comparar(X, y, "verdad lineal"), comparar(X, y_nl, "verdad no lineal")]).set_index("escenario")
print("\n", comp.round(3).to_string())
assert comp.loc["verdad lineal", "R2 prueba (lineal)"] >= comp.loc["verdad lineal", "R2 prueba (bosque)"]            # el modelo correcto gana (y ademas se interpreta)
assert comp.loc["verdad no lineal", "R2 prueba (bosque)"] > comp.loc["verdad no lineal", "R2 prueba (lineal)"] + 0.3  # cuando la forma es mas compleja, gana el flexible

# (c) Significativo pero irrelevante
n_g = 200_000
xg = rng.normal(size=n_g); yg = 0.01 * xg + rng.normal(size=n_g)
m_g = sm.OLS(yg, sm.add_constant(xg)).fit()
print(f"\n(c) n = {n_g:,}: pendiente = {m_g.params[1]:.4f} (IC95 {m_g.conf_int()[1, 0]:.4f} a {m_g.conf_int()[1, 1]:.4f}) | valor p = {m_g.pvalues[1]:.2e} | R^2 = {m_g.rsquared:.5f}")
assert m_g.pvalues[1] < 0.001 and m_g.rsquared < 0.001

fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
ax[0].errorbar(range(4), beta_hat, yerr=[beta_hat - ic[:, 0], ic[:, 1] - beta_hat], fmt="o", color=PALETA[0], capsize=5, label="estimado con IC 95 %")
ax[0].scatter(range(4), beta_real, marker="x", color=PALETA[1], s=70, label="verdadero", zorder=3)
ax[0].set_xticks(range(4)); ax[0].set_xticklabels(["intercepto", "x1", "x2", "x3"]); ax[0].set_title("Inferencia: coeficientes con incertidumbre", fontweight="bold", fontsize=10); ax[0].legend(fontsize=8)
comp.plot.bar(ax=ax[1], color=[PALETA[0], PALETA[2]], rot=0); ax[1].set_ylabel("R² en datos de prueba"); ax[1].set_title("Predicción: depende de la forma verdadera", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** Los mínimos cuadrados desde cero reproducen **exactamente** los coeficientes, errores estándar y valores p de *statsmodels*, y los coeficientes de *scikit-learn* (que **no** informa errores estándar ni valores p: está pensado para predecir). Los intervalos de confianza contienen a los valores verdaderos y el coeficiente de `x3` (sin efecto) es compatible con cero (con otras semillas, `x3` resulta «significativo» en ≈ 5 % de las ocasiones: justo el nivel $\alpha$, no un defecto del modelo). En **predicción**, el bosque no es mejor «por ser más moderno»: con **verdad lineal** el modelo lineal (correcto y además interpretable) iguala o supera al bosque; con una **verdad no lineal** el bosque gana por mucho, pero **sin** decirnos *cuánto* cambia $Y$ por unidad de $X$. Y el apartado (c) muestra que un valor p diminuto (con 200 000 datos) puede acompañar un efecto **irrelevante**: $R^2\approx 0.0001$. Por eso se reportan **tamaños del efecto e intervalos**, no solo «p < 0.05».

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Sesgo frente a variabilidad** | $\text{ECM}=\text{varianza}+\text{sesgo}^2$. Más datos reducen la **varianza** ($\propto 1/\sqrt{n}$), **no el sesgo**; el error total se aplana en el valor del sesgo |
| **Sesgos de muestreo** | Cobertura, no respuesta, autoselección, supervivencia y medición. La **ponderación** $1/\pi_i$ corrige solo lo que depende de variables **observadas** |
| **Variabilidad y múltiples pruebas** | Cuadruplicar $n$ divide el error entre 2. Con $m$ pruebas, $\text{FWER}=1-(1-\alpha)^m$ ($\approx 64\,\%$ con $m=20$); Bonferroni lo controla. El valor p **no** es la probabilidad de $H_0$ |
| **Correlación y causalidad** | $r$ mide asociación **lineal**; $r\approx 0$ no implica independencia. Un **confusor** crea asociación sin causa (se corrige con ajuste o **aleatorización**); un **colisionador** la crea al condicionar |
| **Paradoja de Simpson** | La tendencia de cada subgrupo puede invertirse al agregar: el agregado es un promedio **ponderado**. Se **estandariza** y se decide según la **historia causal** (confusor frente a mediador) |
| **Inferencia frente a predicción** | *statsmodels* da coeficientes con **incertidumbre**; *scikit-learn* optimiza el **error fuera de muestra**. El modelo correcto gana cuando la verdad es simple; el flexible, cuando no. **Significancia $\neq$ relevancia** |
