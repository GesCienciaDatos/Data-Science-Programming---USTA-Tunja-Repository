## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **«para no estadísticos»** del [cuaderno 00 del Módulo 00 (Introducción y resumen del curso)](../00_Introduccion_y_Resumen_del_Curso.ipynb). Imagina que cocinas una **olla enorme de sopa** y quieres saber si está bien de sal. ¿Te tomas toda la olla? No: **revuelves y pruebas una cucharada**. Si la cucharada sale bien de sal, **supones** que toda la olla también. Eso es, en el fondo, la estadística: **sacar conclusiones sobre el todo mirando solo una parte**, y saber **cuánto fiarse** de esa parte.

Al terminar podrás explicar:
1. Por qué **una cucharada cambia un poco cada vez**, y por qué **una cucharada más grande** se parece más a la olla.
2. Que **no todos los datos son iguales**: hay datos que se **cuentan**, que se **ordenan** o que se **promedian**, y cada uno pide un resumen distinto.
3. Por qué **a quién le preguntas** importa tanto como **cuántos** responden.

La sopa y los datos, lado a lado:

| En la cocina | En los datos |
|---|---|
| **La olla completa** | **La población**: todo lo que nos interesa (todos los costales, todas las familias) |
| **La cucharada** | **La muestra**: la parte que sí miramos |
| **El sabor real de la olla** | **El dato verdadero** que queremos conocer (casi nunca lo podemos ver completo) |
| **El sabor de la cucharada** | **Lo que medimos** en la muestra; cambia de cucharada en cucharada |
| **Revolver antes de probar** | **Escoger la muestra con cuidado** (al azar), para que no salga solo de un lado de la olla |

## 1. Probar la Sopa: Una Cucharada No Es la Olla 🥣

En la plaza de mercado hay **1 000 costales de papa**. Queremos saber **cuánto pesa en promedio un costal**, pero pesarlos todos tomaría el día entero. Entonces **pesamos solo unos pocos** (una «cucharada») y sacamos el promedio de esos.

Los pesos son **inventados** (para tener algo con qué jugar), pero el truco es real: como **nosotros** fabricamos los 1 000 costales, **sí conocemos el promedio verdadero** y podemos ver **qué tan cerca** queda cada cucharada. Probamos 40 cucharadas **chicas** (5 costales) y 40 **grandes** (50 costales).

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
costales = rng.normal(50, 4, 1000).round(1)          # peso en kg de 1 000 costales (datos INVENTADOS)
real = costales.mean()                                # el promedio verdadero de toda la olla


def cucharada(n):
    """Pesa n costales al azar y devuelve su promedio."""
    return rng.choice(costales, n, replace=False).mean()


chicas = np.array([cucharada(5) for _ in range(40)])      # 40 cucharadas de 5 costales
grandes = np.array([cucharada(50) for _ in range(40)])    # 40 cucharadas de 50 costales

print(f"Promedio verdadero de la olla: {real:.2f} kg")
print(f"Cucharadas chicas  (5 costales): van de {chicas.min():.1f} a {chicas.max():.1f} kg (amplitud {chicas.max() - chicas.min():.1f})")
print(f"Cucharadas grandes (50 costales): van de {grandes.min():.1f} a {grandes.max():.1f} kg (amplitud {grandes.max() - grandes.min():.1f})")
assert abs(chicas.mean() - real) < 1.5 and abs(grandes.mean() - real) < 0.5          # en promedio, las cucharadas aciertan
assert (grandes.max() - grandes.min()) < 0.6 * (chicas.max() - chicas.min())          # las grandes se parecen MUCHO mas entre si

fig, ax = plt.subplots(figsize=(8, 3.8))
ax.scatter(chicas, np.full(40, 1) + rng.uniform(-0.12, 0.12, 40), color="#dc2626", alpha=0.7, label="cucharadas de 5")
ax.scatter(grandes, np.full(40, 2) + rng.uniform(-0.12, 0.12, 40), color="#16a34a", alpha=0.7, label="cucharadas de 50")
ax.axvline(real, color="black", ls="--"); ax.text(real + 0.1, 2.45, "promedio verdadero", fontsize=9)
ax.set_yticks([1, 2]); ax.set_yticklabels(["5 costales", "50 costales"]); ax.set_ylim(0.4, 2.7); ax.set_xlabel("promedio de la cucharada (kg)")
ax.set_title("Cada cucharada sale distinta; las más grandes se parecen más a la olla", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** ninguna cucharada acierta **exactamente**, pero **en promedio caen alrededor del valor verdadero**. Con 5 costales el resultado **baila bastante** (más de 5 kg de amplitud entre la cucharada más baja y la más alta); con 50 costales **baila menos de la mitad** (≈ 2.6 kg). Más cucharada = menos «suerte». Pero ojo: no hace falta probar **toda** la olla; con una muestra bien escogida basta, y **saber cuánto puede fallar** es parte del trabajo del estadístico.

## 2. Los Datos No Son Todos Iguales: Contar, Ordenar o Promediar 📦

No todo lo que se anota en una libreta se puede **promediar**. Don José, el de la tienda, anota tres cosas de sus clientes:

- **El barrio** (Norte, Centro, Sur): son **etiquetas**. No hay orden ni «más» ni «menos». Lo único que tiene sentido es **cuál se repite más**.
- **La talla de ruana** (S, M, L): se **pueden ordenar** (S < M < L), pero la distancia entre S y M **no** es la misma que entre M y L. Sirve **la del medio**.
- **Los kilos de papa que compró**: son **números de verdad**. Aquí sí: sumar, restar y **sacar el promedio** tiene sentido.

> ⚠️ **Ojo:** que algo esté escrito con un número no lo vuelve un número «de verdad». Si numeras los barrios como 1, 2 y 3, **el promedio «2» no es ningún barrio**. Y **20 °C no es «el doble de caliente» que 10 °C**, porque el cero de los grados centígrados es un punto escogido por comodidad.

Hacemos las tres cuentas con ocho clientes de la tienda (inventados).

````python
tienda = pd.DataFrame({
    "barrio": ["Norte", "Centro", "Norte", "Sur", "Norte", "Centro", "Sur", "Norte"],
    "talla_ruana": ["S", "M", "S", "M", "M", "L", "M", "L"],
    "kilos_papa": [2.0, 5.5, 1.0, 3.0, 2.5, 6.0, 3.5, 1.5],
})
tallas = ["S", "M", "L"]                                           # el orden que SI tienen las tallas
codigos = tienda["talla_ruana"].map({t: i for i, t in enumerate(tallas)})

barrio_tipico = tienda["barrio"].mode()[0]                         # etiquetas -> la que mas se repite
talla_del_medio = tallas[int(np.median(codigos))]                  # ordenables -> la del medio
promedio_kilos = tienda["kilos_papa"].mean()                       # numeros de verdad -> promedio
print(tienda.to_string())
print(f"\nBarrio más repetido: {barrio_tipico} | talla del medio: {talla_del_medio} | promedio de kilos: {promedio_kilos:.3f}")
assert barrio_tipico == "Norte" and talla_del_medio == "M" and promedio_kilos == 25 / 8

# Los grados NO se pueden dividir: 20 C frente a 10 C
razon_c = 20 / 10
razon_k = (20 + 273.15) / (10 + 273.15)                            # en kelvin el cero es el cero de verdad
print(f"20 °C / 10 °C = {razon_c:.2f}, pero con el cero verdadero (kelvin) la razón es solo {razon_k:.3f}: no es 'el doble de caliente'.")
assert razon_c == 2 and abs(razon_k - 1.035) < 0.001
````

---
**Lección:** lo primero que hay que preguntarse con una columna de datos es **¿qué clase de dato es?** Con las **etiquetas** se cuenta; con lo **ordenable** se mira el del medio; con los **números de verdad** se promedia. Y algo escrito con números **no siempre** es un número «de verdad» (códigos de barrio, cédulas, el «1 a 5» de una encuesta de satisfacción).

## 3. ¿A Quién le Preguntas? La Puerta de la Tienda 🚪

Queremos saber **cuántos litros de leche al día** produce, en promedio, una familia de **tres veredas** (300 familias). Preguntarle a todas es caro. Tres maneras de escoger a 30:

1. **Los de la puerta:** las primeras 30 familias que se ven **desde la carretera** (las que quedan más cerca y son más fáciles de encontrar).
2. **Un sorteo entre todas:** como sacar 30 papelitos de un sombrero.
3. **Diez por vereda:** sorteamos 10 familias dentro de **cada** vereda.

Las familias cercanas a la carretera suelen estar mejor equipadas y producir más (**supuesto del ejemplo**, con datos inventados). Veamos qué pasa con cada método.

````python
rng = np.random.default_rng(7)
# Tres veredas de 100 familias: litros de leche por dia (INVENTADOS). La 1.a esta sobre la carretera, la 3.a lejos.
veredas = {"cerca de la carretera": rng.normal(12, 2, 100), "intermedia": rng.normal(10, 2, 100), "lejana": rng.normal(7, 2, 100)}
todas = np.concatenate(list(veredas.values()))
real_leche = todas.mean()


def de_la_puerta():
    return veredas["cerca de la carretera"][:30].mean()                                  # las 30 mas visibles


def sorteo_total():
    return rng.choice(todas, 30, replace=False).mean()


def diez_por_vereda():
    return np.mean([rng.choice(v, 10, replace=False).mean() for v in veredas.values()])


intentos = 300
res = pd.DataFrame({
    "promedio de los intentos (L/día)": [de_la_puerta(), np.mean([sorteo_total() for _ in range(intentos)]), np.mean([diez_por_vereda() for _ in range(intentos)])],
}, index=["los de la puerta", "sorteo entre todas", "10 por vereda"])
res["se aleja del verdadero"] = res["promedio de los intentos (L/día)"] - real_leche
print(f"Promedio verdadero de las 300 familias: {real_leche:.2f} L/día\n")
print(res.round(2).to_string())
assert abs(res.loc["los de la puerta", "se aleja del verdadero"]) > 1.0                    # la puerta se equivoca MUCHO
assert abs(res.loc["sorteo entre todas", "se aleja del verdadero"]) < 0.3                  # el sorteo acierta en promedio
assert abs(res.loc["10 por vereda", "se aleja del verdadero"]) < 0.3

fig, ax = plt.subplots(figsize=(7.5, 3.8))
ax.bar(res.index, res["promedio de los intentos (L/día)"], color=["#dc2626", "#16a34a", "#2563eb"])
ax.axhline(real_leche, color="black", ls="--"); ax.text(2.45, real_leche + 0.3, "promedio verdadero", fontsize=9, ha="right")
ax.set_ylabel("litros por día"); ax.set_ylim(0, 14); ax.set_title("Preguntar solo a los de la puerta sale caro: se equivoca siempre para el mismo lado", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()
````

## Resumen en una frase 🎯

> **La estadística nos deja conocer «la olla» probando solo «una cucharada»; sirve si la cucharada se escoge al azar, si usamos el resumen que le corresponde a cada tipo de dato y si sabemos cuánto puede fallar.**

### Lo que aprendiste hoy:
- ✅ **Población y muestra:** cada cucharada sale distinta, pero las **más grandes se parecen más** a la olla (amplitud de ≈ 5.6 kg con 5 costales frente a ≈ 2.6 kg con 50).
- ✅ **Tipos de datos:** **etiquetas** se cuentan, **lo ordenable** se mira por el medio y **los números de verdad** se promedian; 20 °C no es «el doble» de 10 °C.
- ✅ **Quién responde:** preguntar solo a «los de la puerta» se equivoca siempre para el mismo lado; el **sorteo** acierta en promedio.
