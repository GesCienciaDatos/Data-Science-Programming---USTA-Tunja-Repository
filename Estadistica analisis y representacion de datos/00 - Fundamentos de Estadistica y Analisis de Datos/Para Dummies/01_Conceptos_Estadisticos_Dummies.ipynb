Aquí vemos **tres trampas** muy comunes:

| Trampa | Cómo se ve | La pregunta que la desarma |
|---|---|---|
| **El azar parece un patrón** | «Salieron 7 caras de 10 monedas: ¡la moneda está cargada!» | ¿Con pocos datos esto pasa **seguido** por pura suerte? |
| **Dos cosas suben juntas, ¿una causa la otra?** | «Cuando se venden más ruanas, hay más gripa» | ¿Hay **una tercera cosa** (el frío) que mueve a las dos? |
| **El resultado de cada grupo dice una cosa y el total otra** | «La flota A es más puntual en cada tipo de día, pero el total dice que la B es mejor» | ¿**Cuántos** casos de cada tipo tiene cada uno? |

## 1. El Azar También Hace de las Suyas 🪙

Lanzas **10 monedas** y salen **7 caras o más**. ¿La moneda está cargada? Antes de acusarla, hagamos la prueba: **10 000 personas** lanzan 10 monedas **buenas** (50 % cara). ¿Cuántas ven 7 caras o más **solo por suerte**? Después repetimos con **1 000 lanzamientos** por persona.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
personas = 10_000
caras_10 = (rng.random((personas, 10)) < 0.5).sum(axis=1)              # caras en 10 lanzamientos de una moneda BUENA
caras_1000 = (rng.random((personas, 1000)) < 0.5).sum(axis=1)          # caras en 1 000 lanzamientos

seguido_10 = (caras_10 >= 7).mean()                                    # 7 caras o mas de 10
seguido_1000 = (caras_1000 >= 700).mean()                              # 70 % de caras o mas de 1 000
print(f"Con 10 lanzamientos, el {seguido_10:.1%} de las personas ve 7 o más caras (con una moneda perfecta).")
print(f"Con 1 000 lanzamientos, el {seguido_1000:.1%} ve 70 % de caras o más.")
assert abs(seguido_10 - 176 / 1024) < 0.015                            # exacto: 176/1024 = 17.2 %
assert seguido_1000 == 0

fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
ax[0].hist(caras_10 / 10, bins=np.arange(-0.05, 1.15, 0.1), color="#dc2626", alpha=0.8)
ax[0].set_title("10 lanzamientos: se ve de todo", fontweight="bold", fontsize=10); ax[0].set_xlabel("fracción de caras")
ax[1].hist(caras_1000 / 1000, bins=30, color="#16a34a", alpha=0.8); ax[1].set_xlim(0, 1)
ax[1].set_title("1 000 lanzamientos: casi siempre cerca de 0.5", fontweight="bold", fontsize=10); ax[1].set_xlabel("fracción de caras")
plt.tight_layout(); plt.show()
````

---
**Lección:** con **10 monedas buenas**, ver **7 o más caras le pasa a casi 1 de cada 6 personas** (≈ 17 %): no es señal de trampa. Con **1 000** lanzamientos, un 70 % de caras **ya no pasa nunca** por suerte. Moraleja: **con pocos datos, el azar produce resultados que parecen raros**; antes de decir «¡hay un patrón!» hay que preguntarse **cuántos datos hay** y **si eso pasaría seguido por pura suerte**.

## 2. Van Juntas, pero No Una Causa la Otra: Ruanas y Gripa 🧣

En 1 000 días de una ciudad fría (datos **inventados**) anotamos **cuántas ruanas se venden** y **cuántos casos de gripa** se reportan. Los dos números **suben y bajan juntos**. ¿Las ruanas dan gripa? No: **los días muy fríos** la gente **compra ruana** y, además, **se resfría más**. El **frío** es la tercera cosa que mueve a las dos.

¿Cómo comprobarlo? **Comparando días igual de fríos**: si dentro de un grupo de días con **el mismo frío** las dos cosas ya **no** se mueven juntas, entonces el frío lo explicaba todo. La «fuerza» con que dos cosas se mueven juntas se mide con un número entre -1 y 1 (la **correlación**): cerca de 1, suben y bajan juntas; cerca de 0, no hay relación en línea recta.

````python
rng = np.random.default_rng(3)
dias = 1000
frio = rng.normal(0, 1, dias)                                          # que tan frio estuvo el dia (la "tercera cosa")
ruanas = 50 + 10 * frio + rng.normal(0, 5, dias)                       # ruanas vendidas: dependen del frio
gripa = 30 + 6 * frio + rng.normal(0, 4, dias)                         # casos de gripa: tambien dependen del frio (NO de las ruanas)

fuerza_total = np.corrcoef(ruanas, gripa)[0, 1]
mismo_frio = np.abs(frio) < 0.25                                       # solo los dias "igual de frios" (un frio parecido)
fuerza_mismo_frio = np.corrcoef(ruanas[mismo_frio], gripa[mismo_frio])[0, 1]
print(f"Todos los días: ruanas y gripa se mueven juntas con fuerza {fuerza_total:.2f}.")
print(f"Solo {mismo_frio.sum()} días igual de fríos: la fuerza baja a {fuerza_mismo_frio:.2f}  (casi nada).")
assert fuerza_total > 0.6 and abs(fuerza_mismo_frio) < 0.2 and mismo_frio.sum() > 100

fig, ax = plt.subplots(figsize=(7.5, 4.2))
sc = ax.scatter(ruanas, gripa, c=frio, cmap="coolwarm_r", s=14)
ax.scatter(ruanas[mismo_frio], gripa[mismo_frio], facecolors="none", edgecolors="black", s=40, label="días igual de fríos")
plt.colorbar(sc, ax=ax, label="qué tan frío (más azul = más frío)")
ax.set_xlabel("ruanas vendidas"); ax.set_ylabel("casos de gripa"); ax.legend(loc="upper left")
ax.set_title("Se mueven juntas por el frío, no porque una cause la otra", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** ruanas y gripa **suben juntas** (fuerza ≈ 0.72), pero **comparando días igual de fríos** la relación **casi desaparece**. Cuando dos cosas se mueven juntas hay **cuatro explicaciones posibles**: que **una cause la otra**, que **la otra cause la una**, que **una tercera cosa mueva a las dos** (como el frío) o que sea **pura casualidad**. Para saber cuál es, hay que **pensar** y, mejor aún, **hacer un experimento** (repartir las ruanas al azar y mirar si cambia la gripa).

## 3. Cuando el Total Dice lo Contrario de Cada Grupo: Dos Flotas de Buses 🚌

Dos flotas, **Norte** y **Sur** (nombres **inventados**), registran cuántas veces llegaron **a tiempo** en **días secos** y en **días de lluvia** (cuando es más difícil ser puntual). Mira bien los números.

````python
buses = pd.DataFrame({
    "flota": ["Norte", "Norte", "Sur", "Sur"],
    "dia": ["seco", "lluvia", "seco", "lluvia"],
    "a_tiempo": [38, 180, 276, 20],
    "viajes": [40, 300, 300, 40],
})
buses["puntualidad"] = buses["a_tiempo"] / buses["viajes"]
por_dia = buses.pivot(index="dia", columns="flota", values="puntualidad")
total = buses.groupby("flota")[["a_tiempo", "viajes"]].sum()
total["puntualidad"] = total["a_tiempo"] / total["viajes"]
print("Puntualidad por tipo de día:\n", (por_dia * 100).round(1).to_string())
print("\nPuntualidad TOTAL:\n", (total["puntualidad"] * 100).round(1).to_string())
print("\nViajes de cada flota por tipo de día:\n", buses.pivot(index="flota", columns="dia", values="viajes").to_string())

assert (por_dia["Norte"] > por_dia["Sur"]).all()                       # Norte gana en CADA tipo de dia
assert total.loc["Norte", "puntualidad"] < total.loc["Sur", "puntualidad"]   # ...pero pierde en el total

fig, ax = plt.subplots(figsize=(7.5, 4))
x = np.arange(3); w = 0.35
for i, f in enumerate(["Norte", "Sur"]):
    ax.bar(x + i * w - w / 2, [por_dia.loc["seco", f], por_dia.loc["lluvia", f], total.loc[f, "puntualidad"]], w, color=["#2563eb", "#f59e0b"][i], label=f"flota {f}")
ax.set_xticks(x); ax.set_xticklabels(["día seco", "día de lluvia", "TOTAL"]); ax.set_ylabel("fracción de viajes a tiempo"); ax.set_ylim(0, 1.1); ax.legend()
ax.set_title("Norte gana en cada tipo de día y pierde en el total", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````
