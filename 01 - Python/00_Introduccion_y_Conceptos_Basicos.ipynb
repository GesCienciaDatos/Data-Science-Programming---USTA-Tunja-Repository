## ¿Qué es la Programación y el Pensamiento Computacional? 💻

La **programación** es el proceso de diseñar, codificar, depurar y mantener instrucciones precisas que una computadora ejecuta para resolver un problema o automatizar una tarea.

Para un científico de datos, programar no es simplemente memorizar sintaxis, sino desarrollar **pensamiento computacional**:
1. **Descomposición**: Dividir un problema complejo (ej. predecir la deserción de clientes) en subproblemas manejables (cargar datos, imputar nulos, entrenar modelo, evaluar métricas).
2. **Reconocimiento de Patrones**: Identificar tendencias, similitudes o regularidades en los datos o en los procesos algorítmicos.
3. **Abstracción**: Filtrar los detalles irrelevantes para concentrarse en las variables e hipótesis verdaderamente determinantes.
4. **Diseño de Algoritmos**: Establecer una secuencia lógica de pasos reproducibles para transformar entradas (*inputs*) en resultados (*outputs*).

## ¿Por qué Python es el Lenguaje Líder en Ciencia de Datos? 🐍

Creado por **Guido van Rossum** a principios de los años 90, Python se ha convertido en el estándar indiscutible de la industria para Ciencia de Datos, Inteligencia Artificial e Ingeniería de Datos gracias a:

- **Sintaxis Clara y Expresiva**: Diseñado bajo el principio de que *el código se lee muchas más veces de las que se escribe*. Su sintaxis cercana al pseudocódigo permite a los analistas centrarse en los datos y modelos, no en complejidades de bajo nivel.
- **Ecosistema Científico Incomparable**: Posee un conjunto cohesionado de bibliotecas de alto rendimiento escritas en C/C++/Fortran con interfaces limpias en Python:
  - *Computación Numérica*: **NumPy**, **SciPy**
  - *Manipulación de Datos*: **Pandas**, **Polars**
  - *Visualización*: **Matplotlib**, **Seaborn**, **Plotly**
  - *Machine Learning y Deep Learning*: **Scikit-Learn**, **PyTorch**, **TensorFlow**, **XGBoost**
- **Comunidad Global y Soporte Continuo**: Extensa documentación, miles de paquetes de código abierto a través de `PyPI` (`pip`) y soporte de las mayores organizaciones tecnológicas del mundo.

---
## Características Técnicas Clave ⚙️

- **Lenguaje Interpretado**: Se ejecuta instrucción por instrucción mediante el intérprete de Python (CPython), lo que posibilita flujos de trabajo altamente iterativos y exploración interactiva mediante Jupyter Notebooks.
- **Tipado Dinámico y Fuerte**: Las variables no requieren una declaración explícita de tipo de dato (el tipo se asocia al objeto en tiempo de ejecución), pero las operaciones incompatibles entre tipos distintos (como sumar un entero y un texto) no se convierten implícitamente, evitando errores silenciosos.
- **Multiparadigma**: Soporta programación imperativa, estructurada, funcional y orientada a objetos (POO).
- **Gestión Automática de Memoria**: Cuenta con un recolector de basura (*Garbage Collector*) basado en conteo de referencias y detección de ciclos, liberando al programador de la gestión manual de memoria.
