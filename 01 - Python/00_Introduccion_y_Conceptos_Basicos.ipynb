## ¿Qué es la programación? 🍳
*   La programación es el proceso de escribir instrucciones que una computadora puede ejecutar para realizar una tarea específica.
*   Implica dividir un problema en pasos más pequeños y manejables, creando una secuencia de instrucciones o algoritmos que pueden resolver el problema.
*   Un algoritmo es un conjunto de instrucciones que una computadora puede seguir para completar una tarea.
*   Se puede comparar con una receta de cocina, donde cada paso es una instrucción específica que, seguida en orden, produce el resultado deseado.
*   En programación, los algoritmos se escriben utilizando lenguajes como Python, que nos permiten escribir instrucciones que la máquina puede entender y ejecutar.

## Python **🐍**

Python es un lenguaje de programación de propósito general, de alto nivel y tipado dinámico, que se ha consolidado como el **lenguaje rey en el ecosistema de la Ciencia de Datos e Inteligencia Artificial**. 

Su popularidad no es casualidad, se basa en los siguientes pilares:

* **Legibilidad y Sintaxis Limpia:** Su código fue diseñado para leerse de forma casi idéntica al idioma inglés, reduciendo drásticamente la curva de aprendizaje en comparación con lenguajes como C++ o Java.
* **Ecosistema Gigante (Librerías):** Posee una colección inigualable de paquetes altamente optimizados para datos, como Pandas, NumPy, Scikit-Learn y TensorFlow.
* **Comunidad Inmensa:** Si tienes un problema o un error, alguien más en el mundo ya lo resolvió en plataformas como StackOverflow.
* **Versatilidad:** Sirve tanto para entrenar una red neuronal compleja como para automatizar el envío de un correo electrónico o construir un servidor web.

La instrucción más simple en Python es la directiva `print` - simplemente imprime una línea en pantalla.

### 🤓 Características Técnicas de Python
* **Lenguaje Interpretado:** Python ejecuta las instrucciones línea por línea directamente en el procesador sin necesidad de ser "compilado" previamente (a diferencia de C). Esto facilita inmensamente la fase de experimentación típica de la Ciencia de Datos.
* **Tipado Dinámico:** No necesitas declararle a Python si una variable contendrá texto o números, él lo infiere automáticamente al vuelo.
* **Multiparadigma:** Permite programar de forma procedural (paso a paso), funcional y orientada a objetos.
* **Open Source:** Es completamente gratuito, de código abierto y su desarrollo es sostenido por la comunidad (Python Software Foundation).
