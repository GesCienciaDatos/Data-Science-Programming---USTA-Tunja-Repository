# Colecciones en Python: Listas, Tuplas y Conjuntos 🚀

Manejar múltiples datos es el pan de cada día en Ciencia de Datos. En este notebook dominaremos las tres colecciones (estructuras de datos) más importantes que complementan a los diccionarios: **Listas, Tuplas y Conjuntos (Sets)**, desde lo más básico hasta trucos de rendimiento y comprensión.

---

## 1. Nivel Básico: Las Listas (Mutables y Ordenadas) 🌱

Las listas son colecciones ordenadas que pueden cambiar en el tiempo (son *mutables*). Permiten datos duplicados y de diferentes tipos.

````python
# Creación
frutas = ["Manzana", "Banana", "Cereza"]
mixta = [1, "Texto", True, 3.14]

# Acceso por índice
print("Primera fruta:", frutas[0])
print("Última fruta:", frutas[-1])

# Mutabilidad (Cambiar, añadir, eliminar)
frutas[1] = "Kiwi"             # Modificar
frutas.append("Naranja")       # Añadir al final
frutas.insert(1, "Pera")       # Insertar en una posición específica

eliminada = frutas.pop()       # Elimina y retorna el último elemento
frutas.remove("Manzana")       # Elimina por valor (falla si no existe)

print("\nLista actual:", frutas)
print("Fruta eliminada con pop:", eliminada)
````

## 2. Nivel Intermedio: Las Tuplas (Inmutables) y el Desempaquetado 🛠️

Las tuplas son como las listas, pero **no se pueden modificar** una vez creadas (*inmutables*). Son más rápidas, seguras para datos fijos y pueden usarse como claves en diccionarios.

````python
# Creación (Se usan paréntesis, aunque son opcionales si hay comas)
coordenadas = (4.5, -74.0)
punto_3d = 10, 20, 30  # Esto también es una tupla!

# Acceso (Igual que las listas)
print("Latitud:", coordenadas[0])

# Intentar modificar arroja error (TypeError):
# coordenadas[0] = 5.0 

# Desempaquetado (Unpacking)
x, y, z = punto_3d
print(f"\nDesempaquetado -> X:{x}, Y:{y}, Z:{z}")

# Desempaquetado avanzado usando *
primero, *medio, ultimo = (1, 2, 3, 4, 5)
print("\nPrimero:", primero)
print("Medio (se vuelve lista):", medio)
print("Último:", ultimo)
````

## 3. Nivel Avanzado: Los Conjuntos (Sets) y Valores Únicos 🧠

Un *Set* es una colección **no ordenada** de elementos **únicos**. Son perfectos para eliminar duplicados de una lista de forma instantánea.

````python
# Creación (Se usan llaves como los diccionarios, pero sin formato clave:valor)
colores = {"Rojo", "Azul", "Verde", "Rojo", "Azul"}

# Los duplicados se eliminan automáticamente
print("Conjunto de colores:", colores)

# Convertir lista a conjunto (Truco para eliminar duplicados)
ids_usuarios = [101, 102, 101, 103, 104, 102]
ids_unicos = set(ids_usuarios)
print("\nIDs Únicos:", ids_unicos)

# Modificar conjuntos
ids_unicos.add(105)        # Añadir un elemento
ids_unicos.discard(101)    # Eliminar (no da error si no existe, a diferencia de .remove())
print("IDs Modificados:", ids_unicos)
````

## 4. Nivel Pro: Teoría de Conjuntos y Búsqueda Ultrarrápida 💎

Los Sets permiten operaciones matemáticas súper eficientes. Además, buscar si un elemento existe (`in`) en un Set es de orden **O(1)** (casi instantáneo sin importar el tamaño), mientras que en una lista es **O(N)**.

````python
grupo_a = {"Juan", "Ana", "Luis", "Maria"}
grupo_b = {"Pedro", "Ana", "Carlos", "Juan"}

# Unión (Todos los elementos sin repetir) -> Operador |
print("Unión:", grupo_a | grupo_b)

# Intersección (Elementos en común) -> Operador &
print("Intersección:", grupo_a & grupo_b)

# Diferencia (Elementos de A que no están en B) -> Operador -
print("Diferencia (A - B):", grupo_a - grupo_b)

# Diferencia Simétrica (Elementos que están en uno u otro, pero no en ambos) -> Operador ^
print("Diferencia Simétrica:", grupo_a ^ grupo_b)

# Eficiencia de Búsqueda (Se recomienda usar Sets si vas a hacer muchos 'in')
print("\n¿Está Carlos en el Grupo A?", "Carlos" in grupo_a)
````

## 5. Nivel Experto: List Comprehensions, zip y enumerate 🏆

Escribir código *Pythónico* implica usar List Comprehensions (una línea) en lugar de largos bucles `for`, e iterar eficientemente con las funciones nativas `zip()` y `enumerate()`.

````python
# List Comprehension: [resultado for elemento in iterable if condicion]
numeros = [1, 2, 3, 4, 5, 6]
cuadrados_pares = [x**2 for x in numeros if x % 2 == 0]
print("Comprehension (Cuadrados de pares):", cuadrados_pares)

# Enumerate: Iterar obteniendo el índice y el valor simultáneamente
nombres = ["Ana", "Juan", "Luis"]
print("\n--- Enumerate ---")
for indice, nombre in enumerate(nombres, start=1):
    print(f"{indice}. {nombre}")

# Zip: Iterar dos (o más) listas en paralelo
edades = [28, 30, 25]
print("\n--- Zip ---")
for nombre, edad in zip(nombres, edades):
    print(f"{nombre} tiene {edad} años")
    
# Truco Maestro: Convertir dos listas en un diccionario en 1 línea usando zip
diccionario_personas = dict(zip(nombres, edades))
print("\nDiccionario creado con zip:", diccionario_personas)
````
