## ¿Qué es Feature Engineering? ⚙️

La **Ingeniería de Características (*Feature Engineering*)** es el proceso de transformar datos crudos (*raw data*) en representaciones y formatos numéricos y estructurados que faciliten y optimicen el aprendizaje de los algoritmos de Machine Learning.

Comprende tres actividades principales:
1. **Crear nuevas variables (*Feature Creation*)**: Construir nuevos indicadores combinando o transformando atributos existentes para capturar mejor las relaciones del dominio del problema.
2. **Transformar variables (*Feature Transformation*)**: Adaptar tipos de datos, escalas, distribuciones y representaciones (por ejemplo, codificar variables cualitativas o reducir dimensionalidad).
3. **Seleccionar variables relevantes (*Feature Selection*)**: Identificar el subconjunto óptimo de características que aportan señal predictiva y descartar aquellas que introducen ruido o redundancia.

> *«Al final del día, algunos proyectos de Machine Learning tienen éxito y otros fracasan. ¿Qué marca la diferencia? El factor más importante, con diferencia, son las características utilizadas.»*
> 
> — **Pedro Domingos**, *A Few Useful Things to Know about Machine Learning*

---
## ¿Por qué es fundamental en Ciencia de Datos? 🎯

Aunque los algoritmos modernos (como Gradient Boosting o Redes Neuronales) son capaces de aproximar patrones no lineales complejos, **la calidad, estructura y pertinencia de los datos de entrada limitan el rendimiento máximo alcanzable** (principio de *Garbage In, Garbage Out*).

Un buen proceso de Feature Engineering permite:
- **Mejorar la precisión del modelo**: Descubre relaciones no lineales que algoritmos lineales simples no pueden detectar por sí mismos.
- **Reducir la complejidad y el sobreajuste (*Overfitting*)**: Eliminar variables ruidosas permite entrenar modelos más generalizables y rápidos.
- **Facilitar la interpretabilidad**: Características bien diseñadas permiten explicar a partes interesadas del negocio el motivo detrás de las predicciones.
- **Aprovechar el conocimiento experto del dominio**: Traducir reglas de negocio, leyes físicas o comportamientos económicos en variables numéricas concretas.
