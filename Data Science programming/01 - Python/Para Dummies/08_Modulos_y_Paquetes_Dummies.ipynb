## 1. La caja de herramientas del carpintero 🧰

Imagina un carpintero que empieza un mueble nuevo. No fabrica el martillo, el destornillador ni la sierra desde cero cada vez — los saca de su **caja de herramientas**, ya construidos y listos para usar.

En Python pasa exactamente lo mismo:

- Un **módulo** es como una herramienta individual: un archivo `.py` que alguien ya escribió, con funciones y datos listos para usar (por ejemplo, el módulo `math` trae funciones matemáticas).
- Un **paquete** es como la caja completa: una carpeta que agrupa varios módulos relacionados (por ejemplo, `pandas` agrupa muchos módulos para trabajar con tablas de datos).

> 🧠 **Para recordar:** si tuvieras que reinventar el martillo cada vez que quieres clavar un clavo, nunca terminarías el mueble. Los módulos existen para que tú te concentres en el mueble (tu proyecto), no en fabricar las herramientas.

## 2. ¿Cómo le pido a Python que me preste una herramienta? 🙋

La palabra mágica es `import`. Pero así como puedes pedir prestada *toda* la caja de herramientas o *solo el martillo*, Python te da varias formas de importar:

| Forma | Analogía | Ejemplo |
|---|---|---|
| `import modulo` | Traes la caja completa. Para usar algo, dices "de la caja, dame el martillo": `modulo.herramienta()` | `import math` → `math.sqrt(9)` |
| `from modulo import herramienta` | Sacas solo el martillo de la caja y lo dejas sobre tu mesa. Ya no dices "de la caja" — solo `herramienta()` | `from math import sqrt` → `sqrt(9)` |
| `import modulo as apodo` | Le pones un apodo corto a la caja porque su nombre es largo. | `import pandas as pd` → `pd.DataFrame()` |
| `from modulo import *` | Sacas **todas** las herramientas de la caja y las riegas sobre la mesa, sin etiquetas. ⚠️ No se recomienda: es fácil que se mezclen con tus propias cosas. | `from math import *` |

En la próxima celda vamos a probar la primera forma: importar un módulo completo de la librería estándar de Python (las herramientas que ya vienen incluidas, sin instalar nada).

````python
# Importamos el módulo 'random' completo (viene incluido en Python, no se instala nada)
import random

# Como lo importamos completo, usamos la notación de punto: modulo.herramienta()
numero_azar = random.randint(1, 100)
print(f'Tu número de la suerte es: {numero_azar}')
````

### 🤔 ¿Qué acaba de pasar?

- `import random` trajo la caja completa llamada `random` (el módulo que genera números al azar).
- Para usar cualquier herramienta de esa caja escribimos primero el nombre de la caja: `random.randint(...)`.
- Si vuelves a ejecutar la celda, el número cambia — porque es aleatorio.

````python
# Ahora probamos 'from ... import ...' y el alias 'as'
from math import pi          # sacamos SOLO el número pi de la caja 'math'
import datetime as dt        # le ponemos un apodo corto al módulo 'datetime'

radio = 5
area_circulo = pi * radio ** 2
print(f'El área del círculo es: {area_circulo:.2f}')

hoy = dt.date.today()
print(f'Hoy es: {hoy}')
````

### 🤔 ¿Qué acaba de pasar?

- `from math import pi` sacó únicamente el valor `pi` de la caja `math` — por eso lo usamos directo, sin escribir `math.pi`.
- `import datetime as dt` le puso el apodo `dt` al módulo `datetime`, así que escribimos `dt.date.today()` en vez del nombre completo.
- Ambas formas son válidas: la eliges según qué tan larga sea la caja y cuántas herramientas de ella necesites.

## 3. ¿Y si la herramienta que necesito no viene incluida? 🛒

Python trae "pilas incluidas": módulos como `math`, `random` o `datetime` ya vienen listos. Pero para tareas de Ciencia de Datos casi siempre necesitas paquetes que **no** vienen por defecto, como `pandas` o `numpy`. A esos primero hay que "comprarlos" (instalarlos) con una herramienta llamada `pip`, y solo después puedes importarlos.

```bash
pip install pandas
```

```python
import pandas as pd
```

> 💬 **Analogía:** la librería estándar es la caja de herramientas que ya viene con la casa cuando la compras. `pip install` es ir a la ferretería a comprar una herramienta especializada que no venía incluida.

## 4. Construyendo tu propia caja de herramientas 🧱

También puedes ser tú quien fabrica el módulo. Cualquier archivo `.py` que tú escribas puede ser importado por otro archivo tuyo, tal como importas `math` o `random`.

Vamos a crear un archivo llamado `operaciones.py` con una función adentro, y luego lo vamos a "importar" como si fuera un módulo cualquiera.

````python
%%writefile operaciones.py
# Contenido del archivo operaciones.py
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b
````

````python
# Ahora importamos NUESTRO propio módulo, igual que importaríamos 'math'
import operaciones

resultado_suma = operaciones.sumar(10, 5)
resultado_resta = operaciones.restar(10, 5)

print(f'10 + 5 = {resultado_suma}')
print(f'10 - 5 = {resultado_resta}')
````

### 🤔 ¿Qué acaba de pasar?

- `%%writefile operaciones.py` es un truco especial de Jupyter para guardar el contenido de la celda en un archivo real llamado `operaciones.py`, en la misma carpeta del cuaderno.
- Una vez existe ese archivo, Python lo trata exactamente igual que a `math` o `random`: `import operaciones` lo trae, y `operaciones.sumar(...)` usa la función que definiste adentro.
- Esta es la base de cómo se organizan los proyectos reales: en vez de un solo archivo gigante, se reparte el código en varios archivos pequeños que se importan entre sí.

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Módulo | Un archivo `.py` con funciones y datos listos para usar — una herramienta individual. |
| Paquete | Una carpeta con varios módulos relacionados — la caja completa de herramientas. |
| `import modulo` | Traes la caja completa; usas `modulo.herramienta()`. |
| `from modulo import x` | Sacas solo una herramienta específica; la usas directo como `x()`. |
| `import modulo as apodo` | Le pones un apodo corto al módulo. |
| `from modulo import *` | Trae todo sin etiquetas — evítalo, puede chocar con tus propias variables. |
| Librería estándar | Módulos que ya vienen incluidos en Python (`math`, `random`, `datetime`...). |
| `pip install` | Cómo "compras" paquetes que no vienen incluidos (`pandas`, `numpy`...). |
