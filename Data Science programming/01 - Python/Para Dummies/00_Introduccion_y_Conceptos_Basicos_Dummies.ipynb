## 1. Programar es como escribir una receta de cocina 👩‍🍳

Imagina que le explicas a alguien, paso a paso, cómo preparar arroz:
1. Pon agua a hervir.
2. Añade el arroz.
3. Espera 15 minutos.
4. Retira del fuego y sirve.

Eso es exactamente lo que es **programar**: escribir instrucciones muy precisas, en un orden exacto, para que "alguien" (en este caso, el computador) las siga sin ambigüedad. La diferencia es que el computador no "intuye" nada — si te saltas un paso o lo escribes mal, no cocina el arroz.

### Los 4 trucos mentales de cualquier programador (pensamiento computacional)

| Truco | En la cocina | En Ciencia de Datos |
|---|---|---|
| **Descomponer** | Dividir "hacer el almuerzo" en "hacer arroz" + "hacer ensalada" + "poner la mesa" | Dividir "predecir el precio de una casa" en "conseguir datos" + "limpiarlos" + "entrenar un modelo" |
| **Reconocer patrones** | Notar que siempre que el arroz se pega, usaste poca agua | Notar que las casas con más habitaciones tienden a costar más |
| **Abstraer** | No te importa la marca de la olla, solo que hierva agua | No te importa el nombre exacto del cliente, solo las variables que explican su comportamiento |
| **Diseñar el algoritmo** | Escribir la receta completa, en orden | Escribir el código que ejecuta cada paso del análisis, en orden |

> 🧠 **Para recordar:** un algoritmo es simplemente "una receta bien escrita". Si puedes explicarle una receta a alguien sin dejar nada a la imaginación, ya piensas como programador.

## 2. ¿Por qué Python y no otro idioma? 🐍

Existen cientos de lenguajes de programación, igual que existen cientos de idiomas humanos. Python se volvió el favorito de la Ciencia de Datos por razones muy concretas:

- 🗣️ **Se lee casi como español/inglés simple.** Comparado con otros lenguajes, Python evita símbolos raros. Escribir `if edad >= 18:` se lee casi tal cual se dice en voz alta.
- 🧰 **Viene con "cajas de herramientas" ya hechas.** No tienes que inventar la rueda: `NumPy` hace cuentas con números, `Pandas` organiza tablas de datos, `Matplotlib` dibuja gráficas — tú solo las usas.
- 🌍 **Millones de personas ya lo usan y ayudan.** Si te atoras, casi cualquier error que tengas ya lo tuvo otra persona antes, y la solución está documentada en internet.
- ⚡ **Puedes probar cosas al instante.** No necesitas "compilar" nada complicado — escribes una línea, la ejecutas, ves el resultado. Por eso los cuadernos Jupyter (como este) funcionan tan bien para aprender.

> 💬 **Analogía:** si programar en general es como cocinar, Python es como cocinar con una cocina que ya tiene todos los electrodomésticos enchufados y las recetas más comunes ya escritas en tarjetitas — tú solo las sigues y las combinas.

## 3. Tres palabras técnicas que sonarán mucho (explicadas sin miedo) ⚙️

En el cuaderno principal viste una tabla con características "técnicas" de Python. Aquí las traducimos:

| Palabra técnica | Lo que realmente significa | Ejemplo cotidiano |
|---|---|---|
| **Interpretado** | El computador lee y ejecuta tu código línea por línea, como alguien leyendo una receta en voz alta mientras cocina — no necesita "traducir todo el libro" antes de empezar. | Vas leyendo la receta y cocinando a la vez, no memorizas el libro completo primero. |
| **Tipado dinámico** | No tienes que avisarle a Python de antemano si una variable va a guardar un número o una palabra — él lo descubre solo, en el momento. | Una caja de cartón sin etiqueta: hoy guardas ahí manzanas, mañana guardas papeles. La caja no te obliga a decidir de antemano. |
| **Tipado fuerte** | Python sí te va a **reclamar** si intentas mezclar cosas que no tiene sentido mezclar (como sumar un número con una palabra). No hace conversiones "raras" a escondidas. | Si intentas sumar "3 manzanas" + "una silla", cualquier persona sensata te dice que eso no se puede sumar así como así. Python piensa igual. |

Vamos a **comprobar esto último con código real** más abajo, para que no se quede solo en teoría.

## 4. Verifiquemos que todo esté listo 🛠️

Antes de escribir código "de verdad", confirmemos que el entorno (Python + este cuaderno) están funcionando. Ejecuta la siguiente celda con `Shift + Enter`.

```python
import sys
import platform

print("✅ Entorno de Python configurado correctamente.")
print(f"🐍 Versión de Python: {platform.python_version()}")
print(f"💻 Sistema Operativo: {platform.system()} {platform.release()}")
```

## 5. Tu primera instrucción de verdad 📢

Vamos a guardar información en "cajas con nombre" (esto se llama **variables**) y luego mostrarla.

```python
mensaje = '¡Bienvenido a la programación en Ciencia de Datos!'
estudiantes = 25
nota_aprobacion = 3.5

print(f'📢 {mensaje}')
print(f'👥 Número de estudiantes inscritos: {estudiantes}')
print(f'🎯 Nota mínima para aprobar: {nota_aprobacion}')
```

### 🤔 ¿Qué acaba de pasar?

- `mensaje`, `estudiantes` y `nota_aprobacion` son **variables**: cajas con nombre donde guardamos un valor para usarlo después.
- Nota que `mensaje` guarda texto (entre comillas), `estudiantes` guarda un número entero, y `nota_aprobacion` guarda un número con decimales. **Nunca le dijimos a Python qué tipo de dato iba en cada caja** — él lo dedujo solo. Eso es el "tipado dinámico" del que hablamos arriba, ya en acción.

```python
# Esta celda genera un error A PROPÓSITO — es parte del aprendizaje, no un bug.
# Descomenta la siguiente línea y ejecútala para ver el mensaje de error de Python:

# resultado = "5 estudiantes" + 3
```

Si descomentas y ejecutas la línea anterior, Python te devuelve un `TypeError` explicando que no puede sumar un texto (`str`) con un número (`int`). **Esto es una ventaja, no un problema**: Python prefiere detenerse y avisarte del error, en vez de adivinar y darte un resultado incorrecto sin que te des cuenta.
