## 1. Los datos crudos son como ingredientes sin pelar 🥔

Imagina que quieres cocinar un puré de papa. Si le entregas al cocinero las papas **tal como salieron de la tierra** — con cáscara, con tierra pegada, sin cortar — el resultado será malo, sin importar qué tan buena sea la olla o la estufa.

Antes de cocinar, alguien tiene que **pelar, lavar y cortar** las papas. Eso es justo lo que hace la **Ingeniería de Características (Feature Engineering)**, pero con datos: toma información "cruda" (columnas de texto, fechas sin procesar, números en escalas raras) y la transforma en columnas limpias, numéricas y bien organizadas — listas para que el modelo de Machine Learning (la "estufa") pueda cocinar algo bueno.

> 📌 **Para recordar:** Feature Engineering = pelar y cortar los ingredientes antes de cocinar. El modelo (la estufa) solo puede ser tan bueno como los ingredientes que le des.

---
## 2. ¿Por qué le damos tanta importancia? 🏆

Un experto muy reconocido en Machine Learning, Andrew Ng, dijo algo parecido a esto:

> *"Aplicar Machine Learning es, en el fondo, hacer Feature Engineering. Es la parte más difícil, más creativa, y la que realmente separa un modelo mediocre de uno ganador."*

Con la analogía de la cocina: puedes tener la mejor estufa del mundo (el algoritmo más avanzado), pero si tus ingredientes están mal preparados, la comida sabrá mal. En cambio, con ingredientes bien preparados, hasta una estufa sencilla puede producir un plato delicioso.

En la práctica, esto significa que **mejorar tus datos de entrada** suele dar más beneficio que saltar a un modelo más complicado.

## 3. Los 4 grandes trucos que veremos en esta serie 🧰

| Truco | En una frase | Ejemplo cotidiano |
|---|---|---|
| Codificar categorías | Convertir palabras (ej. "rojo", "azul") en números que el modelo pueda entender | Traducir un menú a otro idioma para que el cocinero lo entienda |

## 5. Un ejemplo pequeño: crear una columna nueva más útil 🧮

Imagina una tabla con casas: su área en metros cuadrados y su precio. El precio total por sí solo no nos dice si una casa es "cara" o "barata" comparada con otras — depende de su tamaño. Vamos a crear una nueva columna, el **precio por metro cuadrado**, que es mucho más comparable entre casas de tamaños distintos. Esto es Feature Engineering: no conseguimos datos nuevos, simplemente **combinamos los que ya teníamos** de una forma más informativa.

```python
casas = pd.DataFrame({
    "barrio": ["El Prado", "El Prado", "La Colina", "La Colina", "El Prado"],
    "area_m2": [45, 120, 60, 200, 80],
    "precio_millones": [180, 420, 210, 640, 300]
})

# Feature nueva: precio por metro cuadrado
casas["precio_por_m2"] = casas["precio_millones"] / casas["area_m2"]

casas
```

### 🤔 ¿Qué acaba de pasar?

- Creamos una tabla (`DataFrame`) con 5 casas: su barrio, su área y su precio.
- `casas["precio_millones"] / casas["area_m2"]` divide, fila por fila (de forma vectorizada, ¿recuerdas el módulo de NumPy?), el precio entre el área.
- La nueva columna `precio_por_m2` nos permite comparar casas de tamaños distintos de forma justa: dos casas pueden costar lo mismo en total, pero una puede ser mucho más cara "por metro" que la otra.
- Combinar columnas con una operación matemática es una de las formas más simples y más usadas de Feature Engineering.

## 6. ¿Y la columna `barrio`? 🏘️

Fíjate que la columna `barrio` contiene texto ("El Prado", "La Colina"), no números. La mayoría de los modelos de Machine Learning **no entienden texto** directamente — solo números.
