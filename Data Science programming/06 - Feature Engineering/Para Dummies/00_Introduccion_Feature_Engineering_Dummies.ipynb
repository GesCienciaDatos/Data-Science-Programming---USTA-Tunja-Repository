## ¿Qué es Feature Engineering y por qué es el Superpoder de la Ciencia de Datos? ⚡

**Feature Engineering (Ingeniería de Características)** es el proceso de transformar datos crudos en variables (features) que hagan que los algoritmos de Machine Learning aprendan mejor y más rápido.

### 🍳 La Analogía del Chef Gourmet
- **Datos crudos:** Traer papas con tierra, carne sin cortar y cebolla entera. Si metes todo directo al horno, el resultado será incomible.
- **Feature Engineering:** Pelar las papas, cortar la carne en medallones, caramelizar la cebolla y sazonar en proporciones ideales. Ahora cualquier horno básico cocinará un platillo de 5 estrellas.

> *"Mejores variables superan a mejores algoritmos casi siempre." — Proverbio de Machine Learning.*

### Tipos Principales de Transformaciones
1. **Codificación Categórica:** Convertir texto en números comprensibles (One-Hot, Ordinal, Target Encoding).
2. **Creación de Ratios e Interacciones:** Combinar variables (ej. Precio / Metro Cuadrado).
3. **Transformaciones No Lineales:** Logaritmo, raíz cuadrada para des-sesgar datos.
4. **Reducción de Dimensionalidad:** Comprimir 50 variables en las 3 más informativas (PCA).
5. **Selección:** Quedarnos solo con las variables que realmente aportan señal y no ruido.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Dataset de ejemplo: Inmuebles sin preparar vs preparados
np.random.seed(42)
n = 100
area = np.random.uniform(40, 200, n)
habitaciones = np.random.randint(1, 5, n)
antiguedad = np.random.uniform(0, 30, n)
precio = 50000000 + area * 3500000 + habitaciones * 15000000 - antiguedad * 2000000 + np.random.normal(0, 20000000, n)

df_casas = pd.DataFrame({
    'Area_m2': area,
    'Habitaciones': habitaciones,
    'Antiguedad_anios': antiguedad,
    'Precio_COP': precio
})

print('📋 Dataset Crudo:')
display(df_casas.head())
```

```python
# Creamos nuevas características inteligentes
df_fe = df_casas.copy()
df_fe['Precio_por_m2'] = df_fe['Precio_COP'] / df_fe['Area_m2']
df_fe['Area_por_Habitacion'] = df_fe['Area_m2'] / df_fe['Habitaciones']
df_fe['Es_Nuevo'] = (df_fe['Antiguedad_anios'] <= 3).astype(int)
df_fe['Log_Precio'] = np.log1p(df_fe['Precio_COP'])

print('✨ Dataset con Feature Engineering aplicado:')
display(df_fe.head())

print('\n📈 Correlaciones con el Precio:')
print(df_fe.corr()['Precio_COP'].sort_values(ascending=False).round(3))
```

> **🔑 Regla de Oro**
>
> Nunca hagas Feature Engineering sobre todo el dataset a la vez si involucra estadísticas globales (medias, varianzas, imputaciones).
> ¡Debes separar primero en Entrenamiento (Train) y Prueba (Test) para evitar la temida fuga de datos (Data Leakage)!
