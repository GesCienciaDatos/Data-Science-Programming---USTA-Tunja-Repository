## 1. El clóset lleno de cosas inútiles 🗄️

Imagina un clóset enorme donde guardas de todo: la ropa que sí usas, pero también cosas que nunca te pones, cables sueltos sin saber para qué sirven, y cajas vacías. Cuando necesitas encontrar algo rápido, ese desorden **te hace perder tiempo**, aunque en teoría "más cosas" debería significar "más opciones".

A un modelo de Machine Learning le pasa algo parecido con las columnas de datos: agregar columnas *informativas* ayuda, pero agregar columnas de puro ruido (que no tienen relación real con lo que quieres predecir) **empeora** el modelo — le cuesta más entrenar, tarda más, y puede terminar "aprendiendo" patrones falsos que solo existen por casualidad en esos datos de ruido. A este fenómeno se le llama la **maldición de la dimensionalidad**.

> 📌 **Para recordar:** el rendimiento de un modelo mejora al agregar columnas útiles, pero después de cierto punto, agregar más columnas (sobre todo si son ruido) empieza a perjudicarlo.

---
## 2. Tres maneras de decidir qué guardar y qué botar 🗂️

Existen tres grandes familias de métodos para seleccionar características. Aquí va la versión simplificada de cada una:

| Familia | Idea central | Analogía |
|---|---|---|
| **Filtro (Filter)** | Mide cada columna con una fórmula estadística, **sin entrenar ningún modelo**. Rápido y simple. | Revisar el clóset y descartar a simple vista lo que claramente no sirve (rotos, sin uso). |
| **Envoltura (Wrapper)** | Prueba distintos subconjuntos de columnas **entrenando el modelo una y otra vez**, y se queda con el que mejor funcione. Preciso, pero lento. | Probarte distintas combinaciones de ropa completas para ver cuál luce mejor — funciona, pero toma mucho tiempo. |
| **Integrado (Embedded)** | El propio algoritmo decide qué columnas usar **mientras se entrena** (ej. Lasso, árboles de decisión). | Un asesor de imagen que va armando tu outfit y descartando prendas sobre la marcha, todo en un mismo proceso. |

En este cuaderno nos concentraremos en el método de filtro más usado al inicio de cualquier proyecto: la **Información Mutua**.

---
## 3. Información Mutua: el juego de las 20 preguntas 🕵️

Imagina el juego de "20 preguntas": alguien piensa un objeto y tú debes adivinarlo haciendo preguntas de Sí/No. Si la primera pregunta reduce muchísimo las posibilidades (ej. "¿es un animal?"), es una **muy buena** pregunta. Si apenas reduce las opciones, es una pregunta **poco útil**.

La **Información Mutua (Mutual Information, MI)** hace exactamente eso, pero con columnas de datos: mide **cuánta incertidumbre sobre tu variable objetivo se reduce** si conoces el valor de una columna.

- $MI = 0$: la columna no te dice absolutamente nada sobre el objetivo (como preguntar algo totalmente irrelevante).
- $MI$ alto: conocer esa columna reduce muchísimo la incertidumbre (como una gran pregunta de 20 preguntas).

La gran ventaja de la Información Mutua frente a la correlación de toda la vida es que la correlación **solo** detecta relaciones en línea recta, mientras que la Información Mutua detecta **cualquier tipo de relación**: curvas, forma de "U", patrones periódicos, lo que sea.

```python
import numpy as np
import pandas as pd
from sklearn.feature_selection import mutual_info_regression

np.random.seed(1)
n = 200

# Una columna SI relacionada con el precio: el tamano de la casa
tamano_m2 = np.random.normal(120, 30, n)

# Una columna que NO tiene ninguna relacion real con el precio (puro ruido)
variable_ruido = np.random.normal(0, 1, n)

# El precio depende del tamano (con algo de ruido natural), pero NO de "variable_ruido"
precio = tamano_m2 * 3.5 + np.random.normal(0, 15, n)

casas = pd.DataFrame({
    "tamano_m2": tamano_m2,
    "variable_ruido": variable_ruido,
    "precio": precio,
})

X = casas[["tamano_m2", "variable_ruido"]]
y = casas["precio"]

mi_scores = mutual_info_regression(X, y, random_state=0)
mi_scores = pd.Series(mi_scores, index=X.columns, name="MI Score").sort_values(ascending=False)
mi_scores
```

### 🤔 ¿Qué acaba de pasar?

- Construimos a propósito dos columnas: `tamano_m2`, que sí determina el precio, y `variable_ruido`, que es puro azar sin ninguna relación con `precio`.
- `mutual_info_regression(X, y)` calcula la Información Mutua entre cada columna de `X` y el objetivo `y`.
- El resultado debería mostrar `tamano_m2` con un puntaje claramente más alto que `variable_ruido` — la Información Mutua "descubrió" cuál columna es realmente útil, exactamente como esperaríamos.

---
## 4. La ventaja frente a la correlación: relaciones en forma de "U" 📈

Ahora veamos el verdadero superpoder de la Información Mutua: detectar relaciones que **no** son una línea recta, algo que la correlación de Pearson no puede hacer.

Imagina una columna $x$ y un objetivo $y = x^2$ (una parábola perfecta, en forma de "U"). Si $x$ va de valores negativos a positivos, la correlación lineal sale casi en cero (porque no hay una tendencia "hacia arriba" ni "hacia abajo" consistente), aunque en realidad $x$ determina completamente a $y$.

```python
x = np.linspace(-3, 3, 60)
ruido = np.random.normal(0, 0.3, size=x.shape)
y_cuadratica = x ** 2 + ruido

correlacion = np.corrcoef(x, y_cuadratica)[0, 1]
mi_score = mutual_info_regression(x.reshape(-1, 1), y_cuadratica, random_state=0)[0]

print(f"Correlacion de Pearson (relacion lineal): {correlacion:.3f}")
print(f"Informacion Mutua (cualquier tipo de relacion): {mi_score:.3f}")
```

### 🤔 ¿Qué acaba de pasar?

- Creamos `y_cuadratica = x**2 + ruido`: una relación perfectamente clara, pero en forma de parábola, no de línea recta.
- La **correlación de Pearson** debería salir muy cerca de 0, "engañándonos" y sugiriendo que `x` no tiene relación con `y_cuadratica` — un error grave si solo mirásemos ese número.
- La **Información Mutua**, en cambio, debería salir claramente por encima de 0, confirmando que sí existe una relación fuerte (aunque no sea una línea recta).
- **Lección clave:** antes de descartar una columna solo porque su correlación es baja, vale la pena revisar su Información Mutua — podría estar escondiendo una relación en forma de curva.

## 5. Cuidado con la fuga de datos (data leakage) 🛡️

Igual que con las transformaciones agrupadas del cuaderno anterior, la selección de características debe hacerse **solo con los datos de entrenamiento**. Si calculas la Información Mutua (o cualquier otro método de selección) usando también los datos de prueba, esos datos de prueba habrán "influido" en qué columnas elegiste, y tus resultados de evaluación serán engañosamente optimistas.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import SelectKBest

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

# Definimos el selector con Informacion Mutua y le pedimos que se quede con 1 columna
selector = SelectKBest(
    score_func=lambda X_, y_: mutual_info_regression(X_, y_, random_state=0),
    k=1,
)

# Ajustamos (fit) el selector SOLO con los datos de entrenamiento
selector.fit(X_train, y_train)

columna_elegida = X_train.columns[selector.get_support()]
print("Columna seleccionada usando solo entrenamiento:", list(columna_elegida))

# Aplicamos la MISMA decision (ya aprendida) tanto a entrenamiento como a prueba
X_train_sel = selector.transform(X_train)
X_test_sel = selector.transform(X_test)
print("Forma de X_train_sel:", X_train_sel.shape)
print("Forma de X_test_sel:", X_test_sel.shape)
```

### 🤔 ¿Qué acaba de pasar?

- `train_test_split` separa los datos en entrenamiento (70%) y prueba (30%), tal como hicimos en cuadernos anteriores del curso.
- `selector.fit(X_train, y_train)` calcula la Información Mutua y decide qué columna conservar **usando únicamente los datos de entrenamiento** — el conjunto de prueba nunca participa en esa decisión.
- `selector.transform(...)` aplica esa misma decisión (ya tomada) tanto a `X_train` como a `X_test`, sin recalcular nada con los datos de prueba.
- El resultado debería confirmar que `tamano_m2` fue la columna elegida, descartando `variable_ruido` — y lo hicimos de la forma correcta, sin fuga de datos.

## 6. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Maldición de la dimensionalidad | Agregar columnas de puro ruido empeora el modelo, aunque en teoría "más datos" suene mejor. |
| Métodos de Filtro | Miden cada columna con una fórmula estadística, sin entrenar ningún modelo — rápidos y simples. |
| Métodos de Envoltura (Wrapper) | Prueban subconjuntos de columnas entrenando el modelo repetidamente — precisos, pero lentos. |
| Métodos Integrados (Embedded) | El propio algoritmo (ej. Lasso, árboles) decide qué columnas usar mientras se entrena. |
| Información Mutua | Mide cuánta incertidumbre sobre el objetivo se reduce al conocer una columna — detecta relaciones lineales y no lineales. |
| Fuga de datos | La selección de características se ajusta solo con los datos de entrenamiento, nunca con los de prueba. |
