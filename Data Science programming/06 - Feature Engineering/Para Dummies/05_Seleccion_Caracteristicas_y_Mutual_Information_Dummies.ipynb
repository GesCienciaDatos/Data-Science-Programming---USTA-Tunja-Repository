## Empacar la maleta: menos es más 🎒

Cuando empacas una maleta para un viaje corto, no te llevas todo tu clóset — eliges solo lo que realmente vas a usar. Llevar de más solo hace la maleta más pesada y difícil de cargar, sin ayudarte en nada.

Con los datos pasa lo mismo. Meter 200 columnas a un modelo no lo hace más inteligente: lo hace **más lento**, más propenso a **memorizar ruido** en vez de aprender patrones reales (*sobreajuste*), y más difícil de interpretar. **Seleccionar características** es el proceso de decidir, con criterios objetivos, cuáles columnas realmente vale la pena conservar.

---
## 1. La maldición de la dimensionalidad 📉

Buscar a un amigo perdido en un pueblo pequeño es fácil: pocas calles, pocos lugares donde mirar. Buscarlo en una ciudad gigantesca, con millones de rincones posibles, es muchísimo más difícil — aunque tengas más "espacio" para explorar, en la práctica te pierdes.

Algo parecido les pasa a los modelos: agregar características ayuda **al principio**, pero después de cierto punto, cada columna extra agrega más "espacio vacío" y ruido que información útil, y el rendimiento del modelo empieza a **empeorar**. A este fenómeno se le llama la **maldición de la dimensionalidad**.

---
## 2. Tres formas de elegir qué conservar 🗂️

| Familia | ¿Cómo decide? | Velocidad | Analogía |
|---|---|---|---|
| **Filtro (*Filter*)** | Mide qué tan relacionada está cada columna con el objetivo, sin entrenar ningún modelo | ⚡ Muy rápida | Descartar ropa que "obviamente" no combina, antes de empacar |
| **Envolvente (*Wrapper*)** | Prueba distintos subconjuntos de columnas entrenando modelos una y otra vez | 🐢 Lenta | Armar la maleta completa, pesarla, y ajustar quitando/poniendo cosas |
| **Integrado (*Embedded*)** | El propio modelo, mientras aprende, le baja el peso a las columnas poco útiles | ⚖️ Media | Una maleta inteligente que descarta sola lo que no usas |

En este cuaderno nos enfocamos en el método de filtro más popular: la **Información Mutua**.

---
## 3. Información mutua: el juego de las veinte preguntas 🕵️

Imagina que estás jugando "veinte preguntas" para adivinar un número entre 1 y 1,000. Sin ninguna pista, necesitas muchas preguntas. Pero si alguien te dice **"el número es par"**, de inmediato descartas la mitad de las posibilidades — te ahorraste varias preguntas.

La **información mutua** mide exactamente eso: **cuánta incertidumbre sobre la variable objetivo se reduce al conocer el valor de una característica**.

- Si `MI = 0`, esa pista no te ayuda absolutamente nada — la característica y el objetivo son independientes.
- Entre más alto el valor de `MI`, más "preguntas te ahorras" al conocer esa columna — es decir, más útil es para predecir.

A diferencia de la correlación (que solo detecta relaciones en línea recta), la información mutua detecta **cualquier tipo de relación**: lineal, curva, en forma de U, lo que sea.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import mutual_info_classif

cancer = load_breast_cancer()
X = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y = cancer.target

# Calculamos la información mutua de cada columna respecto al diagnóstico
mi_scores = mutual_info_classif(X, y, random_state=42)
mi_series = pd.Series(mi_scores, index=X.columns).sort_values(ascending=True)

plt.figure(figsize=(8, 5))
mi_series.tail(8).plot(kind='barh', color='#0284c7', edgecolor='white')
plt.title('Las 8 características más informativas (diagnóstico de cáncer)')
plt.xlabel('Puntaje de Información Mutua')
plt.grid(True, alpha=0.3)
plt.show()
```

### 🤔 ¿Qué acaba de pasar?

- `mutual_info_classif(X, y)` calculó, para cada una de las 30 columnas del dataset, cuánto "reduce la incertidumbre" sobre si un tumor es benigno o maligno.
- Ordenamos los resultados y graficamos las 8 columnas con mayor puntaje: son las candidatas más fuertes para quedarse en el modelo.
- Una columna con puntaje cercano a cero podría eliminarse sin perder casi nada de poder predictivo — igual que dejar en casa la ropa que nunca usas en un viaje.

---
## 4. Cuidado: una columna "aburrida" puede ser clave en pareja 🧂💧

La sal sola no es comida. El agua sola tampoco. Pero juntas, en las proporciones correctas, son la base de casi cualquier receta. Con las características pasa algo parecido: una columna puede tener información mutua **individual** muy baja, y aun así ser decisiva cuando se combina con otra.

Veámoslo con un ejemplo pequeño: dos variables que, por separado, no dicen casi nada sobre el objetivo, pero que combinadas (con una regla tipo "si ambas son altas, o ambas son bajas") sí lo predicen perfectamente.

```python
np.random.seed(1)
n = 500
x1 = np.random.choice([0, 1], size=n)
x2 = np.random.choice([0, 1], size=n)

# El objetivo depende de la COMBINACION de x1 y x2 (tipo XOR), no de cada una por separado
y_interaccion = (x1 == x2).astype(int)

X_interaccion = pd.DataFrame({'x1': x1, 'x2': x2})
mi_individual = mutual_info_classif(X_interaccion, y_interaccion, discrete_features=True, random_state=42)

print('Información mutua de cada variable POR SEPARADO:')
print(pd.Series(mi_individual, index=['x1', 'x2']).round(3))
```

### 🤔 ¿Qué acaba de pasar?

- Construimos a propósito un objetivo (`y_interaccion`) que depende de si `x1` y `x2` son **iguales o distintas** — una relación de pareja, no de una sola columna.
- Al medir la información mutua de `x1` y `x2` por separado, ambas salen con un puntaje muy bajo (cercano a cero): ¡parecería que ninguna sirve!
- Pero en realidad, juntas explican el objetivo **perfectamente**. Por eso la información mutua (que evalúa una columna a la vez) es un excelente primer filtro, pero nunca debe ser la única herramienta: siempre vale la pena combinarla con el conocimiento del problema antes de descartar una variable para siempre.

---
## 5. No hagas trampa: selecciona solo con el conjunto de entrenamiento 🚫🔓

Igual que con el resto de la ingeniería de características, seleccionar columnas usando **todo** el dataset (entrenamiento + prueba) antes de dividirlo es una forma de fuga de datos (*data leakage*): el conjunto de prueba "opina" sobre qué columnas conservar, y eso infla artificialmente qué tan bien luce tu modelo.

La regla es simple: primero divides, y el ranking de características se calcula **solo** con los datos de entrenamiento.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import SelectKBest

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Seleccionamos las 5 columnas mas informativas usando SOLO el entrenamiento
selector = SelectKBest(score_func=mutual_info_classif, k=5)
selector.fit(X_train, y_train)

columnas_elegidas = X_train.columns[selector.get_support()]
print('Columnas seleccionadas (calculado solo con X_train):')
print(list(columnas_elegidas))

# Aplicamos la MISMA seleccion, ya decidida, al conjunto de prueba
X_test_reducido = selector.transform(X_test)
print(f'\nForma de X_test tras aplicar la seleccion: {X_test_reducido.shape}')
```

### 🤔 ¿Qué acaba de pasar?

- `selector.fit(X_train, y_train)` calcula el ranking de información mutua **solo** con los datos de entrenamiento — el conjunto de prueba nunca participa en esta decisión.
- `selector.transform(X_test)` simplemente aplica esa misma selección (ya decidida) al conjunto de prueba, sin recalcular nada con él.
- Este patrón — decidir con entrenamiento, aplicar sobre prueba — es el mismo que debes usar para cualquier paso de selección o creación de características que dependa de estadísticas del dataset.

## 6. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Selección de características | Elegir qué columnas conservar para no cargar al modelo con ruido |
| Maldición de la dimensionalidad | Agregar columnas ayuda... hasta que empieza a perjudicar |
| Filtro / Wrapper / Embebido | Tres familias de métodos, de más rápida a más costosa |
| Información mutua | Cuánta incertidumbre sobre el objetivo se reduce al conocer una columna |
| Interacciones | Dos columnas "aburridas" por separado pueden ser clave juntas |
| Selección sin fuga de datos | Calcula el ranking solo con entrenamiento; aplícalo luego a prueba |
