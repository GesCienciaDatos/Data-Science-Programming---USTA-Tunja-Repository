## 1. Adivinar un número, no una categoría 🔢

Imagina que un amigo te pide dos favores distintos:

- **Favor A:** "Adivina cuánto va a costar mi cuenta del restaurante." → la respuesta es un **número** que puede ser casi cualquier valor: $32,450, $58,900, $101,200...
- **Favor B:** "Adivina si voy a pedir pollo, carne o pasta." → la respuesta es una **categoría** de una lista cerrada: {pollo, carne, pasta}.

El Favor A es un problema de **Regresión**: predecir una cantidad continua (un número que puede tomar casi cualquier valor decimal). El Favor B es un problema de **Clasificación**, y lo verás más adelante en el curso.

> 📌 **Para recordar:** si la pregunta es "¿cuánto?" o "¿cuál será el valor de...?", casi siempre es Regresión. Si la pregunta es "¿cuál de estas opciones?", es Clasificación.

Este módulo completo trata sobre cómo enseñarle a la computadora a "adivinar números" a partir de ejemplos pasados — eso es exactamente lo que hace la **Regresión Lineal**, el modelo más famoso de todos.

## 2. La Analogía de la Cuenta del Restaurante 🍽️

Piensa en un restaurante donde, por experiencia, sabes más o menos cómo se calcula la cuenta: hay un cargo fijo por la mesa (el "cubierto") y luego un valor aproximado por cada persona que se sienta a comer.

Si el cubierto cuesta $10,000 COP y cada persona gasta en promedio $25,000 COP:

$$\text{Cuenta} = 10000 + 25000 \times (\text{Comensales})$$

Esa fórmula — un valor fijo más un valor que crece proporcionalmente con otra cantidad — es exactamente la idea detrás de la **Regresión Lineal Simple**. En el mundo real nunca conocemos esos $10,000 y $25,000 de antemano: los **aprendemos** observando muchas cuentas reales de mesas con diferente número de comensales.

Vamos a simular justo eso: generamos cuentas "reales" (con algo de variación aleatoria, como pasa en la vida real) y le pedimos a la computadora que **encuentre** esa fórmula por sí sola.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

np.random.seed(42)
comensales = np.array([2, 3, 4, 5, 6, 8, 10, 12]).reshape(-1, 1)
cuenta_cop = 10000 + 25000 * comensales.flatten() + np.random.normal(0, 15000, len(comensales))

modelo = LinearRegression().fit(comensales, cuenta_cop)
x_pred = np.linspace(1, 14, 100).reshape(-1, 1)
y_pred = modelo.predict(x_pred)

plt.figure(figsize=(8, 5))
plt.scatter(comensales, cuenta_cop, color='#ef4444', s=80, label='Cuentas Reales Observadas')
plt.plot(x_pred, y_pred, color='#2563eb', linewidth=2.5, label=f'Línea Ajustada: {modelo.coef_[0]:,.0f}x + {modelo.intercept_:,.0f}')
plt.xlabel('Número de Comensales')
plt.ylabel('Cuenta Total (COP)')
plt.title('Regresión Lineal: Prediciendo la Cuenta', fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### 🤔 ¿Qué acaba de pasar?

- `comensales` guarda el número de personas en cada mesa observada (2, 3, 4... hasta 12).
- `cuenta_cop` simula las cuentas reales: usamos la fórmula "10,000 + 25,000 por persona" pero le sumamos ruido aleatorio (`np.random.normal`), porque en la vida real nunca dos mesas con el mismo número de comensales pagan exactamente lo mismo.
- `LinearRegression().fit(comensales, cuenta_cop)` es la parte mágica: le mostramos a `scikit-learn` los pares (comensales, cuenta) y él **encuentra solo** los valores del cubierto (`intercept_`) y el gasto por persona (`coef_`) que mejor explican esos datos.
- La línea azul del gráfico es la "fórmula aprendida"; los puntos rojos son las cuentas reales observadas. El objetivo del modelo es que la línea pase lo más cerca posible de todos los puntos.

## 3. ¿Por qué no usar siempre modelos súper complicados? 💡

Hoy existen modelos de Machine Learning mucho más sofisticados que la regresión lineal (redes neuronales, bosques de árboles, etc.). Sin embargo, la Regresión sigue siendo una de las herramientas más usadas en Ciencia de Datos por razones muy prácticas:

1. **Se explica fácil:** puedes decirle a tu jefe "por cada comensal adicional, la cuenta sube en promedio $25,000" — un modelo de caja negra no te da esa frase tan clara.
2. **Es rapidísima:** entrenar y usar una regresión toma fracciones de segundo, incluso con miles de datos.
3. **Es el punto de partida obligatorio:** antes de usar un modelo complicado, primero pruebas con regresión. Si el modelo complicado no le gana claramente a la regresión, ¡no vale la pena la complicación extra!

```python
# Usamos el modelo ya entrenado para predecir la cuenta de una mesa nueva
comensales_nuevos = np.array([[7]])
cuenta_estimada = modelo.predict(comensales_nuevos)

print(f"Para una mesa de {comensales_nuevos[0][0]} comensales, el modelo estima una cuenta de ${cuenta_estimada[0]:,.0f} COP")
print(f"Fórmula aprendida: Cuenta = {modelo.intercept_:,.0f} + {modelo.coef_[0]:,.0f} x Comensales")
```

### 🤔 ¿Qué acaba de pasar?

- `modelo.predict(...)` usa la fórmula ya aprendida para calcular la cuenta de una mesa que **nunca vio** durante el entrenamiento (7 comensales).
- `modelo.intercept_` es el "cubierto" que la computadora dedujo de los datos (debería estar cerca de $10,000, aunque con ruido nunca será exacto).
- `modelo.coef_[0]` es el "gasto por persona" que dedujo (debería estar cerca de $25,000).
- Esto es justo lo que hace la regresión en cualquier problema real: aprende una fórmula con datos del pasado y la usa para **predecir** casos nuevos.

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Regresión | Predecir un número (cantidad continua), no una categoría. |
| Aprendizaje Supervisado | El modelo aprende viendo ejemplos pasados donde ya conocemos la respuesta correcta. |
| `LinearRegression().fit(X, y)` | Le pide a scikit-learn que encuentre la mejor línea (o fórmula) que explica los datos. |
| `intercept_` | El valor fijo de la fórmula (el "cubierto" del restaurante). |
| `coef_` | Cuánto cambia la predicción por cada unidad de la variable de entrada. |
| `.predict(...)` | Usa la fórmula aprendida para estimar el valor de casos nuevos. |
