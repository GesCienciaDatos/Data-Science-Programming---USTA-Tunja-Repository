## 1. El estudiante que memoriza el examen 📚

Imagina dos estudiantes preparándose para un examen de matemáticas:

- **Estudiante A** memoriza, palabra por palabra, las 20 preguntas del examen de práctica, incluyendo hasta los errores de tipeo del profesor. En el examen de práctica saca 100%. Pero el examen real tiene preguntas distintas... y este estudiante reprueba, porque no entendió los conceptos, solo memorizó.
- **Estudiante B** ni siquiera estudia. No entiende los conceptos ni memoriza nada. Reprueba tanto el examen de práctica como el real.

En Machine Learning pasa lo mismo:
- El **Estudiante A** es un modelo con **sobreajuste (*overfitting*)**: le va increíble con los datos de entrenamiento (los "vio" y los memorizó) pero le va mal con datos nuevos.
- El **Estudiante B** es un modelo con **subajuste (*underfitting*)**: es tan simple que ni siquiera logra aprender el patrón, así que le va mal en ambos casos.

Lo que buscamos siempre es un término medio: un modelo que **entienda el patrón general** sin memorizar el ruido de los datos de entrenamiento.

## 2. El balance sesgo-varianza 🎯

Volvamos al arquero del cuaderno anterior, pero ahora imagina que dispara varias rondas de flechas:

- **Alto sesgo (Bias):** todas las flechas caen agrupadas, pero lejos del centro — el arquero apunta mal de forma consistente. Esto es un modelo **demasiado simple** (como el Estudiante B): siempre se equivoca de la misma manera.
- **Alta varianza (Variance):** las flechas caen en cualquier parte del blanco, muy dispersas — el arquero es inconsistente. Esto es un modelo **demasiado flexible/complejo** (como el Estudiante A): funciona distinto según qué datos usó para entrenar.

El objetivo ideal es un arquero con **bajo sesgo y baja varianza**: flechas agrupadas justo en el centro. En la práctica, mejorar uno de los dos casi siempre empeora un poco el otro — a eso se le llama el **compromiso (trade-off) sesgo-varianza**.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

np.random.seed(42)
X = np.linspace(-3, 3, 60).reshape(-1, 1)
y = 0.5 * X.flatten() ** 2 + np.random.normal(0, 1.5, 60)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

for grado in [1, 2, 15]:
    modelo = make_pipeline(PolynomialFeatures(grado), LinearRegression())
    modelo.fit(X_train, y_train)
    r2_train = r2_score(y_train, modelo.predict(X_train))
    r2_test = r2_score(y_test, modelo.predict(X_test))
    print(f"Grado {grado:2d} -> R² Entrenamiento: {r2_train:.3f} | R² Prueba: {r2_test:.3f}")
```

### 🤔 ¿Qué acaba de pasar?

- Generamos datos con una curva real (cuadrática) más ruido aleatorio.
- Probamos tres modelos de distinta complejidad: una línea recta (grado 1, demasiado simple), una parábola (grado 2, la complejidad correcta) y una curva de grado 15 (extremadamente flexible).
- Fíjate en el patrón: el modelo de grado 15 debería tener un R² de entrenamiento altísimo (memorizó el ruido) pero un R² de prueba mucho peor que el de grado 2 — la señal clásica de **sobreajuste**.
- El modelo de grado 1 debería tener un R² bajo en ambos conjuntos — la señal clásica de **subajuste**, porque una línea recta no puede capturar una curva.
- El modelo de grado 2 es el que mejor equilibrio logra, porque coincide con la forma real de los datos.

## 3. Cuando dos variables se parecen demasiado: Multicolinealidad 👯

Imagina que quieres predecir cuánto gasta una familia al mes usando dos variables: "ingreso mensual en pesos" e "ingreso mensual en dólares". Son básicamente la misma información contada de dos formas distintas (una es la otra multiplicada por la tasa de cambio).

Si le pides al modelo que reparta el "crédito" del efecto del ingreso entre esas dos variables casi idénticas, el modelo se confunde: no sabe si darle todo el peso a una, a la otra, o repartirlo — y pequeños cambios en los datos pueden hacer que el resultado cambie mucho. A esto se le llama **Multicolinealidad**: dos o más variables predictoras que están muy correlacionadas entre sí.

> 📌 **Importante:** la multicolinealidad casi no afecta qué tan buena es la predicción final, pero sí arruina la interpretación — ya no puedes confiar en el peso individual de cada variable "gemela".

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

np.random.seed(0)
n = 200
ingreso_pesos = np.random.normal(3_000_000, 500_000, n)
ingreso_dolares = ingreso_pesos / 4000 + np.random.normal(0, 5, n)  # casi la misma variable
edad = np.random.randint(20, 65, n)

df_demo = pd.DataFrame({'ingreso_pesos': ingreso_pesos, 'ingreso_dolares': ingreso_dolares, 'edad': edad})

vif = pd.DataFrame({
    'Variable': df_demo.columns,
    'VIF': [variance_inflation_factor(df_demo.values, i) for i in range(df_demo.shape[1])]
})
print(vif)
```

### 🤔 ¿Qué acaba de pasar?

- Creamos `ingreso_dolares` como (casi) una copia matemática de `ingreso_pesos` — a propósito, para simular dos variables "gemelas".
- El **VIF (Factor de Inflación de la Varianza)** mide, para cada variable, qué tanto se puede predecir a partir de las demás. Cuanto más alto, más "redundante" es esa variable frente al resto.
- Como regla práctica: un VIF cercano a 1 es perfecto (la variable aporta información única), y un VIF mayor a 5 o 10 es una señal de alerta de multicolinealidad problemática.
- Nota cómo `ingreso_pesos` e `ingreso_dolares` deberían tener un VIF disparado (porque una casi determina a la otra), mientras que `edad` — que no tiene relación con el ingreso — debería tener un VIF bajo, cercano a 1.

## 4. ¿Qué hago si detecto estos problemas? 🛠️

| Problema | Señal | Qué hacer |
|---|---|---|
| Sobreajuste | R² de entrenamiento mucho mayor que el de prueba | Simplificar el modelo, conseguir más datos, o usar regularización (lo verás en el próximo cuaderno). |
| Subajuste | R² bajo tanto en entrenamiento como en prueba | Usar un modelo más flexible, o agregar variables/curvas que capturen mejor el patrón. |
| Multicolinealidad | VIF alto (mayor a 5) en una o más variables | Eliminar una de las variables "gemelas", combinarlas en una sola, o usar regularización tipo Ridge. |

La buena noticia es que estas tres señales de alerta son fáciles de calcular con muy pocas líneas de código, como acabas de ver.

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Sobreajuste (Overfitting) | El modelo memoriza el entrenamiento pero falla con datos nuevos. |
| Subajuste (Underfitting) | El modelo es tan simple que ni siquiera aprende el patrón. |
| Sesgo (Bias) | Qué tan lejos, en promedio, apunta mal el modelo (por ser demasiado simple). |
| Varianza (Variance) | Qué tan inconsistente es el modelo ante distintos datos de entrenamiento (por ser demasiado flexible). |
| Multicolinealidad | Dos o más variables predictoras que se parecen demasiado entre sí. |
| VIF | Número que mide qué tan redundante es una variable frente a las demás (mayor a 5 es alerta). |
