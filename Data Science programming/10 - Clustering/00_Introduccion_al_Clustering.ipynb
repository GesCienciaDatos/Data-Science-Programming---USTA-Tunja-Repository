## 1. ¿Qué es el Clustering? 🧠

El **clustering** (agrupamiento) es una técnica de **Aprendizaje No Supervisado (*Unsupervised Learning*)** cuyo objetivo es agrupar puntos de datos similares con base en sus características (*features*), sin conocer de antemano ninguna etiqueta o clase objetivo.

Formalmente, un algoritmo de clustering busca una partición de las observaciones que:

* **Maximice la similitud intra-cluster**: los puntos dentro de un mismo grupo deben parecerse entre sí tanto como sea posible.
* **Minimice la similitud inter-cluster**: los puntos de grupos distintos deben ser tan diferentes entre sí como sea posible.

A diferencia del aprendizaje supervisado, el clustering **no utiliza una variable de salida (*target*) etiquetada** — el algoritmo descubre patrones y estructuras ocultas directamente a partir de la geometría o distribución de los datos.

### Aprendizaje No Supervisado vs. Supervisado

| Criterio | Aprendizaje Supervisado | Aprendizaje No Supervisado (Clustering) |
|---|---|---|
| **Etiquetas ($y$)** | Disponibles durante el entrenamiento | No disponibles |
| **Objetivo** | Predecir una variable objetivo conocida | Descubrir estructura/patrones ocultos |
| **Evaluación** | Métricas directas (exactitud, RMSE, etc.) frente a $y$ real | Métricas indirectas (cohesión, separación) — no hay "verdad" contra la cual comparar |
| **Ejemplos de algoritmos** | Regresión Logística, Árboles de Decisión, SVM | K-Means, DBSCAN, Clustering Jerárquico |
| **Pregunta típica** | "¿Este cliente comprará el producto?" | "¿Qué grupos naturales de clientes existen?" |

## 2. Vista Previa: El Resultado de un Algoritmo de Clustering 👀

Antes de profundizar en la teoría, veamos visualmente qué produce un algoritmo de clustering. Generaremos un conjunto de datos sintético en dos dimensiones con `make_blobs` — **sin ninguna etiqueta** — y observaremos primero los puntos "en crudo", y luego el resultado de agruparlos.

````python
from sklearn.datasets import make_blobs

X_demo, y_true_demo = make_blobs(
    n_samples=300, centers=4, cluster_std=0.85, random_state=42
)

plt.figure()
plt.scatter(X_demo[:, 0], X_demo[:, 1], s=25, color="#475569", alpha=0.7)
plt.title("Datos Sin Etiquetar: ¿Cuántos Grupos Observa? 🤔", fontweight='bold')
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.show()
````

````python
from sklearn.cluster import KMeans

kmeans_demo = KMeans(n_clusters=4, n_init=10, random_state=42)
labels_demo = kmeans_demo.fit_predict(X_demo)

plt.figure()
plt.scatter(X_demo[:, 0], X_demo[:, 1], c=labels_demo, cmap="viridis", s=25, alpha=0.8)
plt.scatter(
    kmeans_demo.cluster_centers_[:, 0], kmeans_demo.cluster_centers_[:, 1],
    c="red", marker="X", s=200, edgecolor="black", label="Centroides"
)
plt.title("Resultado del Clustering (K-Means, k=4)", fontweight='bold')
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.legend()
plt.show()
````

El algoritmo `KMeans` (que estudiaremos en detalle en el cuaderno **02**) recibió únicamente las coordenadas de los puntos — sin ninguna etiqueta $y$ — y aun así logró recuperar una partición razonable en 4 grupos, cada uno representado por su centroide (marcado con ✕ roja). Esta es la esencia del clustering: descubrir estructura a partir de la geometría de los datos.

## 3. Aplicaciones Reales del Clustering 🌍

El clustering es una de las herramientas más utilizadas en la práctica de la ciencia de datos, con aplicaciones que abarcan prácticamente todas las industrias:

* 🎯 **Segmentación de Clientes (*Customer Segmentation*):** Agrupar clientes con comportamientos, preferencias o características demográficas similares para diseñar campañas de marketing dirigidas y servicios personalizados.
* 🚨 **Detección de Anomalías (*Anomaly Detection*):** Al agrupar los patrones "normales" de comportamiento, los puntos que no encajan en ningún grupo se destacan como posibles anomalías, errores o fraudes.
* 🖼️ **Procesamiento de Imágenes y Señales:** Agrupar píxeles o patrones de señal similares — útil para reconocimiento de imágenes, compresión y reducción de ruido.
* 📄 **Clasificación de Documentos y Modelado de Tópicos:** Agrupar documentos o textos con contenido similar, facilitando la clasificación y organización de grandes corpus textuales.
* 🧬 **Ingeniería de Características (*Feature Engineering*):** La pertenencia a un clúster puede convertirse en una nueva variable (*feature*) para alimentar modelos supervisados posteriores.
* 🎧 **Sistemas de Recomendación:** Agrupar usuarios con gustos similares para recomendar contenido basado en lo que a usuarios parecidos les ha gustado.

### Aplicaciones en Agricultura 🌾

* **Agricultura de Precisión:** Identificar zonas homogéneas dentro de un mismo cultivo con base en propiedades del suelo, clima y salud de las plantas — permitiendo riego, fertilización y aplicación de pesticidas focalizados.
* **Análisis de Imágenes Satelitales:** Clasificar tipos de cultivo y patrones de uso del suelo a partir de imágenes satelitales.
* **Planeación de Rotación de Cultivos:** Agrupar cultivos con requerimientos de crecimiento similares para planear rotaciones que mantengan la salud del suelo y reduzcan la acumulación de plagas y enfermedades.
