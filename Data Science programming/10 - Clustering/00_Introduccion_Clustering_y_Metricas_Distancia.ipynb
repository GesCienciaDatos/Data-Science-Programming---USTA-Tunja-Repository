### 1. ¿Qué es el Aprendizaje No Supervisado y el Clustering? 🌐

En el **Aprendizaje No Supervisado (*Unsupervised Learning*)**, nuestro conjunto de datos $D = \{x_1, x_2, \dots, x_n\}$ **no contiene etiquetas objetivo** ($y$). El objetivo no es predecir un valor continuo ni una clase conocida, sino **descubrir patrones latentes, estructuras intrínsecas y subgrupos naturales** en los datos.

El **Clustering (Agrupamiento)** consiste en particionar las observaciones en $k$ grupos homogéneos denominados *clusters*, tales que:
1. **Alta homogeneidad intra-cluster:** Las observaciones dentro del mismo cluster sean lo más parecidas posible.
2. **Alta separación inter-cluster:** Las observaciones de clusters distintos sean lo más diferentes posible.

#### Casos de Uso del Mundo Real 💼
* 🎧 **Sistemas de Recomendación:** Agrupamiento de canciones y perfiles de escucha (ej. *Spotify Daily Mix*).

### 2. Espacios Métricos y Fórmulas de Distancia 📏

Para cuantificar qué tan "parecidas" son dos observaciones $p = (p_1, \dots, p_d)$ y $q = (q_1, \dots, q_d)$ en $\mathbb{R}^d$, definimos funciones de distancia métrica:

#### 1. Distancia Euclidiana ($L_2$ Norm)
La distancia geométrica en línea recta en el espacio euclidiano:
$$d_{\text{Euclidiana}}(p, q) = \|p - q\|_2 = \sqrt{\sum_{i=1}^d (p_i - q_i)^2}$$

#### 2. Distancia Manhattan / Cityblock ($L_1$ Norm)
La distancia recorrida a lo largo de ejes perpendiculares (análoga a las manzanas de una ciudad):
$$d_{\text{Manhattan}}(p, q) = \|p - q\|_1 = \sum_{i=1}^d |p_i - q_i|$$

<div align="center">
  <img src="images/manhattan_distance.svg" width="280" alt="Manhattan vs Euclidean Distance" style="margin: 10px 0;"/>
</div>

#### 3. Similitud y Distancia del Coseno (*Cosine Similarity*)
Mide el ángulo entre dos vectores independientemente de su magnitud (ideal para texto y embeddings de alta dimensión):
$$\text{Sim}_{\text{Coseno}}(u, v) = rac{u \cdot v}{\|u\|_2 \|v\|_2} = rac{\sum u_i v_i}{\sqrt{\sum u_i^2}\sqrt{\sum v_i^2}}, \quad d_{\text{Coseno}}(u, v) = 1 - \text{Sim}_{\text{Coseno}}(u, v)$$


````python
from sklearn.metrics.pairwise import euclidean_distances, manhattan_distances, cosine_distances

# Definir dos perfiles de clientes: [Edad, Ingreso_k, Puntuacion_Gasto]
cliente_A = np.array([[25, 40, 80]])
cliente_B = np.array([[28, 42, 75]])
cliente_C = np.array([[60, 110, 20]])

print("--- Comparación de Distancias Métricas ---")
print(f"Distancia Euclidiana (A vs B): {euclidean_distances(cliente_A, cliente_B)[0,0]:.4f}")
print(f"Distancia Euclidiana (A vs C): {euclidean_distances(cliente_A, cliente_C)[0,0]:.4f}")
print(f"Distancia Manhattan (A vs B):  {manhattan_distances(cliente_A, cliente_B)[0,0]:.4f}")
print(f"Distancia Coseno (A vs B):     {cosine_distances(cliente_A, cliente_B)[0,0]:.4f}")
````

### 3. La Importancia Crítica del Escalado de Características ⚖️

> [!WARNING]
> La mayoría de los algoritmos de clustering dependen directamente de distancias métricas. Si una variable tiene un rango de 0 a \100,000 (ej. Salario) y otra de 0 a 5 (ej. Número de hijos), la variable de mayor escala **dominará por completo el cálculo de distancias**, anulando el efecto de las demás.


````python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Demostración del impacto del escalado
datos_ejemplo = np.array([
    [25, 30000],  # Joven, salario medio
    [27, 32000],  # Joven similar
    [65, 30000]   # Adulto mayor, mismo salario
])

scaler_std = StandardScaler()
datos_escalados = scaler_std.fit_transform(datos_ejemplo)

print("Distancia Euclidiana SIN escalar (P1 vs P2):", euclidean_distances([datos_ejemplo[0]], [datos_ejemplo[1]])[0,0])
print("Distancia Euclidiana SIN escalar (P1 vs P3):", euclidean_distances([datos_ejemplo[0]], [datos_ejemplo[2]])[0,0])
print("\n--- Con StandardScaler ---")
print("Distancia CON escalado (P1 vs P2):", euclidean_distances([datos_escalados[0]], [datos_escalados[1]])[0,0])
print("Distancia CON escalado (P1 vs P3):", euclidean_distances([datos_escalados[0]], [datos_escalados[2]])[0,0])
````

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics.pairwise import euclidean_distances

# 1. Seleccionar muestra de 5 clientes
muestra = df_mall[['Age', 'Annual_Income_k', 'Spending_Score']].iloc[:5]

# 2. Normalizar con MinMaxScaler
scaler = MinMaxScaler()
muestra_norm = scaler.fit_transform(muestra)

# 3. Calcular matriz de distancias euclidianas
mat_dist = euclidean_distances(muestra_norm)

print("Matriz de Distancias Euclidianas Normalizada:")
df_dist = pd.DataFrame(mat_dist, index=[f"C_{i+1}" for i in range(5)], columns=[f"C_{i+1}" for i in range(5)])
print(df_dist.round(4))

# 4. Encontrar el par con menor distancia (excluyendo la diagonal)
np.fill_diagonal(mat_dist, np.inf)
min_i, min_j = np.unravel_index(np.argmin(mat_dist), mat_dist.shape)
print(f"\nLos clientes más cercanos son Cliente {min_i+1} y Cliente {min_j+1} con distancia {mat_dist[min_i, min_j]:.4f}")

### 5. Resumen y Conclusiones del Cuaderno 00 📌

1. **Aprendizaje No Supervisado:** Explora estructuras intrínsecas en $X$ sin etiquetas $y$. Es fundamental para segmentación, recomendación y detección de anomalías.
2. **Métricas de Distancia:** La Euclidiana ($L_2$) mide proximidad geométrica esférica, la Manhattan ($L_1$) es robusta a valores extremos en altas dimensiones y el Coseno mide alineación angular.
3. **Escalado Mandatorio:** Antes de ejecutar cualquier algoritmo de clustering basado en distancias, las variables continuas deben ser escaladas (`StandardScaler` o `MinMaxScaler`).
