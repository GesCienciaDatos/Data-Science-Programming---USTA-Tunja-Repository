````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os, urllib.request
import warnings
warnings.filterwarnings('ignore')

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (8.5, 4.5)
plt.rcParams['font.size'] = 10

def load_dataset(filename, module_folder="10 - Clustering"):
    local_path = os.path.join(os.getcwd(), "data", filename)
    if os.path.exists(local_path):
        return local_path
    
    parent_path = os.path.join(os.getcwd(), "..", module_folder, "data", filename)
    if os.path.exists(parent_path):
        return parent_path

    raw_url = f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/Data%20Science%20programming/{module_folder.replace(' ', '%20')}/data/{filename}"
    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    if not os.path.exists(target_path):
        urllib.request.urlretrieve(raw_url, target_path)
    return target_path

print("🚀 Entorno configurado exitosamente para el Módulo 10: Clustering.")
````

## 🌟 ¿Qué es el Agrupamiento Jerárquico? La Analogía del Árbol Genealógico 🌳

Imagina organizar a los invitados en una gran fiesta de bodas:
1. Al principio, cada persona está sola.
2. Primero se juntan las parejas (los más cercanos).
3. Luego esas parejas se juntan con sus primos o amigos del colegio formando una pequeña familia.
4. Luego las familias se juntan con el lado del novio o de la novia.
5. Al final, todos pertenecen al gran grupo de la fiesta.

👉 El **Dendrograma** es simplemente el dibujo de ese árbol genealógico. Si trazamos una línea con unas tijeras a cierta altura, ¡decidimos cuántas mesas armar!


````python
from scipy.cluster.hierarchy import linkage, dendrogram

path_data = load_dataset('mall_customers.csv', '10 - Clustering')
df = pd.read_csv(path_data)

# Tomamos 20 personas para un árbol fácil de leer
muestra = df[['Annual_Income_k', 'Spending_Score']].iloc[:20]
Z = linkage(muestra, method='ward')

plt.figure(figsize=(9, 4.5))
dendrogram(Z, labels=muestra.index.values, leaf_rotation=0)
plt.title('Árbol de Familias de Clientes (Dendrograma)', fontweight='bold')
plt.ylabel('Distancia / Diferencia entre personas')
plt.xlabel('Número de Cliente')
plt.axhline(y=70, color='r', linestyle='--', label='Tijera de Corte')
plt.legend()
plt.show()
````
