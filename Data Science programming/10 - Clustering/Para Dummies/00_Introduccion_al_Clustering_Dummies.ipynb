## 1. Ordenar sin que nadie te diga las categorías 🧦

Imagina que te entregan una canasta enorme con calcetines de muchos colores y tamaños, todos revueltos, **sin ninguna etiqueta** que diga "aquí van los negros" o "aquí los de niño". Aun así, tú puedes organizarlos: separas los que se parecen entre sí (color parecido, tamaño parecido) en montoncitos, sin que nadie te haya dicho de antemano cuántos montones debía haber ni cómo se llamaba cada uno.

Eso es exactamente lo que hace el **clustering** (agrupamiento) con datos: mirar un montón de observaciones (clientes, animales, canciones...) y separarlas en grupos donde las cosas de un mismo grupo se parezcan mucho entre sí, y se parezcan poco a las de otros grupos — **sin que nadie le diga de antemano cuáles son los grupos correctos**.

> 📌 **Para recordar:** en clustering no existe una "respuesta correcta" dada de antemano (como sí ocurre cuando entrenamos un modelo para predecir "sí compra / no compra"). El algoritmo descubre los grupos por sí mismo, mirando qué tan parecidos son los datos entre sí.

## 2. "Supervisado" vs. "No supervisado" en una frase 🧑‍🏫

En cursos anteriores usaste algoritmos que **aprenden con la respuesta correcta delante**, como un estudiante que estudia con las respuestas del examen ya resueltas: le muestras miles de ejemplos de "estas son las características de un cliente, y sabemos que sí compró" y el modelo aprende a predecir eso. A esto se le llama **aprendizaje supervisado**.

El clustering es distinto: es como un estudiante al que le entregan un montón de fotos de animales **sin decirle qué animal es cada uno**, y aun así logra notar "estas fotos se parecen entre sí, seguramente son del mismo tipo de animal, y estas otras son de otro tipo". No hay "respuesta correcta" que el algoritmo esté tratando de adivinar — solo agrupa por parecido. Por eso se le llama **aprendizaje no supervisado**.

| | Aprendizaje Supervisado | Clustering (No Supervisado) |
|---|---|---|
| ¿Tiene "respuestas correctas" de antemano? | Sí (etiquetas) | No |
| ¿Qué hace? | Aprende a predecir una etiqueta conocida | Descubre grupos que nadie había nombrado |
| Pregunta típica | "¿Este cliente va a comprar?" | "¿Qué tipos de clientes tengo?" |

## 3. Viendo el resultado de un clustering, en vivo 👀

Antes de entrar en teoría, veamos qué produce en la práctica un algoritmo de clustering. Vamos a generar puntos de ejemplo agrupados en 4 "nubes" — sin decirle al algoritmo cuáles son esas nubes — y a pedirle que las descubra solo.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# Generamos "a mano" 4 nubes de puntos, como si fueran 4 grupos de clientes
centro_1 = np.random.normal(loc=[2, 2], scale=0.5, size=(30, 2))
centro_2 = np.random.normal(loc=[8, 2], scale=0.5, size=(30, 2))
centro_3 = np.random.normal(loc=[2, 8], scale=0.5, size=(30, 2))
centro_4 = np.random.normal(loc=[8, 8], scale=0.5, size=(30, 2))

puntos = np.vstack([centro_1, centro_2, centro_3, centro_4])

plt.figure(figsize=(6, 5))
plt.scatter(puntos[:, 0], puntos[:, 1], color="#475569", s=30, alpha=0.8)
plt.title("Puntos sin agrupar: ¿cuántos grupos ves? 🤔")
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.show()
```

### 🤔 ¿Qué acaba de pasar?

- Creamos 4 "nubes" de puntos alrededor de 4 centros distintos, para simular 4 grupos naturales de datos (como 4 tipos de clientes).
- Al graficarlos, el ojo humano detecta fácilmente los 4 grupos porque están bien separados. Pero el computador, por sí solo, todavía no sabe que existen esos 4 grupos — solo ve una lista de coordenadas.
- Ahora vamos a pedirle a un algoritmo de clustering (`KMeans`, que veremos a fondo en el cuaderno 02) que descubra esos grupos sin que le digamos dónde están.

```python
from sklearn.cluster import KMeans

modelo = KMeans(n_clusters=4, n_init=10, random_state=42)
grupo_asignado = modelo.fit_predict(puntos)

plt.figure(figsize=(6, 5))
plt.scatter(puntos[:, 0], puntos[:, 1], c=grupo_asignado, cmap="viridis", s=30, alpha=0.85)
plt.scatter(
    modelo.cluster_centers_[:, 0], modelo.cluster_centers_[:, 1],
    c="red", marker="X", s=200, edgecolor="black", label="Centro de cada grupo"
)
plt.title("Resultado del Clustering (4 grupos descubiertos)")
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.legend()
plt.show()
```

### 🤔 ¿Qué acaba de pasar?

- Le entregamos a `KMeans` únicamente las coordenadas de los puntos — **nunca le dijimos** cuáles pertenecían a cada grupo.
- `fit_predict` hace dos cosas a la vez: "aprende" dónde están los grupos (`fit`) y nos devuelve, para cada punto, a qué grupo quedó asignado (`predict`).
- El resultado colorea cada punto según el grupo que el algoritmo descubrió, y marca con una ✕ roja el "centro" de cada grupo (llamado *centroide*).
- Esta es la esencia del clustering: descubrir estructura a partir de la posición de los datos, sin que nadie le diera pistas de antemano.

## 4. ¿Para qué sirve esto en la vida real? 🌍

El clustering aparece por todas partes, aunque no lo notemos:

* 🎯 **Segmentación de clientes:** agrupar clientes parecidos para diseñar promociones dirigidas (justo lo que haremos con `mall_customers.csv` en este módulo).
* 🚨 **Detectar algo raro:** si agrupamos el comportamiento "normal" de miles de usuarios, una transacción que no encaja en ningún grupo puede ser un fraude.
* 🎧 **Listas de reproducción automáticas:** el "Daily Mix" de Spotify agrupa canciones y usuarios con gustos parecidos para armar listas personalizadas.
* 🌾 **Agricultura de precisión:** agrupar zonas de un cultivo según el estado del suelo para regar o fertilizar solo donde hace falta.
* 📄 **Organizar documentos:** agrupar textos parecidos para clasificar noticias, correos o reseñas automáticamente.

En todos los casos, la idea de fondo es la misma: **encontrar grupos naturales sin que alguien los defina a mano primero**.

A lo largo de este módulo usaremos un dataset real y sencillo: `mall_customers.csv`, con 200 clientes de un centro comercial y columnas como `Age` (edad), `Annual_Income_k` (ingreso anual en miles) y `Spending_Score` (qué tanto gasta, de 1 a 100).

Vamos a cargarlo y a graficar el ingreso contra el puntaje de gasto — **sin aplicar ningún algoritmo todavía** — solo para practicar lo que hicimos en el cuaderno principal: mirar los datos "a ojo" antes de agrupar formalmente.

```python
df_mall = pd.read_csv('../data/mall_customers.csv')

print("Filas y columnas:", df_mall.shape)
df_mall.head()
```

## 6. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Clustering | Agrupar datos parecidos entre sí, sin que nadie diga de antemano cuáles son los grupos. |
| Aprendizaje no supervisado | No hay "respuesta correcta" (etiqueta) — el algoritmo descubre la estructura solo. |
| Similar / distinto | El objetivo es que lo de un mismo grupo se parezca mucho, y lo de grupos distintos se parezca poco. |
| Centroide | El "centro" que representa a un grupo (lo veremos a fondo en el cuaderno 02). |
| Usos reales | Segmentar clientes, detectar fraudes, armar listas de música, agricultura de precisión, y más. |
