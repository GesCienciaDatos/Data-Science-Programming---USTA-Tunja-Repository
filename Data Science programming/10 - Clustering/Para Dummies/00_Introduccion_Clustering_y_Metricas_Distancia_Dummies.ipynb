## 🌟 ¿Qué es el Clustering? La Analogía del Armario Desordenado 👕👗

Imagina que abres tu armario y toda tu ropa está mezclada en una sola montaña gigante.
* **No tienes etiquetas** que digan *"esto es de fiesta"* o *"esto es de hacer deporte"*.
* Lo que haces de forma natural es **juntar las cosas parecidas**: todas las camisetas juntas en un cajón, los pantalones en otro y los abrigos colgados.

👉 **Eso es el Clustering:** Enseñar a la computadora a encontrar grupos naturales de cosas que se parecen entre sí, ¡sin que nadie le diga de antemano qué son!


---
## 🗺️ ¿Cómo mide la computadora el parecido? Las Distancias

Para saber si dos clientes o canciones se parecen, medimos qué tan "cerca" están en un mapa:

1. **Distancia en Línea Recta (Euclidiana):** Como si volaras en dron directamente de un punto a otro.
2. **Distancia en Manzanas de la Ciudad (Manhattan):** Como caminar por una ciudad cuadra por cuadra doblando en las esquinas.
<div align="center"><img src="../images/manhattan_distance.svg" width="260"/></div>

3. **Distancia del Coseno:** Compara el "estilo" o dirección de los gustos, sin importar si una persona compra mucho o poco.


````python
import pandas as pd
from sklearn.metrics.pairwise import euclidean_distances

# Comparemos a 3 personas según sus minutos escuchados de Rock y Pop al mes
amigos = pd.DataFrame({
    'Minutos_Rock': [120, 115, 10],
    'Minutos_Pop': [15, 20, 180]
}, index=['Ana', 'Beto', 'Carlos'])

print("--- Minutos de Música al Mes ---")
print(amigos)

distancias = euclidean_distances(amigos)
print("\nDistancia entre Ana y Beto:", round(distancias[0, 1], 2), "(¡Muy parecidos!)")
print("Distancia entre Ana y Carlos:", round(distancias[0, 2], 2), "(¡Gustos totalmente opuestos!)")
````

Llega una nueva usuaria llamada **Diana** con `Minutos_Rock = 110` y `Minutos_Pop = 25`.
¿A quién se parece más Diana: a Ana o a Carlos?


````python
# =========================================================================
# TU SOLUCIÓN: Reto Dummies 1 - Diana vs el Mundo
# =========================================================================

# diana = [[110, 25]]
# ...
````

```python
diana = [[110, 25]]
dist_ana = euclidean_distances(diana, [amigos.loc['Ana']])[0,0]
dist_carlos = euclidean_distances(diana, [amigos.loc['Carlos']])[0,0]

print(f"Distancia de Diana a Ana: {dist_ana:.2f}")
print(f"Distancia de Diana a Carlos: {dist_carlos:.2f}")

if dist_ana < dist_carlos:
    print("👉 Diana tiene gustos similares a Ana. ¡Recomendarle Rock!")
else:
    print("👉 Diana tiene gustos similares a Carlos. ¡Recomendarle Pop!")
```
