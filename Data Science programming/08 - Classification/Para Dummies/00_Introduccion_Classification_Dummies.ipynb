## 1. Regresión vs Clasificación — Números vs Etiquetas 🏷️

Hasta ahora en el curso hemos predicho **números** (precios, temperaturas, salarios). Eso es **regresión**.

Pero muchas preguntas del mundo real no tienen una respuesta numérica — tienen una **etiqueta**:

| Pregunta | Respuesta numérica (regresión) | Respuesta de etiqueta (clasificación) |
|---|---|---|
| ¿Cuánto costará esta casa? | $350,000,000 | — |
| ¿Este correo es spam? | — | Spam / No spam |
| ¿El paciente tiene cardiopatía? | — | Sí / No |
| ¿Qué especie de flor es esta? | — | Setosa / Versicolor / Virginica |
| ¿El cliente cancelará el servicio? | — | Cancela / No cancela |

> 🎯 **Clasificación** = predecir a qué **categoría** pertenece algo. Es como el semáforo: no dice "qué tan verde está", dice **verde, amarillo o rojo**.

## 2. Por qué la regresión lineal falla en clasificación ❌

Imagina que quieres predecir si alguien tiene cardiopatía (0 = No, 1 = Sí) usando su edad.

Si usas regresión lineal, el modelo puede predecir valores como -0.3 o 1.7. ¿Qué significa una probabilidad de 170%? ¡Nada! Las probabilidades solo pueden estar entre 0% y 100%.

Necesitamos algo que **"comprima"** cualquier número al rango [0, 1]. Eso es la función Sigmoide.

````python
import numpy as np
import matplotlib.pyplot as plt

# La función Sigmoide: convierte cualquier número a probabilidad entre 0 y 1
def sigmoide(z):
    return 1 / (1 + np.exp(-z))

z = np.linspace(-10, 10, 300)
prob = sigmoide(z)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Línea recta (regresión lineal) — puede salirse del rango [0,1]
axes[0].plot(z, 0.05 * z + 0.5, color='#ef4444', linewidth=2.5, label='Regresión lineal')
axes[0].axhline(0, color='gray', linestyle='--', alpha=0.5)
axes[0].axhline(1, color='gray', linestyle='--', alpha=0.5)
axes[0].fill_between(z, 0, 1, alpha=0.08, color='green', label='Zona válida [0,1]')
axes[0].set_ylim(-0.5, 1.5)
axes[0].set_title('❌ Regresión Lineal para clasificación\n(puede predecir -0.3 o 1.7 — ¡inválido!)', fontweight='bold')
axes[0].legend()
axes[0].set_xlabel('Variable de entrada (ej. edad)')
axes[0].set_ylabel('Predicción')

# Sigmoide — siempre entre 0 y 1
axes[1].plot(z, prob, color='#f59e0b', linewidth=3, label='Función Sigmoide σ(z)')
axes[1].axhline(0.5, color='#6366f1', linestyle='--', linewidth=1.5, label='Umbral 50%')
axes[1].axhline(0, color='gray', linestyle=':', alpha=0.5)
axes[1].axhline(1, color='gray', linestyle=':', alpha=0.5)
axes[1].fill_between(z, 0.5, prob, where=prob >= 0.5, alpha=0.15, color='#10b981', label='Clase 1 (Positivo)')
axes[1].fill_between(z, prob, 0.5, where=prob < 0.5, alpha=0.15, color='#ef4444', label='Clase 0 (Negativo)')
axes[1].set_title('✅ Función Sigmoide (la curva en S)\n(siempre entre 0% y 100%)', fontweight='bold')
axes[1].legend()
axes[1].set_xlabel('Combinación lineal de variables')
axes[1].set_ylabel('Probabilidad P(Y=1)')

plt.suptitle('¿Por qué no usar regresión lineal para clasificar?', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n💡 La Sigmoide convierte cualquier número real en una probabilidad entre 0 y 1.")
print("   Si z = 0  → σ(0) = 0.50 (50% de probabilidad, incertidumbre máxima)")
print("   Si z = 5  → σ(5) ≈ 0.99 (99% de probabilidad de ser clase 1)")
print("   Si z = -5 → σ(-5) ≈ 0.01 (1% de probabilidad, casi seguro clase 0)")
````

## 3. El clasificador de correo Spam — El ejemplo más intuitivo 📧

Imagina que entrenas un modelo con miles de correos marcados como Spam o No-Spam. El modelo aprende que:

- Si el correo contiene "GANASTE", "GRATIS", "¡URGENTE!" → probabilidad alta de ser spam.
- Si el correo tiene un nombre conocido, texto coherente, no tiene signos de exclamación en todo → probabilidad baja de ser spam.

```
  Correo → [Extracción de características] → [Modelo Logístico] → P(spam) = 0.97 → 🚨 SPAM
  Correo → [Extracción de características] → [Modelo Logístico] → P(spam) = 0.03 → ✅ NO SPAM
```

El **umbral de decisión** es normalmente 0.5 (50%): si P > 0.5 → Spam, si P ≤ 0.5 → No Spam.

````python
import os, urllib.parse, urllib.request
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

def load_dataset(filename, module_name="08 - Classification"):
    candidates = [f"data/{filename}", f"../{module_name}/data/{filename}", f"{module_name}/data/{filename}", filename]
    for path in candidates:
        if os.path.exists(path):
            return path
    os.makedirs("data", exist_ok=True)
    target_path = f"data/{filename}"
    url = f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{urllib.parse.quote(module_name)}/data/{urllib.parse.quote(filename)}"
    urllib.request.urlretrieve(url, target_path)
    print(f"✅ Dataset '{filename}' descargado.")
    return target_path

df = pd.read_csv(load_dataset('heart_disease.csv'))
print(f"Dataset: {df.shape[0]} pacientes, {df.shape[1]} columnas")
print(f"Variable objetivo: '{df.columns[-1]}'")
print(df[df.columns[-1]].value_counts().rename({0:'Sin cardiopatía', 1:'Con cardiopatía'}))
````
