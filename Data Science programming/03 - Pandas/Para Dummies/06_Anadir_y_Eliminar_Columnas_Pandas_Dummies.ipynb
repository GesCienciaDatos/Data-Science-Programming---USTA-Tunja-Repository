## Agregar y Borrar Columnas sin Miedo 🗑️

- **Añadir:** `df['nueva_columna'] = valores`
- **Eliminar con `drop()`:** `df.drop(columns=['columna_inutil'])` (usa `axis=1` o el parámetro `columns`).
- **`inplace=True` vs Reasignar:** Recomendamos reasignar `df = df.drop(...)` para mantener el código claro y sin efectos secundarios inesperados.


```python
import pandas as pd

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Nombre": ["Producto A", "Producto B", "Producto C"],
    "Columna_Inutil": ["x", "y", "z"],
    "Precio": [100, 200, 300]
})

print("Tabla original:")
display(df)

# Eliminamos la columna inútil:
df = df.drop(columns=["Columna_Inutil"])
print("\nTabla después de eliminar 'Columna_Inutil':")
display(df)
```
