## 1. Un DataFrame es como una hoja de cálculo con superpoderes 📊

Imagina una hoja de cálculo de Excel con columnas: `nombre`, `ciudad`, `edad`. Un DataFrame de Pandas es básicamente eso, pero puedes manipularlo con código en vez de hacer clic y arrastrar.

Y aquí viene la analogía clave: **un DataFrame se comporta como un diccionario de Python** donde cada "llave" es el nombre de una columna, y el "valor" es toda esa columna de datos (una `Series`). Por eso, para leer, crear o borrar columnas usamos una sintaxis muy parecida a la de los diccionarios: corchetes `[ ]` con el nombre de la columna adentro.

Vamos a crear una pequeña tabla de ciudades para practicar con datos reales.

````python
import pandas as pd

ciudades = pd.DataFrame(
    [
        ['Tunja', 195000, 118, 'Boyacá'],
        ['Bogotá', 7412566, 1636, 'Cundinamarca'],
        ['Medellín', 2529403, 380, 'Antioquia'],
        ['Cali', 2227642, 564, 'Valle del Cauca'],
    ],
    columns=['nombre', 'poblacion', 'area_km2', 'departamento']
)
ciudades
````

### 🤔 ¿Qué acaba de pasar?

- Creamos un DataFrame de 4 filas y 4 columnas, describiendo algunas ciudades de Colombia.
- Cada columna (`nombre`, `poblacion`, `area_km2`, `departamento`) es como una "llave" del diccionario gigante que es la tabla.
- Ahora vamos a **agregarle** una columna nueva, tal como agregaríamos una llave nueva a un diccionario.

---
## 2. Añadir una columna: como agregar una llave a un diccionario ➕

Para crear una columna nueva, simplemente le asignas un valor a un nombre de columna que todavía no existe: `tabla['columna_nueva'] = ...`.

Por defecto, **la columna nueva siempre se agrega al final (a la derecha)** de la tabla. Vamos a crear una columna con la población expresada en millones.

````python
# Convertimos la población a millones de habitantes
ciudades['poblacion (millones)'] = ciudades['poblacion'] / 1_000_000

# También podemos crear columnas booleanas evaluando una condición
ciudades['es_grande'] = ciudades['poblacion'] > 1_000_000
ciudades
````

### 🤔 ¿Qué acaba de pasar?

- `ciudades['poblacion (millones)']` no existía antes, así que Pandas la crea como columna nueva; el valor se calcula **para cada fila a la vez**, sin escribir ningún bucle.
- `ciudades['es_grande'] = ciudades['poblacion'] > 1_000_000` crea otra columna nueva, esta vez con valores `True`/`False`, evaluando la condición fila por fila.
- Ambas columnas nuevas aparecieron **al final**, como los últimos puestos en la mesa.

---
## 3. Insertar en un lugar exacto: `insert()` 📍

¿Y si no quieres que la columna nueva vaya siempre al final? Imagina que en una fila de sillas quieres meter una silla extra justo en la posición 3, no al final de la fila. Para eso existe el método `insert()`, que recibe **la posición** donde quieres la columna.

````python
densidad = ciudades['poblacion'] / ciudades['area_km2']

# Insertamos "densidad" en la posición 2 (osea, como la 3ra columna)
ciudades.insert(2, 'densidad_hab_km2', densidad)
ciudades
````

### 🤔 ¿Qué acaba de pasar?

- `insert(2, 'densidad_hab_km2', densidad)` le dice a Pandas: "mete esta nueva columna justo en la posición número 2" (recuerda que en Python contamos desde 0, así que la posición 2 es el tercer lugar).
- El resto de columnas se corren hacia la derecha para hacerle espacio, igual que cuando alguien se cuela en una fila y todos los demás se recorren un puesto.

---
## 4. Eliminar columnas: `drop()` 🗑️

Para quitar una columna usamos `drop()`. Hay un detalle **muy importante**: debes indicar `axis=1` para decirle a Pandas "estoy borrando una columna, no una fila" (`axis=0` sería una fila).

> 📌 **Para recordar:** `axis=0` apunta hacia abajo (filas), `axis=1` apunta hacia la derecha (columnas).

````python
ciudades.drop('es_grande', axis=1)
````

⚠️ **¡Cuidado! El DataFrame original no cambió.** Casi todas las operaciones "destructivas" de Pandas, por seguridad, te devuelven una **copia** con el cambio aplicado, pero dejan la tabla original intacta — como una fotocopia editada, mientras el documento original sigue guardado sin tocar.

Si de verdad quieres que el borrado sea permanente, tienes dos caminos válidos: **reasignar la variable** (sobrescribir la tabla con el resultado) o usar el parámetro **`inplace=True`**, que le dice a Pandas "modifica esta tabla directamente, no me devuelvas una copia".

````python
# Camino 1: reasignar la variable
ciudades = ciudades.drop('es_grande', axis=1)

# Camino 2: usar inplace=True
ciudades.drop('poblacion (millones)', axis=1, inplace=True)
ciudades
````

### 🤔 ¿Qué acaba de pasar?

- Con `ciudades = ciudades.drop(...)` le pedimos la copia editada y la guardamos encima de la variable original.
- Con `inplace=True`, Pandas edita la tabla "en el sitio", sin generar una copia extra — un poco más rápido y con menos gasto de memoria RAM, pero ojo: ya no puedes "deshacer" fácilmente.

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| Añadir columna | `tabla['nueva'] = valores` — igual que agregar una llave a un diccionario. Se agrega al final. |
| `insert(pos, nombre, valores)` | Inserta la columna nueva exactamente en la posición que le indiques. |
| `drop(columna, axis=1)` | Elimina una columna. `axis=1` = columnas, `axis=0` = filas. |
| Copia vs. original | Por defecto, `drop()` devuelve una copia editada; el DataFrame original sigue intacto. |
| Borrado permanente | Reasigna la variable (`tabla = tabla.drop(...)`) o usa `inplace=True`. |
