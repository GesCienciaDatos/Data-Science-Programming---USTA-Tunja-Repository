## Reorganizando la Estructura de tu DataFrame 🗂️

A menudo necesitas reestructurar tus datos: eliminar columnas innecesarias,
añadir nuevas, o filtrar filas que no sirven.

### Las Operaciones Más Comunes

| Operación | Código | Notas |
|---|---|---|
| Eliminar columna | `df.drop('col', axis=1)` | `axis=1` para columnas |
| Eliminar fila | `df.drop(indice, axis=0)` | `axis=0` para filas |
| Añadir columna | `df['nueva'] = ...` | Al final del DataFrame |
| Renombrar columna | `df.rename(columns={'viejo': 'nuevo'})` | Mapa de renombrado |
| Reiniciar índice | `df.reset_index(drop=True)` | `drop=True` elimina el antiguo |


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'ID': [101, 102, 103, 104, 105],
    'Nombre': ['Ana García', 'Luis Torres', 'María Pérez', 'Carlos López', 'Sofía Ruiz'],
    'Edad': [28, 35, 24, 42, 31],
    'Email': ['ana@email.com', 'luis@email.com', 'maria@email.com', 'carlos@email.com', 'sofia@email.com'],
    'Salario': [3200000, 5800000, 2900000, 7100000, 4300000],
    'Nota_Interna': ['OK', 'OK', 'REVISAR', 'OK', 'OK'],  # columna que eliminaremos
})

print('DataFrame original:')
display(df)
```

```python
# Eliminar columnas innecesarias
df_limpio = df.drop(columns=['Nota_Interna', 'Email'])  # Más claro que axis=1
print('Sin columnas confidenciales:')
display(df_limpio)

# Renombrar columnas
df_limpio = df_limpio.rename(columns={'Nombre': 'Nombre_Completo', 'Salario': 'Salario_COP'})
print('\nCon columnas renombradas:')
display(df_limpio)
```

```python
# Añadir columnas y filtrar filas
df_limpio['Salario_USD'] = (df_limpio['Salario_COP'] / 4200).round(0).astype(int)
df_limpio['Rango'] = pd.cut(df_limpio['Edad'],
                             bins=[0, 30, 40, 100],
                             labels=['Joven', 'Adulto', 'Senior'])

print('Con nuevas columnas calculadas:')
display(df_limpio)

# Eliminar filas con índice específico
df_sin_id103 = df_limpio[df_limpio['ID'] != 103].reset_index(drop=True)
print('\nSin el empleado ID 103:')
display(df_sin_id103)
```

> **⚠️ inplace=True — Con Precaución**
>
> Muchas operaciones de Pandas tienen la opción `inplace=True`.
> 
> - `df.drop('col', axis=1, inplace=True)` → modifica `df` directamente (sin devolver copia)
> - `df.drop('col', axis=1)` → devuelve un **nuevo** DataFrame sin modificar `df`
> 
> **Recomendación:** Evita `inplace=True` — es mejor asignar el resultado a una variable
> para poder deshacer cambios fácilmente.

**Pregunta:** Si usas `df.drop('Edad', axis=1)` pero no asignas el resultado a ninguna variable, ¿qué pasa con `df`?

<details>
<summary>💡 Ver respuesta</summary>

`df` **permanece igual**, sin ningún cambio.

La función `drop()` devuelve un **nuevo DataFrame** sin la columna 'Edad',
pero si no lo capturas con `df = df.drop(...)` o `df_nuevo = df.drop(...)`,
el resultado se descarta.
