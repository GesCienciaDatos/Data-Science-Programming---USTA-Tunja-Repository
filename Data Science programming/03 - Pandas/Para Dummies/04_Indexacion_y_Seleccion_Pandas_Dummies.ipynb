## ¿Qué vamos a aprender aquí? 🎈

Ya sabemos hacerle un chequeo rápido a cualquier tabla. El siguiente paso natural es aprender a **elegir exactamente** el pedazo de datos que necesitamos — una columna, una fila, un grupo de filas que cumplan cierta condición.

Al terminar podrás:
1. Sacar una columna con el punto (`.`) o con corchetes (`[]`), y cortar filas en "rebanadas" (*slicing*).
2. Usar `iloc` para seleccionar por **posición** y `loc` para seleccionar por **etiqueta**.
3. Filtrar filas con preguntas de Sí/No (indexación booleana) usando `&` (Y) y `|` (O).
4. Usar los atajos `isin()` y `notnull()`.

````python
import pandas as pd
import numpy as np

# Recuperamos la misma tabla de mascotas del cuaderno anterior
mascotas = pd.DataFrame({
    'nombre': ['Rocky', 'Luna', 'Milo', 'Nina', 'Toby', 'Coco', 'Max', 'Bella', 'Simon', 'Lola'],
    'especie': ['Perro', 'Gato', 'Perro', 'Gato', 'Perro', 'Ave', 'Gato', 'Perro', 'Gato', 'Conejo'],
    'edad_meses': [8, 14, 3, 24, 60, 6, 36, 2, 18, 9],
    'precio': [450000, 300000, 500000, 250000, np.nan, 80000, 320000, 600000, np.nan, 120000],
    'vacunado': [True, True, False, True, True, False, True, False, True, True]
})

mascotas
````

---
## 1. Sacar una columna, y cortar en rebanadas ✂️

Piensa en el DataFrame como un edificio de casilleros con nombre. Hay dos formas de pedir el contenido de un casillero:

- **Como atributo (punto):** `mascotas.especie` — rápido, pero **frágil**: no funciona si el nombre tiene espacios, empieza con número, o choca con un método que ya existe en Pandas.
- **Con corchetes:** `mascotas['especie']` — un poco más de tecleo, pero **siempre funciona**. Es la forma recomendada.

Sobre una `Series`, el *slicing* `serie[inicio:fin]` funciona igual que en una lista de Python. Sobre un `DataFrame` completo, `df[inicio:fin]` es un atajo que corta **filas**.

````python
# Punto vs. corchetes — ambos devuelven lo mismo
display(mascotas.especie)
print("Tipo con corchetes:", type(mascotas['especie']))

# Slicing sobre una Series: nombres en las posiciones 2 a 5
print("\n--- Nombres, posiciones 2 a 5 ---")
print(mascotas['nombre'][2:6])

# Slicing directo sobre el DataFrame completo -> corta filas
print("\n--- DataFrame recortado, filas 2 a 5 ---")
display(mascotas[2:6])
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas.especie` y `mascotas['especie']` devuelven exactamente lo mismo: una `Series`. Si quieres sacar **varias** columnas a la vez, usa una lista dentro de los corchetes: `mascotas[['especie', 'precio']]` — ahí sí obtienes un DataFrame más pequeño.
- `mascotas['nombre'][2:6]` primero saca la columna y luego la recorta por posición (recuerda: el límite final `6` **no** se incluye, como en cualquier lista de Python).
- `mascotas[2:6]` hace lo mismo pero sobre la tabla entera: filas en esas posiciones, con todas sus columnas. Para selecciones más precisas (filas **y** columnas específicas a la vez) necesitamos `loc` e `iloc`.

---
## 2. Selección experta: `iloc` (posición) vs. `loc` (etiqueta) 🗺️

- **`iloc`** (*Integer Location*): piensas en **posiciones**, como contar asientos en un cine — el primero es la posición 0, sin importar cómo se llamen las filas o columnas.
- **`loc`** (*Location*): piensas en **nombres (etiquetas)** — el nombre de la columna, o la etiqueta del índice de la fila.

La sintaxis siempre es `df.iloc[filas, columnas]` / `df.loc[filas, columnas]`.

⚠️ **Cuidado:** a diferencia del slicing normal de Python, en `loc` el límite final **sí se incluye**.

````python
# iloc: por posición numérica
print("--- iloc: filas 0,2,5 (posiciones), columnas 0 y 2 ---")
display(mascotas.iloc[[0, 2, 5], [0, 2]])

# loc: por etiqueta
print("--- loc: filas etiquetadas 0 a 3 (¡el 3 SÍ se incluye!), columnas por nombre ---")
display(mascotas.loc[0:3, ['nombre', 'edad_meses']])
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas.iloc[[0, 2, 5], [0, 2]]` trajo las filas en las posiciones 0, 2 y 5, y las columnas en las posiciones 0 y 2 — puedes pasar listas de posiciones, no tienen que ser consecutivas.
- `mascotas.loc[0:3, ...]` trajo las filas `0, 1, 2, 3` — cuatro filas, porque `loc` **incluye** el límite final. Con `iloc[0:3]` habrías obtenido solo `0, 1, 2` (tres filas).

**Regla mental:** `iloc` = posiciones tipo "asiento de cine" (el final no se incluye). `loc` = nombres/etiquetas (el final sí se incluye).

## 3. Preguntas de Sí/No: indexación booleana 🎭

Lo más común en la vida real es seleccionar filas que **cumplan una condición** — por ejemplo, "solo los gatos vacunados". El truco: una comparación como `mascotas['especie'] == 'Gato'` no filtra nada por sí sola — devuelve una **máscara booleana** (una Serie de `True`/`False`, una por fila). Metemos esa máscara dentro de `loc` para filtrar de verdad, y podemos combinar condiciones con `&` (Y) o `|` (O) — siempre entre paréntesis.

````python
# La máscara booleana, y su uso para filtrar
print("--- Solo los gatos ---")
display(mascotas.loc[mascotas['especie'] == 'Gato'])

# Combinando condiciones: & (Y) y | (O)
print("--- Gatos QUE ADEMÁS estén vacunados (&) ---")
display(mascotas.loc[(mascotas['especie'] == 'Gato') & (mascotas['vacunado'] == True)])

print("--- Perros O que cuesten más de 300.000 (|) ---")
display(mascotas.loc[(mascotas['especie'] == 'Perro') | (mascotas['precio'] > 300000)])
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas['especie'] == 'Gato'` comparó, fila por fila, el valor de `especie` contra `'Gato'` — sin filtrar nada todavía. `mascotas.loc[máscara]` se quedó **solo** con las filas donde esa comparación dio `True`.
- Con `&` solo sobrevivieron las filas donde **ambas** condiciones eran `True` a la vez. Con `|` sobrevivió cualquier fila que cumpliera **al menos una**.
- Nota que `Toby` y `Simon` tienen `precio` igual a `NaN` — al comparar `NaN > 300000`, Pandas simplemente devuelve `False` (no sabe si cumple, así que asume que no).

---
## 4. Atajos útiles: `isin()` y `notnull()` 🧰

- **`isin(lista)`**: `True` si el valor está en la lista que le pases — mucho más corto que encadenar varios `|`.
- **`notnull()`** (y su opuesto `isnull()`): `True` donde el dato **sí** existe (o **no** existe) — perfecto para encontrar datos faltantes.

````python
# isin: mascotas que sean Perro o Ave (sin escribir dos condiciones con |)
print("--- Perros o aves ---")
display(mascotas.loc[mascotas['especie'].isin(['Perro', 'Ave'])])

# notnull: mascotas que SÍ tienen precio registrado
print("--- Mascotas con precio conocido ---")
display(mascotas.loc[mascotas['precio'].notnull()])
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas['especie'].isin(['Perro', 'Ave'])` equivale a `(especie == 'Perro') | (especie == 'Ave')`, pero más corto y legible cuando la lista crece.
- `mascotas['precio'].notnull()` filtró exactamente las 8 filas con precio válido — las mismas que `describe()` contaba en el cuaderno anterior. Su opuesto, `isnull()`, te daría justo las 2 filas con `NaN`.

## 5. Resumen relámpago ⚡

| Herramienta | ¿Para qué sirve? |
|---|---|
| `df.col` (punto) | Acceso rápido a una columna, pero frágil (nombres con espacios o que chocan con métodos fallan). |
| `df['col']` | Acceso seguro y recomendado a una columna. |
| `df[a:b]` | Slicing directo sobre el DataFrame: corta **filas**. |
| `df.iloc[filas, cols]` | Selección por **posición** numérica; el límite final no se incluye. |
| `df.loc[filas, cols]` | Selección por **etiqueta**; el límite final sí se incluye. |
| Máscara booleana + `loc` | `df['col'] == valor` da `True`/`False`; `df.loc[máscara]` filtra de verdad. |
| `&` / `\|` | Combina condiciones (Y / O); cada condición va entre paréntesis. |
| `isin(lista)` / `notnull()` | Atajos para "está en esta lista" y "el dato sí existe". |
