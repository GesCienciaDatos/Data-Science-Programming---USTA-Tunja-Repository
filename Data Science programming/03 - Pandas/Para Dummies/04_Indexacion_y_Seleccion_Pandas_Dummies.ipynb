## La Batalla Clásica: `loc` vs `iloc` ⚔️

Es la pregunta que más confunde a principiantes. La respuesta es simple:

| Selector | 💡 Significado | ¿Cómo busca? | Ejemplo |
|---|---|---|---|
| **`iloc`** | **i**nteger **loc**ation | Por **número de posición** (0, 1, 2...) | `df.iloc[0:5, 1:3]` |
| **`loc`** | **l**abel **loc**ation | Por **nombre de etiqueta** o condición | `df.loc[df['Edad'] >= 18]` |

### 🗺️ La Analogía del Edificio
- **`iloc`** = decirle al ascensor el **número de piso** (Piso 3)
- **`loc`** = decirle al ascensor el **nombre del piso** (Oficinas de Contabilidad)


```python
import pandas as pd

df = pd.DataFrame({
    'Nombre':  ['Laura', 'Camilo', 'Daniela', 'Esteban', 'María'],
    'Edad':    [23, 31, 19, 45, 28],
    'Ciudad':  ['Tunja', 'Bogotá', 'Tunja', 'Medellín', 'Cali'],
    'Salario': [2500000, 4200000, 1800000, 5600000, 3100000]
}, index=['usr_01', 'usr_02', 'usr_03', 'usr_04', 'usr_05'])

print('Dataset completo:')
display(df)
```

```python
# ── ILOC: Selección por posición numérica ────────────────────────────
print('1. iloc[0] → Primera fila (índice 0):')
display(df.iloc[0])

print('\n2. iloc[1:4] → Filas 1, 2 y 3 (como slicing de listas):')
display(df.iloc[1:4])

print('\n3. iloc[0:3, 0:2] → Primeras 3 filas, primeras 2 columnas:')
display(df.iloc[0:3, 0:2])
```

```python
# ── LOC: Selección por etiqueta o condición ─────────────────────────
print('1. loc["usr_02"] → Fila con etiqueta usr_02:')
display(df.loc['usr_02'])

print('\n2. loc[["usr_01", "usr_03"], ["Nombre", "Salario"]] → Filas y columnas por nombre:')
display(df.loc[['usr_01', 'usr_03'], ['Nombre', 'Salario']])

print('\n3. Filtrado condicional — Personas de Tunja con salario > 2M:')
filtro = (df['Ciudad'] == 'Tunja') & (df['Salario'] > 2000000)
display(df.loc[filtro])
```

```python
# ── Selección de columnas — Forma simple ──────────────────────────────
print('Seleccionar una sola columna → devuelve Series:')
print(df['Nombre'])

print('\nSeleccionar múltiples columnas → devuelve DataFrame:')
display(df[['Nombre', 'Ciudad', 'Salario']])

# Filtro booleano directo
print('\nPersonas mayores de 25 años:')
display(df[df['Edad'] > 25])
```

> **⚠️ Error Clásico**
>
> No confundas `df['col']` (una columna) con `df[['col']]` (un DataFrame de una columna).
> 
> - `df['Nombre']` → devuelve una **`Series`** (1D)
> - `df[['Nombre']]` → devuelve un **`DataFrame`** (2D, con una sola columna)

### ✅ Autocomprobación Rápida

**Pregunta:** Tienes un DataFrame con índice personalizado ('a', 'b', 'c'...). ¿Qué pasa si usas `df.iloc[0]` vs `df.loc['a']`?

<details>
<summary>💡 Ver respuesta</summary>

- `df.iloc[0]` → devuelve la **primera fila por posición** (siempre el índice numérico 0, independientemente del nombre)
- `df.loc['a']` → devuelve la fila con **etiqueta 'a'** (busca por nombre)

Si el índice es `('a', 'b', 'c')`, ambas devuelven la misma fila,
pero si el índice empieza en `'b'`, `iloc[0]` devuelve 'b' y `loc['a']` da error.

</details>
