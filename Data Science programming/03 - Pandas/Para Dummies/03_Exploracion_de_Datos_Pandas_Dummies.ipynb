## Explorando un Dataset por Primera Vez 🔍

Cuando recibes un dataset nuevo, lo primero es **conocerlo**. Es como conocer a una persona:
¿Cuántos datos tiene? ¿Qué tipo de información guarda? ¿Tiene datos faltantes? ¿Hay valores raros?

### Tu Kit de Supervivencia para Exploración

| Función | ¿Qué te dice? |
|---|---|
| `df.shape` | Dimensiones: (filas, columnas) |
| `df.head(n)` | Las primeras n filas |
| `df.tail(n)` | Las últimas n filas |
| `df.info()` | Tipos de datos + valores no nulos por columna |
| `df.describe()` | Estadísticas de columnas numéricas |
| `df.isnull().sum()` | Cuenta de valores faltantes por columna |
| `df['col'].value_counts()` | Frecuencia de cada valor único |
| `df.nunique()` | Número de valores únicos por columna |


```python
import pandas as pd
import numpy as np

# Dataset simulado de una empresa de ventas
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'ID_Venta': range(1001, 1001+n),
    'Vendedor': np.random.choice(['Ana', 'Luis', 'María', 'Carlos', 'Sofía'], n),
    'Región': np.random.choice(['Norte', 'Sur', 'Centro', 'Occidente'], n),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Auriculares'], n),
    'Precio': np.random.randint(50000, 3000000, n),
    'Descuento': np.random.choice([0, 5, 10, 15, 20, None], n),
    'Unidades': np.random.randint(1, 20, n),
})

print('📋 EXPLORACIÓN INICIAL DEL DATASET')
print('='*45)
print(f'Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas')
print(f'\nPrimeras 5 filas:')
display(df.head())
```

```python
# Información de tipos y nulos — ¡el diagnóstico rápido!
print('📊 Información del DataFrame:')
df.info()

print('\n🔍 Valores faltantes por columna:')
nulos = df.isnull().sum()
pct_nulos = (df.isnull().sum() / len(df) * 100).round(2)
resumen_nulos = pd.DataFrame({'Nulos': nulos, '% del Total': pct_nulos})
print(resumen_nulos[resumen_nulos['Nulos'] > 0])
```

```python
# Estadísticas descriptivas
print('📈 Estadísticas descriptivas (columnas numéricas):')
display(df.describe().round(2))

print('\n🏪 Distribución de ventas por Región:')
print(df['Región'].value_counts())

print('\n🏅 Top 3 vendedores por volumen de ventas:')
print(df['Vendedor'].value_counts().head(3))
```

> **🎯 Rutina de Exploración Profesional**
>
> Cuando recibas un dataset nuevo, sigue siempre este orden:
> 
> 1. `df.shape` — ¿Qué tamaño tiene?
> 2. `df.head()` y `df.tail()` — ¿Qué aspecto tienen los datos?
> 3. `df.info()` — ¿Tipos correctos? ¿Hay nulos?
> 4. `df.describe()` — ¿Hay valores extremos o incoherentes?
> 5. `df['cat_col'].value_counts()` — ¿Distribución de categorías?

### ✅ Autocomprobación Rápida

**Pregunta:** Ejecutas `df.info()` y ves que la columna 'Precio' tiene tipo `object` en lugar de `float64`. ¿Qué podría significar?

<details>
<summary>💡 Ver respuesta</summary>

Significa que **la columna Precio tiene valores no numéricos** mezclados:

- Puede haber el símbolo `$` o separadores de miles como `.` ó `,` (ej: `'$2.500.000'`)
- Puede haber texto como `'N/A'` o `'-'` interpretado como cadena
- Puede haber espacios en blanco alrededor de los números

Solución: `df['Precio'] = df['Precio'].str.replace('[^0-9.]', '', regex=True).astype(float)`

</details>
