## 1. Nuestra tienda de mascotas de práctica 🐾

Para explorar necesitamos algo que explorar. Imagina que trabajas en una tienda de mascotas y te pasan el inventario en una tabla: nombre, especie, edad, precio y si ya está vacunada.

Vamos a crear esa tabla nosotros mismos, tal como si acabara de llegar del sistema de la tienda. A propósito dejamos **dos precios sin registrar** (`NaN`), porque en la vida real casi ningún dataset llega perfecto — siempre hay datos faltantes que debemos detectar.

````python
import pandas as pd
import numpy as np

mascotas = pd.DataFrame({
    'nombre': ['Rocky', 'Luna', 'Milo', 'Nina', 'Toby', 'Coco', 'Max', 'Bella', 'Simon', 'Lola'],
    'especie': ['Perro', 'Gato', 'Perro', 'Gato', 'Perro', 'Ave', 'Gato', 'Perro', 'Gato', 'Conejo'],
    'edad_meses': [8, 14, 3, 24, 60, 6, 36, 2, 18, 9],
    'precio': [450000, 300000, 500000, 250000, np.nan, 80000, 320000, 600000, np.nan, 120000],
    'vacunado': [True, True, False, True, True, False, True, False, True, True]
})

mascotas
````

### 🤔 ¿Qué acaba de pasar?

- Creamos un `DataFrame` a partir de un diccionario de Python: cada llave se convierte en una columna, y cada lista en los valores de esa columna, fila por fila.
- `np.nan` (de la librería NumPy) es la forma estándar de representar "no sé este dato" en Pandas — no es cero, no es texto vacío, es simplemente "falta".
- Esta tabla de 10 mascotas será nuestro laboratorio de pruebas durante todo el cuaderno.

---
## 2. Conociendo la tabla: medidas y primeras miradas 📏👀

Antes de leer una tabla larga, es útil preguntar cosas básicas, como cuando llegas a un salón nuevo: ¿cuántos puestos hay? ¿cómo se llaman las filas y las columnas? ¿qué pinta tienen los primeros y los últimos registros?

- `shape` responde "¿cuántas filas y columnas tiene esto?", devolviendo una tupla `(filas, columnas)`.
- `index` / `columns` responden "¿cómo se llama cada fila / columna?".
- `head(n)` / `tail(n)` te muestran las primeras / últimas `n` filas (5 por defecto), como hojear las primeras y últimas páginas de un cuaderno sin leerlo completo.

````python
print("Dimensiones (filas, columnas):", mascotas.shape)
print("Nombres de columnas:", list(mascotas.columns))

print("\n--- Primeras 3 filas ---")
display(mascotas.head(3))

print("--- Últimas 2 filas ---")
display(mascotas.tail(2))
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas.shape` nos dice `(10, 5)`: 10 mascotas (filas) y 5 datos por mascota (columnas). Fíjate que `shape` **no lleva paréntesis** — es un atributo, no una función, así que no calcula nada nuevo, solo reporta lo que ya existe.
- `mascotas.columns` nos da la lista de nombres de columna, en el mismo orden en que aparecen en la tabla.
- `head(3)` y `tail(2)` devuelven DataFrames nuevos y más pequeños — no modifican `mascotas`, solo nos "prestan" una vista parcial para revisar rápido.

## 3. La radiografía y las estadísticas: `info()` y `describe()` 🩻📊

- `info()` es como pedirle una radiografía completa a la tabla: te muestra, columna por columna, cuántos valores no están vacíos y de qué tipo es cada una.
- `describe()` va un paso más allá y calcula estadísticas automáticas — es "inteligente" según el tipo de columna: para columnas numéricas calcula promedio, mínimo, máximo, etc.

````python
mascotas.info()

print("\n--- describe() (solo columnas numéricas por defecto) ---")
display(mascotas.describe())
````

### 🤔 ¿Qué acaba de pasar?

- En `info()`, la columna **Non-Null Count** es clave para detectar datos faltantes: fíjate que `precio` muestra `8 non-null` en vez de `10` — esos son justo los dos precios que dejamos como `NaN` al crear la tabla.
- La columna **Dtype** te dice el tipo interno de cada columna: `int64` para enteros, `float64` para decimales (`precio` es `float64` justo *por* tener valores faltantes), `bool` para verdadero/falso, y `object` para texto.
- En `describe()`, Pandas ignoró las columnas de texto y booleanas, y nos dio `count`, `mean`, `std` (desviación estándar), `min`, los cuartiles y `max` de `edad_meses` y `precio`. Fíjate que `precio` muestra `count = 8`, no 10 — `describe()` también ignora los `NaN` en el cálculo. Si quisieras que describa **todas** las columnas (incluyendo texto), le pasarías `describe(include='all')`.

## 4. Preguntas puntuales: `mean()`, `unique()` y `value_counts()` 🧮

A veces no quieres todo el resumen, solo **una** respuesta concreta. Para eso existen funciones específicas sobre una sola columna (una `Series`):

- `mean()`: el promedio.
- `unique()`: la lista de valores distintos, sin repetir.
- `value_counts()`: cuántas veces aparece cada valor distinto — el "conteo de popularidad".

````python
print("Precio promedio:", mascotas['precio'].mean())
print("Especies distintas:", mascotas['especie'].unique())

print("\nConteo por especie:")
print(mascotas['especie'].value_counts())
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas['precio'].mean()` calculó el promedio usando solo los 8 precios que sí conocemos — los `NaN` no cuentan ni suman ni restan al total.
- `mascotas['especie'].unique()` devolvió cada especie **una sola vez**, sin importar cuántas mascotas de esa especie haya.
- `mascotas['especie'].value_counts()` fue un paso más allá: ordena las especies de la más común a la menos común junto con su conteo — 'Perro' y 'Gato' aparecen 4 veces cada uno, mientras que 'Ave' y 'Conejo' solo una.

## 5. Resumen relámpago ⚡

| Herramienta | ¿Para qué sirve? |
|---|---|
| `shape` | Dimensiones de la tabla: `(filas, columnas)`. |
| `index` / `columns` | Etiquetas de las filas y de las columnas. |
| `head(n)` / `tail(n)` | Primeras / últimas `n` filas (5 por defecto). |
| `info()` | "Radiografía": tipo de dato y conteo de valores no nulos por columna. |
| `describe()` | Estadísticas automáticas — numéricas por defecto, o todas con `include='all'`. |
| `mean()` | Promedio de una columna numérica (ignora los `NaN`). |
| `unique()` | Lista de valores distintos de una columna. |
| `value_counts()` | Cuántas veces aparece cada valor distinto, ordenado de mayor a menor. |
