## Modificando Datos con Pandas ✏️

Una vez que tienes los datos cargados, **necesitarás modificarlos**: corregir errores,
actualizar valores, crear nuevas columnas calculadas, etc.

### Los 3 Pilares de la Modificación

1. **Crear columna nueva** → `df['nueva_col'] = expresión`
2. **Modificar valores existentes** → `df.loc[condición, 'col'] = nuevo_valor`
3. **Reemplazar valores específicos** → `df['col'].replace({viejo: nuevo})`


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Auriculares'],
    'Precio_COP': [2500000, 85000, 120000, 850000, 350000],
    'Descuento_pct': [10, 0, 5, 15, 20],
    'Stock': [10, 50, 30, 15, 25],
    'Categoría': ['electrónica', 'periferico', 'periferico', 'electrónica', 'electrónica']
})

print('Dataset original:')
display(df)
```

```python
# 1. Crear nuevas columnas calculadas
df['Precio_USD'] = (df['Precio_COP'] / 4200).round(2)  # Tasa de cambio aproximada
df['Precio_Final'] = df['Precio_COP'] * (1 - df['Descuento_pct'] / 100)
df['Valor_Inventario'] = df['Precio_Final'] * df['Stock']

print('Con nuevas columnas calculadas:')
display(df[['Producto', 'Precio_COP', 'Descuento_pct', 'Precio_Final', 'Valor_Inventario']])
```

```python
# 2. Modificar valores específicos con loc
# El teclado tuvo un error de precio — corrección
df.loc[df['Producto'] == 'Teclado', 'Precio_COP'] = 135000
df.loc[df['Producto'] == 'Teclado', 'Precio_Final'] = 135000 * 0.95  # 5% descuento

print('Después de corregir el precio del Teclado:')
display(df[['Producto', 'Precio_COP', 'Precio_Final']])
```

```python
# 3. Estandarizar valores con replace y str
print('Categorías antes de estandarizar:', df['Categoría'].unique())

df['Categoría'] = df['Categoría'].replace({
    'electrónica': 'Electrónica',
    'periferico': 'Periférico'
})
print('Categorías después:', df['Categoría'].unique())

# 4. Columna condicional con np.where
df['Estado_Stock'] = np.where(df['Stock'] >= 20, '✅ Suficiente', '⚠️ Bajo')
display(df[['Producto', 'Stock', 'Estado_Stock', 'Categoría']])
```

### ✅ Autocomprobación Rápida

**Pregunta:** ¿Cuál es la diferencia entre `df['col'] = valor` y `df.loc[condición, 'col'] = valor`?

<details>
<summary>💡 Ver respuesta</summary>

- `df['col'] = valor` → **modifica TODA la columna** (reemplaza todos los valores)
- `df.loc[condición, 'col'] = valor` → modifica **solo las filas donde la condición es True**

Ejemplo: `df.loc[df['Stock'] == 0, 'Estado'] = 'Agotado'` solo cambia las filas con Stock=0.

</details>
