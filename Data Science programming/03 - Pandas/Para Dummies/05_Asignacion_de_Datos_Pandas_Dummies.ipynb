## ¿Qué vamos a aprender aquí? 🎈

Ya sabemos explorar una tabla y seleccionar exactamente el pedazo que nos interesa. El último paso natural es aprender a **modificar** esos datos: crear columnas nuevas y cambiar valores existentes.

Al terminar podrás:
1. Asignar un mismo valor a toda una columna (existente o nueva).
2. Asignar una lista de valores distintos, uno por fila.
3. Cambiar **solo algunas** filas que cumplan una condición, usando `loc`.

---
## 1. Nuestra tienda de mascotas, una vez más 🐾

Seguimos con la misma tabla de los cuadernos anteriores. Esta vez no solo la vamos a leer: la vamos a **editar**.

````python
import pandas as pd
import numpy as np

mascotas = pd.DataFrame({
    'nombre': ['Rocky', 'Luna', 'Milo', 'Nina', 'Toby', 'Coco', 'Max', 'Bella', 'Simon', 'Lola'],
    'especie': ['Perro', 'Gato', 'Perro', 'Gato', 'Perro', 'Ave', 'Gato', 'Perro', 'Gato', 'Conejo'],
    'edad_meses': [8, 14, 3, 24, 60, 6, 36, 2, 18, 9],
    'precio': [450000, 300000, 500000, 250000, np.nan, 80000, 320000, 600000, np.nan, 120000],
    'vacunado': [True, True, False, True, True, False, True, False, True, True]
})

mascotas
````

---
## 2. Asignar un valor igual para todos 🏷️

La forma más simple de crear o sobrescribir una columna: le asignas **un solo valor**, y Pandas lo repite automáticamente en todas las filas. Es como pegarle la misma etiqueta a cada producto de un estante — no importa cuántos productos haya, todos reciben la misma etiqueta.

Si el nombre de columna que usas **no existe todavía**, Pandas la crea; si **ya existe**, la sobrescribe por completo.

````python
# Creamos una columna nueva llamada 'tienda', igual para todas las filas
mascotas['tienda'] = 'Mundo Animal'
mascotas[['nombre', 'tienda']]
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas['tienda'] = 'Mundo Animal'` le dijo a Pandas: "crea (o sobrescribe) la columna `tienda`, y pon ese mismo texto en las 10 filas".
- Como `tienda` no existía antes, Pandas la agregó al final del DataFrame automáticamente — no hace falta ningún paso extra para "declarar" una columna nueva.

---
## 3. Asignar una lista completa de valores 📋

También puedes asignar una lista (o cualquier secuencia, como un `range`), siempre y cuando tenga **exactamente** la misma cantidad de elementos que filas tiene el DataFrame — uno por fila, en orden.

````python
# Un número de turno de atención, del 1 al 10, en el mismo orden de las filas
mascotas['turno'] = range(1, len(mascotas) + 1)
mascotas[['nombre', 'turno']]
````

### 🤔 ¿Qué acaba de pasar?

- `range(1, len(mascotas) + 1)` genera los números `1, 2, 3, ..., 10` — exactamente tantos como filas tiene `mascotas` (`len(mascotas)` es 10).
- Pandas asignó el primer número de la secuencia a la primera fila, el segundo a la segunda fila, y así sucesivamente. Si la lista tuviera menos o más elementos que filas, Pandas te lanzaría un error — las longitudes **deben** coincidir.

---
## 4. Cambiar solo algunas filas: asignación condicional con `loc` 🎯

Lo más común en la práctica no es cambiar toda la columna, sino **solo las filas que cumplen una condición** — por ejemplo, "a los animales más baratos, súbeles el precio mínimo". Para esto combinamos las máscaras booleanas del cuaderno anterior con `loc`, pero ahora del lado izquierdo de un `=`.

⚠️ **Importante:** para modificar filas condicionalmente, siempre usa `df.loc[condición, 'columna'] = valor`. Si intentas hacerlo solo con corchetes, corres el riesgo de modificar una *copia* de los datos en vez de la tabla original, y Pandas te lo advertirá con un `SettingWithCopyWarning`.

````python
print("Precio mínimo antes:", mascotas['precio'].min())

# A todas las mascotas que cuesten menos de 150.000, les subimos el precio a 150.000
mascotas.loc[mascotas['precio'] < 150000, 'precio'] = 150000

print("Precio mínimo después:", mascotas['precio'].min())
mascotas[['nombre', 'precio']]
````

### 🤔 ¿Qué acaba de pasar?

- `mascotas['precio'] < 150000` construyó la máscara booleana: `True` solo en las filas cuyo precio es menor a 150.000 (recuerda: las filas con `NaN` dan `False` en esta comparación, así que no se tocan).
- `mascotas.loc[máscara, 'precio'] = 150000` le dijo a Pandas: "en las filas donde la máscara es `True`, y **solo** en la columna `precio`, reemplaza el valor por 150.000".
- El resto de la tabla (nombres, especies, las filas que ya costaban 150.000 o más) quedó completamente intacto — la asignación con `loc` es quirúrgica, no afecta nada fuera de lo que le pediste.

## 5. Resumen relámpago ⚡

| Forma de asignar | ¿Cuándo usarla? |
|---|---|
| `df['col'] = valor_único` | Poner el mismo valor en toda una columna (existente o nueva). |
| `df['col'] = lista_o_range` | Poner un valor distinto por fila; la longitud debe coincidir exactamente con el número de filas. |
| `df.loc[condición, 'col'] = valor` | Cambiar solo las filas que cumplen una condición, sin tocar el resto de la tabla. |

**Regla de oro:** siempre que la asignación dependa de una condición, pásala por `loc` — nunca directamente con corchetes sueltos.
