## ¿Cómo actualizar datos en una tabla? ✏️

Podemos modificar una celda individual, una columna completa o actualizar solo las filas que cumplan una condición específica.


```python
import pandas as pd

df_empleados = pd.DataFrame({
    "Empleado": ["Javier", "Diana", "Felipe", "Karen"],
    "Salario_Base": [2000000, 3500000, 1800000, 4200000],
    "Desempenio": ["Alto", "Excelente", "Medio", "Excelente"]
})

# 1. Asignar un bono del 10% a TODOS:
df_empleados["Bono_General"] = df_empleados["Salario_Base"] * 0.10

# 2. Asignar un bono especial SOLO a quienes tuvieron desempeño 'Excelente':
df_empleados["Bono_Especial"] = 0
df_empleados.loc[df_empleados["Desempenio"] == "Excelente", "Bono_Especial"] = 500000

# 3. Calcular el salario total final:
df_empleados["Salario_Total"] = df_empleados["Salario_Base"] + df_empleados["Bono_General"] + df_empleados["Bono_Especial"]

display(df_empleados)
```
