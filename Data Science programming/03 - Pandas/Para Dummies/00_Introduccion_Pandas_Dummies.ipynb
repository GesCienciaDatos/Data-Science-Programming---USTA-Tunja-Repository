## 1. Pandas es Excel, pero programable 📊

Imagina que en vez de abrir Excel y hacer clic para ordenar, filtrar o sumar columnas, pudieras **escribir instrucciones** para que el computador haga exactamente eso, y las volviera a ejecutar en segundos sobre un archivo de 200,000 filas sin quejarse.

Eso es, en esencia, **Pandas**: una biblioteca de Python para trabajar con **datos en forma de tabla** (filas y columnas, como una hoja de cálculo o una tabla de base de datos). Fue creada en 2008 por Wes McKinney, quien necesitaba analizar datos financieros más rápido de lo que Excel le permitía.

> 📌 **Para recordar:** si NumPy es la calculadora superrápida para números "sueltos", Pandas es la hoja de cálculo superrápida para números organizados en tablas.

---
## 2. Pandas no inventó la rueda: usa NumPy por debajo ⚙️

En el cuaderno de NumPy vimos que un array (`ndarray`) es como una huevera: ordenada, compacta y muy rápida de procesar. Pandas toma esa idea y le agrega dos cosas que a una huevera común le faltan:

- **Etiquetas:** en vez de que cada compartimento se identifique solo por un número (0, 1, 2...), le puedes poner un nombre ("Bogotá", "2024-01-01", "Cliente A"...).
- **Manejo de "huecos":** si a una tabla real le falta un dato (algo muy común en la vida real), Pandas lo representa con `NaN` ("Not a Number") y sabe cómo trabajar alrededor de esos huecos sin romperse.

Por dentro, cada columna de una tabla de Pandas sigue siendo, básicamente, un array de NumPy con una etiqueta pegada encima. Por eso Pandas hereda la velocidad de NumPy, pero se siente mucho más cómodo para trabajar con datos "del mundo real".

## 3. Los dos pilares: Series y DataFrame 🧱

Todo en Pandas se construye con dos piezas:

| Pieza | ¿Qué es? | Analogía |
|---|---|---|
| **Series** | Una sola columna de datos, con una etiqueta por cada valor. | Una fila de casilleros con nombre, todos guardando el mismo tipo de cosa (por ejemplo, solo precios). |
| **DataFrame** | Varias Series "pegadas" una junto a otra, formando una tabla. | Una hoja de cálculo completa: varias columnas, cada una con su propio nombre. |

No necesitas memorizar esto todavía — lo vas a ver y a tocar en el siguiente cuaderno. Por ahora, quédate con la idea de que **un DataFrame es, literalmente, un grupo de Series compartiendo la misma tabla**.

```python
# Antes de nada, importamos pandas (casi siempre con el apodo "pd")
import pandas as pd

print("Pandas está listo para usarse.")
print(f"Versión instalada: {pd.__version__}")
```

### 🤔 ¿Qué acaba de pasar?

- `import pandas as pd` carga la biblioteca y le pone el apodo `pd`, que es una convención tan universal que casi todo el mundo la usa (igual que `np` para NumPy).
- `pd.__version__` nos deja ver qué versión tenemos instalada — útil cuando algo funciona distinto en un computador y en otro.
- Todavía no hemos creado ninguna tabla: solo dejamos la herramienta lista sobre la mesa.

```python
# Creamos nuestra primera tabla (DataFrame) a partir de un diccionario de Python
datos = {
    'Ciudad': ['Bogotá', 'Medellín', 'Tunja', 'Cali'],
    'Poblacion_miles': [7900, 2600, 180, 2200],
    'Sede_USTA': [True, True, True, False]
}

df_ciudades = pd.DataFrame(datos)
df_ciudades
```

### 🤔 ¿Qué acaba de pasar?

- El diccionario `datos` tiene tres "llaves" (`'Ciudad'`, `'Poblacion_miles'`, `'Sede_USTA'`): cada una se convierte en una **columna** del DataFrame.
- Cada lista dentro del diccionario se convierte en los **valores** de esa columna, fila por fila.
- `pd.DataFrame(datos)` empaqueta todo eso en una tabla real, con un índice numérico automático (0, 1, 2, 3) a la izquierda.
- Si en un notebook de Jupyter la última línea de una celda es una tabla, Jupyter la muestra automáticamente con un formato bonito — por eso ves una tabla en vez de texto plano.

## 4. ¿Por qué tanto alboroto con Pandas? 🚀

Podrías preguntarte: "¿y por qué no simplemente uso listas y diccionarios de Python, como hasta ahora?" Estas son algunas razones por las que casi cualquier analista de datos usa Pandas todos los días:

- **Alineación automática:** si combinas dos tablas, Pandas hace coincidir las filas por su etiqueta automáticamente, en vez de que tú tengas que hacerlo a mano.
- **Datos faltantes sin dramas:** si a una fila le falta un dato, Pandas lo marca como `NaN` y sigue funcionando, sin que el programa se caiga.
- **Operaciones tipo "hoja de cálculo avanzada":** agrupar, ordenar, cruzar tablas (como un `VLOOKUP` con esteroides) en una sola línea de código.
- **Lee y escribe casi cualquier formato:** CSV, Excel, JSON, bases de datos SQL... con una sola función cada vez.

En los siguientes cuadernos iremos probando cada una de estas ventajas con ejemplos pequeños.

## 5. El camino típico de un proyecto de datos con Pandas 🗺️

Casi cualquier análisis de datos sigue, más o menos, el mismo camino:

1. **Traer los datos** (leer un CSV, un Excel, una base de datos).
2. **Mirar qué hay** (¿cuántas filas y columnas tengo? ¿qué tipo de datos son?).
3. **Limpiar** (¿hay datos faltantes o repetidos? ¿hay errores?).
4. **Transformar** (crear columnas nuevas, agrupar, combinar tablas).
5. **Usar los resultados** (gráficas, reportes, o alimentar un modelo de Machine Learning).

Pandas es la herramienta principal para los pasos 1 a 4 — y eso es, básicamente, todo lo que vamos a recorrer en este módulo.
