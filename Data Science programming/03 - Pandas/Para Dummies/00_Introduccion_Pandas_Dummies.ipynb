## ¿Qué es Pandas y por qué lo amarás? 🐼

Si alguna vez has usado **Microsoft Excel o Google Sheets**, ¡ya tienes el 70% de la intuición necesaria para entender Pandas!

### 💡 La Analogía de Excel con Esteroides:
- **Excel Tradicional:** Es genial para hojas pequeñas, pero cuando abres un archivo con 500,000 filas se congela, se cierra solo y pierdes tu trabajo. Además, si quieres repetir el mismo reporte cada lunes, tienes que volver a hacer los mismos clics manualmente.
- **Pandas:** Es una librería de Python que manipula tablas de millones de filas en fracciones de segundo. Y lo mejor: escribes el código una sola vez y lo automatizas para siempre con un solo clic.

---
### Las Dos Estructuras Rey de Pandas:
1. **Series (1D):** Es **una sola columna** de datos (como la lista de edades de tus clientes).
2. **DataFrame (2D):** Es **la tabla completa** con filas y columnas (como una hoja de cálculo completa con Nombre, Edad, Ciudad y Saldo).
