## ¿Qué es Pandas y por qué lo amarás? 🐼

Si alguna vez has usado **Microsoft Excel o Google Sheets**, ¡ya tienes el 70% de la intuición para entender Pandas!

### 💡 Excel con Esteroides — La Gran Analogía

| Característica | 📊 Excel Tradicional | 🐼 Pandas |
|---|---|---|
| **Filas máximas** | ~1,000,000 (y se congela) | Millones sin problema |
| **Automatización** | Macros complicadas | Una sola línea de Python |
| **Lectura de archivos** | Solo .xlsx, .csv | CSV, Excel, SQL, JSON, Parquet, Feather... |
| **Velocidad** | Lento en datos grandes | Ultra rápido |
| **Reproducibilidad** | Difícil, clicks manuales | 100% reproducible con código |

### 🏗️ Las Dos Estructuras Fundamentales

1. **`pd.Series`** — Una sola columna de datos, como una lista con etiquetas.
2. **`pd.DataFrame`** — La tabla completa: filas + columnas (como la hoja de cálculo, pero con superpoderes).


```python
import pandas as pd
import numpy as np

print(f'✅ Pandas versión: {pd.__version__}')
print(f'✅ NumPy versión:  {np.__version__}')

# Creamos una Series simple — lista de notas de estudiantes
notas = pd.Series([4.5, 3.8, 4.9, 2.7, 4.1],
                   index=['Laura', 'Camilo', 'Daniela', 'Esteban', 'María'],
                   name='Calificaciones')
print('\n📊 Serie de Notas:')
print(notas)
print(f'  Tipo: {type(notas)}')
```

```python
# Creamos un DataFrame — tabla completa de estudiantes
df_estudiantes = pd.DataFrame({
    'Nombre': ['Laura', 'Camilo', 'Daniela', 'Esteban', 'María'],
    'Nota':   [4.5, 3.8, 4.9, 2.7, 4.1],
    'Ciudad': ['Tunja', 'Bogotá', 'Tunja', 'Medellín', 'Cali'],
    'Aprobado': [True, True, True, False, True]
})

print('📋 DataFrame de Estudiantes:')
display(df_estudiantes)
print(f'\nDimensiones: {df_estudiantes.shape[0]} filas x {df_estudiantes.shape[1]} columnas')
print(f'Columnas: {list(df_estudiantes.columns)}')
```

> **🔑 Truco de Oro**
>
> Recuerda siempre: **Series = una columna** con índice.  
> **DataFrame = múltiples Series pegadas lado a lado**.
> 
> Puedes extraer una Serie de un DataFrame así:  
> `df['Nombre']` → devuelve la columna 'Nombre' como una Serie.

**Pregunta:** ¿Cuál es la diferencia principal entre una `pd.Series` y un `pd.DataFrame`?

<details>
<summary>💡 Ver respuesta</summary>

- Una **Series** es **unidimensional** (1 sola columna de datos con un índice).
- Un **DataFrame** es **bidimensional** (múltiples columnas, como una tabla).
- Técnicamente, un DataFrame es un diccionario de Series.
