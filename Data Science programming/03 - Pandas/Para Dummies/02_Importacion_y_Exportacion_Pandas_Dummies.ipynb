## Cómo Leer y Guardar Datos con Pandas 📂

En la vida real, **nunca escribes los datos a mano** en Python. Los datos vienen de:
archivos CSV, hojas de Excel, bases de datos SQL, APIs JSON, etc.

### Guía Rápida de Lectura

| Formato | Función | Parámetro clave |
|---|---|---|
| CSV | `pd.read_csv('file.csv')` | `sep=','`, `encoding='utf-8'` |
| Excel | `pd.read_excel('file.xlsx')` | `sheet_name=0` |
| JSON | `pd.read_json('file.json')` | `orient='records'` |
| SQL | `pd.read_sql(query, conn)` | Requiere `sqlalchemy` |
| Parquet | `pd.read_parquet('file.parquet')` | Ultra rápido, big data |


```python
import pandas as pd
import io

# Simulamos un CSV en memoria (igual que leer un archivo real)
csv_simulado = """Nombre,Edad,Ciudad,Salario
Laura Pérez,28,Tunja,3200000
Carlos Gómez,35,Bogotá,5800000
Valentina Ríos,24,Cali,2900000
Miguel Torres,42,Medellín,7100000
Sofía Herrera,31,Tunja,4300000
"""

df = pd.read_csv(io.StringIO(csv_simulado))
print('✅ Datos cargados desde CSV:')
display(df)
print(f'\nForma: {df.shape}')
print(f'Columnas: {list(df.columns)}')
```

```python
# Parámetros útiles de read_csv
csv_con_separador_punto_coma = """Nombre;Edad;Ciudad
Ana;25;Tunja
Pedro;30;Bogotá
"""

# Cuando el CSV usa ';' como separador (común en archivos de Excel europeos y colombianos)
df_puntocoma = pd.read_csv(io.StringIO(csv_con_separador_punto_coma), sep=';')
print('CSV con separador punto y coma:')
display(df_puntocoma)

# Exportar a CSV
df.to_csv('mi_dataset.csv', index=False, encoding='utf-8')
print('\n✅ Archivo guardado como mi_dataset.csv (index=False evita guardar el índice numérico)')
```

```python
# Lectura con dtypes explícitos y manejo de NaN
csv_con_nulos = """ID,Temperatura,Humedad,Lluvia
1,22.5,65,No
2,,70,Sí
3,19.8,,No
4,24.1,80,Sí
"""

df_clima = pd.read_csv(io.StringIO(csv_con_nulos))
print('Dataset con valores faltantes:')
display(df_clima)
print(f'\n🔍 Valores nulos por columna:')
print(df_clima.isnull().sum())
```

> **💡 Tip Profesional: index=False**
>
> Siempre usa `df.to_csv('archivo.csv', index=False)` al exportar.
> 
> Sin `index=False`, Pandas guarda una columna extra con números 0,1,2...
> que no necesitas y confunde cuando vuelves a cargar el archivo.


---
### ✅ Autocomprobación Rápida

**Pregunta:** ¿Por qué deberías usar `index=False` cuando guardas un DataFrame con `to_csv()`?

<details>
<summary>💡 Ver respuesta</summary>

Porque sin `index=False`, Pandas añade una **columna adicional** con el índice numérico (0, 1, 2...)
que no pertenece a tus datos originales.

Al volver a cargar el archivo, esa columna extra aparecería como `Unnamed: 0`, causando confusión.

</details>
