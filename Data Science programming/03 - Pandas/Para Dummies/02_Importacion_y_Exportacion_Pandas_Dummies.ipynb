## 1. ¿Qué es un archivo CSV? 📄

Imagina una hoja de cálculo, pero guardada de la forma más simple posible: sin colores, sin fórmulas, solo texto plano donde cada fila es una línea, y cada columna está separada por una coma. Así de simple:

```csv
Ciudad,Poblacion_miles
Bogotá,7900
Tunja,180
```

**CSV** significa *Comma-Separated Values* ("valores separados por coma"). Es, por lejos, el formato más común para compartir datos tabulares, porque cualquier programa (Excel, Google Sheets, Pandas, hasta un bloc de notas) puede abrirlo.

> 📌 **Para recordar:** un CSV es a una tabla lo que una lista de la compra escrita a mano es a un carrito de supermercado — el contenido "en crudo", sin decoración.

## 2. Primero, creemos un archivo para practicar 🧪

Para no depender de descargar nada de internet, vamos a crear nuestra propia tabla pequeña y **guardarla** en un archivo CSV con `DataFrame.to_csv()`. Así tendremos un archivo real con el cual practicar a leer datos, tal como harías con cualquier dataset del mundo real.

````python
import pandas as pd

# Creamos una tabla pequeña de práctica
frutas = pd.DataFrame({
    'Fruta': ['Manzana', 'Banana', 'Cereza', 'Durazno'],
    'Precio': [1500, 800, 4200, 2600],
    'Stock_kg': [30, 45, 12, 20]
})

# La guardamos en un archivo CSV, sin incluir el índice numérico automático
frutas.to_csv('frutas_practica.csv', index=False)
print("Archivo 'frutas_practica.csv' guardado con éxito.")
````

### 🤔 ¿Qué acaba de pasar?

- `frutas.to_csv('frutas_practica.csv', ...)` toma el DataFrame que teníamos en memoria y lo escribe en un archivo de texto plano, separado por comas, en la misma carpeta donde corre este cuaderno.
- El parámetro `index=False` le dice a Pandas "no guardes el número de fila (0, 1, 2, 3) como si fuera una columna más" — si no lo pusiéramos, el archivo tendría una columna extra sin nombre al principio.
- A partir de aquí, `frutas_practica.csv` existe como un archivo real en el disco — exactamente como si alguien nos lo hubiera enviado por correo.

---
## 3. Leyendo el archivo de vuelta con `read_csv` 📖

Ahora hagamos lo que se hace el 90% de las veces en un proyecto real: **leer** un archivo que ya existe, usando `pd.read_csv()`.

````python
# Leemos el archivo que acabamos de crear
frutas_leidas = pd.read_csv('frutas_practica.csv')
frutas_leidas
````

### 🤔 ¿Qué acaba de pasar?

- `pd.read_csv('frutas_practica.csv')` abre el archivo, entiende que las comas separan columnas, y arma un DataFrame completo — con tipos de datos ya detectados automáticamente (`Precio` y `Stock_kg` como números, `Fruta` como texto).
- Pandas volvió a poner un índice numérico automático (0, 1, 2, 3) porque el archivo no tenía ninguna columna marcada como "esta es el índice" — nosotros la quitamos a propósito en el paso anterior con `index=False`.
- Este es el flujo más común en la práctica: `to_csv()` para guardar, `read_csv()` para volver a cargar.

---
## 4. Controlando la lectura: `index_col` 🎛️

A veces el archivo CSV sí trae una columna pensada para ser el índice (por ejemplo, un identificador único de cada fila). Si no le decimos nada a Pandas, la tratará como una columna normal. El parámetro `index_col` le dice **cuál columna usar como índice** en vez de crear uno nuevo desde cero.

Vamos a guardar de nuevo el archivo, esta vez sí incluyendo el índice, para ver la diferencia.

````python
# Esta vez SÍ guardamos el índice (el nombre de la fruta) como columna
frutas_indexadas = frutas.set_index('Fruta')
frutas_indexadas.to_csv('frutas_practica_v2.csv')

# Leemos indicando que la primera columna del archivo debe ser el índice
frutas_v2 = pd.read_csv('frutas_practica_v2.csv', index_col=0)
frutas_v2
````

### 🤔 ¿Qué acaba de pasar?

- `frutas.set_index('Fruta')` le dice a Pandas "usa la columna `Fruta` como etiqueta de fila en vez de columna normal" — ahora cada fila se identifica por su nombre de fruta.
- Al guardar ese DataFrame con `to_csv()` (sin `index=False` esta vez), el archivo sí incluye esa etiqueta como primera columna.
- Al leerlo con `index_col=0` le decimos a Pandas "la columna en la posición 0 (la primera) no es un dato más: úsala como índice de filas".
- Resultado: en vez de un índice numérico 0, 1, 2, 3, tenemos como índice el nombre de cada fruta — mucho más fácil de leer.

## 5. Pandas lee mucho más que solo CSV 🌐

El CSV es el formato más simple, pero Pandas tiene funciones "hermanas" para casi cualquier formato que te encuentres en el mundo real:

| Formato | Leer | Escribir |
|---|---|---|
| CSV | `pd.read_csv()` | `df.to_csv()` |
| Excel (`.xlsx`) | `pd.read_excel()` | `df.to_excel()` |
| JSON | `pd.read_json()` | `df.to_json()` |
| Base de datos SQL | `pd.read_sql()` | `df.to_sql()` |
| Parquet (formato comprimido, usado en Big Data) | `pd.read_parquet()` | `df.to_parquet()` |

La buena noticia: **todas se usan de forma muy parecida** — casi siempre `pd.read_algo('ruta_del_archivo')` para traer datos, y `df.to_algo('ruta_del_archivo')` para guardarlos.

## 6. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| CSV | Un archivo de texto plano donde cada línea es una fila y las comas separan columnas. |
| `to_csv()` | Guarda un DataFrame en un archivo CSV en el disco. |
| `read_csv()` | Lee un archivo CSV y lo convierte en un DataFrame. |
| `index=False` (al guardar) | Evita guardar el número de fila automático como si fuera una columna más. |
| `index_col=0` (al leer) | Le dice a Pandas que use la primera columna del archivo como índice de filas. |
| `nrows=N` (al leer) | Carga solo las primeras N filas — útil para dar un vistazo rápido a archivos grandes. |
| Pandas lee más que CSV | Excel, JSON, SQL y Parquet siguen el mismo patrón: `read_algo()` / `to_algo()`. |
