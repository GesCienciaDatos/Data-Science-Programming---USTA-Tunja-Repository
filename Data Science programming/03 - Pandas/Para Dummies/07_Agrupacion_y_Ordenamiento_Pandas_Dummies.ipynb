## Agrupando Datos como un Pro con groupby 📊

**`groupby`** es la función más poderosa de Pandas para análisis. Es el equivalente de las
**tablas dinámicas de Excel** (Pivot Tables), pero infinitamente más flexible.

### La Fórmula Mágica de groupby

```python
df.groupby('columna_agrupadora')['columna_a_calcular'].función_agregadora()
```

### Funciones de Agregación Más Comunes

| Función | ¿Qué calcula? |
|---|---|
| `.sum()` | Suma total |
| `.mean()` | Promedio |
| `.count()` | Conteo de elementos |
| `.min()` / `.max()` | Mínimo / Máximo |
| `.std()` | Desviación estándar |
| `.agg({'col1': 'sum', 'col2': 'mean'})` | Múltiples agregaciones |


````python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 80
df = pd.DataFrame({
    'Vendedor': np.random.choice(['Ana', 'Luis', 'María', 'Carlos'], n),
    'Región': np.random.choice(['Norte', 'Sur', 'Centro'], n),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Monitor'], n),
    'Ventas': np.random.randint(500000, 5000000, n),
    'Unidades': np.random.randint(1, 15, n),
})

print(f'Dataset: {len(df)} registros de ventas')
print('\nPrimeros registros:')
display(df.head())
````

````python
# ── Agregaciones básicas ─────────────────────────────────────────────
print('Total de ventas por Vendedor:')
ventas_por_vendedor = df.groupby('Vendedor')['Ventas'].sum().sort_values(ascending=False)
print(ventas_por_vendedor)

print('\nPromedio de ventas por Región:')
promedio_region = df.groupby('Región')['Ventas'].mean().round(0).astype(int)
print(promedio_region)
````

````python
# ── Múltiples agregaciones con agg ────────────────────────────────────
resumen = df.groupby('Vendedor').agg(
    Total_Ventas=('Ventas', 'sum'),
    Promedio_Venta=('Ventas', 'mean'),
    Total_Unidades=('Unidades', 'sum'),
    Num_Transacciones=('Ventas', 'count')
).round(0).astype(int)

resumen = resumen.sort_values('Total_Ventas', ascending=False)
print('Resumen completo por Vendedor:')
display(resumen)
````

````python
# ── Ordenamiento con sort_values ──────────────────────────────────────
print('Top 10 ventas individuales (ordenadas de mayor a menor):')
display(df.sort_values('Ventas', ascending=False).head(10))

print('\nOrdenar por múltiples columnas:')
display(df.sort_values(['Región', 'Ventas'], ascending=[True, False]).head(8))
````

**Pregunta:** ¿Qué hace exactamente `df.groupby('Región')['Ventas'].mean()`?

<details>
<summary>💡 Ver respuesta</summary>

1. **`groupby('Región')`** → Agrupa todas las filas que tienen el mismo valor en la columna 'Región'
2. **`['Ventas']`** → Selecciona solo la columna 'Ventas' dentro de cada grupo
3. **`.mean()`** → Calcula el **promedio** de Ventas para cada grupo

Resultado: una Series con el promedio de ventas para cada Región.
