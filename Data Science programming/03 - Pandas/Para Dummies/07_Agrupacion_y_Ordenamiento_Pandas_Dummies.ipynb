## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **"para no ingenieros"** del módulo 07 de Pandas. Si el cuaderno principal usó la palabra `groupby()` y sentiste que era abstracto, aquí lo vamos a explicar con una analogía muy concreta: **hacer montones** de cosas parecidas y luego medir cada montón.

No necesitas ningún dataset complicado — usaremos una lista pequeña de frutas vendidas en un mercado.

## 1. Agrupar es hacer montones sobre la mesa 🍎🍌🍊

Imagina que vuelcas una canasta de frutas mezcladas sobre una mesa: manzanas, bananos y naranjas, todas revueltas. Lo primero que harías para entenderlas mejor sería **separarlas en montones**: un montón de manzanas, uno de bananos, uno de naranjas.

Eso es exactamente lo que hace `groupby()`: agarra todas las filas de una tabla y las separa en "montones" según el valor de una columna (por ejemplo, la columna `fruta`). Una vez que tienes los montones armados, puedes preguntarle cosas a cada uno: ¿cuántas hay en este montón?, ¿cuál es el precio promedio de este montón?, ¿cuál es la más cara de este montón?

Vamos a construir nuestra "canasta de frutas" como un DataFrame.

````python
import pandas as pd

ventas = pd.DataFrame({
    'fruta': ['manzana', 'banano', 'naranja', 'manzana', 'banano', 'manzana', 'naranja'],
    'ciudad': ['Tunja', 'Tunja', 'Bogotá', 'Bogotá', 'Bogotá', 'Tunja', 'Tunja'],
    'precio': [1500, 800, 1200, 1600, 850, 1400, 1100],
    'cantidad': [10, 20, 15, 8, 25, 12, 18],
})
ventas
````

### 🤔 ¿Qué acaba de pasar?

- Creamos una tabla de ventas con 7 filas, mezclando 3 frutas distintas en 2 ciudades.
- Fíjate que la columna `fruta` se repite: hay varias filas de `manzana`, varias de `banano`, varias de `naranja`. Justo esas repeticiones son las que `groupby()` va a usar para armar los montones.

---
## 2. Armando el primer montón: `groupby()` + una función 🗂️

Para contar cuántas filas hay en cada montón de fruta, agrupamos por `fruta` y aplicamos `count()`:

````python
ventas.groupby('fruta')['cantidad'].count()
````

### 🤔 ¿Qué acaba de pasar?

- `ventas.groupby('fruta')` arma 3 montones invisibles: uno con todas las filas de `manzana`, uno con las de `banano`, uno con las de `naranja`.
- `['cantidad']` le dice "de cada montón, mírame solo la columna `cantidad`".
- `.count()` cuenta cuántos valores hay en cada montón.

Puedes cambiar `count()` por **cualquier función de resumen**: `mean()` para el promedio, `min()` para el mínimo, `max()` para el máximo, `sum()` para la suma, y muchas más.

````python
# Precio promedio de cada fruta
print(ventas.groupby('fruta')['precio'].mean())

# Venta con menor cantidad dentro de cada montón de fruta
print(ventas.groupby('fruta')['cantidad'].min())
````

### 🤔 ¿Qué acaba de pasar?

- Ambas líneas arman los mismos 3 montones de fruta, pero les preguntan cosas distintas: una calcula el **promedio** de `precio`, la otra busca el **mínimo** de `cantidad`.
- Esta es la gran ventaja de `groupby()`: una vez armados los montones, puedes aplicarles cualquier función de resumen que necesites.

---
## 3. Montones dentro de montones: agrupar por varias columnas 🧩

También puedes pedir montones más finos, agrupando por **más de una columna a la vez**. Es como separar primero por tipo de fruta, y dentro de cada montón de fruta, separar otra vez por ciudad.

Para eso, le pasamos una **lista** de columnas a `groupby()`:

````python
ventas.groupby(['fruta', 'ciudad'])['precio'].mean()
````

### 🤔 ¿Qué acaba de pasar?

- Ahora los montones son más específicos: "manzana en Bogotá", "manzana en Tunja", "banano en Bogotá", etc.
- El resultado tiene un índice de dos niveles (fruta y ciudad) — a esto se le llama `MultiIndex`, pero por ahora basta con saber que es una tabla de resultados organizada en dos capas.

---
## 4. Poniendo los montones en orden: `sort_values()` 🔀

Cuando agrupamos, el resultado no siempre viene ordenado de la forma más útil para leerlo. Para eso usamos `sort_values()`, que ordena de menor a mayor (o de mayor a menor, si le pasamos `ascending=False`).

Por ejemplo: ¿cuál es la fruta con el precio promedio más alto?

````python
precio_promedio = ventas.groupby('fruta')['precio'].mean()
precio_promedio.sort_values(ascending=False)
````

### 🤔 ¿Qué acaba de pasar?

- Primero armamos los montones y calculamos el precio promedio de cada uno (igual que antes).
- Luego `sort_values(ascending=False)` reordena esa lista de resultados, poniendo el valor **más alto primero**.
- Si quisiéramos del más barato al más caro, usaríamos `ascending=True` (que además es el valor por defecto, así que ni haría falta escribirlo).

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| `groupby('col')` | Arma "montones" de filas según los valores repetidos de una columna. |
| `groupby(...)['col2'].funcion()` | De cada montón, toma la columna `col2` y le aplica una función de resumen (`count`, `mean`, `min`, `max`, `sum`...). |
| `groupby(['col1', 'col2'])` | Arma montones dentro de montones, agrupando por varias columnas a la vez. |
| `sort_values()` | Ordena un resultado de menor a mayor (por defecto) o de mayor a menor con `ascending=False`. |
