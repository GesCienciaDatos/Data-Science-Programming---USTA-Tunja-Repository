## Construyendo Series y DataFrames desde Cero 🏗️

Pandas te permite crear tablas de datos de **múltiples fuentes**:
desde listas de Python, desde diccionarios, desde archivos CSV, desde bases de datos SQL, etc.

### Formas de Crear un DataFrame

| Fuente | Código | Cuándo usarlo |
|---|---|---|
| Diccionario Python | `pd.DataFrame({'col': [...]})` | Datos escritos a mano |
| Lista de diccionarios | `pd.DataFrame([{...}, {...}])` | Datos de APIs JSON |
| Array de NumPy | `pd.DataFrame(np.array(...))` | Computación científica |
| Archivo CSV | `pd.read_csv('archivo.csv')` | La fuente más común |


```python
import pandas as pd
import numpy as np

# Forma 1: Desde diccionario (la más común)
df1 = pd.DataFrame({
    'Producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor'],
    'Precio':   [2500000, 85000, 120000, 850000],
    'Stock':    [10, 50, 30, 15]
})
print('Forma 1 - Desde diccionario:')
display(df1)
```

```python
# Forma 2: Desde lista de diccionarios (común en APIs REST)
registros = [
    {'Nombre': 'Ana', 'Edad': 25, 'Ciudad': 'Tunja'},
    {'Nombre': 'Luis', 'Edad': 32, 'Ciudad': 'Bogotá'},
    {'Nombre': 'Sara', 'Edad': 28, 'Ciudad': 'Medellín'},
]
df2 = pd.DataFrame(registros)
print('Forma 2 - Desde lista de diccionarios:')
display(df2)

# Inspeccionando el DataFrame
print(f'\nTipos de datos por columna:')
print(df2.dtypes)
```

```python
# Forma 3: Series personalizadas con índice propio
ventas_enero = pd.Series({'Laptop': 5, 'Mouse': 20, 'Teclado': 12, 'Monitor': 3},
                          name='Enero')
ventas_febrero = pd.Series({'Laptop': 8, 'Mouse': 25, 'Teclado': 9, 'Monitor': 7},
                            name='Febrero')

# Uniendo Series en un DataFrame
df_ventas = pd.DataFrame([ventas_enero, ventas_febrero])
print('Ventas por mes y producto:')
display(df_ventas)
print(f'\nTotal vendido por producto:')
print(df_ventas.sum())
```
