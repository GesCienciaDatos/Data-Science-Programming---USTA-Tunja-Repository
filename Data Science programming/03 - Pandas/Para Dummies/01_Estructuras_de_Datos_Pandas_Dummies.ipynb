## 1. La Series: una fila de casilleros con nombre 🏷️

Piensa en los casilleros de un gimnasio o de un colegio: cada uno tiene un número o un nombre pegado en la puerta, y adentro guarda **un solo tipo de cosa** (por ejemplo, solo mochilas).

Una **Series** de Pandas es exactamente eso: una columna de datos donde cada valor tiene una etiqueta al lado (el **índice**). Todos los valores de una Series son del mismo tipo — solo números, o solo texto, nunca una mezcla (igual que un array de NumPy, pero con nombres en vez de solo números de posición).

La forma más simple de crear una es:

```python
s = pd.Series(datos, index=etiquetas)
```

Vamos a probarlo.

````python
# Es buena práctica activar esto en Jupyter para ver varias salidas por celda
try:
    from IPython.core.interactiveshell import InteractiveShell
    InteractiveShell.ast_node_interactivity = 'all'
except Exception:
    pass

import pandas as pd

# Una Series muy simple, sin etiquetas propias (Pandas pone 0, 1, 2... por defecto)
temperaturas = pd.Series([18, 21, 19, 23])
print("Sin etiquetas propias:")
print(temperaturas)

# Ahora sí, con etiquetas propias y un nombre para la Series completa
ventas = pd.Series(
    [30, 35, 40],
    index=['Enero', 'Febrero', 'Marzo'],
    name='Ventas_2024'
)
print("\nCon etiquetas propias y nombre:")
ventas
````

### 🤔 ¿Qué acaba de pasar?

- `pd.Series([18, 21, 19, 23])` guarda esos 4 números en una sola columna etiquetada. Como no le dimos etiquetas propias, Pandas usó números de posición automáticos: 0, 1, 2, 3 — igual que el índice por defecto de una lista de Python.
- En `ventas`, el parámetro `index=['Enero', 'Febrero', 'Marzo']` reemplaza los casilleros numerados por casilleros con nombre, y `name='Ventas_2024'` le pone un rótulo a toda la Series — algo así como el nombre que tendría esta columna si viviera dentro de una tabla más grande (un DataFrame).
- A la izquierda de cada valor ves el índice (la "puerta del casillero"); a la derecha, el dato. Recuerda: **una Series es, en el fondo, una sola columna de un futuro DataFrame.**

## 2. Una Series es dos cosas a la vez 🎭

Lo interesante de una Series es que se comporta como **dos estructuras que ya conoces**, al mismo tiempo:

- **Como un array de NumPy:** puedes multiplicarla, sumarla, o "rebanarla" (slicing) con `[inicio:fin]`.
- **Como un diccionario de Python:** puedes pedir un valor directamente por su etiqueta, en vez de por su posición.

Vamos a comprobar ambos comportamientos con la Series de ventas que ya creamos.

````python
# Comportamiento tipo array: slicing (rebanado) — de la posición 0 a la 1 (sin incluir la 2)
print("Rebanado (como array):")
print(ventas[0:2])

# Comportamiento tipo diccionario: pedir un valor por su etiqueta
print("\nAcceso por etiqueta (como diccionario):")
print(ventas['Febrero'])
````

### 🤔 ¿Qué acaba de pasar?

- `ventas[0:2]` corta la Series igual que cortarías una lista — pero como es una Series, el resultado **conserva las etiquetas** de las filas que quedaron (`Enero` y `Febrero`), no solo los números.
- `ventas['Febrero']` no usa una posición (0, 1, 2...) sino el nombre de la etiqueta, tal como harías con `un_diccionario['Febrero']`.
- Esta doble personalidad es justo lo que hace tan cómoda a una Series: tienes la velocidad de un array y la comodidad de buscar por nombre de un diccionario.

## 3. El DataFrame: varias Series pegadas 📊

Si una Series es una fila de casilleros con nombre, un **DataFrame** es un **armario completo**: varias filas de casilleros (columnas) puestas una junto a otra, todas compartiendo las mismas "puertas" (el índice de filas).

La forma más común de crear uno es a partir de un diccionario, donde cada llave se vuelve una columna. Y, tal como con la Series, le podemos poner nombres propios a las filas con el parámetro `index` en vez de dejar los números automáticos.

````python
# DataFrame a partir de un diccionario de listas (índice automático)
reseñas = pd.DataFrame({
    'Le_gusto': ['Sí', 'No'],
    'Comentario': ['Me encantó.', 'Muy flojo.']
})
print("Con índice automático:")
print(reseñas)

# El mismo DataFrame, pero con un índice de filas con nombre propio
reseñas_con_indice = pd.DataFrame(
    {'Le_gusto': ['Sí', 'No'], 'Comentario': ['Me encantó.', 'Muy flojo.']},
    index=['Producto A', 'Producto B']
)
print("\nCon índice propio:")
reseñas_con_indice
````

### 🤔 ¿Qué acaba de pasar?

- Cada llave del diccionario (`'Le_gusto'`, `'Comentario'`) se convirtió en el nombre de una columna, y cada lista se convirtió en los valores de esa columna.
- En `reseñas`, como no dimos un índice propio, Pandas asignó números automáticos a las filas (0, 1) — el mismo comportamiento por defecto que vimos con la Series.
- En `reseñas_con_indice`, el parámetro `index=['Producto A', 'Producto B']` reemplaza esos números por etiquetas con significado — cada fila del armario ahora tiene una "puerta con nombre".
- También puedes crear un DataFrame con una lista de listas y el parámetro `columns` para nombrar las columnas — útil cuando tus datos ya vienen "fila por fila" en vez de "columna por columna".

## 4. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| `Series` | Una sola columna etiquetada de datos, todos del mismo tipo. |
| `index` | Las "etiquetas de las puertas" — el nombre de cada fila (por defecto, números 0, 1, 2...). |
| Series como array | Se puede rebanar (`serie[0:2]`) y conserva las etiquetas del recorte. |
| Series como diccionario | Se puede consultar un valor directo por su etiqueta (`serie['Febrero']`). |
| `DataFrame` | Varias Series pegadas, formando una tabla completa con filas y columnas. |
| Crear un DataFrame | Lo más común: un diccionario de listas, con o sin `index` personalizado. |
