## 2. Merge: como armar un rompecabezas por una pieza en común 🧩

Imagina que tienes dos listas en papel: una con el nombre y el correo de cada estudiante, y otra con el nombre y la nota final de cada estudiante. Para juntar la información en una sola hoja, buscarías la columna que **tienen en común** (el nombre) y emparejarías fila con fila.

Eso es justo lo que hace `merge()`: busca una o varias columnas en común entre dos tablas, y arma una tabla nueva emparejando las filas que coinciden — igual que los `JOIN` que ya conocerás si alguna vez usaste bases de datos SQL.

Vamos a crear dos tablas pequeñas para practicar: una con estudiantes y su ciudad, y otra con las notas de algunos de esos estudiantes.

````python
import pandas as pd

estudiantes = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'nombre': ['Ana', 'Beto', 'Cata', 'Dario'],
    'ciudad': ['Tunja', 'Duitama', 'Tunja', 'Sogamoso'],
})

notas = pd.DataFrame({
    'id': [1, 2, 3, 5],
    'nota_final': [4.5, 3.8, 4.1, 4.9],
})

print(estudiantes)
print()
print(notas)
````

### 🤔 ¿Qué acaba de pasar?

- `estudiantes` tiene 4 filas (ids 1 a 4); `notas` tiene 4 filas, pero con los ids 1, 2, 3 y **5** (no el 4).
- Fíjate que el `id` es la columna en común entre ambas tablas: esa será la "pieza" que usaremos para unirlas.

#### Inner join (por defecto): solo lo que coincide en ambas tablas

Si no le decimos nada más, `merge()` busca la columna en común y se queda **solo con los ids que existen en las dos tablas a la vez**.

````python
inner = pd.merge(estudiantes, notas, on='id')
inner
````

### 🤔 ¿Qué acaba de pasar?

El resultado solo tiene 3 filas: los ids **1, 2 y 3**, porque son los únicos que aparecen en `estudiantes` **y** en `notas` al mismo tiempo. El id 4 (que solo está en `estudiantes`) y el id 5 (que solo está en `notas`) se quedaron **fuera**. Ese descarte de lo que no coincide es la característica del `inner join`.

#### Los otros tres sabores: `outer`, `left` y `right`

- `how="outer"`: conserva **todas** las filas de ambas tablas, rellenando con `NaN` (= "no hay dato") lo que falte.
- `how="left"`: conserva **todas** las filas de la tabla de la izquierda (la primera), aunque no tengan pareja en la derecha.
- `how="right"`: lo mismo, pero priorizando la tabla de la derecha (la segunda).

````python
outer = pd.merge(estudiantes, notas, on='id', how='outer')
left = pd.merge(estudiantes, notas, on='id', how='left')

print('outer:')
print(outer)
print()
print('left:')
print(left)
````

### 🤔 ¿Qué acaba de pasar?

- `outer` tiene 5 filas: los ids 1 a 5, todos, con `NaN` donde faltaba información (el id 4 no tiene `nota_final`, el id 5 no tiene `nombre` ni `ciudad`).
- `left` tiene 4 filas: las mismas 4 de `estudiantes` (ids 1 a 4), incluido el id 4 con `NaN` en `nota_final`. El id 5, que solo existía en `notas`, no aparece — `left` siempre prioriza la tabla de la izquierda.

## 3. Join: la versión simplificada, pero por índice 🗃️

`join()` hace algo muy parecido a `merge()`, pero en vez de buscar una columna en común, combina las tablas usando **el índice** de cada una (el número o etiqueta que identifica cada fila).

Para poder usar `join()` con nuestras tablas, primero pongamos `id` como índice de ambas:

````python
estudiantes_idx = estudiantes.set_index('id')
notas_idx = notas.set_index('id')

unido = estudiantes_idx.join(notas_idx)
unido
````

### 🤔 ¿Qué acaba de pasar?

- `set_index('id')` convierte la columna `id` en el índice de cada tabla (en vez de una columna normal).
- `estudiantes_idx.join(notas_idx)` combina ambas tablas emparejando filas que compartan el mismo índice — por defecto, `join()` se comporta como un `left join`, así que conserva todas las filas de `estudiantes_idx`.
- Si dos tablas tuvieran columnas con el mismo nombre, usarías los parámetros `lsuffix` y `rsuffix` para diferenciarlas (por ejemplo, `nombre_izq` y `nombre_der`).

## 4. Concat: apilar tablas, sin buscar coincidencias 📚

`merge()` y `join()` buscan **coincidencias** entre columnas o índices. `concat()` es distinto: simplemente **apila** las tablas una encima de la otra (o una al lado de la otra), sin preguntarse si los valores coinciden — como poner un libro nuevo encima de una pila de libros ya existente.

#### Apilar filas (una tabla debajo de la otra)

Imagina que llega un grupo nuevo de estudiantes y quieres agregarlo a la tabla que ya tenías:

````python
nuevos_estudiantes = pd.DataFrame({
    'id': [6, 7],
    'nombre': ['Elena', 'Fabio'],
    'ciudad': ['Tunja', 'Duitama'],
})

todos = pd.concat([estudiantes, nuevos_estudiantes])
print(todos)

# Con ignore_index=True el índice queda limpio y continuo
todos_reindexados = pd.concat([estudiantes, nuevos_estudiantes], ignore_index=True)
print(todos_reindexados)

# Apilar columnas (lado a lado) se hace con axis='columns'
contacto = pd.DataFrame({'telefono': ['3001', '3002', '3003', '3004']})
lado_a_lado = pd.concat([estudiantes, contacto], axis='columns')
print(lado_a_lado)
````

### 🤔 ¿Qué acaba de pasar?

- `pd.concat([estudiantes, nuevos_estudiantes])` simplemente pega la segunda tabla debajo de la primera. Fíjate que el índice (la numeración de la izquierda) **no se reinicia**: sigue el 0, 1, 2, 3 de la primera tabla y luego repite 0, 1 para la segunda.
- Con `ignore_index=True`, Pandas descarta los índices originales y numera todo de nuevo desde 0, de forma continua.
- Con `axis='columns'`, `concat()` en cambio pega la tabla `contacto` **a la derecha** de `estudiantes`, alineando fila por fila según el índice — como poner dos hojas una junto a la otra sobre la mesa, en vez de apilarlas. A diferencia de `merge()`, aquí no se busca ninguna columna en común: se confía en que las filas ya estén en el mismo orden.

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| `merge()` | Une dos tablas buscando columnas en común, como los `JOIN` de SQL. |
| `how="inner"` | Se queda solo con lo que coincide en ambas tablas (comportamiento por defecto). |
| `how="outer"` | Conserva todo de ambas tablas, rellenando huecos con `NaN`. |
| `how="left"` / `how="right"` | Prioriza todas las filas de una de las dos tablas. |
| `join()` | Como `merge()`, pero combina usando el **índice** de cada tabla en vez de una columna. |
| `concat()` | Apila tablas una encima de otra (`axis=0`) o una al lado de otra (`axis="columns"`), sin buscar coincidencias. |
| `ignore_index=True` | Al apilar con `concat`, reinicia la numeración del índice para que quede limpia. |
