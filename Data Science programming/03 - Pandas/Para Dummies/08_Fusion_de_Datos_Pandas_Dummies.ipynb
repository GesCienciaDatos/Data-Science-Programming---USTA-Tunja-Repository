## Uniendo Tablas como en SQL: merge y concat 🔗

En el mundo real, los datos **nunca viven en una sola tabla**. Tienes una tabla de clientes,
otra de pedidos, otra de productos. Para analizarlos juntos, debes **unirlos (join)**.

### Tipos de JOIN — La Visualización de los Círculos

| Tipo | `how=` | ¿Qué incluye? |
|---|---|---|
| Inner Join | `'inner'` | Solo filas que existen en **ambas** tablas |
| Left Join | `'left'` | Todas las filas de la **izquierda** + coincidencias de la derecha |
| Right Join | `'right'` | Todas las filas de la **derecha** + coincidencias de la izquierda |
| Outer Join | `'outer'` | Todas las filas de **ambas** tablas |


```python
import pandas as pd
import numpy as np

# Tabla de Clientes
clientes = pd.DataFrame({
    'ID_Cliente': [1, 2, 3, 4, 5],
    'Nombre': ['Laura Pérez', 'Camilo Gómez', 'Daniela Ríos', 'Esteban Cruz', 'María López'],
    'Ciudad': ['Tunja', 'Bogotá', 'Cali', 'Medellín', 'Tunja']
})

# Tabla de Pedidos
pedidos = pd.DataFrame({
    'ID_Pedido': [101, 102, 103, 104, 105, 106],
    'ID_Cliente': [1, 2, 1, 3, 6, 2],  # El ID 6 no existe en clientes
    'Producto': ['Laptop', 'Mouse', 'Monitor', 'Teclado', 'Auriculares', 'Webcam'],
    'Total': [2500000, 85000, 850000, 120000, 350000, 180000]
})

print('Tabla Clientes:'); display(clientes)
print('\nTabla Pedidos:'); display(pedidos)
```

```python
# INNER JOIN — solo clientes con pedidos
inner = pd.merge(clientes, pedidos, on='ID_Cliente', how='inner')
print(f'INNER JOIN ({len(inner)} filas) — Solo coincidencias:')
display(inner)
```

```python
# LEFT JOIN — todos los clientes, con o sin pedidos
left = pd.merge(clientes, pedidos, on='ID_Cliente', how='left')
print(f'LEFT JOIN ({len(left)} filas) — Todos los clientes:')
display(left)

# Clientes sin pedidos
sin_pedidos = left[left['ID_Pedido'].isnull()]
print(f'\nClientes sin pedidos: {len(sin_pedidos)}')
print(sin_pedidos[['Nombre', 'Ciudad']])
```

```python
# CONCAT — apilar DataFrames verticalmente
ventas_q1 = pd.DataFrame({'Vendedor': ['Ana', 'Luis'], 'Ventas': [1000000, 1500000], 'Trimestre': ['Q1', 'Q1']})
ventas_q2 = pd.DataFrame({'Vendedor': ['Ana', 'María'], 'Ventas': [1200000, 900000], 'Trimestre': ['Q2', 'Q2']})

ventas_total = pd.concat([ventas_q1, ventas_q2], ignore_index=True)
print('Ventas anuales (Q1 + Q2 apilados):')
display(ventas_total)
print(f'\nTotal vendido por vendedor:')
print(ventas_total.groupby('Vendedor')['Ventas'].sum())
```

> **🔑 ¿Cuándo usar merge vs concat?**
>
> - **`pd.merge()`** → Para unir tablas **horizontalmente** (como un JOIN de SQL), combinando columnas
> - **`pd.concat()`** → Para apilar tablas **verticalmente** (agregar más filas del mismo formato)
> 
> Ejemplo: concat para unir ventas de enero + febrero (mismas columnas, más filas).
