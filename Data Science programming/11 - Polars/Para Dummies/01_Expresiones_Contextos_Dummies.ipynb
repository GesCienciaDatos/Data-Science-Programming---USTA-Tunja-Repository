## 1. La Regla de Oro: En Polars no cambias la tabla, creas 'Recetas' 📜

En Pandas solías escribir `df['nueva'] = df['precio'] * 1.19`.
En Polars le dices a la computadora: *"Oye Polars, prepara una columna llamada `precio_iva` multiplicando `pl.col('precio') * 1.19`"*.

Solo necesitas dominar **3 contextos** — imagínalos como **3 modos distintos** de una misma herramienta, como si fuera un buffet:
1. **`select()`:** la vitrina del buffet — "muéstrame solo estos platos, olvida el resto".
2. **`with_columns()`:** el mesero — "tráeme platos extra a la mesa, sin quitarme los que ya tenía".
3. **`filter()`:** el portero de la fiesta — "solo dejo pasar a quien cumpla este requisito".

Los tres usan el mismo vocabulario (`pl.col(...)`), pero cada uno responde una pregunta distinta.

```python
import polars as pl
import os

import os, urllib.request, urllib.parse

def load_dataset(filename, module_name="11 - Polars"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo genera automáticamente.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("Data Science programming", module_name, "data", filename),
        os.path.join("..", "Data Science programming", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path
            
    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Science programming/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)
    
    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]
    
    print(f"📥 Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=3) as response:
                if response.status == 200:
                    with open(target_path, 'wb') as out_f:
                        out_f.write(response.read())
                    print(f"✅ Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue
            
    print(f"⚙️ Generando '{filename}' sintéticamente para ejecución inmediata...")
    import polars as pl
    import numpy as np
    np.random.seed(42)
    
    n_clientes = 1000
    df_c = pl.DataFrame({
        'id_cliente': [f'CLI-{i:04d}' for i in range(1, n_clientes + 1)],
        'nombre': [f'Cliente_{i}' for i in range(1, n_clientes + 1)],
        'segmento': np.random.choice(['Corporativo', 'Pyme', 'Consumo', 'Gobierno'], n_clientes),
        'edad': np.random.randint(18, 70, n_clientes),
        'ciudad_residencia': np.random.choice(['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Bucaramanga'], n_clientes),
        'ingreso_anual': np.random.normal(45000000, 15000000, n_clientes).round(2)
    })
    
    n_ventas = 60000
    cats = ['Tecnología', 'Mobiliario', 'Material de Oficina', 'Servicios']
    prods = ['Laptop Pro', 'Monitor 4K', 'Silla Ergonómica', 'Escritorio', 'Papel A4', 'Tóner', 'Mantenimiento']
    ciudades = ['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Barranquilla']
    
    cant = np.random.randint(1, 10, n_ventas)
    pu = np.random.choice([25000.0, 120000.0, 450000.0, 1200000.0, 3500000.0], n_ventas)
    desc = np.random.choice([0.0, 0.05, 0.10, 0.15], n_ventas)
    tot = (cant * pu * (1 - desc)).round(2)
    
    df_v = pl.DataFrame({
        'id_venta': [f'VNT-{i:06d}' for i in range(1, n_ventas + 1)],
        'fecha': [f'2024-{np.random.randint(1,13):02d}-{np.random.randint(1,29):02d}' for _ in range(n_ventas)],
        'id_cliente': np.random.choice(df_c['id_cliente'], n_ventas),
        'categoria': np.random.choice(cats, n_ventas),
        'producto': np.random.choice(prods, n_ventas),
        'cantidad': cant,
        'precio_unitario': pu,
        'descuento': desc,
        'ciudad_venta': np.random.choice(ciudades, n_ventas),
        'total_venta': tot
    })
    
    c_csv_path = os.path.join("data", "clientes.csv")
    c_pq_path = os.path.join("data", "clientes.parquet")
    v_csv_path = os.path.join("data", "ventas.csv")
    v_pq_path = os.path.join("data", "ventas.parquet")
    
    if not os.path.exists(c_csv_path): df_c.write_csv(c_csv_path)
    if not os.path.exists(c_pq_path): df_c.write_parquet(c_pq_path)
    if not os.path.exists(v_csv_path): df_v.write_csv(v_csv_path)
    if not os.path.exists(v_pq_path): df_v.write_parquet(v_pq_path)
    
    print(f"✅ Datasets preparados exitosamente en '{target_path}'.")
    return target_path

ruta = load_dataset("ventas.csv")
df = pl.read_csv(ruta)
print(f"¡Cargamos {len(df):,} ventas sin esfuerzo!")
```

### 🤔 ¿Qué acaba de pasar?

- El `if` de la ruta es un "plan B": si el cuaderno se ejecuta desde `Para Dummies/`, sube un nivel con `".."` para encontrar la carpeta `data/`; si no la encuentra ahí, prueba directamente en `data/` (por si lo ejecutas desde otra ubicación).
- `pl.read_csv(ruta)` cargó las 60,000 filas de `ventas.csv` en la variable `df`, que usaremos en todo el cuaderno.
- Nota que **no imprimimos la tabla completa**: solo confirmamos que cargó. Eso lo haremos en la próxima sección con nuestro primer "modo".

---
## 2. Modo Vitrina: `select()` — "Muéstrame solo esto" 🍽️

`select()` te devuelve **una tabla nueva** con únicamente las columnas o cálculos que le pidas. La tabla original (`df`) queda intacta — es como pedir un plato del buffet sin que eso vacíe la cocina.

```python
vista_rapida = df.select([
    pl.col("producto"),
    pl.col("categoria"),
    pl.col("total_venta"),
    (pl.col("total_venta") * 0.30).round(2).alias("ganancia_estimada_30pct")
])
display(vista_rapida.head(5))
```

### 🤔 ¿Qué acaba de pasar?

- Le pedimos a `select()` 3 columnas que ya existían (`producto`, `categoria`, `total_venta`) y 1 columna que **no existía**: la calculamos al vuelo multiplicando `total_venta` por `0.30` y la bautizamos con `.alias(...)`.
- El resultado, `vista_rapida`, es una tabla **completamente nueva** con solo esas 4 columnas — las demás (como `fecha` o `id_cliente`) simplemente no aparecen, aunque siguen existiendo intactas en `df`.
- Si vuelves a ejecutar `display(df.head())`, verás que `df` no cambió en absoluto. `select()` nunca modifica la tabla original.

---
## 3. Modo Mesero: `with_columns()` — "Agrega esto sin quitar nada" ➕

```python
# Agregamos el cálculo de IVA (19%) y clasificamos si la compra fue barata o cara
df_magico = df.with_columns([
    (pl.col("precio_unitario") * 0.19).round(2).alias("valor_iva"),
    pl.when(pl.col("total_venta") > 3000000)
      .then(pl.lit("Compra Premium 💎"))
      .otherwise(pl.lit("Compra Normal 🛒"))
      .alias("tipo_cliente")
])

display(df_magico.select(["producto", "precio_unitario", "valor_iva", "total_venta", "tipo_cliente"]).head(5))
```

### 🤔 ¿Qué acaba de pasar?

- A diferencia de `select()`, aquí `df_magico` conserva **todas** las columnas originales de `df`, más las 2 nuevas (`valor_iva` y `tipo_cliente`). Por eso al final usamos `.select([...])` aparte, solo para no imprimir las 10 columnas de una vez.
- `pl.when(condición).then(valor_si_verdadero).otherwise(valor_si_falso)` es como un semáforo: si la condición se cumple, toma el camino de `.then()`; si no, toma el de `.otherwise()`. Puedes encadenar varios `.when()` seguidos para más de 2 caminos, como un árbol de decisiones.
- Fíjate que esto se calculó para las 60,000 filas **a la vez**, sin ningún bucle `for` — Polars reparte el trabajo entre los núcleos de tu procesador automáticamente.

---
## 4. Modo Portero: `filter()` — "Solo pasa si cumples esto" 🔎

```python
# Queremos solo ventas de Boyacá (Tunja, Duitama, Sogamoso) con cantidad >= 10
ventas_boyaca = df.filter(
    pl.col("ciudad_venta").is_in(["Tunja", "Duitama", "Sogamoso"]) &
    (pl.col("cantidad") >= 10)
)
print(f"Total de ventas masivas en Boyacá: {ventas_boyaca.height}")
display(ventas_boyaca.head(3))
```

### 🤔 ¿Qué acaba de pasar?

- `.is_in([...])` revisa, fila por fila, si el valor de `ciudad_venta` está dentro de esa lista de 3 ciudades — es un atajo para no escribir `(ciudad == "Tunja") | (ciudad == "Duitama") | (ciudad == "Sogamoso")`.
- El símbolo `&` combina las dos condiciones con un "Y": solo pasan las filas donde **ambas** son ciertas (ciudad de Boyacá **Y** cantidad ≥ 10). El "portero" solo deja entrar a quien cumple las dos reglas del club.
- A diferencia de `select()` y `with_columns()`, `filter()` sí puede **reducir el número de filas**: de 60,000 originales, `ventas_boyaca` se quedó solo con las que cumplieron el requisito.

## 5. Bonus: Texto y Fechas en un Vistazo 🔤📅

Dos herramientas muy usadas viven "dentro" de cada columna: `.str` (para texto) y `.dt` (para fechas). Se usan igual que cualquier otra transformación, solo que primero hay que "despertar" la fecha con `.str.to_datetime()`, porque en el CSV llega como texto plano.

```python
df_fechas = df.with_columns([
    pl.col("fecha").str.to_datetime(),
    pl.col("producto").str.to_uppercase().alias("producto_grande")
]).with_columns(
    pl.col("fecha").dt.month().alias("mes_de_compra")
)
display(df_fechas.select(["fecha", "mes_de_compra", "producto_grande"]).head(4))
```

### 🤔 ¿Qué acaba de pasar?

- `pl.col("fecha").str.to_datetime()` convierte el texto `"2024-03-15 10:00:00"` en una fecha *de verdad* que Polars entiende — antes de esto, Polars solo veía letras y números sueltos, no un calendario.
- Una vez convertida, `.dt.month()` puede "preguntarle" a la fecha en qué mes ocurrió, sin que tengas que cortar el texto a mano con `[5:7]` como harías con strings normales.
- `.str.to_uppercase()` es la versión de texto de lo mismo: un "namespace" (`.str`) lleno de herramientas específicas para columnas de tipo texto.

## 6. Mini Reto: Combina los 3 Modos 🕵️

Antes de ver el código, piensa: ¿cómo mostrarías solo las ventas de la categoría `"Muebles"` con `precio_unitario` mayor a 500,000, agregando una columna que diga si el `descuento` fue mayor al 10%? Cuando tengas una idea, ejecuta la celda de abajo y compara.

```python
reto = (
    df
    .filter((pl.col("categoria") == "Muebles") & (pl.col("precio_unitario") > 500000))
    .with_columns((pl.col("descuento") > 0.10).alias("descuento_grande"))
    .select(["producto", "precio_unitario", "descuento", "descuento_grande"])
)
print(f"Muebles caros encontrados: {reto.height}")
display(reto.head(5))
```

### 🤔 ¿Qué acaba de pasar?

- Encadenamos los 3 modos en una sola "receta": primero el **portero** (`filter`) redujo las filas a solo muebles caros, luego el **mesero** (`with_columns`) agregó una columna extra de verdadero/falso, y al final la **vitrina** (`select`) mostró solo las 4 columnas que nos interesaban.
- El orden importa por legibilidad y eficiencia: filtrar primero (menos filas) antes de calcular columnas nuevas suele ser más natural, aunque Polars igual optimiza internamente.

## 7. Resumen Relámpago ⚡

| Modo | Comando | En una frase |
|---|---|---|
| Vitrina | `select()` | Muestra solo lo que pides; el resto desaparece de la vista (no de `df`). |
| Mesero | `with_columns()` | Agrega columnas nuevas sin quitar las que ya había. |
| Portero | `filter()` | Deja pasar solo las filas que cumplen una condición. |
| `pl.when/then/otherwise` | — | El semáforo de Polars: condición → un camino u otro. |
| `.is_in([...])` | — | Atajo para "está en esta lista de valores". |
| `.str` / `.dt` | — | Cajas de herramientas para texto y fechas dentro de una columna. |
