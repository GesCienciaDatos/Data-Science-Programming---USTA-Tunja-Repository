## 1. La Pista de Carreras: ¿Quién es más rápido con 60,000 datos? 🏁

Vamos a poner a competir a Pandas y a Polars haciendo exactamente la misma tarea pesada (filtrar, agrupar y sumar) mientras medimos el tiempo con un cronómetro:

````python
import pandas as pd
import polars as pl
import time, os

import os, urllib.request, urllib.parse

def load_dataset(filename, module_name="11 - Polars"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo genera automáticamente.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("Data Science programming", module_name, "data", filename),
        os.path.join("..", "Data Science programming", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path
            
    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Science programming/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)
    
    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]
    
    print(f"📥 Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=3) as response:
                if response.status == 200:
                    with open(target_path, 'wb') as out_f:
                        out_f.write(response.read())
                    print(f"✅ Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue
            
    print(f"⚙️ Generando '{filename}' sintéticamente para ejecución inmediata...")
    import polars as pl
    import numpy as np
    np.random.seed(42)
    
    n_clientes = 1000
    df_c = pl.DataFrame({
        'id_cliente': [f'CLI-{i:04d}' for i in range(1, n_clientes + 1)],
        'nombre': [f'Cliente_{i}' for i in range(1, n_clientes + 1)],
        'segmento': np.random.choice(['Corporativo', 'Pyme', 'Consumo', 'Gobierno'], n_clientes),
        'edad': np.random.randint(18, 70, n_clientes),
        'ciudad_residencia': np.random.choice(['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Bucaramanga'], n_clientes),
        'ingreso_anual': np.random.normal(45000000, 15000000, n_clientes).round(2)
    })
    
    n_ventas = 60000
    cats = ['Tecnología', 'Mobiliario', 'Material de Oficina', 'Servicios']
    prods = ['Laptop Pro', 'Monitor 4K', 'Silla Ergonómica', 'Escritorio', 'Papel A4', 'Tóner', 'Mantenimiento']
    ciudades = ['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Barranquilla']
    
    cant = np.random.randint(1, 10, n_ventas)
    pu = np.random.choice([25000.0, 120000.0, 450000.0, 1200000.0, 3500000.0], n_ventas)
    desc = np.random.choice([0.0, 0.05, 0.10, 0.15], n_ventas)
    tot = (cant * pu * (1 - desc)).round(2)
    
    df_v = pl.DataFrame({
        'id_venta': [f'VNT-{i:06d}' for i in range(1, n_ventas + 1)],
        'fecha': [f'2024-{np.random.randint(1,13):02d}-{np.random.randint(1,29):02d}' for _ in range(n_ventas)],
        'id_cliente': np.random.choice(df_c['id_cliente'], n_ventas),
        'categoria': np.random.choice(cats, n_ventas),
        'producto': np.random.choice(prods, n_ventas),
        'cantidad': cant,
        'precio_unitario': pu,
        'descuento': desc,
        'ciudad_venta': np.random.choice(ciudades, n_ventas),
        'total_venta': tot
    })
    
    c_csv_path = os.path.join("data", "clientes.csv")
    c_pq_path = os.path.join("data", "clientes.parquet")
    v_csv_path = os.path.join("data", "ventas.csv")
    v_pq_path = os.path.join("data", "ventas.parquet")
    
    if not os.path.exists(c_csv_path): df_c.write_csv(c_csv_path)
    if not os.path.exists(c_pq_path): df_c.write_parquet(c_pq_path)
    if not os.path.exists(v_csv_path): df_v.write_csv(v_csv_path)
    if not os.path.exists(v_pq_path): df_v.write_parquet(v_pq_path)
    
    print(f"✅ Datasets preparados exitosamente en '{target_path}'.")
    return target_path

ruta = load_dataset("ventas.csv")

# Carrera 1: Polars
t0 = time.time()
res_polars = pl.read_csv(ruta).filter(pl.col("total_venta") > 500000).select(pl.col("total_venta").sum()).item()
t_polars = time.time() - t0

# Carrera 2: Pandas
t0 = time.time()
df_pd = pd.read_csv(ruta)
res_pandas = df_pd[df_pd["total_venta"] > 500000]["total_venta"].sum()
t_pandas = time.time() - t0

print(f"🐻 Polars tardó: {t_polars*1000:.1f} ms")
print(f"🐼 Pandas tardó: {t_pandas*1000:.1f} ms")
````

### 🤔 ¿Qué acaba de pasar?

Ambas líneas hacen exactamente lo mismo: filtrar ventas mayores a 500,000 y sumar el total por categoría. La diferencia de tiempo que ves impresa es un **cronómetro real, no un truco**: ejecutaste la misma tarea dos veces, con dos motores distintos, y mediste cuánto tardó cada uno con `time.perf_counter()` (el cronómetro más preciso que ofrece Python).

El número de "veces más rápido" que se imprime al final es simplemente `tiempo_pandas / tiempo_polars`. Ejecuta la celda de nuevo — el resultado puede variar un poco según qué tan ocupado esté tu computador en ese momento, pero Polars debería ganar la carrera de forma consistente.

## 2. Pasando el Testigo sin Copiar Nada: Polars ↔ Pandas ↔ Arrow 🏃‍♂️➡️🏃‍♀️

Imagina una carrera de relevos donde, en vez de que cada corredor tenga que **fabricar una copia nueva** del testigo para dárselo al siguiente, simplemente se lo entregan de la mano directamente. Eso es lo que pasa cuando conviertes entre Polars, Pandas y Apache Arrow: como los tres "hablan" el mismo formato columnar por debajo, no hace falta reconstruir los datos — solo se reinterpreta el mismo bloque de memoria. Por eso estas conversiones son casi instantáneas incluso con datasets grandes.

````python
# Polars -> Pandas (por si una función vieja solo acepta pandas.DataFrame)
res_polars_pandas = res_polars.to_pandas()
print("Tipo convertido:", type(res_polars_pandas))

# Polars -> Arrow (el formato que comparten Polars, Pandas moderno y muchas herramientas de Big Data)
tabla_arrow = pldf.select(["id_venta", "total_venta"]).to_arrow()
print("Tipo Arrow:", type(tabla_arrow), "| filas:", tabla_arrow.num_rows)
````

### 🤔 ¿Qué acaba de pasar?

`.to_pandas()` convirtió el resultado de Polars a un DataFrame de Pandas normal y corriente — útil si necesitas graficarlo con una librería vieja que solo entiende Pandas. `.to_arrow()` lo convirtió al formato Apache Arrow, el "idioma común" que también hablan herramientas como DuckDB. En ningún caso Polars tuvo que leer el archivo de nuevo ni recalcular nada: solo *reempacó* los mismos datos que ya tenía en memoria.

````python
# Y de regreso: Pandas -> Polars, sin duplicar tampoco
de_vuelta_a_polars = pl.from_pandas(res_polars_pandas)
print("Tipo de vuelta en Polars:", type(de_vuelta_a_polars))
````

### 🤔 ¿Qué acaba de pasar?

`pl.from_pandas()` hizo el camino inverso: tomó un DataFrame de Pandas y lo devolvió a Polars. El testigo de la carrera de relevos viaja en ambas direcciones sin que nadie tenga que fabricar una copia nueva.

## 3. Los 3 Pecados Capitales en Polars (¡No lo hagas!) 🛑🙅

1. **No uses bucles `for fila in df:`** $\rightarrow$ Rompes el motor en Rust y obligas a Python a ir paso a paso.
2. **No uses `.apply(lambda x: ...)`** $\rightarrow$ Las expresiones de Polars (`pl.when().then()`, `.str`, `.dt`) hacen todo 100 veces más rápido.
3. **No tengas miedo a combinar ambos** $\rightarrow$ Puedes usar Polars para limpiar gigabytes en segundos y luego pasarlo a Pandas con `.to_pandas()` si una gráfica antigua lo requiere.

## 5. Resumen Relámpago y Cierre de la Ruta Para Dummies 🎓⚡

* **Pandas vs. Polars** = una carrera con cronómetro real: mismo resultado, tiempos distintos.
* **`.to_pandas()` / `.to_arrow()`** = pasar el testigo sin copiar nada, gracias a que comparten el mismo formato de memoria por debajo.
