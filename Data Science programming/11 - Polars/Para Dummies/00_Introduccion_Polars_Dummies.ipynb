## 1. La Analogía del Tráfico: ¿Por qué Pandas se cansa y Polars vuela? 🚗💨

Imagina que tienes que transportar 1 millón de paquetes:
* **Pandas 🐼** es como un camión de mudanzas que hace todo un solo conductor (monohilo). Cada vez que le pides cambiar algo, descarga todo en el suelo, lo copia en otra caja y vuelve a cargar (duplicando la memoria RAM). Si la carga es muy pesada, el camión se apaga.
* **Polars 🐻‍❄️** es un convoy de bólidos de Fórmula 1 diseñados con fibra de carbono (lenguaje **Rust**). Cada núcleo de tu procesador es un piloto trabajando al mismo tiempo. Además, organiza los paquetes en columnas ordenaditas (Apache Arrow) para que encontrarlos tome milisegundos sin duplicar nada.

```python
import polars as pl
import os

print(f"🐻 Polars versión: {pl.__version__}")

import os, urllib.request, urllib.parse

def load_dataset(filename, module_name="11 - Polars"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo genera automáticamente.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("Data Science programming", module_name, "data", filename),
        os.path.join("..", "Data Science programming", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path
            
    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Science programming/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)
    
    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]
    
    print(f"📥 Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=3) as response:
                if response.status == 200:
                    with open(target_path, 'wb') as out_f:
                        out_f.write(response.read())
                    print(f"✅ Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue
            
    print(f"⚙️ Generando '{filename}' sintéticamente para ejecución inmediata...")
    import polars as pl
    import numpy as np
    np.random.seed(42)
    
    n_clientes = 1000
    df_c = pl.DataFrame({
        'id_cliente': [f'CLI-{i:04d}' for i in range(1, n_clientes + 1)],
        'nombre': [f'Cliente_{i}' for i in range(1, n_clientes + 1)],
        'segmento': np.random.choice(['Corporativo', 'Pyme', 'Consumo', 'Gobierno'], n_clientes),
        'edad': np.random.randint(18, 70, n_clientes),
        'ciudad_residencia': np.random.choice(['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Bucaramanga'], n_clientes),
        'ingreso_anual': np.random.normal(45000000, 15000000, n_clientes).round(2)
    })
    
    n_ventas = 60000
    cats = ['Tecnología', 'Mobiliario', 'Material de Oficina', 'Servicios']
    prods = ['Laptop Pro', 'Monitor 4K', 'Silla Ergonómica', 'Escritorio', 'Papel A4', 'Tóner', 'Mantenimiento']
    ciudades = ['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Barranquilla']
    
    cant = np.random.randint(1, 10, n_ventas)
    pu = np.random.choice([25000.0, 120000.0, 450000.0, 1200000.0, 3500000.0], n_ventas)
    desc = np.random.choice([0.0, 0.05, 0.10, 0.15], n_ventas)
    tot = (cant * pu * (1 - desc)).round(2)
    
    df_v = pl.DataFrame({
        'id_venta': [f'VNT-{i:06d}' for i in range(1, n_ventas + 1)],
        'fecha': [f'2024-{np.random.randint(1,13):02d}-{np.random.randint(1,29):02d}' for _ in range(n_ventas)],
        'id_cliente': np.random.choice(df_c['id_cliente'], n_ventas),
        'categoria': np.random.choice(cats, n_ventas),
        'producto': np.random.choice(prods, n_ventas),
        'cantidad': cant,
        'precio_unitario': pu,
        'descuento': desc,
        'ciudad_venta': np.random.choice(ciudades, n_ventas),
        'total_venta': tot
    })
    
    c_csv_path = os.path.join("data", "clientes.csv")
    c_pq_path = os.path.join("data", "clientes.parquet")
    v_csv_path = os.path.join("data", "ventas.csv")
    v_pq_path = os.path.join("data", "ventas.parquet")
    
    if not os.path.exists(c_csv_path): df_c.write_csv(c_csv_path)
    if not os.path.exists(c_pq_path): df_c.write_parquet(c_pq_path)
    if not os.path.exists(v_csv_path): df_v.write_csv(v_csv_path)
    if not os.path.exists(v_pq_path): df_v.write_parquet(v_pq_path)
    
    print(f"✅ Datasets preparados exitosamente en '{target_path}'.")
    return target_path
```

### 🤔 ¿Qué acaba de pasar?

- `import polars as pl` trae la biblioteca completa, y la apodamos `pl` (así la vas a ver escrita en todos lados, incluyendo la documentación oficial).
- `pl.__version__` simplemente te dice qué versión tienes instalada — útil para asegurarte de que sigues estas instrucciones con una versión reciente.
- No necesitaste instalar Rust, compilar nada ni configurar hilos manualmente: todo el paralelismo del que hablamos en la analogía del tráfico ya viene activado por defecto apenas importas la librería.

## 2. Creando tu Primera Tabla sin Dolores de Cabeza 🧸🧱

En Polars no hay que preocuparse por el temido `df.index` de Pandas (¡adiós a los errores de `reset_index`!). Cada fila tiene su posición natural y punto.

```python
# Mi primer DataFrame en Polars
amigos = pl.DataFrame({
    "nombre": ["Mateo", "Valentina", "Santiago", "Camila"],
    "edad": [22, 25, 28, 24],
    "cafe_al_dia": [2, 1, 4, 3],
    "le_gusta_python": [True, True, True, True]
})

print("Nuestra tabla de amigos:")
display(amigos)
print(f"¿Cuántos amigos hay?: {amigos.height} filas")
print(f"¿Cuántas características medimos?: {amigos.width} columnas")
```

### 🤔 ¿Qué acaba de pasar?

- Le dimos a `pl.DataFrame(...)` un diccionario de Python: cada *llave* (`"nombre"`, `"edad"`, ...) se convierte en el nombre de una columna, y cada *lista* de valores se convierte en los datos de esa columna.
- `amigos.height` te dice cuántas **filas** tiene la tabla (cuántos amigos hay).
- `amigos.width` te dice cuántas **columnas** tiene (cuántas características medimos de cada amigo).
- Fíjate que no tuvimos que crear ningún índice a mano: cada fila simplemente "es" la fila 0, 1, 2, 3... por su posición, como en una lista normal de Python.

## 3. Polars es un Poco Quisquilloso con los Tipos (¡y eso es bueno!) 🧦

Imagina un organizador de cajones de ropa: un cajón es solo para medias, otro solo para camisetas. Si intentas meter un zapato en el cajón de las medias, un buen organizador **te detiene de inmediato** y te dice "oye, esto no va aquí" — en vez de dejarte guardar el zapato a la fuerza y que después, cuando busques medias a oscuras, te encuentres con una sorpresa incómoda.

Polars funciona igual con los datos: cada columna tiene un tipo único y fijo (todos números enteros, todos textos, etc.). Si intentas meter un dato que no encaja, Polars **te avisa de inmediato con un error claro**, en vez de guardarlo "como pueda" y dejar que el problema explote después, muchos pasos más adelante, sin avisar.

Pandas, en cambio, suele ser más permisivo: si mezclas números y texto en una misma columna, simplemente los guarda juntos sin quejarse — y ese "sin quejarse" es justamente lo peligroso.

```python
import pandas as pd

datos_con_error = [10, 20, "treinta", 40]

# Pandas no se queja: mezcla todo en una columna de tipo "object"
serie_pandas = pd.Series(datos_con_error)
print("Pandas guardó la mezcla sin avisar. Tipo resultante:", serie_pandas.dtype)

# Polars sí se queja, y de inmediato
try:
    serie_polars = pl.Series("valores", datos_con_error, dtype=pl.Int64)
except Exception as error:
    print("\n¡Polars se detuvo a tiempo!")
    print("Tipo de error:", type(error).__name__)
```

### 🤔 ¿Qué acaba de pasar?

- Le pasamos a ambas librerías la misma lista "sucia": tres números y un texto (`"treinta"`) colado entre ellos.
- Pandas la aceptó sin protestar y la guardó como tipo `object`, un cajón "de todo un poco" que ya no te avisará si más adelante hay otro dato raro.
- Polars, al pedirle explícitamente que la columna sea de números enteros (`dtype=pl.Int64`), revisó cada valor y **se detuvo de inmediato** al encontrar `"treinta"`, señalando exactamente cuál fue el problema.
- Esto parece "más molesto" al principio, pero en la práctica te ahorra horas de andar buscando por qué un cálculo dio un resultado extraño tres notebooks después.

## 5. El Formato que Llega Ya Empacado: Parquet 📦⚡

Sigamos con la analogía de la mudanza: un archivo **CSV** es como una pila de hojas sueltas de papel — hay que leerlas una por una para saber qué contiene cada columna. Un archivo **Parquet**, en cambio, es como una caja de mudanza que **ya viene organizada, etiquetada y comprimida** por dentro: Polars no tiene que "adivinar" nada, simplemente abre la caja y ya sabe exactamente qué hay en cada compartimento.

Por eso, para el mismo dataset, Parquet casi siempre pesa menos en el disco duro y se lee más rápido que el CSV equivalente. Comprobémoslo:

```python
import time

ruta_parquet = load_dataset("ventas.parquet")
if not os.path.exists(ruta_parquet):
    ruta_parquet = load_dataset("ventas.parquet")

t0 = time.time()
_ = pl.read_csv(ruta)
t_csv = time.time() - t0

t0 = time.time()
df_ventas_parquet = pl.read_parquet(ruta_parquet)
t_parquet = time.time() - t0

print(f"CSV tardó:     {t_csv * 1000:.1f} ms")
print(f"Parquet tardó: {t_parquet * 1000:.1f} ms")
print(f"\nCSV pesa:     {os.path.getsize(ruta) / 1_048_576:.2f} MB")
print(f"Parquet pesa: {os.path.getsize(ruta_parquet) / 1_048_576:.2f} MB")
```

### 🤔 ¿Qué acaba de pasar?

- Leímos el mismo dataset de ventas dos veces: una desde `.csv` y otra desde `.parquet`, y medimos cuánto tardó cada una con el módulo `time`.
- Normalmente Parquet gana la carrera por buen margen, y ocupa bastante menos espacio en disco, porque además de venir "ya organizado" también viene comprimido.
- Los tiempos exactos dependen de tu computador, pero la tendencia (Parquet más rápido y más liviano) se mantiene casi siempre, y la diferencia se nota muchísimo más con archivos de varios gigabytes.

## 6. Mini Reto: Explora a los Clientes 🕵️

Antes de ver el código, piensa: si tuvieras la tabla `clientes.csv` (con columnas como `segmento`, `edad` e `ingreso_anual`), ¿cómo harías para ver **cuántos clientes hay de cada segmento**? Cuando tengas una idea, ejecuta la celda de abajo y compara.

```python
ruta_clientes = load_dataset("clientes.csv")
if not os.path.exists(ruta_clientes):
    ruta_clientes = load_dataset("clientes.csv")

df_clientes = pl.read_csv(ruta_clientes)
print(f"Clientes cargados: {df_clientes.height:,}")

# ¿Cuántos clientes hay por segmento?
conteo_segmentos = df_clientes["segmento"].value_counts()
display(conteo_segmentos)
```

### 🤔 ¿Qué acaba de pasar?

- `df_clientes["segmento"]` selecciona una sola columna como una `Serie` (fíjate en los corchetes simples, como si fuera un diccionario de Python).
- `.value_counts()` cuenta cuántas veces aparece cada valor distinto: perfecto para responder "¿cuántos clientes hay de cada segmento?" en una sola línea, sin bucles ni contadores manuales.
- Con datasets de miles (o millones) de filas, este tipo de conteo es justo donde más se nota la velocidad de Polars: reparte el trabajo entre varios núcleos sin que tengas que pensar en ello.

## 7. Resumen Relámpago ⚡

| Idea | En una frase |
|---|---|
| Polars | Un "bólido de F1" para procesar tablas de datos: Rust + Apache Arrow + todos los núcleos de tu CPU trabajando a la vez. |
| `pl.DataFrame` | Una tabla de datos sin índice explícito: cada fila vive en su posición natural. |
| `.height` / `.width` | Cuántas filas / cuántas columnas tiene la tabla. |
| Tipado estricto | Cada columna tiene un único tipo de dato, y Polars te avisa de inmediato si algo no encaja, en vez de ocultarlo. |
| CSV | Hojas sueltas de papel: hay que leerlas e interpretarlas una por una. |
| Parquet | Una caja de mudanza ya organizada, etiquetada y comprimida: se lee más rápido y pesa menos. |
| `.value_counts()` | Cuenta cuántas veces aparece cada valor distinto de una columna, en una sola línea. |
