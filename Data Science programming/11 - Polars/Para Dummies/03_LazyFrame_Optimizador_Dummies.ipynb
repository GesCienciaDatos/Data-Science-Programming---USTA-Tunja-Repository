## 1. ¿Por qué ser Perezoso te hace un Genio? 🦥💡

Imagina que vas a un restaurante:
* **Modo Eager (Pandas):** Le pides al mesero un vaso de agua. Va a la cocina, te lo trae. Le pides el menú. Va a la cocina, te lo trae. Le pides el pan. Va a la cocina, te lo trae. ¡10 viajes agotadores!
* **Modo Lazy (Polars LazyFrame):** Le anotas en una hoja todo lo que vas a pedir: agua, entrada, plato fuerte y postre. El mesero mira tu lista completa, la optimiza y te trae todo en una sola bandeja perfecta.

En Polars, cambiamos `read_csv` por `scan_csv`. ¡Polars no lee nada hasta que escribes `.collect()`!

````python
import polars as pl
import os

import os, urllib.request, urllib.parse

def load_dataset(filename, module_name="11 - Polars"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo genera automáticamente.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("Data Science programming", module_name, "data", filename),
        os.path.join("..", "Data Science programming", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path
            
    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Science programming/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)
    
    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]
    
    print(f"📥 Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=3) as response:
                if response.status == 200:
                    with open(target_path, 'wb') as out_f:
                        out_f.write(response.read())
                    print(f"✅ Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue
            
    print(f"⚙️ Generando '{filename}' sintéticamente para ejecución inmediata...")
    import polars as pl
    import numpy as np
    np.random.seed(42)
    
    n_clientes = 1000
    df_c = pl.DataFrame({
        'id_cliente': [f'CLI-{i:04d}' for i in range(1, n_clientes + 1)],
        'nombre': [f'Cliente_{i}' for i in range(1, n_clientes + 1)],
        'segmento': np.random.choice(['Corporativo', 'Pyme', 'Consumo', 'Gobierno'], n_clientes),
        'edad': np.random.randint(18, 70, n_clientes),
        'ciudad_residencia': np.random.choice(['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Bucaramanga'], n_clientes),
        'ingreso_anual': np.random.normal(45000000, 15000000, n_clientes).round(2)
    })
    
    n_ventas = 60000
    cats = ['Tecnología', 'Mobiliario', 'Material de Oficina', 'Servicios']
    prods = ['Laptop Pro', 'Monitor 4K', 'Silla Ergonómica', 'Escritorio', 'Papel A4', 'Tóner', 'Mantenimiento']
    ciudades = ['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Barranquilla']
    
    cant = np.random.randint(1, 10, n_ventas)
    pu = np.random.choice([25000.0, 120000.0, 450000.0, 1200000.0, 3500000.0], n_ventas)
    desc = np.random.choice([0.0, 0.05, 0.10, 0.15], n_ventas)
    tot = (cant * pu * (1 - desc)).round(2)
    
    df_v = pl.DataFrame({
        'id_venta': [f'VNT-{i:06d}' for i in range(1, n_ventas + 1)],
        'fecha': [f'2024-{np.random.randint(1,13):02d}-{np.random.randint(1,29):02d}' for _ in range(n_ventas)],
        'id_cliente': np.random.choice(df_c['id_cliente'], n_ventas),
        'categoria': np.random.choice(cats, n_ventas),
        'producto': np.random.choice(prods, n_ventas),
        'cantidad': cant,
        'precio_unitario': pu,
        'descuento': desc,
        'ciudad_venta': np.random.choice(ciudades, n_ventas),
        'total_venta': tot
    })
    
    c_csv_path = os.path.join("data", "clientes.csv")
    c_pq_path = os.path.join("data", "clientes.parquet")
    v_csv_path = os.path.join("data", "ventas.csv")
    v_pq_path = os.path.join("data", "ventas.parquet")
    
    if not os.path.exists(c_csv_path): df_c.write_csv(c_csv_path)
    if not os.path.exists(c_pq_path): df_c.write_parquet(c_pq_path)
    if not os.path.exists(v_csv_path): df_v.write_csv(v_csv_path)
    if not os.path.exists(v_pq_path): df_v.write_parquet(v_pq_path)
    
    print(f"✅ Datasets preparados exitosamente en '{target_path}'.")
    return target_path

ruta = load_dataset("ventas.csv")
lazy_plan = pl.scan_csv(ruta)
print("Plan perezoso listo:", type(lazy_plan))
````

### 🤔 ¿Qué acaba de pasar?

Al imprimir `lazy_plan` **no viste ni una fila de datos**. Viste una especie de "recibo" con el plan: de dónde va a leer y qué va a hacer. Es exactamente como anotar tu lista de compras en un papel: todavía no has ido al supermercado, solo tienes el plan escrito.

`pl.scan_csv()` es la versión perezosa de `pl.read_csv()`: en vez de la palabra "leer" (read), usamos la palabra "escanear" (scan) — Polars le echa un vistazo rápido al archivo para conocer sus columnas, pero no carga los datos todavía.

## 2. El Optimizador: tu Asistente que Reordena la Lista de Pendientes 🧑‍💼📋

Imagina que le dictas a un asistente muy inteligente una lista de tareas desordenada: "ve a comprar pan, pero antes revisa que la tienda tenga descuento, y de paso trae solo pan integral". Un asistente torpe haría los pasos en el orden exacto que se los dijiste, con viajes de más. Un asistente inteligente **reorganiza mentalmente** la lista para hacer el mínimo esfuerzo: revisa el descuento y el tipo de pan en el mismo viaje, antes de decidir si entra a la tienda.

Eso hace el optimizador de Polars con tu código: no importa en qué orden escribas `.filter()` y `.select()`, Polars los reorganiza para tocar el disco lo menos posible.

````python
# 2. Le damos una lista de tareas
pedido = (
    lazy_plan
    .filter(pl.col("ciudad_venta") == "Tunja")
    .select(["id_venta", "producto", "total_venta"])
    .filter(pl.col("total_venta") > 1000000)
)

# 3. ¡Ahora sí! Con .collect() Polars ejecuta todo optimizado en un parpadeo
resultado_final = pedido.collect()
print(f"Filas obtenidas: {resultado_final.height}")
display(resultado_final.head(4))
````

### 🤔 ¿Qué acaba de pasar?

Escribimos tres pasos (`.filter()`, `.select()`, `.filter()`) pero Polars **no los ejecutó uno por uno como si fueran instrucciones sueltas**. Esperó a ver la lista completa (`.collect()`) y luego decidió la forma más rápida de hacerlo todo en un solo recorrido de los datos: leer del disco solo las columnas que pedimos y solo las filas que cumplen ambos filtros.

Si hubiéramos usado Pandas con `pd.read_csv()` + filtros normales, Pandas habría cargado el CSV completo en memoria (las 10 columnas y las 60,000 filas) **antes** de aplicar cualquier filtro. Polars, en modo lazy, ni siquiera lee las columnas que no necesita.

## 3. Espiando el Plan del Asistente con `.explain()` 🕵️‍♀️

Si quieres ver *exactamente* qué decidió hacer tu asistente inteligente antes de que mueva un dedo, pídele el plan con `.explain()`. No hace falta entender cada palabra técnica: fíjate solo en dos cosas:

* `PROJECT` → cuántas columnas del total va a leer (menos es mejor).
* `SELECTION` → qué condiciones va a aplicar mientras lee, en vez de después.

````python
print(pedido.explain())
````

### 🤔 ¿Qué acaba de pasar?

El plan te muestra algo como `PROJECT 3/10 COLUMNS` y `SELECTION: ...`. Traducido: de las 10 columnas de `ventas.csv`, Polars solo va a leer 3 (las que pediste en `.select()`), y va a aplicar tus dos condiciones de `.filter()` **en el mismo momento** en que lee el archivo, no en un paso aparte después. Es la prueba en blanco y negro de que tu "asistente" (el optimizador) sí reorganizó la lista de tareas antes de moverse.

## 4. El Motor de Streaming: Tomar Agua de una Manguera, no Tragarte el Mar 🌊🚰

Cuando el dataset es tan grande que no cabe cómodo en la memoria RAM de tu computador, Polars puede procesarlo **por partes**, como beber de una manguera con calma, en vez de intentar tragarte el océano entero de un solo golpe (lo que ahogaría a cualquier computador con poca memoria).

Esto se activa cambiando el motor de ejecución al llamar `.collect()`:

```python
resultado = pedido.collect(engine="streaming")
```

````python
# El motor "streaming" procesa el archivo por lotes en vez de cargarlo todo de golpe
resultado_streaming = pedido.collect(engine="streaming")
print("Filas obtenidas con streaming:", resultado_streaming.height)
print("¿Es el mismo resultado que con el motor normal?", resultado_streaming.equals(resultado_final))
````

### 🤔 ¿Qué acaba de pasar?

Con un archivo de 60,000 filas no vas a notar ninguna diferencia de velocidad o memoria — es lo bastante pequeño para caber en RAM sin problema. Pero si este mismo código corriera sobre un archivo de 500 millones de filas, `engine="streaming"` sería la diferencia entre que tu programa termine bien o se quede sin memoria (*Out of Memory*) y se caiga. Es la misma idea de la manguera: Polars procesa un trago de datos, los agrega a lo que ya llevaba calculado, suelta ese trago de memoria y sigue con el siguiente.

## 6. Resumen Relámpago ⚡

* **`pl.scan_csv()`** = anotar la lista de compras, sin ir todavía al supermercado.
* **`.collect()`** = ir por fin al supermercado, con la lista ya optimizada.
* El **optimizador** reordena tu lista de tareas solito, para hacer el menor número de viajes posible.
* **`.explain()`** = pedirle al asistente que te muestre el plan antes de moverse.
* **`engine="streaming"`** = beber de a poco con una manguera, en vez de tragarte el océano entero — útil cuando los datos no caben en la memoria.
