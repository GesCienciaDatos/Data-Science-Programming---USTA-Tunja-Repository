## 1. `group_by()`: El Jefe que Organiza los Equipos 👔

Hacer resúmenes en Polars es súper intuitivo. Usamos `group_by("columna")` y dentro de `.agg([...])` metemos todas las preguntas que tengamos. Imagina a `group_by` como un jefe que separa a todos los empleados en equipos (uno por cada valor distinto de la columna) y luego le pide a cada equipo un reporte con `.agg()`.

```python
import polars as pl
import os

import os, urllib.request, urllib.parse

def load_dataset(filename, module_name="11 - Polars"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo genera automáticamente.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("Data Science programming", module_name, "data", filename),
        os.path.join("..", "Data Science programming", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path
            
    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Science programming/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)
    
    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]
    
    print(f"📥 Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=3) as response:
                if response.status == 200:
                    with open(target_path, 'wb') as out_f:
                        out_f.write(response.read())
                    print(f"✅ Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue
            
    print(f"⚙️ Generando '{filename}' sintéticamente para ejecución inmediata...")
    import polars as pl
    import numpy as np
    np.random.seed(42)
    
    n_clientes = 1000
    df_c = pl.DataFrame({
        'id_cliente': [f'CLI-{i:04d}' for i in range(1, n_clientes + 1)],
        'nombre': [f'Cliente_{i}' for i in range(1, n_clientes + 1)],
        'segmento': np.random.choice(['Corporativo', 'Pyme', 'Consumo', 'Gobierno'], n_clientes),
        'edad': np.random.randint(18, 70, n_clientes),
        'ciudad_residencia': np.random.choice(['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Bucaramanga'], n_clientes),
        'ingreso_anual': np.random.normal(45000000, 15000000, n_clientes).round(2)
    })
    
    n_ventas = 60000
    cats = ['Tecnología', 'Mobiliario', 'Material de Oficina', 'Servicios']
    prods = ['Laptop Pro', 'Monitor 4K', 'Silla Ergonómica', 'Escritorio', 'Papel A4', 'Tóner', 'Mantenimiento']
    ciudades = ['Tunja', 'Bogotá', 'Medellín', 'Cali', 'Barranquilla']
    
    cant = np.random.randint(1, 10, n_ventas)
    pu = np.random.choice([25000.0, 120000.0, 450000.0, 1200000.0, 3500000.0], n_ventas)
    desc = np.random.choice([0.0, 0.05, 0.10, 0.15], n_ventas)
    tot = (cant * pu * (1 - desc)).round(2)
    
    df_v = pl.DataFrame({
        'id_venta': [f'VNT-{i:06d}' for i in range(1, n_ventas + 1)],
        'fecha': [f'2024-{np.random.randint(1,13):02d}-{np.random.randint(1,29):02d}' for _ in range(n_ventas)],
        'id_cliente': np.random.choice(df_c['id_cliente'], n_ventas),
        'categoria': np.random.choice(cats, n_ventas),
        'producto': np.random.choice(prods, n_ventas),
        'cantidad': cant,
        'precio_unitario': pu,
        'descuento': desc,
        'ciudad_venta': np.random.choice(ciudades, n_ventas),
        'total_venta': tot
    })
    
    c_csv_path = os.path.join("data", "clientes.csv")
    c_pq_path = os.path.join("data", "clientes.parquet")
    v_csv_path = os.path.join("data", "ventas.csv")
    v_pq_path = os.path.join("data", "ventas.parquet")
    
    if not os.path.exists(c_csv_path): df_c.write_csv(c_csv_path)
    if not os.path.exists(c_pq_path): df_c.write_parquet(c_pq_path)
    if not os.path.exists(v_csv_path): df_v.write_csv(v_csv_path)
    if not os.path.exists(v_pq_path): df_v.write_parquet(v_pq_path)
    
    print(f"✅ Datasets preparados exitosamente en '{target_path}'.")
    return target_path

ruta_v = load_dataset("ventas.csv")
ruta_c = load_dataset("clientes.csv")
df_ventas = pl.read_csv(ruta_v)
df_clientes = pl.read_csv(ruta_c)
print(f"Ventas: {len(df_ventas):,} | Clientes: {len(df_clientes):,}")
```

### 🤔 ¿Qué acaba de pasar?

- `group_by("ciudad_venta")` separó las 60,000 filas en 7 "equipos" (uno por ciudad). Todavía no calcula nada: solo organiza.
- `.agg([...])` es donde le pedimos el reporte a cada equipo: cuántas ventas tuvo (`pl.len()`), cuánto dinero movió en total (`.sum()`) y cuál fue el promedio de unidades por venta (`.mean()`).
- El resultado tiene **una sola fila por ciudad** — de 60,000 filas originales pasamos a solo 7. Eso es lo que distingue a `group_by`: resume, no conserva cada fila individual.
- `.sort("dinero_total", descending=True)` ordena el reporte de la ciudad que más vendió a la que menos.

## 2. `join()`: Juntando Dos Listas por un Código en Común 🧩

Imagina que tienes dos hojas de Excel: una con **ventas** (qué se vendió) y otra con **clientes** (quién es cada comprador). Ambas comparten una columna `id_cliente`, como un carné de identidad. Un `join()` es literalmente "grapar" las dos hojas, fila por fila, haciendo coincidir ese carné.

```python
ventas_con_cliente = df_ventas.join(
    df_clientes,
    on="id_cliente",
    how="left"
)
print(f"Ventas originales: {df_ventas.shape} -> Después del join: {ventas_con_cliente.shape}")
display(ventas_con_cliente.select(["id_venta", "id_cliente", "nombre", "segmento", "total_venta"]).head(4))
```

### 🤔 ¿Qué acaba de pasar?

- `on="id_cliente"` le dice a Polars: "usa esta columna como el carné de identidad para emparejar filas entre las dos tablas".
- `how="left"` significa "conserva **todas** las filas de `df_ventas` (la tabla de la izquierda), y pégales los datos de `df_clientes` que hagan match". Si algún `id_cliente` no existiera en `clientes`, esas columnas quedarían vacías (`null`) en vez de perder la venta.
- Fíjate en el `.shape`: seguimos teniendo el mismo número de filas que en `df_ventas`, pero ahora con más columnas (las de `clientes` pegadas al lado). Ese es el efecto de un `left` join: nunca pierdes filas de la tabla izquierda.

### 2.1. Variantes de `join`: no siempre quieres pegar columnas

A veces solo quieres **filtrar** una tabla usando otra como criterio, sin fusionar sus columnas. Para eso existen `"semi"` (¿quién SÍ hace match?) y `"anti"` (¿quién NO hace match?).

```python
# Clientes del segmento "Corporativo"
clientes_corp = df_clientes.filter(pl.col("segmento") == "Corporativo")

# SEMI: ventas hechas por esos clientes corporativos (sin traer sus columnas)
ventas_de_corp = df_ventas.join(clientes_corp, on="id_cliente", how="semi")
print(f"Ventas de clientes Corporativo: {ventas_de_corp.height}")
print("Columnas del resultado (iguales a las de ventas, sin nada nuevo):", ventas_de_corp.columns)

# ANTI: ventas hechas por clientes que NO son corporativos
ventas_no_corp = df_ventas.join(clientes_corp, on="id_cliente", how="anti")
print(f"\nVentas de clientes que NO son Corporativo: {ventas_no_corp.height}")
```

### 🤔 ¿Qué acaba de pasar?

- `"semi"` es como preguntar en la puerta: "¿tu carné está en esta otra lista?" — si sí, pasas, pero no te dan nada extra (no se pegan columnas de `clientes`). Por eso `ventas_de_corp` tiene exactamente las mismas columnas que `df_ventas`.
- `"anti"` es lo opuesto: solo pasan los que **no** están en la otra lista.
- Nota que `ventas_de_corp.height + ventas_no_corp.height` debería dar el total de `df_ventas` — entre ambos joins cubrimos absolutamente todas las filas, sin traslapes.

## 3. El Superpoder Secreto: `.over()` (Ventana sin mareos) 🪟

¿Alguna vez quisiste calcular qué porcentaje del total de su categoría representa cada venta, sin tener que hacer un `join` con otra tabla ni perder ninguna fila con `group_by`? ¡Con `.over()` lo haces en 1 línea!

**Analogía:** imagina que te comparas con el promedio de tu salón de clase, pero **sin dejar de ser tú mismo** en la lista — no te conviertes en "el promedio del salón", solo te enteras de cómo te comparas con él, fila por fila.

```python
calculo_genial = df_ventas.select([
    pl.col("id_venta"),
    pl.col("categoria"),
    pl.col("total_venta"),
    # Promedio de esa categoría pegado directamente en la fila
    pl.col("total_venta").mean().over("categoria").round(2).alias("promedio_de_su_categoria"),
    # Comparación: ¿Estuvo por encima o por debajo del promedio?
    (pl.col("total_venta") - pl.col("total_venta").mean().over("categoria")).round(2).alias("diferencia_con_media")
])
display(calculo_genial.head(6))
```

### 🤔 ¿Qué acaba de pasar?

- `.over("categoria")` le dice a Polars: "calcula este promedio **por categoría**, pero pégalo de vuelta en cada fila individual, sin resumir la tabla". Por eso seguimos teniendo tantas filas como ventas — nada se colapsó, a diferencia de `group_by()`.
- Cada venta ahora "sabe" cuál es el promedio de su propia categoría (`promedio_de_su_categoria`), sin que tuvieras que hacer un `group_by` aparte y luego pegarlo con un `join`.
- `diferencia_con_media` te dice, fila por fila, si esa venta estuvo por encima (positivo) o por debajo (negativo) del promedio de su categoría — perfecto para detectar ventas "raras" o destacadas sin perder detalle.

## 4. Mini Reto: ¿Quién es el Mejor Cliente de Cada Ciudad? 🏆

Antes de ver el código, piensa: si quisieras saber cuánto gastó cada cliente en total, y luego compararlo contra los demás clientes **de su misma ciudad**, ¿qué combinarías: `join`, `group_by` y/o `.over()`? Cuando tengas una idea, ejecuta la celda de abajo y compara.

```python
gasto_por_cliente = (
    df_ventas
    .join(df_clientes, on="id_cliente", how="left")
    .group_by(["ciudad_residencia", "nombre"])
    .agg(pl.col("total_venta").sum().round(2).alias("gasto_total"))
)

mejor_cliente_por_ciudad = (
    gasto_por_cliente
    .filter(
        pl.col("gasto_total") == pl.col("gasto_total").max().over("ciudad_residencia")
    )
    .sort("ciudad_residencia")
)
display(mejor_cliente_por_ciudad)
```

### 🤔 ¿Qué acaba de pasar?

- Primero usamos `join` para pegarle a cada venta el nombre y la ciudad del cliente que la hizo.
- Luego `group_by(["ciudad_residencia", "nombre"])` + `.agg()` resumió esas ventas en **un gasto total por cliente**, ya con una sola fila por persona.
- Finalmente, `.over("ciudad_residencia")` calculó el gasto máximo **dentro de cada ciudad** y lo comparamos con `==` contra el gasto de cada cliente — así nos quedamos solo con quien empató con ese máximo: el mejor cliente de cada ciudad, sin perder de vista a qué ciudad pertenece cada uno.

## 5. Resumen Relámpago ⚡

| Idea | En una frase |
|---|---|
| `group_by(...).agg(...)` | El jefe que separa en equipos y les pide un reporte: una fila de salida por grupo. |
| `join(..., on=..., how="left")` | Grapar dos tablas por un código en común, sin perder filas de la tabla izquierda. |
| `join(..., how="semi")` | Filtra la tabla izquierda por quién SÍ tiene match, sin pegar columnas nuevas. |
| `join(..., how="anti")` | Filtra la tabla izquierda por quién NO tiene match. |
| `.over("columna")` | Te comparas con tu grupo (promedio, máximo, ranking...) sin dejar de ser tú mismo en la lista — ninguna fila desaparece. |
