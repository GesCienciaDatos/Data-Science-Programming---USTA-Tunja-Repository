## ¿Por qué los Datos Rara Vez Están Listos para Usar? 🧹

**El secreto sucio de la Ciencia de Datos:** Los científicos de datos pasan entre el **60-80% del tiempo**
limpiando y preparando datos. El modelo es solo el 20-40% del trabajo.

### ¿Por Qué los Datos Son 'Sucios'?

| Problema | Ejemplo | Consecuencia |
|---|---|---|
| **Valores faltantes** | Encuesta con preguntas en blanco | El modelo se rompe o sesga |
| **Outliers** | Salario de $500,000,000 en una empresa pequeña | Distorsiona la distribución |
| **Inconsistencias** | 'Bogota', 'BOGOTÁ', 'bogota' | Se tratan como 3 ciudades distintas |
| **Escala diferente** | Edad (0-100) vs Ingreso (0-10,000,000) | Algunos modelos se confunden |
| **Duplicados** | El mismo cliente registrado 3 veces | Sesgo en estadísticas |

### El Pipeline Estándar de Preparación de Datos

```
1. Carga → 2. Exploración (EDA) → 3. Limpieza → 4. Transformación → 5. Escalado → 6. Listo para ML
```


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Simulamos un dataset 'sucio' realista de una empresa
np.random.seed(42)
n = 150
df_sucio = pd.DataFrame({
    'ID': range(1, n+1),
    'Nombre': [f'Empleado_{i}' for i in range(1, n+1)],
    'Edad': np.concatenate([np.random.randint(22, 60, n-5), [None, None, -5, 200, 999]]),
    'Salario': np.concatenate([np.random.normal(4500000, 1200000, n-5),
                                [None, None, 0, 50000000, 4800000]]),
    'Ciudad': np.concatenate([np.random.choice(['Tunja', 'Bogotá', 'Medellín'], n-3),
                               ['TUNJA', 'bogota', '  Medellín ']]),
    'Departamento': np.random.choice(['TI', 'Ventas', 'RRHH', None], n),
})

print('🚨 DIAGNÓSTICO DEL DATASET SUCIO:')
print(f'  Filas: {df_sucio.shape[0]} | Columnas: {df_sucio.shape[1]}')
print(f'  Nulos: {df_sucio.isnull().sum().sum()}')
print(f'  Edades negativas o imposibles: {(df_sucio["Edad"] <= 0).sum()}')
print(f'  Ciudades únicas: {df_sucio["Ciudad"].nunique()} (¡debería ser 3!)')
display(df_sucio.tail(8))
````

````python
# El pipeline de limpieza
df = df_sucio.copy()

# 1. Corregir inconsistencias de texto
df['Ciudad'] = df['Ciudad'].str.strip().str.title()

# 2. Eliminar valores imposibles
df = df[df['Edad'].between(18, 80, inclusive='both') | df['Edad'].isnull()]
df = df[df['Salario'].between(500000, 20000000) | df['Salario'].isnull()]

# 3. Imputar nulos con la mediana
df['Edad'] = df['Edad'].fillna(df['Edad'].median())
df['Salario'] = df['Salario'].fillna(df['Salario'].median())
df['Departamento'] = df['Departamento'].fillna('Sin Asignar')

# 4. Verificar resultado
print('✅ DATASET LIMPIO:')
print(f'  Filas: {df.shape[0]} | Nulos: {df.isnull().sum().sum()}')
print(f'  Ciudades únicas: {df["Ciudad"].nunique()} ✓')
display(df.describe()[['Edad', 'Salario']].round(0))
````
