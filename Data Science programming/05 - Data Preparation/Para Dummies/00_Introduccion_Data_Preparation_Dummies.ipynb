## 1. La analogía de la cocina 👩‍🍳

Imagina que eres chef y quieres preparar un delicioso plato. Tienes los ingredientes pero:

- Algunas frutas están podridas *(valores erróneos o extremos)*.
- Falta la sal en la lista *(valores faltantes)*.
- Las medidas vienen en tazas y en gramos mezcladas *(escalas inconsistentes)*.
- Un ingrediente está escrito "Zanoria" en lugar de "Zanahoria" *(errores tipográficos)*.

¿Qué haces antes de cocinar? **Limpias y organizas los ingredientes.**

Eso es exactamente la **Preparación de Datos**: antes de entrenar cualquier modelo de Machine Learning, los datos deben estar limpios, consistentes y en el mismo formato.

> 🧑‍🍳 **Para recordar:** Un chef puede tener la mejor receta del mundo, pero si los ingredientes están en mal estado, el plato saldrá horrible. Lo mismo pasa con los modelos de IA y los datos sucios.

---
## 2. El principio GIGO — Basura Entra, Basura Sale 🗑️

**GIGO** viene del inglés *"Garbage In, Garbage Out"* (Basura entra, Basura sale). Es el principio más importante de la Ciencia de Datos:

```
  Datos sucios   →   Modelo inteligente   →   Predicciones basura
      🗑️                   🤖                        💩

  Datos limpios  →   Modelo inteligente   →   Predicciones útiles
      ✨                   🤖                        🎯
```

Ningún algoritmo, por sofisticado que sea (ni siquiera ChatGPT), puede sacar buenas predicciones de datos malos. La calidad del modelo está limitada por la calidad de los datos.

> 💡 **Dato real:** Los científicos de datos profesionales dedican entre el **70% y el 80% de su tiempo** a limpiar y preparar datos, y solo el 20% a entrenar y ajustar modelos. ¡La limpieza es el trabajo real!

## 4. El pipeline de limpieza — El orden importa 🗺️

Limpiar datos tiene un orden lógico. Siempre sigue este pipeline:

```
 📂 1. Cargar los datos
        ↓
 🔍 2. Diagnóstico (¿qué está mal? ¿cuánto?)
        ↓
 🕳️ 3. Tratar valores faltantes (rellenar o eliminar)
        ↓
 🏷️ 4. Corregir tipos de dato (fechas, números mal leídos como texto)
        ↓
 ✏️ 5. Corregir inconsistencias (errores tipográficos, mayúsculas/minúsculas)
        ↓
 📏 6. Escalar variables numéricas (ponerlas en la misma escala)
        ↓
 ✅ 7. Datos listos para el modelo
```

> ⚠️ **El orden importa:** No puedes escalar variables que aún tienen NaN (valores vacíos). Primero se tratan los nulos, luego se escala.

````python
# Demostración mínima: cómo se ve un dataset "sucio" vs uno "limpio"
import pandas as pd
import numpy as np

# Dataset sucio (como llegan los datos del mundo real)
datos_sucios = pd.DataFrame({
    'nombre':  ['Ana',   'Luis',  'María', 'CARLOS', 'María'],
    'edad':    [25,       np.nan,  30,      22,       30],
    'salario': [3000000,  5000000, np.nan,  2500000,  3100000],
    'fecha':   ['2024-01-15', '15/01/2024', 'Jan 15 2024', '2024-01-15', None]
})

print("🗑️ Dataset SUCIO (como llega del sistema):")
print(datos_sucios.to_string())
print(f"\nProblemas detectados:")
print(f"  • Nulos: {datos_sucios.isnull().sum().sum()} valores vacíos")
print(f"  • Nombres inconsistentes: 'María' aparece dos veces, 'CARLOS' en mayúsculas")
print(f"  • Fechas en 3 formatos distintos")
````

````python
# Dataset limpio (después del proceso de Data Preparation)
datos_limpios = pd.DataFrame({
    'nombre':  ['Ana',   'Luis',  'María', 'Carlos'],  # Sin duplicados, nombres estandarizados
    'edad':    [25,       27.5,    30,      22],        # NaN rellenado con la media
    'salario': [3000000,  5000000, 3050000, 2500000],   # NaN rellenado con la mediana
    'fecha':   pd.to_datetime(['2024-01-15', '2024-01-15', '2024-01-15', '2024-01-15'])
})

print("✨ Dataset LIMPIO (listo para el modelo):")
print(datos_limpios.to_string())
print(f"\n✅ Sin nulos: {datos_limpios.isnull().sum().sum()} valores vacíos")
print(f"✅ Sin duplicados de nombres")
print(f"✅ Fechas en un único formato estándar")
````
