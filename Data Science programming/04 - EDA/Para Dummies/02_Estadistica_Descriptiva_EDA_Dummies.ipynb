## 1. La trampa del promedio — Bill Gates en la cafetería ☕

Imagina que en una cafetería hay 9 personas con salarios de $1,200,000 al mes. Luego entra Bill Gates (salario mensual: ~$10,000,000,000).

¿Cuál sería el salario **promedio** ahora? ¡Parecería que todos son millonarios! Pero claramente ese número no representa a nadie en la cafetería.

Eso se llama **sesgo por valores extremos (outliers)**. El promedio es muy sensible a valores muy grandes o muy pequeños.

````python
import pandas as pd
import numpy as np

salarios = [1_200_000] * 9 + [10_000_000_000]  # 9 personas normales + Bill Gates

s = pd.Series(salarios)
print(f"💰 Salario promedio (media):  ${s.mean():,.0f}")
print(f"💰 Salario mediana:           ${s.median():,.0f}")
print(f"\n¿Cuál representa mejor a los 9 empleados de la cafetería?")
print(f"→ La MEDIANA: ${s.median():,.0f} (que es el salario real de la mayoría)")
````

### 🧠 Regla de oro:
| Situación | Usa... |
|---|---|
| Datos sin valores extremos (notas de un examen) | **Media (promedio)** |
| Datos con valores extremos (salarios, precios de casas) | **Mediana** |

> 💡 **Para recordar:** La mediana es simplemente el valor del **medio** cuando ordenas todos los datos de menor a mayor. Bill Gates no puede "jalar" la mediana hacia arriba porque está demasiado lejos de los demás.

## 2. La desviación estándar — ¿Qué tan juntos están los datos? 📏

La desviación estándar mide **qué tan dispersos están los datos alrededor del promedio**. Piensa en esto:

- **Grupo A de estudiantes:** todos sacaron entre 4.0 y 5.0 → poca dispersión (desviación estándar pequeña).
- **Grupo B de estudiantes:** algunos sacaron 0.0 y otros 10.0 → mucha dispersión (desviación estándar grande).

Ambos grupos podrían tener el mismo promedio (5.0), pero son muy diferentes entre sí.

````python
import os, urllib.parse, urllib.request

def load_dataset(filename, module_name="04 - EDA"):
    candidates = [f"data/{filename}", f"../{module_name}/data/{filename}", f"{module_name}/data/{filename}", filename]
    for path in candidates:
        if os.path.exists(path):
            return path
    os.makedirs("data", exist_ok=True)
    target_path = f"data/{filename}"
    url = f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{urllib.parse.quote(module_name)}/data/{urllib.parse.quote(filename)}"
    urllib.request.urlretrieve(url, target_path)
    print(f"✅ Dataset '{filename}' descargado.")
    return target_path

df = pd.read_csv(load_dataset('StudentsPerformance.csv'))

# Comparamos dos grupos de notas
notas_mat = df['math score']
print(f"📊 Notas de Matemáticas:")
print(f"   Promedio:            {notas_mat.mean():.2f}")
print(f"   Mediana:             {notas_mat.median():.2f}")
print(f"   Desviación estándar: {notas_mat.std():.2f}")
print(f"   Mínimo:              {notas_mat.min()}")
print(f"   Máximo:              {notas_mat.max()}")
````

## 3. Los cuartiles — Dividiendo los datos en 4 partes iguales 🍕

Imagina que ordenas a todos los estudiantes de menor a mayor nota y los divides en 4 grupos iguales (como cuartos de pizza):

```
 0%   ←— 25% —→  Q1  ←— 25% —→  Q2(mediana)  ←— 25% —→  Q3  ←— 25% —→  100%
 min                                                                        max
```

- **Q1 (percentil 25):** El 25% de los estudiantes tiene una nota menor que este valor.
- **Q2 (percentil 50 = mediana):** La mitad de los estudiantes está por debajo.
- **Q3 (percentil 75):** El 75% de los estudiantes tiene una nota menor que este valor.
- **IQR = Q3 - Q1:** El rango del 50% central de los datos. Se usa para detectar valores atípicos.

````python
Q1 = notas_mat.quantile(0.25)
Q2 = notas_mat.quantile(0.50)
Q3 = notas_mat.quantile(0.75)
IQR = Q3 - Q1

print(f"🍕 Cuartiles de las notas de Matemáticas:")
print(f"   Q1 (25%):  {Q1} — El 25% de los estudiantes sacó menos de esto")
print(f"   Q2 (50%):  {Q2} — La mitad de los estudiantes sacó menos de esto")
print(f"   Q3 (75%):  {Q3} — El 75% de los estudiantes sacó menos de esto")
print(f"   IQR:       {IQR} — El rango del 50% central de los datos")

# Límites para detectar outliers con la regla del IQR
lim_inferior = Q1 - 1.5 * IQR
lim_superior = Q3 + 1.5 * IQR
outliers = notas_mat[(notas_mat < lim_inferior) | (notas_mat > lim_superior)]
print(f"\n⚠️  Posibles valores atípicos (outliers): {len(outliers)} estudiantes")
````

## 4. Asimetría (Skewness) — ¿La cola del gráfico va a la izquierda o a la derecha? 🐠

Cuando graficas la distribución de los datos, la forma puede ser:

| Forma | Nombre técnico | Qué significa |
|---|---|---|
| Campana simétrica | Sin asimetría (skewness ≈ 0) | Los datos se distribuyen igual a ambos lados del promedio |
| Cola larga a la derecha | Asimetría positiva (skewness > 0) | Hay algunos valores muy altos que "jalan" la cola |
| Cola larga a la izquierda | Asimetría negativa (skewness < 0) | Hay algunos valores muy bajos que "jalan" la cola |

> 💡 **Ejemplo cotidiano:** Los salarios tienen asimetría positiva (muy pocos millonarios jalan la cola hacia la derecha). Las notas de exámenes muy fáciles tienen asimetría negativa (casi todos sacan notas altas).

````python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
columnas = ['math score', 'reading score', 'writing score']
colores = ['#f59e0b', '#10b981', '#6366f1']

for ax, col, color in zip(axes, columnas, colores):
    df[col].hist(bins=20, ax=ax, color=color, alpha=0.8, edgecolor='white')
    ax.set_title(f'{col}\nSkewness: {df[col].skew():.3f}', fontweight='bold')
    ax.set_xlabel('Puntaje')
    ax.set_ylabel('Frecuencia')
    ax.axvline(df[col].mean(), color='red', linestyle='--', linewidth=2, label='Media')
    ax.axvline(df[col].median(), color='blue', linestyle='-', linewidth=2, label='Mediana')
    ax.legend()

plt.suptitle('Distribución de puntajes — ¿La media y la mediana coinciden?', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("\n📌 Cuando la media (rojo) ≠ mediana (azul), los datos tienen asimetría.")
````

---
## 5. Resumen visual completo con `.describe()` 📋

````python
resumen = df[columnas].describe().round(2)
print("📊 Resumen estadístico completo de los tres puntajes:")
display(resumen)

print("\n🔎 Interpretación rápida:")
for col in columnas:
    media = df[col].mean()
    mediana = df[col].median()
    diferencia = abs(media - mediana)
    print(f"   {col}: media={media:.1f}, mediana={mediana:.1f}, diferencia={diferencia:.1f} ({'asimétrico' if diferencia > 2 else 'casi simétrico'})")
````
