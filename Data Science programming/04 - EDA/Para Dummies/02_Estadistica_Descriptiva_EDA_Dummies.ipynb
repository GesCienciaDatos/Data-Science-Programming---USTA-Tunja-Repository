## Estadística Descriptiva: Los Números que Cuentan la Historia 📊

Antes de hacer una sola gráfica, los números te dicen mucho. Aquí están las métricas clave:

### Las Medidas de Centralidad — ¿Dónde está el 'centro'?

| Medida | Fórmula | Sensible a Outliers | Cuándo Usar |
|---|---|---|---|
| **Media** (promedio) | Σx / n | ⚠️ SÍ, mucho | Variables sin outliers extremos |
| **Mediana** | Valor del medio | ✅ NO | Variables con outliers (salarios, precios) |
| **Moda** | Valor más frecuente | ✅ NO | Variables categóricas o discretas |

### Las Medidas de Dispersión — ¿Qué tan 'esparcidos' están?

| Medida | ¿Qué mide? |
|---|---|
| **Rango** | Máx - Mín |
| **Varianza** | Promedio de desviaciones cuadradas |
| **Desv. Estándar** | Raíz cuadrada de la varianza |
| **IQR** | Percentil 75 - Percentil 25 |


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Dataset de salarios con outliers realistas
np.random.seed(42)
salarios_normales = np.random.normal(4000000, 1500000, 100)
salarios_outliers = [25000000, 30000000, 28000000]  # CEO y directivos
salarios = np.concatenate([salarios_normales, salarios_outliers])
salarios = np.maximum(salarios, 1000000)  # mínimo 1M

df = pd.DataFrame({'Salario': salarios, 'Departamento': np.random.choice(
    ['TI', 'Marketing', 'Ventas', 'RRHH', 'Dirección'], len(salarios))})
df.loc[df.index[-3:], 'Departamento'] = 'Dirección'

media = df['Salario'].mean()
mediana = df['Salario'].median()

print(f'📊 Estadísticas de Salarios:')
print(f'  Media:    ${media:,.0f}')
print(f'  Mediana:  ${mediana:,.0f}')
print(f'  Diferencia: ${media - mediana:,.0f} (impacto de outliers en la media)')
```

```python
# Describe completo
print('📋 Resumen estadístico completo:')
resumen = df.groupby('Departamento')['Salario'].agg(['mean', 'median', 'std', 'count'])
resumen.columns = ['Media', 'Mediana', 'Desv.Std', 'N']
resumen = resumen.astype({'N': int}).round(0)
display(resumen)
```

```python
# Visualización: Media vs Mediana con outliers
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Histograma con media y mediana
axes[0].hist(df['Salario'], bins=25, color='steelblue', edgecolor='white', alpha=0.7)
axes[0].axvline(media, color='red', linewidth=2, label=f'Media: ${media/1e6:.1f}M')
axes[0].axvline(mediana, color='green', linewidth=2, linestyle='--', label=f'Mediana: ${mediana/1e6:.1f}M')
axes[0].set_title('Distribución de Salarios', fontweight='bold')
axes[0].set_xlabel('Salario (COP)')
axes[0].legend()

# Boxplot por departamento
depts = df.groupby('Departamento')['Salario'].apply(list)
axes[1].boxplot(depts.values, labels=depts.index, vert=True)
axes[1].set_title('Boxplot por Departamento', fontweight='bold')
axes[1].set_ylabel('Salario (COP)')
axes[1].tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.show()

print('💡 Observa cómo la Media es jalada hacia arriba por los salarios de Dirección.')
print('   La Mediana es más representativa del salario típico del empleado promedio.')
```

**Pregunta:** Tienes un dataset de precios de inmuebles en Bogotá. ¿Usarías la media o la mediana como 'precio típico'? ¿Por qué?

<details>
<summary>💡 Ver respuesta</summary>

**La mediana** es la medida más apropiada para precios de inmuebles.

Los precios de inmuebles tienen una distribución **sesgada a la derecha** (algunos penthouse y
edificios de lujo tienen precios 10-100x superiores al promedio).

Estos outliers **inflan la media** haciéndola no representativa.
La mediana ignora los extremos y refleja el precio del inmueble 'de en medio'.
