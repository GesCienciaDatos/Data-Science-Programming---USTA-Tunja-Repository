## 1. El EDA es como llegar a la escena del crimen 🔍

Imagina que eres un detective que llega a investigar un caso. Lo primero que haces **no** es sacar conclusiones — primero inspeccionas la escena:

- ¿Cuántas personas estuvieron aquí? *(¿cuántas filas tiene el dataset?)*
- ¿Qué objetos hay en la habitación? *(¿qué columnas hay?)*
- ¿Hay algo roto, faltante o fuera de lugar? *(¿hay datos nulos o erróneos?)*
- ¿Qué patrones o pistas llaman la atención? *(¿hay correlaciones o tendencias?)*

Eso es exactamente el **EDA (Análisis Exploratorio de Datos)**: inspeccionar los datos antes de sacar cualquier conclusión o construir cualquier modelo.

> 🕵️ **Para recordar:** Un buen detective nunca asume quién fue el culpable antes de revisar todas las pistas. Un buen científico de datos nunca construye un modelo sin explorar primero sus datos.

## 2. ¿Por qué no basta con mirar los números? 🤔

Aquí va el truco más famoso de la estadística. Mira esta tabla:

| Grupo | Promedio de X | Promedio de Y |
|:---:|:---:|:---:|
| A | 9.0 | 7.5 |
| B | 9.0 | 7.5 |
| C | 9.0 | 7.5 |
| D | 9.0 | 7.5 |

Los 4 grupos tienen **exactamente los mismos promedios**. ¿Son iguales entonces?

**¡NO!** Cuando los graficamos, resultan ser completamente diferentes:
- El grupo A: línea recta (relación lineal perfecta)
- El grupo B: una curva (relación cuadrática)
- El grupo C: línea con un dato atípico extremo
- El grupo D: columna vertical (todos los X son iguales, excepto uno)

Esto se llama el **Cuarteto de Anscombe** y es la lección más importante del EDA:

> ⚠️ **Los números resumen pueden mentirte. Los gráficos no mienten.**

## 3. Las 4 grandes preguntas del EDA 🏛️

Cuando haces EDA siempre estás respondiendo 4 preguntas fundamentales:

| # | Pregunta | En lenguaje técnico | Ejemplo cotidiano |
|:---:|---|---|---|
| 1 | **¿Qué forma tienen los datos?** | Dimensiones, tipos de variables | ¿Cuántas filas y columnas tiene mi tabla de clientes? |
| 2 | **¿Qué tan buenos son los datos?** | Nulos, duplicados, errores | ¿Hay clientes sin teléfono registrado? |
| 3 | **¿Cómo se comporta cada variable?** | Distribuciones, promedios, medianas | ¿Cuál es el rango de edades de mis clientes? |
| 4 | **¿Cómo se relacionan las variables?** | Correlaciones, patrones cruzados | ¿Los clientes mayores compran más caro? |

> 💡 **Truco mental:** Piensa en el EDA como hacerle una revisión médica a tus datos. El médico mide tu peso, talla, presión y hace preguntas antes de diagnosticar — no te opera sin revisar nada.

## 4. El flujo de trabajo de un EDA real 🗺️

Todo EDA sigue más o menos este camino:

```
  📂 Cargar datos
       ↓
  🔎 Inspección inicial (¿cuántas filas? ¿qué columnas? ¿qué tipos?)
       ↓
  🧹 Auditoría de calidad (¿nulos? ¿duplicados? ¿valores raros?)
       ↓
  📊 Estadística descriptiva (promedios, medianas, rangos)
       ↓
  🎨 Visualización (histogramas, boxplots, scatter plots, heatmaps)
       ↓
  💡 Hipótesis y conclusiones (¿qué encontré? ¿qué debo investigar más?)
```
