## ¿Qué es el EDA y por qué es lo primero que hace un Data Scientist? 🔍

**EDA (Exploratory Data Analysis)** es la fase de **detective de datos**: antes de construir cualquier modelo,
exploramos, visualizamos y entendemos los datos para descubrir:

- 🧩 ¿Qué patrones interesantes hay?
- 🚨 ¿Qué problemas de calidad existen? (nulos, duplicados, outliers)
- 📊 ¿Qué distribución tienen las variables?
- 🔗 ¿Qué relaciones hay entre variables?

### 🎯 La Analogía del Médico

Un médico **nunca receta** sin antes hacer una historia clínica completa:
temperatura, presión, síntomas, análisis de sangre... Solo después diagnostica.

Un Data Scientist **nunca construye un modelo** sin antes hacer EDA completo:
estadísticas, visualizaciones, distribuciones... Solo después modela.

### Las 4 Preguntas Clave del EDA

1. **¿Qué hay aquí?** → Estructura: filas, columnas, tipos
2. **¿Está completo?** → Nulos, duplicados, datos faltantes
3. **¿Qué forma tiene?** → Distribuciones, histogramas, boxplots
4. **¿Qué se relaciona?** → Correlaciones, scatter plots, groupby


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Dataset clásico de Iris — perfecto para EDA introductorio
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['especie'] = pd.Categorical.from_codes(iris.target, iris.target_names)

print('🌸 Dataset Iris — Información Básica')
print('='*40)
print(f'Filas: {df.shape[0]}  |  Columnas: {df.shape[1]}')
print(f'Especies: {df["especie"].unique()}')
print(f'Nulos: {df.isnull().sum().sum()}')
print(f'Duplicados: {df.duplicated().sum()}')
display(df.head())
```

```python
# Un EDA express — el panorama completo en pocas líneas
print('📈 Estadísticas Descriptivas:')
display(df.describe().round(2))

print('\n🌺 Distribución por Especie:')
print(df['especie'].value_counts())

print('\n🔗 Correlaciones entre Variables:')
print(df.select_dtypes(include='number').corr().round(3))
```

```python
# Visualización rápida del EDA
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histograma de longitud de sépalo
axes[0].hist(df['sepal length (cm)'], bins=20, color='steelblue', edgecolor='white')
axes[0].set_title('Distribución: Longitud del Sépalo', fontweight='bold')
axes[0].set_xlabel('Longitud (cm)')
axes[0].set_ylabel('Frecuencia')

# Boxplot por especie
colores = {'setosa': 'tomato', 'versicolor': 'steelblue', 'virginica': 'mediumseagreen'}
for especie, color in colores.items():
    data = df[df['especie'] == especie]['sepal length (cm)']
    axes[1].boxplot(data, positions=[list(colores.keys()).index(especie)],
                    widths=0.4, patch_artist=True,
                    boxprops=dict(facecolor=color, alpha=0.7))

axes[1].set_xticks([0, 1, 2])
axes[1].set_xticklabels(list(colores.keys()))
axes[1].set_title('Longitud del Sépalo por Especie', fontweight='bold')
axes[1].set_ylabel('Longitud (cm)')

plt.tight_layout()
plt.show()
```

**Pregunta:** ¿Por qué es crucial hacer EDA ANTES de construir un modelo de Machine Learning?

<details>
<summary>💡 Ver respuesta</summary>

1. **Detectar problemas de calidad** (nulos, outliers) que rompen los modelos
2. **Entender las distribuciones** para elegir el modelo correcto
3. **Descubrir correlaciones** que revelan qué variables son predictoras
4. **Identificar sesgos** en los datos antes de que contaminen el modelo

Un modelo entrenado con datos sucios produce predicciones sucias: **Garbage In, Garbage Out**.
