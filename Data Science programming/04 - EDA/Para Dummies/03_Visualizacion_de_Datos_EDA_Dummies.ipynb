````python
import gzip, base64
_DATA_EMBEDDED = {
    "quartets.csv": "H4sIAMq5mGoC/0WSSU5EMQxE930Wy/rOZOcI/wK9Z8EBQI0Et6dcDmJX5fHJiXzLj3x8vX2+3l8PE7sk9BpyP5qELN0Tsot1cZ0BPWSjIgxyihl079BLDBndC9plobrlkJAhQ1tGt1jDeI2MY40jES21yZSpK6cTYCtm3UUQRxMh1GmKwT01ITZXHAq0pE6Kpei6C6NXmBj7xInhp5kc46wgiOvIzN8tuPCgWKvCgnE1oyGN8zz/OFGFBXQFTRJN7ZuGTMCedIRaOlq5uo73cgTJh3gSComVGrfBWrfUg2fLpU/wpB6eerHmYtzZ22bqEEsApdlMTA7Nv4COzalWq2On+QUB8v1SNQIAAA==",
    "StudentsPerformance.csv": "H4sIAMq5mGoC/61dyZIdN5K8z1fQdJnLmDX25Uiq+CE1ZDVJM4olK5ZmbP5+6iWIAMAEkLG8y2tKbKXlAgQiPNw9/vjy9OPz08sf//XHy+Onp389vX798e3Tt9f/e/sXfz++PP14ffz+7vvT/zx9f/f873dPn//59Pj67fnH299+/+fHp69v//v69PP13d8vT2//7+Ov3n16/ufl59Pb3/z1+Pr13c9Pzy+3f3h5evz87ccX+Of/ffn22v75P/7499Nfj99vf/Hl5fmfv999ePvTfz9++vr0/fnlP3+++/z05eXp9rc/Xx9/fH58+fz2xx/PP27/Jhr4cefr/Hn7b57/enq7q+/fn778dolPz3/9/f3p9en255DffrJ6+0lpfj9/Pf58fbq4m+MC2d9+7NtVhmu8f/vT48+fz5++Pb4+9Zf599v//Ovl9nKfuku5+Pbjbz/O/X6p/WPVNxNuP+n24+dPNL2byYX07a3YcrXphXCvOCV4Oeb3J5pcZfFabm/Y3W7G5t8v8vD2p6/fvtxW3dfn5+/nawwf3MFPnD/V9lK/bsfeninc7smr2VdCvmJ/u4y/3Yw3s6dCXuZ4Od6UN4R4pvMVwu3zpNsHj9MFjPvSZdUdm1Kd7+M9cjP542lseT3rrX35SLlsgV/L93SRq//+WLbpdpEUNssfs1r07VL29l7M9FbOb2WzgN1tfztT/jR9PYRoU9besZj19baavORQ33SwqBix25kePpia3ctl5HMOnsiKF46d7oWHxQG1eazoIJaq2da6/Pg1HFu4GCaMLp6rvB4/fz24ozfEbncxNlZU9SdM7uMBGSWCgUstroJ7IeWNOuoBVy9h68I99s9vl/hI+Lw+1NgZPG/1qxoX7CnIfMRcIN++azp+zOwCV4fRcOZrOE/y/OtgkxAHV7Obz4yJwse5cpz8nnJPu63th+QRl2gtbi7WNeAjOqFYXOrXcjw2O/6AWGQnR9gJSZRJ1sN8trZpZ4TPsN/CfL9RXpSCd+7Qn2+RO5W4GudZD2XnlK/nS4Sbfj3UmXwcODHcZ50fi7MEykCK+ENIuIWm5Jbh5TI+lVV0ezMuoePTbvfCORYSq9I6buhIOpzhJE9Hhrms1P5ExbbhgTS84YhZOovS5ngqvSgmyAfB8dV9zaK5x8BR3JRvn2mn47COUw0hhMpi/pZuoUPfFpBWs7VIzRHzOrd7T7ktm+HHCYGII8Eqx4gXHSPHFkuqLFDsET5ZArn7gqtD5DI2Rqj+9XIdYV60u61FG8sPEX+q96IGCEuSPZTPlMpxJvlWx9IJq7PsgXZWp5ogz+soJEAHUX9WYVLBteM96ekRQtr+5YXH+sJ5habL8KOFC6CcQnqf/F/dj70tSHPLHsw0Cl2Uh0PYV/VoS6w9P6wAPQRHVI21qBdDkiezB0B7fLUz9IeFD8tH9+XnGm7bbTIFmLHjFG09nF7ioiXDq/WhzHD2cELiUWwcaycZ1CUWUEmoyPUM1H8gQuBRLxCC96gkMQHaEQWnX3kzYfNmkOCNXx862AzRxPKN334CCwDSkPJobq07IKK27iavmDhOzPX1nnsUnFuqoW9W7HygRT8PMOsi+hHO0SU2/4DdWuVQj4vbeaCtw3Yd5hcrF4hYlHb1atIOBySEi6POybr0DS6PhtXdKDjPI60s7a+SEyxoxQzG7VNHw8dpUymS1K9fIa5YNoIthxX+FF+sZQPtUoXu8Czw7OA274mFTBytp8xqXYVNtfyeUHQ7ADkiE2L3uutyTqMXYZ/mvFtKqJ6eq293hfgjCvf62T0gG6JIWI4Ks4HKceWxhp2huRveAyDGaxCdII1slrGQsutrSP11OrPJHoWJYKc0DWyWW+rReeWHgjET5LiR2aw/QmqeY2r4qNXqoSiozv2woXgnhIENsEiycRup1PW5/Il7L3kNUtMB2SN2HiCRU4i1srlSagi8kuyACDyP5O7W2163gpEfLq2BfVqy7HU9ts6Iykdsa/q4EyGOVvpUQbi7jkLpgD+C5GiI0FqOWlBVl2cyi9yLtp4dZCh+ilNfnnYR6r7I6pYn6G3GJHgnpmEelh2KC9PILXcRIaWt3dp5f5RVF/kK4okbbi11W/EkCWWx6ZgodKyyvq4IV2GCgqfmdKkGAqMk7bPuAhBGZItihXZGVjY4RCAFh2Ka73k88TLAgZHYNNCjOK4/4g92hNYCqkV0c3JxY2EIjqyy9NQ3jWmRFNGCbdnCqZQrk3vDUpRKbmWnbFVM9VWauLaWpUwg/8hejsIia3GVkgOtpLxK9ryu6SeKSX7VPatcblqZcIrhiX/ODdeKa4Lme+wNlTzETisGdKcqu44+wUMmGuKbtKy9beHVKPx6vIKz/Y6/td0mJxpeqZ45ZPcT7bjShVnv65TZlrox3mOzABl0hUrjM4scr0IC6dZa33FL8rnUPwRo0AXMQbyi1asNmkfhu4U8DQP0Grn0bcwUq+R0XrrGgueUgAWZCXseLiFFDVN+GUk+4BK8IhFxu8DKaRENCN3iELplwK9uoUs7Az4/0FJdt8y6kTcT4VJGGHRDw5C8gPJWqVerbIsIReUB26Wl8T0+XPiymiP26KlTlcwpBeETAFBBfiodYaN8uSDokXZiqigQh5UQGxfoD4V+LSa6ZqCEGllMbGKYmLC8yQVi3DYrPsafJQpqsXTQq7kJFESXqRqsbRRD7/byWLmGVl7KYX0RrB2EDzad2LVvpe5yvivIjgP/20fIi7UMcS7Rwi7SOw5KlhrTJwlY3K5jZDFZagWxidNE7wOdwldOjkVgxcH7oWNr8AUbRePkFx04Eg/0N7UGD6wrH1sJFNunWqh8OiuTpDk/6GNYhD4ztB7IEonSaFXLPgGesJs6lqO4hewLBsFpPMSwTlcJyUFRAdt6JLN9FErKE6q+jgnHGGhRJkYb8Bz9zLJBid3iCqCqyOxzDvBOYPM3z94Dahm/rhMxBzCqFiyecmipqSSZZKawbzcREjoFn94Sz4dePltVQix1WQYKgWVH8rLi3LQw/pPII4xhGjUJZXrKcBkngJA7BnMSuEuUxc9qHITmtiFsGzbyvsuCCq9ZoyTLgS5OCW3VxHNpKwpiHkuJ2MUVMsl40OPbDQxO6aa5QXksg5rTFdR8fWarwUoEBQ5dQZRJb1oPtGovNBaOgDFcy8WLThGq5nOQtWkeWH1QPQtTT3EWdAzrBUQzDcjgtqJFugs7cCkYJ1yBg8xiHT8Qd0XJY01JZhn95UHKEZa9AOTLyQN3lUlVd3mRDRNb+UVNtfIfwjmLeIgampbonb5RzQXYL7c2f0nV7uqOPKwaS9zYZzDaioTvA9YapqksEUq2awcO1BlqoUvDOoOdr60eR6Bu7byCcq2iRDLeCFLOrQ3cpbg0QtUapHzwKqy6h1NNjnAtyeMFkGzMBV/4b5eAdRcDvSI/9bRikPYhangnnKULjKulh4IYVM4JtZCo42vO4OGOvKwIhgRzHoIoeW9cg+XEQicMLh6iLRJBgB2SiCJVLKrCVDpNPKSL61Kcfj2agjrpQaouSBXDYOZ4v16klmDdqsMc5bQdB1ks0SZyyPTmtF28lvWoDgqKr9nuC6WHZKrRqlBKaIGuHzHp/Y5gGRcriGLdaoFfgc6KVjQPtcyIKFTyCO3NxEcHTsTGsOqOE1FnD7YnTEtD3bFi6FLZs/YT6K1sCLHJvZNgBbRlncV1lW+sIctobJ6ZqKv8FqX0gbohZm5vS3csAiFgYqGAsIJWu+4aQ7xnKgtPTdtbOFOjJlky9HO5qmoO0sHtdegshf8Kkd5fwX+X1VljY2l+hFBLiINiPlNsp7xYS28gabGCDCNvrX6IOEWB/LyASFMPGQKRZsdPLHWPpJnjxsaSxEfauGo9biLXDLV/woPhUygMimsG04hLnutXqQGrtRKlw9jtEqGj1fWCLt/s1QXRLvJBVMAwAymEYU1sOp8ysZo/VtVF9lzlfNccDZKu5mhGymY+eLMmMeM6XmGXrRNGUABAm5RIiVIq67T54rgueKn3lYhWoge3Ia4DSe0JoGWMV5Lj6ld/vacvNdCa2Rs49nRW1RiD0U4vIdcwnTBPeH5yeHXYStSXrrolRF8xw+cpl4I1YUXBuxUDjqNBsW2PQ7Mttdgmzi6zCYu1957coA2O9tFWpnRzpQPOC6X58cj3t18jzPgncjuvSbycpNRRsShMBeZATVIeLVeCf2YMSmx4u/CFdgXboa++8imL2A63qrfDiiLfDyCEgajEB+AKIWdOqyRXsE2QmD1LjzhIXvVit30k+UynzsVUFovKpezGb/w6E2uNLi9qCbk7cph8rGkdDhLeKtvY5kr1aWZlxEdywRhB6+wzV3cdgY0dWWdGDGvjR+rW8hFaNkHKjQhQ3swjEFrpWYgAmj+czCt4Q1ZUUJT0ME8DzwfSkVpd09h+Jgoouk76oTzMQQqR1cTydhex0Cing1km3gtEMYXJlVcnKMqYw3WljWBwll2T7sjWLH7XLKCcxA1H4XTmTsqahfXfA+dQb6mzqGOkBpNqZk+1E7IlgfCi64RJNWN2iPNyXw3fMInMcgZopOQsVLlcMSoow7EaysjgKaZh4/PNnbrxVe4+/d1kBq9OtmQwQg3v7lCuuNgZIzOqldM5VI5YxZ+Lqomm+7thL3Gp0EC2cEKnXxWN1wAYMniZ7M/DgW9ldwTDTLwTevRV8d88icZ3TMySTIf678tulVZvLsILVmwqTmwau0CXq5/HYczRNsop5uGTOyFRLfpBjsubltlMNwT9zGrjhx6usWWm7WY7Xj5T2jEnSOvP5+FQFlKmkp3Sbkl3FDOwDsR8+xDXNSXejiSvUzL89L4wfDJyz8LFbnSxgM4IveJoxHZ2Ds5fzwSefqFyv8BeseptY6aNP8l9p4cRY1h2zeahNS8sJIgr9zGkh1WGSJaFkdlbcFtNEtSwt7WdOMahCQyxNaST7GW3eaveszNLD9O0HXd71Eby2p0JJzKLQErTwrlk0Xf9UyZFznUGMlJO5DFtzc7ntr2nFGDRDLZ3kyMVf6m4totFvaGsdwwNcvvDQpuICW8F2w37lhPzYbLOVluFGme5dGgmT8hQwxaR+X43UefcE4RyqdAZIQrtPKqFC3sUkQcbgqgks/LCwPVnBvvGrE5yZrWBIizwE4cEP5plyZ9BTxFkAxbhQt6xK6ZqNMdOmjUUt0Ge2KmO/iYDRWEkk4/c+b4n0X9TUsqmqwY/6JUFlspu2KisUxrKuaxErGQNLRbFZGvBw8zlk3gb76X9L44c5TorWelo3wy6xyARzYa1MSW+Fxy18FxoU2ySucNR3PJDww/GUHit5qaRbsnv5mc9ECgkbtuHxU88MV1njnHWhLSetEFTW3s/PBDN9Xcge6dFi5EIn1cvNba3XNRTrBHNpApg3oI2wtr0FOphLjRJ8m4AIqR+8aWijFXmJMLoYugsJZmcGNeNHpPMeA1dU1DqVReubB+Iar00bbw+oHGE5UoiJBZZD7Ubd3ZtHnj2bD5eo114PjDb2Y4wZQyp49fIfcYDbFQlcK8NHc1G6IVimnl1lGWl5moGHxlGKDvDCjg3LnaDWu8hCAx6wQonu0ZAuyk7WZncMkQ5CSSEwblPxh7Ug9yA41qph+HeEqQltT6j5ihuTp44JREJ7MAU/Tq5f5CVqjKZVGqzZ7Js3Lfpcmt29RFc5zIrZ+imbsSskKByrIBfLa7IXwcNmXb8EyGMVxHk6WZJ4KK1eBczYj7QY1KC/kZKMuSssR8UG/nqJqE4AUpaTm9fZ99yez92HLApPHSPam9BAaRkbkHDagx38o5RQuGSHU4StnCpOfywfTGC7YRUMp7kMZ4vqw1wQWj327XpNQ0zKDtMs63sBrK/3TRYiXGki9dTKjiZ7Rj0eloLeoiEhUeVOrGn9cRyvKlmeZz1EENcCXgUkjmwJtIP3Aqz6TDgiy7Vadd4djhukIaJ+wquDdImGoWdx4boDYJ0CTS2AYiZYyoyPFMCJl4SNCHtzgkLL8PUO4EpenJDbEZKSlpoN/lHTHKD4tgMkxXXhTfvnL8plX/z4w8YcsWi9RztptonHD+tYDCSXlApPLxAdOvSYHvGHCufOiEeF7SOGoKp4s8V604YSRyNgfidEfXXHTzuQlcCMQ9zq6oLkfbsb7Xn3eGt0CMb5x761WbjyY4K6DDUBWdrd77CgVAeP1ZxmXZh8JBh+fNUp8cV05ym64eWRhboAqBcC+YOQla3ob7TGqBp05hFWuTdwWTAgyZlM6oWVxvlqzWMs+D3eanhoMQ9tyOzkc/tuCD+vacifSsxGoqDuGWw4Y+oYqPup4RRfAerjk6foeDYuJfbFBN3t8HNTT3M7Rz5gTghqPILBGaXpDp8dZaGKoZZC/3KA8JiOO3HP8hWQo0GzcSKABr+lZ+wQLCzhR+EIdHk5DZJgQeDNi9mJ7aGKPNw9JSPj6tk4pq5iimDY4vvnmXrNmcZrAw/cea7aTkhG1cBu26SHYND1Gmw3T0IDgVa1FLeRQJFUeJx4lLYTX2jiCBHNgl9oIkehhUyKc4ncHpLxsS3cMxOxIieZNNRB3kbU4HmR8k35nauGOHLN9ex4DEvZ6dzNoK5gV7tJsXjD6qQO/crOcVFD77JFDTorCNZiaMIVnPgZJ4TL8k9Z2FqMzMS345qA8g5Q/e6jm8UKu5jXDeP8WhB3tkvUmeM2bVTJWGjtikoiQ/ARTegKWIJvxUM1iw4qd5MdcDNGs1LTxSS4sheVcTEFo3a8OKIncwqIEIfGputUUJIZIn9InR7gxabIundiEM0VQQSID7xaBBWxQors7rhW6nEn5zwXDNfvitOB6Amang+r2q3mf+MbMraq5dE4GMr8F8Iws+f9DD3lakksztGLAruBjf66DhJZsOeAitJDQaaue4esbAwclWdvMRce6H5eS/YqmgHtQI45rl+/yMtE4v+HoEnxmFoJ8tMsrkAOIwv3GLOmoU0Q+Yl6fJuxA3Ry7jNt4p3pAUvMQ1K0Rx3rSk8rTsP4CN7Tnlzy7jD4RN9NyFoWmlQTK3NgkhPTRr8Rl1CShmWllxUzXvuRpTySygfu9Avc73ozCYjl0th4aSPXOhlPgI4OKZV/yDLMMsu2OVWCzB1IPBGXoVO/c64gQTTDGO+D07mRiEH0+crg7t/VhyK0QnVXHotoVZgTiMbgtW0SnpoWok0iUFdQSVkfDN0+mGhOKVLqe8wjbRQHcOGiXIJEIT1eKaPjHeVwQksKanx5cJKkT5Qq1MT53tEyegGZQDd2BkEAXPXtctNV2zbYGgENzHWgyOHUFgW/DDRTxCVQuvMCqa2HouwdGx4MxOPrLhoHOJ9AknYjKSm2St6/kjJcwaSp2JAvGWShxWJ495cuqNFrpSozECIS9EsFr81uzmmNHqyt2ujUQrwqdceAuh0ugkjvYCkonZej0RvGhsH73bGAVazhjWHjFS15OHhmBZ9vtsVArdbWIXzyYk4OqafigfwGGmzyBGxoxvQvhmOhoxXbqqAJU4R1UKPm0oaW5GsacYXDfj1YqC1yRa9sEFXvSqJ4NCCFHIw07JhMW08GER6tCJ7974vjesIAR5c3oIXwKaXogg0thjNxlYfLyBobFrJNmuW0oFHNgimI/2xSFrQMvTMwY518Ny6ZsNZgqR1gox3XmjOjUrCsU+d5o47D9ynjp7OaGAUe5Q09UghWA/7rt15DwjMDACf6NQrZewcBP2ABrG2RCT8SIA8uA9ztmLYAcR47biv3EvknKadg0Hzh5NZsTXTcSNnkQRgIkWsK9aiZVHOcSkRsnhh+XvlyGY9nYYqKdA7uh8N97MG5N9OOEaxy7wjkyalc52hOB+aS/HvTBADtEwl083W1rwQ4BLq+LmE90AZHfm6KOgNbgaFEyZwuM3nvnwtZsep5eCovs1jZLF8oxrcK4R9XQVll+YVS20CtBO1q33jPEQ6Ke7sfmeXtv5Up0G16C1TqqQIpAfPDDtFNukXLUUCcyxM2TXo1DD6Dl8WxAnXeZ+wGc/FZQ48b8gQU1adbxZSvLQy4nZTHSdepKthMEFgTzNteG3Gnidb/8/5zE5iDAZkJzKBwAgTJc/iOVomkR3YUUYJocJB2yk4psfK2fZFbdyd0XLiaDagHt6hIuuFORbN0KxOkVjbCRGIswmuZensqqFbn3dnOvZ1lwHo8Wq6F3oK8QID+xONpMSFUQpHG+pDZ/osGFAL3PCt5AKv+8rdZCzufOSQBgdP7jThrudyn3GpsQWnzDdKCm7nkYlHoyLkHDx8wydoh3sBtBZUlywwZshGoFH5KNAONSZevIvwtvNgkbE8S620c+ZAibjHUSLMze467bUM1jC2Akk20Jt1Z71gvFsEck0/OF2PKCG/6hwyqP3wM8jil7Je3BdzG7UFagZNyzlE4p9FHo42G9W78dWoIQe6mzvLsxpJoO7ZJCmoYQ1tUhVzNxr4wDwFel2scfA94/XNu/GuXjKwIfi1rhR9tHkYkzAPdhS9Sryajkkxby2yD4vVVmwkNK3EkZYmx8lwgGA2SfhIEVguSUn4OxEkEprvo9FNQ0lcyo3P3SwAvidpsc4wS74fmdGaurambHGXMklPYY0P2LPBp80cAOTSgbo2stxfz1/eTSEwInTaJgAmcTfR2W4aAFdWnEZtIL2fXiUUW14lTkWxGqREm9YYzcaNkjLabEGDJva0OklGkItW9A7GR2M+blAmC+CVjnjqRFi3bx6y9xj+aTocXj6XyzZJg2dHfiCN+CxBwDMkNI7bKWs+R6uJysgDEdrIWcsaHR6o2QufU3RHUw3jWiQGP+mq1UBxaQlm4ZVA6kMXl303dZVDe4MGsLK8i0Is2P3UKNK2hSk5QT4YywPdzLBlSL6zyKR3RoYGWtrYG6JftoXhkE6zkfFFaFt60eMnycXB9YCRSqSB9EjPHWNn+y+ZrgAEgJWzJYHyFTuHVOGUZrW2lyMe2wrqh0SlXPUH0ZZnTzE9BiXkam4qcV27OAy0lk3+cDAszXuumexmEiZxFbTdutizGATOwaBGfw8GWFzrmPCU525Yg2aTgJolWhS6k5bJOIFFtjGdWF08d8YBduy5r6WOw5Nzo6LrukWyUrjAjGEz6RifhPhuDMod8JAqpZuXNYwubZjiIUQ/qTgMNmTWD2CVNLeUJUsYL3RoHCHGYgQ7WQyduqFd7DFbTS2QOPP6zkJEyycu+biWRFKdCDXT1vXU7i0AfZZrnkxHKecPCU/dScIgRSQ9zOhhSHJPu6MhdBxvzTbrQmay3kYz+8TJrX+bBc7FL5rqICgkF+OSk6MWR8hHyhHiY6cdYVAGstu5/RFqDzDDT1ZKW3d7VB+vUiq0J8f11jMAXGXBqMK4ObpQ91GtdWSDRDDj2y7Tbz9o0JjO3Hqx2FD1W52QRZIG7lIcJ5eFokobYgRuM6F4nu5t1fnENWUyXT+KzPQ9+Z+1yUfMoY/BLvQahE8VEogPo4QoEP1gD49EtBbuTP3KkRD40zCAVgLYlVavkupjXN4JdXD7vY0jE/VVgwNOW5SRrApJz0xTAeq89+CXHs24U86DLiCJ0qySevrqoMJs0KU6v8tY/uAFDUs5yfjvBQ71C9E2QaDZCjItoZq7jlrJmChRZ2eJxUdOAV9BC9UOzcE2oNnzm+ZekXR48SzeCF0+RE68GzMWuoeU+rk1Um0SS3P9dmQkcbCmWfv/ooiEdjcwkuSUW4ZvZFHDurMT9tLBtxmYlrIBETAPK0aJ3Wpb1m8f/f8Bny5zxmQZAQA=",
}

import os, urllib.parse, urllib.request
import pandas as pd
import matplotlib.pyplot as plt

def load_dataset(filename, module_name="04 - EDA"):
    candidates = [f"data/{filename}", f"../{module_name}/data/{filename}", f"{module_name}/data/{filename}", filename]
    for path in candidates:
        if os.path.exists(path):
            return path
    os.makedirs("data", exist_ok=True)
    target_path = f"data/{filename}"
    url = f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{urllib.parse.quote(f"Data Science programming/{module_name}")}/data/{urllib.parse.quote(filename)}"
    urllib.request.urlretrieve(url, target_path)
    return target_path

quartets = pd.read_csv(load_dataset('quartets.csv'), index_col=0)

# Primero mostramos que las estadísticas son iguales
print("📊 Estadísticas del Cuarteto de Anscombe:")
print(quartets.groupby('quartet')[['x', 'y']].agg(['mean', 'std']).round(3))
print("\n🤯 ¿Todos iguales? ¡Sí! Pero ahora míralos graficados...")
````

````python
# Ahora los graficamos — ¡sorpresa!
grupos = quartets['quartet'].unique()
colores = ['#f59e0b', '#10b981', '#6366f1', '#ef4444']
descripciones = [
    'Relación LINEAL perfecta',
    'Relación CURVA (cuadrática)',
    'Línea con un OUTLIER extremo',
    'Columna vertical + un punto lejano'
]

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, (grupo, color, desc) in enumerate(zip(grupos, colores, descripciones)):
    subset = quartets[quartets['quartet'] == grupo]
    axes[i].scatter(subset['x'], subset['y'], color=color, s=80, alpha=0.9, edgecolors='white', linewidth=1.5)
    axes[i].set_title(f'Grupo {grupo}: {desc}', fontweight='bold', fontsize=12)
    axes[i].set_xlabel('X')
    axes[i].set_ylabel('Y')
    axes[i].set_xlim(0, 20)
    axes[i].set_ylim(0, 14)
    # Línea de tendencia
    m, b = np.polyfit(subset['x'], subset['y'], 1)
    x_line = range(0, 21)
    axes[i].plot(x_line, [m*x+b for x in x_line], 'gray', linestyle='--', alpha=0.6, label='Tendencia')
    axes[i].text(0.05, 0.95, f'Media X={subset["x"].mean():.1f}\nMedia Y={subset["y"].mean():.1f}',
                transform=axes[i].transAxes, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.4), fontsize=9)

plt.suptitle('🔍 Cuarteto de Anscombe — Mismas estadísticas, formas completamente diferentes',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

import numpy as np  # Aseguramos que numpy está importado arriba
````

## 2. El Histograma — ¿Cómo se reparten los datos? 📊

Un histograma divide el rango de valores en "cajitas" (bins) y cuenta cuántos datos caen en cada cajita. Es como organizar la asistencia a un concierto por rangos de edad:

- Cajita 1: 15-20 años → 500 personas
- Cajita 2: 21-25 años → 1200 personas
- Cajita 3: 26-30 años → 800 personas
- ...

De un vistazo, ves dónde está concentrado el público.

````python
df = pd.read_csv(load_dataset('StudentsPerformance.csv'))

fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df['math score'], bins=20, color='#f59e0b', alpha=0.85, edgecolor='white', linewidth=1.5)
ax.axvline(df['math score'].mean(), color='#ef4444', linestyle='--', linewidth=2.5, label=f'Media: {df["math score"].mean():.1f}')
ax.axvline(df['math score'].median(), color='#6366f1', linestyle='-', linewidth=2.5, label=f'Mediana: {df["math score"].median():.1f}')
ax.set_title('Histograma de notas de Matemáticas', fontsize=14, fontweight='bold')
ax.set_xlabel('Puntaje')
ax.set_ylabel('Cantidad de estudiantes')
ax.legend()
plt.tight_layout()
plt.show()

print("\n💡 Cuando la barra más alta está hacia el centro → distribución aproximadamente normal (campana).")
print("   Cuando está desplazada → hay asimetría (skewness).")
````

## 3. El Boxplot (caja y bigotes) — El resumen visual de los cuartiles 📦

El boxplot es como el `.describe()` pero en forma de imagen:

```
         ┌──────────────────┐
   ──────┤       CAJA       ├──────   ← Bigotes (hasta 1.5×IQR)
         └──────────────────┘
         ↑      ↑           ↑
         Q1   Mediana       Q3
                                  ○  ← Outlier (punto individual)
```

````python
fig, ax = plt.subplots(figsize=(10, 5))

datos = [df['math score'], df['reading score'], df['writing score']]
etiquetas = ['Matemáticas', 'Lectura', 'Escritura']
bp = ax.boxplot(datos, labels=etiquetas, patch_artist=True,
                boxprops=dict(facecolor='#fef3c7', color='#b45309'),
                medianprops=dict(color='#ef4444', linewidth=2.5),
                whiskerprops=dict(color='#b45309', linewidth=1.5),
                capprops=dict(color='#b45309', linewidth=2),
                flierprops=dict(marker='o', markerfacecolor='#f59e0b', markersize=5, alpha=0.5))

ax.set_title('Boxplot comparativo de los tres puntajes', fontsize=14, fontweight='bold')
ax.set_ylabel('Puntaje')
ax.set_ylim(0, 110)
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("\n💡 Los puntos fuera de los bigotes son los outliers — valores inusualmente altos o bajos.")
print("   La línea roja dentro de la caja es la MEDIANA (no el promedio).")
````

## 4. El Scatter Plot — ¿Se relacionan dos variables? 🔗

El gráfico de dispersión muestra cada dato como un punto en un plano. Si los puntos forman una línea diagonal, hay una **correlación**:
- **Diagonal hacia arriba (↗):** correlación positiva — cuando sube X también sube Y.
- **Diagonal hacia abajo (↘):** correlación negativa — cuando sube X baja Y.
- **Nube dispersa sin forma:** no hay correlación.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Correlación fuerte
axes[0].scatter(df['reading score'], df['writing score'], alpha=0.4, color='#6366f1', s=30)
axes[0].set_title(f'Lectura vs Escritura\nCorrelación: {df["reading score"].corr(df["writing score"]):.2f} (fuerte)', fontweight='bold')
axes[0].set_xlabel('Puntaje Lectura')
axes[0].set_ylabel('Puntaje Escritura')

# Correlación moderada
axes[1].scatter(df['math score'], df['reading score'], alpha=0.4, color='#10b981', s=30)
axes[1].set_title(f'Matemáticas vs Lectura\nCorrelación: {df["math score"].corr(df["reading score"]):.2f} (moderada)', fontweight='bold')
axes[1].set_xlabel('Puntaje Matemáticas')
axes[1].set_ylabel('Puntaje Lectura')

plt.suptitle('¿Qué estudiantes buenos en lectura también son buenos escribiendo?', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n💡 Correlación cercana a 1.0 → fuerte relación positiva")
print("   Correlación cercana a 0.0 → sin relación")
print("   Correlación cercana a -1.0 → fuerte relación negativa (cuando sube uno, baja el otro)")
````

## 5. Resumen: ¿Qué gráfico usar? 🗺️

| Pregunta | Gráfico |
|---|---|
| ¿Cómo se distribuye una variable numérica? | **Histograma** |
| ¿Hay outliers? ¿Dónde está la mediana? | **Boxplot** |
| ¿Hay relación entre dos variables numéricas? | **Scatter plot** |
| ¿Cuántos hay de cada categoría? | **Gráfico de barras** |
