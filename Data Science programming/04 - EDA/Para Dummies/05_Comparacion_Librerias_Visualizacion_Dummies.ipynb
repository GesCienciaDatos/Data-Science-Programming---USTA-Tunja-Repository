## 1. El mismo gráfico de barras en las tres librerías 📊

Vamos a graficar la **cantidad de estudiantes por nivel educativo de los padres** usando las tres herramientas. Observa la diferencia en cantidad de código y en estética.

````python
conteo = df['parental level of education'].value_counts()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# --- 1. Pandas Plotting (el más rápido, el más básico) ---
conteo.plot.bar(ax=axes[0], color='steelblue', rot=30)
axes[0].set_title('🐼 Pandas Plotting\n(1 línea de código, estética básica)', fontweight='bold')
axes[0].set_xlabel('')

# --- 2. Matplotlib (control total, más código) ---
colores = plt.cm.Blues([0.4, 0.5, 0.6, 0.7, 0.8, 0.9])
axes[1].barh(conteo.index, conteo.values, color=colores)
axes[1].set_title('🎨 Matplotlib\n(más código, control total)', fontweight='bold')
axes[1].set_xlabel('Cantidad de estudiantes')
for i, v in enumerate(conteo.values):
    axes[1].text(v + 5, i, str(v), va='center', fontsize=9)

# --- 3. Seaborn (equilibrio entre facilidad y estética) ---
sns.countplot(data=df, y='parental level of education', ax=axes[2],
              order=conteo.index, palette='YlOrBr')
axes[2].set_title('🌊 Seaborn\n(código mínimo, resultado premium)', fontweight='bold')
axes[2].set_xlabel('Cantidad de estudiantes')
axes[2].set_ylabel('')

plt.suptitle('El mismo gráfico de barras en las 3 librerías', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 2. ¿Por qué el gráfico de torta es (casi siempre) una mala idea? 🍕

El gráfico de torta (pie chart) tiene un problema fundamental: **el ojo humano es muy malo comparando ángulos**. Es mucho más fácil comparar alturas de barras que tamaños de porciones.

Solo usa torta cuando tienes **2 o 3 categorías máximo** y la diferencia es obvia.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Torta — difícil de comparar
conteo.plot.pie(ax=axes[0], autopct='%1.1f%%', startangle=90,
                colors=plt.cm.Wistia.colors[:len(conteo)])
axes[0].set_title('❌ Gráfico de torta\n(¿Cuánto mayor es "some college" que "associate\'s degree"?)', fontweight='bold')
axes[0].set_ylabel('')

# Barras — fácil de comparar
conteo.sort_values().plot.barh(ax=axes[1], color='#f59e0b')
axes[1].set_title('✅ Gráfico de barras horizontales\n(La diferencia es inmediatamente visible)', fontweight='bold')
axes[1].set_xlabel('Cantidad de estudiantes')

plt.tight_layout()
plt.show()
print("\n💡 Regla de oro: Barras > Torta en casi todos los casos con más de 2 categorías.")
````

## 3. Línea vs Barras vs Dispersión — ¿Cuándo usar cada uno? 📈

| Tipo de variable | Pregunta | Gráfico ideal |
|---|---|---|
| Una variable numérica | ¿Cómo se distribuye? | Histograma o KDE |
| Una variable categórica | ¿Cuántos de cada tipo? | Barras (`countplot`) |
| Numérica vs Numérica | ¿Hay relación? | Dispersión (`scatterplot`) |
| Numérica a lo largo del tiempo | ¿Hay tendencia? | Línea (`lineplot`) |
| Numérica por categoría | ¿Cómo varía entre grupos? | Boxplot o Violinplot |
| Dos o más numéricas + categoría | ¿Cómo interactúan? | `pairplot` con `hue` |

````python
# Demostración: el puntaje promedio de matemáticas varía según el nivel educativo de los padres
promedios = df.groupby('parental level of education')['math score'].mean().sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Línea — NO tiene sentido aquí (no son datos temporales)
axes[0].plot(range(len(promedios)), promedios.values, marker='o', color='#ef4444', linewidth=2)
axes[0].set_xticks(range(len(promedios)))
axes[0].set_xticklabels(promedios.index, rotation=30, ha='right', fontsize=8)
axes[0].set_title('❌ Línea aplicada a categorías\n(¿Tiene sentido conectar con una línea?)', fontweight='bold')
axes[0].set_ylabel('Promedio nota matemáticas')

# Barras — SÍ tiene sentido aquí
sns.barplot(x=promedios.values, y=promedios.index, ax=axes[1], palette='YlOrBr')
axes[1].set_title('✅ Barras para comparar categorías\n(Inmediatamente claro)', fontweight='bold')
axes[1].set_xlabel('Promedio nota matemáticas')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
print("\n💡 El gráfico de líneas es para secuencias ordenadas (tiempo, iteraciones, etc.)")
print("   Las categorías no tienen orden natural → usa barras.")
````
