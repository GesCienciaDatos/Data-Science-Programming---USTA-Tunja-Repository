## Comparación Práctica: La Misma Gráfica en 3 Librerías 🏆

La mejor manera de entender las diferencias es ver el **mismo gráfico en las 3 librerías**.
Haremos un gráfico de dispersión (scatter plot) con colores por categoría:


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Dataset de Pingüinos (alternativa a Iris, más moderno)
try:
    import seaborn as sns
    df = sns.load_dataset('penguins').dropna()
    print(f'Dataset Pingüinos: {df.shape}')
    display(df.head(3))
except:
    # Fallback: datos simulados
    np.random.seed(42)
    df = pd.DataFrame({
        'bill_length_mm': np.random.normal(45, 5, 150),
        'bill_depth_mm': np.random.normal(17, 3, 150),
        'flipper_length_mm': np.random.normal(200, 15, 150),
        'body_mass_g': np.random.normal(4000, 800, 150),
        'species': np.random.choice(['Adelie', 'Chinstrap', 'Gentoo'], 150),
    })
    print('Dataset simulado de pingüinos creado.')
```

```python
# ── VERSIÓN 1: Matplotlib (máximo control, más código) ───────────────
fig, ax = plt.subplots(figsize=(8, 5))

colores = {'Adelie': '#e63946', 'Chinstrap': '#457b9d', 'Gentoo': '#2a9d8f'}
for especie, color in colores.items():
    mask = df['species'] == especie
    ax.scatter(df[mask]['bill_length_mm'], df[mask]['flipper_length_mm'],
               c=color, label=especie, alpha=0.6, s=60, edgecolors='white', linewidths=0.5)

ax.set_xlabel('Largo del Pico (mm)', fontsize=11)
ax.set_ylabel('Largo de Aleta (mm)', fontsize=11)
ax.set_title('Matplotlib: Pico vs Aleta en Pingüinos', fontweight='bold', fontsize=13)
ax.legend(title='Especie', framealpha=0.8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print(f'Líneas de código: ~15 | Interactivo: ❌ | Personalizable: ⭐⭐⭐⭐⭐')
```

```python
# ── VERSIÓN 2: Seaborn (conciso, estadístico) ───────────────────────
try:
    import seaborn as sns
    sns.set_theme(style='whitegrid')

    ax = sns.scatterplot(data=df,
                          x='bill_length_mm', y='flipper_length_mm',
                          hue='species', style='species',
                          palette='Set2', s=70, alpha=0.8)
    ax.set_title('Seaborn: Pico vs Aleta en Pingüinos', fontweight='bold')
    ax.set_xlabel('Largo del Pico (mm)')
    ax.set_ylabel('Largo de Aleta (mm)')
    plt.tight_layout()
    plt.show()
    print(f'Líneas de código: ~8 | Interactivo: ❌ | Personalizable: ⭐⭐⭐⭐')
except ImportError:
    print('Instala seaborn: pip install seaborn')
```

```python
# ── VERSIÓN 3: Plotly (interactivo, web-ready) ──────────────────────
try:
    import plotly.express as px
    fig = px.scatter(df,
                     x='bill_length_mm', y='flipper_length_mm',
                     color='species', symbol='species',
                     title='Plotly: Pico vs Aleta en Pingüinos (¡Interactivo!)',
                     labels={'bill_length_mm': 'Largo del Pico (mm)',
                             'flipper_length_mm': 'Largo de Aleta (mm)'},
                     template='plotly_white',
                     hover_data=['body_mass_g'])
    fig.show()
    print(f'Líneas de código: ~7 | Interactivo: ✅ | Personalizable: ⭐⭐⭐⭐')
except ImportError:
    print('Instala plotly: pip install plotly')
```
