## La Guía de Decisión para Visualizar 🎨

- **Usa Pandas Plot (`df['col'].plot()`)** cuando estés en medio de la exploración rápida y solo quieras ver algo en 3 segundos.
- **Usa Seaborn (`sns.grafico()`)** para el 90% de tus análisis estadísticos y tableros profesionales.
- **Usa Matplotlib (`plt.plot()`)** cuando necesites un ajuste milimétrico especial (cambiar una fuente específica, agregar una flecha con texto o personalizar una leyenda).


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ventas = pd.Series([100, 150, 120, 210, 190], index=["Ene", "Feb", "Mar", "Abr", "May"])

# Enfoque rápido con Pandas:
ventas.plot(kind="bar", color="#38bdf8", figsize=(8, 4), title="Ventas Mensuales Rápidas")
plt.ylabel("Millones COP")
plt.show()
```
