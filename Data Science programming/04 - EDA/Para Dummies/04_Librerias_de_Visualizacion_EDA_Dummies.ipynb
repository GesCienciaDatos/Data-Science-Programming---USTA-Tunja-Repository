## ¿Qué vamos a aprender aquí? 🎈

Seaborn es como tener un diseñador gráfico profesional que sabe estadística. Con pocas líneas de código produces gráficos que en Matplotlib te tomarían horas.

Aprenderás a usar:
1. **`histplot`** — histograma mejorado con curva de densidad.
2. **`boxplot` y `violinplot`** — para comparar grupos.
3. **`countplot`** — para contar categorías.
4. **`heatmap`** — el famoso mapa de calor de correlaciones.
5. **`pairplot`** — el gráfico más poderoso para explorar todo de golpe.

```python
import os, urllib.parse, urllib.request
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def load_dataset(filename, module_name="04 - EDA"):
    candidates = [f"data/{filename}", f"../{module_name}/data/{filename}", f"{module_name}/data/{filename}", filename]
    for path in candidates:
        if os.path.exists(path):
            return path
    os.makedirs("data", exist_ok=True)
    target_path = f"data/{filename}"
    url = f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{urllib.parse.quote(module_name)}/data/{urllib.parse.quote(filename)}"
    urllib.request.urlretrieve(url, target_path)
    return target_path

df = pd.read_csv(load_dataset('StudentsPerformance.csv'))
tips = sns.load_dataset('tips')   # Dataset integrado de propinas en un restaurante
iris = sns.load_dataset('iris')   # Dataset integrado de flores

sns.set_theme(style='whitegrid', palette='muted')
print("✅ Datos y librerías listos.")
```

---
## 1. `histplot` — El histograma con curva de densidad 📊

La curva de densidad (KDE) es una versión suavizada del histograma — en lugar de barras cuadradas, dibuja una curva fluida que muestra dónde se concentran los datos. Es como pasar del pixel al vector.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Solo histograma
sns.histplot(df['math score'], bins=20, ax=axes[0], color='#f59e0b')
axes[0].set_title('Histograma simple', fontweight='bold')

# Histograma + curva KDE + separación por género
sns.histplot(data=df, x='math score', hue='gender', bins=20, kde=True, ax=axes[1], palette=['#6366f1', '#f59e0b'])
axes[1].set_title('Histograma + KDE + separación por género', fontweight='bold')

plt.suptitle('histplot: del histograma simple al comparativo', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
print("\n💡 El parámetro 'hue' separa los datos por categoría — aquí por género.")
print("   'kde=True' agrega la curva de densidad suavizada encima del histograma.")
```

---
## 2. `boxplot` vs `violinplot` — ¿Cuándo usar cada uno? 🎻

- **Boxplot:** muestra mediana, cuartiles y outliers. Conciso y directo.
- **Violinplot:** muestra todo lo que muestra el boxplot PLUS la forma completa de la distribución (si hay una sola "joroba" o varias). Más rico en información, más difícil de leer para principiantes.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x='lunch', y='math score', hue='gender',
            ax=axes[0], palette=['#6366f1', '#f59e0b'])
axes[0].set_title('Boxplot: Nota de Matemáticas\npor tipo de almuerzo y género', fontweight='bold')

sns.violinplot(data=df, x='lunch', y='math score', hue='gender',
               ax=axes[1], palette=['#6366f1', '#f59e0b'], split=True)
axes[1].set_title('Violinplot: misma información\npero con la forma de distribución', fontweight='bold')

plt.tight_layout()
plt.show()
print("\n💡 ¿Qué observas? Los estudiantes con 'standard lunch' tienden a tener notas más altas.")
print("   El violinplot muestra que la distribución es bastante simétrica en ambos grupos.")
```

---
## 3. `countplot` — Contar cuántos hay de cada categoría 🧮

El `countplot` es el equivalente a `value_counts()` pero en forma de gráfico. Perfecto para variables categóricas.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df, x='race/ethnicity', hue='gender', ax=axes[0],
              palette=['#6366f1', '#f59e0b'], order=df['race/ethnicity'].value_counts().index)
axes[0].set_title('Cantidad por grupo étnico y género', fontweight='bold')
axes[0].set_xlabel('Grupo étnico')

sns.countplot(data=df, x='parental level of education', ax=axes[1], color='#10b981',
              order=df['parental level of education'].value_counts().index)
axes[1].set_title('Nivel educativo de los padres', fontweight='bold')
axes[1].set_xlabel('Nivel educativo')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

---
## 4. `heatmap` — El mapa de calor de correlaciones 🌡️

El mapa de calor es una tabla donde cada celda se colorea según el valor que contiene. En el contexto del EDA, lo usamos para mostrar la **matriz de correlación**: qué tan relacionadas están las variables numéricas entre sí.

**Cómo leer los colores:**
- 🔴 Rojo intenso (cerca de 1.0) → correlación positiva fuerte.
- ⚪ Blanco (cerca de 0.0) → sin correlación.
- 🔵 Azul intenso (cerca de -1.0) → correlación negativa fuerte.

```python
numericas = df[['math score', 'reading score', 'writing score']]
correlacion = numericas.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(correlacion, annot=True, fmt='.2f', cmap='RdYlGn',
            vmin=-1, vmax=1, linewidths=1, square=True,
            annot_kws={'size': 14, 'weight': 'bold'})
plt.title('Mapa de calor de correlaciones entre puntajes', fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()

print("\n💡 Interpretación:")
print(f"   Lectura ↔ Escritura: {correlacion.loc['reading score','writing score']:.2f} → correlación MUY fuerte")
print(f"   Matemáticas ↔ Lectura: {correlacion.loc['math score','reading score']:.2f} → correlación moderada")
print("   Un estudiante que lee bien, también escribe bien. ¡Tiene mucho sentido!")
```

---
## 5. `pairplot` — El comando exploratorio más poderoso 🚀

`pairplot` genera automáticamente una cuadrícula completa con **todos los scatter plots posibles** entre variables numéricas, más los histogramas en la diagonal. Con una sola línea de código ves todo el panorama.

```python
g = sns.pairplot(df[['math score', 'reading score', 'writing score', 'gender']],
                 hue='gender', palette=['#6366f1', '#f59e0b'],
                 plot_kws={'alpha': 0.5, 's': 30},
                 diag_kind='kde')
g.figure.suptitle('Pairplot: Vista completa de relaciones entre todos los puntajes', y=1.02, fontsize=13, fontweight='bold')
plt.show()
print("\n💡 La diagonal muestra la distribución de cada variable por separado.")
print("   El triángulo superior e inferior muestran la relación entre cada par de variables.")
```

## 6. Resumen: Seaborn en una tabla 📋

| Función | Para qué | Ejemplo de uso |
|---|---|---|
| `histplot` | Distribución de una variable numérica | Cómo se distribuyen las notas |
| `boxplot` | Mediana, IQR y outliers por categoría | Notas por género |
| `violinplot` | Como boxplot + forma de distribución | Notas por tipo de almuerzo |
| `countplot` | Contar categorías | Cuántos estudiantes por grupo étnico |
| `heatmap` | Matriz de correlación en colores | ¿Qué variables se relacionan? |
| `pairplot` | Vista completa de todas las relaciones | Al inicio del EDA |
