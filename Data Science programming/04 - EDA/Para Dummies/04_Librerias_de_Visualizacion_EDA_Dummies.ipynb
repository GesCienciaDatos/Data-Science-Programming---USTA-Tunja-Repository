## ¿Qué librería de visualización usar? 🎨

En Python hay 3 librerías principales de visualización y cada una tiene su lugar:

| Librería | 💡 Analogía | Cuándo Usar | Nivel de Código |
|---|---|---|---|
| **Matplotlib** | El lienzo en blanco — control total | Gráficas personalizadas, publicaciones | ⚙️ Alto |
| **Seaborn** | Matplotlib con plantillas hermosas | EDA estadístico, correlaciones | 🟡 Medio |
| **Plotly** | Gráficas interactivas con zoom/hover | Dashboards, presentaciones, web | ✅ Bajo-Medio |


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Intentar importar seaborn y plotly
try:
    import seaborn as sns
    print('✅ Seaborn disponible')
except ImportError:
    print('⚠️ Instala con: pip install seaborn')

try:
    import plotly.express as px
    print('✅ Plotly disponible')
except ImportError:
    print('⚠️ Instala con: pip install plotly')

# Dataset para demostración
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=['sepalo_largo','sepalo_ancho','petalo_largo','petalo_ancho'])
df['especie'] = pd.Categorical.from_codes(iris.target, iris.target_names)
print(f'\nDataset Iris listo: {df.shape}')
```

```python
# ── MATPLOTLIB — El más básico pero el más flexible ─────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
fig.suptitle('Matplotlib — El Clásico', fontweight='bold', fontsize=13)

# Scatter manual
colores_esp = {'setosa': 'red', 'versicolor': 'blue', 'virginica': 'green'}
for especie, color in colores_esp.items():
    mask = df['especie'] == especie
    axes[0].scatter(df[mask]['sepalo_largo'], df[mask]['petalo_largo'],
                    c=color, label=especie, alpha=0.6, s=50)
axes[0].set_xlabel('Largo del Sépalo'); axes[0].set_ylabel('Largo del Pétalo')
axes[0].legend(); axes[0].set_title('Scatter: Sépalo vs Pétalo')

# Histogramas superpuestos
for especie, color in colores_esp.items():
    data = df[df['especie']==especie]['sepalo_largo']
    axes[1].hist(data, alpha=0.5, color=color, label=especie, bins=15)
axes[1].set_xlabel('Largo del Sépalo'); axes[1].set_ylabel('Frecuencia')
axes[1].legend(); axes[1].set_title('Histogramas por Especie')

plt.tight_layout()
plt.show()
```

```python
# ── SEABORN — Estadístico y Bonito ──────────────────────────────────
try:
    import seaborn as sns
    sns.set_theme(style='whitegrid', palette='muted')

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle('Seaborn — Estadístico y Elegante', fontweight='bold', fontsize=13)

    # Boxplot con una línea
    sns.boxplot(data=df, x='especie', y='petalo_largo', palette='muted', ax=axes[0])
    axes[0].set_title('Boxplot por Especie')

    # Pairplot simplificado (solo 2 variables)
    sns.scatterplot(data=df, x='sepalo_largo', y='petalo_largo',
                    hue='especie', style='especie', ax=axes[1])
    axes[1].set_title('Scatter con Seaborn')

    plt.tight_layout()
    plt.show()
    print('💡 Seaborn: mucho más conciso que Matplotlib para gráficas estadísticas.')
except ImportError:
    print('Instala seaborn con: pip install seaborn')
```
