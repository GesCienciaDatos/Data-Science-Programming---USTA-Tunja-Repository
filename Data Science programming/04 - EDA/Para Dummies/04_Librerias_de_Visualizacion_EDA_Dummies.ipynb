## Gráficos Avanzados: Pairplots y Mapas de Calor (Heatmaps) 🔥

- **Pairplot:** Genera una cuadrícula que cruza automáticamente **todas las variables numéricas contra todas**.
- **Mapa de Calor de Correlación:** Muestra qué tan unidas están las variables mediante colores cálidos (cercanos a $+1$) o fríos (cercanos a $-1$).


```python
import seaborn as sns
import matplotlib.pyplot as plt

df_iris = sns.load_dataset("iris")

# 1. Matriz de Correlación:
matriz_corr = df_iris.corr(numeric_only=True)

plt.figure(figsize=(7, 5))
sns.heatmap(matriz_corr, annot=True, cmap="Blues", fmt=".2f", linewidths=1)
plt.title("Mapa de Calor de Correlación")
plt.show()
```
