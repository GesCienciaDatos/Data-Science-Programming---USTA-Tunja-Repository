## Tu Checklist de Exploración Preliminar ✅

Cuando recibes un dataset nuevo, hay una **checklist de 5 puntos** que todo Data Scientist ejecuta:

1. **Tamaño y Forma** → ¿Cuántos datos tengo?
2. **Tipos de Variables** → ¿Numéricas, categóricas, fechas, texto?
3. **Valores Faltantes** → ¿Cuántos nulos hay y en qué columnas?
4. **Valores Únicos** → ¿Hay columnas con demasiada variación? ¿O con muy poca?
5. **Valores Duplicados** → ¿Hay filas repetidas?


```python
import pandas as pd
import numpy as np

# Dataset simulado con problemas reales
np.random.seed(7)
n = 200
df = pd.DataFrame({
    'ID': list(range(1, n+1)) + [5, 17, 88],  # 3 duplicados
    'Temperatura': np.concatenate([np.random.normal(22, 4, n), [None, None, None]]),
    'Humedad': np.concatenate([np.random.uniform(40, 90, n), [None, 71.2, 65.0]]),
    'Ciudad': np.concatenate([np.random.choice(['Tunja', 'Bogotá', 'Cali', 'Medellín'], n),
                               ['TUNJA', 'bogota', 'CALI']]),  # inconsistencias
    'Lluvia': np.concatenate([np.random.choice(['Sí', 'No'], n), ['si', 'NO', 'Sí']]),
})[:n+3]

print(f'Dataset: {df.shape[0]} filas x {df.shape[1]} columnas')
display(df.head(8))
```

```python
# Punto 1 y 2: Tamaño y tipos
print('📐 FORMA Y TIPOS:')
print(f'  Filas: {df.shape[0]} | Columnas: {df.shape[1]}')
print('\n  Tipos por columna:')
print(df.dtypes)

# Punto 3: Valores faltantes
print('\n🔍 VALORES FALTANTES:')
nulos = df.isnull().sum()
pct = (nulos / len(df) * 100).round(1)
reporte_nulos = pd.DataFrame({'Nulos': nulos, 'Porcentaje': pct})
print(reporte_nulos[reporte_nulos['Nulos'] > 0])
```

```python
# Punto 4: Valores únicos
print('🔢 VALORES ÚNICOS POR COLUMNA:')
for col in df.columns:
    nunique = df[col].nunique()
    print(f'  {col}: {nunique} valores únicos', end='')
    if nunique <= 6:
        print(f' → {df[col].unique()}')
    else:
        print()

# Punto 5: Duplicados
print(f'\n👥 DUPLICADOS:')
print(f'  Filas completamente duplicadas: {df.duplicated().sum()}')
print(f'  IDs duplicados: {df["ID"].duplicated().sum()}')

# Inconsistencias en variables categóricas
print(f'\n⚠️ INCONSISTENCIAS EN CIUDAD:')
print(df['Ciudad'].value_counts())
```

> **🛠️ Cómo Corregir Inconsistencias de Texto**
>
> Cuando una columna tiene 'Tunja', 'TUNJA', 'tunja' — son la misma ciudad pero Python las trata diferente.
> 
> Solución rápida: `df['Ciudad'] = df['Ciudad'].str.strip().str.title()`
> 
> - `.str.strip()` → elimina espacios al inicio/final
> - `.str.title()` → convierte a 'Primer Letra Mayúscula'
