## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **"para no ingenieros"** del módulo 01 de EDA. Aprenderemos a hacer la primera inspección de cualquier dataset nuevo, como cuando abres una caja por primera vez y antes de usarla miras qué hay adentro.

Al terminar sabrás usar:
1. `.shape` — para saber cuántas filas y columnas tiene la tabla.
2. `.info()` — para saber de qué tipo es cada columna y si hay datos faltantes.
3. `.head()` y `.sample()` — para ver ejemplos reales de los datos.
4. `.isnull().sum()` — para contar cuántos espacios en blanco hay por columna.
5. `.describe()` — para obtener el resumen estadístico básico.

---
## 1. La analogía de la lista de asistencia 📋

Piensa en un dataset como una **lista de asistencia de un colegio**:

- Cada **fila** es un estudiante.
- Cada **columna** es una característica: nombre, edad, nota de matemáticas, ciudad de origen...

Cuando el rector llega y le entregan la lista, lo primero que hace es:
1. Contar cuántos estudiantes hay *(`.shape`)*.
2. Ver qué información tiene la lista *(`.info()`)*.
3. Hojear algunos registros al azar *(`.sample()`)*.
4. Verificar si hay espacios en blanco (campos sin llenar) *(`.isnull().sum()`)*.

Eso es exactamente lo que haremos con código.

````python
import os
import urllib.parse
import urllib.request
import pandas as pd

# 🚀 Función para cargar datasets de forma segura (local o Google Colab)
def load_dataset(filename, module_name="04 - EDA"):
    candidates = [
        f"data/{filename}",
        f"../{module_name}/data/{filename}",
        f"{module_name}/data/{filename}",
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path
    os.makedirs("data", exist_ok=True)
    target_path = f"data/{filename}"
    encoded_module = urllib.parse.quote(module_name)
    encoded_file = urllib.parse.quote(filename)
    url_main = f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_module}/data/{encoded_file}"
    print(f"📥 Descargando dataset '{filename}' desde el repositorio...")
    try:
        urllib.request.urlretrieve(url_main, target_path)
    except Exception:
        url_master = url_main.replace("/main/", "/master/")
        urllib.request.urlretrieve(url_master, target_path)
    print(f"✅ Dataset '{filename}' cargado exitosamente.")
    return target_path

# Cargamos el dataset de rendimiento de estudiantes
file_path = load_dataset('StudentsPerformance.csv', '04 - EDA')
df = pd.read_csv(file_path)
print("\n¡Dataset cargado! Primeras filas:")
df.head()
````

---
## 2. ¿Cuántos hay? → `.shape` 📐

`.shape` devuelve una pareja `(número_de_filas, número_de_columnas)`. Es como preguntar: *"¿cuántos estudiantes hay y cuántas cosas sabemos de cada uno?"*

````python
filas, columnas = df.shape
print(f"📊 El dataset tiene {filas} estudiantes (filas) y {columnas} características (columnas).")
````

---
## 3. ¿Qué tipo de información tenemos? → `.info()` 🗂️

`.info()` es como la ficha técnica de la lista: muestra el nombre de cada columna, qué tipo de dato guarda (número, texto, fecha) y cuántos registros válidos tiene (es decir, cuántos NO están en blanco).

````python
df.info()
````

### 🤔 ¿Cómo leer el resultado de `.info()`?

- **`Non-Null Count`**: cuántas filas tienen dato. Si ves un número menor que el total de filas, hay valores faltantes en esa columna.
- **`Dtype`**: el tipo de dato.
  - `object` → texto (palabras, categorías)
  - `int64` → número entero (sin decimales)
  - `float64` → número con decimales

> 💡 **Tip:** Si una columna que debería ser número aparece como `object`, significa que hay algo raro (quizás letras mezcladas con números) y habrá que limpiarla.

---
## 4. Ver ejemplos reales → `.head()` y `.sample()` 👀

````python
# .head(n) muestra las primeras n filas
print("=== Primeras 5 filas ===")
display(df.head(5))

# .sample(n) muestra n filas al azar — ¡más representativo!
print("\n=== 5 filas aleatorias ===")
display(df.sample(5, random_state=42))
````

### ¿Cuándo usar `.head()` vs `.sample()`?

- Usa **`.head()`** para un vistazo rápido del inicio de la tabla.
- Usa **`.sample()`** cuando los datos están ordenados (por fecha, por ID) y quieres ver la variedad real — las primeras filas podrían ser todas del mismo tipo.

---
## 5. ¿Hay espacios en blanco? → `.isnull().sum()` 🕳️

Un valor nulo (`NaN` en Python) es simplemente un campo vacío — como una celda en blanco en Excel. `.isnull()` crea una tabla de `True`/`False` (verdadero donde hay nulo), y `.sum()` suma los `True` para darte el total de huecos por columna.

````python
nulos = df.isnull().sum()
print("🕳️ Valores nulos por columna:")
print(nulos)

total_nulos = nulos.sum()
print(f"\n📌 Total de celdas vacías en todo el dataset: {total_nulos}")
if total_nulos == 0:
    print("✅ ¡Perfecto! No hay valores faltantes en este dataset.")
else:
    print("⚠️ Hay valores faltantes que deberías analizar antes de modelar.")
````

## 6. El resumen estadístico básico → `.describe()` 📈

`.describe()` es el equivalente a pedirle al médico un "chequeo básico" de salud de tus datos numéricos. Te da:

| Campo | ¿Qué significa? |
|---|---|
| `count` | Cuántas filas válidas tiene esa columna |
| `mean` | El promedio |
| `std` | Qué tan dispersos están los datos (desviación estándar) |
| `min` | El valor más pequeño |
| `25%` | El 25% de los datos está por debajo de este valor |
| `50%` | La mitad exacta (mediana) |
| `75%` | El 75% de los datos está por debajo de este valor |
| `max` | El valor más grande |

````python
df.describe()
````

### 🤔 ¿Cómo detectar algo raro con `.describe()`?

Busca:
- **`min` negativo** en columnas que no debería tenerlo (ej. edad = -5 → error de digitación).
- **`max` muy alejado del `75%`** → posible valor atípico extremo.
- **`count` menor que el total de filas** → hay nulos en esa columna.

## 7. Tu checklist de exploración inicial ✅

Cada vez que recibas un dataset nuevo, ejecuta estos 5 pasos en orden:

```python
df.shape           # ¿Cuántas filas y columnas?
df.info()          # ¿Qué tipos de datos? ¿Hay nulos?
df.head()          # ¿Cómo se ven los datos reales?
df.isnull().sum()  # ¿Cuántos valores faltantes por columna?
df.describe()      # ¿Cuál es el resumen estadístico?
```
