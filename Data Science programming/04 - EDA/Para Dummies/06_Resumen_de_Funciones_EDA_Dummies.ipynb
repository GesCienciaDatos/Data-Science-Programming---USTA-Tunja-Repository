## Chuleta Rápida de EDA para el Bolsillo 📑

```python
# 1. Primer Vistazo:
df.shape                 # Filas y columnas
df.head(10)              # Primeras 10 filas
df.info()                # Tipos de datos y nulos
df.describe()            # Estadísticas descriptivas

# 2. Conteo y Frecuencias:
df['categoria'].value_counts()           # Cantidad por categoría
df['categoria'].value_counts(normalize=True) * 100  # Porcentaje

# 3. Agrupaciones:
df.groupby('ciudad')['ventas'].agg(['count', 'mean', 'std', 'max'])

# 4. Gráficos Clave:
sns.histplot(df['columna'], kde=True)    # Histograma con curva
sns.boxplot(x='categoria', y='valor', data=df) # Comparar grupos
sns.scatterplot(x='var1', y='var2', hue='categoria', data=df) # Cruce 2D
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm') # Correlación
```
