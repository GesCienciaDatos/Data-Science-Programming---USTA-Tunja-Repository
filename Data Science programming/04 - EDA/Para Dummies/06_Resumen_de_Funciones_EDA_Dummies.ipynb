## Tu Hoja de Ruta Completa de EDA 🗺️

Este notebook es tu **cheat sheet** de referencia: todas las funciones de EDA en un solo lugar.
Guárdalo y úsalo como referencia en cada proyecto de datos.


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Dataset completo para demostración
from sklearn.datasets import load_breast_cancer
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['diagnostico'] = pd.Categorical.from_codes(cancer.target, ['maligno', 'benigno'])

print(f'Dataset Cáncer de Mama: {df.shape[0]} muestras x {df.shape[1]} variables')
print(f'Variables: {list(df.columns[:5])} ...')
````

````python
# ═══════════════════════════════════════════════
#  CHEAT SHEET COMPLETO DE EDA
# ═══════════════════════════════════════════════

print('1️⃣ FORMA Y ESTRUCTURA')
print(f'   df.shape = {df.shape}')
print(f'   df.dtypes:\n{df.dtypes.head(5).to_string()}\n   ...')

print('\n2️⃣ VALORES FALTANTES')
print(f'   df.isnull().sum().sum() = {df.isnull().sum().sum()} (cero = perfecto)')

print('\n3️⃣ ESTADÍSTICAS DESCRIPTIVAS')
display(df.describe().iloc[:, :4].round(2))  # Primeras 4 columnas para brevedad
````

````python
print('4️⃣ CORRELACIONES FUERTES (|r| > 0.8)')
corr = df.select_dtypes(include='number').corr()
# Encontrar pares altamente correlacionados
pares_altos = []
for i in range(len(corr.columns)):
    for j in range(i+1, len(corr.columns)):
        val = abs(corr.iloc[i, j])
        if val > 0.8:
            pares_altos.append((corr.columns[i], corr.columns[j], round(corr.iloc[i,j], 3)))

print(f'   Pares con correlación > 0.8: {len(pares_altos)}')
for a, b, v in pares_altos[:5]:
    print(f'   {a} ↔ {b}: r={v}')

print('\n5️⃣ BALANCE DE CLASES')
print(df['diagnostico'].value_counts())
pct_maligno = (df['diagnostico'] == 'maligno').mean() * 100
print(f'   → {pct_maligno:.1f}% maligno, {100-pct_maligno:.1f}% benigno')
````

````python
# Resumen visual final
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
fig.suptitle('EDA Resumen Visual — Dataset Cáncer de Mama', fontweight='bold')

# 1. Balance de clases
counts = df['diagnostico'].value_counts()
axes[0].pie(counts.values, labels=counts.index, autopct='%1.1f%%',
            colors=['#ef4444', '#22c55e'], startangle=90,
            wedgeprops={'edgecolor': 'white', 'linewidth': 2})
axes[0].set_title('Balance de Clases')

# 2. Histograma de feature más informativa
for clase, color in [('maligno','#ef4444'),('benigno','#22c55e')]:
    data = df[df['diagnostico']==clase]['mean radius']
    axes[1].hist(data, alpha=0.6, bins=20, color=color, label=clase)
axes[1].set_xlabel('Radio Medio'); axes[1].set_ylabel('Frecuencia')
axes[1].set_title('Radio Medio por Diagnóstico'); axes[1].legend()

# 3. Boxplot
for i, (clase, color) in enumerate([('maligno','#ef4444'),('benigno','#22c55e')]):
    data = df[df['diagnostico']==clase]['mean concavity']
    bp = axes[2].boxplot(data, positions=[i], widths=0.5, patch_artist=True,
                          boxprops=dict(facecolor=color, alpha=0.7))
axes[2].set_xticks([0,1]); axes[2].set_xticklabels(['Maligno','Benigno'])
axes[2].set_ylabel('Concavidad Media'); axes[2].set_title('Concavidad por Diagnóstico')

plt.tight_layout()
plt.show()
````
