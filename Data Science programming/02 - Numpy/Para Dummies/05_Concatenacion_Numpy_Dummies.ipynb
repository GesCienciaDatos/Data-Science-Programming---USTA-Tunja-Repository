## 1. Pegando hojas de un cuaderno: `np.concatenate()` 📎

Imagina que tienes dos hojas con una tabla de datos (edad y salario de algunas personas), y quieres unirlas en una sola tabla más grande. Tienes dos formas de pegarlas:

* **Una debajo de la otra** (`axis=0`): agregas más **filas**, es decir, más personas. Para esto, ambas hojas deben tener el mismo número de columnas.
* **Una al lado de la otra** (`axis=1`): agregas más **columnas**, es decir, más variables sobre las mismas personas. Para esto, ambas hojas deben tener el mismo número de filas.

`np.concatenate([hoja1, hoja2], axis=...)` hace exactamente eso.

```python
import numpy as np

grupo_a = np.array([[25, 1500], [30, 2200]])              # 2 personas, 2 variables
grupo_b = np.array([[22, 1200], [45, 3500], [38, 2800]])  # 3 personas, 2 variables

# Pegamos "una tabla debajo de la otra": más personas
base_completa = np.concatenate([grupo_a, grupo_b], axis=0)
print("Base con las 5 personas (axis=0):\n", base_completa)
print("Forma final:", base_completa.shape)

edad_salario = np.array([[25, 1500], [30, 2200]])  # 2 personas, 2 variables
antiguedad = np.array([[3], [7]])                  # 2 personas, 1 variable nueva

# Pegamos "una tabla al lado de la otra": más variables para las mismas personas
base_con_antiguedad = np.concatenate([edad_salario, antiguedad], axis=1)
print("\nBase con la nueva columna de antigüedad (axis=1):\n", base_con_antiguedad)
```

### 🤔 ¿Qué acaba de pasar?

- `grupo_a` tiene 2 personas y `grupo_b` tiene 3, pero **ambos tienen 2 columnas** (edad y salario). Por eso podemos pegarlos con `axis=0`: el resultado tiene 2+3=5 filas (personas) y sigue teniendo 2 columnas.
- En el segundo ejemplo, `edad_salario` y `antiguedad` tienen **las mismas 2 filas** (mismas personas), así que podemos pegarlos con `axis=1`: el resultado sigue teniendo 2 filas, pero ahora con 2+1=3 columnas.
- La regla general: en `axis=0` deben coincidir las columnas; en `axis=1` deben coincidir las filas. Si no coinciden, NumPy te lanzará un error.

## 2. Los atajos `np.vstack()` y `np.hstack()` 🧱

`np.vstack()` (*vertical stack*) y `np.hstack()` (*horizontal stack*) son atajos de `np.concatenate()` para los casos más comunes, sin tener que escribir `axis=...`:

* `np.vstack([A, B])` equivale a `np.concatenate([A, B], axis=0)` — apilar hacia abajo.
* `np.hstack([A, B])` equivale a `np.concatenate([A, B], axis=1)` — apilar hacia la derecha.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Apilado vertical (vstack) - más filas:\n", np.vstack([A, B]))
print("\nApilado horizontal (hstack) - más columnas:\n", np.hstack([A, B]))
```

### 🤔 ¿Qué acaba de pasar?

- `np.vstack([A, B])` puso la matriz `B` justo debajo de `A`, dando una matriz de 4 filas x 2 columnas — el mismo resultado que hubieras obtenido con `np.concatenate([A, B], axis=0)`.
- `np.hstack([A, B])` puso `B` a la derecha de `A`, dando una matriz de 2 filas x 4 columnas — igual que `np.concatenate([A, B], axis=1)`.
- En la práctica, muchas personas prefieren `vstack`/`hstack` porque son más cortos de escribir y más fáciles de leer cuando ya sabes en qué dirección quieres apilar.

## 3. Apilar creando una dimensión nueva: `np.stack()` 🥞

Todo lo anterior *reutilizaba* un eje que ya existía (filas o columnas). `np.stack()` es distinto: **crea un eje totalmente nuevo**, como si apilaras varias fotos transparentes una encima de la otra para formar un paquete.

Por ejemplo, si tienes 3 fotos en blanco y negro (matrices 2D) del mismo tamaño y las apilas con `np.stack()`, obtienes un "paquete" 3D — exactamente lo que se hace en una imagen a color, donde se apilan los canales Rojo, Verde y Azul.

```python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

# Apilamos dos vectores 1D: se crea una dimensión nueva
paquete = np.stack([vector_a, vector_b])
print("Vectores apilados en un 'paquete' 2D:\n", paquete)
print("Forma:", paquete.shape, "  (antes cada vector era solo (3,))")

# Con axis=-1 los "emparejamos" en columnas en vez de filas
paquete_columnas = np.stack([vector_a, vector_b], axis=-1)
print("\nApilados como columnas (axis=-1):\n", paquete_columnas)
```

### 🤔 ¿Qué acaba de pasar?

- `vector_a` y `vector_b` eran cada uno un array 1D con forma `(3,)`. Al apilarlos con `np.stack()`, NumPy creó una dimensión nueva: el resultado tiene forma `(2, 3)` — "2 vectores de 3 elementos cada uno".
- Esto es distinto de `np.concatenate` o `np.vstack`: si hubiéramos usado `np.concatenate([vector_a, vector_b])` sobre dos vectores 1D, habríamos obtenido un solo vector largo de 6 elementos, **no** una matriz nueva.
- Cambiar `axis=-1` cambia cómo se acomodan los datos dentro del nuevo eje: en este caso, en vez de dos filas obtuvimos pares (columna por columna).

## 4. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| `np.concatenate([a, b], axis=0)` | Pega arrays agregando más filas (mismas columnas). |
| `np.concatenate([a, b], axis=1)` | Pega arrays agregando más columnas (mismas filas). |
| `np.vstack([a, b])` | Atajo de `concatenate(..., axis=0)`: apilar hacia abajo. |
| `np.hstack([a, b])` | Atajo de `concatenate(..., axis=1)`: apilar hacia la derecha. |
| `np.stack([a, b])` | Crea un **eje nuevo**, como apilar fotos transparentes en un paquete. |
