## 1. Una lista de Python es un cajón de sastre 🗄️

Piensa en el cajón de la cocina donde guardas "de todo un poco": un control remoto, unas tijeras, un cargador, un tenedor suelto... Los objetos no tienen un tamaño fijo ni un orden claro, así que para buscar algo tienes que ir revisando uno por uno.

Una **lista de Python** funciona de forma parecida por dentro: puede mezclar números, texto y otras listas, y Python en realidad no guarda los valores pegados uno junto al otro en la memoria del computador — guarda *notitas con la dirección* de dónde está cada valor real. Leer la lista completa significa saltar de dirección en dirección, como ir de cajón en cajón buscando cada cosa.

Un **array de NumPy** (`ndarray`) es más parecido a una huevera: todos los compartimentos son del mismo tamaño y guardan el mismo tipo de cosa — todos números enteros, o todos números decimales, nunca una mezcla. Como todo está ordenado, del mismo tipo y **pegado en un solo bloque de memoria**, el computador puede leerlo y procesarlo de un solo tirón, sin ir saltando de dirección en dirección.

> 📌 **Para recordar:** lista de Python = cajón desordenado con notitas de dirección. Array de NumPy = huevera ordenada, uniforme y compacta.

Por eso, cuando trabajamos con muchos números (miles, millones), NumPy no solo es más cómodo: es muchísimo más rápido.

## 2. Uno por uno vs. todos a la vez ⚡

Si tuvieras que sellar 1,000 hojas con un sello de "Aprobado", tienes dos opciones:

- **Sellar hoja por hoja, a mano:** tomas una hoja, la sellas, la dejas, tomas la siguiente... 1,000 veces.
- **Usar una máquina que sella 1,000 hojas de un solo golpe** porque están perfectamente apiladas y alineadas.

Un bucle `for` de Python que recorre una lista elemento por elemento es como sellar hoja por hoja: funciona, pero es lento cuando hay muchos datos. NumPy, en cambio, aprovecha que sus arrays están "perfectamente apilados" en memoria (como vimos con la huevera) para aplicar la operación a **todos los elementos de un solo golpe**. A esto se le llama **vectorizar** una operación.

Vamos a comprobarlo con un experimento real: multiplicar por 2 un montón de números, primero con una lista y luego con un array de NumPy, y medir cuánto tarda cada uno.

```python
import numpy as np
import time

n = 200_000
lista_py = list(range(n))
array_np = np.arange(n)

# Multiplicar cada número por 2, con una lista (uno por uno)
t0 = time.time()
res_py = [x * 2 for x in lista_py]
t_py = time.time() - t0

# Multiplicar cada número por 2, con un array de NumPy (todos a la vez)
t0 = time.time()
res_np = array_np * 2
t_np = time.time() - t0

print(f"Lista de Python (uno por uno):  {t_py * 1000:.2f} ms")
print(f"Array de NumPy (todos a la vez): {t_np * 1000:.2f} ms")
print(f"NumPy fue aproximadamente {t_py / t_np:.1f} veces más rápido")
```

### 🤔 ¿Qué acaba de pasar?

- `list(range(n))` crea una lista normal de Python con 200,000 números; `np.arange(n)` crea un array de NumPy con los mismos números, pero guardados en bloque.
- `[x * 2 for x in lista_py]` es un bucle disfrazado: Python visita cada número de la lista, uno por uno, lo multiplica y guarda el resultado.
- `array_np * 2` no tiene ningún bucle escrito por nosotros: le decimos a NumPy "multiplica por 2" una sola vez, y él se encarga de aplicarlo a los 200,000 números internamente, en código C ya optimizado.
- Los tiempos pueden variar según el computador, pero NumPy debería salir claramente más rápido — normalmente entre 10 y 100 veces, dependiendo del equipo. Esa diferencia crece muchísimo más con millones de datos, que es justo lo normal en Ciencia de Datos.

## 3. Conociendo a la huevera por dentro: el `ndarray` 🥚

Ya vimos *por qué* NumPy es rápido. Ahora veamos *qué es exactamente* lo que estamos creando cuando escribimos `np.array([...])`. A ese objeto se le llama **`ndarray`** (*N-dimensional array*, o "array de N dimensiones"), y tiene tres datos que siempre vale la pena revisar:

| Dato | Pregunta que responde | Analogía con la huevera |
|---|---|---|
| `dtype` | ¿De qué tipo son los elementos? | ¿Los compartimentos guardan huevos grandes o pequeños? |
| `shape` | ¿Cuántos elementos hay, organizados cómo? | ¿Es una huevera de 1 fila x 6 puestos, o de 2 filas x 6? |
| `ndim` | ¿Cuántas "dimensiones" tiene esa organización? | Una fila de hueveras es 1D; una caja con varias hueveras apiladas ya sería 2D |

Vamos a crearlo y a preguntarle directamente esos tres datos.

```python
# Creamos un array sencillo de una sola dimensión
arr = np.array([10, 20, 30, 40, 50])
print("Array original:", arr)

# Operación vectorizada: multiplicar por 2 y sumar 5, a todos los elementos a la vez
resultado = arr * 2 + 5
print("Array transformado:", resultado)

print(f"Dimensiones (ndim): {arr.ndim}")
print(f"Forma (shape): {arr.shape}")
print(f"Tipo de dato (dtype): {arr.dtype}")
```

### 🤔 ¿Qué acaba de pasar?

- `np.array([10, 20, 30, 40, 50])` empaqueta esos 5 números en un `ndarray` de una sola dimensión.
- `arr * 2 + 5` es la misma idea de "sellar todas las hojas de un golpe": a **cada** número del array se le multiplica por 2 y se le suma 5, sin escribir ningún bucle.
- `arr.ndim` es `1` porque el array es "una sola fila" de números (una dimensión).
- `arr.shape` es `(5,)`: significa "5 elementos en esa única dimensión". Cuando lleguemos a tablas de datos (filas y columnas), el `shape` tendrá dos números, por ejemplo `(3, 5)`.
- `arr.dtype` te dice el tipo exacto que NumPy eligió para guardar esos números (normalmente `int64`, es decir, números enteros). Recuerda: **todos** los elementos del array comparten ese mismo tipo — esa es justo la regla de la huevera.

## 4. ¿Por qué me debería importar todo esto? 🧭

Porque **casi todo** lo que usarás en el resto del curso está construido sobre NumPy, aunque no lo veas directamente:

- `Pandas` (tablas de datos) guarda sus columnas internamente como arrays de NumPy.
- `Scikit-Learn` (Machine Learning) espera que le entregues los datos como arrays de NumPy.
- `Matplotlib` (gráficas) dibuja a partir de arrays de NumPy.

Es decir: cuando entiendes qué es un `ndarray`, su `shape` y su `dtype`, estás entendiendo el "idioma común" que hablan casi todas las herramientas de Ciencia de Datos en Python — no solo NumPy.

## 5. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| NumPy | La "caja de herramientas" numérica sobre la que se construye casi todo el ecosistema de Ciencia de Datos en Python. |
| Lista de Python | Un cajón desordenado: guarda "notitas de dirección", no los valores pegados uno junto al otro. |
| `ndarray` | Una huevera ordenada: todos sus elementos son del mismo tipo y están en un bloque compacto de memoria. |
| Vectorizar | Aplicarle una operación a **todos** los elementos de un array a la vez, sin escribir un bucle `for`. |
| `dtype` | El tipo de dato que comparten todos los elementos del array (ej. `int64`). |
| `shape` | La "forma" del array: cuántos elementos hay y cómo están organizados (ej. `(5,)`). |
| `ndim` | El número de dimensiones del array (1D, 2D, 3D...). |
