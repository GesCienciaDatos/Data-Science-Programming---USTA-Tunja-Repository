## 1. La máquina de números al azar, con receta 🎲

Cuando tiras un dado de verdad, el resultado es impredecible. Pero en programación, muchas veces necesitamos números "al azar" que sean **reproducibles** — por ejemplo, para que un compañero pueda ejecutar tu mismo código y obtener exactamente los mismos resultados, y así comprobar que todo funciona igual.

Para eso, NumPy usa una "receta" llamada **semilla** (`seed`). Si usas la misma semilla, siempre obtienes la misma secuencia de números "aleatorios" — es como si la máquina de azar no fuera realmente impredecible, sino que sigue fielmente una receta que tú eliges.

```python
generador = np.random.default_rng(seed=42)
```

Con ese generador puedes pedir distintos "sabores" de números al azar: enteros, decimales, o números que siguen la famosa campana de Gauss (distribución normal).

````python
import numpy as np

generador = np.random.default_rng(seed=42)

# 5 enteros al azar entre 1 y 6 (como tirar un dado 5 veces)
dados = generador.integers(low=1, high=7, size=5)
print("Tiradas de dado:", dados)

# 5 decimales al azar entre 0 y 1
decimales = generador.uniform(low=0, high=1, size=5)
print("Decimales uniformes:", decimales)

# 5 números siguiendo la campana de Gauss (media=0, desviación=1)
normales = generador.normal(loc=0, scale=1, size=5)
print("Números 'campana de Gauss':", normales)
````

### 🤔 ¿Qué acaba de pasar?

- `np.random.default_rng(seed=42)` crea el generador con la receta número 42. Si vuelves a ejecutar esta celda, obtendrás exactamente los mismos números — pero si cambias el `42` por otro número, la secuencia completa cambia.
- `generador.integers(low=1, high=7, size=5)` simula 5 tiradas de dado: recuerda que `high` **no se incluye**, por eso ponemos 7 para permitir hasta el 6.
- `generador.uniform(low=0, high=1, size=5)` reparte los números al azar de forma pareja entre 0 y 1.
- `generador.normal(loc=0, scale=1, size=5)` genera números que se agrupan más cerca de la media (`loc`) y se alejan cada vez menos entre más te acercas a los extremos — la famosa "campana".

## 2. Contando votos: `np.unique()` 🗳️

Imagina que tienes los resultados de una encuesta y quieres saber qué respuestas distintas aparecieron y cuántas veces se repitió cada una — como contar votos en una elección. Para eso está `np.unique()`.

````python
# Simulamos 20 tiradas de un dado de 6 caras
tiradas = np.random.default_rng(seed=777).integers(1, 7, size=20)
print("Tiradas:", tiradas)

# ¿Qué caras salieron, y cuántas veces cada una?
caras, frecuencias = np.unique(tiradas, return_counts=True)
print("\nCaras que aparecieron:", caras)
print("Cuántas veces salió cada una:", frecuencias)

for cara, veces in zip(caras, frecuencias):
    print(f"El número {cara} salió {veces} veces")
````

### 🤔 ¿Qué acaba de pasar?

- `np.unique(tiradas)` por sí solo te daría solo la lista de valores distintos que aparecieron, ordenados de menor a mayor.
- Con `return_counts=True`, además obtienes un segundo array con la cantidad de veces que se repitió cada uno de esos valores — en el mismo orden.
- `zip(caras, frecuencias)` simplemente empareja cada cara con su frecuencia para poder imprimirlas juntas, como una tabla de resultados de la encuesta.

## 3. Dimensiones "de mentira": `expand_dims` y `squeeze` 🗜️

Algunos modelos (sobre todo de Machine Learning y Deep Learning) son muy exigentes con la forma exacta que debe tener la información que reciben. A veces piden un dato con una dimensión extra que **no aporta información nueva**, solo cumple con el formato esperado — como poner un solo zapato dentro de una caja de zapatos vacía extra, solo porque la caja "debe" tener esa forma.

* **`np.expand_dims(array, axis=...)`** agrega una dimensión de tamaño 1 en la posición que indiques.
* **`np.squeeze(array)`** hace lo contrario: elimina todas las dimensiones "de mentira" (las que miden 1).

````python
datos = np.array([10, 20, 30])
print("Forma original:", datos.shape)

# Agregamos una dimensión "de mentira" al inicio (por ejemplo, un "lote" de 1 elemento)
con_dimension_extra = np.expand_dims(datos, axis=0)
print("Con dimensión extra al inicio:", con_dimension_extra.shape)

# Agregamos una dimensión "de mentira" al final
con_dimension_final = np.expand_dims(datos, axis=-1)
print("Con dimensión extra al final:", con_dimension_final.shape)

# squeeze() elimina TODAS las dimensiones de tamaño 1
de_vuelta_a_la_normalidad = np.squeeze(con_dimension_extra)
print("Después de squeeze():", de_vuelta_a_la_normalidad.shape)
````

### 🤔 ¿Qué acaba de pasar?

- `datos` empieza con forma `(3,)`: un simple vector de 3 elementos.
- `np.expand_dims(datos, axis=0)` le agrega una dimensión de tamaño 1 justo al inicio, dejando la forma en `(1, 3)` — los mismos 3 datos, mismo contenido, solo "empacados" en una caja adicional.
- `np.expand_dims(datos, axis=-1)` hace lo mismo pero al final: `(3, 1)`.
- `np.squeeze()` revisa el array y elimina **todas** las dimensiones que midan exactamente 1, devolviendo la forma a `(3,)` — como sacar el zapato de la caja extra y quedarte solo con lo esencial.

## 4. Resumen relámpago ⚡

| Idea | En una frase |
|---|---|
| `np.random.default_rng(seed)` | Crea un generador de números "al azar" reproducible, siguiendo una receta (semilla). |
| `.integers()`, `.uniform()`, `.normal()` | Distintos "sabores" de números aleatorios: enteros, decimales parejos, o campana de Gauss. |
| `np.unique(array)` | Devuelve los valores distintos que aparecen en el array, ordenados. |
| `np.unique(array, return_counts=True)` | Además cuenta cuántas veces se repite cada valor distinto. |
| `np.expand_dims(array, axis)` | Agrega una dimensión "de mentira" (tamaño 1) para cumplir con un formato esperado. |
| `np.squeeze(array)` | Elimina todas las dimensiones "de mentira" (tamaño 1). |
