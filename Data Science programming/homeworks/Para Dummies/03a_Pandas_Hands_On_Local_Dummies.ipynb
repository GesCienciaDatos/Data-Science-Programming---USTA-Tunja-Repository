## Ejemplo resuelto: tu primera tabla con Pandas 📋

Veamos el patrón completo con un ejemplo mini: una tabla de 3 frutas y su precio.


````python
import pandas as pd

datos_frutas = {
    'Fruta': ['Manzana', 'Banano', 'Mango'],
    'Precio': [2500, 1200, 3000]
}

df_frutas = pd.DataFrame(datos_frutas)
print(df_frutas)

# Seleccionar solo la columna Precio
print(df_frutas['Precio'])

# Filtrar frutas de más de 2000
print(df_frutas[df_frutas['Precio'] > 2000])
````

### 🤔 ¿Qué acaba de pasar?

- Un diccionario de Python (`{'Fruta': [...], 'Precio': [...]}`) se convierte directamente en una tabla con `pd.DataFrame(...)`: cada llave se vuelve una columna.
- `df_frutas['Precio']` selecciona una sola columna (se llama una `Series`).
- `df_frutas[df_frutas['Precio'] > 2000]` es un **filtro**: dentro de los corchetes armamos una condición que da `True`/`False` por fila, y Pandas se queda solo con las filas `True`.

Este mismo patrón — crear, seleccionar, filtrar — es el que vas a usar en todos los ejercicios.

## Resumen relámpago ⚡

| Código | Para qué sirve |
|---|---|
| `pd.DataFrame(diccionario)` | Convertir un diccionario de Python en una tabla |
| `df.head(n)` | Ver solo las primeras `n` filas |
| `df['Columna']` | Seleccionar una columna |
| `df[condicion]` | Filtrar filas que cumplen una condición |
| `df['Nueva'] = ...` | Crear una columna nueva a partir de otras |
| `df['Columna'].mean()` | Calcular el promedio de una columna |
