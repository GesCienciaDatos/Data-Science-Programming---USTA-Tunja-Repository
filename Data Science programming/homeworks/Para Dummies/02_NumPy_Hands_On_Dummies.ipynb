## Ejemplo resuelto: crear y consultar un array 🧱

Antes de los ejercicios, veamos el patrón que vas a repetir una y otra vez: **crear un array con una función de NumPy, y luego mirar su forma con `.shape`.**


````python
import numpy as np

# Creamos un array de puros unos, de tamaño 6
ejemplo = np.ones(6)
print("Array:", ejemplo)
print("Forma (shape):", ejemplo.shape)

# Lo convertimos en una matriz de 2 filas x 3 columnas con reshape
ejemplo_matriz = ejemplo.reshape(2, 3)
print("Como matriz 2x3:")
print(ejemplo_matriz)
````

### 🤔 ¿Qué acaba de pasar?

- `np.ones(6)` crea un array 1D de 6 elementos, todos en `1.0`.
- `.shape` te dice "cuántos elementos hay en cada dimensión" — aquí, `(6,)`.
- `.reshape(2, 3)` reorganiza los mismos 6 valores en 2 filas y 3 columnas, sin cambiar ni un solo número.

Casi todos los ejercicios de este taller son variaciones de "crea un array así" + "pregúntale algo" (su forma, su mínimo, su promedio...).

## Resumen relámpago ⚡

| Función | Para qué sirve |
|---|---|
| `np.zeros(n)` / `np.ones(n)` | Crear un vector de ceros o de unos |
| `np.arange(a, b)` | Crear un vector con los enteros entre `a` y `b` |
| `Z[::-1]` | Invertir un array |
| `.reshape(f, c)` | Reacomodar un array en `f` filas y `c` columnas |
| `np.nonzero(array)` | Encontrar las posiciones distintas de cero |
| `np.eye(n)` | Crear la matriz identidad de tamaño `n` |
| `np.random.random(forma)` | Crear un array de decimales aleatorios entre 0 y 1 |
| `.min()`, `.max()`, `.mean()` | Mínimo, máximo y promedio de un array |
| `array[1:-1, 1:-1]` | Seleccionar "el interior" de una matriz (todo menos el borde) |
| `np.pad(array, ...)` | Añadir un borde relleno alrededor de un array |
| `np.diag(valores, k=...)` | Crear una matriz con valores en una diagonal (desplazada con `k`) |
| `array[:, i]` / `array[:n, i:j]` | Seleccionar una columna completa, o un rango de filas/columnas |
