## Ejemplo resuelto: de la idea al código 🧩

Antes de empezar, veamos un mini-problema parecido resuelto de principio a fin: **pedirle un número al usuario y usarlo en una fórmula**.

Vamos a calcular el perímetro de un cuadrado a partir del lado que escriba la persona que usa el programa.


````python
# 1. Pedimos el dato al usuario. input() SIEMPRE devuelve texto (str),
#    por eso lo convertimos a número con float().
lado_texto = "4.5"  # <- en un programa real sería: input("Lado del cuadrado: ")
lado = float(lado_texto)

# 2. Aplicamos la fórmula
perimetro = lado * 4

# 3. Mostramos el resultado con un mensaje claro
print(f"El perímetro de un cuadrado de lado {lado} es {perimetro}")
````

### 🤔 ¿Qué acaba de pasar?

- `input()` siempre entrega texto, aunque la persona escriba números — por eso el primer paso casi siempre es convertir con `float()` o `int()`.
- Guardamos el resultado en una variable con un nombre que explica qué es (`perimetro`, no `x`).
- El f-string `f"...{variable}..."` inserta el valor dentro del texto sin tener que concatenar con `+`.

Con este mismo patrón — **leer, convertir, calcular, mostrar** — vas a resolver casi todos los ejercicios de este taller.

    if x == y == z:

def char_frequency(str1):
    frecuencias = {}
    for letra in str1:
        if letra in frecuencias:
            frecuencias[letra] += 1
        else:
            frecuencias[letra] = 1
    return frecuencias

def count_even_odd(numbers):
    pares, impares = 0, 0
    for n in numbers:
        if n % 2 == 0:
            pares += 1
        else:

    return pares, impares

## Resumen relámpago ⚡

| Herramienta | Para qué sirve |
|---|---|
| `float(input(...))` | Leer un número escrito por el usuario |
| `texto.split(',')` | Partir un texto en pedazos usando una coma (u otro separador) |
| `"%d/%d" % (a, b)` | Insertar valores dentro de un texto con formato clásico |
| `if x == y == z:` | Comparar varias variables a la vez |
| `n in lista` | Preguntar si un valor está dentro de una lista/tupla |
| `dict[llave] = dict.get(llave, 0) + 1` | Contar apariciones con un diccionario |
| `numero % 2 == 0` | Saber si un número es par |
| `texto[::-1]` | Invertir un texto |
| `all(condicion for x in lista)` | Comprobar que TODOS los elementos cumplan una condición |
| `break` | Salir de un bucle `for`/`while` antes de que termine solo |
