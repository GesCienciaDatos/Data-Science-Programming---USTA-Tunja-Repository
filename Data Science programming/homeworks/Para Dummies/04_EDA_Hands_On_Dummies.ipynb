## ¿Qué vamos a practicar aquí? 🎯

Este taller guiado recorre los mismos 8 ejercicios del **Taller Práctico 04**, pero con pistas y soluciones desplegables para que puedas comparar tu código. La idea de fondo es siempre la misma: aplicar una **checklist de 5 puntos** para conocer un dataset nuevo antes de usarlo en cualquier análisis o modelo.

1. **¿Qué forma tiene la tabla?** — cuántas filas, cuántas columnas y de qué tipo (`shape`, `info()`).
2. **¿Hay huecos?** — valores nulos que debamos tener en cuenta (`isnull()`).
3. **¿Cómo son los números, en resumen?** — media, mínimo, máximo, cuartiles (`describe()`).
4. **¿Cómo se ve cada variable por separado?** — histogramas y conteos por categoría.
5. **¿Cómo se relacionan las variables entre sí?** — comparaciones por grupo (boxplot) y correlación (heatmap).

Como en el taller principal, usaremos el clásico dataset del **Titanic**, que viene incluido en la librería Seaborn.

## Ejemplo resuelto: aplicando la checklist a una tabla mini 🍎

Antes de ir al Titanic, practiquemos los primeros tres puntos de la checklist (forma, nulos y estadísticas) sobre una tabla pequeña e inventada: las ventas de una frutería.


````python
import pandas as pd
import numpy as np

ventas = pd.DataFrame({
    'fruta': ['manzana', 'banano', 'pera', 'uva', 'manzana', 'banano'],
    'precio': [2000, 1500, np.nan, 3000, 2100, 1450],
    'unidades_vendidas': [50, 80, 30, 20, 45, 90]
})

print(f"Forma de la tabla (filas, columnas): {ventas.shape}")
print("\nTipos de dato por columna:")
print(ventas.dtypes)

print("\nNulos por columna:")
print(ventas.isnull().sum())

print("\nEstadísticas descriptivas:")
print(ventas.describe())
````

### 🤔 ¿Qué acaba de pasar?

- `ventas.shape` devuelve una pareja `(filas, columnas)` — aquí `(6, 3)`.
- `ventas.dtypes` te dice qué tipo de dato guarda cada columna: `fruta` es texto (`object`), `precio` y `unidades_vendidas` son números.
- `ventas.isnull().sum()` cuenta los `NaN` de cada columna — verás que `precio` tiene exactamente 1 nulo (la fila de la pera).
- `ventas.describe()` solo resume las columnas **numéricas**: te da el promedio, la desviación estándar, el mínimo, el máximo y los cuartiles. La columna `fruta` (texto) no aparece ahí.

Con estos tres pasos ya sabes, en segundos, el tamaño de la tabla, si tiene huecos y cómo se comportan sus números. Ahora hagamos lo mismo — y un poco más — con un dataset real.

## Resumen relámpago ⚡

| Código | Para qué sirve |
|---|---|
| `sns.load_dataset('titanic')` | Cargar un dataset de ejemplo incluido en Seaborn |
| `df.info()` | Ver tipos de dato, cantidad de filas y nulos por columna |
| `df.describe()` | Obtener estadísticas descriptivas de las columnas numéricas |
| `sns.histplot(data=df, x='col')` | Ver la distribución de una variable numérica |
| `sns.countplot(data=df, x='col')` | Contar cuántas veces aparece cada categoría |
| `sns.boxplot(data=df, x='cat', y='num')` | Comparar una variable numérica entre categorías |
| `df.select_dtypes(include='number')` | Quedarte solo con las columnas numéricas |
| `sns.heatmap(matriz, annot=True)` | Visualizar una matriz de correlación como mapa de calor |
