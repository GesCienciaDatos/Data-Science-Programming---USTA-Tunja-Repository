## Taller 03b: Importación y Análisis Exploratorio 🌐

Practicaremos la carga de datos masivos y filtros con `.loc` y `groupby`.


```python
import pandas as pd
import seaborn as sns

# Usamos el dataset de pingüinos de Palmer como ejemplo de dataset del mundo real:
df_penguins = sns.load_dataset("penguins")

print("1. Primeras 5 filas:")
display(df_penguins.head())

print("\n2. Conteo de pingüinos por isla:")
display(df_penguins["island"].value_counts())

print("\n3. Peso corporal promedio (en gramos) por especie:")
display(df_penguins.groupby("species")["body_mass_g"].mean().round(1))
```
