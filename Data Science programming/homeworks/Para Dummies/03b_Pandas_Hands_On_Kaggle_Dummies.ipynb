## Ejemplo resuelto: inspeccionar una tabla desconocida 🔍

Cuando te entregan una tabla que nunca has visto, el primer reflejo siempre es el mismo: mirarla, y preguntarle su tamaño y sus tipos de datos.


````python
import pandas as pd
import numpy as np

df_ejemplo = pd.DataFrame({
    'producto': ['Camisa', 'Pantalón', 'Zapatos', 'Media'],
    'precio': [45000, 89000, np.nan, 8000],
    'categoria': ['Ropa', 'Ropa', 'Calzado', 'Ropa']
})

print(df_ejemplo.shape)      # (filas, columnas)
print(df_ejemplo.dtypes)     # tipo de cada columna
print(df_ejemplo.isnull().sum())  # nulos por columna
````

### 🤔 ¿Qué acaba de pasar?

- `.shape` devuelve `(filas, columnas)` — una forma rápida de saber "qué tan grande" es la tabla.
- `.dtypes` te dice si cada columna es texto (`object`), número entero (`int64`) o decimal (`float64`) — importante porque no puedes promediar una columna de texto.
- `.isnull()` marca con `True` cada celda vacía; `.sum()` las cuenta por columna. Aquí, `precio` debería mostrar 1 nulo.

## Resumen relámpago ⚡

| Código | Para qué sirve |
|---|---|
| `df.head()` / `df.tail()` | Ver las primeras / últimas filas |
| `df.info()` | Tipos de dato y valores no-nulos por columna |
| `df.describe()` | Estadísticas de las columnas numéricas |
| `df.isnull().sum()` | Contar valores nulos por columna |
| `df['col'].fillna(valor)` | Rellenar nulos con un valor |
| `df.dropna()` | Eliminar filas con nulos |
| `df.groupby('cat')['num'].mean()` | Promedio de una columna, agrupado por categoría |
| `df.sort_values('col', ascending=False)` | Ordenar la tabla |
| `df.to_csv('archivo.csv', index=False)` | Guardar la tabla como CSV |
