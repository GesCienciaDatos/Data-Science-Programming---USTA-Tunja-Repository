## Ejemplo resuelto: una tabla con un hueco 🕳️

Veamos el patrón completo — detectar el nulo, y rellenarlo — en una tabla mínima de 4 filas.


````python
import pandas as pd
import numpy as np

df_ejemplo = pd.DataFrame({
    'producto': ['A', 'B', 'C', 'D'],
    'precio': [1000, np.nan, 1500, 1200]
})

print("Nulos por columna:")
print(df_ejemplo.isnull().sum())

df_ejemplo['precio'] = df_ejemplo['precio'].fillna(df_ejemplo['precio'].median())
print("Después de rellenar con la mediana:")
print(df_ejemplo)
````

### 🤔 ¿Qué acaba de pasar?

- `np.nan` representa "no hay dato" — es el valor que Pandas reconoce como nulo.
- `.isnull().sum()` cuenta cuántos `NaN` hay en cada columna, sin tener que revisarlo a ojo.
- Rellenamos con la **mediana** (`.median()`) en vez del promedio porque la mediana no se deja engañar tan fácil por valores extremos (*outliers*).

## Resumen relámpago ⚡

| Código | Para qué sirve |
|---|---|
| `df.isnull().sum()` | Contar nulos por columna |
| `df['col'].fillna(df['col'].median())` | Rellenar nulos con la mediana |
| `df.drop(columns=['col'])` | Eliminar una columna completa |
| `MinMaxScaler().fit_transform(df[['col']])` | Escalar una columna al rango [0, 1] |
| `pd.to_datetime(columna)` | Convertir texto en fecha real |
| `columna.dt.year` | Extraer el año de una columna de fechas |
| `.str.lower().str.strip()` | Estandarizar texto (minúsculas, sin espacios extra) |
