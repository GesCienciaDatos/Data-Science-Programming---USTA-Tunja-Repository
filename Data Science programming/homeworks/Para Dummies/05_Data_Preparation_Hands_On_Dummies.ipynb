## Taller 05: Taller Integrador de Preparación de Datos 🧼

Pondremos en práctica la imputación de nulos y el escalado de características en un pipeline completo.


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Dataset simulado de candidatos con nulos y escalas dispares:
df_candidatos = pd.DataFrame({
    "Candidato": ["Ana", "Luis", "Carlos", "Diana"],
    "Experiencia_Anios": [3, np.nan, 8, 5],
    "Puntaje_Prueba": [85, 90, np.nan, 78],
    "Pretension_Salarial": [3500000, 4200000, 6000000, 4800000]
})

print("1. Dataset Original con Nulos:")
display(df_candidatos)

# Paso 1: Imputar nulos con la media
df_imputado = df_candidatos.copy()
df_imputado["Experiencia_Anios"] = df_imputado["Experiencia_Anios"].fillna(df_imputado["Experiencia_Anios"].mean())
df_imputado["Puntaje_Prueba"] = df_imputado["Puntaje_Prueba"].fillna(df_imputado["Puntaje_Prueba"].mean())

print("\n2. Dataset Imputado:")
display(df_imputado)

# Paso 2: Escalar variables numéricas
num_cols = ["Experiencia_Anios", "Puntaje_Prueba", "Pretension_Salarial"]
scaler = StandardScaler()
df_imputado[num_cols] = scaler.fit_transform(df_imputado[num_cols])

print("\n3. Dataset Limpio y Escalado (Listo para Modelado):")
display(df_imputado.round(2))
```
