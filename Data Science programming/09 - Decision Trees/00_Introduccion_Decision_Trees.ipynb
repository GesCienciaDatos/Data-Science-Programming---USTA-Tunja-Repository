## 1. Fundamentos: ¿Qué es un Árbol de Decisión? 🌲

Un **Árbol de Decisión (*Decision Tree*)** es un modelo de aprendizaje supervisado no paramétrico que particiona recursivamente el espacio de características en regiones rectangulares disjuntas. A cada región resultante se le asigna un valor predictivo constante:

* **Para Clasificación:** Se etiqueta cada región con la **clase mayoritaria (*majority voting*)** de los puntos de entrenamiento contenidos en ella.
* **Para Regresión:** Se asigna a cada región el **promedio aritmético (*mean*)** de los valores objetivo de las observaciones que caen en dicha región.

Estructuralmente, un árbol de decisión se compone de:
1. **Nodo Raíz (*Root Node*):** El punto de partida que contiene la totalidad de los datos y realiza la primera división.
2. **Nodos Internos (*Internal/Split Nodes*):** Evalúan una condición lógica binaria sobre una variable específica ($X_j \le t_j$).
3. **Ramas (*Branches*):** Conectan las decisiones según si la condición es Verdadera o Falsa.
4. **Nodos Hoja o Terminales (*Leaf Nodes*):** No poseen divisiones posteriores; albergan la predicción final asignada a cualquier muestra que alcance dicho nodo.

## 2. ¿Cómo se Construyen los Árboles? El Enfoque Goloso (*Greedy Approach*) 📐

Para entrenar un árbol de decisión, el algoritmo adopta una estrategia heurística conocida como (*greedy approach*)mediante partición binaria recursiva:

Algoritmo de Aprendizaje de un Árbol de Decisión</h3>
Inicio: Comenzar con un árbol vacío que abarca todo el espacio de características sin dividir.
Selección Óptima:</b> Elegir el predictor 'óptimo' $X_j$ y el umbral de corte 'óptimo' $t_j \in \mathbb{R}$ aplicando un criterio de división</b> (como la reducción de impureza de Gini, Entropía o MSE).
Partición Binaria: Dividir el nodo actual en dos regiones hijas ($R_1$ y $R_2$) según la regla $X_j \le t_j$.
Recursión:Repetir el proceso recursivamente en cada nuevo nodo secundario hasta satisfacer una condición de parada</b>.


> 💡 **¿Por qué es (greedy)?** En cada paso, el algoritmo toma la mejor decisión local inmediata (la división que más reduce la impureza en ese instante), sin verificar si esa división conducirá al árbol globalmente óptimo en pasos posteriores.

## 3. Ventajas y Limitaciones de los Árboles de Decisión ⚖️

| Dimensión | Ventajas Clave 🟢 | Desventajas y Limitaciones 🔴 |
|---|---|---|
| **Interpretabilidad** | Extraordinariamente intuitivos; las decisiones pueden representarse visualmente como un diagrama de flujo o reglas lógicas *if-then*. | Si el árbol crece demasiado, se vuelve ilegible y difícil de interpretar. |
| **Preparación de Datos** | Requieren muy poco preprocesamiento. No son sensibles a la escala de las variables (no necesitan estandarización ni normalización). | La implementación de Scikit-Learn requiere variables cuantitativas (no soporta directamente texto crudo sin encoding). |
| **No Linealidad** | Capturan relaciones e interacciones complejas no lineales entre variables de forma natural. | Solo crean fronteras ortogonales (paralelas a los ejes coordenados). |
| **Estabilidad** | Fáciles y rápidos de ajustar. | **Alta inestabilidad:** una pequeña perturbación en los datos de entrenamiento puede alterar radicalmente la estructura de todo el árbol. |
| **Riesgo de Sobreajuste** | Si se podan o ensamblan, ofrecen excelente capacidad de generalización. | Si no se restringen, **memorizan el ruido al 100%** (*overfitting* severo). |
| **Desbalance de Clases** | Soportan esquemas de ponderación de clases integrados. | Tienden a generar árboles sesgados a favor de la clase mayoritaria si no se equilibra el dataset. |

## 4. Manejo de Conjuntos de Datos Desbalanceados (*Class-Imbalanced Datasets*) ⚖️

Los algoritmos de Machine Learning pueden sesgarse gravemente si una clase domina sobre las demás (por ejemplo, en detección de fraude o diagnóstico médico, donde la clase positiva es muy minoritaria). Un árbol de decisión tenderá a clasificar todo en la clase mayoritaria para maximizar la exactitud aparente.

### Estrategias de Compensación en Scikit-Learn:
1. **Ponderación con `class_weight='balanced'`:** Ajusta automáticamente pesos inversamente proporcionales a las frecuencias de clase en los datos de entrada según la fórmula:
   $$w_k = \frac{N}{K \cdot N_k}$$
   donde $N$ es el total de muestras, $K$ el número de clases y $N_k$ la cantidad de muestras de la clase $k$.
2. **Ponderación por Muestra (`sample_weight`):** Permite asignar un vector de pesos individuales a cada fila en el método `.fit(X, y, sample_weight=pesos)`.

## 5. Implementación con Scikit-Learn: `DecisionTreeClassifier` y `DecisionTreeRegressor` 💻

Crear un árbol en Scikit-Learn es directo:
```python
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor

# Para Clasificación:
arbol_clf = DecisionTreeClassifier(max_depth=3, random_state=42)
arbol_clf.fit(X_train, y_train)

# Para Regresión:
arbol_reg = DecisionTreeRegressor(max_depth=3, random_state=42)
arbol_reg.fit(X_train, y_train)
```

A continuación, entrenamos un árbol de clasificación sobre el conjunto de datos `Iris` y visualizamos sus reglas tanto en texto como de forma gráfica:

````python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

# 1. Cargar dataset de flores Iris
iris = load_iris()
X_tr, X_ts, y_tr, y_ts = train_test_split(
    iris.data, iris.target, test_size=0.30, random_state=42, stratify=iris.target
)

# 2. Instanciar y ajustar árbol con profundidad controlada
modelo_iris = DecisionTreeClassifier(max_depth=3, criterion='gini', random_state=42)
modelo_iris.fit(X_tr, y_tr)

# 3. Evaluar exactitud
y_pred = modelo_iris.predict(X_ts)
print(f"Exactitud en Prueba (Iris): {accuracy_score(y_ts, y_pred) * 100:.2f}%")

# 4. Visualización en texto legible de las reglas lógicas
print("\n" + "="*50)
print("REGLAS LÓGICAS EXTRAÍDAS DEL ÁRBOL:")
print("="*50)
print(export_text(modelo_iris, feature_names=list(iris.feature_names)))
````

````python
# 5. Visualización gráfica del Árbol de Decisión
plt.figure(figsize=(14, 7), dpi=100)
plot_tree(
    modelo_iris,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Estructura Gráfica del Árbol de Decisión (Iris Dataset)", fontsize=14, fontweight='bold')
plt.show()
````
