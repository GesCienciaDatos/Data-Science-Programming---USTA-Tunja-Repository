## Objetivos de Aprendizaje 🎯

1. División Train / Test con `train_test_split` estratificado.
2. Validación Cruzada $K$-Fold con `KFold`, `StratifiedKFold`, `TimeSeriesSplit`.
3. Catálogo de Métricas en `sklearn.metrics`: `accuracy_score`, `precision_score`, `recall_score`, `f1_score`, `confusion_matrix`.

```python
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()
X_tr, X_te, y_tr, y_te = train_test_split(cancer.data, cancer.target, test_size=0.2, stratify=cancer.target, random_state=42)

print(f"Train samples: {len(X_tr)} | Test samples: {len(X_te)}")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
print("Folds estratificados generados:", skf.get_n_splits(X_tr, y_tr))
```
