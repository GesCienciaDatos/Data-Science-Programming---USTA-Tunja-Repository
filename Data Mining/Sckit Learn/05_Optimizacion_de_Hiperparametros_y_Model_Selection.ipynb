## Objetivos de Aprendizaje 🎯

1. Búsqueda en Grilla Exhaustiva con `GridSearchCV`.
2. Búsqueda Aleatoria Eficiente con `RandomizedSearchCV`.
3. Diagnóstico de Sesgo vs Varianza con `validation_curve` y `learning_curve`.

```python
import numpy as np
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.datasets import make_regression

X, y = make_regression(n_samples=200, n_features=3, noise=5.0, random_state=42)

pipe = Pipeline([
    ('poly', PolynomialFeatures()),
    ('scaler', StandardScaler())
])

param_grid = {'poly__degree': [1, 2, 3], 'poly__include_bias': [True, False]}
grid = GridSearchCV(pipe, param_grid, cv=3)
grid.fit(X, y)

print("Mejores parámetros:", grid.best_params_)
print("Mejor score CV    :", grid.best_score_)
```
