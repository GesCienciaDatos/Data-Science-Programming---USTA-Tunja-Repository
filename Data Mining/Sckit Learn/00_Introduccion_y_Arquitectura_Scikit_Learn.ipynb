## 1. Principios de Diseño e Inspección de Estimadores 📐

La arquitectura de Scikit-Learn expone los parámetros aprendidos durante `fit()` mediante atributos con sufijo de guion bajo (`_`).

```python
from sklearn.preprocessing import StandardScaler

X_demo = np.array([
    [10.0, 200.0],
    [20.0, 300.0],
    [30.0, 400.0],
    [40.0, 500.0],
    [50.0, 600.0]
])

scaler = StandardScaler(with_mean=True, with_std=True)
scaler.fit(X_demo)

print("Medias aprendidas (mu)        :", scaler.mean_)
print("Varianzas aprendidas (sigma^2):", scaler.var_)
print("Desviaciones estándar (sigma) :", scaler.scale_)

X_scaled = scaler.transform(X_demo)
print("\nMatriz Transformada:\n", X_scaled)
```
