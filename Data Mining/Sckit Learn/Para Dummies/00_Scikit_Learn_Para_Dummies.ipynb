## ¿Qué es Scikit-Learn en Español Sencillo? 🧩

Imagina que quieres armar un carro de juguete con bloques de **LEGO**. Cada bloque tiene una función específica:
* Un bloque son las ruedas (para escalar números).
* Otro bloque es la carrocería (para limpiar datos nulos).
* Otro bloque es el motor (para aprender patrones).

**Scikit-Learn** es exactamente eso: una caja gigante de bloques de LEGO de software diseñados para que todos encajen perfectamente entre sí.

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

edades = np.array([[10], [20], [30], [40]])
escalador = MinMaxScaler()
edades_0_a_1 = escalador.fit_transform(edades)

print("Edades Originales:\n", edades)
print("Edades convertidas de 0 a 1:\n", edades_0_a_1)
```
