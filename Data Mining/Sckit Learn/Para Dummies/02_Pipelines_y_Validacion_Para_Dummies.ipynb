## ¿Qué es un Pipeline? La Banda Transportadora Automática 🏭

Imagina una fábrica de botellas:
1. La botella entra en la banda.
2. La estación 1 lava la botella (`SimpleImputer`).
3. La estación 2 le pega la etiqueta (`StandardScaler`).

Un **Pipeline** en Scikit-Learn conecta todas estas estaciones para que nunca tengas que hacer los pasos manualmente uno por uno.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import numpy as np

banda = Pipeline([
    ('lavar', SimpleImputer(strategy='mean')),
    ('escalar', StandardScaler())
])

datos = np.array([[10], [np.nan], [30], [40]])
datos_procesados = banda.fit_transform(datos)
print("Datos procesados automáticamente:\n", datos_procesados)
```
