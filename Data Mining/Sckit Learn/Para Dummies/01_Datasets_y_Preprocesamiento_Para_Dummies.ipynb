## Limpiar Datos es como Cocinar una Receta 🍳

Antes de meter un pastel al horno, debes:
1. Sacar los ingredientes (**Cargar Datasets** con `load_iris`).
2. Rellenar los huecos si te falta harina (**Imputar Nulos** con `SimpleImputer`).
3. Pesar todo en la misma balanza (**Escalar Números** con `StandardScaler`).

```python
import numpy as np
from sklearn.impute import SimpleImputer

ingredientes = np.array([[100.0], [np.nan], [300.0], [400.0]])
filtro = SimpleImputer(strategy='mean')
ingredientes_listos = filtro.fit_transform(ingredientes)

print("Ingredientes listos y sin huecos:\n", ingredientes_listos)
```
