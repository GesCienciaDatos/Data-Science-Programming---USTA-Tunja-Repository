## Objetivos de Aprendizaje 🎯

1. Prevención sistemática de **Data Leakage** con `Pipeline`.
2. Ensamblaje heterogéneo por columnas con `ColumnTransformer`.
3. Serialización y persistencia con `joblib`.

```python
import pandas as pd, numpy as np, joblib, tempfile, os
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df_sample = pd.DataFrame({
    'Edad': [25.0, np.nan, 45.0, 35.0, 52.0],
    'Salario': [3200.0, 4800.0, np.nan, 5100.0, 8900.0],
    'Ciudad': ['Tunja', 'Bogotá', 'Tunja', 'Duitama', 'Bogotá']
})

num_pipe = Pipeline([('imp', SimpleImputer(strategy='median')), ('scl', StandardScaler())])
cat_pipe = Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('ohe', OneHotEncoder(sparse_output=False))])

preprocessor = ColumnTransformer([
    ('num', num_pipe, ['Edad', 'Salario']),
    ('cat', cat_pipe, ['Ciudad'])
])

X_out = preprocessor.fit_transform(df_sample)
print("Columnas de salida:", preprocessor.get_feature_names_out())
print("Matriz procesada:\n", X_out)
```
