## Objetivos de Aprendizaje 🎯

En este cuaderno aprenderás a utilizar el submódulo `sklearn.datasets` para:
1. Comprender la estructura y los métodos del objeto contenedor `Bunch`.
2. Cargar y explorar los conjuntos de datos clásicos de prueba (*Toy Datasets*): Iris, Wine, Diabetes, Digits y Breast Cancer.
3. Descargar conjuntos de datos del mundo real (*Real World Datasets*) como California Housing.
4. Generar conjuntos de datos sintéticos controlados con `make_classification`, `make_regression`, `make_blobs` y `make_moons`.
5. Convertir nativamente los datasets a DataFrames de `pandas` utilizando el parámetro `as_frame=True`.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.datasets import load_iris, fetch_california_housing, make_classification, make_blobs

# 1. Carga de Iris con Pandas DataFrame nativo
iris = load_iris(as_frame=True)
print("Iris shape:", iris.frame.shape)
print("Columnas  :", list(iris.frame.columns))
iris.frame.head()
```

```python
# 2. Generadores Sintéticos Controlados
X_clf, y_clf = make_classification(n_samples=250, n_features=2, n_informative=2, n_redundant=0, random_state=42)
X_blobs, y_blobs = make_blobs(n_samples=250, centers=3, random_state=42)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(X_clf[:, 0], X_clf[:, 1], c=y_clf, cmap='coolwarm', edgecolor='k', alpha=0.8)
axes[0].set_title("make_classification")
axes[1].scatter(X_blobs[:, 0], X_blobs[:, 1], c=y_blobs, cmap='viridis', edgecolor='k', alpha=0.8)
axes[1].set_title("make_blobs (3 Clusters)")
plt.tight_layout()
plt.show()
```
