## 1. La Gran Analogía: El Cuaderno de Ejercicios ya Corregido 📖✅

En el colegio, un buen cuaderno de ejercicios resueltos te muestra el problema *y* la respuesta correcta. Estudias comparando tu propio intento con esa respuesta, ajustas lo que entendiste mal, y repites con el siguiente ejercicio. Eso es justo lo que hace un algoritmo de **aprendizaje supervisado**: recibe una tabla de datos donde cada fila ya tiene su "respuesta correcta" (la **variable objetivo**), y ajusta sus reglas internas comparando lo que predice contra esa respuesta.

Pero no todas las "respuestas correctas" son del mismo tipo. En este módulo vas a trabajar con dos ejemplos reales, cada uno con un tipo de respuesta distinto:

| | ¿Qué tipo de respuesta busca? | Pregunta típica | Ejemplo de este cuaderno |
|---|---|---|---|
| **Clasificación** | Una categoría (una etiqueta entre pocas opciones) | ¿En qué cajón cae esto? | `clientes_credito.csv`: ¿el cliente tiene `riesgo_default` (Sí/No)? |
| **Regresión** | Un número (puede tomar cualquier valor dentro de un rango) | ¿Qué cantidad va aquí? | `inmuebles_precios.csv`: ¿cuánto vale el inmueble en `precio_millones`? |

Todo lo que sigue en este cuaderno aplica **por igual** a los dos casos: no importa si al final se predice una categoría o un número, las reglas del juego (cómo se estudia, cómo se examina, qué es memorizar vs. entender) son exactamente las mismas.

````python
import os, sys, urllib.request, urllib.parse, gzip, base64

_DATA_EMBEDDED = {
    "clientes_credito.csv": "",
    "inmuebles_precios.csv": "",
    "transacciones_fraude.csv": "",
}

def load_dataset(filename: str) -> str:
    candidates = [
        os.path.join("data", filename),
        os.path.join("..", "data", filename),
        os.path.join("02 - Clasificacion y Regresion", "data", filename),
        os.path.join("Data Mining", "02 - Clasificacion y Regresion", "data", filename),
        os.path.join("..", "02 - Clasificacion y Regresion", "data", filename),
        filename,
    ]
    for p in candidates:
        if os.path.exists(p):
            return p
    os.makedirs("data", exist_ok=True)
    target = os.path.join("data", filename)
    base_url = "https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/Data%20Mining/02%20-%20Clasificacion%20y%20Regresion/data/"
    url = base_url + urllib.parse.quote(filename)
    try:
        urllib.request.urlretrieve(url, target, timeout=3)
        return target
    except Exception:
        pass
    if filename in _DATA_EMBEDDED:
        raw = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target, "wb") as f:
            f.write(raw)
        return target
    raise FileNotFoundError(f"No se pudo obtener el dataset: {filename}")

import pandas as pd
import os

credito = pd.read_csv(load_dataset("clientes_credito.csv"))
inmuebles = pd.read_csv(load_dataset("inmuebles_precios.csv"))

print(f"clientes_credito.csv  -> {credito.shape[0]} clientes, objetivo: riesgo_default (categoria)")
print(f"inmuebles_precios.csv -> {inmuebles.shape[0]} inmuebles, objetivo: precio_millones (numero)")

print("\nCategorias de riesgo_default (0 = sin riesgo, 1 = riesgo de incumplir):")
print(credito["riesgo_default"].value_counts(normalize=True).round(3))

print("\nRango de precio_millones (numero continuo):")
print(inmuebles["precio_millones"].describe().round(2))
````

### 🤔 ¿Qué acaba de pasar?
`riesgo_default` solo puede valer **0 o 1** — dos cajones posibles, ni uno más: eso es clasificación. `precio_millones` en cambio puede ser 58.69, 336.18, 1343.15... cualquier número dentro de un rango: eso es regresión. Fíjate que `riesgo_default` está casi perfectamente repartido (52% / 48%), así que ninguna de las dos categorías es "rara" en este dataset; más adelante, en otros cuadernos del curso, verás ejemplos donde una categoría es muchísimo más escasa que la otra.

## 2. El Examen Sorpresa: por qué no puedes calificarte con lo que ya estudiaste 📝

Imagina que estudias para un examen con un libro de ejercicios... y luego el profesor te examina **con las mismas preguntas exactas** del libro. Sacarías una nota perfecta, pero esa nota no diría nada sobre si realmente aprendiste algo o solo memorizaste las respuestas.

Por eso, antes de entrenar cualquier modelo, se separan los datos en dos partes:

* **Conjunto de entrenamiento (`train`):** el "libro de ejercicios" — el modelo lo ve y aprende de él.
* **Conjunto de prueba (`test`):** el "examen sorpresa" — el modelo **nunca** lo ve durante el entrenamiento, así que evaluarlo ahí sí dice algo real sobre qué tan bien generaliza.

En `scikit-learn`, `train_test_split` hace exactamente esa separación por ti. La siguiente celda la usa sobre `clientes_credito.csv` y entrena un modelo sencillo para comparar la nota que saca "estudiando" contra la que saca en el "examen sorpresa".

````python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score

columnas_predictoras = ["edad", "ingreso_mensual", "antiguedad_laboral_meses", "score_credito", "ratio_deuda_ingreso"]
X = credito[columnas_predictoras]
y = credito["riesgo_default"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

print(f"Filas para 'estudiar' (train): {X_train.shape[0]}")
print(f"Filas para el 'examen sorpresa' (test): {X_test.shape[0]}")

modelo = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
modelo.fit(X_train, y_train)

nota_estudio = accuracy_score(y_train, modelo.predict(X_train))
nota_examen = accuracy_score(y_test, modelo.predict(X_test))

print(f"\n'Nota' en los datos que ya vio (train): {nota_estudio:.4f}")
print(f"'Nota' en el examen sorpresa (test):     {nota_examen:.4f}")
````

### 🤔 ¿Qué acaba de pasar?
De los 1200 clientes, 900 (75%) se usaron para "estudiar" y 300 (25%) se guardaron como examen sorpresa que el modelo jamás vio durante el entrenamiento. La nota en el examen sorpresa (0.7267) es apenas un poco más baja que la nota estudiando (0.7511) — una diferencia pequeña y sana. Si el modelo hubiera memorizado el libro de ejercicios en vez de entenderlo, verías una nota casi perfecta en `train` y una nota mucho peor en `test`. Justo ese fenómeno es el que exploras a fondo en la siguiente sección.

## 3. Memorizar vs. Entender: el Dilema Sesgo-Varianza 🧠

Piensa en dos tipos de estudiante:

* **El que memoriza:** se aprende de memoria cada pregunta y respuesta exacta del libro. En un examen con preguntas idénticas, saca 10. Pero si el examen cambia un poco las preguntas, se hunde — nunca entendió el patrón, solo copió los casos puntuales.
* **El que entiende el patrón:** capta la idea general detrás de los ejercicios. Puede no sacar 10 perfecto en el libro (porque no memorizó cada detalle), pero le va bien en preguntas nuevas porque de verdad entendió el concepto.

En Machine Learning, "el que memoriza" es un modelo con **alta varianza** (demasiado flexible, se ajusta al ruido de los datos de entrenamiento) y "el que no entiende nada" es un modelo con **alto sesgo** (demasiado simple, ni siquiera capta el patrón real). El objetivo siempre es encontrar un punto intermedio. La siguiente celda lo demuestra con dos modelos reales sobre `inmuebles_precios.csv`, usando solo el área del inmueble para exagerar el efecto.

````python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score

X_area = inmuebles[["area_m2"]]
y_precio = inmuebles["precio_millones"]
Xa_train, Xa_test, ya_train, ya_test = train_test_split(X_area, y_precio, test_size=0.25, random_state=42)

# El estudiante que "entiende el patron general": una recta simple
modelo_simple = LinearRegression().fit(Xa_train, ya_train)
r2_simple_train = r2_score(ya_train, modelo_simple.predict(Xa_train))
r2_simple_test = r2_score(ya_test, modelo_simple.predict(Xa_test))

# El estudiante que "memoriza": copia el precio del vecino mas parecido (k=1)
modelo_memoriza = KNeighborsRegressor(n_neighbors=1).fit(Xa_train, ya_train)
r2_memoriza_train = r2_score(ya_train, modelo_memoriza.predict(Xa_train))
r2_memoriza_test = r2_score(ya_test, modelo_memoriza.predict(Xa_test))

print("Modelo simple (LinearRegression):")
print(f"  R2 estudiando (train) = {r2_simple_train:.4f}   R2 examen sorpresa (test) = {r2_simple_test:.4f}")
print("\nModelo que memoriza (KNN con k=1):")
print(f"  R2 estudiando (train) = {r2_memoriza_train:.4f}   R2 examen sorpresa (test) = {r2_memoriza_test:.4f}")
````

### 🤔 ¿Qué acaba de pasar?
El modelo que "memoriza" (k=1) obtiene un R² decente estudiando, copiando el precio del inmueble más parecido que ya vio (0.4736) — pero en el examen sorpresa su R² se vuelve **negativo (-0.5786)**, peor que si simplemente hubiera respondido "el precio promedio" para todo. Memorizó los inmuebles del libro de ejercicios, no aprendió qué hace que un inmueble valga más o menos. El modelo simple (una recta) tiene un R² más modesto tanto en train (0.2809) como en test (0.2743) — casi idéntico entre los dos — porque nunca tuvo la flexibilidad para memorizar nada: lo que ves en el examen es prácticamente lo mismo que viste estudiando, aunque el nivel general sea más bajo. Ningún R² aquí es "bueno" todavía (solo se usó una variable), pero el patrón —memorizar vs. entender— queda clarísimo en la brecha entre train y test.

## 4. Varios Exámenes Sorpresa en Vez de Uno Solo: Validación Cruzada 🔁

Un solo examen sorpresa puede tener, por pura casualidad, preguntas más fáciles o más difíciles que el promedio. Si ese único examen te tocó "fácil", tu nota final se ve mejor de lo que en realidad mereces; si te tocó "difícil", peor. ¿Cómo estar más seguro de tu nota real? **Haciendo varios exámenes sorpresa distintos y promediando el resultado.**

Eso es la **validación cruzada**: en vez de partir los datos una sola vez, se parten en varios grupos ("folds"). Por turnos, cada grupo hace de examen sorpresa mientras el modelo estudia con los demás. Al final tienes varias notas — no solo una — y puedes calcular su promedio y qué tanto varían entre sí.

````python
from sklearn.model_selection import cross_val_score, KFold

print("Nota del examen sorpresa segun que datos le tocaron (una sola particion, distintas semillas):")
for semilla in range(5):
    Xs_train, Xs_test, ys_train, ys_test = train_test_split(X, y, test_size=0.25, random_state=semilla, stratify=y)
    m = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
    m.fit(Xs_train, ys_train)
    print(f"  semilla={semilla}: nota = {m.score(Xs_test, ys_test):.4f}")

modelo_cv = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
kf = KFold(n_splits=5, shuffle=True, random_state=42)
notas_cv = cross_val_score(modelo_cv, X, y, cv=kf, scoring="accuracy")

print(f"\nCinco examenes sorpresa distintos (5-fold): {notas_cv.round(4)}")
print(f"Promedio: {notas_cv.mean():.4f}   Variacion entre examenes (desviacion estandar): {notas_cv.std():.4f}")
````

### 🤔 ¿Qué acaba de pasar?
Con un único examen sorpresa, la nota varió entre 0.70 y 0.77 solo por cambiar la semilla aleatoria que decide qué filas caen en el examen — ¡una diferencia de 7 puntos porcentuales sin cambiar nada del modelo! Al hacer 5 exámenes sorpresa distintos (5-fold) y promediar, se obtiene una nota más confiable (0.7458) junto con una medida de qué tanto varía esa nota entre exámenes (0.0137, bastante poco). Esa es la ventaja real de la validación cruzada: no solo da un promedio más estable, sino que además te dice qué tan seguro puedes estar de ese promedio.
