## ¿Qué vamos a aprender aquí? 🕵️

Imagina que estás jugando a las **20 preguntas**: alguien piensa en un cliente de una empresa de telecomunicaciones, y tú tienes que adivinar si ese cliente se va a ir (*churn*) o se va a quedar, haciendo solamente preguntas de sí o no. Eso —ni más ni menos— es lo que hace un **árbol de decisión**.

Al terminar este cuaderno vas a poder responder:
1. ¿Cómo "juega" un árbol de decisión a las 20 preguntas con datos reales?
2. ¿Por qué es tan fácil explicarle una decisión del árbol a alguien sin formación técnica?
3. ¿Por qué dos árboles entrenados con datos parecidos pueden terminar haciendo preguntas distintas?

## 1. La Gran Analogía: El Juego de las 20 Preguntas 🕵️‍♀️

En las 20 preguntas, ganas rápido si haces las preguntas correctas en el orden correcto: primero las que más dividen las posibilidades, después las que afinan el detalle. Un árbol de decisión hace exactamente eso con los datos de un cliente:

* **Pregunta 1:** ¿Tiene contrato *Mes a Mes* (sin compromiso de permanencia)? → Si NO, ya es mucho menos probable que se vaya.
* **Pregunta 2 (solo si respondió que sí a la 1):** ¿Lleva pocos meses como cliente (por ejemplo, 24 meses o menos)? → Si SÍ, sube todavía más el riesgo.
* **Pregunta 3 (solo si respondió que sí a la 1 y a la 2):** ¿Paga una factura mensual alta? → Si SÍ, el riesgo de que se vaya es el más alto de todos.

Cada pregunta reduce el universo de clientes posibles, igual que en el juego. Al final de la cadena de preguntas llegas a un grupo de clientes muy parecidos entre sí, y le asignas a todo ese grupo la misma respuesta ("se va" o "se queda") — esa es la predicción del árbol. En la jerga técnica, cada pregunta es un **nodo**, la primera pregunta es la **raíz**, y el grupo final donde ya no se pregunta más es una **hoja**.

---
## 2. Construyamos un árbol de verdad con clientes reales 📡

Vamos a usar datos reales de 2500 clientes de telecomunicaciones (`abandono_clientes_telecom.csv`) para dejar que un árbol de decisión "juegue" a las 20 preguntas de verdad, en vez de solo imaginarlo.

```python
import os, urllib.request, urllib.parse, gzip, base64

_DATA_EMBEDDED = {
    "abandono_clientes_telecom.csv": "",
}

def load_dataset(filename: str) -> str:
    candidates = [
        os.path.join("data", filename),
        os.path.join("..", "data", filename),
        os.path.join("04 - Arboles de Decision y Bosques Aleatorios", "data", filename),
        os.path.join("Data Mining", "04 - Arboles de Decision y Bosques Aleatorios", "data", filename),
        os.path.join("..", "04 - Arboles de Decision y Bosques Aleatorios", "data", filename),
        filename,
    ]
    for p in candidates:
        if os.path.exists(p):
            return p
    os.makedirs("data", exist_ok=True)
    target = os.path.join("data", filename)
    base_url = "https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/Data%20Mining/04%20-%20Arboles%20de%20Decision%20y%20Bosques%20Aleatorios/data/"
    url = base_url + urllib.parse.quote(filename)
    try:
        urllib.request.urlretrieve(url, target, timeout=3)
        return target
    except Exception:
        pass
    if filename in _DATA_EMBEDDED:
        raw = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target, "wb") as f:
            f.write(raw)
        return target
    raise FileNotFoundError(f"No se pudo obtener el dataset: {filename}")

import pandas as pd, numpy as np, matplotlib.pyplot as plt, os
from sklearn.tree import DecisionTreeClassifier, plot_tree

clientes = pd.read_csv(load_dataset("abandono_clientes_telecom.csv"))
print(f"Clientes cargados: {clientes.shape[0]} filas, {clientes.shape[1]} columnas")
print(clientes[["antiguedad_meses", "cargo_mensual", "tipo_contrato", "churn"]].head(3))

print("\nDe cada 100 clientes, cuantos se van (churn=1) sin hacer ninguna pregunta todavia?")
print((clientes["churn"].value_counts(normalize=True) * 100).round(2))
```

### 🤔 ¿Qué acaba de pasar?
Cargamos los datos de 2500 clientes. Solo con estas columnas ya se pueden imaginar preguntas útiles: `tipo_contrato`, `antiguedad_meses`, `cargo_mensual`. Y confirmamos el punto de partida del "juego": sin hacer ninguna pregunta, el 28.88% de los clientes se va — ese es el número que el árbol va a intentar mejorar con sus preguntas. Antes de construir el árbol falta un paso técnico: convertir las columnas de texto (como `tipo_contrato`) en números, porque `scikit-learn` no puede hacer preguntas directamente sobre palabras.

```python
datos = clientes.copy()
datos["contrato_mes_a_mes"] = (datos["tipo_contrato"] == "Mes_a_Mes").astype(int)
datos["factura_electronica_num"] = datos["factura_electronica"].map({"No": 0, "Si": 1})
datos["tiene_soporte"] = (datos["soporte_tecnico"] == "Si").astype(int)

columnas_modelo = ["antiguedad_meses", "cargo_mensual", "cargos_totales",
                    "contrato_mes_a_mes", "factura_electronica_num", "tiene_soporte"]
X = datos[columnas_modelo]
y = datos["churn"]

print(X.head(3))
```

### 🤔 ¿Qué acaba de pasar?
Convertimos tres columnas de texto en números de 0 y 1 (por ejemplo, `contrato_mes_a_mes` vale 1 si el cliente tiene ese tipo de contrato, y 0 si no). El árbol sigue "preguntando" exactamente lo mismo que antes, solo que ahora la pregunta se ve como `contrato_mes_a_mes <= 0.5` en lugar de `¿tipo_contrato es "Mes_a_Mes"?`. *(Nota honesta: `soporte_tecnico` en realidad tiene tres categorías —`No`, `Si` y `Sin_Internet`— pero para mantener este cuaderno simple la resumimos en "sí tiene / no tiene"; el cuaderno técnico completo del módulo sí distingue las tres.)*

---
## 3. El primer árbol: dejando que el algoritmo elija las preguntas 🌳

Vamos a limitar el árbol a apenas 2 niveles de preguntas (`max_depth=2`) para poder verlo completo de un vistazo. El algoritmo que usa `scikit-learn` para elegir automáticamente "las mejores preguntas" se llama **CART**, y en cada paso prueba muchísimas preguntas posibles y se queda con la que mejor separa a los clientes que se van de los que se quedan.

```python
arbol = DecisionTreeClassifier(max_depth=2, random_state=42)
arbol.fit(X, y)

print(f"Exactitud del arbol sobre estos mismos clientes: {arbol.score(X, y) * 100:.2f}%")
print(f"Numero de preguntas (nodos de division): {arbol.tree_.node_count - arbol.get_n_leaves()}")
print(f"Numero de grupos finales (hojas): {arbol.get_n_leaves()}")
```

### 🤔 ¿Qué acaba de pasar?
El árbol, sin que nadie le dijera qué preguntar, decidió por sí mismo hacer solo 2 rondas de preguntas y terminó acertando el 74.16% de las veces sobre estos mismos clientes. Con apenas 4 grupos finales (hojas), ya mejora bastante el 71.12% de acierto que tendrías si simplemente dijeras "nadie se va" para todos.

---
## 4. Viendo el árbol completo 👀

La ventaja de un árbol pequeño es que se puede dibujar completo y leerlo de un vistazo.

```python
plt.figure(figsize=(14, 6))
plot_tree(arbol, feature_names=columnas_modelo, class_names=["Se_queda", "Se_va"],
          filled=True, rounded=True, fontsize=9)
plt.title("Nuestro primer arbol -- prediciendo el abandono de clientes")
plt.show()
```

### 🤔 ¿Qué acaba de pasar?
Ahí está el árbol completo. Traduzcamos dos caminos reales a lenguaje de gerente, sin tecnicismos:

* **Camino de ALTO riesgo:** *"Si el cliente tiene contrato mes a mes Y lleva 24.5 meses o menos como cliente, entonces (de 472 clientes con ese perfil) el 58.05% tiene probabilidad de irse."* Más de la mitad se va — un grupo prioritario para cualquier campaña de retención.
* **Camino de BAJO riesgo:** *"Si el cliente NO tiene contrato mes a mes Y lleva más de 24.5 meses como cliente, entonces (de 754 clientes con ese perfil) apenas el 5.97% tiene probabilidad de irse."* Ese grupo casi no necesita atención especial.

Nadie tuvo que programar estas reglas a mano: el árbol las descubrió solo, probando preguntas hasta encontrar las que mejor separaban a un grupo del otro.

## 5. ¿Por qué un árbol es tan fácil de explicar? 💬

Compara lo que acabas de leer arriba con la típica respuesta de un modelo "caja negra" (como una red neuronal compleja): *"el modelo le asignó a este cliente una probabilidad de 0.58 de irse, calculada combinando cientos de números internos sin un significado directo."* Técnicamente correcto, pero inútil para explicárselo a un gerente o a un cliente.

Con un árbol, en cambio, la explicación ES la regla: "se va porque tiene contrato mes a mes y es nuevo" es algo que cualquier persona entiende, cuestiona y hasta puede accionar (por ejemplo, ofreciéndole a ese cliente un descuento por firmar un contrato de un año). Esa capacidad de mostrar el "por qué" en lenguaje humano, sin caja negra de por medio, es la razón número uno por la que los árboles de decisión siguen siendo tan populares en negocios, medicina y otros campos donde alguien necesita justificar una decisión.

## 6. El lado incómodo: dos "grupos de estudiantes" pueden concluir cosas distintas 🎓

Imagina que le das el mismo cuestionario a dos grupos distintos de estudiantes que estudiaron exactamente el mismo tema. Sería raro que los dos grupos llegaran a EXACTAMENTE las mismas conclusiones sobre qué era lo más importante — sobre todo si nadie les puso un límite de qué tan detallados podían ser sus argumentos.

A los árboles de decisión les pasa algo parecido: si entrenas el mismo algoritmo dos veces, sobre dos muestras distintas (pero parecidas) de los mismos clientes, puede terminar haciendo preguntas distintas — sobre todo en las preguntas que vienen después de la primera. Vamos a comprobarlo entrenando el mismo árbol sobre dos muestras remuestreadas de los mismos 2500 clientes.

```python
fig, ejes = plt.subplots(1, 2, figsize=(18, 6))

for eje, semilla in zip(ejes, [0, 1]):
    rng = np.random.RandomState(semilla)
    idx_muestra = rng.choice(len(X), size=len(X), replace=True)
    X_muestra, y_muestra = X.iloc[idx_muestra], y.iloc[idx_muestra]

    arbol_muestra = DecisionTreeClassifier(max_depth=2, random_state=42)
    arbol_muestra.fit(X_muestra, y_muestra)

    plot_tree(arbol_muestra, feature_names=columnas_modelo, class_names=["Se_queda", "Se_va"],
              filled=True, rounded=True, fontsize=8, ax=eje)
    eje.set_title(f"Grupo de estudiantes {semilla + 1}")

plt.tight_layout()
plt.show()
```

### 🤔 ¿Qué acaba de pasar?
Ambos árboles se entrenaron con el mismo algoritmo, sobre muestras muy parecidas (remuestreadas de los mismos 2500 clientes), y ambos empiezan igual: primero preguntan si el contrato es mes a mes. Pero fíjate en la SEGUNDA pregunta, dentro del grupo con contrato mes a mes: el "Grupo de estudiantes 1" pregunta por `cargo_mensual` (cuánto paga cada mes), mientras que el "Grupo de estudiantes 2" pregunta por `antiguedad_meses` (cuántos meses lleva como cliente). Son preguntas distintas sobre el mismo grupo de clientes, y las dos parecen igual de razonables — el algoritmo encontró un "empate técnico" entre ambas, y una muestra apenas distinta bastó para inclinar la balanza hacia un lado u otro.

Con solo 2 niveles el efecto todavía es manejable. Pero en el cuaderno técnico de este módulo se muestra que, si al árbol no se le pone NINGÚN límite de profundidad, este problema se dispara: ese árbol logra memorizar perfectamente a los 2500 clientes de entrenamiento (100% de acierto) pero, frente a clientes nuevos que nunca vio, su acierto cae a apenas 69.28% — peor que el árbol sencillo de este cuaderno. Es la versión extrema de "estudiar de memoria las respuestas del examen anterior" en lugar de entender el tema: funciona perfecto en el examen que ya viste, y falla en el que no.

## Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **Árbol de decisión** | Juega a las 20 preguntas con los datos: encadena preguntas de sí/no hasta llegar a una predicción. |
| **Raíz, nodos, hojas** | La raíz es la primera pregunta; cada nodo es una pregunta nueva; las hojas son los grupos finales con la predicción. |
| **CART** | El algoritmo que elige automáticamente, en cada paso, cuál es "la mejor pregunta posible" en ese momento. |
| **Ventaja: interpretabilidad** | Cada predicción se explica con una regla en español simple — nada de caja negra. |
| **Debilidad: inestabilidad** | Muestras parecidas de los mismos datos pueden llevar al árbol a hacer preguntas distintas, sobre todo sin límite de profundidad. |
