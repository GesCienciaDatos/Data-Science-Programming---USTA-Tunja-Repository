## ¿Qué vamos a aprender aquí? 🎈

Cuando un cliente o tu jefe se acerca a pedirte un proyecto con datos, casi siempre te pedirá uno de **cinco tipos de magia diferentes**.

En este cuaderno conocerás a los 5 superhéroes de la minería de datos y aprenderás a identificar al instante a cuál de ellos llamar según el problema. Al final, pondrás a prueba tu instinto de detective con un catálogo real de 8 proyectos de minería de datos.

---
## 1. La Galería de los 5 Superhéroes 🦸‍♀️

### 1. El Clasificador (El Seleccionador de Sombrero) 🎩
* **Qué hace:** Responde preguntas de Sí/No o mete las cosas en cajas con etiquetas.
* **Ejemplo:** ¿Este correo es Spam o No Spam? ¿Este tumor es benigno o maligno?

### 2. El Adivino de Números (El Regresor) 📈
* **Qué hace:** No adivina una categoría, ¡adivina un número exacto continuo!
* **Ejemplo:** ¿A cuántos millones de pesos se venderá este apartamento en Tunja el próximo mes?

### 3. El Agrupador Misterioso (Clustering) 🧩
* **Qué hace:** Junta a personas o cosas que se parecen entre sí, sin que nadie le haya dicho las etiquetas de antemano.
* **Ejemplo:** Spotify organizando a sus millones de oyentes en tribus musicales (los rockeros de los 80s, los amantes del reggaetón nocturno).

### 4. El Detective de Parejas (Reglas de Asociación) 🛒
* **Qué hace:** Descubre qué cosas viajan siempre juntas en el carrito de compras.
* **Ejemplo:** "Si alguien compra café y leche, hay un 85% de probabilidad de que también lleve pan".

### 5. El Perro Guardián (Detección de Anomalías) 🚨
* **Qué hace:** Ladra cuando algo huele raro o se sale completamente de lo normal.
* **Ejemplo:** Tu tarjeta de crédito se usa a las 3:00 am en Rusia para comprar 5 iPhones, cuando tú estás durmiendo en Boyacá.

## 2. ¿Cómo Saber a Cuál Llamar? El Framework de la Pregunta Clave 🔑

No necesitas memorizar a los 5 superhéroes de memoria: basta con hacerte una pregunta y dejar que ella te lleve al héroe correcto.

| Tu pregunta interna | Superhéroe a invocar |
|---|---|
| "¿La respuesta es una categoría (Sí/No, Spam/No Spam)?" | El Clasificador 🎩 |
| "¿La respuesta es un número exacto (precio, cantidad, tiempo)?" | El Adivino de Números 📈 |
| "¿Quiero descubrir grupos parecidos, sin que nadie me diga las categorías de antemano?" | El Agrupador Misterioso 🧩 |
| "¿Quiero saber qué cosas aparecen siempre juntas?" | El Detective de Parejas 🛒 |
| "¿Quiero que algo me avise cuando aparezca lo rarísimo, lo que no encaja?" | El Perro Guardián 🚨 |

> 💡 **El truco maestro:** la pregunta madre es "¿ya tengo ejemplos pasados con la respuesta correcta anotada (una etiqueta)?". Si la respuesta es SÍ, buscas al Clasificador o al Adivino de Números (tareas **supervisadas**). Si es NO, buscas al Agrupador, al Detective o al Perro Guardián (tareas **no supervisadas**), porque ahí nadie te dio la respuesta de antemano.

## 4. Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **¿Etiqueta histórica?** | Si sí, es tarea supervisada: Clasificación (categoría) o Regresión (número). |
| **¿Sin etiqueta previa?** | Es tarea no supervisada: Clustering, Reglas de Asociación o Anomalías. |
| **Clasificación** | Responde preguntas de "¿A o B?" con ejemplos ya etiquetados. |
| **Regresión** | Adivina un número exacto, no una categoría. |
| **Clustering** | Descubre grupos parecidos sin que nadie diga las categorías de antemano. |
| **Reglas de Asociación** | Encuentra combos que viajan juntos ("si compra X, también compra Y"). |
| **Detección de Anomalías** | Ladra cuando algo se sale por completo de lo normal. |
