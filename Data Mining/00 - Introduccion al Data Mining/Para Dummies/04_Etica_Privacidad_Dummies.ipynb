## 1. La Falsa Seguridad: Borrar el Nombre no es Suficiente 🙅‍♂️

Imagina que en un pueblo pequeño alguien dice:
> *"No voy a decir el nombre por respeto, pero la persona que se enfermó es el único pelirrojo de 2 metros que vive frente a la iglesia".*

¿Protegiste su privacidad? ¡Para nada! Todo el pueblo sabe exactamente de quién estás hablando.

En datos pasa igual: aunque borres el nombre y la cédula, si dejas el código postal, la edad y el género, **puedes identificar con nombre y apellido al 87% de las personas**.

A esos datos "inocentes" (dónde vives, tu edad, tu color de cabello) que juntos te delatan, los expertos los llaman **cuasi-identificadores**. Ya volveremos sobre ellos en la Sección 4.

## 2. Dos Formas de Disfrazar a Alguien: el Diario Tachado vs. el Diario con Apodo 🕶️📔

Imagina un diario de secretos de un salón de clase. Hay dos formas de protegerlo:

* **Diario Tachado (Anonimización):** Tachas los nombres con marcador negro, para siempre. Nadie —ni tú mismo— puede volver a saber de quién es cada secreto. Es la protección más fuerte, pero también pierdes la posibilidad de, por ejemplo, seguir la historia de una sola persona a lo largo de varias páginas.
* **Diario con Apodo (Seudonimización):** En lugar de tachar, le pones a cada quien un apodo fijo (ej. "Estudiante-7"). No sabes quién es "Estudiante-7" en la vida real, pero cada vez que esa persona escribe, aparece con el mismo apodo. Puedes seguir su historia completa sin saber su nombre — y si alguien con la "llave secreta" del apodo necesita revertirlo (por una orden legal, por ejemplo), es posible.

En minería de datos casi siempre usamos la **seudonimización**: queremos poder decir "este cliente compró tres veces este mes" sin saber quién es exactamente esa persona.

## 3. La Solución del Detective: La Seudonimización 🔐

En lugar de usar el documento de identidad real, lo convertimos en un código secreto enredado (un *Hash*) que no se puede descifrar hacia atrás. Vamos a probarlo con clientes reales del banco que ya conoces del cuaderno anterior:

```python
import os, urllib.request, urllib.parse, gzip, base64

_DATA_EMBEDDED = {
    "auditoria_calidad_datos.csv": ""
}

def load_dataset(filename, module_name="00 - Introduccion al Data Mining"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo restaura automáticamente de forma exacta.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("..", "..", "data", filename),
        os.path.join("Data Mining", module_name, "data", filename),
        os.path.join("..", "Data Mining", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path

    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Mining/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)

    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]

    print(f"[INFO] Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=2) as response:
                if response.status == 200:
                    with open(target_path, "wb") as out_f:
                        out_f.write(response.read())
                    print(f"[OK] Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue

    if filename in _DATA_EMBEDDED:
        print(f"[INFO] Restaurando '{filename}' de forma autónoma...")
        data_bytes = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target_path, "wb") as out_f:
            out_f.write(data_bytes)
        print(f"[OK] Dataset '{filename}' preparado exitosamente en '{target_path}'.")
        return target_path

    raise FileNotFoundError(f"No se pudo encontrar ni generar el dataset '{filename}'.")


import os
import hashlib
import pandas as pd

def disfrazar_cedula(identificador):
    # Agregamos una palabra secreta (sal) para que nadie pueda hackearlo
    secreto = f"{identificador}_secreto_usta"
    disfraz = hashlib.sha256(secreto.encode()).hexdigest()[:10]
    return f"AGENTE-{disfraz}"

ruta_clientes = load_dataset("auditoria_calidad_datos.csv")
df_clientes = pd.read_csv(ruta_clientes)

muestra = df_clientes[['id_cliente']].head(5).copy()
muestra['id_disfrazado'] = muestra['id_cliente'].apply(disfrazar_cedula)

print("Identificadores reales vs. disfrazados:")
display(muestra)
print(f"\n¿El mismo cliente siempre recibe el mismo disfraz? {disfrazar_cedula('CLI-00001') == disfrazar_cedula('CLI-00001')}")
```

### 🤔 ¿Para qué sirve esto?
El científico de datos puede hacer toda su analítica y saber que el cliente `AGENTE-4f9a1c8b2d` gastó tal cantidad de dinero, pero jamás sabrá cuál es su verdadero `id_cliente` sin la palabra secreta. Y como el mismo `id_cliente` siempre produce el mismo disfraz, ¡se puede seguir analizando a esa persona sin descubrir jamás quién es!

## 4. ¿Cuántos Vecinos Comparten tus Mismos Datos? — El Secreto del k-Anonimato 🔍👥

Volvamos al pueblo de la Sección 1. Si eres el **único** pelirrojo de 2 metros frente a la iglesia, cualquiera te reconoce con solo dos datos. Pero si en el pueblo viven **50 personas** con esa misma descripción, estás escondido entre la multitud: nadie puede señalar a una persona específica.

A ese número de "compañeros indistinguibles" los expertos lo llaman **k**. Se dice que una base de datos cumple **k-anonimato** cuando toda combinación de cuasi-identificadores (ej. edad + ciudad) la comparten, como mínimo, `k` personas.

Vamos a medir el k real de nuestra base del banco, usando `edad` y `ciudad` como cuasi-identificadores:

```python
df_priv = df_clientes[['id_cliente', 'edad', 'ciudad']].dropna()
k_deseado = 5  # queremos que cada combinación la compartan al menos 5 personas

# ¿Cuántas personas comparten exactamente la misma edad y la misma ciudad?
grupos = df_priv.groupby(['edad', 'ciudad']).size().reset_index(name='cuantos_comparten')
vulnerables = (grupos['cuantos_comparten'] < k_deseado).sum()
print(f"Con la edad EXACTA: {vulnerables} de {len(grupos)} combinaciones tienen menos de {k_deseado} personas (fáciles de reidentificar)")
display(grupos.sort_values('cuantos_comparten').head(5))

# Truco de generalización: en vez de la edad exacta, usamos rangos de 10 años
df_priv['rango_edad'] = pd.cut(df_priv['edad'], bins=range(0, 111, 10), right=False).astype(str)
grupos_generalizados = df_priv.groupby(['rango_edad', 'ciudad']).size().reset_index(name='cuantos_comparten')
vulnerables_gen = (grupos_generalizados['cuantos_comparten'] < k_deseado).sum()
print(f"\nCon RANGOS de 10 años: {vulnerables_gen} de {len(grupos_generalizados)} combinaciones tienen menos de {k_deseado} personas")
```

### 🤔 ¿Qué nos dicen estos números?
Con la edad exacta, un montón de combinaciones de edad + ciudad las comparte muy poca gente — esas personas son fáciles de reidentificar, igual que el pelirrojo de 2 metros. Al cambiar la edad exacta por un rango (20-30, 30-40...), agrupamos a más gente bajo la misma "etiqueta" y el número de combinaciones vulnerables cae en picada: hay muchos más vecinos parecidos entre los cuales esconderse. Esa técnica se llama **generalización**, y es la herramienta favorita de los expertos en privacidad.

(Ojo: unos pocos casos seguirán solos sin importar cuánto generalices — por ejemplo, alguien con `edad = 999`, uno de los "mentirosos" que cazamos en el Cuaderno 03. La privacidad, igual que la calidad, empieza por tener datos limpios.)

## 5. Cuando el Computador Aprende a Ser Injusto: el Sesgo Algorítmico ⚠️

En 2018, una empresa gigante de comercio electrónico probó un sistema que leía hojas de vida automáticamente. El sistema aprendió, de 10 años de contrataciones pasadas, que "ingeniero" se parecía más a currículums de hombres... y empezó a **rechazar automáticamente a las mujeres**, aunque nadie le hubiera dicho explícitamente "discrimina por género".

Lo más peligroso: ni siquiera hace falta usar la variable "género" o "raza" directamente. Un dato aparentemente inocente como el **código postal** o el **barrio** puede funcionar como un sustituto silencioso (un *proxy*) de la etnia o el nivel socioeconómico de alguien. El modelo nunca "ve" la raza, pero igual termina discriminando por ella a través de la puerta trasera.

---
## 6. La Regla de Oro del Científico de Datos Santo Tomás ⚔️

> *"No uses tus conocimientos de datos para construir algoritmos que tú mismo no aceptarías que juzgaran a tu propia familia".*

Si un modelo niega un crédito o un empleo a alguien por ser mujer, por vivir en un barrio humilde o por su etnia, no es un modelo inteligente: **es un modelo discriminatorio que debe ser corregido de inmediato**.

## 7. Habeas Data: Tus Datos son Tuyos 📜🇨🇴

En Colombia existe una ley (la **Ley 1581 de 2012**, conocida como **Habeas Data**) que dice, en cristiano, algo muy simple:

* Nadie puede usar tus datos personales **sin tu permiso**.
* Solo pueden usarlos **para lo que te dijeron que los iban a usar** (si autorizaste tus datos para que te manden la factura, no pueden usarlos también para venderte seguros sin preguntarte de nuevo).
* Tienes derecho a **saber qué datos tienen de ti**, a **pedir que los corrijan** si están mal, y a **pedir que los borren**.

Es exactamente lo mismo que exigirías si alguien tuviera una libreta con tus secretos: quieres saber qué anotaron, corregir lo que está mal, y poder pedir que arranquen la página si ya no quieres que la tengan.

## Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **Anonimización** | Tachar el nombre para siempre: nadie, ni tú, puede revertirlo. |
| **Seudonimización** | Cambiar el nombre por un apodo fijo (hash): se puede seguir el rastro sin saber quién es. |
| **Cuasi-identificadores** | Datos "inocentes" (código postal, edad, género) que juntos delatan a alguien. |
| **k-Anonimato** | Cuántas personas comparten exactamente tus mismos cuasi-identificadores; entre más grande el k, más escondido estás. |
| **Sesgo algorítmico** | El modelo repite (y amplifica) la discriminación que ya existía en los datos históricos. |
| **Habeas Data (Ley 1581)** | En Colombia, tus datos son tuyos: exigen permiso, tienen un fin específico, y puedes corregirlos o borrarlos. |
