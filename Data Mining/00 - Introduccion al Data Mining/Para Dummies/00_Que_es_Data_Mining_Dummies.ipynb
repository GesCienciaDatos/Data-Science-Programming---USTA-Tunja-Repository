## 1. La Gran Analogía: El Minero del Río 🤠🏞️

Imagina a un minero con una batea parado a la orilla de un río en Boyacá:
* Llena su batea con **10 kilos de piedras, lodo, hojas secas y arena** (esos son los **Datos Crudos** que genera un banco o una tienda todos los días).
* El minero lava la tierra, retira las piedras grandes y quita el lodo (eso es **Limpieza de Datos**).
* Mueve la batea con paciencia hasta que, en el fondo, brillan **3 pequeñas pepitas de oro puro** (eso es el **Conocimiento Accionable**).

> 💡 **En resumen:** La Minería de Datos es el arte de colar millones de filas aburridas de una base de datos para encontrar la pepita de oro: un patrón que te ayude a vender más, salvar vidas o evitar fraudes.

## 3. Pongamos las Manos en la Masa: Tu Primer Experimento 🧪

Vamos a crear una pequeña montaña de datos de compras de un supermercado para encontrar la "pepita de oro" con código muy simple. Ejecuta la siguiente celda con `Shift + Enter`:

````python
import pandas as pd

# Creamos una tabla con 6 clientes ficticios
datos = {
    'Cliente': ['Carlos', 'Ana', 'Pedro', 'Luisa', 'Marta', 'Jorge'],
    'Visitas_al_Mes': [1, 12, 2, 15, 14, 1],
    'Gasto_Total_COP': [25000, 480000, 32000, 620000, 540000, 18000]
}

df = pd.DataFrame(datos)
print("📋 Esta es nuestra pequeña montaña de datos:")
display(df)
````

### 🤔 ¿Qué acaba de pasar?
Acabas de crear un **DataFrame**, que es simplemente la versión en Python de una hoja de cálculo de Excel con filas y columnas.

---
## 4. Ahora Minemos el Oro: Filtrando los Clientes Estrella ⭐

Vamos a pedirle al computador que descubra quiénes son los clientes que visitan la tienda más de 10 veces al mes y gastan más de $400,000 COP:

````python
# Encontramos la pepita de oro
pepitas_de_oro = df[(df['Visitas_al_Mes'] > 10) & (df['Gasto_Total_COP'] > 400000)]

print("✨ ¡PEPITA DE ORO ENCONTRADA!")
print("Estos son los clientes fieles que mantienen vivo el negocio:")
display(pepitas_de_oro[['Cliente', 'Gasto_Total_COP']])
````

### 🤔 ¿Qué significa esto para el negocio?
En lugar de gastar dinero imprimiendo volantes para todo el mundo, el gerente del supermercado puede enviarle un regalo de agradecimiento únicamente a Ana, Luisa y Marta para asegurarse de que nunca se vayan a la competencia.
