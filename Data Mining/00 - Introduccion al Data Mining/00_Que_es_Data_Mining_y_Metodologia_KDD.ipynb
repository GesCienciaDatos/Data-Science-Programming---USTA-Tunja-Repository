## 1. Definición Canónica y Evolución Histórica de la Minería de Datos 📚

A principios de la década de 1990, el auge de los sistemas transaccionales relacionales (RDBMS) y los almacenes de datos (*Data Warehouses*) generó una paradoja descrita por John Naisbitt: ***"Nos ahogamos en información, pero estamos hambrientos de conocimiento"***.

En este contexto, **Fayyad, Piatetsky-Shapiro y Smyth (1996)** establecieron la definición científica canónica de la **Minería de Datos (*Data Mining*)**:

> *"El proceso no trivial de identificar patrones válidos, novedosos, potencialmente útiles y comprensibles a partir de grandes volúmenes de datos."*

### 🔬 Las Cuatro Propiedades Fundamentales del Conocimiento Minado:

| Propiedad | Definición Rigurosa | Riesgo si no se cumple |
|---|---|---|
| **Válido (*Valid*)** | El patrón debe sostenerse con significancia estadística sobre datos nuevos y no vistos. | **Sobreajuste (*Overfitting*):** El modelo memorizó el ruido aleatorio del pasado y fracasará en producción. |
| **Novedoso (*Novel*)** | El patrón debe aportar conocimiento inédito o contraintuitivo que los expertos del dominio no conocían. | **Trivialidad:** Descubrir que "los pacientes embarazados son mujeres" no aporta ningún valor al negocio. |
| **Potencialmente Útil (*Potentially Useful*)** | Debe permitir tomar decisiones operativas, clínicas o estratégicas cuantificables. | **Inutilidad Operativa:** Hallazgos teóricos abstractos que no se pueden monetizar ni aplicar en el mundo real. |
| **Comprensible (*Understandable*)** | El patrón debe ser interpretable directa o indirectamente por seres humanos. | **Caja Negra Ingobernable:** Modelos que deciden sin posibilidad de auditoría regulatoria o justificación legal. |

## 2. El Proceso KDD (*Knowledge Discovery in Databases*) al Detalle 🔄

Existe una confusión conceptual frecuente: confundir 'Data Mining' con todo el proyecto. Rigurosamente, la Minería de Datos es solo **la fase algorítmica** dentro del proceso integral de descubrimiento de conocimiento en bases de datos (**KDD**):

```mermaid
graph LR
    A[Datos Crudos] --> B(1. Selección)
    B --> C(2. Limpieza y Preprocesamiento)
    C --> D(3. Transformación y Reducción)
    D --> E(4. Minería de Datos)
    E --> F(5. Interpretación y Evaluación)
    F --> G[Conocimiento Accionable]
```

### Fases del Marco KDD:
1. **Selección (*Data Selection*):** Identificación de fuentes heterogéneas (bases SQL, logs, APIs, archivos planos) y extracción de una muestra representativa de la población objetivo.
2. **Limpieza y Preprocesamiento (*Data Preprocessing*):** Remoción de ruido, gestión de valores faltantes (MCAR, MAR, MNAR) y resolución de inconsistencias semánticas.
3. **Transformación y Reducción (*Data Transformation*):** Normalización métrica, ingeniería de características derivadas, agregaciones temporales y reducción dimensional (PCA).
4. **Minería de Datos (*Data Mining*):** Aplicación de algoritmos de clasificación, regresión, agrupamiento (*clustering*), reglas de asociación o detección de anomalías.
5. **Interpretación y Evaluación (*Interpretation / Evaluation*):** Validación cuantitativa (métricas estadísticas, curvas ROC, lift) y traducción del modelo a conocimiento estratégico para los tomadores de decisiones.

## 3. Matriz Comparativa: Bases de Datos vs. Estadística vs. Data Mining vs. Machine Learning ⚖️

| Dimensión | Bases de Datos (SQL / OLAP) | Estadística Clásica | Minería de Datos (Data Mining) | Machine Learning Moderno |
|---|---|---|---|---|
| **Pregunta Típica** | "¿Cuáles fueron las ventas totales de la tienda de Tunja en mayo?" | "¿Es la media de ventas significativamente distinta entre dos regiones al 95% de confianza?" | "¿Qué patrones ocultos de compra distinguen a los clientes que cancelarán su suscripción?" | "¿Cómo minimizamos la pérdida de entropía cruzada en un espacio de 500 dimensiones?" |
| **Tipo de Razonamiento** | Deductivo (recuperación exacta de hechos) | Confirmatorio (prueba formal de hipótesis previas) | Inductivo y Exploratorio (búsqueda orientada a patrones en los datos) | Predictivo y Adaptativo (optimización matemática de representaciones) |
| **Supuestos sobre los Datos** | Estructura relacional rígida | Supuestos distributivos estrictos (normalidad, homocedasticidad) | Mínimos supuestos; tolerante a ruido y distribuciones multimodales | Basado en minimización del riesgo empírico y generalización |
| **Volumen Típico** | Gigabytes a Terabytes de transacciones | Muestras pequeñas o medianas controladas | Grandes repositorios masivos, no homogéneos | Muestras grandes a Big Data masivo |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os, warnings
warnings.filterwarnings('ignore')

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

# Configuración visual institucional
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (9, 4.5)
plt.rcParams['font.size'] = 10

print("🚀 Entorno de Minería de Datos configurado con éxito.")

import os, urllib.request, urllib.parse, gzip, base64

_DATA_EMBEDDED = {
    "auditoria_calidad_datos.csv": ""
}

def load_dataset(filename, module_name="00 - Introduccion al Data Mining"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo restaura automáticamente de forma exacta.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("..", "..", "data", filename),
        os.path.join("Data Mining", module_name, "data", filename),
        os.path.join("..", "Data Mining", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path

    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Mining/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)

    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]

    print(f"[INFO] Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=2) as response:
                if response.status == 200:
                    with open(target_path, "wb") as out_f:
                        out_f.write(response.read())
                    print(f"[OK] Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue

    if filename in _DATA_EMBEDDED:
        print(f"[INFO] Restaurando '{filename}' de forma autónoma...")
        data_bytes = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target_path, "wb") as out_f:
            out_f.write(data_bytes)
        print(f"[OK] Dataset '{filename}' preparado exitosamente en '{target_path}'.")
        return target_path

    raise FileNotFoundError(f"No se pudo encontrar ni generar el dataset '{filename}'.")
````

## 4. Laboratorio Práctico: El Proceso KDD Paso a Paso en Python 💻🔬

Simularemos el ciclo completo de KDD sobre una base de clientes bancarios para descubrir qué variables y comportamientos determinan el abandono de productos (*Churn*).

````python
# =============================================================================
# FASE 1 KDD: SELECCIÓN DE DATOS
# =============================================================================
np.random.seed(42)
N = 1000

# Datos crudos extraídos de logs y tablas transaccionales
edad = np.random.normal(40, 12, N).round()
saldo_cuenta = np.random.exponential(15000000, N).round(2)
num_productos = np.random.choice([1, 2, 3, 4], N, p=[0.5, 0.35, 0.1, 0.05])
reclamaciones = np.random.poisson(0.8, N)

# Inyectar ruido intencional (valores atípicos y nulos)
edad[np.random.choice(N, 20, replace=False)] = np.nan
edad[np.random.choice(N, 5, replace=False)] = -8 # Error de digitación

# Patrón oculto latente (Fórmula de Abandono)
prob_churn = 0.10 + 0.35 * (reclamaciones >= 2) - 0.20 * (num_productos >= 2) + 0.15 * (saldo_cuenta < 5000000)
prob_churn = np.clip(prob_churn, 0.02, 0.95)
abandono = (np.random.rand(N) < prob_churn).astype(int)

df_kdd = pd.DataFrame({
    'id_cliente': [f'CLI-{i:05d}' for i in range(1, N + 1)],
    'edad': edad,
    'saldo_cuenta': saldo_cuenta,
    'num_productos': num_productos,
    'reclamaciones_soporte': reclamaciones,
    'abandono': abandono
})

print("Fase 1 KDD (Selección): Primeras 5 filas de los datos crudos:")
display(df_kdd.head(5))
print(f"Tasa general de abandono detectada: {df_kdd['abandono'].mean()*100:.2f}%")
````

````python
# =============================================================================
# FASE 2 KDD: LIMPIEZA Y PREPROCESAMIENTO
# =============================================================================
print("Auditoría previa a la limpieza:")
print(f"• Valores nulos en edad: {df_kdd['edad'].isnull().sum()}")
print(f"• Edades anómalas (< 0): {(df_kdd['edad'] < 0).sum()}")

# Saneamiento
df_kdd_clean = df_kdd.copy()
df_kdd_clean.loc[df_kdd_clean['edad'] < 0, 'edad'] = np.nan
mediana_edad = df_kdd_clean['edad'].median()
df_kdd_clean['edad'] = df_kdd_clean['edad'].fillna(mediana_edad)

print(f"\nLimpieza completada: edad imputada con la mediana ({mediana_edad:.0f} años). Nulos restantes: {df_kdd_clean['edad'].isnull().sum()}")
````

````python
# =============================================================================
# FASE 3 KDD: TRANSFORMACIÓN Y REDUCCIÓN
# =============================================================================
# Creación de ratios e ingeniería de características derivadas
df_kdd_clean['saldo_por_producto'] = (df_kdd_clean['saldo_cuenta'] / df_kdd_clean['num_productos']).round(2)
df_kdd_clean['cliente_alto_riesgo'] = ((df_kdd_clean['reclamaciones_soporte'] >= 2) & (df_kdd_clean['saldo_cuenta'] < 5000000)).astype(int)

print("Fase 3 KDD (Transformación): Nuevas características creadas:")
display(df_kdd_clean[['saldo_cuenta', 'num_productos', 'saldo_por_producto', 'cliente_alto_riesgo', 'abandono']].head(5))
````

````python
# =============================================================================
# FASE 4 KDD: MINERÍA DE DATOS (DATA MINING)
# =============================================================================
from sklearn.tree import DecisionTreeClassifier, export_text

X_kdd = df_kdd_clean[['edad', 'saldo_cuenta', 'num_productos', 'reclamaciones_soporte']]
y_kdd = df_kdd_clean['abandono']

# Aplicamos Minería de Datos mediante un Árbol CART para extraer reglas explicables
arbol_mineria = DecisionTreeClassifier(max_depth=3, random_state=42)
arbol_mineria.fit(X_kdd, y_kdd)

print("Fase 4 KDD (Minería de Datos): Reglas lógicas descubiertas por el modelo:")
print(export_text(arbol_mineria, feature_names=list(X_kdd.columns)))
````

````python
# =============================================================================
# FASE 5 KDD: INTERPRETACIÓN Y EVALUACIÓN DEL CONOCIMIENTO
# =============================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

# 1. Análisis bivariado del impacto de las reclamaciones en el abandono
sns.barplot(data=df_kdd_clean, x='reclamaciones_soporte', y='abandono', ax=axes[0], palette='Reds_d', ci=None)
axes[0].set_title('Tasa de Abandono vs. Reclamaciones a Soporte', fontweight='bold')
axes[0].set_ylabel('Probabilidad de Abandono (Churn)')
axes[0].set_xlabel('Número de Reclamaciones')

# 2. Impacto de la diversificación de productos
sns.barplot(data=df_kdd_clean, x='num_productos', y='abandono', ax=axes[1], palette='Blues_d', ci=None)
axes[1].set_title('Efecto Protector de la Tenencia de Productos', fontweight='bold')
axes[1].set_ylabel('Probabilidad de Abandono (Churn)')
axes[1].set_xlabel('Cantidad de Productos Contratados')

plt.tight_layout()
plt.show()

print("💡 HALLAZGO DE CONOCIMIENTO (Fase 5 KDD):")
print("1. Clientes con 2 o más reclamaciones multiplican por 4 su probabilidad de fuga.")
print("2. Clientes con 2 o más productos bancarios reducen a menos de la mitad su tasa de abandono.")
print("Acción de Negocio: Crear protocolo de retención automática al recibir la segunda queja.")
````

### 4.1 Segundo Caso: Auditoría KDD sobre Datos Reales del Módulo 🔍📋

El laboratorio anterior usó datos **sintéticos** (generados con `numpy.random`), perfectos para enseñar el concepto pero irrealmente limpios. En un proyecto real, las **Fases 1 y 2 del KDD** (Selección y Limpieza) casi nunca son tan sencillas.

Vamos a repetir esas dos primeras fases sobre `auditoria_calidad_datos.csv`, un archivo real de este módulo con clientes de varias ciudades colombianas, para exponer los problemas de calidad típicos que un analista debe resolver **antes** de poder minar cualquier patrón.

````python
# =============================================================================
# FASE 1 KDD (Caso Real): SELECCIÓN Y DIAGNÓSTICO INICIAL
# =============================================================================
df_auditoria = pd.read_csv(load_dataset("auditoria_calidad_datos.csv"))

print("Fase 1 KDD (Selección) - Caso Real: dimensiones y primeras filas:")
display(df_auditoria.head(5))
print(f"\nRegistros totales: {df_auditoria.shape[0]} | Columnas: {df_auditoria.shape[1]}")

print("\nDiagnóstico de calidad ANTES de limpiar:")
print(f"- Valores nulos por columna:\n{df_auditoria.isnull().sum().to_string()}")
print(f"\n- Edades fuera de rango humano (< 0 o > 110): {((df_auditoria['edad'] < 0) | (df_auditoria['edad'] > 110)).sum()}")
print(f"- Variantes distintas de 'ciudad' encontradas: {sorted(df_auditoria['ciudad'].dropna().unique())}")
print(f"- Variantes distintas de 'cliente_activo' encontradas: {sorted(df_auditoria['cliente_activo'].dropna().unique())}")
````

````python
# =============================================================================
# FASE 2 KDD (Caso Real): LIMPIEZA Y ESTANDARIZACIÓN
# =============================================================================
df_auditoria_clean = df_auditoria.copy()

# 1. Estandarizar nombres de ciudad (mayúsculas/minúsculas y valor "Desconocido")
mapa_ciudades = {
    'tunja': 'Tunja', 'TUNJA': 'Tunja',
    'BOGOTA': 'Bogotá',
    'Desconocido': np.nan
}
df_auditoria_clean['ciudad'] = df_auditoria_clean['ciudad'].str.strip().replace(mapa_ciudades)

# 2. Estandarizar la variable booleana 'cliente_activo' (Si/SI/1/TRUE -> 1 ; No/0 -> 0)
mapa_activo = {'Si': 1, 'SI': 1, 'TRUE': 1, '1': 1, 'No': 0, '0': 0}
df_auditoria_clean['cliente_activo_bin'] = df_auditoria_clean['cliente_activo'].map(mapa_activo)

# 3. Sanear edades imposibles y valores monetarios negativos
df_auditoria_clean.loc[(df_auditoria_clean['edad'] < 0) | (df_auditoria_clean['edad'] > 110), 'edad'] = np.nan
df_auditoria_clean.loc[df_auditoria_clean['ingreso_mensual'] < 0, 'ingreso_mensual'] = np.nan
df_auditoria_clean.loc[df_auditoria_clean['gasto_anual'] < 0, 'gasto_anual'] = np.nan

# 4. Imputación con la mediana para las columnas numéricas afectadas
for col in ['edad', 'ingreso_mensual', 'gasto_anual']:
    df_auditoria_clean[col] = df_auditoria_clean[col].fillna(df_auditoria_clean[col].median())

print("Fase 2 KDD (Limpieza) - Caso Real: resultado tras estandarizar y sanear:")
print(f"- Ciudades únicas antes: {df_auditoria['ciudad'].nunique(dropna=True)} -> después: {df_auditoria_clean['ciudad'].nunique(dropna=True)}")
cols_numericas = ['edad', 'ingreso_mensual', 'gasto_anual']
print(f"- Nulos en columnas numéricas antes: {df_auditoria[cols_numericas].isnull().sum().sum()} -> después: {df_auditoria_clean[cols_numericas].isnull().sum().sum()}")
display(df_auditoria_clean.head(5))

print("\n💡 Lección clave: la Fase 2 del KDD no es un simple 'dropna()'. Exige decisiones de negocio")
print("   (¿qué ciudad representa 'tunja'? ¿qué codificación de 'Si' es la correcta?) antes de imputar nada.")
````

### 4.2 Completando el Ciclo KDD sobre Datos Reales: Fases 3 a 5 🔧🌳

La Sección 4.1 dejó los datos de auditoría limpios y estandarizados (`df_auditoria_clean`), pero el ciclo KDD no termina en la limpieza. Para cerrar el ejemplo con datos reales replicamos, sobre los mismos clientes, la misma lógica de la Fase 3 (Transformación), Fase 4 (Minería) y Fase 5 (Interpretación) que se aplicó al caso sintético de la Sección 4 — y de paso resolvemos dos problemas de calidad que la limpieza rápida de la Sección 4.1 no cubrió: **registros duplicados** y **valores extremos** de ingreso y gasto.

````python
# =============================================================================
# FASE 3 KDD (Caso Real, continuación): TRANSFORMACIÓN Y REDUCCIÓN
# =============================================================================
df_auditoria_final = df_auditoria_clean.drop_duplicates(subset='id_cliente', keep='first').copy()
print(f"Registros duplicados por id_cliente eliminados: {df_auditoria_clean.shape[0] - df_auditoria_final.shape[0]}")

# La limpieza de la Sección 4.1 imputó nulos, pero no acotó valores extremos como
# un ingreso mensual de $85'000.000 COP (probable error de digitación de un cero de más).
for col in ['ingreso_mensual', 'gasto_anual']:
    q1, q99 = df_auditoria_final[col].quantile([0.01, 0.99])
    n_atipicos = ((df_auditoria_final[col] < q1) | (df_auditoria_final[col] > q99)).sum()
    df_auditoria_final[col] = df_auditoria_final[col].clip(lower=q1, upper=q99)
    print(f"- '{col}': {n_atipicos} valores extremos acotados al percentil [1, 99].")

# Ingeniería de características derivadas
df_auditoria_final['ratio_gasto_ingreso'] = (df_auditoria_final['gasto_anual'] / (df_auditoria_final['ingreso_mensual'] * 12)).round(3)
df_auditoria_final['segmento_edad'] = pd.cut(df_auditoria_final['edad'], bins=[0, 30, 45, 60, 110], labels=['18-30', '31-45', '46-60', '61+'])

print("\nFase 3 KDD (Transformación) - Caso Real: nuevas variables creadas:")
display(df_auditoria_final[['edad', 'segmento_edad', 'ingreso_mensual', 'gasto_anual', 'ratio_gasto_ingreso', 'cliente_activo_bin']].head(5))
````

````python
# =============================================================================
# FASE 4 KDD (Caso Real, continuación): MINERÍA DE DATOS
# =============================================================================
from sklearn.tree import DecisionTreeClassifier, export_text

features_reales = ['edad', 'ingreso_mensual', 'gasto_anual', 'ratio_gasto_ingreso']
df_modelo_real = df_auditoria_final.dropna(subset=features_reales + ['cliente_activo_bin'])

X_real = df_modelo_real[features_reales]
y_real = df_modelo_real['cliente_activo_bin']

arbol_real = DecisionTreeClassifier(max_depth=3, min_samples_leaf=30, random_state=42)
arbol_real.fit(X_real, y_real)

print(f"Fase 4 KDD (Minería) - Caso Real: árbol entrenado con {len(df_modelo_real)} clientes reales para predecir 'cliente_activo_bin'.")
print(export_text(arbol_real, feature_names=features_reales))
````

````python
# =============================================================================
# FASE 5 KDD (Caso Real, continuación): INTERPRETACIÓN Y EVALUACIÓN
# =============================================================================
importancias_real = pd.Series(arbol_real.feature_importances_, index=features_reales).sort_values(ascending=False)

plt.figure(figsize=(8, 4))
sns.barplot(x=importancias_real.values, y=importancias_real.index, hue=importancias_real.index, palette='Oranges_r', legend=False)
plt.title('Fase 5 KDD: Importancia de Variables en la Actividad del Cliente (Datos Reales)', fontweight='bold')
plt.xlabel('Importancia (Gini)')
plt.tight_layout()
plt.show()

print("Fase 5 KDD (Interpretación) - Caso Real:")
print(f"- Variable más determinante para el árbol: '{importancias_real.index[0]}'.")
print("- A diferencia del caso sintético (con una señal de abandono diseñada deliberadamente),")
print("  en datos reales las fronteras entre cliente activo/inactivo suelen ser mucho más difusas:")
print("  este árbol es una primera hipótesis exploratoria, no un modelo listo para producción.")
````
