## 1. El Estándar Industrial: Metodología CRISP-DM 🌐🏭

A finales de la década de 1990, un consorcio formado por **DaimlerChrysler, SPSS, NCR y OHRA** financiado por la Unión Europea desarrolló **CRISP-DM (*Cross-Industry Standard Process for Data Mining* )** con el objetivo de convertir la minería de datos en una disciplina de ingeniería repetible, estructurada y orientada al valor comercial.

A pesar de los avances tecnológicos, encuestas internacionales de KDnuggets confirman que **CRISP-DM sigue siendo el marco de referencia más utilizado (más del 45% de la industria)**.

```mermaid
graph TD
    A[1. Comprensión del Negocio] <--> B[2. Comprensión de los Datos]
    B --> C[3. Preparación de los Datos]
    C <--> D[4. Modelado]
    D --> E[5. Evaluación]
    E -->|Si cumple objetivos| F[6. Despliegue en Producción]
    E -.->|Si no cumple objetivos| A
```

### Las Seis Fases de CRISP-DM Explicadas al Detalle:

1. **Comprensión del Negocio (*Business Understanding*):**
   - Determinar los objetivos del negocio y qué problema real se desea resolver.
   - Traducir la necesidad empresarial en una definición técnica de minería de datos.
   - Establecer los **criterios de éxito de negocio** (ej. reducir la tasa de morosidad en 3 puntos porcentuales o generar un ROI del 25%) y los **criterios de éxito técnico** (ej. ROC-AUC $\ge 0.82$).

2. **Comprensión de los Datos (*Data Understanding*):**
   - Recolección inicial de fuentes de datos (internas, externas, APIs, logs transaccionales).
   - Exploración estadística y visual para identificar problemas de calidad (nulos, ruido, tipos de datos inconsistentes).

3. **Preparación de los Datos (*Data Preparation*):**
   - Es la etapa más exigente: **suele consumir entre el 60% y el 80% del tiempo total del proyecto**.
   - Incluye selección de tablas y columnas, limpieza de valores atípicos, imputación, ingeniería de características derivadas y escalado.

4. **Modelado (*Modeling*):**
   - Selección de las técnicas algorítmicas adecuadas para la tarea.
   - Diseño de la estrategia de partición y validación (K-Fold estratificado).
   - Calibración y optimización sistemática de hiperparámetros.

5. **Evaluación (*Evaluation*):**
   - Contrastar los resultados obtenidos frente a los objetivos iniciales de negocio.
   - Validar que no existan trampas metodológicas (como fuga de datos / *data leakage*).
   - Decidir formalmente: ¿Se aprueba para despliegue o se itera de nuevo a la fase 1?

6. **Despliegue (*Deployment*):**
   - Integración del modelo en la operación diaria (microservicios REST, batch pipelines, dashboards interactivos).
   - Plan de monitoreo continuo para detectar desvío de datos (*Data Drift*) y degradación conceptual (*Concept Drift*).

## 2. La Metodología SEMMA del SAS Institute 📊🔍

Desarrollada por el **SAS Institute**, la metodología **SEMMA** organiza el trabajo analítico directamente desde la perspectiva del flujo de datos en el software analítico:

| Fase SEMMA | Significado | Equivalente en CRISP-DM |
|---|---|---|
| **S (Sample)** | Muestreo estadístico representativo de grandes volúmenes. | Comprensión de los Datos |
| **E (Explore)** | Exploración visual, correlaciones y detección de tendencias. | Comprensión de los Datos |
| **M (Modify)** | Transformación, imputación, normalización y codificación. | Preparación de los Datos |
| **M (Model)** | Ajuste de modelos predictivos y descriptivos. | Modelado |
| **A (Assess)** | Evaluación del rendimiento y precisión de los modelos. | Evaluación |

> [!IMPORTANT]
> **Diferencia Crítica:** A diferencia de CRISP-DM, SEMMA **no incluye explícitamente las fases de Comprensión del Negocio ni Despliegue**, centrándose exclusivamente en la secuencia técnica de experimentación en laboratorio.

### 2.1 SEMMA en Acción: Mini-Caso con los Datos Reales de Auditoría 🔬📊

Para que la tabla anterior deje de ser solo teoría, replicamos las cinco fases de SEMMA sobre `auditoria_calidad_datos.csv`, el mismo archivo de clientes con problemas de calidad reales que se auditó en el Cuaderno 00. A diferencia del Caso #1 (Banco Andino) de la Sección 3, aquí no existe una variable objetivo histórica: es un ejercicio de **descubrimiento descriptivo**, el terreno donde SEMMA se aplica con más frecuencia en la práctica del SAS Institute.

````python
# =============================================================================
# SEMMA FASE S (Sample) + E (Explore)
# =============================================================================
df_audit_semma = pd.read_csv(load_dataset("auditoria_calidad_datos.csv"))

# S (Sample): tomamos una muestra aleatoria representativa del 70% para acelerar la exploración
muestra_semma = df_audit_semma.sample(frac=0.7, random_state=42)
print(f"SEMMA Fase S (Sample): muestra de {len(muestra_semma)} de {len(df_audit_semma)} clientes ({len(muestra_semma)/len(df_audit_semma)*100:.0f}%).")

# E (Explore): estadísticas descriptivas de variables numéricas antes de transformar nada
print("\nSEMMA Fase E (Explore): estadísticas descriptivas de la muestra:")
display(muestra_semma[['edad', 'ingreso_mensual', 'gasto_anual']].describe().round(2))
````

````python
# =============================================================================
# SEMMA FASE M (Modify) + M (Model)
# =============================================================================
df_audit_semma_clean = muestra_semma.copy()
df_audit_semma_clean['ciudad'] = df_audit_semma_clean['ciudad'].str.strip().replace(
    {'tunja': 'Tunja', 'TUNJA': 'Tunja', 'BOGOTA': 'Bogotá', 'Desconocido': np.nan}
)
df_audit_semma_clean.loc[(df_audit_semma_clean['edad'] < 0) | (df_audit_semma_clean['edad'] > 110), 'edad'] = np.nan
for col in ['edad', 'ingreso_mensual', 'gasto_anual']:
    df_audit_semma_clean[col] = df_audit_semma_clean[col].fillna(df_audit_semma_clean[col].median())
    q1, q99 = df_audit_semma_clean[col].quantile([0.01, 0.99])
    df_audit_semma_clean[col] = df_audit_semma_clean[col].clip(lower=q1, upper=q99)

# Modify también incluye estandarizar antes de un algoritmo basado en distancias
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X_semma = StandardScaler().fit_transform(df_audit_semma_clean[['ingreso_mensual', 'gasto_anual']])

# M (Model): modelo descriptivo K-Means para descubrir segmentos naturales de clientes
modelo_semma = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_semma)
df_audit_semma_clean['segmento'] = modelo_semma.labels_

print("SEMMA Fase M (Modify): variables saneadas, estandarizadas y acotadas.")
print("SEMMA Fase M (Model): 3 segmentos descubiertos con K-Means sobre ingreso y gasto.")
````

````python
# =============================================================================
# SEMMA FASE A (Assess)
# =============================================================================
from sklearn.metrics import silhouette_score

sil_semma = silhouette_score(X_semma, modelo_semma.labels_)

plt.figure(figsize=(8, 4.5))
sns.scatterplot(data=df_audit_semma_clean, x='ingreso_mensual', y='gasto_anual', hue='segmento', palette='viridis', s=40)
plt.title('SEMMA Fase A (Assess): Segmentos de Clientes Reales Descubiertos', fontweight='bold')
plt.xlabel('Ingreso Mensual (COP)')
plt.ylabel('Gasto Anual (COP)')
plt.tight_layout()
plt.show()

print(f"SEMMA Fase A (Assess): Coeficiente de Silueta = {sil_semma:.3f} (más cercano a 1 indica segmentos mejor separados).")
display(df_audit_semma_clean.groupby('segmento')[['ingreso_mensual', 'gasto_anual']].mean().round(0))
````

## 3. Simulación Empresarial: Las 6 Fases de CRISP-DM en Código 💻🏢

Aplicaremos las 6 fases de CRISP-DM para resolver el Caso #1 de nuestro portafolio: **Predicción de Riesgo Crediticio en el Banco Andino**.

````python
# =============================================================================
# CRISP-DM FASE 1: COMPRENSIÓN DEL NEGOCIO (Business Understanding)
# =============================================================================
# Criterios fijados por el comité de crédito del banco:
# - El costo de prestarle a alguien que no paga (Falso Negativo) es de $15,000,000 COP.
# - El costo de rechazar a un buen cliente (Falso Positivo) es de $1,200,000 COP (ganancia de intereses perdida).
# - Objetivo: Maximizar el ahorro neto financiero de la cartera mediante un corte óptimo.

COSTO_FN = 15000000.0  # Prestarle a un moroso
COSTO_FP = 1200000.0   # Rechazar a un buen pagador

print("CRISP-DM Fase 1 definida:")
print(f"• Costo por Falso Negativo (Default no detectado): ${COSTO_FN:,.0f} COP")
print(f"• Costo por Falso Positivo (Cliente bueno rechazado): ${COSTO_FP:,.0f} COP")
````

````python
# =============================================================================
# CRISP-DM FASE 2: COMPRENSIÓN DE LOS DATOS (Data Understanding)
# =============================================================================
np.random.seed(42)
N = 1200
ingresos = np.random.normal(5200000, 1800000, N).clip(min=1300000)
deuda_actual = np.random.normal(12000000, 8000000, N).clip(min=500000)
score_datacredito = np.random.normal(680, 95, N).clip(min=300, max=850)
retrasos_previos = np.random.poisson(1.2, N)

# Relación de default oculta
dti = deuda_actual / (ingresos * 12)
score_riesgo = 2.0 * (dti > 0.4) + 1.8 * (retrasos_previos >= 2) - 0.006 * (score_datacredito - 500)
p_default = 1 / (1 + np.exp(-score_riesgo))
default_real = (np.random.rand(N) < p_default).astype(int)

df_credito = pd.DataFrame({
    'ingresos_mensuales': ingresos,
    'deuda_actual': deuda_actual,
    'score_datacredito': score_datacredito,
    'retrasos_previos': retrasos_previos,
    'default_real': default_real
})

print("CRISP-DM Fase 2 (Comprensión): Estadísticas Descriptivas:")
display(df_credito.describe().round(2))
print(f"Tasa de morosidad en la muestra: {df_credito['default_real'].mean()*100:.2f}%")
````

### 3.1 Profundizando la Fase 2: Exploración Visual (Conexión con SEMMA "Explore") 📊

La tabla de `describe()` es útil pero oculta relaciones entre variables. La fase **E (Explore)** de SEMMA —equivalente a la "Comprensión de los Datos" de CRISP-DM— exige mirar las distribuciones y su relación con el objetivo antes de modelar.

````python
# =============================================================================
# EXPLORACIÓN VISUAL COMPLEMENTARIA (Fase 2 CRISP-DM / Fase "Explore" de SEMMA)
# =============================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

sns.boxplot(data=df_credito, x='default_real', y='score_datacredito', ax=axes[0], palette=['#16a34a', '#dc2626'])
axes[0].set_title('Score DataCrédito según Estado de Default', fontweight='bold')
axes[0].set_xlabel('Default (0 = Al día, 1 = Moroso)')

sns.boxplot(data=df_credito, x='default_real', y='retrasos_previos', ax=axes[1], palette=['#16a34a', '#dc2626'])
axes[1].set_title('Retrasos Previos según Estado de Default', fontweight='bold')
axes[1].set_xlabel('Default (0 = Al día, 1 = Moroso)')

plt.tight_layout()
plt.show()

print("Hallazgo exploratorio: los clientes en default tienden a tener menor score y más retrasos previos,")
print("confirmando visualmente los coeficientes que la regresión logística aprenderá en la Fase 4.")
````

````python
# =============================================================================
# CRISP-DM FASE 3: PREPARACIÓN DE LOS DATOS (Data Preparation)
# =============================================================================
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Ingeniería de características
df_credito['ratio_deuda_ingreso'] = df_credito['deuda_actual'] / (df_credito['ingresos_mensuales'] * 12)
features = ['ingresos_mensuales', 'score_datacredito', 'retrasos_previos', 'ratio_deuda_ingreso']

X_train, X_test, y_train, y_test = train_test_split(
    df_credito[features], df_credito['default_real'], test_size=0.30, random_state=42, stratify=df_credito['default_real']
)

scaler = StandardScaler()
X_train_std = scaler.fit_transform(X_train)
X_test_std = scaler.transform(X_test)

print(f"CRISP-DM Fase 3 (Preparación): Conjunto Train: {X_train_std.shape} | Test: {X_test_std.shape}")
````

````python
# =============================================================================
# CRISP-DM FASE 4: MODELADO (Modeling)
# =============================================================================
from sklearn.linear_model import LogisticRegression

# Entrenamos clasificador probabilístico
modelo_banco = LogisticRegression(random_state=42)
modelo_banco.fit(X_train_std, y_train)

# Obtenemos probabilidades de default en test
p_test = modelo_banco.predict_proba(X_test_std)[:, 1]
print("CRISP-DM Fase 4 (Modelado): Modelo entrenado exitosamente.")
````

````python
# =============================================================================
# CRISP-DM FASE 5: EVALUACIÓN (Evaluation en Términos de Negocio)
# =============================================================================
# Buscamos el umbral de decisión que MINIMIZA la pérdida económica total del banco
umbrales = np.linspace(0.1, 0.9, 81)
perdidas_totales = []

for umbral in umbrales:
    pred_u = (p_test >= umbral).astype(int)
    fn = ((pred_u == 0) & (y_test == 1)).sum() # Prestar a quien no paga
    fp = ((pred_u == 1) & (y_test == 0)).sum() # Rechazar a buen cliente
    costo_total = fn * COSTO_FN + fp * COSTO_FP
    perdidas_totales.append(costo_total)

idx_opt = np.argmin(perdidas_totales)
umbral_opt = umbrales[idx_opt]
ahorro_maximo = perdidas_totales[np.abs(umbrales - 0.5).argmin()] - perdidas_totales[idx_opt]

plt.figure(figsize=(9, 4.5))
plt.plot(umbrales, np.array(perdidas_totales) / 1e6, color='#d97706', lw=2.5, label='Pérdida Financiera Total')
plt.axvline(umbral_opt, color='#1e3a8a', linestyle='--', lw=2, label=f'Umbral Financiero Óptimo ({umbral_opt:.2f})')
plt.axvline(0.5, color='gray', linestyle=':', label='Umbral Estándar (0.50)')
plt.title('CRISP-DM Fase 5: Evaluación de Costo Financiero Total vs. Umbral de Decisión', fontweight='bold')
plt.xlabel('Umbral de Probabilidad de Default para Rechazar Crédito')
plt.ylabel('Pérdida Total del Portafolio (Millones COP)')
plt.legend()
plt.show()

print("CRISP-DM Fase 5 (Evaluación): Criterio de Éxito de Negocio Cumplido:")
print(f"• Al calibrar el umbral a {umbral_opt:.2f}, el banco ahorra ${ahorro_maximo:,.0f} COP en comparación con el umbral por defecto (0.50).")
````

````python
# =============================================================================
# CRISP-DM FASE 6: DESPLIEGUE (Deployment Pipeline Simulado)
# =============================================================================
def evaluar_nuevo_solicitante(ingreso, score, retrasos, ratio_deuda):
    """Función de inferencia lista para ser expuesta en una API REST de crédito."""
    vector_crudo = np.array([[ingreso, score, retrasos, ratio_deuda]])
    vector_std = scaler.transform(vector_crudo)
    probabilidad = modelo_banco.predict_proba(vector_std)[0, 1]
    decision = "RECHAZADO ❌" if probabilidad >= umbral_opt else "APROBADO ✅"
    return {
        'probabilidad_default': round(float(probabilidad), 4),
        'decision_credito': decision,
        'umbral_aplicado': umbral_opt
    }

# Prueba con dos clientes nuevos en tiempo real:
cliente_1 = evaluar_nuevo_solicitante(ingreso=6500000, score=720, retrasos=0, ratio_deuda=0.15)
cliente_2 = evaluar_nuevo_solicitante(ingreso=2800000, score=540, retrasos=3, ratio_deuda=0.55)

print("CRISP-DM Fase 6 (Despliegue): Inferencia en Producción:")
print("• Solicitante 1:", cliente_1)
print("• Solicitante 2:", cliente_2)
````
