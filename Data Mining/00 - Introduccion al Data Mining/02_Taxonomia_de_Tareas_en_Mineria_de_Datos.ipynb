## 1. La Taxonomía Formal de la Minería de Datos 🧭

En la literatura de Ciencia de Datos y Minería de Datos (Tan, Steinbach, Kumar & Karpatne, 2018; Han, Kamber & Pei, 2011), las tareas analíticas se dividen en **dos grandes familias** según la disponibilidad o no de una variable objetivo (*target*):

```mermaid
graph TD
    Root[Tareas de Data Mining] --> Pred[A. Tareas Predictivas Supervisadas]
    Root --> Desc[B. Tareas Descriptivas No Supervisadas]
    
    Pred --> Clf[1. Clasificación: Salida Discreta/Categoría]
    Pred --> Reg[2. Regresión: Salida Cuantitativa Continua]
    
    Desc --> Clust[3. Clustering: Segmentación Natural]
    Desc --> Assoc[4. Reglas de Asociación: Co-ocurrencias]
    Desc --> Anom[5. Detección de Anomalías: Desviaciones Outliers]
```

### Comparación Estructural de las Cinco Tareas:

| Tarea | Paradigma | Tipo de Salida | Pregunta de Negocio Típica | Algoritmos Representativos |
|---|---|---|---|---|
| **1. Clasificación** | Supervisada | Variable discreta / categórica | "¿Este cliente pagará el crédito o entrará en mora?" | Árboles CART, Random Forest, SVM, Regresión Logística, k-NN, Redes Neuronales |
| **2. Regresión** | Supervisada | Variable continua cuantitativa | "¿Cuál será el volumen de ventas en dólares el próximo trimestre?" | Regresión Lineal, Ridge, Lasso, SVR, MLPRegressor, Gradient Boosting |
| **3. Agrupamiento (*Clustering*)** | No Supervisada | Asignación a conglomerado | "¿Cómo se dividen naturalmente mis clientes según hábitos de compra?" | K-Means, K-Medoids, DBSCAN, Clustering Jerárquico Aglomerativo |
| **4. Reglas de Asociación** | No Supervisada | Implicaciones $X \implies Y$ | "¿Qué medicamentos o productos se compran conjuntamente?" | Apriori, FP-Growth, Eclat |
| **5. Detección de Anomalías** | No / Semi Supervisada | Puntuación de rareza (Normal vs. Atípico) | "¿Cuál de estas 10,000 transferencias bancarias es un fraude informático?" | Isolation Forest, One-Class SVM, Local Outlier Factor (LOF) |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os, warnings
warnings.filterwarnings('ignore')

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 5)

print("🚀 Entorno listo para explorar la Taxonomía de Data Mining.")

import os, urllib.request, urllib.parse, gzip, base64

_DATA_EMBEDDED = {
    "auditoria_calidad_datos.csv": "",
    "proyectos_mineria_negocio.csv": "H4sIAAN0mGoC/31TO24UQRDNLfkOFRqpE8AY0sUYQQBCNrlVO107Lquna9yflTkCxyB04Ggz0rkYr3sXgRBysOrZnvq836i/Hjibk2lOktnZ+laKbs0Vne26cBK+ljxLYs/HR8/dW46D0Sp6jeY+eolFNzpworu6PNAQFDeSqeCUSBwK05xszWsN6tmTF4pGM49oyZWGtDx4LUY3WFcEg9RIAnqW3b1ORpNkdx449y2DLrtIVxVwtpo7oheu/50kDewt0/L9HuPcO8lDXSc94AIEX4eC97lKALDBwJdTFrqtsd0H/DYaAR0Is0wcmVAXmCbNE9OAi1zYXcoINK1olW3QA6ST1QzgSZ8dH710F4PNUpIF9yWJlwEgQlNBcq/F1DXIY0KNZEH2yCBJYmqjD9TxdEY3lnCuGVytNW5CvTX6RgV2wZJSU0f0e/S5wY5YIctpEy0ZfQVbcK2xq2fg51ZjqljWMGGwdCC5NuEzTctuqwEiYNcpcbqr0nLQ+TbJLBWemsVGWUGlaKvooGWr3KyquUjSONLJZ/vLKoM0r/bpCTatld0ni3AK+UpUEsfMwx4gFkVC8JDDbgGihMWEwgDmHAQYEjEWhyYXTuos8nZ5RLlHev6BeoP0lYr29u7AFQlB3PbuNTOjTRyWR87HR2fug+VZCxquONIlb1jCnxAitwrrEAyebEy8QRd5HXtDcxk7uXkbl52vwWgtUUeEHmHm0J6eDPRrd8nzrO4iF4QgtY/hoIDccwsKOE4wuYW2cYcYntPyo6c1SPuusHsWr4hMlnH5GfsMlGBjC08TKGD2/5Pzxr3nNHExb+5KxqmxST2Ym3aPyOeu48zLY/69YGuhTl182raGjDXgtNGmeVPpsHwfchvRWkD/6bj8AigmRQ2bBAAA"
}

def load_dataset(filename, module_name="00 - Introduccion al Data Mining"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo restaura automáticamente de forma exacta.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("..", "..", "data", filename),
        os.path.join("Data Mining", module_name, "data", filename),
        os.path.join("..", "Data Mining", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path

    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Mining/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)

    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]

    print(f"[INFO] Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=2) as response:
                if response.status == 200:
                    with open(target_path, "wb") as out_f:
                        out_f.write(response.read())
                    print(f"[OK] Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue

    if filename in _DATA_EMBEDDED:
        print(f"[INFO] Restaurando '{filename}' de forma autónoma...")
        data_bytes = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target_path, "wb") as out_f:
            out_f.write(data_bytes)
        print(f"[OK] Dataset '{filename}' preparado exitosamente en '{target_path}'.")
        return target_path

    raise FileNotFoundError(f"No se pudo encontrar ni generar el dataset '{filename}'.")
````

## 2. Demostración Práctica: Las 5 Tareas en un Mismo Entorno Empresarial 💻🏢

Generamos un conjunto de datos transaccionales con 800 clientes de retail con información de edad, compras anuales, satisfacción y productos adquiridos:

````python
np.random.seed(42)
n = 800

edad = np.random.randint(18, 70, n)
ingresos_k = np.random.normal(50, 15, n).clip(min=15, max=120)
gasto_anual_k = (0.6 * ingresos_k + np.random.normal(0, 8, n)).clip(min=5)
frecuencia_visitas = np.random.poisson(6, n)

# 1. Variable objetivo discreta: Cliente Premium (0: No, 1: Sí)
es_premium = ((gasto_anual_k > 35) & (frecuencia_visitas >= 6)).astype(int)

df_demo = pd.DataFrame({
    'edad': edad,
    'ingresos_k_usd': ingresos_k.round(1),
    'gasto_anual_k_usd': gasto_anual_k.round(1),
    'frecuencia_visitas': frecuencia_visitas,
    'cliente_premium': es_premium
})

print("Datos del Ecosistema Empresarial:")
display(df_demo.head(5))
````

````python
# =============================================================================
# TAREA 1: CLASIFICACIÓN SUPERVISADA (Predecir si un cliente es Premium)
# =============================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X_clf = df_demo[['ingresos_k_usd', 'frecuencia_visitas']]
y_clf = df_demo['cliente_premium']

clf_modelo = LogisticRegression().fit(X_clf, y_clf)
acc_clf = clf_modelo.score(X_clf, y_clf)
print(f"✅ [TAREA 1: CLASIFICACIÓN] Exactitud del modelo logístico: {acc_clf*100:.2f}%")
````

````python
# =============================================================================
# TAREA 2: REGRESIÓN SUPERVISADA (Estimar el Gasto Anual Continuo)
# =============================================================================
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score

X_reg = df_demo[['ingresos_k_usd', 'edad', 'frecuencia_visitas']]
y_reg = df_demo['gasto_anual_k_usd']

reg_modelo = Ridge().fit(X_reg, y_reg)
r2 = r2_score(y_reg, reg_modelo.predict(X_reg))
print(f"✅ [TAREA 2: REGRESIÓN] Coeficiente de Determinación R²: {r2:.4f}")
````

````python
# =============================================================================
# TAREA 3: CLUSTERING NO SUPERVISADO (Descubrir Arquetipos sin Etiquetas)
# =============================================================================
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X_clust_std = StandardScaler().fit_transform(df_demo[['ingresos_k_usd', 'gasto_anual_k_usd']])
kmeans = KMeans(n_clusters=3, random_state=42).fit(X_clust_std)
df_demo['cluster'] = kmeans.labels_

print(f"✅ [TAREA 3: CLUSTERING] Clientes agrupados en {len(set(kmeans.labels_))} conglomerados naturales:")
display(df_demo.groupby('cluster')[['ingresos_k_usd', 'gasto_anual_k_usd']].mean().round(2))
````

````python
# =============================================================================
# TAREA 4: REGLAS DE ASOCIACIÓN (Afinidad Conjunta de Productos)
# =============================================================================
# Simulación de transacciones de compra
transacciones = [
    ['Cafe', 'Leche', 'Pan'],
    ['Cafe', 'Pan'],
    ['Leche', 'Cereal'],
    ['Cafe', 'Leche', 'Pan', 'Mantequilla'],
    ['Cerveza', 'Pañales'],
    ['Cerveza', 'Pañales', 'Papas'],
    ['Cafe', 'Pan', 'Mantequilla']
]

# Cálculo manual de Soporte y Confianza para: {Cafe, Pan} => {Leche}
total_t = len(transacciones)
n_cafe_pan = sum(1 for t in transacciones if 'Cafe' in t and 'Pan' in t)
n_cafe_pan_leche = sum(1 for t in transacciones if 'Cafe' in t and 'Pan' in t and 'Leche' in t)

soporte = n_cafe_pan_leche / total_t
confianza = n_cafe_pan_leche / n_cafe_pan
print(f"✅ [TAREA 4: REGLAS DE ASOCIACIÓN] Regla: {{Cafe, Pan}} => {{Leche}}")
print(f"• Soporte  : {soporte*100:.1f}%")
print(f"• Confianza: {confianza*100:.1f}%")
````

````python
# =============================================================================
# TAREA 5: DETECCIÓN DE ANOMALÍAS (Aislamiento de Casos Atípicos / Outliers)
# =============================================================================
from sklearn.ensemble import IsolationForest

iso = IsolationForest(contamination=0.03, random_state=42)
anomalias = iso.fit_predict(X_clust_std)
df_demo['es_anomalia'] = (anomalias == -1).astype(int)

print(f"✅ [TAREA 5: ANOMALÍAS] Clientes sospechosos aislados: {df_demo['es_anomalia'].sum()} observaciones")
````

````python
# =============================================================================
# VISUALIZACIÓN INTEGRADA DE LAS 5 TAREAS
# =============================================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Panel 1: Clasificación (Premium vs Estándar)
sns.scatterplot(data=df_demo, x='ingresos_k_usd', y='gasto_anual_k_usd', hue='cliente_premium', palette=['#94a3b8', '#dc2626'], ax=axes[0], s=50)
axes[0].set_title('1. Clasificación: Clientes Premium', fontweight='bold')

# Panel 2: Clustering (3 Conglomerados)
sns.scatterplot(data=df_demo, x='ingresos_k_usd', y='gasto_anual_k_usd', hue='cluster', palette='viridis', ax=axes[1], s=50)
axes[1].set_title('2. Clustering: 3 Grupos Descubiertos', fontweight='bold')

# Panel 3: Detección de Anomalías (Outliers)
sns.scatterplot(data=df_demo, x='ingresos_k_usd', y='gasto_anual_k_usd', hue='es_anomalia', palette=['#2563eb', '#f59e0b'], ax=axes[2], s=50)
axes[2].set_title('3. Anomalías: Valores Atípicos Aislados', fontweight='bold')

plt.tight_layout()
plt.show()
````

### 2.2 Bonus: Aplicando las Tareas 2 y 3 a Datos Reales (con sus Problemas de Calidad) 🧹📈

Los ejemplos de la Sección 2 usaron datos sintéticos perfectamente limpios. En la práctica, antes de aplicar cualquier tarea de la taxonomía hay que pasar primero por las Fases 1 y 2 del proceso KDD (Cuaderno 00). Repetimos aquí, de forma abreviada, la **Tarea 2 (Regresión)** y la **Tarea 3 (Clustering)** sobre `auditoria_calidad_datos.csv`, el archivo de clientes reales de este módulo.

````python
# =============================================================================
# LIMPIEZA MÍNIMA (Fases 1-2 KDD) SOBRE DATOS REALES
# =============================================================================
df_real = pd.read_csv(load_dataset("auditoria_calidad_datos.csv")).drop_duplicates(subset='id_cliente')
df_real.loc[(df_real['edad'] < 0) | (df_real['edad'] > 110), 'edad'] = np.nan
for col in ['edad', 'ingreso_mensual', 'gasto_anual']:
    df_real[col] = df_real[col].fillna(df_real[col].median())
    q1, q99 = df_real[col].quantile([0.01, 0.99])
    df_real[col] = df_real[col].clip(lower=q1, upper=q99)

# =============================================================================
# TAREA 2 (REVISITADA): REGRESIÓN SOBRE EL GASTO ANUAL REAL
# =============================================================================
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score

X_reg_real = df_real[['edad', 'ingreso_mensual']]
y_reg_real = df_real['gasto_anual']

reg_real = Ridge().fit(X_reg_real, y_reg_real)
r2_real = r2_score(y_reg_real, reg_real.predict(X_reg_real))
print(f"✅ [TAREA 2 · DATOS REALES] R² prediciendo gasto_anual desde edad e ingreso_mensual: {r2_real:.4f}")
print("   Un R² tan bajo también es informativo: a diferencia del caso sintético, el gasto real")
print("   de estos clientes no depende linealmente solo de su edad e ingreso — haría falta más")
print("   información de negocio (categoría de producto, antigüedad, canal) para explicarlo.")
````

````python
# =============================================================================
# TAREA 3 (REVISITADA): CLUSTERING SOBRE CLIENTES REALES
# =============================================================================
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X_clust_real = StandardScaler().fit_transform(df_real[['ingreso_mensual', 'gasto_anual']])
kmeans_real = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_clust_real)
df_real['cluster'] = kmeans_real.labels_

plt.figure(figsize=(8, 4.5))
sns.scatterplot(data=df_real, x='ingreso_mensual', y='gasto_anual', hue='cluster', palette='viridis', s=35)
plt.title('Tarea 3 · Datos Reales: Segmentos de Clientes por Ingreso y Gasto', fontweight='bold')
plt.xlabel('Ingreso Mensual (COP)')
plt.ylabel('Gasto Anual (COP)')
plt.tight_layout()
plt.show()

print(f"✅ [TAREA 3 · DATOS REALES] Clientes agrupados en {df_real['cluster'].nunique()} conglomerados naturales:")
display(df_real.groupby('cluster')[['ingreso_mensual', 'gasto_anual']].mean().round(0))
````

## 4. Resumen Ejecutivo del Cuaderno 02 📌

1. **Supervisado vs. No Supervisado:** Si tienes una columna $y$ histórica que quieres predecir, estás ante **Clasificación** (categórica) o **Regresión** (continua). Si solo tienes descriptores $X$ y quieres encontrar estructura intrínseca, estás ante **Clustering**, **Asociación** o **Anomalías**.
2. **Complementariedad:** En proyectos reales, las tareas se combinan en cascada: por ejemplo, primero se agrupan clientes con *Clustering* y luego se entrena un modelo de *Clasificación* específico para cada segmento.
3. **La taxonomía se valida en el mundo real:** el catálogo de 8 casos de negocio de la Sección 2.1 confirmó que la pregunta correcta ("¿existe una etiqueta histórica?", "¿la salida es discreta o continua?") basta para ubicar cualquier problema nuevo dentro de las 5 familias.
4. **La taxonomía sobrevive el contacto con datos sucios:** la Sección 2.2 repitió la Tarea 2 (Regresión) y la Tarea 3 (Clustering) sobre `auditoria_calidad_datos.csv` ya limpio; el R² casi nulo de la regresión no invalida la tarea elegida, solo confirma que las variables disponibles no bastan para explicar el gasto — una lección de negocio tan válida como un buen ajuste.
