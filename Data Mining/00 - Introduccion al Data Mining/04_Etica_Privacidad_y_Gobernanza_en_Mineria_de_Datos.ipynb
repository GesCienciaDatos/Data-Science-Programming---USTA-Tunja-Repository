## 1. El Dilema Ético y el Sesgo Algorítmico (*Algorithmic Bias*) ⚠️

Existe el mito de que los algoritmos de Machine Learning y Minería de Datos son matemáticamente neutrales y objetivos. La realidad es contundente: **los algoritmos no aprenden lo que debería ser, sino lo que históricamente ha sido**.

Si los datos de entrenamiento contienen discriminación histórica humana, el algoritmo optimizará su función de costo perpetuando e incluso amplificando ese sesgo de forma sistemática y a escala masiva.

### Casos Emblemáticos de Falla Ética en la Industria:
1. **Amazon AI Recruitment Tool (2018):** Un modelo de filtrado de hojas de vida penalizaba sistemáticamente a candidatas mujeres para puestos de ingeniería, debido a que el modelo fue entrenado con 10 años de currículums predominantemente masculinos.
2. **COMPAS en el Sistema Penal de EE.UU. (2016):** El algoritmo asignaba puntuaciones de riesgo de reincidencia falsamente elevadas al doble de reclusos afroamericanos respecto a caucásicos con historiales equivalentes.
3. **Credit Scoring y Código Postal (Redlining Digital):** Aunque no se incluya la etnia o género, el código postal o nivel socioeconómico de la manzana actúa como un proxy indirecto de discriminación (*Proxy Variable*).

### Definiciones Matemáticas de Equidad (*Fairness*):

* **Paridad Demográfica (*Demographic Parity*):** La tasa de decisión positiva debe ser idéntica para todos los grupos protegidos:
  $$P(\hat{Y}=1 \mid A=0) = P(\hat{Y}=1 \mid A=1)$$
* **Igualdad de Oportunidades (*Equal Opportunity*):** La tasa de verdaderos positivos (Sensibilidad) debe ser igual entre grupos protegidos:
  $$P(\hat{Y}=1 \mid A=0, Y=1) = P(\hat{Y}=1 \mid A=1, Y=1)$$

## 2. Privacidad de Datos y Ataques de Reidentificación 🔐🕵️

### El Experimento Histórico de Latanya Sweeney (2000):
El gobernador de Massachusetts aseguró al público que la base de datos médica de empleados públicos estaba "completamente anonimizada" porque se habían eliminado nombres y números de seguro social.

La investigadora **Latanya Sweeney** compró el registro electoral público de Cambridge por $20 USD y cruzó ambas bases utilizando únicamente tres atributos inocuos conocidos como **Cuasi-Identificadores (*Quasi-Identifiers*)**:
1. Código Postal (5 dígitos)
2. Fecha de Nacimiento (Día/Mes/Año)
3. Género

> 💥 **El Hallazgo:** ¡Solo una persona en toda la ciudad de Cambridge compartía el código postal, fecha de nacimiento y género del gobernador! Sweeney localizó su historial médico completo con diagnósticos y prescripciones y se lo envió por correo a su despacho.
> Demostró formalmente que el **87% de toda la población de Estados Unidos** es unívocamente identificable combinando solo esos tres datos.

### Modelos Matemáticos de Protección de la Privacidad:

| Modelo | Definición Formal | Mecanismo de Defensa |
|---|---|---|
| **$k$-Anonimato ($k$-Anonymity)** | Cada combinación de cuasi-identificadores en la base debe ser idéntica para al menos $k$ personas. | Generalización (ej. agrupar edades en rangos [30-40]) y supresión de columnas. |
| **$l$-Diversidad ($l$-Diversity)** | Dentro de cada grupo de $k$ individuos indistinguibles, el atributo sensible (ej. enfermedad) debe tener al menos $l$ valores distintos bien distribuidos. | Evita ataques de homogeneidad (si los 5 miembros de un grupo tienen cáncer, el $k$-anonimato falla). |
| **Privacidad Diferencial ($\varepsilon$-DP)** | Garantiza matemáticamente que la presencia o ausencia de un individuo en la base no altera significativamente el resultado de ninguna consulta analítica. | Inyección de ruido aleatorio calibrado de Laplace o Gauss sobre los resultados agregados. |

| Marco Legal | Jurisdicción | Principios Clave para Minería de Datos |

Prohibición estricta de tratamiento de datos sensibles sin consentimiento previo y explícito. |
| **GDPR (Reglamento General de Protección de Datos)** | Unión Europea (Impacto Global) | Artículo 22: "Derecho a no ser objeto de una decisión basada únicamente en el tratamiento automatizado".
````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import hashlib, os, warnings
warnings.filterwarnings('ignore')

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (9, 4.5)

print("🚀 Entorno de Ética, Privacidad y Gobernanza configurado con éxito.")

import os, urllib.request, urllib.parse, gzip, base64

_DATA_EMBEDDED = {
    "auditoria_calidad_datos.csv": ""
}

def load_dataset(filename, module_name="00 - Introduccion al Data Mining"):
    """
    Carga o descarga de forma segura el dataset para ejecución local o en Google Colab.
    Si no se encuentra localmente ni en GitHub, lo restaura automáticamente de forma exacta.
    """
    candidates = [
        os.path.join("data", filename),
        os.path.join(module_name, "data", filename),
        os.path.join("..", "data", filename),
        os.path.join("..", module_name, "data", filename),
        os.path.join("..", "..", "data", filename),
        os.path.join("Data Mining", module_name, "data", filename),
        os.path.join("..", "Data Mining", module_name, "data", filename),
        filename
    ]
    for path in candidates:
        if os.path.exists(path):
            return path

    os.makedirs("data", exist_ok=True)
    target_path = os.path.join("data", filename)
    folder_path = f"Data Mining/{module_name}"
    encoded_folder = urllib.parse.quote(folder_path)
    encoded_file = urllib.parse.quote(filename)

    urls = [
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/main/{encoded_folder}/data/{encoded_file}",
        f"https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/master/{encoded_folder}/data/{encoded_file}"
    ]

    print(f"[INFO] Intentando descargar '{filename}' desde el repositorio oficial...")
    for url in urls:
        try:
            with urllib.request.urlopen(url, timeout=2) as response:
                if response.status == 200:
                    with open(target_path, "wb") as out_f:
                        out_f.write(response.read())
                    print(f"[OK] Dataset '{filename}' descargado exitosamente.")
                    return target_path
        except Exception:
            continue

    if filename in _DATA_EMBEDDED:
        print(f"[INFO] Restaurando '{filename}' de forma autónoma...")
        data_bytes = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target_path, "wb") as out_f:
            out_f.write(data_bytes)
        print(f"[OK] Dataset '{filename}' preparado exitosamente en '{target_path}'.")
        return target_path

    raise FileNotFoundError(f"No se pudo encontrar ni generar el dataset '{filename}'.")
````

## 4. Laboratorio 1: Seudonimización Irreversible con Hashing Criptográfico y Sal (*Salt*) 💻🔐

Una simple función hash (`SHA-256`) puede vulnerarse mediante **tablas arcoíris (*Rainbow Tables*)** si los números de documento son secuenciales. La técnica obligatoria según el estándar GDPR y la SIC es utilizar **Hashing Criptográfico con Sal Secreta (*Salted Hashing*)**:

````python
def seudonimizar_identificador(documento, sal_secreta="USTA_TUNJA_CIENCIA_DATOS_2026"):
    """
    Transforma un identificador personal en un token seudonimizado irreversible.
    """
    texto_completo = f"{documento}_{sal_secreta}"
    # Generamos hash SHA-256 de 256 bits (64 caracteres hexadecimales)
    token = hashlib.sha256(texto_completo.encode('utf-8')).hexdigest()[:16]
    return token

# Demostración con cédulas colombianas reales simuladas
cedulas_ejemplo = ["1052384910", "1052384911", "79543210", "24890123"]

df_seguro = pd.DataFrame({
    'cedula_original (Dato Sensible)': cedulas_ejemplo,
    'token_seudonimizado (Apto para Minería)': [seudonimizar_identificador(c) for c in cedulas_ejemplo]
})

print("Resultado de Seudonimización Criptográfica:")
display(df_seguro)
print("💡 El token es consistente (la misma cédula siempre da el mismo hash para hacer joins), pero matemáticamente imposible de revertir a la cédula original sin la sal secreta.")
````

## 5. Laboratorio 2: Verificación y Refuerzo de k-Anonimato sobre Datos Reales 🔐📊

La Sección 2 introdujo el $k$-anonimato de forma teórica. Lo llevamos ahora a la práctica sobre `auditoria_calidad_datos.csv`, usando `edad` y `ciudad` como **cuasi-identificadores**. Un dataset satisface $k$-anonimato si toda combinación de cuasi-identificadores está compartida por al menos $k$ individuos; en caso contrario, un atacante que conozca esos dos atributos de una persona puede aislarla en la base —tal como hizo Sweeney con el gobernador de Massachusetts.

````python
# Cuasi-identificadores tal como fueron capturados originalmente (edad exacta x ciudad)
path_audit = load_dataset("auditoria_calidad_datos.csv")
df_priv = pd.read_csv(path_audit)[['id_cliente', 'edad', 'ciudad']].dropna()
df_priv['ciudad'] = df_priv['ciudad'].astype(str).str.strip().str.upper()

grupos_qi = df_priv.groupby(['edad', 'ciudad']).size().reset_index(name='tamano_grupo')
k_minimo = grupos_qi['tamano_grupo'].min()
k_deseado = 5
n_grupos_vulnerables = (grupos_qi['tamano_grupo'] < k_deseado).sum()

print("="*65)
print("🔓 RIESGO DE REIDENTIFICACIÓN CON CUASI-IDENTIFICADORES CRUDOS:")
print(f"• k mínimo observado (edad exacta x ciudad)  : {k_minimo}")
print(f"• Combinaciones con menos de k={k_deseado} personas    : {n_grupos_vulnerables} de {len(grupos_qi)}")
print("="*65)
display(grupos_qi.sort_values('tamano_grupo').head(5))
````

````python
# Generalización: agrupamos la edad exacta en rangos decenales (técnica estándar de k-anonimato)
df_priv['edad_generalizada'] = pd.cut(df_priv['edad'], bins=range(0, 111, 10), right=False).astype(str)

grupos_qi_gen = df_priv.groupby(['edad_generalizada', 'ciudad']).size().reset_index(name='tamano_grupo')
k_minimo_gen = grupos_qi_gen['tamano_grupo'].min()
n_grupos_vulnerables_gen = (grupos_qi_gen['tamano_grupo'] < k_deseado).sum()

print("="*65)
print("🔒 RIESGO TRAS GENERALIZACIÓN DE LA EDAD (RANGOS DE 10 AÑOS):")
print(f"• k mínimo tras generalización                : {k_minimo_gen}")
print(f"• Combinaciones con menos de k={k_deseado} personas    : {n_grupos_vulnerables_gen} de {len(grupos_qi_gen)}")
print("="*65)

if n_grupos_vulnerables_gen > 0:
    print("⚠️ Aún quedan combinaciones vulnerables: se requeriría generalizar también 'ciudad' (ej. por región) o suprimir esos registros residuales.")
else:
    print("✅ El dataset generalizado satisface k-anonimato para el umbral definido.")
````

## 6. Laboratorio 3: Auditoría y Mitigación de Sesgo Demográfico (*Fairness Audit*) ⚖️

Simulamos un modelo de otorgamiento de créditos bancarios para evaluar si existe discriminación por género:

````python
np.random.seed(42)
n_solicitudes = 2000

genero = np.random.choice(['Masculino', 'Femenino'], n_solicitudes, p=[0.52, 0.48])
score_financiero = np.random.normal(650, 70, n_solicitudes)

# Inyección de sesgo histórico en los datos de aprobación
prob_aprobacion = (score_financiero > 640).astype(float)
# Penalización artificial histórica hacia las mujeres
prob_aprobacion[genero == 'Femenino'] *= 0.72
aprobado = (np.random.rand(n_solicitudes) < prob_aprobacion).astype(int)

df_auditoria_etica = pd.DataFrame({
    'genero': genero,
    'score_financiero': score_financiero.round(1),
    'credito_aprobado': aprobado
})

# Cálculo de la Métrica de Impacto Dispar (Disparate Impact Ratio - Regla del 80% de la EEOC)
tasa_hombres = df_auditoria_etica[df_auditoria_etica['genero'] == 'Masculino']['credito_aprobado'].mean()
tasa_mujeres = df_auditoria_etica[df_auditoria_etica['genero'] == 'Femenino']['credito_aprobado'].mean()
disparate_impact = tasa_mujeres / tasa_hombres

print("="*65)
print("📊 AUDITORÍA DE EQUIDAD DEMOGRÁFICA (FAIRNESS AUDIT):")
print(f"• Tasa de Aprobación en Hombres: {tasa_hombres*100:.2f}%")
print(f"• Tasa de Aprobación en Mujeres: {tasa_mujeres*100:.2f}%")
print(f"• Disparate Impact Ratio (DIR)  : {disparate_impact:.3f}")
print("="*65)

if disparate_impact < 0.80:
    print("🚨 ALERTA LEGAL: El modelo viola la regla del 80% de no discriminación (DIR < 0.80). El modelo NO puede salir a producción.")
else:
    print("✅ Cumplimiento de paridad demográfica dentro de márgenes aceptables.")
````

````python
# Visualización del sesgo por grupo protegido
plt.figure(figsize=(7, 4.5))
sns.barplot(data=df_auditoria_etica, x='genero', y='credito_aprobado', palette=['#2563eb', '#ec4899'], ci=None)
plt.axhline(tasa_hombres * 0.80, color='red', linestyle='--', label=f'Umbral Mínimo Legal (80% de Hombres: {tasa_hombres*0.80*100:.1f}%)')
plt.title('Disparidad Demográfica en Aprobación de Créditos', fontweight='bold')
plt.ylabel('Tasa de Aprobación')
plt.xlabel('Grupo Protegido (Género)')
plt.legend()
plt.show()
````

## 7. Lista de Chequeo Ético para Científicos de Datos (*Ethics Checklist*) 📋✅

Antes de pasar cualquier modelo de minería a un entorno productivo, valida los siguientes 6 puntos:

1. [ ] **Consentimiento Informado:** ¿Los titulares autorizaron explícitamente el tratamiento de sus datos para esta finalidad analítica (Habeas Data)?
2. [ ] **Supresión de Identificadores Directos:** ¿Se eliminaron nombres, teléfonos y direcciones, aplicando seudonimización criptográfica?
3. [ ] **Auditoría de Variables Proxy:** ¿Se excluyeron códigos postales, zonas residenciales o apellidos que sirvan como sustitutos de género, raza o nivel socioeconómico?
4. [ ] **Test de Impacto Dispar:** ¿El ratio de selección del grupo minoritario supera el 80% respecto al grupo mayoritario?
5. [ ] **Explicabilidad de Decisiones:** ¿Existe un mecanismo para explicarle a un usuario exactamente por qué fue rechazado su crédito o servicio?
6. [ ] **Monitoreo de Deriva Ética:** ¿Se cuenta con alertas en producción si el modelo comienza a sesgarse con datos nuevos?
