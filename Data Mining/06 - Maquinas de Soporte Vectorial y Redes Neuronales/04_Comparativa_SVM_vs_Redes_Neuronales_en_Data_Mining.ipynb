## 1. Introducción: Cerrando el Módulo con una Confrontación Empírica ⚖️🔬

Los Cuadernos **00** y **01** de este módulo agotaron la familia de las Máquinas de Soporte Vectorial (margen máximo, margen blando, y el truco del kernel para fronteras no lineales y regresión). Los Cuadernos **02** y **03** hicieron lo mismo con las Redes Neuronales (propagación hacia adelante, retropropagación, optimizadores y regularización). Ambas familias resuelven, en el fondo, el mismo tipo de problema — clasificación y regresión sobre datos tabulares — pero llegan ahí por caminos matemáticos completamente distintos. Este cuaderno cierra el módulo confrontando ambas familias directamente, no con argumentos teóricos sueltos, sino con **evidencia empírica real** sobre los mismos datasets usados en los cuadernos anteriores.

La tabla siguiente resume las diferencias estructurales que el resto del cuaderno pondrá a prueba una por una:

| Propiedad | Support Vector Machines (SVM) | Redes Neuronales Artificiales (ANN / MLP) |
|---|---|---|
| **Espacio de Optimización** | Convexa (Garantiza mínimo global único) | No Convexa (Susceptible a mínimos locales y puntos de silla) |
| **Tamaño Muestral Ideal** | Muestras pequeñas/medianas ($N < 50,000$), alta dimensión | Muestras grandes a masivas ($N > 100,000$) |
| **Escalabilidad** | $O(N^2)$ a $O(N^3)$ (computacionalmente costoso en Big Data) | $O(N)$ mediante Minibatch SGD / Adam |
| **Hiperparámetros Clave** | $C, \gamma, \text{kernel}$ | Topología, Tasa de aprendizaje, $\alpha$, Épocas, Activación |
| **Interpretabilidad** | Vectores de soporte delimitan la frontera | Caja negra con representaciones latentes distribuidas |

Cuatro preguntas concretas organizan el resto del cuaderno: **(2)** ¿la ventaja teórica de convexidad de SVM se traduce en reproducibilidad real frente a la no convexidad de una red?, **(3)** ¿la complejidad $O(N^2)$–$O(N^3)$ de SVM realmente se nota en tiempos de entrenamiento reales frente a una red?, **(4)** ¿cómo maneja cada familia el aumento de dimensiones?, y **(5)** en un enfrentamiento directo y ajustado (`GridSearchCV` para ambos) sobre dos datasets reales del módulo, ¿cuál gana?

## 2. Optimización: Superficie Convexa vs. No Convexa — ¿Determinismo Real? 🎲

El problema de optimización que resuelve `SVC` (minimizar $\frac{1}{2}\|w\|^2 + C\sum \xi_i$, Cuaderno 00) es **convexo**: tiene, para un mismo conjunto de datos y unos mismos hiperparámetros, un **único mínimo global**, sin importar desde dónde arranque el algoritmo de optimización. El problema que resuelve `MLPClassifier` —minimizar una función de pérdida sobre miles de pesos organizados en capas, conectados por transformaciones no lineales (Cuaderno 03)— es, en cambio, **no convexo**: su superficie de pérdida tiene múltiples mínimos locales, puntos de silla y mesetas, y el punto exacto donde el descenso de gradiente termina depende de dónde empezó — es decir, de la **inicialización aleatoria de los pesos**, controlada en `scikit-learn` por el parámetro `random_state`.

En teoría, esto predice algo muy concreto y verificable: entrenar la **misma** arquitectura de `MLPClassifier` varias veces, cambiando solo `random_state`, debería producir modelos ligeramente distintos (distintas pérdidas finales, distintas exactitudes); entrenar el **mismo** `SVC` varias veces, cambiando solo `random_state`, debería producir exactamente el mismo modelo cada vez —la solución convexa no depende del punto de partida—. **Lectura:** entrenar exactamente la misma arquitectura (`hidden_layer_sizes=(16, 8)`) siete veces, cambiando únicamente `random_state`, produce **siete resultados distintos**: la exactitud de prueba oscila entre **94.15%** y **95.91%** (rango de 1.75 puntos porcentuales), con una media de **95.07%** y una desviación estándar de 0.53 puntos. Nada de esto es un error ni ruido de medición: es la consecuencia directa y esperada de una superficie de pérdida no convexa — cada inicialización aleatoria de pesos arranca el descenso de gradiente en un punto distinto de esa superficie, y termina en un mínimo local ligeramente distinto cada vez. La celda siguiente repite exactamente el mismo experimento, pero con `SVC`.

````python
print("--- SVC(kernel='rbf'): mismo modelo, distinto random_state ---")
svc_accs = []
for s in semillas:
    modelo = SVC(kernel="rbf", C=1.0, gamma="scale", random_state=s)
    modelo.fit(X_diag_tr_s, y_diag_tr)
    acc = modelo.score(X_diag_te_s, y_diag_te)
    svc_accs.append(acc)
    print(f"random_state={s}   exactitud_prueba={acc:.4f}")

print(f"\nExactitud SVC -> media={np.mean(svc_accs):.4f}  desv.est.={np.std(svc_accs):.4f}  "
      f"min={np.min(svc_accs):.4f}  max={np.max(svc_accs):.4f}  rango={np.max(svc_accs)-np.min(svc_accs):.4f}")
````

**Lectura:** el contraste es total. Las siete corridas de `SVC` con distinto `random_state` producen **exactamente el mismo resultado**: 95.32% de exactitud de prueba, sin ninguna variación (desviación estándar = 0.0000). Esto confirma la teoría de forma directa: el `random_state` de `SVC` solo afecta procesos auxiliares internos ajenos al problema de optimización principal (como el barajado usado para calibrar probabilidades cuando `probability=True`, que aquí no se activó); la solución del problema convexo de márgenes es **determinista** dado un mismo conjunto de datos e hiperparámetros — no existe "punto de partida" que pueda cambiar a dónde converge. El siguiente gráfico compara visualmente la dispersión real de ambos conjuntos de resultados.

## 3. Escalabilidad: Tiempo de Entrenamiento vs. Tamaño de Muestra ⏱️

La tabla de la Sección 1 afirma que SVM escala aproximadamente $O(N^2)$ a $O(N^3)$ con el número de muestras $N$ (el costo de construir y operar sobre la matriz de kernel, de tamaño $N \times N$), mientras que una red neuronal entrenada con descenso de gradiente por mini-lotes escala mucho mejor, cerca de $O(N)$. Esta sección lo mide directamente, no lo asume: se entrena `SVC(kernel="rbf")` y `MLPClassifier` sobre submuestras crecientes de `rendimiento_industrial_no_lineal.csv` (2000 filas en total), cronometrando cada ajuste con `time.perf_counter()`.

El dataset original tiene una variable objetivo continua (`eficiencia_proceso`, usada como regresión en el Cuaderno 01). Para poder comparar tiempos de entrenamiento de **clasificadores** en igualdad de condiciones (el objetivo de este experimento es medir tiempo, no exactitud), se binariza `eficiencia_proceso` respecto a su mediana: `eficiencia_alta = 1` si está por encima de la mediana, `0` en caso contrario — un split balanceado por construcción que no favorece a ningún modelo. Como el tiempo de reloj (`time.perf_counter()`) es sensible al ruido del sistema operativo (otros procesos, variaciones de caché), cada ajuste se repite **5 veces** por tamaño y se conserva el tiempo **mínimo** observado — la práctica estándar de microbenchmarking para aislar el costo real del algoritmo del ruido circunstancial de una sola medición.

````python
df_ind = pd.read_csv(load_dataset("rendimiento_industrial_no_lineal.csv"))

mediana_efic = df_ind["eficiencia_proceso"].median()
df_ind["eficiencia_alta"] = (df_ind["eficiencia_proceso"] >= mediana_efic).astype(int)

print("Forma del dataset:", df_ind.shape)
print("Mediana de eficiencia_proceso:", round(mediana_efic, 4))
print("\nDistribucion de 'eficiencia_alta':")
print(df_ind["eficiencia_alta"].value_counts().sort_index())

feat_ind = ["temperatura_c", "presion_bar", "flujo_l_min", "rpm_motor", "concentracion_reactivo"]
X_ind = df_ind[feat_ind].values
y_ind = df_ind["eficiencia_alta"].values
````

````python
tamanos = [200, 500, 1000, 2000]
resultados_tiempo = []
REPETICIONES = 5  # se repite cada ajuste y se conserva el tiempo minimo (microbenchmarking estandar)

rng = np.random.RandomState(42)
indices_barajados = rng.permutation(len(X_ind))

for n in tamanos:
    idx = indices_barajados[:n]
    Xs, ys = X_ind[idx], y_ind[idx]
    Xs_tr, Xs_te, ys_tr, ys_te = train_test_split(Xs, ys, test_size=0.3, random_state=42, stratify=ys)

    scaler_n = StandardScaler().fit(Xs_tr)
    Xs_tr_s = scaler_n.transform(Xs_tr)
    Xs_te_s = scaler_n.transform(Xs_te)

    tiempos_svc = []
    for _ in range(REPETICIONES):
        t0 = time.perf_counter()
        svc_n = SVC(kernel="rbf", C=1.0, gamma="scale")
        svc_n.fit(Xs_tr_s, ys_tr)
        tiempos_svc.append(time.perf_counter() - t0)
    t_svc = min(tiempos_svc)
    acc_svc = svc_n.score(Xs_te_s, ys_te)

    tiempos_mlp = []
    for _ in range(REPETICIONES):
        t0 = time.perf_counter()
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            mlp_n = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=1000, random_state=42)
            mlp_n.fit(Xs_tr_s, ys_tr)
        tiempos_mlp.append(time.perf_counter() - t0)
    t_mlp = min(tiempos_mlp)
    acc_mlp = mlp_n.score(Xs_te_s, ys_te)

    resultados_tiempo.append((n, t_svc, acc_svc, t_mlp, acc_mlp))
    print(f"N={n:5d}   SVC (min de {REPETICIONES}): {t_svc:.4f}s (acc={acc_svc:.4f})   "
          f"MLP (min de {REPETICIONES}): {t_mlp:.4f}s (acc={acc_mlp:.4f})")

ratio_svc = resultados_tiempo[-1][1] / resultados_tiempo[0][1]
ratio_mlp = resultados_tiempo[-1][3] / resultados_tiempo[0][3]
print(f"\nCrecimiento del tiempo, N=200 -> N=2000 (factor 10x en N):")
print(f"  SVC: tiempo x{ratio_svc:.2f}")
print(f"  MLP: tiempo x{ratio_mlp:.2f}")
````

**Lectura:** en términos **absolutos**, `MLPClassifier` es más lento que `SVC` en las cuatro submuestras — más de 100 veces más lento en tiempo bruto en $N=2000$ (2.99s de MLP frente a 0.025s de SVC) — porque una red que itera cientos de veces sobre los datos tiene un costo fijo por iteración considerablemente mayor que resolver el problema cuadrático de SVM en este rango modesto de tamaños. Pero la pregunta de esta sección no es "¿quién es más rápido en total?", sino "¿cómo **crece** cada uno al multiplicar $N$ por 10?" — y ahí la respuesta favorece la teoría: al pasar de $N=200$ a $N=2000$ (10 veces más datos), el tiempo mínimo de `SVC` se multiplica por **15.06**, mientras que el de `MLPClassifier` se multiplica por solo **7.28**. Si SVM escalara linealmente (como MLP se acerca a hacerlo aquí), su tiempo debería crecer también cerca de 10x; en cambio crece más rápido que $N$ — coherente con una complejidad superlineal ($O(N^2)$ o peor), aunque a esta escala modesta (miles, no millones de filas) todavía está lejos de ser prohibitivo. Repetir este mismo experimento varias veces mostró que el factor exacto varía un poco de una corrida a otra (ruido normal de medir tiempo de reloj, típicamente entre 8x y 15x para SVC y entre 6x y 8x para MLP), pero el patrón cualitativo —SVC crece proporcionalmente más rápido que MLP al aumentar $N$— se mantuvo consistente en todas las repeticiones. La honestidad del resultado importa: a $N=2000$, SVM sigue siendo, en términos absolutos, la opción más rápida — la ventaja teórica de las redes neuronales en escalabilidad solo se vuelve decisiva en el régimen de $N$ mucho más grande ($N > 100{,}000$, según la Sección 1) que ningún dataset de este módulo alcanza. El siguiente gráfico visualiza ambas curvas de crecimiento.

````python
tam_arr = [r[0] for r in resultados_tiempo]
t_svc_arr = [r[1] for r in resultados_tiempo]
t_mlp_arr = [r[3] for r in resultados_tiempo]

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

axes[0].plot(tam_arr, t_svc_arr, marker="o", color="#64748b", label="SVC (kernel RBF)")
axes[0].plot(tam_arr, t_mlp_arr, marker="s", color="#8b5cf6", label="MLPClassifier")
axes[0].set_xlabel("N (numero de muestras)")
axes[0].set_ylabel("Tiempo de entrenamiento (s)")
axes[0].set_title("Tiempo absoluto vs. N")
axes[0].legend()

t_svc_norm = np.array(t_svc_arr) / t_svc_arr[0]
t_mlp_norm = np.array(t_mlp_arr) / t_mlp_arr[0]
n_norm = np.array(tam_arr) / tam_arr[0]
axes[1].plot(tam_arr, n_norm, marker="^", color="#16a34a", linestyle="--", label="Crecimiento lineal (referencia)")
axes[1].plot(tam_arr, t_svc_norm, marker="o", color="#64748b", label="SVC (normalizado)")
axes[1].plot(tam_arr, t_mlp_norm, marker="s", color="#8b5cf6", label="MLP (normalizado)")
axes[1].set_xlabel("N (numero de muestras)")
axes[1].set_ylabel("Tiempo relativo a N=200")
axes[1].set_title("Crecimiento relativo vs. N")
axes[1].legend()

plt.tight_layout()
plt.show()
````

El panel derecho es el más revelador: normalizando cada curva contra su propio tiempo en $N=200$ (para poder comparar **tasas de crecimiento** en la misma escala), la curva de SVC (gris) se despega claramente por encima de la línea de referencia de crecimiento lineal (verde punteada) a medida que $N$ crece, mientras que la curva de MLP (violeta) se mantiene mucho más cerca de esa referencia lineal. ## 4. Dimensionalidad: Cómo Maneja Cada Modelo el Espacio de Variables 📐

Más allá del número de muestras $N$, el número de **variables** (dimensiones) también afecta de forma distinta a cada familia. Una SVM con kernel RBF (Cuaderno 01) puede funcionar razonablemente bien incluso en espacios de alta dimensión, precisamente gracias al truco del kernel: la frontera de decisión depende solo de productos punto entre pares de puntos, no de construir explícitamente combinaciones de variables, así que agregar variables no dispara necesariamente el número de parámetros que hay que aprender — el modelo "solo" necesita encontrar los vectores de soporte adecuados en ese espacio, sin importar cuán grande sea. Esto hace de SVM una opción competitiva en problemas de alta dimensión con relativamente pocas muestras (el caso clásico es la clasificación de texto o de datos genómicos, donde el número de variables puede superar ampliamente al número de observaciones).

Una red neuronal profunda, en cambio, aprende sus propias transformaciones no lineales **a partir de los datos** (Cuadernos 02 y 03): cada variable adicional en la capa de entrada agrega una columna completa a la primera matriz de pesos $W^{[1]}$, y cada neurona oculta adicional agrega más parámetros todavía. Esa capacidad extra es una ventaja cuando hay suficientes datos para aprovecharla, pero se convierte en un riesgo de sobreajuste cuando el número de variables es grande en relación con el número de muestras disponibles — hay, literalmente, más pesos que ajustar por cada ejemplo de entrenamiento, y sin suficientes datos para restringirlos, el modelo puede memorizar en lugar de generalizar (el mismo fenómeno que la Sección 5 del Cuaderno 03 mostró con `alpha` demasiado bajo, solo que aquí el origen no es la falta de regularización sino la falta de datos). Ningún dataset de este módulo alcanza una dimensionalidad extrema (el máximo es `diagnostico_celular.csv`, con 8 variables), así que esta sección queda como nota conceptual en lugar de forzar un experimento sobre datos que no ilustran genuinamente el fenómeno; la lección práctica para minería de datos es que, ante un problema de **muchas variables y relativamente pocas observaciones**, SVM con kernel suele ser un punto de partida más seguro, mientras que una red neuronal profunda rinde mejor cuando el volumen de datos disponible escala junto con la dimensionalidad.

## 5. Benchmark Empírico Cabeza a Cabeza: Dos Datasets Reales 🥊

Las secciones anteriores probaron propiedades aisladas (reproducibilidad, tiempo). Esta sección hace lo que en la práctica de minería de datos realmente importa: tomar **dos** de los datasets ya usados en este módulo — `diagnostico_celular.csv` (caso limpio, binario, moderadamente desbalanceado) y `sensores_maquinaria.csv` (caso desbalanceado 82/18) — y enfrentar un `SVC` y un `MLPClassifier` **ambos optimizados** con `GridSearchCV` bajo la misma metodología: la misma partición 70/30 estratificada, el mismo escalado, la misma validación cruzada de 5 folds, y una rejilla de búsqueda de tamaño comparable para cada uno. Se registra exactitud, F1 (sobre la clase positiva) y tiempo de entrenamiento del modelo final ya optimizado.

````python
def benchmark_modelo(df, columna_objetivo, columnas_excluir, nombre_dataset):
    features = [c for c in df.columns if c != columna_objetivo and c not in columnas_excluir]
    X = df[features].values
    y = df[columna_objetivo].values

    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
    scaler = StandardScaler().fit(X_tr)
    X_tr_s = scaler.transform(X_tr)
    X_te_s = scaler.transform(X_te)

    # --- SVC optimizado con GridSearchCV ---
    rejilla_svc = {"C": [0.1, 1, 10, 100], "gamma": ["scale", 0.01, 0.1, 1]}
    gs_svc = GridSearchCV(SVC(kernel="rbf"), rejilla_svc, cv=5, scoring="f1", n_jobs=-1)
    gs_svc.fit(X_tr_s, y_tr)
    mejor_svc = gs_svc.best_estimator_

    t0 = time.perf_counter()
    mejor_svc.fit(X_tr_s, y_tr)
    t_svc = time.perf_counter() - t0
    pred_svc = mejor_svc.predict(X_te_s)

    # --- MLPClassifier optimizado con GridSearchCV ---
    rejilla_mlp = {"hidden_layer_sizes": [(16, 8), (32, 16)], "alpha": [0.0001, 0.001, 0.01]}
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        gs_mlp = GridSearchCV(
            MLPClassifier(max_iter=1000, random_state=42, solver="adam"),
            rejilla_mlp, cv=5, scoring="f1", n_jobs=-1,
        )
        gs_mlp.fit(X_tr_s, y_tr)
    mejor_mlp = gs_mlp.best_estimator_

    t0 = time.perf_counter()
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        mejor_mlp.fit(X_tr_s, y_tr)
    t_mlp = time.perf_counter() - t0
    pred_mlp = mejor_mlp.predict(X_te_s)

    print(f"=== {nombre_dataset} ===")
    print(f"SVC   -> mejores params: {gs_svc.best_params_}")
    print(f"        acc={accuracy_score(y_te, pred_svc):.4f}  f1={f1_score(y_te, pred_svc):.4f}  "
          f"recall={recall_score(y_te, pred_svc):.4f}  precision={precision_score(y_te, pred_svc):.4f}  "
          f"tiempo_ajuste={t_svc:.4f}s")
    print(f"MLP   -> mejores params: {gs_mlp.best_params_}")
    print(f"        acc={accuracy_score(y_te, pred_mlp):.4f}  f1={f1_score(y_te, pred_mlp):.4f}  "
          f"recall={recall_score(y_te, pred_mlp):.4f}  precision={precision_score(y_te, pred_mlp):.4f}  "
          f"tiempo_ajuste={t_mlp:.4f}s")

    return {
        "dataset": nombre_dataset,
        "svc_acc": accuracy_score(y_te, pred_svc), "svc_f1": f1_score(y_te, pred_svc),
        "svc_tiempo": t_svc, "svc_params": gs_svc.best_params_,
        "mlp_acc": accuracy_score(y_te, pred_mlp), "mlp_f1": f1_score(y_te, pred_mlp),
        "mlp_tiempo": t_mlp, "mlp_params": gs_mlp.best_params_,
    }


resultado_diag = benchmark_modelo(df_diag, "diagnostico", [], "diagnostico_celular.csv")
````

**Lectura:** sobre `diagnostico_celular.csv`, `GridSearchCV` eligió `C=100, gamma=0.01` para SVC y `alpha=0.0001, hidden_layer_sizes=(16, 8)` para MLP. El resultado favorece a la red neuronal en ambas métricas de calidad: **95.91%** de exactitud y F1 de **0.9671** para MLP, frente a **94.74%** de exactitud y F1 de **0.9573** para SVC — una diferencia modesta pero consistente (MLP también gana en *recall*: 96.26% vs. 94.39%). En tiempo de entrenamiento del modelo final, sin embargo, la diferencia es enorme: SVC ajusta en apenas **0.0074s**, frente a **0.9140s** de MLP (más de 100 veces más lento) — coherente con lo observado en la Sección 3, aunque aquí el dataset es aún más pequeño (569 filas) que el usado en el experimento de escalabilidad.

## 6. Tabla de Síntesis: ¿Cuándo Conviene Cada Modelo? 📋

Con la evidencia empírica de las Secciones 2, 3 y 5 ya sobre la mesa, la siguiente tabla consolida el criterio de decisión práctico para minería de datos:

| Dimensión | Support Vector Machine (SVM) | Red Neuronal (MLP) |
|---|---|---|
| **Convexidad / Reproducibilidad** | Convexa: mínimo global único, resultado determinista (Sección 2: desv.est.=0.0000 entre semillas) | No convexa: múltiples mínimos locales, resultado variable entre semillas (Sección 2: desv.est.=0.0053, rango de 1.75 puntos) |
| **Escalabilidad en $N$** | Más lenta al crecer $N$ ($\times$15.06 el tiempo al multiplicar $N$ por 10, Sección 3); rápida en términos absolutos a escala moderada | Crece más suavemente con $N$ ($\times$7.28 en el mismo experimento), pero con un costo fijo por iteración más alto |
| **Dimensionalidad alta, pocos datos** | Favorecida — el kernel evita construir explícitamente el espacio expandido | Riesgo de sobreajuste — más parámetros que aprender por variable agregada |
| **Interpretabilidad** | Media — los vectores de soporte y el hiperplano tienen significado geométrico directo | Baja — pesos distribuidos en capas ocultas sin interpretación directa |
| **Necesidad de escalado de variables** | Alta — el kernel se basa en distancias (Cuaderno 00, Sección 9: sin escalar, la exactitud cayó de 71.56% a 34.89%) | Alta — el descenso de gradiente es igual de sensible a la escala |
| **Sensibilidad a hiperparámetros** | Moderada ($C$, $\gamma$, kernel) — rejillas de búsqueda pequeñas suelen bastar | Alta (arquitectura, `alpha`, tasa de aprendizaje, `early_stopping`) — más combinaciones posibles, más superficie para ajustar |
| **Escenario típico donde conviene** | $N$ pequeño-mediano, pocos recursos de cómputo, se requiere reproducibilidad estricta o resultados auditable | Datos abundantes, se puede tolerar variabilidad entre entrenamientos, y una ganancia pequeña de desempeño justifica el costo computacional |

Ninguna fila de esta tabla es una regla absoluta — como mostró la Sección 5, ambos modelos pueden llegar a desempeños muy cercanos sobre el mismo dataset. La tabla resume **tendencias estructurales**, no un veredicto único; el criterio de decisión final depende del problema de negocio concreto (una idea que el taller evaluativo de este módulo desarrolla con más profundidad).

**Conclusión:** en este tercer dataset, el resultado es distinto al patrón de la Sección 5 — SVM y MLP quedan prácticamente **empatados** en calidad (ambos con 99.17% de exactitud; F1 de 0.9916 y 0.9917, una diferencia de una diezmilésima) en lugar de que la red neuronal gane con un margen claro. Lo que sí se mantiene, y con una diferencia todavía más marcada, es el tiempo: SVM ajusta en 0.0084s frente a 1.0093s de MLP — más de **100 veces más rápido**. La explicación más plausible está en la naturaleza del propio dataset: `geometria_lunas.csv` tiene solo **2 variables** y una geometría curva pero relativamente simple (dos medialunas entrelazadas), un problema que el kernel RBF de SVM —diseñado precisamente para curvar fronteras de forma eficiente en espacios de baja dimensión (Cuaderno 01, Sección 4)— resuelve casi tan bien como una red con mucha más capacidad de representación. Cuando el problema no exige la flexibilidad adicional de una arquitectura profunda, esa flexibilidad extra no se traduce en mejor desempeño — solo en más tiempo de cómputo. Es la misma lección que ya había aparecido en el Cuaderno 02 (Sección 9): más capacidad no ayuda cuando el cuello de botella no es la capacidad del modelo, sino la complejidad real del problema.

## 8. Resumen Ejecutivo del Módulo 06 🎓

Este cuaderno cerró el Módulo 06 confrontando SVM y Redes Neuronales con evidencia empírica real, no solo con la tabla teórica de la Sección 1:

- **La convexidad de SVM se traduce en reproducibilidad real:** siete corridas de `SVC` con distinto `random_state` dieron el mismo resultado exacto (95.32%, desv.est.=0.0000), mientras que siete corridas de `MLPClassifier` con la misma arquitectura variaron entre 94.15% y 95.91% (desv.est.=0.53 puntos) — la consecuencia directa y medible de optimizar una superficie no convexa.
- **La escalabilidad teórica de SVM se confirmó, aunque con matices:** al multiplicar $N$ por 10 (de 200 a 2000 filas), el tiempo mínimo de `SVC` creció $\times$15.06 frente a $\times$7.28 de `MLPClassifier` — una tasa de crecimiento más rápida y consistente entre repeticiones, coherente con $O(N^2)$–$O(N^3)$, aunque en términos absolutos SVM siguió siendo más rápida en todo el rango modesto de este experimento.
- **En alta dimensión con pocos datos, el truco del kernel favorece a SVM** por diseño, mientras que una red profunda necesita más datos para justificar su capacidad extra — una nota conceptual, sin dataset de este módulo que la exigiera empíricamente.
- **En un benchmark cabeza a cabeza sobre dos datasets reales** (`diagnostico_celular.csv` y `sensores_maquinaria.csv`), la red neuronal optimizada superó a la SVM optimizada en exactitud y F1 en ambos casos, por márgenes modestos (0.8 a 1.2 puntos), a costa de ser entre 100 y 250 veces más lenta de entrenar — y en un tercer dataset más simple (`geometria_lunas.csv`), ambas quedaron prácticamente empatadas en calidad, con SVM más de 100 veces más rápida.
- **Ningún modelo domina universalmente:** la Sección 6 consolidó estas tendencias en una tabla de síntesis que depende del contexto —tamaño de datos, dimensionalidad, necesidad de reproducibilidad, presupuesto de cómputo— más que de una preferencia absoluta.

**La historia completa del Módulo 06:** el Cuaderno **00** introdujo el margen máximo y el margen blando de SVM sobre `clientes_segmentacion_compleja.csv`, mostrando que escalar las variables no es opcional (sin `StandardScaler`, la exactitud cayó de 71.56% a 34.89%). El Cuaderno **01** extendió SVM a fronteras no lineales mediante el truco del kernel — el kernel RBF elevó la exactitud sobre `geometria_lunas.csv` de 90.00% (lineal) a 97.92%, y SVR alcanzó $R^2=0.9836$ sobre `rendimiento_industrial_no_lineal.csv`, frente a 0.8429 de una regresión lineal simple. Los Cuadernos **02** y **03** cambiaron por completo de familia: el 02 construyó el perceptrón multicapa desde su ecuación fundamental hasta un `MLPClassifier` con 98.61% de exactitud sobre `sensores_maquinaria.csv`, y el 03 abrió la caja negra del entrenamiento —retropropagación, optimizador Adam (95.91% vs. 93.57% de SGD plano sobre `diagnostico_celular.csv`), regularización L2, y parada temprana—. Este Cuaderno **04** cerró el módulo confrontando ambas familias directamente: ninguna gana siempre, pero cada una tiene un perfil de fortalezas medible y reproducible —convexidad y velocidad de entrenamiento para SVM, capacidad de representación y una ligera ventaja de desempeño en los datasets probados para la red neuronal— que, junto con el contexto de negocio del problema concreto, es lo que debe guiar la elección en la práctica de minería de datos.
