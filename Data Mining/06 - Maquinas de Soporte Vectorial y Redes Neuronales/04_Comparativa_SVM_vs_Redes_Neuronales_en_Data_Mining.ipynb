## 1. Matriz de Confrontación Tecnológica: SVM vs. ANN 🥊

| Propiedad | Support Vector Machines (SVM) | Redes Neuronales Artificiales (ANN / MLP) |
|---|---|---|
| **Espacio de Optimización** | Convexa (Garantiza mínimo global único) | No Convexa (Susceptible a mínimos locales y puntos de silla) |
| **Tamaño Muestral Ideal** | Muestras pequeñas/medianas ($N < 50,000$), alta dimensión | Muestras grandes a masivas ($N > 100,000$) |
| **Escalabilidad** | $O(N^2)$ a $O(N^3)$ (computacionalmente costoso en Big Data) | $O(N)$ mediante Minibatch SGD / Adam |
| **Hiperparámetros Clave** | $C, \gamma, \text{kernel}$ | Topología, Tasa de aprendizaje, $\alpha$, Épocas, Activación |
| **Interpretabilidad** | Vectores de soporte delimitan la frontera | Caja negra con representaciones latentes distribuidas |
