## 1. La Varianza de un Solo Árbol vs. La Estabilidad de un Bosque 🌲🌲

Un árbol de decisión individual sin podar posee **muy bajo sesgo pero altísima varianza** (una perturbación de un único dato de entrenamiento altera drásticamente toda la jerarquía de divisiones).

Random Forest promedia $B$ árboles descorrelacionados, reduciendo la varianza asintóticamente a $\rho \sigma^2 + \frac{1-\rho}{B} \sigma^2$ (donde $\rho$ es la correlación promedio entre árboles).

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, os
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

df_bench = pd.read_csv(os.path.join("data", "riesgo_credito_benchmark.csv"))
X = df_bench.drop(columns=['clase_objetivo'])
y = df_bench['clase_objetivo']

X_tr, X_ts, y_tr, y_ts = train_test_split(X, y, test_size=0.3, random_state=42)
dt = DecisionTreeClassifier(random_state=42).fit(X_tr, y_tr)
rf = RandomForestClassifier(n_estimators=100, random_state=42).fit(X_tr, y_tr)

print(f"Exactitud Árbol Individual -> Train: {dt.score(X_tr, y_tr)*100:.1f}% | Test: {dt.score(X_ts, y_ts)*100:.1f}%")
print(f"Exactitud Random Forest     -> Train: {rf.score(X_tr, y_tr)*100:.1f}% | Test: {rf.score(X_ts, y_ts)*100:.1f}%")
```
