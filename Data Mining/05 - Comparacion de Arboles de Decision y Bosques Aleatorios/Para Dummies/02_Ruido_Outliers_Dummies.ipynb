## ¿Qué vamos a aprender aquí? 🔍

El Cuaderno 01 de esta ruta comparó a un árbol de decisión y a un Random Forest en su capacidad de **explicarse**. Este cuaderno cambia de terreno: ¿qué pasa cuando los datos de entrenamiento no son perfectos? En el mundo real, los datos llegan con errores — un valor absurdo capturado por accidente, una etiqueta mal escrita por una persona cansada. Este cuaderno pone a prueba, con datos reales, quién aguanta mejor esas imperfecciones.

Al terminar vas a poder responder:
1. ¿Qué le pasa a un árbol de decisión cuando un solo valor de una variable es absurdamente grande?
2. ¿Qué le pasa a un Random Forest en la misma situación?
3. ¿Es cierto que "el bosque siempre gana" ante datos sucios, o la realidad es más interesante que eso?

---
## 1. La Gran Analogía: El Estudiante que Memorizó la Guía de Estudio 📝

Imagina un estudiante que se prepara para un examen memorizando, palabra por palabra, una guía de repaso con 100 preguntas de práctica.

* Si **una sola** de esas 100 preguntas tenía la respuesta mal escrita (un error de imprenta absurdo, tipo "la capital de Francia es Marte"), un estudiante que memoriza sin razonar puede repetir ese error el día del examen real, como si fuera un hecho verdadero.
* Si en cambio el estudiante le pregunta a un **grupo de 100 compañeros** sobre cada tema y promedia sus respuestas, un solo compañero que memorizó mal una respuesta pesa muy poco frente al consenso del grupo.

Un **árbol de decisión** aprende de los datos de entrenamiento de forma parecida al estudiante que memoriza solo: construye sus reglas mirando exactamente los datos que le tocaron, sin filtrar nada. Un **Random Forest** es el grupo de 100 compañeros: cada árbol ve una muestra distinta de los datos, así que un dato raro rara vez contamina la opinión de todos a la vez.

Esa es la intuición. Este cuaderno la pone a prueba con dos tipos distintos de "errores en la guía de estudio" — y el resultado, como vas a ver, no es tan simple como "el grupo siempre gana".

---
## 2. Por Qué Esto Importa en la Vida Real 🌍

Los datos nunca llegan perfectos. Dos tipos de error son especialmente comunes:

* **Un valor absurdo (outlier):** un sensor que falla por un instante, alguien que digita un ingreso mensual con un cero de más, un campo capturado a mano con un error de dedo. El dato existe, pero su valor no tiene sentido.
* **Una etiqueta mal puesta (ruido de etiqueta):** en el dataset de crédito que se usa en todo este módulo, esto sería un caso donde por un error humano al registrar el resultado de un préstamo, una solicitud que en realidad era "bajo riesgo" queda anotada como "alto riesgo" (o al revés) — literalmente, le "dijeron mal la respuesta" al modelo durante el entrenamiento.

Ambos tipos de error llegan solos, sin avisar, a cualquier conjunto de datos real. La pregunta de este cuaderno es simple: si entrenamos con datos así de sucios, ¿qué modelo mantiene mejor su exactitud sobre datos limpios y nuevos?

```python
import os, urllib.request, urllib.parse, gzip, base64
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

_DATA_EMBEDDED = {
    "riesgo_credito_benchmark.csv": "",
}

def load_dataset(filename: str) -> str:
    candidates = [
        os.path.join("..", "data", filename),
        os.path.join("data", filename),
        os.path.join("05 - Comparacion de Arboles de Decision y Bosques Aleatorios", "data", filename),
        os.path.join("Data Mining", "05 - Comparacion de Arboles de Decision y Bosques Aleatorios", "data", filename),
        os.path.join("..", "..", "05 - Comparacion de Arboles de Decision y Bosques Aleatorios", "data", filename),
        filename,
    ]
    for p in candidates:
        if os.path.exists(p):
            return p
    os.makedirs("data", exist_ok=True)
    target = os.path.join("data", filename)
    base_url = ("https://raw.githubusercontent.com/sazuniga06/Data-Science-Programming---USTA-Tunja-Repository/"
                "main/Data%20Mining/05%20-%20Comparacion%20de%20Arboles%20de%20Decision%20y%20Bosques%20Aleatorios/data/")
    url = base_url + urllib.parse.quote(filename)
    try:
        urllib.request.urlretrieve(url, target, timeout=3)
        return target
    except Exception:
        pass
    if filename in _DATA_EMBEDDED:
        raw = gzip.decompress(base64.b64decode(_DATA_EMBEDDED[filename]))
        with open(target, "wb") as f:
            f.write(raw)
        return target
    raise FileNotFoundError(f"No se pudo obtener el dataset: {filename}")

np.random.seed(42)
df_bench = pd.read_csv(load_dataset("riesgo_credito_benchmark.csv"))
X = df_bench.drop(columns=["clase_objetivo"])
y = df_bench["clase_objetivo"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
print("Filas y columnas:", df_bench.shape)
print("Balance de clases:", df_bench["clase_objetivo"].value_counts().to_dict())
print("X_train:", X_train.shape, " X_test:", X_test.shape)
```

### 🤔 ¿Qué acaba de pasar?
Cargamos las 2000 solicitudes de crédito (`load_dataset` busca primero el archivo en disco; si no lo encuentra, lo descarga desde el repositorio del curso; y si tampoco hay internet, lo reconstruye desde una copia comprimida incluida en este mismo cuaderno) y separamos 1400 para entrenar y 600 para probar, con la misma partición (`random_state=42`) usada en el resto del módulo. El conjunto de prueba queda **limpio** — nunca se va a contaminar en ninguno de los dos experimentos siguientes, para que la comparación sea justa.

---
## 3. Línea Base: Antes de Ensuciar Nada ✅

Antes de contaminar cualquier dato, hay que ver cómo se comportan ambos modelos con los datos **limpios** — esta es la referencia contra la que se compara todo lo demás.

```python
tree_base = DecisionTreeClassifier(random_state=0).fit(X_train, y_train)
rf_base = RandomForestClassifier(n_estimators=100, random_state=0, n_jobs=-1).fit(X_train, y_train)

acc_tree_base = accuracy_score(y_test, tree_base.predict(X_test))
acc_rf_base = accuracy_score(y_test, rf_base.predict(X_test))

print(f"Linea base (datos 100% limpios):")
print(f"  Arbol individual : {acc_tree_base:.4f}")
print(f"  Random Forest    : {acc_rf_base:.4f}")
```

### 🤔 ¿Qué acaba de pasar?
Con datos limpios, el Random Forest ya arranca con una exactitud considerablemente mayor (**0.8033**) que el árbol individual (**0.6850**) — la misma ventaja de estabilidad que se vio en el Cuaderno 00. Ahora sí podemos ensuciar los datos y ver cómo reacciona cada uno.

---
## 4. Experimento 1: Un Dato Absurdo en la Solicitud 📉

Simulamos el error del sensor que falla: tomamos un 15% de las solicitudes de entrenamiento, elegidas al azar, y les cambiamos el valor de `feature_01` por un número disparatado (entre 10 y 15 desviaciones estándar del promedio — recordemos que las variables están estandarizadas, así que esto es un valor extremo de verdad). El resto de cada solicitud, incluida su etiqueta real, queda intacto.

Para no depender de la suerte de una sola corrida, el experimento se repite 10 veces con distintas filas "dañadas" cada vez, y se promedia la exactitud resultante sobre el mismo conjunto de prueba limpio de siempre.

```python
def contaminar_outliers(X_train, frac, seed=0):
    """Reemplaza feature_01 por un valor extremo (magnitud 10-15, signo aleatorio)
    en una fraccion `frac` de filas elegidas al azar."""
    Xc = X_train.copy()
    rng = np.random.RandomState(seed)
    n = len(Xc)
    n_out = int(round(frac * n))
    idx = rng.choice(Xc.index, size=n_out, replace=False)
    signs = rng.choice([-1, 1], size=n_out)
    magnitudes = rng.uniform(10, 15, size=n_out)
    Xc.loc[idx, "feature_01"] = signs * magnitudes
    return Xc


N_REPS = 10
FRAC_OUTLIERS = 0.15

accs_tree_out, accs_rf_out = [], []
for rep in range(N_REPS):
    Xc = contaminar_outliers(X_train, FRAC_OUTLIERS, seed=rep)
    t = DecisionTreeClassifier(random_state=0).fit(Xc, y_train)
    r = RandomForestClassifier(n_estimators=100, random_state=0, n_jobs=-1).fit(Xc, y_train)
    accs_tree_out.append(accuracy_score(y_test, t.predict(X_test)))
    accs_rf_out.append(accuracy_score(y_test, r.predict(X_test)))

acc_tree_out15 = np.mean(accs_tree_out)
acc_rf_out15 = np.mean(accs_rf_out)

print(f"Outliers en el 15% de las filas de entrenamiento (promedio de {N_REPS} repeticiones):")
print(f"  Arbol : {acc_tree_out15:.4f}  (cambio vs. linea base: {acc_tree_out15 - acc_tree_base:+.4f})")
print(f"  Bosque: {acc_rf_out15:.4f}  (cambio vs. linea base: {acc_rf_out15 - acc_rf_base:+.4f})")
```

### 🤔 ¿Qué acaba de pasar?
Esto sorprende: el **árbol individual no empeora — mejora levemente** (+0.0198 frente a la línea base), mientras que el **Random Forest sí se degrada un poco** (-0.0117). ¿Por qué? Un valor absurdo en una sola variable es fácil de "encerrar": el árbol puede crear una rama que aísla ese caso raro en su propia hoja, sin tener que cambiar el resto de sus reglas. El bosque, en cambio, construye 100 árboles distintos sobre submuestras distintas de los datos — el outlier termina apareciendo, de formas ligeramente distintas, en varios de ellos, y el promedio se ve un poco afectado. No es el resultado que la analogía inicial hacía esperar.

---
## 5. Experimento 2: Le Dijeron Mal la Respuesta 🏷️

Ahora simulamos el otro tipo de error: tomamos un 20% de las etiquetas de entrenamiento, elegidas al azar, y las **invertimos** (una solicitud que en realidad era "bajo riesgo" queda marcada como "alto riesgo", y viceversa). Ya no es un valor raro en una variable — es la "respuesta correcta" la que está mal, para uno de cada cinco casos.

Igual que antes, se repite 10 veces con distintas filas "mal etiquetadas" cada vez, y se promedia.

```python
def contaminar_etiquetas(y_train, frac, seed=0):
    """Invierte la etiqueta (0<->1) de una fraccion `frac` de filas elegidas al azar."""
    yc = y_train.copy()
    rng = np.random.RandomState(seed)
    n = len(yc)
    n_flip = int(round(frac * n))
    idx = rng.choice(yc.index, size=n_flip, replace=False)
    yc.loc[idx] = 1 - yc.loc[idx]
    return yc


FRAC_LABELS = 0.20

accs_tree_lbl, accs_rf_lbl = [], []
for rep in range(N_REPS):
    yc = contaminar_etiquetas(y_train, FRAC_LABELS, seed=rep)
    t = DecisionTreeClassifier(random_state=0).fit(X_train, yc)
    r = RandomForestClassifier(n_estimators=100, random_state=0, n_jobs=-1).fit(X_train, yc)
    accs_tree_lbl.append(accuracy_score(y_test, t.predict(X_test)))
    accs_rf_lbl.append(accuracy_score(y_test, r.predict(X_test)))

acc_tree_lbl20 = np.mean(accs_tree_lbl)
acc_rf_lbl20 = np.mean(accs_rf_lbl)

print(f"Etiquetas invertidas en el 20% de las filas de entrenamiento (promedio de {N_REPS} repeticiones):")
print(f"  Arbol : {acc_tree_lbl20:.4f}  (cambio vs. linea base: {acc_tree_lbl20 - acc_tree_base:+.4f},"
      f" {(acc_tree_lbl20 - acc_tree_base) / acc_tree_base * 100:.1f}% relativo)")
print(f"  Bosque: {acc_rf_lbl20:.4f}  (cambio vs. linea base: {acc_rf_lbl20 - acc_rf_base:+.4f},"
      f" {(acc_rf_lbl20 - acc_rf_base) / acc_rf_base * 100:.1f}% relativo)")
```

### 🤔 ¿Qué acaba de pasar?
Aquí sí aparece la ventaja del bosque, aunque de forma modesta: **ambos modelos empeoran** con las etiquetas mal puestas, pero el árbol cae un poco más (**-10.4%** relativo) que el Random Forest (**-8.0%** relativo). A diferencia del experimento anterior, una etiqueta equivocada no se puede "encerrar" en una sola rama — contamina directamente la regla que se aprende ahí. El bosque amortigua un poco el daño porque promedia muchas opiniones entrenadas sobre submuestras distintas, pero no lo elimina: ningún modelo es inmune a que le enseñen mal la respuesta correcta.

```python
fig, ax = plt.subplots(figsize=(8, 5))
grupos = ["Sin contaminar", "Outlier 15%\n(feature_01)", "Etiquetas 20%\ninvertidas"]
valores_arbol = [acc_tree_base, acc_tree_out15, acc_tree_lbl20]
valores_bosque = [acc_rf_base, acc_rf_out15, acc_rf_lbl20]

x = np.arange(len(grupos))
ancho = 0.35
ax.bar(x - ancho / 2, valores_arbol, ancho, label="Arbol individual", color="#c2410c")
ax.bar(x + ancho / 2, valores_bosque, ancho, label="Random Forest", color="#0ea5e9")
ax.set_xticks(x)
ax.set_xticklabels(grupos)
ax.set_ylabel("Exactitud sobre X_test (limpio)")
ax.set_title("Exactitud segun el tipo de dato sucio inyectado")
ax.legend()
plt.tight_layout()
plt.show()
```

### 🤔 La verdad completa (sin simplificar)

| Experimento | ¿Quién se ve más afectado? | Por qué (razonamiento simple) |
|---|---|---|
| **Outlier en una variable (15%)** | El **bosque** — el árbol prácticamente no se inmuta (incluso mejora un poco) | El árbol aísla el valor absurdo en una rama propia, sin tocar el resto de sus reglas; el bosque reparte ese dato raro entre varios de sus 100 árboles, de forma menos "limpia" |
| **Etiquetas mal puestas (20%)** | El **árbol** — ambos empeoran, pero el árbol cae más | Una etiqueta equivocada no se puede aislar en una sola rama; el bosque amortigua el daño al promediar muchas opiniones, pero no lo elimina |

**La lección real (no la simplificada):** no es cierto que "el Random Forest siempre gana cuando los datos están sucios". Ante un valor absurdo aislado en una sola variable, el árbol individual se defendió sorprendentemente bien — mejor, de hecho, que el bosque. La ventaja de robustez del bosque es real, pero **específica**: aparece con más claridad cuando el problema es la calidad de las etiquetas, no un valor extremo suelto en una variable.

## Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **La analogía** | El árbol memoriza solo, sin filtrar; el bosque promedia 100 opiniones entrenadas sobre muestras distintas. |
| **Outlier aislado en una variable** | El árbol lo absorbe bien (incluso mejora un poco); el Random Forest se degrada levemente. Un resultado que sorprende. |
| **Etiquetas mal puestas** | Ambos modelos empeoran, pero el bosque cae un poco menos (-8.0% vs. -10.4% relativo al 20% de ruido). |
| **La lección real** | La ventaja de robustez del bosque no es universal — depende del tipo de error en los datos, no es "el bosque siempre gana". |
