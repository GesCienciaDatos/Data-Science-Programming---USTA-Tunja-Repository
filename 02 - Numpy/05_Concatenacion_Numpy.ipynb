## Concatenación y Apilamiento de Arrays 

En el mundo real, los datos rara vez vienen limpios y en un solo archivo. Usualmente tendrás que combinar datos de diferentes fuentes: las ventas de enero por un lado, las de febrero por otro.

NumPy ofrece dos operaciones fundamentales para unir (merge) múltiples arrays en un único array final: **Concatenación** (unión en un eje existente) y **Apilamiento o Stacking** (creación de un eje completamente nuevo).

---
### 1. Concatenación (Unión en un Eje Existente) 🔗

La función `np.concatenate()` toma una secuencia de arrays y los une a lo largo de un eje o dimensión **que ya existe**. Piensa en ello como tomar dos trenes y pegarlos vagón con vagón para hacer un tren más largo. El resultado sigue siendo un tren (misma cantidad de dimensiones).

**Regla Matemática Estricta:**
Para poder concatenar dos arrays, **absolutamente todos sus ejes deben tener la misma longitud**, con la única excepción del eje sobre el cual estás haciendo la concatenación. 

Por ejemplo, si tienes una matriz `[A, B, C, N1]` y otra `[A, B, C, N2]`, puedes concatenarlas en el último eje para obtener `[A, B, C, N1+N2]`. ¡Si `A`, `B` o `C` difieren, NumPy explotará!

````python
# Configuración interactiva
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'
import numpy as np

# --- EJEMPLO 1D ---
arr1 = np.arange(1, 4) # [1, 2, 3] (Forma: 3,)
arr2 = np.arange(4, 7) # [4, 5, 6] (Forma: 3,)

print("Formas 1D:", arr1.shape, arr2.shape)

# Concatenamos vectores 1D [3] + [3] -> [6]
unidos_1d = np.concatenate([arr1, arr2], axis=0)
print("Vectores concatenados:", unidos_1d)
````

````python
# --- EJEMPLO 2D ---
arr3 = np.arange(1, 7).reshape(2, 3)
arr4 = np.arange(7, 13).reshape(2, 3)

print("Matriz 1:\n", arr3)
print("Matriz 2:\n", arr4)
print("\nFormas 2D:", arr3.shape, arr4.shape)

# Concatenación en el Eje 0 (Hacia abajo / Filas)
# [2, 3] + [2, 3] -> [4, 3]
concat_axis_0 = np.concatenate([arr3, arr4], axis=0)
print("\nConcatenación axis=0 (Pegado Vertical):\n", concat_axis_0)
print("Nueva forma:", concat_axis_0.shape)

# Concatenación en el Eje 1 (Hacia la derecha / Columnas)
# [2, 3] + [2, 3] -> [2, 6]
concat_axis_1 = np.concatenate([arr3, arr4], axis=1)
print("\nConcatenación axis=1 (Pegado Horizontal):\n", concat_axis_1)
print("Nueva forma:", concat_axis_1.shape)
````

### 2. Apilamiento / Stacking (Unión en un NUEVO Eje) 🥞

A diferencia de la concatenación, el Apilamiento (`np.stack()`) crea una **dimensión completamente nueva**. 
Piensa en ello como tomar varias hojas de papel 2D (matrices) y "apilarlas" físicamente una encima de la otra para crear un libro 3D (un tensor).

**Regla Matemática Estricta:**
Para que NumPy pueda crear esta nueva dimensión perfecta, los arrays originales deben ser **100% idénticos en su forma** (mismas dimensiones, mismos elementos en cada dimensión).

Si apilas dos matrices de forma `[A, B, C, D]` en el último eje, obtendrás una estructura superior de forma `[A, B, C, D, 2]`.


````python
# --- EJEMPLO 1D a 2D ---
# Apilar dos vectores 1D crea una matriz 2D
# arr1 y arr2 son [1,2,3] y [4,5,6] (Forma: 3,)

# Stacking en el eje 0 (Crea la dimensión al principio)
# [3] y [3] -> [2, 3]
stack_axis_0 = np.stack([arr1, arr2], axis=0)
print("Stack axis=0 (Crea filas):\n", stack_axis_0)
print("Forma:", stack_axis_0.shape)

# Stacking en el eje -1 (Último eje, crea la dimensión al final)
# [3] y [3] -> [3, 2]
stack_axis_1 = np.stack([arr1, arr2], axis=-1)
print("\nStack axis=-1 (Crea columnas pareadas):\n", stack_axis_1)
print("Forma:", stack_axis_1.shape)
````

````python
# --- EJEMPLO 2D a 3D ---
# arr3 y arr4 son matrices 2x3. Al apilarlas, pasaremos al mundo 3D.

# [2, 3] y [2, 3] -> [2, 2, 3] 
# (Se crea una dimensión '2' al inicio, representando "2 matrices, cada una de 2x3")
stack_2d_ax0 = np.stack([arr3, arr4], axis=0)
print("Stack 3D (axis=0):\n", stack_2d_ax0)
print("Forma:", stack_2d_ax0.shape)

# [2, 3] y [2, 3] -> [2, 3, 2]
# (Se emparejan los elementos individuales uno detrás de otro)
stack_2d_ax_last = np.stack([arr3, arr4], axis=-1)
print("\nStack 3D (axis=-1):\n", stack_2d_ax_last)
print("Forma:", stack_2d_ax_last.shape)
````
