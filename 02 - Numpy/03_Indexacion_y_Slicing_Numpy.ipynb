## Indexación y Slicing en NumPy 

Extraer partes de tus datos es una de las tareas fundamentales en el análisis de datos. Ya sea que quieras los primeros 10 registros, una columna específica, o todos los valores que cumplan cierta condición.

NumPy toma la sintaxis nativa de Python y la potencia al máximo. En este cuaderno exploraremos la indexación 1D, el acceso a matrices 2D, y técnicas avanzadas como *Fancy Indexing* y las *Máscaras Booleanas*.




---
### 1. Indexación Básica y Slicing (1D) 📏

Los arrays unidimensionales (vectores) se comportan de manera idéntica a las listas nativas de Python. 
Utilizan la sintaxis estándar de rebanado (slicing): **`array[inicio : fin : paso]`**

*   **`inicio`**: Desde dónde cortar (incluido).
*   **`fin`**: Hasta dónde cortar (excluido).
*   **`paso`**: De a cuántos elementos saltar.

*Nota: Al igual que en Python, el índice siempre comienza en cero `0`.*

````python
# Configurar salidas interactivas
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'
import numpy as np

# Creamos un array del 0 al 9, y elevamos todo al cubo (**3)
a = np.arange(10) ** 3
print("Array original:", a)

print("Un solo elemento a[2]:", a[2])
print("Rango a[2:5]:", a[2:5]) # Trae los índices 2, 3 y 4

# Slicing con Asignación (¡Poderoso!)
# Desde el inicio hasta el 6 (excluido), saltando de 2 en 2, reemplázalos por 1000
a[:6:2] = 1000
print("Array modificado:", a)

# Truco ninja: Voltear el array con paso negativo
print("Array invertido:", a[::-1])
````

### 2. Indexación Multidimensional (Matrices) 🧊

A diferencia de las listas anidadas de Python, donde tendrías que usar doble corchete `lista[fila][columna]`, en NumPy es mucho más limpio y eficiente: usamos **un solo par de corchetes separados por comas**: `matriz[fila, columna]`.

Esta sintaxis separada por comas aplica también para el slicing: `matriz[inicio_fila:fin_fila, inicio_col:fin_col]`.


````python
# Creamos una matriz de 5 filas y 4 columnas
b = np.array([[ 0,  1,  2,  3],
              [10, 11, 12, 13],
              [20, 21, 22, 23],
              [30, 31, 32, 33],
              [40, 41, 42, 43]])

print("Elemento exacto b[2, 3]:", b[2, 3]) # Fila 2, Columna 3

print("\nExtraer toda la SEGUNDA COLUMNA:")
print(b[:, 1]) # El ':' solo significa "tráeme TODAS las filas"

print("\nExtraer un Sub-bloque (Filas 1 a 2, TODAS las columnas):")
print(b[1:3, :])
````

**Iterando sobre Matrices**

Si utilizas un bucle `for` tradicional en una matriz 2D, NumPy recorrerá el primer eje (te entregará fila por fila). Si por alguna razón necesitas iterar absolutamente por cada número individual, NumPy te ofrece el atributo `.flat`.


````python
print("Iterar normal:")
for fila in b:
    print(fila) # Imprime 5 filas

print("\nIterar aplanado con .flat:")
for elemento in list(b.flat)[:6]: # Limitado a los 6 primeros para no llenar la pantalla
    print(elemento)
````

### 3. Fancy Indexing (Indexación Mágica) 🧗

*Nota: El ícono del escalador significa que este es un tema de nivel intermedio/avanzado.*

El **Fancy Indexing** es una característica exclusiva de NumPy (las listas normales de Python explotarían si intentas hacer esto). Te permite extraer elementos saltados pasando un *array (o lista) de índices*, en lugar de un bloque continuo con `:`.

Si quieres extraer la coordenada `[0,0]`, la `[1,1]` y la `[2,2]` al mismo tiempo, le pasas a NumPy una lista con todas las filas `[0, 1, 2]` y otra lista con todas las columnas `[0, 1, 2]`.


````python
# Creamos un array del 1 al 16, y lo moldeamos a 4x4
arr = np.arange(1, 17).reshape([4, 4])
print("Matriz Original:\n", arr)

# 0  1  2  3
# -----------
# 1  2  3  4   | 0
# 5  6  7  8   | 1
# 9  10 11 12  | 2
# 13 14 15 16  | 3

print("\nExtrayendo la diagonal principal usando Fancy Indexing:")
diagonal = arr[[0, 1, 2, 3], [0, 1, 2, 3]]
print(diagonal)
````

### 4. Indexación Booleana (Máscaras) 🧗

Este es, por mucho, **el tipo de indexación que más usarás en Ciencia de Datos** (es la base matemática sobre la que funciona el filtrado en la librería *Pandas*).

La idea es simple pero brillante: en lugar de buscar índices con números exactos `[0, 2]`, pasamos un array lógico de `True` y `False` que funciona como un "filtro" o "máscara". NumPy extraerá o modificará únicamente las posiciones donde encuentre un `True`.


````python
arr2 = np.arange(1, 17).reshape([4, 4])

# 1. Creamos una condición. ¿Qué números son pares?
mascara = (arr2 % 2 == 0)
print("Máscara Booleana (True = Par):\n", mascara)

# 2. Le pasamos la máscara como índice al array
pares = arr2[mascara]
print("\nSolo los números pares:", pares)

# 3. Podemos sobrescribir datos usando la máscara
# Reemplazaremos todos los impares por 0
arr2[arr2 % 2 != 0] = 0
print("\nMatriz con impares apagados:\n", arr2)
````

**La función `np.where()`**

A veces no quieres el dato extraído, sino saber **dónde (en qué índices numéricos)** se cumple la condición. Para esto existe `np.where()`. Retorna las coordenadas de los `True`.


````python
# ¿Dónde están los números mayores a 10 en la matriz original?
indices_mayores_10 = np.where(arr2 > 10)
print("Índices de elementos > 10:", indices_mayores_10)
````
