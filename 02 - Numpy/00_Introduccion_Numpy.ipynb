## ¿Qué es NumPy? 🧊

NumPy es una biblioteca de Python que permite realizar cálculos numéricos. Piensa en el álgebra lineal: NumPy es la biblioteca de Python para ello. Trata sobre matrices y vectores, y permite realizar operaciones matemáticas con ellos.

En el corazón de NumPy se encuentra un tipo de dato básico llamado el **array de NumPy**. Un array de NumPy puede tener múltiples dimensiones, lo que le permite representar cantidades como vectores (1D), matrices (2D) o arrays de dimensiones superiores como los tensores. Un array de NumPy requiere que todos sus elementos sean de un único tipo de dato.

¡Aprender NumPy es un gran primer paso para adentrarse en el campo del Machine Learning y la Ciencia de Datos!

En el aprendizaje automático, los algoritmos cruciales y las estructuras de datos dependen de cálculos matriciales, los cuales son manejados de manera eficiente por NumPy. Dentro de Python, NumPy es una de las bibliotecas más importantes para la ciencia de datos y el aprendizaje automático.

NumPy está optimizado para el cálculo rápido porque está basado en código C precompilado.

### ¿Cuál es la diferencia entre una lista de Python y un array de NumPy?

NumPy ofrece una enorme variedad de formas rápidas y eficientes de crear arrays y manipular datos numéricos en su interior. Mientras que una lista de Python puede contener diferentes tipos de datos dentro de una misma lista, todos los elementos en un array de NumPy deben ser **homogéneos**. Las operaciones matemáticas destinadas a ser realizadas en arrays serían extremadamente ineficientes si los arrays no fueran homogéneos.


---
### ¿Por qué usar NumPy?

Los arrays de NumPy son más rápidos y compactos que las listas de Python. Un array consume menos memoria y es muy conveniente de usar. NumPy utiliza mucha menos memoria para almacenar datos y proporciona un mecanismo para especificar los tipos de datos. Esto permite que el código pueda ser optimizado aún más.
