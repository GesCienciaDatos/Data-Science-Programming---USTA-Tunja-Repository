## Temas Avanzados: Aleatoriedad, Únicos y Expansión 

En este cuaderno abordaremos tres herramientas vitales para cualquier científico de datos que trabaje con algoritmos numéricos o de Machine Learning: la generación de números pseudo-aleatorios, la extracción de valores únicos (frecuencias) y la manipulación avanzada de dimensiones "falsas".



---
### 1. Números Aleatorios (Random Numbers) 🎲

La generación de números aleatorios es la base de casi todo en Machine Learning: desde inicializar los pesos de una Red Neuronal Artificial, hasta mezclar (shuffle) un dataset para evitar sesgos, o separar los datos de entrenamiento y validación.

A nivel informático, la aleatoriedad perfecta no existe. Hablamos de números **pseudo-aleatorios**, generados por un algoritmo matemático complejo. Lo brillante de esto es que si le pasamos una "semilla" (`seed`) al algoritmo, la secuencia "aleatoria" será **exactamente la misma** cada vez que ejecutes el código, permitiendo que tus experimentos científicos sean 100% reproducibles.

*Nota: La forma moderna y recomendada (best practice) de generar números aleatorios en NumPy es instanciando un generador con `np.random.default_rng()`.*


````python
# Configuración interactiva
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'
import numpy as np

# Instanciamos el Generador con una semilla (seed = 16)
rng = np.random.default_rng(seed=16)

# 1. Números Enteros Aleatorios
# El límite inferior es inclusivo, el superior es EXCLUSIVO. (Va del 0 al 4).
matriz_enteros = rng.integers(low=0, high=5, size=(2, 4))
print("Enteros (0 a 4):\n", matriz_enteros)

# 2. Distribución Uniforme (Flotantes)
# Cualquier número entre 0.0 y 1.0 tiene exactamente la misma probabilidad de salir
un_solo_numero = rng.uniform(low=0.0, high=1.0)
print("\nUn solo número (Distribución Uniforme):", un_solo_numero)

# 3. Distribución Normal (Campana de Gauss)
# loc = media (mean), scale = Desviación Estándar (std)
matriz_normal = rng.normal(loc=0.0, scale=1.0, size=(2, 4))
print("\nDistribución Normal (Media 0, Std 1):\n", matriz_normal)
````

### 2. Elementos Únicos y Conteos 🔍

En análisis exploratorio de datos (EDA), una de las primeras cosas que querrás saber sobre una columna categórica (e.g., el estrato social, el género, la ciudad) es: **¿Cuáles son las categorías únicas y cuántas personas hay en cada una?**

La función `np.unique()` es la navaja suiza para esto. Es mucho más poderosa que la función `set()` nativa de Python porque, además de darte los elementos únicos y ordenados, te permite extraer **los índices originales** y **la frecuencia** (conteo) de cada elemento.


````python
# Imaginemos que este array representa las edades de un grupo de personas
a = np.array([11, 11, 12, 13, 14, 15, 16, 17, 12, 13, 11, 14, 18, 19, 20])

# Obtener solo los valores únicos
valores_unicos = np.unique(a)
print("Valores únicos ordenados:", valores_unicos)

# Obtener los únicos y sus índices en el array original (la primera vez que aparece cada número)
unicos, indices = np.unique(a, return_index=True)
print("\nÍndice donde aparece el número '12' por primera vez:", indices[1]) # '12' es el segundo valor único

# Obtener la FRECUENCIA (conteos) de cada elemento
unicos, conteos = np.unique(a, return_counts=True)
print("\nFrecuencia (veces que aparece cada edad):")
for u, c in zip(unicos, conteos):
    print(f"La edad {u} aparece {c} veces.")
````

### 3. Expandiendo y Comprimiendo Dimensiones (Expand & Squeeze) 🗜️

En bibliotecas de Deep Learning como TensorFlow o PyTorch, a menudo te encontrarás con errores de dimensión. Por ejemplo, una red neuronal convolucional (CNN) *exige* que las imágenes tengan 3 dimensiones: `[Alto, Ancho, Canales de Color]`. 

¿Qué pasa si tienes una imagen en blanco y negro (escala de grises) de `[28, 28]`? ¡La red neuronal fallará porque espera 3 dimensiones! Para solucionar esto, necesitas añadir una dimensión "falsa" o "vacía" de tamaño 1, convirtiéndola en `[28, 28, 1]`. Los datos siguen siendo exactamente los mismos, pero la **arquitectura dimensional** ha cambiado.

<img src="images/expand_squeeze.png" width="700"/>


````python
# Recordatorio: Un vector 1D no es ni "fila" ni "columna" matemáticamente. Es solo un vector.
arr = np.arange(5)
print("Forma original:", arr.shape)

# Podemos convertirlo artificialmente en un "Vector Fila" (1x5) o "Vector Columna" (5x1) usando reshape
print("Vector Fila:", arr.reshape(1, -1).shape)
print("Vector Columna:", arr.reshape(-1, 1).shape)
````

````python
# --- EXPANDIENDO (Añadiendo Dimensiones Falsas) ---
matriz_2d = np.arange(9).reshape(3, 3)
print("\nMatriz original:\n", matriz_2d)
print("Forma 2D:", matriz_2d.shape)

# Método 1: Usando reshape y agregando un '1'
expandida_res = np.reshape(matriz_2d, [3, 3, 1])
print("Expandida con reshape:", expandida_res.shape)

# Método 2: Usando np.expand_dims (El método explícito preferido)
# axis=-1 significa "agrega la dimensión vacía al final"
expandida = np.expand_dims(matriz_2d, axis=-1)
print("Expandida con expand_dims:", expandida.shape)


# --- COMPRIMIENDO (Eliminando Dimensiones Falsas) ---
# La función np.squeeze() destruye todas las dimensiones cuyo tamaño sea exactamente '1'
comprimida = np.squeeze(expandida, axis=-1)
print("\nComprimida con squeeze (Volvió a la normalidad):", comprimida.shape)
````
