## Operaciones con Arrays 

Una vez que has creado tus arrays, el siguiente paso lógico es comenzar a trabajar y operar matemáticamente con ellos. NumPy está diseñado específicamente para que este proceso sea increíblemente veloz e intuitivo.

En esta sección exploraremos la aritmética básica, el concepto mágico del "broadcasting", las reducciones estadísticas y el álgebra matricial.




---
### 1. Operaciones Básicas  ➕ ➖ ✖️ ➗

Es crucial entender que los operadores aritméticos básicos en NumPy se aplican **elemento por elemento** (element-wise). 

En el Python estándar, si usas el símbolo `+` en dos listas, Python las concatena (las une una detrás de otra). En NumPy, si usas `+` en dos matrices, NumPy **sumará matemáticamente las posiciones correspondientes** (por ejemplo, la posición `[0,0]` de la matriz A con la posición `[0,0]` de la matriz B).

¡Esta característica vectorizada es la que nos ahorra tener que escribir largos e ineficientes bucles `for`!

````python
# Importar numpy y configurar salidas interactivas
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = 'all'
import numpy as np

# Creamos dos arrays del mismo tamaño
data = np.array([1, 2])
ones = np.ones(2, dtype=int)

print("Suma (Elemento a elemento):")
print(data + ones)

print("\nResta:")
print(data - ones)

print("\nMultiplicación (NO matricial, sino posición por posición):")
print(data * data)

print("\nDivisión:")
print(data / data)
````

### 2. Broadcasting (Transmisión) 📡

El término **Broadcasting** (transmisión) describe la forma mágica y eficiente en la que NumPy maneja operaciones aritméticas entre arrays de *diferentes formas o tamaños*.

Normalmente, pensaríamos que para operar dos arrays deben tener dimensiones idénticas. Sin embargo, bajo ciertas restricciones lógicas, NumPy "transmite" o estira virtualmente el array más pequeño para que coincida con la forma del array más grande. 

**¿Por qué esto es genial?**
Lo hace a nivel de lenguaje C en segundo plano, sin tener que duplicar innecesariamente los datos en la memoria RAM. Evita copias de datos y bucles, llevando a implementaciones de algoritmos ultrarrápidos.

<img src="images/broadcasting.png" alt="NumPy Broadcasting" width="600"/>

El caso más común en el que usamos broadcasting es al operar una matriz contra un único número (un vector contra un escalar). Por ejemplo, si tienes una matriz de distancias en Millas y deseas convertirlo todo a Kilómetros (multiplicando por `1.6`), NumPy expandirá ese `1.6` para multiplicarlo por cada elemento.


````python
# Un array (vector) y un escalar numérico
distancias_millas = np.array([1.0, 2.0, 5.0, 10.0])

# El número 1.6 hace 'broadcast' internamente a [1.6, 1.6, 1.6, 1.6] de forma ultrarrápida
distancias_km = distancias_millas * 1.6
print("Conversión a KM:", distancias_km)
````

⚠️ **Nota sobre la compatibilidad:** Las dimensiones de tu array deben ser compatibles. Esto sucede cuando son iguales, o cuando una de las dimensiones es `1`. Si las dimensiones son totalmente asimétricas e incompatibles (ej. multiplicar una matriz 3x3 con un vector de 2 elementos), obtendrás un `ValueError`.

### 3. Operaciones de Agregación (Reducción) 📉

Las funciones de agregación toman una estructura de datos masiva y la **colapsan o reducen** a un solo número (o a un vector pequeño) extrayendo información estadística vital de los datos en bruto.

NumPy está repleto de estas funciones. Además de `.min()`, `.max()`, y `.sum()`, puedes usar fácilmente `.mean()` (promedio), `.prod()` (multiplicación de todos los elementos), y `.std()` (desviación estándar).

Por defecto, si no le especificas nada, estas funciones **aplastarán** la matriz entera y te devolverán un estadístico global sobre todos los números contenidos en ella.


````python
a = np.array([
    [0.45053314, 0.17296777, 0.34376245, 0.5510652],
    [0.54627315, 0.05093587, 0.40067661, 0.55645993],
    [0.12697628, 0.82485143, 0.26590556, 0.56917101]
])

print("Suma total de toda la matriz:", a.sum())
print("Valor mínimo en toda la matriz:", a.min())
print("Promedio global de toda la matriz:", a.mean())
````

**El parámetro Mágico: `axis`**

Rara vez querrás el promedio de toda tu hoja de cálculo. Generalmente querrás "el promedio de cada columna" (ej. temperatura media diaria) o "la suma de cada fila" (ej. notas totales por estudiante).

El parámetro `axis` (eje) controla cuál dimensión será "colapsada" matemáticamente:
*   `axis=0`: Recorre la matriz verticalmente hacia abajo. Colapsa las filas, entregando un resultado **por columna**.
*   `axis=1`: Recorre la matriz horizontalmente. Colapsa las columnas, entregando un resultado **por fila**.


````python
# Retorna el valor mínimo contenido en CADA COLUMNA
# Al haber 4 columnas, obtenemos 4 resultados.
print("Mínimos por Columna (axis=0):", a.min(axis=0))

# Retorna la suma contenida en CADA FILA
# Al haber 3 filas, obtenemos 3 resultados.
print("Suma por Fila (axis=1):", a.sum(axis=1))
````

### 4. Operaciones Matriciales (Álgebra Lineal) 📐

Uno de los errores más comunes de los analistas que provienen de Matlab o R es creer que el operador asterisco `*` realizará el clásico "producto punto" del álgebra lineal. **No lo hace.** En NumPy, el `*` siempre hará un producto *elemento a elemento*.

Para realizar el auténtico **producto de matrices** (donde multiplicas filas por columnas matemáticamente), debes utilizar el operador **`@`** (introducido en Python 3.5) o invocar el método matricial **`.dot()`**.

Además, con mucha frecuencia necesitarás rotar o pivotar tu matriz, intercambiando las filas por las columnas. Esto se conoce como la matriz **transpuesta** y en NumPy puedes obtenerla instantáneamente llamando al atributo **`.T`**.


````python
A = np.array([[1, 1], [0, 1]])
B = np.array([[2, 0], [3, 4]])

print("Producto elemento a elemento (A * B):")
print(A * B)

print("\nProducto de Matrices de Álgebra Lineal (A @ B):")
print(A @ B)

print("\nTransposición (A.T):")
print("Matriz A original:\n", A)
print("Matriz A transpuesta:\n", A.T)
````
