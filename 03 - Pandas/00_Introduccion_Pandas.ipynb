## ¿Qué es Pandas? 🐼
Pandas es una biblioteca de Python de código abierto que proporciona estructuras de datos y herramientas de análisis de datos fáciles de usar para la manipulación y el análisis de datos. Es una de las bibliotecas más utilizadas para el análisis de datos en el ecosistema de Python y es particularmente popular en campos como la ciencia de datos y el aprendizaje automático (machine learning).

En el nivel más básico, los objetos de Pandas pueden considerarse como versiones mejoradas de los arreglos de NumPy, en los que las filas y columnas se identifican con etiquetas en lugar de simples índices enteros. Esto mejora la flexibilidad en la manipulación de datos y permite realizar operaciones similares a las de bases de datos en los arreglos.

🤓 Para información más detallada, [lee aquí](https://pandas.pydata.org/docs/getting_started/overview.html)
