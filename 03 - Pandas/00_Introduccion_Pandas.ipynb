## ¿Qué es Pandas y cuál es su ventaja frente a NumPy? 🐼

**Pandas** (nombre derivado de *Panel Data*) es una biblioteca de código abierto construida sobre NumPy que introduce estructuras de datos de alto nivel con **etiquetas semánticas** (nombres de columnas e índices en filas).

Mientras que NumPy está optimizado para matrices numéricas homogéneas de álgebra lineal, los datos del mundo real son **heterogéneos** (combinan números, texto, fechas y valores nulos). Pandas llena este vacío proporcionando:

1. **Estructuras Tabulares Flexibles**: El `DataFrame` permite que cada columna tenga su propio tipo de dato (`int`, `float`, `object`, `datetime`, `category`).
2. **Alineación Automática por Índice**: Las operaciones entre estructuras se alinean automáticamente por sus etiquetas, evitando desajustes accidentales.
3. **Manejo Nativo de Datos Faltantes**: Detección, exclusión e imputación integrada de valores `NaN` / `None`.
4. **Operaciones Relacionales y de Agregación**: Capacidad completa para ejecutar operaciones de estilo SQL (`groupby`, `merge`, `pivot_table`, `melt`) a la velocidad de código C optimizado.

---
## Las Dos Estructuras Fundamentales 🧱

- **`pd.Series`**: Arreglo unidimensional etiquetado capaz de contener cualquier tipo de dato. Es la representación de una columna individual en una tabla.
- **`pd.DataFrame`**: Estructura de datos tabular bidimensional con filas y columnas etiquetadas. Es análogo a una hoja de cálculo de Excel o una tabla de base de datos relacional (SQL).
