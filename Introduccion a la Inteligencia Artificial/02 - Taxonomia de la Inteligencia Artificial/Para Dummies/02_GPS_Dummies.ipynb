En 1957, dos científicos crearon uno de los primeros programas de la historia que intentaba resolver **cualquier rompecabezas**, no solo uno en particular. Le pusieron el nombre más directo posible: **Solucionador General de Problemas** (*General Problem Solver*, o **GPS**).

## 1. La Gran Analogía: El Viajero con un Mapa a Medias 🗺️🚗

Imagina que quieres manejar desde tu casa hasta la playa, pero solo tienes un mapa a medias: sabes dónde estás, sabes a dónde quieres llegar, y conoces algunas "movidas" posibles (tomar tal carretera, cruzar tal puente). No sabes de antemano cuál es la ruta exacta.

¿Qué harías? Probablemente:
1. Mirarías el mapa y notarías **qué tan lejos** estás todavía de la playa.
2. Elegirías la carretera que **más te acerque** a la playa.
3. Si esa carretera tiene un peaje que no puedes pagar todavía, primero resolverías **eso** (conseguir monedas) antes de seguir.
4. Repetirías el proceso hasta llegar.

¡Eso es *exactamente* lo que hace un GPS (el programa, no el de tu carro) para resolver cualquier rompecabezas!

---
## 2. La Receta del Detective: Medios y Fines 🕵️

| Paso de la receta | En español de la calle |
|---|---|
| **Comparar** | "¿En qué se parece mi situación actual a la meta, y en qué no?" |
| **Elegir una movida** | "¿Qué acción me acerca más a la meta?" |
| **Revisar requisitos** | "¿Puedo hacer esa movida ya, o necesito resolver algo antes?" |
| **Resolver lo pendiente primero** | Si falta algo, se convierte en un mini-rompecabezas aparte |
| **Actuar y repetir** | Se hace la movida y se vuelve a comparar, hasta llegar a la meta |

A esta receta los científicos la llamaron **análisis medios-fines**: usar los "medios" (las movidas disponibles) para achicar la diferencia con el "fin" (la meta).

---
## 3. ¿Por qué este Detective fue Tan Importante? 🏆

Antes del GPS, cada programa se escribía **a la medida** de un solo problema (uno para jugar damas, otro para geometría, etc). El GPS fue de los primeros en decir: "dame CUALQUIER rompecabezas descrito de cierta forma (dónde empiezo, adónde quiero llegar, qué movidas tengo) y yo lo resuelvo con la misma receta".

Fue como inventar una **navaja suiza** en lugar de una caja llena de herramientas especializadas: la misma "receta" servía para muchos rompecabezas distintos, con solo cambiar los datos de entrada.

---
## 4. Pero... ¿Qué le Faltó a Este Detective? 🤔

* **Se perdía en rompecabezas grandes:** con pocos movimientos posibles, el detective era genial. Pero si el rompecabezas tenía millones de combinaciones posibles, se tardaba una eternidad revisándolas casi todas.
* **Alguien tenía que explicarle el rompecabezas a mano:** el detective no "entendía" el problema por sí solo; un humano tenía que traducírselo primero a su formato especial.
* **No aprendía de la experiencia:** resolver el mismo tipo de rompecabezas mil veces no lo hacía ni un poquito más rápido la vez 1001.

Por eso, muchos años después, aparecieron otro tipo de IAs que en lugar de seguir una receta fija, **aprenden de muchísimos ejemplos** — como el "Dr. Aprendizaje" que conociste en el primer cuaderno de este módulo.

## 5. Manos a la Obra: El Rompecabezas de las Jarras de Agua 🏺💧

Vamos a programar a nuestro propio mini-detective para resolver un rompecabezas clásico:

> Tienes una jarra que aguanta **4 litros** y otra que aguanta **3 litros**. Ninguna tiene marcas de medición y ambas están vacías. **¿Cómo consigues exactamente 2 litros** en alguna de las dos, usando solo llenar, vaciar y verter entre ellas?

Para nuestro detective:
* **"Dónde estoy":** cuánta agua hay en cada jarra ahora mismo, por ejemplo `(0, 0)` (las dos vacías).
* **"Adónde quiero llegar":** cualquier momento en que una jarra tenga exactamente 2 litros.
* **"Movidas posibles":** llenar una jarra, vaciarla, o verter el agua de una a la otra.

Ejecuta la siguiente celda con `Shift + Enter` y observa cómo el detective encuentra el camino solo, probando movidas sistemáticamente:

````python
from collections import deque

CAPACIDAD_JARRA_A, CAPACIDAD_JARRA_B = 4, 3
ESTADO_INICIAL = (0, 0)
es_la_meta = lambda estado: estado[0] == 2 or estado[1] == 2

def llenar_A(estado):
    a, b = estado
    return (CAPACIDAD_JARRA_A, b)

def llenar_B(estado):
    a, b = estado
    return (a, CAPACIDAD_JARRA_B)

def vaciar_A(estado):
    a, b = estado
    return (0, b)

def vaciar_B(estado):
    a, b = estado
    return (a, 0)

def verter_A_en_B(estado):
    a, b = estado
    transferido = min(a, CAPACIDAD_JARRA_B - b)
    return (a - transferido, b + transferido)

def verter_B_en_A(estado):
    a, b = estado
    transferido = min(b, CAPACIDAD_JARRA_A - a)
    return (a + transferido, b - transferido)

movidas_posibles = {
    "Llenar jarra A": llenar_A,
    "Llenar jarra B": llenar_B,
    "Vaciar jarra A": vaciar_A,
    "Vaciar jarra B": vaciar_B,
    "Verter A en B": verter_A_en_B,
    "Verter B en A": verter_B_en_A,
}

def detective_resuelve(estado_inicial, es_meta, movidas):
    caminos_por_explorar = deque([[estado_inicial]])
    ya_visitados = {estado_inicial}
    while caminos_por_explorar:
        camino = caminos_por_explorar.popleft()
        estado_actual = camino[-1]
        if es_meta(estado_actual):
            return camino
        for nombre_movida, hacer_movida in movidas.items():
            nuevo_estado = hacer_movida(estado_actual)
            if nuevo_estado not in ya_visitados:
                ya_visitados.add(nuevo_estado)
                caminos_por_explorar.append(camino + [nuevo_estado])
    return None

solucion = detective_resuelve(ESTADO_INICIAL, es_la_meta, movidas_posibles)

print("🕵️ El detective arranca en:", ESTADO_INICIAL, "(las dos jarras vacías)")
print("🎯 Meta: que alguna jarra tenga exactamente 2 litros\n")

if solucion:
    print(f"✅ ¡Encontró la solución en {len(solucion) - 1} movidas!\n")
    for i, estado in enumerate(solucion):
        etiqueta = "Inicio" if i == 0 else f"Movida {i}"
        print(f"{etiqueta}: Jarra A={estado[0]}L, Jarra B={estado[1]}L")
else:
    print("❌ El detective no encontró solución.")
````

### 🤔 ¿Qué acaba de pasar?

El detective fue probando, movida por movida (`llenar`, `vaciar`, `verter`), todos los caminos posibles hasta encontrar uno donde alguna jarra quedara con exactamente 2 litros. La lista de pasos que imprimió es, ni más ni menos, **la solución completa** al rompecabezas: la misma que encontrarías tú si te sentaras con lápiz y papel a probar combinaciones — solo que el detective la encontró en una fracción de segundo.

## 6. Otro Rompecabezas para el Detective: La Torre de 3 Discos 🗼

Probemos con otro clásico: la **Torre de Hanoi**. Tienes 3 discos de distinto tamaño apilados en el primer poste (del más grande abajo al más chiquito arriba), y debes moverlos todos al tercer poste, usando el segundo como apoyo, **sin poner nunca un disco grande sobre uno más chiquito**.

Igual que con las jarras, cada movida es un pasito que acerca al detective a la meta (todos los discos en el poste correcto).

````python
def torre_de_hanoi(cuantos_discos, origen, destino, apoyo, lista_de_movidas):
    if cuantos_discos == 0:
        return
    torre_de_hanoi(cuantos_discos - 1, origen, apoyo, destino, lista_de_movidas)
    lista_de_movidas.append((cuantos_discos, origen, destino))
    torre_de_hanoi(cuantos_discos - 1, apoyo, destino, origen, lista_de_movidas)

movidas = []
torre_de_hanoi(3, "Poste 1", "Poste 3", "Poste 2", movidas)

print("🗼 El detective resuelve la Torre de Hanoi de 3 discos:\n")
for i, (disco, origen, destino) in enumerate(movidas, start=1):
    print(f"Movida {i}: mover disco {disco} de {origen} a {destino}")

print(f"\nTotal de movidas: {len(movidas)}")
````

## 7. El Detective de 1957 vs. la IA de Hoy 🆚

| | El Detective (GPS, 1957) | La IA de Hoy (Aprendizaje Automático) |
|---|---|---|
| **¿Cómo "sabe" qué hacer?** | Alguien le explicó las reglas del rompecabezas a mano | Aprendió viendo muchísimos ejemplos parecidos |
| **¿Se le puede seguir el rastro?** | Sí, cada pasito queda anotado y se puede explicar | A veces es más difícil de explicar por qué decidió algo |
| **¿Sirve para rompecabezas gigantes?** | No muy bien, se pierde entre tantas combinaciones | Mucho mejor, aunque necesita muchos datos y cómputo |

Los dos enfoques **no compiten**: hoy en día muchas aplicaciones (como la logística de un camión de entregas) combinan un poco de cada uno.
