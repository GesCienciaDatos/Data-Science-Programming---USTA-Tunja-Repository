## 3. Conozcamos al Dr. Aprendizaje 🩺

El Dr. Aprendizaje no memoriza reglas fijas: aprende **comparando ejemplos parecidos**. Es como cuando aprendiste a distinguir una manzana de una uva sin que nadie te diera una fórmula matemática: simplemente viste muchas frutas y notaste el patrón (tamaño, textura).

Vamos a construir, con Python bien sencillo, un "Dr. Aprendizaje" de bolsillo que adivina qué fruta es una nueva, comparándola con frutas que ya conoce.

````python
# El Dr. Aprendizaje "recuerda" frutas que ya vio antes (peso en gramos y qué tan rugosa es de 0 a 10)
frutas_que_ya_conoce = [
    ("manzana", 150, 2), ("manzana", 130, 3), ("manzana", 160, 2),
    ("naranja", 170, 6), ("naranja", 180, 7), ("naranja", 165, 6),
    ("uva", 5, 1), ("uva", 6, 1), ("uva", 4, 2),
]

def que_tan_parecidas_son(fruta_1, fruta_2):
    # Una "regla de distancia" bien simple: entre más parecidas, más chiquito el número
    return ((fruta_1[0] - fruta_2[0]) ** 2 + (fruta_1[1] - fruta_2[1]) ** 2) ** 0.5

def adivinar_fruta(fruta_nueva, cuantas_comparar=3):
    mas_parecidas = sorted(
        frutas_que_ya_conoce,
        key=lambda f: que_tan_parecidas_son((f[1], f[2]), fruta_nueva)
    )[:cuantas_comparar]
    nombres = [f[0] for f in mas_parecidas]
    return max(set(nombres), key=nombres.count), nombres

fruta_misteriosa = (155, 2.5)  # peso, rugosidad
adivinanza, comparadas_con = adivinar_fruta(fruta_misteriosa)

print("🍇 Fruta misteriosa: pesa", fruta_misteriosa[0], "gramos y tiene rugosidad", fruta_misteriosa[1])
print("🔍 El Dr. Aprendizaje la comparó con:", comparadas_con)
print("✅ ¡Adivinó que es una...", adivinanza, "!")
````

## 4. Conozcamos al Traductor de Idiomas 🗣️

Este especialista no traduce entre español e inglés (bueno, también puede), sino que entiende **el tono** de lo que escribes. Es el mismo tipo de tecnología detrás de "¿esta reseña es buena o mala?".

Le vamos a dar una frase y va a decidir si suena positiva, negativa o neutral, buscando palabras clave que ya conoce.

````python
palabras_que_conoce_el_traductor = {
    "excelente": 2, "increíble": 2, "bueno": 1, "genial": 1,
    "malo": -1, "terrible": -2, "aburrido": -1, "lento": -1,
}

def separar_en_palabras(frase):
    return frase.lower().replace(",", "").replace(".", "").split()

def opinar_sobre_la_frase(frase):
    palabras = separar_en_palabras(frase)
    puntos = sum(palabras_que_conoce_el_traductor.get(p, 0) for p in palabras)
    if puntos > 0:
        return "¡Le gustó! 😀"
    elif puntos < 0:
        return "No le gustó 😞"
    else:
        return "Le da igual 😐"

frase_de_prueba = "El curso es excelente, aunque el material se sintió un poco lento al inicio."
print("Frase:", frase_de_prueba)
print("Opinión del Traductor de Idiomas:", opinar_sobre_la_frase(frase_de_prueba))
````

## 5. Conozcamos al Oftalmólogo (Visión por Computador) 👁️

Este especialista "mira" imágenes representadas como una cuadrícula de números (entre más oscuro el número, más oscuro el pixel). Vamos a mostrarle una fotico muy pequeñita, mitad oscura y mitad clara, y va a encontrar **dónde está el borde** entre ambas mitades — el primer paso para "reconocer" algo en una imagen.

````python
fotico_pequenita = [
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
]

lupa_para_bordes = [
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1],
]

def pasar_la_lupa(imagen, lupa):
    alto, ancho = len(imagen), len(imagen[0])
    resultado = [[0] * ancho for _ in range(alto)]
    for i in range(1, alto - 1):
        for j in range(1, ancho - 1):
            suma = 0
            for ki in range(-1, 2):
                for kj in range(-1, 2):
                    suma += imagen[i + ki][j + kj] * lupa[ki + 1][kj + 1]
            resultado[i][j] = suma
    return resultado

bordes_encontrados = pasar_la_lupa(fotico_pequenita, lupa_para_bordes)

print("📷 La fotico (números grandes = más clarito):")
for fila in fotico_pequenita:
    print(fila)

print("\n🔎 Lo que 've' el Oftalmólogo (los números grandes marcan dónde hay un borde):")
for fila in bordes_encontrados:
    print(fila)
````

## 6. El Doctor con Manual, el Coordinador de Agenda y el Artista 📜🗺️🎨

* **El Doctor con Manual (Sistemas Expertos)** no "aprende": sigue un recetario fijo de reglas "si pasa esto, entonces haz esto otro", como cuando llamas a soporte técnico y te preguntan "¿está encendido el router?" antes de cualquier otra cosa.
* **El Coordinador de Agenda (Planificación)** no decide UNA acción, sino **el orden completo de pasos** para llegar a un objetivo, como cuando el GPS del carro arma toda la ruta antes de que arranques.
* **El Artista (IA Generativa)** no clasifica ni ordena nada: **inventa cosas nuevas** parecidas a lo que ha visto, como cuando le pides una imagen o un texto que nunca había existido antes.

Vamos a construirle un mini-recetario al Doctor con Manual: le daremos unos datos del clima y él decidirá, siguiendo reglas fijas, si recomienda llevar paraguas.

````python
clima_de_hoy = {
    "probabilidad_lluvia": 70,   # en porcentaje
    "hace_mucho_viento": False,
}

def doctor_con_manual(clima):
    if clima["probabilidad_lluvia"] >= 60 and clima["hace_mucho_viento"]:
        return "☔🌬️ Lleva paraguas reforzado, ¡va a llover con viento!"
    elif clima["probabilidad_lluvia"] >= 60:
        return "☔ Lleva paraguas, hay buena probabilidad de lluvia."
    elif clima["probabilidad_lluvia"] >= 30:
        return "🤔 Quizás lleves un paraguas plegable, por si las moscas."
    else:
        return "☀️ No necesitas paraguas hoy."

recomendacion = doctor_con_manual(clima_de_hoy)
print("Datos del clima:", clima_de_hoy)
print("Recomendación del Doctor con Manual:", recomendacion)
````
