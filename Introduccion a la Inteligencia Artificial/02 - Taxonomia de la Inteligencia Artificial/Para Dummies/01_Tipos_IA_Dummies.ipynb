## 1. La Gran Analogía: La Escalera de la Inteligencia 🪜

Imagina una escalera con tres escalones:



* **Escalón 1 (IA Débil):** una IA que hace **una sola cosa** muy bien, como un jugador de ajedrez que solo sabe jugar ajedrez. **Toda la IA que existe hoy está en este escalón**, incluyendo los asistentes conversacionales más avanzados.
* **Escalón 2 (IA General):** una IA que pudiera aprender **cualquier cosa**, tan flexible como una persona. Todavía **no existe**; es una meta de investigación.
* **Escalón 3 (Superinteligencia):** una IA que fuera **más lista que todos los humanos juntos** en casi todo. Es 100% especulación, tema de películas y debates filosóficos.

---
## 2. La Guía Rápida de la Escalera 📋

| Escalón | Nombre técnico | ¿Existe ya? | Ejemplo |
|---|---|---|---|
| 1️⃣ | IA Débil / Estrecha | ✅ Sí, toda la IA actual | Asistentes virtuales, apps de recomendación, filtros de spam |
| 2️⃣ | IA General (AGI) | ❌ No | Ninguno confirmado |
| 3️⃣ | Superinteligencia (ASI) | ❌ No | Ninguno, es especulación |

## 3. Comprobemos que la IA de Hoy es "de un Solo Truco" 🎩

Vamos a construir una mini-IA que detecta correos molestos (spam) buscando palabras sospechosas. Funciona genial... **para eso, y solo para eso**. Si le pides que haga otra cosa completamente distinta, no sabe qué hacer — justo como un jugador de ajedrez al que le pides que cocine.

````python
palabras_sospechosas = {"gratis", "gane", "premio", "urgente"}

def detector_de_correos_molestos(correo):
    correo_en_minusculas = correo.lower()
    pistas_encontradas = [p for p in palabras_sospechosas if p in correo_en_minusculas]
    return len(pistas_encontradas) > 0, pistas_encontradas

correo_sospechoso = "¡Gane un premio gratis ya mismo!"
correo_normal = "Hola, ¿nos vemos mañana para el proyecto?"

for correo in (correo_sospechoso, correo_normal):
    es_molesto, pistas = detector_de_correos_molestos(correo)
    print(f"Correo: '{correo}' -> ¿Molesto? {es_molesto} (pistas: {pistas})")

# Ahora intentemos pedirle algo TOTALMENTE distinto: que sume dos números
print("\n🎩 Esta IA solo sabe hacer UN truco. Si le pedimos otra cosa...")
try:
    detector_de_correos_molestos(4 + 5)
except Exception as error:
    print(f"❌ No supo qué hacer: {type(error).__name__}. Justo como un jugador de ajedrez al que le piden cocinar.")
````

## 4. Los 4 "Niveles de Memoria" de una IA 🎮

Aparte de qué tan lista es una IA (la escalera de arriba), también podemos preguntarnos **cómo "recuerda" las cosas** mientras decide. Un investigador llamado Arend Hintze propuso 4 niveles, como si fueran niveles de un videojuego:

```mermaid
graph LR
    N1[Nivel 1: Sin memoria] --> N2[Nivel 2: Memoria de corto plazo]
    N2 --> N3[Nivel 3: Entiende a otros - aún no existe]
    N3 --> N4[Nivel 4: Se conoce a sí misma - pura ficción]
```

Solo los **Niveles 1 y 2** existen en la vida real hoy en día.

---
## 5. Nivel 1: Sin Memoria (Máquina Reactiva) ⚡

Este tipo de IA **no recuerda nada de antes**. Cada vez que le preguntas algo, responde solo mirando lo que tiene justo en frente, como un guardia de seguridad que solo reacciona a lo que ve en el momento, sin importar quién entró ayer.

* **Ejemplo real:** *Deep Blue*, la IA que le ganó al campeón mundial de ajedrez en 1997. Evaluaba el tablero actual y ya, sin "recordar" partidas anteriores.

Construyamos un mini-guardia de seguridad sin memoria: decide una acción solo mirando la temperatura del momento, sin importar el orden en que le lleguen los datos.

````python
def guardia_sin_memoria(temperatura_actual):
    # Decide SOLO con el dato de ahora mismo, nunca recuerda lecturas anteriores
    if temperatura_actual > 26:
        return "Encender aire acondicionado"
    elif temperatura_actual < 18:
        return "Encender calefacción"
    else:
        return "Dejar todo como está"

lecturas = [22, 28, 15, 24, 30, 17]

print("El guardia sin memoria decide mirando SOLO la lectura actual:\n")
for temperatura in lecturas:
    print(f"Temperatura ahora: {temperatura}°C -> Decisión: {guardia_sin_memoria(temperatura)}")

print("\n🔎 Si cambiamos el orden de llegada, cada decisión sigue siendo la misma:")
for temperatura in sorted(lecturas):
    print(f"Temperatura ahora: {temperatura}°C -> Decisión: {guardia_sin_memoria(temperatura)}")
````

## 6. Nivel 2: Memoria de Corto Plazo (Memoria Limitada) 🧾

Esta IA **sí recuerda un ratito**, como cuando te acuerdas de las últimas 3 jugadas de un juego de mesa para decidir tu siguiente movimiento, pero no te acuerdas de la partida de la semana pasada. **Casi toda la IA moderna** —incluyendo los carros que se manejan solos— funciona así.

Construyamos un mini-agente que recuerda solo las últimas 3 lecturas de velocidad para decidir si debe frenar.

````python
from collections import deque

class AgenteConMemoriaCorta:
    def __init__(self, cuantas_recuerda=3):
        self.recuerdos = deque(maxlen=cuantas_recuerda)

    def decidir(self, velocidad_actual):
        self.recuerdos.append(velocidad_actual)
        promedio = sum(self.recuerdos) / len(self.recuerdos)
        if promedio > 80:
            return f"promedio reciente {promedio:.1f} km/h -> ¡frenar un poco!"
        return f"promedio reciente {promedio:.1f} km/h -> todo tranquilo"

agente = AgenteConMemoriaCorta(cuantas_recuerda=3)
lecturas_velocidad = [70, 75, 90, 95, 100, 60, 55]

print("El agente con memoria corta SÍ recuerda las últimas 3 lecturas:\n")
for v in lecturas_velocidad:
    print(f"Nueva lectura: {v} km/h | Recuerda: {list(agente.recuerdos)} -> {agente.decidir(v)}")
````

## 7. Niveles 3 y 4: Lo que Todavía No Existe 🔮

* **Nivel 3 (Entiende a otros / Teoría de la Mente):** sería una IA capaz de imaginarse qué está pensando o sintiendo otra persona, como cuando notas que un amigo está triste aunque no te lo haya dicho. Hoy en día **no existe** una IA que haga esto de verdad; apenas hay intentos muy limitados en robots sociales.
* **Nivel 4 (Se conoce a sí misma / Autoconciencia):** sería una IA consciente de que ella misma existe, como cuando tú piensas "yo estoy pensando". Esto es **pura especulación** y ni siquiera los científicos se ponen de acuerdo sobre si algún día será posible o cómo lo comprobaríamos.

> 🧸 No, ninguna IA de hoy —ni la más avanzada— "siente" ni "es consciente" de sí misma. Eso sigue siendo terreno de la ciencia ficción.

````python
catalogo = [
    {"ejemplo": "Deep Blue (ajedrez)", "escalon": "IA Débil", "nivel": "Sin memoria"},
    {"ejemplo": "Termostato inteligente", "escalon": "IA Débil", "nivel": "Sin memoria"},
    {"ejemplo": "Carro que se maneja solo", "escalon": "IA Débil", "nivel": "Memoria de corto plazo"},
    {"ejemplo": "App de recomendaciones", "escalon": "IA Débil", "nivel": "Memoria de corto plazo"},
]

for item in catalogo:
    print(f"🔹 {item['ejemplo']}: {item['escalon']} + {item['nivel']}")
````
