## 1. Dos Ejes de Clasificación 🧭

Cuando hablamos de "tipos de IA" en realidad estamos combinando **dos clasificaciones independientes**: una basada en *qué tan capaz* es el sistema y otra basada en *cómo funciona internamente* desde el punto de vista cognitivo.


Ambas clasificaciones son complementarias: un mismo sistema (por ejemplo, un vehículo autónomo) se puede describir a la vez como **IA Débil** (por capacidad) y de **Memoria Limitada** (por funcionalidad).

---
## 2. Clasificación por Capacidad: Débil, General y Superinteligencia ⚖️

| Tipo | Definición | ¿Existe hoy? | Ejemplos |
|---|---|---|---|
| **IA Débil / Estrecha (ANI)** | Diseñada y entrenada para **una tarea específica**; no puede transferir su "inteligencia" a otro dominio. | ✅ Sí, es toda la IA que existe actualmente. | Asistentes virtuales, sistemas de recomendación, detectores de spam, modelos de lenguaje conversacional. |
| **IA General (AGI)** | Un sistema hipotético con capacidad de razonar, aprender y resolver **cualquier tarea intelectual** al nivel de un humano. | ❌ No existe todavía; es un objetivo de investigación activo y debatido. | Ninguno confirmado; tema de investigación y especulación en la comunidad científica. |
| **Superinteligencia (ASI)** | Un sistema hipotético que **superaría** la capacidad cognitiva humana en prácticamente todos los dominios. | ❌ Puramente especulativo. | Ninguno; concepto discutido en filosofía de la IA y estudios de riesgo existencial. |

> ⚠️ Toda la IA que usas hoy —incluyendo los modelos de lenguaje más avanzados— es **IA Débil**: por más sorprendente que parezca su desempeño en lenguaje, no puede "transferirse" de forma autónoma para, por ejemplo, operar un quirófano sin haber sido diseñada explícitamente para ello.

````python
palabras_spam = {"gratis", "gane", "premio", "urgente", "haga clic"}

def detector_spam(correo):
    correo_normalizado = correo.lower()
    coincidencias = [p for p in palabras_spam if p in correo_normalizado]
    return len(coincidencias) > 0, coincidencias

correo_1 = "¡Gane un premio gratis, haga clic ya!"
correo_2 = "Hola equipo, adjunto el informe del proyecto de la semana."

for correo in (correo_1, correo_2):
    es_spam, pistas = detector_spam(correo)
    print(f"Correo: '{correo}'")
    print(f"  -> ¿Es spam? {es_spam} (pistas: {pistas})\n")

# Esta misma IA "estrecha" es incapaz de resolver una tarea distinta,
# aunque sea mucho más simple, porque solo fue diseñada para detectar spam:
try:
    resultado = detector_spam(4 + 5)  # Intentamos pedirle que "sume" en vez de clasificar texto
except Exception as error:
    print(f"❌ El sistema falla fuera de su dominio estrecho: {type(error).__name__}: {error}")
````

## 3. ¿Por qué la IA General (AGI) sigue siendo especulativa? 🔮

La demostración anterior evidencia el punto central de la IA Débil: un sistema entrenado para una tarea **falla o simplemente no aplica** fuera de su dominio. Alcanzar una IA General implicaría resolver, entre otros, los siguientes retos abiertos:

| Reto abierto | ¿Por qué es difícil? |
|---|---|
| **Transferencia de aprendizaje general** | Los sistemas actuales no reutilizan su "conocimiento" entre dominios muy distintos sin reentrenamiento. |
| **Razonamiento de sentido común** | Entender implícitos cotidianos (causalidad, física intuitiva) sigue siendo limitado. |
| **Aprendizaje continuo** | La mayoría de modelos no aprenden de forma incremental después de su entrenamiento inicial. |
| **Autonomía de objetivos** | Definir y perseguir metas propias de forma segura y alineada con valores humanos. |

No existe consenso científico sobre **cuándo** (o si) se logrará la AGI; las estimaciones de expertos varían entre décadas y siglos, y algunos consideran que podría no ser alcanzable con los paradigmas actuales.

## 4. Clasificación por Funcionalidad: La Taxonomía de Arend Hintze (2016) 🧠

El investigador **Arend Hintze** propuso en 2016 una clasificación de la IA en **cuatro tipos**, ordenados según su sofisticación cognitiva. A diferencia de la clasificación por capacidad, esta taxonomía describe **cómo procesa la información** el sistema, sin importar si es una IA débil o (hipotéticamente) general.

```mermaid

    R[Tipo 1: Máquinas Reactivas] --> M[Tipo 2: Memoria Limitada]
    M --> TM[Tipo 3: Teoría de la Mente]
```

> 💡 Solo los **Tipos 1 y 2** existen en sistemas reales hoy en día. Los Tipos 3 y 4 son, por ahora, terreno de la investigación teórica y la ciencia ficción.

## 5. Tipo 1 — Máquinas Reactivas (*Reactive Machines*) ⚡

Son los sistemas más simples: **reaccionan a la percepción actual** sin ningún tipo de memoria del pasado ni capacidad de usar experiencias previas para decisiones futuras.

* **Ejemplo histórico:** *Deep Blue*, el sistema de IBM que venció al campeón mundial de ajedrez Garry Kasparov en 1997. Evaluaba el tablero actual y calculaba el mejor movimiento, pero no "recordaba" partidas anteriores para mejorar.
* **Ejemplo cotidiano:** un termostato programado o un filtro de spam basado en reglas fijas.

Implementemos un agente puramente reactivo para comprobar que su decisión depende **únicamente** de la percepción del instante presente.

````python
def agente_reactivo(temperatura_actual):
    """Decide una acción basándose ÚNICAMENTE en la percepción actual, sin memoria del pasado."""
    if temperatura_actual > 26:
        return "Encender aire acondicionado"
    elif temperatura_actual < 18:
        return "Encender calefacción"
    else:
        return "Mantener sistema apagado"

lecturas_de_temperatura = [22, 28, 15, 24, 30, 17]

print("Simulación de un agente reactivo (sin memoria de estados anteriores):\n")
for t in lecturas_de_temperatura:
    accion = agente_reactivo(t)
    print(f"Percepción actual: {t}°C -> Acción: {accion}")

print("\n🔎 Nota: el agente decide igual sin importar el orden de las lecturas.")
print("Si reordenamos la lista, cada decisión individual no cambia:")
for t in sorted(lecturas_de_temperatura):
    print(f"Percepción actual: {t}°C -> Acción: {agente_reactivo(t)}")
````

## 6. Tipo 2 — Memoria Limitada (*Limited Memory*) 🧾

Estos sistemas **sí usan experiencias pasadas** (generalmente una ventana reciente de observaciones) para tomar decisiones, aunque esa memoria es transitoria y no se acumula indefinidamente como en un humano. **La inmensa mayoría de los sistemas de Aprendizaje Automático y Deep Learning actuales pertenecen a este tipo.**

* **Ejemplo:** un vehículo autónomo que observa la trayectoria reciente de los vehículos cercanos (últimos segundos) para predecir su comportamiento futuro.
* **Ejemplo cotidiano:** un sistema de recomendación que ajusta sus sugerencias según tus interacciones recientes.

Implementemos un agente de memoria limitada usando una ventana deslizante de observaciones.

````python
from collections import deque

class AgenteMemoriaLimitada:
    def __init__(self, ventana=3):
        self.historial = deque(maxlen=ventana)

    def percibir_y_decidir(self, velocidad_actual):
        self.historial.append(velocidad_actual)
        promedio_reciente = sum(self.historial) / len(self.historial)
        if promedio_reciente > 80:
            return f"Promedio reciente={promedio_reciente:.1f} km/h -> Reducir velocidad (posible exceso sostenido)"
        else:
            return f"Promedio reciente={promedio_reciente:.1f} km/h -> Mantener velocidad actual"

agente = AgenteMemoriaLimitada(ventana=3)
lecturas_velocidad = [70, 75, 90, 95, 100, 60, 55]

print("Simulación de un agente de memoria limitada (usa las últimas 3 lecturas):\n")
for v in lecturas_velocidad:
    decision = agente.percibir_y_decidir(v)
    print(f"Nueva lectura: {v} km/h | Historial: {list(agente.historial)} | Decisión -> {decision}")
````

## 7. Tipo 3 — Teoría de la Mente (*Theory of Mind*) 🎭

Un sistema de este tipo debería ser capaz de **modelar los estados mentales de otros agentes**: sus creencias, intenciones, emociones y cómo estas influyen en su comportamiento — algo que los humanos hacemos de forma natural desde la infancia.

* **Estado actual:** es un área de investigación activa (por ejemplo, en robótica social e interacción humano-computador), pero **no existen sistemas** que posean una teoría de la mente genuina y generalizable; a lo sumo hay aproximaciones muy limitadas y específicas de dominio.
* **Por qué es difícil:** requiere representar no solo el entorno físico, sino también los **estados internos inferidos** de otros agentes, algo mucho más abstracto que procesar sensores.

---
## 8. Tipo 4 — Autoconciencia (*Self-Awareness*) 🪞

Es el tipo más avanzado y **puramente hipotético** de la taxonomía: un sistema con **conciencia de su propia existencia**, sus propios estados internos, y capaz de razonar sobre sí mismo como un agente entre otros agentes.

> ⚠️ Este tipo pertenece por completo al terreno filosófico y especulativo. No existe consenso científico sobre si la autoconciencia artificial es siquiera posible, ni sobre cómo se podría verificar de forma objetiva si un sistema la posee ("el problema de la conciencia" en filosofía de la mente).
