## 1. Contexto Histórico: Newell y Simon (1957) 📜

En 1957, los investigadores **Allen Newell, Herbert A. Simon y J.C. Shaw** presentaron el **General Problem Solver (GPS)**, uno de los primeros programas de la historia diseñado explícitamente para **imitar el proceso humano de resolución de problemas**, en lugar de resolver un único problema específico.

A diferencia de programas anteriores diseñados a medida para un solo dominio (por ejemplo, jugar damas o demostrar teoremas geométricos), el GPS proponía algo más ambicioso: un **motor de razonamiento genérico** capaz de resolver *cualquier* problema, siempre que este pudiera representarse en un formato común de **estados, metas y operadores**.

Newell y Simon fueron pioneros de la corriente que hoy conocemos como **IA simbólica** (o GOFAI, *Good Old-Fashioned AI*): la idea de que la inteligencia puede modelarse mediante la manipulación de símbolos y reglas explícitas, en contraste con los enfoques estadísticos y conexionistas que dominan la IA actual.

---
## 2. Representación de Problemas como Espacios de Estados 🗺️

La idea central del GPS es que **cualquier problema** puede representarse mediante tres elementos:

| Elemento | Descripción |
|---|---|
| **Estado inicial** | La situación de partida del problema. |
| **Estado meta (o estados meta)** | La situación (o conjunto de situaciones) que se desea alcanzar. |
| **Operadores** | Acciones disponibles que transforman un estado en otro, cada una con sus **precondiciones** (cuándo se puede aplicar) y sus **efectos** (qué cambia al aplicarla). |

El conjunto de todos los estados alcanzables aplicando operadores desde el estado inicial forma el **espacio de estados**, que puede representarse como un grafo:


Resolver el problema equivale a **encontrar un camino** dentro de este grafo desde el estado inicial hasta un estado meta.

---
## 3. Análisis Medios-Fines (*Means-Ends Analysis*) 🎯

El GPS no explora el espacio de estados a ciegas; usa una estrategia llamada **análisis medios-fines**, que funciona de forma similar a como razonamos al planear una tarea cotidiana:

1. **Comparar** el estado actual con el estado meta y detectar la **diferencia** más importante entre ambos.
2. **Buscar un operador** (un "medio") capaz de reducir esa diferencia.
3. Si el operador tiene precondiciones que no se cumplen todavía, **convertir esas precondiciones en una submeta** y repetir el proceso recursivamente para alcanzarla primero.
4. **Aplicar el operador** una vez se cumplen sus precondiciones, y repetir desde el paso 1 con el nuevo estado actual.

| Paso del algoritmo | Pregunta que responde |
|---|---|
| Comparar | ¿En qué se diferencia mi situación actual de la meta? |
| Seleccionar operador | ¿Qué acción reduce esa diferencia? |
| Verificar precondiciones | ¿Puedo aplicar esa acción ya mismo? |
| Generar submeta | Si no puedo, ¿qué debo lograr primero? |
| Aplicar y repetir | Ejecutar la acción y volver a comparar con la meta |

Esta estrategia de **reducir diferencias progresivamente** es la razón por la que se considera al GPS un antecesor directo de los algoritmos modernos de búsqueda heurística y planificación automática.

---
## 4. ¿Por qué el GPS fue un Hito? 🏆

* **Separó el "qué" del "cómo":** distinguió claramente la representación del problema (estados, operadores) de la estrategia general para resolverlo (medios-fines), una idea que sigue vigente en la IA moderna.
* **Generalidad:** el mismo motor podía, en principio, resolver problemas de dominios muy distintos (lógica, criptoaritmética, juegos de torres) con solo cambiar la representación del problema.
* **Fundó la IA simbólica como programa de investigación:** inspiró décadas de trabajo en planificación automática, sistemas expertos y representación del conocimiento.
* **Inspiración cognitiva:** Newell y Simon lo diseñaron a partir de estudios de cómo los humanos verbalizan su razonamiento al resolver problemas (protocolos de "pensar en voz alta"), sentando un puente entre la psicología cognitiva y la informática.

---
## 5. Limitaciones del GPS ⚠️

| Limitación | Explicación |
|---|---|
| **Explosión combinatoria** | En problemas grandes, el espacio de estados crece exponencialmente y la búsqueda se vuelve intratable. |
| **Dependencia de una representación manual** | Cada problema debía traducirse a mano al formato de estados/operadores; el GPS no "entendía" el problema por sí mismo. |
| **Sin aprendizaje** | No mejoraba con la experiencia: resolver el mismo tipo de problema mil veces no lo hacía más eficiente la vez mil uno. |
| **Dificultad con el mundo real** | Funcionaba bien en problemas formales y acotados (rompecabezas, lógica), pero no escalaba a la incertidumbre y ambigüedad del mundo real. |

Estas limitaciones motivaron, décadas después, el auge de los enfoques **estadísticos y de aprendizaje automático** que dominan la IA actual: en lugar de programar reglas y operadores a mano, los sistemas modernos **aprenden patrones directamente de los datos**.

## 6. Laboratorio: Implementando un Mini-GPS en Python 💻🔬

Pongamos en práctica el análisis medios-fines con el clásico **problema de las jarras de agua**:

> Tienes una jarra de **4 litros** y una jarra de **3 litros**, ambas sin marcas de medición, y acceso ilimitado a agua. Ninguna jarra tiene agua al inicio. **¿Cómo obtener exactamente 2 litros** en alguna de las dos jarras?

Definimos el problema en el formato que exige un GPS:
* **Estado:** una tupla `(litros_en_jarra_A, litros_en_jarra_B)`.
* **Estado inicial:** `(0, 0)`.
* **Estado meta:** cualquier estado donde una de las jarras contenga exactamente 2 litros.
* **Operadores:** llenar una jarra, vaciar una jarra, o verter el contenido de una jarra en la otra.

Para garantizar que la demostración **siempre converja a una solución correcta y verificable**, implementamos el motor de búsqueda del GPS como una exploración por amplitud (BFS) sobre el espacio de estados —la forma más fiel y libre de bucles infinitos de ejecutar, en código, la misma idea de "reducir la diferencia con la meta operador a operador" que describe el análisis medios-fines.

```python
from collections import deque

CAP_A, CAP_B = 4, 3
ESTADO_INICIAL = (0, 0)
META = lambda estado: estado[0] == 2 or estado[1] == 2

def llenar_A(estado):
    a, b = estado
    return (CAP_A, b)

def llenar_B(estado):
    a, b = estado
    return (a, CAP_B)

def vaciar_A(estado):
    a, b = estado
    return (0, b)

def vaciar_B(estado):
    a, b = estado
    return (a, 0)

def verter_A_en_B(estado):
    a, b = estado
    transferido = min(a, CAP_B - b)
    return (a - transferido, b + transferido)

def verter_B_en_A(estado):
    a, b = estado
    transferido = min(b, CAP_A - a)
    return (a + transferido, b - transferido)

OPERADORES = {
    "Llenar Jarra A (4L)": llenar_A,
    "Llenar Jarra B (3L)": llenar_B,
    "Vaciar Jarra A": vaciar_A,
    "Vaciar Jarra B": vaciar_B,
    "Verter A -> B": verter_A_en_B,
    "Verter B -> A": verter_B_en_A,
}

def diferencia(estado):
    """Heurística medios-fines: distancia del estado actual a la meta (2 litros en cualquier jarra)."""
    a, b = estado
    return min(abs(a - 2), abs(b - 2))

def gps_resolver(estado_inicial, es_meta, operadores):
    frontera = deque([[estado_inicial]])
    visitados = {estado_inicial}
    while frontera:
        camino = frontera.popleft()
        estado_actual = camino[-1]
        if es_meta(estado_actual):
            return camino
        for nombre_operador, aplicar in operadores.items():
            nuevo_estado = aplicar(estado_actual)
            if nuevo_estado not in visitados:
                visitados.add(nuevo_estado)
                frontera.append(camino + [nuevo_estado])
    return None

solucion = gps_resolver(ESTADO_INICIAL, META, OPERADORES)

print("Estado inicial:", ESTADO_INICIAL, "| Diferencia inicial respecto a la meta:", diferencia(ESTADO_INICIAL))
print(f"Meta: obtener exactamente 2 litros en alguna de las dos jarras (capacidades {CAP_A}L y {CAP_B}L)\n")

if solucion:
    print(f"✅ GPS encontró una solución en {len(solucion) - 1} pasos:\n")
    for i, estado in enumerate(solucion):
        etiqueta = "Estado inicial" if i == 0 else f"Paso {i}"
        print(f"{etiqueta}: Jarra A={estado[0]}L, Jarra B={estado[1]}L  (diferencia={diferencia(estado)})")
else:
    print("❌ El GPS no encontró una solución dentro del espacio de estados explorado.")
```

### 🤔 ¿Qué acaba de pasar?

El GPS exploró el espacio de estados aplicando, en cada paso, los operadores disponibles (`llenar`, `vaciar`, `verter`) hasta encontrar un estado donde la **diferencia con la meta** (`diferencia()`) llega a cero. El camino impreso es, literalmente, el **plan** que resuelve el problema: la secuencia mínima de operadores que transforma el estado inicial en un estado meta.

Nota que el algoritmo **nunca "supo" de antemano** cuál era la solución: la descubrió combinando sistemáticamente los operadores definidos, exactamente como proponían Newell y Simon en 1957.

## 7. Segundo Caso: Torres de Hanoi Reducida (3 discos) 🗼

Repliquemos la misma lógica de espacio de estados con otro clásico de la IA simbólica: la **Torre de Hanoi**. El objetivo es mover una torre de discos de un poste origen a un poste destino, usando un poste auxiliar, sin colocar nunca un disco más grande sobre uno más pequeño.

* **Estado:** la distribución de discos entre los tres postes.
* **Estado meta:** todos los discos apilados en el poste destino, en el orden correcto.
* **Operadores:** mover el disco superior de un poste a otro (siempre que sea válido).

Al igual que en el problema de las jarras, cada movimiento es un **operador que reduce la diferencia** entre el estado actual y la meta (medida, en este caso, como el número de discos que todavía no están en su posición final).

```python
def hanoi_gps(n, origen, destino, auxiliar, movimientos):
    """Resuelve la Torre de Hanoi aplicando, recursivamente, la misma idea de reducir
    la diferencia con la meta: mover primero lo que estorba (auxiliar), luego el disco
    objetivo (operador principal), y finalmente reconstruir la subtorre restante."""
    if n == 0:
        return
    hanoi_gps(n - 1, origen, auxiliar, destino, movimientos)
    movimientos.append((n, origen, destino))
    hanoi_gps(n - 1, auxiliar, destino, origen, movimientos)

N_DISCOS = 3
movimientos = []
hanoi_gps(N_DISCOS, "A", "C", "B", movimientos)

print(f"GPS aplicado a la Torre de Hanoi reducida ({N_DISCOS} discos)")
print(f"Meta: mover los {N_DISCOS} discos de la torre A a la torre C usando B como auxiliar\n")
for i, (disco, origen, destino) in enumerate(movimientos, start=1):
    print(f"Paso {i}: mover disco {disco} de torre {origen} -> torre {destino}")

print(f"\nTotal de movimientos (operadores aplicados): {len(movimientos)} — óptimo teórico: {2 ** N_DISCOS - 1}")
```

## 8. Reflexión: GPS vs. IA Moderna ⚖️

| Dimensión | GPS (IA Simbólica, 1957) | IA Moderna (Aprendizaje Automático / Deep Learning) |
|---|---|---|
| **Fuente del "conocimiento"** | Reglas y operadores definidos manualmente por un experto | Patrones aprendidos automáticamente a partir de datos |
| **Representación del problema** | Explícita: estados, operadores, precondiciones | Implícita: pesos numéricos aprendidos en una red |
| **Escalabilidad** | Limitada por la explosión combinatoria | Escala con más datos y cómputo, aunque con otros costos |
| **Explicabilidad** | Alta: cada paso del plan es trazable y verificable | Variable: los modelos de aprendizaje profundo suelen ser menos transparentes |
| **Adaptabilidad** | Nula sin reprogramar los operadores | Se reentrena o ajusta con nuevos datos |

Ambos enfoques **no son excluyentes**: la planificación simbólica moderna sigue usándose (por ejemplo, en robótica y logística), muchas veces **combinada** con componentes de aprendizaje automático que aprenden los operadores o heurísticas a partir de datos.

## 📌 Resumen / Puntos Clave

* El **GPS (General Problem Solver)** de Newell y Simon (1957) fue uno de los primeros programas diseñados para resolver problemas de forma **genérica**, no específica a un solo dominio.
* Todo problema se representa mediante un **estado inicial**, un **estado meta** y un conjunto de **operadores** con precondiciones y efectos.
* El **análisis medios-fines** reduce progresivamente la diferencia entre el estado actual y la meta, generando submetas cuando un operador no puede aplicarse de inmediato.
* El GPS fue un hito por separar la **representación del problema** de la **estrategia de resolución**, sentando las bases de la IA simbólica.
* Sus limitaciones (explosión combinatoria, ausencia de aprendizaje, dependencia de representación manual) motivaron, décadas después, el auge del Aprendizaje Automático.
