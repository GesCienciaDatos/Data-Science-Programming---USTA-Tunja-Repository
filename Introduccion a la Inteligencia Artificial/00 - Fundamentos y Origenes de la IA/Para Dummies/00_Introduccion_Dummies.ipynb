## 1. La Gran Analogía: el Loro Súper Entrenado 🦜

Imagina un loro que ha escuchado tantas conversaciones que ya sabe **qué decir en cada momento**, aunque no entienda ni una palabra de lo que dice.

* Si le dices "Buenos días", el loro responde "¡Buenos días, buenos días!"
* Si aplaudes, silba una canción.
* Pero si le preguntas "¿Cuánto es 2 más 2?", simplemente repite cualquier frase que ya conoce, porque no sabe sumar.

> 💡 **En resumen:** muchas cosas que llamamos "Inteligencia Artificial" se parecen más a este loro que a un genio pensante: siguen patrones aprendidos muy bien, pero eso no siempre significa que "entienden" lo que hacen. Justamente por eso, definir qué es "inteligencia" en una máquina es tan complicado.

## 2. Los 4 Sabores de "Ser Inteligente" para una Máquina 🍦

Dos científicos muy famosos, Russell y Norvig, notaron que cuando la gente dice "inteligencia artificial", en realidad se refiere a 4 cosas distintas. Aquí te las traducimos al español de la calle:

| Sabor | En palabras sencillas | Ejemplo cotidiano |
|---|---|---|
| **Actuar como Humano** | Que la máquina "actúe" tan parecido a un humano que no notes la diferencia. | Un chatbot que conversa tan bien que crees que es tu amigo. |
| **Pensar como Humano** | Que la máquina "piense" copiando cómo funciona el cerebro humano por dentro. | Un programa que imita cómo memorizas una lista del mercado. |
| **Pensar de Forma Perfecta (Racional)** | Que la máquina razone siguiendo reglas de lógica sin ningún error, como Spock de Star Trek. | Un programa que demuestra "si A entonces B" sin equivocarse jamás. |
| **Actuar de Forma Inteligente (Racional)** | Que la máquina simplemente tome la mejor decisión posible para lograr su meta, sin importar si "piensa" como humano. | El GPS de tu carro, que calcula la ruta más rápida a tu destino. |

> 🎯 Hoy en día, la mayoría de la IA que usamos (asistentes de voz, recomendadores de Netflix, mapas) sigue el último sabor: **actuar de forma inteligente para cumplir un objetivo**.

---
## 3. Pongamos las Manos en la Masa 🧪

Vamos a crear un pequeño programa que, al leer la descripción de una máquina, adivina a cuál de los 4 "sabores" se parece más. Ejecuta la celda con `Shift + Enter`:

````python
def adivinar_sabor_de_ia(descripcion):
    """Adivina el 'sabor' de IA según algunas palabras clave del texto."""
    descripcion = descripcion.lower()

    if "engañar" in descripcion or "parece humano" in descripcion:
        return "🗣️ Actuar como Humano"
    elif "cerebro" in descripcion or "piensa como" in descripcion:
        return "🧠 Pensar como Humano"
    elif "lógica" in descripcion or "nunca se equivoca" in descripcion:
        return "📐 Pensar Perfecto (Racional)"
    elif "mejor decisión" in descripcion or "cumple su meta" in descripcion:
        return "🎯 Actuar Inteligente (Racional)"
    else:
        return "🤷 No estoy segura, cuéntame más detalles"


ejemplos = [
    "Un chatbot que conversa tan bien que parece humano.",
    "Un programa que intenta funcionar como el cerebro de una persona.",
    "Un robot que calcula la mejor decisión para llegar rápido a su meta.",
]

for ejemplo in ejemplos:
    print(f"📋 {ejemplo}")
    print(f"   -> {adivinar_sabor_de_ia(ejemplo)}\n")
````

### 🤔 ¿Qué acaba de pasar?
Le diste al programa una frase, y él buscó "pistas" (palabras clave) para adivinar de qué sabor de IA le estás hablando. Así de simple: sin magia, solo comparación de palabras.
