## 1. La Fiesta de Fundación: Verano de 1956 🎉

Imagina un grupo de amigos muy inteligentes que se reúnen un verano completo en una universidad, convencidos de que en pocos años construirían una máquina tan lista como un humano. > 💡 **En resumen:** 1956 es como el cumpleaños oficial de la Inteligencia Artificial. Antes de eso ya había ideas sueltas, pero ahí nació el nombre y el grupo que le daría forma a la disciplina.

````python
# Los invitados más famosos de la "fiesta de fundación" de la IA (1956)

invitados_famosos = {
    "John McCarthy": "Puso el nombre 'Inteligencia Artificial' e inventó el lenguaje LISP.",
    "Marvin Minsky": "Uno de los primeros en experimentar con redes neuronales.",
    "Claude Shannon": "El papá de la Teoría de la Información.",
}

print("🎉 Invitados estrella de la fiesta de Dartmouth (1956):\n")
for nombre, aporte in invitados_famosos.items():
    print(f"- {nombre}: {aporte}")
````

## 2. Cuando se Apaga la Fiesta: los "Inviernos de la IA" ❄️🥶

Después de tanta emoción, la gente prometió demasiado ("¡en 10 años las máquinas traducirán idiomas perfectamente!") y no cumplió a tiempo. Cuando eso pasa, quienes ponen el dinero (gobiernos y empresas) se enojan y **cierran la billetera**. A esto se le llama un "invierno de la IA": de repente, nadie quiere financiar más investigación.

Esto pasó **dos veces**:
* **Primer invierno (1966-1974):** un informe muy crítico bajó el ánimo y el dinero se acabó.
* **Segundo invierno (1987-1993):** después de un segundo boom (que veremos abajo), volvió a apagarse la fiesta.

> 🎢 Así es la montaña rusa: sube la emoción, sube el dinero... y de repente, todo baja de golpe.

## 3. Los Libros de Recetas Gigantes: Sistemas Expertos (Años 80) 📖🧑‍⚕️

En los años 80 se pusieron de moda los **sistemas expertos**: programas que funcionan como un **libro de recetas gigante** escrito por un experto humano. Cada "receta" es una regla del tipo: *"SI el paciente tiene fiebre y tos, ENTONCES sospecha de gripe"*.

El problema: si aparecía un caso nuevo que no estaba en el libro, el programa simplemente **no sabía qué hacer**, porque no aprendía por sí solo — solo seguía las recetas ya escritas.

````python
# Un "libro de recetas" de juguete para recomendar qué ropa usar según el clima.
# Así funcionaban (de forma muy simplificada) los sistemas expertos de los años 80.

def libro_de_recetas_clima(sintomas):
    if sintomas.get("lluvia"):
        return "🌧️ Receta: lleva paraguas y chaqueta impermeable."
    elif sintomas.get("mucho_sol") and sintomas.get("calor"):
        return "☀️ Receta: usa gorra, gafas de sol y ropa ligera."
    elif sintomas.get("frio"):
        return "🧥 Receta: abrígate con chaqueta y bufanda."
    else:
        return "🤷 No hay una receta clara para este clima. Pregúntale a un humano."


casos = [
    {"lluvia": True},
    {"mucho_sol": True, "calor": True},
    {"nublado": True},  # este caso no está en el libro de recetas
]

for caso in casos:
    print(f"Clima: {caso}")
    print(f"  {libro_de_recetas_clima(caso)}\n")
````

### 🤔 ¿Qué pasó con el último caso?
Cuando el clima no encajaba en ninguna "receta" escrita de antemano, el programa se quedó sin respuesta útil. Justo esa era la debilidad de los sistemas expertos: no podían improvisar fuera de su libro de recetas.

## 4. La IA Aprende a Ver: 2012 👀

Durante muchos años, hacer que una computadora reconociera una foto de un gato fue sorprendentemente difícil. En 2012, gracias a tres cosas que se juntaron —**muchísimas fotos etiquetadas**, **computadoras más rápidas (GPUs)** y **mejores técnicas**— un programa llamado **AlexNet** dejó a todos con la boca abierta al reconocer imágenes muchísimo mejor que cualquier intento anterior. Ese momento reinició por completo el interés (y el dinero) en la IA.

## 5. Los Loros Gigantes: la Era de los LLMs (2020s) 🦜📚

Recuerda a nuestro loro súper entrenado del primer cuaderno. Ahora imagina un loro que ha "escuchado" (leído) una porción enorme de todo internet: libros, artículos, conversaciones. Eso es, en esencia, un **modelo de lenguaje grande (LLM)**: un programa entrenado con cantidades gigantescas de texto para predecir qué palabra sigue después de otra, y que resulta sorprendentemente útil para conversar, escribir y programar.

A finales de 2022, uno de estos "loros gigantes" (ChatGPT) se hizo público y millones de personas empezaron a usarlo de la noche a la mañana.
