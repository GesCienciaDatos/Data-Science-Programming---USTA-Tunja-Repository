## 1. El Juego de "Adivina Quién es la Máquina" 🕵️

Imagina el juego de mesa "¿Quién es quién?", pero por chat de texto. Hay tres participantes en cuartos separados:

* **Tú (el interrogador):** escribes preguntas y lees las respuestas, sin ver ni escuchar a nadie.
* **Jugador A:** una persona real.
* **Jugador B:** una máquina.

Tu única misión es, después de un rato de "chatear" con ambos, adivinar **cuál de los dos es la máquina**. Si te equivocas tan seguido como cuando los dos son personas reales, ¡la máquina "ganó" el juego!

> 💡 **En resumen:** en 1950, Alan Turing propuso que en vez de discutir filosóficamente "¿pueden pensar las máquinas?", mejor jugáramos este juego y viéramos qué tan bien "engañan" a un humano.

## 2. Construyamos Nuestra Propia Máquina de Juguete 🧪

Vamos a programar un chatbot muy simple que solo reconoce algunas palabras clave y responde con frases ya preparadas — sin entender absolutamente nada de lo que le dices. Así funcionaban los primeros chatbots de la historia (como "Eliza", de 1966).

````python
import random

class BotDeJuguete:
    """Un chatbot que solo 'reconoce' palabras clave y responde con frases
    ya preparadas. No entiende nada, solo compara texto."""

    def __init__(self):
        self.reglas = [
            (["hola", "buenas"], ["¡Hola! ¿Cómo estás hoy?"]),
            (["triste"], ["Vaya, lo siento. ¿Qué pasó?"]),
            (["feliz", "bien"], ["¡Qué bueno! Cuéntame más."]),
            (["nombre"], ["Prefiero no hablar de mi nombre, hablemos de ti."]),
        ]
        self.respuestas_genericas = ["Cuéntame más...", "Interesante, ¿por qué?"]

    def responder(self, mensaje):
        mensaje = mensaje.lower()
        for palabras, respuestas in self.reglas:
            if any(p in mensaje for p in palabras):
                return random.choice(respuestas)
        return random.choice(self.respuestas_genericas)


random.seed(7)
bot = BotDeJuguete()

preguntas = [
    "Hola, ¿cómo estás?",
    "Hoy me siento un poco triste.",
    "¿Cuál es tu nombre?",
]

print("🕵️ Simulación del juego de adivinar quién es la máquina:\n")
for pregunta in preguntas:
    print(f"Tú: {pregunta}")
    print(f"Jugador B: {bot.responder(pregunta)}\n")
````

### 🤔 ¿Lo engañarías fácilmente?
En una charla corta y casual, este bot podría pasar sin levantar sospechas. Pero probemos con una pregunta más difícil, fuera de sus reglas...

````python
pregunta_dificil = "¿Cuánto es 347 por 289?"
print(f"Tú: {pregunta_dificil}")
print(f"Jugador B: {bot.responder(pregunta_dificil)}")
print("\n⚠️ El bot no calculó nada: solo devolvió una frase genérica porque")
print("   no encontró ninguna palabra clave conocida. ¡Ahí se nota el truco!")
````

## 3. La Crítica: el Señor Encerrado con un Diccionario de Chino 🈳🚪

Un filósofo llamado **John Searle** dijo en 1980: "¡Un momento! Aunque una máquina logre engañarte, eso no prueba que *entienda* nada."

Su ejemplo: imagina a un señor que **no habla chino**, encerrado en un cuarto. Le pasan papelitos con símbolos chinos por debajo de la puerta. Él tiene un **manual gigante en español** que le dice: "si ves este símbolo, dibuja este otro símbolo como respuesta". Siguiendo el manual al pie de la letra, logra responder perfectamente en chino... **sin entender ni una palabra**.

> 💡 Para alguien afuera, parece que el cuarto "sabe chino". Pero adentro solo hay un señor copiando símbolos de un manual. Searle dice que muchos chatbots son exactamente así: **muy buenos copiando patrones, sin entender nada de verdad**.

### ⚖️ Comparando las dos ideas

| Idea | ¿Qué dice? | ¿Cómo se relaciona con nuestro bot? |
|---|---|---|
| **Juego de Turing** | Si logras engañar a un humano en el chat, ya cuenta como "inteligente". | Nuestro bot podría engañar en preguntas simples y cortas. |
| **Cuarto Chino de Searle** | Engañar no es lo mismo que entender. | Nuestro bot es literalmente eso: un manual de reglas sin comprensión real. |

## 4. ¿Por Qué Sigue Importando Esto Hoy? 🔍

Hoy en día existen chatbots mucho más avanzados que pueden mantener conversaciones larguísimas sin que notes que son una máquina. Pero los expertos recuerdan que:

* Engañar a alguien en un chat **no prueba** que la máquina "piensa" de verdad.
* El resultado depende mucho de qué tan buenas sean las preguntas de quien interroga.
* Hoy existen pruebas más específicas (como problemas de sentido común) para medir mejor las capacidades reales de una IA, en lugar de solo "si parece humana".
