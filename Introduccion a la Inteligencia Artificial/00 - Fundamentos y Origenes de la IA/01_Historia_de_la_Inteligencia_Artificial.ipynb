## 1.4 Historia de la Inteligencia Artificial 📖

### El Nacimiento Oficial: la Conferencia de Dartmouth (1956)

En el verano de 1956, un grupo de investigadores se reunió en el Dartmouth College (EE. UU.) para un taller de dos meses. Fue **John McCarthy** quien acuñó el término *"Artificial Intelligence"* en la propuesta de esta conferencia, con una afirmación de un optimismo notable:

> *"Todo aspecto del aprendizaje o cualquier otra característica de la inteligencia puede, en principio, describirse de forma tan precisa que se puede construir una máquina que la simule."*

Este taller no produjo avances técnicos inmediatos, pero sí fundó un **campo de investigación formal**, con un nombre, una comunidad y una agenda de preguntas compartidas.

````python
# Estructura de datos simple con los asistentes clave de la Conferencia de Dartmouth (1956)
# y su aporte posterior más reconocido dentro de la IA.

fundadores_dartmouth = [
    {"nombre": "John McCarthy", "aporte": "Acuñó el término 'Inteligencia Artificial' y organizó el taller; luego creó el lenguaje LISP."},
    {"nombre": "Marvin Minsky", "aporte": "Pionero en redes neuronales tempranas y razonamiento simbólico; cofundó el laboratorio de IA del MIT."},
    {"nombre": "Claude Shannon", "aporte": "Padre de la Teoría de la Información; aportó rigor matemático a la comunicación y el cómputo."},
    {"nombre": "Nathaniel Rochester", "aporte": "Diseñador de la IBM 701; exploró de forma temprana las redes neuronales artificiales."},
]

print("🏛️ Taller de Dartmouth (verano de 1956) — participantes clave:\n")
for f in fundadores_dartmouth:
    print(f"- {f['nombre']}: {f['aporte']}")
````

### Los "Inviernos de la IA" ❄️

Tras el optimismo inicial (se predijo que en 10 años una máquina sería campeona mundial de ajedrez o traduciría idiomas perfectamente), la disciplina sufrió dos grandes crisis de financiación y credibilidad conocidas como **"inviernos de la IA"**:

* **Primer invierno (1974–1980):** provocado en parte por el *Informe Lighthill* (1973) en el Reino Unido, que criticó duramente la falta de resultados prácticos de la IA simbólica frente a las promesas hechas. Los gobiernos recortaron drásticamente la financiación.
* **Segundo invierno (1987–1993):** tras el auge comercial de los sistemas expertos (que veremos a continuación), el colapso del mercado de las "máquinas LISP" especializadas y el desencanto con el mantenimiento costoso de estos sistemas provocaron una segunda ola de escepticismo y recortes.

> 💡 **Lección histórica:** los inviernos de la IA no ocurrieron por falta de ideas, sino por una **brecha entre las expectativas prometidas y los resultados entregables** con el cómputo y los datos disponibles en cada época.

### Sistemas Expertos de los años 80 🧑‍⚕️📋

Durante el auge de los años 80, la IA comercial se centró en los **sistemas expertos**: programas que codifican el conocimiento de un especialista humano en una **base de conocimiento** de reglas del tipo *"SI (condición) ENTONCES (conclusión)"*, junto con un **motor de inferencia** que las aplica sobre un caso concreto.

El ejemplo más citado es **MYCIN** (Universidad de Stanford), un sistema experto para diagnosticar infecciones bacterianas y recomendar antibióticos, que en pruebas controladas llegó a igualar o superar a médicos junior. Otro caso de éxito comercial fue **XCON**, usado por Digital Equipment Corporation para configurar automáticamente pedidos de computadoras.

Su limitación principal: estos sistemas **no aprendían** de nuevos casos ni generalizaban bien fuera del dominio exacto para el que fueron programados a mano, y mantener miles de reglas escritas por humanos se volvió costoso e inmanejable.

````python
# Un sistema experto de juguete (estilo años 80) para diagnosticar el estado
# de una planta de jardín, con una base de conocimiento de reglas SI-ENTONCES
# y un motor de inferencia muy simple (la primera regla que aplica, gana).

def sistema_experto_jardin(sintomas: dict) -> str:
    base_reglas = [
        (lambda s: s.get("hojas_amarillas") and s.get("tierra_humeda"),
         "Diagnóstico: exceso de riego. Recomendación: reduce la frecuencia de riego."),
        (lambda s: s.get("hojas_marchitas") and not s.get("tierra_humeda"),
         "Diagnóstico: falta de riego. Recomendación: riega la planta pronto."),
        (lambda s: s.get("manchas_blancas"),
         "Diagnóstico: posible mildiu polvoso. Recomendación: aplica fungicida."),
        (lambda s: s.get("hojas_amarillas") and not s.get("tierra_humeda"),
         "Diagnóstico: posible deficiencia de nutrientes. Recomendación: revisa el abono."),
    ]
    for condicion, diagnostico in base_reglas:
        if condicion(sintomas):
            return diagnostico
    return "Sin diagnóstico concluyente: se requiere inspección de un experto humano."


casos_de_prueba = [
    {"hojas_amarillas": True, "tierra_humeda": True},
    {"hojas_marchitas": True, "tierra_humeda": False},
    {"manchas_blancas": True, "tierra_humeda": True},
]

for i, caso in enumerate(casos_de_prueba, start=1):
    print(f"Caso {i}: {caso}")
    print(f"  {sistema_experto_jardin(caso)}\n")
````

### El Resurgimiento: Deep Learning y AlexNet (2012) 🚀

Tras décadas de progreso lento en redes neuronales (una idea que existía desde los años 50), tres factores convergieron alrededor de 2012 para producir un resurgimiento masivo:

1. **Big Data:** conjuntos de datos etiquetados a gran escala, como *ImageNet* (más de un millón de imágenes clasificadas en miles de categorías).
2. **Cómputo paralelo (GPUs):** las tarjetas gráficas, diseñadas originalmente para videojuegos, resultaron ideales para entrenar redes neuronales profundas mucho más rápido.
3. **Mejoras algorítmicas:** funciones de activación, regularización (*dropout*) y arquitecturas más profundas.

En 2012, la red **AlexNet** (Krizhevsky, Sutskever y Hinton) redujo drásticamente la tasa de error en el reto anual *ImageNet Large Scale Visual Recognition Challenge*, superando por un margen enorme a los métodos clásicos de visión artificial. Este momento se considera el detonante del auge moderno del **Deep Learning**.

---
### La Era de los LLMs (años 2020) 🤖💬

La arquitectura **Transformer**, presentada en 2017 en el influyente artículo *"Attention Is All You Need"*, permitió entrenar modelos de lenguaje sobre cantidades masivas de texto de forma mucho más eficiente que las arquitecturas anteriores.

Esto dio lugar a los **Modelos de Lenguaje Grandes (LLMs)**: sistemas entrenados con enormes volúmenes de texto de internet, capaces de generar lenguaje natural coherente, responder preguntas, programar y razonar sobre una variedad sorprendente de tareas. El lanzamiento público de **ChatGPT** a finales de 2022 marcó la entrada masiva de la IA generativa a la vida cotidiana, un fenómeno que se sigue expandiendo hacia modelos multimodales (texto, imagen, audio, video) en la actualidad.

````python
# Representamos los grandes hitos de la historia de la IA como una lista de diccionarios
# y los clasificamos según si representaron un "auge" o un "invierno" para la disciplina.

hitos_historia_ia = [
    {"anio": "1956", "hito": "Conferencia de Dartmouth", "tipo": "auge"},
    {"anio": "1966-1974", "hito": "Primer Invierno de la IA", "tipo": "invierno"},
    {"anio": "1980-1987", "hito": "Auge de los Sistemas Expertos", "tipo": "auge"},
    {"anio": "1987-1993", "hito": "Segundo Invierno de la IA", "tipo": "invierno"},
    {"anio": "1997", "hito": "Deep Blue vence a Kasparov", "tipo": "auge"},
    {"anio": "2012", "hito": "AlexNet gana ImageNet (Deep Learning)", "tipo": "auge"},
    {"anio": "2017", "hito": "Arquitectura Transformer", "tipo": "auge"},
    {"anio": "2020-2026", "hito": "Era de los LLMs", "tipo": "auge"},
]

print("📜 Línea de tiempo de la Inteligencia Artificial:\n")
for h in hitos_historia_ia:
    icono = "📈" if h["tipo"] == "auge" else "📉"
    print(f"{icono} {h['anio']:<12} {h['hito']}")

total_auges = sum(1 for h in hitos_historia_ia if h["tipo"] == "auge")
total_inviernos = sum(1 for h in hitos_historia_ia if h["tipo"] == "invierno")
print(f"\nResumen: {total_auges} periodos de auge y {total_inviernos} inviernos registrados en esta línea de tiempo.")
````
