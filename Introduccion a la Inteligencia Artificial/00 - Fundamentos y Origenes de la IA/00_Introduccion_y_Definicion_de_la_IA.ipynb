## 1.1 Introducción a la IA como disciplina 📚

La **Inteligencia Artificial (IA)** es el campo de la informática dedicado al diseño de sistemas capaces de realizar tareas que, cuando las ejecuta un ser humano, requieren de lo que llamamos "inteligencia": percibir, razonar, aprender, planificar, comunicarse y actuar sobre un entorno.

A diferencia de otras ramas de la computación centradas en un tipo de problema concreto (bases de datos, redes, gráficos), la IA es **profundamente interdisciplinaria**: toma prestado de la filosofía sus preguntas sobre la mente, de las matemáticas su formalismo lógico y probabilístico, de la neurociencia su inspiración biológica, de la lingüística su interés por el lenguaje, de la economía su teoría de la decisión racional y de la ingeniería su vocación de construir sistemas que funcionen.

> 💡 **Idea clave:** la IA no es un algoritmo ni una librería de Python. Es un **programa de investigación** de más de 70 años que busca responder una pregunta filosófica con herramientas de ingeniería: *¿se puede construir una máquina que actúe con inteligencia?*

````python
# Un pequeño mapa de las disciplinas que alimentan a la Inteligencia Artificial.
# No es una librería externa: es solo una estructura de datos para razonar sobre el tema.

disciplinas_fundadoras = {
    "Filosofía": "Aporta las preguntas sobre la mente, el conocimiento y el razonamiento válido.",
    "Matemáticas": "Aporta la lógica formal, la probabilidad y la teoría de la computación.",
    "Economía": "Aporta la teoría de la decisión racional bajo incertidumbre.",
    "Neurociencia": "Aporta el modelo biológico de cómo procesa información el cerebro.",
    "Psicología": "Aporta teorías sobre cómo perciben, aprenden y razonan los seres humanos.",
    "Lingüística": "Aporta modelos formales del lenguaje y la comunicación.",
    "Ingeniería de Control": "Aporta la idea de sistemas que se autorregulan para alcanzar un objetivo.",
}

print("🧩 Disciplinas que convergen en la Inteligencia Artificial:\n")
for disciplina, aporte in disciplinas_fundadoras.items():
    print(f"- {disciplina}: {aporte}")

print(f"\nTotal de disciplinas fundadoras consideradas: {len(disciplinas_fundadoras)}")
````

## 1.3 Definición de la IA: ¿por qué no hay una sola definición? 🧭

Si le preguntas a diez expertos "¿qué es la Inteligencia Artificial?", es probable que obtengas diez respuestas distintas — y las diez pueden ser correctas desde su propio marco. En su libro de referencia *Artificial Intelligence: A Modern Approach*, **Stuart Russell y Peter Norvig** organizan las definiciones históricas de la IA en una tabla de **dos ejes**:

* **Eje del proceso vs. el comportamiento:** ¿nos importa que el sistema *piense* de cierta forma, o solo que *actúe* de cierta forma?
* **Eje humano vs. racional:** ¿el estándar de comparación es imitar al ser humano (con sus errores incluidos), o alcanzar el ideal matemático de racionalidad?

De cruzar estos dos ejes surgen los **cuatro enfoques clásicos** para definir la IA.

### 📊 El cuadro de los cuatro enfoques (Russell & Norvig)

| Enfoque | Estándar de comparación | Pregunta que responde | Ejemplo histórico representativo |
|---|---|---|---|
| **Pensar Humanamente** | El proceso cognitivo humano | ¿El sistema razona *como* lo haría una persona? | Ciencia cognitiva, modelos de arquitectura mental (ej. SOAR). |
| **Pensar Racionalmente** | Las "leyes del pensamiento" (lógica formal) | ¿El sistema razona con *validez lógica* perfecta? | Sistemas basados en lógica y demostradores de teoremas. |
| **Actuar Humanamente** | El comportamiento observable de un humano | ¿El sistema *se comporta* de forma indistinguible de una persona? | El Test de Turing (lo veremos en el Cuaderno 02). |
| **Actuar Racionalmente** | El agente que toma la mejor decisión posible | ¿El sistema *actúa* para maximizar el logro de sus objetivos? | Agentes inteligentes modernos (el enfoque dominante hoy). |

> 🎯 El enfoque **"Actuar Racionalmente"** es el que domina la IA moderna, porque es medible, no exige replicar los errores humanos y se puede formalizar matemáticamente como un **agente** que percibe su entorno y actúa para maximizar una medida de desempeño.

## 🔬 Laboratorio: clasificando sistemas de IA según su enfoque

Vamos a construir una función muy simple que, dada la **descripción** de un sistema de IA, intenta identificar a cuál de los cuatro enfoques se acerca más, basándose en palabras clave presentes en la descripción. Es una simplificación didáctica (en la práctica un mismo sistema puede combinar varios enfoques), pero ilustra cómo los cuatro marcos conceptuales se traducen en objetivos de diseño distintos.

````python
def clasificar_enfoque(descripcion: str) -> str:
    """Clasifica de forma heurística la descripción de un sistema de IA
    según los 4 enfoques de definición de Russell & Norvig (1995).
    No usa ninguna librería externa: solo coincidencia de palabras clave.
    """
    descripcion = descripcion.lower()

    reglas = [
        (["engañar", "indistinguible", "conversa como", "parece humano"],
         "Actuar Humanamente (Test de Turing)"),
        (["modela la mente", "cognitiv", "psicológic", "como piensa el cerebro"],
         "Pensar Humanamente (Ciencia Cognitiva)"),
        (["lógica formal", "silogismo", "demuestra teoremas", "inferencia deductiva"],
         "Pensar Racionalmente (Leyes del Pensamiento)"),
        (["maximiza", "optimiza", "mejor decisión", "agente racional", "objetivo"],
         "Actuar Racionalmente (Agente Racional)"),
    ]

    for palabras_clave, enfoque in reglas:
        if any(palabra in descripcion for palabra in palabras_clave):
            return enfoque
    return "Enfoque no determinado (la descripción es ambigua)"


sistemas_ejemplo = [
    "Un chatbot diseñado para conversar de forma tan natural que un evaluador humano no note que habla con una máquina.",
    "Un programa que intenta modelar cómo el cerebro humano resuelve problemas de memoria a corto plazo.",
    "Un motor de inferencia que demuestra teoremas matemáticos aplicando reglas de lógica formal paso a paso.",
    "Un agente de conducción autónoma que optimiza continuamente sus acciones para tomar la mejor decisión y maximizar la seguridad.",
]

for sistema in sistemas_ejemplo:
    print(f"Sistema: {sistema}")
    print(f"  -> Enfoque dominante: {clasificar_enfoque(sistema)}\n")
````

### 🤔 ¿Qué nos muestra este ejercicio?

Cada sistema fue diseñado con un **criterio de éxito distinto**: el primero se evalúa por si "engaña" a un humano, el segundo por su fidelidad al proceso cognitivo, el tercero por su corrección lógica, y el cuarto por qué tan bien logra su objetivo. Ninguno de los cuatro es "más correcto" que los demás — simplemente responden a **preguntas distintas** sobre qué significa que una máquina sea inteligente.
