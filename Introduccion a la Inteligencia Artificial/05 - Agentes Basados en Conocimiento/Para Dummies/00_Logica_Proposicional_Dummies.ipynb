---
## 1. La Gran Analogía: El Detective de Bolsillo 🕵️

Imagina un detective de bolsillo (nuestro "agente") que solo sabe dos cosas:

* **Regla:** "Si llueve, entonces el señor lleva paraguas."
* **Pista:** "Hoy llueve."

Sin que nadie le diga nada más, el detective puede concluir con toda seguridad: **"¡El señor lleva paraguas!"**. Eso no es magia, es **lógica**: combinar lo que sabe (reglas + pistas) para deducir cosas nuevas.

En este cuaderno, ese detective de bolsillo se llama `logic.py`, un archivito de Python que programamos nosotros mismos, sin usar ninguna librería mágica de terceros. 🧸


---
## 2. Las Piezas del Rompecabezas 🧩

Para que la computadora entienda nuestras reglas, necesitamos traducirlas a "bloques" muy simples:

| Bloque | Lo que significa | Ejemplo cotidiano |
|---|---|---|
| **Símbolo** | Un hecho que puede ser cierto o falso | "Llueve", "Paraguas" |
| **NOT (¬)** | "No es cierto que..." | "No llueve" |
| **AND (∧)** | "...y también..." | "Llueve Y hace frío" |
| **OR (∨)** | "...o bien..." | "Llueve O hace sol" |
| **IF-THEN (→)** | "Si... entonces..." | "Si llueve, entonces llevo paraguas" |

Cada uno de estos bloques es una "piecita de Lego" en Python: se combinan entre sí para construir frases cada vez más complejas.


```python
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))  # logic.py vive un nivel arriba (Para Dummies/ -> módulo)

from logic import Symbol, Not, And, Or, Implication

# Creamos nuestras "piecitas": cada Symbol es un hecho simple.
Llueve = Symbol("Llueve")
Paraguas = Symbol("Paraguas")

# Las combinamos como si fueran piezas de Lego para armar una regla.
regla = Implication(Llueve, Paraguas)

print("Nuestra regla en palabras normales:", "Si llueve, entonces llevo paraguas.")
print("Nuestra regla traducida a lógica:  ", regla.formula())
```

---
## 3. ¿Qué es un "mundo posible"? 🌍

Antes de que la computadora saque conclusiones, imagina que ella dibuja **todos los universos alternativos posibles**. Con solo 2 hechos ("Llueve" y "Paraguas"), hay exactamente **4 universos** posibles:

| Universo | ¿Llueve? | ¿Lleva paraguas? |
|---|---|---|
| 1 | No | No |
| 2 | No | Sí |
| 3 | Sí | No |
| 4 | Sí | Sí |

La computadora revisa **los 4** uno por uno (¡sin saltarse ninguno!) para asegurarse de que su conclusión sea válida en TODOS los universos donde lo que sabe es verdad. Eso, con nombre elegante, se llama **verificación de modelos**, pero tú puedes pensarlo simplemente como *"revisar todas las posibilidades sin trampa"*.


```python
from logic import enumerate_models

# Le pedimos a Python que nos "dibuje" los 4 universos posibles.
for universo in enumerate_models(regla.symbols()):
    resultado = regla.evaluate(universo)
    print(f"Universo {universo}  ->  ¿se cumple la regla? {resultado}")
```

Fíjate que la regla es falsa en un **único** universo: aquel donde llueve pero NO se lleva paraguas (justo el que rompe la promesa). En todos los demás, la regla se sostiene sin problema.

---
## 4. El truco final: sumar hechos y preguntar 🔮

Ahora viene la parte divertida. Le vamos a **contar** (`TELL`) al detective dos cosas:
1. La regla: "si llueve, llevo paraguas".
2. El hecho: "hoy llueve".

Y luego le vamos a **preguntar** (`ASK`): *"¿entonces hoy llevo paraguas?"*


```python
from logic import model_check

# TELL: le contamos al detective todo lo que sabemos, juntándolo con AND.
lo_que_sabemos = And(
    Implication(Llueve, Paraguas),  # la regla
    Llueve                            # el hecho de hoy
)

print("Todo lo que sabe el detective:", lo_que_sabemos.formula())

# ASK: le preguntamos si puede deducir que llevamos paraguas.
respuesta = model_check(lo_que_sabemos, Paraguas)
print("\n¿El detective puede asegurar que hoy llevamos paraguas? ->", respuesta)
```

¡Exacto! El detective revisó **todos** los universos posibles por su cuenta y, en todos aquellos donde lo que sabe es verdad, resultó que también llevamos paraguas. Por eso puede responder `True` con total seguridad, sin que nadie se lo haya dicho directamente. 🎉
