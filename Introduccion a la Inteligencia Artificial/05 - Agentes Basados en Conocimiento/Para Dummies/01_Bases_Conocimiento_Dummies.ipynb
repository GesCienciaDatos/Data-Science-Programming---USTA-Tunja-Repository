## 1. La Receta de 3 Pasos 🍳

Armar cualquier historia lógica en Python siempre sigue la misma receta, sin importar qué tan complicada sea:

1. **Nombra los hechos** (los "personajes" de la historia): `Symbol("Nombre")`.
2. **Escribe las reglas** que los conectan, usando las piezas de Lego (`And`, `Or`, `Not`, `Implication`).
3. **Junta todo** en una sola caja con `And(*todo)` y pregúntale a la caja con `model_check`.

Con esta receta, vamos a resolver dos historias que ya dejamos listas en archivos de Python: `medico_o_clase.py` y `programando.py`.


---
## 2. Historia 1: ¿La doctora está de turno o en clase? 🩺📚

Camila estudia la Especialización y también es médica. Un mismo día, **solo puede pasar una cosa**: o está de turno en el hospital, o está en su clase de anatomía — nunca las dos al tiempo. Y resulta que hoy es miércoles, día en que **siempre** tiene esa clase.

¿Puede nuestro detective de Python deducir, sin que se lo digamos directamente, que hoy Camila **no** está de turno? Vamos a preguntarle:


````python
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))  # logic.py vive un nivel arriba (Para Dummies/ -> módulo)

import medico_o_clase as historia1
from logic import Not, model_check

print("Lo que sabe el detective:", historia1.knowledge.formula())

print("\n¿Podemos asegurar que Camila NO está de turno hoy?",
      model_check(historia1.knowledge, Not(historia1.MD)))

print("¿Podemos asegurar que Camila tiene clase de anatomía hoy?",
      model_check(historia1.knowledge, historia1.CL))
````

¡El detective lo dedujo perfecto! Como sabía que Camila SÍ tiene clase hoy, y también sabía que "de turno" y "en clase" nunca pasan juntas, la única posibilidad que queda en pie es que **no** está de turno. Eso es **razonar por descarte** — como cuando eliminas opciones en un examen de selección múltiple hasta quedarte con la única que tiene sentido.

## 3. Historia 2: ¿Llueve, entonces la clase es virtual? 🌧️💻

Esta historia tiene tres reglas encadenadas, como fichas de dominó:

1. Hay salida al campo **solo si** no llueve.
2. Si **no** hay salida al campo, la clase se hace virtual.
3. Hoy... ¡está lloviendo!

Vamos a ver si el detective logra "tumbar" todas las fichas hasta llegar a la conclusión final.


````python
import programando as historia2

print("Lo que sabe el detective:", historia2.knowledge.formula())

print("\n¿Hay salida al campo hoy?", model_check(historia2.knowledge, historia2.CLASE_CAMPO))
print("¿La clase se dicta de forma virtual?", model_check(historia2.knowledge, historia2.SESION_VIRTUAL))
````

```mermaid
flowchart LR
    A["Ficha 1:\n'Hoy llueve'"] --> B["Ficha 2:\n'Entonces NO hay salida al campo'"]
    B --> C["Ficha 3:\n'Entonces la clase es virtual'"]
```

¡Tres fichas de dominó cayendo en cadena! Ninguna regla por sí sola nos decía directamente "la clase es virtual", pero **combinándolas** el detective lo dedujo sin ayuda. Así funciona un agente basado en conocimiento de verdad: encadena piezas pequeñas para llegar a conclusiones grandes.

## 4. Un detective que aprende cosas nuevas sobre la marcha 🔎

Por último, hagamos un mini-detective al que le podamos ir **contando cosas nuevas poco a poco** (como si fuera aprendiendo en tiempo real) y preguntarle después de cada dato nuevo.


````python
from logic import Symbol, And, Implication

class Detective:
    def __init__(self):
        self.cosas_que_sabe = []

    def cuentame(self, algo_nuevo):
        self.cosas_que_sabe.append(algo_nuevo)

    def pregunta(self, algo):
        todo = And(*self.cosas_que_sabe) if len(self.cosas_que_sabe) > 1 else self.cosas_que_sabe[0]
        return model_check(todo, algo)


Estudio = Symbol("Estudio")
Practico = Symbol("Practico")
Apruebo = Symbol("Apruebo")

d = Detective()
d.cuentame(Implication(And(Estudio, Practico), Apruebo))

print("Antes de saber nada más -> ¿Aprobará?", "no se puede saber aún (no hay suficientes datos)")

d.cuentame(Estudio)
print("Después de saber que SÍ estudió -> ¿Podemos afirmar que aprobará?", d.pregunta(Apruebo))

d.cuentame(Practico)
print("Después de saber que TAMBIÉN practicó -> ¿Podemos afirmar que aprobará?", d.pregunta(Apruebo))
````
