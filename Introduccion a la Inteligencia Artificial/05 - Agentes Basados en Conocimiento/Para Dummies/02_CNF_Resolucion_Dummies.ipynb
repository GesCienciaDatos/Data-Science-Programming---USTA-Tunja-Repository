## ¿Qué vamos a aprender aquí? 🎈

Revisar **todos** los universos posibles funciona muy bien... hasta que hay demasiados hechos y el detective se demora una eternidad. En este cuaderno aprenderás un **atajo profesional** llamado *resolución*, que llega a la misma conclusión mucho más rápido, y lo pondremos a prueba resolviendo dos acertijos completos. ¡Vamos!

## 1. El Problema de Revisar TODO 🐌

Recuerda: si tenemos $n$ hechos, hay $2^n$ universos posibles que revisar. Con pocos hechos eso es rapidísimo... pero mira qué pasa si la historia crece:


```python
import sys, os, time
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))  # logic.py vive un nivel arriba (Para Dummies/ -> módulo)

from logic import Symbol, And, model_check

for cantidad_de_hechos in [5, 10, 15, 18]:
    hechos = [Symbol(f"Hecho{i}") for i in range(cantidad_de_hechos)]
    historia = And(*hechos)

    inicio = time.time()
    model_check(historia, hechos[0])
    duracion = time.time() - inicio

    print(f"Con {cantidad_de_hechos:>2} hechos hay que revisar {2**cantidad_de_hechos:>7,} universos  ({duracion:.3f} segundos)")
```

¿Ves cómo el tiempo se dispara? Si tuviéramos 50 hechos (algo nada raro en la vida real), ¡el detective tardaría más que la edad del universo revisando universo por universo! Por eso necesitamos un atajo más inteligente.

## 2. El Atajo: Traducir todo a "bloques SI-O" 🧱

El truco se llama **resolución**, y su primer paso es traducir cualquier historia a una forma muy ordenada donde todo queda escrito como *"esto O esto O esto"*, unido con "Y". A esa forma ordenada la llamamos, en la jerga técnica, **CNF**, pero tú puedes pensarla simplemente como *"la lista de pistas separadas, bien organizaditas"*.

`logic.py` ya sabe hacer esa traducción automáticamente con la función `to_cnf()`. No necesitas entender la matemática detrás, solo confiar en que la computadora lo hace bien (¡y ya lo comprobamos en el cuaderno estándar!).


```python
from logic import Symbol, Not, And, Or, Implication, to_cnf, to_clauses, formula_of_clause

A = Symbol("A")
B = Symbol("B")

historia = Implication(A, B)  # "Si A entonces B"
traducida = to_cnf(historia)

print("Historia original:", historia.formula())
print("Traducida (CNF)  :", traducida.formula())
```

## 3. El Truco de Tachar Pistas Contradictorias ✂️

Una vez que todo está en "bloques SI-O", el detective busca dos pistas que se contradigan directamente (una dice "sí, P" y otra dice "no, P") y las combina en una pista nueva, más corta. Repite esto una y otra vez. Si en algún momento logra tachar TODO y quedarse con una pista vacía (¡una contradicción total!), eso confirma que la conclusión que estaba buscando es verdadera.

Es como resolver un sudoku tachando posibilidades imposibles hasta que solo queda una respuesta. 🧩

## 4. Acertijo 1: ¿Quién ganó el partido? ⚽

Un partido entre el equipo Local y el Visitante terminó, y solo pudo pasar UNA de tres cosas: ganó el Local, ganó el Visitante, o hubo empate. Dos pistas periodísticas nos ayudan a descartar:

* **Pista 1:** hubo un gol en el minuto 90 → no fue empate.
* **Pista 2:** el equipo Local no anotó ningún gol en todo el partido → no ganó el Local.

Vamos a resolverlo con el atajo rápido (resolución) y compararlo con el método "revisar todo" (model checking), para comprobar que ambos coinciden:


```python
import futbol as acertijo1
from logic import model_check, resolution_entails

print("Pistas:", acertijo1.knowledge.formula())

pregunta = acertijo1.VISITANTE
print("\n¿Ganó el equipo Visitante?")
print("  Revisando TODOS los universos (lento) ->", model_check(acertijo1.knowledge, pregunta))
print("  Usando el atajo rápido (resolución)   ->", resolution_entails(acertijo1.knowledge, pregunta))
```

## 5. Acertijo 2: ¿Quién se perdió en la excursión? 🧭

Durante una salida de campo, uno de tres estudiantes (Ana, Beto o Caro) se perdió. El guía tiene dos testimonios:

* **Testimonio 1:** vio a Beto cruzando el río a las 3pm → Beto no se perdió.
* **Testimonio 2:** Ana mandó un mensaje confirmando que llegó al campamento → Ana no se perdió.

¿Adivinas quién fue? Comprobémoslo con el mismo atajo:


```python
import perdidos as acertijo2

print("Pistas:", acertijo2.knowledge.formula())

pregunta2 = acertijo2.PC
print("\n¿Se perdió Caro?")
print("  Revisando TODOS los universos (lento) ->", model_check(acertijo2.knowledge, pregunta2))
print("  Usando el atajo rápido (resolución)   ->", resolution_entails(acertijo2.knowledge, pregunta2))
```

## Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **El problema de revisar todo** | Con muchos hechos, revisar universo por universo se vuelve imposiblemente lento. |
| **CNF ("bloques SI-O")** | Una forma ordenada de escribir cualquier historia como pistas separadas. |
| **Resolución** | Tachar pistas contradictorias hasta encontrar una contradicción total, lo que confirma la respuesta. |
| **Dos caminos, un mismo destino** | `model_check` y `resolution_entails` siempre llegan a la misma conclusión — solo que uno es mucho más rápido con historias grandes. |
