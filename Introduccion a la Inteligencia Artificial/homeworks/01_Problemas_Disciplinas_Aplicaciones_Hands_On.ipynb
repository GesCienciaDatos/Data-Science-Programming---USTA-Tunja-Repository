### 📖 Contexto: Criterios de Aptitud para IA (Sección 1.6.4)

Antes de resolver el Reto 3, ten presentes algunos de los criterios típicos que determinan si un problema es apto para ser resuelto con IA:

- **Disponibilidad y calidad de los datos:** ¿existen suficientes datos históricos representativos y bien etiquetados?
- **Ausencia de reglas explícitas claras:** ¿el problema es tan complejo o cambiante que un humano no podría codificar reglas fijas a mano?
- **Tolerancia al error:** ¿el sistema puede equivocarse ocasionalmente sin consecuencias catastróficas, o se requiere precisión absoluta?
- **Valor del patrón a descubrir:** ¿encontrar el patrón genera un beneficio medible (económico, social, operativo)?
- **Naturaleza dinámica del problema:** ¿las condiciones cambian con el tiempo, requiriendo que el sistema se adapte o reentrene?
