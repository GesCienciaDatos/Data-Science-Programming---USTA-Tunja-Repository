### 📌 Reto 1: Los 4 Enfoques de la IA — ¡Traducidos! 🧸

Antes de empezar, aquí tienes una "chuleta" con los 4 enfoques en palabras simples:

| Enfoque | En palabras simples |
|---|---|
| **Pensar como humano** | El sistema intenta *razonar por dentro* parecido a como razona una persona (ej. simular el proceso mental). |
| **Actuar como humano** | Al sistema no le importa razonar "como persona" por dentro, solo que **por fuera** parezca humano (ej. pasar el Test de Turing). |
| **Pensar racionalmente** | El sistema usa **lógica y reglas correctas** para llegar a conclusiones, sin importar si se parece a un humano. |
| **Actuar racionalmente** | El sistema simplemente **toma la mejor decisión posible** para lograr su objetivo, sin importar cómo "piense" por dentro. |

**Ejemplo ya resuelto — Un termostato inteligente que aprende tus horarios:**
- *Descripción:* ajusta la temperatura de la casa automáticamente para gastar menos energía, aprendiendo tus rutinas.
- *Enfoque dominante:* "Actuar racionalmente" (toma la mejor decisión para lograr un objetivo: ahorrar energía y mantener el confort).
