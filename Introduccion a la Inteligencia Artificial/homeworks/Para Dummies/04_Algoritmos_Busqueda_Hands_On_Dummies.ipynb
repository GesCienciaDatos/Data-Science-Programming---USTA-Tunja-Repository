## 🎈 ¿Qué vamos a hacer en este taller?

Vas a jugar a ser un explorador dentro de un sistema de carpetas y archivos, probando 4 formas distintas de buscar un archivo:

1. 🕳️ **DFS:** te metes por un camino hasta el fondo antes de probar otro.
2. 🌊 **BFS:** exploras "en anillos", nivel por nivel.
3. 🧭 **GBFS:** usas una brújula (heurística) que apunta hacia el objetivo.
4. ⭐ **A\*:** combinas la brújula con la distancia ya recorrida, para no perderte pero tampoco dar rodeos.
