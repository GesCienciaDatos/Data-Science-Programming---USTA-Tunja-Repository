### 📖 Antes de empezar: ¿qué hace que un problema sea "bueno" para IA? 🧸

- ¿Hay **suficientes datos** de ejemplo para aprender el patrón?
- ¿Es **demasiado complicado** escribir reglas fijas a mano?
- ¿Un error ocasional del sistema es **tolerable** o sería grave?
- ¿Vale la pena el esfuerzo, es decir, **el patrón encontrado sirve para algo**?
- ¿El problema **cambia con el tiempo** (por eso convendría que el sistema aprenda en vez de seguir reglas fijas)?
