## 1. Introducción: La IA no es Neutral ⚖️

Durante años se vendió la idea de que los sistemas de IA, al estar basados en matemáticas y estadística, eran objetivos por diseño. La evidencia acumulada en la última década demuestra lo contrario: **un sistema de IA es tan neutral como los datos y las decisiones humanas que lo construyeron**.

Este cuaderno recorre seis dimensiones éticas que todo científico de datos debe dominar antes de desplegar un modelo en producción:

| # | Dimensión | Pregunta que responde |
|---|---|---|
| 1 | Sesgo algorítmico | ¿El modelo trata distinto a grupos que deberían tratarse igual? |
| 2 | Privacidad | ¿Qué tanto sabe el modelo sobre las personas, y con su consentimiento? |
| 3 | Transparencia / Explicabilidad | ¿Podemos explicar por qué el modelo decidió lo que decidió? |
| 4 | Responsabilidad | ¿Quién responde cuando el modelo se equivoca? |
| 5 | Impacto laboral | ¿Qué empleos transforma, reemplaza o crea la automatización? |
| 6 | Marco regulatorio | ¿Qué exige la ley, hoy y en el futuro cercano? |

## 2. Sesgo Algorítmico (*Algorithmic Bias*) ⚠️

Un modelo de IA no "decide" en el vacío: optimiza una función de costo sobre datos históricos. Si esos datos encierran discriminación estructural, el modelo la aprende, la reproduce y —peor aún— la presenta con la falsa autoridad de "lo dijo el algoritmo".

### Casos que marcaron el debate público:

| Caso | Año | Qué ocurrió |
|---|---|---|
| **Amazon – Reclutamiento con IA** | 2018 | El modelo penalizaba currículums que incluían la palabra "mujeres" (p. ej. "club de ajedrez de mujeres"), entrenado con 10 años de CVs mayoritariamente masculinos. |
| **COMPAS – Justicia Penal (EE.UU.)** | 2016 | Un algoritmo de predicción de reincidencia asignaba puntajes de riesgo falsamente altos a personas afroamericanas con el doble de frecuencia que a personas blancas con historiales equivalentes (investigación de ProPublica). |
| **Gender Shades – Reconocimiento Facial** | 2018 | Joy Buolamwini (MIT Media Lab) demostró que sistemas comerciales de reconocimiento facial tenían hasta 34 % más error en mujeres de piel oscura que en hombres de piel clara. |
| **Redlining digital – Scoring crediticio** | — | Aunque no se use raza o género explícitamente, variables como el código postal actúan como *proxy* indirecto, heredando patrones de segregación histórica. |

### ¿De dónde viene el sesgo?
- **Sesgo histórico:** los datos reflejan una desigualdad que ya existía en el mundo.
- **Sesgo de representación:** algunos grupos están sub-representados en el conjunto de entrenamiento.
- **Sesgo de medición:** las variables elegidas miden distinto según el grupo (p. ej. arrestos ≠ delitos cometidos).
- **Sesgo de agregación:** un solo modelo "para todos" ignora diferencias legítimas entre subpoblaciones.

### 💻 Laboratorio: Sesgo Colado por una Variable Proxy

Vamos a construir un caso sintético de otorgamiento de crédito. El modelo **nunca recibe** la variable protegida (`grupo_protegido`) como insumo — pero sí recibe el **puntaje del código postal**, una variable "neutral" en apariencia que, en este escenario simulado, está correlacionada históricamente con el grupo. Veremos cómo el sesgo se cuela de todas formas.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 2000

# Variable protegida (NO se usará como insumo del modelo)
grupo = rng.choice(['Grupo A', 'Grupo B'], size=n, p=[0.5, 0.5])

# Variable "neutral" en apariencia: puntaje de código postal (proxy correlacionado con el grupo)
# En este escenario simulado, el Grupo B ha sido históricamente segregado hacia
# barrios con menor puntaje crediticio de referencia
codigo_postal_score = np.where(
    grupo == 'Grupo B',
    rng.normal(40, 12, n),   # barrios con score históricamente más bajo
    rng.normal(65, 12, n)    # barrios con score históricamente más alto
)
codigo_postal_score = np.clip(codigo_postal_score, 0, 100)

# Ingreso real y antigüedad laboral: NO dependen del grupo (son equivalentes entre grupos)
ingreso_mensual = rng.normal(3_000_000, 600_000, n)
antiguedad_laboral = rng.integers(0, 20, n)

df = pd.DataFrame({
    'grupo_protegido': grupo,                     # excluido del modelo
    'codigo_postal_score': codigo_postal_score,    # proxy indirecto del grupo
    'ingreso_mensual': ingreso_mensual,
    'antiguedad_laboral': antiguedad_laboral
})
df.head()
````

````python
# Clasificador simple: aprueba el crédito si un puntaje combinado supera un umbral.
# El puntaje SÍ incluye 'codigo_postal_score', aunque el equipo de datos nunca
# usó 'grupo_protegido' de forma directa: el sesgo se cuela por la variable proxy.

def puntaje_credito(row):
    return (
        0.5 * row['codigo_postal_score'] +
        0.00003 * row['ingreso_mensual'] +
        1.2 * row['antiguedad_laboral']
    )

df['puntaje'] = df.apply(puntaje_credito, axis=1)
umbral = df['puntaje'].median()
df['credito_aprobado'] = (df['puntaje'] >= umbral).astype(int)

tasa_aprobacion = df.groupby('grupo_protegido')['credito_aprobado'].mean()
print("Tasa de aprobación por grupo protegido:")
print((tasa_aprobacion * 100).round(1).astype(str) + " %")

diferencia_paridad = tasa_aprobacion.max() - tasa_aprobacion.min()
razon_impacto = tasa_aprobacion.min() / tasa_aprobacion.max()
print(f"\nDiferencia de paridad demográfica: {diferencia_paridad:.2%}")
print("Regla de referencia ('regla del 80%', EEOC): una razón < 0.8 sugiere impacto adverso.")
print(f"Razón de impacto adverso observada: {razon_impacto:.2f}")
````

````python
plt.figure(figsize=(6, 4))
tasa_aprobacion.plot(kind='bar', color=['#4f46e5', '#f59e0b'])
plt.ylabel('Tasa de aprobación de crédito')
plt.title('Impacto dispar por grupo protegido\n(el modelo nunca "vio" el grupo, pero lo infirió vía código postal)')
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

## 3. Privacidad y Vigilancia de Datos 🔐

Los sistemas de IA modernos —especialmente los modelos de lenguaje y los sistemas de recomendación— se entrenan y operan con volúmenes de datos personales sin precedentes: ubicación, hábitos de consumo, conversaciones, biometría facial y de voz.

### Tensiones centrales:

| Tensión | Descripción |
|---|---|
| **Consentimiento informado vs. Términos y Condiciones ilegibles** | Nadie lee 40 páginas de términos legales; el "consentimiento" es a menudo una ficción jurídica. |
| **Recolección incidental** | Asistentes de voz y cámaras "siempre activas" capturan datos de terceros que nunca dieron su consentimiento. |
| **Reidentificación** | Datos "anonimizados" pueden reidentificarse cruzando variables aparentemente inocuas (Latanya Sweeney, 1997: 87 % de la población de EE.UU. es identificable solo con código postal, fecha de nacimiento y sexo). |
| **Vigilancia algorítmica** | Reconocimiento facial en espacios públicos y *scoring* social plantean riesgos para la libertad de expresión y de movimiento. |
| **Dark patterns** | Interfaces diseñadas para inducir a compartir más datos de los que la persona desearía. |

**Principio guía:** minimización de datos — recolectar solo lo estrictamente necesario para el propósito declarado (principio central del GDPR y de la Ley 1581 de 2012 en Colombia).

## 4. Transparencia y Explicabilidad (*Explainable AI — XAI*) 🔍

Existe una tensión estructural entre **desempeño** y **explicabilidad**: los modelos más precisos (redes neuronales profundas, ensambles) suelen ser los más opacos ("cajas negras"), mientras que los modelos más transparentes (árboles pequeños, regresión lineal) suelen sacrificar precisión.

| Tipo de modelo | Explicabilidad | Ejemplos |
|---|---|---|
| **Caja de cristal** (*glass-box*) | Alta — la lógica es legible directamente | Regresión lineal/logística, árboles de decisión poco profundos, reglas *if-then* |
| **Caja negra** (*black-box*) | Baja — requiere técnicas externas para interpretarse | Redes neuronales profundas, *random forests* grandes, *gradient boosting* |

### Técnicas post-hoc para abrir la caja negra:
- **LIME** (*Local Interpretable Model-agnostic Explanations*): aproxima localmente una predicción compleja con un modelo simple e interpretable.
- **SHAP** (*SHapley Additive exPlanations*): reparte la contribución de cada variable a una predicción usando teoría de juegos cooperativos (valores de Shapley).
- **Mapas de calor / *saliency maps*:** resaltan qué píxeles o palabras influyeron más en la decisión de un modelo de visión o de lenguaje.

**¿Por qué importa?** En sectores regulados (crédito, salud, justicia), la ley suele exigir poder explicar una decisión automatizada a la persona afectada — no basta con que el modelo "funcione".

## 5. Responsabilidad ante Decisiones Automatizadas 👤⚖️

Cuando un sistema de IA causa un daño (deniega un crédito injustamente, diagnostica mal, rechaza una hoja de vida), surge la llamada **brecha de responsabilidad** (*accountability gap*): ¿quién responde — la empresa que lo desplegó, quien lo programó, quien etiquetó los datos, o el propio sistema?

| Actor | Nivel de responsabilidad típico |
|---|---|
| **Organización que despliega el modelo** | Responsabilidad legal principal: decide implementarlo en producción y sobre quién. |
| **Equipo de ciencia de datos** | Responsabilidad técnica y ética: diseño, validación y documentación del modelo. |
| **Proveedor del modelo / API externa** | Responsabilidad contractual parcial, usualmente limitada por términos de servicio. |
| **El algoritmo** | **Ninguna.** Un modelo no es sujeto de derechos ni de obligaciones legales. |

### Principio de "humano en el bucle" (*Human-in-the-loop*)
Para decisiones de alto impacto (crédito, empleo, justicia, salud), las buenas prácticas —y cada vez más la ley— exigen que una persona humana pueda revisar, cuestionar y revertir la decisión automatizada antes de que produzca efectos irreversibles.

## 6. Impacto Laboral y Económico de la IA 💼

La automatización impulsada por IA no elimina "el empleo" de forma uniforme: transforma **tareas específicas** dentro de las ocupaciones.

| Categoría de tareas | Ejemplos | Efecto esperado |
|---|---|---|
| **Rutinarias y predecibles** | Digitación de datos, clasificación documental, atención de primer nivel | Alta automatización |
| **Creativas asistidas** | Diseño gráfico, redacción, programación | Transformación: la IA como copiloto, no reemplazo total |
| **Alto contacto humano / juicio ético** | Enfermería, negociación, liderazgo, trabajo social | Baja automatización directa |
| **Nuevas ocupaciones** | Ingeniería de *prompts*, auditoría algorítmica, curaduría de datos | Creación de empleo especializado |

**Dato de referencia:** el Foro Económico Mundial (*Future of Jobs Report*) proyecta que, aunque la IA desplazará millones de puestos rutinarios en esta década, también creará un número comparable de roles nuevos — el reto central es la **reconversión laboral (*reskilling*)**, no solo el balance neto de empleos.

## 7. Marcos Regulatorios Globales 📜

Ante estos riesgos, distintas jurisdicciones han empezado a regular la IA con enfoques diferentes:

| Marco | Jurisdicción | Enfoque |
|---|---|---|
| **Ley de IA de la Unión Europea (*EU AI Act*, 2024)** | Unión Europea | Enfoque basado en **niveles de riesgo**: a mayor riesgo, mayores obligaciones. |
| **NIST AI Risk Management Framework** | EE.UU. (voluntario) | Marco de gestión de riesgos por principios: gobernar, mapear, medir, gestionar. |
| **Principios de IA de la OCDE** | Internacional (38+ países) | Transparencia, robustez, rendición de cuentas e IA centrada en el ser humano. |
| **CONPES 3975 / Ley 1581 de 2012 (Habeas Data)** | Colombia | Política nacional de transformación digital e IA + protección de datos personales. |

### Pirámide de riesgo de la Ley de IA de la UE:

```mermaid

    A["🔴 Riesgo Inaceptable — PROHIBIDO<br/>scoring social, manipulación subliminal, IA policial predictiva masiva"] --> B["🟠 Alto Riesgo — Regulación estricta<br/>selección de personal, crédito, salud, justicia, infraestructura crítica"]
```

A mayor riesgo para los derechos fundamentales de las personas, mayores son las obligaciones de documentación, auditoría humana y transparencia exigidas antes de poner el sistema en producción.

## Resumen / Puntos Clave 📌

- La neutralidad algorítmica es un mito: todo modelo de IA hereda los sesgos de sus datos y de las decisiones de diseño de quienes lo construyen.
- El sesgo puede colarse incluso sin usar la variable protegida directamente, a través de **variables proxy** (código postal, nombre, barrio).
- La privacidad exige minimización de datos y consentimiento genuino, no solo "términos y condiciones" formales.
- Existe una tensión entre desempeño y explicabilidad; técnicas como LIME y SHAP ayudan a abrir la caja negra cuando el contexto lo exige.
- La responsabilidad legal y ética recae siempre en personas y organizaciones — nunca en el algoritmo.
- La automatización transforma tareas, no ocupaciones completas; el reto central es la reconversión laboral.
- Marcos como la Ley de IA de la UE regulan por **niveles de riesgo**, exigiendo mayor control cuanto mayor sea el impacto potencial sobre las personas.
