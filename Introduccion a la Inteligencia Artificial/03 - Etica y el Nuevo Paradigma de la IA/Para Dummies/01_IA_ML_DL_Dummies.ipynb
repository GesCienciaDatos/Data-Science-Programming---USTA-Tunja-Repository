## ¿Qué vamos a aprender aquí? 🎈

Seguro has escuchado usar "Inteligencia Artificial", "Machine Learning" y "Deep Learning" como si fueran la misma cosa con nombres distintos. **¡No lo son!** Son como esas muñecas rusas (*matryoshkas*) que se meten una dentro de la otra. En este cuaderno vas a entender, sin fórmulas raras:

1. Cómo se "meten" estos tres términos uno dentro del otro.
2. Cuándo conviene usar la muñeca grande, la mediana o la chiquita.
3. Por qué a la muñeca más pequeña (Deep Learning) le dicen "profunda".
4. Vamos a construir, con puro código sencillo, un cerebrito de juguete que resuelve un acertijo que un "cerebrito" más simple no puede resolver.

## 1. Las Muñecas Rusas del Cerebro Digital 🪆

Imagina tres muñecas rusas, una dentro de la otra:

- La muñeca **más grande** es la **Inteligencia Artificial**: cualquier cosa que haga que una máquina "parezca" lista — hasta un simple sistema de reglas tipo "si pasa esto, haz esto otro".
- Adentro de esa, hay una muñeca **mediana**: el **Machine Learning**. Son las máquinas que, en lugar de que les programemos cada regla a mano, **aprenden solas** mirando ejemplos.
- Y adentro de esa, la muñeca **más chiquita**: el **Deep Learning**. Son máquinas que aprenden usando "cerebritos" hechos de muchas capas conectadas (redes neuronales), inspiradas —muy de lejos— en cómo funcionan las neuronas de verdad.


**Frase para recordar:** toda muñeca chiquita (DL) es también mediana (ML), y toda mediana (ML) es también grande (IA) — pero no al revés: no toda IA usa Machine Learning, y no todo Machine Learning es Deep Learning.

## 2. Tabla de Diferencias en Cristiano 📋

| | IA (la grande) | Machine Learning (la mediana) | Deep Learning (la chiquita) |
|---|---|---|---|
| **¿Cómo "piensa"?** | A veces con reglas fijas, escritas por una persona | Mirando ejemplos y encontrando patrones | Con un cerebrito de muchas capas |
| **¿Necesita muchos ejemplos?** | No siempre | Algunos cientos o miles | Miles o millones |
| **¿Se puede explicar fácil por qué decidió algo?** | Sí, casi siempre | Depende, pero suele ser explicable | Difícil: es como una "caja negra" |
| **Ejemplo de la vida real** | Un semáforo con reglas fijas de tiempo | Un sistema que aprende a detectar clientes que se van a ir de un servicio | Un sistema que reconoce tu cara en una foto |

## 3. ¿Cuándo Uso Cuál? 🤔

| Situación | ¿Qué muñeca uso? |
|---|---|
| Tengo pocos datos, pero reglas claras del negocio | La muñeca grande: **reglas de IA** |
| Tengo datos en una tabla (filas y columnas) y necesito explicar cada decisión | La muñeca mediana: **Machine Learning** |
| Tengo fotos, audios, videos o textos largos | La muñeca chiquita: **Deep Learning** |
| Tengo un computador viejito, sin mucha potencia | Mejor la mediana; la chiquita necesita mucha "fuerza bruta" de cómputo |

## 4. ¿Qué es "Profundo"? El Semáforo que Necesitaba Ayuda 🚦

Imagina un guardia de seguridad solitario que solo puede decidir con **una sola regla** a la vez ("si veo esto, digo sí; si no, digo no"). Ese guardia es como un **perceptrón simple**: solo puede resolver acertijos sencillos, donde basta con trazar una única línea para separar las respuestas correctas de las incorrectas.

Pero hay acertijos donde **una sola línea no alcanza**. Para esos casos, contratamos un **equipo de guardias** que se pasan opiniones entre sí antes de dar la respuesta final — eso es una **capa oculta**. Entre más "capas" de guardias trabajando en equipo, más "profundo" es el cerebrito, y más acertijos complicados puede resolver.

Vamos a construir, con puro `numpy` (sin ayuditas de librerías avanzadas), un cerebrito de juguete que resuelve justo uno de esos acertijos que un guardia solitario no puede resolver: el acertijo de las **luces que se prenden solo si los switches son diferentes** (en la jerga técnica, esto se llama "XOR").

```python
import numpy as np

# El acertijo: dos switches (0 = apagado, 1 = encendido).
# La luz se prende (1) SOLO si los switches están en posiciones DIFERENTES.
switches = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
luz_esperada = np.array([[0], [1], [1], [0]])

print("Combinaciones de switches:\n", switches)
print("\n¿Se prende la luz? (esperado):", luz_esperada.ravel())
print("\nOjo: no existe una sola línea recta que separe los 'sí' de los 'no'")
print("en este acertijo -> un guardia solitario (perceptrón simple) no puede resolverlo.")
```

```python
def sigmoide(z):
    return 1 / (1 + np.exp(-z))

def derivada_sigmoide(a):
    return a * (1 - a)

rng = np.random.default_rng(7)

# Nuestro "equipo de guardias": 2 entradas -> 4 guardias en la capa oculta -> 1 respuesta final
W1 = rng.normal(0, 1, (2, 4))
b1 = np.zeros((1, 4))
W2 = rng.normal(0, 1, (4, 1))
b2 = np.zeros((1, 1))

tasa_aprendizaje = 0.5
for ronda in range(10000):
    # El equipo de guardias opina (capa oculta) y luego se da la respuesta final
    opinion_equipo = sigmoide(switches @ W1 + b1)
    respuesta_final = sigmoide(opinion_equipo @ W2 + b2)

    # Comparamos con la respuesta correcta y ajustamos un poquito los "criterios" de cada guardia
    error = respuesta_final - luz_esperada
    delta2 = error * derivada_sigmoide(respuesta_final)
    delta1 = (delta2 @ W2.T) * derivada_sigmoide(opinion_equipo)

    W2 -= tasa_aprendizaje * (opinion_equipo.T @ delta2)
    b2 -= tasa_aprendizaje * delta2.sum(axis=0, keepdims=True)
    W1 -= tasa_aprendizaje * (switches.T @ delta1)
    b1 -= tasa_aprendizaje * delta1.sum(axis=0, keepdims=True)

opinion_equipo, respuesta_final = sigmoide(switches @ W1 + b1), sigmoide(sigmoide(switches @ W1 + b1) @ W2 + b2)
print("Respuesta del equipo de guardias (redondeada):", respuesta_final.round(3).ravel())
print("Clasificación final:                          ", (respuesta_final > 0.5).astype(int).ravel())
print("Lo que se esperaba:                            ", luz_esperada.ravel())
```

### 🤔 ¿Qué aprendimos con esto?

El equipo de guardias (la capa oculta) sí logra resolver el acertijo, mientras que un guardia solitario (sin capa oculta) se queda atascado y nunca acierta las cuatro combinaciones. Esa es, en chiquito, la diferencia entre Machine Learning "básico" y Deep Learning: **agregar capas le da al cerebrito la capacidad de resolver acertijos más retorcidos**, a cambio de necesitar más ejemplos y más potencia de cómputo para entrenarlo.

## Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **IA** | La muñeca rusa más grande: cualquier máquina que "parezca" lista. |
| **Machine Learning** | La muñeca mediana: aprende mirando ejemplos, en vez de reglas escritas a mano. |
| **Deep Learning** | La muñeca chiquita: aprende con un cerebrito de muchas capas conectadas. |
| **"Profundo"** | Se refiere a cuántas capas de "guardias" tiene el cerebrito — más capas, más acertijos difíciles puede resolver. |
| **XOR** | El acertijo clásico que un guardia solitario no puede resolver, pero un equipo (capa oculta) sí. |
