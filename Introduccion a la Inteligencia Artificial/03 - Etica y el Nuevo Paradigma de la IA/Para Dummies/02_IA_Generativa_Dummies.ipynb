## ¿Qué vamos a aprender aquí? 🎈

Hasta ahora hemos hablado de robots que **clasifican** cosas (¿es spam o no es spam?, ¿aprueba el crédito o no?). Pero hay otro tipo de robot, mucho más de moda últimamente: el que **inventa cosas nuevas** — textos, dibujos, canciones. En este cuaderno, sin tecnicismos, vas a entender:

1. La diferencia entre un robot "juez" y un robot "pintor".
2. Qué tipo de cosas puede inventar un robot pintor/escritor.
3. Cómo, en el fondo, un robot que "escribe" solo está adivinando la palabra más probable que sigue — igual que un loro muy, muy leído.
4. Vamos a construir nuestro propio "loro de juguete" con puro código sencillo.

## 1. Dos Tipos de Robots: El Juez y el Pintor ⚖️🎨

- El **robot juez** (IA "discriminativa" o predictiva) mira algo que ya existe y decide a qué categoría pertenece: "esta foto es un gato", "este correo es spam", "a esta persona le aprobamos el crédito". Su trabajo es **clasificar o predecir**.
- El **robot pintor** (IA "generativa") no clasifica nada que ya exista: **inventa algo nuevo desde cero** — un texto, una imagen, una canción — que se parece mucho a los ejemplos que vio mientras aprendía, pero que nunca existió exactamente igual antes.

| | Robot Juez (Discriminativo) | Robot Pintor (Generativo) |
|---|---|---|
| ¿Qué hace? | Decide a qué categoría pertenece algo | Inventa algo nuevo, parecido a lo que aprendió |
| Pregunta típica | "¿Esto es A o B?" | "¿Cómo se vería/sonaría/leería un ejemplo nuevo de esto?" |
| Ejemplo | Detector de correos basura | Robot que escribe cuentos o dibuja imágenes |

## 2. ¿Qué Puede Inventar? 🎁

| Tipo de cosa | ¿Cómo se llama la técnica? | Familias de robots famosos (a nivel de nombre) |
|---|---|---|
| **Texto** | Adivina la próxima palabra, una y otra vez | GPT, Llama, Claude, Gemini |
| **Imágenes** | Empieza con "ruido" (como estática de tele) y lo va limpiando hasta que aparece un dibujo | Stable Diffusion, DALL·E, Midjourney |
| **Audio / Música** | Parecido a las imágenes, pero con sonido | MusicLM, voces sintéticas |
| **Código de programación** | Adivina la próxima línea de código | GitHub Copilot |

## 3. El Loro Súper Leído 🦜

Imagina un loro que leyó **toda** una biblioteca gigante. Ese loro no memorizó cada libro palabra por palabra, pero aprendió tan bien "qué palabra suele venir después de cuál otra" que, si le dices las primeras dos palabras de una frase, puede seguir inventando una frase completa que suena coherente — palabra tras palabra, adivinando siempre "cuál es la palabra más probable que sigue".

Eso es, en el fondo, lo que hacen los robots que escriben texto: **son loros estadísticos gigantes**. Los robots que dibujan funcionan distinto: son más parecidos a un escultor que parte de un bloque de mármol lleno de "ruido" y, poquito a poco, le va quitando pedacitos hasta que aparece una figura reconocible que coincide con lo que le pediste.

## 4. Manos a la Obra: Nuestro Propio Loro de Juguete 🧸

Vamos a construir, con puro Python sencillo (sin ningún robot gigante de verdad), un "loro de juguete" que aprende de un puñado de frases y luego inventa frases nuevas adivinando, palabra por palabra, cuál suele venir después de cuál.

```python
import random

# El "libro" que le vamos a dar a leer a nuestro loro (inventado, chiquito)
libro = """
la inteligencia artificial aprende patrones a partir de datos
el aprendizaje profundo usa redes neuronales con muchas capas
la inteligencia artificial generativa produce contenido nuevo
el aprendizaje automatico aprende patrones sin reglas explicitas
la inteligencia artificial etica exige transparencia y responsabilidad
""".split()

# El loro se fija en parejas de palabras y anota qué palabra vino justo después
memoria_del_loro = {}
for i in range(len(libro) - 2):
    pareja = (libro[i], libro[i + 1])
    palabra_siguiente = libro[i + 2]
    memoria_del_loro.setdefault(pareja, []).append(palabra_siguiente)

print(f"El loro aprendió {len(memoria_del_loro)} 'parejitas de palabras' distintas.")
print("Ejemplo de lo que recuerda:", list(memoria_del_loro.items())[0])
```

```python
def el_loro_inventa(primeras_dos_palabras, cuantas_palabras_mas=15, semilla=3):
    generador = random.Random(semilla)
    frase = list(primeras_dos_palabras)
    for _ in range(cuantas_palabras_mas):
        pareja_actual = tuple(frase[-2:])
        posibles_siguientes = memoria_del_loro.get(pareja_actual)
        if not posibles_siguientes:
            break  # el loro nunca escuchó esta combinación -> mejor se calla
        frase.append(generador.choice(posibles_siguientes))
    return ' '.join(frase)

print("El loro inventa:", el_loro_inventa(('la', 'inteligencia')))
```

### 🤔 ¿Qué aprendimos?

Nuestro loro solo se fija en las **2 últimas palabras** antes de adivinar la siguiente. Los robots-loro de verdad (como los que escriben textos larguísimos) se fijan en **miles** de palabras de contexto a la vez, y aprendieron leyendo una cantidad de texto casi imposible de imaginar. Pero la idea de fondo es exactamente la misma que programamos aquí: **adivinar, palabra por palabra, qué es lo más probable que siga**.

## 5. Cuando el Loro se Inventa Cosas: Alucinaciones y Caras Falsas 👻🎭

| Riesgo | En criollo |
|---|---|
| **Alucinaciones** | El loro, en vez de decir "no sé", se inventa una respuesta que suena muy convincente pero es falsa. |
| **Deepfakes** | Videos o audios falsos, hechos por un robot pintor, donde parece que una persona real dijo o hizo algo que nunca dijo ni hizo. |
| **Desinformación en cantidad** | Como el robot puede escribir mil textos falsos en un minuto, es fácil inundar internet de mentiras. |
| **Derechos de autor** | El robot pintor aprendió mirando obras de artistas de verdad; a veces "copia" un estilo sin permiso ni crédito. |
| **Prejuicios heredados** | Si el robot aprendió de ejemplos con prejuicios (como vimos en el Cuaderno 00), sus dibujos o textos también los repiten. |

## 6. Reglas de Oro para Usar Estos Robots con Cabeza 🧠✅

| Regla | Por qué ayuda |
|---|---|
| **Siempre revisa lo que el robot te dio** | Un humano atento detecta la mayoría de los inventos falsos antes de que hagan daño. |
| **Marca el contenido hecho por robots** | Así la gente sabe que algo es sintético y no una grabación real. |
| **Pide fuentes, no solo respuestas** | Si el robot te dice de dónde sacó la información, es más fácil verificarla. |
| **Ten reglas claras en tu organización** | Decidir de antemano para qué sí y para qué no se usará el robot evita sorpresas. |
| **Enseña a la gente a reconocer contenido falso** | Tan importante como frenar al robot es enseñarle al público a dudar y verificar. |
