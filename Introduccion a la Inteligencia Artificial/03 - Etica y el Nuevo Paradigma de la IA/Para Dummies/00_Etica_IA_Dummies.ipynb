Mucha gente cree que, como un computador no tiene sentimientos, sus decisiones son automáticamente justas. **¡Falso de toda falsedad!** Un robot que aprende observando el pasado puede aprender también las mañas, los prejuicios y las injusticias del pasado. ## 1. El Robot Aprendiz que Copia Mañas 🦜

Imagina un loro muy inteligente que creció escuchando conversaciones en un barrio donde, por costumbre injusta, siempre le daban el mejor pedazo de pan a un solo grupo de vecinos. Si un día le pides al loro que reparta el pan "como siempre se ha hecho", el loro va a repetir la misma injusticia — no porque sea malo, sino porque **aprendió del ejemplo que le dieron**.

Así funcionan muchos sistemas de IA: aprenden de datos del pasado, y si el pasado fue injusto con algún grupo de personas, el robot va a repetir esa injusticia, pero ahora a una velocidad y una escala mucho mayores.

### Casos que pasaron de verdad:

| Caso | En criollo |
|---|---|
| **Robot de contratación de Amazon** | Aprendió mirando 10 años de hojas de vida, casi todas de hombres, y empezó a rechazar hojas de vida que mencionaran "mujeres" (como "club de ajedrez de mujeres"). |
| **Robot de riesgo penal COMPAS** | Le ponía "cara de peligroso" el doble de veces a personas afrodescendientes que a personas blancas con historiales parecidos. |
| **Robots que reconocen caras** | Le costaba mucho más reconocer correctamente a mujeres de piel oscura que a hombres de piel clara, porque casi no vio fotos de ese grupo mientras aprendía. |

```python
# Vamos a "fabricar" un mini ejemplo de esto que acabamos de contar,
# usando solo números inventados (nadie real está en estos datos).
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 2000

# Imaginemos dos barrios: "Barrio Sol" y "Barrio Luna" (nombres inventados)
barrio = rng.choice(['Barrio Sol', 'Barrio Luna'], size=n, p=[0.5, 0.5])

# El "puntaje del código postal": una variable que suena inocente,
# pero que en nuestra historia inventada quedó ligada históricamente al barrio
puntaje_barrio = np.where(
    barrio == 'Barrio Luna',
    rng.normal(40, 12, n),   # a este barrio, históricamente, le fue peor
    rng.normal(65, 12, n)    # a este barrio, históricamente, le fue mejor
)
puntaje_barrio = np.clip(puntaje_barrio, 0, 100)

datos = pd.DataFrame({'barrio': barrio, 'puntaje_barrio': puntaje_barrio})
datos.head()
```

### 🤔 ¿Qué pasamos por alto?

Fíjate: en ningún momento le dijimos al robot "fíjate en el barrio de la persona". Solo le dimos un "puntaje" que suena técnico y neutral. Pero como ese puntaje **ya venía cargado con la injusticia del pasado**, el robot termina discriminando igual, aunque nunca haya "visto" directamente el nombre del barrio. A esto los expertos le llaman usar una **variable disfrazada (proxy)**.

## 2. El Diario que Todos Pueden Leer 📔

Ahora hablemos de privacidad. Imagina que llevas un diario secreto, pero cada vez que hablas con tu asistente de voz, con tu buscador o con tus redes sociales, arrancas una hojita y se la regalas a un desconocido. Poco a poco, ese desconocido junta tantas hojitas que termina sabiendo de ti más que tu mejor amigo: dónde vives, qué te gusta comprar, a qué horas duermes.

- **El problema del "acepto sin leer":** casi nadie lee los 40 páginas de términos y condiciones antes de darle clic a "Aceptar". Eso no es un consentimiento real, es una formalidad.
- **El problema del vecino curioso:** aunque un dato "no tenga tu nombre", cruzando dos o tres pistas (tu barrio, tu edad, tu género) casi siempre se puede adivinar quién eres — como armar un rompecabezas con pocas piezas.

**Regla de oro:** solo se debería pedir la información estrictamente necesaria para lo que se va a usar — ni una hojita de diario más.

## 3. La Caja Mágica que no Explica Nada 🎩

Hay dos tipos de robots, según qué tan bien podamos "ver" cómo piensan:

| Tipo de robot | ¿Podemos ver cómo piensa? |
|---|---|
| **Frasco de vidrio** | Sí: es una receta de cocina con pasos clarísimos ("si el ingreso es mayor a X, aprueba"). |
| **Caja negra mágica** | No: mete la mano, saca una respuesta, ¡y ni el mago sabe explicar exactamente por qué salió esa respuesta y no otra! |

Los robots más "listos" (redes neuronales gigantes) suelen ser cajas negras mágicas: aciertan mucho, pero cuesta muchísimo explicar el "por qué" de cada respuesta. Por eso, en cosas importantes de verdad (como decidir si te dan un crédito o no), la ley exige poder explicarle a la persona afectada qué pasó, no solo decirle "confía en la caja mágica".

## 4. ¿Quién Paga los Platos Rotos? 💥

Si contratas a un robot mayordomo y este rompe un jarrón carísimo, ¿quién paga? ¿El robot? ¡El robot no tiene bolsillo ni cuenta bancaria!

| ¿Quién es? | ¿Responde por el error? |
|---|---|
| La empresa que puso al robot a trabajar | Sí, la principal responsable. |
| Las personas que programaron y entrenaron al robot | Sí, responsabilidad técnica y ética. |
| El robot mismo | **No.** Un robot no puede ir a la cárcel ni pagar una multa. |

**Regla de oro #2:** para decisiones muy importantes (crédito, salud, empleo, justicia), siempre debería haber una **persona humana** que pueda revisar y, si hace falta, corregir lo que dijo el robot antes de que sea demasiado tarde.

## 5. El Empleo del Futuro: ¿Amigo o Enemigo? 🤝🤖

La IA no llega a "quitarle el trabajo a todo el mundo" de un solo golpe: cambia **tareas puntuales** dentro de cada trabajo.

| Tipo de tarea | Ejemplo | ¿Qué pasa? |
|---|---|---|
| Repetitiva | Digitar datos, ordenar papeles | Se automatiza mucho |
| Creativa con ayuda | Diseñar, escribir, programar | El robot ayuda, no reemplaza del todo |
| Con mucho trato humano | Enfermería, liderazgo, negociación | Difícil de automatizar |
| Totalmente nueva | Entrenar y auditar robots | ¡Se crean empleos que antes no existían! |

El reto más grande no es "cuántos empleos desaparecen", sino **aprender cosas nuevas a tiempo** para no quedarse atrás (a esto se le llama *reskilling* o reconversión laboral).

## 6. Las Reglas del Juego (Semáforo de la Ley) 🚦

Igual que un semáforo, algunos países (como los de la Unión Europea) decidieron ponerle "colores de riesgo" a los usos de la IA:

| Color | Significado | Ejemplo |
|---|---|---|
| 🔴 **Rojo** | Prohibido totalmente | Vigilar y puntuar el comportamiento de toda la población ("scoring social") |
| 🟠 **Naranja** | Permitido, pero con muchas reglas | Elegir a quién contratar, a quién dar un crédito, diagnósticos médicos |
| 🟡 **Amarillo** | Permitido, avisando que es un robot | Chatbots, videos generados por IA (*deepfakes*) |
| 🟢 **Verde** | Libre, sin reglas especiales | Filtros de spam, videojuegos con IA |

Mientras más daño le pueda hacer un uso de la IA a una persona real, más "luces rojas y naranjas" enciende la ley.
