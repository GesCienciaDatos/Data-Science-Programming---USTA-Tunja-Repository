## 1. Introducción: Tres Círculos, un Solo Campo 🎯

La Inteligencia Artificial, el Aprendizaje Automático (*Machine Learning*) y el Aprendizaje Profundo (*Deep Learning*) **no son sinónimos** ni campos independientes: son **conjuntos anidados**, donde cada uno es un subconjunto especializado del anterior.

- La **Inteligencia Artificial (IA)** es el campo más amplio: cualquier sistema que exhiba un comportamiento que asociaríamos con "inteligencia" — desde sistemas basados en reglas escritas a mano hasta los modelos más sofisticados de hoy.
- El **Aprendizaje Automático (*Machine Learning*, ML)** es un subconjunto de la IA: sistemas que **aprenden patrones a partir de datos**, en lugar de seguir reglas programadas explícitamente.
- El **Aprendizaje Profundo (*Deep Learning*, DL)** es un subconjunto del ML: usa **redes neuronales con múltiples capas** para aprender representaciones jerárquicas de los datos de forma automática.

 

Es decir: **todo Deep Learning es Machine Learning, y todo Machine Learning es Inteligencia Artificial — pero no al revés.**

## 2. Diferencias Conceptuales

| Dimensión | IA (general) | Machine Learning | Deep Learning |
|---|---|---|---|
| **Definición** | Cualquier sistema que simula comportamiento inteligente | Sistemas que aprenden patrones desde datos | ML con redes neuronales de múltiples capas |
| **Cómo "aprende"** | Puede no aprender: reglas fijas escritas por humanos | Ajusta parámetros minimizando el error sobre datos | Ajusta millones de parámetros vía descenso de gradiente y retropropagación |
| **Ingeniería de características** | No aplica | Manual: un experto diseña las variables (*features*) | Automática: las capas aprenden representaciones jerárquicas |
| **Volumen de datos típico** | Variable | Moderado (cientos a miles de registros) | Grande a masivo (miles a millones de registros) |
| **Poder de cómputo requerido** | Bajo | Bajo–moderado | Alto (usualmente GPU/TPU) |
| **Interpretabilidad** | Alta (reglas explícitas) | Media–alta, según el algoritmo | Baja ("caja negra") |
| **Ejemplos** | Sistema experto médico de los años 80, motor de ajedrez con reglas | Regresión logística para *churn*, árboles de decisión para crédito | Reconocimiento de imágenes, traducción automática, modelos de lenguaje |

## 3. ¿Cuándo Usar Cada Enfoque? 🧭

| Escenario | Enfoque recomendado | Razón |
|---|---|---|
| Pocos datos (< 1,000 registros) y reglas de negocio claras | **IA basada en reglas** | No hay suficientes datos para "aprender" patrones confiables. |
| Datos tabulares estructurados, se necesita explicar cada decisión (crédito, salud) | **Machine Learning clásico** | Modelos como árboles o regresión logística son precisos y auditables. |
| Imágenes, audio, texto libre, video | **Deep Learning** | Las redes profundas extraen automáticamente patrones visuales/semánticos complejos. |
| Recursos computacionales muy limitados (dispositivo embebido) | **ML clásico o DL comprimido** | El DL completo exige GPU/memoria que puede no estar disponible. |
| Se necesita una solución explicable ante un regulador | **ML interpretable o IA basada en reglas** | El DL suele sacrificar transparencia a cambio de desempeño. |

## 4. ¿Qué Significa "Profundo" en Deep Learning? 🔬

La palabra "profundo" no es solo mercadeo: se refiere literalmente al **número de capas** que atraviesa la información entre la entrada y la salida de la red.

- Una red con **una sola capa** (perceptrón simple) solo puede aprender fronteras de decisión **lineales**: no puede resolver problemas donde las clases no son separables con una línea recta.
- Al agregar **capas ocultas** con funciones de activación no lineales, la red puede combinar fronteras simples para construir fronteras de decisión arbitrariamente complejas — esto es lo que permite "aprender" representaciones jerárquicas (bordes → formas → objetos, en visión; letras → palabras → sintaxis, en texto).

El ejemplo clásico para demostrar esta limitación —y su solución— es el problema **XOR**, que vamos a resolver a continuación construyendo una red neuronal desde cero, sin ningún framework.

## 5. Laboratorio: Una Red Neuronal Mínima en NumPy Puro — El Problema XOR 💻

El operador lógico XOR ("o exclusivo") devuelve 1 solo cuando sus dos entradas son diferentes. Es el ejemplo histórico (Minsky & Papert, 1969) que mostró que un **perceptrón de una sola capa no puede resolver problemas no linealmente separables** — ninguna línea recta separa las clases en el plano.

Vamos a implementar, usando exclusivamente `numpy` (sin TensorFlow, PyTorch ni scikit-learn), una red neuronal con **una capa oculta** que sí resuelve XOR, para comprobar tangiblemente por qué la "profundidad" importa.

````python
import numpy as np

# Tabla de verdad de XOR: el problema que un perceptrón simple NO puede resolver
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

print("Entradas (X):\n", X)
print("\nSalida esperada XOR (y):\n", y.ravel())
print("\nNota: no existe una sola línea recta en el plano 2D que separe")
print("las clases 0 y 1 de este conjunto -> por eso un perceptrón simple falla.")
````

````python
def sigmoide(z):
    return 1 / (1 + np.exp(-z))

def derivada_sigmoide(a):
    # 'a' ya es sigmoide(z); la derivada se expresa en función de la propia salida
    return a * (1 - a)

rng = np.random.default_rng(7)
n_entrada, n_oculta, n_salida = 2, 4, 1

# Pesos y sesgos inicializados aleatoriamente (pequeños, para estabilidad numérica)
W1 = rng.normal(0, 1, (n_entrada, n_oculta))
b1 = np.zeros((1, n_oculta))
W2 = rng.normal(0, 1, (n_oculta, n_salida))
b2 = np.zeros((1, n_salida))

def forward(X):
    z1 = X @ W1 + b1
    a1 = sigmoide(z1)          # capa oculta: aquí ocurre la "magia" no lineal
    z2 = a1 @ W2 + b2
    a2 = sigmoide(z2)          # capa de salida
    return a1, a2

print("Arquitectura: 2 entradas -> 4 neuronas ocultas -> 1 salida")
print("W1 shape:", W1.shape, "| W2 shape:", W2.shape)
````

````python
tasa_aprendizaje = 0.5
epocas = 10000

for epoca in range(epocas):
    # --- Forward ---
    a1, a2 = forward(X)

    # --- Error (MSE) ---
    error = a2 - y

    # --- Backward (regla de la cadena, implementada a mano) ---
    delta2 = error * derivada_sigmoide(a2)
    dW2 = a1.T @ delta2
    db2 = delta2.sum(axis=0, keepdims=True)

    delta1 = (delta2 @ W2.T) * derivada_sigmoide(a1)
    dW1 = X.T @ delta1
    db1 = delta1.sum(axis=0, keepdims=True)

    # --- Actualización de parámetros (descenso de gradiente) ---
    W2 -= tasa_aprendizaje * dW2
    b2 -= tasa_aprendizaje * db2
    W1 -= tasa_aprendizaje * dW1
    b1 -= tasa_aprendizaje * db1

    if epoca % 2000 == 0:
        perdida = np.mean(error ** 2)
        print(f"Época {epoca:5d} | Pérdida (MSE): {perdida:.4f}")

_, predicciones = forward(X)
print("\nPredicciones finales (probabilidad):", predicciones.round(3).ravel())
print("Clasificación final (umbral 0.5):    ", (predicciones > 0.5).astype(int).ravel())
print("Valores esperados:                   ", y.ravel())
````

````python
# Contraste: un perceptrón de UNA sola capa (SIN capa oculta) intentando aprender XOR
W_simple = rng.normal(0, 1, (n_entrada, n_salida))
b_simple = np.zeros((1, n_salida))

for epoca in range(epocas):
    a_simple = sigmoide(X @ W_simple + b_simple)
    error_simple = a_simple - y
    delta_simple = error_simple * derivada_sigmoide(a_simple)
    W_simple -= tasa_aprendizaje * (X.T @ delta_simple)
    b_simple -= tasa_aprendizaje * delta_simple.sum(axis=0, keepdims=True)

pred_simple = sigmoide(X @ W_simple + b_simple)
print("Perceptrón SIN capa oculta -> predicciones:", pred_simple.round(3).ravel())
print("Clasificación:", (pred_simple > 0.5).astype(int).ravel(), " | Esperado:", y.ravel())
print("\n-> El perceptrón simple no logra converger a la solución correcta:")
print("   confirma que se necesita al menos una capa oculta ('profundidad')")
print("   para resolver un problema no linealmente separable como XOR.")
````

## 6. Relación con los Círculos Anidados 🔄

Este laboratorio ilustra exactamente por qué DL es un subconjunto especializado de ML, y no un simple sinónimo:

- **El perceptrón simple** (una sola capa, sin capa oculta) es un modelo de **Machine Learning clásico**: puede aprender relaciones lineales, pero se queda corto ante problemas como XOR.
- **La red con una capa oculta** ya es, técnicamente, una red neuronal "profunda" en miniatura: al apilar transformaciones no lineales, logra separar clases que ninguna línea recta podría separar.
- Los modelos de Deep Learning modernos (visión por computador, modelos de lenguaje) llevan esta misma idea —apilar capas para construir representaciones cada vez más abstractas— a arquitecturas con docenas o cientos de capas y miles de millones de parámetros.

La diferencia entre "un perceptrón" y "una red profunda" no es de naturaleza, sino de **grado**: más capas, más parámetros, más capacidad de representar funciones complejas — a costa de más datos, más cómputo y menos interpretabilidad.

## Resumen / Puntos Clave 📌

- IA, ML y DL son **conjuntos anidados** (IA ⊃ ML ⊃ DL), no sinónimos intercambiables.
- El ML aprende patrones desde datos con ingeniería de características manual; el DL automatiza esa extracción de características mediante capas apiladas.
- La elección del enfoque depende de: volumen de datos, tipo de datos (tabular vs. imagen/audio/texto), necesidad de interpretabilidad y recursos computacionales disponibles.
- "Profundo" se refiere literalmente al número de capas de una red neuronal: más capas permiten aprender fronteras de decisión no lineales cada vez más complejas.
- El problema XOR demuestra, con código real y sin frameworks, por qué un perceptrón de una sola capa es insuficiente y por qué agregar una capa oculta resuelve el problema.
