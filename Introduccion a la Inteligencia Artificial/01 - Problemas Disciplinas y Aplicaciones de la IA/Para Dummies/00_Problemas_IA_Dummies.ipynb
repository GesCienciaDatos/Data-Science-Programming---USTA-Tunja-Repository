## 1. El Doctor Robot: Cómo la IA Aprendió a "Diagnosticar" 🩺🤖

Imagina un doctor recién graduado que en lugar de memorizar libros enteros, lleva una **libretica con reglas** escritas por los mejores médicos del mundo:

> "Si el paciente tiene fiebre alta Y tos con flema Y dolor en el pecho... probablemente sea neumonía."

Eso, ni más ni menos, es lo que hacía **MYCIN**, uno de los primeros "doctores robot" de la historia (años 70). No pensaba como un humano: solo **revisaba su libretica de reglas** muy, muy rápido, y le decía al médico de verdad: "yo creo que es esto, con tal nivel de seguridad, pero tú decides".

Vamos a construir nuestra propia mini-libretica de reglas médicas en código.

````python
# Nuestra "libretica de reglas" del doctor robot 📖
# Cada regla dice: SI el paciente tiene estos síntomas, ENTONCES sospecha de esta enfermedad

libretica_del_doctor = [
    {"sintomas": {"fiebre", "tos_con_flema", "dolor_pecho"}, "sospecha": "Neumonía", "seguridad": 0.85},
    {"sintomas": {"fiebre", "tos_seca", "sin_olfato"}, "sospecha": "Gripe/Virus", "seguridad": 0.75},
    {"sintomas": {"fiebre", "dolor_cabeza_fuerte", "cuello_rigido"}, "sospecha": "¡Alerta! Posible meningitis", "seguridad": 0.90},
]

# Síntomas del paciente de hoy
paciente_de_hoy = {"fiebre", "tos_con_flema", "dolor_pecho"}

print("🩺 El doctor robot está revisando su libretica...\n")
for regla in libretica_del_doctor:
    coincidencias = regla["sintomas"] & paciente_de_hoy
    if coincidencias:
        print(f"👉 Coincide con: {regla['sospecha']} (seguridad: {regla['seguridad']*100:.0f}%)")
````

### 🤔 ¿Qué acaba de pasar?

El "doctor robot" no adivinó nada: simplemente **comparó los síntomas del paciente contra su libretica de reglas** y avisó cuáles coincidían. Así de simple (y así de poderoso) es un sistema experto: conocimiento humano convertido en reglas que una máquina puede revisar en milisegundos.

## 2. El Cajero que Nunca Duerme: Supermercados y Cámaras Inteligentes 🛒👀

Cuando pasas tus productos por la caja de autopago, hay una **báscula escondida** debajo de la bolsa. Si el peso no coincide con lo que escaneaste... ¡algo anda raro! Puede ser un error, o puede ser que se te "olvidó" escanear algo.

Vamos a simular esa báscula inteligente.

````python
# Peso esperado de cada producto (en gramos)
pesos_productos = {"manzana": 180, "leche": 1030, "chocolate": 100}

# Un cliente escaneó estos productos...
escaneado = ["manzana", "manzana", "leche"]
peso_esperado = sum(pesos_productos[p] for p in escaneado)

# ...pero la báscula midió esto:
peso_real_bascula = 2400  # ¡mucho más de lo esperado!

diferencia = peso_real_bascula - peso_esperado
print(f"📦 Peso esperado según lo escaneado: {peso_esperado} g")
print(f"⚖️  Peso real medido por la báscula: {peso_real_bascula} g")

if abs(diferencia) > 150:
    print("🚨 ¡Algo no cuadra! Se llama a un empleado para revisar la bolsa.")
else:
    print("✅ Todo en orden, puedes pagar tranquilo.")
````

### 🤔 ¿Qué significa esto?

El sistema **no te acusa de nada**: solo nota que el peso no cuadra y pide ayuda humana. Así evita errores injustos y, al mismo tiempo, cuida el inventario de la tienda.

## 3. El Ajedrez: El Gimnasio Donde la IA Hizo Músculo ♟️💪

En 1997, una computadora llamada **Deep Blue** le ganó una partida al campeón mundial de ajedrez, Garry Kasparov. Fue una noticia mundial, ¡como si una máquina le hubiera ganado a Messi en fútbol!

¿Por qué el ajedrez y no, por ejemplo, cocinar una sopa? Porque el ajedrez tiene **reglas clarísimas** y un tablero que ambos jugadores ven por completo: es el escenario perfecto para que una máquina practique "pensar varios pasos adelante".

La técnica que usaba Deep Blue se llama **Minimax**: para cada jugada posible, la máquina se pregunta *"si yo juego esto, ¿cuál sería la peor respuesta de mi rival?"* y elige la jugada donde esa "peor respuesta" sea la menos mala de todas.

````python
# Simulamos 3 posibles jugadas del "robot ajedrecista" y el peor resultado que
# el rival le podría provocar en cada una (números inventados, solo para ilustrar).

jugadas_posibles = {
    "Mover_Torre":   [3, 5],   # los peores resultados posibles si elige esta jugada
    "Mover_Alfil":   [1, -2],
    "Mover_Reina":   [7, 4],
}

print("♟️  El robot ajedrecista analiza el peor escenario de cada jugada:\n")
peor_resultado_por_jugada = {}
for jugada, resultados_posibles in jugadas_posibles.items():
    peor_resultado_por_jugada[jugada] = min(resultados_posibles)
    print(f"{jugada:15s} -> en el peor caso, obtiene: {peor_resultado_por_jugada[jugada]}")

mejor_jugada = max(peor_resultado_por_jugada, key=peor_resultado_por_jugada.get)
print(f"\n✅ El robot elige '{mejor_jugada}': es la jugada con el 'peor caso' menos malo de todas.")
````

### 🤔 ¿Y esto de qué sirve en la vida real?

Ese mismo truco de "pensar el peor escenario antes de decidir" lo usan hoy los autos autónomos (¿qué pasa si el peatón cruza de repente?) y los sistemas financieros (¿qué pasa si el mercado cae?). El ajedrez fue solo el primer gimnasio.

## 4. ¿Cuándo SÍ Vale la Pena Construir un Sistema Experto? ✅

No siempre conviene construir un "doctor robot" para cada problema. Es como llamar a un plomero: solo tiene sentido si el problema es de plomería, si hay un plomero de verdad disponible, y si puede explicarte qué va a hacer. Estas son las 4 preguntas clave:

| Pregunta | En palabras simples |
|---|---|
| ¿El tema está bien delimitado? | ¿Es un problema específico, o "de todo un poco"? |
| ¿Hay expertos humanos disponibles? | ¿Existe alguien que ya sepa resolver esto bien? |
| ¿Se puede explicar en reglas simples? | ¿El experto puede decir "si pasa esto, haz esto otro"? |
| ¿Un error se puede corregir? | ¿Si la máquina se equivoca, pasa algo grave e irreversible? |

### 🎯 Tu turno: completa el checklist

Vamos a construir un pequeño "semáforo" que decide si conviene o no construir un sistema experto, según cuántas de las 4 preguntas se respondan con "sí".

````python
### TU CÓDIGO AQUÍ ###
# Completa la función: cuenta cuántas respuestas son True y decide el semáforo.
# Pista: 4 síes -> "🟢 Sí, constrúyelo"; 3 síes -> "🟡 Con supervisión humana"; menos -> "🔴 Mejor no, todavía"

def semaforo_sistema_experto(tema_delimitado, hay_expertos, se_explica_en_reglas, error_corregible):
    respuestas = [tema_delimitado, hay_expertos, se_explica_en_reglas, error_corregible]
    total_si = None       # <- cuenta cuántas son True
    veredicto = None      # <- decide el semáforo según el total
    return total_si, veredicto


# Probemos con el caso del doctor robot (Sección 1)
total_si, veredicto = semaforo_sistema_experto(
    tema_delimitado=True,
    hay_expertos=True,
    se_explica_en_reglas=True,
    error_corregible=True,
)
print(f"Respuestas afirmativas: {total_si}/4 -> {veredicto}")
````

```python
        veredicto = "🟡 Con supervisión humana"
    else:
        veredicto = "🔴 Mejor no, todavía"

    return total_si, veredicto
```

Con las 4 respuestas en `True` (como el doctor robot MYCIN), el semáforo queda en verde. Si cambiamos, por ejemplo, `error_corregible=False`, el semáforo baja a amarillo: la IA puede ayudar, pero siempre con un humano vigilando de cerca.
