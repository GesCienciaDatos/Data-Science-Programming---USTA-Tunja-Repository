## 1. La Banda de Rock de la IA 🎸

Imagina una banda de rock donde cada integrante viene de un mundo distinto:

| Integrante | Lo que aporta a la banda (IA) |
|---|---|
| 🧮 **Matemáticas** | El bajista: pone el ritmo y la estructura (fórmulas, probabilidad). |
| 🧠 **Neurociencia** | El baterista: inspira el "pulso" biológico (cómo funciona el cerebro). |
| 🗣️ **Lingüística** | El vocalista: le da voz y lenguaje a la banda (cómo hablan los chatbots). |
| 🤔 **Filosofía** | El compositor: se pregunta el "por qué" de cada canción (¿puede pensar una máquina?). |
| 💰 **Economía** | El manager: decide qué conviene más hacer (elegir la mejor decisión posible). |
| ⚙️ **Ingeniería de Control** | El técnico de sonido: ajusta todo en tiempo real para que no se desafine (robots que se autocorrigen). |

Sin cualquiera de estos integrantes, ¡la banda sonaría incompleta! Así de multidisciplinaria es la IA.

````python
# El "manager de la banda" (economía) decide qué conviene hacer, comparando
# las ganancias posibles de cada opción según qué tan probable es cada una.

opciones_de_gira = {
    "Tocar_en_Tunja":  [(0.8, 5), (0.2, -1)],   # alta probabilidad de éxito, ganancia moderada
    "Tocar_en_Bogota": [(0.5, 15), (0.5, -3)],  # arriesgado, pero puede pagar mucho más
}


def ganancia_esperada(resultados_posibles):
    return sum(probabilidad * ganancia for probabilidad, ganancia in resultados_posibles)


print("💰 El manager evalúa qué conviene más:\n")
for opcion, resultados in opciones_de_gira.items():
    print(f"{opcion:18s} -> Ganancia esperada: {ganancia_esperada(resultados):+.1f}")

mejor_opcion = max(opciones_de_gira, key=lambda o: ganancia_esperada(opciones_de_gira[o]))
print(f"\n🎤 El manager recomienda: '{mejor_opcion}'")
````

### 🤔 ¿Y esto qué tiene que ver con IA?

Así piensan muchos sistemas de IA: no adivinan al azar, sino que **calculan qué opción conviene más** sumando lo bueno y lo malo de cada posibilidad, exactamente como haría un manager decidiendo la próxima ciudad de la gira.

## 2. ¿Por Qué Todo el Mundo Habla de IA? La Nueva Electricidad 💡

Hace 150 años, la electricidad cambió absolutamente todo: la fábrica, el hospital, la casa, la escuela. No era "solo una tecnología más", era algo que **se metía en todos lados**. Hoy en día, muchos expertos dicen que la IA está haciendo exactamente lo mismo.

| ¿Dónde se nota? | Ejemplo cotidiano |
|---|---|
| 💼 En el trabajo | Menos tareas repetitivas, más tiempo para pensar. |
| 🏥 En la salud | Diagnósticos más rápidos, incluso en pueblos sin especialistas. |
| 🔬 En la ciencia | Descubrimientos que antes tomaban años, ahora toman días. |

**En pocas palabras:** así como nadie hoy diría "yo no uso electricidad", en unos años será raro decir "yo no uso IA en nada".

````python
# Un cálculo sencillo: ¿cuánto "mejora" cada área de la vida al usar IA?
# (números inventados solo para ilustrar la idea de impacto transversal)

areas_de_vida = {
    "Trabajo":    {"antes": 100, "mejora_pct": 0.18},
    "Salud":      {"antes": 100, "mejora_pct": 0.22},
    "Educacion":  {"antes": 100, "mejora_pct": 0.12},
}

print("📈 ¿Cuánto mejora cada área al incorporar IA?\n")
for area, datos in areas_de_vida.items():
    despues = datos["antes"] * (1 + datos["mejora_pct"])
    print(f"{area:10s}: antes {datos['antes']:.0f} -> con IA {despues:.1f}  (+{datos['mejora_pct']*100:.0f}%)")

print("\n💡 Nota cómo la mejora aparece en áreas que no tienen nada que ver entre sí.")
print("Esa es la señal de una tecnología que, como la electricidad, se mete en todos lados.")
````
