## 1. La Navaja Suiza de la IA 🔧

| Lugar | ¿Qué hace la IA ahí? | Analogía |
|---|---|---|
| 🏥 Hospital | Ayuda a decidir a quién atender primero. | Una enfermera que revisa signos vitales antes de que llegue el médico. |
| 🏦 Banco | Detecta compras raras en tu tarjeta. | Un guardia que nota cuando algo no cuadra. |
| 🚗 Carretera | Ayuda al carro a "ver" y reaccionar. | Un copiloto que nunca se distrae. |
| 🎓 Escuela | Adapta las tareas a tu ritmo de aprendizaje. | Un tutor personal paciente. |
| 🎬 Cine/Streaming | Te recomienda qué ver según tus gustos. | Un amigo que sabe exactamente qué película te va a gustar. |

Vamos a explorar tres de estos lugares con código real y sencillo.

---
## 2. En el Hospital 🏥

Cuando llegan varios pacientes a urgencias al mismo tiempo, alguien debe decidir a quién atender primero. Un sistema de apoyo revisa los signos vitales y sugiere un orden, aunque **la decisión final siempre es del personal médico**.

````python
# Signos vitales de 2 pacientes que acaban de llegar
pacientes = [
    {"nombre": "Paciente 1", "pulso": 130, "oxigeno": 89},
    {"nombre": "Paciente 2", "pulso": 78,  "oxigeno": 97},
]

for p in pacientes:
    puntos = 0
    if p["pulso"] > 120:
        puntos += 2
    if p["oxigeno"] < 90:
        puntos += 2

    urgencia = "🔴 Atender primero" if puntos >= 3 else "🟢 Puede esperar un poco"
    print(f"{p['nombre']}: pulso={p['pulso']}, oxígeno={p['oxigeno']}% -> {urgencia}")
````

---
## 3. En el Banco 🏦

Si de repente alguien hace una compra gigante en una ciudad donde nunca has estado, el banco sospecha y te avisa.

````python
# Gasto habitual de un cliente
gasto_habitual = 80_000  # pesos COP, en promedio

# Nueva compra registrada
nueva_compra = 1_500_000

if nueva_compra > gasto_habitual * 3:
    print(f"🚨 Compra de ${nueva_compra:,} COP es mucho más alta de lo habitual (${gasto_habitual:,} COP).")
    print("El banco te envía un mensaje para confirmar que sí fuiste tú.")
else:
    print("✅ Compra dentro de lo normal, no se envía ninguna alerta.")
````

## 4. En la Carretera, la Escuela y el Cine 🚗🎓🎬

Los carros autónomos no son "todo o nada": existen **niveles** de autonomía, como un videojuego con distintas dificultades:

| Nivel | ¿Qué tanto conduce solo el carro? |
|---|---|
| 0 | Nada, el humano hace todo. |
| 1-2 | Ayuda con frenos o dirección, pero el humano sigue al mando. |
| 3 | Conduce solo en situaciones específicas (ej. atascos), pero el humano debe estar listo. |
| 4-5 | Conduce solo casi siempre o siempre, casi sin intervención humana. |

Y para cerrar, veamos una mini-recomendación de películas, el mismo tipo de IA que usan las plataformas de streaming.

````python
# Gustos de dos amigos por género de película (escala 1-5)
gustos = {
    "Tu_Amigo": {"Accion": 5, "Comedia": 2, "Terror": 4},
    "Tu":       {"Accion": 4, "Comedia": 1, "Terror": 5},
}

generos_comunes = gustos["Tu_Amigo"].keys() & gustos["Tu"].keys()
diferencia = sum(abs(gustos["Tu_Amigo"][g] - gustos["Tu"][g]) for g in generos_comunes)
parecido = round(1 / (1 + diferencia), 2)

print(f"🎬 Qué tan parecidos son sus gustos: {parecido} (más cerca de 1 = más parecidos)")

genero_favorito_amigo = max(gustos["Tu_Amigo"], key=gustos["Tu_Amigo"].get)
print(f"🍿 Como sus gustos se parecen, te recomendamos: películas de '{genero_favorito_amigo}'")
````
