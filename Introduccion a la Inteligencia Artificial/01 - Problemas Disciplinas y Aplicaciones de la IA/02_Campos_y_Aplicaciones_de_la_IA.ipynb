## 2. Salud: Del Diagnóstico Manual al Apoyo Algorítmico 🏥

En salud, la IA se usa principalmente para **priorizar la atención** y **apoyar (no reemplazar) el criterio médico**. Un ejemplo típico es un sistema de **triage automatizado**: un algoritmo que revisa los signos vitales de varios pacientes y sugiere un orden de atención según el riesgo estimado.

A continuación simulamos un mini sistema de triage basado en reglas de umbral (no un modelo entrenado, pero sí el mismo principio de clasificación de riesgo que emplean sistemas reales de apoyo clínico).

````python
# =============================================================================
# SIMULACIÓN: SISTEMA DE TRIAGE AUTOMATIZADO EN URGENCIAS (basado en reglas de umbral)
# =============================================================================

pacientes = [
    {"id": "P001", "frecuencia_cardiaca": 130, "saturacion_oxigeno": 89, "temperatura": 39.5},
    {"id": "P002", "frecuencia_cardiaca": 78,  "saturacion_oxigeno": 97, "temperatura": 36.8},
    {"id": "P003", "frecuencia_cardiaca": 105, "saturacion_oxigeno": 93, "temperatura": 38.2},
]


def calcular_riesgo(paciente):
    """Puntaje de riesgo simple: suma puntos por cada signo vital fuera de rango normal.
    Es una versión simplificada de los sistemas de alerta temprana (Early Warning Score)
    que usan hospitales reales como capa de apoyo, nunca como reemplazo del criterio clínico."""
    puntos = 0
    if paciente["frecuencia_cardiaca"] > 120:
        puntos += 3
    elif paciente["frecuencia_cardiaca"] > 100:
        puntos += 1

    if paciente["saturacion_oxigeno"] < 90:
        puntos += 3
    elif paciente["saturacion_oxigeno"] < 95:
        puntos += 1

    if paciente["temperatura"] > 39.0:
        puntos += 2
    elif paciente["temperatura"] > 38.0:
        puntos += 1

    return puntos


niveles = {0: "Verde (rutina)", 1: "Verde (rutina)", 2: "Amarillo (prioritario)",
           3: "Amarillo (prioritario)", 4: "Naranja (urgente)"}

resultados = []
for paciente in pacientes:
    riesgo = calcular_riesgo(paciente)
    nivel = niveles.get(riesgo, "Rojo (crítico, atención inmediata)")
    resultados.append((paciente["id"], riesgo, nivel))

resultados.sort(key=lambda r: r[1], reverse=True)

print("🏥 Orden sugerido de atención (de mayor a menor riesgo):\n")
for id_paciente, riesgo, nivel in resultados:
    print(f"{id_paciente} | Puntaje de riesgo: {riesgo} | Nivel: {nivel}")

print("\n⚠️ Recordatorio: este puntaje es solo una capa de apoyo. La decisión clínica final")
print("siempre corresponde al personal de salud, tal como vimos con MYCIN en el cuaderno anterior.")
````

## 3. Finanzas: Detección de Fraude en Tiempo Real 🏦

En el sector financiero, la IA opera sobre **millones de transacciones por segundo**, buscando patrones que se desvíen del comportamiento habitual de cada cliente. A continuación simulamos una versión simplificada de este tipo de sistema, comparando cada transacción contra el patrón histórico del cliente.

````python
# =============================================================================
# SIMULACIÓN: DETECCIÓN DE FRAUDE BASADA EN DESVIACIÓN DEL PATRÓN HISTÓRICO
# =============================================================================

# Patrón histórico de gasto promedio y ciudad habitual por cliente
perfil_historico = {
    "CLI-01": {"gasto_promedio": 80_000, "ciudad_habitual": "Tunja"},
    "CLI-02": {"gasto_promedio": 250_000, "ciudad_habitual": "Bogotá"},
}

# Transacciones nuevas a evaluar
transacciones_nuevas = [
    {"cliente": "CLI-01", "monto": 95_000, "ciudad": "Tunja"},
    {"cliente": "CLI-01", "monto": 1_800_000, "ciudad": "Miami"},   # fuera de patrón
    {"cliente": "CLI-02", "monto": 260_000, "ciudad": "Bogotá"},
]


def evaluar_transaccion(transaccion):
    perfil = perfil_historico[transaccion["cliente"]]
    monto_atipico = transaccion["monto"] > perfil["gasto_promedio"] * 3
    ciudad_atipica = transaccion["ciudad"] != perfil["ciudad_habitual"]

    if monto_atipico and ciudad_atipica:
        return "🚨 ALERTA ALTA: bloquear y verificar con el cliente"
    elif monto_atipico or ciudad_atipica:
        return "⚠️ ALERTA MEDIA: notificar al cliente por otro canal"
    else:
        return "✅ Transacción normal"


print("🏦 Evaluación de transacciones en tiempo real:\n")
for t in transacciones_nuevas:
    veredicto = evaluar_transaccion(t)
    print(f"Cliente {t['cliente']} | Monto: ${t['monto']:,} COP | Ciudad: {t['ciudad']:8s} | {veredicto}")
````

## 4. Transporte Autónomo: Niveles de Autonomía 🚗

La Sociedad de Ingenieros Automotrices (SAE International) define **6 niveles de autonomía vehicular**, un estándar que ayuda a entender qué tan "autónoma" es realmente una tecnología de conducción:

| Nivel SAE | Nombre | ¿Quién conduce? | Ejemplo |
|---|---|---|---|
| **0** | Sin automatización | El humano, 100% del tiempo. | Vehículo convencional sin asistencias. |
| **1** | Asistencia al conductor | El humano, con una asistencia puntual (frenado o dirección). | Control crucero adaptativo. |
| **2** | Automatización parcial | El humano supervisa; el sistema controla dirección y velocidad simultáneamente. | Autopilot en autopista, manos cerca del volante. |
| **3** | Automatización condicional | El sistema conduce en condiciones específicas; el humano debe estar listo para retomar el control. | Conducción autónoma en atascos de autopista. |
| **4** | Alta automatización | El sistema conduce solo dentro de un área o condición definida (*geofencing*), sin intervención humana. | Robotaxis en zonas urbanas delimitadas. |
| **5** | Automatización total | El sistema conduce en cualquier condición; no se requiere volante ni pedales. | Aún en fase experimental a gran escala. |

Detrás de estos niveles hay una combinación de **visión por computador** (cámaras), **sensores LiDAR y radar** (percepción del entorno) y **aprendizaje por refuerzo** (toma de decisiones de conducción entrenada mediante prueba y error en simulación).

## 5. Educación, Entretenimiento e Industria 4.0 🎓🎬🏭

* **Educación:** los **sistemas de tutoría inteligente** adaptan el ritmo y el contenido según el desempeño de cada estudiante, similar a como un profesor particular ajustaría su explicación.
* **Entretenimiento:** los **sistemas de recomendación** (streaming de música y video) analizan el comportamiento de consumo para sugerir contenido afín.
* **Industria 4.0:** el **mantenimiento predictivo** usa sensores IoT y detección de anomalías para anticipar fallas de maquinaria antes de que ocurran, evitando paradas de producción costosas.
* **Agricultura de precisión:** drones equipados con visión por computador monitorean cultivos para detectar plagas o déficit de riego en etapas tempranas.

A continuación simulamos un **motor de recomendación de contenido** muy simplificado, basado en similitud de intereses entre usuarios (el mismo principio detrás del filtrado colaborativo).

````python
# =============================================================================
# SIMULACIÓN: MOTOR DE RECOMENDACIÓN DE CONTENIDO (filtrado colaborativo simplificado)
# =============================================================================

# Calificaciones de usuarios a distintos géneros de contenido (escala 1-5)
calificaciones = {
    "Ana":    {"Ciencia_Ficcion": 5, "Documental": 4, "Comedia": 2},
    "Carlos": {"Ciencia_Ficcion": 4, "Documental": 5, "Comedia": 1},
    "Marta":  {"Ciencia_Ficcion": 1, "Documental": 2, "Comedia": 5},
}

usuario_objetivo = "Ana"


def similitud(usuario_a, usuario_b):
    """Similitud simplificada: distancia inversa entre calificaciones compartidas.
    A menor diferencia total, mayor similitud de gustos entre usuarios."""
    generos_comunes = calificaciones[usuario_a].keys() & calificaciones[usuario_b].keys()
    diferencia_total = sum(abs(calificaciones[usuario_a][g] - calificaciones[usuario_b][g]) for g in generos_comunes)
    return 1 / (1 + diferencia_total)  # normalizamos para que valores altos = más parecido


print(f"🎬 Buscando el usuario más parecido a '{usuario_objetivo}' para recomendarle contenido:\n")
similitudes = {}
for otro_usuario in calificaciones:
    if otro_usuario != usuario_objetivo:
        similitudes[otro_usuario] = round(similitud(usuario_objetivo, otro_usuario), 3)
        print(f"Similitud entre {usuario_objetivo} y {otro_usuario}: {similitudes[otro_usuario]}")

usuario_mas_parecido = max(similitudes, key=similitudes.get)
genero_top = max(calificaciones[usuario_mas_parecido], key=calificaciones[usuario_mas_parecido].get)

print(f"\n✨ Usuario más parecido a {usuario_objetivo}: {usuario_mas_parecido}")
print(f"📺 Recomendación para {usuario_objetivo}: contenido de género '{genero_top}' "
      f"(el favorito de {usuario_mas_parecido}, con calificación {calificaciones[usuario_mas_parecido][genero_top]}/5)")
````
