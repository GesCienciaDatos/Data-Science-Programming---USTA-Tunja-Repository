## 1. ¿Qué Hace que un Problema Sea "Apto" para la IA? 🧭

No todos los problemas se benefician de la Inteligencia Artificial. Un problema es un buen candidato cuando reúne, en mayor o menor medida, las siguientes condiciones:

| Característica | Descripción | Ejemplo |
|---|---|---|
| **Complejidad algorítmica alta** | No existe una fórmula cerrada o un procedimiento determinístico simple que lo resuelva. | Reconocer una cara en una foto no se puede programar con un `if`. |
| **Gran volumen de datos o de posibilidades** | El espacio de búsqueda o el volumen de información supera la capacidad de análisis manual. | Millones de transacciones bancarias diarias. |
| **Presencia de patrones (aunque ocultos)** | Existe una regularidad subyacente que puede aprenderse, aunque no sea evidente a simple vista. | Hábitos de compra que predicen fuga de clientes. |
| **Tolerancia razonable al error** | El costo de una predicción incorrecta es asumible y corregible, no catastrófico e irreversible. | Recomendar una película equivocada vs. diagnosticar mal una cirugía. |
| **Expertise humana replicable en reglas o ejemplos** | Existen expertos humanos cuyo conocimiento puede formalizarse como reglas o cuyo comportamiento puede aprenderse por ejemplos. | Médicos, jugadores de ajedrez, analistas de fraude. |

## 2. 1.6.1 Diagnóstico Médico: Los Primeros Sistemas Expertos 🩺

Uno de los primeros y más influyentes casos de éxito de la IA aplicada fue **MYCIN** (Universidad de Stanford, década de 1970), un sistema experto diseñado para **diagnosticar infecciones bacterianas de la sangre** y recomendar antibióticos.

### ¿Por qué el diagnóstico médico fue un problema tan atractivo para la IA temprana?

* Existían **expertos humanos** (médicos infectólogos) cuyo razonamiento podía traducirse en reglas del tipo `SI ... ENTONCES ...`.
* El dominio, aunque complejo, era **delimitable**: un conjunto acotado de bacterias, síntomas y tratamientos.
* Había **incertidumbre inherente** (los síntomas no siempre son concluyentes), lo que motivó a MYCIN a introducir los famosos **factores de certeza (*certainty factors*)**: números entre -1 y 1 que representan qué tan seguro está el sistema de una hipótesis.
* El costo de un mal diagnóstico es alto, por lo que MYCIN nunca decidía solo: **siempre explicaba su razonamiento** al médico, quien tomaba la decisión final.

| Componente de MYCIN | Función |
|---|---|
| **Base de conocimiento** | ~600 reglas de producción `SI síntomas/cultivos ENTONCES bacteria probable` capturadas de infectólogos expertos. |
| **Motor de inferencia** | Encadena reglas hacia atrás (*backward chaining*) para llegar a un diagnóstico a partir de la evidencia disponible. |
| **Factores de certeza** | Cuantifican la confianza de cada regla y combinan evidencia parcial o contradictoria. |
| **Módulo de explicación** | Responde "¿por qué?" y "¿cómo?" llegó el sistema a una conclusión, clave para la confianza clínica. |

A continuación construimos una **maqueta simplificada** de este tipo de sistema experto, con reglas explícitas y factores de certeza, para diagnosticar un caso de infección respiratoria de ejemplo.

````python
# =============================================================================
# MAQUETA DE SISTEMA EXPERTO MÉDICO (inspirado en MYCIN)
# Reglas explícitas tipo SI-ENTONCES con factores de certeza (CF entre 0 y 1)
# =============================================================================

# Base de conocimiento: cada regla asocia un conjunto de síntomas con una
# enfermedad candidata y un factor de certeza (qué tan confiable es la regla).
base_de_reglas = [
    {
        "id": "R1",
        "si": {"fiebre_alta", "tos_con_flema", "dolor_toracico"},
        "entonces": "Neumonía bacteriana",
        "cf": 0.85,
        "recomendacion": "Iniciar antibiótico de amplio espectro y radiografía de tórax.",
    },
    {
        "id": "R2",
        "si": {"fiebre_alta", "tos_seca", "perdida_olfato"},
        "entonces": "Infección viral respiratoria",
        "cf": 0.75,
        "recomendacion": "Aislamiento, manejo sintomático y prueba diagnóstica viral.",
    },
    {
        "id": "R3",
        "si": {"tos_con_flema", "sibilancias", "dificultad_respiratoria"},
        "entonces": "Exacerbación de asma / EPOC",
        "cf": 0.70,
        "recomendacion": "Broncodilatador y evaluación de función pulmonar.",
    },
    {
        "id": "R4",
        "si": {"fiebre_alta", "rigidez_cuello", "dolor_cabeza_intenso"},
        "entonces": "Sospecha de meningitis bacteriana",
        "cf": 0.90,
        "recomendacion": "URGENCIA: remitir de inmediato, no esperar más pruebas ambulatorias.",
    },
]


def diagnosticar(sintomas_paciente):
    """Motor de inferencia simplificado: evalúa cada regla contra los síntomas
    presentados y calcula un puntaje de coincidencia ponderado por el CF."""
    sintomas_paciente = set(sintomas_paciente)
    hipotesis = []
    for regla in base_de_reglas:
        interseccion = regla["si"] & sintomas_paciente
        cobertura = len(interseccion) / len(regla["si"])  # % de síntomas que coinciden
        if cobertura > 0:
            certeza_final = round(cobertura * regla["cf"], 2)
            hipotesis.append((regla["entonces"], certeza_final, regla["recomendacion"], regla["id"]))
    # Ordenamos de mayor a menor certeza, como haría el módulo de explicación de MYCIN
    hipotesis.sort(key=lambda h: h[1], reverse=True)
    return hipotesis


# --- Caso clínico de ejemplo ---
sintomas_caso = {"fiebre_alta", "tos_con_flema", "dolor_toracico", "dificultad_respiratoria"}

print("🩺 Síntomas reportados:", sintomas_caso)
print("\nDiagnósticos hipotéticos ordenados por certeza:\n")
for diagnostico, certeza, recomendacion, regla_id in diagnosticar(sintomas_caso):
    print(f"[{regla_id}] {diagnostico:35s} | Certeza: {certeza:.0%} | Recomendación: {recomendacion}")
````

### 🔍 ¿Qué acabamos de demostrar?

El código anterior **no es "inteligente" en un sentido mágico**: es un motor de reglas explícitas, exactamente como los primeros sistemas expertos médicos. Su valor no está en la sofisticación matemática, sino en que:

* **Captura conocimiento experto de forma explicable**: cada conclusión viene acompañada de la regla que la originó.
* **Cuantifica la incertidumbre** en lugar de dar respuestas binarias.
* **Prioriza casos críticos** (como la sospecha de meningitis) mediante factores de certeza altos.

Este es precisamente el tipo de problema apto para IA simbólica: dominio delimitado, expertos disponibles y reglas explicitables. Lo veremos formalizado como criterio general en la Sección 5.

## 3. 1.6.2 Supermercados y Cajeros Automáticos: Visión por Computador en la Vida Diaria 🛒

Otro terreno fértil para la IA temprana y contemporánea son los **entornos comerciales de alto volumen**: supermercados, cajas de autopago y sistemas antifraude.

### ¿Por qué estos problemas son aptos para IA?

* **Volumen masivo de transacciones**: ningún equipo humano puede revisar manualmente millones de compras diarias.
* **Patrones de compra repetitivos y aprendibles**: ciertos productos se compran juntos, ciertos horarios concentran ciertos perfiles de cliente.
* **Visión por computador** permite identificar productos en la banda del autopago (reconocimiento de patrones visuales) y detectar comportamientos anómalos, como productos "escaneados" que no coinciden con el peso registrado en la báscula (*bagging area*).
* **Tolerancia al error moderada**: un falso positivo de fraude genera una revisión humana, no una consecuencia irreversible, lo que hace manejable el riesgo.

| Aplicación | Técnica de IA típica | Qué patrón detecta |
|---|---|---|
| Cajero de autopago (*self-checkout*) | Visión por computador + verificación de peso | Discrepancia entre el producto escaneado y el peso físico registrado. |
| Recomendación de productos | Reglas de asociación / filtrado colaborativo | "Los clientes que compran X también compran Y". |
| Prevención de pérdidas (*shrinkage*) | Detección de anomalías | Patrones de escaneo inusuales (p. ej. "escaneo fallido intencional"). |
| Gestión de inventario | Series de tiempo / pronóstico de demanda | Estacionalidad y tendencias de consumo por sucursal. |

A continuación simulamos, con una tabla de compras y una regla de auditoría, cómo un sistema de autopago podría **señalar carritos sospechosos** para revisión humana.

````python
# =============================================================================
# SIMULACIÓN: DETECCIÓN DE CARRITOS SOSPECHOSOS EN CAJA DE AUTOPAGO
# Comparamos el peso esperado (según productos escaneados) contra el peso real
# medido por la báscula del área de empaque (bagging area).
# =============================================================================

# Catálogo simplificado del supermercado: peso promedio esperado por unidad (gramos)
catalogo_pesos = {
    "manzana": 180,
    "leche_1L": 1030,
    "arroz_1kg": 1000,
    "chocolate_barra": 100,
    "detergente_1L": 1100,
}

# Carritos de ejemplo escaneados por distintos clientes en cajas de autopago
carritos = [
    {
        "cliente": "Carrito_A",
        "items_escaneados": ["manzana", "manzana", "leche_1L"],
        "peso_bascula_g": 1395,  # muy cercano al esperado
    },
    {
        "cliente": "Carrito_B",
        "items_escaneados": ["arroz_1kg", "chocolate_barra"],
        "peso_bascula_g": 2450,  # sospechosamente más pesado de lo escaneado
    },
    {
        "cliente": "Carrito_C",
        "items_escaneados": ["detergente_1L"],
        "peso_bascula_g": 1090,
    },
]

TOLERANCIA_GRAMOS = 150  # margen de error aceptable del sensor de báscula

for carrito in carritos:
    peso_esperado = sum(catalogo_pesos[item] for item in carrito["items_escaneados"])
    diferencia = carrito["peso_bascula_g"] - peso_esperado
    sospechoso = abs(diferencia) > TOLERANCIA_GRAMOS

    estado = "🚨 REVISIÓN HUMANA REQUERIDA" if sospechoso else "✅ Peso conforme"
    print(f"{carrito['cliente']:12s} | Esperado: {peso_esperado:5d} g | Báscula: {carrito['peso_bascula_g']:5d} g | "
          f"Diferencia: {diferencia:+6d} g | {estado}")
````

### 🔍 Lectura del resultado

El `Carrito_B` dispara una alerta porque el peso real supera por mucho lo que los productos escaneados deberían pesar: probablemente hay un artículo sin escanear en la bolsa. Nótese que **el sistema no acusa a nadie de robar**: solo señala una discrepancia estadística para que una persona la revise. Esta separación entre *detección automática* y *decisión humana final* es un patrón recurrente en aplicaciones de IA de alto impacto social.

## 4. 1.6.3 El Ajedrez como Laboratorio de IA ♟️

El ajedrez ocupa un lugar histórico único en la investigación de IA: durante más de 40 años fue el **"banco de pruebas" (testbed)** preferido para validar algoritmos de búsqueda y toma de decisiones.

### ¿Por qué el ajedrez?

1. **Reglas perfectamente definidas y de información completa**: a diferencia del mundo real, ambos jugadores conocen exactamente el estado del tablero en todo momento. Esto simplifica el problema y permite compararlo matemáticamente.
2. **Espacio de búsqueda enorme pero acotado**: se estima que el "árbol de juego" del ajedrez tiene más de $10^{120}$ posibles partidas (el número de Shannon), suficientemente grande para exigir algoritmos inteligentes, pero formalmente definido.
3. **Métrica de éxito objetiva**: ganar, perder o empatar. No hay ambigüedad en la evaluación del desempeño, algo raro en problemas de IA aplicados al mundo real.
4. **Hito simbólico y mediático**: cuando **Deep Blue de IBM venció al campeón mundial Garry Kasparov en 1997**, se consolidó ante el público la idea de que las máquinas podían superar a los mejores expertos humanos en tareas cognitivas complejas.

### El algoritmo detrás de Deep Blue: búsqueda Minimax

Deep Blue no "entendía" el ajedrez como un humano: **exploraba millones de jugadas posibles por segundo** usando el algoritmo **Minimax** (con optimización *alpha-beta pruning*):

* En cada turno, el algoritmo simula: *"si yo juego esto, ¿qué jugaría mi oponente para minimizar mi ventaja?"*
* Construye un árbol de jugadas futuras y elige la que **maximiza el peor escenario posible** (de ahí "mini-max": minimizar la pérdida máxima).

```mermaid
graph TD
    A["Turno de la Máquina (MAX)"] --> B1[Jugada 1]
    A --> B2[Jugada 2]
    A --> B3[Jugada 3]
    B1 --> C1["Respuesta rival (MIN): peor caso = 2"]
    B2 --> C2["Respuesta rival (MIN): peor caso = 5"]
    B3 --> C3["Respuesta rival (MIN): peor caso = 1"]
    C2 --> D["✅ Se elige Jugada 2: maximiza el peor caso (5 > 2 > 1)"]
```

Del ajedrez, la IA "se graduó" hacia juegos con espacios de búsqueda aún mayores (Go, con AlphaGo en 2016) y, después, hacia problemas abiertos del mundo real, donde ya no hay reglas perfectas ni información completa.

````python
# =============================================================================
# ALGORITMO MINIMAX SOBRE UN ÁRBOL DE JUEGO SIMPLIFICADO
# No es ajedrez real (el árbol real es intratable a mano), pero ilustra
# exactamente el mismo principio de decisión que usó Deep Blue.
# =============================================================================

# Árbol de juego de juguete: cada lista de listas representa las jugadas
# posibles y, en las hojas, el "valor" final de la posición para la máquina
# (valores más altos = mejor para la máquina que juega como MAX).
arbol_de_juego = [
    [[3, 5], [6, 9]],   # rama si la máquina juega "Jugada 1"
    [[1, 2], [0, -1]],  # rama si la máquina juega "Jugada 2"
    [[7, 4], [8, 2]],   # rama si la máquina juega "Jugada 3"
]


def minimax(nodo, profundidad, es_turno_max):
    """Minimax recursivo clásico: alterna entre maximizar (la máquina)
    y minimizar (el rival) en cada nivel del árbol de jugadas."""
    # Caso base: llegamos a un valor hoja (fin de la simulación)
    if isinstance(nodo, (int, float)):
        return nodo

    if es_turno_max:
        return max(minimax(hijo, profundidad + 1, False) for hijo in nodo)
    else:
        return min(minimax(hijo, profundidad + 1, True) for hijo in nodo)


print("Evaluando el árbol de decisión con Minimax (como haría un motor de ajedrez simplificado):\n")
mejores_valores = []
for i, jugada in enumerate(arbol_de_juego, start=1):
    valor = minimax(jugada, 0, es_turno_max=False)  # el rival responde minimizando
    mejores_valores.append(valor)
    print(f"Jugada {i}: en el peor de los casos (si el rival juega óptimo), el resultado es {valor}")

mejor_jugada = mejores_valores.index(max(mejores_valores)) + 1
print(f"\n♟️  Decisión Minimax: la máquina elige la Jugada {mejor_jugada} "
      f"(garantiza el mejor resultado posible en el peor escenario: {max(mejores_valores)}).")
````

## 5. 1.6.4 ¿Qué Tipos de Problemas Son Aptos para Sistemas Expertos? 🧠

Después de revisar tres casos concretos, podemos consolidar los **criterios formales** que la literatura clásica de IA (Feigenbaum, Buchanan y otros pioneros de los sistemas expertos en Stanford) utiliza para decidir si un problema es candidato a resolverse con un sistema experto basado en reglas:

| Criterio | Pregunta clave | Ejemplo que lo cumple | Ejemplo que lo incumple |
|---|---|---|---|
| **Dominio bien delimitado** | ¿El problema tiene fronteras claras y un vocabulario acotado? | Diagnóstico de infecciones bacterianas específicas. | "Dar consejos de vida en general". |
| **Expertos humanos disponibles** | ¿Existen personas que resuelven bien este problema hoy y pueden articular *cómo* lo hacen? | Infectólogos, maestros de ajedrez, analistas de fraude. | Tareas donde nadie logra explicar su intuición. |
| **Conocimiento explicitable en reglas** | ¿El razonamiento experto puede escribirse como reglas `SI-ENTONCES` o heurísticas? | "Si hay rigidez de cuello y fiebre alta, sospechar meningitis". | Reconocer el "estilo" de un pintor a simple vista. |
| **Tolerancia razonable al error** | ¿Un error del sistema es detectable, corregible y no catastrófico si hay supervisión humana? | Sugerencia de diagnóstico revisada por un médico. | Decisiones totalmente autónomas sin supervisión en contextos de vida o muerte. |
| **Tarea repetitiva o de alto volumen** | ¿Vale la pena automatizar por la frecuencia o escala del problema? | Millones de transacciones de tarjeta al día. | Una decisión estratégica única e irrepetible. |


````python
def evaluar_aptitud_ia(dominio_delimitado, expertos_disponibles,
````

## Resumen / Puntos Clave 📌

* Un problema es apto para IA cuando combina **complejidad algorítmica**, **datos o expertise disponible**, **patrones aprendibles** y **tolerancia razonable al error**.
* **1.6.1 Diagnóstico médico (MYCIN):** los primeros sistemas expertos formalizaron el conocimiento médico en reglas con factores de certeza, siempre bajo supervisión humana final.
* **1.6.2 Supermercados y cajeros automáticos:** la visión por computador y la detección de anomalías permiten operar a un volumen que ningún equipo humano podría auditar manualmente.
* **1.6.3 El ajedrez:** gracias a sus reglas completas y su espacio de búsqueda gigantesco pero formal, fue el laboratorio ideal para validar algoritmos como **Minimax**, y el triunfo de Deep Blue sobre Kasparov (1997) marcó un hito histórico de la IA.
* **1.6.4 Criterios para sistemas expertos:** dominio delimitado, expertos disponibles, conocimiento explicitable en reglas y tolerancia al error son las cuatro preguntas que todo científico de datos debería hacerse antes de construir un sistema basado en IA.
