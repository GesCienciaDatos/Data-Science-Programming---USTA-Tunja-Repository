## ¿Qué vamos a aprender aquí? 🎈

En el cuaderno anterior conocimos la oficina de *Constructora Horizonte Andino* y los conceptos de "pendientes por revisar" y "ya revisados". Ahora vamos a **programar de verdad** dos formas distintas de recorrer esa oficina buscando `informe_final.pdf`, sin ninguna pista, sólo abriendo carpetas:

* 🕳️ **DFS** ("Depth-First Search"): te metes hasta el fondo de un armario antes de mirar el siguiente.
* 🌊 **BFS** ("Breadth-First Search"): miras un poquito de cada armario antes de profundizar en ninguno.

Vamos a construir la oficina completa (más grande que la del cuaderno anterior) y a comparar cómo busca cada uno.

````python
# La oficina completa de Constructora Horizonte Andino, como un diccionario de Python.
# Cada carpeta es un diccionario; cada archivo es una llave apuntando a None.

oficina = {
    "Gerencia": {
        "Actas": {"acta_enero.docx": None, "acta_febrero.docx": None},
        "Estrategia": {"plan_2026.pptx": None, "presupuesto.xlsx": None},
    },
    "Finanzas": {
        "Contabilidad": {"balance.xlsx": None, "Auditorias": {"auditoria_2024.pdf": None}},
        "Tesoreria": {"flujo_caja.xlsx": None},
    },
    "ProyectosIngenieria": {
        "ProyectoNorte": {
            "planos.dwg": None,
            "Informes": {"informe_avance.pdf": None, "informe_final.pdf": None},  # <- ¡aquí está!
        },
        "ProyectoSur": {
            "Informes": {"informe_preliminar.pdf": None},
        },
    },
    "RecursosHumanos": {"nomina.xlsx": None, "contrato_empleado.pdf": None},
}

print("La oficina completa tiene", len(oficina), "armarios principales:", list(oficina.keys()))
````

---
## 1. DFS: Meterse Hasta el Fondo Primero 🕳️

DFS funciona como una **pila de carpetas en tus manos**: abres la última carpeta que agarraste, y sólo cuando esa rama se acaba, vuelves a la anterior. En Python, eso se programa con una **lista que usamos como pila** (agregamos y quitamos siempre del mismo extremo, el final).

````python
def buscar_dfs(mapa, nombre_objetivo):
    """DFS: se mete hasta el fondo de cada carpeta antes de probar la siguiente."""
    pila = [([nombre], nombre, contenido) for nombre, contenido in mapa.items()]
    orden_visitado = []

    while pila:
        camino, nombre_actual, contenido_actual = pila.pop()  # <- saca el ÚLTIMO agregado
        orden_visitado.append(nombre_actual)

        if nombre_actual == nombre_objetivo:
            return camino, orden_visitado

        if contenido_actual is not None:  # es una carpeta, tiene más cosas adentro
            for nombre_hijo, contenido_hijo in contenido_actual.items():
                pila.append((camino + [nombre_hijo], nombre_hijo, contenido_hijo))

    return None, orden_visitado


objetivo = "informe_final.pdf"
camino_dfs, orden_dfs = buscar_dfs(oficina, objetivo)

print("🕳️ Resultado de DFS:")
print("   Camino encontrado:", " → ".join(camino_dfs))
print(f"   Carpetas/archivos revisados: {len(orden_dfs)}")
print("   Orden en que se revisaron:", orden_dfs)
````

---
## 2. BFS: Un Poquito de Cada Armario Primero 🌊

BFS funciona como una **fila de espera** (el primero en llegar es el primero en atenderse): revisamos TODOS los armarios de un mismo nivel antes de asomarnos a las carpetas que están un nivel más adentro.

````python
from collections import deque

def buscar_bfs(mapa, nombre_objetivo):
    """BFS: revisa todo un nivel antes de profundizar al siguiente."""
    cola = deque([([nombre], nombre, contenido) for nombre, contenido in mapa.items()])
    orden_visitado = []

    while cola:
        camino, nombre_actual, contenido_actual = cola.popleft()  # <- saca el PRIMERO agregado
        orden_visitado.append(nombre_actual)

        if nombre_actual == nombre_objetivo:
            return camino, orden_visitado

        if contenido_actual is not None:
            for nombre_hijo, contenido_hijo in contenido_actual.items():
                cola.append((camino + [nombre_hijo], nombre_hijo, contenido_hijo))

    return None, orden_visitado


camino_bfs, orden_bfs = buscar_bfs(oficina, objetivo)

print("🌊 Resultado de BFS:")
print("   Camino encontrado:", " → ".join(camino_bfs))
print(f"   Carpetas/archivos revisados: {len(orden_bfs)}")
print("   Orden en que se revisaron:", orden_bfs)
````

### 🤔 ¿Qué acaba de pasar?
Fíjate que **ambos encontraron el mismo camino** hacia `informe_final.pdf` (¡tenía que ser así, sólo hay un camino posible en esta oficina!), pero probablemente uno de los dos tuvo que revisar **más** carpetas y archivos que el otro antes de encontrarlo. Eso depende de en qué posición está la carpeta correcta dentro de la lista de armarios.

## 🎯 Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **DFS** | Usa una **pila**: se mete hasta el fondo de una carpeta antes de probar la siguiente. Gasta poca memoria. |
| **BFS** | Usa una **fila (cola)**: revisa todo un nivel completo antes de profundizar. Puede gastar mucha memoria. |
| **En una oficina sin atajos** | DFS y BFS siempre encuentran el **mismo camino** (sólo existe uno), pero pueden revisar una cantidad distinta de carpetas para llegar a él. |
| **Ninguno usa el nombre del archivo** | Ambos son búsquedas "a ciegas": no les importa si buscas `informe_final.pdf` o `nomina.xlsx`, exploran exactamente en el mismo orden. |
