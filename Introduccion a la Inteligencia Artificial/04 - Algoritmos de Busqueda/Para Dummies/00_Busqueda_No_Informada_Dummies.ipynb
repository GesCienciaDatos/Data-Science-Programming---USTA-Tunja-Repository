## ¿Qué vamos a aprender aquí? 🎈

Imagina que trabajas en una oficina enorme, con archivadores, carpetas y sub-carpetas por todas partes, y tu jefe te pide: *"Búscame el informe final del Proyecto Norte"*. No hay ningún buscador mágico: tienes que abrir carpeta por carpeta hasta encontrarlo.

¡Eso es exactamente lo que hace una computadora cuando "busca" algo! En este cuaderno, sin fórmulas ni jerga complicada, entenderás:
1. Qué es una "frontera" y un nodo "explorado" (con una analogía de carpetas de oficina).
2. La diferencia entre buscar a ciegas y buscar siguiendo pistas.
3. Por qué a veces conviene ahorrar memoria y otras veces conviene ser más rápido.

---
## 1. La Gran Analogía: Buscando en la Oficina 🏢

Piensa en la empresa ficticia *Constructora Horizonte Andino*. Su oficina central tiene 4 grandes armarios: **Gerencia**, **Finanzas**, **Proyectos de Ingeniería** y **Recursos Humanos**. Dentro de "Proyectos de Ingeniería" hay más carpetas, y dentro de esas, más carpetas todavía... hasta llegar al papel que buscas: `informe_final.pdf`.

Cuando buscas así, en tu cabeza (o en un cuaderno de notas) llevas dos listas:

* 📋 **La lista de "pendientes por revisar"** (a esto los científicos de datos lo llaman **la frontera**): armarios y carpetas que ya viste, pero que todavía no has abierto.
* ✅ **La lista de "ya revisados"** (a esto lo llaman **el conjunto de explorados**): carpetas que ya abriste y de las que ya anotaste todo lo que había adentro, para no perder tiempo revisándolas dos veces.

Y por supuesto, tienes un objetivo: el papel exacto que estás buscando, al que llamamos el **estado meta**.

```python
# Una versión MUY simplificada de la oficina, para practicar con Python.
# (En el cuaderno 01 construiremos la versión completa y "de verdad".)

oficina = {
    "Gerencia": {"acta_reunion.docx": None, "presupuesto.xlsx": None},
    "Finanzas": {"factura_marzo.pdf": None, "recibo_pago.pdf": None},
    "ProyectosIngenieria": {
        "informe_avance.pdf": None,
        "informe_final.pdf": None,   # <-- ¡Lo que estamos buscando!
    },
    "RecursosHumanos": {"nomina.xlsx": None, "contrato_empleado.pdf": None},
}

print("Armarios disponibles en la oficina:", list(oficina.keys()))
```

### 🤔 ¿Qué acabamos de crear?
Un **diccionario de Python**, que aquí funciona como un mapa de la oficina: cada llave es el nombre de un armario o carpeta, y su contenido puede ser otro diccionario (más carpetas adentro) o `None` (que quiere decir "esto ya es un papel/archivo, no una carpeta").

---
## 2. Viendo la "Frontera" y los "Explorados" en Acción 🎬

Vamos a simular, paso a paso y a mano, cómo un buscador revisaría el primer armario.

```python
pendientes_por_revisar = ["Gerencia", "Finanzas", "ProyectosIngenieria", "RecursosHumanos"]
ya_revisados = []

# Revisamos el primer armario de la lista de pendientes
armario_actual = pendientes_por_revisar.pop(0)
ya_revisados.append(armario_actual)

contenido = list(oficina[armario_actual].keys())
print(f"📂 Abrimos el armario: '{armario_actual}'")
print(f"   Adentro encontramos: {contenido}")
print(f"\n📋 Pendientes por revisar ahora: {pendientes_por_revisar}")
print(f"✅ Armarios ya revisados: {ya_revisados}")
print(f"\n¿Encontramos 'informe_final.pdf' en este armario? {'informe_final.pdf' in contenido}")
```

## 3. Buscar a Ciegas vs. Buscar con Pistas 🕵️

Hay dos grandes formas de buscar:

| | 🙈 Buscar a Ciegas | 🔍 Buscar con Pistas |
|---|---|---|
| **¿Cómo decide qué abrir primero?** | Abre las carpetas en el orden en que las encuentra, sin fijarse en el nombre. | Se fija en los nombres: si busca "informe_final.pdf", primero abre la carpeta que se llama "Informes". |
| **Analogía** | Como cuando alguien nunca ha estado en la oficina y abre armario por armario al azar. | Como cuando un compañero con experiencia te dice "yo empezaría por la carpeta que dice Informes". |
| **En este módulo** | Lo veremos en el próximo cuaderno (**buscar de un tirón hasta el fondo** o **de a poquitos por todos lados**). | Lo veremos en los cuadernos 02 y 03, con "pistas" basadas en el nombre del archivo. |

En el cuaderno 01 aprenderemos las dos formas de buscar a ciegas: **DFS** (te metes hasta el fondo de un armario antes de mirar el siguiente) y **BFS** (miras un poquito de cada armario antes de profundizar en ninguno).

---
## 4. ¿Cuaderno Chiquito o Cuaderno Gigante? El Dilema de la Memoria ⚖️

Imagina que llevas un cuaderno de notas para anotar qué te falta por revisar.

* Si vas **de a poquitos por todos lados** (como BFS), tu cuaderno de pendientes puede llenarse RAPIDÍSIMO, porque anotas TODAS las carpetas de TODOS los armarios antes de profundizar en ninguna.
* Si vas **hasta el fondo de un armario primero** (como DFS), tu cuaderno se mantiene pequeño: sólo necesitas recordar el camino que llevas en ESE momento, no todas las alternativas que dejaste atrás.

```python
# ¿Qué tan rápido crece la lista de pendientes según cuántas carpetas
# tenga en promedio cada armario ("ramificación") y qué tan profundo busquemos?

carpetas_por_armario = 4
print(f"{'Qué tan profundo':<18} | {'Cuaderno GIGANTE (revisar todo un nivel)':<42} | {'Cuaderno CHIQUITO (una sola rama)'}")
print("-" * 90)
for nivel in range(1, 5):
    cuaderno_gigante = carpetas_por_armario ** nivel
    cuaderno_chiquito = carpetas_por_armario * nivel
    print(f"{nivel:<18} | {cuaderno_gigante:<42} | {cuaderno_chiquito}")

print("\n💡 El cuaderno gigante crece muchísimo más rápido: ¡ese es el precio de revisar todo, nivel por nivel!")
```

## 5. Cuidado con las Oficinas "Trampa" 🌀

Un último aviso antes del próximo cuaderno: imagina que la oficina fuera **infinita** (nunca se acaban las carpetas) o que tuviera **puertas giratorias** que te devuelven a una carpeta que ya visitaste, sin que te des cuenta. Si no llevas tu lista de "ya revisados" muy bien organizada, ¡podrías darte vueltas para siempre buscando el mismo papel sin nunca encontrarlo!

Por eso, todo buen algoritmo de búsqueda SIEMPRE lleva su lista de "ya revisados" (lo vimos en la Sección 2) para nunca abrir dos veces el mismo armario.

---
## 🎯 Resumen Relámpago ⚡

| Idea Clave | Explicación en 5 segundos |
|---|---|
| **Frontera** | La lista de "pendientes por revisar". |
| **Explorados** | La lista de "armarios ya revisados", para no repetir trabajo. |
| **Buscar a ciegas** | No usa ninguna pista del nombre de lo que buscas (DFS, BFS). |
| **Buscar con pistas** | Usa el nombre del archivo para decidir qué revisar primero (GBFS, A*). |
| **El dilema memoria/tiempo** | Ir "de a poquitos por todos lados" (BFS) gasta más cuaderno; ir "hasta el fondo" (DFS) gasta menos cuaderno, pero puede tardar más si se equivoca de armario. |
